```python

from google.colab import drive
from pathlib import Path
import pandas as pd

drive.mount("/content/gdrive")

records = [
    ("Sleep", "Ages 13-17: aim for 8-10 hours of sleep daily.",
     "CDC", "https://www.cdc.gov/sleep/about/"),
    ("Rest", "Aim for 1-2 days off weekly from sport-specific training and competition.",
     "AAP", "https://doi.org/10.1542/peds.2023-065129"),
    ("Training load", "Increase weekly training volume gradually.",
     "AAP", "https://doi.org/10.1542/peds.2023-065129"),
    ("Fatigue", "Persistent fatigue, mood changes, or declining performance warrant review for possible overtraining.",
     "AAP", "https://doi.org/10.1542/peds.2023-065129"),
    ("Injury", "Repetitive stress without adequate recovery can cause overuse injuries.",
     "AAP", "https://doi.org/10.1542/peds.2023-065129")
]

recovery_df = pd.DataFrame(
    records, columns=["topic", "guidance", "source", "source_url"]
)

path = Path("/content/gdrive/MyDrive/ChampionsAthleticMap/data/processed/recovery_guidance.csv")
recovery_df.to_csv(path, index=False)

print("Records saved:", len(recovery_df))
print("File exists:", path.exists())
display(recovery_df)
```
