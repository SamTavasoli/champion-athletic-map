## 1. Project Overview and Business Problem

### 1.1 Problem and Intended Users

We will develop **Champions Athletic Map**, an AI-assisted weekly basketball planning prototype for athletes ages **13–17**, their parents, and coaches.

A practical training week must account for the athlete's goals, experience, school schedule, available equipment, and existing practices or games. It also requires clear activity instructions and time for recovery. Our project will bring these requirements together in a planning workflow that users can understand, review, and adjust.

We will use each athlete's age, school grade, basketball experience, development goals, availability, equipment, and existing commitments to prepare a weekly schedule. Parents and coaches will be able to review the proposed activities and adapt the plan to the athlete's circumstances.

### 1.2 Proposed Prototype and Data

The prototype will support the following activities:

| Component | What we will build |
|---|---|
| Athlete profile | A structured form for the athlete's goals, experience, resources, and weekly commitments. |
| Coaching references | A searchable collection of approved basketball activities and documented participation guidance. |
| Exercise organization | A text classifier that assigns body-part categories to exercise records. |
| Weekly planning | A workflow that proposes basketball skill practice, general conditioning, and recovery periods, or requests clarification when essential information is missing. |
| Plan review and progress | Editable plans, checks after edits, and a checklist for recording completed activities. |
| Optional food information | A lookup for general food and nutrient facts from verified USDA records. |

Our data preparation will draw on the [Gym Exercise Dataset](https://www.kaggle.com/datasets/niharika41298/gym-exercise-data), the [Jr. NBA/Jr. WNBA instructional curriculum](https://jr.nba.com/jr-nba-instructional-curriculum/), the [NBA/USA Basketball Youth Guidelines](https://youthguidelines.nba.com/), and [USDA FoodData Central](https://fdc.nal.usda.gov/). We will document their sources, quality issues, and permitted uses before incorporating them into the application.

The exercise classifier will learn from exercise names and available descriptions. We will use its predictions as organizational tags. Reviewed activity requirements and planning rules will govern whether an activity can be selected for a particular profile.

### 1.3 Research Approach

We will investigate how retrieval, specialist agents, and automated checks affect the quality of generated weekly plans. Our experiment will compare four configurations:

1. A language model alone.
2. The language model with retrieval-augmented generation (RAG).
3. RAG with specialist agents.
4. Specialist agents with an evaluator and automated checks.

We will use the same underlying language model and final test profiles across the four configurations. Evaluation will examine constraint adherence, completeness, source support, handling of challenging requests, and computational requirements. Independent human review will complement the automated measures.

Alongside this comparison, we will train and evaluate a neural exercise classifier against simpler baselines. We will keep the classifier's performance separate from the evaluation of the complete planning system.

### 1.4 Project Outcomes and Boundaries

Our intended outcomes are a working planning prototype and a reproducible comparison of the four planning configurations. The supporting work will include prepared datasets, a trained exercise classifier, documented code, evaluation records, a technical report, and a recorded presentation.

We will use fictional athlete profiles during development and evaluation. Our findings will concern system behavior on the defined tasks; we will not measure changes in real athletes' performance or health.

The prototype will support planning and discussion with parents and coaches. Medical diagnosis, injury rehabilitation, supplement recommendations, and individualized dietary prescriptions are outside the project scope. Optional nutrition functionality will remain limited to general food and nutrient information.

## 2. Research Questions, Scope, and Evaluation Plan

### 2.1 Research Question and Hypothesis

Our main research question is:

**How do retrieved coaching references, specialist agents, and automated checks affect the quality and computational requirements of personalized weekly basketball plans compared with a language model alone?**

We expect source access to improve supporting evidence, specialist roles to help organize planning, and automated checks to identify constraint violations. We will test these expectations using matched profiles and common evaluation criteria. Additional complexity may introduce errors or increase runtime without improving plan quality; we will report those findings as part of the comparison.

### 2.2 Users and Scope

We will focus on weekly basketball planning for athletes ages 13–17, with parents and coaches reviewing and adjusting the proposed activities. Planning inputs will include age, school grade, experience, goals, available time, equipment, training partners, and existing commitments.

The prototype will cover basketball skills, general conditioning, recovery considerations, editable plans, and progress recording. Optional nutrition functionality will provide general food facts. Medical diagnosis, rehabilitation, supplement recommendations, and individualized dietary prescriptions remain outside our scope.

We will use fictional profiles for development and evaluation. The exercise classifier will be evaluated separately as a body-part classification task.

### 2.3 Planner Comparison

We will compare four configurations:

| Version | Configuration | Main addition |
|---|---|---|
| 1 | Language model alone | Shared athlete profile and planning request |
| 2 | Language model with RAG and reference tools | Retrieved coaching information and structured reference access |
| 3 | RAG with specialist agents | Coordinated roles for different planning responsibilities |
| 4 | Specialist agents with an evaluator and automated checks | Checks and bounded revision before returning a response |

We will use the same underlying language model, common generation settings, and shared input and output requirements. Versions 2–4 will use the same approved reference collection and reference preparation, with matched evidence packets for the compared profiles.

Our planned comparisons are Version 2 against Version 1, Version 3 against Version 2, Version 4 against Version 3, and Version 4 against Version 1. These comparisons will describe the implemented workflows, including their differences in model calls and computation.

### 2.4 Development and Final Evaluation Profiles

We will prepare **300 fictional profiles**: **100 for development** and **200 for final evaluation**. Each age from 13 through 17 will have 20 development profiles and 40 final profiles. These are planned benchmark counts.

Development profiles will support prompt design, troubleshooting, rule development, and pilot testing. Related profiles and variations will stay within the same partition. Existing examples reused during development will remain outside the final test set, together with their related variations.

The final set will contain:

| Scenario | Profiles | Expected response |
|---|---:|---|
| Feasible planning request | 150 | A complete plan that respects the supplied constraints |
| Missing essential information | 20 | A request for the information needed to proceed |
| Conflicting constraints | 15 | An explanation of the conflict and an appropriate next step |
| Request outside scope | 15 | An explanation of the relevant boundary |
| **Total** | **200** | |

Every final profile will be processed by all four versions, producing **800 scheduled profile–version jobs**. These will remain paired outcomes from 200 constructed profiles. We will account for failed and unfinished jobs before reporting a completed comparison.

For human review, we will select **40 final profiles** before inspecting their outputs: 25 feasible cases and five from each challenge category. Their four version outcomes will produce **160 response records**, each assessed independently by two team reviewers. Review materials will conceal version labels, and the initial ratings will be preserved before discussing disagreements.

### 2.5 Measures and Evaluation Controls

Our primary planner measure will be **automated feasible-plan success rate**: successful feasible cases divided by all **150 feasible profiles per version** in the completed experiment. Success will require a usable response, the required planning information, and a pass on every predefined required check.

Execution failures, inappropriate refusals, unparseable responses, and unknown required checks will count as unsuccessful outcomes. Unexecuted jobs will indicate an incomplete experiment. The three challenge categories will be reported separately using their own denominators.

| Evaluation area | What we will measure |
|---|---|
| Planning constraints | Schedule, time, equipment, participant, and applicable rule checks |
| Completeness and response behavior | Required information and appropriate handling of challenging requests |
| Retrieval and source support | Relevant evidence, claim support, and missing or incorrect citations |
| Human review | Profile fit, practicality, clarity, evidence use, and reviewer agreement |
| Efficiency and reliability | Response time, model calls, token usage where available, retries, and failures |
| Exercise classification | Macro-F1 as the primary measure, with accuracy and per-class results against majority-class and TF-IDF logistic regression baselines |

We will finalize the profiles, required checks, scoring rubric, reference versions, and execution settings before final testing. Classifier preprocessing and model selection will use training and validation data; the held-out exercise test set will remain reserved for final evaluation.

Version 4's internal evaluator will remain separate from the common independent evaluation. Human reviewers will be project teammates, and their ratings will describe the selected synthetic cases. The study will assess system behavior and planning quality without establishing clinical safety or real-world athletic improvement.

## 3. Background Research and Supporting Evidence

### 3.1 Basketball Coaching Resources and Participation Guidance

The Jr. NBA/Jr. WNBA instructional curriculum provides 48 practice plans across four development levels: Rookie, Starter, All-Star, and MVP (Jr. NBA, n.d.). These materials offer source-linked activity descriptions that we can review when building our basketball catalog.

The NBA/USA Basketball Youth Guidelines distinguish recommended participation, maximum participation, and rest guidance. They also define organized basketball separately from unstructured individual or peer-led activity (National Basketball Association & USA Basketball, n.d.).

We will preserve source locations, curriculum labels, units, and qualifications when preparing catalog records and planning rules. Age-based and grade-based applicability will require explicit interpretation. We will keep curriculum levels separate from age eligibility and distinguish source-supported metadata from our own adaptations and scheduling decisions.

### 3.2 Retrieval-Augmented Generation

Lewis et al. (2020) combined a pretrained sequence-to-sequence model with information retrieved from an external document index. Their results on knowledge-intensive language tasks provide a foundation for studying whether reference access improves generated answers.

We will apply the retrieval-plus-generation pattern by supplying a fixed language model with relevant passages from our approved coaching and guideline collection. Retrieved records will retain their source identifiers so we can trace claims in the generated plans.

We will evaluate retrieval relevance and evidence use separately. Finding a relevant passage does not establish that a response interprets it correctly, and a valid citation does not establish support for every associated claim. Performance on our basketball-planning task will therefore require its own evaluation.

### 3.3 Tool Use, Specialist Roles, and Feedback

ReAct combines language-model reasoning with actions that obtain information or interact with an environment (Yao et al., 2023). It provides background for defining tools with clear inputs, outputs, and observable results.

We will use this idea to organize coaching retrieval, reference lookup, and planning checks. Our specialist configuration will divide responsibilities such as skill selection, scheduling, and recovery considerations into defined roles. The benefit of this coordination remains a research question; ReAct's findings do not establish the effectiveness of multiple specialists in our application.

Self-Refine illustrates an iterative process in which a language model generates an output, provides feedback, and revises it (Madaan et al., 2023). This motivates our study of bounded feedback and revision in Version 4. We will combine the planned evaluator with explicit automated checks and record the additional computation. Its internal feedback will remain separate from the independent evaluation used to compare versions.

### 3.4 Exercise Text Classification

Joulin et al. (2017) showed that compact text representations can support efficient supervised classification. Their fastText approach combines learned feature representations with a linear classifier, providing useful background for a small exercise-text model and meaningful baseline comparisons.

We will train a neural classifier from scratch using exercise names and available descriptions, and compare it with majority-class and TF-IDF logistic regression baselines. Model selection will use validation results, with macro-F1 as the primary measure. Related exercise records will remain within the same data partition.

The predicted body-part categories will help organize the exercise catalog. Activity eligibility and profile compatibility will be assessed through separately reviewed requirements and planning rules.

### 3.5 Food Composition References

USDA FoodData Central provides food and nutrient records through distinct data types, including Foundation Foods and SR Legacy (U.S. Department of Agriculture, Agricultural Research Service, n.d.). These sources have different releases and documentation requirements.

We will review the prepared Foundation Foods extract and any additional release selected for the food lookup. Preparation will preserve food identifiers, nutrient definitions, units, reference amounts, and release information. Records from older food tables will require verified provenance and relationships before reuse.

The optional lookup will provide general food facts. Individualized dietary prescriptions, supplement recommendations, and medical nutrition planning remain outside the project scope.

### 3.6 Relationship to Our Study

The literature and reference materials serve different purposes in our project:

| Evidence or reference | Role in our work |
|---|---|
| Jr. NBA curriculum | Source material for activity descriptions and practical requirements |
| NBA/USA Basketball guidelines | Participation definitions, planning constraints, and qualifications |
| RAG research | Background for supplying external evidence during generation |
| ReAct and Self-Refine | Background for tool use, feedback, and revision |
| Efficient text-classification research | Background for the exercise classifier and baseline comparison |
| USDA FoodData Central | Documented food composition information |

We will test how these components work together using the evaluation design in Step 2. Our findings will describe the implemented workflows, including their errors and computational tradeoffs. Published results from other tasks will inform our design without becoming assumed performance gains or evidence of real-world athletic benefit.

## 4. System Architecture and Data Flow

### 4.1 Architecture Overview

We will separate data preparation, model training, planning, application behavior, and research evaluation into components with clear interfaces. Shared Python functions will support both the notebook and the application.

The application will use Version 4 to generate and review plans. The experiment runner will call all four planner versions directly under the evaluation protocol.

| Component | Responsibility | Main output |
|---|---|---|
| Data preparation | Review source records, clean task-specific tables, and preserve provenance and decisions | Versioned datasets and review records |
| Exercise classifier | Train, select, save, and later load the exercise-text model | Predicted categories with their label origin |
| Reference tools | Prepare coaching evidence and access reviewed rules, exercise records, and optional food facts | Source-linked reference packets and structured results |
| Planner workflows | Coordinate the permitted model calls for Versions 1–4 | Returned response and execution records |
| Internal review | Apply Version 4's coded checks, evaluator, and bounded revisions | Check findings and workflow status |
| User application | Collect profiles, display responses, support edits, and record completion | Editable plan versions and progress records |
| Research evaluation | Assess saved outcomes using the common independent protocol | Comparable scores, review files, and resource measurements |

The table summarizes our planned components. Reference access applies to Versions 2–4, and the application uses Version 4.

### 4.2 Data and Classifier Preparation

We will keep exercise records, coaching material, planning rules, food tables, fictional profiles, and evaluation records separate according to their purpose. The existing master file will serve as a heterogeneous record index. Training tables will be built from task-specific inputs with documented relationships.

For exercise classification, we will review labels and text, group related records, and create training, validation, and test partitions. Preprocessing and learned text features will be fitted on training data. We will compare majority-class and TF-IDF logistic regression baselines with a neural classifier, select settings using validation data, and reserve the test set for final evaluation.

We will save the selected model, preprocessing components, and label mapping. Application use will load these artifacts for inference. Original, reviewed, and predicted labels will remain distinguishable, and activity eligibility will depend on separate catalog reviews and planning rules.

The reference collection will contain approved material with stable source identifiers. Evaluation profiles, expected answers, reviewer notes, and generated plans will remain outside the retrieval index. Held-out exercise records will also remain outside the development reference collection under the classifier's evaluation boundary.

### 4.3 Planner Workflows and Reference Access

All versions will use the same base language model, common generation settings, shared profile representation, and response requirements.

| Version | Reference access | Generation workflow |
|---|---|---|
| 1 | No retrieval, structured lookup results, or classifier suggestions | One planner response |
| 2 | Approved retrieval and reference tools | One planner response using the prepared reference packet |
| 3 | The same reference preparation as Version 2 | Specialist contributions followed by one coordinator response |
| 4 | The same reference preparation and initial specialist workflow as Version 3 | Coded checks, evaluator feedback, and bounded coordinator revisions |

For each matched profile, Versions 2–4 will receive the same prepared reference packet, including retrieval and lookup results. We will retain its identifier and checksum. The initial design will reuse that packet across specialist and evaluator calls.

Mandatory rule applicability will come from the reviewed structured register. Food calculations will use verified values and units in code. Retrieval will supply explanatory evidence alongside these structured tools.

The shared input adapter will standardize field names and types while preserving missing information and conflicting commitments. Case labels, expected answers, and independent scoring feedback will remain outside the planner inputs.

### 4.4 Shared Responses and Application Editing

We will preserve raw model output and record parsing results separately. Parsing will check the agreed structure without silently repairing plan content.

| Response element | Information to preserve |
|---|---|
| Identity | Profile, run, planner version, and relevant artifact versions |
| Response type | Plan, clarification, constraint conflict, or outside-scope response |
| Plan content | Activities, days, allocated durations, equipment, participants, and relevant locations |
| Explanations and evidence | Activity descriptions, source references where available, and optional verified food facts |
| Questions and unresolved issues | Missing information, conflicts, and limitations requiring attention |
| Execution and review status | Parsing failures, timeouts, tool errors, and Version 4's internal review outcome |

Required fields will depend on the response type. Clarification and boundary responses will not require a complete weekly schedule. Execution failures will be recorded separately from valid response types. Source identifiers may be absent where unavailable, including in Version 1.

The application will preserve the original returned Version 4 response before creating an editable copy. Each substantive edit will create a new version, rerun the full coded check set, and mark the earlier evaluator assessment as outdated. A separate review action will assess the edited plan without silently rewriting the user's changes.

Checklist entries will record reported completion. Edited plans and progress records will remain separate from benchmark responses and training data.

### 4.5 Independent Evaluation

The experiment runner will process the final profiles through each version, save the returned outcomes before user edits, and apply the frozen evaluation protocol. It will preserve unsuccessful and unfinished jobs so that reported results can be reconciled with the run manifest.

Version 4's internal acceptance decision will be recorded separately from independent performance. Research scores will not trigger corrective model calls in Versions 1–3. Human review will use the predefined response sample and concealed version labels described in Step 2.

Classifier, retrieval, planner, and human-review results will remain separate measures. We will compare complete workflows and report their model calls and resource use alongside output quality.

### 4.6 Traceability and Failure Handling

Run records will retain profile IDs, model and prompt versions, reference-packet IDs, tool results, raw and parsed responses, specialist contributions, revision history, timing, and usage where available.

We will define model-call limits, timeouts, transport retries, and revision budgets before final evaluation. State will be reset between independent cases. Failed calls and parsing problems will remain explicit, and any permitted attempt will count toward recorded resource use.

Source refresh, data preparation, classifier training, and final experiment execution will be selected explicitly. Routine result reproduction will use saved, verified artifacts. This separation will let us update a component while preserving the records needed to explain and reproduce the reported experiment.

```python
# Step 7-1
# We reuse the verified project environment from Step 6.
import csv
import json
import os
import platform
import shutil
import subprocess
from pathlib import Path

feasibility_report = {"status": "incomplete", "notes": []}
if globals().get("environment_report", {}).get("status") != "passed":
    raise RuntimeError("Run the Step 6 environment check successfully first.")
if not isinstance(globals().get("project_root"), Path):
    raise RuntimeError("The project path is unavailable. Rerun Step 6.")

print("Step 7 — Hardware and data access")
print(f"Logical CPU cores: {os.cpu_count()}")
feasibility_report["logical_cpu_cores"] = os.cpu_count()

# We query Windows hardware without changing system settings.
powershell = shutil.which("powershell.exe") or shutil.which("pwsh.exe")
if platform.system() == "Windows" and powershell:
    command = """
    $ErrorActionPreference = 'Stop'
    $system = Get-CimInstance Win32_OperatingSystem
    $processor = Get-CimInstance Win32_Processor | Select-Object -First 1
    [PSCustomObject]@{
        CPU = $processor.Name
        RAM_GiB = [math]::Round($system.TotalVisibleMemorySize / 1MB, 2)
        AvailableRAM_GiB = [math]::Round($system.FreePhysicalMemory / 1MB, 2)
    } | ConvertTo-Json -Compress
    """
    try:
        query = subprocess.run(
            [powershell, "-NoProfile", "-Command", command],
            capture_output=True, text=True, timeout=25, check=False,
        )
        if query.returncode != 0:
            raise RuntimeError(query.stderr.strip()[:300] or "Hardware query failed.")
        hardware = json.loads(query.stdout)
        if not isinstance(hardware, dict):
            raise ValueError("Hardware query returned an unexpected format.")
        feasibility_report["hardware"] = hardware
        for name, value in hardware.items():
            print(f"{name}: {value}")
    except (OSError, ValueError, RuntimeError, subprocess.TimeoutExpired) as error:
        feasibility_report["notes"].append(f"Hardware details unavailable: {error}")
else:
    feasibility_report["notes"].append("Detailed Windows hardware query unavailable.")

disk = shutil.disk_usage(project_root)
feasibility_report["disk_free_GiB"] = round(disk.free / 1024**3, 2)
print(f"Free storage: {feasibility_report['disk_free_GiB']} GiB")

package = project_root / "Champions_Athletic_Agent_Ready_Dataset _Package"
expected_files = (
    "champions_master_agent_ready.csv", "exercise_clean.csv",
    "athlete_profile_clean.csv", "basketball_season_clean.csv",
    "player_bio_clean.csv", "player_career_clean.csv", "food_category_clean.csv",
    "food_portion_clean.csv", "food_sample_link_clean.csv",
    "nutrient_result_clean.csv", "data_quality_audit.csv",
)
file_presence = {name: (package / name).is_file() for name in expected_files}
feasibility_report["cleaned_files"] = file_presence
print(f"\nCleaned package files: {sum(file_presence.values())}/{len(expected_files)}")
for name, present in file_presence.items():
    if not present:
        print(f"Missing: {name}")

# We require a readable exercise sample for the planned classifier.
with (package / "exercise_clean.csv").open(encoding="utf-8-sig", newline="") as handle:
    reader = csv.DictReader(handle)
    columns = reader.fieldnames
    sample = next(reader, None)
    if not columns or not sample or not any(sample.values()):
        raise ValueError("The exercise CSV needs a header and a nonempty data row.")
feasibility_report["exercise_columns"] = columns
print(f"Exercise sample readable: {len(columns)} columns")

historical_inputs = (
    "megaGymDataset(1).csv", "Players(1).csv", "player_data(1).csv",
    "Seasons_Stats(1).csv", "sub_sample_food(1).csv", "sub_sample_result(1).csv",
    "food_category(1).csv", "food_portion(1).csv", "athlete_test_profiles.csv",
)
locations = (project_root, project_root / "data/raw")
available_inputs = {
    name: any((location / name).is_file() for location in locations)
    for name in historical_inputs
}
feasibility_report["historical_inputs"] = available_inputs
print(f"Historical input files in checked locations: {sum(available_inputs.values())}/9")
for folder in ("data/processed", "data/references", "data/synthetic"):
    count = sum(1 for _ in (project_root / folder).rglob("*.csv"))
    print(f"{folder}: {count} CSV files")
for note in feasibility_report["notes"]:
    print(f"Note: {note}")
feasibility_report["local_access_status"] = "passed"
```

```python
# Step 7-2
# We test a small training update and one local model request.
import time
from urllib.request import Request, urlopen

if globals().get("feasibility_report", {}).get("local_access_status") != "passed":
    raise RuntimeError("Run the Step 7 hardware and data cell successfully first.")
feasibility_report["status"] = "incomplete"
execution_failures = []


def check_torch_execution():
    import torch

    use_cuda = torch.cuda.is_available()
    device = torch.device("cuda:0" if use_cuda else "cpu")
    devices = list(range(torch.cuda.device_count())) if use_cuda else []
    with torch.random.fork_rng(devices=devices):
        torch.manual_seed(42)
        a = torch.randn(1024, 1024, device=device)
        b = torch.randn(1024, 1024, device=device)
        torch.mm(a, b)  # We warm up the operation before timing it.
        if use_cuda:
            torch.cuda.synchronize(device)
        start = time.perf_counter()
        product = torch.mm(a, b)
        if use_cuda:
            torch.cuda.synchronize(device)
        elapsed = time.perf_counter() - start
        if not torch.isfinite(product).all().item():
            raise ValueError("Matrix multiplication produced nonfinite values.")

        model = torch.nn.Sequential(
            torch.nn.Linear(8, 16), torch.nn.ReLU(), torch.nn.Linear(16, 3)
        ).to(device)
        x = torch.randn(32, 8, device=device)
        y = torch.arange(32, device=device) % 3
        before = [parameter.detach().clone() for parameter in model.parameters()]
        optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
        optimizer.zero_grad()
        loss = torch.nn.functional.cross_entropy(model(x), y)
        if not torch.isfinite(loss).item():
            raise ValueError("Training loss is nonfinite.")
        loss.backward()
        if not all(p.grad is not None and torch.isfinite(p.grad).all().item()
                   for p in model.parameters()):
            raise ValueError("Training gradients are missing or nonfinite.")
        optimizer.step()
        updated = any(not torch.equal(old, new)
                      for old, new in zip(before, model.parameters()))
        if not updated or not all(torch.isfinite(p).all().item() for p in model.parameters()):
            raise ValueError("The optimization step did not produce a finite parameter update.")
        return {
            "status": "passed", "version": torch.__version__, "device": str(device),
            "device_name": torch.cuda.get_device_name(0) if use_cuda else "CPU",
            "gpu_memory_GiB": torch.cuda.get_device_properties(0).total_memory / 1024**3 if use_cuda else None,
            "compiled_cuda": torch.version.cuda, "matrix_seconds": elapsed,
            "training_loss": loss.item(), "finite_gradients": True, "parameters_updated": updated,
        }


try:
    feasibility_report["torch"] = check_torch_execution()
    print("PyTorch execution")
    for name, value in feasibility_report["torch"].items():
        print(f"{name}: {value}")
except Exception as error:
    feasibility_report["torch"] = {"status": "failed", "error": str(error)}
    execution_failures.append(f"PyTorch: {type(error).__name__}: {error}")


def ollama_json(endpoint, payload=None, timeout=15):
    request = Request(
        f"http://127.0.0.1:11434/api/{endpoint}",
        data=json.dumps(payload).encode("utf-8") if payload is not None else None,
        headers={"Content-Type": "application/json"},
        method="POST" if payload is not None else "GET",
    )
    with urlopen(request, timeout=timeout) as response:
        result = json.loads(response.read().decode("utf-8"))
    if not isinstance(result, dict) or result.get("error"):
        raise ValueError(f"Ollama returned an API error or unexpected response: {result}")
    return result


def check_model_response(result, token_limit):
    text = result.get("response", "")
    parsed = json.loads(text)
    valid_structure = (
        isinstance(parsed, dict) and set(parsed) == {"day_1", "day_2", "day_3"}
        and all(isinstance(value, str) and value.strip() for value in parsed.values())
    )
    reason = result.get("done_reason")
    count = result.get("eval_count")
    count_known = type(count) is int and count >= 0
    limit_hit = reason == "length" or (
        reason is None and count_known and count >= token_limit
    )
    stop_verified = reason == "stop" or (
        reason is None and count_known and 0 < count < token_limit
    )
    complete = result.get("done") is True and stop_verified and not limit_hit
    return parsed, valid_structure, complete, limit_hit


model_name = "llama3.1:8b"
prompt = """We are testing a basketball planning interface with a fictional profile.
Athlete: age 15, goal: ball handling, three sessions of 30 minutes each.
Available: one basketball and a clear practice area. No hoop, cones, or partner.
Suggest three short solo practice themes using only those resources.
Return JSON with exactly day_1, day_2, day_3, each containing a nonempty string.
Exclude medical and nutrition advice."""
settings = {"temperature": 0, "seed": 42, "num_predict": 256, "num_ctx": 4096}
request_payload = {
    "model": model_name, "prompt": prompt, "stream": False,
    "format": "json", "options": settings,
}
model_report = {"status": "incomplete", "request": request_payload, "content_review": "pending"}
feasibility_report["ollama"] = model_report
try:
    model_report["server_version"] = ollama_json("version").get("version")
    installed = ollama_json("tags").get("models", [])
    selected = next((item for item in installed
                     if model_name in (item.get("name"), item.get("model"))), None)
    if selected is None:
        raise RuntimeError(f"{model_name} is not installed in the local Ollama server.")
    model_report["model_identity"] = selected
    if not selected.get("digest"):
        raise ValueError("The installed model digest is unavailable.")
    start = time.perf_counter()
    result = ollama_json("generate", request_payload, timeout=180)
    model_report["wall_seconds"] = time.perf_counter() - start
    model_report["raw_response"] = result
    parsed, valid, complete, limit_hit = check_model_response(result, settings["num_predict"])
    model_report.update({"parsed_response": parsed, "valid_structure": valid,
                         "completion_verified": complete, "generation_limit_hit": limit_hit})
    print("\nLocal Ollama request")
    print(f"Server: {model_report['server_version']} | Model: {model_name}")
    print(f"Digest: {selected['digest']}")
    print(f"Quantization: {selected.get('details', {}).get('quantization_level', 'unreported')}")
    print(f"Settings: {settings}")
    print(json.dumps(parsed, indent=2))
    print(f"JSON structure: {valid} | Completion: {complete} | Generation limit hit: {limit_hit}")
    print(f"Wall time: {model_report['wall_seconds']:.2f} seconds")
    if result.get("total_duration") is not None:
        print(f"Ollama total time: {result['total_duration'] / 1e9:.2f} seconds")
    print(f"Input/output tokens: {result.get('prompt_eval_count')} / {result.get('eval_count')}")
    print(f"Stop reason: {result.get('done_reason', 'unreported')}")
    print("Content review: pending inspection of the returned themes.")
    if not valid or not complete:
        raise ValueError("The response did not pass the structure and completion checks.")
    model_report["status"] = "passed"
except Exception as error:
    model_report.update({"status": "failed", "error": str(error)})
    if "raw_response" in model_report:
        print(f"Returned text: {model_report['raw_response'].get('response', '')[:1500]}")
    execution_failures.append(f"Ollama: {type(error).__name__}: {error}")

feasibility_report["execution_failures"] = execution_failures
feasibility_report["status"] = "passed" if not execution_failures else "needs attention"
if execution_failures:
    for finding in execution_failures:
        print(f"\nNeeds attention: {finding}")
    raise RuntimeError("Resolve the execution findings and rerun this cell.")
print("\nInitial execution checks passed. Full planning evaluation remains pending.")
```

### 8.5 Review Decisions and Usage Terms

We will preserve existing source descriptions, dates, permissions, limitations, issue statuses, and additional documentation columns. Current file measurements will be written to the audit reports rather than silently replacing recorded hashes or source-verification dates.

A changed required input will require review of its version and lineage before the register is updated. Changes to saved later outputs will be reported for reconciliation in their producing steps. Existing artifact-bound decisions will require review whenever their underlying contents change.

FoodData Central documents CC0 terms for its data. We will retain that source-policy information while separately checking whether our prepared files are traceable to the stated USDA release. We will keep the older food tables' unresolved provenance separate from this policy.

For exercise data, coaching content, and other references, we will document the applicable terms for storage, transformation, model use, retrieval, display, and redistribution. Attribution and permission will remain separate records. This audit will preserve pending decisions and will not assign catalog approval or planner eligibility.

### 8.6 Outputs and Handoff

We will use three code cells to define the source records, inspect files and relationships, and save the results.

| Output | Purpose |
|---|---|
| `data/source_manifest.csv` | Curated source records and declared lineage |
| `results/provenance_followups.csv` | Open and completed provenance work, with existing statuses preserved |
| `results/source_file_inventory.csv` | Current paths, hashes, counts, availability, and comparisons with recorded facts |
| `results/source_lineage_checks.csv` | Required and informational check results |
| `results/source_manifest_changes.csv` | Proposed and applied source-register changes |
| `results/step08_source_audit_summary.json` | Audit scope, counts, issue statuses, and final outcome |

We will validate the records before updating the source and issue ledgers. Changed documentation will be backed up, and a failed output update will trigger restoration of the files already written. If required source checks fail, we will save diagnostic reports while leaving those two ledgers unchanged.

The code will inspect local files without refreshing publisher pages or downloading new datasets. A local audit timestamp will remain separate from acquisition dates and recorded source-verification dates.

Step 9 will use this register and its current inventory to load the relevant datasets and review their quality. We will finalize this step's observations after inspecting the execution output.

```python
# Step 8-1

# We keep source descriptions separate from the file measurements below.
import csv
import hashlib
import io
import json
import os
import tempfile
from datetime import datetime, timezone
from pathlib import Path, PureWindowsPath

step08_report = {"status": "incomplete"}
ledger_snapshots = {}
if globals().get("environment_report", {}).get("status") != "passed":
    raise RuntimeError("Run the Step 6 environment check successfully first.")
if not isinstance(globals().get("project_root"), Path):
    raise RuntimeError("Rerun Step 6 to identify the project folder.")
root = project_root.resolve()
audited_at = datetime.now(timezone.utc).isoformat()
FIELDS = (
    "source_id title asset_type intended_role publisher source_url version_or_release "
    "acquisition_date verification_date file_status source_filename relative_path "
    "file_size_bytes sha256 reported_records record_count_basis parent_source_ids "
    "transformation_script source_commit license_or_terms permission_status limitations next_action"
).split()
ISSUE_FIELDS = ["issue_id", "priority", "issue", "next_action", "status"]
package_rel = "Champions_Athletic_Agent_Ready_Dataset _Package"
PACKAGE_COMMIT = "1076a81f3b4f71f69e38cfbc43575358652c611a"
INTEGRATION_COMMIT = "c1631ed735684e292563037f7c8ceca72a5766b7"
RECOVERY_COMMIT = "c42c9f352c575b37c4dd967cca5f64b78eeb9069"


def project_path(relative):
    value = Path(relative)
    if value.is_absolute() or PureWindowsPath(relative).drive:
        raise ValueError(f"Use a project-relative path: {relative}")
    path = (root / value).resolve()
    if not path.is_relative_to(root):
        raise ValueError(f"Path leaves our project: {relative}")
    return path


def read_table(path):
    with path.open(encoding="utf-8-sig", newline="") as handle:
        reader = csv.DictReader(handle)
        header = reader.fieldnames or []
        if not header or any(not name.strip() for name in header) or len(header) != len(set(header)):
            raise ValueError(f"Invalid CSV header: {path}")
        rows = list(reader)
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f"CSV row width differs from its header: {path}")
    return rows, header


def load_ledger(path, fields, key):
    if not path.exists():
        ledger_snapshots[path] = None
        return {}, list(fields)
    before = path.read_bytes()
    rows, header = read_table(path)
    if path.read_bytes() != before:
        raise RuntimeError(f"Ledger changed while being loaded: {path}")
    ledger_snapshots[path] = before
    if not set(fields).issubset(header):
        raise ValueError(f"Required columns missing: {path}")
    ids = [row[key] for row in rows]
    if any(not item.strip() or item != item.strip() for item in ids) or len(ids) != len(set(ids)):
        raise ValueError(f"Empty, padded, or duplicate {key}: {path}")
    return {row[key]: row for row in rows}, header


def sha256_file(path):
    digest = hashlib.sha256()
    with path.open("rb") as handle:
        for block in iter(lambda: handle.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()


def source_record(source_id, **values):
    row = dict.fromkeys(FIELDS, "")
    row.update(source_id=source_id, publisher="Project team", permission_status="pending_review")
    row.update(values)
    return row


audit_rows, _ = read_table(project_path(f"{package_rel}/data_quality_audit.csv"))
audit_by_type = {row["dataset"]: row for row in audit_rows}
if len(audit_by_type) != len(audit_rows):
    raise ValueError("The cleaning audit contains duplicate dataset names.")
historical = {
    "exercise": "megaGymDataset(1).csv", "player_bio": "Players(1).csv",
    "player_career": "player_data(1).csv", "basketball_season": "Seasons_Stats(1).csv",
    "food_sample_link": "sub_sample_food(1).csv", "nutrient_result": "sub_sample_result(1).csv",
    "food_category": "food_category(1).csv", "food_portion": "food_portion(1).csv",
    "athlete_profile": "athlete_test_profiles.csv",
}
if not set(historical).issubset(audit_by_type):
    raise ValueError("The cleaning audit does not describe all expected prepared datasets.")
specifications = {}
required_ids = set()
later_steps = {"catalog_basketball_v01": "12", "catalog_basketball_v02": "12",
               "local_exercise_catalog": "10-11", "local_basketball_drills": "12"}
for kind, filename in historical.items():
    specifications[f"raw_{kind}"] = source_record(
        f"raw_{kind}", title=f"Historical input: {kind}", asset_type="original_input",
        intended_role=kind, publisher="Unverified", source_filename=filename,
        relative_path=f"data/raw/{filename}", limitations="Original release and acquisition remain unverified.",
        next_action="Locate the input and document its origin; profile inputs are fictional examples.",
    )
    count = int(audit_by_type[kind]["clean_rows"])
    if count < 0:
        raise ValueError(f"Invalid cleaning-audit count: {kind}")
    sid = f"derived_{kind}"
    specifications[sid] = source_record(
        sid, title=f"Prepared {kind} dataset", asset_type="derived_dataset", intended_role=kind,
        source_filename=f"{kind}_clean.csv", relative_path=f"{package_rel}/{kind}_clean.csv",
        reported_records=str(count), record_count_basis="existing_data_quality_audit",
        parent_source_ids=f"raw_{kind}", source_commit=PACKAGE_COMMIT,
        transformation_script="Champions Athletic data-cleaning pipeline.py",
        limitations="Original input lineage and usage terms require review.",
        next_action="Verify source documentation and task-specific quality.",
    )
    required_ids.add(sid)
parents = ";".join(f"derived_{kind}" for kind in historical)
for sid, filename, count, basis, kind in (
    ("derived_master", "champions_master_agent_ready.csv", sum(int(audit_by_type[k]["clean_rows"]) for k in historical), "sum_of_existing_audit_counts", "derived_catalog"),
    ("derived_audit", "data_quality_audit.csv", len(audit_rows), "audit_entries_read", "derived_report"),
):
    specifications[sid] = source_record(
        sid, title=filename, asset_type=kind, intended_role="source_documentation_and_inventory",
        source_filename=filename, relative_path=f"{package_rel}/{filename}",
        reported_records=str(count), record_count_basis=basis, parent_source_ids=parents,
        source_commit=PACKAGE_COMMIT, transformation_script="Champions Athletic data-cleaning pipeline.py",
        limitations="The master is a heterogeneous index; the audit is a preparation record.",
        next_action="Compare the recorded counts with independent file inspection.",
    )
    required_ids.add(sid)

references = [
    ("reference_jr_nba", "Jr. NBA/Jr. WNBA curriculum", "Jr. NBA", "https://jr.nba.com/jr-nba-instructional-curriculum/", "basketball_coaching", ""),
    ("reference_youth_guidelines", "Youth basketball guidelines", "NBA and USA Basketball", "https://youthguidelines.nba.com/", "participation_guidance", ""),
    ("reference_usda", "FoodData Central", "USDA ARS", "https://fdc.nal.usda.gov/download-datasets/", "food_reference", ""),
    ("reference_usda_foundation_2025", "Foundation Foods preparation source", "USDA ARS", "https://fdc.nal.usda.gov/download-datasets/", "nutrition_source", "2025-04-24 (recorded preparation release; archive verification pending)"),
    ("reference_cdc_sleep", "CDC sleep guidance", "CDC", "https://www.cdc.gov/sleep/about/", "sleep_reference", ""),
    ("reference_aap_overuse", "Overuse, overtraining, and burnout", "American Academy of Pediatrics", "https://doi.org/10.1542/peds.2023-065129", "recovery_reference", ""),
    ("reference_usda_sr_legacy_2018", "SR Legacy planned reference", "USDA ARS", "https://fdc.nal.usda.gov/download-datasets/", "planned_food_reference", "April 2018; acquisition not established"),
]
for sid, title, publisher, url, role, version in references:
    specifications[sid] = source_record(
        sid, title=title, publisher=publisher, source_url=url, intended_role=role,
        asset_type="external_reference", version_or_release=version, file_status="reference_not_downloaded",
        license_or_terms="CC0 1.0 for verified FoodData Central data" if sid.startswith("reference_usda") else "Reuse terms require review",
        permission_status="source_terms_documented" if sid.startswith("reference_usda") else "pending_review",
        limitations="This registration does not establish a saved source archive or verified local derivation.",
        next_action="Document source evidence, applicability, and permitted use.",
    )

# Legacy source IDs remain stable so that existing lineage links keep working.
assets = [
    ("aditi_jr_nba_drills", "data/processed/jr_nba_drills_master.csv", "derived_dataset", "reference_jr_nba", "notebooks/06_JrNBA_Combined_Dataset.ipynb", INTEGRATION_COMMIT),
    ("aditi_jr_nba_rag", "data/processed/jr_nba_rag_documents.json", "derived_rag_documents", "aditi_jr_nba_drills", "notebooks/07_JrNBA_RAG_Preparation.ipynb", INTEGRATION_COMMIT),
    ("aditi_usda_foundation_nutrition", "data/processed/usda_foundation_nutrition.csv", "derived_dataset", "reference_usda_foundation_2025", "notebooks/08_Nutrition_Data_Preparation.ipynb", INTEGRATION_COMMIT),
    ("aditi_usda_quality_report", "data/processed/usda_nutrition_quality_report.csv", "derived_report", "aditi_usda_foundation_nutrition", "notebooks/08_Nutrition_Data_Preparation.ipynb", INTEGRATION_COMMIT),
    ("aditi_recovery_guidance", "data/processed/recovery_guidance.csv", "derived_reference", "reference_cdc_sleep;reference_aap_overuse", "notebooks/09_Recovery_Data_Preparation.ipynb", INTEGRATION_COMMIT),
    ("aditi_synthetic_athletes", "data/synthetic/synthetic_athlete_profiles.csv", "synthetic_dataset", "", "notebooks/03_Generate_Athlete_Profiles.ipynb", INTEGRATION_COMMIT),
    ("recovery_rules_v01", "data/references/recovery/youth_basketball_recovery_rules_v0_1.csv", "derived_reference", "reference_youth_guidelines;reference_cdc_sleep;reference_aap_overuse", "manual source-linked curation", RECOVERY_COMMIT),
    ("recovery_rules_readme_v01", "data/references/recovery/YOUTH_BASKETBALL_RECOVERY_RULES_README.md", "derived_documentation", "recovery_rules_v01", "manual documentation", RECOVERY_COMMIT),
    ("catalog_basketball_v01", "data/processed/basketball_drill_catalog_v0_1.csv", "derived_catalog", "reference_jr_nba", "Step 12", ""),
    ("catalog_basketball_v02", "data/processed/basketball_drill_catalog_v0_2_REVIEW_ONLY.csv", "derived_catalog", "catalog_basketball_v01;reference_jr_nba", "Step 12", ""),
    ("local_exercise_catalog", "data/processed/exercise_catalog.csv", "derived_catalog", "derived_exercise", "Steps 10-11", ""),
    ("local_basketball_drills", "data/processed/basketball_drills.csv", "derived_dataset", "reference_jr_nba", "Step 12", ""),
]
for sid, relative, kind, parents, transform, commit in assets:
    specifications[sid] = source_record(
        sid, title=Path(relative).name, asset_type=kind, intended_role=kind,
        source_filename=Path(relative).name, relative_path=relative, parent_source_ids=parents,
        transformation_script=transform, source_commit=commit,
        source_url=f"https://github.com/SamTavasoli/champion-athletic-map/blob/{commit}/{relative}" if commit else "",
        limitations="Prepared or curated material; approval and planner eligibility require separate review.",
        next_action="Review lineage, intended use, and content in the relevant data step.",
    )
    if sid not in later_steps:
        required_ids.add(sid)
specifications["aditi_synthetic_athletes"].update(
    version_or_release="seed=42", permission_status="team_generated",
    license_or_terms="Fictional project-generated examples", limitations="Development inputs, not real athlete observations or a reserved final test set.",
)

manifest_path = project_path("data/source_manifest.csv")
issues_path = project_path("results/provenance_followups.csv")
previous_sources, manifest_header = load_ledger(manifest_path, FIELDS, "source_id")
source_rows = {sid: dict(row) for sid, row in previous_sources.items()}
for sid, specification in specifications.items():
    if sid not in source_rows:
        source_rows[sid] = {name: specification.get(name, "") for name in manifest_header}
    elif not source_rows[sid]["relative_path"] and specification["relative_path"]:
        source_rows[sid]["relative_path"] = specification["relative_path"]
print(f"Step 8 — Source registration: {len(previous_sources)} existing, {len(source_rows)} after registration")
step08_report["registration_complete"] = True
```

```python
# Step 8-2

# We measure local files without renewing source or permission decisions.
if not globals().get("step08_report", {}).get("registration_complete"):
    raise RuntimeError("Complete the source registration cell first.")
inventory = []
source_checks = []
file_cache = {}
step08_report["inspection_complete"] = False
step08_report["status"] = "inspecting"


def add_check(name, passed, detail, required=True):
    source_checks.append({"check": name, "passed": bool(passed),
                          "required": required, "detail": str(detail)})


def check_parent_graph(records):
    complete, active = set(), set()

    def visit(sid):
        if sid in active:
            raise ValueError(f"Source lineage contains a cycle at {sid}.")
        if sid in complete:
            return
        active.add(sid)
        for parent in filter(None, (p.strip() for p in records[sid]["parent_source_ids"].split(";"))):
            if parent not in records:
                raise ValueError(f"Unknown parent {parent} for {sid}.")
            visit(parent)
        active.remove(sid)
        complete.add(sid)

    for sid in records:
        visit(sid)


def observe_file(path):
    if path in file_cache:
        return file_cache[path]
    before = path.stat()
    digest = sha256_file(path)
    count = None
    if path.suffix.lower() == ".csv":
        with path.open(encoding="utf-8-sig", newline="") as handle:
            reader = csv.reader(handle)
            header = next(reader, [])
            if not header or len(header) != len(set(header)) or any(not name.strip() for name in header):
                raise ValueError("CSV header is empty or contains duplicate/blank names.")
            count = 0
            for row in reader:
                if not row:
                    continue
                if len(row) != len(header):
                    raise ValueError("A CSV row differs in width from its header.")
                count += 1
    elif path.suffix.lower() == ".json":
        with path.open(encoding="utf-8-sig") as handle:
            value = json.load(handle)
        count = len(value) if isinstance(value, list) else None
    after = path.stat()
    if (before.st_size, before.st_mtime_ns) != (after.st_size, after.st_mtime_ns):
        raise ValueError("File changed during inspection.")
    observed = {"observed_sha256": digest, "observed_records": count,
                "file_size_bytes": after.st_size}
    file_cache[path] = observed
    return observed


check_parent_graph(source_rows)
add_check("source_parent_graph", True, "All parents resolve; the graph is acyclic.")
for sid, row in source_rows.items():
    relative = row["relative_path"]
    entry = {"source_id": sid, "declared_path": relative,
             "producer_step": later_steps.get(sid, ""), "required_now": sid in required_ids,
             "registered_sha256": row["sha256"], "registered_records": row["reported_records"],
             "observed_path": "", "observed_sha256": "", "observed_records": "",
             "file_size_bytes": "", "hash_status": "not_checked", "count_status": "not_checked",
             "error": "", "file_audited_at": audited_at}
    candidates = [project_path(relative)] if relative else []
    if sid in {f"raw_{kind}" for kind in historical}:
        candidates += [project_path(row["source_filename"]), project_path(f"data/raw/{row['source_filename']}")]
    matches = list(dict.fromkeys(path for path in candidates if path.is_file()))
    if not matches:
        entry["availability"] = (
            "external_reference_not_refreshed" if row["asset_type"] == "external_reference"
            else "expected_later" if sid in later_steps else "not_found_in_checked_locations"
        )
        add_check(f"file_access:{sid}", sid not in required_ids, entry["availability"], sid in required_ids)
    else:
        try:
            observations = [observe_file(path) for path in matches]
            if len({item["observed_sha256"] for item in observations}) > 1:
                raise ValueError("Different files occupy the alternative declared input locations.")
            entry.update(observations[0], availability="present", observed_path=matches[0].relative_to(root).as_posix())
            entry["hash_status"] = "not_previously_recorded" if not row["sha256"] else (
                "unchanged" if row["sha256"] == entry["observed_sha256"] else "changed"
            )
            count = entry["observed_records"]
            entry["count_status"] = "not_applicable" if count is None else (
                "not_previously_recorded" if not row["reported_records"] else
                "matches" if row["reported_records"].isdigit() and int(row["reported_records"]) == count else "differs"
            )
            add_check(f"file_access:{sid}", True, entry["observed_path"], sid in required_ids)
            add_check(f"recorded_file_facts:{sid}", entry["hash_status"] != "changed" and entry["count_status"] != "differs",
                      f"hash={entry['hash_status']}; count={entry['count_status']}", sid in required_ids)
        except (OSError, ValueError, csv.Error) as error:
            entry.update(availability="inspection_failed", error=str(error))
            add_check(f"file_access:{sid}", False, error, sid in required_ids)
    inventory.append(entry)
observed_by_id = {entry["source_id"]: entry for entry in inventory}


def local_records(sid):
    entry = observed_by_id[sid]
    if entry["availability"] != "present":
        raise ValueError(f"Cannot inspect records for {sid}: {entry['availability']}")
    path = project_path(entry["observed_path"])
    if path.suffix.lower() == ".csv":
        return read_table(path)[0]
    with path.open(encoding="utf-8-sig") as handle:
        rows = json.load(handle)
    if not isinstance(rows, list) or not all(isinstance(row, dict) for row in rows):
        raise ValueError(f"Expected a list of records: {sid}")
    return rows


def unique_ids(rows, field):
    values = [row.get(field) for row in rows]
    if not all(isinstance(value, str) and value and value == value.strip() for value in values):
        return False
    return bool(values) and all(values) and len(values) == len(set(values))


try:
    drills = local_records("aditi_jr_nba_drills")
    rag = local_records("aditi_jr_nba_rag")
    matches = unique_ids(drills, "drill_id") and unique_ids(rag, "drill_id") and (
        {r["drill_id"] for r in drills} == {r["drill_id"] for r in rag}
    )
    add_check("drill_rag_identifier_correspondence", matches, f"{len(drills)} drill rows; {len(rag)} derived documents")
    profiles = local_records("aditi_synthetic_athletes")
    synthetic = unique_ids(profiles, "athlete_id") and all(r.get("is_synthetic", "").strip().lower() == "true" for r in profiles)
    add_check("fictional_profile_identifiers", synthetic, f"{len(profiles)} development examples; final-test independence is not established here")
    recovery = local_records("recovery_rules_v01")
    add_check("recovery_rule_identifiers", unique_ids(recovery, "rule_id"), f"{len(recovery)} rule records")
    add_check("recovery_v01_snapshot", observed_by_id["recovery_rules_v01"]["observed_sha256"] ==
              "d4ef1ab35329e747b980d19fb0339c28ee3dff41e5b60724490193b0dab46ac6", "Pinned v0.1 reference bytes; applicability remains under review")
except (KeyError, ValueError, OSError, TypeError) as error:
    add_check("related_record_inspection", False, error)

# Only newly registered files receive initial measured facts in the source ledger.
# Existing hashes, counts, dates, permission decisions, and extra columns are retained.
for sid, row in source_rows.items():
    entry = observed_by_id[sid]
    if sid not in previous_sources and entry["availability"] == "present":
        row.update(file_status="present", file_size_bytes=str(entry["file_size_bytes"]),
                   sha256=entry["observed_sha256"])
        if entry["observed_records"] is not None and not row["reported_records"]:
            row.update(reported_records=str(entry["observed_records"]), record_count_basis="initial_local_file_count")
    elif sid not in previous_sources and not row["file_status"]:
        row["file_status"] = entry["availability"]

issue_specs = [
    ("P01", "high", "Historical input files unavailable", "Locate inputs and document their source or generation history."),
    ("P02", "high", "Exercise dataset release and terms", "Verify the acquired release and its license evidence."),
    ("P03", "high", "Original food-table lineage", "Verify source identifiers, nutrient units, and transformations."),
    ("P04", "medium", "Historical basketball source provenance", "Document the publisher, release, and intended analytical role."),
    ("P05", "high", "Original fictional-profile origins", "Document how the earlier prepared examples were generated."),
    ("P06", "high", "Coaching-content reuse", "Review storage, display, and redistribution terms."),
    ("P07", "medium", "Original row lineage", "Distinguish original record identifiers from cleaned row positions."),
    ("P08", "high", "Basketball source-occurrence overlap", "Reconcile source occurrences and candidate identities in Step 12."),
    ("P09", "high", "Coaching summaries and adaptations", "Document source wording, summaries, adaptations, and permitted use."),
    ("P10", "high", "Nutrition quality", "Review missing and unusual nutrient values during Step 9."),
    ("P11", "medium", "Synthetic evaluation coverage", "Preserve development examples and isolate the final test set."),
    ("P12", "high", "Participation-rule applicability", "Verify age, grade, scope, and interpretation in Step 13."),
    ("P13", "medium", "Preparation notebook portability", "Replace machine-specific paths through reviewed implementation changes."),
    ("P14", "medium", "Foundation Foods source archive", "Verify the recorded archive release and preparation lineage."),
]
previous_issues, issue_header = load_ledger(issues_path, ISSUE_FIELDS, "issue_id")
issues = {sid: dict(row) for sid, row in previous_issues.items()}
for sid, priority, description, action in issue_specs:
    if sid not in issues:
        issues[sid] = dict.fromkeys(issue_header, "")
        issues[sid].update(issue_id=sid, priority=priority, issue=description, next_action=action, status="open")

step08_report.update(inspection_complete=True, audited_at=audited_at)
failed_checks = [check for check in source_checks if check["required"] and not check["passed"]]
print(f"Registered records: {len(source_rows)} | Local files inspected: {sum(e['availability'] == 'present' for e in inventory)}")
print("Independent file counts:", {label: observed_by_id[sid]["observed_records"] for label, sid in (
    ("exercise", "derived_exercise"), ("prepared_master", "derived_master"),
    ("food_reference", "aditi_usda_foundation_nutrition"), ("recovery_guidance", "aditi_recovery_guidance"))})
print(f"Required checks passed: {sum(c['required'] and c['passed'] for c in source_checks)} | Failed: {len(failed_checks)}")
for check in source_checks:
    if ":" not in check["check"]:
        print(f"{check['check']}: {'passed' if check['passed'] else 'needs attention'} — {check['detail']}")
for check in source_checks:
    if not check["passed"]:
        print(f"{'Needs attention' if check['required'] else 'Informational'}: {check['check']} — {check['detail']}")
```

```python
# Step 8-3

# We save the audit and preserve earlier versions of changed documentation.
if not globals().get("step08_report", {}).get("inspection_complete"):
    raise RuntimeError("Complete the source inspection cell first.")
step08_report["status"] = "preparing_outputs"


def csv_content(rows, header):
    buffer = io.StringIO(newline="")
    writer = csv.DictWriter(buffer, fieldnames=header, extrasaction="raise")
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode("utf-8")


def replace_bytes(path, content):
    path.parent.mkdir(parents=True, exist_ok=True)
    descriptor, temporary = tempfile.mkstemp(dir=path.parent)
    try:
        with os.fdopen(descriptor, "wb") as handle:
            handle.write(content)
        os.replace(temporary, path)
    finally:
        Path(temporary).unlink(missing_ok=True)


def save_bundle(outputs):
    prior = {path: path.read_bytes() if path.exists() else None for path in outputs}
    for path, loaded in ledger_snapshots.items():
        if path in outputs and prior[path] != loaded:
            raise RuntimeError(f"Documentation changed since loading; rerun Step 8: {path}")
    changed = [path for path, content in outputs.items() if prior[path] != content]
    backup_root = project_path("results/backups/step08") / datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S_%fZ")
    for path in changed:
        if prior[path] is not None:
            backup = backup_root / path.relative_to(root)
            backup.parent.mkdir(parents=True, exist_ok=True)
            backup.write_bytes(prior[path])
    written = []
    try:
        for path in changed:
            written.append(path)
            replace_bytes(path, outputs[path])
            if path.read_bytes() != outputs[path]:
                raise OSError(f"Saved bytes differ from the prepared output: {path}")
    except Exception as error:
        restoration_errors = []
        for path in reversed(written):
            try:
                if prior[path] is None:
                    path.unlink(missing_ok=True)
                else:
                    replace_bytes(path, prior[path])
            except Exception as restore_error:
                restoration_errors.append(f"{path}: {restore_error}")
        if restoration_errors:
            raise RuntimeError("Output update failed; some files could not be restored: " + "; ".join(restoration_errors)) from error
        raise
    return {path.relative_to(root).as_posix(): "unchanged" if path not in changed else
            "updated" if prior[path] is not None else "created" for path in outputs}


changes = []
for sid, row in source_rows.items():
    old = previous_sources.get(sid, {})
    for field in manifest_header:
        before, after = old.get(field, ""), row[field]
        if before != after:
            changes.append({"source_id": sid, "field": field, "previous_value": before,
                            "proposed_value": after, "applied": not failed_checks})

status_counts = {}
for issue in issues.values():
    status = issue["status"].strip().lower() or "unspecified"
    status_counts[status] = status_counts.get(status, 0) + 1
audit_summary = dict(step08_report)
audit_summary.update(
    status="passed" if not failed_checks else "needs attention",
    registered_sources=len(source_rows), existing_source_ids_preserved=len(previous_sources),
    new_source_ids=[sid for sid in source_rows if sid not in previous_sources],
    local_files_inspected=sum(e["availability"] == "present" for e in inventory),
    historical_inputs_not_found=sum(e["source_id"].startswith("raw_") and e["availability"] != "present" for e in inventory),
    later_outputs={sid: observed_by_id[sid]["availability"] for sid in later_steps},
    tracked_issues=len(issues), issue_status_counts=status_counts,
    required_checks_passed=sum(c["required"] and c["passed"] for c in source_checks),
    required_checks_failed=len(failed_checks), existing_permission_decisions_changed=0,
    external_pages_refreshed_by_code=False,
)
outputs = {
    project_path("results/source_file_inventory.csv"): csv_content(inventory, list(inventory[0])),
    project_path("results/source_lineage_checks.csv"): csv_content(source_checks, ["check", "passed", "required", "detail"]),
    project_path("results/source_manifest_changes.csv"): csv_content(changes, ["source_id", "field", "previous_value", "proposed_value", "applied"]),
    project_path("results/step08_source_audit_summary.json"): (json.dumps(audit_summary, indent=2) + "\n").encode("utf-8"),
}
if not failed_checks:
    outputs[manifest_path] = csv_content(source_rows.values(), manifest_header)
    outputs[issues_path] = csv_content(issues.values(), issue_header)
step08_report["status"] = "saving"
saved_actions = save_bundle(outputs)
step08_report.update(audit_summary)
for relative, action in saved_actions.items():
    print(f"{action.capitalize()}: {relative}")
print(f"Tracked issues: {len(issues)} | Statuses: {status_counts}")
print(f"Historical inputs not found: {audit_summary['historical_inputs_not_found']}")
print(f"New source IDs: {audit_summary['new_source_ids'] or 'none'}")
print("Later-output availability:", json.dumps(audit_summary["later_outputs"], indent=2))
if failed_checks:
    raise RuntimeError("Source checks need attention. Diagnostic reports were saved; the source and issue ledgers were not updated.")
print("Step 8 source audit passed. Source permissions, content approval, and planner eligibility remain separate reviews.")
```

### 9.3 General Quality Measures

| Check area | What we will record |
|---|---|
| Structure | Readability, headers, row widths, required columns, and JSON record shapes |

| Relationships | Drill-to-document correspondence and prepared-file links in the master |

We will count cells, records, files, columns, and links separately. A missing-cell count will not be reported as the number of affected records.

Blank values and common null-like markers will be screened consistently. Explicit `unknown` values will remain separate. Literal `none` in injury-status or equipment fields can describe reported absence, so we will preserve that distinction. Numerical zero will remain a measured value rather than a missing-value substitute.

Missing optional information will require interpretation according to its purpose. For example, absent exercise ratings, blank profile scenario notes, and conditionally unused recovery bounds have different implications.

```python
# Step 9-1

# We load the current source inventory before examining dataset quality.
import csv
import hashlib
import io
import json
import math
import os
import re
import tempfile
from collections import Counter, defaultdict
from datetime import datetime, timezone

step09_report = {"status": "incomplete", "loading_complete": False, "domain_complete": False}
if globals().get("environment_report", {}).get("status") != "passed":
    raise RuntimeError("Run the Step 6 environment check successfully first.")
if globals().get("step08_report", {}).get("status") != "passed":
    raise RuntimeError("Complete the Step 8 source audit successfully first.")
if not all(callable(globals().get(name)) for name in ("project_path", "read_table", "sha256_file", "replace_bytes")):
    raise RuntimeError("Run Step 8 to load our shared source-inspection functions.")
root = project_root.resolve()
quality_audited_at = datetime.now(timezone.utc).isoformat()
csv.field_size_limit(10_000_000)
checks, inventory, column_profiles, category_counts, quality_examples = [], [], [], [], []
tables, observed, example_counts = {}, {}, Counter()
master_types, master_positions, master_issues = Counter(), defaultdict(set), Counter()
control_snapshots, inspected_file_states = {}, {}


def quality_check(sid, name, status, affected=0, details="", action="", required=False, unit="records"):
    checks.append({"source_id": sid, "check_name": name, "status": status,
                   "affected_count": affected, "details": details, "next_action": action,
                   "required": required, "affected_unit": unit})


def example(sid, finding, row_number, identifier, detail):
    key = (sid, finding)
    if example_counts[key] < 5:
        quality_examples.append({"source_id": sid, "finding": finding,
                                 "record_position": row_number, "identifier": identifier,
                                 "details": str(detail)[:600]})
        example_counts[key] += 1


def distribution(sid, column, values):
    counts = Counter("[blank]" if value is None or str(value).strip() == "" else str(value).strip() for value in values)
    category_counts.extend({"source_id": sid, "column": column, "value": value, "count": count}
                           for value, count in sorted(counts.items()))


def is_missing(value, field=""):
    if value is None:
        return True
    token = str(value).strip().lower()
    if token == "none" and field in {"injury_status", "equipment"}:
        return False
    return token in {"", "null", "none", "nan", "n/a"}


def finite_number(value):
    if isinstance(value, bool) or is_missing(value):
        return None
    try:
        number = float(value)
        return number if math.isfinite(number) else None
    except (TypeError, ValueError, OverflowError):
        return None


def boolean_value(value):
    token = str(value).strip().lower()
    return True if token in {"yes", "true", "1"} else False if token in {"no", "false", "0"} else None


def normalize_name(value):
    return " ".join(re.findall(r"[a-z0-9]+", str(value).lower()))


def strict_json(text):
    def unique_object(pairs):
        result = {}
        for key, value in pairs:
            if key in result:
                raise ValueError(f"Repeated JSON key: {key}")
            result[key] = value
        return result

    def finite_float(value):
        result = float(value)
        if not math.isfinite(result):
            raise ValueError("Nonfinite JSON number")
        return result

    def invalid_constant(value):
        raise ValueError(f"Nonstandard JSON constant: {value}")

    return json.loads(text, object_pairs_hook=unique_object,
                      parse_float=finite_float, parse_constant=invalid_constant)


control_paths = [project_path(name) for name in (
    "data/source_manifest.csv", "results/source_file_inventory.csv", "results/step08_source_audit_summary.json")]
for path in control_paths:
    control_snapshots[path] = path.read_bytes()
manifest, _ = read_table(control_paths[0])
source_inventory, _ = read_table(control_paths[1])
source_summary = strict_json(control_snapshots[control_paths[2]].decode("utf-8-sig"))
if not isinstance(source_summary, dict):
    raise ValueError("The Step 8 summary must be a JSON object.")
inventory_fields = {"source_id", "observed_path", "observed_sha256", "observed_records",
                    "file_audited_at", "required_now", "producer_step"}
if not all(inventory_fields.issubset(row) and boolean_value(row["required_now"]) is not None for row in source_inventory):
    raise ValueError("The Step 8 inventory has missing fields or an invalid required-input marker.")
for rows in (manifest, source_inventory):
    identifiers = [row["source_id"] for row in rows]
    if any(not value or value != value.strip() for value in identifiers) or len(identifiers) != len(set(identifiers)):
        raise ValueError("Source records need nonempty, unique identifiers.")
if (source_summary.get("status") != "passed" or source_summary.get("registered_sources") != len(manifest)
        or {r["source_id"] for r in manifest} != {r["source_id"] for r in source_inventory}
        or any(r["file_audited_at"] != source_summary.get("audited_at") for r in source_inventory)):
    raise RuntimeError("The source register and Step 8 audit are inconsistent. Rerun Step 8 first.")
if any(path.read_bytes() != content for path, content in control_snapshots.items()):
    raise RuntimeError("Source documentation changed while loading. Rerun this cell.")
manifest_by_id = {row["source_id"]: row for row in manifest}
step08_by_id = {row["source_id"]: row for row in source_inventory}

# These fields support the checks we perform; additional fields remain intact.
REQUIRED_COLUMNS = {
    "derived_exercise": "exercise_name description body_part exercise_type equipment difficulty_level",
    "derived_athlete_profile": "athlete_id age available_days goal equipment injury_status",
    "derived_player_bio": "player_name height weight",
    "derived_player_career": "player_name year_start year_end",
    "derived_basketball_season": "season_year player_name team",
    "derived_food_category": "id code description",
    "derived_food_portion": "id fdc_id amount gram_weight",
    "derived_food_sample_link": "fdc_id fdc_id_of_sample_food",
    "derived_nutrient_result": "food_nutrient_id adjusted_amount lab_method_id nutrient_name",
    "derived_audit": "dataset clean_rows clean_columns",
    "derived_master": "record_id entity_type source_file source_row entity_label agent_text data_json",
    "aditi_jr_nba_drills": "drill_id drill_name skill_category curriculum_level duration equipment source_document source_page solo_adaptation_possible",
    "aditi_jr_nba_rag": "drill_id text",
    "aditi_usda_foundation_nutrition": "fdc_id description protein_g fat_g carbohydrate_g fiber_g calories_kcal energy_method data_type",
    "aditi_usda_quality_report": "column non_missing missing missing_pct",
    "aditi_recovery_guidance": "topic guidance source source_url",
    "aditi_synthetic_athletes": "athlete_id age sport position experience_level primary_goal available_days_per_week equipment injury_status scenario_type is_synthetic",
    "recovery_rules_v01": "rule_id metric rule_type age_min age_max grade_min grade_max value_min value_max unit source_url verification_status",
    "catalog_basketball_v01": "catalog_candidate_id drill_name scope_candidate catalog_approval_status planner_eligible",
    "catalog_basketball_v02": "catalog_candidate_id drill_name scope_candidate catalog_approval_status planner_eligible source_record_class",
    "local_exercise_catalog": "catalog_id exercise_name body_part",
    "local_basketball_drills": "candidate_id record_kind source_url planner_eligible",
}
ID_FIELDS = {
    "derived_master": "record_id", "derived_athlete_profile": "athlete_id",
    "aditi_jr_nba_drills": "drill_id", "aditi_jr_nba_rag": "drill_id",
    "aditi_usda_foundation_nutrition": "fdc_id", "aditi_usda_quality_report": "column",
    "aditi_synthetic_athletes": "athlete_id", "recovery_rules_v01": "rule_id",
    "catalog_basketball_v01": "catalog_candidate_id", "catalog_basketball_v02": "catalog_candidate_id",
    "local_exercise_catalog": "catalog_id", "local_basketball_drills": "candidate_id",
}
SMALL_TABLES = set(REQUIRED_COLUMNS) - {
    "derived_master", "derived_player_bio", "derived_player_career", "derived_basketball_season",
    "derived_food_category", "derived_food_portion", "derived_food_sample_link", "derived_nutrient_result"}
master_parents = [sid.strip() for sid in manifest_by_id["derived_master"]["parent_source_ids"].split(";") if sid.strip()]
if not master_parents or any(sid not in manifest_by_id for sid in master_parents):
    raise ValueError("The master register needs resolvable prepared-domain parents.")
domain_sources = {sid.removeprefix("derived_"): sid for sid in master_parents}
expected_source_files = {domain: manifest_by_id.get(f"raw_{domain}", {}).get("source_filename", "")
                         for domain in domain_sources}


def inspect_master_row(row, position):
    kind = row.get("entity_type", "")
    master_types[kind] += 1
    if kind not in domain_sources:
        master_issues["unexpected_entity_type"] += 1
    if any(is_missing(row.get(field)) for field in ("entity_type", "source_file", "source_row", "entity_label")):
        master_issues["missing_lineage_fields"] += 1
    if kind in expected_source_files and row.get("source_file") != expected_source_files[kind]:
        master_issues["source_filename_mismatch"] += 1
    source_position = finite_number(row.get("source_row"))
    if source_position is None or not source_position.is_integer() or source_position < 1:
        master_issues["invalid_cleaned_position"] += 1
    elif int(source_position) in master_positions[kind]:
        master_issues["repeated_cleaned_position"] += 1
    else:
        master_positions[kind].add(int(source_position))
    try:
        payload = strict_json(row.get("data_json", ""))
        if not isinstance(payload, dict):
            raise ValueError("Payload must be an object")
        if not payload:
            master_issues["empty_json_payload"] += 1
            example("derived_master", "empty_json_payload", position, row.get("record_id", ""),
                    f"Prepared source: {row.get('source_file')}; position: {row.get('source_row')}")
    except (ValueError, TypeError, RecursionError) as error:
        master_issues["invalid_json_payload"] += 1
        example("derived_master", "invalid_json_payload", position, row.get("record_id", ""), error)


def scan_records(sid, columns, records, csv_rows=True):
    missing_columns = set(REQUIRED_COLUMNS.get(sid, "").split()) - set(columns)
    schema_declared = sid in REQUIRED_COLUMNS
    quality_check(sid, "required_columns", "not_checked" if not schema_declared else "failed" if missing_columns else "pass", len(missing_columns),
                  "No task-specific schema is declared." if not schema_declared else ", ".join(sorted(missing_columns)) or "All fields needed for these checks are present.", required=schema_declared, unit="columns")
    counts = Counter()
    absent, unknown, literal_none = Counter(), Counter(), Counter()
    seen_rows, seen_ids, retained = set(), set(), []
    identifier_field = ID_FIELDS.get(sid)
    for position, row in enumerate(records, 1):
        counts["rows"] += 1
        if csv_rows and (None in row or any(value is None for value in row.values())):
            counts["malformed_rows"] += 1
            example(sid, "malformed_row", position, "", "CSV width differs from the header.")
            continue
        counts["inspected_rows"] += 1
        signature = hashlib.blake2b(json.dumps([row.get(name) for name in columns], ensure_ascii=False,
                                               separators=(",", ":"), allow_nan=False).encode(), digest_size=16).digest()
        counts["duplicate_rows"] += signature in seen_rows
        seen_rows.add(signature)
        for name in columns:
            value = row.get(name)
            absent[name] += is_missing(value, name)
            unknown[name] += isinstance(value, str) and value.strip().lower() == "unknown"
            literal_none[name] += isinstance(value, str) and value.strip().lower() == "none"
        if identifier_field:
            value = row.get(identifier_field)
            if not isinstance(value, str) or is_missing(value) or value.strip().lower() == "unknown":
                counts["missing_ids"] += 1
            elif value.strip() in seen_ids:
                counts["repeated_ids"] += 1
            else:
                seen_ids.add(value.strip())
        if sid == "derived_master":
            inspect_master_row(row, position)
        if sid in SMALL_TABLES:
            retained.append(row)
    quality_check(sid, "nonempty_dataset", "pass" if counts["rows"] else "failed", 0 if counts["rows"] else 1, required=True, unit="datasets")
    quality_check(sid, "record_structure", "failed" if counts["malformed_rows"] else "pass", counts["malformed_rows"], required=True)
    quality_check(sid, "exact_row_duplicate_screen", "needs_review" if counts["duplicate_rows"] else "pass", counts["duplicate_rows"],
                  "Repeated full-row fingerprints; no records were removed.")
    missing_total = sum(absent.values())
    conditional = {"age_min", "age_max", "grade_min", "grade_max", "value_min", "value_max"}
    conditional_only = sid == "recovery_rules_v01" and all(not count or name in conditional for name, count in absent.items())
    quality_check(sid, "missing_values", "needs_review" if missing_total and not conditional_only else "pass", missing_total,
                  "Missing/null-like cells; optional and conditional fields need domain-specific interpretation.", unit="cells")
    if sum(unknown.values()):
        quality_check(sid, "explicit_unknown_values", "needs_review", sum(unknown.values()), "Literal unknown is preserved separately from missing values.", unit="cells")
    if identifier_field:
        bad_ids = counts["missing_ids"] + counts["repeated_ids"]
        quality_check(sid, "identifier_uniqueness", "failed" if bad_ids else "pass", bad_ids,
                      f"{identifier_field}: missing={counts['missing_ids']}; repeated={counts['repeated_ids']}", required=True)
    for name in columns:
        column_profiles.append({"source_id": sid, "column": name, "inspected_rows": counts["inspected_rows"],
                                "missing_cells": absent[name], "explicit_unknown_cells": unknown[name],
                                "literal_none_cells": literal_none[name]})
    if sid in SMALL_TABLES:
        tables[sid] = {"columns": columns, "rows": retained}
    return {"rows": counts["rows"], "columns": columns, "missing": dict(absent), "unknown": dict(unknown)}


print(f"Step 9 — Dataset loading: {len(manifest)} registered source records")
for item in manifest:
    sid = item["source_id"]
    anchor = step08_by_id[sid]
    relative = anchor["observed_path"] or item["relative_path"]
    summary = {"source_id": sid, "filename": item["source_filename"], "relative_path": relative,
               "load_status": "not_checked", "actual_rows": "", "columns": "", "file_size_bytes": "",
               "sha256_match": "not_checked", "notes": "", "file_format": "",
               "expected_records": item["reported_records"], "observed_sha256": "",
               "step08_sha256_match": "not_checked", "step08_rows": anchor["observed_records"],
               "required_input": boolean_value(anchor["required_now"]) is True, "producer_step": anchor["producer_step"]}
    if item["asset_type"] == "external_reference":
        summary["load_status"] = "external_reference"
        quality_check(sid, "publisher_content_review", "not_checked", details="External source content and permissions are outside this local quality inspection.", unit="sources")
    elif item["asset_type"] == "original_input":
        candidate_paths = [project_path(value) for value in
                           (relative, item["source_filename"], f"data/raw/{item['source_filename']}") if value]
        exists = any(path.is_file() for path in candidate_paths)
        summary["load_status"] = "present_not_inspected" if exists else "missing_original"
        quality_check(sid, "historical_input_availability", "not_checked" if exists else "needs_review", 0 if exists else 1,
                      "Historical publisher or fictional-profile input; provenance remains unresolved.", unit="files")
    elif not relative or not project_path(relative).is_file():
        expected_later = bool(anchor["producer_step"]) and not summary["required_input"]
        summary["load_status"] = "expected_later" if expected_later else "missing"
        quality_check(sid, "file_availability", "not_applicable" if expected_later else "failed", 0 if expected_later else 1,
                      "Declared later output is not required here." if expected_later else "Registered local input is unavailable.", required=not expected_later, unit="files")
    else:
        path = project_path(relative)
        try:
            before = path.stat()
            digest = sha256_file(path)
            summary.update(observed_sha256=digest, file_size_bytes=before.st_size, file_format=path.suffix.lower().lstrip("."))
            current_match = digest == anchor["observed_sha256"]
            summary["step08_sha256_match"] = "yes" if current_match else "no"
            quality_check(sid, "step08_snapshot_integrity", "pass" if current_match else "failed", 0 if current_match else 1,
                          "Current bytes compared with the latest Step 8 file observation.", action="Rerun Step 8 after reconciling changed inputs." if not current_match else "", required=True, unit="files")
            registered_match = digest == item["sha256"] if item["sha256"] else None
            summary["sha256_match"] = "yes" if registered_match else "no" if registered_match is False else "not_recorded"
            quality_check(sid, "historical_registration_hash", "pass" if registered_match else "needs_review", 0 if registered_match else 1,
                          "Historical registration is kept separate from the current run anchor.", unit="files")
            if path.suffix.lower() == ".csv":
                with path.open(encoding="utf-8-sig", newline="") as handle:
                    reader = csv.DictReader(handle, strict=True)
                    columns = reader.fieldnames or []
                    if not columns or len(columns) != len(set(columns)) or any(not name.strip() for name in columns):
                        raise ValueError("Missing, blank, or duplicate CSV headers.")
                    observed[sid] = scan_records(sid, columns, reader)
            elif path.suffix.lower() == ".json":
                value = strict_json(path.read_text(encoding="utf-8-sig"))
                if isinstance(value, list):
                    if not all(isinstance(row, dict) for row in value):
                        raise ValueError("JSON table must contain objects.")
                    columns = sorted({key for row in value for key in row})
                    observed[sid] = scan_records(sid, columns, value, csv_rows=False)
                elif sid == "aditi_jr_nba_rag":
                    raise ValueError("The RAG source must be a list of records.")
            if sid in observed:
                summary.update(load_status="loaded", actual_rows=observed[sid]["rows"], columns=len(observed[sid]["columns"]))
                expected = finite_number(anchor["observed_records"])
                matched = expected is not None and expected.is_integer() and int(expected) == summary["actual_rows"]
                quality_check(sid, "step08_row_count", "pass" if matched else "failed", 0 if matched else 1,
                              f"Step 8={anchor['observed_records']}; current={summary['actual_rows']}", required=True, unit="datasets")
            else:
                summary["load_status"] = "verified_document"
            after = path.stat()
            stable = (before.st_size, before.st_mtime_ns) == (after.st_size, after.st_mtime_ns)
            inspected_file_states[path] = (after.st_size, after.st_mtime_ns)
            quality_check(sid, "file_stability_during_inspection", "pass" if stable else "failed", 0 if stable else 1, required=True, unit="files")
            row_display = summary["actual_rows"] if summary["actual_rows"] != "" else "document"
            print(f"{summary['filename']}: rows={row_display}; columns={summary['columns']}; Step 8 hash={summary['step08_sha256_match']}")
        except (OSError, UnicodeError, csv.Error, ValueError, KeyError, TypeError, RecursionError) as error:
            summary.update(load_status="failed", notes=f"{type(error).__name__}: {error}")
            quality_check(sid, "file_loading", "failed", 1, summary["notes"], required=True, unit="files")
    inventory.append(summary)
step09_report["loading_complete"] = True
step09_report["status"] = "loaded_for_review"
loading_checkpoint = {"checks": len(checks), "category_counts": len(category_counts), "quality_examples": len(quality_examples)}
```

```python
# Step 9-2

# We review each data family without changing its source values.
from collections import Counter
from datetime import date
import json
import math
from urllib.parse import urlparse

if isinstance(globals().get("step09_report"), dict):
    step09_report["domain_complete"] = False
if not globals().get("step09_report", {}).get("loading_complete"):
    raise RuntimeError("Complete the Step 9 loading cell first.")
step09_report.update(domain_complete=False, status="reviewing")
for name, length in loading_checkpoint.items():
    del globals()[name][length:]
example_counts = Counter((row["source_id"], row["finding"]) for row in quality_examples)
domain_metrics = {}


def review_rows(sid):
    table = tables.get(sid)
    if table is None:
        quality_check(sid, "domain_review", "not_checked", details="No loaded table is available for this review.")
        return None
    return table["rows"]


def text_value(value):
    return value.strip() if isinstance(value, str) else ""


def review_condition(sid, name, affected, details, action="", required=False, unit="records"):
    status = "failed" if affected and required else "needs_review" if affected else "pass"
    quality_check(sid, name, status, affected, details, action, required=required, unit=unit)


def source_url_present(value):
    try:
        parsed = urlparse(text_value(value))
        return parsed.scheme == "https" and bool(parsed.netloc)
    except ValueError:
        return False


# Exercise labels and text support the separate classification task.
sid = "derived_exercise"
exercise = review_rows(sid)
if exercise is not None:
    missing_names = sum(is_missing(r.get("exercise_name")) for r in exercise)
    missing_labels = sum(is_missing(r.get("body_part")) for r in exercise)
    missing_descriptions = sum(is_missing(r.get("description")) for r in exercise)
    labels = Counter(text_value(r.get("body_part")) for r in exercise if not is_missing(r.get("body_part")))
    names = Counter(normalize_name(r.get("exercise_name", "")) for r in exercise if not is_missing(r.get("exercise_name")))
    repeated_names = sum(count - 1 for count in names.values() if count > 1)
    label_sets = {}
    for row in exercise:
        key = normalize_name(row.get("exercise_name", ""))
        if key and not is_missing(row.get("body_part")):
            label_sets.setdefault(key, set()).add(text_value(row.get("body_part")))
    conflicting_name_keys = sorted(key for key, values in label_sets.items() if len(values) > 1)
    distribution(sid, "repeated_normalized_name_key", (key for key, count in names.items() if count > 1 for _ in range(count)))
    for key in conflicting_name_keys:
        example(sid, "normalized_name_label_conflicts", "", key, ", ".join(sorted(label_sets[key])))
    for column in ("body_part", "exercise_type", "difficulty_level", "equipment"):
        distribution(sid, column, (r.get(column, "") for r in exercise))
    review_condition(sid, "classification_names_and_labels", missing_names + missing_labels,
                     f"Missing names={missing_names}; missing body-part labels={missing_labels}.",
                     "Resolve missing classification identifiers or labels in Step 10.", required=True, unit="cells")
    review_condition(sid, "exercise_description_availability", missing_descriptions,
                     "Descriptions are reviewed separately from optional ratings.", "Document text-preparation choices in Step 10.")
    review_condition(sid, "normalized_exercise_name_overlap", repeated_names,
                     "Additional records sharing a normalized name; a name match alone does not establish a duplicate.",
                     "Compare descriptions, labels, and identifiers before any consolidation.")
    review_condition(sid, "normalized_name_label_conflicts", len(conflicting_name_keys),
                     "Different body-part labels share a normalized exercise-name key.",
                     "Review the source and group related names before classifier splitting; retain original labels.", unit="groups")
    largest_two = sum(count for _, count in labels.most_common(2))
    domain_metrics["exercise"] = {
        "records": len(exercise), "body_part_categories": len(labels), "label_counts": dict(labels),
        "missing_names": missing_names, "missing_labels": missing_labels,
        "missing_descriptions": missing_descriptions, "additional_normalized_name_matches": repeated_names,
        "repeated_normalized_name_groups": sum(count > 1 for count in names.values()),
        "different_label_name_groups": len(conflicting_name_keys), "different_label_name_keys": conflicting_name_keys,
        "largest_two_label_fraction": largest_two / len(exercise) if exercise else None,
    }

catalog = review_rows("local_exercise_catalog")
if catalog is not None and exercise is not None:
    sid = "local_exercise_catalog"
    if "source_cleaned_record_position" not in tables[sid]["columns"]:
        quality_check(sid, "cleaned_exercise_position_links", "not_checked",
                      details="This catalog schema does not provide cleaned-record positions.",
                      action="Review the catalog's documented lineage in Step 10.")
    else:
        positions, invalid_links, value_differences = [], 0, 0
        for row in catalog:
            position = finite_number(row.get("source_cleaned_record_position"))
            if position is None or not position.is_integer() or not 1 <= position <= len(exercise):
                invalid_links += 1
                continue
            positions.append(int(position))
            original = exercise[int(position) - 1]
            if any(text_value(row.get(field)) != text_value(original.get(field))
                   for field in ("exercise_name", "body_part")):
                value_differences += 1
        mapping_gaps = len(set(range(1, len(exercise) + 1)).symmetric_difference(positions))
        repeated_positions = len(positions) - len(set(positions))
        review_condition(sid, "cleaned_exercise_position_links", invalid_links + mapping_gaps + repeated_positions,
                         f"Invalid positions={invalid_links}; coverage differences={mapping_gaps}; repeated positions={repeated_positions}.",
                         "Reconcile the saved catalog version with its cleaned source in Step 10.", unit="link findings")
        review_condition(sid, "preserved_exercise_names_and_labels", value_differences,
                         "Original name and label fields are compared at the documented cleaned positions.",
                         "Keep source labels separate from reviewed and predicted labels.")
        domain_metrics["exercise_catalog"] = {"records": len(catalog), "invalid_position_links": invalid_links,
                                                "mapping_gaps": mapping_gaps, "repeated_positions": repeated_positions,
                                                "source_value_differences": value_differences}


# Fictional examples remain development inputs until the benchmark partitions are established.
profile_settings = (
    ("derived_athlete_profile", "available_days", "goal", False),
    ("aditi_synthetic_athletes", "available_days_per_week", "primary_goal", True),
)
for sid, days_field, goal_field, declared_generator in profile_settings:
    rows = review_rows(sid)
    if rows is None:
        continue
    invalid_ranges, invalid_markers, unexpected_categories = 0, 0, 0
    for number, row in enumerate(rows, 1):
        age, days = finite_number(row.get("age")), finite_number(row.get(days_field))
        day_bounds = (2, 5) if declared_generator else (0, 7)
        if (age is None or not age.is_integer() or not 13 <= age <= 17
                or days is None or not days.is_integer() or not day_bounds[0] <= days <= day_bounds[1]):
            invalid_ranges += 1
            example(sid, "profile_ranges", number, row.get("athlete_id", ""), f"age={row.get('age')}; {days_field}={row.get(days_field)}")
        if declared_generator and boolean_value(row.get("is_synthetic")) is not True:
            invalid_markers += 1
        if declared_generator and (
            text_value(row.get("sport")).casefold() != "basketball"
            or text_value(row.get("experience_level")).casefold() not in {"beginner", "intermediate", "advanced"}
            or text_value(row.get("position")).casefold() not in {"point guard", "shooting guard", "small forward", "power forward", "center"}
        ):
            unexpected_categories += 1
    for column in ("age", "experience_level", "position", days_field, "injury_status"):
        distribution(sid, column, (r.get(column, "") for r in rows))
    if declared_generator:
        distribution(sid, "scenario_type", (r.get("scenario_type", "") for r in rows))
        review_condition(sid, "synthetic_data_marker", invalid_markers, "Each declared synthetic example must retain its explicit marker.",
                         "Review the source labeling before these profiles are used.", required=True)
        review_condition(sid, "profile_category_conventions", unexpected_categories,
                         "Current generator conventions: basketball, three experience levels, and five positions.",
                         "Document new categories or correct inconsistent source values.")
    review_condition(sid, "profile_integer_ranges", invalid_ranges,
                     f"Current snapshot expects integer ages 13–17 and integer availability {day_bounds[0]}–{day_bounds[1]} days.",
                     "Review these examples without silently correcting their constraints.")
    available_columns = set(tables[sid]["columns"])
    planning_fields = {
        "school_grade": {"grade", "school_grade"},
        "session_minutes": {"session_minutes", "minutes_per_session", "available_minutes_per_session"},
        "existing_commitments": {"existing_commitments", "school_team_commitments", "organized_basketball_hours_per_week"},
        "sleep_information": {"sleep_hours", "sleep_hours_per_night"},
        "annual_participation": {"organized_basketball_months_per_year", "annual_participation_history"},
    }
    absent_features = [name for name, aliases in planning_fields.items() if not available_columns.intersection(aliases)]
    review_condition(sid, "planning_input_coverage", len(absent_features),
                     "Planned field groups absent from this schema: " + (", ".join(absent_features) or "none"),
                     "Extend documented profile inputs and reserve an independent final evaluation set.", unit="field groups")
    domain_metrics[sid] = {
        "records": len(rows), "invalid_integer_ranges": invalid_ranges, "invalid_synthetic_markers": invalid_markers,
        "unexpected_generator_categories": unexpected_categories, "missing_planning_field_groups": absent_features,
        "unknown_injury_statuses": sum(text_value(r.get("injury_status")).casefold() == "unknown" for r in rows),
        "missing_goals": sum(is_missing(r.get(goal_field)) for r in rows),
        "age_counts": dict(Counter(text_value(r.get("age")) for r in rows)),
        "scenario_counts": dict(Counter(text_value(r.get("scenario_type")) for r in rows)) if declared_generator else {},
    }


# Nutrition completeness, numeric validity, and nonnegative values are separate measures.
sid = "aditi_usda_foundation_nutrition"
foods = review_rows(sid)
if foods is not None:
    numeric_fields = ("protein_g", "fat_g", "carbohydrate_g", "fiber_g", "energy_general_kcal", "energy_specific_kcal", "calories_kcal")
    core = ("calories_kcal", "protein_g", "fat_g", "carbohydrate_g")
    missing_counts, invalid_counts, negative_counts = Counter(), Counter(), Counter()
    present_core, finite_core, nonnegative_core, energy_inconsistencies, invalid_dates = 0, 0, 0, 0, 0
    for number, row in enumerate(foods, 1):
        parsed = {field: finite_number(row.get(field)) for field in numeric_fields}
        present_core += all(not is_missing(row.get(field)) for field in core)
        complete = all(parsed[field] is not None for field in core)
        finite_core += complete
        nonnegative_core += complete and all(parsed[field] >= 0 for field in core)
        for field in numeric_fields:
            value = row.get(field)
            if is_missing(value):
                missing_counts[field] += 1
            elif parsed[field] is None:
                invalid_counts[field] += 1
                example(sid, f"invalid_numeric:{field}", number, row.get("fdc_id", ""), str(value))
            elif parsed[field] < 0:
                negative_counts[field] += 1
                example(sid, f"negative_numeric:{field}", number, row.get("fdc_id", ""), str(value))
        method = text_value(row.get("energy_method"))
        selected = parsed["calories_kcal"]
        expected = parsed["energy_specific_kcal"] if method == "Atwater Specific (2048)" else parsed["energy_general_kcal"] if method == "Atwater General (2047)" else None
        if (method == "Missing" and selected is not None) or (method not in {"Missing", "Atwater Specific (2048)", "Atwater General (2047)"}) or (
            method != "Missing" and (selected is None or expected is None or not math.isclose(selected, expected, rel_tol=1e-9, abs_tol=1e-6))
        ):
            energy_inconsistencies += 1
        if not is_missing(row.get("publication_date")):
            try:
                date.fromisoformat(text_value(row.get("publication_date")))
            except ValueError:
                invalid_dates += 1
    for field in numeric_fields:
        review_condition(sid, f"numeric_finiteness:{field}", invalid_counts[field],
                         "Nonmissing measurements must parse as finite numbers.", "Review the original nutrient measurement and units.")
        review_condition(sid, f"negative_measurements:{field}", negative_counts[field],
                         "Negative measurements are preserved for source-level investigation.", "Check source values and calculation conventions; do not replace them with zero.")
    review_condition(sid, "finite_core_nutrition", len(foods) - finite_core,
                     f"Present core fields={present_core}; finite core values={finite_core}; finite nonnegative core values={nonnegative_core}; total={len(foods)}.",
                     "Keep incomplete foods distinguishable from measured zero values.")
    review_condition(sid, "selected_energy_consistency", energy_inconsistencies,
                     "Selected calories are compared with the energy method named in this prepared table.", "Review preparation logic before selecting or substituting energy measures.")
    review_condition(sid, "publication_date_format", invalid_dates, "Nonmissing publication dates should use an ISO date.")
    for field in ("energy_method", "source", "data_type"):
        distribution(sid, field, (r.get(field, "") for r in foods))
    quality_check(sid, "nutrient_units_and_reference_quantity", "not_checked", details="Column names describe grams/kcal; original archive units and reference quantities still require provenance review.",
                  action="Verify the source archive and quantity basis before food calculations.", unit="datasets")
    domain_metrics["nutrition"] = {
        "records": len(foods), "core_fields_present": present_core, "finite_core_complete": finite_core,
        "finite_core_incomplete": len(foods) - finite_core, "nonnegative_finite_core_complete": nonnegative_core,
        "missing_numeric_fields": dict(missing_counts), "invalid_numeric_fields": dict(invalid_counts),
        "negative_numeric_fields": dict(negative_counts), "energy_inconsistencies": energy_inconsistencies,
    }
    report_rows = review_rows("aditi_usda_quality_report")
    if report_rows is not None:
        differences = 0
        columns = set(tables[sid]["columns"])
        for row in report_rows:
            field = row.get("column")
            measured_missing = sum(is_missing(food.get(field)) for food in foods) if field in columns else None
            reported_missing, reported_present = finite_number(row.get("missing")), finite_number(row.get("non_missing"))
            if measured_missing is None or reported_missing != measured_missing or reported_present != len(foods) - measured_missing:
                differences += 1
        review_condition("aditi_usda_quality_report", "nutrition_report_correspondence", differences,
                         "Prepared report counts are compared with this run's field measurements.", "Review differing count definitions or preparation versions.")


# Matching retrieval identifiers establish a relationship, not coaching approval.
drills = review_rows("aditi_jr_nba_drills")
rag = review_rows("aditi_jr_nba_rag")
if drills is not None:
    for field in ("skill_category", "curriculum_level", "group_requirement", "solo_adaptation_possible"):
        distribution("aditi_jr_nba_drills", field, (row.get(field, "") for row in drills))
    missing_metadata = {field: sum(is_missing(row.get(field)) for row in drills) for field in (
        "drill_name", "duration", "equipment", "instructions_summary", "group_requirement", "source_document", "source_page")}
    review_condition("aditi_jr_nba_drills", "drill_source_metadata", sum(missing_metadata.values()),
                     json.dumps(missing_metadata, sort_keys=True), "Verify instructional sections, requirements, and any proposed adaptations in Step 12.", unit="cells")
    drill_ids = [text_value(row.get("drill_id")) for row in drills]
    names = {normalize_name(row.get("drill_name", "")) for row in drills if not is_missing(row.get("drill_name"))}
    domain_metrics["drills"] = {"source_occurrences": len(drills), "normalized_names": len(names), "missing_metadata": missing_metadata}
    if rag is not None:
        invalid_documents = sum(not isinstance(row.get("drill_id"), str) or not text_value(row.get("drill_id"))
                                or not isinstance(row.get("text"), str) or not text_value(row.get("text")) for row in rag)
        rag_ids = [text_value(row.get("drill_id")) for row in rag]
        linkage_problem = len(set(drill_ids).symmetric_difference(rag_ids)) + abs(len(drill_ids) - len(rag_ids))
        review_condition("aditi_jr_nba_rag", "retrieval_document_text", invalid_documents,
                         "Each document requires string drill_id and nonempty text.", required=True)
        review_condition("aditi_jr_nba_rag", "drill_document_correspondence", linkage_problem,
                         f"Structured occurrences={len(drill_ids)}; documents={len(rag_ids)}.", required=True)
        domain_metrics["drills"].update(rag_documents=len(rag), linkage_problems=linkage_problem)

for sid in ("catalog_basketball_v01", "catalog_basketball_v02", "local_basketball_drills"):
    rows = review_rows(sid)
    if rows is None:
        continue
    unapproved_eligible, unknown_eligibility = 0, 0
    for number, row in enumerate(rows, 1):
        eligible = boolean_value(row.get("planner_eligible"))
        approval = text_value(row.get("catalog_approval_status")).casefold()
        if eligible is None:
            unknown_eligibility += 1
        if eligible is True and approval != "approved":
            unapproved_eligible += 1
            example(sid, "unapproved_eligible", number, row.get("catalog_candidate_id", row.get("candidate_id", "")), f"approval={approval or 'not documented'}")
    review_condition(sid, "planner_approval_gate", unapproved_eligible,
                     "Only an explicitly approved record may be marked planner eligible.", "Keep unapproved candidates out of planning inputs.", required=True)
    review_condition(sid, "eligibility_encoding", unknown_eligibility,
                     "Blank or unrecognized eligibility values remain unresolved and are not treated as eligible.")
    for field in ("scope_candidate", "skill_area", "catalog_approval_status", "record_kind"):
        if field in tables[sid]["columns"]:
            distribution(sid, field, (row.get(field, "") for row in rows))
    domain_metrics[sid] = {"records": len(rows), "unapproved_eligible": unapproved_eligible, "unknown_eligibility": unknown_eligibility}
    if sid == "catalog_basketball_v02" and drills is not None:
        catalog_names = {normalize_name(row.get("drill_name", "")) for row in rows if not is_missing(row.get("drill_name"))}
        overlap = len(names.intersection(catalog_names))
        review_condition("aditi_jr_nba_drills", "catalog_normalized_name_overlap", overlap,
                         f"Imported unique names={len(names)}; overlapping catalog names={overlap}. Names alone do not resolve identity.", "Compare source occurrences and variations in Step 12.", unit="names")
        domain_metrics["drills"]["catalog_name_overlap"] = overlap


# Rule bounds are checked according to their declared type; applicability stays under review.
guidance = review_rows("aditi_recovery_guidance")
if guidance is not None:
    missing_source = sum(not source_url_present(row.get("source_url")) or is_missing(row.get("guidance")) for row in guidance)
    review_condition("aditi_recovery_guidance", "guidance_source_fields", missing_source,
                     "Guidance needs readable text and an HTTPS source reference.", "Verify the original section before applying the guidance.")
    domain_metrics["recovery_guidance"] = {"records": len(guidance), "incomplete_source_fields": missing_source}

rules = review_rows("recovery_rules_v01")
if rules is not None:
    bound_types = {"maximum": "upper", "minimum": "lower", "recommended_range": "range",
                   "recommended_exact": "exact", "published_range_9_to_10": "range"}
    invalid_bounds, invalid_segments, incomplete_sources = 0, 0, 0
    for number, row in enumerate(rules, 1):
        kind = text_value(row.get("rule_type"))
        shape = bound_types.get(kind)
        low, high = finite_number(row.get("value_min")), finite_number(row.get("value_max"))
        malformed = shape is None or (shape == "upper" and high is None) or (shape == "lower" and low is None)
        malformed |= shape in {"range", "exact"} and (low is None or high is None)
        malformed |= any(not is_missing(row.get(field)) and finite_number(row.get(field)) is None for field in ("value_min", "value_max"))
        malformed |= any(value is not None and value < 0 for value in (low, high))
        malformed |= low is not None and high is not None and low > high
        malformed |= shape == "exact" and low is not None and high is not None and low != high
        invalid_bounds += bool(malformed)
        for lower, upper in (("age_min", "age_max"), ("grade_min", "grade_max")):
            present = [not is_missing(row.get(field)) for field in (lower, upper)]
            values = [finite_number(row.get(field)) for field in (lower, upper)]
            if any(present) and (not all(present) or any(value is None or not value.is_integer() or value < 0 for value in values)
                                 or (all(value is not None for value in values) and values[0] > values[1])):
                invalid_segments += 1
                example("recovery_rules_v01", "rule_segments", number, row.get("rule_id", ""), f"{lower}/{upper} need review")
        if not source_url_present(row.get("source_url")) or any(is_missing(row.get(field)) for field in ("metric", "unit", "verification_status")):
            incomplete_sources += 1
        if malformed:
            example("recovery_rules_v01", "rule_bounds", number, row.get("rule_id", ""), f"type={kind}; bounds={row.get('value_min')}/{row.get('value_max')}")
    review_condition("recovery_rules_v01", "conditional_rule_bounds", invalid_bounds,
                     "Maximum, minimum, range, exact, and published-range types have different required bounds.", "Resolve structure and source interpretation in Step 13.")
    review_condition("recovery_rules_v01", "conditional_rule_segments", invalid_segments,
                     "Age and grade pairs are checked when supplied; a general rule may omit both pairs.", unit="field pairs")
    review_condition("recovery_rules_v01", "rule_source_fields", incomplete_sources,
                     "Source URLs, metrics, units, and recorded verification status are inspected without activating rules.")
    for field in ("rule_type", "metric", "segment", "verification_status"):
        distribution("recovery_rules_v01", field, (row.get(field, "") for row in rules))
    domain_metrics["recovery_rules"] = {"records": len(rules), "invalid_bounds": invalid_bounds,
                                         "invalid_segment_pairs": invalid_segments, "incomplete_source_fields": incomplete_sources}

# We compare the heterogeneous master with actual prepared-domain counts.
master_loaded = "derived_master" in observed
for finding in ("unexpected_entity_type", "missing_lineage_fields", "source_filename_mismatch",
                "invalid_cleaned_position", "repeated_cleaned_position", "invalid_json_payload"):
    affected = master_issues[finding]
    quality_check("derived_master", finding, "failed" if affected or not master_loaded else "pass",
                  affected if master_loaded else 1, "Structural source links and JSON objects are checked; original publisher rows remain unverified.",
                  required=True)
master_domain_metrics = {}
audit_rows = tables.get("derived_audit", {}).get("rows", [])
audit_groups = Counter(row.get("dataset", "") for row in audit_rows)
quality_check("derived_audit", "unique_audit_domains", "failed" if any(not name or n != 1 for name, n in audit_groups.items()) else "pass",
              sum(n - 1 for n in audit_groups.values() if n > 1), required=True, unit="domains")
audit_by_domain = {row.get("dataset", ""): row for row in audit_rows}
for domain, sid in domain_sources.items():
    actual = observed.get(sid, {}).get("rows")
    master_count = master_types.get(domain, 0)
    matched = actual is not None and master_loaded and actual == master_count
    quality_check("derived_master", f"prepared_domain_count:{domain}", "pass" if matched else "failed",
                  abs(actual - master_count) if actual is not None else 1,
                  f"Prepared rows={actual}; master rows={master_count}.", required=True)
    positions = master_positions.get(domain, set())
    if actual is not None:
        valid_positions = {value for value in positions if 1 <= value <= actual}
        position_gaps = actual - len(valid_positions) + len(positions) - len(valid_positions)
        quality_check("derived_master", f"cleaned_position_coverage:{domain}", "failed" if position_gaps else "pass",
                      position_gaps, "Positions must cover each one-based prepared-file row exactly once.", required=True, unit="positions")
        reported = finite_number(audit_by_domain.get(domain, {}).get("clean_rows"))
        audit_match = reported is not None and reported.is_integer() and int(reported) == actual
        quality_check(sid, "earlier_audit_row_count", "pass" if audit_match else "needs_review", 0 if audit_match else 1,
                      f"Earlier audit={reported}; current prepared file={actual}.", unit="datasets")
    master_domain_metrics[domain] = {"prepared_rows": actual, "master_rows": master_count}
distribution("derived_master", "entity_type", (name for name, count in master_types.items() for _ in range(count)))
quality_check("derived_master", "empty_json_payloads", "needs_review" if master_issues["empty_json_payload"] else "pass",
              master_issues["empty_json_payload"], "Missing source values can be omitted from payloads; empty objects still require review.")
quality_check("derived_master", "payload_value_identity", "not_checked", details="Payload values were not compared cell-by-cell with prepared tables; numeric formatting and omitted-null handling need a documented comparison rule.", unit="datasets")
domain_metrics["master"] = {"records": observed.get("derived_master", {}).get("rows"), "entity_counts": dict(master_types),
                            "domains": master_domain_metrics, "structural_findings": dict(master_issues)}
step09_report["domain_complete"] = True
domain_checkpoint = len(checks)
print("Domain reviews completed. Quality findings and content-review requirements remain recorded separately.")
```

```python
# Step 9-3

# We save the findings before enforcing the required inspection outcome.
if not globals().get("step09_report", {}).get("domain_complete"):
    raise RuntimeError("Complete the Step 9 loading and domain-review cells first.")
step09_report["status"] = "preparing_reports"
del checks[domain_checkpoint:]
for path, snapshot in control_snapshots.items():
    if path.read_bytes() != snapshot:
        raise RuntimeError(f"Source documentation changed during inspection. Rerun Step 9: {path}")
for path, expected_state in inspected_file_states.items():
    try:
        current = path.stat()
        stable = (current.st_size, current.st_mtime_ns) == expected_state
    except OSError:
        stable = False
    quality_check("inspection", f"input_unchanged_before_save:{path.relative_to(root).as_posix()}",
                  "pass" if stable else "failed", 0 if stable else 1,
                  "File metadata compared with the end of its inspection.", required=True, unit="files")


def quality_csv(rows, header):
    buffer = io.StringIO(newline="")
    writer = csv.DictWriter(buffer, fieldnames=header, extrasaction="raise")
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode("utf-8")


def save_quality_bundle(outputs):
    prior = {path: path.read_bytes() if path.exists() else None for path in outputs}
    changed = [path for path, content in outputs.items() if prior[path] != content]
    backup_root = project_path("results/backups/step09") / datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S_%fZ")
    for path in changed:
        if prior[path] is not None:
            backup = backup_root / path.relative_to(root)
            backup.parent.mkdir(parents=True, exist_ok=True)
            backup.write_bytes(prior[path])
    written = []
    try:
        for path in changed:
            written.append(path)
            replace_bytes(path, outputs[path])
            if path.read_bytes() != outputs[path]:
                raise OSError(f"Saved output differs from prepared bytes: {path}")
    except Exception as error:
        restore_errors = []
        for path in reversed(written):
            try:
                if prior[path] is None:
                    path.unlink(missing_ok=True)
                else:
                    replace_bytes(path, prior[path])
            except Exception as restore_error:
                restore_errors.append(f"{path}: {restore_error}")
        if restore_errors:
            raise RuntimeError("Some reports could not be restored: " + "; ".join(restore_errors)) from error
        raise
    return {path.relative_to(root).as_posix(): "unchanged" if path not in changed else
            "updated" if prior[path] is not None else "created" for path in outputs}


status_counts = Counter(row["status"] for row in checks)
required_failures = [row for row in checks if row["required"] and row["status"] == "failed"]
flagged = [row for row in checks if row["status"] in {"failed", "needs_review"}]
outcome = "needs_attention" if required_failures else "completed_with_reviews" if flagged else "passed"
summary = dict(step09_report)
summary.update(
    status=outcome, audited_at=quality_audited_at,
    manifest_records=len(manifest), local_files_inspected=sum(row["load_status"] in {"loaded", "verified_document"} for row in inventory),
    structured_datasets_loaded=sum(row["load_status"] == "loaded" for row in inventory),
    historical_inputs_not_found=sum(row["load_status"] == "missing_original" for row in inventory),
    expected_later_outputs=sum(row["load_status"] == "expected_later" for row in inventory),
    check_status_counts=dict(status_counts), total_checks=len(checks), required_failures=len(required_failures),
    source_control_hashes={path.relative_to(root).as_posix(): hashlib.sha256(content).hexdigest() for path, content in control_snapshots.items()},
    domain_metrics=domain_metrics, examples_per_finding_limit=5,
)
outputs = {
    project_path("results/data_inventory.csv"): quality_csv(inventory, list(inventory[0])),
    project_path("results/initial_quality_checks.csv"): quality_csv(checks, ["source_id", "check_name", "status", "affected_count", "details", "next_action", "required", "affected_unit"]),
    project_path("results/column_quality_profile.csv"): quality_csv(column_profiles, ["source_id", "column", "inspected_rows", "missing_cells", "explicit_unknown_cells", "literal_none_cells"]),
    project_path("results/data_category_counts.csv"): quality_csv(category_counts, ["source_id", "column", "value", "count"]),
    project_path("results/data_quality_examples.csv"): quality_csv(quality_examples, ["source_id", "finding", "record_position", "identifier", "details"]),
    project_path("results/step09_quality_summary.json"): (json.dumps(summary, indent=2, allow_nan=False) + "\n").encode("utf-8"),
}
protected_inputs = set(control_snapshots) | {project_path(row["relative_path"]) for row in inventory if row["relative_path"]}
overlap = set(outputs).intersection(protected_inputs)
if overlap:
    raise RuntimeError("A generated report target is also registered as an input; retain a separate source snapshot: " + ", ".join(str(path) for path in sorted(overlap)))
actions = save_quality_bundle(outputs)
step09_report.update(summary)
print(f"Step 9 — Quality summary: {len(inventory)} source entries; {summary['local_files_inspected']} local files; {summary['structured_datasets_loaded']} tables")
print(f"Historical input gaps: {summary['historical_inputs_not_found']} | Expected later outputs absent: {summary['expected_later_outputs']}")
print(f"Check results: {dict(status_counts)} | Required failures: {len(required_failures)}")
print("Exercise summary:", domain_metrics.get("exercise", {}))
print("Nutrition summary:", domain_metrics.get("nutrition", {}))
print("Master entity counts:", dict(master_types))
print("All findings requiring review:")
for row in flagged:
    print(f"{row['source_id']} | {row['check_name']} | {row['status']} | {row['affected_count']} {row['affected_unit']} | {row['details']}")
for relative, action in actions.items():
    print(f"{action.capitalize()}: {relative}")
if required_failures:
    raise RuntimeError("Required Step 9 checks failed. Diagnostic reports were saved; resolve the findings before continuing.")
print("Step 9 inspection completed. Recorded quality findings remain for the following preparation and review steps.")
```

## 10. Exercise Preparation, Label Review, and Grouping

### 10.1 Purpose and Source Identity

We will prepare a source-linked exercise catalog and record the relationships needed for later classification experiments. Our verified cleaned source contains 2,909 records, ten columns, and 17 body-part categories.

Step 9 identified missing descriptions, related names, and different labels within some name groups. We will carry these findings into preparation while preserving the source records and earlier review work.

The cleaned exercise file will remain our parent dataset. Its checksum and dimensions must agree with the completed Step 9 inspection before we build new outputs. The original publisher file remains unavailable in the checked locations, so our preparation will retain that provenance limitation.

Existing identifiers such as `exercise_clean:000001` will continue to identify one-based cleaned-file positions within this source version. A changed or reordered parent file will require explicit reconciliation before those identifiers can be reused.

### 10.2 Prepared Fields and Classifier Text

We will copy every source column into the catalog and create separate prepared fields. Preparation will compact whitespace and standardize categorical capitalization while preserving the original values.

| Information | Role |
|---|---|
| Prepared exercise name and available description | Inputs used to construct `classifier_text` |
| Prepared body-part label | Classification target, retained separately from model inputs |
| Original source columns | Traceability and comparison with the parent dataset |
| Equipment, difficulty, and exercise type | Catalog metadata and review context |
| Record identifiers and source checksums | Provenance |
| Comparison keys and split-group identifiers | Relationship tracking for later dataset partitioning |
| Review fields | Recorded findings, annotations, and decisions |

The model-text function will use an explicit two-field input list: prepared name and prepared description. Existing `agent_text`, derived goal tags, body-part targets, identifiers, and review metadata will remain outside that construction.

We will preserve meaningful wording, punctuation, numbers, and naturally occurring anatomical terms in the feature text. Comparison normalization will be recorded separately. Learned text transformations will be fitted using training data during the modeling steps.

### 10.3 Missing Information and Preliminary Candidacy

We will interpret missing information according to its purpose:

| Field | Preparation decision |
|---|---|
| Name | Flag an unusable name for review and withhold preliminary candidacy |
| Description | Preserve the missing value and retain a name-only candidate when its name and label are available |
| Body-part label | Preserve uncertainty and withhold preliminary supervised candidacy |
| Equipment or difficulty | Preserve unavailable information and record a review finding |
| Ratings | Retain as optional source metadata |
| Review-history fields | Leave empty when no earlier decision exists |

We will distinguish explicit absence from missing equipment information. Descriptions, requirements, and labels will not be filled with invented values.

The preliminary candidate indicator will describe whether usable name and label fields are present. Final class inclusion, reviewed targets, and dataset partitions will be decided in the later preparation and modeling steps. Classification candidacy will remain separate from approval for youth activity recommendations.

### 10.4 Comparison Relationships and Split Groups

We will retain the earlier strict name key and add the broader name key used in Step 9. This will make the difference between the two comparison methods explicit.

| Comparison | Purpose |
|---|---|
| Strict name key | Unicode, whitespace, and case comparison while retaining punctuation |
| Broader name key | Alphanumeric comparison consistent with Step 9 |
| Shared nonempty description | Identify repeated instructional text across names or variations |
| Strict and broader classifier-input comparisons | Review repeated or closely normalized name-and-description inputs |
| Content comparisons excluding the target | Examine prepared content without letting different labels hide a match |

We will form conservative split groups from matching broader names, shared nonempty descriptions, and identical normalized classifier input. Connected records will remain in the same group even when their relationship is indirect.

Labels, equipment, and difficulty will not determine split-group membership. Label differences will be measured after grouping and recorded for review. Metadata-based content comparisons will remain separate review relationships.

Missing descriptions will not link records together. Group identifiers will be deterministic for the source version, preparation recipe, and member identifiers.

These groups will guide later partitioning to reduce overlap between training and evaluation examples. They will preserve all exercise records and will not establish canonical identity, remove duplicates, or assign training, validation, or test membership at this step.

### 10.5 Label Review and Existing Decisions

We will retain original body-part labels and keep their prepared formatting separate. Different labels within a comparison group may reflect variations, multiple muscle groups, or source inconsistencies. We will record the evidence before deciding whether a correction is justified.

A low-support alert of fewer than ten source examples will identify classes needing attention. It will not automatically remove or relabel their records. Distinct split groups per label will provide additional context for later partitioning.

Existing catalog annotations will be matched by `catalog_id`. Review-queue entries will be matched by catalog ID and issue type. We will preserve additional annotation columns and custom findings.

Where evidence remains equivalent, earlier review statuses will be carried forward. A decision whose supporting evidence changes will be marked for renewed review, with its earlier status and evidence retained. A finding no longer detected by the current recipe will remain available as history without being automatically closed.

If an existing catalog contains a substantive edit to an original or protected prepared value, we will stop for reconciliation. Regeneration will not silently overwrite a reviewed correction or convert a pending decision into approval.

Observed findings and unresolved decisions will be counted separately. A reviewed finding can remain visible as a property of the data while its recorded handling decision is preserved.

### 10.6 Outputs and Producer Record

We will produce:

| Output | Purpose |
|---|---|
| `data/processed/exercise_catalog.csv` | Source values, prepared text, preliminary candidacy, comparison groups, and review fields |
| `results/exercise_cleaning_log.csv` | Formatting transformations and retained preparation history |
| `results/exercise_review_queue.csv` | Current, custom, and historical findings with evidence-bound review states |
| `results/exercise_group_membership.csv` | Comparison relationships and the complete split-group mapping |
| `results/exercise_review_carryover.csv` | How earlier annotations and decisions were preserved or marked for renewed review |
| `results/step10_exercise_preparation.json` | Recipe, parent-source identity, output hashes and schemas, checks, metrics, and previous-version references |

The producer record will identify the newly prepared catalog for Step 11. Step 8's historical catalog checksum will remain part of the earlier source record. The new artifact's identity will be established by its producing step and parent-source link.

We will validate record retention, identifier uniqueness, source-value preservation, feature construction, and group consistency before saving. Existing outputs will be backed up before replacement. Intervening edits will stop the save, and a failed update will trigger restoration of affected files.

Manual review edits can be incorporated by rerunning these cells. The producer record will then capture the updated output hashes alongside preserved decisions.

### 10.7 Handoff to Later Steps

The four code cells will verify and prepare the source, build comparison groups, reconcile earlier reviews, and save the artifacts.

```python
# Step 10-1

# We prepare a derived exercise catalog while retaining its source values.
import csv
import hashlib
import io
import json
import re
import unicodedata
from collections import Counter, defaultdict
from datetime import datetime, timezone

step10_report = {"status": "incomplete", "prepared": False, "grouped": False, "review_merge_complete": False}
if globals().get("step09_report", {}).get("status") not in {"passed", "completed_with_reviews"}:
    raise RuntimeError("Complete the Step 9 quality inspection before preparing exercises.")
if not all(callable(globals().get(name)) for name in ("project_path", "read_table", "sha256_file", "strict_json", "is_missing", "normalize_name", "replace_bytes")):
    raise RuntimeError("Run the preceding source and quality cells to load our shared functions.")
root = project_root.resolve()
RECIPE_VERSION = "exercise-preparation-0.2.0"
LOW_SUPPORT_ALERT = 10
preparation_started_at = datetime.now(timezone.utc).isoformat()
control_files = [project_path(name) for name in (
    "data/source_manifest.csv", "results/data_inventory.csv", "results/step09_quality_summary.json")]
preparation_controls = {path: path.read_bytes() for path in control_files}
source_register, _ = read_table(control_files[0])
quality_inventory, _ = read_table(control_files[1])
quality_summary = strict_json(preparation_controls[control_files[2]].decode("utf-8-sig"))
if not isinstance(quality_summary, dict) or quality_summary.get("status") not in {"passed", "completed_with_reviews"} or quality_summary.get("required_failures") != 0:
    raise RuntimeError("The saved Step 9 inspection has unresolved required failures.")
for relative, expected_hash in quality_summary["source_control_hashes"].items():
    path = project_path(relative)
    contents = path.read_bytes()
    if hashlib.sha256(contents).hexdigest() != expected_hash:
        raise RuntimeError(f"Source documentation changed after Step 9: {relative}")
    preparation_controls[path] = contents
sources = [row for row in source_register if row["source_id"] == "derived_exercise"]
observations = [row for row in quality_inventory if row["source_id"] == "derived_exercise"]
if len(sources) != 1 or len(observations) != 1 or observations[0]["load_status"] != "loaded":
    raise ValueError("Expected one successfully inspected cleaned exercise source.")
exercise_source = sources[0]
source_relative = observations[0]["relative_path"]
source_path = project_path(source_relative)
source_hash = sha256_file(source_path)
if source_hash != observations[0]["observed_sha256"] or source_hash != exercise_source["sha256"]:
    raise RuntimeError("The exercise source differs from its verified version. Reconcile it through Steps 8–9 first.")
exercise_rows, source_columns = read_table(source_path)
if sha256_file(source_path) != source_hash:
    raise RuntimeError("The exercise source changed while being loaded.")
PREPARED_FIELDS = ["exercise_name", "description", "body_part", "exercise_type", "equipment", "difficulty_level"]
if not set(PREPARED_FIELDS).issubset(source_columns) or not exercise_rows:
    raise ValueError("The exercise source is empty or lacks required preparation fields.")
if len(exercise_rows) != int(observations[0]["actual_rows"]) or len(exercise_rows) != int(exercise_source["reported_records"]):
    raise ValueError("Exercise row counts differ from the recorded source version.")

catalog_path = project_path("data/processed/exercise_catalog.csv")
log_path = project_path("results/exercise_cleaning_log.csv")
queue_path = project_path("results/exercise_review_queue.csv")
groups_path = project_path("results/exercise_group_membership.csv")
review_audit_path = project_path("results/exercise_review_carryover.csv")
handoff_path = project_path("results/step10_exercise_preparation.json")
preparation_targets = [catalog_path, log_path, queue_path, groups_path, review_audit_path, handoff_path]
if len(set(preparation_targets)) != len(preparation_targets):
    raise ValueError("Preparation artifacts must use distinct output paths.")
if set(preparation_targets).intersection(set(preparation_controls) | {source_path}):
    raise ValueError("A preparation output would overwrite a source or control input.")
output_snapshots = {path: path.read_bytes() if path.exists() else None for path in preparation_targets}


def existing_table(path):
    if output_snapshots[path] is None:
        return None, []
    rows, header = read_table(path)
    if path.read_bytes() != output_snapshots[path]:
        raise RuntimeError(f"An output changed while loading: {path}")
    return rows, header


existing_catalog, existing_catalog_columns = existing_table(catalog_path)
existing_queue, existing_queue_columns = existing_table(queue_path)
existing_log, existing_log_columns = existing_table(log_path)


def compact_text(value):
    return re.sub(r"\s+", " ", value).strip()


def prepare_value(field, value):
    text = compact_text(value)
    if is_missing(text, field):
        return ""
    return text.casefold() if field in {"body_part", "exercise_type", "equipment", "difficulty_level"} else text


def strict_comparison_key(value):
    return unicodedata.normalize("NFKC", compact_text(value)).casefold()


def evidence_digest(value):
    content = json.dumps(value, ensure_ascii=False, sort_keys=True, separators=(",", ":"))
    return hashlib.sha256(content.encode("utf-8")).hexdigest()


def classifier_input(row):
    return "\n\n".join(value for value in (row["exercise_name_prepared"], row["description_prepared"]) if value)


def make_issue(row, issue_type, priority, evidence, action, context=None):
    return {"catalog_id": row["catalog_id"], "issue_type": issue_type, "priority": priority,
            "evidence": evidence, "suggested_action": action, "review_status": "pending",
            "evidence_sha256": evidence_digest({"source_record": row["source_record_evidence_sha256"],
                                               "issue_type": issue_type, "context": context or {}}),
            "finding_state": "observed", "prior_review_status": "", "prior_evidence_sha256": ""}


def build_prepared_exercises(rows, original_columns, source_relative, source_hash, recipe_version):
    derived_columns = {
        "catalog_id", "source_cleaned_record_position", "source_file", "source_sha256",
        *(f"{name}_prepared" for name in PREPARED_FIELDS), "name_comparison_key", "label_review_status",
        "preliminary_classifier_candidate", "review_flags", "classifier_text", "text_availability",
        "name_group_key", "preparation_version", "source_record_evidence_sha256", "label_review_evidence_sha256",
        "prior_label_review_status", "prior_label_review_evidence_sha256", "split_group_id", "split_group_size", "split_group_label_count",
        "name_group_id", "name_group_size", "name_group_label_count", "description_group_id", "description_group_size",
    }
    if derived_columns.intersection(original_columns):
        raise ValueError("Source columns conflict with derived catalog field names.")
    prepared, log, queue = [], [], []
    labels = Counter(prepare_value("body_part", row["body_part"]) for row in rows)
    for position, original in enumerate(rows, 1):
        row = dict(original)
        row.update(catalog_id=f"exercise_clean:{position:06d}", source_cleaned_record_position=str(position),
                   source_file=source_relative, source_sha256=source_hash, preparation_version=recipe_version,
                   label_review_status="original_source_verification_pending", prior_label_review_status="",
                   source_record_evidence_sha256=evidence_digest({"source_sha256": source_hash,
                       "position": position, "values": [original[name] for name in original_columns]}),
                   label_review_evidence_sha256="", review_flags="")
        for field in PREPARED_FIELDS:
            value = prepare_value(field, original[field])
            row[f"{field}_prepared"] = value
            if value != original[field]:
                log.append({"catalog_id": row["catalog_id"], "field": field, "source_value": original[field],
                            "prepared_value": value, "reason": "Whitespace/category normalization or null-like marker",
                            "decision": "Prepared copy only; source preserved"})
        name, description, label = (row[f"{field}_prepared"] for field in ("exercise_name", "description", "body_part"))
        row["name_comparison_key"] = strict_comparison_key(name)
        row["name_group_key"] = normalize_name(name)
        row["classifier_text"] = classifier_input(row)
        row["text_availability"] = "name_and_description" if name and description else "name_only" if name else "unusable_name"
        row["preliminary_classifier_candidate"] = "yes" if row["name_group_key"] and label else "no"
        findings = []
        if not row["name_group_key"]:
            findings.append(("missing_exercise_name", "high", "No usable exercise title", "Review source before classifier inclusion"))
        if not description:
            findings.append(("missing_description", "medium", "Exercise description unavailable", "Retain title-only candidate; do not invent text"))
        if not label:
            findings.append(("missing_body_part", "high", "Body-part label unavailable", "Exclude from supervised targets pending review"))
        for field in ("equipment", "difficulty_level"):
            if not row[f"{field}_prepared"]:
                findings.append((f"missing_{field}", "medium", f"{field} unavailable", "Preserve uncertainty and review before planning"))
        if label and labels[label] < LOW_SUPPORT_ALERT:
            findings.append(("low_support_label", "medium", f"{label}: {labels[label]} examples", "Review class eligibility in Step 15"))
        for issue_type, priority, evidence, action in findings:
            queue.append(make_issue(row, issue_type, priority, evidence, action,
                                    {"prepared_name": name, "prepared_label": label, "evidence": evidence}))
        prepared.append(row)
    return prepared, log, queue


catalog, cleaning_log, generated_queue = build_prepared_exercises(
    exercise_rows, source_columns, source_relative, source_hash, RECIPE_VERSION)
step10_report.update(prepared=True, status="prepared_for_grouping")
print(f"Step 10 — Prepared {len(catalog)} source-linked exercise records")
print(f"Source columns: {len(source_columns)} | Formatting changes: {len(cleaning_log)}")
print("Classifier text uses prepared exercise names and available descriptions.")
```

```python
# Step 10-2

# We keep comparison groups separate from decisions about exercise identity.
import hashlib
import json
import re
import unicodedata
from collections import defaultdict


def build_exercise_groups(catalog, source_hash, recipe_version):
    """Add derived grouping fields and return deterministic membership records."""
    if not isinstance(source_hash, str) or not source_hash.strip():
        raise ValueError("A source hash is required for versioned exercise groups.")
    if not isinstance(recipe_version, str) or not recipe_version.strip():
        raise ValueError("A preparation recipe version is required.")
    required = {"catalog_id", "exercise_name_prepared", "description_prepared",
                "exercise_type_prepared", "equipment_prepared", "difficulty_level_prepared",
                "body_part_prepared", "name_group_key", "classifier_text"}
    for row in catalog:
        if not isinstance(row, dict) or not required.issubset(row):
            raise ValueError("A catalog record lacks fields required for grouping.")
        if any(not isinstance(row[name], str) for name in required):
            raise ValueError("Grouping input fields must retain their prepared text values.")
        if not row["catalog_id"] or row["catalog_id"] != row["catalog_id"].strip():
            raise ValueError("Catalog identifiers must be nonempty and unpadded.")
    if len({row["catalog_id"] for row in catalog}) != len(catalog):
        raise ValueError("Catalog identifiers must be unique before grouping.")

    def strict_text(value):
        return unicodedata.normalize("NFKC", re.sub(r"\s+", " ", value).strip()).casefold()

    def broad_text(value):
        return " ".join(re.findall(r"[a-z0-9]+", value.lower()))

    def canonical(value):
        return json.dumps(value, ensure_ascii=False, sort_keys=True, separators=(",", ":"))

    def digest(value):
        return hashlib.sha256(canonical(value).encode("utf-8")).hexdigest()

    id_payloads = {}

    def group_id(relation, comparison_digest):
        payload = canonical({"recipe_version": recipe_version, "source_sha256": source_hash,
                             "relation_type": relation, "comparison_digest": comparison_digest})
        identifier = "exgroup:" + hashlib.sha256(payload.encode("utf-8")).hexdigest()
        if identifier in id_payloads and id_payloads[identifier] != payload:
            raise RuntimeError("A group identifier collision was detected.")
        id_payloads[identifier] = payload
        return identifier

    ordered = sorted(catalog, key=lambda row: row["catalog_id"])
    by_id = {row["catalog_id"]: row for row in ordered}
    parent = {rid: rid for rid in by_id}

    def find(rid):
        while parent[rid] != rid:
            parent[rid] = parent[parent[rid]]
            rid = parent[rid]
        return rid

    def join(left, right):
        a, b = find(left), find(right)
        if a != b:
            parent[max(a, b)] = min(a, b)

    relation_order = (
        "broad_name", "shared_description", "strict_classifier_input",
        "broad_classifier_input", "strict_content_without_label", "broad_content_without_label",
    )
    relations = {name: defaultdict(list) for name in relation_order}
    null_description_keys = {"", "nan", "null", "none", "n/a"}
    content_fields = ("exercise_name_prepared", "description_prepared", "exercise_type_prepared",
                      "equipment_prepared", "difficulty_level_prepared")
    for row in ordered:
        rid = row["catalog_id"]
        name_key = broad_text(row["exercise_name_prepared"])
        if row["name_group_key"] != name_key:
            raise ValueError(f"The broader name key differs from the declared rule: {rid}")
        if name_key:
            relations["broad_name"][name_key].append(rid)
        description_key = strict_text(row["description_prepared"])
        if description_key not in null_description_keys:
            relations["shared_description"][description_key].append(rid)
        strict_input = strict_text(row["classifier_text"])
        if strict_input:
            relations["strict_classifier_input"][strict_input].append(rid)
        broad_input = broad_text(row["classifier_text"])
        if broad_input:
            relations["broad_classifier_input"][broad_input].append(rid)
        strict_content = tuple(strict_text(row[name]) for name in content_fields)
        broad_content = tuple(broad_text(row[name]) for name in content_fields)
        relations["strict_content_without_label"][strict_content].append(rid)
        relations["broad_content_without_label"][broad_content].append(rid)

    # We link repeated names, shared nonempty descriptions, and identical input text.
    # Labels, equipment, and difficulty never determine split-group membership.
    split_relations = {"broad_name", "shared_description", "strict_classifier_input"}
    for relation in relation_order:
        if relation in split_relations:
            for members in relations[relation].values():
                for rid in members[1:]:
                    join(members[0], rid)

    components = defaultdict(list)
    for rid in by_id:
        components[find(rid)].append(rid)

    def labels_for(members):
        return sorted({by_id[rid]["body_part_prepared"] for rid in members
                       if by_id[rid]["body_part_prepared"].strip()})

    updates = {rid: {"name_group_id": "", "name_group_size": "0", "name_group_label_count": "0",
                     "description_group_id": "", "description_group_size": "0"} for rid in by_id}
    group_rows = []
    relation_metrics = {}
    for relation in relation_order:
        repeated = [(key, members) for key, members in relations[relation].items() if len(members) > 1]
        relation_metrics[relation] = {
            "repeated_groups": len(repeated),
            "records_in_repeated_groups": sum(len(members) for _, members in repeated),
            "additional_records": sum(len(members) - 1 for _, members in repeated),
            "different_label_groups": sum(len(labels_for(members)) > 1 for _, members in repeated),
            "largest_group": max((len(members) for _, members in repeated), default=0),
        }
        for key, members in sorted(relations[relation].items(), key=lambda item: canonical(item[0])):
            members = sorted(members)
            comparison_digest = digest(key)
            identifier = group_id(relation, comparison_digest)
            labels = labels_for(members)
            if relation == "broad_name":
                for rid in members:
                    updates[rid].update(name_group_id=identifier, name_group_size=str(len(members)),
                                        name_group_label_count=str(len(labels)))
            elif relation == "shared_description":
                for rid in members:
                    updates[rid].update(description_group_id=identifier, description_group_size=str(len(members)))
            if len(members) > 1:
                for rid in members:
                    group_rows.append({"relation_type": relation, "group_id": identifier, "catalog_id": rid,
                                       "member_count": len(members), "labels_json": canonical(labels),
                                       "comparison_digest": comparison_digest,
                                       "link_role": "split_link" if relation in split_relations else "review_only"})

    multiple = [members for members in components.values() if len(members) > 1]
    split_metrics = {
        "total_groups": len(components), "singleton_groups": len(components) - len(multiple),
        "multi_record_groups": len(multiple), "records_in_multi_record_groups": sum(map(len, multiple)),
        "largest_group": max((len(members) for members in components.values()), default=0),
        "different_label_groups": sum(len(labels_for(members)) > 1 for members in components.values()),
        "total_memberships": len(catalog),
    }
    for members in sorted(components.values(), key=lambda values: tuple(sorted(values))):
        members = sorted(members)
        comparison_digest = digest(members)
        identifier = group_id("split_component", comparison_digest)
        labels = labels_for(members)
        for rid in members:
            updates[rid].update(split_group_id=identifier, split_group_size=str(len(members)),
                                split_group_label_count=str(len(labels)))
            group_rows.append({"relation_type": "split_component", "group_id": identifier, "catalog_id": rid,
                               "member_count": len(members), "labels_json": canonical(labels),
                               "comparison_digest": comparison_digest, "link_role": "split_link"})

    for row in catalog:
        row.update(updates[row["catalog_id"]])
    metrics = {"source_sha256": source_hash, "recipe_version": recipe_version,
               "relation_metrics": relation_metrics, "split_metrics": split_metrics}
    return group_rows, metrics

# We retain the records and expose their comparison and split relationships.
if isinstance(globals().get("step10_report"), dict):
    step10_report.update(grouped=False, review_merge_complete=False)
if not globals().get("step10_report", {}).get("prepared"):
    raise RuntimeError("Complete the exercise preparation cell first.")

# We recreate generated findings on reruns instead of appending duplicate entries.
catalog, cleaning_log, generated_queue = build_prepared_exercises(
    exercise_rows, source_columns, source_relative, source_hash, RECIPE_VERSION)
group_membership, grouping_metrics = build_exercise_groups(catalog, source_hash, RECIPE_VERSION)
catalog_by_id = {row["catalog_id"]: row for row in catalog}
relation_groups = defaultdict(list)
for membership in group_membership:
    relation_groups[(membership["relation_type"], membership["group_id"])].append(membership)
detectors = {
    "broad_name": ("repeated_normalized_name", "medium", "Review related names without merging records."),
    "shared_description": ("shared_description", "medium", "Keep shared descriptions together in later partitions; review their source context."),
    "strict_classifier_input": ("identical_classifier_input", "medium", "Review repeated model-input text and preserve its split grouping."),
    "broad_classifier_input": ("normalized_classifier_input", "medium", "Review input text that matches after comparison normalization."),
    "strict_content_without_label": ("duplicate_meaningful_content", "medium", "Compare equivalent source content before any consolidation."),
}
for (relation, group_id), members in sorted(relation_groups.items()):
    if len(members) <= 1:
        continue
    labels = json.loads(members[0]["labels_json"])
    context = {"relation_type": relation, "group_id": group_id,
               "catalog_ids": sorted(item["catalog_id"] for item in members), "labels": labels}
    if relation in detectors:
        issue_type, priority, action = detectors[relation]
        for member in members:
            row = catalog_by_id[member["catalog_id"]]
            generated_queue.append(make_issue(row, issue_type, priority,
                f"{relation}: {len(members)} records; group={group_id}; labels={labels}", action, context))
    if relation == "broad_name" and len(labels) > 1:
        for member in members:
            row = catalog_by_id[member["catalog_id"]]
            generated_queue.append(make_issue(row, "conflicting_labels_for_name", "high",
                f"Name group has labels {labels}; group={group_id}",
                "Review source labels and exercise variations without choosing a replacement label automatically.", context))
    if relation == "split_component" and len(labels) > 1:
        for member in members:
            row = catalog_by_id[member["catalog_id"]]
            generated_queue.append(make_issue(row, "mixed_label_split_group", "high",
                f"Conservative split group contains labels {labels}; group={group_id}",
                "Review label evidence while keeping connected records in one later partition.", context))

for row in catalog:
    row["label_review_evidence_sha256"] = evidence_digest({
        "source_record": row["source_record_evidence_sha256"], "label": row["body_part_prepared"],
        "name_group": row["name_group_id"], "split_group": row["split_group_id"],
        "name_group_label_count": row["name_group_label_count"], "split_group_label_count": row["split_group_label_count"],
    })
class_groups = defaultdict(set)
for row in catalog:
    if row["body_part_prepared"]:
        class_groups[row["body_part_prepared"]].add(row["split_group_id"])
grouping_metrics["distinct_split_groups_by_label"] = {label: len(values) for label, values in sorted(class_groups.items())}
strict_names = Counter(row["name_comparison_key"] for row in catalog if row["name_comparison_key"])
grouping_metrics["strict_name_repeated_groups"] = sum(count > 1 for count in strict_names.values())
step10_report.update(grouped=True, status="grouped_for_review")
print("Comparison groups:", grouping_metrics["relation_metrics"])
print("Conservative split groups:", grouping_metrics["split_metrics"])
print("We recorded comparison links for later splitting; exercise records remain separate.")
```

```python
# Step 10-3

# We carry review work forward only when its source and evidence still agree.
from collections import Counter
import json


def merge_exercise_reviews(catalog, queue, existing_catalog, existing_queue,
                          source_columns, source_hash):
    """Merge annotations into generated copies without applying source corrections.

    None denotes a missing previous artifact; an empty supplied artifact is
    treated as an existing artifact. Inputs are left unchanged.
    """
    audit = []
    metrics = Counter()
    prepared_fields = [f"{field}_prepared" for field in (
        "exercise_name", "description", "body_part", "exercise_type",
        "equipment", "difficulty_level")]
    unresolved = {"", "pending", "pending_review", "open", "unreviewed", "not_reviewed",
                  "needs_review", "needs_re_review", "awaiting_review", "review_required",
                  "original_source_verification_pending", "source_verification_pending"}
    legacy_issues = {"missing_description", "missing_equipment", "low_support_label",
                     "missing_exercise_name", "missing_body_part", "missing_difficulty_level"}
    automated_issues = legacy_issues | {"repeated_normalized_name", "conflicting_labels_for_name",
                                        "duplicate_meaningful_content", "shared_description",
                                        "mixed_label_split_group", "identical_classifier_input",
                                        "normalized_classifier_input"}

    def text(row, field):
        value = row.get(field, "")
        if not isinstance(value, str):
            raise ValueError(f"Review field {field!r} must contain text.")
        return value

    def record(entity, key, change, before, after, reason):
        audit.append({"entity_type": entity, "key": key, "change": change,
                      "previous_value": str(before), "current_value": str(after), "reason": reason})

    def index_catalog(rows, label):
        indexed = {}
        for row in rows:
            key = text(row, "catalog_id")
            if not key or key != key.strip() or key in indexed:
                raise ValueError(f"{label} contains an empty, padded, or duplicate catalog ID: {key!r}")
            indexed[key] = row
        return indexed

    def index_queue(rows, label, valid_ids):
        indexed = {}
        for row in rows:
            key = (text(row, "catalog_id"), text(row, "issue_type"))
            if any(not value or value != value.strip() for value in key) or key in indexed:
                raise ValueError(f"{label} contains an empty, padded, or duplicate issue key: {key!r}")
            if key[0] not in valid_ids:
                raise ValueError(f"{label} refers to an unknown catalog ID: {key[0]}")
            indexed[key] = row
        return indexed

    generated_catalog = index_catalog(catalog, "Generated catalog")
    if any(text(row, "source_sha256") != source_hash for row in generated_catalog.values()):
        raise ValueError("Generated catalog source hashes do not match the verified parent source.")
    prior_catalog = index_catalog(existing_catalog or [], "Existing catalog")
    source_bound = existing_catalog is not None
    if source_bound and set(prior_catalog) != set(generated_catalog):
        raise ValueError("Existing catalog IDs differ from the generated source-derived records; reconcile them before updating.")
    if source_bound:
        for key, previous in prior_catalog.items():
            generated = generated_catalog[key]
            if text(previous, "source_sha256") != source_hash:
                raise ValueError(f"Existing catalog source identity differs for {key}; use an explicit source-version migration.")
            for field in list(source_columns) + prepared_fields + ["source_cleaned_record_position"]:
                if field not in previous or text(previous, field) != text(generated, field):
                    raise ValueError(f"Existing source or prepared value differs for {key}, field {field}; preserve and reconcile the edit.")

    merged_catalog = []
    catalog_history = {"prior_label_review_status", "prior_label_review_evidence_sha256"}
    for generated in catalog:
        row = dict(generated)
        key = text(row, "catalog_id")
        row.setdefault("prior_label_review_status", "")
        row.setdefault("prior_label_review_evidence_sha256", "")
        current_fingerprint = text(row, "label_review_evidence_sha256")
        if not current_fingerprint:
            raise ValueError(f"Generated label-review evidence is missing for {key}.")
        previous = prior_catalog.get(key)
        if previous is None:
            metrics["new_catalog_records"] += 1
        else:
            metrics["existing_catalog_records"] += 1
            for field, value in previous.items():
                if field not in generated or field in catalog_history:
                    row[field] = value
                    if field not in generated and field not in catalog_history and value:
                        record("catalog", key, "annotation_preserved", "", value, f"Preserved column: {field}")
            old_status = text(previous, "label_review_status")
            old_fingerprint = text(previous, "label_review_evidence_sha256")
            same_evidence = bool(old_fingerprint) and old_fingerprint == current_fingerprint
            if same_evidence or old_status.strip().lower() in unresolved:
                row["label_review_status"] = old_status
                metrics["catalog_statuses_carried"] += 1
                metrics["catalog_fingerprint_matches" if same_evidence else "catalog_unresolved_statuses_carried"] += 1
                if old_status != generated.get("label_review_status", ""):
                    record("catalog", key, "review_status_preserved", generated.get("label_review_status", ""), old_status,
                           "Evidence matches, or the carried status remains unresolved.")
            else:
                row.update(label_review_status="needs_re_review", prior_label_review_status=old_status,
                           prior_label_review_evidence_sha256=old_fingerprint)
                metrics["stale_catalog_decisions"] += 1
                record("catalog", key, "review_requires_recheck", old_status, "needs_re_review",
                       "The earlier decision has no matching current evidence fingerprint.")
            if old_fingerprint != current_fingerprint:
                record("catalog", key, "evidence_fingerprint_updated", old_fingerprint, current_fingerprint,
                       "Current evidence is recorded without declaring its label independently verified.")
        merged_catalog.append(row)

    generated_queue = index_queue(queue, "Generated queue", set(generated_catalog))
    prior_queue = index_queue(existing_queue or [], "Existing queue", set(generated_catalog))
    merged_queue = []
    queue_history = {"prior_review_status", "prior_evidence_sha256", "prior_evidence"}
    for key, generated in generated_queue.items():
        row = dict(generated)
        row.setdefault("prior_review_status", "")
        row.setdefault("prior_evidence_sha256", "")
        row.setdefault("prior_evidence", "")
        row["finding_state"] = "observed"
        current_fingerprint = text(row, "evidence_sha256")
        if not current_fingerprint:
            raise ValueError(f"Generated finding evidence is missing for {key!r}.")
        previous = prior_queue.get(key)
        audit_key = json.dumps(list(key), ensure_ascii=False)
        if previous is None:
            row.setdefault("review_status", "pending")
            metrics["new_observed_findings"] += 1
            record("queue", audit_key, "new_finding", "", text(row, "review_status"), "Current recipe detected this finding.")
        else:
            for field, value in previous.items():
                if field not in generated or field in queue_history:
                    row[field] = value
                    if field not in generated and field not in queue_history and value:
                        record("queue", audit_key, "annotation_preserved", "", value, f"Preserved column: {field}")
            row["finding_state"] = "observed"
            old_status = text(previous, "review_status")
            old_fingerprint = text(previous, "evidence_sha256")
            same_evidence = bool(old_fingerprint) and old_fingerprint == current_fingerprint
            legacy_equivalent = (not old_fingerprint and key[1] in legacy_issues and source_bound
                                 and text(previous, "evidence") == text(generated, "evidence"))
            if same_evidence or legacy_equivalent or old_status.strip().lower() in unresolved:
                row["review_status"] = old_status
                metrics["queue_statuses_carried"] += 1
                if same_evidence:
                    metrics["queue_fingerprint_matches"] += 1
                elif legacy_equivalent:
                    metrics["legacy_equivalent_findings_carried"] += 1
                else:
                    metrics["queue_unresolved_statuses_carried"] += 1
                if old_status != generated.get("review_status", ""):
                    record("queue", audit_key, "review_status_preserved", generated.get("review_status", ""), old_status,
                           "Current evidence agrees, a source-bound legacy finding agrees, or the status remains unresolved.")
            else:
                row.update(review_status="needs_re_review", prior_review_status=old_status,
                           prior_evidence_sha256=old_fingerprint, prior_evidence=text(previous, "evidence"))
                metrics["stale_queue_decisions"] += 1
                record("queue", audit_key, "review_requires_recheck", old_status, "needs_re_review",
                       "The prior decision does not have matching current evidence.")
            if old_fingerprint != current_fingerprint:
                record("queue", audit_key, "evidence_fingerprint_updated", old_fingerprint, current_fingerprint,
                       "The generated finding now records its current evidence fingerprint.")
        merged_queue.append(row)

    for key, previous in prior_queue.items():
        if key in generated_queue:
            continue
        row = dict(previous)
        row.setdefault("prior_review_status", "")
        row.setdefault("prior_evidence_sha256", "")
        row.setdefault("prior_evidence", "")
        row.setdefault("evidence_sha256", "")
        prior_state = text(previous, "finding_state")
        automated = key[1] in automated_issues
        row["finding_state"] = "not_detected_by_current_recipe" if automated else "retained_custom"
        old_status = text(previous, "review_status")
        if not automated and not source_bound and old_status.strip().lower() not in unresolved:
            row.update(review_status="needs_re_review", prior_review_status=old_status,
                       prior_evidence_sha256=text(previous, "evidence_sha256"), prior_evidence=text(previous, "evidence"))
            metrics["stale_queue_decisions"] += 1
            record("queue", json.dumps(list(key), ensure_ascii=False), "review_requires_recheck", old_status,
                   "needs_re_review", "The retained custom decision lacks a verified previous catalog binding.")
        metrics["historical_automated_findings" if automated else "retained_custom_findings"] += 1
        record("queue", json.dumps(list(key), ensure_ascii=False), "prior_finding_retained", prior_state,
               row["finding_state"], "The prior finding remains recorded; its review status was not automatically closed.")
        merged_queue.append(row)

    # We retain a consistent set of annotation columns across each output table.
    for rows in (merged_catalog, merged_queue):
        columns = list(dict.fromkeys(field for row in rows for field in row))
        for row in rows:
            for field in columns:
                row.setdefault(field, "")
    result_metrics = dict(metrics)
    for name in ("new_catalog_records", "existing_catalog_records", "catalog_statuses_carried",
                 "catalog_fingerprint_matches", "catalog_unresolved_statuses_carried", "stale_catalog_decisions",
                 "new_observed_findings", "queue_statuses_carried", "queue_fingerprint_matches",
                 "legacy_equivalent_findings_carried", "queue_unresolved_statuses_carried", "stale_queue_decisions",
                 "historical_automated_findings", "retained_custom_findings"):
        result_metrics.setdefault(name, 0)
    result_metrics.update(catalog_records=len(merged_catalog), queue_records=len(merged_queue),
                          catalog_status_counts=dict(Counter(row["label_review_status"] for row in merged_catalog)),
                          queue_status_counts=dict(Counter(row["review_status"] for row in merged_queue)),
                          finding_state_counts=dict(Counter(row["finding_state"] for row in merged_queue)))
    return merged_catalog, merged_queue, audit, result_metrics

# We carry review decisions and annotations forward before saving new artifacts.
if isinstance(globals().get("step10_report"), dict):
    step10_report["review_merge_complete"] = False
if not globals().get("step10_report", {}).get("grouped"):
    raise RuntimeError("Complete the grouping cell before reconciling reviews.")

catalog, review_queue, review_carryover, review_metrics = merge_exercise_reviews(
    catalog, generated_queue, existing_catalog, existing_queue, source_columns, source_hash)
active_states = {"observed", "retained_custom"}
active_findings = [row for row in review_queue if row["finding_state"] in active_states]
flags_by_id = defaultdict(set)
for finding in active_findings:
    flags_by_id[finding["catalog_id"]].add(finding["issue_type"])
for row in catalog:
    row["review_flags"] = ";".join(sorted(flags_by_id[row["catalog_id"]]))

# Earlier formatting entries remain available as history if a recipe stops emitting them.
log_index = {(row["catalog_id"], row["field"]): row for row in cleaning_log}
current_log_keys = set(log_index)
if len(log_index) != len(cleaning_log):
    raise ValueError("Generated formatting entries have duplicate keys.")
catalog_ids = {row["catalog_id"] for row in catalog}
old_log_keys = set()
for previous in existing_log or []:
    key = (previous.get("catalog_id", ""), previous.get("field", ""))
    if not all(key) or key in old_log_keys or key[0] not in catalog_ids:
        raise ValueError("Existing formatting log has an invalid or repeated record/field key.")
    old_log_keys.add(key)
    if key in log_index:
        for field, value in previous.items():
            if field not in log_index[key]:
                log_index[key][field] = value
    else:
        retained = dict(previous)
        retained["record_state"] = "historical_preparation"
        log_index[key] = retained
cleaning_log = list(log_index.values())
for row in cleaning_log:
    row["record_state"] = "observed" if (row["catalog_id"], row["field"]) in current_log_keys else "historical_preparation"

if len(catalog) != len(exercise_rows) or len(catalog_ids) != len(exercise_rows):
    raise RuntimeError("Record retention or catalog-identifier uniqueness failed.")
for row in catalog:
    position = int(row["source_cleaned_record_position"])
    original = exercise_rows[position - 1]
    if any(row[name] != original[name] for name in source_columns):
        raise RuntimeError(f"An original source field changed: {row['catalog_id']}")
    if row["classifier_text"] != classifier_input(row):
        raise RuntimeError(f"Classifier text differs from the allowed source fields: {row['catalog_id']}")
    if row["source_sha256"] != source_hash or row["preparation_version"] != RECIPE_VERSION:
        raise RuntimeError("Catalog source or recipe identity is inconsistent.")
if sha256_file(source_path) != source_hash:
    raise RuntimeError("The exercise source changed during preparation. Rerun after source reconciliation.")
split_members = [row for row in group_membership if row["relation_type"] == "split_component"]
if len(split_members) != len(catalog) or {row["catalog_id"] for row in split_members} != catalog_ids:
    raise RuntimeError("Each catalog record must have exactly one conservative split-group membership.")
split_by_id = {row["catalog_id"]: row["split_group_id"] for row in catalog}
for (relation, group_id), members in relation_groups.items():
    if members[0]["link_role"] == "split_link" and len({split_by_id[row["catalog_id"]] for row in members}) != 1:
        raise RuntimeError(f"A linked comparison group was separated: {relation} / {group_id}")
if any(row["description_group_id"] for row in catalog if not row["description_prepared"]):
    raise RuntimeError("Missing descriptions must not become a shared-description group.")
earlier_name_groups = quality_summary.get("domain_metrics", {}).get("exercise", {}).get("repeated_normalized_name_groups")
if earlier_name_groups is not None and earlier_name_groups != grouping_metrics["relation_metrics"]["broad_name"]["repeated_groups"]:
    raise RuntimeError("The broader name comparison differs from Step 9; review the recipe before saving.")

labels = Counter(row["body_part_prepared"] for row in catalog if row["body_part_prepared"])
observed_issue_counts = Counter(row["issue_type"] for row in review_queue if row["finding_state"] == "observed")
resolved_statuses = {"resolved", "closed", "accepted", "approved", "dismissed", "reviewed", "retained_as_is", "accepted_as_is", "not_an_issue"}
unresolved_findings = [row for row in active_findings if row["review_status"].strip().lower() not in resolved_statuses]
preparation_metrics = {
    "source_records": len(exercise_rows), "catalog_records": len(catalog), "label_counts": dict(labels),
    "preliminary_classifier_candidates": sum(row["preliminary_classifier_candidate"] == "yes" for row in catalog),
    "text_availability": dict(Counter(row["text_availability"] for row in catalog)),
    "current_formatting_changes": sum(row["record_state"] == "observed" for row in cleaning_log),
    "historical_formatting_entries": sum(row["record_state"] != "observed" for row in cleaning_log),
    "records_with_observed_or_custom_flags": sum(bool(row["review_flags"]) for row in catalog),
    "review_queue_entries": len(review_queue), "active_review_entries": len(active_findings),
    "unresolved_active_entries": len(unresolved_findings), "observed_issue_counts": dict(observed_issue_counts),
}
step10_report.update(review_merge_complete=True, status="reconciled_for_save")
print("Preparation findings:", preparation_metrics)
print("Review carryover:", review_metrics)
print("Original source fields retained; classifier input construction verified.")
```

```python
# Step 10-4

# We publish the prepared files together with their producer and review records.
if not globals().get("step10_report", {}).get("review_merge_complete"):
    raise RuntimeError("Complete the preparation, grouping, and review-reconciliation cells first.")
step10_report["status"] = "preparing_outputs"
for path, expected in preparation_controls.items():
    if path.read_bytes() != expected:
        raise RuntimeError(f"Source documentation changed after loading. Rerun Step 10: {path}")
if sha256_file(source_path) != source_hash:
    raise RuntimeError("The exercise source changed before saving. Reconcile its version first.")
for path, expected in output_snapshots.items():
    current = path.read_bytes() if path.exists() else None
    if current != expected:
        raise RuntimeError(f"An output or review file changed after loading. Rerun Step 10 to retain the edit: {path}")


def output_columns(rows, initial):
    return list(dict.fromkeys([*initial, *(field for row in rows for field in row)]))


def preparation_csv(rows, columns):
    buffer = io.StringIO(newline="")
    writer = csv.DictWriter(buffer, fieldnames=columns, extrasaction="raise")
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode("utf-8")


legacy_catalog_columns = source_columns + [
    "catalog_id", "source_cleaned_record_position", "source_file", "source_sha256",
    *(f"{field}_prepared" for field in PREPARED_FIELDS), "name_comparison_key", "label_review_status",
    "preliminary_classifier_candidate", "review_flags",
]
catalog_header = output_columns(catalog, existing_catalog_columns or legacy_catalog_columns)
queue_header = output_columns(review_queue, existing_queue_columns or [
    "catalog_id", "issue_type", "priority", "evidence", "suggested_action", "review_status"])
log_header = output_columns(cleaning_log, [*(existing_log_columns or [
    "catalog_id", "field", "source_value", "prepared_value", "reason", "decision"]), "record_state"])
group_header = ["relation_type", "group_id", "catalog_id", "member_count", "labels_json", "comparison_digest", "link_role"]
carryover_header = ["entity_type", "key", "change", "previous_value", "current_value", "reason"]
artifacts = {
    "catalog": (catalog_path, catalog, catalog_header),
    "cleaning_log": (log_path, cleaning_log, log_header),
    "review_queue": (queue_path, review_queue, queue_header),
    "group_membership": (groups_path, group_membership, group_header),
    "review_carryover": (review_audit_path, review_carryover, carryover_header),
}
outputs = {path: preparation_csv(rows, columns) for path, rows, columns in artifacts.values()}
backup_root = project_path("results/backups/step10") / datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S_%fZ")
previous_artifacts = {}
for path in preparation_targets:
    previous = output_snapshots[path]
    if previous is not None:
        changed = path not in outputs or outputs[path] != previous
        previous_artifacts[path.relative_to(root).as_posix()] = {
            "sha256": hashlib.sha256(previous).hexdigest(),
            "backup_path": (backup_root / path.relative_to(root)).relative_to(root).as_posix() if changed else "",
        }
handoff = {
    "schema_version": "step10-exercise-preparation-1", "recipe_version": RECIPE_VERSION,
    "prepared_at": preparation_started_at,
    "status": "prepared_with_reviews" if preparation_metrics["unresolved_active_entries"] else "prepared",
    "parent_source": {"source_id": "derived_exercise", "relative_path": source_relative,
                      "sha256": source_hash, "rows": len(exercise_rows), "columns": source_columns},
    "upstream_document_hashes": {path.relative_to(root).as_posix(): hashlib.sha256(content).hexdigest()
                                  for path, content in preparation_controls.items()},
    "artifacts": {name: {"relative_path": path.relative_to(root).as_posix(),
                         "sha256": hashlib.sha256(outputs[path]).hexdigest(),
                         "rows": len(rows), "columns": columns}
                  for name, (path, rows, columns) in artifacts.items()},
    "previous_artifacts": previous_artifacts,
    "preparation_metrics": preparation_metrics, "grouping_metrics": grouping_metrics, "review_metrics": review_metrics,
    "feature_contract": {"input_fields": ["exercise_name_prepared", "description_prepared"],
                         "model_text_field": "classifier_text", "target_field": "body_part_prepared",
                         "split_group_field": "split_group_id", "final_splits_created": False,
                         "excluded_from_features": ["body_part", "body_part_prepared", "goal_tag", "agent_text",
                                                    "catalog_id", "equipment", "difficulty_level", "review_flags",
                                                    "label_review_status", "split_group_label_count", "source_metadata"]},
    "verified_checks": {"all_source_records_retained": True, "source_fields_unchanged": True,
                        "original_body_part_labels_unchanged": True, "unique_catalog_ids": True,
                        "classifier_text_allowlist": True, "comparison_links_kept_within_groups": True,
                        "blank_descriptions_not_grouped_together": True},
}
outputs[handoff_path] = (json.dumps(handoff, ensure_ascii=False, indent=2, allow_nan=False) + "\n").encode("utf-8")
changed_paths = [path for path, content in outputs.items() if output_snapshots[path] != content]

# We back up old bytes before replacing any target and restore all affected files on failure.
for path in changed_paths:
    if output_snapshots[path] is not None:
        backup = backup_root / path.relative_to(root)
        backup.parent.mkdir(parents=True, exist_ok=True)
        backup.write_bytes(output_snapshots[path])
written = []
try:
    for path in changed_paths:
        written.append(path)
        replace_bytes(path, outputs[path])
        if path.read_bytes() != outputs[path]:
            raise OSError(f"Saved bytes differ from the prepared artifact: {path}")
except Exception as error:
    restoration_errors = []
    for path in reversed(written):
        try:
            if output_snapshots[path] is None:
                path.unlink(missing_ok=True)
            else:
                replace_bytes(path, output_snapshots[path])
        except Exception as restore_error:
            restoration_errors.append(f"{path}: {restore_error}")
    if restoration_errors:
        raise RuntimeError("Some preparation files could not be restored: " + "; ".join(restoration_errors)) from error
    raise
step10_report.update(handoff)
for path in outputs:
    action = "unchanged" if path not in changed_paths else "updated" if output_snapshots[path] is not None else "created"
    print(f"{action.capitalize()}: {path.relative_to(root).as_posix()}")
print(f"Exercise records retained: {len(catalog)} | Catalog columns: {len(catalog_header)}")
print(f"Preliminary candidates: {preparation_metrics['preliminary_classifier_candidates']} | Active unresolved findings: {preparation_metrics['unresolved_active_entries']}")
print("Body-part label counts:", dict(labels))
print("Step 10 preparation completed. The producer record identifies the current catalog for Step 11.")
```

Bias-corrected Cramér's V will summarize categorical association, following Bergsma's method [1]. We will report the number of categories and expected-cell sparsity alongside the coefficient. ### 11.6 Food-Reference Quality and Relationships

We will describe the recorded protein, fat, carbohydrate, fiber, and energy fields. Food identity, reference amount, and suitability for application use will remain separate verification questions.

Our numerical screening will distinguish:

| State | Treatment in This Analysis |
|---|---|
| Missing value | Preserve it and report coverage |
| Invalid numerical text | Record it as an anomaly |
| Nonfinite number, including numeric NaN or infinity | Record it separately and exclude it from numerical summaries |
| Finite negative measurement | Preserve it for source investigation and exclude it from the affected nonnegative analysis view |
| Finite nonnegative measurement | Include it in the applicable descriptive analysis |

This numeric classification gives explicit numeric NaN its own nonfinite state; Step 9's general missing-token screen used a broader missing category. The source values will remain unchanged under both approaches.

We will report completeness for the four core fields: recorded energy, protein, fat, and carbohydrate. Finite core completeness and finite nonnegative core completeness will be separate measures. Fiber coverage will also be reported independently.

Fieldwise descriptive statistics will use finite nonnegative values. Pearson correlation will describe linear association, while Spearman correlation will describe rank association [2, 3]. Each pair will use only records where both relevant fields are finite and nonnegative, with its own eligible and excluded counts. An unavailable third nutrient will not automatically exclude an otherwise usable pair.

We will require at least ten eligible pairs for a reported coefficient and leave constant-input correlations unavailable. This is an operational reporting threshold, not a guarantee of statistical reliability. We will report coefficients without p-values.

We will compare alternative energy fields within the same food record and check whether the selected energy agrees with its recorded method. Differences between energy conventions will not automatically be classified as errors. Correlation with energy can reflect mathematical dependence on nutrient values; these relationships do not establish dietary benefit or athletic performance effects.

```python
# Step 11-1

# We verify the saved preparation and quality records before starting our analysis.
import csv
import hashlib
import io
import json
import math
import re
from collections import Counter, defaultdict
from datetime import datetime, timezone
from pathlib import Path, PureWindowsPath

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt

EDA_RECIPE = "integrated-eda-0.2.0"
SHOW_EDA_FIGURES = True
eda_state = {"status": "incomplete", "inputs_verified": False, "exercise_complete": False,
             "supporting_complete": False, "integrated_complete": False}
if globals().get("step10_report", {}).get("status") not in {"prepared", "prepared_with_reviews"}:
    raise RuntimeError("Complete the four Step 10 cells before starting Step 11.")
if not all(callable(globals().get(name)) for name in ("read_table", "sha256_file", "strict_json", "replace_bytes")):
    raise RuntimeError("Run Steps 8–10 in this kernel to load our shared inspection functions.")
eda_root = Path(project_root).resolve()
eda_started_at = datetime.now(timezone.utc).isoformat()


def eda_path(relative):
    value = Path(relative)
    if value.is_absolute() or PureWindowsPath(relative).drive:
        raise ValueError(f"Expected a project-relative path: {relative}")
    path = (eda_root / value).resolve()
    if not path.is_relative_to(eda_root):
        raise ValueError(f"Path leaves our project: {relative}")
    return path


eda_controls, eda_file_hashes, eda_checks = {}, {}, []


def eda_check(name, condition, detail):
    eda_checks.append({"check_name": name, "status": "pass" if condition else "failed", "details": detail})
    if not condition:
        raise RuntimeError(f"Step 11 input check failed: {name}. {detail}")


def eda_control(relative, expected_hash=None):
    path = eda_path(relative)
    content = path.read_bytes()
    digest = hashlib.sha256(content).hexdigest()
    if expected_hash is not None and digest != expected_hash:
        raise RuntimeError(f"A preceding-step document changed: {relative}. Rerun the affected preceding steps.")
    eda_controls[path] = content
    eda_file_hashes[path] = digest
    return content


eda_preparation = strict_json(eda_control("results/step10_exercise_preparation.json").decode("utf-8-sig"))
eda_check("step10_producer", isinstance(eda_preparation, dict)
          and eda_preparation.get("schema_version") == "step10-exercise-preparation-1"
          and eda_preparation.get("status") in {"prepared", "prepared_with_reviews"},
          "The current catalog must have a completed Step 10 producer record.")
for eda_relative, eda_expected in eda_preparation["upstream_document_hashes"].items():
    eda_control(eda_relative, eda_expected)
eda_quality = strict_json(eda_controls[eda_path("results/step09_quality_summary.json")].decode("utf-8-sig"))
eda_check("step09_quality", eda_quality.get("status") in {"passed", "completed_with_reviews"}
          and eda_quality.get("required_failures") == 0,
          "The saved quality inspection must have no required failures.")
for eda_relative, eda_expected in eda_quality["source_control_hashes"].items():
    eda_control(eda_relative, eda_expected)
eda_manifest, _ = read_table(eda_path("data/source_manifest.csv"))
eda_inventory, _ = read_table(eda_path("results/source_file_inventory.csv"))
eda_quality_inventory, _ = read_table(eda_path("results/data_inventory.csv"))
eda_quality_checks_path = eda_path("results/initial_quality_checks.csv")
eda_control("results/initial_quality_checks.csv")
eda_prior_checks, _ = read_table(eda_quality_checks_path)
eda_check("quality_report_counts", len(eda_prior_checks) == eda_quality["total_checks"]
          and dict(Counter(row["status"] for row in eda_prior_checks)) == eda_quality["check_status_counts"],
          "The quality-check table agrees with its saved summary counters.")
eda_sources = {row["source_id"]: row for row in eda_manifest}
eda_inspected = {row["source_id"]: row for row in eda_quality_inventory}
eda_check("source_register_ids", len(eda_sources) == len(eda_manifest)
          and len(eda_inspected) == len(eda_quality_inventory)
          and set(eda_sources) == set(eda_inspected)
          and len(eda_sources) == eda_quality["manifest_records"],
          "Source identifiers are unique and resolve across the preceding reports.")

# The Step 10 producer supersedes the earlier catalog snapshot, not its source parent.
eda_required_sources = {
    "derived_exercise", "derived_athlete_profile", "aditi_jr_nba_drills", "aditi_jr_nba_rag",
    "aditi_usda_foundation_nutrition", "aditi_synthetic_athletes", "aditi_recovery_guidance", "recovery_rules_v01",
}
eda_optional_sources = {"catalog_basketball_v01", "catalog_basketball_v02", "local_basketball_drills"}
eda_tables_by_source, eda_context_headers, eda_input_rows = {}, {}, []
for eda_sid, eda_entry in eda_inspected.items():
    eda_use = "supporting analysis" if eda_sid in eda_required_sources else (
        "optional historical snapshot" if eda_sid in eda_optional_sources else "verified Step 9 context")
    if eda_sid == "local_exercise_catalog":
        continue
    eda_relative = eda_entry["relative_path"]
    eda_loaded = eda_entry["load_status"] in {"loaded", "verified_document"}
    eda_input_rows.append({"source_id": eda_sid, "relative_path": eda_relative,
                           "authority": "Step 9 inspection", "role": eda_use,
                           "availability": eda_entry["load_status"], "records": eda_entry["actual_rows"],
                           "sha256": eda_entry["observed_sha256"]})
    if not eda_loaded:
        if eda_sid in eda_required_sources:
            raise RuntimeError(f"A required analysis source was not loaded in Step 9: {eda_sid}")
        continue
    eda_input_path = eda_path(eda_relative)
    eda_digest = sha256_file(eda_input_path)
    eda_check(f"source_identity:{eda_sid}", eda_digest == eda_entry["observed_sha256"],
              "Current source bytes agree with the completed Step 9 inspection.")
    eda_file_hashes[eda_input_path] = eda_digest
    if eda_sid == "derived_master":
        with eda_input_path.open(encoding="utf-8-sig", newline="") as eda_handle:
            eda_context_headers[eda_sid] = next(csv.reader(eda_handle))
    if eda_sid not in eda_required_sources | eda_optional_sources:
        continue
    if eda_input_path.suffix.lower() == ".csv":
        eda_rows, eda_columns = read_table(eda_input_path)
    elif eda_input_path.suffix.lower() == ".json":
        eda_rows = strict_json(eda_input_path.read_text(encoding="utf-8-sig"))
        if not isinstance(eda_rows, list) or not all(isinstance(row, dict) for row in eda_rows):
            raise ValueError(f"Expected a list of document objects: {eda_relative}")
        eda_columns = sorted({field for row in eda_rows for field in row})
    else:
        raise ValueError(f"Unsupported structured analysis source: {eda_relative}")
    eda_check(f"source_dimensions:{eda_sid}", len(eda_rows) == int(eda_entry["actual_rows"])
              and len(eda_columns) == int(eda_entry["columns"]),
              "Loaded dimensions agree with the Step 9 inspection.")
    eda_tables_by_source[eda_sid] = {"columns": eda_columns, "rows": eda_rows}
eda_check("analysis_sources", eda_required_sources.issubset(eda_tables_by_source),
          "All required domain tables are available; Step 12 outputs remain optional.")

eda_artifacts = {}
for eda_name, eda_record in eda_preparation["artifacts"].items():
    eda_input_path = eda_path(eda_record["relative_path"])
    eda_digest = sha256_file(eda_input_path)
    eda_check(f"step10_artifact:{eda_name}", eda_digest == eda_record["sha256"],
              "Prepared artifacts must match their current producer record; rerun Step 10 after review edits.")
    eda_rows, eda_columns = read_table(eda_input_path)
    eda_check(f"step10_dimensions:{eda_name}", len(eda_rows) == eda_record["rows"]
              and eda_columns == eda_record["columns"], "Prepared table dimensions and field order agree.")
    eda_file_hashes[eda_input_path] = eda_digest
    eda_artifacts[eda_name] = {"columns": eda_columns, "rows": eda_rows}
    eda_input_rows.append({"source_id": f"step10:{eda_name}", "relative_path": eda_record["relative_path"],
                           "authority": "Step 10 producer", "role": "current exercise preparation",
                           "availability": "loaded", "records": len(eda_rows), "sha256": eda_digest})
eda_catalog = eda_artifacts["catalog"]["rows"]
eda_queue = eda_artifacts["review_queue"]["rows"]
eda_group_rows = eda_artifacts["group_membership"]["rows"]
eda_parent = eda_tables_by_source["derived_exercise"]["rows"]
eda_contract = eda_preparation["feature_contract"]
eda_check("feature_contract", eda_contract["input_fields"] == ["exercise_name_prepared", "description_prepared"]
          and eda_contract["target_field"] == "body_part_prepared"
          and eda_contract["model_text_field"] == "classifier_text"
          and eda_contract["split_group_field"] == "split_group_id"
          and eda_contract["final_splits_created"] is False,
          "Step 11 uses the documented text fields, separate target, and unsplit group mapping.")
eda_parent_record = eda_preparation["parent_source"]
eda_check("exercise_parent", eda_parent_record["sha256"] == eda_inspected["derived_exercise"]["observed_sha256"]
          and eda_parent_record["rows"] == len(eda_parent) == len(eda_catalog)
          and eda_parent_record["columns"] == eda_tables_by_source["derived_exercise"]["columns"],
          "The prepared catalog links to the verified cleaned exercise source.")
eda_ids = {row["catalog_id"] for row in eda_catalog}
eda_positions = [int(row["source_cleaned_record_position"]) for row in eda_catalog]
eda_check("catalog_identifiers", len(eda_ids) == len(eda_catalog)
          and sorted(eda_positions) == list(range(1, len(eda_parent) + 1)),
          "Catalog identifiers and cleaned-file positions retain one-to-one source coverage.")
for eda_row in eda_catalog:
    eda_position = int(eda_row["source_cleaned_record_position"])
    eda_original = eda_parent[eda_position - 1]
    if (any(eda_row[field] != eda_original[field] for field in eda_parent_record["columns"])
            or eda_row["catalog_id"] != f"exercise_clean:{eda_position:06d}"
            or eda_row["source_sha256"] != eda_parent_record["sha256"]
            or eda_row["classifier_text"] != "\n\n".join(eda_row[field] for field in eda_contract["input_fields"] if eda_row[field])):
        raise RuntimeError(f"Exercise source values or allowed classifier text differ: {eda_row['catalog_id']}")
eda_check("source_and_text_preservation", True, "Original source values and the two-field classifier input were independently checked.")
eda_queue_keys = [(row["catalog_id"], row["issue_type"]) for row in eda_queue]
eda_check("review_identifiers", len(set(eda_queue_keys)) == len(eda_queue_keys)
          and all(row["catalog_id"] in eda_ids for row in eda_queue), "Review entries resolve uniquely to catalog records.")
eda_split_members = [row for row in eda_group_rows if row["relation_type"] == "split_component"]
eda_split_by_id = {row["catalog_id"]: row["split_group_id"] for row in eda_catalog}
eda_check("split_membership", len(eda_split_members) == len(eda_catalog)
          and {row["catalog_id"] for row in eda_split_members} == eda_ids
          and all(eda_split_by_id[row["catalog_id"]] == row["group_id"] for row in eda_split_members),
          "Every exercise has exactly one recorded split-group membership.")
eda_link_groups = defaultdict(set)
for eda_row in eda_group_rows:
    if eda_row["catalog_id"] not in eda_ids:
        raise RuntimeError("A comparison membership has an unknown exercise identifier.")
    if eda_row["link_role"] == "split_link":
        eda_link_groups[(eda_row["relation_type"], eda_row["group_id"])].add(eda_split_by_id[eda_row["catalog_id"]])
eda_check("comparison_links", all(len(groups) == 1 for groups in eda_link_groups.values())
          and all(not row["description_group_id"] for row in eda_catalog if not row["description_prepared"]),
          "Linked records remain together; missing descriptions do not form a shared-description group.")

# We remember prior report bytes so an edit made during analysis cannot be overwritten.
eda_table_dir, eda_figure_dir = eda_path("results/eda"), eda_path("results/figures")
eda_summary_path = eda_path("results/step11_eda_summary.json")
eda_existing_outputs = {
    path.resolve(): path.read_bytes()
    for folder, pattern in ((eda_table_dir, "*.csv"), (eda_figure_dir, "*.png"))
    if folder.exists() for path in folder.glob(pattern) if path.is_file()
}
eda_existing_outputs[eda_summary_path] = eda_summary_path.read_bytes() if eda_summary_path.exists() else None
eda_state.update(status="inputs_verified", inputs_verified=True)
print(f"Step 11 — Verified {len(eda_catalog):,} prepared exercise records and {len(eda_tables_by_source) - 1} supporting tables")
print(f"Step 9 context: {eda_quality['local_files_inspected']} local files; {eda_quality['historical_inputs_not_found']} historical input gaps")
print("Current exercise artifacts match Step 10. Later basketball snapshots are optional and reported separately.")
```

```python
# Step 11-2

# We describe the prepared exercises without changing records or review decisions.
import json
import math
import re
from collections import Counter, defaultdict


def analyze_exercises(catalog_rows, queue_rows, group_rows, source_rows):
    """Return reproducible exercise summaries using the verified Step 10 artifacts."""
    catalog = [dict(row) for row in catalog_rows]
    source = [dict(row) for row in source_rows]
    queue = [dict(row) for row in queue_rows]
    memberships = [dict(row) for row in group_rows]
    required = {
        "catalog_id", "body_part_prepared", "exercise_name_prepared", "description_prepared",
        "equipment_prepared", "difficulty_level_prepared", "exercise_type_prepared",
        "classifier_text", "preliminary_classifier_candidate", "label_review_status", "split_group_id",
    }
    if any(not required.issubset(row) for row in catalog):
        raise ValueError("Exercise analysis requires the complete Step 10 catalog schema.")
    identifiers = [row["catalog_id"] for row in catalog]
    if any(not value for value in identifiers) or len(set(identifiers)) != len(identifiers):
        raise ValueError("Exercise analysis requires nonempty, unique catalog identifiers.")
    if len(source) != len(catalog):
        raise ValueError("The exercise source and prepared catalog must retain the same record count.")
    known_ids = set(identifiers)
    if any(row.get("catalog_id") not in known_ids for row in queue + memberships):
        raise ValueError("Exercise review and group records must resolve to the prepared catalog.")
    catalog.sort(key=lambda row: row["catalog_id"])
    queue.sort(key=lambda row: (row["catalog_id"], row.get("issue_type", "")))
    memberships.sort(key=lambda row: (row["relation_type"], row["group_id"], row["catalog_id"]))
    unresolved_exceptions = {
        "resolved", "closed", "accepted", "approved", "dismissed", "reviewed", "retained_as_is",
        "accepted_as_is", "not_an_issue",
    }
    active_states = {"observed", "retained_custom"}
    missing_tokens = {"", "null", "none", "nan", "n/a"}
    unknown_tokens = {"unknown", "unspecified", "not known", "not specified"}
    tables, findings = {}, []
    n = len(catalog)

    def state(value, field=""):
        text = str(value or "").strip().casefold()
        base = field.removesuffix("_prepared")
        if text == "none" and base == "equipment":
            return "present"
        if text in missing_tokens:
            return "missing"
        return "explicit_unknown" if text in unknown_tokens else "present"

    def percent(count, denominator):
        return round(100 * count / denominator, 4) if denominator else None

    def category(value, field=""):
        return "[missing]" if state(value, field) == "missing" else str(value).strip()

    def add_table(name, columns, rows):
        tables[name] = {"columns": columns, "rows": rows}

    def word_count(value):
        return len(re.findall(r"\b[\w'-]+\b", str(value or "")))

    def quantile(values, fraction):
        if not values:
            return None
        ordered = sorted(values)
        index = (len(ordered) - 1) * fraction
        lower, upper = math.floor(index), math.ceil(index)
        return ordered[lower] + (ordered[upper] - ordered[lower]) * (index - lower)

    def statistics(values):
        values = list(values)
        return {
            "records": len(values), "mean_words": round(sum(values) / len(values), 4) if values else None,
            "median_words": quantile(values, 0.5), "p25_words": quantile(values, 0.25),
            "p75_words": quantile(values, 0.75), "p95_words": quantile(values, 0.95),
            "maximum_words": max(values) if values else None,
        }

    def finding(identifier, observation, implication, limitation):
        findings.append({"finding_id": identifier, "domain": "exercise", "observation": observation,
                         "implication": implication, "limitation": limitation})

    by_id = {row["catalog_id"]: row for row in catalog}
    labels = Counter(category(row["body_part_prepared"], "body_part") for row in catalog)
    available = {row["catalog_id"]: state(row["description_prepared"], "description") != "missing"
                 for row in catalog}
    label_records = defaultdict(list)
    for row in catalog:
        label_records[category(row["body_part_prepared"], "body_part")].append(row)
    classes = [{"body_part": label, "records": count, "percent": percent(count, n)}
               for label, count in sorted(labels.items(), key=lambda item: (-item[1], item[0]))]
    add_table("exercise_body_part_distribution.csv", ["body_part", "records", "percent"], classes)

    lengths = [{"catalog_id": row["catalog_id"], "body_part": category(row["body_part_prepared"], "body_part"),
                "split_group_id": row["split_group_id"], "has_description": available[row["catalog_id"]],
                "name_words": word_count(row["exercise_name_prepared"]),
                "description_words": word_count(row["description_prepared"]) if available[row["catalog_id"]] else 0,
                "classifier_words": word_count(row["classifier_text"])} for row in catalog]
    add_table("exercise_text_lengths.csv", ["catalog_id", "body_part", "split_group_id", "has_description",
                                          "name_words", "description_words", "classifier_words"], lengths)
    statistics_columns = ["population", "records", "mean_words", "median_words", "p25_words", "p75_words",
                          "p95_words", "maximum_words"]
    length_stats = []
    for population, field, rows in (
        ("exercise_name_all", "name_words", lengths),
        ("description_all_including_missing", "description_words", lengths),
        ("description_available_only", "description_words", [row for row in lengths if row["has_description"]]),
        ("classifier_text_all", "classifier_words", lengths),
    ):
        length_stats.append({"population": population, **statistics(row[field] for row in rows)})
    add_table("exercise_text_length_statistics.csv", statistics_columns, length_stats)
    by_label_length = []
    for label in sorted(labels):
        rows = [row for row in lengths if row["body_part"] == label]
        for population, field, subset in (
            ("exercise_name_all", "name_words", rows),
            ("description_available_only", "description_words", [row for row in rows if row["has_description"]]),
            ("classifier_text_all", "classifier_words", rows),
        ):
            by_label_length.append({"body_part": label, "population": population,
                                    **statistics(row[field] for row in subset)})
    add_table("exercise_label_text_statistics.csv", ["body_part", *statistics_columns], by_label_length)

    coverage = []
    for label, rows in label_records.items():
        missing = sum(not available[row["catalog_id"]] for row in rows)
        coverage.append({"body_part": label, "records": len(rows), "missing_descriptions": missing,
                         "available_descriptions": len(rows) - missing, "missing_percent": percent(missing, len(rows))})
    coverage.sort(key=lambda row: (-row["missing_percent"], row["body_part"]))
    add_table("exercise_description_coverage.csv", ["body_part", "records", "missing_descriptions",
                                                  "available_descriptions", "missing_percent"], coverage)

    completeness = []
    source_fields = list(dict.fromkeys(field for row in source for field in row))
    prepared_fields = ["exercise_name_prepared", "description_prepared", "body_part_prepared",
                       "exercise_type_prepared", "equipment_prepared", "difficulty_level_prepared", "classifier_text"]
    for scope, rows, fields in (("source", source, source_fields), ("prepared", catalog, prepared_fields)):
        for field in fields:
            counts = Counter(state(row.get(field, ""), field) for row in rows)
            completeness.append({"scope": scope, "field": field, "records": len(rows),
                                 "missing": counts["missing"], "explicit_unknown": counts["explicit_unknown"],
                                 "present": counts["present"], "missing_percent": percent(counts["missing"], len(rows))})
    add_table("exercise_field_completeness.csv", ["scope", "field", "records", "missing", "explicit_unknown",
                                                "present", "missing_percent"], completeness)
    for field, key, filename in (
        ("equipment_prepared", "equipment", "exercise_equipment_distribution.csv"),
        ("difficulty_level_prepared", "difficulty", "exercise_difficulty_distribution.csv"),
        ("exercise_type_prepared", "exercise_type", "exercise_type_distribution.csv"),
    ):
        counts = Counter(category(row[field], field) for row in catalog)
        rows = [{key: value, "records": count, "percent": percent(count, n),
                 "value_state": "missing" if value == "[missing]" else state(value, field)}
                for value, count in sorted(counts.items(), key=lambda item: (-item[1], item[0]))]
        add_table(filename, [key, "records", "percent", "value_state"], rows)

    # We count current findings separately from preserved history and handling decisions.
    active = [row for row in queue if row.get("finding_state") in active_states]
    unresolved = [row for row in active if row.get("review_status", "").strip().casefold() not in unresolved_exceptions]
    observed = [row for row in queue if row.get("finding_state") == "observed"]
    issues = []
    for issue in sorted({row["issue_type"] for row in queue}):
        rows = [row for row in queue if row["issue_type"] == issue]
        active_rows = [row for row in rows if row.get("finding_state") in active_states]
        unresolved_rows = [row for row in active_rows if row.get("review_status", "").strip().casefold() not in unresolved_exceptions]
        issues.append({"issue_type": issue, "queue_entries": len(rows),
                       "observed_entries": sum(row.get("finding_state") == "observed" for row in rows),
                       "retained_custom_entries": sum(row.get("finding_state") == "retained_custom" for row in rows),
                       "historical_entries": len(rows) - len(active_rows), "active_entries": len(active_rows),
                       "unresolved_entries": len(unresolved_rows),
                       "distinct_active_records": len({row["catalog_id"] for row in active_rows}),
                       "distinct_unresolved_records": len({row["catalog_id"] for row in unresolved_rows})})
    add_table("exercise_review_issues.csv", ["issue_type", "queue_entries", "observed_entries", "retained_custom_entries",
                                           "historical_entries", "active_entries", "unresolved_entries",
                                           "distinct_active_records", "distinct_unresolved_records"], issues)
    statuses = []
    for status, count in sorted(Counter(row.get("label_review_status", "") for row in catalog).items()):
        statuses.append({"entity": "catalog", "finding_state": "", "review_status": status,
                         "entries": count, "distinct_records": count, "active": None, "unresolved": None})
    queue_statuses = defaultdict(list)
    for row in queue:
        queue_statuses[(row.get("finding_state", ""), row.get("review_status", ""))].append(row)
    for (finding_state, status), rows in sorted(queue_statuses.items()):
        is_active = finding_state in active_states
        statuses.append({"entity": "queue", "finding_state": finding_state, "review_status": status,
                         "entries": len(rows), "distinct_records": len({row["catalog_id"] for row in rows}),
                         "active": is_active,
                         "unresolved": is_active and status.strip().casefold() not in unresolved_exceptions})
    add_table("exercise_review_status.csv", ["entity", "finding_state", "review_status", "entries", "distinct_records",
                                           "active", "unresolved"], statuses)

    # We summarize the saved group membership rather than recreating identities in EDA.
    grouped_memberships = defaultdict(list)
    for row in memberships:
        grouped_memberships[(row["relation_type"], row["group_id"])].append(row)
    split_groups = {group_id: rows for (relation, group_id), rows in grouped_memberships.items()
                    if relation == "split_component"}
    split_members = [row["catalog_id"] for rows in split_groups.values() for row in rows]
    if len(split_members) != n or set(split_members) != known_ids:
        raise ValueError("Every exercise must have exactly one saved split-component membership.")
    for (relation, group_id), rows in grouped_memberships.items():
        ids = [row["catalog_id"] for row in rows]
        if len(set(ids)) != len(ids) or any(int(row["member_count"]) != len(rows) for row in rows):
            raise ValueError("An exercise comparison group has duplicate members or inconsistent size.")
        if relation == "split_component" and any(by_id[rid]["split_group_id"] != group_id for rid in ids):
            raise ValueError("The catalog and saved split-component memberships disagree.")
    comparison_rows = []
    relation_names = ["broad_name", "shared_description", "strict_classifier_input", "broad_classifier_input",
                      "strict_content_without_label", "broad_content_without_label"]
    for relation in relation_names:
        groups = [rows for (name, _), rows in grouped_memberships.items() if name == relation]
        comparison_rows.append({"relation_type": relation, "repeated_groups": len(groups),
                                "records_in_repeated_groups": sum(len(rows) for rows in groups),
                                "additional_records": sum(len(rows) - 1 for rows in groups),
                                "different_label_groups": sum(len({by_id[row["catalog_id"]]["body_part_prepared"]
                                                                   for row in rows}) > 1 for rows in groups),
                                "largest_group": max((len(rows) for rows in groups), default=0),
                                "link_role": "split_link" if relation in {"broad_name", "shared_description", "strict_classifier_input"}
                                             else "review_only"})
    add_table("exercise_comparison_summary.csv", ["relation_type", "repeated_groups", "records_in_repeated_groups",
                                                "additional_records", "different_label_groups", "largest_group", "link_role"], comparison_rows)
    group_summary = []
    for group_id, rows in sorted(split_groups.items()):
        ids = [row["catalog_id"] for row in rows]
        group_labels = sorted({by_id[rid]["body_part_prepared"] for rid in ids})
        group_summary.append({"split_group_id": group_id, "member_count": len(ids), "label_count": len(group_labels),
                              "labels_json": json.dumps(group_labels, ensure_ascii=False, separators=(",", ":")),
                              "name_only_records": sum(not available[rid] for rid in ids),
                              "described_records": sum(available[rid] for rid in ids)})
    add_table("exercise_split_group_summary.csv", ["split_group_id", "member_count", "label_count", "labels_json",
                                                 "name_only_records", "described_records"], group_summary)
    size_counts = Counter(row["member_count"] for row in group_summary)
    add_table("exercise_split_group_size_distribution.csv", ["member_count", "groups", "records", "percent_of_records"],
              [{"member_count": size, "groups": count, "records": size * count,
                "percent_of_records": percent(size * count, n)} for size, count in sorted(size_counts.items())])
    group_label_counts = {row["split_group_id"]: row["label_count"] for row in group_summary}
    unresolved_ids = {row["catalog_id"] for row in unresolved}
    label_readiness = []
    for label, rows in sorted(label_records.items()):
        within_label_groups = Counter(row["split_group_id"] for row in rows)
        group_ids = set(within_label_groups)
        mixed_ids = {group_id for group_id in group_ids if group_label_counts[group_id] > 1}
        described_count = sum(available[row["catalog_id"]] for row in rows)
        label_readiness.append({"body_part": label, "records": len(rows), "percent": percent(len(rows), n),
                                "described_records": described_count, "name_only_records": len(rows) - described_count,
                                "description_missing_percent": percent(len(rows) - described_count, len(rows)),
                                "distinct_split_groups": len(group_ids), "mixed_label_split_groups": len(mixed_ids),
                                "single_label_split_groups": len(group_ids - mixed_ids),
                                "largest_within_label_split_group": max(within_label_groups.values(), default=0),
                                "largest_group_share_of_label_percent": percent(max(within_label_groups.values(), default=0), len(rows)),
                                "records_with_unresolved_findings": sum(row["catalog_id"] in unresolved_ids for row in rows),
                                "low_support_under_10_records": len(rows) < 10})
    add_table("exercise_label_readiness.csv", ["body_part", "records", "percent", "described_records", "name_only_records",
                                             "description_missing_percent", "distinct_split_groups", "mixed_label_split_groups",
                                             "single_label_split_groups", "largest_within_label_split_group",
                                             "largest_group_share_of_label_percent", "records_with_unresolved_findings", "low_support_under_10_records"], label_readiness)

    # We use associations to describe this catalog, not to select additional model inputs.
    categorical_values = {
        "body_part": [category(row["body_part_prepared"], "body_part") for row in catalog],
        "description_availability": ["available" if available[row["catalog_id"]] else "unavailable" for row in catalog],
        "equipment": [category(row["equipment_prepared"], "equipment") for row in catalog],
        "difficulty": [category(row["difficulty_level_prepared"], "difficulty") for row in catalog],
        "exercise_type": [category(row["exercise_type_prepared"], "exercise_type") for row in catalog],
    }
    associations, crosstabs = [], []
    pairs = [("body_part", right) for right in ("description_availability", "equipment", "difficulty", "exercise_type")]
    pairs += [("equipment", "difficulty"), ("exercise_type", "equipment")]
    for left_field, right_field in pairs:
        pairs_values = list(zip(categorical_values[left_field], categorical_values[right_field]))
        cells = Counter(pairs_values)
        left_counts = Counter(left for left, _ in pairs_values)
        right_counts = Counter(right for _, right in pairs_values)
        r, k = len(left_counts), len(right_counts)
        chi_square, expected_values = 0.0, []
        for left, left_total in sorted(left_counts.items()):
            for right, right_total in sorted(right_counts.items()):
                count = cells[(left, right)]
                expected = left_total * right_total / n
                expected_values.append(expected)
                chi_square += (count - expected) ** 2 / expected
                if left_field == "body_part":
                    crosstabs.append({"left_field": left_field, "left_value": left,
                                      "right_field": right_field, "right_value": right, "records": count,
                                      "percent_within_left": percent(count, left_total)})
        correction = ((r - 1) * (k - 1) / (n - 1)) if n > 1 else None
        corrected_phi = max(0.0, chi_square / n - correction) if correction is not None else None
        corrected_r = r - ((r - 1) ** 2 / (n - 1)) if n > 1 else None
        corrected_k = k - ((k - 1) ** 2 / (n - 1)) if n > 1 else None
        denominator = min(corrected_r - 1, corrected_k - 1) if n > 1 else None
        corrected_v = (math.sqrt(corrected_phi / denominator)
                       if denominator is not None and denominator > 0 else None)
        associations.append({"left_field": left_field, "right_field": right_field, "paired_records": n,
                             "left_categories": r, "right_categories": k,
                             "observed_nonzero_cells": sum(count > 0 for count in cells.values()),
                             "possible_cells": r * k, "chi_square_descriptive": chi_square if n else None,
                             "bias_corrected_cramers_v": min(1.0, corrected_v) if corrected_v is not None else None,
                             "minimum_expected_cell": min(expected_values) if expected_values else None,
                             "expected_cells_below_5": sum(value < 5 for value in expected_values),
                             "percent_expected_cells_below_5": percent(sum(value < 5 for value in expected_values), len(expected_values)),
                             "missing_category_included": True,
                             "interpretation": "Descriptive catalog association; related records and sparse cells limit inference."})
    add_table("exercise_categorical_associations.csv", ["left_field", "right_field", "paired_records", "left_categories",
                                                       "right_categories", "observed_nonzero_cells", "possible_cells",
                                                       "chi_square_descriptive", "bias_corrected_cramers_v", "minimum_expected_cell",
                                                       "expected_cells_below_5", "percent_expected_cells_below_5",
                                                       "missing_category_included", "interpretation"], associations)
    add_table("exercise_label_metadata_crosstabs.csv", ["left_field", "left_value", "right_field", "right_value",
                                                      "records", "percent_within_left"], crosstabs)

    current_formatting_changes = sum(str(row.get(field, "")) != str(row.get(field + "_prepared", ""))
                                     for row in catalog for field in
                                     ("exercise_name", "description", "body_part", "exercise_type", "equipment", "difficulty_level"))
    eligible_counts = Counter(row["body_part_prepared"] for row in catalog
                              if row["preliminary_classifier_candidate"].strip().casefold() == "yes"
                              and state(row["body_part_prepared"], "body_part") == "present")
    eligible_n = sum(eligible_counts.values())
    largest_two = sum(count for _, count in labels.most_common(2))
    ratio = max(eligible_counts.values()) / min(eligible_counts.values()) if eligible_counts else None
    mixed = [row for row in group_summary if row["label_count"] > 1]
    metrics = {
        "source_records": len(source), "catalog_records": n, "body_part_categories": len(labels),
        "preliminary_classifier_candidates": sum(row["preliminary_classifier_candidate"].strip().casefold() == "yes" for row in catalog),
        "described_records": sum(available.values()), "name_only_records": n - sum(available.values()),
        "missing_description_percent": percent(n - sum(available.values()), n),
        "current_formatting_changes": current_formatting_changes,
        "records_with_active_findings": len({row["catalog_id"] for row in active}),
        "records_with_unresolved_findings": len(unresolved_ids), "review_queue_entries": len(queue),
        "observed_review_entries": len(observed), "active_review_entries": len(active),
        "unresolved_active_entries": len(unresolved), "historical_review_entries": len(queue) - len(active),
        "split_groups": len(split_groups), "singleton_split_groups": sum(row["member_count"] == 1 for row in group_summary),
        "multi_record_split_groups": sum(row["member_count"] > 1 for row in group_summary),
        "records_in_multi_record_split_groups": sum(row["member_count"] for row in group_summary if row["member_count"] > 1),
        "largest_split_group": max((row["member_count"] for row in group_summary), default=0),
        "mixed_label_split_groups": len(mixed), "records_in_mixed_label_split_groups": sum(row["member_count"] for row in mixed),
        "largest_two_label_percent": percent(largest_two, n), "largest_to_smallest_label_ratio": ratio,
        "majority_reference": {"population": "preliminary candidates with present source labels",
                               "records": eligible_n, "majority_labels": sorted(label for label, count in eligible_counts.items()
                                                                                 if count == max(eligible_counts.values())) if eligible_counts else [],
                               "majority_records": max(eligible_counts.values()) if eligible_counts else 0,
                               "descriptive_accuracy": max(eligible_counts.values()) / eligible_n if eligible_n else None,
                               "evaluation_status": "Full-catalog descriptive reference; no holdout predictions or measured model performance."},
        "class_counts": dict(sorted(labels.items())), "low_support_labels": sorted(label for label, count in labels.items() if count < 10),
        "relationship_counts_are_overlapping": True,
        "per_label_split_group_counts_can_overlap": True,
    }
    preparation = [
        ("catalog_records", n), ("preliminary_classifier_candidates", metrics["preliminary_classifier_candidates"]),
        ("formatting_changes", current_formatting_changes), ("flagged_distinct_records", metrics["records_with_active_findings"]),
        ("review_queue_entries", len(queue)), ("body_part_categories", len(labels)),
        ("missing_descriptions", metrics["name_only_records"]),
        ("missing_equipment", sum(state(row["equipment_prepared"], "equipment") == "missing" for row in catalog)),
        ("unresolved_active_entries", len(unresolved)), ("split_groups", len(split_groups)),
    ]
    add_table("exercise_preparation_summary.csv", ["measure", "count"], [{"measure": key, "count": value} for key, value in preparation])
    legacy_findings = [
        ("EX01", "missing_descriptions", metrics["name_only_records"], "exercise_description_coverage.csv", "Assess name-only and described records separately during modeling."),
        ("EX02", "largest_to_smallest_class_ratio", ratio, "exercise_body_part_distribution.csv", "Review class and linked-group support before partitioning."),
        ("EX03", "records_with_active_findings", metrics["records_with_active_findings"], "exercise_review_issues.csv", "Retain observed findings separately from unresolved handling decisions."),
        ("EX04", "mixed_label_split_groups", len(mixed), "exercise_split_group_summary.csv", "Keep linked records together and review label disagreements."),
    ]
    add_table("exercise_eda_findings.csv", ["finding_id", "metric", "value", "evidence", "follow_up"],
              [{"finding_id": fid, "metric": metric, "value": value, "evidence": evidence, "follow_up": action}
               for fid, metric, value, evidence, action in legacy_findings])
    finding("EX01", f"{metrics['name_only_records']} of {n} exercises lack an available description.",
            "Report text availability by label and assess name-only performance separately.",
            "A missing description is not evidence that an exercise is unusable or unsuitable for a particular athlete.")
    finding("EX02", f"The two largest source-label categories account for {metrics['largest_two_label_percent']}% of records; the largest/smallest ratio is {ratio}.",
            "Use per-class metrics and training-only balancing; examine distinct split-group support before final partitions.",
            "Source-label counts describe this collected catalog, not population exercise prevalence or validated muscle targeting.")
    finding("EX03", f"{len(split_groups)} split groups retain {n} exercises, including {len(mixed)} mixed-label groups.",
            "Use group-aware partitioning and investigate label conflicts before freezing training targets.",
            "Similarity links are conservative split protection and do not establish canonical exercise identity or statistical independence.")
    finding("EX04", f"The review queue contains {len(queue)} entries; {len(unresolved)} active entries remain unresolved across {len(unresolved_ids)} exercises.",
            "Track review progress without treating overlapping findings as independent records or automatically excluding flagged examples.",
            "A preserved handling decision does not independently verify the original publisher label or youth suitability.")
    finding("EX05", "Categorical associations and cross-tabs describe the source labels, text availability, and catalog metadata.",
            "Inspect uneven documentation and collection patterns while retaining the name-and-description feature contract.",
            "No p-values or causal claims are made; sparse categories and related records limit interpretation, and metadata association does not authorize new predictors.")
    return tables, metrics, findings

if not eda_state.get("inputs_verified"):
    raise RuntimeError("Complete the Step 11 input-verification cell first.")
eda_state.update(exercise_complete=False, integrated_complete=False, status="analyzing")
eda_checks[:] = [row for row in eda_checks if not row["check_name"].startswith("exercise_summary:")
                and row["check_name"] != "split_group_count"]
eda_exercise_tables, eda_exercise_metrics, eda_exercise_findings = analyze_exercises(
    eda_catalog, eda_queue, eda_group_rows, eda_parent)
eda_expected_preparation = eda_preparation["preparation_metrics"]
for eda_current, eda_saved in (
    ("catalog_records", "catalog_records"), ("name_only_records", "name_only"),
    ("review_queue_entries", "review_queue_entries"), ("unresolved_active_entries", "unresolved_active_entries"),
):
    eda_expected_value = (eda_expected_preparation["text_availability"].get(eda_saved, 0)
                          if eda_saved == "name_only" else eda_expected_preparation[eda_saved])
    eda_check(f"exercise_summary:{eda_current}", eda_exercise_metrics[eda_current] == eda_expected_value,
              "Recomputed exercise measures agree with Step 10.")
eda_check("split_group_count", eda_exercise_metrics["split_groups"] == eda_preparation["grouping_metrics"]["split_metrics"]["total_groups"],
          "The EDA uses the exact saved split mapping.")
eda_state["exercise_complete"] = True
print("Exercise class and text coverage")
print(pd.DataFrame(eda_exercise_tables["exercise_label_readiness.csv"]["rows"])[
    ["body_part", "records", "described_records", "name_only_records", "distinct_split_groups", "largest_within_label_split_group"]
].to_string(index=False))
print(f"Class size ratio: {eda_exercise_metrics['largest_to_smallest_label_ratio']:.2f} | Largest two labels: {eda_exercise_metrics['largest_two_label_percent']:.2f}%")
print("Full-catalog majority reference:", eda_exercise_metrics["majority_reference"])
print("Text-length summaries (word-pattern counts)")
print(pd.DataFrame(eda_exercise_tables["exercise_text_length_statistics.csv"]["rows"]).round(2).to_string(index=False))
print("Linked groups:", {key: eda_exercise_metrics[key] for key in (
    "split_groups", "multi_record_split_groups", "records_in_multi_record_split_groups", "mixed_label_split_groups")})
print(f"Review entries: {eda_exercise_metrics['review_queue_entries']} | Active unresolved: {eda_exercise_metrics['unresolved_active_entries']} | Distinct affected records: {eda_exercise_metrics['records_with_unresolved_findings']}")
print("Descriptive categorical associations")
print(pd.DataFrame(eda_exercise_tables["exercise_categorical_associations.csv"]["rows"])[
    ["left_field", "right_field", "bias_corrected_cramers_v", "percent_expected_cells_below_5"]
].round(4).to_string(index=False))
```

```python
# Step 11-3

def analyze_supporting_sources(tables_by_source_id, source_inventory, step09_summary):
    """Describe reference and development data without changing their approval state."""
    import math
    import re
    from collections import Counter, defaultdict
    from itertools import combinations
    from statistics import mean, median, pstdev

    reports, metrics, findings = {}, {}, []
    missing_tokens = {"", "none", "null", "n/a", "na"}
    unknown_tokens = {"unknown", "unspecified", "not known", "not provided"}

    def text(value):
        return "" if value is None else str(value).strip()

    def missing(value, field=""):
        token = text(value).lower()
        return token in missing_tokens and not (token == "none" and field in {"equipment", "injury_status"})

    def numeric_state(value):
        if missing(value):
            return "missing", None
        try:
            number = float(value)
        except (TypeError, ValueError, OverflowError):
            return "invalid_numeric", None
        if not math.isfinite(number):
            return "nonfinite", None
        return ("negative" if number < 0 else "nonnegative_finite"), number

    def eligible(value):
        state, number = numeric_state(value)
        return number if state == "nonnegative_finite" else None

    def source_rows(sid):
        return list(tables_by_source_id.get(sid, {}).get("rows", []))

    def columns(sid):
        return list(tables_by_source_id.get(sid, {}).get("columns", []))

    def table(name, header, rows):
        reports[name + ".csv"] = {"columns": header, "rows": rows}

    def finding(identifier, domain, observation, implication, limitation):
        findings.append({"finding_id": identifier, "domain": domain, "observation": observation,
                         "implication": implication, "limitation": limitation})

    def percent(count, total):
        return round(100 * count / total, 4) if total else ""

    def quantile(values, fraction):
        if not values:
            return ""
        ordered = sorted(values)
        position = (len(ordered) - 1) * fraction
        lower = int(position)
        upper = min(lower + 1, len(ordered) - 1)
        return ordered[lower] + (ordered[upper] - ordered[lower]) * (position - lower)

    def descriptive(values):
        return {"eligible_records": len(values), "mean": mean(values) if values else "",
                "standard_deviation_population": pstdev(values) if values else "",
                "minimum": min(values) if values else "", "q1": quantile(values, .25),
                "median": median(values) if values else "", "q3": quantile(values, .75),
                "maximum": max(values) if values else ""}

    def category(value, field=""):
        if missing(value, field):
            return "[missing]"
        if text(value).lower() in unknown_tokens:
            return "[unknown]"
        if ((text(value).lower() == "none" and field in {"injury_status", "equipment"})
                or (field == "injury_status" and text(value).lower() in {"none_reported", "no injury reported", "no reported injury"})):
            return "[reported none]"
        return text(value)

    def frequencies(sid, field, unit):
        rows = source_rows(sid)
        counts = Counter(category(row.get(field), field) for row in rows)
        return [{"source_id": sid, "field": field, "value": value, "records": count,
                 "denominator_records": len(rows), "percent": percent(count, len(rows)), "record_unit": unit}
                for value, count in sorted(counts.items(), key=lambda pair: (-pair[1], pair[0]))]

    frequency_header = ["source_id", "field", "value", "records", "denominator_records", "percent", "record_unit"]

    # We screen each nutrient separately and retain every source value.
    nutrition_sid = "aditi_usda_foundation_nutrition"
    food = source_rows(nutrition_sid)
    nutrients = ["protein_g", "fat_g", "carbohydrate_g", "fiber_g", "energy_general_kcal",
                 "energy_specific_kcal", "calories_kcal"]
    core = ["calories_kcal", "protein_g", "fat_g", "carbohydrate_g"]
    completeness, statistics, anomalies = [], [], []
    for field in nutrients:
        states = Counter(numeric_state(row.get(field))[0] for row in food)
        completeness.append({"source_id": nutrition_sid, "nutrient": field, "total_records": len(food),
                             "missing_records": states["missing"], "missing_pct": percent(states["missing"], len(food)),
                             "invalid_numeric_records": states["invalid_numeric"], "nonfinite_records": states["nonfinite"],
                             "negative_records": states["negative"],
                             "finite_records": states["negative"] + states["nonnegative_finite"],
                             "finite_nonnegative_records": states["nonnegative_finite"]})
        values = [number for row in food if (number := eligible(row.get(field))) is not None]
        statistics.append({"source_id": nutrition_sid, "nutrient": field,
                           "analysis_subset": "fieldwise finite nonnegative values", **descriptive(values),
                           "reference_amount_status": "Source reference amount requires independent confirmation before serving calculations."})
        for position, row in enumerate(food, 1):
            state, _ = numeric_state(row.get(field))
            if state in {"negative", "invalid_numeric", "nonfinite"}:
                anomalies.append({"source_id": nutrition_sid, "source_row": position, "fdc_id": text(row.get("fdc_id")),
                                  "field": field, "source_value": text(row.get(field)), "finding": state,
                                  "handling": "Preserved in source; excluded from the affected numeric analysis view."})
    finite_core = sum(all(numeric_state(row.get(field))[0] in {"negative", "nonnegative_finite"} for field in core) for row in food)
    nonnegative_core = sum(all(eligible(row.get(field)) is not None for field in core) for row in food)
    core_present = sum(all(not missing(row.get(field)) for field in core) for row in food)
    metrics["nutrition"] = {"records": len(food), "core_fields_present": core_present,
                            "finite_core_complete": finite_core, "finite_core_incomplete": len(food) - finite_core,
                            "nonnegative_finite_core_complete": nonnegative_core,
                            "negative_measurement_records": len({r["source_row"] for r in anomalies if r["finding"] == "negative"}),
                            "invalid_or_nonfinite_measurement_cells": sum(r["finding"] != "negative" for r in anomalies)}
    table("nutrition_field_completeness", ["source_id", "nutrient", "total_records", "missing_records", "missing_pct",
          "invalid_numeric_records", "nonfinite_records", "negative_records", "finite_records", "finite_nonnegative_records"], completeness)
    table("nutrition_descriptive_statistics", ["source_id", "nutrient", "analysis_subset", "eligible_records", "mean",
          "standard_deviation_population", "minimum", "q1", "median", "q3", "maximum", "reference_amount_status"], statistics)

    # Pairwise samples keep usable measurements even when another nutrient is absent.
    def ranks(values):
        ordered = sorted(range(len(values)), key=lambda i: values[i])
        result = [0.] * len(values)
        start = 0
        while start < len(values):
            stop = start + 1
            while stop < len(values) and values[ordered[stop]] == values[ordered[start]]:
                stop += 1
            for index in ordered[start:stop]:
                result[index] = (start + 1 + stop) / 2
            start = stop
        return result

    def pearson(left, right):
        scale_left, scale_right = max(map(abs, left), default=0), max(map(abs, right), default=0)
        if not scale_left or not scale_right:
            return None
        left = [value / scale_left for value in left]
        right = [value / scale_right for value in right]
        center_left, center_right = mean(left), mean(right)
        numerator = sum((a - center_left) * (b - center_right) for a, b in zip(left, right))
        denominator = math.sqrt(sum((a - center_left) ** 2 for a in left) * sum((b - center_right) ** 2 for b in right))
        return max(-1., min(1., numerator / denominator)) if denominator else None

    correlations = []
    for first, second in combinations(["protein_g", "fat_g", "carbohydrate_g", "fiber_g", "calories_kcal"], 2):
        pairs = [(eligible(row.get(first)), eligible(row.get(second))) for row in food]
        pairs = [(a, b) for a, b in pairs if a is not None and b is not None]
        left, right = [p[0] for p in pairs], [p[1] for p in pairs]
        for method in ("pearson", "spearman"):
            status, value = "insufficient_pairs", None
            if len(pairs) >= 10:
                value = pearson(left, right) if method == "pearson" else pearson(ranks(left), ranks(right))
                status = "computed" if value is not None else "constant_input"
            correlations.append({"source_id": nutrition_sid, "first_field": first, "second_field": second,
                                 "method": method, "eligible_pair_records": len(pairs), "excluded_records": len(food) - len(pairs),
                                 "coefficient": value if value is not None else "", "status": status,
                                 "eligibility": "Both fields finite and nonnegative; pairwise sample; minimum 10 pairs.",
                                 "interpretation_limit": "Selected reference foods; derived energy may depend on macronutrients; no causal or athlete-outcome inference."})
    table("nutrition_pairwise_correlations", ["source_id", "first_field", "second_field", "method", "eligible_pair_records",
          "excluded_records", "coefficient", "status", "eligibility", "interpretation_limit"], correlations)
    table("nutrition_energy_methods", frequency_header, frequencies(nutrition_sid, "energy_method", "food_records"))
    paired_energy = []
    for first, second in [("energy_general_kcal", "energy_specific_kcal"), ("calories_kcal", "energy_specific_kcal"),
                          ("calories_kcal", "energy_general_kcal")]:
        pairs = [(eligible(row.get(first)), eligible(row.get(second))) for row in food]
        differences = [a - b for a, b in pairs if a is not None and b is not None]
        paired_energy.append({"source_id": nutrition_sid, "first_field": first, "second_field": second,
                              "eligible_pair_records": len(differences), "excluded_records": len(food) - len(differences),
                              "mean_signed_difference": mean(differences) if differences else "",
                              "median_signed_difference": median(differences) if differences else "",
                              "mean_absolute_difference": mean(abs(value) for value in differences) if differences else "",
                              "maximum_absolute_difference": max(map(abs, differences)) if differences else "",
                              "interpretation_limit": "Differences between energy conventions are not automatically errors; all pairs use the same source row."})
    energy_inconsistent, energy_uncheckable, energy_checked = 0, 0, 0
    for position, row in enumerate(food, 1):
        method = text(row.get("energy_method"))
        field = "energy_specific_kcal" if method == "Atwater Specific (2048)" else "energy_general_kcal" if method == "Atwater General (2047)" else None
        actual = eligible(row.get("calories_kcal"))
        expected = eligible(row.get(field)) if field else None
        if field and actual is not None and expected is not None:
            energy_checked += 1
            if not math.isclose(actual, expected, rel_tol=1e-9, abs_tol=1e-8):
                energy_inconsistent += 1
                anomalies.append({"source_id": nutrition_sid, "source_row": position, "fdc_id": text(row.get("fdc_id")),
                                  "field": "calories_kcal", "source_value": text(row.get("calories_kcal")),
                                  "finding": "selected_energy_method_mismatch",
                                  "handling": "Investigate recorded energy_method and its selected source energy field."})
        else:
            energy_uncheckable += 1
    metrics["nutrition"].update(energy_method_rows_checked=energy_checked, energy_method_rows_uncheckable=energy_uncheckable,
                                energy_method_inconsistencies=energy_inconsistent)
    table("nutrition_energy_comparison", ["source_id", "first_field", "second_field", "eligible_pair_records", "excluded_records",
          "mean_signed_difference", "median_signed_difference", "mean_absolute_difference", "maximum_absolute_difference", "interpretation_limit"], paired_energy)
    table("nutrition_anomaly_review", ["source_id", "source_row", "fdc_id", "field", "source_value", "finding", "handling"], anomalies)
    finding("SUP-NUTRITION", "nutrition", f"{len(food)} food records; {finite_core} have finite core values and {nonnegative_core} have finite nonnegative core values.",
            "Use explicit numeric eligibility masks and report the sample size for each comparison.",
            "Completeness and nonnegativity do not independently verify food identity, reference amount, or suitability for an individual athlete.")
    finding("SUP-NUTRITION-DEPENDENCE", "nutrition", "Energy and macronutrient associations were described using finite nonnegative pairs.",
            "Interpret food-composition patterns in the context of the recorded energy convention.",
            "Derived food energy can be mathematically related to macronutrients; these are not independent athlete outcomes or evidence of benefit.")

    # We describe the two fictional profile sources separately.
    profile_ids = ["derived_athlete_profile", "aditi_synthetic_athletes"]
    planning_aliases = {
        "age": ["age"], "school_grade": ["school_grade", "grade", "grade_level"],
        "experience": ["experience_level"], "goal": ["goal", "primary_goal"],
        "available_days": ["available_days", "available_days_per_week"],
        "session_minutes": ["session_minutes", "session_duration_minutes", "minutes_per_session"],
        "existing_commitments": ["existing_commitments", "existing_basketball_commitments", "current_training_schedule", "organized_basketball_hours_per_week"],
        "sleep_information": ["sleep_information", "sleep_hours", "sleep_hours_per_night"],
        "annual_participation": ["annual_participation", "organized_basketball_months_per_year", "months_per_year"],
        "equipment": ["equipment"], "injury_status": ["injury_status"],
    }
    profile_distributions, profile_coverage, goal_cross, day_cross = [], [], [], []
    profile_metrics = {}
    for sid in profile_ids:
        rows, header = source_rows(sid), columns(sid)
        if sid not in tables_by_source_id:
            profile_metrics[sid] = {"availability": "not_loaded", "records": 0}
            continue
        for field in ("age", "sport", "position", "experience_level", "goal", "primary_goal", "available_days", "available_days_per_week",
                      "equipment", "injury_status", "scenario_type", "is_synthetic"):
            if field in header:
                profile_distributions.extend(frequencies(sid, field, "fictional_development_profiles"))
        for group, aliases in planning_aliases.items():
            found = [field for field in aliases if field in header]
            present = unknown = absent_values = 0
            for row in rows:
                values = [row.get(field) for field in found]
                if any(not missing(value, group) and text(value).lower() not in unknown_tokens for value in values):
                    present += 1
                elif any(text(value).lower() in unknown_tokens for value in values):
                    unknown += 1
                else:
                    absent_values += 1
            profile_coverage.append({"source_id": sid, "planning_field_group": group,
                                     "schema_presence": "present" if found else "absent", "source_fields": ";".join(found),
                                     "total_records": len(rows), "nonmissing_nonunknown_records": present,
                                     "explicit_unknown_records": unknown, "missing_value_records": absent_values if found else 0,
                                     "schema_absent_records": len(rows) if not found else 0,
                                     "interpretation_limit": "Recorded presence is not verification or a complete planning constraint; available days do not establish total workload."})
        goal_field = "primary_goal" if "primary_goal" in header else "goal"
        day_field = "available_days_per_week" if "available_days_per_week" in header else "available_days"
        for destination, second in ((goal_cross, goal_field), (day_cross, day_field)):
            cells = Counter((category(row.get("experience_level")), category(row.get(second))) for row in rows)
            level_totals = Counter(category(row.get("experience_level")) for row in rows)
            for (level, value), count in sorted(cells.items()):
                destination.append({"source_id": sid, "experience_level": level, "second_field": second, "value": value,
                                    "profiles": count, "experience_level_profiles": level_totals[level],
                                    "within_level_percent": percent(count, level_totals[level]),
                                    "interpretation_limit": "Generator scenario coverage only; fictional development sample, not population inference."})
        profile_metrics[sid] = {"availability": "loaded", "records": len(rows),
                                "unknown_injury_statuses": sum(text(row.get("injury_status")).lower() in unknown_tokens for row in rows),
                                "reported_no_injury_statuses": sum(text(row.get("injury_status")).lower() in {"none", "none_reported", "no injury reported", "no reported injury"} for row in rows),
                                "schema_absent_planning_groups": [name for name, aliases in planning_aliases.items() if not set(aliases).intersection(header)],
                                "evaluation_role": "fictional_development_examples; final-test independence not established"}
    metrics["profiles"] = profile_metrics
    table("synthetic_athlete_distributions", frequency_header, profile_distributions)
    cross_header = ["source_id", "experience_level", "second_field", "value", "profiles", "experience_level_profiles", "within_level_percent", "interpretation_limit"]
    table("synthetic_goal_by_experience", cross_header, goal_cross)
    table("synthetic_days_by_experience", cross_header, day_cross)
    table("profile_planning_input_coverage", ["source_id", "planning_field_group", "schema_presence", "source_fields", "total_records",
          "nonmissing_nonunknown_records", "explicit_unknown_records", "missing_value_records", "schema_absent_records", "interpretation_limit"], profile_coverage)
    finding("SUP-FICTIONAL-PROFILES", "profiles", "; ".join(f"{sid}: {values['records']} records" for sid, values in profile_metrics.items()),
            "Use the separate sources to examine request and constraint coverage and identify missing planner inputs.",
            "Fictional development examples do not establish population prevalence, validated drill labels, or an independent final test set.")

    # We keep occurrence counts, name groups, snapshots, and planner decisions distinct.
    drill_sid = "aditi_jr_nba_drills"
    drills, rag = source_rows(drill_sid), source_rows("aditi_jr_nba_rag")
    name_key = lambda value: " ".join(re.findall(r"[a-z0-9]+", text(value).lower()))
    drill_names = {name_key(row.get("drill_name")) for row in drills if name_key(row.get("drill_name"))}
    drill_ids = Counter(text(row.get("drill_id")) for row in drills)
    rag_ids = Counter(text(row.get("drill_id")) for row in rag)
    traceability = [{"metric": "imported_source_occurrences", "value": len(drills), "unit": "source_occurrences"},
                    {"metric": "imported_normalized_name_groups", "value": len(drill_names), "unit": "name_groups"},
                    {"metric": "rag_documents", "value": len(rag), "unit": "derived_documents"},
                    {"metric": "duplicate_drill_identifier_excess", "value": sum(n - 1 for key, n in drill_ids.items() if key and n > 1), "unit": "records"},
                    {"metric": "duplicate_rag_identifier_excess", "value": sum(n - 1 for key, n in rag_ids.items() if key and n > 1), "unit": "records"},
                    {"metric": "drill_ids_without_rag", "value": len(set(drill_ids) - set(rag_ids)), "unit": "identifiers"},
                    {"metric": "rag_ids_without_drill", "value": len(set(rag_ids) - set(drill_ids)), "unit": "identifiers"},
                    {"metric": "blank_rag_text", "value": sum(missing(row.get("text")) for row in rag), "unit": "documents"}]
    for row in traceability:
        row.update(source_id=drill_sid, interpretation_limit="Identifier correspondence is not independent source verification or retrieval-quality evaluation.")
    table("basketball_rag_traceability", ["source_id", "metric", "value", "unit", "interpretation_limit"], traceability)
    snapshots = {"catalog_basketball_v01": "candidate_groups_v0.1_historical", "catalog_basketball_v02": "candidate_groups_v0.2_historical",
                 "local_basketball_drills": "local_candidate_items"}
    skill_rows = frequencies(drill_sid, "skill_category", "source_occurrences")
    source_rows_report = frequencies(drill_sid, "source_document", "source_occurrences")
    level_rows = frequencies(drill_sid, "curriculum_level", "source_occurrences")
    participant_rows = frequencies(drill_sid, "group_requirement", "source_occurrences")
    catalog_summary, overlaps = [], []
    optional_metrics = {}
    for sid, unit in snapshots.items():
        rows, header = source_rows(sid), columns(sid)
        optional_metrics[sid] = {
            "availability": "loaded_historical_snapshot" if sid in tables_by_source_id else "not_loaded_optional",
            "records": len(rows),
            "name_comparison_status": "not_loaded" if sid not in tables_by_source_id else "compared_names" if "drill_name" in header else "not_comparable_schema",
            "comparable_catalog_name_groups": None,
        }
        if sid not in tables_by_source_id:
            continue
        for field in ("scope_candidate", "record_kind", "metadata_status", "distinctness_status", "permission_status", "catalog_approval_status", "planner_eligible", "review_status"):
            if field in header:
                catalog_summary.extend(frequencies(sid, field, unit))
        for destination, choices in ((skill_rows, ["skill_area", "proposed_primary_skill"]),
                                     (source_rows_report, ["source_plan", "source_id"]),
                                     (level_rows, ["curriculum_levels", "publisher_curriculum_level"]),
                                     (participant_rows, ["participants", "participant_format"])):
            field = next((field for field in choices if field in header), None)
            if field:
                destination.extend(frequencies(sid, field, unit))
        if "drill_name" in header:
            names = defaultdict(list)
            for row in rows:
                key = name_key(row.get("drill_name"))
                if key:
                    names[key].append(text(row.get("catalog_candidate_id")))
            optional_metrics[sid]["comparable_catalog_name_groups"] = len(names)
            for key in sorted(drill_names):
                overlaps.append({"imported_source_id": drill_sid, "historical_catalog_source_id": sid, "normalized_name": key,
                                 "matched_candidate_groups": len(names.get(key, [])), "candidate_ids": ";".join(names.get(key, [])),
                                 "interpretation_limit": "Name correspondence only; snapshots overlap and are not additive or canonical identity decisions."})
    table("basketball_skill_coverage", frequency_header, skill_rows)
    table("basketball_source_coverage", frequency_header, source_rows_report)
    table("basketball_level_coverage", frequency_header, level_rows)
    table("basketball_participant_requirements", frequency_header, participant_rows)
    table("basketball_catalog_review_summary", frequency_header, catalog_summary)
    table("basketball_name_overlap_review", ["imported_source_id", "historical_catalog_source_id", "normalized_name", "matched_candidate_groups", "candidate_ids", "interpretation_limit"], overlaps)
    metrics["basketball"] = {"imported_source_occurrences": len(drills), "imported_normalized_name_groups": len(drill_names),
                             "rag_documents": len(rag), "optional_historical_sources": optional_metrics,
                             "name_overlaps_by_snapshot": {
                                 sid: sum(row["matched_candidate_groups"] > 0 for row in overlaps if row["historical_catalog_source_id"] == sid)
                                 if optional_metrics[sid]["name_comparison_status"] == "compared_names" else None
                                 for sid in snapshots}}
    finding("SUP-BASKETBALL", "basketball", f"{len(drills)} imported activity occurrences form {len(drill_names)} normalized name groups and link to {len(rag)} derived RAG documents.",
            "Review coaching coverage, identities, permissions, and requirements in Step 12; evaluate retrieval separately.",
            "Derived RAG documents are another representation of the same activities, not additional independent drills or source evidence. Historical catalog versions are not additive.")

    # Recovery tables describe recorded guidance; they do not activate planning rules.
    guidance_sid, rules_sid = "aditi_recovery_guidance", "recovery_rules_v01"
    guidance, rules = source_rows(guidance_sid), source_rows(rules_sid)
    table("recovery_guidance_topics", frequency_header, frequencies(guidance_sid, "topic", "guidance_records"))
    table("recovery_rule_types", frequency_header, frequencies(rules_sid, "rule_type", "reference_rules"))
    rule_groups = defaultdict(list)
    for row in rules:
        rule_groups[tuple(text(row.get(field)) for field in ("segment", "metric", "rule_type", "unit"))].append(row)
    coverage = []
    for key, members in sorted(rule_groups.items()):
        coverage.append({"source_id": rules_sid, "segment": key[0], "metric": key[1], "rule_type": key[2], "unit": key[3],
                         "rules": len(members), "source_urls_present": sum(text(row.get("source_url")).startswith("https://") for row in members),
                         "lower_bounds_present": sum(not missing(row.get("value_min")) for row in members),
                         "upper_bounds_present": sum(not missing(row.get("value_max")) for row in members),
                         "verification_statuses": ";".join(sorted({text(row.get("verification_status")) for row in members})),
                         "interpretation_limit": "Blank bounds can be conditional; source population, age/grade overlaps, and individual applicability need Step 13 review."})
    table("recovery_rule_coverage", ["source_id", "segment", "metric", "rule_type", "unit", "rules", "source_urls_present", "lower_bounds_present", "upper_bounds_present", "verification_statuses", "interpretation_limit"], coverage)
    checked_recovery = step09_summary.get("domain_metrics", {}).get("recovery_rules", {})
    metrics["recovery"] = {"guidance_records": len(guidance), "rule_records": len(rules),
                           "recorded_verification_status_counts": dict(Counter(text(row.get("verification_status")) for row in rules)),
                           "step09_conditional_schema_findings": {key: checked_recovery.get(key) for key in ("invalid_bounds", "invalid_segment_pairs", "incomplete_source_fields")},
                           "planner_rules_activated_by_this_analysis": 0}
    finding("SUP-RECOVERY", "recovery", f"{len(guidance)} guidance records and {len(rules)} structured reference rules were described.",
            "Use Step 13 to reconcile population segments, evidence, and planning inputs before implementation.",
            "These are reference rules, not measured recovery outcomes. Recorded verification wording is not an individual prescription or approval by this analysis.")

    master = step09_summary.get("domain_metrics", {}).get("master", {})
    master_rows = [{"entity_type": domain, "master_records": count,
                    "prepared_domain_records": master.get("domains", {}).get(domain, {}).get("prepared_rows", ""),
                    "role": "heterogeneous source index; historical/context/reference domains remain separate",
                    "interpretation_limit": "Domain totals are not independent athlete examples or rows of a jointly labeled training matrix."}
                   for domain, count in sorted(master.get("entity_counts", {}).items())]
    table("master_entity_coverage", ["entity_type", "master_records", "prepared_domain_records", "role", "interpretation_limit"], master_rows)
    provenance = [{"source_id": row.get("source_id", ""), "availability_at_step08": row.get("availability", ""),
                   "observed_records_at_step08": row.get("observed_records", ""), "hash_status_at_step08": row.get("hash_status", ""),
                   "count_status_at_step08": row.get("count_status", ""), "producer_step": row.get("producer_step", ""),
                   "interpretation_limit": "Earlier audit snapshot; producer records supersede checksums for regenerated later outputs. Availability alone is not permissions or content approval."}
                  for row in source_inventory]
    table("source_provenance_coverage", ["source_id", "availability_at_step08", "observed_records_at_step08", "hash_status_at_step08", "count_status_at_step08", "producer_step", "interpretation_limit"], provenance)
    historical_gaps = sum(text(row.get("source_id")).startswith("raw_") and not text(row.get("observed_path")) for row in source_inventory)
    metrics["master"] = {"records": master.get("records", 0), "entity_counts": master.get("entity_counts", {}),
                         "structural_findings_from_step09": master.get("structural_findings", {})}
    metrics["provenance"] = {"registered_source_entries": len(source_inventory), "historical_input_gaps": historical_gaps}
    finding("SUP-MASTER", "master", f"The verified Step 9 summary records {master.get('records', 0)} master-index rows across {len(master_rows)} entity types.",
            "Keep source domains separate and choose inputs according to each project component.",
            "Historical player records and nutrition tables do not supply validated youth drill-to-profile outcomes or a joint supervised target.")
    finding("SUP-PROVENANCE", "provenance", f"{historical_gaps} historical inputs were absent from the checked Step 8 locations.",
            "Retain the lineage limitation and resolve publisher inputs and permissions before release.",
            "This local analysis does not refresh publisher pages, validate source acquisition, or grant usage rights.")
    return reports, metrics, findings

if not eda_state.get("inputs_verified"):
    raise RuntimeError("Complete the Step 11 input-verification cell first.")
eda_state.update(supporting_complete=False, integrated_complete=False, status="analyzing")
eda_checks[:] = [row for row in eda_checks if not row["check_name"].startswith("nutrition_summary:")
                and row["check_name"] not in {"master_index_counts", "drill_document_correspondence"}]
eda_supporting_tables, eda_supporting_metrics, eda_supporting_findings = analyze_supporting_sources(
    eda_tables_by_source, eda_inventory, eda_quality)
for eda_measure in ("records", "finite_core_complete", "finite_core_incomplete", "nonnegative_finite_core_complete"):
    eda_check(f"nutrition_summary:{eda_measure}", eda_supporting_metrics["nutrition"][eda_measure]
              == eda_quality["domain_metrics"]["nutrition"][eda_measure],
              "The recomputed nutrition screening agrees with the verified Step 9 summary.")
eda_check("master_index_counts", sum(eda_supporting_metrics["master"]["entity_counts"].values())
          == eda_supporting_metrics["master"]["records"], "The separate domain totals reconcile to the verified master index.")
eda_traceability = {row["metric"]: row["value"] for row in eda_supporting_tables["basketball_rag_traceability.csv"]["rows"]}
eda_check("drill_document_correspondence", all(eda_traceability[metric] == 0 for metric in (
    "duplicate_drill_identifier_excess", "duplicate_rag_identifier_excess", "drill_ids_without_rag", "rag_ids_without_drill", "blank_rag_text")),
    "Imported occurrences and their derived documents have one-to-one identifier correspondence.")
eda_state["supporting_complete"] = True
print("Food-reference screening:", eda_supporting_metrics["nutrition"])
print("Nutrient coverage and numerical states")
print(pd.DataFrame(eda_supporting_tables["nutrition_field_completeness.csv"]["rows"])[
    ["nutrient", "missing_records", "nonfinite_records", "invalid_numeric_records", "negative_records", "finite_nonnegative_records"]
].to_string(index=False))
print("Fieldwise nutrient summaries from finite nonnegative values")
print(pd.DataFrame(eda_supporting_tables["nutrition_descriptive_statistics.csv"]["rows"])[
    ["nutrient", "eligible_records", "mean", "median", "minimum", "maximum"]
].round(3).to_string(index=False))
print("Paired energy-field comparisons (same food record)")
print(pd.DataFrame(eda_supporting_tables["nutrition_energy_comparison.csv"]["rows"])[
    ["first_field", "second_field", "eligible_pair_records", "mean_signed_difference", "mean_absolute_difference"]
].round(3).to_string(index=False))
print("Food-composition correlations with recorded calories")
eda_correlation_excerpt = [row for row in eda_supporting_tables["nutrition_pairwise_correlations.csv"]["rows"]
                           if row["second_field"] == "calories_kcal"]
print(pd.DataFrame(eda_correlation_excerpt)[
    ["first_field", "method", "eligible_pair_records", "coefficient", "status"]
].round(4).to_string(index=False))
print("Fictional profile sources:", eda_supporting_metrics["profiles"])
print("Basketball source coverage:", eda_supporting_metrics["basketball"])
print("Recovery reference coverage:", eda_supporting_metrics["recovery"])
print("Master index rows:", eda_supporting_metrics["master"]["records"], "| Historical input gaps:", eda_supporting_metrics["provenance"]["historical_input_gaps"])
```

#### Variable Relationships and Exploratory Findings

Our exercise EDA examined source labels, description availability, equipment, difficulty, exercise type, and text length. Bias-corrected Cramér's V was **0.1492** for body-part label and description availability, **0.1982** for label and equipment, **0.2172** for label and difficulty, and **0.1945** for label and exercise type. Exercise names averaged **3.86 words**. The 1,359 available descriptions averaged **54.48 words**, with a median of 54 and a maximum of 108. Constructed classifier text averaged 29.31 words, with a median of seven. These results reflect the mixture of name-only and described records and support examining performance by text availability. The counts describe word patterns rather than model-tokenizer lengths.

The largest source class accounts for **22.69%** of preliminary labeled candidates. This is a full-catalog majority reference, not a measured model accuracy or held-out baseline result.

```python
# Step 12-1 — Verify our saved basketball catalog and upstream EDA
# This cell reads local files and leaves source records and review decisions intact.
import csv
import hashlib
import io
import json
import re
from collections import Counter
from pathlib import Path, PureWindowsPath

step12_report = {"status": "checking", "inputs_verified": False}
bb12_checks, bb12_observed_hashes, bb12_expected_hashes = [], {}, {}


def bb12_require(condition, message):
    if not condition:
        step12_report.update(status="blocked", inputs_verified=False, reason=message)
        raise RuntimeError("Step 12-1: " + message)


def bb12_check(name, condition, detail):
    bb12_checks.append({"check_name": name, "status": "pass" if condition else "failed"})
    bb12_require(condition, detail)


# Reuse our established root, or locate it from the saved Step 11 record.
bb12_hint = globals().get("project_root")
bb12_roots = ([Path(bb12_hint).resolve()] if bb12_hint is not None else [])
bb12_cwd = Path.cwd().resolve()
bb12_roots.extend([bb12_cwd, *bb12_cwd.parents])
bb12_root = next((p for p in bb12_roots
                  if (p / "results/step11_eda_summary.json").is_file()), None)
bb12_require(bb12_root is not None,
             "Run inside our project after completing Step 11; its saved producer record was not found.")


def bb12_path(reference):
    bb12_require(isinstance(reference, str) and bool(reference), "An artifact path is empty or invalid.")
    normalized = reference.replace("\\", "/")
    relative = Path(normalized)
    bb12_require(not relative.is_absolute() and not PureWindowsPath(reference).drive,
                 "Artifact paths must be relative to our project: " + reference)
    path = (bb12_root / relative).resolve()
    bb12_require(path.is_relative_to(bb12_root), "An artifact path leaves our project: " + reference)
    return path


def bb12_read(reference, expected=None):
    path = bb12_path(reference)
    bb12_require(path.is_file(), "Missing local artifact: " + reference)
    content = path.read_bytes()
    digest = hashlib.sha256(content).hexdigest()
    if expected is not None:
        bb12_require(isinstance(expected, str) and re.fullmatch(r"[0-9a-f]{64}", expected),
                     "Invalid recorded SHA-256: " + reference)
        bb12_require(reference not in bb12_expected_hashes or bb12_expected_hashes[reference] == expected,
                     "Producer records disagree about: " + reference)
        bb12_expected_hashes[reference] = expected
        bb12_require(digest == expected,
                     "Saved checksum differs: " + reference + ". Share this message before rerunning earlier steps.")
    bb12_observed_hashes[path.relative_to(bb12_root).as_posix()] = digest
    return content


def bb12_pairs(pairs):
    result = {}
    for key, value in pairs:
        bb12_require(key not in result, "A JSON object contains a repeated key: " + key)
        result[key] = value
    return result


def bb12_json(reference):
    try:
        value = json.loads(bb12_read(reference).decode("utf-8-sig"), object_pairs_hook=bb12_pairs)
    except (UnicodeError, json.JSONDecodeError) as error:
        bb12_require(False, "Unreadable JSON: " + reference + ". " + str(error))
    bb12_require(isinstance(value, dict), "Expected a JSON object: " + reference)
    return value


def bb12_csv(reference, required=()):
    reader = csv.DictReader(io.StringIO(bb12_read(reference).decode("utf-8-sig"), newline=""))
    columns = list(reader.fieldnames or [])
    rows = list(reader)
    bb12_require(bool(columns) and all(columns) and len(columns) == len(set(columns))
                 and set(required).issubset(columns), "Missing or invalid CSV columns: " + reference)
    bb12_require(all(None not in row and all(v is not None for v in row.values()) for row in rows),
                 "Malformed CSV records: " + reference)
    return rows, columns


def bb12_index(rows, field, label):
    values = [row[field] for row in rows]
    bb12_require(all(isinstance(value, str) and value.strip() for value in values)
                 and len(values) == len(set(values)),
                 "Missing or duplicate " + field + " in " + label)
    return dict(zip(values, rows))


def bb12_ids(value):
    return {part.strip() for part in value.split(";") if part.strip()}


# Confirm the completed Step 11 run, its inspected inputs, and its saved reports.
bb12_eda = bb12_json("results/step11_eda_summary.json")
bb12_check("step11_producer", bb12_eda.get("schema_version") == "step11-integrated-eda-1"
           and bb12_eda.get("status") in {"passed", "completed_with_reviews"},
           "Step 11 needs a completed producer record.")
bb12_prior_checks = bb12_eda.get("verified_checks", [])
bb12_check("step11_checks", isinstance(bb12_prior_checks, list) and bool(bb12_prior_checks)
           and all(isinstance(row, dict) and row.get("status") == "pass" for row in bb12_prior_checks),
           "Step 11 contains missing or failed verification checks.")
bb12_check("step11_manifest", isinstance(bb12_eda.get("input_hashes"), dict)
           and bool(bb12_eda["input_hashes"]) and isinstance(bb12_eda.get("artifacts"), dict)
           and bool(bb12_eda["artifacts"]), "Step 11 has no complete artifact inventory.")
for bb12_ref, bb12_digest in bb12_eda["input_hashes"].items():
    bb12_read(bb12_ref, bb12_digest)
for bb12_ref, bb12_metadata in bb12_eda["artifacts"].items():
    bb12_require(isinstance(bb12_metadata, dict) and "sha256" in bb12_metadata,
                 "Invalid Step 11 artifact entry: " + bb12_ref)
    bb12_read(bb12_ref, bb12_metadata["sha256"])

# Resolve imported inputs through the source register rather than copying them.
bb12_manifest, _ = bb12_csv("data/source_manifest.csv", ("source_id", "relative_path"))
bb12_sources = bb12_index(bb12_manifest, "source_id", "source register")
bb12_imported = {}
for bb12_sid in ("aditi_jr_nba_drills", "aditi_jr_nba_rag"):
    bb12_require(bb12_sid in bb12_sources, "Required basketball source is not registered: " + bb12_sid)
    bb12_ref = bb12_sources[bb12_sid]["relative_path"]
    bb12_require(bb12_ref in bb12_eda["input_hashes"], "Basketball source was not verified by Step 11: " + bb12_sid)
    if bb12_sid == "aditi_jr_nba_drills":
        bb12_rows, bb12_columns = bb12_csv(bb12_ref, ("drill_id", "drill_name"))
    else:
        bb12_rows = json.loads(bb12_read(bb12_ref).decode("utf-8-sig"), object_pairs_hook=bb12_pairs)
        bb12_require(isinstance(bb12_rows, list) and all(isinstance(row, dict) and "drill_id" in row
                     for row in bb12_rows), "Expected RAG document objects with drill_id fields.")
        bb12_columns = sorted({key for row in bb12_rows for key in row})
    bb12_imported[bb12_sid] = {"path": bb12_ref, "rows": bb12_rows, "columns": bb12_columns}
bb12_drills = bb12_index(bb12_imported["aditi_jr_nba_drills"]["rows"], "drill_id", "imported activities")
bb12_documents = bb12_index(bb12_imported["aditi_jr_nba_rag"]["rows"], "drill_id", "RAG documents")
bb12_check("imported_document_links", set(bb12_drills) == set(bb12_documents),
           "Imported activity IDs and RAG document IDs do not correspond exactly.")

# Inspect the latest checkpoint recorded in our notebook; do not rebuild older versions.
bb12_manifests = {}
for bb12_step in ("12-64", "12-65", "12-66"):
    bb12_ref = "results/basketball_step" + bb12_step.replace("-", "_") + "_summary_v0_3_9.json"
    bb12_record = bb12_json(bb12_ref)
    bb12_check("producer:" + bb12_step,
               bb12_record.get("step") == bb12_step and bb12_record.get("catalog_version") == "0.3.9",
               "The recorded basketball checkpoint does not identify " + bb12_step + " / v0.3.9.")
    for bb12_kind in ("input_sha256", "output_sha256"):
        bb12_mapping = bb12_record.get(bb12_kind)
        bb12_require(isinstance(bb12_mapping, dict) and bool(bb12_mapping),
                     "Missing checkpoint hash inventory: " + bb12_step + " / " + bb12_kind)
        for bb12_ref, bb12_digest in bb12_mapping.items():
            bb12_read(bb12_ref.replace("\\", "/"), bb12_digest)
    bb12_manifests[bb12_step] = bb12_record
bb12_checkpoint = bb12_manifests["12-66"]
bb12_check("historical_checkpoint", bb12_checkpoint.get("integration_checkpoint_status") == "passed"
           and bb12_checkpoint.get("audit_checks_failed") == 0,
           "The saved Step 12 integration checkpoint did not pass.")

bb12_catalog_ref = "data/processed/basketball_drill_catalog_v0_3_9_REVIEW_ONLY.csv"
bb12_registry_ref = "results/basketball_drill_source_registry_v0_3_9.csv"
bb12_review_ref = "results/basketball_wave3_draft_metadata_review_v0_3_9.csv"
bb12_catalog, bb12_catalog_columns = bb12_csv(bb12_catalog_ref, (
    "catalog_candidate_id", "catalog_version", "scope_candidate", "skill_area", "evidence_ids",
    "aditi_evidence_ids", "total_evidence_records", "source_identity_review_cases",
    "approved_drill_id", "catalog_approval_status", "planner_eligible"))
bb12_registry, bb12_registry_columns = bb12_csv(bb12_registry_ref, ("evidence_id", "catalog_candidate_id"))
bb12_review, bb12_review_columns = bb12_csv(bb12_review_ref, (
    "candidate_group_id", "field", "reviewer_decision", "human_confirmation_recorded"))
bb12_candidates = bb12_index(bb12_catalog, "catalog_candidate_id", "catalog")
bb12_evidence = bb12_index(bb12_registry, "evidence_id", "source registry")
bb12_check("catalog_version", all(row["catalog_version"] == "0.3.9" for row in bb12_catalog),
           "Catalog rows have mixed or unexpected versions.")
bb12_check("registry_candidate_links", all(row["catalog_candidate_id"] in bb12_candidates
           for row in bb12_registry), "A source occurrence points to an unknown candidate.")
for bb12_cid, bb12_row in bb12_candidates.items():
    bb12_declared = bb12_ids(bb12_row["evidence_ids"]) | bb12_ids(bb12_row["aditi_evidence_ids"])
    bb12_actual = {key for key, row in bb12_evidence.items() if row["catalog_candidate_id"] == bb12_cid}
    bb12_check("source_links:" + bb12_cid, bool(bb12_actual) and bb12_declared == bb12_actual
               and str(len(bb12_actual)) == bb12_row["total_evidence_records"],
               "Catalog/source membership differs for " + bb12_cid)
bb12_scope = Counter(row["scope_candidate"] for row in bb12_catalog)
bb12_check("scope_labels", set(bb12_scope).issubset({"basketball_drill_candidate", "general_warmup_candidate"}),
           "An unrecognized scope needs review before counting basketball candidates.")
bb12_metrics = {
    "catalog_candidate_groups": len(bb12_catalog),
    "registered_source_occurrences": len(bb12_registry),
    "basketball_candidate_groups": bb12_scope["basketball_drill_candidate"],
    "general_warmup_candidate_groups": bb12_scope["general_warmup_candidate"],
    "human_field_confirmations_pending": sum(row["human_confirmation_recorded"].strip().casefold() == "no"
                                            for row in bb12_review),
    "explicit_identity_cases_pending": len(set().union(*(bb12_ids(row["source_identity_review_cases"])
                                                         for row in bb12_catalog))),
    "catalog_approvals": sum(row["catalog_approval_status"].strip().casefold() == "approved" for row in bb12_catalog),
    "planner_eligible": sum(row["planner_eligible"].strip().casefold() in {"yes", "true", "1"} for row in bb12_catalog),
}
bb12_check("review_links", all(row["candidate_group_id"] in bb12_candidates
           and row["field"] in bb12_catalog_columns for row in bb12_review),
           "A pending field review refers to an unknown candidate or catalog column.")
bb12_check("pending_confirmation_state", all(row["human_confirmation_recorded"].strip().casefold() == "no"
           and row["reviewer_decision"].strip().casefold() == "pending" for row in bb12_review),
           "Review decisions changed since this draft; share the output so they can be preserved.")
for bb12_key, bb12_value in bb12_metrics.items():
    bb12_check("checkpoint_count:" + bb12_key, bb12_checkpoint.get(bb12_key) == bb12_value,
               "Current counts disagree with the saved checkpoint: " + bb12_key)
bb12_check("unapproved_draft", bb12_metrics["catalog_approvals"] == bb12_metrics["planner_eligible"] == 0
           and all(not row["approved_drill_id"].strip()
                   and row["catalog_approval_status"].strip().casefold() == "pending"
                   and row["planner_eligible"].strip().casefold() == "no" for row in bb12_catalog),
           "Approval states differ from the recorded draft; share the output before continuing.")

# Detect edits made while the inspection was running. No project files are written.
for bb12_ref, bb12_digest in list(bb12_observed_hashes.items()):
    bb12_read(bb12_ref, bb12_digest)
step12_report.update(status="inputs_verified", inputs_verified=True, catalog_version="0.3.9",
                     metrics=bb12_metrics, verified_checks=bb12_checks,
                     input_hashes=dict(bb12_observed_hashes), project_files_written=0,
                     source_refresh_performed=False, approval_decisions_created=0)
print("Step 12-1 — current local inputs verified")
print(f"Step 11: {len(bb12_prior_checks)} saved checks; {len(bb12_eda['artifacts'])} saved reports verified")
print(f"Imported activities / RAG documents: {len(bb12_drills)} / {len(bb12_documents)}; exact ID correspondence")
print(f"Catalog v0.3.9: {len(bb12_catalog)} candidates / {len(bb12_registry)} source occurrences")
print(f"Basketball candidates / general warm-ups: {bb12_metrics['basketball_candidate_groups']} / {bb12_metrics['general_warmup_candidate_groups']}")
print(f"Pending field confirmations / explicit identity cases: {bb12_metrics['human_field_confirmations_pending']} / {bb12_metrics['explicit_identity_cases_pending']}")
print(f"Approved / planner eligible: {bb12_metrics['catalog_approvals']} / {bb12_metrics['planner_eligible']}")
print(f"Local files inspected: {len(bb12_observed_hashes)} | Verification checks passed: {len(bb12_checks)}")
print("Project files written: 0 | Source bytes unchanged")
print("Next: inspect this output before preparing the revised catalog schema and review worklist.")
```

```python
# Step 12-2 — Assess our catalog schema and organize the review work
# We prepare derived reports in memory; saving follows after local output review.
import copy
from collections import Counter, defaultdict

if not globals().get("step12_report", {}).get("inputs_verified"):
    raise RuntimeError("Run the revised Step 12-1 successfully before this cell.")
step12_report.update(status="assessing_schema", schema_review_complete=False)
step12_tables = {}
bb12s_before = dict(step12_report["input_hashes"])
for bb12s_ref, bb12s_digest in bb12s_before.items():
    bb12_read(bb12s_ref, bb12s_digest)

# Reload inspected records so later kernel edits cannot silently change our inputs.
bb12s_catalog, bb12s_catalog_columns = bb12_csv(bb12_catalog_ref)
bb12s_registry, bb12s_registry_columns = bb12_csv(bb12_registry_ref)
bb12s_reviews, bb12s_review_columns = bb12_csv(bb12_review_ref)
bb12s_sources = bb12_index(bb12_csv("data/source_manifest.csv")[0], "source_id", "source register")
bb12s_import_ref = bb12s_sources["aditi_jr_nba_drills"]["relative_path"]
bb12s_imports, bb12s_import_columns = bb12_csv(bb12s_import_ref)
bb12s_candidates = bb12_index(bb12s_catalog, "catalog_candidate_id", "catalog")
bb12s_evidence = bb12_index(bb12s_registry, "evidence_id", "source registry")
bb12s_checks = []


def bb12s_check(name, condition, detail):
    bb12s_checks.append({"check_name": name, "status": "pass" if condition else "failed"})
    bb12_require(condition, detail)


def bb12s_state(value):
    text = re.sub(r"\s+", " ", value).strip().casefold().rstrip(".")
    if not text:
        return "blank_in_catalog"
    if text in {"not specified", "not stated", "unspecified", "unknown", "not provided", "n/a", "na"}:
        return "recorded_unspecified"
    if text in {"see source practice plan", "see source", "refer to source", "see practice plan"}:
        return "reference_only"
    if text == "not explicitly required" or text.endswith("| not explicitly required"):
        return "recorded_requirement_uncertain"
    return "recorded_value_present"


# Imported values stay occurrence-level and retain their original wording.
# The registry supplies identity; names are never used as a join key.
bb12s_occurrences = []
for bb12s_raw in bb12s_imports:
    bb12s_eid = "ADI-" + bb12s_raw["drill_id"]
    bb12s_check("imported_registry_link:" + bb12s_raw["drill_id"], bb12s_eid in bb12s_evidence,
                 "Imported activity has no registered source occurrence: " + bb12s_raw["drill_id"])
    bb12s_source = bb12s_evidence[bb12s_eid]
    bb12s_cid = bb12s_source["catalog_candidate_id"]
    bb12s_occurrences.append({
        "imported_drill_id": bb12s_raw["drill_id"], "evidence_id": bb12s_eid,
        "catalog_candidate_id": bb12s_cid, "catalog_drill_name": bb12s_candidates[bb12s_cid]["drill_name"],
        "registered_drill_name": bb12s_source["drill_name"],
        "source_url": bb12s_source["source_url"], "input_artifact": bb12s_import_ref,
        "input_sha256": bb12s_before[bb12s_import_ref],
        "imported_content_review_status": "pending_instruction_and_annotation_review",
        "adaptation_authorship_status": "publisher_or_preparer_authorship_not_established_by_this_cell",
        **{"imported_" + field: value for field, value in bb12s_raw.items() if field != "drill_id"},
    })
bb12s_check("imported_occurrence_identity", len({r["evidence_id"] for r in bb12s_occurrences}) == len(bb12s_imports),
             "Imported evidence links must be unique.")
bb12s_imported_candidates = {r["catalog_candidate_id"] for r in bb12s_occurrences}

# This is a mapping specification, not an application of imported values.
bb12s_mapping_specs = [
    ("drill_id", "source occurrence identity", "evidence_id", "Use the saved ADI- identifier correspondence; preserve imported drill_id."),
    ("drill_name", "original activity name", "drill_name", "Keep raw name separately; preserve reviewed catalog and registry names."),
    ("skill_category", "imported skill label", "skill_area", "Retain raw category; reconcile any classification difference before remapping."),
    ("curriculum_level", "publisher curriculum label", "curriculum_levels", "Preserve occurrence level; candidate memberships may overlap."),
    ("duration", "mixed prescription text", "source_prescription", "Keep source wording; do not derive a standalone duration automatically."),
    ("equipment", "imported equipment annotation", "equipment", "Compare with reviewed occurrence evidence; do not overwrite reconciled requirements."),
    ("instructions_summary", "instructional summary", "instructions", "Retain per occurrence with source links; instructional accuracy and intended use need review."),
    ("source_page", "recorded document page", "source_page", "Preserve document context; a PDF page and website location are different locators."),
    ("group_requirement", "participant arrangement", "participants", "Preserve qualitative roles; do not infer numeric minimum participants."),
    ("primary_goal", "activity goal", "primary_goal", "Retain as imported annotation until its source basis is reviewed."),
    ("difficulty_modification", "modification annotation", "difficulty_modification", "Preserve separately from base instructions and establish provenance."),
    ("source_document", "recorded source document", "source_document", "Retain occurrence document identity alongside the registry URL."),
    ("solo_adaptation_possible", "imported solo-adaptation flag", "solo_adaptation_possible", "Preserve the original flag; it does not establish approved solo feasibility."),
    ("solo_adaptation_note", "solo-adaptation annotation", "solo_adaptation", "Preserve separately; publisher authorship and practical feasibility remain to be reviewed."),
]
bb12s_check("mapping_field_coverage", {r[0] for r in bb12s_mapping_specs} == set(bb12s_import_columns),
             "Imported schema changed; update the explicit mapping before continuing.")
bb12s_registry_fields = {"drill_id", "source_page", "source_document"}
bb12s_mappings = [{"input_field": field, "concept": concept, "proposed_destination": destination,
                  "destination_table": "registry" if field in bb12s_registry_fields else "catalog",
                  "destination_column_present": "yes" if destination in (
                      bb12s_registry_columns if field in bb12s_registry_fields else bb12s_catalog_columns) else "no",
                  "imported_nonblank_records": sum(bool(row[field].strip()) for row in bb12s_imports),
                  "mapping_policy": policy, "application_status": "not_applied"}
                 for field, concept, destination, policy in bb12s_mapping_specs]

# Missing columns and missing evidence are different questions. The imported
# 'duration' field contains mixed prescriptions, not a verified duration field.
bb12s_specs = [
    ("candidate_identity", "catalog_candidate_id", "catalog", "", "Preserve current stable candidate IDs."),
    ("approved_drill_identity", "approved_drill_id", "catalog", "", "Assign only after canonical identity and approval criteria are satisfied."),
    ("source_identity", "evidence_id", "registry", "drill_id", "Use explicit registered occurrence IDs."),
    ("primary_skill", "skill_area", "catalog", "skill_category", "Preserve current labels pending any documented reconciliation."),
    ("equipment", "equipment", "catalog", "equipment", "Retain source-occurrence requirements and unresolved differences."),
    ("participants", "participants", "catalog", "group_requirement", "Keep qualitative arrangements distinct from numeric minima."),
    ("space", "space", "catalog", "", "Inspect recorded requirements; avoid inferring dimensions."),
    ("source_prescription", "source_prescription", "catalog", "duration", "Preserve mixed source prescriptions as text."),
    ("source_url", "source_url", "registry", "", "Preserve registry source locations."),
    ("source_instructions", "instructions", "catalog", "instructions_summary", "Prepare source-linked summaries with explicit pending review."),
    ("primary_goal", "primary_goal", "catalog", "primary_goal", "Keep imported goal wording and evidence provenance."),
    ("secondary_skills", "secondary_skills", "catalog", "", "Leave unknown unless reviewed evidence supports additional labels."),
    ("numeric_minimum_participants", "minimum_participants", "catalog", "", "Do not infer a number from qualitative group labels."),
    ("separate_duration", "duration", "catalog", "", "Separate elapsed time only when the source and units support it."),
    ("separate_rest_interval", "rest", "catalog", "", "Leave unknown where no reviewed rest prescription is available."),
    ("difficulty_modifications", "difficulty_modification", "catalog", "difficulty_modification", "Review modification provenance and retain the base activity separately."),
    ("solo_adaptations", "solo_adaptation", "catalog", "solo_adaptation_note", "Retain imported notes without treating them as approved alternatives."),
    ("adaptation_provenance", "adaptation_provenance", "catalog", "", "Imported file lineage is known; publisher authorship of adaptations is not established."),
]
bb12s_schema = []
for concept, field, table, imported_field, handling in bb12s_specs:
    rows, columns = (bb12s_catalog, bb12s_catalog_columns) if table == "catalog" else (bb12s_registry, bb12s_registry_columns)
    bb12s_schema.append({
        "planned_concept": concept, "table_audited": table, "expected_column": field,
        "column_present": "yes" if field in columns else "no", "table_records": len(rows),
        "current_nonblank_records": sum(bool(row[field].strip()) for row in rows) if field in columns else "",
        "related_imported_field": imported_field,
        "imported_nonblank_records": sum(bool(row[imported_field].strip()) for row in bb12s_imports) if imported_field else "",
        "linked_imported_candidates": len({o["catalog_candidate_id"] for o in bb12s_occurrences
                                           if imported_field and o.get("imported_" + imported_field, "").strip()}) if imported_field else "",
        "handling": handling,
        "evidence_limit": "Presence is not verification; other project sources may contain additional evidence.",
    })
bb12s_gaps = [r for r in bb12s_schema if r["column_present"] == "no"]

# Preserve the existing ledger verbatim; the new worklist only references it.
step12_carried_field_reviews = copy.deepcopy(bb12s_reviews)
bb12s_work = []


def bb12s_task(kind, unit, **values):
    task_id = "S12-" + hashlib.sha256((kind + "|" + unit).encode("utf-8")).hexdigest()[:20]
    row = dict.fromkeys(("task_id", "task_type", "catalog_candidate_ids", "evidence_ids", "concept_or_field",
                        "scope", "observed_state", "current_value", "proposed_value", "source_artifact",
                        "source_record_key", "recorded_decision", "task_status", "next_action"), "")
    row.update(task_id=task_id, task_type=kind, task_status="open", **values)
    bb12s_work.append(row)


for row in bb12s_reviews:
    cid, eid, field = row["candidate_group_id"], row["evidence_id"], row["field"]
    bb12s_task("existing_field_confirmation", "|".join((cid, eid, field)),
                catalog_candidate_ids=cid, evidence_ids=eid, concept_or_field=field,
                scope=bb12s_candidates[cid]["scope_candidate"], observed_state="draft_applied_confirmation_pending",
                current_value=bb12s_candidates[cid][field], proposed_value=row["proposed_value"],
                source_artifact=bb12_review_ref, source_record_key="|".join((cid, eid, field)),
                recorded_decision=row["reviewer_decision"],
                next_action="Review the existing evidence and record the field decision in a controlled continuation of the ledger.")
bb12s_cases = defaultdict(set)
for row in bb12s_catalog:
    for case in bb12_ids(row["source_identity_review_cases"]):
        bb12s_cases[case].add(row["catalog_candidate_id"])
for case, members in sorted(bb12s_cases.items()):
    bb12s_task("existing_identity_case", case, catalog_candidate_ids=";".join(sorted(members)),
                concept_or_field="canonical_identity", scope="case_level", observed_state="identity_case_pending",
                source_artifact=bb12_catalog_ref, source_record_key=case,
                next_action="Compare the linked instructional evidence and record a canonical or variation decision.")
for row in bb12s_gaps:
    bb12s_task("schema_concept", row["planned_concept"], concept_or_field=row["planned_concept"], scope="schema_level",
                observed_state="no_dedicated_column", source_artifact=bb12_catalog_ref,
                source_record_key=row["expected_column"], next_action=row["handling"])

bb12s_core = ("equipment", "participants", "space", "source_prescription")
bb12s_states = ("blank_in_catalog", "recorded_unspecified", "reference_only",
                 "recorded_requirement_uncertain", "recorded_value_present")
bb12s_scope_rows = {"all_candidates": bb12s_catalog}
for scope in ("basketball_drill_candidate", "general_warmup_candidate"):
    bb12s_scope_rows[scope] = [row for row in bb12s_catalog if row["scope_candidate"] == scope]
bb12s_completeness = []
for scope, rows in bb12s_scope_rows.items():
    for field in bb12s_core:
        counts = Counter(bb12s_state(row[field]) for row in rows)
        for state in bb12s_states:
            bb12s_completeness.append({"scope": scope, "field": field, "record_state": state,
                                       "records": counts[state], "scope_records": len(rows)})
for row in bb12s_catalog:
    for field in bb12s_core:
        state = bb12s_state(row[field])
        if state != "recorded_value_present":
            bb12s_task("metadata_evidence", row["catalog_candidate_id"] + "|" + field,
                        catalog_candidate_ids=row["catalog_candidate_id"],
                        evidence_ids=";".join(sorted(bb12_ids(row["evidence_ids"]) | bb12_ids(row["aditi_evidence_ids"]))),
                        concept_or_field=field, scope=row["scope_candidate"], observed_state=state,
                        current_value=row[field], source_artifact=bb12_catalog_ref,
                        source_record_key=row["catalog_candidate_id"] + "|" + field,
                        next_action="Inspect the occurrence evidence; record a supported requirement or explicit unresolved state.")
for row in bb12s_occurrences:
    bb12s_task("imported_annotation_review", row["evidence_id"],
                catalog_candidate_ids=row["catalog_candidate_id"], evidence_ids=row["evidence_id"],
                concept_or_field="instructions;goal;modification;solo_adaptation",
                scope=bb12s_candidates[row["catalog_candidate_id"]]["scope_candidate"],
                observed_state="imported_annotations_pending_review", source_artifact=bb12s_import_ref,
                source_record_key=row["imported_drill_id"],
                next_action="Review accuracy, authorship, intended-use permissions, and any discrepancy with reconciled source evidence.")

bb12s_gate_fields = ("metadata_status", "heading_screen", "section_screen", "provenance_review",
                     "requirements_review", "distinctness_status", "canonical_linkage_review_status",
                     "permission_status", "reproduction_permission", "youth_planning_status", "catalog_approval_status",
                     "planner_eligible", "wave2_user_decision_status", "wave2_deferred_fields",
                     "wave3_metadata_decision_status", "wave3_requirement_draft_status")
bb12s_check("review_dimension_columns", set(bb12s_gate_fields).issubset(bb12s_catalog_columns),
             "The existing review-state schema changed.")
bb12s_candidate_states = [{"catalog_candidate_id": row["catalog_candidate_id"], "drill_name": row["drill_name"],
                           "scope": row["scope_candidate"],
                           "linked_imported_occurrences": sum(o["catalog_candidate_id"] == row["catalog_candidate_id"]
                                                              for o in bb12s_occurrences),
                           "open_worklist_tasks": sum(row["catalog_candidate_id"] in bb12_ids(t["catalog_candidate_ids"])
                                                      for t in bb12s_work),
                           **{field: row[field] for field in bb12s_gate_fields}}
                          for row in bb12s_catalog]

# Verify that schema assessment has not silently changed existing decisions.
bb12s_check("carried_ledger_preserved", step12_carried_field_reviews == bb12s_reviews,
             "The carried-forward field ledger changed during assessment.")
bb12s_check("unique_tasks", len({row["task_id"] for row in bb12s_work}) == len(bb12s_work),
             "Review worklist identifiers must be unique.")
bb12s_counts = Counter(row["task_type"] for row in bb12s_work)
bb12s_check("pending_fields_preserved", bb12s_counts["existing_field_confirmation"]
             == step12_report["metrics"]["human_field_confirmations_pending"], "Pending field confirmations were lost or duplicated.")
bb12s_check("identity_cases_preserved", bb12s_counts["existing_identity_case"]
             == step12_report["metrics"]["explicit_identity_cases_pending"], "Identity cases were lost or duplicated.")
for scope, rows in bb12s_scope_rows.items():
    observed = {field: sum(not row[field].strip() for row in rows) for field in bb12s_core}
    bb12s_check("blank_counts:" + scope, observed == bb12_manifests["12-65"]["blank_core_metadata_by_scope"][scope],
                 "Recomputed blank counts differ from the verified completeness checkpoint: " + scope)
for bb12s_ref, bb12s_digest in bb12s_before.items():
    bb12_read(bb12s_ref, bb12s_digest)


def bb12s_table(rows):
    bb12_require(bool(rows), "An assessment report unexpectedly contains no rows.")
    return {"columns": list(rows[0]), "rows": rows}


step12_tables = {
    "basketball_schema_assessment.csv": bb12s_table(bb12s_schema),
    "basketball_imported_field_mapping.csv": bb12s_table(bb12s_mappings),
    "basketball_imported_occurrence_metadata.csv": bb12s_table(bb12s_occurrences),
    "basketball_preserved_field_reviews.csv": bb12s_table(step12_carried_field_reviews),
    "basketball_review_worklist.csv": bb12s_table(bb12s_work),
    "basketball_candidate_review_states.csv": bb12s_table(bb12s_candidate_states),
    "basketball_metadata_states.csv": bb12s_table(bb12s_completeness),
}
step12_report.update(status="schema_review_prepared", schema_review_complete=True,
                     schema_metrics={"planned_concepts": len(bb12s_schema), "missing_dedicated_columns": len(bb12s_gaps),
                                     "imported_fields_mapped": len(bb12s_mappings), "imported_occurrences": len(bb12s_occurrences),
                                     "imported_candidates": len(bb12s_imported_candidates),
                                     "worklist_tasks": len(bb12s_work), "worklist_task_counts": dict(bb12s_counts)},
                     schema_verified_checks=bb12s_checks, assessment_tables_prepared=len(step12_tables),
                     catalog_rows_changed=0, review_decisions_created=0, project_files_written=0)
print("Step 12-2 — schema assessment and review worklist prepared")
print(f"Planned concepts assessed / missing dedicated columns: {len(bb12s_schema)} / {len(bb12s_gaps)}")
print("Missing concepts: " + ", ".join(row["planned_concept"] for row in bb12s_gaps))
print(f"Imported fields mapped: {len(bb12s_mappings)} | Source occurrences / candidate groups: {len(bb12s_occurrences)} / {len(bb12s_imported_candidates)}")
print("Review worklist tasks by type:")
for kind, count in sorted(bb12s_counts.items()):
    print(f"  {kind}: {count}")
print(f"Candidate review-state rows / prepared evidence tables: {len(bb12s_candidate_states)} / {len(step12_tables)}")
print("Basketball-only blank fields:", {field: sum(not row[field].strip() for row in bb12s_scope_rows["basketball_drill_candidate"])
                                             for field in bb12s_core})
print(f"Assessment checks passed: {len(bb12s_checks)} | Existing field ledger preserved exactly")
print("Catalog rows changed / review decisions created / project files written: 0 / 0 / 0")
print("All inspected source bytes unchanged. Reports are prepared in memory and have not yet been saved.")
print("Task counts overlap across candidates; they do not represent distinct drills or verified content defects.")
print("Next: run Step 12-3 to save and verify this assessment bundle.")
```

```python
# Step 12-3 — Save and verify our schema-assessment reports
# Each distinct bundle has its own directory; existing catalog/review files stay intact.
import os

if not globals().get("step12_report", {}).get("schema_review_complete"):
    raise RuntimeError("Complete the revised Step 12-2 before saving its reports.")
step12_report.update(status="saving_schema_assessment", schema_reports_saved=False)
bb12p_expected = {
    "basketball_schema_assessment.csv", "basketball_imported_field_mapping.csv",
    "basketball_imported_occurrence_metadata.csv", "basketball_preserved_field_reviews.csv",
    "basketball_review_worklist.csv", "basketball_candidate_review_states.csv", "basketball_metadata_states.csv",
}
bb12_require(set(step12_tables) == bb12p_expected, "The assessment report collection is incomplete or unexpected.")


def bb12p_csv(table):
    buffer = io.StringIO(newline="")
    writer = csv.DictWriter(buffer, fieldnames=table["columns"], extrasaction="raise", lineterminator="\n")
    writer.writeheader()
    writer.writerows(table["rows"])
    return buffer.getvalue().encode("utf-8")


def bb12p_json(value):
    return (json.dumps(value, ensure_ascii=False, indent=2, sort_keys=True, allow_nan=False) + "\n").encode("utf-8")


def bb12p_verify_inputs():
    for reference, digest in bb12s_before.items():
        bb12_read(reference, digest)


bb12p_verify_inputs()
bb12_require(step12_tables["basketball_preserved_field_reviews.csv"]["rows"] == bb12s_reviews,
             "The carried-forward field ledger changed in memory after Step 12-2.")
bb12p_tables = {name: bb12p_csv(table) for name, table in step12_tables.items()}
bb12p_summary = {
    "schema_version": "step12-schema-assessment-1", "recipe_version": "schema-assessment-0.1.0",
    "status": "prepared_with_reviews", "baseline_catalog_version": "0.3.9",
    "scope": {"catalog_rows_changed": 0, "review_decisions_created": 0, "canonical_merges": 0,
              "publisher_sources_refreshed": False, "new_planner_eligible_records": 0,
              "full_step12_approval_objective_complete": False,
              "worklist_scope": "Schema concepts, carried-forward field/identity cases, nonspecific core metadata, and imported annotation reviews.",
              "remaining_approval_dimensions": "Candidate review-state table preserves additional provenance, requirements, permission, distinctness, youth-rule, and approval work."},
    "baseline_metrics": step12_report["metrics"], "schema_metrics": step12_report["schema_metrics"],
    "input_hashes": dict(sorted(bb12s_before.items())),
    "verified_checks": bb12s_checks,
    "methods": {
        "identity_join": "Imported drill_id maps to its saved ADI- evidence_id; candidate membership comes from the verified source registry. No name join.",
        "annotations": "Raw imported wording is retained per occurrence. Accuracy, adaptation authorship, and intended-use permissions remain under review.",
        "schema": "A missing dedicated column does not establish absence of evidence from other sources. Numeric durations and participant minima are not inferred.",
        "worklist": "Stable task IDs identify type and review unit. Task counts overlap across candidates and are not counts of distinct drills or verified errors.",
        "existing_decisions": "The 68-row field ledger is copied verbatim; prior candidate review-state fields are carried forward without new decisions.",
    },
    "artifacts": {name: {"sha256": hashlib.sha256(content).hexdigest(), "bytes": len(content),
                         "rows": len(step12_tables[name]["rows"]), "columns": step12_tables[name]["columns"]}
                  for name, content in sorted(bb12p_tables.items())},
}
# The directory is determined by report contents and inputs, so identical reruns
# reuse the same files. No previous report is replaced to resolve a difference.
bb12p_summary_bytes = bb12p_json(bb12p_summary)
bb12p_bundle_digest = hashlib.sha256(bb12p_summary_bytes).hexdigest()
bb12p_relative = "results/basketball_catalog_review/schema_assessment_" + bb12p_bundle_digest[:16]
bb12p_folder = bb12_path(bb12p_relative)
bb12p_payloads = {bb12p_folder / name: content for name, content in bb12p_tables.items()}
bb12p_producer = bb12p_folder / "step12_schema_assessment_summary.json"
bb12p_payloads[bb12p_producer] = bb12p_summary_bytes
bb12_require(not set(bb12p_payloads).intersection({bb12_path(r) for r in bb12s_before}),
             "A report destination overlaps a verified input.")
for path, content in bb12p_payloads.items():
    bb12_require(not path.exists() or (path.is_file() and path.read_bytes() == content),
                 "A different report already occupies this destination: " + path.relative_to(bb12_root).as_posix())

bb12p_created = []
bb12p_folder.mkdir(parents=True, exist_ok=True)
try:
    # Exclusive creation prevents an intervening user edit from being overwritten.
    # The producer record is last, after all table files have been written.
    for path, content in bb12p_payloads.items():
        if path.exists():
            bb12_require(path.is_file() and path.read_bytes() == content, "An existing bundle artifact changed during saving.")
            continue
        try:
            with path.open("xb") as handle:
                bb12p_created.append(path)
                handle.write(content)
                handle.flush()
                os.fsync(handle.fileno())
        except FileExistsError:
            bb12_require(path.is_file() and path.read_bytes() == content, "A bundle destination appeared with different content.")
    for path, content in bb12p_payloads.items():
        bb12_require(path.is_file() and path.read_bytes() == content,
                     "Saved assessment report failed verification: " + path.name)
    bb12p_verify_inputs()
except Exception:
    # Remove only files this attempt created, provided their bytes still match.
    # Preserve any file that another process edited while this cell ran.
    bb12p_retained = []
    for path in reversed(bb12p_created):
        if path.is_file():
            if path.read_bytes() == bb12p_payloads[path]:
                path.unlink()
            else:
                bb12p_retained.append(path.relative_to(bb12_root).as_posix())
    step12_report.update(status="save_failed", schema_reports_saved=False,
                         save_attempt_files_retained=bb12p_retained)
    raise

step12_report.update(status="schema_assessment_saved_with_reviews", schema_reports_saved=True,
                     schema_bundle_directory=bb12p_relative, schema_bundle_digest=bb12p_bundle_digest,
                     schema_producer_record=bb12p_producer.relative_to(bb12_root).as_posix(),
                     schema_report_files_created=len(bb12p_created),
                     project_files_written=len(bb12p_created), source_input_files_changed=0)
print("Step 12-3 — schema-assessment bundle saved and verified")
print("Bundle directory:", bb12p_relative)
print(f"Evidence tables / producer records: {len(bb12p_tables)} / 1")
print(f"Files created / identical files reused: {len(bb12p_created)} / {len(bb12p_payloads) - len(bb12p_created)}")
print("Producer record:", step12_report["schema_producer_record"])
print(f"Verified upstream files unchanged: {len(bb12s_before)}")
print("Catalog rows changed / review decisions created / new planner-eligible records: 0 / 0 / 0")
print("Full Step 12 approval objective remains open. Reports do not approve imported annotations or resolve identity cases.")
print("Next: review the Step 12-2 and Step 12-3 outputs before preparing the source-linked catalog schema.")
```

```python
# Step 12-4 — Prepare the source-linked candidate and occurrence schemas
# Original catalog fields remain intact. Imported annotations stay occurrence-level.
if not globals().get("step12_report", {}).get("schema_reports_saved") or not step12_report.get("inputs_verified"):
    raise RuntimeError("Complete the revised Step 12-3 before preparing the linked schema.")
step12_report.update(status="preparing_linked_schema", linked_schema_prepared=False)
step12_linked_tables = {}
bb12l_inputs = dict(step12_report["input_hashes"])
bb12l_producer_ref = step12_report["schema_producer_record"]
bb12_read(bb12l_producer_ref, step12_report["schema_bundle_digest"])
bb12l_assessment = bb12_json(bb12l_producer_ref)
bb12_require(bb12l_assessment.get("schema_version") == "step12-schema-assessment-1"
             and bb12l_assessment.get("input_hashes") == bb12l_inputs,
             "The saved assessment does not match our verified input baseline.")
bb12l_inputs[bb12l_producer_ref] = step12_report["schema_bundle_digest"]
bb12l_assessment_dir = Path(bb12l_producer_ref).parent.as_posix()
bb12l_saved = {}
for name, metadata in bb12l_assessment["artifacts"].items():
    bb12_require(Path(name).name == name, "Assessment artifact names must be simple filenames.")
    reference = bb12l_assessment_dir + "/" + name
    bb12_read(reference, metadata["sha256"])
    rows, columns = bb12_csv(reference)
    bb12_require(len(rows) == metadata["rows"] and columns == metadata["columns"],
                 "The saved assessment dimensions differ: " + name)
    bb12l_inputs[reference] = metadata["sha256"]
    bb12l_saved[name] = rows
for reference, digest in bb12l_inputs.items():
    bb12_read(reference, digest)

bb12l_catalog, bb12l_catalog_columns = bb12_csv(bb12_catalog_ref)
bb12l_registry, bb12l_registry_columns = bb12_csv(bb12_registry_ref)
bb12l_candidates = bb12_index(bb12l_catalog, "catalog_candidate_id", "catalog")
bb12l_original_evidence = bb12_index(bb12l_registry, "evidence_id", "source registry")
bb12l_imported = bb12l_saved["basketball_imported_occurrence_metadata.csv"]
bb12l_annotations = bb12_index(bb12l_imported, "evidence_id", "imported occurrence report")
bb12l_worklist = bb12l_saved["basketball_review_worklist.csv"]
bb12l_field_reviews = bb12l_saved["basketball_preserved_field_reviews.csv"]
bb12l_version = "basketball-source-linked-schema-1"
bb12l_specs = [
    ("instructions", "source_instructions", "text", "instructions_summary", "Imported summary; instructional accuracy and permitted use need review."),
    ("primary_goal", "primary_goal", "text", "primary_goal", "Imported goal; retain occurrence identity until reviewed adoption."),
    ("secondary_skills", "secondary_skills", "text", "", "Unknown unless additional reviewed labels are established."),
    ("minimum_participants", "numeric_minimum_participants", "nullable_integer", "", "Qualitative arrangements do not establish a numeric minimum."),
    ("duration", "separate_duration", "nullable_number", "", "Elapsed duration needs source support and an explicit duration_unit."),
    ("rest", "separate_rest_interval", "nullable_number", "", "Rest needs source support and an explicit rest_unit."),
    ("difficulty_modification", "difficulty_modifications", "text", "difficulty_modification", "Imported modification; authorship and applicability remain under review."),
    ("solo_adaptation", "solo_adaptations", "text", "solo_adaptation_note", "Imported alternative; provenance and practical feasibility remain under review."),
    ("adaptation_provenance", "adaptation_provenance", "text", "", "Record imported lineage while leaving publisher authorship unverified."),
]
bb12l_fields = [row[0] for row in bb12l_specs]
bb12_require(not set(bb12l_fields).intersection(bb12l_catalog_columns)
             and not set(bb12l_fields).intersection(bb12l_registry_columns),
             "A planned extension field already exists; inspect the schema before replacing it.")
bb12l_checks = []


def bb12l_check(name, condition, detail):
    bb12l_checks.append({"check_name": name, "status": "pass" if condition else "failed", "details": detail})
    bb12_require(condition, detail)


def bb12l_json_text(value):
    return json.dumps(value, ensure_ascii=False, sort_keys=True, separators=(",", ":"))


# Candidate fields are reserved for reviewed canonical values. No source-specific
# annotation is silently promoted into those fields or combined across occurrences.
bb12l_candidate_rows = []
for original in bb12l_catalog:
    cid = original["catalog_candidate_id"]
    linked = sorted(eid for eid, row in bb12l_original_evidence.items() if row["catalog_candidate_id"] == cid)
    imported_ids = [eid for eid in linked if eid in bb12l_annotations]
    row = dict(original)
    row.update({field: "" for field in bb12l_fields})
    row.update(schema_version=bb12l_version, schema_preparation_status="extended_review_only",
               duration_unit="", rest_unit="", linked_source_evidence_ids=";".join(linked),
               source_annotation_evidence_ids=";".join(imported_ids), source_annotation_occurrences=str(len(imported_ids)),
               extended_field_review_status=bb12l_json_text({f: "not_populated_pending_canonical_review" for f in bb12l_fields}))
    bb12l_candidate_rows.append(row)

bb12l_occurrence_rows, bb12l_lineage = [], []
for original in bb12l_registry:
    eid, cid = original["evidence_id"], original["catalog_candidate_id"]
    annotation = bb12l_annotations.get(eid)
    row = dict(original)
    row.update({field: "" for field in bb12l_fields})
    statuses = {field: "not_established_by_schema_preparation" for field in bb12l_fields}
    if annotation is not None:
        bb12_require(annotation["catalog_candidate_id"] == cid,
                     "An imported annotation disagrees with registered membership: " + eid)
        for field, concept, datatype, imported_field, handling in bb12l_specs:
            if imported_field:
                row[field] = annotation["imported_" + imported_field]
                statuses[field] = "imported_annotation_pending_review" if row[field].strip() else "missing_imported_value"
        row["adaptation_provenance"] = "Imported dataset lineage recorded; publisher authorship of adaptations remains unverified."
        statuses["adaptation_provenance"] = "import_lineage_recorded_authorship_unverified"
    row.update(schema_version=bb12l_version, candidate_scope=bb12l_candidates[cid]["scope_candidate"],
               duration_unit="", rest_unit="", planner_eligible="no",
               imported_annotation_available="yes" if annotation else "no",
               annotation_input_artifact=annotation["input_artifact"] if annotation else "",
               annotation_input_sha256=annotation["input_sha256"] if annotation else "",
               imported_drill_id=annotation["imported_drill_id"] if annotation else "",
               imported_original_drill_name=annotation["imported_drill_name"] if annotation else "",
               imported_original_prescription=annotation["imported_duration"] if annotation else "",
               imported_solo_adaptation_possible=annotation["imported_solo_adaptation_possible"] if annotation else "",
               extended_field_review_status=bb12l_json_text(statuses))
    bb12l_occurrence_rows.append(row)
    for field, concept, datatype, imported_field, handling in bb12l_specs:
        bb12l_lineage.append({
            "evidence_id": eid, "catalog_candidate_id": cid, "field": field, "planned_concept": concept,
            "value": row[field], "value_state": ("generated_lineage_assertion" if field == "adaptation_provenance" and annotation
                                                  else "recorded_annotation" if row[field].strip() else "unknown_in_this_schema"),
            "review_status": statuses[field],
            "input_artifact": annotation["input_artifact"] if annotation and row[field].strip() else "",
            "input_record_id": annotation["imported_drill_id"] if annotation and row[field].strip() else "",
            "input_field": imported_field if row[field].strip() else "",
            "provenance_note": "Lineage assertion generated from saved input metadata; authorship not inferred."
                               if field == "adaptation_provenance" and annotation else "",
        })

# Group recorded locators, without presenting them as newly verified web sources.
bb12l_locator_groups = defaultdict(list)
for row in bb12l_registry:
    key = tuple(row[f] for f in ("source_url", "source_document", "source_page", "source_level", "practice_plan"))
    bb12l_locator_groups[key].append(row)
bb12l_source_index = []
for key, members in sorted(bb12l_locator_groups.items()):
    bb12l_source_index.append({
        "source_locator_id": "S12-SOURCE-" + hashlib.sha256(bb12l_json_text(key).encode()).hexdigest()[:20],
        **dict(zip(("source_url", "source_document", "source_page", "curriculum_level", "practice_plan"), key)),
        "evidence_ids": ";".join(sorted(r["evidence_id"] for r in members)),
        "catalog_candidate_ids": ";".join(sorted({r["catalog_candidate_id"] for r in members})),
        "source_occurrences": str(len(members)),
        "publisher_attribution": "Jr. NBA (recorded project source attribution)",
        "retrieval_date": "", "verification_date": "",
        "recorded_locator_statuses": ";".join(sorted({r["source_locator_status"] for r in members})),
        "recorded_permission_statuses": ";".join(sorted({r["permissions_status"] for r in members})),
        "fresh_source_verification_performed": "no", "input_artifact": bb12_registry_ref,
    })
bb12l_dictionary = [{"field": field, "planned_concept": concept, "data_type": datatype,
                     "candidate_value_policy": "Reviewed canonical value only; left unknown by this preparation.",
                     "occurrence_value_policy": handling, "imported_field": imported_field,
                     "unknown_csv_representation": "empty string", "numeric_conversion_performed": "no",
                     "unit_field": "duration_unit" if field == "duration" else "rest_unit" if field == "rest" else "",
                     "review_state_field": "extended_field_review_status", "schema_version": bb12l_version}
                    for field, concept, datatype, imported_field, handling in bb12l_specs]
bb12l_extended_worklist = []
for original in bb12l_worklist:
    row = dict(original)
    row["schema_preparation_action"] = "representation_added_content_review_pending" if row["task_type"] == "schema_concept" else "existing_task_preserved"
    row["implementation_artifacts"] = "basketball_candidate_schema_REVIEW_ONLY.csv;basketball_source_occurrence_schema_REVIEW_ONLY.csv" if row["task_type"] == "schema_concept" else ""
    bb12l_extended_worklist.append(row)
bb12l_coverage = []
for table, rows in (("candidate_schema", bb12l_candidate_rows), ("source_occurrence_schema", bb12l_occurrence_rows),
                    ("imported_source_occurrences", [r for r in bb12l_occurrence_rows if r["imported_annotation_available"] == "yes"])):
    for field in bb12l_fields:
        present = sum(bool(r[field].strip()) for r in rows)
        bb12l_coverage.append({"table_scope": table, "field": field, "records": str(len(rows)),
                               "nonblank_values": str(present), "unknown_values": str(len(rows) - present),
                               "newly_verified_values": "0", "interpretation": "Nonblank occurrence annotations remain pending review."})

# Exact comparisons preserve all original columns, including prior corrections.
bb12l_check("original_catalog_fields", [{k: r[k] for k in bb12l_catalog_columns} for r in bb12l_candidate_rows] == bb12l_catalog,
             "Every original catalog field and row remains unchanged.")
bb12l_check("original_registry_fields", [{k: r[k] for k in bb12l_registry_columns} for r in bb12l_occurrence_rows] == bb12l_registry,
             "Every original occurrence field and row remains unchanged.")
bb12l_check("canonical_fields_unknown", all(not r[f] for r in bb12l_candidate_rows for f in bb12l_fields),
             "Candidate canonical fields remain unknown pending review.")
bb12l_check("unknown_numeric_fields", all(not r[f] for r in bb12l_candidate_rows + bb12l_occurrence_rows
                                         for f in ("minimum_participants", "duration", "rest", "duration_unit", "rest_unit")),
             "No numeric minima, elapsed durations, rest values, or units were inferred.")
bb12l_check("all_planned_fields_present", all(set(bb12l_fields).issubset(r) for r in bb12l_candidate_rows + bb12l_occurrence_rows),
             "All nine planned concepts have dedicated fields in the linked schema.")
bb12l_check("source_membership", {r["evidence_id"]: r["catalog_candidate_id"] for r in bb12l_occurrence_rows}
             == {r["evidence_id"]: r["catalog_candidate_id"] for r in bb12l_registry},
             "All source-to-candidate memberships are preserved.")
bb12l_check("imported_occurrences_preserved", {r["evidence_id"] for r in bb12l_occurrence_rows if r["imported_annotation_available"] == "yes"}
             == set(bb12l_annotations), "Imported occurrences remain separate and complete.")
bb12l_check("annotation_input_lineage", all(r["input_sha256"] == bb12l_inputs.get(r["input_artifact"])
                                          for r in bb12l_imported), "Imported annotation lineage matches the inspected input bytes.")
bb12l_check("unique_field_lineage", len(bb12l_lineage) == len(bb12l_registry) * len(bb12l_fields)
             == len({(r["evidence_id"], r["field"]) for r in bb12l_lineage}),
             "Each planned field has exactly one lineage record per occurrence.")
bb12l_check("unique_source_locators", len({r["source_locator_id"] for r in bb12l_source_index}) == len(bb12l_source_index),
             "Recorded source-locator identifiers are unique.")
bb12l_check("source_index_membership", Counter(eid for row in bb12l_source_index for eid in bb12_ids(row["evidence_ids"]))
             == Counter(bb12l_original_evidence.keys()), "The source index includes each occurrence exactly once.")
bb12l_check("worklist_preserved", [{k: r[k] for k in original} for r, original in zip(bb12l_extended_worklist, bb12l_worklist)]
             == bb12l_worklist and len(bb12l_extended_worklist) == len(bb12l_worklist),
             "All original worklist fields and task statuses are preserved.")
bb12l_check("field_reviews_preserved", bb12_csv(bb12_review_ref)[0] == bb12l_field_reviews,
             "The saved field ledger remains exactly as inspected.")
bb12l_check("no_new_eligibility", all(r["planner_eligible"] == "no" for r in bb12l_candidate_rows + bb12l_occurrence_rows),
             "No candidate or source occurrence becomes planner eligible.")
bb12l_check("no_new_approval", all(r["catalog_approval_status"] == "pending" and not r["approved_drill_id"] for r in bb12l_candidate_rows),
             "Catalog approvals and approved identifiers remain unchanged.")
for reference, digest in bb12l_inputs.items():
    bb12_read(reference, digest)
step12_linked_tables = {
    "basketball_candidate_schema_REVIEW_ONLY.csv": bb12s_table(bb12l_candidate_rows),
    "basketball_source_occurrence_schema_REVIEW_ONLY.csv": bb12s_table(bb12l_occurrence_rows),
    "basketball_schema_field_dictionary.csv": bb12s_table(bb12l_dictionary),
    "basketball_extended_field_lineage.csv": bb12s_table(bb12l_lineage),
    "basketball_recorded_source_index.csv": bb12s_table(bb12l_source_index),
    "basketball_schema_coverage.csv": bb12s_table(bb12l_coverage),
    "basketball_review_worklist_with_schema_status.csv": bb12s_table(bb12l_extended_worklist),
    "basketball_linked_schema_checks.csv": bb12s_table(bb12l_checks),
}
bb12l_metrics = {
    "catalog_candidates": len(bb12l_candidate_rows), "source_occurrences": len(bb12l_occurrence_rows),
    "original_catalog_columns_preserved": len(bb12l_catalog_columns), "original_registry_columns_preserved": len(bb12l_registry_columns),
    "planned_concepts_represented": len(bb12l_fields), "imported_annotation_occurrences": len(bb12l_annotations),
    "candidates_with_imported_annotations": len({r["catalog_candidate_id"] for r in bb12l_imported}),
    "recorded_source_locators": len(bb12l_source_index), "field_lineage_rows": len(bb12l_lineage),
    "open_worklist_tasks": len(bb12l_worklist), "checks_passed": len(bb12l_checks),
    "canonical_values_newly_verified": 0, "approvals_created": 0, "planner_eligible": 0,
}
step12_report.update(status="linked_schema_prepared", linked_schema_prepared=True, linked_schema_metrics=bb12l_metrics,
                     linked_schema_input_hashes=bb12l_inputs, linked_schema_reports_saved=False,
                     linked_schema_table_fingerprints={name: hashlib.sha256(bb12p_csv(table)).hexdigest()
                                                       for name, table in step12_linked_tables.items()})
print("Step 12-4 — source-linked schema prepared")
print(f"Candidate rows / source occurrences: {len(bb12l_candidate_rows)} / {len(bb12l_occurrence_rows)}")
print(f"Original catalog / registry columns preserved exactly: {len(bb12l_catalog_columns)} / {len(bb12l_registry_columns)}")
print(f"New planned concepts represented: {len(bb12l_fields)} | Reviewed canonical values added: 0")
print(f"Imported annotations: {len(bb12l_annotations)} occurrences across {bb12l_metrics['candidates_with_imported_annotations']} candidates; all pending review")
print(f"Recorded source locators / field-lineage rows: {len(bb12l_source_index)} / {len(bb12l_lineage)}")
print(f"Open worklist tasks preserved: {len(bb12l_worklist)} | Original field ledger unchanged")
print(f"Schema checks passed / prepared tables: {len(bb12l_checks)} / {len(step12_linked_tables)}")
print("Numeric durations, rest, and participant minima inferred: 0")
print("Catalog approvals / planner eligible: 0 / 0 | Project files written: 0")
print("Next: run Step 12-5 to save, verify, and prepare the Step 13 handoff.")
```

## 13. Youth Participation Guidelines and Planning Rules

### 13.1 Purpose and Scope

We will prepare a reviewed, versioned rule package for evaluating weekly basketball plans for athletes ages 13–17. The rules will help us identify scheduling conflicts, missing information, activity requirements, and departures from the selected participation guidance.

We will define the evidence, applicability, expected behavior, and evaluation treatment for each adopted check. These records will support later planner and validator implementation. Passing the checks will describe adherence to the documented conditions; it will not establish medical clearance or individual training suitability.

### 13.2 Starting Inputs and Connection to Earlier Steps

We will review the prepared recovery references and structured rule records carried forward from earlier steps. Our starting inventory includes five guidance records and seventeen structured participation and recovery records.

| Input | Role in this step |
|---|---|
| `data/processed/recovery_guidance.csv` | Prepared guidance summaries and source references |
| `data/references/recovery/youth_basketball_recovery_rules_v0_1.csv` | Existing structured rule records and their recorded verification labels |
| Source registration and quality records | File identity, provenance, and earlier inspection findings |
| Step 12 source-linked schema and handoff | Activity requirements, unknown values, and unresolved catalog reviews |
| Step 2 evaluation plan | Common outcomes, required checks, and evaluation controls |

We will confirm the current files and producer records before using them. We will preserve prior verification labels and record new source, applicability, and adoption decisions separately. Existing records will enter our review process rather than automatically becoming active checks.

The Step 12 catalog remains review-only, with no planner-eligible drills. We will carry that state forward. Rule preparation can proceed while catalog reviews remain open; runtime activation and drill eligibility will require their own recorded decisions.

### 13.3 Distinguish Rule Origins and Enforcement Decisions

We will identify three origins for our checks:

| Rule origin | What it represents | Example |
|---|---|---|
| Published guidance | A statement supported by an identified external reference | Participation or rest guidance |
| User constraint | A requirement supplied for the athlete | Available days, equipment, or existing practices |
| Project policy | An application or evaluation decision defined by our project | Request clarification when an essential input is missing |

We will also distinguish the published statement type from our enforcement decision. A source may describe a recommendation, maximum, minimum, range, or qualitative caution. We will separately decide whether the corresponding application check is required, advisory, or review-only.

A recommendation will therefore retain its source meaning even when we adopt a particular application response. We will document the reason for that response and avoid presenting our implementation choice as a publisher requirement.

### 13.4 Review Participation Guidance and Applicability

Our primary basketball participation reference will be the NBA and USA Basketball *Youth Basketball Guidelines* [1]. Its participation and rest tables include:

| Source category | Maximum organized basketball per week | Minimum rest days per week |
|---|---:|---:|
| Ages 12–14 | 10 hours | 1 |
| Grades 9–12 | 14 hours | 1 |

We will retain the source's distinction between recommended participation and maximum limits. Maximums will remain upper limits rather than training targets. We will preserve its qualifications concerning multiple games, concurrent sports, camps, and recovery [1].

The source distinguishes organized games, practices, and focused structured training from genuinely unstructured individual or peer-led play [1]. We will document how app-assigned basketball sessions are classified as a project interpretation, with uncertain classifications flagged for review.

We will collect both age and school grade, using an identified planning date. We will record the conditions and logic for each source segment. Where categories overlap or the available information does not resolve applicability, we will retain an Unknown result and a review question rather than automatically choosing the more permissive limit.

The two-row table is a summary, not our complete rule inventory. We will also review the candidate rule families below and their supporting sources:

| Rule family | Information needed for review or evaluation |
|---|---|
| Weekly participation | Classified activities and their durations across the planning window |
| Game and practice recommendations | Event type, frequency, session length, and applicable source segment |
| Weekly rest | Days containing organized basketball and the relevant definition of rest |
| Annual participation | Reported participation history and the applicable annual measurement period |
| Sleep | Relevant source segment and reported sleep information |
| Qualified situations | Tournament, camp, other-sport, and recovery context |

We will preserve qualitative guidance as explanations or review flags where the source does not establish a numerical comparison. We will check additional references recorded in our inputs before adopting rules based on them.

### 13.5 Prepare the Structured Rule Register

We will organize the reviewed material into records with explicit evidence and implementation fields:

| Information group | What we will record |
|---|---|
| Identity and origin | Stable rule ID, version, category, description, and rule origin |
| Applicability | Source segment, age or grade conditions, condition logic, required inputs, and activity scope |
| Measurement | Metric, comparison operator, threshold or range, unit, aggregation method, and time window |
| Source meaning | Recommendation, maximum, minimum, range, or qualitative caution |
| Qualifications | Definitions, footnotes, exceptions, and unresolved interpretations |
| Evidence | Source ID, URL, section or table location, verification date, and available integrity information |
| Application decision | Required, advisory, or review-only status; intended response; and rationale |
| Review and use states | Preparation, reviewer decision, adoption status, and activation status |
| Evaluation contract | Expected outcomes, Unknown and Not applicable handling, and test references |

We will distinguish a draft record, an accepted interpretation, and an enabled runtime check. Source verification alone will not activate a rule. We will preserve the history of rejected, deferred, revised, and superseded records.

We will link candidate records and draft tests to their supporting evidence and record the required review decisions before adopting a rule.

### 13.6 Evaluate the Complete Schedule and Activity Requirements

We will account for existing commitments together with proposed sessions. Each scheduled activity will have an identifier, calendar date, activity type, planned duration, and participation classification. Where overlap checks require start and end times, we will request or record those values and the relevant time zone.

We will define an explicit seven-day planning window and document the scope of each check. A weekly result will remain separate from claims about annual participation or longer-term workload.

We will distinguish three kinds of timing information:

| Timing information | How we will use it |
|---|---|
| Source prescription | Retain published repetitions, rounds, intervals, or other task instructions |
| Planner-assigned session duration | Record the proposed time budget separately from source-prescribed elapsed time |
| Reported existing commitment | Include its supplied duration in the relevant schedule totals |

Step 12 left separate source durations, rest intervals, and numeric participant minima unknown. We will preserve those unknowns. We will validate supplied durations and units, and we will not turn missing or invalid values into zero.

We will avoid counting a session and its component drills twice. Activity identifiers and parent–component relationships will identify which durations contribute to each total.

Our planned checks will examine available days and time, overlapping commitments, equipment, space, participants, existing practices and games, and applicable participation and rest conditions. Other sports and conditioning will remain visible with their own classifications.

If known commitments already establish a conflict, we will report it even when other information is incomplete. We will preserve existing commitments and explain the limitation before proposing additional activity.

### 13.7 Represent Check Outcomes and Missing Information

Each check will return one of four outcomes:

| Outcome | Meaning |
|---|---|
| Met | Available evidence supports the applicable condition |
| Violated | Available evidence establishes a conflict with the applicable condition |
| Unknown | Necessary information, applicability, or interpretation remains insufficient |
| Not applicable | Available evidence establishes that the rule does not apply |

Each result will identify the rule version, relevant inputs, explanation, and unresolved information. We will distinguish missing profile or history inputs, missing planner output, uncertain activity classification, and unresolved rule interpretation.

Incomplete information will not erase an established violation. For example, known activities already exceeding an adopted maximum can establish a conflict even if an additional session has an unknown duration. Where the available evidence cannot establish the result, the outcome will remain Unknown.

We will require a reason for Not applicable. Missing age, grade, duration, or other required information will not by itself establish inapplicability.

Annual participation and actual sleep cannot be established from a proposed weekly schedule alone. We will use relevant reported history where available and otherwise retain Unknown outcomes or clearly identified advisory reminders.

### 13.8 Handle Requests Outside the Project Scope

Our prototype will cover basketball planning, general conditioning, recovery considerations, and optional general food facts within the scope established in Step 2.

As project policies, we will exclude injury rehabilitation programs, supplement recommendations, and individualized dietary prescriptions. Requests involving pain, injury, or medical restrictions will receive appropriate clarification, a scope explanation, or a recommendation to obtain qualified professional guidance.

We will preserve the distinction between an unknown injury state, a reported absence of injury, and a documented restriction. We will not interpret a reported absence of injury as medical clearance.

### 13.9 Review the Rules and Define Test Cases

One teammate will prepare each proposed rule and another will review its source, applicability, interpretation, enforcement decision, and expected behavior. We will record both roles, the review date, the outcome, and the rationale. Unresolved disagreements will remain visible until a decision is recorded.

We will define expected test outcomes before assessing an implementation. The test specifications will cover:

- Values below, equal to, and above numerical boundaries.
- Age and grade boundaries, overlapping segments, and missing or conflicting information.
- Unit conversions, incomplete durations, and invalid numerical input.
- Existing commitments that already exceed a limit or leave no room for additional sessions.
- Weekly rest, schedule overlap, and planning-window boundaries.
- Misclassified activities and duplicate counting of sessions and component drills.
- Annual history, reported sleep, and other information unavailable from a weekly plan alone.
- Camps, tournaments, and qualified situations requiring interpretation.
- Requests outside scope and user edits made after plan generation.

We will distinguish review of the rule and test specifications from execution of the later validator. Generated tests will not be described as independently reviewed until the required review is recorded.

### 13.10 Apply Common Evaluation Rules to All Four Planner Versions

We will freeze the adopted rule versions, case-level required and applicable checks, expected clarification behavior, and scoring treatment before final evaluation. Applicability will be determined from the evaluation case and the rule contract, rather than from a planner omitting information.

For the feasible cases defined in Step 2, automated success will require a usable response and success on every predefined required applicable check. An Unknown result on such a check, including one caused by missing required planner output, will count as unsuccessful. The primary denominator will remain all 150 feasible profiles per version.

We will distinguish this from shared missing profile information. Essential missing inputs will have a predefined clarification response and challenge-category treatment. History-dependent or advisory checks that lack optional inputs will remain reported as Unknown under their frozen policy; they will not silently become passes or newly introduced mandatory failures.

We will report Met, Violated, Unknown, and Not applicable counts with check coverage and stated denominators. Omitting required plan information will not improve a version's score by removing its checks from evaluation.

The same independent evaluation contract will assess all four planner versions. Version 4 may use checks during generation and bounded revision, while the final research scores will come from the common evaluation process. We will preserve human ratings and automated results separately, following Step 2.

We will keep evaluation profiles, rule-test answer keys, and expected final answers outside the retrieval knowledge base.

### 13.11 Expected Outputs and Version Preservation

We will preserve existing input files and review history. Before writing, we will inspect current destinations and use versioned outputs or controlled replacement with prior-file preservation.

| Planned artifact | Contents |
|---|---|
| Versioned planning-rule package | Rules, source references, applicability, enforcement decisions, and review/use states |
| `data/processed/planning_rules.json` | A controlled current-package export when an adopted version is ready |

| `results/planning_rule_review.csv` | Preparation and review decisions, revisions, deferrals, and unresolved questions |
| Applicability and issue report | Required inputs, ambiguous cases, and remaining interpretation work |
| Versioned test specifications | Input cases and independently reviewed expected outcomes |
| Step 13 producer record and handoff | Input/output identities, checksums, verified checks, adopted versions, limitations, and next actions |

We will document later corrections and identify affected tests or evaluation results. Final evaluation will use a frozen rule package and scoring contract.

### 13.12 Completion Criteria and Remaining Dependencies

We will consider the Step 13 preparation checkpoint complete when:

1. Input identity and the relationship to earlier source records have been checked.
2. The candidate rule inventory, source evidence, and applicability conditions are documented.
3. Source guidance, user constraints, and project policies remain distinguishable.
4. Review and adoption decisions are recorded, with pending items preserved.

If adoption requirements remain unmet, we will report a prepared package with open reviews rather than declare those rules adopted. We will write a separate Observations section after examining the actual outputs.

Completion of this rule-preparation checkpoint will remain separate from runtime activation, the later validator's execution results, and Step 12's target of at least 100 approved distinct drills. We will continue to preserve the catalog's approval and planner-eligibility controls.

```python
# Step 13-1 — Verify our current handoff and inventory planning-rule references
# We retain source statements and prior verification labels; no rules are adopted.
import csv
import hashlib
import io
import json
import os
import re
from collections import Counter
from decimal import Decimal, InvalidOperation
from pathlib import Path, PureWindowsPath
from urllib.parse import urlsplit

# Normally resolved from the completed Step 12-5 in this kernel. If several saved
# bundles exist after a restart, set this to the desired project-relative producer.
STEP12_PRODUCER_REFERENCE = ""
step13_report = {"status": "checking_inputs", "inventory_saved": False}
p13_hashes, p13_expected, p13_checks = {}, {}, []


def p13_require(condition, message):
    if not condition:
        step13_report.update(status="blocked", inventory_saved=False, reason=message)
        raise RuntimeError("Step 13-1: " + message)


def p13_check(name, condition, details):
    p13_checks.append({"check_name": name, "status": "pass" if condition else "failed", "details": details})
    p13_require(condition, details)


p13_hint = globals().get("project_root")
p13_cwd = Path.cwd().resolve()
p13_roots = ([Path(p13_hint).resolve()] if p13_hint is not None else []) + [p13_cwd, *p13_cwd.parents]
p13_root = next((p for p in p13_roots if (p / "data/source_manifest.csv").is_file()
                 and (p / "results").is_dir()), None)
p13_require(p13_root is not None, "Run inside our project with its saved source register and results.")


def p13_path(reference):
    p13_require(isinstance(reference, str) and bool(reference), "Invalid empty artifact path.")
    value = Path(reference.replace("\\", "/"))
    p13_require(not value.is_absolute() and not PureWindowsPath(reference).drive,
                "Expected a project-relative path: " + reference)
    path = (p13_root / value).resolve()
    p13_require(path.is_relative_to(p13_root), "Artifact path leaves our project: " + reference)
    return path


def p13_read(reference, expected=None):
    path = p13_path(reference)
    p13_require(path.is_file(), "Missing local input: " + reference)
    content = path.read_bytes()
    digest = hashlib.sha256(content).hexdigest()
    key = path.relative_to(p13_root).as_posix()
    p13_require(key not in p13_hashes or p13_hashes[key] == digest,
                "An inspected input changed during inventory: " + key)
    if expected is not None:
        p13_require(isinstance(expected, str) and re.fullmatch(r"[0-9a-f]{64}", expected),
                    "Invalid producer checksum: " + key)
        p13_require(key not in p13_expected or p13_expected[key] == expected,
                    "Producer records disagree about: " + key)
        p13_require(digest == expected, "Saved checksum differs: " + key + ". Share this error before rerunning earlier producers.")
        p13_expected[key] = expected
    p13_hashes[key] = digest
    return content


def p13_pairs(pairs):
    result = {}
    for key, value in pairs:
        p13_require(key not in result, "Repeated JSON key: " + key)
        result[key] = value
    return result


def p13_json(reference, expected=None):
    result = json.loads(p13_read(reference, expected).decode("utf-8-sig"), object_pairs_hook=p13_pairs)
    p13_require(isinstance(result, dict), "Expected a JSON object: " + reference)
    return result


def p13_csv(reference, required=()):
    reader = csv.DictReader(io.StringIO(p13_read(reference).decode("utf-8-sig"), newline=""))
    columns = list(reader.fieldnames or [])
    rows = list(reader)
    p13_require(bool(columns) and all(columns) and len(columns) == len(set(columns)) and set(required).issubset(columns),
                "Invalid or incomplete CSV schema: " + reference)
    p13_require(all(None not in row and all(v is not None for v in row.values()) for row in rows),
                "Malformed CSV row: " + reference)
    return rows, columns


def p13_unique(rows, field, label):
    keys = [row[field] for row in rows]
    p13_require(all(isinstance(k, str) and k.strip() for k in keys) and len(keys) == len(set(keys)),
                "Missing or duplicate " + field + " in " + label)
    return dict(zip(keys, rows))


# Bind to the revised source-linked checkpoint, not the historical 66-cell workflow.
p13_state12 = globals().get("step12_report", {})
p13_producer_ref = STEP12_PRODUCER_REFERENCE or (
    p13_state12.get("linked_schema_producer_record", "") if p13_state12.get("linked_schema_reports_saved") else "")
if not p13_producer_ref:
    p13_options = sorted((p13_root / "results/basketball_catalog_review").glob(
        "source_linked_schema_*/step12_source_linked_schema_summary.json"))
    if len(p13_options) != 1:
        print("Available Step 12 producers:", [p.relative_to(p13_root).as_posix() for p in p13_options])
        p13_require(False, "Run Step 12-5, or select one producer using STEP12_PRODUCER_REFERENCE at the top of this cell.")
    p13_producer_ref = p13_options[0].relative_to(p13_root).as_posix()
p13_anchor = p13_state12.get("linked_schema_bundle_digest") if (
    p13_producer_ref == p13_state12.get("linked_schema_producer_record")) else None
p13_producer = p13_json(p13_producer_ref, p13_anchor)
p13_check("current_step12_producer", p13_producer.get("schema_version") == "step12-source-linked-schema-producer-1"
          and p13_producer.get("status") == "integration_checkpoint_passed_with_reviews",
          "The current source-linked Step 12 integration checkpoint is identified.")
p13_check("step12_saved_checks", bool(p13_producer.get("verified_checks")) and all(
    r.get("status") == "pass" for r in p13_producer["verified_checks"]), "The saved linked-schema checks all passed.")
p13_require(isinstance(p13_producer.get("input_hashes"), dict) and bool(p13_producer["input_hashes"])
             and isinstance(p13_producer.get("artifacts"), dict) and bool(p13_producer["artifacts"]),
             "The Step 12 producer is missing its input or output inventory.")
p13_require({"step12_step13_source_linked_handoff.json", "basketball_candidate_schema_REVIEW_ONLY.csv"}
             .issubset(p13_producer["artifacts"]), "The declared Step 12 outputs omit the current handoff or candidate schema.")
for p13_ref, p13_digest in p13_producer["input_hashes"].items():
    p13_read(p13_ref, p13_digest)
p13_folder12 = Path(p13_producer_ref).parent.as_posix()
for p13_name, p13_metadata in p13_producer["artifacts"].items():
    p13_require(Path(p13_name).name == p13_name, "Step 12 output names must be simple filenames.")
    p13_read(p13_folder12 + "/" + p13_name, p13_metadata["sha256"])
p13_handoff_ref = p13_folder12 + "/step12_step13_source_linked_handoff.json"
p13_handoff = p13_json(p13_handoff_ref)
p13_check("current_step12_handoff", p13_handoff.get("schema_version") == "step12-linked-schema-step13-handoff-1"
          and p13_handoff.get("integration_checkpoint_status") == "passed"
          and p13_handoff.get("step13_rule_development_can_proceed") is True,
          "The revised handoff permits rule development.")
p13_candidate_name = p13_handoff["bundle_artifacts"]["candidates"]
p13_require(Path(p13_candidate_name).name == p13_candidate_name and p13_candidate_name in p13_producer["artifacts"],
             "The handoff candidate schema must be a declared artifact within its bundle.")
p13_catalog_ref = p13_folder12 + "/" + p13_candidate_name
p13_catalog, _ = p13_csv(p13_catalog_ref, ("catalog_candidate_id", "catalog_approval_status", "planner_eligible"))
p13_unique(p13_catalog, "catalog_candidate_id", "linked candidate schema")
p13_check("catalog_exclusions", len(p13_catalog) == p13_producer["metrics"]["catalog_candidates"]
          and p13_handoff.get("approved_distinct_drills") == p13_handoff.get("planner_eligible") == 0
          and all(r["catalog_approval_status"].strip().casefold() == "pending"
                  and r["planner_eligible"].strip().casefold() == "no" for r in p13_catalog),
          "The review-only catalog and zero planner eligibility are preserved.")

# Resolve reference files through the inspected register and verify exact bytes.
p13_require("data/source_manifest.csv" in p13_producer["input_hashes"]
             and "results/step11_eda_summary.json" in p13_producer["input_hashes"],
             "The current handoff must include the source register and Step 11 producer.")
p13_manifest, _ = p13_csv("data/source_manifest.csv", ("source_id", "relative_path", "sha256"))
p13_sources = p13_unique(p13_manifest, "source_id", "source register")
p13_refs = {}
for p13_role, p13_sid in {"structured_rules": "recovery_rules_v01", "general_guidance": "aditi_recovery_guidance",
                         "readme": "recovery_rules_readme_v01"}.items():
    p13_require(p13_sid in p13_sources, "A required reference is not registered: " + p13_sid)
    p13_ref = p13_sources[p13_sid]["relative_path"].replace("\\", "/")
    p13_require(p13_ref in p13_producer["input_hashes"], "Reference was not tracked by the current handoff: " + p13_ref)
    p13_read(p13_ref, p13_producer["input_hashes"][p13_ref])
    if p13_sources[p13_sid]["sha256"]:
        p13_read(p13_ref, p13_sources[p13_sid]["sha256"])
    p13_refs[p13_role] = p13_ref
p13_eda = p13_json("results/step11_eda_summary.json")
p13_recovery_metrics = p13_eda["supporting_metrics"]["recovery"]
p13_rules, p13_rule_columns = p13_csv(p13_refs["structured_rules"], (
    "rule_id", "domain", "segment", "age_min", "age_max", "grade_min", "grade_max", "metric", "rule_type",
    "value_min", "value_max", "unit", "source_organization", "source_url", "source_section",
    "applicability_notes", "planner_handling", "verification_status"))
p13_guidance, p13_guidance_columns = p13_csv(p13_refs["general_guidance"], ("topic", "guidance", "source", "source_url"))
p13_check("reference_dimensions", len(p13_rules) == p13_recovery_metrics["rule_records"]
          and len(p13_guidance) == p13_recovery_metrics["guidance_records"],
          "Reference counts agree with the verified Step 11 inventory.")
p13_unique(p13_rules, "rule_id", "structured references")
p13_check("reference_readme", bool(p13_read(p13_refs["readme"]).decode("utf-8-sig").strip()),
          "The registered README is available and nonempty.")


def p13_number(value, label):
    if not value.strip():
        return None
    try:
        number = Decimal(value)
    except InvalidOperation as error:
        raise ValueError("Invalid numeric bound: " + label) from error
    p13_require(number.is_finite() and number >= 0, "Invalid numeric bound: " + label)
    return number


def p13_pair(row, low_name, high_name, integer=False):
    low = p13_number(row[low_name], row["rule_id"] + "/" + low_name)
    high = p13_number(row[high_name], row["rule_id"] + "/" + high_name)
    p13_require(low is None or high is None or low <= high, "Reversed bounds: " + row["rule_id"])
    if integer:
        p13_require((low is None) == (high is None) and all(v == v.to_integral_value() for v in (low, high) if v is not None),
                    "Invalid or incomplete age/grade bounds: " + row["rule_id"])
    return low, high


for p13_row in p13_rules:
    p13_require(all(p13_row[k].strip() for k in ("segment", "metric", "rule_type", "unit", "source_organization",
                                                "source_url", "source_section", "applicability_notes")),
                "Incomplete structured reference: " + p13_row["rule_id"])
    p13_pair(p13_row, "age_min", "age_max", integer=True)
    p13_pair(p13_row, "grade_min", "grade_max", integer=True)
    p13_low, p13_high = p13_pair(p13_row, "value_min", "value_max")
    p13_kind = p13_row["rule_type"]
    p13_require(p13_kind in {"maximum", "minimum", "recommended_range", "recommended_exact", "published_range_9_to_10"},
                "An unfamiliar source rule type needs schema review: " + p13_row["rule_id"])
    p13_require((p13_kind != "maximum" or (p13_low is None and p13_high is not None))
                and (p13_kind != "minimum" or (p13_low is not None and p13_high is None))
                and (p13_kind not in {"recommended_range", "recommended_exact", "published_range_9_to_10"}
                     or (p13_low is not None and p13_high is not None))
                and (p13_kind != "recommended_exact" or p13_low == p13_high),
                "Unexpected numeric bound shape: " + p13_row["rule_id"])
for p13_row in p13_guidance:
    p13_require(all(p13_row[k].strip() for k in ("topic", "guidance", "source", "source_url")), "Incomplete general guidance record.")
p13_check("reference_shapes", True, "Numeric bounds, required reference fields, and intentional one-sided values pass structural checks.")
p13_urls = sorted({r["source_url"] for r in p13_rules + p13_guidance})
for p13_url in p13_urls:
    p13_parts = urlsplit(p13_url)
    p13_require(p13_parts.scheme == "https" and bool(p13_parts.hostname) and not p13_parts.username and not p13_parts.password,
                "Invalid reference URL: " + p13_url)
p13_source_ids = {url: "GUID-SRC-" + hashlib.sha256(url.encode()).hexdigest()[:20] for url in p13_urls}

# Interpretation flags are screening prompts, not adopted rule definitions.
def p13_interpretation_flags(row, general=False):
    flags = ['current_source_and_qualifications_review_pending', 'peer_review_and_adoption_pending']
    if general:
        flags.append('unstructured_guidance_do_not_infer_numeric_cutoff')
        if row['topic'] == 'Sleep':
            flags += ['reported_sleep_history_required', 'review_relation_to_structured_sleep_records']
        if row['source'] == 'AAP':
            flags.append('broad_sport_guidance_not_basketball_specific')
        if row['topic'] == 'Rest':
            flags += ['review_relation_to_AAP_structured_rest_record', 'do_not_treat_rest_range_as_ceiling']
        return flags
    metric = row['metric']
    if row['rule_type'].startswith('recommended'):
        flags.append('recommendation_not_mandatory_participation_requirement')
    if row['rule_type'] == 'maximum':
        flags.append('maximum_is_not_training_target_or_clearance')
    if row['source_organization'] == 'NBA and USA Basketball':
        flags += ['review_age_grade_overlap_without_assuming_precedence', 'review_publisher_table_coverage_and_footnotes']
    if 'resolve using published player segmentation' in row['applicability_notes']:
        flags.append('upstream_segmentation_precedence_note_needs_interpretation_review')
    if metric == 'organized_basketball_hours_per_week':
        flags.append('structured_training_classification_and_complete_schedule_required')
    if metric == 'organized_basketball_months_per_year':
        flags.append('annual_history_required_not_inferable_from_weekly_plan')
    if row['rule_id'] == 'NBA_G9_12_MAX_MONTHS':
        flags.append('published_maximum_range_lower_endpoint_is_not_minimum_participation')
    if metric.startswith('sleep_'):
        flags += ['reported_sleep_history_required', 'preserve_nightly_versus_daily_measurement_period']
    if row['rule_id'] == 'NBA_A12_SLEEP':
        flags.append('outside_project_target_ages_13_to_17')
    if row['rule_id'] == 'AAP_WEEKLY_SPORT_REST':
        flags += ['broad_sport_guidance_not_basketball_specific', 'do_not_treat_upper_rest_value_as_ceiling']
    if row['rule_id'] == 'NBA_A12_14_MAX_HOURS':
        flags.append('review_concurrent_sport_qualification')
    return flags

p13_candidates = []
for p13_record_kind, p13_rows in (("structured_reference", p13_rules), ("general_guidance", p13_guidance)):
    for p13_index, p13_row in enumerate(p13_rows, start=1):
        p13_general = p13_record_kind == "general_guidance"
        p13_identity = ("RECOVERY-GUIDANCE-" + hashlib.sha256(
            (p13_row["source_url"] + "|" + p13_row["topic"]).encode()).hexdigest()[:20]) if p13_general else p13_row["rule_id"]
        p13_candidates.append({
            "candidate_rule_id": p13_identity, "record_kind": p13_record_kind, "rule_origin": "published_guidance",
            "input_reference": p13_refs["general_guidance" if p13_general else "structured_rules"],
            "input_row_number": p13_index, "source_id": p13_source_ids[p13_row["source_url"]],
            "metric_or_topic": p13_row["topic"] if p13_general else p13_row["metric"],
            "source_rule_type": "unstructured_guidance" if p13_general else p13_row["rule_type"],
            "source_unit": "" if p13_general else p13_row["unit"], "imported_record": dict(p13_row),
            "upstream_verification_assertion": "" if p13_general else p13_row["verification_status"],
            "interpretation_flags": p13_interpretation_flags(p13_row, general=p13_general),
            "current_source_review_status": "pending", "interpretation_review_status": "pending", "peer_review_status": "pending",
            "adoption_status": "not_adopted", "runtime_activation_status": "inactive", "enabled_for_planner": False,
            "application_enforcement": None, "comparison_operator": None, "adopted_threshold": None,
            "applicability_definition_status": "pending", "evaluation_treatment_status": "pending",
        })
p13_check("unique_candidate_ids", len({r["candidate_rule_id"] for r in p13_candidates}) == len(p13_candidates),
          "Structured IDs and content-derived guidance IDs are unique.")
p13_check("original_records_preserved", [r["imported_record"] for r in p13_candidates if r["record_kind"] == "structured_reference"] == p13_rules
          and [r["imported_record"] for r in p13_candidates if r["record_kind"] == "general_guidance"] == p13_guidance,
          "Every original source field and verification assertion is preserved.")
p13_check("no_adoption_or_activation", all(not r["enabled_for_planner"] and r["adoption_status"] == "not_adopted"
          and r["application_enforcement"] is None for r in p13_candidates), "No rule is adopted, enforced, or enabled.")
p13_source_rows = [{"source_id": p13_source_ids[url], "source_url": url,
                    "candidate_records": sum(r["source_id"] == p13_source_ids[url] for r in p13_candidates),
                    "source_organization_labels": ";".join(sorted({r.get("source_organization", r.get("source", ""))
                                                                  for r in p13_rules + p13_guidance if r["source_url"] == url})),
                    "current_source_review_status": "pending", "source_refresh_performed": "no",
                    "alias_or_duplicate_source_review": "pending", "intended_use_review": "pending"}
                   for url in p13_urls]
p13_check("source_memberships", len({r["source_id"] for r in p13_source_rows}) == len(p13_urls)
          and sum(r["candidate_records"] for r in p13_source_rows) == len(p13_candidates),
          "Every candidate has a stable source-URL identity; aliases remain separate.")
p13_review_rows = [{"candidate_rule_id": r["candidate_rule_id"], "record_kind": r["record_kind"], "source_id": r["source_id"],
                    "source_url": r["imported_record"]["source_url"], "metric_or_topic": r["metric_or_topic"],
                    "source_rule_type": r["source_rule_type"], "source_unit": r["source_unit"],
                    "upstream_verification_assertion": r["upstream_verification_assertion"],
                    "interpretation_flags": ";".join(r["interpretation_flags"]), "source_review_status": "pending",
                    "interpretation_review_status": "pending", "peer_review_status": "pending", "reviewer_decision": "pending",
                    "reviewer": "", "review_date": "", "review_evidence_note": "", "application_enforcement": "",
                    "adoption_status": "not_adopted", "runtime_activation_status": "inactive", "enabled_for_planner": "no"}
                   for r in p13_candidates]

# Existing review ledgers remain separate inputs; this inventory neither replaces
# them nor invalidates their decisions. Migration needs its own lineage review.
p13_prior_ledgers = []
for p13_ref in ("results/planning_rule_review_v0_1.csv", "results/planning_rule_review.csv"):
    if p13_path(p13_ref).is_file():
        p13_read(p13_ref)
        p13_prior_ledgers.append(p13_ref)
p13_inputs = dict(p13_hashes)
p13_metrics = {"structured_reference_records": len(p13_rules), "general_guidance_records": len(p13_guidance),
               "candidate_records": len(p13_candidates), "source_urls": len(p13_urls), "adopted_rules": 0, "enabled_rules": 0,
               "existing_review_ledgers_detected": len(p13_prior_ledgers), "inventory_checks_passed": len(p13_checks)}
p13_register = {"schema_version": "planning-rule-candidate-inventory-1", "status": "candidate_inventory_only",
                "target_age_min": 13, "target_age_max": 17, "records": p13_candidates,
                "rule_origins_planned": ["published_guidance", "user_constraint", "project_policy"],
                "user_constraints_or_project_policies_created": False, "source_coverage_complete": False,
                "current_source_verification_performed": False, "existing_review_ledgers": p13_prior_ledgers,
                "prior_review_migration_status": "not_performed; existing decisions remain in their original ledgers"}


def p13_json_bytes(value):
    return (json.dumps(value, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False) + "\n").encode("utf-8")


def p13_csv_bytes(rows):
    buffer = io.StringIO(newline="")
    writer = csv.DictWriter(buffer, fieldnames=list(rows[0]), extrasaction="raise", lineterminator="\n")
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode("utf-8")


def p13_verify_inputs():
    for reference, digest in p13_inputs.items():
        p13_read(reference, digest)


p13_payload_names = {
    "planning_rule_candidates_REVIEW_ONLY.json": p13_json_bytes(p13_register),
    "planning_rule_review.csv": p13_csv_bytes(p13_review_rows),
    "planning_rule_source_index.csv": p13_csv_bytes(p13_source_rows),
    "planning_rule_inventory_checks.csv": p13_csv_bytes(p13_checks),
}
p13_summary = {"schema_version": "step13-rule-inventory-producer-1", "recipe_version": "rule-inventory-0.2.0",
               "status": "inventory_prepared_with_reviews", "step12_producer": p13_producer_ref, "step12_handoff": p13_handoff_ref,
               "metrics": p13_metrics, "input_hashes": dict(sorted(p13_inputs.items())), "verified_checks": p13_checks,
               "existing_review_ledgers": p13_prior_ledgers, "prior_review_migration_performed": False,
               "scope": {"source_refresh_performed": False, "source_coverage_complete": False, "rules_adopted": 0,
                         "rules_enabled": 0, "drill_approvals_created": 0, "catalog_fields_changed": 0,
                         "user_constraints_or_project_policies_created": False, "full_step13_complete": False},
               "artifacts": {name: {"sha256": hashlib.sha256(data).hexdigest(), "bytes": len(data)}
                             for name, data in sorted(p13_payload_names.items())}}
p13_summary_bytes = p13_json_bytes(p13_summary)
p13_digest = hashlib.sha256(p13_summary_bytes).hexdigest()
p13_bundle_ref = "results/planning_rule_inventory/inventory_" + p13_digest[:16]
p13_bundle = p13_path(p13_bundle_ref)
p13_payloads = {p13_bundle / name: data for name, data in p13_payload_names.items()}
p13_producer_path = p13_bundle / "step13_rule_inventory_summary.json"
p13_payloads[p13_producer_path] = p13_summary_bytes
p13_verify_inputs()
p13_require(not set(p13_payloads).intersection({p13_path(r) for r in p13_inputs}), "Inventory outputs overlap inspected inputs.")
for p13_path_out, p13_data in p13_payloads.items():
    p13_require(not p13_path_out.exists() or (p13_path_out.is_file() and p13_path_out.read_bytes() == p13_data),
                "A different inventory artifact occupies: " + p13_path_out.relative_to(p13_root).as_posix())
p13_created = []
p13_bundle.mkdir(parents=True, exist_ok=True)
try:
    for p13_path_out, p13_data in p13_payloads.items():
        # Verify data files and unchanged inputs before publishing the producer last.
        if p13_path_out == p13_producer_path:
            for p13_name, p13_content in p13_payload_names.items():
                p13_require((p13_bundle / p13_name).read_bytes() == p13_content, "A staged inventory artifact changed.")
            p13_verify_inputs()
        if p13_path_out.exists():
            p13_require(p13_path_out.is_file() and p13_path_out.read_bytes() == p13_data, "An inventory artifact changed during saving.")
            continue
        try:
            with p13_path_out.open("xb") as p13_handle:
                p13_created.append(p13_path_out)
                p13_handle.write(p13_data)
                p13_handle.flush()
                os.fsync(p13_handle.fileno())
        except FileExistsError:
            p13_require(p13_path_out.is_file() and p13_path_out.read_bytes() == p13_data,
                        "An inventory destination appeared with different bytes.")
    for p13_path_out, p13_data in p13_payloads.items():
        p13_require(p13_path_out.read_bytes() == p13_data, "Saved inventory verification failed: " + p13_path_out.name)
    p13_verify_inputs()
except Exception:
    p13_retained = []
    for p13_path_out in reversed(p13_created):
        if p13_path_out.is_file():
            if p13_path_out.read_bytes() == p13_payloads[p13_path_out]:
                p13_path_out.unlink()
            else:
                p13_retained.append(p13_path_out.relative_to(p13_root).as_posix())
    step13_report.update(status="save_failed", inventory_saved=False, save_attempt_files_retained=p13_retained)
    raise
step13_report.update(status="inventory_prepared_with_reviews", inventory_saved=True, metrics=p13_metrics,
                     inventory_bundle=p13_bundle_ref, inventory_bundle_digest=p13_digest,
                     producer_record=p13_producer_path.relative_to(p13_root).as_posix(), input_hashes=p13_inputs,
                     source_reference_paths=p13_refs, step12_producer=p13_producer_ref, step12_handoff=p13_handoff_ref)
print("Step 13-1 — current handoff verified and planning-rule inventory saved")
print("Step 12 producer:", p13_producer_ref)
print(f"Structured references / general guidance / total candidates: {len(p13_rules)} / {len(p13_guidance)} / {len(p13_candidates)}")
print(f"Recorded source URLs: {len(p13_urls)} | Inventory checks passed: {len(p13_checks)}")
print("Source, interpretation, and peer review: pending current inventory review")
print("Inventory rules adopted / enabled: 0 / 0 | Catalog approvals / planner eligible: 0 / 0")
print(f"Existing review ledgers preserved: {len(p13_prior_ledgers)}; prior decisions not migrated by this cell")
print("Bundle directory:", p13_bundle_ref)
print("Producer record:", step13_report["producer_record"])
print(f"Files created / identical files reused: {len(p13_created)} / {len(p13_payloads) - len(p13_created)}")
print(f"Verified input files unchanged: {len(p13_inputs)} | Source refresh performed: No")
print("Next: source-grounded applicability, interpretation, and review decisions before adoption.")
```

### 14.8 Calculate Quantities and Portions Only When Supported

Our default display will use the verified per-100-gram basis. If the requested edible-food mass is known in grams, the calculation will be:

\[
N_{\text{requested}} = N_{100g}\times\frac{M_{\text{grams}}}{100}
\]

Here, \(N_{100g}\) is the recorded nutrient amount in its stated unit, and \(M_{\text{grams}}\) is the selected food's requested mass.

For a household measure, we will use a compatible, food-specific portion record. Its quantity, description, and gram weight must be interpreted together. SR Legacy may carry the household-measure description in `modifier`, and a measure-unit identifier may be a placeholder [2].

We will distinguish two calculations:

- For \(k\) copies of a complete recorded portion weighing \(w\) grams, the requested mass is \(kw\).
- If a verified record represents \(a\) units weighing \(w\) grams and the request is for \(q\) of the same defined unit, the requested mass is \((q/a)w\).

The second calculation requires a known positive \(a\), compatible measure meaning, and usable gram weight. It will not be used to guess the interpretation of an ambiguous portion. We will avoid multiplying a gram weight by its recorded quantity a second time [2, 3].

We will not apply generic cup-to-gram or tablespoon-to-gram assumptions. If household conversion is unsupported, we will retain the per-100-gram display or request a mass in grams. A valid mass-based lookup will remain distinguishable from unsupported household-measure conversion.

We will retain calculation precision internally and document display rounding. A displayed quantity will identify the reference used; it will not recommend that amount for the athlete.

```python
# Step 14-1 — Verify the food-reference baseline and inventory local USDA files
# Read-only: no downloads, extraction, source-register edits, or lookup activation.
import csv
import hashlib
import io
import json
import os
import re
import zipfile
from collections import Counter
from decimal import Decimal, InvalidOperation
from pathlib import Path, PureWindowsPath

STEP13_PRODUCER_REFERENCE = ""  # Usually resolved from the completed Step 13-5.
USDA_EXTRA_DIRECTORIES = []  # Optional explicit paths if downloads are outside the project.
step14_report = {"status": "checking_inputs", "inventory_complete": False, "project_files_written": 0}
fd14_hashes, fd14_expected, fd14_checks = {}, {}, []


def fd14_require(condition, message):
    if not condition:
        step14_report.update(status="blocked", inventory_complete=False, reason=message)
        raise RuntimeError("Step 14-1: " + message)


def fd14_check(name, condition, details):
    fd14_checks.append({"check_name": name, "status": "pass" if condition else "failed", "details": details})
    fd14_require(condition, details)


fd14_hint = globals().get("project_root")
fd14_cwd = Path.cwd().resolve()
fd14_roots = ([Path(fd14_hint).resolve()] if fd14_hint is not None else []) + [fd14_cwd, *fd14_cwd.parents]
fd14_root = next((p for p in fd14_roots if (p / "data/source_manifest.csv").is_file()
                  and (p / "results").is_dir()), None)
fd14_require(fd14_root is not None, "Run inside our project with its saved source register and results.")


def fd14_path(reference):
    fd14_require(isinstance(reference, str) and bool(reference), "An artifact path is empty or invalid.")
    relative = Path(reference.replace("\\", "/"))
    fd14_require(not relative.is_absolute() and not PureWindowsPath(reference).drive,
                  "Expected a project-relative artifact path: " + reference)
    path = (fd14_root / relative).resolve()
    fd14_require(path.is_relative_to(fd14_root), "An artifact path leaves the project: " + reference)
    return path


def fd14_sha(path):
    digest = hashlib.sha256()
    with path.open("rb") as handle:
        for block in iter(lambda: handle.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()


def fd14_track(reference, expected=None):
    path = fd14_path(reference)
    fd14_require(path.is_file(), "Missing inspected input: " + reference)
    key, digest = path.relative_to(fd14_root).as_posix(), fd14_sha(path)
    fd14_require(key not in fd14_hashes or fd14_hashes[key] == digest, "An input changed during inventory: " + key)
    if expected is not None:
        fd14_require(isinstance(expected, str) and re.fullmatch(r"[0-9a-f]{64}", expected), "Invalid recorded SHA-256: " + key)
        fd14_require(key not in fd14_expected or fd14_expected[key] == expected, "Producer hashes disagree: " + key)
        fd14_require(digest == expected, "Input checksum differs: " + key + ". Share this error before rerunning earlier producers.")
        fd14_expected[key] = expected
    fd14_hashes[key] = digest
    return path


def fd14_pairs(pairs):
    result = {}
    for key, value in pairs:
        fd14_require(key not in result, "Repeated JSON key: " + key)
        result[key] = value
    return result


def fd14_json(reference, expected=None):
    value = json.loads(fd14_track(reference, expected).read_text(encoding="utf-8-sig"), object_pairs_hook=fd14_pairs)
    fd14_require(isinstance(value, dict), "Expected a JSON object: " + reference)
    return value


def fd14_csv(reference, required=()):
    with fd14_track(reference).open(encoding="utf-8-sig", newline="") as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    fd14_require(bool(columns) and all(columns) and len(columns) == len(set(columns)) and set(required).issubset(columns),
                  "Invalid or incomplete CSV schema: " + reference)
    fd14_require(all(None not in row and all(v is not None for v in row.values()) for row in rows), "Malformed CSV records: " + reference)
    return rows, columns


def fd14_index(rows, field, label):
    keys = [row[field] for row in rows]
    fd14_require(all(isinstance(k, str) and k.strip() for k in keys) and len(keys) == len(set(keys)),
                  "Missing or duplicate " + field + " in " + label)
    return dict(zip(keys, rows))


# Source coverage and adoption may remain open; their recorded states do not
# prevent this independent food-data inventory from verifying the saved handoff.
fd14_state13 = globals().get("step13_report", {})
fd14_producer_ref = STEP13_PRODUCER_REFERENCE or (
    fd14_state13.get("decision_package_producer", "") if fd14_state13.get("decision_package_saved") else "")
if not fd14_producer_ref:
    fd14_options = sorted((fd14_root / "results/planning_rule_review").glob("decision_tests_*/step13_decision_test_summary.json"))
    if len(fd14_options) != 1:
        print("Available Step 13 preparation producers:", [p.relative_to(fd14_root).as_posix() for p in fd14_options])
        fd14_require(False, "Run Step 13-5 or set STEP13_PRODUCER_REFERENCE to the intended saved producer.")
    fd14_producer_ref = fd14_options[0].relative_to(fd14_root).as_posix()
fd14_anchor = fd14_state13.get("decision_package_digest") if fd14_producer_ref == fd14_state13.get("decision_package_producer") else None
fd14_producer = fd14_json(fd14_producer_ref, fd14_anchor)
fd14_check("step13_producer", fd14_producer.get("schema_version") == "step13-decision-test-producer-1"
            and fd14_producer.get("status") == "preparation_checkpoint_passed_with_open_reviews",
            "The current Step 13 preparation checkpoint is identified.")
fd14_check("step13_checks", bool(fd14_producer.get("verified_checks"))
            and all(r.get("status") == "pass" for r in fd14_producer["verified_checks"]), "Saved Step 13 structural checks passed.")
fd14_require(isinstance(fd14_producer.get("input_hashes"), dict) and bool(fd14_producer["input_hashes"])
              and isinstance(fd14_producer.get("artifacts"), dict) and "step13_preparation_handoff.json" in fd14_producer["artifacts"],
              "The Step 13 producer lacks its input inventory or handoff.")
for fd14_ref, fd14_digest in fd14_producer["input_hashes"].items():
    fd14_track(fd14_ref, fd14_digest)
fd14_parent = Path(fd14_producer_ref).parent.as_posix()
for fd14_name, fd14_metadata in fd14_producer["artifacts"].items():
    fd14_require(Path(fd14_name).name == fd14_name, "Expected a simple producer artifact filename.")
    fd14_track(fd14_parent + "/" + fd14_name, fd14_metadata["sha256"])
fd14_handoff_ref = fd14_parent + "/step13_preparation_handoff.json"
fd14_handoff = fd14_json(fd14_handoff_ref)
fd14_check("step13_handoff", fd14_handoff.get("schema_version") == "step13-preparation-handoff-1"
            and fd14_handoff.get("structural_preparation_checkpoint") == "passed", "Step 13 handoff integrity and scope are preserved.")
for fd14_ref in ("data/source_manifest.csv", "results/step11_eda_summary.json"):
    fd14_require(fd14_ref in fd14_producer["input_hashes"], "The upstream chain omits a required source control: " + fd14_ref)
fd14_manifest, _ = fd14_csv("data/source_manifest.csv", ("source_id", "relative_path", "sha256"))
fd14_sources = fd14_index(fd14_manifest, "source_id", "source register")
fd14_eda = fd14_json("results/step11_eda_summary.json")
fd14_check("step11_food_context", fd14_eda.get("schema_version") == "step11-integrated-eda-1"
            and fd14_eda.get("status") in {"passed", "completed_with_reviews"}, "The saved food EDA is identified.")
fd14_earlier_nutrition = fd14_eda["supporting_metrics"]["nutrition"]


def fd14_registered(source_id):
    fd14_require(source_id in fd14_sources, "A required food source is not registered: " + source_id)
    row = fd14_sources[source_id]
    reference = row["relative_path"].replace("\\", "/")
    fd14_require(reference in fd14_producer["input_hashes"], "A registered food input was not tracked upstream: " + reference)
    fd14_track(reference, fd14_producer["input_hashes"][reference])
    if row["sha256"]:
        fd14_track(reference, row["sha256"])
    return reference


fd14_food_ref = fd14_registered("aditi_usda_foundation_nutrition")
fd14_quality_ref = fd14_registered("aditi_usda_quality_report")
fd14_numeric_fields = ("protein_g", "fat_g", "carbohydrate_g", "fiber_g", "energy_general_kcal", "energy_specific_kcal", "calories_kcal")
fd14_foods, fd14_food_columns = fd14_csv(fd14_food_ref, ("fdc_id", "description", "data_type", "energy_method", *fd14_numeric_fields))
fd14_food_ids = fd14_index(fd14_foods, "fdc_id", "prepared Foundation reference")
fd14_quality, _ = fd14_csv(fd14_quality_ref, ("column", "non_missing", "missing"))
fd14_quality_by_column = fd14_index(fd14_quality, "column", "prepared quality report")
fd14_missing_tokens = {"", "none", "null", "n/a", "na"}


def fd14_numeric(value):
    if value.strip().casefold() in fd14_missing_tokens:
        return "missing", None
    try:
        number = Decimal(value)
    except InvalidOperation:
        return "invalid_numeric", None
    if not number.is_finite():
        return "nonfinite", None
    return ("negative" if number < 0 else "finite_nonnegative"), number


fd14_numbers = [{field: fd14_numeric(row[field]) for field in fd14_numeric_fields} for row in fd14_foods]
fd14_core = ("calories_kcal", "protein_g", "fat_g", "carbohydrate_g")
fd14_finite = {"negative", "finite_nonnegative"}
fd14_current = {
    "records": len(fd14_foods),
    "finite_core_complete": sum(all(row[f][0] in fd14_finite for f in fd14_core) for row in fd14_numbers),
    "finite_core_incomplete": sum(not all(row[f][0] in fd14_finite for f in fd14_core) for row in fd14_numbers),
    "nonnegative_finite_core_complete": sum(all(row[f][0] == "finite_nonnegative" for f in fd14_core) for row in fd14_numbers),
    "negative_measurement_records": sum(any(state == "negative" for state, value in row.values()) for row in fd14_numbers),
    "invalid_or_nonfinite_measurement_cells": sum(state in {"invalid_numeric", "nonfinite"} for row in fd14_numbers for state, value in row.values()),
}
fd14_method_fields = {"Atwater Specific (2048)": "energy_specific_kcal", "Atwater General (2047)": "energy_general_kcal"}
fd14_method_checked, fd14_method_conflicts = 0, 0
for fd14_raw, fd14_num in zip(fd14_foods, fd14_numbers):
    target = fd14_method_fields.get(fd14_raw["energy_method"])
    if target and fd14_num["calories_kcal"][0] in fd14_finite and fd14_num[target][0] in fd14_finite:
        fd14_method_checked += 1
        fd14_method_conflicts += fd14_num["calories_kcal"][1] != fd14_num[target][1]
fd14_current.update(energy_method_rows_checked=fd14_method_checked,
                     energy_method_rows_uncheckable=len(fd14_foods) - fd14_method_checked,
                     energy_method_inconsistencies=fd14_method_conflicts)
for fd14_key, fd14_value in fd14_current.items():
    fd14_check("food_baseline:" + fd14_key, fd14_earlier_nutrition.get(fd14_key) == fd14_value,
                "Recomputed food screening agrees with Step 11: " + fd14_key)
fd14_field_states = {field: dict(Counter(row[field][0] for row in fd14_numbers)) for field in fd14_numeric_fields}
fd14_check("quality_columns", set(fd14_quality_by_column) == set(fd14_food_columns), "The missingness report covers the prepared food columns.")
for fd14_column in fd14_food_columns:
    count = sum(row[fd14_column].strip().casefold() in fd14_missing_tokens for row in fd14_foods)
    report = fd14_quality_by_column[fd14_column]
    fd14_check("reported_missingness:" + fd14_column,
                int(report["missing"]) == count and int(report["non_missing"]) == len(fd14_foods) - count,
                "Reported missing/nonmissing counts agree; nonmissing is not the same as nonnegative eligibility.")
fd14_check("foundation_labels", {r["data_type"] for r in fd14_foods} == {"Foundation Foods"},
            "The existing prepared records retain their Foundation Foods labels.")

# Historical tables remain context: row counts do not establish release compatibility.
fd14_historical = []
for fd14_sid in ("derived_food_category", "derived_food_portion", "derived_food_sample_link", "derived_nutrient_result"):
    if fd14_sid not in fd14_sources or not fd14_sources[fd14_sid]["relative_path"]:
        fd14_historical.append({"source_id": fd14_sid, "availability": "not_registered_with_local_path", "rows": None})
        continue
    reference = fd14_registered(fd14_sid)
    with fd14_path(reference).open(encoding="utf-8-sig", newline="") as handle:
        reader = csv.reader(handle)
        columns = next(reader, [])
        count = 0
        for values in reader:
            fd14_require(len(values) == len(columns), "Malformed historical table row: " + reference)
            count += 1
    fd14_historical.append({"source_id": fd14_sid, "relative_path": reference, "availability": "inspected_historical_input",
                             "rows": count, "columns": columns, "release_compatibility": "not_established"})

# Inspect only explicit project data/download locations and any directories added
# above. New archive/table candidates are sampled, not adopted or fully hashed.
fd14_table_names = {"food.csv", "food_nutrient.csv", "nutrient.csv", "food_category.csv", "food_portion.csv",
                     "measure_unit.csv", "sr_legacy_food.csv", "foundation_food.csv"}
fd14_search_dirs = {fd14_root / name for name in ("data", "downloads", "datasets", "raw_data") if (fd14_root / name).is_dir()}
for row in fd14_historical:
    if row.get("relative_path"):
        fd14_search_dirs.add(fd14_path(row["relative_path"]).parent)
for directory in USDA_EXTRA_DIRECTORIES:
    path = Path(directory).expanduser().resolve()
    if path.is_dir():
        fd14_search_dirs.add(path)
    else:
        print("Additional search directory not found:", str(path))


def fd14_label(path):
    path = path.resolve()
    return path.relative_to(fd14_root).as_posix() if path.is_relative_to(fd14_root) else str(path)


def fd14_candidate_file(path):
    compact = re.sub(r"[^a-z0-9]", "", path.name.casefold())
    return path.name.casefold() in fd14_table_names or (
        path.suffix.casefold() == ".zip" and any(s in compact for s in ("fooddatacentral", "srlegacy", "supporting", "usda", "foundation")))


def fd14_food_sample(reader, columns):
    types = set()
    if "data_type" in columns:
        position = columns.index("data_type")
        for index, values in enumerate(reader):
            if position < len(values):
                types.add(values[position])
            if index >= 24:
                break
    return sorted(types)


fd14_discovered = {p.resolve() for p in fd14_root.iterdir() if p.is_file() and fd14_candidate_file(p)}
for fd14_search in sorted(fd14_search_dirs):
    for folder, directories, filenames in os.walk(fd14_search, followlinks=False):
        directories[:] = [d for d in directories if not d.startswith(".") and d not in {"venv", "node_modules", "__pycache__", "backups"}]
        for name in filenames:
            path = Path(folder) / name
            if path.is_file() and fd14_candidate_file(path):
                fd14_discovered.add(path.resolve())
fd14_archives, fd14_raw_tables = [], []
for path in sorted(fd14_discovered):
    if path.suffix.casefold() == ".zip":
        info = {"path": fd14_label(path), "bytes": path.stat().st_size, "members": [], "sampled_food_data_types": [],
                "filename_suggests_sr_legacy": "srlegacy" in re.sub(r"[^a-z0-9]", "", path.name.casefold()),
                "inspection_scope": "member_listing_and_up_to_25_food_rows_only", "release_verified": False}
        try:
            with zipfile.ZipFile(path) as archive:
                unsafe = []
                for member in archive.infolist():
                    normalized = member.filename.replace("\\", "/")
                    relative = Path(normalized)
                    if relative.is_absolute() or ".." in relative.parts or PureWindowsPath(normalized).drive:
                        unsafe.append(member.filename)
                        continue
                    if not member.is_dir() and relative.name.casefold() in fd14_table_names:
                        info["members"].append(member.filename)
                        if relative.name.casefold() == "food.csv":
                            with archive.open(member) as raw_handle:
                                with io.TextIOWrapper(raw_handle, encoding="utf-8-sig", newline="") as text_handle:
                                    reader = csv.reader(text_handle)
                                    columns = next(reader, [])
                                    info["sampled_food_data_types"] = sorted(set(info["sampled_food_data_types"]) | set(fd14_food_sample(reader, columns)))
                info["unsafe_member_paths"] = len(unsafe)
        except (OSError, UnicodeError, csv.Error, zipfile.BadZipFile, RuntimeError) as error:
            info["inspection_error"] = type(error).__name__ + ": " + str(error)
        fd14_archives.append(info)
    else:
        info = {"path": fd14_label(path), "table_name": path.name.casefold(), "bytes": path.stat().st_size,
                "columns": [], "sampled_food_data_types": [], "release_verified": False}
        try:
            with path.open(encoding="utf-8-sig", newline="") as handle:
                reader = csv.reader(handle)
                info["columns"] = next(reader, [])
                if path.name.casefold() == "food.csv":
                    info["sampled_food_data_types"] = fd14_food_sample(reader, info["columns"])
        except (OSError, UnicodeError, csv.Error) as error:
            info["inspection_error"] = type(error).__name__ + ": " + str(error)
        fd14_raw_tables.append(info)
fd14_sr_archives = [r for r in fd14_archives if r["filename_suggests_sr_legacy"]
                     or any(Path(m.replace("\\", "/")).name.casefold() == "sr_legacy_food.csv" for m in r["members"])
                     or any("srlegacy" in re.sub(r"[^a-z0-9]", "", t.casefold()) or t == "sr_legacy_food" for t in r["sampled_food_data_types"])]
fd14_sr_tables = [r for r in fd14_raw_tables if r["table_name"] == "sr_legacy_food.csv"
                   or any(t in {"sr_legacy_food", "SR Legacy"} for t in r["sampled_food_data_types"])]
if fd14_sr_archives or fd14_sr_tables:
    fd14_availability = "available_candidates_not_release_verified"
elif any(Path(m.replace("\\", "/")).name.casefold() in {"food.csv", "food_nutrient.csv"}
         for r in fd14_archives for m in r["members"]) or any(r["table_name"] in {"food.csv", "food_nutrient.csv"} for r in fd14_raw_tables):
    fd14_availability = "source_classification_pending"
else:
    fd14_availability = "acquisition_needed_in_searched_locations"
for reference, digest in list(fd14_hashes.items()):
    fd14_track(reference, digest)
step14_report.update(status="read_only_inventory_complete", inventory_complete=True, step13_producer=fd14_producer_ref,
                     input_hashes=dict(fd14_hashes), foundation_input=fd14_food_ref, foundation_quality_report=fd14_quality_ref,
                     foundation_metrics=fd14_current, field_numeric_states=fd14_field_states, historical_inputs=fd14_historical,
                     search_directories=[fd14_label(p) for p in sorted(fd14_search_dirs)], archive_candidates=fd14_archives,
                     raw_table_candidates=fd14_raw_tables, intended_source="SR Legacy April 2018",
                     sr_legacy_availability=fd14_availability, lookup_ready=False, verified_checks=fd14_checks,
                     discovery_limitations="Names and samples do not verify release, whole-file contents, supporting-table compatibility, or acquisition provenance.")
print("Step 14-1 — food-reference baseline verified and local USDA inventory completed")
print("Step 13 producer:", fd14_producer_ref)
print(f"Prepared Foundation rows / unique food IDs: {len(fd14_foods)} / {len(fd14_food_ids)}")
print("Recorded data-type labels:", dict(Counter(r["data_type"] for r in fd14_foods)))
print(f"Finite complete core / nonnegative complete core: {fd14_current['finite_core_complete']} / {fd14_current['nonnegative_finite_core_complete']}")
print(f"Negative carbohydrate records: {fd14_field_states['carbohydrate_g'].get('negative', 0)}")
print(f"Missing fiber / selected energy: {fd14_field_states['fiber_g'].get('missing', 0)} / {fd14_field_states['calories_kcal'].get('missing', 0)}")
print(f"Recorded energy-method checks / inconsistencies: {fd14_method_checked} / {fd14_method_conflicts}")
print("Missingness-report counts: consistent | Foundation units/basis/release evidence still needs source-level confirmation")
print("Historical supporting inputs (release compatibility not established):")
for row in fd14_historical:
    print(f"  {row['source_id']}: {row.get('rows')} rows | {row['availability']}")
print("Searched project locations:", step14_report["search_directories"], "; project-root files also checked")
print(f"Local USDA archive candidates / raw table candidates: {len(fd14_archives)} / {len(fd14_raw_tables)}")
for row in fd14_archives:
    print("  ARCHIVE", row["path"], "| relevant members:", row["members"], "| sampled food types:", row["sampled_food_data_types"])
    if row.get("inspection_error") or row.get("unsafe_member_paths"):
        print("    Review:", row.get("inspection_error", ""), "unsafe member paths:", row.get("unsafe_member_paths", 0))
for row in fd14_raw_tables:
    print("  TABLE", row["path"], "| columns:", row["columns"], "| sampled food types:", row["sampled_food_data_types"])
    if row.get("inspection_error"):
        print("    Review:", row["inspection_error"])
print("SR Legacy availability:", fd14_availability)
print(f"Passed consistency checks / unchanged verified source and control files: {len(fd14_checks)} / {len(fd14_hashes)}")
print("Project files written: 0 | Downloads/extraction performed: No | Lookup readiness established: No")
print("Next: review this output to select existing SR Legacy inputs or prepare the required source acquisition.")
```

```python
# Step 14-2 — Acquire the official SR Legacy inputs and save their source evidence
# Run Step 14-1 first. This adds a new raw-data bundle; existing inputs stay frozen.
import shutil
import stat
import tempfile
from datetime import datetime, timezone
from html.parser import HTMLParser
from urllib.parse import quote, unquote, urljoin, urlsplit, urlunsplit
from urllib.request import Request, urlopen

if not globals().get("step14_report", {}).get("inventory_complete") or not callable(globals().get("fd14_sha")):
    raise RuntimeError("Run Step 14-1 successfully in this kernel before Step 14-2.")

STEP14_ACQUISITION_REFERENCE = ""  # Only needed if multiple saved acquisition bundles exist.
fd142_inputs = dict(step14_report["input_hashes"])
fd142_base_ref = "data/raw/usda/sr_legacy_2018_04"
fd142_catalog_url = "https://fdc.nal.usda.gov/download-datasets/"
fd142_sources = [
    {"source_key": "sr_legacy", "publisher_data_type": "SR Legacy", "publisher_catalog_release": "2018-04",
     "url": "https://fdc.nal.usda.gov/fdc-datasets/FoodData_Central_sr_legacy_food_csv_2018-04.zip"},
]
fd142_supporting_spec = {"source_key": "supporting", "publisher_data_type": "Supporting Data", "publisher_catalog_release": "2019-04",
                        "url": "https://fdc.nal.usda.gov/fdc-datasets/FoodData_Central_Supporting_Data_csv_%202019-04-02.zip"}
fd142_created, fd142_reused = 0, 0
step14_report.update(status="acquiring_source_inputs", acquisition_saved=False)


def fd142_require(condition, message):
    if not condition:
        step14_report.update(status="source_acquisition_failed", acquisition_saved=False, reason=message)
        raise RuntimeError("Step 14-2: " + message)


def fd142_verify_upstream():
    for reference, digest in fd142_inputs.items():
        fd142_require(fd14_sha(fd14_path(reference)) == digest, "A frozen input changed: " + reference)


def fd142_json_bytes(value):
    return (json.dumps(value, ensure_ascii=False, indent=2, sort_keys=True, allow_nan=False) + "\n").encode("utf-8")


def fd142_child(base, relative):
    fd142_require(isinstance(relative, str) and bool(relative), "Empty acquisition artifact path.")
    path = Path(relative.replace("\\", "/"))
    fd142_require(not path.is_absolute() and not PureWindowsPath(relative).drive and ".." not in path.parts,
                  "Invalid acquisition artifact path: " + relative)
    target = (base / path).resolve()
    fd142_require(target.is_relative_to(base.resolve()), "Acquisition artifact leaves its bundle.")
    return target


def fd142_verify_bundle(base, summary):
    fd142_require(summary.get("schema_version") == "step14-source-acquisition-producer-1"
                  and summary.get("status") == "raw_inputs_acquired_pending_join_validation",
                  "The saved acquisition producer has an unexpected schema or status.")
    fd142_require(summary.get("input_hashes") == fd142_inputs, "The acquisition belongs to different upstream inputs.")
    fd142_require(isinstance(summary.get("artifacts"), dict) and bool(summary["artifacts"]), "Acquisition artifacts are missing.")
    fd142_require(summary.get("lookup_ready") is False, "This checkpoint must not activate a food lookup.")
    for relative, metadata in summary["artifacts"].items():
        target = fd142_child(base, relative)
        fd142_require(target.is_file() and target.stat().st_size == metadata["bytes"]
                      and fd14_sha(target) == metadata["sha256"], "Saved acquisition bytes differ: " + relative)
    expected_urls = {r["source_key"]: r["url"] for r in [*fd142_sources, fd142_supporting_spec]}
    recorded_urls = {r["source_key"]: r["url"] for r in summary["sources"]}
    fd142_require(recorded_urls.get("sr_legacy") == expected_urls["sr_legacy"]
                  and all(expected_urls.get(k) == v for k, v in recorded_urls.items()), "Saved source URLs differ from the pinned recipe.")
    fd142_verify_upstream()


def fd142_normalize_url(value):
    parts = urlsplit(urljoin(fd142_catalog_url, value))
    return urlunsplit((parts.scheme, parts.netloc, quote(parts.path, safe="/%"), parts.query, ""))


class FD142Links(HTMLParser):
    def __init__(self):
        super().__init__()
        self.links = set()

    def handle_starttag(self, tag, attrs):
        if tag.lower() == "a":
            self.links.update(fd142_normalize_url(value) for key, value in attrs if key.lower() == "href" and value)


def fd142_fetch(url, target, max_bytes):
    request = Request(url, headers={"User-Agent": "ChampionAthleticMap/step14-source-preparation", "Accept-Encoding": "identity"})
    started = datetime.now(timezone.utc).isoformat()
    with urlopen(request, timeout=45) as response:
        fd142_require(response.status == 200 and urlsplit(response.geturl()).scheme == "https", "Unexpected download response: " + url)
        stated = response.headers.get("Content-Length")
        fd142_require(stated is None or int(stated) <= max_bytes, "The source exceeds this recipe's download bound.")
        count = 0
        with target.open("xb") as handle:
            for block in iter(lambda: response.read(1024 * 1024), b""):
                count += len(block)
                fd142_require(count <= max_bytes, "The source exceeded the download bound.")
                handle.write(block)
            handle.flush()
            os.fsync(handle.fileno())
        fd142_require(count > 0 and (stated is None or count == int(stated)), "Incomplete download: " + url)
        return {"requested_url": url, "response_url": response.geturl(), "started_at_utc": started,
                "completed_at_utc": datetime.now(timezone.utc).isoformat(), "bytes": count,
                "sha256": fd14_sha(target), "content_type": response.headers.get("Content-Type"),
                "etag": response.headers.get("ETag"), "last_modified": response.headers.get("Last-Modified")}


def fd142_extract_tables(archive_path, source_key, stage):
    records, seen = [], set()
    with zipfile.ZipFile(archive_path) as archive:
        members = archive.infolist()
        fd142_require(len(members) <= 1000 and sum(m.file_size for m in members) <= 1024 ** 3,
                      "Unexpected archive member count or expansion size.")
        for member in members:
            normalized = member.filename.replace("\\", "/")
            relative = Path(normalized)
            fd142_require(not relative.is_absolute() and ".." not in relative.parts and not PureWindowsPath(normalized).drive
                          and not stat.S_ISLNK(member.external_attr >> 16), "Unsafe archive member: " + member.filename)
            fd142_require(not member.flag_bits & 1, "Encrypted archive member is not supported.")
        fd142_require(archive.testzip() is None, "An archive member failed its CRC check.")
        for member in members:
            if member.is_dir() or Path(member.filename).suffix.casefold() != ".csv":
                continue
            name = Path(member.filename.replace("\\", "/")).name
            fd142_require(name.casefold() not in seen, "Duplicate table basename within archive: " + name)
            seen.add(name.casefold())
            table_ref = "tables/" + source_key + "/" + name
            destination = fd142_child(stage, table_ref)
            destination.parent.mkdir(parents=True, exist_ok=True)
            with archive.open(member) as source, destination.open("xb") as target:
                shutil.copyfileobj(source, target, 1024 * 1024)
            fd142_require(destination.stat().st_size == member.file_size, "Extracted size mismatch: " + name)
            data_types, food_ids, count = Counter(), set(), 0
            with destination.open(encoding="utf-8-sig", newline="") as handle:
                reader = csv.DictReader(handle)
                columns = list(reader.fieldnames or [])
                fd142_require(bool(columns) and all(columns) and len(columns) == len(set(columns)), "Invalid table header: " + name)
                for row in reader:
                    fd142_require(None not in row and all(value is not None for value in row.values()), "Malformed CSV row: " + name)
                    count += 1
                    if name.casefold() == "food.csv":
                        data_types[row.get("data_type", "")] += 1
                        food_id = row.get("fdc_id", "")
                        fd142_require(bool(food_id) and food_id not in food_ids, "Missing or duplicate food ID.")
                        food_ids.add(food_id)
            records.append({"source_key": source_key, "table_name": name, "archive_member": member.filename,
                            "relative_path": table_ref, "rows": count, "columns": columns,
                            "bytes": destination.stat().st_size, "sha256": fd14_sha(destination),
                            "food_data_types": dict(data_types)})
    return records


fd142_verify_upstream()
fd142_hint = STEP14_ACQUISITION_REFERENCE or step14_report.get("acquisition_producer", "")
fd142_options = sorted((fd14_root / fd142_base_ref).glob("acquisition_*/step14_source_acquisition_summary.json"))
if not fd142_hint and len(fd142_options) == 1:
    fd142_hint = fd142_options[0].relative_to(fd14_root).as_posix()
fd142_require(bool(fd142_hint) or not fd142_options,
              "Multiple acquisition bundles exist; set STEP14_ACQUISITION_REFERENCE to the intended producer.")
if fd142_hint:
    fd142_producer = fd14_path(fd142_hint)
    fd142_require(fd142_producer.is_file() and fd142_producer.parent.parent == fd14_path(fd142_base_ref),
                  "Select an acquisition producer under the current raw-source directory.")
    fd142_producer_bytes = fd142_producer.read_bytes()
    fd142_require(fd142_producer.parent.name == "acquisition_" + hashlib.sha256(fd142_producer_bytes).hexdigest()[:16],
                  "The acquisition producer differs from its recorded bundle identity.")
    fd142_summary = json.loads(fd142_producer_bytes.decode("utf-8-sig"), object_pairs_hook=fd14_pairs)
    fd142_bundle = fd142_producer.parent
    fd142_verify_bundle(fd142_bundle, fd142_summary)
    fd142_reused = len(fd142_summary["artifacts"]) + 1
    print("Reusing the saved acquisition after verifying every recorded file; no network requests.")
else:
    fd142_parent = fd14_path(fd142_base_ref)
    fd142_parent.mkdir(parents=True, exist_ok=True)
    fd142_stage = Path(tempfile.mkdtemp(prefix=".step14-acquisition-", dir=fd142_parent))
    try:
        print("Fetching USDA's download catalog and the pinned SR Legacy archive...", flush=True)
        catalog = fd142_fetch(fd142_catalog_url, fd142_stage / "usda_download_catalog.html", 5 * 1024 ** 2)
        parser = FD142Links()
        parser.feed((fd142_stage / "usda_download_catalog.html").read_text(encoding="utf-8-sig"))
        fd142_require(all(r["url"] in parser.links for r in fd142_sources),
                      "The official catalog no longer lists the pinned SR Legacy link. Share this error before substituting another release.")
        source_records, table_records = [], []
        for spec in fd142_sources:
            filename = unquote(urlsplit(spec["url"]).path.rsplit("/", 1)[1])
            relative = "archives/" + spec["source_key"] + "/" + filename
            destination = fd142_child(fd142_stage, relative)
            destination.parent.mkdir(parents=True, exist_ok=True)
            print("Downloading", spec["publisher_data_type"], spec["publisher_catalog_release"], flush=True)
            response = fd142_fetch(spec["url"], destination, 100 * 1024 ** 2)
            source_records.append({**spec, "archive_filename": filename, "relative_path": relative, "response": response})
            table_records.extend(fd142_extract_tables(destination, spec["source_key"], fd142_stage))
        embedded_names = {r["table_name"].casefold() for r in table_records if r["source_key"] == "sr_legacy"}
        if not {"nutrient.csv", "food_category.csv", "measure_unit.csv"}.issubset(embedded_names):
            spec = fd142_supporting_spec
            fd142_require(spec["url"] in parser.links, "The pinned supporting-data fallback is no longer listed by USDA.")
            filename = unquote(urlsplit(spec["url"]).path.rsplit("/", 1)[1])
            relative = "archives/supporting/" + filename
            destination = fd142_child(fd142_stage, relative)
            destination.parent.mkdir(parents=True, exist_ok=True)
            print("Required definitions are missing from the package; downloading separately pinned Supporting Data April 2019.", flush=True)
            response = fd142_fetch(spec["url"], destination, 100 * 1024 ** 2)
            source_records.append({**spec, "archive_filename": filename, "relative_path": relative, "response": response})
            table_records.extend(fd142_extract_tables(destination, "supporting", fd142_stage))
        table_index = {(r["source_key"], r["table_name"].casefold()): r for r in table_records}
        core = {}
        for name, required in {"food.csv": {"fdc_id", "data_type", "description"},
                               "food_nutrient.csv": {"fdc_id", "nutrient_id", "amount"},
                               "nutrient.csv": {"id", "name", "unit_name"}}.items():
            roles = ("sr_legacy", "supporting") if name == "nutrient.csv" else ("sr_legacy",)
            selected = next((table_index[(role, name)] for role in roles if (role, name) in table_index), None)
            fd142_require(selected is not None and selected["rows"] > 0 and required.issubset(selected["columns"]),
                          "Required source table or columns missing: " + name)
            core[name] = selected["relative_path"]
        food_types = table_index[("sr_legacy", "food.csv")]["food_data_types"]
        fd142_require(set(food_types) == {"sr_legacy_food"}, "The SR archive has unexpected food data types.")
        artifacts = {p.relative_to(fd142_stage).as_posix(): {"bytes": p.stat().st_size, "sha256": fd14_sha(p)}
                     for p in sorted(fd142_stage.rglob("*")) if p.is_file()}
        fd142_summary = {"schema_version": "step14-source-acquisition-producer-1", "recipe_version": "source-acquisition-0.1.0",
                        "status": "raw_inputs_acquired_pending_join_validation", "input_hashes": fd142_inputs,
                        "upstream_step13_producer": step14_report["step13_producer"],
                        "catalog_url": fd142_catalog_url, "catalog_response": catalog,
                        "sources": source_records, "tables": table_records, "core_table_paths": core,
                        "artifacts": artifacts, "food_data_types": food_types,
                        "lookup_ready": False, "supporting_key_compatibility_verified": False,
                        "definition_precedence": "Use tables embedded in the SR package first; separately acquired definitions only when absent.",
                        "separate_supporting_archive_used": any(r["source_key"] == "supporting" for r in source_records),
                        "limitations": ["Hashes identify these acquired bytes; they are not independently published USDA checksums.",
                                        "Catalog dates, archive names, and record publication dates remain distinct.",
                                        "Embedded definitions take precedence. Any separately pinned fallback retains its own release; key coverage and semantics remain to be checked.",
                                        "Food subset selection, units/basis review, nutrient eligibility, and portion calculations remain pending."],
                        "verified_checks": [{"check_name": name, "status": "pass"} for name in (
                            "upstream_hashes", "official_catalog_links", "download_lengths", "archive_crc_and_paths",
                            "extracted_table_integrity", "csv_structure", "core_table_columns", "sr_food_ids_and_data_type")],
                        "scope": {"existing_inputs_changed": 0, "source_manifest_changed": False,
                                  "step11_reports_changed": False, "processed_food_tables_created": 0,
                                  "food_lookup_enabled": False, "full_step14_complete": False}}
        summary_bytes = fd142_json_bytes(fd142_summary)
        digest = hashlib.sha256(summary_bytes).hexdigest()
        (fd142_stage / "step14_source_acquisition_summary.json").write_bytes(summary_bytes)
        fd142_verify_bundle(fd142_stage, fd142_summary)
        fd142_bundle = fd142_parent / ("acquisition_" + digest[:16])
        fd142_require(not fd142_bundle.exists(), "Acquisition destination already exists; rerun to verify and reuse it.")
        fd142_stage.rename(fd142_bundle)
        fd142_producer = fd142_bundle / "step14_source_acquisition_summary.json"
        fd142_created = len(artifacts) + 1
    except Exception as error:
        step14_report.update(status="source_acquisition_failed", acquisition_saved=False, reason=str(error))
        raise
    finally:
        if fd142_stage.exists():
            shutil.rmtree(fd142_stage)
fd142_verify_bundle(fd142_bundle, fd142_summary)
fd142_published_digest = fd14_sha(fd142_producer)
fd142_require(fd142_bundle.name == "acquisition_" + fd142_published_digest[:16],
              "The published acquisition producer differs from its bundle identity.")
step14_report.update(status="source_acquisition_saved_pending_join_validation", acquisition_saved=True,
                     acquisition_bundle=fd142_bundle.relative_to(fd14_root).as_posix(),
                     acquisition_producer=fd142_producer.relative_to(fd14_root).as_posix(),
                     acquisition_digest=fd142_published_digest, acquisition_summary=fd142_summary,
                     project_files_written=fd142_created, lookup_ready=False)
print("Step 14-2 — official source inputs acquired and raw bundle verified")
print("Bundle directory:", step14_report["acquisition_bundle"])
print("Producer record:", step14_report["acquisition_producer"])
for source in fd142_summary["sources"]:
    response = source["response"]
    print(f"  {source['publisher_data_type']} catalog release {source['publisher_catalog_release']}: {response['bytes']} archive bytes")
    print("    SHA-256:", response["sha256"], "| acquired UTC:", response["completed_at_utc"])
print("Extracted CSV tables:", len(fd142_summary["tables"]))
for table in fd142_summary["tables"]:
    print(f"  {table['source_key']}/{table['table_name']}: {table['rows']} rows | columns: {table['columns']}")
print("Recorded food data types:", fd142_summary["food_data_types"])
print("Separately downloaded supporting archive used:", fd142_summary["separate_supporting_archive_used"])
print(f"Acquisition checks passed / unchanged upstream inputs: {len(fd142_summary['verified_checks'])} / {len(fd142_inputs)}")
print(f"Files created / verified files reused: {fd142_created} / {fd142_reused}")
print("Supporting key compatibility / units-basis review / food lookup readiness: pending / pending / not established")
print("Original Foundation files, source register, and Step 11 reports unchanged")
print("Next: review this output before validating joins, nutrient definitions, and the reproducible food selection.")
```

```python
# Step 14-3 — Validate the acquired SR source and prepare a quality-evidence bundle
# The source universe is screened here; prototype food selection follows this review.
from collections import defaultdict

if not globals().get("step14_report", {}).get("acquisition_saved"):
    raise RuntimeError("Complete Step 14-2 successfully in this kernel before Step 14-3.")
step14_report.update(status="validating_acquired_source", validation_prepared=False, validation_saved=False)
fd143_checks, fd143_anomalies = [], []


def fd143_require(condition, message):
    if not condition:
        step14_report.update(status="source_validation_blocked", validation_prepared=False, reason=message)
        raise RuntimeError("Step 14-3: " + message)


def fd143_check(name, condition, observed, expected):
    fd143_checks.append({"check_name": name, "status": "pass" if condition else "failed",
                         "observed": str(observed), "expected": str(expected)})
    fd143_require(condition, name + ": observed " + str(observed) + "; expected " + str(expected))


fd143_producer_ref = step14_report["acquisition_producer"]
fd143_producer = fd14_path(fd143_producer_ref)
fd143_producer_bytes = fd143_producer.read_bytes()
fd143_digest = hashlib.sha256(fd143_producer_bytes).hexdigest()
fd143_require(fd143_digest == step14_report["acquisition_digest"]
                and fd143_producer.parent.name == "acquisition_" + fd143_digest[:16], "The acquired source producer changed.")
fd143_acq = json.loads(fd143_producer_bytes.decode("utf-8-sig"), object_pairs_hook=fd14_pairs)
fd142_verify_bundle(fd143_producer.parent, fd143_acq)
fd143_inputs = dict(fd143_acq["input_hashes"])
fd143_inputs[fd143_producer_ref] = fd143_digest
for fd143_relative, fd143_meta in fd143_acq["artifacts"].items():
    path = fd142_child(fd143_producer.parent, fd143_relative)
    fd143_inputs[path.relative_to(fd14_root).as_posix()] = fd143_meta["sha256"]
fd143_table_index = {(r["source_key"], r["table_name"].casefold()): r for r in fd143_acq["tables"]}
fd143_table_refs = {}


def fd143_table(name):
    row = next((fd143_table_index[(key, name)] for key in ("sr_legacy", "supporting")
                if (key, name) in fd143_table_index), None)
    fd143_require(row is not None, "Required table is unavailable: " + name)
    path = fd142_child(fd143_producer.parent, row["relative_path"])
    reference = path.relative_to(fd14_root).as_posix()
    fd143_require(reference in fd143_inputs and fd14_sha(path) == fd143_inputs[reference], "A source table changed: " + name)
    fd143_table_refs[name] = reference
    return path, row


def fd143_rows(name, required):
    path, record = fd143_table(name)
    with path.open(encoding="utf-8-sig", newline="") as handle:
        reader = csv.DictReader(handle)
        fd143_require(set(required).issubset(reader.fieldnames or []), "Required columns are missing: " + name)
        count = 0
        for row in reader:
            fd143_require(None not in row and all(v is not None for v in row.values()), "Malformed table row: " + name)
            count += 1
            yield row
        fd143_check("parsed_rows:" + name, count == record["rows"], count, record["rows"])


def fd143_index(name, key, required):
    rows = list(fd143_rows(name, (key, *required)))
    identifiers = [r[key] for r in rows]
    fd143_check("unique_key:" + name + "." + key, all(k.strip() for k in identifiers)
                 and len(identifiers) == len(set(identifiers)), len(set(identifiers)), len(identifiers))
    return {r[key]: r for r in rows}


fd143_foods = fd143_index("food.csv", "fdc_id", ("data_type", "description", "food_category_id", "publication_date"))
fd143_defs = fd143_index("nutrient.csv", "id", ("name", "unit_name", "nutrient_nbr"))
fd143_categories = fd143_index("food_category.csv", "id", ("code", "description"))
fd143_units = fd143_index("measure_unit.csv", "id", ("name",))
fd143_legacy = fd143_index("sr_legacy_food.csv", "fdc_id", ("NDB_number",))
fd143_derivations = fd143_index("food_nutrient_derivation.csv", "id", ("code", "description", "source_id"))
fd143_nutrient_sources = fd143_index("food_nutrient_source.csv", "id", ("code", "description"))
fd143_check("legacy_food_correspondence", set(fd143_legacy) == set(fd143_foods), len(fd143_legacy), len(fd143_foods))
fd143_check("legacy_ndb_identity", len({r["NDB_number"] for r in fd143_legacy.values()}) == len(fd143_legacy)
             and all(r["NDB_number"].strip() for r in fd143_legacy.values()), "nonempty unique NDB numbers", "nonempty unique NDB numbers")
fd143_check("food_data_type", {r["data_type"] for r in fd143_foods.values()} == {"sr_legacy_food"},
             sorted({r["data_type"] for r in fd143_foods.values()}), ["sr_legacy_food"])
fd143_check("food_description", all(r["description"].strip() for r in fd143_foods.values()), "all nonempty", "all nonempty")
fd143_check("food_category_links", all(r["food_category_id"] in fd143_categories for r in fd143_foods.values()), "all matched", "all matched")
fd143_check("derivation_source_links", all(not r["source_id"] or r["source_id"] in fd143_nutrient_sources for r in fd143_derivations.values()),
             "nonblank references matched", "nonblank references matched")

# Exact definitions, not name matching or an imported Foundation energy policy.
fd143_display_specs = [
    {"display_concept": "energy", "nutrient_id": "1008", "expected_name": "Energy", "expected_unit": "KCAL", "expected_nutrient_nbr": "208"},
    {"display_concept": "protein", "nutrient_id": "1003", "expected_name": "Protein", "expected_unit": "G", "expected_nutrient_nbr": "203"},
    {"display_concept": "total_fat", "nutrient_id": "1004", "expected_name": "Total lipid (fat)", "expected_unit": "G", "expected_nutrient_nbr": "204"},
    {"display_concept": "carbohydrate", "nutrient_id": "1005", "expected_name": "Carbohydrate, by difference", "expected_unit": "G", "expected_nutrient_nbr": "205"},
    {"display_concept": "dietary_fiber", "nutrient_id": "1079", "expected_name": "Fiber, total dietary", "expected_unit": "G", "expected_nutrient_nbr": "291"},
]
fd143_selected_ids = {r["nutrient_id"] for r in fd143_display_specs}
for spec in fd143_display_specs:
    definition = fd143_defs.get(spec["nutrient_id"], {})
    fd143_check("display_definition:" + spec["nutrient_id"], definition.get("name") == spec["expected_name"]
                 and definition.get("unit_name") == spec["expected_unit"]
                 and definition.get("nutrient_nbr") == spec["expected_nutrient_nbr"], definition, spec)


def fd143_numeric(value):
    if value.strip().casefold() in {"", "none", "null", "na", "n/a"}:
        return "missing", None
    try:
        number = Decimal(value)
    except InvalidOperation:
        return "invalid_numeric", None
    if not number.is_finite():
        return "nonfinite", None
    return ("negative" if number < 0 else "zero" if number == 0 else "positive"), number


fd143_stats = defaultdict(Counter)
fd143_nutrient_ids, fd143_pairs, fd143_selected_values = set(), set(), defaultdict(dict)
fd143_joined = fd143_rows_seen = fd143_duplicate_ids = fd143_duplicate_pairs = 0
fd143_orphan_foods = fd143_orphan_nutrients = fd143_orphan_derivations = 0
fd143_missing_derivation = 0
for row in fd143_rows("food_nutrient.csv", ("id", "fdc_id", "nutrient_id", "amount", "derivation_id")):
    fd143_rows_seen += 1
    fd143_require(bool(row["id"].strip()), "Blank food-nutrient record ID.")
    fd143_duplicate_ids += row["id"] in fd143_nutrient_ids
    fd143_nutrient_ids.add(row["id"])
    pair = row["fdc_id"] + ":" + row["nutrient_id"]
    fd143_duplicate_pairs += pair in fd143_pairs
    fd143_pairs.add(pair)
    matched_food = row["fdc_id"] in fd143_foods
    matched_nutrient = row["nutrient_id"] in fd143_defs
    fd143_orphan_foods += not matched_food
    fd143_orphan_nutrients += not matched_nutrient
    fd143_orphan_derivations += bool(row["derivation_id"]) and row["derivation_id"] not in fd143_derivations
    fd143_missing_derivation += not row["derivation_id"]
    fd143_joined += matched_food and matched_nutrient
    state, value = fd143_numeric(row["amount"])
    fd143_stats[row["nutrient_id"]][state] += 1
    if state not in {"positive", "zero"}:
        fd143_anomalies.append({"table": "food_nutrient.csv", "record_id": row["id"], "fdc_id": row["fdc_id"],
                                "nutrient_id": row["nutrient_id"], "field": "amount", "raw_value": row["amount"],
                                "finding": state, "handling": "retain_source_value; exclude_affected_nutrient_from_numeric_display"})
    if row["nutrient_id"] in fd143_selected_ids:
        fd143_selected_values[row["fdc_id"]][row["nutrient_id"]] = {"record_id": row["id"], "amount": row["amount"],
            "numeric_state": state, "derivation_id": row["derivation_id"]}
for name, count in {"duplicate_nutrient_record_ids": fd143_duplicate_ids, "duplicate_food_nutrient_pairs": fd143_duplicate_pairs,
                    "orphan_nutrient_food_refs": fd143_orphan_foods, "orphan_nutrient_definition_refs": fd143_orphan_nutrients,
                    "orphan_nonblank_derivation_refs": fd143_orphan_derivations}.items():
    fd143_check(name, count == 0, count, 0)
fd143_check("nutrient_join_cardinality", fd143_joined == fd143_rows_seen, fd143_joined, fd143_rows_seen)
# Free the largest temporary uniqueness sets; no source records are discarded.
del fd143_nutrient_ids, fd143_pairs

fd143_nutrient_quality = []
for nutrient_id, definition in sorted(fd143_defs.items(), key=lambda pair: int(pair[0])):
    counts = fd143_stats[nutrient_id]
    fd143_nutrient_quality.append({"nutrient_id": nutrient_id, "nutrient_name": definition["name"],
        "unit_name": definition["unit_name"], "nutrient_nbr": definition["nutrient_nbr"], "recorded_rows": sum(counts.values()),
        "positive_rows": counts["positive"], "zero_rows": counts["zero"], "negative_rows": counts["negative"],
        "missing_amount_rows": counts["missing"], "invalid_numeric_rows": counts["invalid_numeric"], "nonfinite_rows": counts["nonfinite"],
        "foods_without_record": len(fd143_foods) - sum(counts.values())})
fd143_dictionary = []
for spec in fd143_display_specs:
    nutrient_id = spec["nutrient_id"]
    fd143_dictionary.append({"display_concept": spec["display_concept"], "nutrient_id": nutrient_id,
        "nutrient_name": fd143_defs[nutrient_id]["name"], "unit_name": fd143_defs[nutrient_id]["unit_name"],
        "nutrient_nbr": fd143_defs[nutrient_id]["nutrient_nbr"], "reference_mass_g": 100, "reference_basis": "edible_food",
        "recorded_rows": sum(fd143_stats[nutrient_id].values()), "numeric_usable_rows": fd143_stats[nutrient_id]["positive"] + fd143_stats[nutrient_id]["zero"],
        "selection_policy": "use_exact_recorded_id; no_4_9_4_recalculation_or_Foundation_fallback" if nutrient_id == "1008" else "use_exact_recorded_id",
        "definition_source": fd143_table_refs["nutrient.csv"], "basis_evidence": "SR_LEGACY_BASIS;FDC_FIELD_BASIS"})

fd143_portions = fd143_index("food_portion.csv", "id", ("fdc_id", "amount", "measure_unit_id", "portion_description", "modifier", "gram_weight", "footnote"))
fd143_check("portion_food_refs", all(r["fdc_id"] in fd143_foods for r in fd143_portions.values()), "all matched", "all matched")
fd143_check("portion_unit_refs", all(r["measure_unit_id"] in fd143_units for r in fd143_portions.values()), "all matched", "all matched")
fd143_portion_review, fd143_portion_counts = [], Counter()
fd143_portion_foods = Counter()
for portion_id, row in fd143_portions.items():
    amount_state, amount = fd143_numeric(row["amount"])
    weight_state, weight = fd143_numeric(row["gram_weight"])
    label = row["modifier"].strip()
    placeholder = row["measure_unit_id"] == "9999"
    numeric_screen = amount_state == "positive" and weight_state == "positive" and bool(label)
    fd143_portion_counts["records"] += 1
    fd143_portion_counts["positive_gram_weight"] += weight_state == "positive"
    fd143_portion_counts["positive_amount"] += amount_state == "positive"
    fd143_portion_counts["household_numeric_screen_pass"] += numeric_screen
    fd143_portion_counts["unit_9999"] += placeholder
    fd143_portion_foods[row["fdc_id"]] += 1
    issues = []
    if amount_state != "positive":
        issues.append("amount_" + amount_state)
    if weight_state != "positive":
        issues.append("gram_weight_" + weight_state)
    if not label:
        issues.append("missing_sr_measure_description")
    if row["footnote"].strip():
        issues.append("source_footnote_requires_context")
    for field, raw, finding in (("amount", row["amount"], amount_state), ("gram_weight", row["gram_weight"], weight_state)):
        if finding != "positive":
            fd143_anomalies.append({"table": "food_portion.csv", "record_id": portion_id, "fdc_id": row["fdc_id"],
                "nutrient_id": "", "field": field, "raw_value": raw, "finding": finding,
                "handling": "retain; block_automatic_household_conversion; preserve_valid_food_facts_and_explicit_gram_scaling"})
    fd143_portion_review.append({"portion_id": portion_id, "fdc_id": row["fdc_id"], "amount_raw": row["amount"],
        "amount_state": amount_state, "gram_weight_raw": row["gram_weight"], "gram_weight_state": weight_state,
        "modifier": row["modifier"], "portion_description": row["portion_description"], "measure_unit_id": row["measure_unit_id"],
        "unit_name": fd143_units[row["measure_unit_id"]]["name"], "expected_sr_unit_placeholder": placeholder,
        "footnote": row["footnote"], "household_numeric_screen_pass": numeric_screen,
        "review_flags": ";".join(issues), "semantic_validation": "not_executed",
        "interpretation_policy": "use_exact_food_bound_source_measure; no_generic_volume_conversion",
        "automatic_household_conversion_enabled": False})

fd143_coverage, fd143_category_counts = [], Counter()
fd143_core_ids = {"1008", "1003", "1004", "1005"}
fd143_core_complete = fd143_fiber_absent = 0
for food_id, food in sorted(fd143_foods.items(), key=lambda pair: int(pair[0])):
    values = fd143_selected_values.get(food_id, {})
    usable = {key for key, row in values.items() if row["numeric_state"] in {"positive", "zero"}}
    complete = fd143_core_ids.issubset(usable)
    fd143_core_complete += complete
    fd143_fiber_absent += "1079" not in values
    fd143_category_counts[food["food_category_id"]] += 1
    fd143_coverage.append({"fdc_id": food_id, "NDB_number": fd143_legacy[food_id]["NDB_number"], "description": food["description"],
        "food_category_id": food["food_category_id"], "food_category": fd143_categories[food["food_category_id"]]["description"],
        "publication_date": food["publication_date"], "source_data_type": food["data_type"], "catalog_release": "2018-04",
        "usable_selected_nutrients": ";".join(sorted(usable, key=int)), "core_four_numeric_complete": complete,
        "fiber_record_present": "1079" in values, "fiber_numeric_usable": "1079" in usable, "portion_records": fd143_portion_foods[food_id],
        "source_record_url": "https://fdc.nal.usda.gov/food-details/" + food_id + "/nutrients",
        "prototype_selected": False})
fd143_category_coverage = [{"food_category_id": key, "code": row["code"], "description": row["description"],
                           "food_records": fd143_category_counts[key]} for key, row in sorted(fd143_categories.items(), key=lambda pair: int(pair[0]))]
fd143_count_manifest = list(fd143_rows("all_downloaded_table_record_counts.csv", ("Table", "Number of Records")))
for row in fd143_count_manifest:
    key = ("sr_legacy", row["Table"] + ".csv")
    actual = fd143_table_index.get(key, {}).get("rows")
    fd143_check("publisher_count:" + row["Table"], actual == int(row["Number of Records"]), actual, row["Number of Records"])

# Keep source evidence as original PDF bytes. The expected fingerprints identify
# the exact public documents inspected for the interpretation cards below.
fd143_reference_specs = [
    {"reference_id": "SR_LEGACY_DOC", "filename": "references/usda_sr_legacy_documentation.pdf",
     "url": "https://www.ars.usda.gov/ARSUserFiles/80400525/Data/SR-Legacy/SR-Legacy_Doc.pdf",
     "sha256": "fdc33442adb5707d448676e92b54f9adeaad63b6df267b683b5fe728307391cf"},
    {"reference_id": "FDC_FIELDS_DOC", "filename": "references/fdc_download_field_descriptions_oct2020.pdf",
     "url": "https://fdc.nal.usda.gov/docs/Download_Field_Descriptions_Oct2020.pdf",
     "sha256": "e21d658d4d3fa583aceaa78e23efd4e64dbf9010fd0210a87b94edbcfb56928a"},
]
fd143_reference_bytes, fd143_reference_records = {}, []
print("Fetching the two official documentation snapshots for the source-evidence package...", flush=True)
for spec in fd143_reference_specs:
    request = Request(spec["url"], headers={"User-Agent": "ChampionAthleticMap/step14-source-validation", "Accept-Encoding": "identity"})
    with urlopen(request, timeout=45) as response:
        data = response.read(5 * 1024 ** 2 + 1)
        fd143_require(response.status == 200 and urlsplit(response.geturl()).scheme == "https"
                        and len(data) <= 5 * 1024 ** 2 and data.startswith(b"%PDF-"), "Unexpected documentation download.")
        digest = hashlib.sha256(data).hexdigest()
        fd143_check("documentation_snapshot:" + spec["reference_id"], digest == spec["sha256"], digest, spec["sha256"])
        fd143_reference_bytes[spec["filename"]] = data
        fd143_reference_records.append({**spec, "bytes": len(data), "response_url": response.geturl(),
            "downloaded_at_utc": datetime.now(timezone.utc).isoformat(), "review_type": "assistant_primary_source_review",
            "interpretation_review_date": "2026-11-08", "interpretation_review_time_zone": "America/Los_Angeles",
            "fingerprint_role": "match_the_document_snapshot_reviewed; not_a_publisher_checksum"})

fd143_evidence_cards = [
    {"evidence_id": "SR_LEGACY_BASIS", "reference_id": "SR_LEGACY_DOC", "locator": "Section 4.5, Table 6; printed p.12 / PDF p.13",
     "short_excerpt": "Amount in 100 g, edible portion", "interpretation": "The SR amount describes 100 grams of edible food."},
    {"evidence_id": "FDC_FIELD_BASIS", "reference_id": "FDC_FIELDS_DOC", "locator": "food_nutrient.amount and nutrient.unit_name; PDF pp.5,10",
     "short_excerpt": "", "interpretation": "FDC nutrient amount uses a 100-gram basis; its unit comes from the nutrient definition."},
    {"evidence_id": "FDC_PORTION_MEANING", "reference_id": "FDC_FIELDS_DOC", "locator": "food_portion; PDF p.7",
     "short_excerpt": "", "interpretation": "Amount counts measure units; SR modifier carries the measure text. Gram weight belongs to the whole recorded measure. ID 9999 is expected for SR."},
]
fd143_metrics = {"source_food_records": len(fd143_foods), "nutrient_value_records": fd143_rows_seen,
    "nutrient_definitions": len(fd143_defs), "nutrients_with_source_values": len([k for k, v in fd143_stats.items() if sum(v.values())]),
    "joined_nutrient_rows": fd143_joined, "core_four_numeric_complete_foods": fd143_core_complete,
    "fiber_absent_foods": fd143_fiber_absent, "portion_records": len(fd143_portions), "foods_with_portions": len(fd143_portion_foods),
    "portion_numeric_screen_pass": fd143_portion_counts["household_numeric_screen_pass"],
    "portion_numeric_screen_fail": len(fd143_portions) - fd143_portion_counts["household_numeric_screen_pass"],
    "portion_expected_unit_9999": fd143_portion_counts["unit_9999"], "recorded_anomaly_rows": len(fd143_anomalies),
    "blank_derivation_references": fd143_missing_derivation, "prototype_selected_foods": 0,
    "alternative_energy_record_counts": {key: sum(fd143_stats[key].values()) for key in ("2047", "2048")}}
fd143_findings = [
    {"finding_id": "SR_FIBER_COVERAGE", "count": fd143_fiber_absent, "meaning": "Foods without a fiber record", "handling": "fiber unavailable; preserve other valid nutrient facts"},
    {"finding_id": "SR_PORTION_NUMERIC_REVIEW", "count": fd143_metrics["portion_numeric_screen_fail"], "meaning": "Portion rows failing positive amount/weight plus measure-description screening", "handling": "retain rows; block automatic household conversion; explicit edible-gram use remains separate"},
    {"finding_id": "SR_PORTION_PLACEHOLDER", "count": fd143_portion_counts["unit_9999"], "meaning": "Expected SR placeholder unit IDs", "handling": "use SR modifier and food-bound portion; this ID alone is not a defect"},
    {"finding_id": "SR_DERIVATION_NOT_RECORDED", "count": fd143_missing_derivation, "meaning": "Nutrient records without a derivation reference", "handling": "keep provenance field unavailable; do not call every nutrient a direct laboratory measurement"},
    {"finding_id": "FOUNDATION_SCOPE", "count": step14_report["foundation_metrics"]["negative_measurement_records"], "meaning": "Earlier Foundation records with a negative value", "handling": "retain Foundation finding separately; do not transfer it to SR records"},
]
fd143_table_inventory = [{"source_key": r["source_key"], "table_name": r["table_name"], "rows": r["rows"],
                         "columns": ";".join(r["columns"]), "sha256": r["sha256"], "archive_member": r["archive_member"]} for r in fd143_acq["tables"]]
fd143_tables = {}


def fd143_add_table(name, rows, empty_columns=()):
    columns = list(rows[0]) if rows else list(empty_columns)
    fd143_require(bool(columns) and all(set(row) == set(columns) for row in rows), "Inconsistent prepared report columns: " + name)
    fd143_tables[name] = {"columns": columns, "rows": rows}


fd143_add_table("usda_source_table_inventory.csv", fd143_table_inventory)
fd143_add_table("usda_join_quality_checks.csv", fd143_checks)
fd143_add_table("usda_nutrient_definition_map.csv", fd143_dictionary)
fd143_add_table("usda_nutrient_quality_by_definition.csv", fd143_nutrient_quality)
fd143_add_table("usda_food_nutrient_coverage.csv", fd143_coverage)
fd143_add_table("usda_category_coverage.csv", fd143_category_coverage)
fd143_add_table("usda_portion_numeric_review.csv", fd143_portion_review)
fd143_add_table("usda_numeric_anomalies.csv", fd143_anomalies, ("table", "record_id", "fdc_id", "nutrient_id", "field", "raw_value", "finding", "handling"))
fd143_add_table("usda_validation_findings.csv", fd143_findings)
fd143_contract = {"schema_version": "step14-food-reference-contract-proposal-1", "source_data_type": "sr_legacy_food",
    "catalog_release": "2018-04", "source_acquisition_producer": fd143_producer_ref, "reference_mass_g": 100,
    "reference_basis": "edible_food", "source_table_paths": fd143_table_refs, "display_definitions": fd143_dictionary,
    "energy_policy": "Return source nutrient 1008 in KCAL; retain derivation metadata; do not invent a universal Atwater method or recompute 4/9/4.",
    "missing_value_policy": "Absent or unusable nutrient stays unavailable; a valid recorded zero remains zero. Eligibility is field-specific.",
    "quantity_policy": {"edible_mass_g": "N100g * requested_grams / 100", "complete_source_portions": "mass = copies * source_gram_weight",
        "same_defined_measure_units": "mass = requested_units / positive_source_amount * source_gram_weight",
        "household_gates": "Match food and portion IDs; use exact measure text; require positive finite amount/weight and handle footnotes; no name-only or generic-volume conversion.",
        "nonpositive_amount": "Block automatic household counts unless row meaning is independently resolved; keep valid per-100g facts and explicit gram scaling."},
    "scope": "Food composition facts; no diet, intake target, supplement, or allergy-safety decisions.",
    "source_terms": {"url": "https://fdc.nal.usda.gov/api-guide/", "locator": "Licensing",
        "review_type": "assistant_official_source_review", "locally_archived": False,
        "policy": "Retain USDA attribution; no extra project permission gate is introduced."},
    "prototype_selection": "pending; full source screening is not a selected prototype reference",
    "lookup_ready": False, "runtime_lookup_implemented": False}
fd143_documents = {
    "usda_documentation_evidence.json": {"schema_version": "step14-documentation-evidence-1", "references": fd143_reference_records,
        "evidence_cards": fd143_evidence_cards, "note": "PDF fingerprints verify the reviewed snapshots; this code does not parse PDFs or perform an independent human review."},
    "usda_reference_contract_proposal.json": fd143_contract,
    "usda_source_register_supplement.json": {"schema_version": "step14-source-register-supplement-1", "manifest_modified": False,
        "acquisition_producer": fd143_producer_ref, "acquisition_sha256": fd143_digest,
        "sources": fd143_acq["sources"], "documentation": fd143_reference_records,
        "lineage_note": "New SR inputs do not establish the provenance of older Foundation or cleaned supporting files."},
    "step14_validation_review_handoff.json": {"schema_version": "step14-validation-review-handoff-1",
        "source_acquisition_producer": fd143_producer_ref, "metrics": fd143_metrics,
        "source_and_join_validation": "passed", "measurement_basis": "official_documentation_snapshots_verified",
        "prototype_selection": "pending", "lookup_calculation_tests": "not_executed", "lookup_ready": False,
        "open_actions": ["Select and freeze prototype food IDs with a transparent scope policy.",
                         "Prepare selected food/nutrient tables with field-specific eligibility.",
                         "Validate supported calculation cases and source comparisons before the Step 24 handoff.",
                         "Keep flagged household measures unavailable for automatic count conversion."]},
}


def fd143_csv_bytes(table):
    handle = io.StringIO(newline="")
    writer = csv.DictWriter(handle, fieldnames=table["columns"], lineterminator="\n", extrasaction="raise")
    writer.writeheader()
    writer.writerows(table["rows"])
    return handle.getvalue().encode("utf-8")


fd143_payload_bytes = {name: fd143_csv_bytes(table) for name, table in fd143_tables.items()}
fd143_payload_bytes.update({name: fd142_json_bytes(doc) for name, doc in fd143_documents.items()})
fd143_payload_bytes.update(fd143_reference_bytes)
fd143_fingerprints = {name: hashlib.sha256(data).hexdigest() for name, data in fd143_payload_bytes.items()}
for reference, digest in fd143_inputs.items():
    fd143_require(fd14_sha(fd14_path(reference)) == digest, "An inspected input changed during validation: " + reference)
step14_report.update(status="source_validation_prepared_selection_pending", validation_prepared=True,
                     validation_metrics=fd143_metrics, validation_artifact_fingerprints=fd143_fingerprints, lookup_ready=False)
print("Step 14-3 — SR source relationships, nutrient definitions, and quality findings prepared")
print(f"Source foods / nutrient rows / nutrient definitions: {len(fd143_foods)} / {fd143_rows_seen} / {len(fd143_defs)}")
print(f"Joined nutrient rows / duplicate food-nutrient pairs: {fd143_joined} / {fd143_duplicate_pairs}")
print(f"Finite nonnegative core-four foods / missing fiber records: {fd143_core_complete} / {fd143_fiber_absent}")
print("Energy display definition: 1008 / Energy / KCAL; no automatic Foundation energy fallback")
print("Alternative 2047 / 2048 source record counts:", fd143_metrics["alternative_energy_record_counts"])
print(f"Portion records / foods with portions: {len(fd143_portions)} / {len(fd143_portion_foods)}")
print(f"Household numeric screening pass / flagged: {fd143_metrics['portion_numeric_screen_pass']} / {fd143_metrics['portion_numeric_screen_fail']}")
print(f"Expected SR unit placeholders: {fd143_portion_counts['unit_9999']} | Placeholders alone are not conversion failures")
print("Food publication dates:", dict(Counter(r["publication_date"] for r in fd143_foods.values())))
print(f"Checks passed / unchanged inspected inputs: {len(fd143_checks)} / {len(fd143_inputs)}")
print(f"Prepared evidence tables / documents / reference PDFs: {len(fd143_tables)} / {len(fd143_documents)} / {len(fd143_reference_bytes)}")
print("Official documentation snapshots matched | Source values changed or imputed by this cell: 0")
print("Prototype foods selected / project files written / lookup enabled: 0 / 0 / No")
print("Next: run Step 14-4 to save and verify this source-validation bundle.")
```

```python
# Step 14-4 — Save and verify our source-validation evidence
# This saves reports and references; selected food exports and lookup use remain pending.
import csv
import hashlib
import io
import json
import os
import re
from pathlib import Path, PureWindowsPath

if not globals().get("step14_report", {}).get("validation_prepared") or not callable(globals().get("fd143_require")):
    raise RuntimeError("Complete Step 14-3 successfully before saving its validation reports.")
step14_report.update(status="saving_source_validation", validation_saved=False)


def fd144_relative(name):
    fd143_require(isinstance(name, str) and bool(name), "An output name is empty or invalid.")
    normalized = name.replace("\\", "/")
    path = Path(normalized)
    fd143_require(not path.is_absolute() and not PureWindowsPath(name).drive
                  and all(part not in {"", ".", ".."} for part in normalized.split("/")),
                  "Expected a safe bundle-relative output name: " + name)
    return normalized


def fd144_csv_bytes(table):
    fd143_require(isinstance(table, dict) and isinstance(table.get("columns"), list)
                  and isinstance(table.get("rows"), list), "A validation table lacks columns or rows.")
    columns, rows = table["columns"], table["rows"]
    fd143_require(bool(columns) and all(isinstance(field, str) and field for field in columns)
                  and len(columns) == len(set(columns)), "A validation table has invalid column names.")
    fd143_require(all(isinstance(row, dict) and set(row) == set(columns) for row in rows),
                  "Validation table row fields disagree with its declared columns.")
    buffer = io.StringIO(newline="")
    writer = csv.DictWriter(buffer, fieldnames=columns, extrasaction="raise", lineterminator="\n")
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode("utf-8")


def fd144_json_bytes(value):
    return (json.dumps(value, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False) + "\n").encode("utf-8")


fd143_require(isinstance(fd143_inputs, dict) and bool(fd143_inputs), "The validation input inventory is empty.")
fd143_require(isinstance(fd143_tables, dict) and isinstance(fd143_documents, dict)
              and isinstance(fd143_reference_bytes, dict) and isinstance(fd143_metrics, dict),
              "A prepared validation collection has an unexpected type.")
fd143_require(isinstance(fd143_checks, list) and bool(fd143_checks)
              and all(isinstance(row, dict) and row.get("status") == "pass" for row in fd143_checks),
              "The prepared validation checks contain missing or failed results.")
fd144_inputs = dict(fd143_inputs)


def fd144_verify_inputs():
    for reference, expected in fd144_inputs.items():
        fd143_require(isinstance(expected, str) and re.fullmatch(r"[0-9a-f]{64}", expected),
                      "An input has an invalid SHA-256: " + str(reference))
        path = fd14_path(reference)
        fd143_require(path.is_file() and fd14_sha(path) == expected,
                      "A validation input changed or is missing: " + reference)


fd144_verify_inputs()
fd144_acquisition_ref = step14_report.get("acquisition_producer", "")
fd143_require(isinstance(fd144_acquisition_ref, str) and fd144_acquisition_ref in fd144_inputs,
              "The validation inventory must include the current source-acquisition producer.")
fd144_payload_names, fd144_artifacts, fd144_portable_names = {}, {}, set()


def fd144_register(name, content, metadata):
    relative = fd144_relative(name)
    fd143_require(relative.casefold() not in fd144_portable_names,
                  "Two prepared outputs share a portable filename: " + relative)
    fd143_require(relative.casefold() != "step14_source_validation_summary.json",
                  "A prepared artifact would replace the validation producer.")
    fd144_portable_names.add(relative.casefold())
    fd144_payload_names[relative] = content
    fd144_artifacts[relative] = {"bytes": len(content), "sha256": hashlib.sha256(content).hexdigest(), **metadata}


for fd144_name, fd144_table in fd143_tables.items():
    fd144_register(fd144_name, fd144_csv_bytes(fd144_table),
                   {"kind": "table", "format": "csv", "rows": len(fd144_table["rows"]),
                    "columns": list(fd144_table["columns"])})
for fd144_name, fd144_document in fd143_documents.items():
    fd143_require(isinstance(fd144_document, dict), "A prepared validation document is not a JSON object.")
    fd144_register(fd144_name, fd144_json_bytes(fd144_document), {"kind": "document", "format": "json"})
for fd144_name, fd144_reference in fd143_reference_bytes.items():
    fd143_require(isinstance(fd144_reference, bytes) and bool(fd144_reference),
                  "A prepared reference document has missing or invalid bytes.")
    fd144_register(fd144_name, fd144_reference, {"kind": "reference_document", "format": "pdf"})
fd143_require(bool(fd144_payload_names), "The validation bundle contains no prepared artifacts.")
fd143_require(isinstance(fd143_fingerprints, dict), "Prepared validation fingerprints are unavailable.")
fd144_expected_fingerprints = {}
for fd144_name, fd144_hash in fd143_fingerprints.items():
    fd144_relative_name = fd144_relative(fd144_name)
    fd143_require(fd144_relative_name not in fd144_expected_fingerprints,
                  "Prepared fingerprints contain duplicate output paths.")
    fd144_expected_fingerprints[fd144_relative_name] = fd144_hash
fd143_require({name: metadata["sha256"] for name, metadata in fd144_artifacts.items()} == fd144_expected_fingerprints,
              "Prepared validation reports or reference bytes changed after Step 14-3.")

# Stable serialization makes identical reruns select the same directory. These
# checks describe the source-validation stage, not an enabled lookup reference.
fd144_summary = {
    "schema_version": "step14-source-validation-producer-1", "recipe_version": "source-validation-save-0.1.0",
    "status": "source_validation_complete_selection_pending",
    "source_acquisition_producer": fd144_acquisition_ref,
    "source_acquisition_digest": fd144_inputs[fd144_acquisition_ref],
    "intended_source": "SR Legacy April 2018",
    "input_hashes": dict(sorted(fd144_inputs.items())), "metrics": fd143_metrics,
    "verified_checks": fd143_checks, "artifacts": dict(sorted(fd144_artifacts.items())),
    "lookup_ready": False, "full_step14_complete": False,
    "scope": {"processed_food_outputs_created": 0, "selected_food_subset_created": False,
              "food_lookup_enabled": False, "existing_inputs_changed": 0,
              "source_manifest_changed": False, "step11_reports_changed": False},
    "methods": {
        "serialization": "UTF-8 CSV with declared column order and LF record endings; sorted-key JSON with finite numbers.",
        "source_checks": "Only the identities, relationships, nutrient definitions, and other conditions recorded in verified_checks are claimed as checked.",
        "reference_basis": "Saved USDA reference documents provide measurement-basis evidence; they do not independently audit every food or nutrient value.",
        "saving": "Exclusive file creation; matching complete bundles are reused; the producer is published after artifact and input verification.",
    },
    "limitations": [
        "Source-validation findings are distinct from food selection, operation-level nutrient eligibility, and supported portion conversion.",
        "No selected processed-food export or active lookup tool is created by this cell.",
        "Only the listed artifacts belong to this validation bundle.",
    ],
}
fd144_summary_bytes = fd144_json_bytes(fd144_summary)
fd144_digest = hashlib.sha256(fd144_summary_bytes).hexdigest()
fd144_bundle_ref = "results/usda_preparation/source_validation_" + fd144_digest[:16]
fd144_bundle = fd14_path(fd144_bundle_ref)
fd144_payloads = {}
for fd144_name, fd144_content in fd144_payload_names.items():
    fd144_target = fd14_path(fd144_bundle_ref + "/" + fd144_name)
    fd143_require(fd144_target.is_relative_to(fd144_bundle), "A validation artifact leaves its bundle.")
    fd144_payloads[fd144_target] = fd144_content
fd144_producer = fd144_bundle / "step14_source_validation_summary.json"
fd144_payloads[fd144_producer] = fd144_summary_bytes
fd143_require(not set(fd144_payloads).intersection({fd14_path(reference) for reference in fd144_inputs}),
              "A validation output overlaps an inspected input.")

# Preserve a partial or edited prior bundle for inspection. A matching complete
# bundle can be reused without replacing files or repeating source acquisition.
fd144_exists = [path.exists() for path in fd144_payloads]
if any(fd144_exists):
    fd143_require(all(fd144_exists), "A partial validation bundle already exists; preserve and inspect it before rerunning.")
    for fd144_target, fd144_content in fd144_payloads.items():
        fd143_require(fd144_target.is_file() and fd144_target.read_bytes() == fd144_content,
                      "An existing validation artifact differs: " + fd144_target.relative_to(fd14_root).as_posix())
fd144_verify_inputs()
fd144_created = []
fd144_bundle.mkdir(parents=True, exist_ok=True)
try:
    for fd144_target, fd144_content in fd144_payloads.items():
        if fd144_target == fd144_producer:
            for fd144_name, fd144_expected in fd144_payload_names.items():
                fd143_require(fd14_path(fd144_bundle_ref + "/" + fd144_name).read_bytes() == fd144_expected,
                              "A validation artifact changed before publishing its producer.")
            fd144_verify_inputs()
        fd143_require(fd144_target.resolve() == fd144_target and fd144_target.is_relative_to(fd144_bundle),
                      "A validation destination changed its location during saving.")
        fd144_target.parent.mkdir(parents=True, exist_ok=True)
        if fd144_target.exists():
            fd143_require(fd144_target.is_file() and fd144_target.read_bytes() == fd144_content,
                          "A validation artifact changed during saving.")
            continue
        try:
            with fd144_target.open("xb") as fd144_handle:
                fd144_created.append(fd144_target)
                fd144_handle.write(fd144_content)
                fd144_handle.flush()
                os.fsync(fd144_handle.fileno())
        except FileExistsError:
            fd143_require(fd144_target.is_file() and fd144_target.read_bytes() == fd144_content,
                          "A validation destination appeared with different content.")
    for fd144_target, fd144_content in fd144_payloads.items():
        fd143_require(fd144_target.is_file() and fd144_target.read_bytes() == fd144_content,
                      "Saved validation verification failed: " + fd144_target.name)
    fd144_verify_inputs()
except Exception:
    fd144_retained, fd144_cleanup_errors = [], []
    for fd144_target in reversed(fd144_created):
        try:
            if fd144_target.is_file():
                if fd144_target.read_bytes() == fd144_payloads[fd144_target]:
                    fd144_target.unlink()
                else:
                    fd144_retained.append(fd144_target.relative_to(fd14_root).as_posix())
        except OSError as fd144_cleanup_error:
            if fd144_target.relative_to(fd14_root).as_posix() not in fd144_retained:
                fd144_retained.append(fd144_target.relative_to(fd14_root).as_posix())
            fd144_cleanup_errors.append(str(fd144_cleanup_error))
    step14_report.update(status="source_validation_save_failed", validation_saved=False,
                         validation_save_attempt_files_retained=fd144_retained,
                         validation_cleanup_errors=fd144_cleanup_errors)
    raise

step14_report.update(status="source_validation_complete_selection_pending", validation_saved=True,
                     validation_bundle=fd144_bundle_ref, validation_producer=fd144_producer.relative_to(fd14_root).as_posix(),
                     validation_digest=fd144_digest, validation_summary=fd144_summary,
                     validation_files_created=len(fd144_created), validation_files_reused=len(fd144_payloads) - len(fd144_created),
                     project_files_written=len(fd144_created), lookup_ready=False, full_step14_complete=False)
print("Step 14-4 — source-validation bundle saved and verified")
print("Bundle directory:", fd144_bundle_ref)
print("Producer record:", step14_report["validation_producer"])
print(f"Saved tables / JSON documents / reference PDFs / producer records: {len(fd143_tables)} / {len(fd143_documents)} / {len(fd143_reference_bytes)} / 1")
print(f"Files created / identical files reused: {len(fd144_created)} / {len(fd144_payloads) - len(fd144_created)}")
print(f"Validation checks passed / unchanged inspected inputs: {len(fd143_checks)} / {len(fd144_inputs)}")
print("Source-table and join findings: recorded in the saved checks and reports")
print("Measurement-basis evidence: saved USDA references; selected-food calculations remain pending")
print("Processed food exports created: 0 | Source register and earlier reports unchanged")
print("Food selection / quantity-conversion validation / lookup readiness: pending / pending / not established")
print("Next: review these outputs before selecting the reproducible food subset and preparing eligible food facts.")
```

```python
# Step 14-5 — Prepare a fixed food reference and execute calculation checks
# Uses verified local sources only. Step 14-6 saves the prepared snapshot.
if not globals().get("step14_report", {}).get("validation_saved"):
    raise RuntimeError("Complete Step 14-4 successfully before Step 14-5.")
step14_report.update(status="preparing_food_reference", reference_prepared=False, reference_saved=False)
fd145_checks = []


def fd145_require(condition, message):
    if not condition:
        step14_report.update(status="food_reference_preparation_blocked", reference_prepared=False, reason=message)
        raise RuntimeError("Step 14-5: " + message)


def fd145_check(name, condition, details):
    fd145_checks.append({"check_name": name, "status": "pass" if condition else "failed", "details": details})
    fd145_require(condition, name + ": " + details)


# These embedded inputs are frozen with this recipe, then saved as separate files.
FD145_SELECTION_CSV = r"""fdc_id,expected_description,selection_family,selection_reason
171265,"Milk, whole, 3.25% milkfat, with added vitamin D",dairy_eggs,"Familiar dairy or egg identity, retaining the exact milk-fat, cheese style, or egg preparation state for factual lookup."
171267,"Milk, reduced fat, fluid, 2% milkfat, with added vitamin A and vitamin D",dairy_eggs,"Familiar dairy or egg identity, retaining the exact milk-fat, cheese style, or egg preparation state for factual lookup."
171269,"Milk, nonfat, fluid, with added vitamin A and vitamin D (fat free or skim)",dairy_eggs,"Familiar dairy or egg identity, retaining the exact milk-fat, cheese style, or egg preparation state for factual lookup."
171284,"Yogurt, plain, whole milk",dairy_eggs,"Familiar dairy or egg identity, retaining the exact milk-fat, cheese style, or egg preparation state for factual lookup."
170894,"Yogurt, Greek, plain, nonfat (Includes foods for USDA's Food Distribution Program)",dairy_eggs,"Familiar dairy or egg identity, retaining the exact milk-fat, cheese style, or egg preparation state for factual lookup."
173414,"Cheese, cheddar (Includes foods for USDA's Food Distribution Program)",dairy_eggs,"Familiar dairy or egg identity, retaining the exact milk-fat, cheese style, or egg preparation state for factual lookup."
170845,"Cheese, mozzarella, whole milk",dairy_eggs,"Familiar dairy or egg identity, retaining the exact milk-fat, cheese style, or egg preparation state for factual lookup."
172182,"Cheese, cottage, lowfat, 2% milkfat",dairy_eggs,"Familiar dairy or egg identity, retaining the exact milk-fat, cheese style, or egg preparation state for factual lookup."
171287,"Egg, whole, raw, fresh",dairy_eggs,"Familiar dairy or egg identity, retaining the exact milk-fat, cheese style, or egg preparation state for factual lookup."
173424,"Egg, whole, cooked, hard-boiled",dairy_eggs,"Familiar dairy or egg identity, retaining the exact milk-fat, cheese style, or egg preparation state for factual lookup."
171688,"Apples, raw, with skin (Includes foods for USDA's Food Distribution Program)",fruits,"Familiar fruit identity, retaining its exact variety, edible-part description, and raw or dried state for factual lookup."
171689,"Apples, raw, without skin",fruits,"Familiar fruit identity, retaining its exact variety, edible-part description, and raw or dried state for factual lookup."
173944,"Bananas, raw",fruits,"Familiar fruit identity, retaining its exact variety, edible-part description, and raw or dried state for factual lookup."
169097,"Oranges, raw, all commercial varieties",fruits,"Familiar fruit identity, retaining its exact variety, edible-part description, and raw or dried state for factual lookup."
167762,"Strawberries, raw",fruits,"Familiar fruit identity, retaining its exact variety, edible-part description, and raw or dried state for factual lookup."
171711,"Blueberries, raw",fruits,"Familiar fruit identity, retaining its exact variety, edible-part description, and raw or dried state for factual lookup."
174683,"Grapes, red or green (European type, such as Thompson seedless), raw",fruits,"Familiar fruit identity, retaining its exact variety, edible-part description, and raw or dried state for factual lookup."
167765,"Watermelon, raw",fruits,"Familiar fruit identity, retaining its exact variety, edible-part description, and raw or dried state for factual lookup."
169928,"Peaches, yellow, raw",fruits,"Familiar fruit identity, retaining its exact variety, edible-part description, and raw or dried state for factual lookup."
169118,"Pears, raw",fruits,"Familiar fruit identity, retaining its exact variety, edible-part description, and raw or dried state for factual lookup."
171705,"Avocados, raw, all commercial varieties",fruits,"Familiar fruit identity, retaining its exact variety, edible-part description, and raw or dried state for factual lookup."
168191,"Dates, medjool",fruits,"Familiar fruit identity, retaining its exact variety, edible-part description, and raw or dried state for factual lookup."
170379,"Broccoli, raw",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
169967,"Broccoli, cooked, boiled, drained, without salt",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
170393,"Carrots, raw",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
170394,"Carrots, cooked, boiled, drained, without salt",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
168462,"Spinach, raw",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
168463,"Spinach, cooked, boiled, drained, without salt",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
170026,"Potatoes, flesh and skin, raw",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
170093,"Potatoes, baked, flesh and skin, without salt",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
168482,"Sweet potato, raw, unprepared (Includes foods for USDA's Food Distribution Program)",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
168483,"Sweet potato, cooked, baked in skin, flesh, without salt",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
170457,"Tomatoes, red, ripe, raw, year round average",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
168409,"Cucumber, with peel, raw",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
169247,"Lettuce, cos or romaine, raw",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
170108,"Peppers, sweet, red, raw",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
170427,"Peppers, sweet, green, raw",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
170000,"Onions, raw",vegetables,"Familiar vegetable identity, retaining raw/cooked, edible-part, and preparation distinctions for factual lookup."
168877,"Rice, white, long-grain, regular, raw, enriched",grains_pasta,"Familiar grain, pasta, oat cereal, or bread identity, retaining dry/cooked and enrichment distinctions for factual lookup."
168878,"Rice, white, long-grain, regular, enriched, cooked",grains_pasta,"Familiar grain, pasta, oat cereal, or bread identity, retaining dry/cooked and enrichment distinctions for factual lookup."
169703,"Rice, brown, long-grain, raw (Includes foods for USDA's Food Distribution Program)",grains_pasta,"Familiar grain, pasta, oat cereal, or bread identity, retaining dry/cooked and enrichment distinctions for factual lookup."
169704,"Rice, brown, long-grain, cooked (Includes foods for USDA's Food Distribution Program)",grains_pasta,"Familiar grain, pasta, oat cereal, or bread identity, retaining dry/cooked and enrichment distinctions for factual lookup."
168874,"Quinoa, uncooked",grains_pasta,"Familiar grain, pasta, oat cereal, or bread identity, retaining dry/cooked and enrichment distinctions for factual lookup."
168917,"Quinoa, cooked",grains_pasta,"Familiar grain, pasta, oat cereal, or bread identity, retaining dry/cooked and enrichment distinctions for factual lookup."
169736,"Pasta, dry, enriched",grains_pasta,"Familiar grain, pasta, oat cereal, or bread identity, retaining dry/cooked and enrichment distinctions for factual lookup."
169737,"Pasta, cooked, enriched, without added salt",grains_pasta,"Familiar grain, pasta, oat cereal, or bread identity, retaining dry/cooked and enrichment distinctions for factual lookup."
173904,"Cereals, oats, regular and quick, not fortified, dry",grains_pasta,"Familiar grain, pasta, oat cereal, or bread identity, retaining dry/cooked and enrichment distinctions for factual lookup."
173905,"Cereals, oats, regular and quick, unenriched, cooked with water (includes boiling and microwaving), without salt",grains_pasta,"Familiar grain, pasta, oat cereal, or bread identity, retaining dry/cooked and enrichment distinctions for factual lookup."
172688,"Bread, whole-wheat, commercially prepared",grains_pasta,"Familiar grain, pasta, oat cereal, or bread identity, retaining dry/cooked and enrichment distinctions for factual lookup."
174924,"Bread, white, commercially prepared (includes soft bread crumbs)",grains_pasta,"Familiar grain, pasta, oat cereal, or bread identity, retaining dry/cooked and enrichment distinctions for factual lookup."
172420,"Lentils, raw",legumes_soy,"Familiar bean, lentil, chickpea, or soy-food identity, retaining dry/cooked or processing distinctions for factual lookup."
172421,"Lentils, mature seeds, cooked, boiled, without salt",legumes_soy,"Familiar bean, lentil, chickpea, or soy-food identity, retaining dry/cooked or processing distinctions for factual lookup."
173756,"Chickpeas (garbanzo beans, bengal gram), mature seeds, raw",legumes_soy,"Familiar bean, lentil, chickpea, or soy-food identity, retaining dry/cooked or processing distinctions for factual lookup."
173757,"Chickpeas (garbanzo beans, bengal gram), mature seeds, cooked, boiled, without salt",legumes_soy,"Familiar bean, lentil, chickpea, or soy-food identity, retaining dry/cooked or processing distinctions for factual lookup."
173734,"Beans, black, mature seeds, raw",legumes_soy,"Familiar bean, lentil, chickpea, or soy-food identity, retaining dry/cooked or processing distinctions for factual lookup."
173735,"Beans, black, mature seeds, cooked, boiled, without salt",legumes_soy,"Familiar bean, lentil, chickpea, or soy-food identity, retaining dry/cooked or processing distinctions for factual lookup."
175193,"Beans, kidney, all types, mature seeds, raw",legumes_soy,"Familiar bean, lentil, chickpea, or soy-food identity, retaining dry/cooked or processing distinctions for factual lookup."
173740,"Beans, kidney, all types, mature seeds, cooked, boiled, without salt",legumes_soy,"Familiar bean, lentil, chickpea, or soy-food identity, retaining dry/cooked or processing distinctions for factual lookup."
172475,"Tofu, raw, firm, prepared with calcium sulfate",legumes_soy,"Familiar bean, lentil, chickpea, or soy-food identity, retaining dry/cooked or processing distinctions for factual lookup."
174272,Tempeh,legumes_soy,Familiar fermented-soy food identity distinct from tofu and mature beans; retained without requiring complete fiber information.
170567,"Nuts, almonds",nuts_seeds,"Familiar nut, peanut, seed, or peanut-butter identity, retaining the exact source preparation rather than merging forms."
170187,"Nuts, walnuts, english",nuts_seeds,"Familiar nut, peanut, seed, or peanut-butter identity, retaining the exact source preparation rather than merging forms."
170162,"Nuts, cashew nuts, raw",nuts_seeds,"Familiar nut, peanut, seed, or peanut-butter identity, retaining the exact source preparation rather than merging forms."
170184,"Nuts, pistachio nuts, raw",nuts_seeds,"Familiar nut, peanut, seed, or peanut-butter identity, retaining the exact source preparation rather than merging forms."
172430,"Peanuts, all types, raw",nuts_seeds,"Familiar nut, peanut, seed, or peanut-butter identity, retaining the exact source preparation rather than merging forms."
172470,"Peanut butter, smooth style, without salt",nuts_seeds,"Familiar nut, peanut, seed, or peanut-butter identity, retaining the exact source preparation rather than merging forms."
170554,"Seeds, chia seeds, dried",nuts_seeds,"Familiar nut, peanut, seed, or peanut-butter identity, retaining the exact source preparation rather than merging forms."
169414,"Seeds, flaxseed",nuts_seeds,"Familiar nut, peanut, seed, or peanut-butter identity, retaining the exact source preparation rather than merging forms."
171077,"Chicken, broiler or fryers, breast, skinless, boneless, meat only, raw",poultry,"Familiar chicken or turkey cut, with paired raw/roasted meat-only records retained as separate factual identities."
171477,"Chicken, broilers or fryers, breast, meat only, cooked, roasted",poultry,"Familiar chicken or turkey cut, with paired raw/roasted meat-only records retained as separate factual identities."
173627,"Chicken, broilers or fryers, dark meat, thigh, meat only, raw",poultry,"Familiar chicken or turkey cut, with paired raw/roasted meat-only records retained as separate factual identities."
172388,"Chicken, broilers or fryers, thigh, meat only, cooked, roasted",poultry,"Familiar chicken or turkey cut, with paired raw/roasted meat-only records retained as separate factual identities."
173614,"Chicken, broilers or fryers, dark meat, drumstick, meat only, raw",poultry,"Familiar chicken or turkey cut, with paired raw/roasted meat-only records retained as separate factual identities."
172376,"Chicken, broilers or fryers, dark meat, drumstick, meat only, cooked, roasted",poultry,"Familiar chicken or turkey cut, with paired raw/roasted meat-only records retained as separate factual identities."
171098,"Turkey, whole, breast, meat only, raw",poultry,"Familiar chicken or turkey cut, with paired raw/roasted meat-only records retained as separate factual identities."
171496,"Turkey, whole, breast, meat only, cooked, roasted",poultry,"Familiar chicken or turkey cut, with paired raw/roasted meat-only records retained as separate factual identities."
174030,"Beef, ground, 90% lean meat / 10% fat, raw",beef_pork,"Familiar ground-beef or pork cut, preserving the source fat/cut and raw/cooked identity without ranking nutritional value."
171794,"Beef, ground, 90% lean meat / 10% fat, crumbles, cooked, pan-browned",beef_pork,"Familiar ground-beef or pork cut, preserving the source fat/cut and raw/cooked identity without ranking nutritional value."
171796,"Beef, ground, 85% lean meat / 15% fat, raw (Includes foods for USDA's Food Distribution Program)",beef_pork,"Familiar ground-beef or pork cut, preserving the source fat/cut and raw/cooked identity without ranking nutritional value."
174034,"Beef, ground, 85% lean meat / 15% fat, crumbles, cooked, pan-browned",beef_pork,"Familiar ground-beef or pork cut, preserving the source fat/cut and raw/cooked identity without ranking nutritional value."
168312,"Pork, fresh, loin, tenderloin, separable lean and fat, raw",beef_pork,"Familiar ground-beef or pork cut, preserving the source fat/cut and raw/cooked identity without ranking nutritional value."
167905,"Pork, fresh, loin, tenderloin, separable lean and fat, cooked, roasted",beef_pork,"Familiar ground-beef or pork cut, preserving the source fat/cut and raw/cooked identity without ranking nutritional value."
168286,"Pork, fresh, loin, center loin (chops), boneless, separable lean and fat, raw",beef_pork,"Familiar ground-beef or pork cut, preserving the source fat/cut and raw/cooked identity without ranking nutritional value."
168298,"Pork, fresh, loin, center loin (chops), boneless, separable lean and fat, cooked, pan-broiled",beef_pork,"Familiar ground-beef or pork cut, preserving the source fat/cut and raw/cooked identity without ranking nutritional value."
175167,"Fish, salmon, Atlantic, farmed, raw",fish_seafood,"Familiar fish or shrimp identity, retaining species, raw/cooked, or canned/drained distinctions for factual lookup."
175168,"Fish, salmon, Atlantic, farmed, cooked, dry heat",fish_seafood,"Familiar fish or shrimp identity, retaining species, raw/cooked, or canned/drained distinctions for factual lookup."
171955,"Fish, cod, Atlantic, raw",fish_seafood,"Familiar fish or shrimp identity, retaining species, raw/cooked, or canned/drained distinctions for factual lookup."
171956,"Fish, cod, Atlantic, cooked, dry heat",fish_seafood,"Familiar fish or shrimp identity, retaining species, raw/cooked, or canned/drained distinctions for factual lookup."
175176,"Fish, tilapia, raw",fish_seafood,"Familiar fish or shrimp identity, retaining species, raw/cooked, or canned/drained distinctions for factual lookup."
175177,"Fish, tilapia, cooked, dry heat",fish_seafood,"Familiar fish or shrimp identity, retaining species, raw/cooked, or canned/drained distinctions for factual lookup."
173709,"Fish, tuna, light, canned in water, drained solids (Includes foods for USDA's Food Distribution Program)",fish_seafood,"Familiar fish or shrimp identity, retaining species, raw/cooked, or canned/drained distinctions for factual lookup."
175180,"Crustaceans, shrimp, cooked",fish_seafood,"Familiar fish or shrimp identity, retaining species, raw/cooked, or canned/drained distinctions for factual lookup."
171413,"Oil, olive, salad or cooking",oils_condiments,Familiar cooking oil or basic condiment identity for factual ingredient lookup; no intake or health preference implied.
172336,"Oil, canola",oils_condiments,Familiar cooking oil or basic condiment identity for factual ingredient lookup; no intake or health preference implied.
171411,"Oil, soybean, salad or cooking",oils_condiments,Familiar cooking oil or basic condiment identity for factual ingredient lookup; no intake or health preference implied.
173469,"Vinegar, cider",oils_condiments,Familiar cooking oil or basic condiment identity for factual ingredient lookup; no intake or health preference implied.
173468,"Salt, table",oils_condiments,Familiar cooking oil or basic condiment identity for factual ingredient lookup; no intake or health preference implied.
173647,"Beverages, water, tap, drinking",beverages,"Plain water, brewed tea, or brewed coffee identity for factual beverage lookup; no beverage-consumption advice implied."
173227,"Beverages, tea, black, brewed, prepared with tap water",beverages,"Plain water, brewed tea, or brewed coffee identity for factual beverage lookup; no beverage-consumption advice implied."
171890,"Beverages, coffee, brewed, prepared with tap water",beverages,"Plain water, brewed tea, or brewed coffee identity for factual beverage lookup; no beverage-consumption advice implied."
"""
FD145_MODULE_SOURCE = r'''"""Source-linked arithmetic helpers; importing or calling them activates no app tool."""
from dataclasses import asdict, dataclass
from decimal import Context, Decimal, DecimalException, localcontext
import re

_UNITS = {"1008": "KCAL", "1003": "G", "1004": "G", "1005": "G", "1079": "G"}
_NUMBER = re.compile(r"[+-]?(?:[0-9]+(?:\.[0-9]*)?|\.[0-9]+)(?:[eE][+-]?[0-9]+)?")
TECHNICAL_LIMITS = {"decimal_precision": 50, "max_significant_digits": 50,
                    "min_adjusted_exponent": -1000, "max_adjusted_exponent": 1000,
                    "max_numeric_text_length": 128}


@dataclass(frozen=True)
class _FoodIdentity:
    fdc_id: str
    description: str
    source_data_type: str
    catalog_release: str


def _identifier(value, label):
    if isinstance(value, bool) or not isinstance(value, (str, int)) or not re.fullmatch(r"[0-9]+", str(value).strip()):
        raise ValueError(label + " must be a positive integer identifier.")
    identifier = str(int(str(value).strip()))
    if identifier == "0":
        raise ValueError(label + " must be positive.")
    return identifier


def _identity(food):
    fields = [str(food.get(key, "")).strip() for key in ("description", "source_data_type", "catalog_release")]
    if not all(fields) or any(food.get(key) is None for key in ("description", "source_data_type", "catalog_release")):
        raise ValueError("Food description, data type, and release must be recorded.")
    if fields[1:] != ["sr_legacy_food", "2018-04"]:
        raise ValueError("These helpers require the selected SR Legacy April 2018 reference.")
    result = asdict(_FoodIdentity(_identifier(food.get("fdc_id"), "fdc_id"), *fields))
    for field in ("NDB_number", "source_record_url", "source_archive_sha256", "source_food_table"):
        if field in food:
            result[field] = None if food[field] is None else str(food[field])
    return result


def _number(value, label, positive=True):
    if isinstance(value, bool) or not isinstance(value, (str, int, float, Decimal)):
        raise ValueError(label + " must be a finite number.")
    text = str(value).strip()
    if len(text) > TECHNICAL_LIMITS["max_numeric_text_length"] or not _NUMBER.fullmatch(text):
        raise ValueError(label + " has an unsupported numeric representation.")
    number = Decimal(text)
    if not number.is_finite() or number < 0 or (positive and number == 0):
        raise ValueError(label + (" must be positive and finite." if positive else " must be nonnegative and finite."))
    if len(number.as_tuple().digits) > TECHNICAL_LIMITS["max_significant_digits"] or not (
            TECHNICAL_LIMITS["min_adjusted_exponent"] <= number.adjusted() <= TECHNICAL_LIMITS["max_adjusted_exponent"]):
        raise ValueError(label + " exceeds the helper's technical numeric limits.")
    return number


def _text(value):
    if value == 0:
        return "0"
    text = format(value, "f") if -50 <= value.adjusted() <= 50 else str(value)
    return text.rstrip("0").rstrip(".") if "." in text and "E" not in text and "e" not in text else text


def food_fact(food, nutrient, grams="100"):
    """Return one eligible fact or its field-level unavailability; requested mass is positive."""
    identity = _identity(food)
    if _identifier(nutrient.get("fdc_id"), "nutrient fdc_id") != identity["fdc_id"]:
        raise ValueError("Nutrient and food identities differ.")
    mass = _number(grams, "Requested edible mass")
    nutrient_id = _identifier(nutrient.get("nutrient_id"), "nutrient_id")
    status = nutrient.get("numeric_status")
    unit = None if nutrient.get("unit_name") is None else str(nutrient["unit_name"])
    reason = status if isinstance(status, str) else "missing_or_invalid_numeric_status"
    result = {**identity, "status": "unavailable", "reason": reason,
              "grams": _text(mass), "amount": None, "unit": unit,
              "nutrient_id": nutrient_id, "nutrient_name": None if nutrient.get("nutrient_name") is None else str(nutrient["nutrient_name"]),
              "source_food_nutrient_id": None if nutrient.get("source_food_nutrient_id") is None else str(nutrient["source_food_nutrient_id"]),
              "amount_per_100g_raw": str(nutrient.get("amount_per_100g_raw", "")),
              "reference_mass_g": None if nutrient.get("reference_mass_g") is None else str(nutrient["reference_mass_g"]),
              "reference_basis": None if nutrient.get("reference_basis") is None else str(nutrient["reference_basis"])}
    for field in ("source_footnote", "source_derivation_id", "source_derivation_code", "source_derivation_description"):
        if field in nutrient:
            result[field] = None if nutrient[field] is None else str(nutrient[field])
    if status != "available":
        return result
    if nutrient_id not in _UNITS or unit != _UNITS[nutrient_id]:
        result["reason"] = "unexpected_unit_or_unsupported_nutrient"
        return result
    if nutrient.get("reference_basis") != "edible_food":
        result["reason"] = "unverified_reference_basis"
        return result
    try:
        reference = _number(nutrient.get("reference_mass_g"), "Reference mass")
        amount = _number(nutrient.get("amount_per_100g_raw"), "Recorded nutrient amount", positive=False)
        source_id = _identifier(nutrient.get("source_food_nutrient_id"), "source_food_nutrient_id")
    except (ValueError, DecimalException):
        result["reason"] = "invalid_available_source_record"
        return result
    if reference != 100:
        result["reason"] = "unsupported_reference_mass"
        return result
    try:
        with localcontext(Context(prec=TECHNICAL_LIMITS["decimal_precision"])):
            calculated = amount * mass / reference
    except DecimalException as error:
        raise ValueError("The nutrient calculation cannot be represented.") from error
    if calculated == 0 and amount != 0:
        raise ValueError("The nutrient calculation underflowed its numeric context.")
    result.update(status="available", reason=None, amount=_text(calculated), source_food_nutrient_id=source_id)
    return result


def household_mass(food, portion, quantity="1", mode="complete_record"):
    """Return grams for copies of a complete record, or quantities of its same defined unit."""
    identity = _identity(food)
    if _identifier(portion.get("fdc_id"), "portion fdc_id") != identity["fdc_id"]:
        raise ValueError("Portion and food identities differ.")
    if mode not in {"complete_record", "same_defined_unit"}:
        raise ValueError("Unknown household conversion mode.")
    flag = portion.get("household_numeric_screen_pass")
    if not (flag is True or isinstance(flag, str) and flag.strip().casefold() == "true"):
        raise ValueError("The portion has not passed its numeric screen.")
    if portion.get("household_eligibility", "available") != "available":
        raise ValueError("The portion is unavailable for household conversion.")
    if "footnote" not in portion or portion["footnote"] is None:
        raise ValueError("The portion footnote state is unknown.")
    modifier = str(portion.get("modifier") or "").strip()
    footnote = str(portion.get("footnote") or "").strip()
    if not modifier or footnote:
        raise ValueError("A missing measure description or unreviewed footnote prevents conversion.")
    count = _number(quantity, "Requested quantity")
    amount = _number(portion.get("amount_raw"), "Recorded portion quantity")
    weight = _number(portion.get("gram_weight_raw"), "Recorded portion gram weight")
    portion_id = _identifier(portion.get("portion_id"), "portion_id")
    measure_id = _identifier(portion.get("measure_unit_id"), "measure_unit_id")
    try:
        with localcontext(Context(prec=TECHNICAL_LIMITS["decimal_precision"])):
            mass = count * weight if mode == "complete_record" else count / amount * weight
    except DecimalException as error:
        raise ValueError("The portion calculation cannot be represented.") from error
    if mass <= 0:
        raise ValueError("The portion calculation underflowed its numeric context.")
    return {**identity, "status": "available", "grams": _text(mass), "quantity": _text(count), "mode": mode,
            "portion_id": portion_id, "amount_raw": str(portion["amount_raw"]),
            "gram_weight_raw": str(portion["gram_weight_raw"]), "modifier": modifier,
            "measure_unit_id": measure_id,
            "unit_name": None if portion.get("unit_name") is None else str(portion["unit_name"]), "footnote": footnote}
'''
FD145_TEST_SOURCE = r'''"""Development checks for exact-source food facts; not planner research scores."""
from decimal import Decimal


def run_tests(module, foods, nutrients, portions, goldens):
    tests, comparisons = [], []
    fact, mass = module["food_fact"], module["household_mass"]

    def record(target, case_id, condition, expected, observed, kind):
        target.append({"case_id": case_id, "status": "pass" if condition else "failed", "case_type": kind,
                       "expected": str(expected), "observed": str(observed)})

    def equal(case_id, expected, observed, kind="calculation"):
        condition = Decimal(str(observed)) == Decimal(str(expected))
        record(tests, case_id, condition, expected, observed, kind)

    def reject(case_id, callback, kind="invalid_input", allow_unavailable=False):
        try:
            result = callback()
        except (ValueError, TypeError):
            record(tests, case_id, True, "rejected", "rejected", kind)
        else:
            unavailable = allow_unavailable and isinstance(result, dict) and result.get("status") == "unavailable" and result.get("amount") is None
            record(tests, case_id, unavailable, "rejected_or_unavailable" if allow_unavailable else "rejected", result, kind)

    def source_portion(row):
        return {"portion_id": row["id"], "fdc_id": row["fdc_id"], "amount_raw": row["amount"],
                "gram_weight_raw": row["gram_weight"], "modifier": row["modifier"],
                "measure_unit_id": row["measure_unit_id"], "unit_name": "undetermined",
                "footnote": row["footnote"], "household_numeric_screen_pass": True,
                "household_eligibility": "unavailable" if row["amount"] == "0" else "available"}

    # These literals were prepared independently of the calculation module.
    for source in goldens["foods"]:
        food_id = source["fdc_id"]
        record(comparisons, "food_identity:" + food_id,
               food_id in foods and foods[food_id]["description"] == source["description"],
               source["description"], foods.get(food_id, {}).get("description"), "source_identity")
        for nutrient in source["source_nutrients"]:
            key = food_id, nutrient["nutrient_id"]
            if nutrient["nutrient_id"] not in {"1008", "1003", "1004", "1005", "1079"}:
                continue
            actual = nutrients.get(key, {})
            expected = nutrient["id"], nutrient["amount"], nutrient["unit_name"]
            observed = actual.get("source_food_nutrient_id"), actual.get("amount_per_100g_raw"), actual.get("unit_name")
            record(comparisons, "source_nutrient:" + ":".join(key), observed == expected, expected, observed, "source_nutrient_record")
    for source in goldens["source_portions"]:
        expected = source_portion(source)
        actual = portions.get(source["id"], {})
        for field in ("fdc_id", "amount_raw", "gram_weight_raw", "modifier", "measure_unit_id"):
            record(comparisons, "source_portion:" + source["id"] + ":" + field,
                   actual.get(field) == expected[field], expected[field], actual.get(field), "source_portion_record")

    for case in goldens["valid_cases"]:
        food = foods[case["fdc_id"]]
        if case["mode"] == "edible_mass_g":
            grams = case["quantity"]
        else:
            mode = {"complete_source_portions": "complete_record", "same_defined_measure_units": "same_defined_unit"}[case["mode"]]
            grams = mass(food, portions[case["portion_id"]], case["quantity"], mode=mode)["grams"]
        equal(case["case_id"] + ":mass", case["expected_mass_g"], grams, "source_calculation")
        for nutrient_id, expected in case["expected_amounts"].items():
            actual = fact(food, nutrients[(food["fdc_id"], nutrient_id)], grams)
            if expected is None:
                record(tests, case["case_id"] + ":" + nutrient_id,
                       actual["status"] == "unavailable" and actual["amount"] is None,
                       "unavailable; no numeric value", actual, "missing_nutrient")
            else:
                record(tests, case["case_id"] + ":" + nutrient_id,
                       actual["status"] == "available" and Decimal(actual["amount"]) == Decimal(expected),
                       expected, actual, "source_calculation")

    banana, energy = foods["173944"], nutrients[("173944", "1008")]
    peanut, peanut_portion = foods["172470"], portions["90673"]
    equal("kcal_identity_not_kj", "89", fact(banana, energy, "100")["amount"], "unit_handling")
    reject("cross_food_nutrient", lambda: fact(banana, nutrients[("172470", "1008")], "100"), "identity")
    reject("cross_food_portion", lambda: mass(banana, peanut_portion, "1"), "identity")
    reject("kcal_field_with_kj_unit", lambda: fact(banana, {**energy, "unit_name": "kJ"}, "100"), "unit_handling", True)
    reject("unsupported_same_name_energy_id", lambda: fact(banana, {**energy, "nutrient_id": "1062", "unit_name": "kJ"}, "100"), "unit_handling", True)
    reject("wrong_reference_mass", lambda: fact(banana, {**energy, "reference_mass_g": "1"}, "100"), "basis", True)
    reject("wrong_reference_basis", lambda: fact(banana, {**energy, "reference_basis": "unknown"}, "100"), "basis", True)
    reject("different_source_type", lambda: fact({**banana, "source_data_type": "foundation_food"}, energy, "100"), "source_scope")
    reject("different_source_release", lambda: fact({**banana, "catalog_release": "2025-04"}, energy, "100"), "source_scope")
    for index, quantity in enumerate(("0", "-1", "NaN", "Infinity", "-Infinity", "unknown", True)):
        reject("invalid_mass_" + str(index), lambda q=quantity: fact(banana, energy, q))
        reject("invalid_household_count_" + str(index), lambda q=quantity: mass(peanut, peanut_portion, q))
    for raw in ("-1", "NaN", "Infinity", "bad"):
        reject("unusable_recorded_amount:" + raw, lambda r=raw: fact(banana, {**energy, "amount_per_100g_raw": r}, "100"), "invalid_source_value", True)
    reject("portion_footnote_gate", lambda: mass(peanut, {**peanut_portion, "footnote": "Unresolved source qualification"}, "1"), "portion_context")
    reject("unknown_portion_mode", lambda: mass(peanut, peanut_portion, "1", mode="generic_cup"), "portion_context")
    reject("missing_measure_description", lambda: mass(peanut, {**peanut_portion, "modifier": ""}, "1"), "portion_context")
    reject("zero_gram_weight", lambda: mass(peanut, {**peanut_portion, "gram_weight_raw": "0"}, "1"), "portion_context")
    reject("negative_source_quantity", lambda: mass(peanut, {**peanut_portion, "amount_raw": "-1"}, "1"), "portion_context")
    reject("zero_quantity_despite_stale_available_flag", lambda: mass(peanut, {**peanut_portion, "amount_raw": "0"}, "1", mode="same_defined_unit"), "portion_context")
    reject("numeric_exponent_technical_bound", lambda: fact(banana, energy, "1e1001"), "technical_numeric_limit")
    reject("numeric_digits_technical_bound", lambda: fact(banana, energy, "1" * 51), "technical_numeric_limit")
    # Arithmetic illustration is explicitly synthetic: source peanut butter is 32g.
    illustration = {**peanut_portion, "gram_weight_raw": "30", "portion_id": "999999999"}
    equal("synthetic_complete_2tbsp30g", "30", mass(peanut, illustration, "1", mode="complete_record")["grams"], "synthetic_contract")
    equal("synthetic_one_tbsp15g", "15", mass(peanut, illustration, "1", mode="same_defined_unit")["grams"], "synthetic_contract")
    equal("actual_complete_2tbsp32g", "32", mass(peanut, peanut_portion, "1", mode="complete_record")["grams"], "source_calculation")
    equal("actual_one_tbsp16g", "16", mass(peanut, peanut_portion, "1", mode="same_defined_unit")["grams"], "source_calculation")
    # The 18 bad source quantities are full-source boundary fixtures, not selected foods.
    for source in goldens["zero_amount_portion_boundary_cases"]:
        food = {"fdc_id": source["fdc_id"], "source_data_type": "sr_legacy_food", "catalog_release": "2018-04",
                "description": source["food_description"]}
        portion = source_portion(source)
        reject("zero_source_amount:" + source["id"], lambda f=food, p=portion: mass(f, p, "1", mode="same_defined_unit"), "full_source_portion_boundary")
    return tests, comparisons


if __name__ == "__main__":
    import csv
    import json
    from pathlib import Path
    import runpy
    import sys

    bundle = Path(__file__).resolve().parents[1]

    def read_csv(name):
        with (bundle / name).open(encoding="utf-8-sig", newline="") as handle:
            return list(csv.DictReader(handle))

    food_rows = {r["fdc_id"]: r for r in read_csv("usda_foods.csv")}
    nutrient_rows = {(r["fdc_id"], r["nutrient_id"]): r for r in read_csv("usda_food_nutrients.csv")}
    portion_rows = {r["portion_id"]: r for r in read_csv("usda_food_portions.csv")}
    golden = json.loads((bundle / "development/usda_independent_source_goldens.json").read_text())
    results, source_checks = run_tests(runpy.run_path(str(bundle / "usda_food_reference.py")), food_rows, nutrient_rows, portion_rows, golden)
    failures = [r for r in [*results, *source_checks] if r["status"] != "pass"]
    print(f"Calculation cases: {len(results)} | Source comparisons: {len(source_checks)} | Failures: {len(failures)}")
    for row in failures:
        print(row)
    sys.exit(bool(failures))
'''
FD145_GOLDENS_JSON = r"""{
  "schema_version": "step14-independent-source-goldens-1",
  "purpose": "Literal regression expectations independently curated from the exact official SR source archive; not production-helper outputs.",
  "archive": {
    "filename": "sr_legacy_2018-04.zip",
    "sha256": "b80817294b8850530aaedf2e515c02593b1824f763a0ff356e5c2081643e6fd0",
    "data_type": "sr_legacy_food",
    "release": "2018-04",
    "publication_date": "2019-04-01"
  },
  "source_tables": {
    "food.csv": {
      "archive_member": "FoodData_Central_sr_legacy_food_csv_2018-04/food.csv",
      "sha256": "de8b0390058db4f102368c9fbdfde144790b480ff4cb0507c0e060ef99805dee"
    },
    "food_nutrient.csv": {
      "archive_member": "FoodData_Central_sr_legacy_food_csv_2018-04/food_nutrient.csv",
      "sha256": "6d07e6fc062b09750c4d95112be5a5cdb56a2346efcfd0c96b7a7d705812d4bd"
    },
    "food_portion.csv": {
      "archive_member": "FoodData_Central_sr_legacy_food_csv_2018-04/food_portion.csv",
      "sha256": "6332e29da61e13f7bd950b759461af73303c76e8b0e64dc9df4e41d5347cf3d1"
    },
    "nutrient.csv": {
      "archive_member": "FoodData_Central_sr_legacy_food_csv_2018-04/nutrient.csv",
      "sha256": "466ab9d011094cc1ad8d939b246693988b56817c42d0e40a98820723695c6084"
    }
  },
  "method": "Expected amounts were manually specified as decimal strings, then separately checked with Decimal against source nutrient amount and source gram mass. Compare unrounded values; display rounding is a separate policy.",
  "foods": [
    {
      "fdc_id": "170379",
      "data_type": "sr_legacy_food",
      "description": "Broccoli, raw",
      "food_category_id": "11",
      "publication_date": "2019-04-01",
      "source_nutrients": [
        {
          "id": "1518850",
          "fdc_id": "170379",
          "nutrient_id": "1003",
          "amount": "2.82",
          "data_points": "23",
          "derivation_id": "43",
          "min": "1.99",
          "max": "3.15",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Protein"
        },
        {
          "id": "1518830",
          "fdc_id": "170379",
          "nutrient_id": "1004",
          "amount": "0.37",
          "data_points": "29",
          "derivation_id": "43",
          "min": "0.33",
          "max": "0.41",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Total lipid (fat)"
        },
        {
          "id": "1518808",
          "fdc_id": "170379",
          "nutrient_id": "1005",
          "amount": "6.64",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Carbohydrate, by difference"
        },
        {
          "id": "1518809",
          "fdc_id": "170379",
          "nutrient_id": "1008",
          "amount": "34",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "KCAL",
          "nutrient_name": "Energy"
        },
        {
          "id": "1518802",
          "fdc_id": "170379",
          "nutrient_id": "1062",
          "amount": "141",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "kJ",
          "nutrient_name": "Energy"
        },
        {
          "id": "1518854",
          "fdc_id": "170379",
          "nutrient_id": "1079",
          "amount": "2.6",
          "data_points": "13",
          "derivation_id": "42",
          "min": "2.2",
          "max": "3.2",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Fiber, total dietary"
        }
      ]
    },
    {
      "fdc_id": "171267",
      "data_type": "sr_legacy_food",
      "description": "Milk, reduced fat, fluid, 2% milkfat, with added vitamin A and vitamin D",
      "food_category_id": "1",
      "publication_date": "2019-04-01",
      "source_nutrients": [
        {
          "id": "1595856",
          "fdc_id": "171267",
          "nutrient_id": "1003",
          "amount": "3.3",
          "data_points": "13",
          "derivation_id": "1",
          "min": "3.18",
          "max": "3.61",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Protein"
        },
        {
          "id": "1595969",
          "fdc_id": "171267",
          "nutrient_id": "1004",
          "amount": "1.98",
          "data_points": "14",
          "derivation_id": "43",
          "min": "1.9",
          "max": "2.06",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Total lipid (fat)"
        },
        {
          "id": "1595920",
          "fdc_id": "171267",
          "nutrient_id": "1005",
          "amount": "4.8",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Carbohydrate, by difference"
        },
        {
          "id": "1595921",
          "fdc_id": "171267",
          "nutrient_id": "1008",
          "amount": "50",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "KCAL",
          "nutrient_name": "Energy"
        },
        {
          "id": "1595911",
          "fdc_id": "171267",
          "nutrient_id": "1062",
          "amount": "210",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "kJ",
          "nutrient_name": "Energy"
        },
        {
          "id": "1595892",
          "fdc_id": "171267",
          "nutrient_id": "1079",
          "amount": "0",
          "data_points": "0",
          "derivation_id": "",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Fiber, total dietary"
        }
      ]
    },
    {
      "fdc_id": "171287",
      "data_type": "sr_legacy_food",
      "description": "Egg, whole, raw, fresh",
      "food_category_id": "1",
      "publication_date": "2019-04-01",
      "source_nutrients": [
        {
          "id": "1597683",
          "fdc_id": "171287",
          "nutrient_id": "1003",
          "amount": "12.56",
          "data_points": "12",
          "derivation_id": "1",
          "min": "11.81",
          "max": "13.06",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Protein"
        },
        {
          "id": "1597714",
          "fdc_id": "171287",
          "nutrient_id": "1004",
          "amount": "9.51",
          "data_points": "12",
          "derivation_id": "1",
          "min": "8.78",
          "max": "10.3",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Total lipid (fat)"
        },
        {
          "id": "1597748",
          "fdc_id": "171287",
          "nutrient_id": "1005",
          "amount": "0.72",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Carbohydrate, by difference"
        },
        {
          "id": "1597626",
          "fdc_id": "171287",
          "nutrient_id": "1008",
          "amount": "143",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "KCAL",
          "nutrient_name": "Energy"
        },
        {
          "id": "1597629",
          "fdc_id": "171287",
          "nutrient_id": "1062",
          "amount": "599",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "kJ",
          "nutrient_name": "Energy"
        },
        {
          "id": "1597756",
          "fdc_id": "171287",
          "nutrient_id": "1079",
          "amount": "0",
          "data_points": "1",
          "derivation_id": "1",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Fiber, total dietary"
        }
      ]
    },
    {
      "fdc_id": "171413",
      "data_type": "sr_legacy_food",
      "description": "Oil, olive, salad or cooking",
      "food_category_id": "4",
      "publication_date": "2019-04-01",
      "source_nutrients": [
        {
          "id": "1607794",
          "fdc_id": "171413",
          "nutrient_id": "1003",
          "amount": "0",
          "data_points": "4",
          "derivation_id": "1",
          "min": "0",
          "max": "0",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Protein"
        },
        {
          "id": "1607766",
          "fdc_id": "171413",
          "nutrient_id": "1004",
          "amount": "100",
          "data_points": "0",
          "derivation_id": "52",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Total lipid (fat)"
        },
        {
          "id": "1607767",
          "fdc_id": "171413",
          "nutrient_id": "1005",
          "amount": "0",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Carbohydrate, by difference"
        },
        {
          "id": "1607768",
          "fdc_id": "171413",
          "nutrient_id": "1008",
          "amount": "884",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "KCAL",
          "nutrient_name": "Energy"
        },
        {
          "id": "1607796",
          "fdc_id": "171413",
          "nutrient_id": "1062",
          "amount": "3699",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "kJ",
          "nutrient_name": "Energy"
        },
        {
          "id": "1607828",
          "fdc_id": "171413",
          "nutrient_id": "1079",
          "amount": "0",
          "data_points": "0",
          "derivation_id": "",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Fiber, total dietary"
        }
      ]
    },
    {
      "fdc_id": "172470",
      "data_type": "sr_legacy_food",
      "description": "Peanut butter, smooth style, without salt",
      "food_category_id": "16",
      "publication_date": "2019-04-01",
      "source_nutrients": [
        {
          "id": "1697629",
          "fdc_id": "172470",
          "nutrient_id": "1003",
          "amount": "22.21",
          "data_points": "18",
          "derivation_id": "1",
          "min": "20.97",
          "max": "23.59",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Protein"
        },
        {
          "id": "1697676",
          "fdc_id": "172470",
          "nutrient_id": "1004",
          "amount": "51.36",
          "data_points": "18",
          "derivation_id": "1",
          "min": "49.67",
          "max": "52.66",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Total lipid (fat)"
        },
        {
          "id": "1697677",
          "fdc_id": "172470",
          "nutrient_id": "1005",
          "amount": "22.31",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Carbohydrate, by difference"
        },
        {
          "id": "1697678",
          "fdc_id": "172470",
          "nutrient_id": "1008",
          "amount": "598",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "KCAL",
          "nutrient_name": "Energy"
        },
        {
          "id": "1697634",
          "fdc_id": "172470",
          "nutrient_id": "1062",
          "amount": "2501",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "kJ",
          "nutrient_name": "Energy"
        },
        {
          "id": "1697635",
          "fdc_id": "172470",
          "nutrient_id": "1079",
          "amount": "5",
          "data_points": "9",
          "derivation_id": "1",
          "min": "3.6",
          "max": "6.1",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Fiber, total dietary"
        }
      ]
    },
    {
      "fdc_id": "173944",
      "data_type": "sr_legacy_food",
      "description": "Bananas, raw",
      "food_category_id": "9",
      "publication_date": "2019-04-01",
      "source_nutrients": [
        {
          "id": "1817943",
          "fdc_id": "173944",
          "nutrient_id": "1003",
          "amount": "1.09",
          "data_points": "12",
          "derivation_id": "1",
          "min": "0.75",
          "max": "1.31",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Protein"
        },
        {
          "id": "1817956",
          "fdc_id": "173944",
          "nutrient_id": "1004",
          "amount": "0.33",
          "data_points": "19",
          "derivation_id": "1",
          "min": "0.07",
          "max": "0.7",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Total lipid (fat)"
        },
        {
          "id": "1817899",
          "fdc_id": "173944",
          "nutrient_id": "1005",
          "amount": "22.84",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Carbohydrate, by difference"
        },
        {
          "id": "1817900",
          "fdc_id": "173944",
          "nutrient_id": "1008",
          "amount": "89",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "KCAL",
          "nutrient_name": "Energy"
        },
        {
          "id": "1817879",
          "fdc_id": "173944",
          "nutrient_id": "1062",
          "amount": "371",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "kJ",
          "nutrient_name": "Energy"
        },
        {
          "id": "1817936",
          "fdc_id": "173944",
          "nutrient_id": "1079",
          "amount": "2.6",
          "data_points": "13",
          "derivation_id": "1",
          "min": "1.7",
          "max": "3.7",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Fiber, total dietary"
        }
      ]
    },
    {
      "fdc_id": "174272",
      "data_type": "sr_legacy_food",
      "description": "Tempeh",
      "food_category_id": "16",
      "publication_date": "2019-04-01",
      "source_nutrients": [
        {
          "id": "1845305",
          "fdc_id": "174272",
          "nutrient_id": "1003",
          "amount": "20.29",
          "data_points": "1",
          "derivation_id": "1",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Protein"
        },
        {
          "id": "1845352",
          "fdc_id": "174272",
          "nutrient_id": "1004",
          "amount": "10.8",
          "data_points": "1",
          "derivation_id": "1",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Total lipid (fat)"
        },
        {
          "id": "1845312",
          "fdc_id": "174272",
          "nutrient_id": "1005",
          "amount": "7.64",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "G",
          "nutrient_name": "Carbohydrate, by difference"
        },
        {
          "id": "1845313",
          "fdc_id": "174272",
          "nutrient_id": "1008",
          "amount": "192",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "KCAL",
          "nutrient_name": "Energy"
        },
        {
          "id": "1845306",
          "fdc_id": "174272",
          "nutrient_id": "1062",
          "amount": "803",
          "data_points": "0",
          "derivation_id": "49",
          "min": "",
          "max": "",
          "median": "",
          "footnote": "",
          "min_year_acquired": "",
          "unit_name": "kJ",
          "nutrient_name": "Energy"
        }
      ]
    }
  ],
  "source_portions": [
    {
      "id": "93515",
      "fdc_id": "173944",
      "seq_num": "5",
      "amount": "1",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "medium (7\" to 7-7/8\" long)",
      "gram_weight": "118",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": ""
    },
    {
      "id": "88320",
      "fdc_id": "171267",
      "seq_num": "1",
      "amount": "1",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "cup",
      "gram_weight": "244",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": ""
    },
    {
      "id": "88374",
      "fdc_id": "171287",
      "seq_num": "1",
      "amount": "1",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "large",
      "gram_weight": "50",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": ""
    },
    {
      "id": "88667",
      "fdc_id": "171413",
      "seq_num": "1",
      "amount": "1",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "tablespoon",
      "gram_weight": "13.5",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": ""
    },
    {
      "id": "90673",
      "fdc_id": "172470",
      "seq_num": "1",
      "amount": "2",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "tbsp",
      "gram_weight": "32",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": ""
    },
    {
      "id": "86532",
      "fdc_id": "170379",
      "seq_num": "5",
      "amount": "0.5",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "cup, chopped or diced",
      "gram_weight": "44",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": ""
    },
    {
      "id": "86528",
      "fdc_id": "170379",
      "seq_num": "1",
      "amount": "1",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "cup chopped",
      "gram_weight": "91",
      "data_points": "9",
      "footnote": "",
      "min_year_acquired": ""
    }
  ],
  "valid_cases": [
    {
      "case_id": "banana_100g_identity",
      "fdc_id": "173944",
      "mode": "edible_mass_g",
      "quantity": "100",
      "expected_mass_g": "100",
      "expected_amounts": {
        "1008": "89",
        "1003": "1.09",
        "1004": "0.33",
        "1005": "22.84",
        "1079": "2.6"
      }
    },
    {
      "case_id": "banana_50g_half",
      "fdc_id": "173944",
      "mode": "edible_mass_g",
      "quantity": "50",
      "expected_mass_g": "50",
      "expected_amounts": {
        "1008": "44.5",
        "1003": "0.545",
        "1004": "0.165",
        "1005": "11.42",
        "1079": "1.3"
      }
    },
    {
      "case_id": "banana_one_medium",
      "fdc_id": "173944",
      "mode": "complete_source_portions",
      "portion_id": "93515",
      "quantity": "1",
      "expected_mass_g": "118",
      "expected_amounts": {
        "1008": "105.02",
        "1003": "1.2862",
        "1004": "0.3894",
        "1005": "26.9512",
        "1079": "3.068"
      }
    },
    {
      "case_id": "milk_one_source_cup",
      "fdc_id": "171267",
      "mode": "complete_source_portions",
      "portion_id": "88320",
      "quantity": "1",
      "expected_mass_g": "244",
      "expected_amounts": {
        "1008": "122",
        "1003": "8.052",
        "1004": "4.8312",
        "1005": "11.712",
        "1079": "0"
      }
    },
    {
      "case_id": "egg_one_large",
      "fdc_id": "171287",
      "mode": "complete_source_portions",
      "portion_id": "88374",
      "quantity": "1",
      "expected_mass_g": "50",
      "expected_amounts": {
        "1008": "71.5",
        "1003": "6.28",
        "1004": "4.755",
        "1005": "0.36",
        "1079": "0"
      }
    },
    {
      "case_id": "olive_oil_one_tablespoon",
      "fdc_id": "171413",
      "mode": "complete_source_portions",
      "portion_id": "88667",
      "quantity": "1",
      "expected_mass_g": "13.5",
      "expected_amounts": {
        "1008": "119.34",
        "1003": "0",
        "1004": "13.5",
        "1005": "0",
        "1079": "0"
      }
    },
    {
      "case_id": "peanut_butter_whole_2tbsp_source_measure",
      "fdc_id": "172470",
      "mode": "complete_source_portions",
      "portion_id": "90673",
      "quantity": "1",
      "expected_mass_g": "32",
      "expected_amounts": {
        "1008": "191.36",
        "1003": "7.1072",
        "1004": "16.4352",
        "1005": "7.1392",
        "1079": "1.6"
      }
    },
    {
      "case_id": "peanut_butter_one_same_defined_tbsp",
      "fdc_id": "172470",
      "mode": "same_defined_measure_units",
      "portion_id": "90673",
      "quantity": "1",
      "expected_mass_g": "16",
      "expected_amounts": {
        "1008": "95.68",
        "1003": "3.5536",
        "1004": "8.2176",
        "1005": "3.5696",
        "1079": "0.8"
      }
    },
    {
      "case_id": "broccoli_whole_halfcup_source_measure",
      "fdc_id": "170379",
      "mode": "complete_source_portions",
      "portion_id": "86532",
      "quantity": "1",
      "expected_mass_g": "44",
      "expected_amounts": {
        "1008": "14.96",
        "1003": "1.2408",
        "1004": "0.1628",
        "1005": "2.9216",
        "1079": "1.144"
      }
    },
    {
      "case_id": "broccoli_one_same_defined_cup",
      "fdc_id": "170379",
      "mode": "same_defined_measure_units",
      "portion_id": "86532",
      "quantity": "1",
      "expected_mass_g": "88",
      "expected_amounts": {
        "1008": "29.92",
        "1003": "2.4816",
        "1004": "0.3256",
        "1005": "5.8432",
        "1079": "2.288"
      }
    },
    {
      "case_id": "tempeh_50g_missing_fiber_preserved",
      "fdc_id": "174272",
      "mode": "edible_mass_g",
      "quantity": "50",
      "expected_mass_g": "50",
      "expected_amounts": {
        "1008": "96",
        "1003": "10.145",
        "1004": "5.4",
        "1005": "3.82",
        "1079": null
      }
    }
  ],
  "specific_boundaries": [
    {
      "case_id": "banana_kcal_not_kj",
      "fdc_id": "173944",
      "energy_kcal_nutrient_id": "1008",
      "expected_kcal_per_100g": "89",
      "distinct_kj_nutrient_id": "1062",
      "source_kj_per_100g": "371",
      "expectation": "The kcal display must not return371 or choose a same-name nutrient by list order."
    },
    {
      "case_id": "food_bound_portion",
      "requested_fdc_id": "173944",
      "portion_id": "90673",
      "actual_portion_fdc_id": "172470",
      "expectation": "Reject cross-food portion lookup before arithmetic."
    },
    {
      "case_id": "no_generic_cup_collapse",
      "fdc_id": "170379",
      "portion_id_a": "86532",
      "amount_a": "0.5",
      "modifier_a": "cup, chopped or diced",
      "gram_weight_a": "44",
      "portion_id_b": "86528",
      "amount_b": "1",
      "modifier_b": "cup chopped",
      "gram_weight_b": "91",
      "expectation": "Preserve exact source measure identity; do not replace row86532 with row86528 during normalization."
    },
    {
      "case_id": "missing_fiber_not_zero",
      "fdc_id": "174272",
      "missing_nutrient_id": "1079",
      "expectation": "Absent fiber remains unavailable while four present valid nutrients remain usable."
    },
    {
      "case_id": "recorded_zero_valid",
      "fdc_id": "171413",
      "nutrient_ids": [
        "1003",
        "1005",
        "1079"
      ],
      "expected_amounts": [
        "0",
        "0",
        "0"
      ],
      "expectation": "Recorded zeros remain available numeric zeros at positive mass."
    }
  ],
  "zero_amount_portion_boundary_cases": [
    {
      "id": "83785",
      "fdc_id": "168789",
      "seq_num": "1",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "cup",
      "gram_weight": "142",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Rennin, chocolate, dry mix, prepared with 2% milk"
    },
    {
      "id": "83786",
      "fdc_id": "168790",
      "seq_num": "1",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "cup",
      "gram_weight": "142",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Rennin, vanilla, dry mix, prepared with 2% milk"
    },
    {
      "id": "83793",
      "fdc_id": "168796",
      "seq_num": "1",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "cup",
      "gram_weight": "141",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Flan, caramel custard, dry mix, prepared with whole milk"
    },
    {
      "id": "84494",
      "fdc_id": "169239",
      "seq_num": "0",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "package (10 oz)",
      "gram_weight": "94",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Kale, frozen, unprepared"
    },
    {
      "id": "85231",
      "fdc_id": "169617",
      "seq_num": "1",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "cup",
      "gram_weight": "142",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Rennin, chocolate, dry mix, prepared with whole milk"
    },
    {
      "id": "85240",
      "fdc_id": "169621",
      "seq_num": "1",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "cup",
      "gram_weight": "141",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Flan, caramel custard, dry mix, prepared with 2% milk"
    },
    {
      "id": "87870",
      "fdc_id": "171056",
      "seq_num": "3",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "chicken, skin only",
      "gram_weight": "72",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Chicken, broilers or fryers, skin only, cooked, stewed"
    },
    {
      "id": "87883",
      "fdc_id": "171062",
      "seq_num": "3",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "chicken, bone removed",
      "gram_weight": "194",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Chicken, broilers or fryers, light meat, meat and skin, raw"
    },
    {
      "id": "87910",
      "fdc_id": "171073",
      "seq_num": "3",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "back, bone and skin removed",
      "gram_weight": "51",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Chicken, broilers or fryers, back, meat only, raw"
    },
    {
      "id": "88407",
      "fdc_id": "171300",
      "seq_num": "6",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "container (4 oz)",
      "gram_weight": "113",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Yogurt, Greek, strawberry, lowfat"
    },
    {
      "id": "88756",
      "fdc_id": "171450",
      "seq_num": "4",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "chicken, bone removed",
      "gram_weight": "299",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Chicken, broilers or fryers, meat and skin, cooked, roasted"
    },
    {
      "id": "88762",
      "fdc_id": "171452",
      "seq_num": "3",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "chicken, skin only",
      "gram_weight": "79",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Chicken, broilers or fryers, skin only, raw"
    },
    {
      "id": "88765",
      "fdc_id": "171453",
      "seq_num": "3",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "chicken, skin only",
      "gram_weight": "190",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Chicken, broilers or fryers, skin only, cooked, fried, batter"
    },
    {
      "id": "88807",
      "fdc_id": "171472",
      "seq_num": "3",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "back, bone and skin removed",
      "gram_weight": "40",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Chicken, broilers or fryers, back, meat only, cooked, roasted"
    },
    {
      "id": "88814",
      "fdc_id": "171475",
      "seq_num": "3",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "breast, bone removed",
      "gram_weight": "140",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Chicken, broilers or fryers, breast, meat and skin, cooked, fried, batter"
    },
    {
      "id": "88848",
      "fdc_id": "171492",
      "seq_num": "3",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "breast, bone removed",
      "gram_weight": "864",
      "data_points": "",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Turkey, all classes, breast, meat and skin, cooked, roasted"
    },
    {
      "id": "90178",
      "fdc_id": "172252",
      "seq_num": "1",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "cup",
      "gram_weight": "7",
      "data_points": "10",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Babyfood, finger snacks, GERBER, GRADUATES, PUFFS, apple and cinnamon"
    },
    {
      "id": "92745",
      "fdc_id": "173509",
      "seq_num": "1",
      "amount": "0",
      "measure_unit_id": "9999",
      "portion_description": "",
      "modifier": "cup",
      "gram_weight": "16",
      "data_points": "12",
      "footnote": "",
      "min_year_acquired": "",
      "food_description": "Babyfood, cereal, Oatmeal, dry, GERBER, SINGLE GRAIN, fortified"
    }
  ],
  "zero_amount_policy": "Reject automatic household-count normalization for all18 source rows; do not divide by amount or infer amount=1. The related food can still supply valid per100g facts and explicit edible-gram scaling.",
  "synthetic_cases_to_keep_separate": [
    "2tbsp30g arithmetic illustration (not the actual peanutbutter source portion, which is2tbsp32g)",
    "User nonfinite/negative/zero/over-limit mass and quantity inputs",
    "Injected incompatible nutrient unit or missing nutrient record",
    "Footnote/context gates"
  ]
}
"""

fd145_validation_ref = step14_report["validation_producer"]
fd145_validation_path = fd14_path(fd145_validation_ref)
fd145_validation_bytes = fd145_validation_path.read_bytes()
fd145_validation_sha = hashlib.sha256(fd145_validation_bytes).hexdigest()
fd145_require(fd145_validation_sha == step14_report["validation_digest"]
                and fd145_validation_path.parent.name == "source_validation_" + fd145_validation_sha[:16],
                "The saved source-validation producer changed.")
fd145_validation = json.loads(fd145_validation_bytes.decode("utf-8-sig"), object_pairs_hook=fd14_pairs)
fd145_require(fd145_validation.get("schema_version") == "step14-source-validation-producer-1"
                and fd145_validation.get("status") == "source_validation_complete_selection_pending"
                and fd145_validation.get("verified_checks")
                and all(r.get("status") == "pass" for r in fd145_validation["verified_checks"]), "Unexpected or failed source-validation checkpoint.")
fd145_inputs = dict(fd145_validation["input_hashes"])
fd145_inputs[fd145_validation_ref] = fd145_validation_sha
for relative, metadata in fd145_validation["artifacts"].items():
    path = fd142_child(fd145_validation_path.parent, relative)
    fd145_require(path.is_file() and fd14_sha(path) == metadata["sha256"] and path.stat().st_size == metadata["bytes"],
                    "A saved validation artifact changed: " + relative)
    fd145_inputs[path.relative_to(fd14_root).as_posix()] = metadata["sha256"]


def fd145_verify_inputs():
    for reference, digest in fd145_inputs.items():
        path = fd14_path(reference)
        fd145_require(path.is_file() and fd14_sha(path) == digest, "An inspected input changed: " + reference)


fd145_verify_inputs()
fd145_contract_path = fd145_validation_path.parent / "usda_reference_contract_proposal.json"
fd145_contract = json.loads(fd145_contract_path.read_text(encoding="utf-8-sig"), object_pairs_hook=fd14_pairs)
fd145_require(fd145_contract["reference_mass_g"] == 100 and fd145_contract["reference_basis"] == "edible_food"
                and fd145_contract["source_data_type"] == "sr_legacy_food", "The measurement-basis contract differs.")
fd145_source_refs = fd145_contract["source_table_paths"]


def fd145_read_rows(name):
    reference = fd145_source_refs[name]
    fd145_require(reference in fd145_inputs, "Source table is outside the verified inventory: " + name)
    with fd14_path(reference).open(encoding="utf-8-sig", newline="") as handle:
        reader = csv.DictReader(handle)
        for row in reader:
            fd145_require(None not in row and all(v is not None for v in row.values()), "Malformed source row: " + name)
            yield row


fd145_acquisition_ref = fd145_validation["source_acquisition_producer"]
fd145_acquisition = json.loads(fd14_path(fd145_acquisition_ref).read_text(encoding="utf-8-sig"), object_pairs_hook=fd14_pairs)
fd145_sr_source = next(r for r in fd145_acquisition["sources"] if r["source_key"] == "sr_legacy")
fd145_archive_sha = fd145_sr_source["response"]["sha256"]
fd145_check("frozen_archive", fd145_archive_sha == "b80817294b8850530aaedf2e515c02593b1824f763a0ff356e5c2081643e6fd0",
              "The explicit selection and independent expected values target the verified SR package.")
fd145_selection = list(csv.DictReader(io.StringIO(FD145_SELECTION_CSV, newline="")))
fd145_ids = {r["fdc_id"] for r in fd145_selection}
fd145_check("fixed_selection_identity", len(fd145_selection) == len(fd145_ids) == 100,
              "Exactly 100 explicit, distinct source IDs; no data-dependent top-up or nutrient ranking.")
fd145_all_foods = {r["fdc_id"]: r for r in fd145_read_rows("food.csv")}
fd145_defs = {r["id"]: r for r in fd145_read_rows("nutrient.csv")}
fd145_cats = {r["id"]: r for r in fd145_read_rows("food_category.csv")}
fd145_ndb = {r["fdc_id"]: r for r in fd145_read_rows("sr_legacy_food.csv")}
fd145_units = {r["id"]: r for r in fd145_read_rows("measure_unit.csv")}
fd145_derivations = {r["id"]: r for r in fd145_read_rows("food_nutrient_derivation.csv")}
fd145_nutrient_sources = {r["id"]: r for r in fd145_read_rows("food_nutrient_source.csv")}
fd145_expected_specs = {r["nutrient_id"]: r for r in fd145_contract["display_definitions"]}
fd145_expected_units = {"1008": "KCAL", "1003": "G", "1004": "G", "1005": "G", "1079": "G"}
fd145_check("definition_ids_and_units", set(fd145_expected_specs) == set(fd145_expected_units)
              and all(fd145_defs[key]["unit_name"] == unit for key, unit in fd145_expected_units.items()),
              "Five exact nutrient IDs retain their source units; kcal and kJ are not interchangeable.")
fd145_food_rows = []
for choice in fd145_selection:
    food_id = choice["fdc_id"]
    fd145_require(food_id in fd145_all_foods and fd145_all_foods[food_id]["description"] == choice["expected_description"],
                    "A selected source identity or description differs: " + food_id)
    source = fd145_all_foods[food_id]
    fd145_require(source["data_type"] == "sr_legacy_food" and source["food_category_id"] in fd145_cats,
                    "Invalid selected food scope or category: " + food_id)
    fd145_food_rows.append({"fdc_id": food_id, "NDB_number": fd145_ndb[food_id]["NDB_number"], "description": source["description"],
        "source_data_type": source["data_type"], "catalog_release": "2018-04", "publication_date": source["publication_date"],
        "food_category_id": source["food_category_id"], "food_category": fd145_cats[source["food_category_id"]]["description"],
        "selection_family": choice["selection_family"], "selection_reason": choice["selection_reason"],
        "preparation_details": "Retained in the exact source description; not normalized or inferred.",
        "source_record_url": "https://fdc.nal.usda.gov/food-details/" + food_id + "/nutrients",
        "source_archive_sha256": fd145_archive_sha, "source_food_table": fd145_source_refs["food.csv"],
        "reference_mass_g": "100", "reference_basis": "edible_food"})
fd145_food_rows.sort(key=lambda r: int(r["fdc_id"]))
fd145_food_index = {r["fdc_id"]: r for r in fd145_food_rows}


def fd145_numeric(value):
    if value.strip().casefold() in {"", "none", "null", "na", "n/a"}:
        return "missing"
    try:
        number = Decimal(value)
    except InvalidOperation:
        return "invalid_numeric"
    if not number.is_finite():
        return "nonfinite"
    return "negative" if number < 0 else "available"


fd145_selected_source = {}
for row in fd145_read_rows("food_nutrient.csv"):
    key = row["fdc_id"], row["nutrient_id"]
    if row["fdc_id"] in fd145_ids and row["nutrient_id"] in fd145_expected_units:
        fd145_require(key not in fd145_selected_source, "Conflicting selected food-nutrient records: " + str(key))
        fd145_selected_source[key] = row
fd145_nutrient_rows = []
for food in fd145_food_rows:
    for nutrient_id in ("1008", "1003", "1004", "1005", "1079"):
        source = fd145_selected_source.get((food["fdc_id"], nutrient_id))
        definition = fd145_defs[nutrient_id]
        status = "missing" if source is None else fd145_numeric(source["amount"])
        if definition["unit_name"] != fd145_expected_units[nutrient_id]:
            status = "unexpected_unit"
        derivation = fd145_derivations.get(source["derivation_id"], {}) if source else {}
        nutrient_source = fd145_nutrient_sources.get(derivation.get("source_id"), {})
        fd145_nutrient_rows.append({"fdc_id": food["fdc_id"], "nutrient_id": nutrient_id, "nutrient_name": definition["name"],
            "unit_name": definition["unit_name"], "amount_per_100g_raw": source["amount"] if source else "", "numeric_status": status,
            "reference_mass_g": "100", "reference_basis": "edible_food", "source_food_nutrient_id": source["id"] if source else "",
            "source_derivation_id": source["derivation_id"] if source else "", "source_derivation_code": derivation.get("code", ""),
            "source_derivation_description": derivation.get("description", ""), "source_nutrient_source_id": derivation.get("source_id", ""),
            "source_nutrient_source_description": nutrient_source.get("description", ""), "source_data_points": source["data_points"] if source else "",
            "source_footnote": source["footnote"] if source else "", "source_food_nutrient_table": fd145_source_refs["food_nutrient.csv"],
            "energy_selection_policy": "recorded_1008_no_recalculation" if nutrient_id == "1008" else "not_applicable"})
fd145_nutrient_index = {(r["fdc_id"], r["nutrient_id"]): r for r in fd145_nutrient_rows}
fd145_check("selected_nutrient_cardinality", len(fd145_nutrient_index) == len(fd145_nutrient_rows) == 5 * len(fd145_food_rows),
              "One slot per selected food and display nutrient; absent source records remain explicit blanks.")
fd145_check("selected_value_preservation", all(r["amount_per_100g_raw"] == fd145_selected_source[(r["fdc_id"], r["nutrient_id"])]["amount"]
              for r in fd145_nutrient_rows if r["source_food_nutrient_id"]), "Every retained amount preserves its original source text.")
fd145_portion_rows = []
for source in fd145_read_rows("food_portion.csv"):
    if source["fdc_id"] not in fd145_ids:
        continue
    issues = []
    amount_state, weight_state = fd145_numeric(source["amount"]), fd145_numeric(source["gram_weight"])
    if amount_state != "available" or Decimal(source["amount"]) <= 0:
        issues.append("source_amount_not_positive_finite")
    if weight_state != "available" or Decimal(source["gram_weight"]) <= 0:
        issues.append("source_gram_weight_not_positive_finite")
    if not source["modifier"].strip():
        issues.append("missing_measure_description")
    if source["footnote"].strip():
        issues.append("source_footnote_requires_context")
    if source["measure_unit_id"] not in fd145_units:
        issues.append("unknown_measure_unit_reference")
    fd145_portion_rows.append({"portion_id": source["id"], "fdc_id": source["fdc_id"], "amount_raw": source["amount"],
        "gram_weight_raw": source["gram_weight"], "modifier": source["modifier"], "portion_description": source["portion_description"],
        "measure_unit_id": source["measure_unit_id"], "unit_name": fd145_units.get(source["measure_unit_id"], {}).get("name", ""),
        "footnote": source["footnote"], "household_numeric_screen_pass": amount_state == "available" and weight_state == "available"
            and Decimal(source["amount"]) > 0 and Decimal(source["gram_weight"]) > 0 and bool(source["modifier"].strip()),
        "household_eligibility": "available" if not issues else "unavailable",
        "eligibility_reasons": ";".join(issues), "interpretation_policy": "exact_food_bound_source_measure; no_generic_volume_mapping",
        "source_food_portion_table": fd145_source_refs["food_portion.csv"]})
fd145_portion_index = {r["portion_id"]: r for r in fd145_portion_rows}
fd145_check("selected_portion_identity", len(fd145_portion_index) == len(fd145_portion_rows), "Selected portions retain unique source IDs.")
for food in fd145_food_rows:
    nutrients = [r for r in fd145_nutrient_rows if r["fdc_id"] == food["fdc_id"]]
    food["available_display_nutrients"] = sum(r["numeric_status"] == "available" for r in nutrients)
    food["unavailable_display_nutrients"] = sum(r["numeric_status"] != "available" for r in nutrients)
    food["per_100g_facts_available"] = food["available_display_nutrients"] > 0
    food["portion_records"] = sum(r["fdc_id"] == food["fdc_id"] for r in fd145_portion_rows)
    food["eligible_household_records"] = sum(r["fdc_id"] == food["fdc_id"] and r["household_eligibility"] == "available" for r in fd145_portion_rows)

fd145_module = {"__name__": "usda_food_reference_preparation"}
exec(compile(FD145_MODULE_SOURCE, "usda_food_reference.py", "exec"), fd145_module)
fd145_tests = {"__name__": "usda_food_reference_development_tests"}
exec(compile(FD145_TEST_SOURCE, "usda_reference_calculation_tests.py", "exec"), fd145_tests)
fd145_goldens = json.loads(FD145_GOLDENS_JSON, object_pairs_hook=fd14_pairs)
fd145_test_rows, fd145_comparison_rows = fd145_tests["run_tests"](
    fd145_module, fd145_food_index, fd145_nutrient_index, fd145_portion_index, fd145_goldens)
fd145_check("executed_calculation_cases", bool(fd145_test_rows) and all(r["status"] == "pass" for r in fd145_test_rows),
              "Executed deterministic source and boundary cases; development tests are not planner evaluation scores.")
fd145_check("independent_source_comparisons", bool(fd145_comparison_rows) and all(r["status"] == "pass" for r in fd145_comparison_rows),
              "Selected records agree with independently prepared literal source comparisons.")
fd145_quality_rows = [{"fdc_id": r["fdc_id"], "nutrient_id": r["nutrient_id"], "finding": r["numeric_status"],
                      "handling": "preserve_source; affected_field_unavailable; other_valid_fields_retained"}
                     for r in fd145_nutrient_rows if r["numeric_status"] != "available"]
fd145_quality_rows += [{"fdc_id": r["fdc_id"], "nutrient_id": "", "finding": "portion:" + r["portion_id"] + ":" + r["eligibility_reasons"],
                       "handling": "retain_portion; automatic_household_conversion_unavailable; explicit_gram_facts_separate"}
                      for r in fd145_portion_rows if r["household_eligibility"] != "available"]
fd145_coverage = []
for category_id in sorted({r["food_category_id"] for r in fd145_food_rows}, key=int):
    group = [r for r in fd145_food_rows if r["food_category_id"] == category_id]
    ids = {r["fdc_id"] for r in group}
    facts = [r for r in fd145_nutrient_rows if r["fdc_id"] in ids]
    fd145_coverage.append({"food_category_id": category_id, "food_category": fd145_cats[category_id]["description"],
        "selected_foods": len(group), "available_nutrient_fields": sum(r["numeric_status"] == "available" for r in facts),
        "unavailable_nutrient_fields": sum(r["numeric_status"] != "available" for r in facts),
        "foods_with_household_records": sum(r["eligible_household_records"] > 0 for r in group)})
fd145_metrics = {"selected_foods": len(fd145_food_rows), "source_foods": len(fd145_all_foods),
    "selection_families": len({r["selection_family"] for r in fd145_food_rows}), "populated_food_categories": len(fd145_coverage),
    "nutrient_slots": len(fd145_nutrient_rows), "recorded_nutrient_values": len(fd145_selected_source),
    "available_nutrient_fields": sum(r["numeric_status"] == "available" for r in fd145_nutrient_rows),
    "unavailable_nutrient_fields": sum(r["numeric_status"] != "available" for r in fd145_nutrient_rows),
    "valid_zero_nutrient_values": sum(r["numeric_status"] == "available" and Decimal(r["amount_per_100g_raw"]) == 0 for r in fd145_nutrient_rows),
    "selected_portion_records": len(fd145_portion_rows), "eligible_household_portions": sum(r["household_eligibility"] == "available" for r in fd145_portion_rows),
    "flagged_household_portions": sum(r["household_eligibility"] != "available" for r in fd145_portion_rows),
    "calculation_tests_executed": len(fd145_test_rows), "calculation_tests_failed": sum(r["status"] != "pass" for r in fd145_test_rows),
    "source_comparisons_executed": len(fd145_comparison_rows), "source_comparisons_failed": sum(r["status"] != "pass" for r in fd145_comparison_rows),
    "full_source_flagged_portions_preserved": fd145_validation["metrics"]["portion_numeric_screen_fail"],
    "source_values_changed_or_imputed": 0}
fd145_tables = {}


def fd145_table(name, rows, columns=()):
    fields = list(rows[0]) if rows else list(columns)
    fd145_require(bool(fields) and all(set(r) == set(fields) for r in rows), "Invalid prepared table schema: " + name)
    fd145_tables[name] = {"columns": fields, "rows": rows}


fd145_table("usda_foods.csv", fd145_food_rows)
fd145_table("usda_food_nutrients.csv", fd145_nutrient_rows)
fd145_table("usda_food_portions.csv", fd145_portion_rows, ("portion_id", "fdc_id", "amount_raw", "gram_weight_raw", "modifier", "portion_description", "measure_unit_id", "unit_name", "footnote", "household_numeric_screen_pass", "household_eligibility", "eligibility_reasons", "interpretation_policy", "source_food_portion_table"))
fd145_table("usda_coverage.csv", fd145_coverage)
fd145_table("usda_quality_findings.csv", fd145_quality_rows, ("fdc_id", "nutrient_id", "finding", "handling"))
fd145_table("development/usda_calculation_test_results.csv", fd145_test_rows)
fd145_table("development/usda_source_comparison_results.csv", fd145_comparison_rows)
fd145_table("development/usda_preparation_checks.csv", fd145_checks)
fd145_policy = {"schema_version": "step14-explicit-food-selection-policy-1", "policy_version": "ordinary-food-demo-1.0.0",
    "selection_type": "explicit_source_ID_and_exact_description_list", "selected_foods": 100,
    "source_archive_sha256": fd145_archive_sha,
    "purpose": "Familiar food-lookup examples with preparation and category variation; a demonstration subset, not a complete food database.",
    "inclusion": "Use the supplied explicit 100-ID list. Preserve exact source descriptions, including raw/cooked/prepared distinctions.",
    "not_used": ["nutrient rankings", "calorie targets", "athlete needs", "complete-fiber or portion requirements", "automatic name merging"],
    "scope_note": "Selected examples are not recommended foods, a balanced diet, or a population-representative sample.",
    "changes": "Any addition, removal, or identity replacement creates a new policy and reference version."}
fd145_dictionary = {"schema_version": "step14-selected-nutrient-definitions-1", "reference_mass_g": 100, "reference_basis": "edible_food",
    "definitions": list(fd145_expected_specs.values()), "energy_policy": fd145_contract["energy_policy"],
    "rounding": "Retain raw source text and Decimal precision in calculations. UI display rounding is a later, explicit presentation policy.",
    "quantity_semantics": "Complete record counts and counts of the same defined measure are distinct; only exact food-bound portion IDs are accepted.",
    "technical_limits": fd145_module.get("TECHNICAL_LIMITS", {}), "validation_producer": fd145_validation_ref}
fd145_handoff = {"schema_version": "step14-step24-food-reference-handoff-1", "metrics": fd145_metrics,
    "validation_producer": fd145_validation_ref, "reference_payload_paths": {"foods": "usda_foods.csv", "nutrients": "usda_food_nutrients.csv",
        "portions": "usda_food_portions.csv", "calculation_module": "usda_food_reference.py", "definitions": "usda_nutrient_definitions.json"},
    "retrieval_scope": "Only the named food/definition/portion files are lookup content. Development cases are software checks, not retrieval documents or planner research answer keys.",
    "data_preparation_complete": True, "application_lookup_enabled": False,
    "earlier_source_findings": {"validation_producer": fd145_validation_ref,
        "full_source_portion_numeric_flags": fd145_validation["metrics"]["portion_numeric_screen_fail"],
        "handling": "Full-source findings remain in the earlier audit; zero flagged selected portions does not resolve unselected source anomalies."},
    "integration_actions": ["Load and verify the selected producer and CSV payload hashes.",
        "Resolve ambiguous descriptions and quantities to exact selected food/portion IDs before arithmetic.",
        "Return food identity, units, reference quantity, missing-field status, and source provenance.",
        "Apply the same frozen food snapshot to all planner configurations receiving the tool."],
    "scope": "Factual food lookup only; no athlete calorie targets, diet prescriptions, supplement advice, or allergy-safety decisions."}
fd145_documents = {"usda_selection_policy.json": fd145_policy, "usda_nutrient_definitions.json": fd145_dictionary,
    "step14_step24_food_reference_handoff.json": fd145_handoff,
    "usda_reference_source_register_supplement.json": {"schema_version": "step14-reference-source-supplement-1",
        "source_manifest_modified": False, "source_acquisition_producer": fd145_acquisition_ref,
        "source_validation_producer": fd145_validation_ref, "source_archive_sha256": fd145_archive_sha,
        "source_table_paths": fd145_source_refs, "selected_fields_preserved": True,
        "source_review": "Source identities, definitions, basis, and literal comparison cases reviewed against official records; no independent human review is claimed."},
    "development/usda_independent_source_goldens.json": fd145_goldens}
fd145_extra_bytes = {"usda_food_reference.py": FD145_MODULE_SOURCE.encode("utf-8"),
    "development/usda_reference_calculation_tests.py": FD145_TEST_SOURCE.encode("utf-8"),
    "usda_food_selection.csv": FD145_SELECTION_CSV.encode("utf-8")}
fd145_payload_bytes = {name: fd144_csv_bytes(table) for name, table in fd145_tables.items()}
fd145_payload_bytes.update({name: fd144_json_bytes(doc) for name, doc in fd145_documents.items()})
fd145_payload_bytes.update(fd145_extra_bytes)
fd145_fingerprints = {name: hashlib.sha256(data).hexdigest() for name, data in fd145_payload_bytes.items()}
fd145_verify_inputs()
step14_report.update(status="food_reference_prepared", reference_prepared=True, reference_metrics=fd145_metrics,
    reference_artifact_fingerprints=fd145_fingerprints, reference_snapshot_ready_for_step24=False, application_lookup_enabled=False)
print("Step 14-5 — fixed food reference prepared and calculation cases executed")
print(f"Selected foods / source foods / source categories: {len(fd145_food_rows)} / {len(fd145_all_foods)} / {len(fd145_coverage)}")
print(f"Nutrient slots / available / unavailable: {len(fd145_nutrient_rows)} / {fd145_metrics['available_nutrient_fields']} / {fd145_metrics['unavailable_nutrient_fields']}")
print("Valid recorded nutrient zeros retained:", fd145_metrics["valid_zero_nutrient_values"])
print(f"Selected household records / eligible / flagged: {len(fd145_portion_rows)} / {fd145_metrics['eligible_household_portions']} / {fd145_metrics['flagged_household_portions']}")
print(f"Executed calculation cases / failures: {len(fd145_test_rows)} / {fd145_metrics['calculation_tests_failed']}")
print(f"Independent source comparisons / failures: {len(fd145_comparison_rows)} / {fd145_metrics['source_comparisons_failed']}")
print(f"Preparation checks passed / unchanged inspected inputs: {len(fd145_checks)} / {len(fd145_inputs)}")
print("Values changed or imputed / project files written / application lookup enabled: 0 / 0 / No")
print("Next: run Step 14-6 to save the frozen reference and Step 24 integration handoff.")
```

The input catalog and existing review history will remain unchanged by this step. Benchmark decisions and split assignments will be saved separately.

### 15.3 Verify and Review the Existing Groups

Step 10 already created deterministic `split_group_id` values. We will use those assignments as our starting point rather than independently create a second grouping system.

The existing groups link records through:

- Matching broader exercise-name keys.
- Repeated nonempty descriptions.
- Identical classifier inputs under the recorded strict comparison rule.

These links are transitive. If one record links to a second record, and the second links to a third, all three belong to the same group.

Labels, equipment, and difficulty did not determine these group assignments. A group can therefore contain records with different body-part labels.

We will verify that every catalog record has one group assignment and that the membership table agrees with the catalog. Missing descriptions must not connect otherwise unrelated exercises.

Our review will examine repeated names, shared descriptions, broader input matches, mixed-label groups, and groups connected through several relationships. The earlier EDA will guide this review.

| Finding | Treatment |
|---|---|
| Related records with supported shared content | Keep them in the same split group |
| Different labels within a group | Review the reason; do not automatically force one label |
| Generic shared wording | Preserve the existing group while examining whether its links are justified |
| Suspected relationship not covered by the current groups | Record the evidence before changing the grouping |
| Unsupported or incorrect group link | Document a versioned grouping correction before allocation |

We will retain eligible records in the initial benchmark rather than automatically remove every repeated name or description. Grouping controls where related records appear; it does not establish that they are duplicate activities.

Any change to the authoritative group mapping will require a documented preparation update, a new version, and refreshed consistency checks. We will not break groups to obtain more convenient split percentages.

Group assignments will be established before filtering records. Excluding a record will not automatically separate the remaining members of its group.

### 15.4 Decide Which Records Can Enter the Benchmark

The classifier will predict the body-part labels available in our exercise data. These labels are not an independently verified anatomical gold standard.

We will preserve their source and review status when defining the benchmark.

A record needs:

- A unique, traceable `catalog_id`.
- A usable exercise name and nonempty classifier input.
- A usable target label.
- A valid split-group assignment.
- No unresolved ambiguity that makes its supervised target unusable.

Missing descriptions alone will not exclude a record. Name-only examples remain part of the task we want to evaluate. Missing equipment or difficulty also will not automatically exclude an example because those fields are not classifier inputs.

The existing review queue contains several kinds of findings. We do not need to close every catalog-review item before creating a classification benchmark. We do need to identify findings that directly affect input validity, target interpretation, or separation between datasets.

For label decisions:

- Different exercises in one group may legitimately have different labels.
- Identical model inputs with conflicting labels require investigation.
- If reliable evidence cannot resolve such a conflict, we will withhold all affected conflicting examples from the benchmark.
- We will not choose a majority label simply to remove a disagreement.
- Any supported label correction will retain the original label, the benchmark target, the reason, and its evidence reference.

The original `body_part_prepared` value will remain in the catalog. The split manifest will record the `target_label` used for the benchmark and any associated decision reference. Without a documented correction, that target will retain the prepared source label.

Records with unresolved original-source verification may remain in the source-label benchmark when their inputs and targets are usable. Their review status will stay visible, and our later conclusions will reflect this limitation.

### 15.5 Define Class Eligibility and Minimum Support

We will assess each class using both record counts and distinct split-group counts after record-level exclusions.

Several related records in one group will not count as several separate groups.

Our initial project policy will require:

| Support requirement | Minimum distinct groups per class |
|---|---:|
| Available before allocation | 10 |
| Assigned to training | 6 |
| Assigned to validation | 2 |
| Assigned to test | 2 |

These are practical benchmark requirements chosen by our team. They do not guarantee reliable estimates or demonstrate statistical independence between groups.

This policy is separate from Step 10’s alert for classes with fewer than ten records.

The earlier neck category had eight records in eight groups. If that support remains unchanged, it will stay in the broader catalog but will not enter this initial classification benchmark.

We will report every excluded class and its reason. We will not merge unrelated body-part categories into an “other” category simply to increase support.

For mixed-label groups, support will be counted for every class represented by an eligible record in that group. We will not assign the whole group a majority label.

The eligible class list will be fixed before generating split candidates. If fewer than two classes qualify, we will stop and reconsider the benchmark design.

### 15.6 Create Reproducible Training, Validation, and Test Sets

We will target the following shares of eligible records:

| Split | Target share | Use |
|---|---:|---|
| Training | 70% | Fit text representations and model parameters |
| Validation | 15% | Compare configurations and select models |
| Test | 15% | Evaluate the frozen models after development |

Groups will remain indivisible. Exact record percentages may therefore be unavailable.

Our initial allocation procedure will use a fixed search over 1,000 candidate allocations:

1. Sort eligible records by `catalog_id` so input order is stable.
2. For candidate number `k`, from 0 through 999, use `GroupShuffleSplit` with a training share of 0.70 and seed `42 + k`.
3. Divide the remaining groups between validation and test using a second `GroupShuffleSplit`, a test share of 0.50, and seed `10042 + k`.
4. Check the resulting record and group counts for every eligible class.
5. Reject any candidate that fails the required separation or minimum-support checks.

`GroupShuffleSplit` allocates proportions of groups, not records, and does not itself stratify by class. We will therefore assess the actual record distribution of each candidate.

A candidate must also keep each overall record share within three percentage points of its target: 67–73% training and 12–18% each for validation and test. Individual class shares may differ more, especially for smaller classes.

Among the candidates that pass, we will select the allocation using this fixed order:

1. Lowest average absolute difference between each class’s actual split shares and the 70/15/15 targets.
2. Lowest average absolute difference between the overall record shares and those targets.
3. Lowest candidate number when the first two values tie.

For the first measure, we will calculate each class’s record share in each partition, compare it with the corresponding target, and average the absolute differences across all eligible classes and all three partitions.

We will save the candidate seeds, check results, balance measures, and selected allocation. No model will be trained or scored to choose the split.

If none of the 1,000 candidates passes, we will report that the bounded procedure found no acceptable allocation. That does not prove that an acceptable allocation is mathematically impossible. Any change to the policy or search procedure will be documented before generating a new version and before model training.

### 15.7 Keep Classifier Inputs Separate from Targets

The classifier input will contain only:

- `exercise_name_prepared`.
- `description_prepared`, when available.

Their recorded construction produces `classifier_text`. We will verify that this text follows the feature contract from Step 10.

The target will be the frozen `target_label` in the split manifest, derived from the prepared source label or a documented benchmark correction.

The input will exclude:

- Body-part target columns.
- Target-bearing `agent_text` and derived fields such as `goal_tag`.
- Equipment, difficulty, and exercise-type metadata.
- Record IDs, group IDs, source information, and review decisions.
- Fields describing group labels or review outcomes.

These fields may support auditing without becoming predictive features.

Naturally occurring anatomical words in an exercise name or description will remain valid input. We will not remove useful original wording simply because it resembles a class label.

### 15.8 Verify the Allocation and Freeze the Benchmark

Before accepting the split, we will check that:

- Every source catalog ID appears exactly once in the manifest.
- Every eligible record has exactly one split assignment.
- Excluded or withheld records have a reason and no training, validation, or test assignment.
- Eligible group IDs do not cross split boundaries.
- Matching inputs under the strict comparison rule do not cross boundaries.
- Broader comparison findings have a recorded handling decision.
- Every included class meets the minimum group support in all three splits.
- Input text, labels, IDs, and group assignments remain aligned.
- Eligible and excluded totals reconcile with the source catalog.
- Original catalog values and earlier review records remain unchanged.

We will report record counts, group counts, class coverage, description availability, and word-length summaries for each split. These summaries will describe the allocation, not guide later changes intended to improve model scores.

Group counts describe our recorded relationships. They do not prove that all remaining groups are statistically independent.

After the checks pass, we will freeze the manifest, eligible class list, and class ordering. The class-to-index mapping will use the eligible labels in alphabetical order. Step 18 will save the corresponding preprocessing artifact without changing that order.

The majority-class baseline, logistic regression model, neural classifier, and any planned input-text comparisons will use the same eligible records and split assignments.

### 15.9 Protect the Test Set During Later Work

Steps 18 and 19 will derive learned text preparation from training records only. Step 18 will build the neural vocabulary and sequence settings; Step 19 will fit the TF-IDF vocabulary and statistics within the complete logistic regression pipeline.

Class weights or resampling decisions will also use training labels only. Validation and test records will not be oversampled.

Validation results will support the model choices in Steps 19–21. The frozen test set will be used for final evaluation in Step 32.

We have already examined the full collected dataset during preparation and EDA. We will describe that honestly. Once the benchmark is frozen, further model-driven decisions will use training and validation data.

We will not choose a new split, remove difficult test examples, or change the class list after inspecting test performance.

If a genuine data error is discovered later, we will preserve the original benchmark, document the correction, and identify which experiments need to be repeated. Corrected results will belong to a new benchmark version.

### 15.10 Expected Outputs and Handoff

We expect to save:

| Output | Required content |
|---|---|
| `data/processed/exercise_groups.csv` | Verified group membership, catalog IDs, and references to the authoritative preparation version |
| `data/processed/exercise_split_manifest.csv` | One row per source record, including group, source label, target label, eligibility, exclusion reason, review reference, and split |
| `results/exercise_benchmark_review_decisions.csv` | Decisions affecting record eligibility, target labels, and grouping, with reasons and evidence |
| `results/exercise_class_eligibility.csv` | Record and group support, class inclusion decisions, thresholds, and per-split support |
| `results/exercise_split_candidates.csv` | Candidate seeds, rejection reasons, balance measures, and selection status |
| `results/exercise_split_summary.csv` | Overall and per-class counts, actual proportions, description coverage, and text-length summaries |
| `results/exercise_split_checks.csv` | Expected and observed values for the verification checks |
| `artifacts/splits/exercise_split_config.json` | The declared support policy, allocation method, search budget, seeds, ranking rules, and tolerances |
| `results/step15_exercise_split_summary.json` | Input and output hashes, schemas, recipe version, selected allocation, eligible class order, checks, limitations, and freeze status |

Existing files will be inspected before replacement. A different earlier version will be preserved rather than silently overwritten. Repeating an unchanged preparation should reproduce the same assignments.

We will consider this step complete when the selected allocation passes its required checks, the saved files agree with the producer record, and another teammate can load the same benchmark without choosing a new split.

The observations will report the actual retained records, eligible classes, exclusions, split sizes, group support, and remaining limitations after the code has run. We will not insert estimated counts or describe the split as complete before verification.

## 16. Fictional Athlete Profiles and Development/Test Separation

### 16.1 Purpose

We will prepare fictional athlete profiles to develop and evaluate our basketball planning application.

Each profile will describe a specific planning situation: what the athlete wants to improve, when they can practice, what resources they have, and which commitments or limitations the planner must consider.

These profiles will serve two purposes:

- Development cases will help us improve prompts, retrieval, application behavior, and evaluation procedures.
- Final test cases will let us compare all four planner versions on the same requests after development decisions are complete.

These are constructed scenarios, not records of real athletes or measured training outcomes. Our results will describe how the application handles the benchmark. They will not show that its plans improve athletic performance.

This dataset is separate from the exercise-classification benchmark prepared in Step 15.

### 16.2 Review Both Existing Profile Collections

We will begin with the two fictional profile sources already examined during EDA:

| Existing collection | Recorded size | Current role |
|---|---:|---|
| Earlier fictional athlete examples | 30 | Possible development material |
| Additional fictional basketball scenarios | 60 | Possible development material |

We will resolve their paths through the source manifest and verify their identifiers, schemas, and recorded file hashes.

The earlier review found missing planning information in both collections, including school grade, session length, existing commitments, sleep information, and annual participation history. Injury information also included unknown values.

Before reusing a record, we will check:

- How it was created and whether its fictional status is clear.
- Whether its request fits our basketball-planning scope.
- Which information is present, missing, unknown, or contradictory.
- Whether it duplicates or closely resembles another example.
- Whether its scenario label agrees with the actual request.

Suitable records may enter the development collection after review. Existing examples and their derivatives will remain outside the final test set.

The existing 90 records will not automatically become 90 approved development cases. Some may require revision, overlap with other records, or fail to fit our coverage plan.

When we add details to an existing fictional example, we will record those details as newly authored scenario information. We will not present them as facts recovered from the original file. The original record will remain unchanged.

The target of 100 development profiles will include any existing examples we accept; it is not an additional 100 profiles on top of them.

### 16.3 Define the Model-Visible Profile

We will use one versioned input schema for all four planner versions.

Each JSONL record will contain a stable `profile_id`, a schema version, and an `input` object. Only the approved contents of `input` will be used to construct the model-visible profile.

| Input component | Information to record |
|---|---|
| Age and school grade | Age in years and explicitly supplied grade |
| Basketball background | Experience level, current skill level, and position when relevant |
| Goals | Specific priorities such as shooting, passing, ball handling, or general conditioning |
| Planning window | Start date, end date, and time zone for one seven-day week |
| Availability | Days and time windows available for additional practice |
| Time limits | Maximum session duration and any stated weekly limit for additional sessions |
| Existing commitments | Practices, games, structured training, other sports, and their scheduled times |
| Equipment | Available items and quantities when those quantities affect the request |
| Space | Court or hoop access, usable area, and any stated restrictions |
| Participants | Total available participants, including the athlete |
| Reported limitations | Relevant restrictions, uncertainty, or requests that need a scope response |
| Optional history | Reported sleep or annual participation information when relevant |
| Athlete request | The natural-language request the application must answer |
| Optional food question | A factual food question, when included in the scenario |

Durations will use minutes. Schedule entries will use explicit dates and start/end times within the recorded time zone. We will choose and save the reference week before generating the scenarios.

Existing commitments will be recorded separately from availability for extra practice. A team practice still matters even if it falls outside the athlete’s stated extra-practice windows.

Activity descriptions must allow the relevant participation rules to be applied. Organized basketball, other sports, general conditioning, and informal activity will remain distinguishable.

We will not infer school grade from age, assume that a hoop means access to a full court, or assume that a training partner is available.

For missing information, the schema will distinguish:

- An absent field: information was not provided.
- `null`: the information is explicitly unknown.
- An empty list: the scenario explicitly confirms that no items or commitments exist.
- Zero or `false`: a stated value, used only where it has a valid meaning.

For example, an unknown practice schedule must not become an empty commitment list.

### 16.4 Set the Development and Final Test Targets

We plan to prepare 100 development profiles and 200 final test profiles.

Our development allocation will provide additional room to examine challenging requests:

| Age | Feasible | Missing information | Conflicting constraints | Outside scope | Total |
|---|---:|---:|---:|---:|---:|
| 13 | 14 | 2 | 2 | 2 | 20 |
| 14 | 14 | 2 | 2 | 2 | 20 |
| 15 | 14 | 2 | 2 | 2 | 20 |
| 16 | 14 | 2 | 2 | 2 | 20 |
| 17 | 14 | 2 | 2 | 2 | 20 |
| **Total** | **70** | **10** | **10** | **10** | **100** |

The final benchmark will retain the planned composition used by the later evaluation steps:

| Age | Feasible | Missing information | Conflicting constraints | Outside scope | Total |
|---|---:|---:|---:|---:|---:|
| 13 | 30 | 4 | 3 | 3 | 40 |
| 14 | 30 | 4 | 3 | 3 | 40 |
| 15 | 30 | 4 | 3 | 3 | 40 |

These are design targets. We will report actual accepted counts after preparation and review.

Balanced age counts describe our experimental design. They do not represent the age distribution of real basketball players.

### 16.5 Define the Scenario Categories

Each profile will have one primary scenario category in its private evaluation record.

| Category | Meaning | Expected response |
|---|---|---|
| Feasible request | Enough information and resources are available for at least one supported plan | Produce a complete plan that respects the stated constraints |
| Missing essential information | A required decision cannot be made without further information | Ask focused clarification questions |
| Conflicting constraints | Known requirements cannot all be satisfied together | Explain the conflict and an appropriate next step |
| Outside scope | The central request falls outside the application’s defined role | Explain the boundary and respond appropriately |

A missing optional detail will not automatically make a request incomplete. Position, for example, may be unnecessary for a general ball-handling plan.

Conflict cases must contain an identifiable contradiction or an established conflict with an adopted planning rule. Limited resources alone do not make a request impossible if a reasonable supported alternative exists.

A gap in our drill collection is also different from a contradiction in the athlete’s request. We will record reference-coverage problems as dataset issues rather than label them as athlete constraint conflicts.

Outside-scope cases may involve requests for injury rehabilitation, supplement recommendations, individualized dietary prescriptions, or unrelated services, according to the project policies developed in Step 13.

Each case must have a clear main issue. Secondary tags may describe additional features, but they will not change the primary category after planner outputs are available.

Missing-age, invalid-format, and outside-age-range inputs will be additional application tests in Step 30. They will not replace the age-balanced final cases above.

### 16.6 Plan Coverage Before Drafting Profiles

We will prepare a coverage matrix before writing or generating the cases.

Alongside the age and scenario quotas, the matrix will track:

- Basketball experience and skill level.
- Training goals.
- Available days and session time.
- Equipment and court access.
- Solo, partner, and group arrangements.
- Existing practices, games, and other activities.
- Relevant age–grade combinations.
- Complete and deliberately incomplete information.
- Requests containing explicit food-facts questions.

We will include ordinary situations as well as challenging ones. Examples include a player with limited equipment, a player with short practice windows, and a player whose existing commitments leave little room for extra training.

Age and grade combinations will be checked for coherence without assuming that every age corresponds to one grade.

Food questions will concern facts supported by the frozen food reference, such as a named food’s recorded nutrient values or a supported quantity calculation. They will not ask the system to calculate an athlete’s nutritional needs.

The development collection must provide at least two suitable food-question cases for the pilot in Step 31. We will record food-question coverage in the final set before its freeze.

We will use a recorded seed, initially 42, for Python-based ordering and selection. If a language model helps draft cases, we will save its identity, prompts, settings, and generated drafts. The reviewed, saved files—not a promise of identical regeneration—will define the benchmark.

### 16.7 Keep Related Profiles Together

We will assign every profile a `family_id`.

A family will connect cases derived from the same underlying example, including paraphrases, renamed copies, and controlled variations of an existing scenario.

Changing a name or adjusting one field does not automatically create an independent case.

Existing profile families will be assigned to development. Any new variation of those examples will remain there.

For new material, we will assign the base family to development or final testing before creating its variations. We will prefer distinct base cases in the final set rather than fill it with paraphrases.

Our checks will examine:

- Exact duplicate inputs.
- Requests with closely repeated wording.
- Matching combinations of meaningful profile facts.
- Recorded parent–child relationships.
- Similarities between new final cases and existing development material.

Shared formatting or the same broad training goal will not, by itself, establish a family.

If a final candidate is found to be a derivative of development material, we will remove it from final consideration and replace it with a distinct case before freezing.

We will report both profile and family counts. Family IDs will also support the paired analysis in Step 35, where related profiles and all four versions’ outputs must remain together.

### 16.8 Prepare Private Expected Behaviors

Each profile will have a separate evaluation record prepared before final planner outputs are examined.

It will contain:

- Profile and family identifiers.
- Primary scenario category.
- Expected response type.
- Constraints that must be respected.
- Applicable adopted rule IDs and their versions.
- Required checks and their justification.
- Missing information that requires clarification.
- Conditions that are optional, unknown, or not applicable.
- Unsupported assumptions to avoid.
- Source or project-policy references supporting the expectation.

We will not define one exact weekly plan as the only correct answer. Several schedules or drill selections may satisfy a request.

For each case classified as feasible, we will keep a private feasibility record showing that at least one plan can satisfy its required conditions. This may be a manually constructed example schedule checked against the available activities, resources, time limits, and adopted rules.

That example will establish that a supported solution exists. It will not become a model answer that every planner must reproduce.

Unknown sleep or annual participation history will not automatically invalidate an otherwise assessable weekly request. The expected record will state whether the missing information is essential to that case or should remain an identified limitation or reminder.

Expected behaviors, feasibility examples, scenario labels, required-check masks, family IDs, reviewer notes, and partition labels will remain outside model-visible inputs and retrieval documents.

Version 4’s internal evaluator may use the public profile and adopted general rules. It must not receive these private case-specific evaluation records. Final scoring will remain a separate process.

### 16.9 Review Cases Before Accepting Them

One teammate will prepare each profile and its expected behavior. A second teammate will check the request, important constraints, category, and supporting evidence.

The review will ask:

- Does the public profile clearly describe the intended situation?
- Are the schedule and resource details internally coherent?
- Is any missing or conflicting information deliberate?
- Does the expected response follow from the supplied facts?
- Are the required checks supported by adopted rules or documented project policies?
- For feasible cases, does the private feasibility record support the classification?
- Has any evaluation information leaked into the model-visible request?
- Are the source and family relationships recorded correctly?

We will save the reviewers’ actual decisions and resolve disagreements before acceptance. Drafting assistance will not substitute for recorded team review.

This is structured project-team review. It is not external coaching or clinical validation.

At the Step 14 checkpoint recorded in this notebook, the basketball catalog had no planner-eligible drills and the planning rules were not yet adopted. We will recheck those statuses before accepting profiles.

We can prepare profile drafts and coverage records now. Final feasibility approval and rule-dependent expectations must wait until the relevant catalog content, rules, and required lookup operations are available and reviewed.

We will not mark all 150 planned-feasible final cases as accepted merely because they have complete-looking fields.

### 16.10 Verify Coverage and Support Later Samples

Before freezing the dataset, we will check that:

- Profile IDs are unique.
- Each profile has one family and one partition.
- No family crosses development and final testing.
- Every accepted profile has one private expected-behavior record.
- Private evaluation fields are absent from model-visible inputs.
- Dates, times, durations, and participant counts follow the schema.
- Age and scenario totals match the declared quotas.

- Feasible cases have accepted feasibility records.

### 16.11 Freeze the Final Set and Keep the Comparison Paired

After review, we will freeze the public inputs, family assignments, private expectations, reference versions, and dataset configuration.

The final profiles will not be used as prompt examples, retrieval documents, classifier training material, or development tests.

## 17. Evaluation Metrics, Scoring Rules, and Human Review Rubric

### 17.1 Purpose

We will define how we judge our models before examining final test results.

Our evaluation has two main questions:

1. How accurately do the exercise classifiers predict the dataset’s body-part labels?
2. How well do the four planning workflows respond to the fictional athlete profiles?

These questions need different evidence. A correct body-part prediction does not establish that an exercise is appropriate for a particular athlete. Similarly, a technically valid weekly schedule does not prove that every instruction is useful or well supported.

For the planners, automated checks will assess requirements that can be measured consistently. Human review will examine meaning, practicality, factual support, and clarity.

This step defines the evaluation design. Actual model scores, planner outcomes, and reviewer ratings will be produced later.

### 17.2 Establish the Evaluation Inputs and Boundaries

The protocol will refer to the versioned artifacts prepared in the earlier steps.

| Artifact | Evaluation role |
|---|---|
| Exercise split manifest and class list from Step 15 | Define the classifier records, targets, groups, and partitions |
| Profile inputs and private expectations from Step 16 | Define planner requests, scenario categories, constraints, and expected behaviors |
| Adopted planning rules from Step 13 | Supply the applicable conditions and their qualifications |
| Approved basketball reference content from Step 12 | Support activity requirements and instructional evidence |
| Frozen food reference from Step 14 | Support factual nutrient checks and permitted quantity calculations |
| Shared input, output, and parsing contracts | Define how responses are submitted and interpreted |

We can prepare the protocol while some dependencies remain open. Final scoring, however, will require the adopted versions and completed reference-dependent expectations.

Draft rules, prepared test specifications, and successful data checks will not be reported as completed model evaluation.

Case-specific expectations will remain private to the independent evaluation process. Planner prompts, retrieval documents, and Version 4’s internal evaluator must not load the expected answers, feasibility examples, or private scoring masks from Step 16.

### 17.3 Define the Classifier Metrics

The majority-class baseline, TF-IDF logistic regression model, and neural classifier will use the same frozen records and eligible class list.

| Measure | Interpretation |
|---|---|
| Macro-F1 — primary measure | Mean of the class-level F1 scores, with equal weight for each included class |
| Accuracy | Correct predictions divided by evaluated records |
| Per-class precision | Correct predictions for a class divided by all predictions of that class |
| Per-class recall | Correct predictions for a class divided by its actual examples |
| Per-class F1 | Harmonic mean of that class’s precision and recall |
| Confusion matrix | Counts showing which labels were predicted for each actual class |
| Support | Record counts and distinct split-group counts |

Implementation will use the fixed class order from Step 15. Macro-F1 will use `average="macro"` and `zero_division=0`. Per-class reporting will use the same label list and zero-division policy.

Confusion matrices will show actual labels in rows and predicted labels in columns. We will save counts and, where useful, row-normalized proportions.

Zero-division results will be accompanied by their support. A numerical zero caused by absent examples is different from observed failure on existing examples.

Validation macro-F1 will guide model selection. Final test metrics will be calculated after configurations and checkpoints have been selected.

The majority baseline will be learned from training labels. The earlier full-catalog majority share will remain an EDA description rather than a measured baseline result.

We will report agreement with the available benchmark labels and retain the source-label limitations documented in Steps 10 and 15.

### 17.4 Define the Primary Planner Measure

Our primary planner measure will be the **automated feasible-plan success rate**.

For each planner version:

**Success rate = successful feasible cases ÷ all 150 feasible final profiles**

A case succeeds only when:

1. The workflow produces its designated returned response.
2. The response can be interpreted under the shared parsing contract.
3. It provides a plan for a feasible request.
4. The plan contains the required information and meets the requested planning scope.
5. Every required applicable automated check is met.

A partial plan will not pass simply because the activities it includes are valid. It must satisfy the completeness requirements established for that request.

The required checks will be defined before final outputs are examined. They must follow from the public request, shared response contract, adopted rules, or documented project policies—not a private preference for one particular answer.

Timeouts, failed executions, missing responses, malformed outputs, inappropriate refusals, incomplete plans, and unknown required checks will prevent success.

A planner cannot pass by omitting activities and producing no activity checks. The absence of the required plan is itself a failure.

All 150 feasible profiles remain in the denominator for each version. Jobs that have not started or are awaiting permitted resumption will remain marked incomplete; we will not describe an unfinished experiment as a completed comparison.

This measure describes compliance with our automated contract. It does not establish training effectiveness or individual medical suitability.

### 17.5 Define Automated Check Results and Coverage

Every check result will retain:

- Run, profile, and planner-version identifiers.
- Rule ID and rule version.
- Applicability and required/advisory status.
- The condition being tested.
- Observed values, units, and evidence references.
- Result and explanation.
- Checker version and any execution error.

We will use the four statuses established in Step 13.

| Status | Meaning |
|---|---|
| Met | Available information supports the condition |
| Violated | Available information establishes a conflict |
| Unknown | Required information or rule interpretation is insufficient |
| Not applicable | The adopted applicability rule establishes that the condition does not apply |

The applicability procedure and profile-based obligations will be fixed before final evaluation. Checks triggered by proposed activities will follow the same rules across versions.

A missing value will not become “not applicable.” If required information is insufficient, the condition will remain unknown and the feasible case will not succeed.

A checker execution error will be recorded separately as an assessment failure. It will keep the affected external evaluation incomplete until resolved, rather than count as evidence that the planner violated the condition.

Advisory unknowns will be reported separately. Missing sleep or annual participation history, for example, will not automatically fail a weekly plan unless that information is essential to the specific request and adopted check.

When one rule examines several activities, we will preserve the individual findings and aggregate them consistently:

- Any established violation makes the parent rule violated.
- Otherwise, unresolved information needed for the rule makes it unknown.
- The rule is met only when all required instances have been assessed successfully.
- Not applicable requires a documented applicability reason.

This keeps comparisons from being dominated by how many small checks a particular response generates.

For expected applicable parent checks:

**Conclusive check coverage = (met + violated) ÷ (met + violated + unknown)**

Justified not-applicable checks will be reported separately. A zero denominator will produce an unavailable ratio with its counts, not an automatic 100%.

Coverage measures whether we reached a conclusion. It is not a pass rate. Required and advisory coverage will be reported separately.

### 17.6 Report Other Planner Outcomes with Their Denominators

The primary result will be supported by separate measures.

| Measure | Population and reporting |
|---|---|
| Plan completion | Complete plans among all 150 feasible cases per version |
| Constraint outcomes | Met, violated, unknown, and not applicable by rule and scenario |
| Check coverage | Conclusive outcomes among expected applicable checks |
| Response-type match | Declared response type compared with the case’s expected category |
| Missing-information handling | Results for the 20 missing-information cases |
| Conflict handling | Results for the 15 conflict cases |
| Scope handling | Results for the 15 outside-scope cases |
| Execution and parsing reliability | Outcomes across all 200 scheduled cases per version |
| Evidence findings | Source validity, factual support, and attribution coverage, with the assessed sample identified |
| Runtime and resource use | Complete workflow measurements, including failed attempts |

Response-type match is a structural measure. A response labeled “clarification” may still ask an irrelevant question or miss the essential information.

Human review will assess that distinction on the selected sample.

We will report the four scenario categories separately rather than create one headline score that could reward refusing feasible requests.

Age-group summaries will show their actual denominators. Under the planned design, each age has 30 feasible cases and ten challenge cases per version.

### 17.7 Assess Evidence and Retrieval Separately

Evidence evaluation will distinguish three questions:

1. Does the supplied reference identify an available source record?
2. Does the evidence support the associated claim?
3. Are important source-dependent claims missing usable attribution?

Automatic reference-ID checks may cover all returned identifiers. Factual support will require structured source comparisons or human review, depending on the claim.

For each evidence measure, we will state what was assessed and its denominator.

Examples include:

- Resolvable distinct reference IDs divided by supplied distinct IDs.
- Supported cited claims divided by cited claims assessed.
- Source-dependent claims with usable attribution divided by source-dependent claims reviewed.

No supplied references means the ID-validity ratio is unavailable with a count of zero. It does not mean perfect citation performance.

When factual claims are present without citations, their accuracy can still be checked against the adopted references. Missing attribution and factual inaccuracy will remain separate findings.

Reviewers will examine relevant age groups, units, quantities, conditions, and qualifications. A valid link alone does not establish support.

We will also distinguish a planner’s scheduling choice from a source prescription. Assigning ten minutes to an activity does not mean that the publisher prescribed that duration.

Retrieval will have its own benchmark in Step 23, separate from the athlete-profile comparison. Its planned measures include Hit@k and Recall@k at 1, 3, and 5; MRR@5; and whether delivered context contains an accepted complete evidence bundle.

Retriever selection will use development Recall@5, followed by MRR@5 for ties and the simpler lexical method if both remain equal. Unsupported questions and evidence-adequacy failures will be reported separately.

Retrieval scores will not substitute for planner scores or factual review of the generated response.

### 17.8 Define the Human Review Rubric

We will use five dimensions, each scored from 0 to 2.

| Dimension | 0 — Does not meet expectations | 1 — Partly meets expectations | 2 — Meets expectations |
|---|---|---|---|
| Response suitability | Misses the central issue or gives an inappropriate response | Addresses the issue but leaves an important part unresolved | Gives the appropriate plan, clarification, conflict explanation, or scope response |
| Profile fit | Ignores or contradicts relevant profile information | Uses some relevant details but misses important constraints or goals | Reflects the information needed for this request |
| Practicality and completeness | Is unusable or substantially incomplete for its purpose | Provides some useful guidance but leaves important gaps | Gives sufficiently complete and practical guidance for the request |
| Evidence accuracy | Contains materially unsupported or misleading claims | Has partial support or omits important qualifications | Agrees with the reviewed evidence and communicates relevant uncertainty |
| Clarity | Is confusing or internally inconsistent | Is understandable but noticeably ambiguous | Is clear, consistent, and easy to follow |

Completeness will be judged relative to the request. A good clarification should identify the essential missing information; it does not need to invent a weekly plan.

Likewise, a correct scope response need not discuss an irrelevant playing position. Profile fit concerns the information that matters to the response.

We will prepare development examples showing the difference between scores 0, 1, and 2. Comments will explain important problems and identify the supporting source when a factual claim is disputed.

A claim that cannot be verified is not automatically proven false. Review comments will distinguish contradiction, insufficient support, and missing qualifications.

We will retain the five dimensions separately. There will be no combined human-quality score or new passing threshold introduced after seeing final results.

### 17.9 Handle Failed Outputs, Non-Applicability, and Missing Ratings

Each dimension entry will have an explicit status.

| Rating status | Stored value |
|---|---|
| Rated | Numeric score of 0, 1, or 2 |
| Not applicable | No numeric score, with a required reason |
| Missing | No numeric score because assessment is unfinished |

The four non-evidence dimensions apply to every benchmark case.

For an executed job that returns no response, empty text, or wholly unusable content, response suitability, profile fit, practicality/completeness, and clarity will receive 0.

Malformed but readable text will be reviewed as received. Reviewers may recognize useful content even though its formatting caused an automated failure. They will not repair the response before rating it.

An inappropriate refusal can be clearly written. Its clarity score will reflect that wording, while suitability and completeness reflect its failure to answer the request.

Evidence accuracy applies whenever substantive factual or instructional claims are present, including claims in refusals and scope responses. A plan does not escape evidence review merely because it lacks citations.

When there are no substantive claims to assess, evidence accuracy may be not applicable. An explicitly requested factual answer that was omitted will remain a documented completeness failure.

A missing reviewer rating is unfinished work. It will not be converted into 0 or not applicable, or filled using another reviewer’s opinion.

Unstarted jobs are incomplete experimental work rather than observed model failures. Their review records will remain pending until the run is resolved under the protocol.

### 17.10 Select and Conduct the Human Review

We will select 40 final profiles before examining generated outputs.

| Scenario category | Profiles selected |
|---|---:|
| Feasible requests | 25 |
| Missing essential information | 5 |
| Conflicting constraints | 5 |
| Outside scope | 5 |
| **Total** | **40** |

Each age from 13 through 17 will contribute eight profiles: five feasible and one from each challenge category.

Selection will use stable profile ordering and a recorded random seed, initially 42, within these age–scenario groups. We will save the selected IDs and family information.

All four version outcomes will be included for every selected profile. This produces 160 response records, each assessed independently by two teammates.

Reviewers will receive:

- The original profile and request.
- The actual returned response, including its citations.
- The rubric.
- Neutral access to the adopted reference materials.

Version labels, workflow traces, internal evaluator findings, automated scores, timing, private scenario labels, and feasibility examples will remain outside the review materials.

Presentation order will be randomized reproducibly. Generated content will remain unchanged, even when its style or references provide clues about the workflow.

Reviewer practice will use development examples. Initial ratings will be saved and locked before discussion or adjudication.

This is independent team rating: reviewers work separately on their first assessments. It is not external coaching or clinical validation, and complete blinding cannot be guaranteed.

The sample deliberately includes a larger share of challenge cases than the full benchmark. We will report its scenario composition rather than present pooled ratings as representative of all 800 planner jobs.

### 17.11 Measure Reviewer Agreement and Paired Differences

Agreement will be calculated separately for each rubric dimension.

For responses with two numeric ratings, we will report:

- Number of jointly rated responses.
- Exact agreement: matching scores divided by jointly rated responses.
- Linearly weighted Cohen’s kappa, using the fixed ordinal order 0, 1, and 2.

Not applicable will not become a fourth ordinal score. We will separately report both-not-applicable cases, one-reviewer-only non-applicability, missing ratings, and each reviewer’s score distribution.

Undefined kappa will be reported as unavailable with an explanation. We will not replace it with zero or perfect agreement. Original ratings and later consensus decisions will remain separate.

Planner comparisons will preserve matching profiles and their family IDs. Our planned comparisons are:

- Version 2 minus Version 1.
- Version 3 minus Version 2.
- Version 4 minus Version 3.
- Version 4 minus Version 1.

Success-rate differences will be reported in percentage points. We will also count cases improved, worsened, or unchanged.

The uncertainty procedure finalized in Step 31 will use 2,000 paired group-bootstrap resamples, seed 42, and individual 95% percentile intervals.

Exercise groups will be resampled for classifier comparisons. Profile families will be resampled for planner comparisons, keeping matched version outcomes together. Metrics will retain their original record-level or profile-level meaning.

These intervals describe sensitivity to the cases in our constructed benchmark. They do not measure real-athlete outcomes or training-seed variability.

### 17.12 Keep Final Scoring Independent of Generation

The four planners will receive the same public profile inputs and shared output requirements. The base language model and common generation settings will remain consistent.

Their intended differences—retrieval, specialist roles, internal evaluation, and revisions—will be recorded as parts of the compared workflows.

Independent scoring will operate on saved returned responses before user edits.

Version 4’s internal acceptance flag will not determine its research result. A returned plan will pass or fail the independent checks regardless of that flag.

If Version 4 reaches its revision limit and returns its last checked candidate, that candidate will be scored under the same rules. A fatal workflow error will remain a failed execution; an earlier diagnostic draft will not silently replace the designated response.

Optional source-ID fields will remain optional under the shared format. Missing attribution may affect evidence findings without becoming an invented formatting requirement.

We will preserve failures and will not select the strongest answer from multiple completed generations unless that selection is explicitly part of the frozen workflow.

Any change prompted by final results will belong to a documented new experiment rather than a silent revision of the current comparison.

### 17.13 Record Runtime and Resource Usage

For each planner job, we will record:

- Total elapsed time and available stage-level timings.
- Logical model calls and transport attempts.
- Retrieval and structured lookup calls.
- Coordinator, specialist, evaluator, and revision activity.
- Input and output tokens where available.
- Revisions, retries, timeout status, and terminal outcome.
- Available device and resource measurements.
- Cost estimates only where a recorded pricing or estimation basis exists.

The total must include the complete workflow, not only its last generation call.

Startup or model-loading measurements will remain separate from request timing. We will identify cache behavior and any common setup policy.

Successful-response latency and time consumed by failures will be reported separately. Unavailable measurements will remain unavailable rather than become zeros.

Timeouts and retry limits will be established using development cases and frozen before final testing.

### 17.14 Verify the Scoring Design and Save the Protocol

Before final evaluation, we will check the scoring implementation using small development examples with independently specified expected outcomes.

These checks will include:

- A complete feasible plan.
- An incomplete plan with otherwise valid activities.
- A missing response that produces no activity checks.
- Values below, equal to, and above an adopted threshold.
- Unknown required information.
- A justified not-applicable condition.
- A valid reference that does not support its claim.
- Factual claims without supplied citations.
- Malformed but readable response text.
- Numeric, not-applicable, and missing human-rating entries.

We expect to save:

| Output | Content |
|---|---|
| `docs/evaluation_protocol.md` | Metrics, populations, denominators, aggregation, comparisons, and failure handling |
| `docs/human_review_rubric.md` | Scoring anchors, applicability rules, and development examples |
| `artifacts/evaluation/scoring_config.json` | Machine-readable metric settings, rule versions, check policies, and protocol status |
| `data/evaluation/human_review_sample.csv` | Selected profile IDs, strata, families, seed, and sampling version |
| `data/evaluation/human_review_template.csv` | Anonymous response and reviewer identifiers, dimensions, rating statuses, scores, and comments |
| `results/evaluation_protocol_checks.csv` | Expected and observed outcomes from scoring checks |
| `results/step17_evaluation_design_summary.json` | Input/output hashes, configuration references, readiness checks, and unresolved dependencies |

Final sample IDs and response assignments will be populated only when the required profile and execution records exist. Templates will not be presented as completed reviews.

This step will establish the evaluation design. The development pilot in Step 31 will confirm operational settings and finalize the protocol before held-out evaluation begins.

We will consider the protocol ready when its definitions are unambiguous, required dependencies are adopted, development scoring checks pass, and the saved configuration identifies the exact version to be used.

Actual scores, ratings, agreement statistics, and runtime findings will be added after the relevant experiments have run.

## 18. Training-Only Text Preprocessing and Feature Preparation

### 18.1 Purpose

In this step, we will prepare the exercise text for our logistic regression and neural classifiers.

Both models will receive the exercise name and its available description. They will use different numerical representations of that text: TF-IDF features for logistic regression and token IDs for the neural network.

Our main goal is to make this preparation reproducible while keeping the training, validation, and test partitions separate. Fixed text rules can be applied consistently across partitions, but vocabulary, word-frequency statistics, and other learned settings must come from training records only.

This step prepares model inputs. It does not train the classifiers or establish that either model performs well.

### 18.2 Verify the Catalog and Frozen Splits

We will begin with the prepared exercise catalog from Step 10 and the frozen split artifacts from Step 15.

Before preparing any features, we will verify that:

- The input files match their recorded versions and hashes.
- Each catalog identifier is unique.
- Eligible manifest records match the corresponding catalog records.
- Group assignments and split membership agree with Step 15.
- No exercise group crosses the training, validation, and test boundaries.
- The target labels and ordered class list match the frozen benchmark configuration.

We will join records by `catalog_id`. We will not assume that two files have the same row order.

The Step 15 manifest will determine eligibility, partition membership, and the benchmark `target_label`. We will not reconstruct those decisions from the original catalog labels or assume that every original class remains eligible.

If a check fails, we will resolve the mismatch before continuing. We will not silently remove records, change labels, or create a new split during preprocessing.

### 18.3 Keep Input Text Separate from Targets and Metadata

Step 10 already created `classifier_text` from the prepared exercise name followed by its available description, separated by two newline characters when both are present.

We will reuse and verify that saved text. The original catalog fields and `classifier_text` will remain unchanged.

| Information | Use in this step |
|---|---|
| Exercise name and available description | Classifier input |
| Benchmark target label from Step 15 | Prediction target |
| Catalog identifier, group identifier, and split | Alignment and auditing |
| Original label and review status | Traceability |
| Source information | Traceability |
| Target-bearing `agent_text` and derived `goal_tag` | Excluded from classifier inputs |
| Separate equipment, difficulty, and exercise-type fields | Excluded from this text-only comparison |

Naturally occurring anatomy or equipment words in the exercise name and description will remain valid input. For example, we will retain “shoulder” when it appears in the original exercise text.

Athlete profiles, basketball reference documents, nutrition records, and private evaluation expectations will not become exercise-classifier features.

The target represents the benchmark’s recorded body-part category. Predicting it does not establish an exercise’s suitability for a particular athlete.

### 18.4 Apply a Shared Text Preparation Policy

We will create a separate `model_text` representation from the saved `classifier_text`.

Our initial preparation will:

1. Apply Unicode NFKC normalization.
2. Replace repeated whitespace with a single space.
3. Remove leading and trailing whitespace.
4. Apply consistent Unicode case folding.

These operations will not change the protected source text.

We will begin without stop-word removal, stemming, lemmatization, or generated descriptions. Words such as “not,” movement directions, anatomy terms, and useful numbers may carry information that we want the models to retain.

A missing description will remain absent. It will not become the literal text `"nan"` or `"None"`, and we will not fill it with an estimated description.

For tokenization, our initial shared expression will be:

`r"(?u)\b\w+(?:[.'’/-]\w+)*\b"`

This keeps word-like tokens and allows internal apostrophes, hyphens, slashes, and decimal points. For example, `one-arm`, `don't`, `1/2`, and `2.5` can remain individual tokens.

We will check representative examples before freezing this policy. These checks will confirm the intended behavior; they will not use test records to choose a more favorable tokenizer.

The normalization and tokenization functions will live in an importable project module so training, saved models, and application predictions can use the same implementation.

### 18.5 Maintain the Training Boundary

Each partition will have a defined role.

| Partition | Permitted use |
|---|---|
| Training | Build learned representations and train model parameters |
| Validation | Apply training-derived preparation and compare development choices |
| Test | Apply frozen preparation during final evaluation in Step 32 |

Vocabulary counts, inverse-document-frequency values, neural vocabulary membership, and the sequence-length setting will not use validation or test records.

Validation results can support the predefined model comparisons in Step 21. They will not add words to a fitted vocabulary or change training-derived statistics.

During Step 18, we will prepare training and validation inputs. Test identity and integrity checks may be performed, but test tokenization, vocabulary-coverage analysis, truncation diagnostics, feature transformation, and predictions will wait until Step 32.

Our earlier dataset-wide EDA will remain documented. We will not describe the project as though the source data had never been inspected.

### 18.6 Define the TF-IDF Configuration

The logistic regression classifier will use TF-IDF features for individual tokens and adjacent token pairs.

We will start with the following fixed configuration.

| Setting | Initial choice |
|---|---|
| Analyzer | Word |
| Tokenizer | Shared tokenizer from Section 18.4 |
| Token pattern | `None`, because tokenization is supplied explicitly |
| Additional lowercasing | Disabled; handled by shared normalization |
| Accent stripping | None |
| N-gram range | Unigrams and bigrams: `(1, 2)` |
| Maximum features | 10,000 |
| Minimum document frequency | 1 document |
| Maximum document frequency | 1.0 |
| Stop words | None |
| Binary term counts | Disabled |
| IDF weighting and smoothing | Enabled |
| Sublinear term-frequency scaling | Disabled |
| Row normalization | L2 |
| Numeric representation | Float64 sparse matrix |

We will supply the custom function as the tokenizer while keeping the word analyzer. This preserves the configured bigram construction.

Step 18 will save this configuration. **Step 19 will fit the complete preprocessing and logistic regression pipeline on training records.** This gives us one clear owner for the fitted TF-IDF vocabulary and IDF values.

Validation inputs will use that fitted pipeline. We will keep the feature matrices sparse and record their dimensions and nonzero counts in the model-training step.

TF-IDF’s L2 row normalization and logistic regression’s L2 regularization serve different purposes. We will record them separately.

The majority-class baseline will use training labels and will not require TF-IDF features.

### 18.7 Build the Neural Vocabulary from Training Text

The neural classifier will represent each token with an integer ID.

We will reserve two separate IDs:

| ID | Meaning |
|---:|---|
| 0 | `PAD`: a position added for batching |
| 1 | `UNK`: a token outside the retained training vocabulary |

Ordinary vocabulary tokens will begin at ID 2.

Our initial vocabulary policy will:

- Count tokens across the full, untruncated training inputs.
- Retain tokens appearing at least twice in training.
- Keep at most 10,000 ordinary tokens.
- Rank tokens by descending training frequency.
- Resolve frequency ties using a fixed lexical order.

The maximum vocabulary size will therefore be 10,002 entries, including `PAD` and `UNK`. The saved artifact will report the actual size.

The frequency threshold refers to total token occurrences, rather than the number of documents containing a token.

Rare training tokens, tokens beyond the vocabulary cap, and unseen validation tokens will map to `UNK`. Validation and test records will never expand the vocabulary.

In Step 20, we will train the embedding weights from scratch. The padding row will remain zero and be excluded from pooling. We will initialize the unknown-token row to zero and allow it to learn when unknown-token positions occur in training.

We will report whether `UNK` actually appears in the training sequences after truncation. If it does not, we will document that its embedding has not learned a fallback representation.

### 18.8 Set Sequence Length, Truncation, and Padding

We will choose the neural sequence-length limit from training token lengths only.

The initial rule will be:

**Maximum sequence length = the training 95th-percentile token length, bounded between 16 and 256 tokens.**

We will calculate the percentile using the nearest-rank method and save the resulting integer. Token lengths will be measured before truncation, after applying the empty-token handling described below.

The exercise name already appears first in the input. We will retain the first tokens up to the saved limit and truncate the remainder on the right.

This keeps the beginning of the text, but it may remove information from long descriptions. We will report how often truncation occurs and how many tokens it removes.

Within each batch, sequences will be padded on the right to the longest retained sequence in that batch, without exceeding the saved limit.

The neural inputs will include:

- Integer token IDs.
- A Boolean mask identifying non-padding positions.
- Retained sequence lengths.
- Integer target class IDs.

For masked mean pooling, padding must be excluded from both the embedding sum and the token count. `UNK` represents an input position and must remain in the count.

An all-padding sequence will be treated as an implementation or input-contract error.

The sequence-length rule and vocabulary will remain fixed during the initial model comparisons in Steps 20 and 21.

### 18.9 Preserve Labels and Handle Difficult Inputs Consistently

We will copy the ordered class list and label-to-index mapping from Step 15.

We will verify that every eligible target maps to one class index and maps back to the same label. An unexpected target will stop preparation for investigation.

Token IDs and target class IDs belong to different mappings. In particular, target class ID 0 is a valid class and must not be ignored because padding also uses token ID 0.

If class weights are used later, they will be calculated from training labels only. We will not resample validation or test records.

Difficult inputs will follow predefined handling rules.

| Situation | Required handling |
|---|---|
| Missing description with a valid exercise name | Retain the name-only input |
| Genuinely empty canonical input | Stop and reconcile the data-contract problem |
| Valid nonempty text producing no tokens | Use one `UNK` position for the neural model and flag the record |
| Valid neural input containing only unknown tokens | Retain it and record the condition |
| Valid all-zero TF-IDF row | Keep the fitted pipeline’s normal prediction and retain the record in evaluation |
| Unexpected label, identifier, or split mismatch | Stop and investigate |

We will not remove difficult examples, extend a vocabulary using held-out text, or substitute a different prediction rule to improve reported scores.

### 18.10 Check Reproducibility and Keep the Comparison Fair

Before training, our checks will confirm that:

- Training and validation record counts match the frozen manifest.
- Text, targets, and identifiers remain aligned.
- Only permitted text reaches the feature preparation functions.
- Repeated preparation produces the same normalized text and token IDs.
- Vocabulary IDs are unique and within the recorded range.
- Padding and unknown tokens have different IDs.
- Sequence lengths, masks, and truncation follow the saved policy.
- Label encoding preserves the frozen class order.
- Every eligible training and validation record remains accounted for.

We will report training and validation token lengths, unknown-token rates, truncation counts, and the number of name-only inputs. Where useful, these diagnostics will also be summarized by class.

TF-IDF feature dimensions and zero-vector checks will be added when the pipeline is fitted in Step 19.

Both classifiers will receive the same underlying exercise text and benchmark records. Their representations will still differ: TF-IDF uses the full prepared text, while the neural model applies the documented sequence limit.

We will acknowledge this difference when interpreting results. A performance gap cannot automatically be attributed to network architecture alone.

### 18.11 Expected Outputs and Handoff

After completing this step, we expect to save:

| Output | Purpose |
|---|---|
| `src/champion_athletic_map/text_preprocessing.py` | Shared normalization, tokenization, and neural input preparation |
| `data/processed/exercise_model_inputs.csv` | Training and validation text, identifiers, targets, and class indices |
| `artifacts/preprocessing/preprocessing_config.json` | Fixed text rules, TF-IDF settings, vocabulary policy, and actual sequence limit |
| `artifacts/preprocessing/label_mapping.json` | Frozen class mapping inherited from Step 15 |
| `artifacts/preprocessing/neural_vocabulary.json` | Training-derived token mapping, retained frequencies, and reserved IDs |
| `results/preprocessing_diagnostics.csv` | Training and validation coverage, length, and truncation summaries |
| `results/preprocessing_checks.csv` | Input, alignment, mapping, and reproducibility checks |
| `results/step18_preprocessing_summary.json` | Producer record linking inputs, settings, checks, and saved outputs |

The model-input table may contain targets for alignment, but training code will select the text and target columns explicitly. Targets and audit fields will never be concatenated into the input.

The producer record will include input and output hashes, the ordered training identifiers, the split-manifest version, software versions, and the actual vocabulary and sequence dimensions.

Step 19 will fit the TF-IDF baseline pipeline. Step 20 will train the neural classifier using the saved vocabulary and sequence policy. Step 22 will package the selected model with its exact preprocessing dependencies.

Passing this step will establish that the development inputs are prepared consistently. Model performance and final test results will be reported in the later evaluation steps.

## 19. Majority-Class and TF-IDF Logistic Regression Baselines

### 19.1 Purpose

We will build two baseline classifiers before training the neural model.

The first will always predict the most common body-part label in the training partition. The second will learn from exercise names and descriptions using TF-IDF features and logistic regression.

These baselines will answer two practical questions:

- Does learning from the exercise text improve on predicting the most common label?
- How well does a relatively simple text classifier perform before we introduce a neural network?

The results will give us a starting point for the later comparison. We will report the findings even if logistic regression performs better than the neural model.

This step evaluates prediction of the benchmark’s body-part labels. It does not establish whether an exercise is appropriate for a particular athlete.

### 19.2 Confirm the Inputs and Evaluation Boundary

We will use the eligible benchmark records and frozen splits from Step 15, together with the preprocessing configuration from Step 18.

The Step 15 manifest will provide the authoritative `target_label`, group assignment, and partition membership. The class mapping will follow its frozen alphabetical class order.

Before fitting either baseline, we will check that:

- Input files match their recorded versions and hashes.
- Catalog identifiers are unique and match the development inputs.
- Exercise text, targets, and identifiers remain aligned.
- Training and validation counts agree with the manifest.
- No exercise group appears in more than one partition.
- Every eligible class has the required training and validation support.
- The label mapping matches the Step 15 configuration.
- Name-only examples remain included where eligible.
- Only the permitted exercise text reaches the classifier.

We will join files by `catalog_id`, rather than depend on row order.

The original source labels remain subject to the review limitations documented earlier. We will describe the targets as benchmark labels rather than imply that all of them have received independent expert confirmation.

Both baselines will use the same training records and will be scored on the same validation records. Final test transformation, diagnostics, predictions, and scoring will remain in Step 32.

### 19.3 Establish the Majority-Class Baseline

We will use `DummyClassifier` with `strategy="most_frequent"`.

This model will count the benchmark labels in the eligible training partition and predict the most frequent label for every validation record. It will ignore the exercise text.

We will record:

- The selected majority label and its class index.
- Its training record count.
- Its proportion of the eligible training records.
- Any tie between classes.

If training counts are tied, our policy will select the tied class with the lowest frozen class index. We will verify that the fitted dummy classifier follows that choice.

The dummy model can receive a constant placeholder input with the required number of rows. It will not require a vocabulary or text features.

Its majority label will be determined once from training records. Validation labels will be used only to score its predictions.

We will calculate validation accuracy and macro-F1 from the actual validation predictions. The majority proportion reported in earlier whole-dataset EDA will not substitute for this result because the eligible records and partition distribution may differ.

### 19.4 Fit the Complete TF-IDF Logistic Regression Pipeline

Our learned baseline will contain three stages:

1. Apply the shared Step 18 normalization to the canonical `classifier_text`.
2. Convert the normalized text into TF-IDF features using the shared tokenizer.
3. Predict the benchmark body-part category with logistic regression.

The pipeline will accept the canonical name-and-description text. Its normalization output must match the `model_text` prepared in Step 18.

Keeping normalization inside the saved pipeline will allow us to reproduce the same behavior when we reload the model or use it in the application.

We will fit the complete pipeline on training records only. Validation will use the existing vocabulary, IDF values, and classifier parameters.

The TF-IDF configuration will remain exactly as defined in Step 18, including:

- Word unigrams and bigrams.
- The shared tokenizer.
- A maximum of 10,000 features.
- A minimum document frequency of one.
- No stop-word removal.
- L2 row normalization.
- Sparse float64 features.

The neural vocabulary will remain separate. Its minimum token-frequency rule and reserved token IDs will not be applied to TF-IDF.

Our initial logistic regression settings will be:

| Setting | Initial choice |
|---|---|
| Regularization | L2 |
| Inverse regularization parameter, `C` | 1.0 |
| Solver | `lbfgs` |
| Class weighting | None |
| Sample weighting | None |
| Stopping tolerance | `1e-4` |
| Maximum iterations | 1,000 |
| Intercept | Enabled |
| Warm start | Disabled |
| Training-data resampling | None |

For more than two eligible classes, we will use the solver’s multinomial classification behavior.

We will check the installed scikit-learn version and use its supported parameters to express the intended L2 regularization. The run record will preserve the actual constructor settings and library version.

These are starting settings. Their performance will be measured rather than assumed.

### 19.5 Check Feature Preparation and Solver Behavior

After fitting, we will record the actual vocabulary size, training and validation matrix dimensions, nonzero feature counts, and zero-vector counts.

The number of learned features may be below the configured maximum.

Our checks will confirm that:

- Training and validation have the same feature columns.
- Validation transformation leaves the fitted vocabulary and IDF values unchanged.
- Feature values, coefficients, and intercepts are finite.
- The classifier’s class indices match the frozen label mapping.
- Every prediction belongs to the eligible class list.
- The feature matrices remain sparse.

A valid nonempty input may produce an all-zero TF-IDF vector when none of its features appears in the learned vocabulary. It will remain in evaluation and receive the pipeline’s normal prediction.

An entirely empty fitted training vocabulary will be treated as a failed preparation condition.

For optimization, we will record fitting time, solver iteration counts, and warnings. A completed `fit` call alone will not be treated as proof that optimization converged.

Our initial convergence policy will be:

1. Fit a fresh pipeline with the 1,000-iteration limit.
2. If it raises a convergence warning, preserve that attempt and refit once from a fresh pipeline with a 5,000-iteration limit.
3. Keep the data, TF-IDF settings, regularization, and tolerance unchanged during that retry.
4. If the retry still does not converge, mark the configuration as unresolved and unsuitable for model selection until the issue is addressed.

The retry will be triggered by optimization behavior, not by validation performance. We will not choose whichever attempt happens to score higher.

The same policy will apply to the logistic regression candidates in Step 21. Any further change will require a documented amendment and consistent reruns of affected comparisons.

### 19.6 Evaluate Both Baselines on Validation Records

We will follow the scoring definitions established in Step 17.

| Measure | Purpose |
|---|---|
| Macro-F1 | Primary measure for comparing classifier configurations |
| Accuracy | Overall proportion of correctly classified records |
| Per-class precision, recall, and F1 | Show performance for individual body-part categories |
| Confusion counts | Show which labels are confused |
| Row-normalized confusion matrix | Show the distribution of predictions within each true class |
| Record and distinct-group support | Describe the amount and grouping of available evidence |

All classification reports will use the complete frozen class list and `zero_division=0`.

Confusion matrices will use true labels as rows and predicted labels as columns. Both axes will follow the same frozen class order.

The main scores will be calculated over records. Distinct-group counts will describe how many exercise groups support those results; they will not silently change the scores into group-weighted metrics.

Every eligible validation record will remain in the denominator, including name-only examples and valid zero-vector inputs.

We will compare the models using unrounded scores. Rounded values will be used only for display.

If a run fails, its status and reason will be recorded. Missing results will remain unavailable rather than become zero scores.

### 19.7 Review Validation Errors

We will link incorrect predictions back to their exercise text and group identifiers.

Our review will consider whether mistakes are associated with:

- Short exercise names or missing descriptions.
- Vocabulary not represented in training.
- Similar movements assigned to different categories.
- Exercises that involve several body parts.
- Classes with relatively few distinct exercise groups.
- Possible inconsistencies in the source labels.

We will distinguish model mistakes from suspected data issues. A model disagreement will not, by itself, justify changing a target.

A confirmed correction will require a documented benchmark revision and consistent reruns. We will not edit validation labels while retaining scores calculated from the earlier version.

Where useful, we will summarize performance for name-only records and records with descriptions. These will be diagnostic breakdowns, while the primary comparison will continue to use the complete validation partition.

The observations will explain what the models learned successfully and where the available text or labels limit the comparison.

### 19.8 Keep Tuning Separate from the Initial Baseline

Step 19 will establish the majority baseline and one initial logistic regression configuration.

In Step 21, we will compare the predefined six logistic regression configurations:

| Parameter | Candidate values |
|---|---|
| `C` | 0.1, 1.0, 10.0 |
| Class weighting | None, balanced |

The Step 18 text preparation and TF-IDF settings will remain fixed throughout that comparison.

Balanced class weights will be derived from training labels only. We will not change the validation distribution.

The initial `C=1.0`, unweighted run from this step is included in that search. It may be reused when its data hashes, software version, preprocessing, model settings, and convergence policy match the recorded experiment.

Validation macro-F1 will remain the selection measure. We will not choose a different metric after seeing which result looks strongest.

These validation results will be development evidence. The separate test comparison will take place after model selection and freezing.

### 19.9 Save Predictions and Verify the Fitted Artifacts

For each baseline, we will save validation predictions with:

- Catalog identifier and exercise group identifier.
- True target label and class index.
- Predicted label and class index.
- Model, configuration, and run identifiers.
- Relevant input or preprocessing flags.

Each model should have exactly one prediction per eligible validation record.

Any saved probability columns will follow the frozen class order. We will describe these as model outputs rather than calibrated confidence estimates. The dummy model’s one-hot probabilities will not be interpreted as certainty about the true label.

We will save the fitted dummy classifier and the complete fitted TF-IDF logistic regression pipeline.

After saving, we will reload the artifacts and check that they reproduce the recorded development predictions. The saved pipeline must use importable preprocessing functions rather than depend on an active notebook session.

The run metadata will identify the ordered training records, data and split hashes, preprocessing version, software versions, actual parameters, feature count, warnings, and timing information.

We will record the shared experiment seed for traceability. We will not claim that it controls optimization randomness for `lbfgs`, which does not use the estimator’s random-state setting.

## 20. Neural Exercise Classifier Design and Training from Scratch

### 20.1 Purpose

We will build a compact neural text classifier that predicts a benchmark body-part category from an exercise name and its available description.

The model will learn its token embeddings and classification weights from scratch using the training partition. Its training history, parameter-update checks, and saved weights will document the neural-model training component of our project.

We will compare its validation results with the majority-class and TF-IDF logistic regression baselines from Step 19. The comparison will help us determine whether this neural approach offers a useful improvement.

### 20.2 Verify the Inputs and Training Boundaries

We will use the frozen benchmark splits from Step 15 and the saved text preparation from Step 18.

Before training, we will verify that:

- Input files and preprocessing artifacts match their recorded hashes.
- Records, targets, and group assignments match the split manifest.
- Inputs contain only the permitted exercise name and description.
- The vocabulary and sequence-length setting were derived from training text.
- Token IDs follow the saved vocabulary.
- Target IDs follow the frozen class mapping.
- Training and validation groups remain separate.
- Name-only examples remain included where eligible.

The Step 15 `target_label` will remain authoritative for this experiment. We will not replace targets with model predictions or assume that every source label has received independent expert confirmation.

The number of output classes will come from the frozen mapping. We will not hard-code the original category count.

Validation will support checkpoint selection and development comparisons. Final test transformation, diagnostics, predictions, and scoring will remain in Step 32.

### 20.3 Define the Initial Network

Our initial architecture will combine trainable embeddings with masked mean pooling and a small classification network.

| Component | Initial design | Purpose |
|---|---|---|
| Token input | Integer IDs from the saved vocabulary | Represent the prepared text |
| Embedding layer | 64 dimensions per token | Learn token representations |
| Masked mean pooling | Average non-padding embeddings | Produce one vector per exercise |
| Hidden layer | 64 units with ReLU activation | Learn combinations of text features |
| Dropout | 0.2 | Apply regularization during training |
| Output layer | One raw score per eligible class | Produce classification scores |

For a batch containing `B` examples, sequence length `L`, and `K` eligible classes, the main shapes will be:

| Tensor | Shape and type |
|---|---|
| Token IDs | `[B, L]`, integer `torch.long` |
| Non-padding mask | `[B, L]`, Boolean |
| Token embeddings | `[B, L, 64]`, floating point |
| Pooled representation | `[B, 64]`, floating point |
| Output logits | `[B, K]`, floating point |
| Targets | `[B]`, integer `torch.long` |

We will use the recorded PyTorch initialization defaults, with the padding and unknown-token exceptions described below. No pretrained embeddings or pretrained classifier weights will be loaded.

The architecture is deliberately small so we can train it within our project resources and examine its behavior.

Mean pooling loses token order. We will therefore describe this as a bag-of-tokens neural classifier, without claiming that it understands the full meaning of exercise instructions.

### 20.4 Handle Padding, Unknown Tokens, and Long Inputs

The model will use the Step 18 vocabulary and sequence policy without rebuilding them.

The reserved token IDs will remain:

- `PAD = 0`
- `UNK = 1`

The embedding layer will use `padding_idx=0`. Its padding row will remain zero, and pooling will exclude padding from both the embedding sum and the token count.

Unknown tokens represent real input positions. They will remain in the pooling denominator even when their embedding is initially zero.

Following Step 18, we will initialize the unknown-token embedding to zero and allow it to learn when `UNK` occurs in training. We will record its actual training occurrence count after truncation. If it never occurs, we will document that its embedding did not receive a training signal.

Inputs will retain the saved name-first order, right truncation, and right padding. We will not change the sequence limit after inspecting validation performance or final test text.

Valid all-unknown inputs will remain in evaluation and be flagged. A valid nonempty input that produces no tokens will follow the saved one-`UNK` fallback.

An all-padding sequence or genuinely empty canonical input will trigger investigation rather than be silently removed.

### 20.5 Set the Loss and Training Configuration

We will use multiclass cross-entropy with the frozen numerical target IDs.

The network will return raw class scores, or logits. These will go directly to the loss function without applying softmax first.

Our initial training configuration will be:

| Setting | Initial value |
|---|---|
| Optimizer | AdamW |
| Learning rate | 0.001 |
| Weight decay | 0.0001 |
| Batch size | 32 |
| Maximum epochs | 30 |
| Class weighting | None |
| Label smoothing | 0 |
| Validation frequency | Once after each complete epoch |
| Checkpoint measure | Validation macro-F1 |
| Early-stopping patience | Five consecutive checks without improvement |
| Initial seed | 42 |
| Model precision | Float32 |

For AdamW, we will record `betas=(0.9, 0.999)`, `eps=1e-8`, `amsgrad=False`, and the remaining effective optimizer settings.

The initial experiment will use a constant learning rate without a scheduler, training-data resampling, or text augmentation.

Token padding and target classes use separate ID systems. **Target class 0 remains a valid category and must not be ignored by the loss because the padding token also uses ID 0.**

Any later class weighting will use training label frequencies only and follow the predefined comparison in Step 21.

### 20.6 Prepare a Reproducible Run

Before creating the recorded model, optimizer, and data loaders, we will set the Python, NumPy, and PyTorch random seeds.

We will use a separately seeded generator for training-example shuffling.

Our initial loader policy will be:

| Setting | Training | Validation |
|---|---|---|
| Shuffle | Yes | No |
| Worker processes | 0 | 0 |
| Drop incomplete final batch | No | No |

Keeping the final batch will ensure that every eligible example is included.

We will select the working CPU or GPU verified earlier and record the device, software versions, and relevant backend settings. Deterministic algorithm checks will be enabled, with cuDNN benchmarking disabled where applicable.

If an operation cannot follow the chosen deterministic settings, we will resolve and document the issue before accepting the run.

These settings support repeatability within the recorded environment. They do not guarantee identical numerical results across different devices or software versions.

### 20.7 Check the Implementation Before Full Training

We will first run a small training-only diagnostic.

The checks will confirm that:

- Token and target IDs are within their permitted ranges.
- Tensor types and output dimensions are correct.
- Loss, gradients, and model parameters remain finite.
- Relevant embedding and classification parameters can receive updates.
- The padding row remains zero.
- Extra padding does not change evaluation-mode predictions beyond the recorded numerical tolerance.
- Pooling excludes padding and counts unknown-token positions correctly.

If needed, a small training-only learning check can help identify an implementation problem. It will remain separate from the reported experiment.

Diagnostic updates can alter weights and consume random numbers. After these checks, we will reset the seeds and create a fresh model, optimizer, and shuffle generator for the recorded run.

### 20.8 Run Separate Training and Validation Loops

During training, we will enable training mode and process the shuffled training batches.

For each batch, we will:

1. Move inputs and targets to the recorded device.
2. Clear gradients from the previous update.
3. Calculate logits and cross-entropy loss.
4. Backpropagate the loss.
5. Check for invalid values.
6. Update the model parameters.

Validation will use evaluation mode with gradient recording disabled. Dropout will be inactive, and validation will not update the model or vocabulary.

We will collect predictions for the complete validation partition and calculate macro-F1 once over those predictions. We will not average separate batch-level F1 scores.

For the initial unweighted loss, epoch loss will equal the summed example losses divided by the total number of examples. This will account correctly for a smaller final batch.

The weighted-loss candidate in Step 21 will require a weight-aware denominator: summed weighted losses divided by the summed target-class weights.

Each epoch record will include:

- Training and validation loss.
- Validation macro-F1 and accuracy.
- Number of training and validation examples processed.
- Epoch duration.
- Whether a new best checkpoint was saved.
- The current early-stopping counter.

Training loss is accumulated while parameters change during the epoch. Validation loss uses the epoch’s resulting model in evaluation mode. We will keep this distinction in mind when interpreting the curves.

Nonfinite values or execution failures will stop the run and remain in the log. An interrupted experiment will retain its incomplete status rather than silently become a valid model-selection candidate.

### 20.9 Save and Restore the Best Checkpoint

Checkpoint selection will use unrounded validation macro-F1.

We will initialize the best score below any valid F1 value so the first valid epoch establishes a checkpoint, including when its macro-F1 is zero.

After each epoch:

- A strictly higher macro-F1 will save a new best checkpoint and reset patience.
- An equal or lower score will retain the earlier checkpoint and increase the counter.
- Training will stop after five consecutive non-improving checks or after 30 epochs.

Validation loss and accuracy will remain diagnostic measures. They will not replace macro-F1 as the checkpoint selection rule.

We will save a copied set of the best weights or serialize them immediately. A reference to the live model state is insufficient because later updates could change it.

After training stops, we will restore the best checkpoint and recompute its complete validation predictions and metrics.

We will verify that the restored result matches the recorded best epoch. The final training epoch will not automatically become the selected model.

### 20.10 Review Learning Behavior and Compare with the Baselines

We will plot training and validation loss alongside validation macro-F1 across epochs.

The observations will address:

- Whether the model learned beyond its initial behavior.
- Whether validation performance improved consistently.
- Whether later training showed signs of overfitting.
- Which classes were most often confused.
- Whether name-only, truncated, or low-coverage inputs were difficult.
- Whether limited class or group support affects interpretation.

The restored checkpoint will be compared with both Step 19 baselines on exactly the same validation records.

We will use the Step 17 scoring conventions: fixed-label macro-F1, accuracy, per-class precision and recall, `zero_division=0`, confusion counts, row-normalized confusion matrices, and record and distinct-group support.

Predictions will use the highest-scoring class. Exact score ties will select the lowest frozen class index.

Any displayed softmax probabilities will be described as uncalibrated model outputs.

These results will measure benchmark body-part classification. They will not establish youth suitability or the effectiveness of a personalized training plan.

This initial experiment will be **candidate N0 with seed 42** in Step 21. It can be reused there when its data, preprocessing, architecture, training settings, and environment match the recorded comparison protocol.

### 21.3 Compare Six Logistic Regression Configurations

| L3 | 1.0 | Balanced |

| L5 | 10.0 | Balanced |

We will evaluate candidates in this order. Smaller `C` values apply stronger regularization.

For each candidate, we will fit a fresh complete normalization, TF-IDF, and logistic regression pipeline using training records only. The TF-IDF settings will remain unchanged.

Balanced class weights will be calculated from the training targets. We will not introduce sample weighting or resampling.

Each fit will follow the convergence policy from Step 19:

2. If a convergence warning occurs, preserve that attempt and refit once from a fresh pipeline with a 5,000-iteration limit.

This retry is part of the optimization policy. It is not another hyperparameter candidate, and its use will not depend on validation scores.

### 21.4 Compare Four Initial Neural Configurations

**Class weight = total training records ÷ (number of eligible classes × training records in that class)**

We will verify that every eligible class has training support and store the weights in the frozen class order.

### 21.6 Select the Research Artifacts and Tagging Candidate

We will distinguish selecting a neural configuration from selecting a saved checkpoint.

### 21.7 Review the Results Without Changing the Rules

- Per-class precision, recall, and F1.
- Confusion counts and row-normalized confusion matrices.
- Record and distinct-group support.
- Name-only, truncated, and low-coverage input cases.
- Neural learning curves and selected epochs.

These diagnostics will help us explain why a configuration performed differently. They will not become replacement selection measures chosen after seeing the results.

Weighted and unweighted loss values represent different objectives. We will use them to examine learning behavior rather than directly rank those configurations by loss.

Our interpretation will also acknowledge the representation difference established in Step 18: logistic regression uses the full prepared text, while the neural model applies the saved sequence limit.

Repeated validation use for early stopping and tuning means the reported validation scores are development evidence. Step 32 will provide the separate assessment of the frozen artifacts.

### 21.8 Record Failures and Control the Search Budget

Every scheduled run will have a recorded status and explanation.

Failed or incomplete runs will have unavailable scores rather than zero scores. They will remain visible in the run log.

A shortlisted neural configuration must have valid results from all three scheduled seeds before its mean can qualify for selection. We will not average only successful runs, replace a failed seed with a new one, or silently promote another configuration into the shortlist.

Valid candidates may be ranked after the scheduled experiments have recorded outcomes. Unresolved implementation errors, fewer than two valid initial neural configurations, or no qualifying three-seed configuration will keep the selection checkpoint open.

A documented failure and an unfinished run will remain distinguishable. We will explain when a ranking includes fewer valid candidates than originally planned.

Corrective reruns will retain the original attempt and record what changed. If a shared implementation error affects several candidates, we will rerun the affected comparisons consistently.

The run summary will distinguish:

- Planned configurations and seed combinations.
- Verified runs reused from earlier steps.
- New fitting attempts.
- Logistic regression convergence retries.
- Completed, failed, and unresolved runs.
- Candidates eligible for selection.

Any extension to the search will require a versioned amendment explaining the reason and additional work. We will not keep adding configurations until a preferred model wins.

### 21.9 Freeze the Selected Models and Decisions

Before packaging, we will verify that the selected artifacts are valid, reload correctly, and reproduce their recorded validation predictions.

We will then freeze:

- Benchmark and split versions.
- Input and source-code hashes.
- Selected configuration and run identifiers.
- The fitted majority-class baseline.
- The selected complete logistic regression pipeline.
- The selected neural seed-42 checkpoint and epoch.
- Preprocessing, vocabulary, sequence settings, and class mapping.
- Validation predictions and selection evidence.
- The tagging-model decision, including a disabled or pending status where applicable.

The comparison models will retain their original training-partition fits. We will not retrain them on combined training and validation data before the planned final evaluation.

The classifier supplied to later planning workflows will use an explicitly identified version. Step 32 results will not be used to switch that classifier while treating the experiment as unchanged.

Any later model revision will belong to a separately documented development cycle.

### 21.10 Expected Outputs and Handoff

After completing this step, we expect to save:

| Output | Contents |
|---|---|
| `artifacts/tuning/search_plan.json` | Candidate order, seeds, fixed settings, selection rules, and search version |
| `artifacts/tuning/runs/` | Fitted pipelines, best neural checkpoints, run configurations, and verified references to reused artifacts |
| `results/tuning/candidate_runs.csv` | Run and attempt identifiers, parameters, reuse information, status, warnings, and failures |
| `results/tuning/validation_metrics.csv` | Validation results for completed runs |
| `results/tuning/neural_seed_summary.csv` | Individual scores, three-seed means, standard deviations, and eligibility |
| `results/tuning/selected_validation_predictions.csv` | Aligned predictions for the selected comparison artifacts |
| `results/tuning/selection_checks.csv` | Search completeness, artifact, alignment, and decision checks |
| `artifacts/model_selection.json` | Selected artifacts, tagging status, scores, and decision reasons |
| `docs/model_selection_summary.md` | Plain-language findings, tradeoffs, and limitations |

The producer record will distinguish a completed search from a provisional ranking. It will also identify any unresolved condition that prevents freezing the selected models.

Step 22 will package the selected artifacts and define their use for catalog tagging.

Actual scores, selected configurations, and conclusions will be added after the experiments have been executed.

## 22. Saving the Classifier and Tagging the Catalog

### 22.1 Purpose

We will package the selected classifiers so our teammates and application can load them without repeating training.

We will also prepare a controlled workflow for suggesting body-part tags for compatible new or unlabeled exercise records.

Three decisions will remain separate:

- Whether the saved model bundle loads and works correctly.
- Whether a suggested body-part tag has been accepted through review.
- Whether an exercise is eligible for use in a youth training plan.

A successful model-loading check will establish the first of these. It will not automatically establish the others.

### 22.2 Verify and Preserve the Selected Models

We will begin with the model-selection record and producer summary from Step 21.

The research bundle will preserve all three comparison artifacts.

| Model | Artifact to preserve |
|---|---|
| Majority-class baseline | The fitted model with its training-derived constant prediction |
| Logistic regression | The selected complete normalization, TF-IDF, and classifier pipeline |
| Neural classifier | The selected configuration’s best seed-42 checkpoint |

We will verify the model identifiers, input hashes, class mapping, preprocessing versions, and validation results against the selection record.

Packaging will preserve the actual selected artifacts. It will not refit a model, change the neural seed, combine checkpoints, or rebuild a vocabulary.

The tagging-model decision will also come directly from Step 21. If tagging was disabled, packaging the research models can still succeed.

If model selection remains incomplete, we may prepare a provisional bundle, but we will record that status rather than declare a verified release.

### 22.3 Package the Models and Their Dependencies

Each release will have a versioned folder under:

`artifacts/classifiers/<release_id>/`

The bundle will contain the fitted models together with the information needed to interpret their inputs and outputs.

| Bundle component | Purpose |
|---|---|
| Fitted majority model | Reproduce the baseline prediction |
| Complete logistic regression pipeline | Preserve normalization, tokenization, vocabulary, IDF values, and classifier parameters |
| Neural state dictionary | Restore the selected learned weights |
| Neural architecture configuration | Reconstruct the correct network dimensions |
| Preprocessing configuration | Reproduce the Step 18 text rules |
| Neural vocabulary | Preserve token-to-ID assignments |
| Sequence and masking settings | Reproduce truncation, padding, and pooling |
| Label mapping | Decode class indices correctly |
| Dependency information | Recreate the recorded model environment |
| Required project source modules | Load custom functions and network definitions outside the notebook |

The reusable implementation will remain under `src/champion_athletic_map/`. The release will include a versioned copy of the required source package with its import paths preserved.

We will record both the Git revision, where available, and the hashes of the actual source files. This will also account for relevant changes that have not yet been committed.

For scikit-learn models, we will use `joblib` with the recorded dependency versions.

For the neural model, we will save its `state_dict` and reconstruct the network from the saved configuration. Loading will use an explicit device mapping, `weights_only=True` where supported, and strict parameter matching.

These artifacts will support inference and evaluation. They will not be described as complete checkpoints for resuming the exact training process.

### 22.4 Create a Versioned Manifest

The bundle’s `manifest.json` will identify:

- Release and model IDs.
- The three research artifacts.
- The selected tagging model, or its disabled or pending status.
- The Step 21 selection record and decision reason.
- Dataset and split versions.
- Preprocessing, vocabulary, sequence, and label-map versions.
- Neural configuration, seed, and selected epoch.
- Validation scores used for selection.
- Source-code revision and file hashes.
- Python and relevant package versions.
- Training, reference-inference, and intended application devices.
- Bundle filenames and their hashes.
- Supported classes, input requirements, and known limitations.

Paths inside the bundle will be relative to its release directory so the package can be moved without depending on one teammate’s folder structure.

The manifest will contain hashes for its payload files. The external Step 22 producer record will hash the completed manifest and verification reports.

We will preserve earlier releases. A repeated run may reuse an identical verified bundle, but it will not silently overwrite a different release.

The application will load the explicit model ID in the manifest rather than assume that the newest model file is the selected one.

### 22.5 Provide a Common Prediction Interface

We will create a reusable interface that accepts an exercise name and an optional description.

It will:

1. Validate the supplied fields.
2. Construct the same name-first canonical text used in Step 10, with two newline characters between nonempty name and description fields.
3. Apply the saved Step 18 preparation exactly once.
4. Run the explicitly selected model.
5. Decode the result with the frozen class mapping.
6. Return the prediction, model version, and relevant handling flags.

For logistic regression, canonical text will go directly into the complete saved pipeline. The neural interface will apply its saved normalization and token preparation. Each route will apply its preparation exactly once.

A missing description will be allowed. A missing or invalid exercise name will return an explained input error.

Identifiers, labels, and review metadata will remain outside the model features.

The response will distinguish a produced prediction from disabled tagging, invalid input, unsupported records, deferred records, or a loading error.

Depending on the model, handling flags may include name-only input, truncation, unknown-token coverage, or an all-zero TF-IDF vector.

If scores are returned, the response will identify their type and class order. Probabilities will remain uncalibrated model outputs rather than verified confidence.

Missing files, incompatible mappings, or failed integrity checks will return a clear error. They will not trigger an undocumented replacement model or majority-class fallback.

### 22.6 Verify Loading in a Fresh Process

We will test loading in a fresh Python process using the recorded dependencies and bundled project modules.

This process must work without relying on fitted objects or function definitions held in notebook memory.

We will compare predictions for the complete saved validation partition with the selected artifacts’ reference predictions.

The checks will confirm that:

- Required files and source modules are available.
- File hashes match the manifest.
- Canonical input construction and normalized text match the saved development inputs.
- Vocabulary, sequence settings, and class ordering are correct.
- Decoded predictions match the recorded results.
- Numerical outputs remain finite and agree within the predefined tolerance.
- Prediction does not alter learned parameters or preprocessing artifacts.
- Neural inference uses evaluation mode with gradients disabled.

Our initial numerical comparison tolerance will be `rtol=1e-5` and `atol=1e-6`. Predicted labels must match exactly in the recorded reference environment.

We will also check the intended application device. A label difference will require investigation even when the underlying scores are numerically close.

Separate fixed synthetic fixtures will check missing descriptions, extra whitespace, long text, unknown tokens, and padding behavior. These fixtures will test input handling, rather than provide additional classification-accuracy evidence.

No held-out test examples will be used for these checks.

### 22.7 Define the Tagging Population

When Step 21 has selected an available tagging model, the initial workflow will cover:

- Eligible training and validation records used for development previews.
- Compatible new general-exercise records needing a body-part suggestion.

Existing source labels will be preserved. We will not bulk-replace them with model predictions.

Training previews will be marked as in-sample outputs. Validation previews will be marked as development outputs. Neither will become additional evidence of generalization.

Held-out test records and their groups will remain outside the tagging pass until the final classifier evaluation.

New records will be screened using the existing conservative grouping policy. Records that match, or may belong to, a protected test group will be deferred. This screening will not change the frozen benchmark or its split assignments.

Known unsupported record types and categories outside the classifier’s demonstrated scope will also be deferred or skipped with an explanation.

The development catalog export will contain only the defined development tagging population. A separate population report will account for processed, skipped, deferred, disabled, and failed records.

If tagging is disabled, the prediction export may contain no new suggestions. We will report that outcome directly rather than generate replacement predictions from another model.

### 22.8 Preserve Label Origins and Review Decisions

The derived catalog will retain existing source values and distinguish them from review decisions and predictions.

| Field | Meaning |
|---|---|
| `source_body_part` | The prepared source-provided label, linked to its original wording |
| `source_verification_status` | The recorded verification status of that source label |
| `reviewed_body_part` | A label adopted through an identifiable review decision |
| `predicted_body_part` | The classifier’s suggested label |
| `prediction_model_version` | The release that produced the suggestion |
| `prediction_input_hash` | The exact input version used for prediction |
| `prediction_status` | Whether prediction was produced, skipped, deferred, disabled, or failed |
| `tag_review_status` | Pending, accepted, rejected, or unresolved for a suggestion |
| `effective_body_part` | The label currently adopted for catalog use |
| `effective_label_origin` | Whether that label comes from the source or documented review |
| `review_decision_id` | The decision supporting an adopted or blocked label |

Our precedence rules will be:

1. An active, documented reviewed label takes precedence.
2. Otherwise, an acceptable source label may remain effective with its source origin and verification status.
3. A pending model suggestion does not automatically become effective.
4. An accepted suggestion becomes a reviewed catalog label only through a recorded decision.
5. An explicit decision rejecting a source label prevents automatic fallback to that rejected value.

A label can be valid within the catalog’s taxonomy while still awaiting independent source verification. We will preserve that distinction.

Model disagreement alone will not establish that a source label is wrong. Replacing an existing label will require an explicit review decision and supporting evidence.

Review decisions will record the reviewer, date, adopted value where applicable, reason, and evidence. Generating a pending worklist entry will not count as a completed human review.

Existing review history will be preserved. New input versions or model releases will receive traceable suggestion records rather than silently inherit earlier acceptance decisions.

The frozen Step 15 benchmark targets will remain separate from these catalog fields. Accepted suggestions will not become new training labels within the current experiment.

### 22.9 Respect the Classifier’s Scope

The classifier will have learned from general exercise names and descriptions. Its results will not establish reliable transfer to every basketball drill.

We will preserve basketball skill categories such as shooting, passing, and ball handling. These describe a different labeling task from the classifier’s body-part categories.

The model will predict only among its frozen supported classes. A returned label will not prove that an unfamiliar exercise belongs within that class set.

Coverage flags and review will remain important for unfamiliar inputs. A high model score will not remove the review requirement.

Accepting a body-part tag will not change equipment requirements, participant counts, source instructions, youth suitability, or planner eligibility.

Those decisions will remain with the catalog and planning-rule review processes established earlier.

### 22.10 Freeze the Catalog Handoff

We will preserve the classifier release, tagging decision, and derived catalog version supplied to later development steps.

Planner workflows using this component will receive the same identified model and catalog snapshot.

The Step 23 handoff will identify which records meet the separate requirements for retrieval and planner use. A predicted or accepted body-part tag alone will not satisfy those requirements.

If no records meet a required approval condition, we will report that limitation and preserve the open review work.

Before final evaluation, Step 31 will freeze the experiment’s model, catalog, and retrieval versions.

Any tagging of formerly held-out records after Step 32 will belong to a separate catalog revision. It will remain outside the frozen planner comparison rather than change the retrieval collection between classifier testing and planner scoring.

Later corrections will retain their history and identify the affected versions.

## 23. RAG Knowledge Base, Document Retrieval, and Retrieval Evaluation

### 23.1 Purpose

We will build a searchable collection of reviewed basketball, exercise, and youth participation references.

When a planner needs supporting information, retrieval will return relevant source passages with identifiers that we can trace back to the original material.

We will evaluate two things separately:

- Whether retrieval finds relevant evidence.
- Whether the context delivered to the language model includes the information needed to interpret that evidence correctly.

Finding a relevant passage does not guarantee that the generated plan will use it correctly. The later planner evaluation will assess that behavior.

### 23.2 Confirm Which Material Can Enter the Collection

We will begin with the source records, review decisions, and handoff documents from the earlier steps.

| Material | Intended role |
|---|---|
| Basketball drill records from Step 12 | Instructions, skill categories, resources, and progression information |
| Reviewed participation references from Step 13 | Published guidance with its applicability, definitions, and qualifications |
| Eligible general-exercise records from Step 22 | Exercise lookup with label origins and review status preserved |
| Adopted structured planning rules | Deterministic checks through the planning tools |
| Frozen USDA records from Step 14 | Food facts through the structured lookup tool |

The final two resources will remain structured tools. We will not depend on passage retrieval to perform nutrient calculations or find every mandatory planning rule.

A preparation bundle will not automatically establish that its contents are approved for retrieval or planning. Each included record must have the required source verification, permitted-use information, and reference-use review.

Drill and exercise records offered for training selection must also meet their separate planner-eligibility requirements.

A reviewed guideline passage may support reference lookup while an executable interpretation remains unadopted. Indexing the passage will not enable that planning rule.

We will build the corpus from an explicit list of eligible records. We will not recursively index every file in the project.

The index will exclude:

- Athlete profiles and private evaluation expectations.
- Retrieval questions, relevance judgments, and expected answers.
- Reviewer rationales and internal decision sheets.
- Generated plans and earlier model responses.
- Protected classifier test records and their groups.
- Material whose permission or required review remains unresolved.

If the available reviewed material cannot support the intended collection, we will report the shortfall and keep readiness open.

### 23.3 Define Evidence Units and Create Passages

We will distinguish an evidence unit from a passage.

An **evidence unit** will represent a reviewed drill record, exercise record, or guideline statement with its necessary supporting context.

A **passage** will be an indexable piece of that unit. One long evidence unit may produce several passages.

This distinction will prevent several fragments from one record from being counted as independent search results.

For passage construction, we will:

- Preserve source wording and meaningful table headings.
- Keep instructions connected to their equipment and participant requirements.
- Preserve age or grade applicability, units, definitions, and footnotes.
- Split long text at paragraph or sentence boundaries where possible.
- Link continuations and required qualifications to their parent evidence unit.
- Preserve confirmed duplicate-source relationships without merging distinct records merely because their wording is similar.

Our initial target will be approximately 150–200 embedding-model tokens for longer passages. Short complete records may remain shorter.

The initial maximum will be 200 encoded tokens, including the heading and model special tokens, provided this fits the loaded model’s actual limit.

For `all-MiniLM-L6-v2`, the model card reports default truncation beyond 256 word pieces. We will verify the loaded model’s `max_seq_length` and measure the actual encoded input rather than rely only on a tokenizer’s separate limit.

Overlong records will be split and linked before encoding. Important source text will not disappear through silent truncation.

A guideline fragment containing a numerical limit will retain a link to any definition or qualification needed to interpret it.

### 23.4 Preserve Metadata and Source Relationships

Each passage will retain:

- Stable source, record, evidence-unit, and passage identifiers.
- Publisher, title, URL, and page or section locator.
- Source version and verification date.
- Relevant applicability and resource metadata.
- Permission, reference-review, and planner-eligibility status.
- Links to required definitions, qualifications, or continuation passages.
- Content hash and corpus version.

We will preserve the distinction between source-supported information and project interpretation.

For example, a published basketball skill level will remain a skill level. It will not become an inferred age band.

Source and passage identifiers will support citation and auditing. Private review notes and expected evaluation answers will not become searchable passage text.

### 23.5 Prepare Two Retrieval Methods

Both methods will search the same eligible passages and use the same evidence-unit mapping.

| Method | Initial approach |
|---|---|
| Lexical retrieval | TF-IDF features and cosine similarity |
| Semantic retrieval | Frozen pretrained embeddings and cosine similarity |

**Lexical retrieval**

We will create a separate retrieval vectorizer using the shared deterministic normalization and tokenization functions.

Its initial settings will be:

- Word unigrams and bigrams.
- Minimum document frequency of one.
- Maximum document frequency of 1.0.
- No stop-word removal.
- No feature-count cap for the initial small corpus.
- Smoothed IDF and ordinary term frequencies.
- L2 normalization and sparse float64 features.

This vectorizer will be fitted on the eligible reference passages. It will remain separate from the exercise classifier’s fitted vectorizer.

Evaluation questions will be transformed using the fitted retrieval vectorizer. They will not contribute vocabulary or IDF values.

**Semantic retrieval**

Our initial candidate will be:

`sentence-transformers/all-MiniLM-L6-v2`

We will verify its license, exact model revision, tokenizer, local availability, and working device before adoption.

The pretrained weights will remain frozen. Queries and passages will use the same recorded encoding approach, without added task prompts for this initial model.

We will use float32, L2-normalized embeddings and cosine similarity. The actual embedding dimensions, model limits, package versions, and device will be recorded.

For this collection, we will score the saved passage embeddings directly. A dedicated vector database will be considered only if collection size or measured performance later justifies it.

### 23.6 Define Query Construction and Ranking

A retrieval request will contain the question or planning goal and any explicit constraints relevant to the search.

Both methods will receive the same query information. We will not add expected source identifiers, private answers, or scoring labels.

Query construction will preserve missing information. It will not infer an athlete’s grade, available equipment, or participant count.

We will check query length before semantic encoding. Oversized queries will return an explained status rather than silently lose constraints. Any later query-decomposition policy will be documented and applied consistently before final freezing.

Filters will use only supported metadata. Known incompatibilities may exclude a record, while unknown requirements will remain visible as uncertainty rather than become invented matches.

Our initial ranking procedure will:

1. Apply the recorded eligibility and applicability filters.
2. Calculate passage similarities.
3. Give each evidence unit its highest passage score.
4. Sort units by descending score.
5. Resolve exact ties using stable evidence-unit identifiers.
6. Return up to five distinct evidence units.

The highest-scoring passage will remain attached to each result. Passage-ID ordering will resolve ties within a unit.

A lexical query with no usable features or no positive similarities will return no lexical matches. We will not fill its results with arbitrary zero-score records.

Semantic results will remain candidate evidence. A similarity score will not establish that the question is answerable.

The initial comparison will not use a tuned similarity cutoff for automatic abstention.

### 23.7 Assemble Complete Reference Context

After ranking, we will recover the source context required to interpret the selected evidence.

For a drill, this may include its instructions and resource requirements. For a guideline, it may include a definition, applicability statement, or footnote.

The context assembler will receive an explicit reference-token budget and the generation tokenizer used by the language model.

Embedding-model token counts and language-model context counts are different measurements. Passage lengths alone will not establish that the full planning request fits.

Within the available budget, we will:

- Process selected evidence units in rank order.
- Add required supporting material.
- Remove repeated shared context while preserving its source links.
- Keep necessary qualifications together with the statement they qualify.
- Record omitted evidence and the reason for omission.

If a required bundle cannot fit, we will omit or flag that bundle rather than remove a qualification that changes its meaning.

Linked supporting material will be recorded as context expansion. It will not silently change the ranked top-five results used for ranking metrics.

We will save both the ranked results and the exact context delivered to the language model.

Step 24 will supply and test the working generation budget. Until that integration is available, ranking development can proceed, but delivered-context checks will remain pending.

### 23.8 Prepare a Separate Retrieval Benchmark

Our target will be 60 reviewed retrieval questions: 30 for development and 30 held out for final assessment.

| Question category | Development | Held out |
|---|---:|---:|
| Answerable drill and resource questions | 10 | 10 |
| Answerable guideline and qualification questions | 10 | 10 |
| Questions unsupported by the identified corpus | 10 | 10 |
| **Total** | **30** | **30** |

These are planned counts. We will confirm them only after the corpus and question reviews support them.

Paraphrases and questions testing the same underlying information need will share a question-family identifier and remain in one partition.

The held-out items will be questions and their judgments. The reviewed reference corpus will remain searchable for both partitions.

One teammate will prepare each question and its evidence mapping, and another will verify it against the sources. Unresolved disagreements will remain open.

Each question record will identify its family, category, partition, review status, and corpus version.

We will store two kinds of judgment:

- Relevant evidence-unit IDs for ranking metrics.
- Accepted complete-evidence bundles identifying the actual statements, passages, definitions, and qualifications needed to answer.

A complete bundle may have acceptable alternatives. Retrieving one fragment with the correct parent ID will not automatically satisfy it.

Development and held-out questions and judgments will be saved separately. Held-out material will remain outside development selection.

“Unsupported” will mean unsupported by the specified corpus version. Corpus changes will require corresponding judgment checks before final freezing.

### 23.9 Calculate Ranking and Context Metrics

For an answerable question, let:

- `R` be its reviewed relevant evidence units.
- `Tₖ` be the first `k` distinct returned units.

We will calculate:

| Measure | Definition |
|---|---|
| Hit@k | 1 when at least one unit in `Tₖ` belongs to `R`; otherwise 0 |
| Recall@k | Number of relevant units in `Tₖ` divided by the number in `R` |
| MRR@5 | Reciprocal rank of the first relevant unit within five results; otherwise 0 |
| Complete-evidence coverage | Whether the delivered context contains an accepted complete evidence bundle |
| Retrieval time | Recorded query-processing and context-assembly time |

Hit and recall will be reported at 1, 3, and 5. Aggregate results will average the per-question values over the reviewed answerable questions.

We will record the actual denominators. Failed answerable requests will remain in them, with zero ranking scores and failed complete-evidence coverage when context assembly was executed.

An assessment that has not yet been run will remain pending rather than receive a fabricated score.

Unsupported questions have no relevant-unit denominator, so ordinary recall will remain not applicable. We will report their candidate-return behavior, no-match behavior, execution failures, and reviewed examples of potentially misleading results separately.

Returning a candidate passage will not itself count as a claim that the corpus supports an answer. An execution failure will not count as successful abstention.

Recall will be measured against our reviewed judgments. Unjudged development results will be flagged for consistent source review across both methods before finalizing the comparison.

Any judgment revision will be versioned and applied to both methods. Corrections discovered after final freezing will be reported separately while preserving the original primary results.

Index-building time will be recorded separately from per-query time.

### 23.10 Select and Freeze the Retriever

The initial selection rule will be:

1. Higher development mean Recall@5.
2. Higher development MRR@5 on an exact Recall@5 tie.
3. Lexical retrieval if both measures tie exactly.

Selection will use full numerical precision.

Complete-evidence coverage, unsupported-question behavior, latency, and context size will remain important diagnostics. They will help us explain the selected method’s limitations rather than become replacement selection measures chosen after seeing the results.

A selected method must also pass the implementation and artifact checks. A software defect will require correction and consistent reruns.

Before final assessment, we will freeze:

- Corpus, passage, and evidence-unit versions.
- Query construction and filtering.
- Lexical settings or semantic model revision.
- Encoding, normalization, and similarity settings.
- Deduplication and tie rules.
- Retrieval depth.
- Context expansion and generation-token budget.
- Any explicitly adopted rejection policy.
- Benchmark questions, judgments, and scoring definitions.

Step 31 will link these components in the final protocol manifest.

The held-out retrieval assessment will run only after that freeze. Its findings will not change the corpus, retriever, prompts, or context rules before the final planner comparison.

Planner Versions 2–4 will use the same adopted retrieval configuration and reference collection.

### 23.11 Verify and Save the Development Package

Our checks will confirm that:

- Every indexed passage has an eligible source and valid locator.
- Protected records and evaluation content are excluded.
- Passage lengths fit the loaded embedding model.
- Evidence-unit and required-context links resolve correctly.
- Matrix or embedding rows align with passage IDs.
- Similarity outputs are finite and ranking is reproducible.
- Duplicate units do not occupy several ranked positions.
- Saved retrieval artifacts reload correctly.
- Benchmark partitions contain no overlapping question families.
- Metrics retain the declared denominators.
- Context omissions and incomplete bundles remain visible.

After development, we expect to save:

| Output | Contents |
|---|---|
| `data/processed/rag_passages.jsonl` | Eligible passage text and source metadata |
| `data/processed/rag_evidence_units.jsonl` | Parent records, passage relationships, and required source context |
| `results/retrieval/corpus_review.csv` | Inclusion decisions, exclusions, and unresolved source issues |

## 24. Language Model Integration and Reference Lookup Tools

### 24.1 Purpose

We will connect our local planning model to the reviewed reference collection and structured lookup tools.

This step will provide the shared foundation for all four planner versions. It will handle model requests, reference preparation, output parsing, context limits, and execution records.

The specialist coordination and revision workflows will be implemented in Steps 25–28. Here, we will establish the common interfaces they need.

We will keep technical integration, source approval, and planning-rule adoption as separate readiness conditions. A working model connection will not automatically make every prepared resource available for planning.

### 24.2 Continue with the Recorded Local Model

Our initial planning model will remain **`llama3.1:8b` through Ollama**, following Step 7.

The Hugging Face model card for Llama 3.1 8B Instruct will provide model-family documentation. The installed Ollama artifact and its digest will identify the model actually used in our experiment.

Before integration, we will verify and record:

- Ollama server version.
- Installed model name and digest.
- Quantization and model details.
- Effective chat template, system defaults, and stop sequences.
- Execution device and any CPU/GPU offloading.
- Context allocation and generation limits.
- Effective decoding settings.

A model tag can change, so the tag alone will not identify the frozen artifact.

We will use Ollama’s `/api/chat` endpoint with explicit message roles. The recorded Ollama template will format those messages once.

Step 7 tested the generation interface. We will therefore run a compatibility check for chat messages and structured output before adopting this adapter.

All four planner versions will use the same selected model artifact, runtime, quantization, and common final-response settings.

### 24.3 Establish the Development Request Settings

Our initial integration settings will be:

| Setting | Development starting point |
|---|---|
| Streaming | Disabled |
| Temperature | 0 |
| Seed | 42 |
| Responses requested per logical generation | 1 |
| Context allocation, `num_ctx` | 8,192 tokens |
| Final-response output limit, `num_predict` | 2,048 tokens |
| Output mode | JSON |
| Concurrent model requests | 1 |

The context and output limits are starting values for integration testing. They are not measurements of verified capacity.

We will test them with representative weekly plans, reference packets, specialist notes, and revision prompts. Any change will receive a configuration version and consistent checks across affected workflows.

Specialist and evaluator calls may use smaller explicit output limits defined in Steps 27–28. Each request will record its actual settings.

The request timeout will be finite and specified before execution. Step 31 will confirm it, along with the final operating limits, using measured development runs.

Temperature zero and a fixed seed support repeatability. We will still record runtime and hardware conditions rather than promise identical outputs across environments.

### 24.4 Preserve the Shared Input Contract

Every planner will receive the same model-visible athlete information for a matched case.

The input adapter will standardize field names and documented units while preserving:

- Missing values.
- Conflicting information.
- Existing practices, games, and other commitments.
- Available days and time.
- Equipment, space, and participant constraints.
- The athlete’s stated goals and experience.
- Any explicitly requested food facts.

It will not invent a school grade, remove a commitment, or replace an unknown requirement with a convenient assumption.

Scenario categories, expected response types, feasibility witnesses, reviewer notes, and scoring expectations will remain private. None will enter a planner prompt, including Version 4’s evaluator prompt.

The shared task instructions will describe the project scope and response requirements. Curated guideline thresholds, catalog records, and lookup results will enter only the versions allowed to receive references.

Each profile–version job will start with fresh messages and workflow state. Keeping model weights resident will be allowed, but earlier cases’ conversations and generated content will not carry into later cases.

### 24.5 Implement the Read-Only Reference Tools

Our initial tools will read frozen local resources.

| Tool | Main arguments | Required result |
|---|---|---|
| `search_references` | Query and supported filters | Ranked evidence, delivered context, source IDs, versions, and context limitations |
| `get_drill` | Canonical drill ID | An eligible drill’s instructions, resources, qualifications, and source |
| `get_exercise` | Catalog ID | An eligible exercise record with label origins and review status |
| `get_planning_rules` | Supplied age and school grade, either nullable | Adopted rules, applicability, required inputs, and qualifications |
| `lookup_food` | Food ID or search text, plus quantity arguments | Food identity, nutrient availability, supported calculations, and source information |

Each tool will have an argument schema and a common response envelope containing:

- Request ID and status.
- Returned data.
- Issues or limitations.
- Source references.
- Data version.

Statuses will distinguish found, ambiguous, not found, not eligible, unavailable, invalid arguments, incomplete context, and execution error where applicable.

A known record awaiting review will remain distinguishable from an unknown identifier.

The tools will not refresh publisher pages, download replacement data, or change source records during generation.

### 24.6 Preserve Resource Eligibility and Rule Applicability

`get_drill` will expose the reviewed planner-use collection. Prepared candidate records will not become available simply because their files exist.

`get_exercise` will preserve source labels, reviewed labels, and classifier suggestions separately. Lookup will not generate a new tag or accept an existing suggestion.

For `get_planning_rules`, operational results will come from adopted and enabled rules. Draft interpretations and pending decisions will not become active limits.

An unavailable or unadopted register will return an explained unavailable status. It will not return an apparently successful empty list suggesting that no restrictions apply.

Missing or conflicting age and grade information may leave applicability unknown. Where categories overlap, the tool will preserve the potentially relevant rules and their qualifications rather than choose the more permissive category.

Reference-approved guideline passages may still be available through retrieval. That will remain separate from adopting executable rules.

Retrieving rules will also remain separate from checking a generated plan. Generation-time coded plan checking will belong to Version 4.

### 24.7 Integrate the Frozen USDA Food Reference

Food lookup will use the producer-identified SR Legacy snapshot prepared in Step 14.

We will verify its manifest, files, and hashes before exposing it through the tool. We will not substitute the earlier Foundation Foods table.

Food identity, nutrient-field availability, and quantity support will remain separate.

The lookup will support these quantity modes:

| Mode | Behavior |
|---|---|
| Per 100 grams | Default reference display |
| Explicit grams | Scale values using a finite, positive gram amount |
| Recorded household portion | Use an eligible food-specific portion and an explicit multiplier |

For supported quantities, arithmetic will use:

**Nutrient amount = recorded amount per 100 grams × selected grams ÷ 100**

A household multiplier will refer to the entire recorded portion. Its gram weight already accounts for the source’s listed amount, so that amount will not be multiplied twice.

The result will preserve the portion description, listed amount, gram weight, selected reference weight, nutrient IDs, units, and availability flags.

We will retain valid recorded zeros and keep missing nutrient values unavailable. Energy will follow Step 14’s selected SR definition, nutrient 1008 with its recorded KCAL unit.

An ambiguous food search will return candidates. Raw and cooked foods will remain distinct.

Calculations will occur in code, with calculation values kept separate from display rounding. Generic cup-to-gram conversions will not be introduced.

Application food-fact displays will use the structured tool values. Model explanations and any repeated numerical claims will remain traceable for assessment.

The tool will provide reference facts without calculating an athlete’s calorie requirements or prescribing a diet.

### 24.8 Prepare References Under Application Control

Our initial implementation will use a deterministic Python function to prepare reference packets before generation.

For matched profiles, Versions 2–4 will use the same public input fields, source versions, retrieval method, and packet-building policy.

The packet will have an identifier and content hash. Each role will receive the same canonical packet; its role-specific instructions will be recorded separately.

Reference preparation may retrieve passages, resolve eligible record IDs, obtain potentially applicable rules, and perform an explicitly requested food lookup.

It will not use private expectations or an additional language-model call to rewrite the query.

Native model-requested tool calling will remain outside the initial design. Model-written requests will not become arbitrary Python, file operations, or web searches.

If native calling is introduced later, we will verify local support and revise the workflow contracts and call budgets before final freezing.

Reference passages and tool results will enter messages as identified data. They will not replace the application’s task instructions.

### 24.9 Enforce the Four Planner Configurations

Access will be controlled in code.

| Version | Information and tools during generation | Workflow |
|---|---|---|
| 1 — LLM alone | Shared instructions and public profile | One planner generation |
| 2 — LLM with references | Shared reference packet and lookup results | One reference-supported planner generation |
| 3 — Specialist workflow | Same reference collection and tools | Specialists followed by a coordinator |
| 4 — Evaluated specialist workflow | Same references, internal evaluation, and coded checks | Specialist draft with bounded revisions |

Version 1 will receive no retrieved passages, lookup results, classifier suggestions, or curated numerical guidance hidden inside the common prompt.

Versions 1 and 2 will each retain one logical generation request per profile. Reference preparation for Version 2 will be recorded separately as application work.

Steps 27–28 will define the specialist, evaluator, and revision calls. We will not introduce extra generations through an undocumented tool exchange or formatting-repair step.

Only Version 4 will receive generation-time candidate checks and internal evaluator feedback.

Independent research evaluation will run on the saved returned responses for every version. Its feedback will not be used to revise those responses.

### 24.10 Define the Conditional Output Structure

We will save a common final-response schema.

| Component | Expected content |
|---|---|
| Response type | Plan, clarification, constraint conflict, or outside scope |
| Planning window | The relevant seven-day period or supplied day labels |
| Weekly schedule | Activities and rest arrangements when a plan is appropriate |
| Activity details | Day, duration, relevant timing, and resource requirements |
| Existing commitments | References to supplied practices, games, and other commitments |
| Activity relationships | Whether an item is a separate session or part of another activity |
| Questions and unresolved issues | Essential gaps or conflicts |
| Explanation | A brief explanation of the response |
| Evidence references | Source or catalog IDs when available |
| Optional food facts | Source-linked factual content where requested and available |

Required fields will depend on the response type. Clarification and outside-scope responses will not require a complete weekly schedule.

Unavailable source IDs may remain absent or null under the schema. Version 1 will not be required to invent identifiers from a collection it cannot access.

We will begin with JSON mode and local schema validation. If schema-constrained decoding is later adopted, we will verify installed-runtime support and apply the same final-response policy across all four versions.

The raw response will be preserved before parsing.

Parsing will check structure, types, and conditional requirements. It will not add durations, alter activities, resolve conflicts, invent citations, or rewrite factual claims.

A malformed completed response will remain a recorded outcome. Version 4 may use its explicitly budgeted revision procedure; Versions 1–3 will not receive hidden repair calls.

### 24.11 Manage the Complete Context Budget

For every model call, we will account for:

- Chat-template and role overhead.
- Task instructions and output requirements.
- Public profile information.
- Reference passages and structured tool results.
- Specialist notes where used.
- The current candidate and evaluation feedback where used.
- Reserved output capacity.

The input allowance will be calculated from the configured context allocation after reserving the output limit and a documented margin.

We will use token accounting verified against the local model and template. A word count or MiniLM token count will not be described as an exact Ollama prompt count.

Development requests will compare local counts with backend usage measurements where available. If exact accounting cannot yet be verified, that limitation will remain visible in context-readiness reporting.

Versions 2–4 will use a shared reference-packet budget. We will size it against their required stages and check each fully assembled request again.

If required content does not fit, the adapter will report the limitation. It will not silently remove original profile facts or source qualifications.

Step 31 will confirm the final context, role-output, and reference budgets after representative workflow tests.

### 24.12 Record Calls, Retries, and Failures

Each logical generation will have a stable job and call identifier.

We will distinguish:

- Logical model calls.
- Backend transport attempts.
- Reference-tool calls.
- Workflow revisions.

The initial transport policy will permit at most one retry for a confirmed connection failure before the request was accepted.

An uncertain timeout or lost response will require resolution of the earlier request’s state before another attempt. We will not leave an unnoticed generation running while starting a replacement.

A completed response will not be regenerated because it is inaccurate, malformed, truncated, or disappointing.

Run records will preserve:

- Public-input and reference-packet hashes.
- Planner version, role, and prompt version.
- Model digest and runtime settings.
- Submitted messages.
- Raw backend response and returned text.
- Parsing and completion status.
- Tool arguments, results, and source versions.
- Stop reason and available token counts.
- Timing, model-loading time, cache information, and errors.
- Retry and revision relationships.

Unavailable measurements will remain null. Estimates will be labeled.

Read-only reference caches may be reused under a recorded policy. Generated answers and private evaluation records will not become reference-cache content.

### 24.13 Verify Integration and Save the Handoff

Development checks will cover:

- Supported gram and household-portion calculations.

- Invalid tool arguments.
- Malformed or generation-limited output.
- Context overflow and backend failure.
- Resetting state between jobs.
- Version 1’s lack of reference access.
- Matching reference packets for Versions 2–4.

These checks will use development profiles, development retrieval questions, and documented lookup fixtures.

We will link the Step 14 calculation evidence and verify that the new lookup wrapper preserves its behavior. Reused evidence will remain distinguishable from newly executed integration tests.

Expected outputs will include:

| Continuation or repair generations | None |

The nominal maximum will be 15 application-tool calls before generation: two searches, one rule lookup, ten catalog lookups, and two food lookups.

Actual calls may be fewer. Context expansion and cache activity will be recorded as preparation work.

Reference preparation will use Python functions. It will not include an additional model call for query rewriting or evidence selection.

The shared transport policy will apply to the planner request. A failed preflight may produce no generation, and a malformed or disappointing completed response will not trigger another one.

### 26.4 Construct the Queries and Merge the Results

We will use two saved query templates.

| Query | Information need |
|---|---|
| Goal and resource query | Basketball goals, experience, equipment, space, and participants |
| Participation and recovery query | Relevant guidance using supplied age, grade, and activity context |

Queries will use public input fields only. Missing information will remain missing rather than become an inferred value.

The templates may retain a general basketball topic when a detail is unavailable. They will not invent a specific goal, school grade, or resource.

We will preserve each query’s separate ranked list.

To combine them, we will use a fixed round-robin rule:

1. Take the first result from the goal/resource search.
2. Take the first result from the participation/recovery search.
3. Continue through their next ranks in the same order.
4. Skip evidence-unit IDs already retained.
5. Stop after both lists are exhausted.

This will produce at most ten distinct evidence units. If one search returns fewer results, the remaining list can continue.

We will retain which queries retrieved each unit, along with their passage IDs, ranks, and scores. Raw scores from different queries will not be treated as calibrated values for cross-query ordering.

Only the supported metadata filters from Step 23 will be applied.

### 26.5 Retrieve Rules, Catalog Records, and Requested Food Facts

**Planning rules**

We will request potentially applicable adopted and enabled rules using the supplied age and grade.

Definitions, units, qualifications, and unresolved applicability will remain attached to the results.

An unadopted register will return an explained unavailable status. It will not appear as a successful empty list implying that no restrictions apply.

This lookup happens before a plan exists. It will not calculate a plan verdict, remove commitments, or choose the expected response type.

**Catalog records**

We will expand unique drill or exercise IDs linked to the retained search results, within the combined ten-record limit.

Only records meeting the lookup’s eligibility conditions will be returned for planning use.

Source instructions, resource requirements, skill categories, and label origins will remain intact. A classifier suggestion will not become an approved label or youth-suitability decision.

We will distinguish source prescriptions from model-selected scheduling choices. Allocating time to an activity will not imply that its source endorsed that exact duration.

**Food facts**

Food lookup will activate only for an explicit food-facts request in the public input.

We will process at most two requested foods in their supplied order. Additional items will be reported as unprocessed because of the limit.

Ambiguous matches will remain ambiguous. The workflow will not choose a candidate or make extra disambiguation calls without further user information.

The tool will use Step 14’s frozen SR Legacy snapshot and supported quantity calculations. Missing nutrients will remain unavailable, recorded zeros will remain zero, and household quantities will require an eligible food-specific portion.

The model will receive the food identity, reference amount, nutrient units, source version, and any limitations.

### 26.6 Assemble One Traceable Reference Packet

Both searches and all lookups will feed one packet with one total reference budget.

The packet will distinguish:

- Retrieved evidence and its source identifiers.
- Eligible drill and exercise details.
- Adopted rules and applicability information.
- Required definitions and qualifications.
- Requested food facts.
- Missing, ambiguous, unavailable, or omitted information.

We will remove repeated content while preserving its source relationships.

Our initial packing order will preserve availability notices and applicable rule bundles first, followed by explicitly requested food facts and the merged activity evidence.

Within those sections, ordering will be deterministic. The exact packing policy will be saved and tested before final freezing.

Necessary qualifications will stay with their statements. If a complete required bundle cannot fit, the packet will record its omission or incomplete status rather than retain a misleading fragment.

We will save the full preparation record separately from the exact text delivered to the model. A retrieved result that was omitted from the prompt will not be described as evidence the model received.

Versions 2–4 will use the same packet builder and frozen resource versions for a matched profile.

The packet’s content hash will cover its canonical model-visible content and configuration. Operational timestamps, attempt IDs, timings, and cache events will remain separate so they do not create artificial content differences.

### 26.7 Generate One Response from the Profile and References

The model request will contain:

1. The shared task instructions.
2. The unchanged public profile.
3. The assembled reference packet.
4. The shared conditional output requirements.

The additional reference-use instructions will ask the planner to:

- Use supplied evidence accurately.
- Preserve conditions attached to published guidance.
- Distinguish source-supported facts from its scheduling choices.
- Use supplied source identifiers when citing them.
- Identify essential gaps and unresolved applicability.
- Avoid presenting unavailable information as established fact.

The packet builder will supply reference information. The planner will decide whether to produce a schedule, ask for clarification, explain a conflict, or explain that the request is outside scope.

We will then make the single logical generation request and preserve its result.

Version 2 will receive no specialist contributions, generated-plan checker findings, or internal evaluator feedback.

Fresh messages and workflow state will be used for every profile. Earlier plans will not become reference material.

### 26.8 Handle Missing Evidence and Technical Failures Separately

A successful lookup may return no match, an ambiguous food, a pending record, or unknown applicability.

These are meaningful data statuses. They can enter the packet so the planner can explain a limitation or ask an appropriate question.

A broken required dependency is different. Missing required artifacts, hash mismatches, incompatible versions, or a failed required retrieval operation will produce a blocked or failed preparation record.

We will not silently discard that failure and run an unreferenced baseline under the Version 2 label.

The configuration will identify required dependencies and the reason codes that distinguish expected availability limitations from technical failure.

A high similarity score will remain a ranking signal. It will not establish that the source answers the request.

After generation, we will preserve raw output before parsing. Malformed JSON, missing fields, truncation, unexpected tool requests, and unsupported citations will remain visible.

The parser will not repair the schedule or replace factual claims. Independent findings will not trigger post-generation lookups or revisions.

### 26.9 Verify the Workflow with Development Cases

We will record the selected cases from the development-profile partition.

Checks will cover:

- Query construction from the correct public fields.
- Exclusion of private expectations and scenario labels.
- Separate search rankings and deterministic merging.
- Deduplication of evidence units and canonical catalog IDs.
- Lookup limits and food-request routing.
- Preservation of rule qualifications and unknown applicability.
- Eligible and pending catalog records.
- Ambiguous foods, missing nutrients, and supported quantities.
- One total context budget.
- Traceability from delivered claims to source material.
- Expected no-match results and technical preparation failures.
- The one-generation budget and absence of hidden revisions.
- Consistent packet content for the later reference-enabled versions.

We will inspect cases where relevant information reaches the model but is ignored or misrepresented.

These checks will help distinguish retrieval errors, context omissions, source limitations, and generation errors.

The 60-question retrieval benchmark from Step 23 will remain separate from the athlete-profile benchmark. A two-query planning packet with up to ten units will not be reported as the single-query Recall@5 assessment.

Held-out retrieval questions and final athlete profiles will remain reserved for their planned evaluations.

### 26.10 Evaluate and Measure the Complete Workflow

We will apply the same independent planner evaluation used for Version 1 after saving the response.

The assessment will cover completion, feasible-plan success, required constraints, handling of challenging requests, and evidence quality.

A valid citation identifier will not automatically establish that its associated claim is supported. We will preserve whether the cited evidence was actually delivered and whether it supports the claim.

Development scores will use the actual evaluated cases and report their denominators.

For the final comparison, all 150 feasible profiles will remain in Version 2’s success denominator, including preparation failures, timeouts, missing responses, malformed plans, and inappropriate refusals.

Required unknown conditions will follow Step 17’s failure policy. Missing-information, conflict, and outside-scope cases will be reported separately.

If rule adoption or scorer verification remains incomplete, affected adherence scores will remain unavailable rather than become automatic passes.

Execution measurements will include:

- Reference preparation.
- Searches and exact lookups.
- Context assembly.
- Cache activity.
- Model generation.
- Transport attempts and failures.
- Total elapsed time and available token usage.

The cache policy will be shared across reference-enabled versions and frozen in the final protocol. Shared preparation will remain visible in the resource accounting.

We will interpret any improvement as evidence about the complete Version 2 configuration, including its additional references and reference-use instructions.

### 26.11 Freeze the Workflow and Save Its Outputs

Before final execution, we will freeze:

- Model digest and shared runtime configuration.
- Query templates and metadata filters.
- Retrieval method, corpus, and evidence-unit mapping.
- Search-merge and deduplication rules.
- Rule, catalog, and food snapshots.
- Lookup limits and food routing.
- Packet ordering, context limits, and cache policy.
- Prompt, output schema, parsing, and failure handling.

The same reference preparation will support Versions 3 and 4. Their additional stages will not silently change the underlying reference collection or obtain extra evidence.

Final execution on the 200 frozen profiles will take place in Step 33.

Expected outputs will include:

| Output | Contents |
|---|---|
| `src/champion_athletic_map/planners/rag.py` | Version 2 workflow using the shared reference builder |
| `docs/prompts/planner_v2.md` | Versioned planner prompt and reference-use instructions |
| `artifacts/planners/version2_config.json` | Shared model references, query settings, lookup limits, and access policy |
| `results/planner_v2/development_manifest.csv` | Selected development cases and coverage |
| `results/planner_v2/development_contexts.jsonl` | Queries, rankings, merged units, lookup results, packets, hashes, and omissions |
| `results/planner_v2/development_runs.jsonl` | Raw responses, parsed outputs, execution status, and measurements |
| `results/planner_v2/development_checks.csv` | Access, packet, context, call-budget, and failure-handling checks |
| `results/planner_v2/development_scores.csv` | Independent development scores where prerequisites are available |
| `docs/planner_v2.md` | Design decisions, findings, and limitations |
| `results/step26_planner_v2_summary.json` | Producer record, hashes, counts, and readiness states |

The summary will distinguish source readiness, packet preparation, generation, parsing, scoring availability, and final-experiment readiness.

We will add measured results after implementation. Step 27 will introduce specialist coordination using the same model and reference foundation.

## 27. Planner Version 3 — RAG with Specialist Agents

### 27.1 Purpose

We will extend the reference-supported planner by dividing the preparation work among specialist roles.

Each specialist will contribute focused notes. A coordinator will then use those notes, the original athlete profile, and the shared references to produce one final response.

Every role will use the same frozen language model with different instructions. We will evaluate whether this division of work improves the plans enough to justify the additional computation.

Version 3 will follow a fixed sequence. Checking and revising a completed coordinator response will be introduced in Step 28.

### 27.2 Verify the Shared Foundation

We will reuse the model adapter and configuration from Step 24 and the reference-packet builder from Step 26.

Before execution, we will verify:

- The selected Ollama model digest and runtime settings.
- Public-profile and final-response schema versions.
- Corpus, retriever, catalog, rule, and food-reference versions.
- Reference-packet configuration and content hash.
- Specialist prompts and note schemas.
- Per-role output limits and context requirements.
- The development-profile manifest.

For a matched profile, Versions 2–4 will use the same canonical reference packet.

The initial Version 3 workflow will add no searches, lookups, or classifier calls between specialist stages.

Prepared sources, accepted tags, and planner eligibility will retain their separate review conditions.

### 27.3 Define the Roles

The initial workflow will contain two core specialists, an optional food-facts specialist, and a coordinator.

| Role | Responsibility | Expected contribution |
|---|---|---|
| Training specialist | Consider basketball goals, experience, resources, and eligible supporting exercises | Candidate activities, resource needs, progression considerations, and source references |
| Participation and recovery specialist | Consider applicable guidance and existing commitments | Relevant conditions, rest considerations, qualifications, conflicts, and missing information |
| Food-facts specialist | Address an explicit food question using supplied lookup results | A source-linked explanation of the available facts and limitations |
| Coordinator | Combine the original information and required specialist notes | One response using the shared conditional output structure |

The participation and recovery specialist will work before the coordinator drafts the plan. It will provide prospective guidance rather than inspect a completed schedule.

The training specialist will distinguish recorded instructions from proposed scheduling or adaptation choices. A model suggestion will not become a source-approved modification.

The food-facts role will explain the structured lookup results. Nutrient and portion arithmetic will remain in the lookup code.

Our initial implementation will use explicit Python functions and the shared adapter. A separate agent framework will not be required.

### 27.4 Keep Specialist Inputs Independent

Each specialist will receive a fresh message list containing:

1. The shared task instructions.
2. Its role instructions.
3. The original public profile.
4. The same canonical reference packet.
5. Its note-format requirements.

Specialists will not receive one another’s generated notes.

The coordinator will receive the original profile, the shared references, and all required valid specialist contributions.

This will keep the primary source information available during coordination instead of making the coordinator depend entirely on model-written summaries.

We will record the packet hash and each role’s actual message hash. Operational timestamps and call IDs will remain separate from the packet content.

Private scenario labels, expected behaviors, feasibility witnesses, reviewer notes, and scoring answers will remain outside every role’s context.

### 27.5 Use a Fixed Execution Sequence

For each profile, the workflow will:

1. Prepare or verify the shared reference packet.
2. Generate the training specialist’s notes.
3. Validate their structure and preserve the raw response.
4. Generate and structurally validate the participation and recovery notes.
5. Run the food-facts role when the public-request routing rule activates it.
6. Check that all required contributions are available.
7. Assemble the coordinator’s complete request.
8. Generate the coordinator response once.
9. Preserve its raw output, parsing status, and execution record.

Calls will initially run sequentially to keep resource use predictable.

A required stage failure will stop the intended sequence. Later stages will be recorded as not started.

Messages and workflow state will reset between profiles. Keeping the model resident will not permit earlier cases’ generated content to enter another job.

We will implement the drafting procedure as a reusable function for Step 28. Each planner job will still have its own recorded execution and costs.

### 27.6 Route the Food-Facts Role from the Public Request

The food role will use the same activation rule as Step 26: an explicit food-facts request in the public input.

A general basketball or fitness goal will not automatically activate it.

The shared two-food lookup limit and request order will remain unchanged. Any additional requested foods will retain the recorded limit notice.

When activated, the role will run even if the lookup returns a valid ambiguous or unavailable result. Its contribution should explain that limitation rather than invent values or choose a food.

Once activated, the food role will be required for that job’s intended workflow.

It will make no additional searches or calculations. It will preserve food identity, units, reference amounts, missing nutrient fields, and supported quantity information.

The corresponding Version 4 workflow will use the same routing rule.

### 27.7 Request Compact, Structured Contributions

Specialists will return notes rather than separate complete weekly plans.

The common note structure will include:

| Field | Expected content |
|---|---|
| Contribution status | Provided, limited information, or outside scope |
| Recommendations | Role-relevant points or candidate activities |
| Evidence references | Supplied source and record IDs supporting factual statements |
| Qualifications | Conditions needed to interpret the recommendations |
| Unresolved issues | Missing information, conflicts, or evidence limitations |
| Clarification questions | Focused questions where needed |
| Brief rationale | A short explanation connecting the contribution to the request |

Application metadata will record the role, input hash, packet hash, and schema version. The model will not need to invent these identifiers.

Lists may be empty when the available information does not support a recommendation.

Local suggestion identifiers will remain distinguishable from canonical source or catalog IDs.

Specialist notes will remain model-generated contributions. We will preserve their raw wording alongside parsed fields.

Structural validation will check whether a contribution can be handed to the coordinator. It will not silently correct a questionable recommendation or treat a self-reported status as verified truth.

A structurally valid contribution reporting insufficient evidence will remain usable.

### 27.8 Bound Role Outputs and Coordinator Context

Our development starting limits will be:

| Role | Initial generated-token limit |
|---|---:|
| Training specialist | 512 |
| Participation and recovery specialist | 512 |
| Food-facts specialist | 512 |
| Coordinator | Shared final-response limit from Step 24 |

These are starting values for testing. We will confirm or revise them through documented development runs and the Step 31 pilot.

Before each call, the adapter will check the complete rendered request against the configured context allowance.

For the coordinator, that includes:

- Shared and coordinator instructions.
- The original public profile.
- The complete shared reference packet.
- All required specialist notes.
- Output-schema instructions.
- Reserved output capacity and the recorded margin.

We will not silently delete substantive notes, remove source qualifications, or add a summarization generation to make the request fit.

If the required context exceeds the limit, we will record the failure and the stage at which it occurred.

### 27.9 Define the Coordinator’s Responsibilities

The coordinator will produce one response using the shared final-response schema.

Its instructions will ask it to:

- Preserve supplied availability, commitments, and resources.
- Select activities relevant to the athlete’s stated goals.
- Combine compatible contributions into a coherent response.
- Resolve disagreements using the original facts and references.
- Keep source qualifications attached to their claims.
- Distinguish evidence from proposed scheduling choices.
- Identify essential uncertainty or unresolved conflicts.
- Use the appropriate plan, clarification, conflict, or outside-scope response.

Agreement between specialists will not establish correctness. Their notes cannot override the public profile or structured source values.

The coordinator may identify mistakes or disagreements while preparing its single response. It will receive no post-generation checker verdict, evaluator feedback, or revision request in Version 3.

Raw coordinator output will be saved before parsing. A malformed final response will remain visible with its parsing status.

### 27.10 Set Call Limits and Failure Handling

The intended successful sequence will use:

| Stage | Logical model calls |
|---|---:|
| Training specialist | 1 |
| Participation and recovery specialist | 1 |
| Food-facts specialist | 0 or 1 |
| Coordinator | 1 |
| **Total** | **3 or 4** |

Early failures may produce fewer calls. Reference preparation and transport attempts will be counted separately.

A valid contribution describing ambiguity or missing evidence will not be treated as a technical failure.

A failed required call, malformed mandatory contribution, or contribution failing the declared completion policy will stop the workflow. Partial outputs will remain available for diagnosis.

We will not replace the failed workflow with a Version 2 answer or issue an unrecorded repair request.

Each role will follow Step 24’s narrow transport policy: at most one retry for confirmed failure before acceptance. Uncertain completion states must be resolved before another request.

Poor recommendations are model-quality findings. They will not, by themselves, authorize another specialist call.

Execution, parsing, and final-response status will remain separate so a completed generation with invalid structure is not confused with a backend failure.

### 27.11 Evaluate the Response and Measure the Workflow

Independent evaluation will run after the coordinator response has been saved.

Its findings will not trigger another specialist call, coordinator pass, or revision.

The coordinator response will be the profile’s research output. Intermediate notes will support workflow analysis rather than count as additional independent planning responses.

Development scores will use the actual evaluated cases. In the final experiment, all 150 feasible profiles will remain in the success denominator, including workflows that fail before returning a plan.

Required unknown checks, incomplete outputs, and challenge-category behavior will follow Step 17.

If source adoption or scorer verification remains incomplete, affected scores will remain unavailable rather than become automatic passes.

We will measure reference preparation, every role call, coordination, parsing, transport attempts, failures, and total elapsed time.

Version 3 changes both role organization and the amount of model computation. We will interpret its results alongside those costs and avoid attributing every difference solely to specialization.

The roles may also share errors because they use the same model.

### 27.12 Verify and Save the Development Workflow

Development checks will cover:

- Correct role routing and execution order.
- Independent specialist message lists.
- Matching reference packets across Versions 2–4.
- Exclusion of private evaluation fields.
- Valid limited-information contributions.
- Malformed or failed required contributions.
- Conflicting but structurally valid recommendations.
- Ambiguous food results.
- Source qualifications reaching the coordinator.
- Context overflow and generation limits.
- Call accounting and transport handling.
- State isolation and coordinator parsing.

Workflow verification and answer quality will be reported separately.

Before final execution, we will freeze role prompts, note schemas, routing, output limits, execution order, context rules, and failure behavior.

Expected outputs will include:

| Output | Contents |
|---|---|
| `src/champion_athletic_map/planners/specialists.py` | Fixed specialist and coordinator workflow |
| `docs/prompts/planner_v3/` | Versioned role and coordinator prompts |
| `artifacts/planners/version3_config.json` | Shared configuration references, routing, role limits, and execution policy |
| `artifacts/planners/version3_note_schema.json` | Specialist contribution structure |
| `results/planner_v3/development_manifest.csv` | Selected development cases and coverage |
| `results/planner_v3/development_agent_outputs.jsonl` | Raw notes, parsed contributions, message hashes, and handoff records |
| `results/planner_v3/development_runs.jsonl` | Coordinator responses and complete execution metadata |
| `results/planner_v3/development_checks.csv` | Workflow, context, access, and failure-handling checks |
| `results/planner_v3/development_scores.csv` | Independent scores where prerequisites are available |
| `docs/planner_v3.md` | Design decisions, findings, costs, and limitations |
| `results/step27_planner_v3_summary.json` | Producer record, hashes, role counts, failures, and readiness states |

The summary will distinguish preparation status, role execution, note validity, returned responses, scoring availability, and final readiness.

Final execution on the 200 frozen profiles will take place in Step 33.

We will add actual findings after implementation. Step 28 will extend this drafting procedure with an internal evaluator, coded checks, and bounded revisions.

## 28. Planner Version 4 — Specialist Agents with an Evaluator and Automated Checks

### 28.1 Purpose

We will extend the specialist workflow by checking its draft and allowing up to two coordinator revisions.

Version 4 will combine coded checks for conditions we can assess directly with a language-model evaluator for issues requiring interpretation.

We will measure whether this process improves the returned response and whether any improvement justifies the additional calls, runtime, and complexity.

Internal acceptance will describe the outcome of our configured procedure. It will not establish medical safety or real-world training effectiveness.

### 28.2 Verify Readiness and Generate the Initial Draft

We will use the same initial drafting procedure as Version 3:

- Training specialist.
- Participation and recovery specialist.
- Food-facts specialist when activated by the public request.
- Coordinator.

The model artifact, role prompts, public profile, reference packet, food routing, and initial generation settings will remain consistent with Step 27.

Version 4 will record its own execution. Earlier Version 3 outputs will not become unrecorded drafting work.

Before enabling the checks, we will verify the adopted rule definitions, required metadata, acceptance policy, and checker implementation.

Prepared test specifications from Step 13 will not count as executed validator tests. We will verify the implemented checks against independently defined expected results.

An unadopted rule register or missing required checker component will keep readiness blocked rather than produce an empty successful assessment.

The drafting function will expose raw coordinator text and separate completion and parsing statuses. A returned draft with a format defect may enter Version 4’s revision path.

A failed mandatory specialist contribution, failed coordinator request, or unresolved backend completion will remain an execution failure.

### 28.3 Implement the Coded Checks

The checker will use the original public profile, current candidate, adopted rules, and declared frozen reference information.

| Check area | What we will assess |
|---|---|
| Output structure | Required fields, types, and conditional response requirements |
| Schedule consistency | Durations, overlaps, availability, and existing commitments |
| Activity accounting | Separate sessions, included drill components, and possible double counting |
| Resources | Documented equipment, space, and participant requirements |
| Participation guidance | Applicable adopted limits and rest conditions |
| Reference identifiers | Whether supplied IDs exist and meet the relevant eligibility conditions |
| Food facts | Reported values, units, reference quantities, and supported calculations |

Every check result will identify:

- Candidate and check IDs.
- Rule and data versions.
- Applicability and its reason.
- Observed information and comparison basis.
- Result and explanation.
- Whether the finding is required or advisory.

Results will use the Step 13 statuses: **met, violated, unknown, and not applicable**.

Missing information will remain unknown. It will not become a recorded zero or an assumed match.

An existing source ID establishes record identity, not support for every associated claim. Missing optional IDs will follow the shared schema policy rather than become an undisclosed structural failure.

No additional searches or undeclared lookup calls will be introduced during revision. Checks will use the fixed information available to the workflow. If necessary supporting information is unavailable, the result will remain unknown.

A checker execution error will be recorded separately from a successfully executed check returning unknown.

### 28.4 Define Required Conditions and Internal Acceptance

We will save an acceptance policy before running the development comparison.

It will identify the expected checks, applicability rules, evaluator issue categories, and required or advisory status of each finding.

Internal acceptance will require:

1. A complete candidate meeting the applicable output structure.
2. A functioning checker with the expected check coverage.
3. Every required applicable coded condition met.
4. A recorded reason for each required not-applicable result.
5. A valid, completed evaluator assessment.
6. No unresolved evaluator issue that blocks acceptance under the policy.

A violated or unknown required condition will block acceptance. Missing check rows, an empty rule register, or zero executed checks will not create a pass.

Advisory findings will remain visible without automatically rejecting the response. For example, unavailable annual history may remain an advisory unknown for a weekly request when the adopted policy defines it that way.

Applicability will follow public profile facts and adopted definitions. Changing the response type to clarification or outside scope will not automatically make every planning requirement disappear.

Appropriate non-plan responses can be accepted under their own conditions:

| Response type | Required basis |
|---|---|
| Clarification | Questions address essential information that is actually missing or unresolved |
| Constraint conflict | The explanation identifies a supported conflict while preserving supplied facts |
| Outside scope | The response follows the adopted scope policy for the actual request |

An internally accepted clarification will remain a clarification. It will not be reported as a successful feasible plan.

### 28.5 Define the Evaluator’s Role

The evaluator will receive:

- The original public athlete profile.
- The current candidate, including raw text when parsing fails.
- The same reference packet.
- Coded check results.
- The fixed evaluator instructions and response schema.

It will examine goal relevance, completeness, contradictory instructions, unsupported claims, and handling of uncertainty.

Its structured report will include:

- Whether the assessment completed.
- Issue category.
- Affected field or activity.
- The observed problem.
- Supporting rule or source where available.
- Relevant uncertainty.
- A concise requested correction.

The application’s fixed policy will determine which issue categories block acceptance. The evaluator will not create new severity rules during a run.

Conclusive coded findings will remain authoritative within their declared scope. The evaluator may identify a separate interpretation concern, but it cannot waive a numerical violation or convert a checker failure into a pass.

Disagreements between coded findings and evaluator statements will remain traceable under the recorded disagreement policy.

We will ask for actionable findings rather than an unsupported overall quality score.

The evaluator will use the same frozen model as the other roles. Its findings may be mistaken and will remain available for later analysis.

A failed, malformed, or incomplete evaluator response will not be interpreted as “no issues.” The initial workflow will stop rather than request an undeclared evaluator repair.

### 28.6 Apply the Bounded Revision Loop

Candidates will have stable identifiers:

- Candidate 0: initial coordinator draft.
- Candidate 1: first revision.
- Candidate 2: second revision.

For each available candidate, we will:

1. Preserve its raw response and completion information.
2. Parse its structure.
3. Run the full applicable coded-check suite.
4. Obtain a fresh evaluator assessment.
5. Apply the fixed acceptance policy.
6. Return the candidate if internally accepted.
7. Otherwise, revise if a revision remains.

Malformed coordinator output will produce a format finding and unknown results for checks that cannot be computed. It may use one of the two revision opportunities.

A checker execution error will stop the workflow. It will not become a content problem for the model to waive.

Each revised candidate will receive fresh checks and evaluation. We will not inspect only the conditions that failed previously, because a revision can introduce new errors.

The loop will stop at the first internally accepted candidate, after the second assessed revision, or when a required stage cannot complete.

### 28.7 Give the Coordinator Focused Revision Inputs

A revision request will contain:

- The unchanged original public profile.
- The original reference packet.
- The original specialist notes.
- The current candidate.
- Current coded findings.
- Current evaluator findings.
- Instructions to return a complete replacement response.

The coordinator will preserve supplied commitments, resources, and source qualifications. It cannot resolve a problem by inventing missing profile facts, changing adopted rules, or pretending unavailable evidence exists.

The specialists will not restart, and food routing will remain tied to the original request.

The complete history will be saved in the logs. Revision prompts will use the current candidate and findings rather than append the entire history indefinitely.

We will not silently shorten substantive notes, discard qualifications, or add a summarization call to fit the prompt.

The initial evaluator output limit will be 768 tokens. Coordinator revisions will use the shared final-response limit. These starting limits will be tested and confirmed in the Step 31 pilot.

### 28.8 Define the Call Budget and Returned Response

The initial drafting procedure will use three calls, or four when the food role is active.

| Stage | Without food role | With food role |
|---|---:|---:|
| Initial drafting | 3 | 4 |
| Maximum evaluator assessments | 3 | 3 |
| Maximum coordinator revisions | 2 | 2 |
| **Maximum logical model calls** | **8** | **9** |

For a completed run:

| Revisions used | Without food role | With food role |
|---|---:|---:|
| 0 | 4 | 5 |
| 1 | 6 | 7 |
| 2 | 8 | 9 |

Failures may stop execution earlier. Reference preparation, coded checks, and transport attempts will be recorded separately.

Each model request will follow Step 24’s narrow pre-acceptance retry policy. Content problems will use only the declared coordinator revisions.

Terminal states will distinguish:

| State | Returned outcome |
|---|---|
| Internally accepted | First candidate satisfying the acceptance policy |
| Revision limit reached | Last fully checked and evaluated candidate, with unresolved findings |
| Execution error | Failed workflow; partial outputs retained for diagnosis |
| Blocked before execution | Required configuration, source adoption, or checker readiness unresolved |

At the revision limit, we will not select an earlier candidate because it appears better or receives a higher external score.

If a required checking, evaluator, or revision stage fails, earlier drafts will remain diagnostic material. They will not silently become a successful final response.

The application will distinguish an accepted response from an unresolved draft.

### 28.9 Preserve Independent Research Evaluation

The final research scorer will assess the returned response afresh using the same independently defined criteria applied to the other versions.

It will not adopt Version 4’s internal acceptance flag, evaluator opinion, or cached check report as the final score.

Verified calculation helpers may be reused, but the final scoring execution and results will remain separate.

Private scenario labels, expected responses, and feasibility witnesses will remain unavailable to the internal checker, evaluator, and revision prompts.

All 150 final feasible profiles will remain in the success denominator. Execution failures and unavailable responses will remain outcomes.

A revision-limit response will still receive independent assessment. It may perform better or worse than its initial draft.

We will report internal acceptance separately from independent performance and examine:

- Incorrect evaluator concerns.
- Internally accepted responses that fail independent assessment.
- Required issues that remain unresolved.
- Revisions that introduce new problems.
- Improvements relative to the added calls and runtime.

Human review will use the designated final responses. Intermediate notes and self-ratings will not become additional independent research results.

### 28.10 Verify the Workflow During Development

Development cases and controlled fixtures will cover:

- Initial acceptance without revision.
- One successful revision.
- Two revisions with unresolved findings.
- Malformed coordinator output entering the revision path.
- A failed mandatory specialist.
- Missing required information and advisory unknowns.
- An unsupported change of response type.
- Incomplete check coverage or an unadopted register.
- A checker exception.
- Invalid or incomplete evaluator output.
- Disagreement with a conclusive coded result.
- A revision introducing a new error.
- Context overflow and uncertain backend completion.
- Correct selection of the terminal response.

We will verify that every returned candidate was assessed, call limits were respected, original facts remained unchanged, and private evaluation information stayed excluded.

Workflow checks and response quality will be reported separately.

Affected adherence scores will remain unavailable when rule adoption or scorer verification is incomplete. Successful execution alone will not establish readiness for final scoring.

- Issue categories, disagreement handling, and acceptance conditions.
- Schemas and per-role output limits.
- Revision and call budgets.
- Failure handling and returned-candidate policy.

Expected outputs will include:

## 29. Prototype Application, Athlete Intake, Editable Plans, and Progress Checklist

### 29.1 Purpose

We will build a Streamlit prototype that lets users enter an athlete profile, generate a weekly basketball plan, inspect its sources, save changes, and record completed activities.

The main generation action will run Version 4 through the modules developed earlier. The four-version research comparison will remain in its separate experiment runner.

We will use fictional development profiles for testing and demonstrations.

The prototype will make the plan’s current status clear. A saved edit, an unresolved draft, and a reviewed plan will remain distinguishable.

### 29.2 Organize the Main Actions

The interface will use a small set of explicit actions.

| Action | Effect |
|---|---|
| Save profile | Commit the entered planning information without generating |
| Generate plan | Save the submitted profile and start one bounded Version 4 workflow |
| Save plan changes | Create a saved plan revision and rerun coded checks |
| Review saved changes | Run fresh coded checks and one evaluator assessment |
| Update progress | Record completion without changing the schedule |
| Download | Export the selected saved version |
| Restore | Validate and load a snapshot without starting generation |

Reviewing saved changes will report findings without rewriting the plan. Generating a new plan will remain a separate action.

Labels and explanations will use ordinary language. Source links and relevant findings will appear beside the plan, while detailed hashes and execution traces will remain in development records.

### 29.3 Build the Athlete Intake Form

We will collect the information needed by the shared planning interface.

| Information | Planned fields |
|---|---|
| Profile basics | Non-identifying label, age, and school grade |
| Basketball background | Experience, skill level, and position where relevant |
| Goals | Basketball skills and supported conditioning priorities |
| Planning period | Week start date and relevant time zone |
| Availability | Available days, time windows, and session limits |
| Existing commitments | Practices, games, other sports, and their timing or duration |
| Resources | Equipment, space, court or hoop access, and participants |
| Additional request | Relevant constraints, clarification needs, or an explicit food-facts question |

Unknown values and confirmed absences will remain different inputs.

For example, an unreported practice schedule will not become an empty schedule. Grade will not be inferred from age.

We will check field types and impossible values without removing legitimate uncertainty or conflicting commitments.

Requests outside the supported scope will receive an appropriate explanation rather than have their inputs silently changed.

Loading a demonstration profile will be an explicit action. It will not replace an active profile without the user choosing to load it.

Form values will remain drafts until submission. The displayed plan will identify the saved profile version it belongs to.

### 29.4 Start and Track Generation Explicitly

Generation will start only from the submitted planning action.

Editing a field, opening a section, downloading a file, or updating a checklist will not start another model workflow.

Before execution, we will:

1. Validate the submitted profile structure.
2. Create an immutable profile snapshot.
3. Assign a job ID and record the configuration.
4. Check Version 4’s required readiness conditions.
5. Mark the job as active before invoking the workflow.

The interface will show progress and prevent duplicate submissions.

Ordinary Streamlit reruns will display the existing job or result rather than submit it again.

Each result will remain attached to its original profile snapshot and job. If the user switches profiles or resets the interface while work finishes, that result will not appear as the new profile’s plan.

A browser reset will not be treated as confirmed cancellation of an active backend request. Uncertain request states will follow Step 24’s handling policy.

If Version 4 is blocked or fails, the application will preserve the submitted information and explain the outcome. It will not silently substitute another planner.

### 29.5 Display the Response and Its Status

A planning response will appear as a readable weekly schedule showing:

- Proposed activities and existing commitments.
- Allocated durations and relevant timing.
- Instructions and resource requirements.
- Rest arrangements.
- Available source links.
- Unresolved findings.

Clarification, conflict, and outside-scope responses will use their appropriate displays rather than produce an empty schedule.

The application will translate technical outcomes into plain labels, such as:

- Plan checked.
- Changes need review.
- More information needed.
- A conflict remains.
- Draft needs attention.
- Planning is unavailable.

A revision-limit response will retain its unresolved status. An execution failure will not be presented as a completed plan.

We will explain what was checked without treating internal acceptance as a guarantee about the plan.

Optional food facts will display the structured lookup values, food identity, reference quantity, units, and source. Model-written explanations will remain distinguishable from those recorded values.

### 29.6 Preserve the Original and Support Controlled Editing

The original returned response will remain unchanged.

A separate working copy will support the initial editing scope:

- Change a proposed activity’s day or time.
- Change its allocated duration.
- Substitute an activity from the eligible catalog.

Existing commitments will be changed through the profile form. They will not be removable as though they were optional proposed sessions.

Source instructions, canonical identifiers, eligibility, and verification fields will remain application-controlled.

Disabling editor columns will support the interface, but backend validation will enforce those restrictions when changes are saved or restored.

A replacement will receive its own authoritative instructions, requirements, and source links. It will not retain the replaced activity’s citation.

Saving will:

1. Validate the proposed changes.
2. Resolve source-controlled fields against the verified catalog.
3. Create a new immutable plan revision when planning content changes.
4. Record the edit and activity-version relationships.
5. Run the full applicable coded-check suite.
6. Mark the earlier model assessment as outdated.

Structurally valid changes that introduce a conflict may remain saved as an unresolved draft. The application will show the findings rather than silently repair the schedule.

Activity identity will not depend on table row positions.

### 29.7 Review the Exact Saved Version

The review action will assess the saved plan, not unsaved editor values.

It will run fresh coded checks and one evaluator call using the Step 28 assessment definitions. It will not start coordinator revisions or modify the user’s changes.

If a changed profile or replacement activity needs different reference information, we will prepare a new, recorded review context from the declared source versions.

This will be a separate interactive assessment job. It will not alter the earlier generation record or frozen research packet.

Every assessment will identify:

- Profile version and planning-content hash.
- Plan version and content hash.
- Reference-packet and source versions.
- Rule and checker versions.
- Evaluator model and prompt versions.
- Acceptance-policy version.

An assessment will be current only for that combination.

After saving an edit, passing coded checks alone will not restore the earlier model review. The interface may show “Schedule checks updated; model review needed.”

A checker or evaluator failure will remain an unavailable or incomplete assessment.

Changes to planning facts or relevant source versions will invalidate the current assessment. Checklist updates and display-only changes will not.

### 29.8 Add a Progress Checklist

Our initial checklist will track top-level training sessions shown in the week, including supplied practices and games and proposed training sessions.

Rest days, non-training commitments, and drill components already included within a session will not add checklist items.

Each trackable session can be marked:

- Pending.
- Completed.
- Skipped.

Progress records will retain the activity occurrence, activity revision, plan version, reported status, and update time.

We will display completed, pending, and skipped counts. Skipped sessions will remain in the total.

If there are no trackable sessions, the interface will state that directly rather than display an undefined or misleading completion percentage.

Progress will carry forward only for unchanged activity versions within the same week.

A changed date, time, duration, or activity definition will create a new activity revision and reset its current progress. A replacement will receive a new identity linked to the replaced activity. Earlier history will remain available.

Display-only row reordering will not erase progress.

Checklist information will remain outside the assessed planning-content hash. Updating it will not change the schedule, invalidate its assessment, or call the model.

These records will describe self-reported completion, not training quality or athletic improvement.

### 29.9 Separate Session State from Shared Resources

Session state will hold the current profile, saved versions, working editor values, assessments, active-job references, and progress.

Streamlit session state is temporary and can reset when the browser connection is replaced. Save-and-restore features will provide continuity; the prototype will not promise permanent account history.

Ollama will manage model residency. Streamlit may cache stateless clients and verified read-only reference artifacts.

Mutable profiles, conversations, generated plans, and progress will remain outside globally shared cached resources.

The initial local server will use a process-wide execution gate for model work. A session-level busy flag alone will not be treated as sufficient to prevent overlapping jobs from different sessions.

Generation and explicit review actions will share that gate.

The execution controller will preserve job ownership and handle completion or failure without attaching one session’s result to another session.

### 29.10 Export and Restore a Coherent Snapshot

We will provide:

- A readable CSV schedule.
- A structured JSON snapshot for restoration.

The snapshot will include the minimal planning profile, selected plan version, activity identities, source references, assessment history, relevant edits, progress, and application/data versions.

Credentials, private benchmark expectations, model weights, and internal reference collections will remain outside the export.

Restoration will be an explicit action. It will not automatically generate a plan or overwrite the active working state.

We will validate the snapshot schema, identifiers, dates, quantities, and activity relationships.

Source-controlled fields will be resolved against the declared available reference versions. Imported approval or eligibility flags will not update the project’s source or review ledgers.

If a required source version is missing or incompatible, the application will show an unresolved historical view rather than pretend that current verification succeeded.

Restored assessments will remain historical. We will recompute coded checks, and a fresh model assessment will remain an explicit user action.

Embedded acceptance flags or hashes will not establish current validity.

If current source information changes the working content, that change will create a new recorded revision while preserving the imported original.

### 29.11 Keep Interactive Use Separate from Research

The application will reuse the planning modules, but interactive sessions will remain separate from the controlled experiment.

The research runner will preserve its returned responses before user edits. App generations, edited plans, and checklist events will not replace those outputs.

They will also not automatically become training examples or evaluation labels.

Demonstrations will use development profiles rather than the final held-out set.

A new app generation or review will receive its own job record. It will not reopen a completed benchmark job or alter the sources frozen for that experiment.

### 29.12 Verify the Complete User Flow

Development checks will confirm that a user can:

- Enter or load a fictional profile.
- Save information without starting generation.
- Start one intended Version 4 workflow.
- Understand clarification, conflict, blocked, and failure outcomes.
- Inspect the schedule and its sources.
- Save edits and see the correct review status.
- Review saved changes without automatic rewriting.
- Update progress without a model call.
- Download and restore a consistent snapshot.
- Reset or switch profiles without mixing results.

We will also check:

- Duplicate submissions and ordinary reruns.
- Late results after a profile switch.
- Two-session state isolation and shared-runtime access.
- Protected-field enforcement.
- Assessment invalidation after planning changes.
- Progress retention and reset rules.
- Empty checklist summaries.
- Imported assessments remaining historical.
- Required source or checker readiness remaining visible.

Workflow checks will be reported separately from model-response quality.

Step 30 will consolidate these checks with the broader integration and failure-handling tests.

### 29.13 Expected Outputs

After completing this step, we expect to prepare:

| Output | Contents |
|---|---|

## 30. Integration Tests, Invalid Requests, and Planning Constraint Checks

### 30.1 Purpose

We will bring together the checks introduced in earlier steps and verify that the project’s components work correctly as a system.

Testing will focus on calculations, reference access, workflow boundaries, application state, failure handling, and result accounting.

Software correctness and model quality will remain separate. A model may produce an incorrect plan while the software correctly records and reports its problems.

This step will establish which behaviors have actually been tested and what remains unresolved before the development pilot.

### 30.2 Organize the Test Suite

We will use complementary test types.

| Test type | Main purpose |
|---|---|
| Focused function tests | Verify calculations, applicability, parsing, versioning, and state updates |
| Integration tests | Verify communication between tools, planners, checks, and execution records |
| Application tests | Verify supported Streamlit interactions |
| Manual browser checks | Inspect actual editing, downloads, restoration, and session behavior |
| Real-backend checks | Verify the selected local Ollama artifact and runtime |

Routine tests will use small fixtures and scripted model responses. They will run without contacting Ollama, downloading models, or refreshing publisher sources.

Real-backend checks will require explicit selection and will have separate records.

Tests will focus on behavior that affects calculations, reference access, application state, and research results.

### 30.3 Prepare Independent Fixtures and Expected Results

Fixtures will contain small profiles, source records, candidate plans, tool responses, and expected outcomes.

Expected results will come from reviewed requirements, source evidence, or hand-worked examples. We will not obtain an expected answer by calling the function being tested.

Each relevant test will link to its requirement or rule, fixture version, expected outcome, and execution evidence.

Draft specifications from Step 13 will remain distinguishable from:

- Reviewed requirements.
- Implemented tests.
- Tests actually executed.
- Operational rules adopted by the project.

Clearly labeled synthetic rules may test software mechanics while source adoption remains pending. Passing those tests will not adopt their values as published guidance.

Tests that simulate damaged files or changed records will use temporary copies. Original sources, frozen artifacts, and review ledgers will remain unchanged.

Development fixtures will exclude final athlete profiles and held-out retrieval questions. Split identifiers and hashes may support integrity checks without transforming or scoring final classifier text.

### 30.4 Test Input Handling and Uncertainty

Input tests will cover:

- Invalid types and impossible values.
- Missing age or school grade.
- Ages outside the prototype’s supported range.
- Missing availability or commitment information.
- Conflicting commitments.
- Negative or nonfinite durations and quantities.
- Dates outside the selected week.
- Unknown equipment, space, or participant availability.
- Ambiguous food descriptions.
- Requests outside the project’s scope.

We will preserve the difference between unknown and confirmed absent information.

An unreported practice schedule must remain different from a confirmed schedule containing no practices.

The application may identify invalid form entries, but the research runner must retain its planned challenging cases and record their outcomes. Frontend filtering must not remove difficult benchmark cases.

Tests will also verify that private scenario labels, expected answers, and feasibility witnesses never enter model prompts or reference packets.

### 30.5 Test Planning Calculations and Rule Boundaries

For each adopted numerical rule, we will test values below, equal to, and above its boundary.

Equality will follow the recorded operator. We will not assume that every maximum or minimum uses the same comparison.

| Area | Cases to verify |
|---|---|
| Participation totals | Existing commitments combined with proposed sessions |
| Activity classification | Organized basketball, other activity types, and unresolved classification |
| Event counting | Duplicate entries, included components, and separate repeated sessions |
| Rest arrangements | Existing basketball commitments on a claimed rest day |
| Applicability | Age/grade overlap, missing inputs, and unresolved interpretation |
| Resources | Available, unavailable, and unknown requirements |
| Time windows | Overlaps, adjacent sessions, invalid intervals, and week boundaries |
| Units | Consistent minutes, hours, and supported quantity conversions |
| Optional history | Unknown annual participation or sleep information under the adopted policy |

Double-counting tests will distinguish three situations:

- The same event entered twice.
- A drill already included within a practice.
- Two separate sessions using the same drill.

A shared drill name or catalog ID will not collapse two legitimate sessions.

For example, a 60-minute practice containing drill components and a separate 30-minute session will contribute 90 minutes.

We will verify that a passing maximum-limit check can coexist with a separate recommendation or advisory finding.

Required unknowns must remain unresolved. Not-applicable results must retain an applicability reason.

A changed response label must not automatically waive relevant requirements.

### 30.6 Test Reference Tools, Retrieval, and Classifier Boundaries

Reference integration tests will verify that:

- Pending records do not become planner eligible.
- An unadopted rule register returns an explained unavailable state.
- Missing evidence remains different from technical failure.
- Versions 2–4 use the same canonical packet for matched inputs.
- Two searches merge in the recorded round-robin order.
- Repeated evidence units and catalog IDs are handled consistently.
- Search, catalog, and food limits are enforced.
- All reference sections share one total context budget.
- Necessary qualifications are preserved or their omission is recorded.
- Packet hashes exclude volatile operational metadata.

Retrieval-metric fixtures will distinguish relevant-unit retrieval from complete delivered evidence.

They will also verify that unsupported questions have recall marked not applicable, while failed answerable requests remain in the declared denominator.

Food-tool tests will cover identity ambiguity, missing values, valid zeros, units, gram scaling, and eligible household portions. Recorded portion amounts must not be multiplied twice.

We will reuse relevant evidence from earlier steps when its code and artifact versions still match, identifying it as reused evidence.

Classifier integration checks will confirm the frozen class mapping, model/preprocessing correspondence, and protected split boundaries. Dedicated synthetic inputs can test training-only fitting without inspecting final test features.

Catalog suggestions must not overwrite frozen benchmark targets or automatically approve an activity.

### 30.7 Verify Planner Access and Call Budgets

A scripted backend will record submitted roles, messages, settings, and call order while returning controlled responses.

We will verify that:

- Version 1 receives no project references or lookup results.
- Versions 1 and 2 make at most their one logical generation request.
- Versions 1–3 receive no post-generation checking feedback.
- Version 3 specialists do not receive peer notes.
- The coordinator receives all required contributions and original inputs.
- No extra searches occur between specialist or revision stages.
- Food routing uses the public request.

### 30.10 Verify Independent Scoring and Result Accounting

### 30.11 Run Controlled and Real-Backend Checks Separately

We will register a `real_backend` pytest marker and exclude those checks from the routine suite.

The planned routine command will be:

`python -m pytest tests/unit tests/integration tests/app -m "not real_backend" --junitxml=results/testing/test_report.xml`

A small set of development requests will inspect:

- Server and model availability.
- Chat formatting and effective settings.
- Reference delivery and context handling.
- Raw response preservation.
- Completion and parsing metadata.
- Timings, token usage, and failure reporting.

We will assess observable contracts rather than require one exact generated schedule or wording.

If a backend check cannot run, it will remain blocked or unexecuted rather than become evidence of backend readiness.

### 30.12 Record Outcomes and Resolve Blocking Defects

- Collected and executed tests.
- Passed assertions.
- Failed assertions and execution errors.
- Skipped or unexecuted checks.
- Expected failures and their unresolved requirements.
- Manual checks.
- Actual backend observations.

A run collecting zero tests will not count as a successful suite. Skipped required checks will not disappear from readiness reporting.

We will avoid broad coverage claims that are not supported by the requirement-to-test mapping and actual execution evidence.

For a confirmed defect, we will record its cause, correction, and affected checks. A focused regression case will be added where useful.

After a change, we will run the affected tests and complete the required routine suite before freezing the experiment.

Blocking defects in calculations, access control, termination, state isolation, and scoring must be resolved before final execution.

Passing controlled tests can coexist with pending source adoption, unavailable live-backend evidence, or incomplete scoring readiness. These conditions will remain separate.

## 31. Development Pilot, Runtime and Cost Checks, and Final Evaluation Protocol

### 31.1 Purpose

We will run a development pilot before beginning the held-out evaluations.

The pilot will help us confirm that all four planner versions execute as intended, fit our available resources, and produce complete records for analysis.

We will use its findings to resolve development problems and confirm the operating limits. We will then freeze the evaluation protocol before examining final outcomes.

Technical diagnostics using fixtures will remain distinguishable from a representative pilot using the adopted project resources.

Required source approvals, rule adoption, checker coverage, and scoring readiness must be resolved before we declare the final experiment ready.

### 31.2 Select Twenty Development Profiles

We will select 20 reviewed profiles from the development partition.

The sample will contain four profiles per age, including two feasible requests per age.

| Age | Feasible | Missing information | Conflicting constraints | Outside scope | Total |
|---:|---:|---:|---:|---:|---:|
| 13 | 2 | 1 | 1 | 0 | 4 |
| 14 | 2 | 1 | 0 | 1 | 4 |
| 15 | 2 | 0 | 1 | 1 | 4 |
| 16 | 2 | 1 | 1 | 0 | 4 |
| 17 | 2 | 1 | 0 | 1 | 4 |
| **Total** | **10** | **4** | **3** | **3** | **20** |

We will include different experience levels, resource restrictions, commitment loads, and context lengths.

At least two selected profiles will contain explicit food-facts requests so we can exercise the optional route.

Selection will use profile metadata and reviewed coverage requirements before examining pilot responses. We will save the selected IDs, family relationships, and selection procedure.

If the development pool cannot support this coverage, we will document and resolve the gap before execution. Final profiles will remain outside that work.

Private categories and expected outcomes may support offline sample selection and scoring. They will not enter model prompts.

### 31.3 Run Matched Cases in a Balanced Order

All four versions will process each selected profile, producing **80 scheduled pilot jobs**.

Each matched case will use the same public input and common model configuration. Versions 2–4 will use matching canonical reference-packet contents.

We will run workflows sequentially on the recorded environment.

To balance execution position, we will use four fixed orders:

| Order | Planner sequence |
|---|---|
| A | 1 → 2 → 4 → 3 |
| B | 2 → 3 → 1 → 4 |
| C | 3 → 4 → 2 → 1 |
| D | 4 → 1 → 3 → 2 |

Within each age group, the four pilot profiles will receive one order each through a recorded seeded assignment.

For the final 40 profiles per age, each order will be assigned ten times. The assignment will be saved before final execution.

Every job will start with fresh messages and workflow state. Model weights and verified read-only indexes may remain loaded.

For primary timing, computed reference-packet and query-result caching will be disabled. Each reference-enabled workflow will incur its own measured preparation while producing the same canonical packet.

Native Ollama caching will be recorded where observable. It will not be assumed absent.

If a development correction changes the configuration, we will preserve the earlier records and rerun affected comparisons. Incompatible configurations will not be pooled into one result.

### 31.4 Count Jobs and Model Calls Separately

The pilot and final comparison will count profile–version jobs separately from language-model generation calls.

| Version | Logical generation calls for a completed workflow |
|---|---|
| Version 1 | 1 |
| Version 2 | 1 |
| Version 3 | 3, or 4 with the food role |
| Version 4 | 4, 6, or 8; or 5, 7, or 9 with the food role |

Version 4’s count depends on whether it uses zero, one, or two revisions.

For one profile processed by all four versions:

**Logical calls = 9 + 2 × food-role flag + 2 × Version 4 revision count**

For the final 200 profiles:

**Total planned logical calls = 1,800 + 2F + 2R**

Here:

- `F` is the number of profiles activating the food role.
- `R` is the total number of Version 4 revisions across those profiles.

The maximum configured generation budget is:

**2,600 + 2F, with an overall ceiling of 3,000 logical calls.**

These figures exclude warm-up calls and transport attempts. Failed workflows may stop earlier.

Embedding computation, classifier inference, reference requests, and coded checks will be measured separately from planner generation calls.

The final experiment will contain **800 scheduled jobs**, not a guarantee of 800 completed responses or only 800 model calls.

### 31.5 Measure Runtime Consistently

We will separate setup from request processing.

Setup records will include loading the selected model and reference artifacts, index preparation where needed, and warm-up activity.

Primary workflow timing will begin before reference preparation and end after the returned response or terminal failure has been recorded and parsed.

For Version 4, that interval will include its coded checks, evaluator calls, and revisions.

Independent research scoring, human review, and interface rendering will remain outside generation-workflow timing.

We will record:

- Reference preparation and context assembly.
- Each model request.
- Internal checking and evaluation.
- Revisions and permitted transport attempts.
- Total elapsed time.
- Completion and failure status.
- Available backend timing and cache measurements.

A monotonic client timer will measure the complete workflow. Ollama’s returned durations will provide separate server-side measurements.

A Python-client PyTorch synchronization call will not be used to time Ollama’s separate process. Direct in-process GPU operations will use appropriate timing methods when measured.

We will report completed-response latency with its denominator, alongside failure counts and time consumed by failed attempts.

Sample counts, medians, ranges, and relevant means will accompany the summaries. Unexpected reloads will remain visible rather than be silently subtracted.

### 31.6 Confirm Context, Memory, and Operating Limits

The pilot will test the full request sizes used by specialists, coordinators, evaluators, and revisions.

We will confirm:

- The local Ollama model digest, template, quantization, and server version.
- Actual execution device and offloading.
- Explicit context allocation.
- Shared final-response output limits.
- Specialist and evaluator limits.
- Reference-packet budget.
- Per-call and per-workflow time limits.
- Search, catalog, and food-lookup caps.
- Transport handling.
- Revision limits and acceptance policy.

Token accounting will include instructions, public profiles, references, notes, current candidates, findings, and template overhead.

We will verify that essential qualifications survive context assembly. Estimates will remain distinguishable from verified token counts.

Memory measurements will identify their source and scope. Client-process PyTorch counters will not be presented as Ollama’s total GPU memory use.

Unavailable measurements will remain unavailable.

A caller timeout will not establish that backend computation has stopped. We will verify how the execution controller resolves uncertain completion before permitting another request.

### 31.7 Project the Final Workload and Cost

We will prepare a typical workload estimate and a maximum configured-call scenario.

The final scenario distribution differs from the pilot, so we will not simply multiply the unweighted pilot average.

| Scenario | Pilot profiles | Final profiles |
|---|---:|---:|
| Feasible | 10 | 150 |
| Missing information | 4 | 20 |
| Conflict | 3 | 15 |
| Outside scope | 3 | 15 |

Where supported by the sample, the runtime projection will multiply each final category count by its corresponding mean workflow time.

We will account for food activation, revision frequency, setup, and failure consumption without adding costs twice when they are already included in measured totals.

Sparse categories will use documented pooled assumptions or ranges. Missing observations will not become zero-cost cases.

The configured call ceiling will describe a workload scenario, not a guaranteed wall-clock bound.

For local execution, we will report elapsed resource time and available usage measurements. Monetary cost will remain unavailable unless we have a documented calculation basis.

If the projection exceeds our practical resources, we will revise development settings, version the change, and repeat affected checks before freezing the experiment.

### 31.8 Confirm the Held-Out Assessments and Human Sample

We will confirm all final evaluation components before examining their outcomes.

| Component | Frozen assessment |
|---|---|
| Exercise classifier | Step 15 test partition and selected comparison artifacts |
| Retrieval | Thirty held-out questions: twenty answerable and ten unsupported |
| Planners | Two hundred profiles processed by all four versions |
| Human review | Forty selected profiles, with all four outcomes reviewed by two teammates |

The planner set will contain 40 profiles per age from 13 through 17:

- 150 feasible requests.
- 20 missing-information cases.
- 15 conflict cases.
- 15 outside-scope cases.

The human-review sample will contain:

| Category | Profiles |
|---|---:|
| Feasible | 25 |
| Missing information | 5 |
| Conflict | 5 |
| Outside scope | 5 |
| **Total** | **40** |

Each age will contribute five feasible profiles and one from each challenge category.

We will freeze those IDs before viewing final responses. All four outcomes, including failures, will remain included, giving 160 response records for two reviewers.

The five rubric dimensions and 0/1/2 anchors will be fixed. Reviewers will practise on development examples before the freeze.

Reviewers will be described accurately as teammates. Anonymous labels will conceal version identities, although response content may still provide clues.

### 31.9 Freeze Metrics, Comparisons, and Uncertainty Rules

The primary classifier measure will remain macro-F1 over the frozen eligible class list.

We will preserve the benchmark `target_label`, class mapping, and documented source-label limitations. We will not describe every source label as independently verified.

The primary planner measure will remain:

**Successful feasible cases ÷ all 150 feasible profiles**

Required unknowns, missing plans, execution failures, and inappropriate refusals will prevent success. Missing check rows or an empty checker will not produce a pass.

Challenge-category results, evidence quality, human ratings, and runtime will remain separate measures.

The planned paired comparisons will be:

- Version 2 minus Version 1.
- Version 3 minus Version 2.
- Version 4 minus Version 3.
- Version 4 minus Version 1.

Step 35 will use 2,000 paired group-level bootstrap resamples, seed 42, and individual 95% percentile intervals.

The sampling units will be profile families for planner analysis and exercise groups for classifier analysis.

Within each declared analysis population, we will sample groups with replacement, retain their members and repeated sampling multiplicities, and keep all compared outcomes paired.

Metrics will remain profile- or record-based. We will not replace them with equally weighted group averages.

The primary feasible-plan analysis will use the feasible cohort. The initial bootstrap will not add age stratification after results are inspected.

Classifier resamples will retain the frozen class list and zero-division policy. Undefined or degenerate results will be explained rather than silently discarded.

The implementation will use explicit group-index resampling and percentile endpoints. Ordinary paired row resampling does not establish group resampling, and library defaults will not replace the declared method.

These will be individual uncertainty summaries within our constructed benchmark. We will emphasize effect sizes and uncertainty without formal significance or real-population claims.

### 31.11 Define Resumption and Amendment Rules

Every final job will have a stable identity connecting its protocol, profile, and planner version. Model stages and attempts will have linked identifiers.

## 32. Final Exercise Classifier Evaluation and Error Analysis

### 32.1 Purpose

We will evaluate our three frozen exercise classifiers on the held-out test partition. This will show how well they predict body-part labels for exercise groups that were kept outside training and validation.

We will compare:

- The majority-class baseline.
- The selected TF-IDF and logistic regression pipeline.
- The selected neural classifier.

We will use the benchmark labels saved in the Step 15 split manifest, including corrections accepted before the evaluation was frozen. We will report their source and review status; they have not all been independently verified by an anatomy expert.

After calculating the results, we will examine common mistakes and difficult inputs. These findings will help us explain the models’ strengths and limitations.

### 32.2 Verify the Protocol, Artifacts, and Test Population

We will begin only after the evaluation protocol from Step 31 has been finalized.

Before opening the test text for inference, we will verify the recorded file hashes, model identities, preprocessing configuration, class mapping, source snapshot, and split manifest.

| Model | Artifact used for final evaluation |
|---|---|
| Majority-class baseline | The fitted baseline whose predicted class was determined from training labels |
| Logistic regression | The selected complete pipeline, fitted on the training partition |
| Neural classifier | The selected configuration’s seed-42 checkpoint from its best validation epoch |

The additional neural training seeds from Step 21 will remain development evidence. We will use the seed-42 artifact selected before testing.

All three models will remain part of this assessment, including when automatic classifier-based tagging was disabled for the application.

We will confirm that:

- Every evaluated record belongs to the eligible test population in the frozen manifest.
- All three models use exactly the same exercise IDs and reference targets.
- Exercise groups do not cross the training, validation, and test partitions.
- The authoritative target is the manifest’s `target_label`.
- Class names, numeric class IDs, and their ordering match the saved mapping.
- No model was refitted on combined training and validation data after selection.
- The execution environment matches the recorded dependency requirements.

We will report the actual number of test records, distinct exercise groups, and included classes. We will also check the Step 15 requirement of at least two distinct test groups per eligible class.

An unexpected missing class or support mismatch will require investigation before headline reporting. We will preserve the frozen population while resolving the discrepancy through the recorded correction process.

Our groups reduce known overlap between related records. They do not establish that every group is statistically independent.

### 32.3 Apply the Saved Preprocessing and Generate Predictions

We will reconstruct each model input from the prepared exercise name and available description, using the same canonical format established in Step 18: nonempty components joined by two newline characters.

The target label and other excluded metadata will remain outside the model input. Exercise and group identifiers will support alignment and reporting.

For logistic regression, we will pass the canonical text through the saved complete pipeline. That pipeline will apply its own saved normalization, vocabulary, TF-IDF transformation, and classifier.

For the neural model, we will use the saved normalizer, tokenizer, training vocabulary, unknown-token handling, sequence limit, truncation, padding, and label mapping. We will process every test record with shuffling disabled and without dropping the final batch.

We will call `model.eval()` and disable gradient recording during inference. Evaluation mode and gradient control will both be applied.

Input padding will remain separate from target labels. Class index zero is a valid predicted class.

We will retain valid difficult inputs, including:

- Exercises with a name but no description.
- Text containing unfamiliar words.
- Inputs whose neural tokens are all unknown.
- Inputs that produce an all-zero TF-IDF vector.

A genuinely empty or corrupted input will be recorded as a data or execution problem. It will not receive invented text or a replacement prediction.

We will measure vocabulary coverage and truncation on the test set to help explain the results. These diagnostics will not change the vocabulary, sequence limit, thresholds, or evaluated population.

### 32.4 Check Prediction Completeness and Alignment

Before calculating the main comparison, we will require exactly one valid prediction from each model for every expected test record.

Our checks will cover:

- Missing, duplicate, and unexpected exercise IDs.
- Mismatched reference labels or group IDs.
- Predictions outside the frozen class list.
- Incorrect conversion between numeric IDs and readable labels.
- Missing batches or silently dropped records.
- Invalid numerical outputs used by the prediction procedure.

We will align results by exercise ID rather than assume that separate output files have the same row order.

If a model fails to produce a complete prediction set, we will preserve its execution records and mark the comparison incomplete. We will not restrict the primary comparison to whichever rows succeeded.

Any technical correction will follow the amendment and resumption policy from Step 31. We will document the defect, preserve earlier evidence, verify the correction, and repeat the affected work consistently.

Low performance alone will not justify changing a model or repeating its evaluation.

### 32.5 Calculate the Final Classification Metrics

Macro-F1 will remain our primary classifier measure.

We will calculate it over the complete frozen class list, giving each included class equal weight in the average.

| Measure | What it tells us |
|---|---|
| Macro-F1 | Average class-level F1, with equal weight for each included class |
| Accuracy | Proportion of test records classified correctly |
| Per-class precision | How often predictions for a class match the reference label |
| Per-class recall | How many reference records in a class the model identifies |
| Per-class F1 | The balance between precision and recall for that class |
| Record support | Number of reference records supporting each class result |
| Distinct-group support | Number of exercise groups containing those reference records |

All models will use the same explicit class list, class ordering, and `zero_division=0` convention.

The primary metrics will be calculated across records. Distinct-group counts will provide additional context; they will not silently change the metric weighting.

A group containing records from several classes can contribute to more than one class’s group count. We will therefore avoid adding per-class group counts and describing the sum as the total number of groups.

We will save results at full numerical precision and round only the displayed tables.

A class that receives no predictions can have undefined precision represented as zero under our scoring convention. We will explain that convention alongside its support.

Missing reference support in the original test partition will remain a population-check issue. Missing support within a later bootstrap resample will be handled using the frozen analysis conventions.

### 32.6 Examine Confusion Patterns

We will create count and row-normalized confusion matrices for each model.

Every matrix will use the same class order:

- Rows will represent reference labels.
- Columns will represent predicted labels.
- Count matrices will show the number of records in each combination.
- Row-normalized matrices will show the prediction proportions within each reference class.

We will use a consistent scale for the normalized figures and include readable class names.

The figures will be saved together with their underlying numerical tables. Class support will remain available so that a pattern based on a small sample is easy to recognize.

We will examine which classes are confused most often and whether the learned models improve on the majority baseline across several classes or mainly within the largest ones.

### 32.7 Review Errors and Input Difficulties

We will connect each prediction to the original input and relevant diagnostic information.

Our review will examine:

- Frequently confused class pairs.
- Exercises with missing descriptions.
- Vocabulary coverage and unknown-token rates.
- All-zero TF-IDF inputs.
- Input length and neural truncation.
- Exercises describing related or multiple muscle groups.
- Cases where logistic regression and the neural classifier disagree.
- Errors concentrated within a small number of exercise groups.

We will report subgroup counts and identify analyses introduced after seeing the results as exploratory.

For illustrative examples, we will record the selection method and include enough context to explain the mistake. Selected examples will support the aggregate findings rather than replace them.

Naturally occurring muscle names are legitimate parts of an exercise description. Their presence alone will not establish target leakage.

If prediction scores are saved, we will identify whether they are probabilities, logits, or another model output and preserve their class ordering. We will not present these scores as calibrated confidence without a separate calibration assessment.

### 32.8 Document Ambiguous or Disputed Labels

Some apparent prediction errors may involve unclear descriptions, multi-muscle movements, or questionable source labels.

A disagreement between a model and the benchmark will not, by itself, establish that the benchmark is wrong.

We will preserve:

- The frozen target and its recorded origin.
- The source description.
- Predictions from all three models.
- The reason for questioning the label.
- Supporting evidence and any subsequent review decision.

Our primary results will retain the original frozen targets.

If a correction is supported by documented review, we may report a separate, versioned sensitivity analysis using the same saved predictions from all three models. The correction will be applied consistently across the comparison.

A proposed correction outside the frozen class list will require an explicit explanation. We will not silently remap it or remove the record.

The original benchmark results will remain available alongside any corrected analysis.

### 32.9 Prepare the Paired Analysis for Step 35

We will save an aligned prediction table containing one row per expected test record, with its exercise ID, group ID, frozen reference label, and predictions from all three models.

The saved records will identify the relevant model artifacts, protocol version, and diagnostic fields.

Step 35 will apply the uncertainty procedure fixed in Step 31:

- 2,000 paired bootstrap resamples.
- Random seed 42.
- Resampling by exercise group.
- Individual 95% percentile intervals.

Each sampled group will contribute all its records. When a group is sampled more than once, its records will contribute with that draw multiplicity.

All models will use the same sampled groups in each resample. Metrics will be recalculated across the resulting records, preserving the fixed class list and zero-division convention.

We will not substitute independent record sampling or an unweighted average of group-level F1 scores.

These intervals will describe uncertainty within our constructed benchmark. Small class support, related exercises, and limitations in the grouping method will remain part of the interpretation.

### 32.10 Explain What the Results Establish

We will report:

- How each learned classifier compares with the majority baseline.
- How the neural classifier compares with logistic regression.
- Which classes and input types remain difficult.
- Whether validation and test rankings differ.
- How source-label quality and available descriptions limit interpretation.

The classifier selected for the application in Step 21 will remain fixed for the current experiment. A different test ranking will be reported as a finding.

This assessment concerns body-part classification within the eligible exercise dataset. It does not establish basketball-drill suitability, appropriate training difficulty for a teenager, or the quality of a weekly plan.

Our discussion will distinguish successful execution from predictive performance. A technically complete evaluation can still reveal weak model results.

### 32.11 Preserve the Remaining Frozen Experiment

The planner experiment was frozen in Step 31.

Classifier test findings will therefore remain outside subsequent changes to the application’s selected model, catalog tags, retrieval collection, prompts, planning rules, and generation settings for this experiment.

If we later prepare catalog predictions for previously held-out records, those predictions will be saved separately from the reference collection used in the frozen planner comparison.

Any new training, revised labels, or application improvements will belong to a documented later development version.

A confirmed implementation defect will follow the protocol amendment process, with the affected results and prior test exposure clearly identified.

### 32.12 Expected Outputs

After completing this step, we expect to save:

- `results/classifier_test/evaluation_manifest.json`: protocol, source, split, environment, preprocessing, and model identities.

## 33. Final Comparison of the Four Planner Versions

We will compare our four frozen planner versions using the same held-out fictional athlete profiles.

We want to find out whether adding references, specialist roles, and evaluation with bounded revision improves the responses. We will also examine the additional runtime and computation required by each workflow.

The comparison will cover:

- Completion of feasible planning requests.
- Adherence to the documented planning requirements.
- Handling of missing information, conflicting constraints, and requests outside our scope.
- Retrieval performance and delivery of complete supporting evidence.
- Execution reliability and resource usage.
- Response quality on the sample prepared for human review.

### 33.7 Calculate Independent Automated Feasible-Plan Success

We will score each designated returned outcome using the external assessment procedure.

Version 4’s internal acceptance decision will remain separate from its research score. The external assessment will recompute the required findings from the returned response, frozen profile inputs, and adopted resources.

Verified calculation helpers may be shared between implementations, but internal pass flags will not replace external scoring.

The primary planner measure will remain:

**Automated feasible-plan success rate = successful feasible cases ÷ all 150 feasible profiles**

For a feasible case to succeed, it must:

- Produce a complete plan under the shared structure and completeness requirements.
- Account for the required profile information and existing commitments.
- Have full expected coverage of the required checks.
- Meet every required applicable condition.
- Include a valid reason for any required check marked not applicable.

Execution failures, missing plans, inappropriate refusals, invalid required structure, required violations, and required unknowns will prevent success.

An empty check list will not produce a pass. The expected checks will come from the profile and frozen assessment policy, rather than whichever response type or activities the model chooses to declare.

An external scoring-system error will be recorded as an assessment problem. We will resolve it through the protocol process before reporting a complete comparison.

All 150 feasible profiles will remain in each version’s primary denominator. Unstarted or unresolved jobs will keep the final assessment incomplete rather than receive invented outcomes.

| Measure | Population or interpretation |
|---|---|

| Required-check outcomes | Met, violated, unknown, and not applicable, with reasons and explicit counts |
| Check coverage | Expected checks that were actually assessed |
| Response-type match | Agreement between the declared type and the frozen scenario category |
| Parsing reliability | Valid and invalid outputs, with missing outputs identified separately |
| Execution reliability | Completed and failed jobs among attempted jobs, with unstarted work shown |
| Reference-ID validity | Supplied identifiers that resolve to the relevant frozen records |
| Resource usage | Recorded work across completed and failed attempts |

Unparseable outputs and missing responses will retain their actual statuses. We will not invent a response type for them.

A correctly named clarification can still ask the wrong question. We will therefore distinguish response-type match from semantic appropriateness.

Similarly, a valid reference identifier will not establish that the associated claim is supported. Optional identifiers may be absent under the shared schema without causing an automatic formatting failure.

We will report missing-information, conflict, and outside-scope results separately, together with descriptive age-group results and their actual denominators.

Human judgments about usefulness and evidence accuracy will remain limited to the responses reviewed in Step 34. Automated findings will describe only what the implemented checks establish.

We will distinguish:

- Logical model calls.
- Transport attempts.
- Searches and other supporting computation.
- Warm-up and setup activity.

Across all four versions, the planned logical-call total for 200 fully completed matched profiles is:

**1,800 + 2F + 2R**

Here, **F** is the number of profiles activating the food role and **R** is the total number of Version 4 revisions. With at most two revisions per profile, the maximum is **2,600 + 2F**, or at most **3,000 logical calls**.

## 34. Independent Human Review and Reviewer Agreement

### 34.1 Purpose

We will use human review to examine qualities that automated checks cannot fully assess. These include whether a response addresses the request, fits the athlete’s situation, gives practical instructions, and accurately uses supporting evidence.

We will also measure how consistently our two reviewers apply the rubric.

Reviewer agreement will help us understand the reliability of our assessment. Agreement alone will not establish that a recommendation is correct or appropriate for real athletes.

This step will assess the responses preserved in Step 33. We will not generate replacement answers or use the reviews to revise the frozen planner versions.

### 34.2 Verify the Selected Review Sample

Each of the 1,600 completed entries will contain a numeric rating or a justified not-applicable decision.

Terminal generation failures will remain in the sample. A selected job that was never executed, or whose execution status remains unresolved, will remain unfinished experiment work rather than become an invented failure record.

### 34.3 Assign Reviewers and Preserve Independent Judgments

The two reviewers identified in the protocol will assess every response separately.

Both reviewers will be project teammates. We will record their relevant involvement in developing the system and describe the assessment as structured team review.

Independent review means that each person will complete their initial ratings without seeing the other person’s ratings or discussing individual responses.

We will provide separate rating files and keep the other reviewer’s judgments unavailable until both submissions have been checked and locked.

Reviewer practice and substantive clarification of the rubric will take place on development examples before the final protocol is frozen.

If final review exposes an unresolved scoring rule, we will document it through the amendment process. We will not improvise different interpretations for individual responses or planner versions.

The third teammate may manage the anonymous identity mapping and help adjudicate disagreements after the original ratings are locked.

### 34.4 Use the Anonymized Materials from Step 33

Each review record will contain:

- The original public profile and request.
- The preserved response, including its citations.
- The frozen review rubric.
- Equal, neutral access to the approved reference collection.

We will use the anonymous response identifiers and saved presentation order prepared in Step 33.

Reviewer materials will exclude:

- Planner-version and model labels.
- Private scenario labels and feasibility witnesses.
- Profile-family metadata.
- Internal specialist messages and evaluator findings.
- Automated scores and acceptance flags.

Anonymous labels will reduce direct identification of the versions, but they cannot guarantee complete blinding. Reviewers may recognize a workflow from its wording, structure, or references. Their prior involvement in the project may also affect their judgments.

### 34.5 Apply the Frozen Five-Dimension Rubric

We will use the rubric developed in Step 17 and finalized in Step 31.

The saved rubric will remain authoritative. Its anchors, applicability instructions, and development examples must match the version recorded in the protocol.

| Dimension | 0 — Does not meet expectations | 1 — Partly meets expectations | 2 — Meets expectations |
|---|---|---|---|
| Response suitability | Gives the wrong kind of response or ignores the central issue. | Addresses the issue partly but misses an important requirement. | Provides an appropriate plan, clarification, conflict explanation, or scope boundary. |
| Profile fit | Ignores or contradicts important profile information. | Uses some details but misses relevant goals or constraints. | Reflects the relevant goals, experience, schedule, and resources. |
| Practicality and completeness | Is unusable or substantially incomplete for the request. | Is partly usable but leaves important gaps. | Provides understandable and sufficiently complete guidance for the request. |
| Evidence accuracy | Contains materially unsupported or misleading claims. | Has partial support but includes important omissions or missing qualifications. | Agrees with the reviewed evidence and communicates relevant uncertainty. |
| Clarity | Contains confusing or inconsistent instructions. | Is understandable but contains noticeable ambiguity. | Provides clear, consistent instructions and explanations. |

Reviewers will judge the response against what the original request needs.

For example, an appropriate clarification should identify the essential missing information. It does not need to invent a complete weekly plan to receive a good completeness rating.

A feasible request that receives an inappropriate refusal will still be assessed against the requirements of that request. The model’s declared response type will not determine which weaknesses reviewers can assess.

We will retain the five dimensions separately. We will not introduce an overall quality score, reviewer average, or passing threshold after seeing the final responses.

### 34.6 Separate Numeric Ratings, Non-Applicability, and Missing Work

Every dimension entry will have an explicit status.

| Status | Meaning | Required entry |
|---|---|---|
| Numeric rating | The dimension applies and has been assessed. | A score of 0, 1, or 2 |
| Not applicable | The dimension does not apply under the frozen rubric. | No numeric score, with a reason |
| Missing rating | The reviewer has not completed the entry. | Remains unfinished until resolved |

A numeric zero will represent an applicable failure. It will not represent missing review work.

A missing model response and an unfinished reviewer rating are different situations. Reviewers must still complete the required assessment of a failed run.

For a feasible request receiving no usable response, the frozen failure rules will assign zero for response suitability and practicality and completeness. The remaining dimensions will follow their recorded applicability instructions.

We will not automatically assign five zeros to every missing or malformed response. Evidence accuracy, for example, may be not applicable when there are no substantive claims to assess.

If malformed response text exists, reviewers will examine its actual content rather than treat formatting failure as proof that every dimension is equally poor.

Applicability disagreements between reviewers will remain valid independent judgments. They will not be treated as incomplete entries simply because the reviewers disagree.

### 34.7 Check Claims and References Consistently

Both reviewers will have access to the same approved references, including when assessing Version 1, which had no reference access during generation.

Reviewers will distinguish:

1. Whether a cited record exists.
2. Whether it supports the associated claim.
3. Whether important source-dependent claims lack required support or attribution.

A valid identifier will not automatically establish evidence accuracy. Missing citations alone will not prove that a factual statement is wrong.

For claims requiring attribution, reviewers will record missing or misleading attribution according to the frozen rubric. A response without citations will not receive automatic credit simply because it contains no invalid identifiers.

Where relevant, reviewers will check:

- Age and grade applicability.
- Participation definitions and source qualifications.
- Units, reference quantities, and portion calculations.
- Activity instructions and resource requirements.
- Whether the source actually prescribed a stated duration or quantity.
- Whether uncertainty or unavailable information is communicated accurately.

We will distinguish a planner’s proposed scheduling choice from a recommendation explicitly stated by a source.

Reviewers will record important contradictions, unsupported assumptions, and evidence problems even when automated checks did not identify them.

### 34.8 Validate and Lock the Original Ratings

We will save one record for every combination of:

**Anonymous response ID × reviewer ID × dimension**

Each record will include:

- Rating status and numeric score, where applicable.
- A reason for a not-applicable decision.
- Brief comments explaining important concerns.
- Relevant source identifiers or locations.
- Rubric version and submission timestamp.

Before locking the ratings, we will check:

- All 160 response records are assigned to both reviewers.
- All 1,600 expected dimension entries are accounted for.
- No rating keys are duplicated.
- Numeric scores belong to the permitted scale.
- Not-applicable entries contain reasons.
- Numeric values and rating statuses agree.
- Rubric and response versions match the assignments.

Missing or invalid entries will return to the original reviewer while version labels and the other reviewer’s ratings remain hidden.

We will keep a correction history and use validation to resolve missing or invalid entries. Valid disagreements will remain in the original ratings.

Once both submissions are complete, we will preserve the original independent ratings with their file hashes and lock records before discussion or unblinding.

### 34.9 Calculate Agreement Separately for Each Dimension

Agreement will use the original independent ratings.

For each dimension, we will identify responses where both reviewers supplied numeric scores and report:

- The number of jointly numeric pairs.
- Each reviewer’s score distribution.
- The number and percentage of exactly matching ratings.
- Linearly weighted Cohen’s kappa.
- The numeric rating-pair count table.

Exact agreement will be calculated as:

**Matching numeric pairs ÷ all jointly numeric pairs**

For weighted kappa, we will retain the fixed label order `[0, 1, 2]` and use linear weighting. A two-point disagreement will receive a larger disagreement penalty than a one-point difference.

The label scale will remain fixed even when one score does not appear in the observed ratings.

We will report applicability separately, including:

- Both reviewers providing numeric ratings.
- Both selecting not applicable.
- Only the first reviewer selecting not applicable.
- Only the second reviewer selecting not applicable.
- Any unresolved missing entries.

Not applicable will not become a fourth ordinal score. Applicability disagreements will remain visible alongside numeric agreement.

If kappa is undefined, we will report it as unavailable with an explanation. Identical constant ratings, for example, can produce 100% exact agreement while leaving kappa undefined.

We will interpret agreement alongside sample counts, rating distributions, and comments. We will not add a new agreement threshold or combine the five dimensions into one kappa value.

### 34.10 Discuss Disagreements and Preserve the Original Evidence

After the independent ratings are locked, we may compare judgments and discuss disagreements.

The third teammate may help adjudicate disputed ratings or applicability decisions using the same rubric and approved evidence.

Any adjudicated assessment will be saved separately and linked to the original ratings. It will identify the decision, supporting reason, participants, and date.

Agreement statistics will continue to use the original independent ratings.

We will clearly distinguish findings based on original ratings from findings using later adjudication. A consensus decision will not overwrite evidence of disagreement.

Possible source, benchmark, or rubric errors will follow the evaluation amendment process. The original experiment and review records will remain available.

### 34.11 Prepare the Analysis Handoff and Describe the Limits

Step 35 will use those links for matched comparisons. Reviewer-facing materials will retain their anonymous identifiers.

- Ratings by dimension and reviewer.

Missing reviewer entries will keep the review incomplete. An undefined kappa with a documented reason will remain a valid reported statistical result.

The summary will keep independent-review completion, agreement calculation, and optional adjudication as separate statuses.

We will add actual observations after the reviews are completed. Step 35 will bring these findings together with the automated results, paired comparisons, uncertainty estimates, and resource measurements.



## 35. Statistical Comparisons, Uncertainty, and Failure Analysis

### 35.1 Purpose

We will bring together the saved classifier, retrieval, planner, and human-review results.

Our analysis will examine how performance changes between versions, how sensitive those differences are to the cases in our benchmark, and where each approach fails. We will also consider the runtime and computation required to produce the responses.

We will follow the evaluation protocol and analysis configuration frozen in Step 31. Primary measures, comparison directions, denominators, and uncertainty methods will remain unchanged after final results are available.

This step will analyze existing records. It will not retrain models or generate replacement responses.

### 35.2 Verify the Analysis Inputs

We will load the saved evidence from Steps 32–34 and verify its identity, completeness, and relationships.

| Input | What we will verify |
|---|---|
| Classifier results | Frozen test IDs, targets, exercise groups, predictions, class mapping, and model artifacts |
| Retrieval results | Thirty held-out questions, frozen judgments, ranked evidence units, delivered content, and execution statuses |
| Planner results | All 800 scheduled jobs, designated outcomes, independent checks, profile families, and protocol versions |
| Human review | Locked original ratings, reviewer assignments, applicability statuses, rubric version, and separate adjudication records |
| Runtime records | Stage timings, model calls, transport attempts, reference preparation, and failed work |
| Analysis configuration | Metrics, comparison directions, populations, resampling settings, and numerical conventions |

We will align records using their identifiers rather than rely on file order.

Attempted generation failures will remain observed outcomes. Unstarted jobs, unresolved executions, missing predictions, and incomplete external scoring will remain unfinished evaluation work.

We will not invent outcomes or restrict the analysis to whichever records happen to be available.

If a component remains incomplete, we will identify which analyses can be reproduced and which conclusions remain pending. An external scoring defect will remain an assessment problem rather than become a model failure.

### 35.3 Keep Measures and Populations Separate

Each result will identify its analysis population and denominator.

| Component | Main reporting population and measure |
|---|---|
| Exercise classification | Frozen eligible test records; macro-F1, accuracy, and class-level results |
| Retrieval ranking | Twenty answerable held-out questions; the recorded ranking measures |
| Unsupported retrieval | Ten unsupported questions; returned-candidate, no-match, and error behavior |
| Feasible planning | All 150 feasible profiles; automated feasible-plan success |
| Challenge handling | Twenty missing-information, fifteen conflict, and fifteen outside-scope profiles, reported separately |
| Human review | Forty selected profiles and all four outcomes, assessed by both reviewers |
| Runtime and resources | The explicitly identified completed, failed, attempted, or scheduled jobs relevant to each summary |

The primary planner success rate will retain all 150 feasible profiles for each version.

Missing responses, inappropriate refusals, required unknowns, and attempted workflow failures will remain unsuccessful feasible outcomes. Version 4’s internal acceptance will not replace the independent result.

We will preserve met, violated, unknown, and justified not-applicable check outcomes. Missing plans and empty check lists will not become successful cases.

Every reported rate will show its numerator and denominator.

### 35.4 Compare Planner Versions on Matched Profiles

We will use the four comparisons fixed in the protocol.

| Comparison | Implemented change being examined |
|---|---|
| Version 2 minus Version 1 | Addition of reference preparation, structured lookups, and reference-use instructions |
| Version 3 minus Version 2 | Organization through specialist contributions and coordination |
| Version 4 minus Version 3 | Addition of coded checks, model evaluation, and bounded revision |
| Version 4 minus Version 1 | Difference between the complete workflow and the language-model-only baseline |

We will compare outcomes from the same profiles.

For each feasible-plan comparison, we will count:

- Profiles where both versions succeed.
- Profiles where only the later version succeeds.
- Profiles where only the earlier version succeeds.
- Profiles where neither succeeds.

Success-rate differences will be reported in percentage points.

For the original 150-profile population:

**Difference in percentage points = 100 × (later-only successes − earlier-only successes) ÷ 150**

These counts will show whether a later workflow fixes earlier failures, introduces new failures, or does both.

The workflows also differ in calls, tokens, and supporting computation. We will interpret their differences as comparisons of complete implementations rather than isolated causal effects.

### 35.5 Apply the Frozen Paired Group Bootstrap

We will use the uncertainty procedure recorded in Step 31:

- 2,000 bootstrap resamples.
- Random seed 42.
- Paired resampling by the appropriate group.
- Individual 95% percentile intervals.

In plain language, we will repeatedly sample groups with replacement and recalculate the results. This will show how the observed differences vary when the benchmark contains a different mix of its existing groups.

The implementation will follow these steps.

1. **Define the analysis population.** For primary planner success, this is the 150 feasible profiles. For classifier analysis, it is the frozen eligible test records.
2. **Identify the groups within that population.** Planner records will use profile families; classifier records will use exercise groups.
3. **Sample groups with replacement.** Each resample will draw the same number of groups as observed in that population.
4. **Include every eligible member of each sampled group.** If a group is drawn twice, its records will contribute twice.
5. **Use identical draws for matched models or versions.** We will calculate their metrics and differences on the same resampled records.
6. **Save the resampled estimates and calculate interval endpoints.** The endpoints will be the recorded 2.5th and 97.5th percentiles.

For feasible-plan analysis, we will first select the 150 feasible profiles and then group them by family. A family containing challenge profiles will contribute only its feasible members.

Metrics will remain at their original level. Planner success will be calculated across sampled profiles, and classifier metrics across sampled exercise records.

When group sizes differ, the number of records in a resample can vary. Each resampled rate will therefore use its actual resampled denominator.

We will preserve repeated contributions rather than deduplicate IDs inside a resample. We will also avoid replacing record-level metrics with equally weighted averages of family success rates or group-level F1 scores.

The saved analysis configuration will identify group ordering, random-number generation, and percentile conventions. We will preserve the actual group-index draws so the calculation can be reproduced.

Ordinary paired row resampling does not implement this group procedure. We will explicitly implement the recorded sampling unit and percentile method rather than rely on software defaults.

### 35.6 Interpret Uncertainty Within Its Limits

We will report observed differences alongside their intervals.

These intervals describe resampling variation within our constructed benchmark and fixed saved outputs. They do not capture every source of uncertainty, including:

- Retraining on different data.
- Changing training or generation seeds.
- Using another model or serving environment.
- Differences between fictional profiles and real athletes.
- Errors or omissions in reference labels and scoring rules.

We will report the number of distinct families or exercise groups supporting each analysis. Grouping reduces known dependence but does not prove statistical independence.

The intervals will be individual intervals, rather than a simultaneous guarantee across all comparisons. We will focus on effect sizes and uncertainty without formal significance claims.

A constant bootstrap distribution will not establish that real-world uncertainty is zero. Undefined or nonfinite calculations will be investigated and reported with reasons.

We will not discard inconvenient resamples, change methods to obtain narrower intervals, or introduce new stratification after inspecting outcomes.

Any required correction will follow the protocol amendment process.

### 35.7 Analyze the Classifier Results

We will compare the fitted majority baseline, selected logistic regression pipeline, and selected neural configuration’s seed-42 checkpoint using their saved predictions.

Macro-F1 will remain primary. Accuracy, class-level precision, recall, F1, record support, distinct-group support, and confusion patterns will provide additional context.

Comparison directions will follow the frozen analysis configuration, covering the learned models against the majority baseline and the neural classifier against logistic regression.

We will label macro-F1 differences on their original 0–1 scale and keep that scale consistent in tables and figures.

Every bootstrap resample will retain the complete frozen class list and `zero_division=0` convention, including when a class is absent from that resample.

This differs from an unexpectedly absent class in the original test partition, which is a contract issue requiring investigation.

We will distinguish test-sample uncertainty from the training-seed variation recorded in Step 21. Resampling saved predictions will not measure retraining variability.

The application classifier selected through validation will remain fixed. A different test ranking will become a reported finding.

Any label-correction sensitivity analysis will remain separate from the primary results and use the documented corrected targets consistently across all models.

### 35.8 Summarize Retrieval and Challenge Handling

We will summarize the development-selected retriever’s held-out results from Step 33.

Ranking measures will use the twenty answerable questions. Executed failures and no-match outcomes will remain in the declared ranking denominators.

The ten unsupported questions will retain separate return, no-match, and error summaries. They will not receive invented perfect recall.

Complete-evidence coverage will concern delivered content and required qualifications. A matching evidence identifier alone will not establish that sufficient information reached the model.

The single-query retrieval benchmark will remain distinct from the planner’s two-query merged reference packet.

For planner challenge cases, we will separate:

- Declared response-type match.
- Relevant automated findings.
- Human judgments of appropriateness on the reviewed sample.

A response can use the expected label while still providing an irrelevant clarification or inadequate conflict explanation.

We will not introduce new retrieval uncertainty procedures or hypothesis tests unless they were already specified in the frozen protocol.

### 35.9 Analyze Human Ratings and Subgroups

We will use the locked original independent ratings from Step 34.

For each dimension and reviewer, we will report the distribution of scores 0, 1, and 2, together with numeric, not-applicable, and missing-entry counts.

The exact-agreement and linearly weighted kappa results will retain their original calculation rules and applicable denominators.

For paired version comparisons, we will match ratings from the same profile, reviewer, and dimension. Among jointly numeric pairs, we will report whether the later version receives a higher, equal, or lower score.

We will show applicability disagreements and the number of available pairs rather than silently exclude them from interpretation.

Original ratings and later adjudication will remain separate. We will not average reviewers, combine dimensions into a new quality score, or introduce a new passing threshold.

Age and scenario summaries will include actual profile, response, and family counts. Newly discovered subgroups will be labeled exploratory.

The human sample overrepresents challenge requests, so its pooled distributions will not estimate the full benchmark’s overall response quality.

Four responses for one profile and two ratings of one response will remain linked observations. They will not become additional independent athletes.

### 35.10 Examine Runtime and Resource Tradeoffs

We will compare measured performance alongside the work required by each workflow.

Our summaries will include:

- End-to-end request time and stage timings.
- Logical model calls and transport attempts.
- Reference preparation and structured lookups.
- Input and output tokens where available.
- Version 4 revision counts.
- Failed work and permitted recovery.
- Available hardware and resource measurements.

We will preserve the setup, caching, and timing boundaries frozen in Step 31. Independent research scoring will remain outside planner request latency.

Completed-response timing will identify its included jobs. Where we show matched timing differences, we will identify the matched population and report excluded failures separately.

Failure time will remain visible so that fast failure does not appear to be efficient successful service.

Unavailable measurements will remain unavailable. Dollar estimates will require a documented calculation basis; local computation will not automatically receive a zero-dollar cost.

We will discuss whether observed improvements appear worthwhile for our prototype without creating a new weighted quality–cost score.

### 35.11 Investigate Failures and Regressions

We will begin with the diagnostic categories defined in the protocol.

Our review will distinguish:

- Execution, timeout, and resource failures.
- Invalid or incomplete outputs.
- Inappropriate response choices.
- Missed profile constraints or existing commitments.
- Required-check violations and unknowns.
- Missing, irrelevant, or incomplete retrieved evidence.
- Unsupported claims and misleading citations.
- Incorrect units or reference-data handling.
- Problems missed or introduced during Version 4 revisions.
- Exercise-classification errors.

Each finding will link to saved case identifiers, responses, check results, and reviewer evidence where available.

A case may receive several diagnostic labels. We will show the denominator and explain that overlapping category percentages can add to more than 100%.

We will distinguish observed symptoms from confirmed causes. For example, an unsupported recommendation does not automatically establish that retrieval was the cause.

Saved revision histories may help explain Version 4’s behavior, but its designated returned outcome will remain authoritative. We will not select an earlier draft because it receives a better retrospective assessment.

Human findings about evidence accuracy will apply to the responses actually reviewed. They will not automatically extend to all 800 scheduled outcomes.

### 35.12 Save Traceable Analysis Outputs

| Output | Purpose |

We will summarize the interpretation in `docs/statistical_analysis_findings.md`.

## 36. Reproducibility and Clean Notebook Execution

### 36.1 Purpose

We will check that another teammate can set up the project, run the documented workflow from a fresh kernel, and reproduce our analysis using the saved experimental records.

We will distinguish three activities:

- Recalculating results from existing evidence.
- Rebuilding data or repeating model training.
- Generating a new set of planner responses.

These activities require different inputs and resources. Our notebook and documentation will make clear which activity is being performed and what its successful completion establishes.

The default workflow will reproduce the analysis while preserving the original experiment.

### 36.2 Define Execution Modes Before Project Operations Begin

| Mode | Purpose | Main behavior |

Ranking measures will use the twenty answerable questions. The ten unsupported questions will retain their separate return, no-match, and error summaries.

We will distinguish three findings:

1. Whether relevant evidence units appear in the ranked results.
2. Whether the delivered context contains complete supporting content and qualifications.
3. Whether the generated response uses that evidence accurately.

A correct evidence identifier will not establish complete context coverage. Complete context will not establish that the model interpreted it correctly.

We will also distinguish the single-query retrieval benchmark from the planner’s two-query reference preparation.

We will describe a link between evidence preparation and a planning problem when the saved records support it. Other explanations will be identified as interpretations.

Source coverage, permission limits, and unresolved review decisions will remain visible in the discussion.

### 37.5 Present the Four Planner Results

For each version, we will report automated feasible-plan success as a count and percentage out of all 150 feasible profiles.

Attempted failures, missing plans, inappropriate refusals, and required unknowns will remain unsuccessful outcomes.

Unstarted jobs or unresolved assessment problems will keep the affected comparison incomplete. They will not disappear from the account of the experiment.

We will report the four planned paired comparisons:

- Version 2 minus Version 1.
- Version 3 minus Version 2.
- Version 4 minus Version 3.
- Version 4 minus Version 1.

Each comparison will include the observed difference in percentage points, its individual uncertainty interval, and the paired success counts from Step 35.

We will discuss improvements, regressions, and uncertain differences. A later version will not be assumed to perform better because its workflow is more complex.

The twenty missing-information, fifteen conflict, and fifteen outside-scope profiles will be reported separately.

We will explain the difference between matching an expected response type and providing an appropriate explanation. Version 4’s internal acceptance will also remain separate from independent research success.

The 800 scheduled jobs will be described as matched outcomes from 200 fictional profiles, with their family relationships preserved in analysis.

### 37.6 Present Human Review and Application Findings

We will report human ratings separately for each dimension and reviewer:

- Response suitability.
- Profile fit.
- Practicality and completeness.
- Evidence accuracy.
- Clarity.

The summary will show rating distributions, applicable counts, not-applicable decisions, exact agreement, and weighted kappa where defined.

Original independent ratings and later adjudication will remain separate.

Human findings will cover forty selected profiles and 160 response records assessed by two teammates. We will explain the sample’s greater proportion of challenge cases and the limits of team-based review.

Reviewer comments and source checks will help explain the ratings. Agreement will describe consistency rather than establish correctness.

For the application, we will summarize verified behavior such as:

- Saving and updating profile information.
- Generating and displaying a response.
- Preserving unresolved issues and reference information.
- Saving edits and updating review status.
- Running the supported recheck actions.
- Tracking session completion.
- Exporting and restoring supported records.

Each application claim will identify whether its evidence comes from a software test, manual interface check, or genuine backend demonstration.

A working interface will establish the tested behavior. It will not establish usability, sustained participation, or improved athletic outcomes.

### 37.7 Explain Runtime and Practical Tradeoffs

We will place performance findings alongside actual final-experiment measurements.

These will include request time, logical model calls, transport attempts, token usage, reference preparation, revisions, and failed work.

Pilot projections will remain distinguishable from final observed usage.

Completed-response timing will show its included population. Failures and their resource use will remain visible.

We will interpret the implemented comparisons accurately:

- Version 2 adds reference preparation, structured lookups, and reference-use instructions together.
- Version 3 changes the organization of generation and uses additional calls.
- Version 4 adds checking, evaluation, and revision with further computation.

The experiment will compare these complete workflows. It will not isolate every contribution of retrieval, role assignment, or extra computation.

We will discuss the size and practical meaning of observed improvements. Small differences will not automatically establish equivalence, and a ranking alone will not explain whether additional work is worthwhile.

Dollar claims will require a documented calculation basis. We will retain unavailable measurements without replacing them with zeros.

### 37.8 Explain Limitations and Their Consequences

We will connect each limitation to the conclusions it restricts.

| Limitation | Consequence for interpretation |
|---|---|
| Fictional athlete profiles | We measure responses to designed cases rather than real participation or athletic outcomes |
| Related profiles and exercise records | Results contain dependence that grouping can address only within the recorded structure |
| Imperfect benchmark labels and expectations | Some apparent errors may involve source ambiguity or assessment mistakes |
| Limited or uneven exercise-class support | Some class results may be unstable or have limited coverage |
| Restricted source coverage | Missing guidance and incomplete metadata can limit retrieval and planning |
| Pending review or adoption decisions | Prepared resources may remain unavailable for operational use |
| Fixed food-reference subset | Lookups cover the selected foods, nutrients, and supported quantities |
| Small human-review sample | Semantic quality findings apply to the reviewed responses |
| Two project-team reviewers | Ratings do not represent a wider population of users or external experts |
| Partial blinding | Wording and references may reveal clues about the workflow |
| Different workflow budgets | Performance differences also involve different amounts of computation |
| Fixed model and serving configuration | Findings may change with other artifacts, settings, or hardware |
| Defined automated checks | Passing establishes the recorded contract rather than complete recommendation quality |
| Limited repeated training or generation | Saved-output uncertainty does not measure every source of model variability |

We will also report unresolved provenance, permission, implementation, and reproducibility issues that affect the findings.

### 39.11 Completion Criteria

We will consider final delivery complete when:

- The presentation explains the problem, implemented methods, findings, and limitations clearly.
- The demonstration accurately shows tested application behavior.
- Required files are consistent with the verified package.
- Reviewers can access the submitted materials.
- The correct submission has been confirmed as received.
- Required team and individual delivery tasks are accounted for.
- The submitted snapshot and confirmation records are preserved.

Any incomplete feature, evaluation component, or required delivery task will remain explicitly identified.

After these actions are performed, we will add the actual presentation, demonstration, and submission observations.

Our closing message will explain what we built, what the evidence shows, where the prototype falls short, and what we would investigate next.
