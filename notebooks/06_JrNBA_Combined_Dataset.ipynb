```python
from google.colab import drive
from pathlib import Path
import pandas as pd

# Mount Google Drive
drive.mount("/content/gdrive")

# Define project directory
PROJECT_DIR = Path(
    "/content/gdrive/MyDrive/ChampionsAthleticMap"
)

DATA_DIR = PROJECT_DIR / "data" / "processed"

# Load both datasets
practice1_df = pd.read_csv(
    DATA_DIR / "jr_nba_starter_practice1.csv"
)

practice2_df = pd.read_csv(
    DATA_DIR / "jr_nba_starter_practice2.csv"
)

# Combine the datasets
combined_df = pd.concat(
    [practice1_df, practice2_df],
    ignore_index=True
)

print("Practice Plan 1:", practice1_df.shape)
print("Practice Plan 2:", practice2_df.shape)
print("Combined dataset:", combined_df.shape)

display(combined_df[[
    "drill_id",
    "drill_name",
    "source_document"
]])
```

```python

# Check missing values in the combined dataset
print("Missing values per column:")
print(combined_df.isna().sum())

# Examine solo adaptation information
display(combined_df[[
    "drill_id",
    "drill_name",
    "solo_adaptation_possible",
    "solo_adaptation_note"
]].head(15))
```

```python
# Find drills with repeated names
duplicate_drills = combined_df[
    combined_df.duplicated(
        subset=["drill_name"],
        keep=False
    )
]

# Display repeated drills and their sources
display(duplicate_drills[[
    "drill_id",
    "drill_name",
    "source_document",
    "source_page",
    "instructions_summary"
]])

print("Repeated records:", len(duplicate_drills))
print("Unique drill names:", combined_df["drill_name"].nunique())
```

```python

# Compare both versions of the repeated drill
duplicate_comparison = combined_df[
    combined_df["drill_name"] == "3 on 3 Numbers Game"
]

# Show all fields without truncating long text
with pd.option_context("display.max_colwidth", None):
    display(duplicate_comparison.T)
```

```python

print("Total drills:", len(combined_df))
print("Unique drill IDs:", combined_df["drill_id"].nunique())
print("Unique drill names:", combined_df["drill_name"].nunique())
print("Missing values:", combined_df.isna().sum().sum())

assert len(combined_df) == 23
assert combined_df["drill_id"].is_unique
assert combined_df.isna().sum().sum() == 0

print("\nCombined dataset validation passed!")
```

```python

from pathlib import Path

# Define the output location
output_path = Path(
    "/content/gdrive/MyDrive/ChampionsAthleticMap/"
    "data/processed/jr_nba_drills_master.csv"
)

# Ensure the folder exists
output_path.parent.mkdir(parents=True, exist_ok=True)

# Save the validated combined dataset
combined_df.to_csv(output_path, index=False)

# Verify the saved file
print("Master dataset saved to:", output_path)
print("File exists:", output_path.exists())
print("Total drills saved:", len(combined_df))
```
