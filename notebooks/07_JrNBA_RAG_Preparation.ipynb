```python

# Store all drill documents
drill_documents = []

# Convert every drill into a text document
for _, drill in drills_df.iterrows():

    drill_text = f"""
Drill: {drill['drill_name']}
Skill Category: {drill['skill_category']}
Level: {drill['curriculum_level']}
Primary Goal: {drill['primary_goal']}
Duration: {drill['duration']}
Equipment: {drill['equipment']}
Group Requirement: {drill['group_requirement']}
Instructions: {drill['instructions_summary']}
Difficulty Modification: {drill['difficulty_modification']}
Solo Adaptation Possible: {drill['solo_adaptation_possible']}
Solo Adaptation: {drill['solo_adaptation_note']}
Source: {drill['source_document']}, page {drill['source_page']}
""".strip()

    drill_documents.append({
        "drill_id": drill["drill_id"],
        "text": drill_text
    })

print("Total RAG documents:", len(drill_documents))

# Display the first document
print("\nFirst document:\n")
print(drill_documents[0]["text"])
```

```python

import json
from pathlib import Path

# Define where to save the RAG documents
output_path = Path(
    "/content/gdrive/MyDrive/ChampionsAthleticMap/"
    "data/processed/jr_nba_rag_documents.json"
)

# Save the documents as JSON
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(drill_documents, f, indent=2, ensure_ascii=False)

# Verify the saved documents
with open(output_path, "r", encoding="utf-8") as f:
    saved_documents = json.load(f)

print("Documents saved:", len(saved_documents))
print("File exists:", output_path.exists())
print("First drill ID:", saved_documents[0]["drill_id"])
```
