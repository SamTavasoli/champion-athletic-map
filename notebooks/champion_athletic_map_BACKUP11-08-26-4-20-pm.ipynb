## 1. Project Overview and Business Problem

### 1.1 Problem and Intended Users

We will develop **Champions Athletic Map**, an AI-assisted weekly basketball planning prototype for athletes ages **13–17**, their parents, and coaches.

A practical training week must account for the athlete's goals, experience, school schedule, available equipment, and existing practices or games. It also requires clear activity instructions and time for recovery. Our project will bring these requirements together in a planning workflow that users can understand, review, and adjust.

We will use each athlete's age, school grade, basketball experience, development goals, availability, equipment, and existing commitments to prepare a weekly schedule. Parents and coaches will be able to review the proposed activities and adapt the plan to the athlete's circumstances.

### 1.2 Proposed Prototype and Data

The prototype will support the following activities:

| Component | What we will build |
|---|---|
| Athlete profile | A structured form for the athlete's goals, experience, resources, and weekly commitments. |
| Coaching references | A searchable collection of approved basketball activities and documented participation guidance. |
| Exercise organization | A text classifier that assigns body-part categories to exercise records. |
| Weekly planning | A workflow that proposes basketball skill practice, general conditioning, and recovery periods, or requests clarification when essential information is missing. |
| Plan review and progress | Editable plans, checks after edits, and a checklist for recording completed activities. |
| Optional food information | A lookup for general food and nutrient facts from verified USDA records. |

Our data preparation will draw on the [Gym Exercise Dataset](https://www.kaggle.com/datasets/niharika41298/gym-exercise-data), the [Jr. NBA/Jr. WNBA instructional curriculum](https://jr.nba.com/jr-nba-instructional-curriculum/), the [NBA/USA Basketball Youth Guidelines](https://youthguidelines.nba.com/), and [USDA FoodData Central](https://fdc.nal.usda.gov/). We will document their sources, quality issues, and permitted uses before incorporating them into the application.

The exercise classifier will learn from exercise names and available descriptions. We will use its predictions as organizational tags. Reviewed activity requirements and planning rules will govern whether an activity can be selected for a particular profile.

### 1.3 Research Approach

We will investigate how retrieval, specialist agents, and automated checks affect the quality of generated weekly plans. Our experiment will compare four configurations:

1. A language model alone.
2. The language model with retrieval-augmented generation (RAG).
3. RAG with specialist agents.
4. Specialist agents with an evaluator and automated checks.

We will use the same underlying language model and final test profiles across the four configurations. Evaluation will examine constraint adherence, completeness, source support, handling of challenging requests, and computational requirements. Independent human review will complement the automated measures.

Alongside this comparison, we will train and evaluate a neural exercise classifier against simpler baselines. We will keep the classifier's performance separate from the evaluation of the complete planning system.

### 1.4 Project Outcomes and Boundaries

Our intended outcomes are a working planning prototype and a reproducible comparison of the four planning configurations. The supporting work will include prepared datasets, a trained exercise classifier, documented code, evaluation records, a technical report, and a recorded presentation.

We will use fictional athlete profiles during development and evaluation. Our findings will concern system behavior on the defined tasks; we will not measure changes in real athletes' performance or health.

The prototype will support planning and discussion with parents and coaches. Medical diagnosis, injury rehabilitation, supplement recommendations, and individualized dietary prescriptions are outside the project scope. Optional nutrition functionality will remain limited to general food and nutrient information.

## 2. Research Questions, Scope, and Evaluation Plan

### 2.1 Research Question and Hypothesis

Our main research question is:

**How do retrieved coaching references, specialist agents, and automated checks affect the quality and computational requirements of personalized weekly basketball plans compared with a language model alone?**

We expect source access to improve supporting evidence, specialist roles to help organize planning, and automated checks to identify constraint violations. We will test these expectations using matched profiles and common evaluation criteria. Additional complexity may introduce errors or increase runtime without improving plan quality; we will report those findings as part of the comparison.

### 2.2 Users and Scope

We will focus on weekly basketball planning for athletes ages 13–17, with parents and coaches reviewing and adjusting the proposed activities. Planning inputs will include age, school grade, experience, goals, available time, equipment, training partners, and existing commitments.

The prototype will cover basketball skills, general conditioning, recovery considerations, editable plans, and progress recording. Optional nutrition functionality will provide general food facts. Medical diagnosis, rehabilitation, supplement recommendations, and individualized dietary prescriptions remain outside our scope.

We will use fictional profiles for development and evaluation. The exercise classifier will be evaluated separately as a body-part classification task.

### 2.3 Planner Comparison

We will compare four configurations:

| Version | Configuration | Main addition |
|---|---|---|
| 1 | Language model alone | Shared athlete profile and planning request |
| 2 | Language model with RAG and reference tools | Retrieved coaching information and structured reference access |
| 3 | RAG with specialist agents | Coordinated roles for different planning responsibilities |
| 4 | Specialist agents with an evaluator and automated checks | Checks and bounded revision before returning a response |

We will use the same underlying language model, common generation settings, and shared input and output requirements. Versions 2–4 will use the same approved reference collection and reference preparation, with matched evidence packets for the compared profiles.

Our planned comparisons are Version 2 against Version 1, Version 3 against Version 2, Version 4 against Version 3, and Version 4 against Version 1. These comparisons will describe the implemented workflows, including their differences in model calls and computation.

### 2.4 Development and Final Evaluation Profiles

We will prepare **300 fictional profiles**: **100 for development** and **200 for final evaluation**. Each age from 13 through 17 will have 20 development profiles and 40 final profiles. These are planned benchmark counts.

Development profiles will support prompt design, troubleshooting, rule development, and pilot testing. Related profiles and variations will stay within the same partition. Existing examples reused during development will remain outside the final test set, together with their related variations.

The final set will contain:

| Scenario | Profiles | Expected response |
|---|---:|---|
| Feasible planning request | 150 | A complete plan that respects the supplied constraints |
| Missing essential information | 20 | A request for the information needed to proceed |
| Conflicting constraints | 15 | An explanation of the conflict and an appropriate next step |
| Request outside scope | 15 | An explanation of the relevant boundary |
| **Total** | **200** | |

Every final profile will be processed by all four versions, producing **800 scheduled profile–version jobs**. These will remain paired outcomes from 200 constructed profiles. We will account for failed and unfinished jobs before reporting a completed comparison.

For human review, we will select **40 final profiles** before inspecting their outputs: 25 feasible cases and five from each challenge category. Their four version outcomes will produce **160 response records**, each assessed independently by two team reviewers. Review materials will conceal version labels, and the initial ratings will be preserved before discussing disagreements.

### 2.5 Measures and Evaluation Controls

Our primary planner measure will be **automated feasible-plan success rate**: successful feasible cases divided by all **150 feasible profiles per version** in the completed experiment. Success will require a usable response, the required planning information, and a pass on every predefined required check.

Execution failures, inappropriate refusals, unparseable responses, and unknown required checks will count as unsuccessful outcomes. Unexecuted jobs will indicate an incomplete experiment. The three challenge categories will be reported separately using their own denominators.

| Evaluation area | What we will measure |
|---|---|
| Planning constraints | Schedule, time, equipment, participant, and applicable rule checks |
| Completeness and response behavior | Required information and appropriate handling of challenging requests |
| Retrieval and source support | Relevant evidence, claim support, and missing or incorrect citations |
| Human review | Profile fit, practicality, clarity, evidence use, and reviewer agreement |
| Efficiency and reliability | Response time, model calls, token usage where available, retries, and failures |
| Exercise classification | Macro-F1 as the primary measure, with accuracy and per-class results against majority-class and TF-IDF logistic regression baselines |

We will finalize the profiles, required checks, scoring rubric, reference versions, and execution settings before final testing. Classifier preprocessing and model selection will use training and validation data; the held-out exercise test set will remain reserved for final evaluation.

Version 4's internal evaluator will remain separate from the common independent evaluation. Human reviewers will be project teammates, and their ratings will describe the selected synthetic cases. The study will assess system behavior and planning quality without establishing clinical safety or real-world athletic improvement.

## 3. Background Research and Supporting Evidence

### 3.1 Basketball Coaching Resources and Participation Guidance

The Jr. NBA/Jr. WNBA instructional curriculum provides 48 practice plans across four development levels: Rookie, Starter, All-Star, and MVP (Jr. NBA, n.d.). These materials offer source-linked activity descriptions that we can review when building our basketball catalog.

The NBA/USA Basketball Youth Guidelines distinguish recommended participation, maximum participation, and rest guidance. They also define organized basketball separately from unstructured individual or peer-led activity (National Basketball Association & USA Basketball, n.d.).

We will preserve source locations, curriculum labels, units, and qualifications when preparing catalog records and planning rules. Age-based and grade-based applicability will require explicit interpretation. We will keep curriculum levels separate from age eligibility and distinguish source-supported metadata from our own adaptations and scheduling decisions.

### 3.2 Retrieval-Augmented Generation

Lewis et al. (2020) combined a pretrained sequence-to-sequence model with information retrieved from an external document index. Their results on knowledge-intensive language tasks provide a foundation for studying whether reference access improves generated answers.

We will apply the retrieval-plus-generation pattern by supplying a fixed language model with relevant passages from our approved coaching and guideline collection. Retrieved records will retain their source identifiers so we can trace claims in the generated plans.

We will evaluate retrieval relevance and evidence use separately. Finding a relevant passage does not establish that a response interprets it correctly, and a valid citation does not establish support for every associated claim. Performance on our basketball-planning task will therefore require its own evaluation.

### 3.3 Tool Use, Specialist Roles, and Feedback

ReAct combines language-model reasoning with actions that obtain information or interact with an environment (Yao et al., 2023). It provides background for defining tools with clear inputs, outputs, and observable results.

We will use this idea to organize coaching retrieval, reference lookup, and planning checks. Our specialist configuration will divide responsibilities such as skill selection, scheduling, and recovery considerations into defined roles. The benefit of this coordination remains a research question; ReAct's findings do not establish the effectiveness of multiple specialists in our application.

Self-Refine illustrates an iterative process in which a language model generates an output, provides feedback, and revises it (Madaan et al., 2023). This motivates our study of bounded feedback and revision in Version 4. We will combine the planned evaluator with explicit automated checks and record the additional computation. Its internal feedback will remain separate from the independent evaluation used to compare versions.

### 3.4 Exercise Text Classification

Joulin et al. (2017) showed that compact text representations can support efficient supervised classification. Their fastText approach combines learned feature representations with a linear classifier, providing useful background for a small exercise-text model and meaningful baseline comparisons.

We will train a neural classifier from scratch using exercise names and available descriptions, and compare it with majority-class and TF-IDF logistic regression baselines. Model selection will use validation results, with macro-F1 as the primary measure. Related exercise records will remain within the same data partition.

The predicted body-part categories will help organize the exercise catalog. Activity eligibility and profile compatibility will be assessed through separately reviewed requirements and planning rules.

### 3.5 Food Composition References

USDA FoodData Central provides food and nutrient records through distinct data types, including Foundation Foods and SR Legacy (U.S. Department of Agriculture, Agricultural Research Service, n.d.). These sources have different releases and documentation requirements.

We will review the prepared Foundation Foods extract and any additional release selected for the food lookup. Preparation will preserve food identifiers, nutrient definitions, units, reference amounts, and release information. Records from older food tables will require verified provenance and relationships before reuse.

The optional lookup will provide general food facts. Individualized dietary prescriptions, supplement recommendations, and medical nutrition planning remain outside the project scope.

### 3.6 Relationship to Our Study

The literature and reference materials serve different purposes in our project:

| Evidence or reference | Role in our work |
|---|---|
| Jr. NBA curriculum | Source material for activity descriptions and practical requirements |
| NBA/USA Basketball guidelines | Participation definitions, planning constraints, and qualifications |
| RAG research | Background for supplying external evidence during generation |
| ReAct and Self-Refine | Background for tool use, feedback, and revision |
| Efficient text-classification research | Background for the exercise classifier and baseline comparison |
| USDA FoodData Central | Documented food composition information |

We will test how these components work together using the evaluation design in Step 2. Our findings will describe the implemented workflows, including their errors and computational tradeoffs. Published results from other tasks will inform our design without becoming assumed performance gains or evidence of real-world athletic benefit.

## 4. System Architecture and Data Flow

### 4.1 Architecture Overview

We will separate data preparation, model training, planning, application behavior, and research evaluation into components with clear interfaces. Shared Python functions will support both the notebook and the application.

The application will use Version 4 to generate and review plans. The experiment runner will call all four planner versions directly under the evaluation protocol.

| Component | Responsibility | Main output |
|---|---|---|
| Data preparation | Review source records, clean task-specific tables, and preserve provenance and decisions | Versioned datasets and review records |
| Exercise classifier | Train, select, save, and later load the exercise-text model | Predicted categories with their label origin |
| Reference tools | Prepare coaching evidence and access reviewed rules, exercise records, and optional food facts | Source-linked reference packets and structured results |
| Planner workflows | Coordinate the permitted model calls for Versions 1–4 | Returned response and execution records |
| Internal review | Apply Version 4's coded checks, evaluator, and bounded revisions | Check findings and workflow status |
| User application | Collect profiles, display responses, support edits, and record completion | Editable plan versions and progress records |
| Research evaluation | Assess saved outcomes using the common independent protocol | Comparable scores, review files, and resource measurements |

The following diagram shows the shared generation and response flow. Reference access applies to Versions 2–4; the application branch uses Version 4.



### 4.2 Data and Classifier Preparation

We will keep exercise records, coaching material, planning rules, food tables, fictional profiles, and evaluation records separate according to their purpose. The existing master file will serve as a heterogeneous record index. Training tables will be built from task-specific inputs with documented relationships.

For exercise classification, we will review labels and text, group related records, and create training, validation, and test partitions. Preprocessing and learned text features will be fitted on training data. We will compare majority-class and TF-IDF logistic regression baselines with a neural classifier, select settings using validation data, and reserve the test set for final evaluation.

We will save the selected model, preprocessing components, and label mapping. Application use will load these artifacts for inference. Original, reviewed, and predicted labels will remain distinguishable, and activity eligibility will depend on separate catalog reviews and planning rules.

The reference collection will contain approved material with stable source identifiers. Evaluation profiles, expected answers, reviewer notes, and generated plans will remain outside the retrieval index. Held-out exercise records will also remain outside the development reference collection under the classifier's evaluation boundary.

### 4.3 Planner Workflows and Reference Access

All versions will use the same base language model, common generation settings, shared profile representation, and response requirements.

| Version | Reference access | Generation workflow |
|---|---|---|
| 1 | No retrieval, structured lookup results, or classifier suggestions | One planner response |
| 2 | Approved retrieval and reference tools | One planner response using the prepared reference packet |
| 3 | The same reference preparation as Version 2 | Specialist contributions followed by one coordinator response |
| 4 | The same reference preparation and initial specialist workflow as Version 3 | Coded checks, evaluator feedback, and bounded coordinator revisions |

For each matched profile, Versions 2–4 will receive the same prepared reference packet, including retrieval and lookup results. We will retain its identifier and checksum. The initial design will reuse that packet across specialist and evaluator calls.

Mandatory rule applicability will come from the reviewed structured register. Food calculations will use verified values and units in code. Retrieval will supply explanatory evidence alongside these structured tools.

The shared input adapter will standardize field names and types while preserving missing information and conflicting commitments. Case labels, expected answers, and independent scoring feedback will remain outside the planner inputs.

### 4.4 Shared Responses and Application Editing

We will preserve raw model output and record parsing results separately. Parsing will check the agreed structure without silently repairing plan content.

| Response element | Information to preserve |
|---|---|
| Identity | Profile, run, planner version, and relevant artifact versions |
| Response type | Plan, clarification, constraint conflict, or outside-scope response |
| Plan content | Activities, days, allocated durations, equipment, participants, and relevant locations |
| Explanations and evidence | Activity descriptions, source references where available, and optional verified food facts |
| Questions and unresolved issues | Missing information, conflicts, and limitations requiring attention |
| Execution and review status | Parsing failures, timeouts, tool errors, and Version 4's internal review outcome |

Required fields will depend on the response type. Clarification and boundary responses will not require a complete weekly schedule. Execution failures will be recorded separately from valid response types. Source identifiers may be absent where unavailable, including in Version 1.

The application will preserve the original returned Version 4 response before creating an editable copy. Each substantive edit will create a new version, rerun relevant coded checks, and mark the earlier evaluator assessment as outdated. A separate review action will assess the edited plan without silently rewriting the user's changes.

Checklist entries will record reported completion. Edited plans and progress records will remain separate from benchmark responses and training data.

### 4.5 Independent Evaluation

The experiment runner will process the final profiles through each version, save the returned outcomes before user edits, and apply the frozen evaluation protocol. It will preserve unsuccessful and unfinished jobs so that reported results can be reconciled with the run manifest.

Version 4's internal acceptance decision will be recorded separately from independent performance. Research scores will not trigger corrective model calls in Versions 1–3. Human review will use the predefined response sample and concealed version labels described in Step 2.

Classifier, retrieval, planner, and human-review results will remain separate measures. We will compare complete workflows and report their model calls and resource use alongside output quality.

### 4.6 Traceability and Failure Handling

Run records will retain profile IDs, model and prompt versions, reference-packet IDs, tool results, raw and parsed responses, specialist contributions, revision history, timing, and usage where available.

We will define model-call limits, timeouts, transport retries, and revision budgets before final evaluation. State will be reset between independent cases. Failed calls and parsing problems will remain explicit, and any permitted attempt will count toward recorded resource use.

Source refresh, data preparation, classifier training, and final experiment execution will be selected explicitly. Routine result reproduction will use saved, verified artifacts. This separation will let us update a component while preserving the records needed to explain and reproduce the reported experiment.

```python
# Step 7 - PyTorch GPU Verification

import torch
import time

print("=" * 60)
print("CHAMPIONS ATHLETIC MAP - PYTORCH GPU VERIFICATION")
print("=" * 60)

print("\n[1] PYTORCH INFORMATION")
print("PyTorch version:", torch.__version__)
print("Compiled CUDA version:", torch.version.cuda)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():

    device = torch.device("cuda:0")
    properties = torch.cuda.get_device_properties(device)

    print("\n[2] GPU INFORMATION")
    print("GPU:", properties.name)
    print(
        "GPU memory (GB):",
        round(properties.total_memory / (1024**3), 2)
    )

    print("\n[3] GPU COMPUTATION TEST")

    torch.manual_seed(42)

    a = torch.randn(1024, 1024, device=device)
    b = torch.randn(1024, 1024, device=device)

    start = time.perf_counter()

    result = torch.matmul(a, b)

    torch.cuda.synchronize()
    elapsed = time.perf_counter() - start

    print("Matrix multiplication:", result.shape)
    print("Output device:", result.device)
    print("Execution time (seconds):", round(elapsed, 4))
    print("Result contains finite values:",
          torch.isfinite(result).all().item())

    del a, b, result
    torch.cuda.empty_cache()

    print("\nGPU COMPUTATION TEST: COMPLETED")

else:
    print("\nCUDA is not available.")
    print("GPU configuration needs further investigation.")

print("\n" + "=" * 60)
print("PYTORCH VERIFICATION COMPLETE")
print("=" * 60)
```

```python
# Step 7 - Local Llama 3.1 Inference Feasibility Test

import json
import time
from urllib.request import Request, urlopen

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - LOCAL LLM FEASIBILITY TEST")
print("=" * 65)

model_name = "llama3.1:8b"

prompt = """
We are testing an AI basketball planning prototype.

Fictional athlete:
- Age: 15
- Sport: Basketball
- Goal: Improve ball handling
- Available days: 3 per week
- Time: 30 minutes per session
- Equipment: Basketball only

Create three simple basketball practice themes.

Return valid JSON with exactly three keys:
day_1, day_2, day_3.

Each value should be a short description.
Do not include medical or nutrition advice.
"""

payload = {
    "model": model_name,
    "prompt": prompt,
    "stream": False,
    "format": "json",
    "options": {
        "temperature": 0,
        "num_predict": 180,
        "num_ctx": 4096
    }
}

request = Request(
    "http://127.0.0.1:11434/api/generate",
    data=json.dumps(payload).encode("utf-8"),
    headers={"Content-Type": "application/json"},
    method="POST"
)

print("\n[1] MODEL INFORMATION")
print("Model:", model_name)
print("Backend: Local Ollama")
print("Temperature: 0")
print("Context setting: 4096 tokens")

print("\n[2] GENERATING TEST RESPONSE")

start = time.perf_counter()

try:
    with urlopen(request, timeout=180) as response:
        result = json.loads(response.read().decode("utf-8"))

    elapsed = time.perf_counter() - start

    generated_text = result.get("response", "")

    print("\n[3] GENERATED RESPONSE")

    try:
        parsed_response = json.loads(generated_text)

        print(json.dumps(parsed_response, indent=2))

        expected_keys = {"day_1", "day_2", "day_3"}

        valid_structure = (
            isinstance(parsed_response, dict)
            and set(parsed_response.keys()) == expected_keys
            and all(
                isinstance(value, str) and value.strip()
                for value in parsed_response.values()
            )
        )

        print("\nValid expected structure:", valid_structure)

    except json.JSONDecodeError:
        print("Response was not valid JSON.")
        print(generated_text[:1000])
        valid_structure = False

    print("\n[4] PERFORMANCE METRICS")
    print("Wall-clock time (seconds):", round(elapsed, 2))

    if result.get("total_duration") is not None:
        print(
            "Ollama total duration (seconds):",
            round(result["total_duration"] / 1e9, 2)
        )

    print("Input tokens:", result.get("prompt_eval_count"))
    print("Generated tokens:", result.get("eval_count"))
    print("Generation completed:", result.get("done"))
    print("Structure verified:", valid_structure)

except Exception as error:
    print("Inference test failed.")
    print("Error type:", type(error).__name__)
    print("Details:", str(error)[:500])

print("\n" + "=" * 65)
print("LOCAL LLM FEASIBILITY TEST FINISHED")
print("=" * 65)
```

```python
# STEP 8 - Unified Source Manifest and Provenance Verification

import csv
import hashlib
import io
import json
import os
import shutil
import tempfile
from datetime import datetime
from pathlib import Path

now = datetime.now().astimezone()
checked_on = now.date().isoformat()

root = next(
    (
        p for p in (
            Path.cwd().resolve(),
            *Path.cwd().resolve().parents
        )
        if (p / ".git").exists()
    ),
    None
)

if root is None:
    raise FileNotFoundError(
        "Open this notebook inside the champion-athletic-map Git repository."
    )

package = (
    root / "Champions_Athletic_Agent_Ready_Dataset _Package"
)

audit_file = package / "data_quality_audit.csv"

if not audit_file.is_file():
    raise FileNotFoundError(
        f"Original cleaning audit not found: {audit_file}"
    )

FIELDS = [
    "source_id",
    "title",
    "asset_type",
    "intended_role",
    "publisher",
    "source_url",
    "version_or_release",
    "acquisition_date",
    "verification_date",
    "file_status",
    "source_filename",
    "relative_path",
    "file_size_bytes",
    "sha256",
    "reported_records",
    "record_count_basis",
    "parent_source_ids",
    "transformation_script",
    "source_commit",
    "license_or_terms",
    "permission_status",
    "limitations",
    "next_action"
]

ISSUE_FIELDS = [
    "issue_id",
    "priority",
    "issue",
    "next_action",
    "status"
]

OLD_COMMIT = "1076a81f3b4f71f69e38cfbc43575358652c611a"
ADITI_SNAPSHOT = "c1631ed735684e292563037f7c8ceca72a5766b7"
RECOVERY_COMMIT = "c42c9f3"


# --------------------------------------------------
# 1. Define File and Source Verification Helpers
# --------------------------------------------------

def read_csv(path):
    with path.open(
        newline="",
        encoding="utf-8-sig"
    ) as f:
        reader = csv.DictReader(f)

        if not reader.fieldnames:
            raise ValueError(
                f"CSV header missing: {path}"
            )

        return list(reader), list(reader.fieldnames)


def checksum(path):
    h = hashlib.sha256()

    with path.open("rb") as f:
        for part in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(part)

    return h.hexdigest()


def file_metadata(path):
    if not path.is_file():
        return {
            "file_status": "not_found_in_checked_locations",
            "relative_path": "",
            "file_size_bytes": "",
            "sha256": ""
        }

    return {
        "file_status": "present",
        "relative_path": path.relative_to(root).as_posix(),
        "file_size_bytes": path.stat().st_size,
        "sha256": checksum(path)
    }


def count_records(path):
    if path.suffix.lower() == ".csv":
        return len(read_csv(path)[0])

    if path.suffix.lower() == ".json":
        with path.open(encoding="utf-8-sig") as f:
            obj = json.load(f)

        if not isinstance(obj, list):
            raise ValueError(
                f"Expected list of records: {path}"
            )

        return len(obj)

    return None


def as_row(**kwargs):
    row = {name: "" for name in FIELDS}
    row.update(kwargs)
    row["verification_date"] = checked_on
    return row


def put_unique(destination, row):
    key = row["source_id"]

    if key in destination:
        raise ValueError(
            f"Duplicate generated source ID: {key}"
        )

    destination[key] = row


# --------------------------------------------------
# 2. Read Existing Cleaning Audit
# --------------------------------------------------

audit_rows, _ = read_csv(audit_file)

if len(audit_rows) != 9:
    raise ValueError(
        f"Expected 9 domain audit rows; found {len(audit_rows)}"
    )

audit_by_type = {
    r["dataset"]: r for r in audit_rows
}

if len(audit_by_type) != len(audit_rows):
    raise ValueError(
        "Duplicate domain names in original audit."
    )

print("=" * 68)
print("CHAMPIONS ATHLETIC MAP - STEP 8")
print("UNIFIED SOURCE MANIFEST AND PROVENANCE")
print("=" * 68)

print("\n[1] PROJECT INFORMATION")
print("Project root:", root)
print("Original package:", package.name)
print("Original source commit:", OLD_COMMIT[:8])
print("Aditi branch snapshot:", ADITI_SNAPSHOT[:8])
print("Recovery reference commit:", RECOVERY_COMMIT)


# --------------------------------------------------
# 3. Register Original Expected Inputs
# --------------------------------------------------

original_inputs = {
    "exercise": "megaGymDataset(1).csv",
    "player_bio": "Players(1).csv",
    "player_career": "player_data(1).csv",
    "basketball_season": "Seasons_Stats(1).csv",
    "food_sample_link": "sub_sample_food(1).csv",
    "nutrient_result": "sub_sample_result(1).csv",
    "food_category": "food_category(1).csv",
    "food_portion": "food_portion(1).csv",
    "athlete_profile": "athlete_test_profiles.csv"
}

records = {}
missing_raw = []

for kind, filename in original_inputs.items():

    candidates = [
        root / "data" / "raw" / filename,
        root / filename
    ]

    path = next(
        (p for p in candidates if p.is_file()),
        candidates[0]
    )

    info = file_metadata(path)

    if info["file_status"] != "present":
        missing_raw.append(filename)

    put_unique(
        records,
        as_row(
            source_id=f"raw_{kind}",
            title=f"Original input: {kind}",
            asset_type="original_input",
            intended_role=kind,
            publisher=(
                "Kaggle dataset: Niharika Pandit"
                if kind == "exercise"
                else "Unverified"
            ),
            source_url=(
                "https://www.kaggle.com/datasets/"
                "niharika41298/gym-exercise-data"
                if kind == "exercise"
                else ""
            ),
            source_filename=filename,
            license_or_terms="Unverified",
            permission_status="pending_review",
            limitations=(
                "Original source release and acquisition not verified."
            ),
            next_action=(
                "Locate original input and verify provenance."
            ),
            **info
        )
    )


# --------------------------------------------------
# 4. Register Original Cleaned Datasets
# --------------------------------------------------

pipeline = "Champions Athletic data-cleaning pipeline.py"

for kind in original_inputs:

    filename = f"{kind}_clean.csv"

    put_unique(
        records,
        as_row(
            source_id=f"derived_{kind}",
            title=f"Cleaned {kind} dataset",
            asset_type="derived_dataset",
            intended_role=kind,
            publisher="Project team (prepared output)",
            version_or_release=OLD_COMMIT[:8],
            source_filename=filename,
            reported_records=(
                audit_by_type.get(kind, {}).get("clean_rows", "")
            ),
            record_count_basis="existing_data_quality_audit",
            parent_source_ids=f"raw_{kind}",
            transformation_script=pipeline,
            source_commit=OLD_COMMIT,
            license_or_terms=(
                "Original-source terms pending verification"
            ),
            permission_status="pending_review",
            limitations=(
                "Original lineage and audit counts require "
                "independent verification."
            ),
            next_action=(
                "Review source and transformation during Step 9."
            ),
            **file_metadata(package / filename)
        )
    )


# --------------------------------------------------
# 5. Register Original Master Dataset and Audit
# --------------------------------------------------

parents = ";".join(
    f"derived_{kind}" for kind in original_inputs
)

reported_total = sum(
    int(row["clean_rows"])
    for row in audit_rows
)

put_unique(
    records,
    as_row(
        source_id="derived_master",
        title="Champions Athletic Agent-Ready Master",
        asset_type="derived_catalog",
        intended_role="retrieval_and_indexing",
        publisher="Project team (prepared output)",
        version_or_release=OLD_COMMIT[:8],
        source_filename="champions_master_agent_ready.csv",
        reported_records=reported_total,
        record_count_basis="sum_of_existing_audit_counts",
        parent_source_ids=parents,
        transformation_script=pipeline,
        source_commit=OLD_COMMIT,
        license_or_terms=(
            "Original-source terms pending verification"
        ),
        permission_status="pending_review",
        limitations=(
            "Heterogeneous union, not a joined training feature table."
        ),
        next_action="Verify record types and counts in Step 9.",
        **file_metadata(
            package / "champions_master_agent_ready.csv"
        )
    )
)

put_unique(
    records,
    as_row(
        source_id="derived_audit",
        title="Existing Data Quality Audit",
        asset_type="derived_report",
        intended_role="preliminary_quality_documentation",
        publisher="Project team (prepared output)",
        version_or_release=OLD_COMMIT[:8],
        source_filename=audit_file.name,
        reported_records=len(audit_rows),
        record_count_basis="audit_entries_read",
        parent_source_ids=parents,
        transformation_script=pipeline,
        source_commit=OLD_COMMIT,
        permission_status="pending_review",
        limitations=(
            "Existing audit is not an independent verification."
        ),
        next_action=(
            "Compare audit with actual data in Step 9."
        ),
        **file_metadata(audit_file)
    )
)


# --------------------------------------------------
# 6. Register Original and New External References
# --------------------------------------------------

reference_specs = [
    (
        "reference_jr_nba",
        "Jr. NBA/Jr. WNBA Curriculum",
        "Jr. NBA",
        "https://jr.nba.com/jr-nba-instructional-curriculum/",
        "basketball_drills",
        "Terms and reuse permission pending review",
        "pending_review"
    ),
    (
        "reference_youth_guidelines",
        "NBA/USA Basketball Youth Guidelines",
        "NBA and USA Basketball",
        "https://youthguidelines.nba.com/",
        "participation_and_recovery_guidelines",
        "Reuse terms pending review",
        "pending_review"
    ),
    (
        "reference_usda",
        "USDA FoodData Central",
        "USDA ARS",
        "https://fdc.nal.usda.gov/download-datasets/",
        "food_and_nutrient_reference",
        "CC0 1.0 for verified USDA FoodData Central data",
        "source_terms_documented"
    ),
    (
        "reference_usda_foundation_2025",
        "USDA Foundation Foods 2025-04-24",
        "USDA FoodData Central",
        "https://fdc.nal.usda.gov/download-datasets/",
        "nutrition_source",
        "CC0 1.0 for verified USDA data",
        "source_terms_documented"
    ),
    (
        "reference_cdc_sleep",
        "CDC Sleep Guidance",
        "CDC",
        "https://www.cdc.gov/sleep/about/",
        "general_sleep_guidance",
        "Reuse terms pending review",
        "pending_review"
    ),
    (
        "reference_aap_overuse",
        "AAP Overuse and Overtraining Guidance",
        "American Academy of Pediatrics",
        "https://doi.org/10.1542/peds.2023-065129",
        "recovery_reference",
        "Reuse terms pending review",
        "pending_review"
    )
]

for (
    source_id, title, publisher, url, role, terms, status
) in reference_specs:

    put_unique(
        records,
        as_row(
            source_id=source_id,
            title=title,
            asset_type="external_reference",
            intended_role=role,
            publisher=publisher,
            source_url=url,
            version_or_release=(
                "2025-04-24"
                if source_id == "reference_usda_foundation_2025"
                else ""
            ),
            file_status="reference_not_downloaded",
            license_or_terms=terms,
            permission_status=status,
            limitations=(
                "Original source document/release evidence "
                "not stored locally."
            ),
            next_action=(
                "Verify applicability, source version and permitted use."
            )
        )
    )


# --------------------------------------------------
# 7. Register Aditi's Datasets and Local Project Assets
# --------------------------------------------------

# Each asset includes:
# Source ID, project-relative path, purpose, asset type,
# parent IDs, transformation reference, Git snapshot,
# and expected record count.

new_assets = [
    (
        "aditi_jr_nba_drills",
        "data/processed/jr_nba_drills_master.csv",
        "basketball_drill_metadata",
        "derived_dataset",
        "reference_jr_nba",
        "notebooks/06_JrNBA_Combined_Dataset.ipynb",
        ADITI_SNAPSHOT,
        23
    ),
    (
        "aditi_jr_nba_rag",
        "data/processed/jr_nba_rag_documents.json",
        "retrieval_documents",
        "derived_rag_documents",
        "aditi_jr_nba_drills",
        "notebooks/07_JrNBA_RAG_Preparation.ipynb",
        ADITI_SNAPSHOT,
        23
    ),
    (
        "aditi_usda_foundation_nutrition",
        "data/processed/usda_foundation_nutrition.csv",
        "nutrition_reference",
        "derived_dataset",
        "reference_usda_foundation_2025",
        "notebooks/08_Nutrition_Data_Preparation.ipynb",
        ADITI_SNAPSHOT,
        411
    ),
    (
        "aditi_usda_quality_report",
        "data/processed/usda_nutrition_quality_report.csv",
        "nutrition_quality_audit",
        "derived_report",
        "aditi_usda_foundation_nutrition",
        "notebooks/08_Nutrition_Data_Preparation.ipynb",
        ADITI_SNAPSHOT,
        13
    ),
    (
        "aditi_recovery_guidance",
        "data/processed/recovery_guidance.csv",
        "recovery_reference",
        "derived_reference",
        "reference_cdc_sleep;reference_aap_overuse",
        "notebooks/09_Recovery_Data_Preparation.ipynb",
        ADITI_SNAPSHOT,
        5
    ),
    (
        "aditi_synthetic_athletes",
        "data/synthetic/synthetic_athlete_profiles.csv",
        "synthetic_planner_inputs",
        "synthetic_dataset",
        "",
        "notebooks/03_Generate_Athlete_Profiles.ipynb",
        ADITI_SNAPSHOT,
        60
    ),
    (
        "recovery_rules_v01",
        "data/references/recovery/"
        "youth_basketball_recovery_rules_v0_1.csv",
        "participation_and_recovery_rules",
        "derived_reference",
        "reference_youth_guidelines;"
        "reference_cdc_sleep;reference_aap_overuse",
        "manual source-linked curation",
        RECOVERY_COMMIT,
        17
    ),
    (
        "recovery_rules_readme_v01",
        "data/references/recovery/"
        "YOUTH_BASKETBALL_RECOVERY_RULES_README.md",
        "recovery_rule_documentation",
        "derived_documentation",
        "recovery_rules_v01",
        "manual documentation",
        RECOVERY_COMMIT,
        None
    ),
    (
        "catalog_basketball_v01",
        "data/processed/basketball_drill_catalog_v0_1.csv",
        "review_only_basketball_candidates",
        "derived_catalog",
        "reference_jr_nba",
        "Step 12 notebook",
        "",
        67
    ),
    (
        "catalog_basketball_v02",
        "data/processed/"
        "basketball_drill_catalog_v0_2_REVIEW_ONLY.csv",
        "review_only_basketball_candidates",
        "derived_catalog",
        "catalog_basketball_v01;reference_jr_nba",
        "Step 12 review package",
        "",
        129
    ),
    (
        "local_exercise_catalog",
        "data/processed/exercise_catalog.csv",
        "exercise_classification_catalog",
        "derived_catalog",
        "derived_exercise",
        "Steps 10-11 notebook",
        "",
        None
    ),
    (
        "local_basketball_drills",
        "data/processed/basketball_drills.csv",
        "basketball_working_table",
        "derived_dataset",
        "reference_jr_nba",
        "Step 12 notebook",
        "",
        None
    )
]

verified_files = []

for (
    sid, relative, role, kind, parent_ids,
    transform, snapshot, expected
) in new_assets:

    path = root / relative

    if not path.is_file():
        raise FileNotFoundError(
            f"Required integrated file missing: {relative}"
        )

    count = count_records(path)

    if expected is not None and count != expected:
        raise ValueError(
            f"{relative}: expected {expected} records, found {count}"
        )

    permissions = "pending_review"
    license_info = "Source reuse terms require review"

    limitations = (
        "Derived/team-curated; verify provenance "
        "and application suitability."
    )

    if sid.startswith("aditi_usda_"):
        permissions = "source_terms_documented"
        license_info = "USDA FoodData Central CC0 1.0"

    if sid == "aditi_synthetic_athletes":
        permissions = "team_generated"
        license_info = (
            "Fictional records generated by project team"
        )
        limitations = (
            "Synthetic inputs are not real athlete "
            "performance observations."
        )

    if sid.startswith("catalog_basketball_"):
        limitations = (
            "REVIEW ONLY: candidate groups are not "
            "approved or planner-eligible."
        )

    if sid == "aditi_jr_nba_rag":
        limitations = (
            "Derived from drill metadata, "
            "not an independent coaching source."
        )

    git_ref = (
        "aditi-development"
        if sid.startswith("aditi_")
        else "sam-development" if snapshot else ""
    )

    github_file = (
        f"https://github.com/SamTavasoli/champion-athletic-map/"
        f"blob/{git_ref}/{relative}"
        if git_ref
        else ""
    )

    put_unique(
        records,
        as_row(
            source_id=sid,
            title=path.name,
            asset_type=kind,
            intended_role=role,
            publisher="Project team (prepared output)",
            source_url=github_file,
            version_or_release=(
                "v0.2 REVIEW_ONLY"
                if sid == "catalog_basketball_v02"
                else (
                    "v0.1"
                    if sid in (
                        "catalog_basketball_v01",
                        "recovery_rules_v01"
                    )
                    else (
                        "seed=42"
                        if sid == "aditi_synthetic_athletes"
                        else ""
                    )
                )
            ),
            source_filename=path.name,
            reported_records=(
                "" if count is None else count
            ),
            record_count_basis=(
                "local_file_count"
                if count is not None
                else "document_not_applicable"
            ),
            parent_source_ids=parent_ids,
            transformation_script=transform,
            source_commit=snapshot,
            license_or_terms=license_info,
            permission_status=permissions,
            limitations=limitations,
            next_action=(
                "Review original source, quality, "
                "and usage permissions."
            ),
            **file_metadata(path)
        )
    )

    verified_files.append((sid, count))


# --------------------------------------------------
# 8. Cross-Check Related Datasets
# --------------------------------------------------

# Jr. NBA drills and RAG document identifiers

drills, _ = read_csv(
    root / "data/processed/jr_nba_drills_master.csv"
)

with (
    root / "data/processed/jr_nba_rag_documents.json"
).open(encoding="utf-8") as f:
    rag = json.load(f)

ids = [r["drill_id"] for r in drills]
rag_ids = [r["drill_id"] for r in rag]

if (
    len(set(ids)) != 23
    or set(ids) != set(rag_ids)
    or len(set(rag_ids)) != 23
):
    raise ValueError(
        "Jr. NBA drill IDs do not match RAG document IDs."
    )


# Synthetic athlete profile verification

profiles, _ = read_csv(
    root / "data/synthetic/synthetic_athlete_profiles.csv"
)

if (
    len({r["athlete_id"] for r in profiles}) != 60
    or not all(
        r["is_synthetic"].strip().lower() == "true"
        for r in profiles
    )
):
    raise ValueError(
        "Synthetic profile ID/indicator verification failed."
    )


# Recovery guideline record verification

recovery_path = (
    root / "data/references/recovery/"
    "youth_basketball_recovery_rules_v0_1.csv"
)

recovery, _ = read_csv(recovery_path)

if len({r["rule_id"] for r in recovery}) != 17:
    raise ValueError(
        "Recovery rule IDs must be unique."
    )

expected_recovery_sha = (
    "d4ef1ab35329e747b980d19fb0339c28"
    "ee3dff41e5b60724490193b0dab46ac6"
)

if checksum(recovery_path) != expected_recovery_sha:
    raise ValueError(
        "Recovery CSV differs from the earlier verified reference."
    )


# USDA completeness and anomaly screening

nutrition, _ = read_csv(
    root / "data/processed/usda_foundation_nutrition.csv"
)

core = (
    "calories_kcal",
    "protein_g",
    "fat_g",
    "carbohydrate_g"
)

complete_nutrition = sum(
    all(r.get(k, "") != "" for k in core)
    for r in nutrition
)

negative_carbs = sum(
    r.get("carbohydrate_g", "") != ""
    and float(r["carbohydrate_g"]) < 0
    for r in nutrition
)


# --------------------------------------------------
# 9. Preserve Existing Manifest Documentation
# --------------------------------------------------

manifest_path = root / "data/source_manifest.csv"

if manifest_path.exists():

    prior_rows, old_columns = read_csv(manifest_path)

    if any(col not in old_columns for col in FIELDS):
        raise ValueError(
            "Existing manifest does not match expected Step 8 schema."
        )

    old_ids = [
        r["source_id"] for r in prior_rows
    ]

    if len(old_ids) != len(set(old_ids)):
        raise ValueError(
            "Duplicate IDs in existing manifest; manual review needed."
        )

    old_map = {
        r["source_id"]: r for r in prior_rows
    }

    for sid, generated in records.items():

        if sid in old_map:
            previous = old_map[sid]

            # Preserve previously documented information.
            for field in (
                "acquisition_date",
                "license_or_terms",
                "permission_status",
                "limitations",
                "next_action"
            ):
                if previous.get(field, ""):
                    generated[field] = previous[field]

    # Preserve any extra records created in earlier work.
    for sid, previous in old_map.items():

        if sid not in records:
            records[sid] = {
                k: previous.get(k, "")
                for k in FIELDS
            }

else:
    old_ids = []


# Verify source relationship references

for row in records.values():

    parent_values = (
        x.strip()
        for x in row["parent_source_ids"].split(";")
    )

    for parent_id in filter(None, parent_values):

        if parent_id not in records:
            raise ValueError(
                f"Unknown parent '{parent_id}' "
                f"for {row['source_id']}"
            )


# --------------------------------------------------
# 10. Define Original and New Provenance Issues
# --------------------------------------------------

original_issues = [
    (
        "P01",
        "high",
        "Original input files unavailable",
        "Locate raw files and verify releases."
    ),
    (
        "P02",
        "high",
        "Gym dataset license and version uncertain",
        "Verify release and license."
    ),
    (
        "P03",
        "high",
        "Original food table lineage unverified",
        "Verify USDA input identifiers and nutrient units."
    ),
    (
        "P04",
        "medium",
        "Historical basketball data source unresolved",
        "Identify publishers and dataset versions."
    ),
    (
        "P05",
        "high",
        "Original 30 athlete profile origins unclear",
        "Verify origin of existing prepared profiles."
    ),
    (
        "P06",
        "high",
        "Coaching content reuse restrictions",
        "Check display, storage and redistribution permissions."
    ),
    (
        "P07",
        "medium",
        "Original row lineage not established",
        "Verify original rows versus cleaned source_row."
    ),
    (
        "P08",
        "high",
        "Basketball drill catalog overlap",
        "Reconcile 23 occurrences with v0.2 source candidates."
    ),
    (
        "P09",
        "high",
        "Jr. NBA reuse and adaptation permissions",
        "Check permitted reuse, citations and adaptations."
    ),
    (
        "P10",
        "high",
        "USDA incomplete and anomalous nutrient values",
        "Review missing fields and negative carbohydrate values."
    ),
    (
        "P11",
        "medium",
        "Synthetic evaluation coverage",
        "Expand development cases and isolate final test profiles."
    ),
    (
        "P12",
        "high",
        "Recovery-rule age/grade applicability",
        "Verify applicability and prioritization before use."
    ),
    (
        "P13",
        "medium",
        "Google Colab-specific notebook paths",
        "Adapt imported notebooks for local VS Code."
    ),
    (
        "P14",
        "medium",
        "Foundation Foods raw source version",
        "Verify archive, nutrient units, and transformation lineage."
    )
]


# --------------------------------------------------
# 11. Preserve Existing Issue Statuses
# --------------------------------------------------

issues_path = root / "results/provenance_followups.csv"

issues = {}

if issues_path.exists():

    previous_issues, prev_cols = read_csv(issues_path)

    if not set(ISSUE_FIELDS).issubset(prev_cols):
        raise ValueError(
            "Existing issue CSV schema differs "
            "from Step 8 requirements."
        )

    for issue in previous_issues:

        if issue["issue_id"] in issues:
            raise ValueError(
                "Duplicate issue IDs in existing report."
            )

        issues[issue["issue_id"]] = {
            k: issue.get(k, "")
            for k in ISSUE_FIELDS
        }

for (
    issue_id, priority, description, action
) in original_issues:

    if issue_id not in issues:

        issues[issue_id] = {
            "issue_id": issue_id,
            "priority": priority,
            "issue": description,
            "next_action": action,
            "status": "open"
        }


# --------------------------------------------------
# 12. Define Safe CSV Writing with Backups
# --------------------------------------------------

def csv_bytes(rows, header):

    buffer = io.StringIO(newline="")

    writer = csv.DictWriter(
        buffer,
        fieldnames=header,
        extrasaction="ignore"
    )

    writer.writeheader()
    writer.writerows(rows)

    return buffer.getvalue().encode("utf-8")


def save_with_backup(path, content):

    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    # Do not rewrite an unchanged file.
    if (
        path.exists()
        and path.read_bytes() == content
    ):
        return "unchanged"

    # Back up the previous version before updating.
    if path.exists():

        backup_dir = (
            root / "results/backups/step08"
        )

        backup_dir.mkdir(
            parents=True,
            exist_ok=True
        )

        stamp = datetime.now().strftime(
            "%Y%m%d_%H%M%S_%f"
        )

        backup = (
            backup_dir /
            f"{path.stem}_before_step08_{stamp}{path.suffix}"
        )

        shutil.copy2(path, backup)

        print(
            "Backup created:",
            backup.relative_to(root)
        )

    # Write through a temporary file.
    temp_name = None

    try:

        with tempfile.NamedTemporaryFile(
            dir=path.parent,
            delete=False
        ) as temp:

            temp_name = temp.name
            temp.write(content)

        os.replace(temp_name, path)

    finally:

        if (
            temp_name
            and os.path.exists(temp_name)
        ):
            os.unlink(temp_name)

    return "updated"


# --------------------------------------------------
# 13. Save Updated Source Manifest and Follow-Ups
# --------------------------------------------------

manifest_action = save_with_backup(
    manifest_path,
    csv_bytes(
        records.values(),
        FIELDS
    )
)

issues_action = save_with_backup(
    issues_path,
    csv_bytes(
        issues.values(),
        ISSUE_FIELDS
    )
)


# --------------------------------------------------
# 14. Print Final Verification Summary
# --------------------------------------------------

print("\n" + "=" * 68)
print("STEP 8 - UNIFIED SOURCE MANIFEST VERIFICATION")
print("=" * 68)

print("\n[2] ORIGINAL DATA SOURCES")
print("Original source entries:", 23)
print("Additional source entries:", 15)
print("Current manifest entries:", len(records))
print("Missing original raw inputs:", len(missing_raw))
print(
    "Original processed master records (existing audit):",
    reported_total
)

print("\n[3] NEWLY INTEGRATED FILES")

print(
    "New local files verified:",
    len(verified_files)
)

for sid, count in verified_files:

    display_count = (
        count if count is not None
        else "document"
    )

    print(
        f"PASS | {sid} | {display_count}"
    )

print("\n[4] CROSS-DATASET VERIFICATION")
print("RAG drill ID match: PASS")
print("Synthetic profile verification: PASS")
print("Recovery ID / checksum validation: PASS")

print(
    "USDA foods with complete core nutrition:",
    complete_nutrition,
    "/",
    len(nutrition)
)

print(
    "USDA negative carbohydrate values for review:",
    negative_carbs
)

print("\n[5] PROVENANCE DOCUMENTATION")
print(
    "Open/tracked provenance issues:",
    len(issues)
)

print(
    "Manifest:",
    manifest_path.relative_to(root),
    "-",
    manifest_action
)

print(
    "Follow-up report:",
    issues_path.relative_to(root),
    "-",
    issues_action
)

print("\nGit commits/push performed: None")
print("OVERALL: PASS")

print("=" * 68)
print("STEP 8 - SOURCE DOCUMENTATION UPDATE COMPLETE")
print("=" * 68)
```

```python

# STEP 9 - Unified Data Loading, Quality Checks, and Flagged Findings
# Standard library only; source files are read-only.
# Rerunnable: reports are backed up before changes.

import csv
import hashlib
import io
import json
import os
import re
import shutil
import tempfile
from collections import Counter
from datetime import datetime
from pathlib import Path

csv.field_size_limit(10_000_000)

root = next(
    (
        p for p in (
            Path.cwd().resolve(),
            *Path.cwd().resolve().parents
        )
        if (p / ".git").exists()
    ),
    None,
)

if root is None:
    raise FileNotFoundError(
        "Open this notebook inside the Git repository."
    )

manifest_path = root / "data/source_manifest.csv"
results_dir = root / "results"
audit_path = (
    root / "Champions_Athletic_Agent_Ready_Dataset _Package"
    / "data_quality_audit.csv"
)

if not manifest_path.is_file() or not audit_path.is_file():
    raise FileNotFoundError(
        "Run Step 8 first; source manifest or audit is missing."
    )


# --------------------------------------------------
# 1. Load Source Documentation
# --------------------------------------------------

def load_csv(path):
    with path.open(
        "r", encoding="utf-8-sig", newline=""
    ) as stream:
        reader = csv.DictReader(stream)

        if not reader.fieldnames:
            raise ValueError(f"Missing CSV header: {path}")

        return list(reader), list(reader.fieldnames)


manifest, _ = load_csv(manifest_path)
audit_rows, _ = load_csv(audit_path)

audit = {r["dataset"]: r for r in audit_rows}

ids = [r["source_id"] for r in manifest]

if len(ids) != len(set(ids)):
    raise ValueError("Duplicate source IDs in Step 8 manifest.")

if len(manifest) != 38:
    raise ValueError(
        f"Expected 38 Step 8 source entries; found {len(manifest)}."
    )

DATA_TYPES = {
    "derived_dataset",
    "derived_catalog",
    "derived_report",
    "derived_reference",
    "synthetic_dataset",
    "derived_rag_documents",
    "derived_documentation",
}

REQUIRED = {
    "derived_exercise": {
        "exercise_name", "body_part", "description"
    },
    "derived_athlete_profile": {
        "athlete_id", "age", "available_days",
        "goal", "equipment", "injury_status"
    },
    "derived_master": {
        "record_id", "entity_type", "source_file",
        "source_row", "entity_label", "agent_text", "data_json"
    },
    "aditi_jr_nba_drills": {
        "drill_id", "drill_name", "skill_category",
        "curriculum_level", "duration", "equipment",
        "source_document", "source_page",
        "solo_adaptation_possible"
    },
    "aditi_usda_foundation_nutrition": {
        "fdc_id", "description", "protein_g", "fat_g",
        "carbohydrate_g", "fiber_g", "calories_kcal",
        "energy_method", "data_type"
    },
    "aditi_usda_quality_report": {
        "column", "non_missing", "missing", "missing_pct"
    },
    "aditi_recovery_guidance": {
        "topic", "guidance", "source", "source_url"
    },
    "aditi_synthetic_athletes": {
        "athlete_id", "age", "sport", "position",
        "experience_level", "primary_goal",
        "available_days_per_week", "equipment",
        "injury_status", "scenario_type", "is_synthetic"
    },
    "recovery_rules_v01": {
        "rule_id", "segment", "metric", "rule_type",
        "value_min", "value_max", "unit", "source_url",
        "applicability_notes", "verification_status"
    },
    "catalog_basketball_v01": {
        "catalog_candidate_id", "drill_name",
        "scope_candidate", "catalog_approval_status",
        "planner_eligible"
    },
    "catalog_basketball_v02": {
        "catalog_candidate_id", "drill_name",
        "scope_candidate", "catalog_approval_status",
        "planner_eligible", "source_record_class"
    },
    "local_exercise_catalog": {
        "catalog_id", "exercise_name", "body_part"
    },
    "local_basketball_drills": {
        "candidate_id", "record_kind", "source_url"
    },
}

ID_FIELDS = {
    "derived_master": "record_id",
    "derived_athlete_profile": "athlete_id",
    "aditi_jr_nba_drills": "drill_id",
    "aditi_usda_foundation_nutrition": "fdc_id",
    "aditi_usda_quality_report": "column",
    "aditi_synthetic_athletes": "athlete_id",
    "recovery_rules_v01": "rule_id",
    "catalog_basketball_v01": "catalog_candidate_id",
    "catalog_basketball_v02": "catalog_candidate_id",
    "local_exercise_catalog": "catalog_id",
    "local_basketball_drills": "candidate_id",
}

MISSING_MARKERS = {"", "null", "none", "nan", "n/a"}

inventory = []
checks = []
master_types = Counter()
exercise_labels = Counter()
observed = {}


# --------------------------------------------------
# 2. Define Verification Helpers
# --------------------------------------------------

def digest_file(path):
    digest = hashlib.sha256()

    with path.open("rb") as f:
        for chunk in iter(
            lambda: f.read(1024 * 1024), b""
        ):
            digest.update(chunk)

    return digest.hexdigest()


def check(
    source_id, name, status, affected=0,
    details="", action=""
):
    checks.append({
        "source_id": source_id,
        "check_name": name,
        "status": status,
        "affected_count": affected,
        "details": details,
        "next_action": action,
    })


def missing(value):
    return (
        value is None
        or str(value).strip().lower() in MISSING_MARKERS
    )


def num(value):
    try:
        return float(str(value).strip())
    except (ValueError, TypeError):
        return None


def normalized(value):
    return " ".join(
        re.findall(r"[a-z0-9]+", str(value).lower())
    )


print("=" * 72)
print("CHAMPIONS ATHLETIC MAP - STEP 9")
print("UNIFIED DATA LOADING, QUALITY CHECKS, AND FLAGGED FINDINGS")
print("=" * 72)

print("Project:", root)
print("Manifest entries:", len(manifest))
print("Domain datasets in original audit:", len(audit))


# --------------------------------------------------
# 3. Inspect All Registered Source Records
# --------------------------------------------------

for item in manifest:

    sid = item["source_id"]
    kind = item["asset_type"]
    rel = item.get("relative_path", "").strip()
    name = item.get("source_filename", "")

    summary = {
        "source_id": sid,
        "filename": name,
        "relative_path": rel,
        "load_status": "not_checked",
        "actual_rows": "",
        "columns": "",
        "file_size_bytes": "",
        "sha256_match": "not_checked",
        "notes": "",
        "file_format": "",
        "expected_records": item.get("reported_records", ""),
        "observed_sha256": "",
    }

    # External references are not local data tables.
    if kind == "external_reference":

        summary["load_status"] = "external_reference"
        summary["notes"] = (
            "External reference documented; "
            "not a local dataset."
        )

        check(
            sid,
            "reference_source",
            "not_checked",
            details=(
                "External content requires separate "
                "source and permissions review."
            ),
            action=(
                "Verify source version, applicability "
                "and reuse conditions."
            )
        )

        inventory.append(summary)
        continue

    # Document original source availability.
    if kind == "original_input" and not rel:

        filename = item.get("source_filename", "")

        candidates = [
            root / "data/raw" / filename,
            root / filename
        ]

        found = next(
            (p for p in candidates if p.is_file()),
            None
        )

        summary["load_status"] = (
            "present_not_inspected"
            if found else "missing_original"
        )

        summary["notes"] = (
            "Original raw input; provenance pending review."
        )

        check(
            sid,
            "original_file_availability",
            "not_checked" if found else "needs_review",
            0 if found else 1,
            f"Original input: {filename}",
            "Locate original publisher file and verify release."
        )

        inventory.append(summary)
        continue

    if kind not in DATA_TYPES:
        summary["load_status"] = "not_applicable"
        inventory.append(summary)
        continue

    if not rel:

        summary["load_status"] = "missing_path"

        check(
            sid, "file_availability", "failed", 1,
            "No local file path in the Step 8 manifest.",
            "Correct the source manifest."
        )

        inventory.append(summary)
        continue

    path = (root / rel).resolve()

    if not path.is_relative_to(root.resolve()):
        raise ValueError(
            f"Manifest path escapes project directory: {rel}"
        )

    summary["file_format"] = path.suffix.lower().lstrip(".")

    if not path.is_file():

        summary["load_status"] = "missing"

        check(
            sid, "file_availability", "failed", 1,
            f"Required file missing: {rel}",
            "Locate or restore the source file."
        )

        inventory.append(summary)
        continue

    summary["file_size_bytes"] = path.stat().st_size

    try:

        # --------------------------------------------------
        # 4. Verify SHA-256 File Integrity
        # --------------------------------------------------

        actual_sha = digest_file(path)

        summary["observed_sha256"] = actual_sha

        expected_sha = item.get(
            "sha256", ""
        ).strip().lower()

        sha_match = (
            actual_sha == expected_sha
            if expected_sha else None
        )

        summary["sha256_match"] = (
            "yes" if sha_match is True
            else "no" if sha_match is False
            else "not_recorded"
        )

        check(
            sid,
            "sha256_integrity",
            "pass" if sha_match is True
            else "failed" if sha_match is False
            else "needs_review",
            0 if sha_match is True else 1,
            "Compared local file SHA-256 with Step 8 manifest.",
            (
                "Investigate the file version before "
                "updating the manifest."
                if sha_match is not True else ""
            )
        )

        # --------------------------------------------------
        # 5. Verify Documentation Files
        # --------------------------------------------------

        if path.suffix.lower() in (".md", ".txt"):

            summary["load_status"] = "verified_document"
            summary["notes"] = (
                "Documentation asset, not a data table."
            )

            check(
                sid,
                "document_available",
                "pass"
            )

            observed[sid] = {
                "rows": None,
                "columns": None
            }

            inventory.append(summary)

            print(
                f"{sid:<34} DOCUMENT "
                f"SHA={summary['sha256_match']}"
            )

            continue

        # --------------------------------------------------
        # 6. Verify JSON / RAG Documents
        # --------------------------------------------------

        if path.suffix.lower() == ".json":

            with path.open(
                "r", encoding="utf-8-sig"
            ) as f:
                documents = json.load(f)

            if not isinstance(documents, list):
                raise ValueError(
                    "Expected JSON list of retrieval documents."
                )

            summary["load_status"] = "loaded"
            summary["actual_rows"] = len(documents)

            summary["columns"] = (
                len(
                    set().union(
                        *(
                            set(d.keys())
                            for d in documents
                            if isinstance(d, dict)
                        )
                    )
                )
                if documents else 0
            )

            missing_keys = sum(
                not isinstance(d, dict)
                or missing(d.get("drill_id"))
                or missing(d.get("text"))
                for d in documents
            )

            rag_ids = [
                d.get("drill_id")
                for d in documents
                if isinstance(d, dict)
            ]

            duplicate_ids = (
                len(rag_ids) - len(set(rag_ids))
            )

            check(
                sid,
                "rag_required_fields",
                "needs_review" if missing_keys else "pass",
                missing_keys,
                "Each RAG record requires drill_id and text.",
                (
                    "Review incomplete retrieval records."
                    if missing_keys else ""
                )
            )

            check(
                sid,
                "identifier_uniqueness",
                "needs_review" if duplicate_ids else "pass",
                duplicate_ids,
                "RAG document identifiers.",
                (
                    "Investigate repeated identifiers."
                    if duplicate_ids else ""
                )
            )

            observed[sid] = {
                "rows": len(documents),
                "ids": rag_ids
            }

        # --------------------------------------------------
        # 7. Inspect CSV Datasets
        # --------------------------------------------------

        elif path.suffix.lower() == ".csv":

            total = 0
            malformed = 0
            exact_duplicates = 0
            invalid_json = 0
            repeated_ids = 0
            absent_ids = 0

            missing_count = Counter()
            unknown_count = Counter()

            seen_signatures = set()
            seen_identifiers = set()

            negative_carbs = 0
            nutrition_complete = 0

            invalid_athlete_ranges = 0
            synthetic_false = 0

            invalid_recovery_rules = 0
            unapproved_eligible = 0

            drill_names = []
            source_ids = []

            with path.open(
                "r",
                encoding="utf-8-sig",
                newline=""
            ) as f:

                reader = csv.DictReader(f)
                columns = list(reader.fieldnames or [])

                if (
                    not columns
                    or len(set(columns)) != len(columns)
                ):
                    raise ValueError(
                        "Missing or repeated CSV headers."
                    )

                required = REQUIRED.get(sid, set())
                absent_columns = sorted(
                    required - set(columns)
                )

                check(
                    sid,
                    "required_columns",
                    "failed" if absent_columns else "pass",
                    len(absent_columns),
                    (
                        ", ".join(absent_columns)
                        if absent_columns
                        else "All required columns present."
                    ),
                    (
                        "Review missing dataset columns."
                        if absent_columns else ""
                    )
                )

                id_field = ID_FIELDS.get(sid)

                # Streaming row-by-row inspection.
                for record in reader:

                    total += 1

                    if (
                        None in record
                        or any(
                            value is None
                            for value in record.values()
                        )
                    ):
                        malformed += 1
                        continue

                    values = [
                        record.get(c, "")
                        for c in columns
                    ]

                    # Memory-conscious full-row fingerprints.
                    signature = hashlib.blake2b(
                        json.dumps(
                            values,
                            ensure_ascii=False,
                            separators=(",", ":")
                        ).encode("utf-8"),
                        digest_size=16
                    ).digest()

                    if signature in seen_signatures:
                        exact_duplicates += 1
                    else:
                        seen_signatures.add(signature)

                    for key, value in record.items():

                        if missing(value):
                            missing_count[key] += 1

                        elif (
                            value.strip().lower()
                            == "unknown"
                        ):
                            unknown_count[key] += 1

                    # Identifier uniqueness.
                    if id_field and id_field in record:

                        identifier = record.get(
                            id_field, ""
                        ).strip()

                        if missing(identifier):
                            absent_ids += 1

                        elif identifier in seen_identifiers:
                            repeated_ids += 1

                        else:
                            seen_identifiers.add(identifier)

                    # Original master dataset checks.
                    if sid == "derived_master":

                        master_types[
                            record.get("entity_type", "")
                        ] += 1

                        try:

                            payload = json.loads(
                                record.get("data_json", "")
                            )

                            if not isinstance(payload, dict):
                                invalid_json += 1

                        except (ValueError, TypeError):
                            invalid_json += 1

                    # Original exercise labels.
                    if sid == "derived_exercise":

                        label = record.get(
                            "body_part", ""
                        )

                        exercise_labels[label] += 1

                    # Jr. NBA drill identifiers and names.
                    if sid == "aditi_jr_nba_drills":

                        drill_names.append(
                            record.get("drill_name", "")
                        )

                        source_ids.append(
                            record.get("drill_id", "")
                        )

                    # USDA Foundation Foods.
                    if sid == "aditi_usda_foundation_nutrition":

                        core_fields = (
                            "calories_kcal",
                            "protein_g",
                            "fat_g",
                            "carbohydrate_g"
                        )

                        if all(
                            not missing(record.get(k))
                            for k in core_fields
                        ):
                            nutrition_complete += 1

                        carbs = num(
                            record.get("carbohydrate_g")
                        )

                        if (
                            carbs is not None
                            and carbs < 0
                        ):
                            negative_carbs += 1

                    # Synthetic athlete profiles.
                    if sid == "aditi_synthetic_athletes":

                        age = num(
                            record.get("age")
                        )

                        days = num(
                            record.get(
                                "available_days_per_week"
                            )
                        )

                        if (
                            age is None
                            or not 13 <= age <= 17
                            or days is None
                            or not 2 <= days <= 5
                        ):
                            invalid_athlete_ranges += 1

                        if (
                            record.get(
                                "is_synthetic", ""
                            ).lower() != "true"
                        ):
                            synthetic_false += 1

                    # Recovery rules.
                    if sid == "recovery_rules_v01":

                        rule_type = record.get(
                            "rule_type", ""
                        )

                        low = num(
                            record.get("value_min")
                        )

                        high = num(
                            record.get("value_max")
                        )

                        if (
                            not record.get(
                                "source_url", ""
                            ).startswith("https://")
                            or not record.get(
                                "metric", ""
                            ).strip()
                            or not record.get(
                                "unit", ""
                            ).strip()
                            or (
                                rule_type == "maximum"
                                and high is None
                            )
                            or (
                                rule_type == "minimum"
                                and low is None
                            )
                            or (
                                rule_type.startswith(
                                    "recommended"
                                )
                                and (
                                    low is None
                                    or high is None
                                )
                            )
                            or (
                                low is not None
                                and high is not None
                                and low > high
                            )
                        ):
                            invalid_recovery_rules += 1

                    # Basketball planner approval gate.
                    if sid in {
                        "catalog_basketball_v01",
                        "catalog_basketball_v02"
                    }:

                        eligible = record.get(
                            "planner_eligible", ""
                        ).strip().lower()

                        approval = record.get(
                            "catalog_approval_status", ""
                        ).strip().lower()

                        if (
                            eligible == "yes"
                            and approval not in {
                                "approved", "accepted"
                            }
                        ):
                            unapproved_eligible += 1

            # --------------------------------------------------
            # 8. Summarize CSV Quality Checks
            # --------------------------------------------------

            summary["load_status"] = "loaded"
            summary["actual_rows"] = total
            summary["columns"] = len(columns)

            check(
                sid,
                "csv_structure",
                "needs_review" if malformed else "pass",
                malformed,
                "Rows with missing or excess CSV fields.",
                (
                    "Investigate malformed records."
                    if malformed else ""
                )
            )

            check(
                sid,
                "exact_duplicates",
                (
                    "needs_review"
                    if exact_duplicates else "pass"
                ),
                exact_duplicates,
                "Repeated full-row values.",
                (
                    "Review duplicate records."
                    if exact_duplicates else ""
                )
            )

            missing_total = sum(
                missing_count.values()
            )

            top_missing = "; ".join(
                f"{key}={count}"
                for key, count
                in missing_count.most_common(5)
            )

            check(
                sid,
                "missing_values",
                (
                    "needs_review"
                    if missing_total else "pass"
                ),
                missing_total,
                (
                    "Missing/null-like cells. "
                    f"Top columns: {top_missing or 'none'}"
                ),
                (
                    "Review required versus optional missing values."
                    if missing_total else ""
                )
            )

            if unknown_count:

                check(
                    sid,
                    "explicit_unknown_values",
                    "needs_review",
                    sum(unknown_count.values()),
                    (
                        "Literal 'unknown' values are kept "
                        "distinct from missing cells."
                    ),
                    "Preserve uncertain status information."
                )

            if id_field:

                identifier_issues = (
                    repeated_ids + absent_ids
                )

                check(
                    sid,
                    "identifier_uniqueness",
                    (
                        "needs_review"
                        if identifier_issues else "pass"
                    ),
                    identifier_issues,
                    (
                        f"Field={id_field}; "
                        f"repeated={repeated_ids}; "
                        f"missing={absent_ids}"
                    ),
                    (
                        "Investigate identifier problems."
                        if identifier_issues else ""
                    )
                )

            if sid == "derived_master":

                check(
                    sid,
                    "json_payload_validity",
                    (
                        "needs_review"
                        if invalid_json else "pass"
                    ),
                    invalid_json,
                    "data_json must contain a valid JSON object.",
                    (
                        "Review invalid JSON payloads."
                        if invalid_json else ""
                    )
                )

            if sid == "aditi_usda_foundation_nutrition":

                incomplete = (
                    total - nutrition_complete
                )

                check(
                    sid,
                    "core_nutrition_completeness",
                    (
                        "needs_review"
                        if incomplete else "pass"
                    ),
                    incomplete,
                    (
                        "Complete calories, protein, fat "
                        "and carbohydrate records: "
                        f"{nutrition_complete}/{total}"
                    ),
                    (
                        "Do not replace missing nutrient "
                        "values with zero."
                    )
                )

                check(
                    sid,
                    "negative_carbohydrates",
                    (
                        "needs_review"
                        if negative_carbs else "pass"
                    ),
                    negative_carbs,
                    (
                        "Negative carbohydrate measurements "
                        "require source review."
                    ),
                    (
                        "Verify original USDA values "
                        "and calculation methods."
                        if negative_carbs else ""
                    )
                )

            if sid == "aditi_synthetic_athletes":

                check(
                    sid,
                    "synthetic_ranges",
                    (
                        "needs_review"
                        if invalid_athlete_ranges else "pass"
                    ),
                    invalid_athlete_ranges,
                    (
                        "Expected age 13-17 and "
                        "training availability 2-5 days."
                    ),
                    (
                        "Review out-of-range profiles."
                        if invalid_athlete_ranges else ""
                    )
                )

                check(
                    sid,
                    "synthetic_indicator",
                    (
                        "failed"
                        if synthetic_false else "pass"
                    ),
                    synthetic_false,
                    (
                        "Every record must be explicitly "
                        "marked as synthetic."
                    ),
                    (
                        "Correct synthetic source labeling."
                        if synthetic_false else ""
                    )
                )

            if sid == "recovery_rules_v01":

                check(
                    sid,
                    "recovery_rule_structure",
                    (
                        "needs_review"
                        if invalid_recovery_rules else "pass"
                    ),
                    invalid_recovery_rules,
                    (
                        "Checks source URLs, metrics, "
                        "units, and numerical rule bounds."
                    ),
                    (
                        "Review rule structure and source."
                        if invalid_recovery_rules else ""
                    )
                )

            if sid in {
                "catalog_basketball_v01",
                "catalog_basketball_v02"
            }:

                check(
                    sid,
                    "planner_approval_gate",
                    (
                        "failed"
                        if unapproved_eligible else "pass"
                    ),
                    unapproved_eligible,
                    (
                        "Unapproved basketball candidates "
                        "must not be planner eligible."
                    ),
                    (
                        "Disable planner eligibility "
                        "until approved."
                        if unapproved_eligible else ""
                    )
                )

            observed[sid] = {
                "rows": total,
                "columns": columns,
                "names": drill_names,
                "ids": source_ids,
            }

        else:
            raise ValueError(
                f"Unsupported local file format: {path.suffix}"
            )

        # --------------------------------------------------
        # 9. Compare Counts Against Source Manifest
        # --------------------------------------------------

        expected = item.get(
            "reported_records", ""
        ).strip()

        if (
            expected
            and summary["actual_rows"] != ""
        ):

            try:
                expected_count = int(expected)

            except ValueError:

                check(
                    sid,
                    "reported_row_count",
                    "failed",
                    1,
                    (
                        "Manifest record count is not "
                        f"an integer: {expected}"
                    ),
                    "Correct the documented record count."
                )

            else:

                difference = abs(
                    int(summary["actual_rows"])
                    - expected_count
                )

                check(
                    sid,
                    "reported_row_count",
                    (
                        "needs_review"
                        if difference else "pass"
                    ),
                    difference,
                    (
                        f"Expected {expected_count}; "
                        f"actual {summary['actual_rows']}."
                    ),
                    (
                        "Investigate count mismatch."
                        if difference else ""
                    )
                )

        print(
            f"{sid:<34}"
            f" rows={str(summary['actual_rows']):>7}"
            f" cols={str(summary['columns']):>3}"
            f" SHA={summary['sha256_match']}"
        )

    except (
        OSError, UnicodeError, csv.Error,
        ValueError, KeyError, TypeError
    ) as error:

        summary["load_status"] = "failed"

        summary["notes"] = (
            f"{type(error).__name__}: {str(error)[:180]}"
        )

        check(
            sid,
            "file_loading",
            "failed",
            1,
            summary["notes"],
            "Inspect source file without modifying it."
        )

        print(
            f"FAILED {sid}: {summary['notes']}"
        )

    inventory.append(summary)


# --------------------------------------------------
# 10. Compare Original Master Dataset With Audit
# --------------------------------------------------

for domain, audit_record in audit.items():

    actual_count = master_types.get(
        domain, 0
    )

    expected_count = int(
        audit_record["clean_rows"]
    )

    difference = abs(
        actual_count - expected_count
    )

    check(
        "derived_master",
        f"entity_count_{domain}",
        (
            "needs_review"
            if difference else "pass"
        ),
        difference,
        (
            f"Master {domain}: "
            f"expected={expected_count}; "
            f"actual={actual_count}"
        ),
        (
            "Investigate master/domain differences."
            if difference else ""
        )
    )


# --------------------------------------------------
# 11. Check Drill / RAG Relationships
# --------------------------------------------------

if (
    "aditi_jr_nba_drills" in observed
    and "aditi_jr_nba_rag" in observed
):

    drill_ids = observed[
        "aditi_jr_nba_drills"
    ]["ids"]

    rag_ids = observed[
        "aditi_jr_nba_rag"
    ]["ids"]

    unmatched = len(
        set(drill_ids).symmetric_difference(
            set(rag_ids)
        )
    )

    check(
        "aditi_jr_nba_rag",
        "drill_to_rag_linkage",
        (
            "needs_review"
            if unmatched else "pass"
        ),
        unmatched,
        (
            f"Drill IDs={len(drill_ids)}; "
            f"RAG IDs={len(rag_ids)}; "
            f"unmatched={unmatched}"
        ),
        (
            "Repair the drill-to-RAG mapping."
            if unmatched else ""
        )
    )


# --------------------------------------------------
# 12. Compare Imported Drill Names With Catalog v0.2
# --------------------------------------------------

if (
    "aditi_jr_nba_drills" in observed
    and "catalog_basketball_v02" in observed
):

    imported_names = observed[
        "aditi_jr_nba_drills"
    ]["names"]

    catalog_path = (
        root / "data/processed/"
        "basketball_drill_catalog_v0_2_REVIEW_ONLY.csv"
    )

    with catalog_path.open(
        "r",
        encoding="utf-8-sig",
        newline=""
    ) as f:

        catalog_names = {
            normalized(row["drill_name"])
            for row in csv.DictReader(f)
        }

    unique_imported_names = {
        normalized(name)
        for name in imported_names
    }

    overlap = len(
        unique_imported_names & catalog_names
    )

    check(
        "aditi_jr_nba_drills",
        "catalog_name_overlap",
        "needs_review",
        overlap,
        (
            "Unique imported names="
            f"{len(unique_imported_names)}; "
            f"exact normalized-name overlaps={overlap}. "
            "Near matches also require review."
        ),
        (
            "Reconcile drill occurrences and variations "
            "in Step 12; do not add counts blindly."
        )
    )


# --------------------------------------------------
# 13. Display Original Master and Exercise Summaries
# --------------------------------------------------

print("\n[ORIGINAL MASTER RECORD TYPES]")

for domain, audit_record in audit.items():

    print(
        f"{domain:<22}"
        f"{master_types.get(domain, 0):>9,}"
        f" / {int(audit_record['clean_rows']):,}"
    )

print("\n[EXERCISE BODY-PART LABELS]")
print("Distinct labels:", len(exercise_labels))

for label, count in exercise_labels.most_common(10):

    print(
        f"  {label or '[blank]':<22}"
        f"{count:>6,}"
    )


# --------------------------------------------------
# 14. Define Report Formats
# --------------------------------------------------

INVENTORY_COLUMNS = [
    "source_id",
    "filename",
    "relative_path",
    "load_status",
    "actual_rows",
    "columns",
    "file_size_bytes",
    "sha256_match",
    "notes",
    "file_format",
    "expected_records",
    "observed_sha256",
]

CHECK_COLUMNS = [
    "source_id",
    "check_name",
    "status",
    "affected_count",
    "details",
    "next_action",
]


def serialize_csv(rows, columns):

    buffer = io.StringIO(newline="")

    writer = csv.DictWriter(
        buffer,
        fieldnames=columns
    )

    writer.writeheader()
    writer.writerows(rows)

    return buffer.getvalue().encode("utf-8")


# --------------------------------------------------
# 15. Safely Save Reports With Backups
# --------------------------------------------------

def safe_write(path, content):

    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    if (
        path.is_file()
        and path.read_bytes() == content
    ):
        return "unchanged"

    if path.exists():

        backup_dir = (
            root / "results/backups/step09"
        )

        backup_dir.mkdir(
            parents=True,
            exist_ok=True
        )

        stamp = datetime.now().strftime(
            "%Y%m%d_%H%M%S_%f"
        )

        backup = (
            backup_dir /
            f"{path.stem}_before_step09_{stamp}{path.suffix}"
        )

        shutil.copy2(path, backup)

        print(
            "Backup created:",
            backup.relative_to(root)
        )

    temp_path = None

    try:

        with tempfile.NamedTemporaryFile(
            dir=path.parent,
            delete=False
        ) as f:

            temp_path = f.name
            f.write(content)

        os.replace(temp_path, path)

    finally:

        if (
            temp_path
            and os.path.exists(temp_path)
        ):
            os.remove(temp_path)

    return "updated"


inventory_file = (
    results_dir / "data_inventory.csv"
)

checks_file = (
    results_dir / "initial_quality_checks.csv"
)

loaded = sum(
    row["load_status"] == "loaded"
    for row in inventory
)

if loaded < 10:
    raise RuntimeError(
        "Too few datasets loaded. Existing reports preserved."
    )

inventory_status = safe_write(
    inventory_file,
    serialize_csv(
        inventory,
        INVENTORY_COLUMNS
    )
)

checks_status = safe_write(
    checks_file,
    serialize_csv(
        checks,
        CHECK_COLUMNS
    )
)


# --------------------------------------------------
# 16. Review All Flagged Quality Checks
# --------------------------------------------------

status_counts = Counter(
    row["status"]
    for row in checks
)

flagged = [
    row for row in checks
    if row["status"] in (
        "needs_review",
        "failed"
    )
]

loaded_files = [
    row for row in inventory
    if row["load_status"] in (
        "loaded",
        "verified_document"
    )
]

failed_checks = [
    row for row in checks
    if row["status"] == "failed"
]


# --------------------------------------------------
# 17. Final Verification Summary
# --------------------------------------------------

print("\n" + "=" * 72)
print("STEP 9 - QUALITY SUMMARY AND FLAGGED FINDINGS")
print("=" * 72)

print("Manifest source records:", len(manifest))
print("Local files inspected:", len(loaded_files))
print("Loaded CSV / JSON datasets:", loaded)

print(
    "Missing original raw inputs:",
    sum(
        row["load_status"] == "missing_original"
        for row in inventory
    )
)

print("Total quality checks:", len(checks))

print(
    "Check status counts:",
    dict(status_counts)
)

print(
    "Checks requiring review:",
    len(flagged)
)

print(
    "Required checks failed:",
    len(failed_checks)
)

print(
    "Inventory report:",
    inventory_file.relative_to(root),
    "-",
    inventory_status
)

print(
    "Quality report:",
    checks_file.relative_to(root),
    "-",
    checks_status
)

# Print the first 25 flagged checks.
# All findings remain available in the CSV report.

for i, row in enumerate(
    flagged[:25], 1
):

    print(
        f"\n[{i}] {row['source_id']} "
        f"| {row['check_name']}"
    )

    print(
        "Status:",
        row["status"],
        "| Affected:",
        row["affected_count"]
    )

    print(
        "Details:",
        row["details"]
    )

    if row["next_action"]:
        print(
            "Next action:",
            row["next_action"]
        )

if len(flagged) > 25:

    print(
        f"\n... {len(flagged) - 25} more flagged "
        f"checks in {checks_file.name}"
    )

print(
    "\nOriginal datasets, source manifest "
    "and code files modified: No"
)

print("Git commit / push performed: No")

print(
    "OVERALL:",
    (
        "REVIEW REQUIRED"
        if failed_checks
        else "COMPLETED WITH TRACKED REVIEWS"
    )
)

print("=" * 72)
```

| Review Metadata | Distinguish intentionally empty review fields from incomplete source information |

We will not invent exercise descriptions, body-part labels, equipment requirements, or difficulty levels.

A missing description will not automatically make an exercise unusable for classification if the exercise name provides meaningful information.

We will distinguish missing equipment information from an explicit indication that no equipment is required.

We will preserve these findings in our exercise review reports.

### 10.4 Duplicate and Similar Records

We will examine exercise records for duplicate content, repeated names, and related exercise variations.

Our checks will distinguish among:

1. **Exact Content Duplicates:** Records with the same meaningful content after documented formatting normalization.
2. **Repeated Exercise Names:** Records with the same normalized name but potentially different descriptions, equipment, or other attributes.
3. **Potential Label Conflicts:** Records sharing a normalized exercise name but having different body-part labels.
4. **Similar Exercises:** Activities with related wording or movement variations that may require grouping.

We will exclude source identifiers and provenance fields from content comparisons because these fields should not prevent equivalent exercise content from being identified.

We will preserve duplicate candidates rather than automatically deleting or merging them.

Repeated exercise names may represent meaningful variations, and individual exercises may involve multiple muscle groups.

We will investigate these relationships before deciding whether records should be retained separately, grouped, or corrected.

We will preserve relevant duplicate relationships for the grouped dataset-splitting process in Step 15.

This will help reduce the risk of near-identical exercise examples appearing in both training and evaluation datasets.

- Potential disagreements between descriptions and assigned labels.

We will retain the original label separately from any proposed correction.

We will not automatically overwrite labels based on exercise names, keyword matching, or AI-generated suggestions.

When a correction is proposed, we will document:

```python
# STEP 10 - Exercise Data Cleaning, Label Screening, and Review Queue


import csv
import hashlib
import re
import unicodedata
from collections import Counter, defaultdict
from pathlib import Path

# --------------------------------------------------
# 1. Locate and verify the cleaned exercise source
# --------------------------------------------------

cwd = Path.cwd().resolve()
root = next((p for p in (cwd, *cwd.parents) if (p / ".git").exists()), None)
if root is None:
    raise FileNotFoundError("Git project root not found.")

manifest_path = root / "data/source_manifest.csv"
if not manifest_path.is_file():
    raise FileNotFoundError("Run the updated Step 8 first: source manifest missing.")

with manifest_path.open(newline="", encoding="utf-8-sig") as f:
    source_records = list(csv.DictReader(f))

matches = [r for r in source_records if r.get("source_id") == "derived_exercise"]
if len(matches) != 1 or not matches[0].get("relative_path"):
    raise ValueError("Expected exactly one derived_exercise entry with a local path.")

source = matches[0]
src = (root / source["relative_path"]).resolve()
if not src.is_relative_to(root) or not src.is_file():
    raise FileNotFoundError("Invalid or missing cleaned exercise source path.")


def sha256_file(path):
    digest = hashlib.sha256()
    with path.open("rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()


source_hash = sha256_file(src)
if source_hash.lower() != source.get("sha256", "").lower():
    raise ValueError("Exercise source SHA-256 does not match the Step 8 manifest.")

with src.open(newline="", encoding="utf-8-sig") as f:
    reader = csv.DictReader(f)
    original_columns = list(reader.fieldnames or [])
    source_rows = list(reader)

fields = [
    "exercise_name", "description", "body_part", "exercise_type",
    "equipment", "difficulty_level",
]
missing_columns = sorted(set(fields) - set(original_columns))
if missing_columns:
    raise ValueError(f"Missing exercise columns: {missing_columns}")
if len(original_columns) != len(set(original_columns)):
    raise ValueError("Repeated source column names detected.")
if any(None in r or any(v is None for v in r.values()) for r in source_rows):
    raise ValueError("Malformed source CSV record detected.")

expected_rows = source.get("reported_records", "").strip()
if expected_rows and len(source_rows) != int(expected_rows):
    raise ValueError("Exercise record count differs from the Step 8 manifest.")

print("=" * 68)
print("CHAMPIONS ATHLETIC MAP - STEP 10")
print("EXERCISE CLEANING, LABEL SCREENING, AND OUTPUT VERIFICATION")
print("=" * 68)
print("Source:", src.relative_to(root).as_posix())
print("SHA-256 matches Step 8: PASS")
print("Source rows:", len(source_rows))
print("Source columns:", len(original_columns))

# --------------------------------------------------
# 2. Prepare helper functions and output schemas
# --------------------------------------------------

categorical_fields = {"body_part", "exercise_type", "equipment", "difficulty_level"}
null_markers = {"", "nan", "null", "n/a"}


def compact(value):
    return re.sub(r"\s+", " ", value or "").strip()


def prepared_value(field, value):
    text = compact(value)
    if text.casefold() in null_markers:
        return ""
    return text.casefold() if field in categorical_fields else text


def comparison_key(value):
    return unicodedata.normalize("NFKC", compact(value)).casefold()


added_columns = [
    "catalog_id", "source_cleaned_record_position", "source_file", "source_sha256",
    "exercise_name_prepared", "description_prepared", "body_part_prepared",
    "exercise_type_prepared", "equipment_prepared", "difficulty_level_prepared",
    "name_comparison_key", "label_review_status",
    "preliminary_classifier_candidate", "review_flags",
]
log_columns = [
    "catalog_id", "field", "source_value", "prepared_value", "reason", "decision",
]
queue_columns = [
    "catalog_id", "issue_type", "priority", "evidence",
    "suggested_action", "review_status",
]

catalog = []
cleaning_log = []
review_queue = []
by_name = defaultdict(list)
by_content = defaultdict(list)
labels = Counter()
issues_by_record = defaultdict(set)


def add_issue(catalog_id, issue_type, priority, evidence, suggested_action):
    review_queue.append({
        "catalog_id": catalog_id,
        "issue_type": issue_type,
        "priority": priority,
        "evidence": evidence,
        "suggested_action": suggested_action,
        "review_status": "pending",
    })
    issues_by_record[catalog_id].add(issue_type)

# --------------------------------------------------
# 3. Prepare a derived copy; preserve all source fields
# --------------------------------------------------

for position, source_row in enumerate(source_rows, start=1):
    record_id = f"exercise_clean:{position:06d}"
    row = dict(source_row)
    row.update({
        "catalog_id": record_id,
        "source_cleaned_record_position": str(position),
        "source_file": source["relative_path"],
        "source_sha256": source_hash,
        "label_review_status": "original_source_verification_pending",
    })

    for field in fields:
        original = source_row[field]
        prepared = prepared_value(field, original)
        row[f"{field}_prepared"] = prepared

        if original != prepared:
            cleaning_log.append({
                "catalog_id": record_id,
                "field": field,
                "source_value": original,
                "prepared_value": prepared,
                "reason": "Whitespace/category normalization or null-like marker",
                "decision": "Prepared copy only; source preserved",
            })

    name = row["exercise_name_prepared"]
    description = row["description_prepared"]
    label = row["body_part_prepared"]
    key = comparison_key(name)

    row["name_comparison_key"] = key
    row["preliminary_classifier_candidate"] = "yes" if name and label else "no"

    if not name:
        add_issue(record_id, "missing_exercise_name", "high",
                  "No usable exercise title", "Review source before classifier inclusion")
    if not description:
        add_issue(record_id, "missing_description", "medium",
                  "Exercise description unavailable",
                  "Retain title-only candidate; do not invent text")
    if not label:
        add_issue(record_id, "missing_body_part", "high",
                  "Body-part label unavailable",
                  "Exclude from supervised targets pending review")

    for field in ("equipment", "difficulty_level"):
        if not row[f"{field}_prepared"]:
            add_issue(record_id, f"missing_{field}", "medium",
                      f"{field} unavailable",
                      "Preserve uncertainty and review before planning")

    if label:
        labels[label] += 1
    if key:
        by_name[key].append(record_id)
        signature = tuple(row[f"{field}_prepared"] for field in fields)
        by_content[signature].append(record_id)

    catalog.append(row)

# --------------------------------------------------
# 4. Screen repeated names, label conflicts, duplicates
# --------------------------------------------------

by_id = {r["catalog_id"]: r for r in catalog}

for key, record_ids in by_name.items():
    if len(record_ids) <= 1:
        continue

    distinct_labels = {by_id[rid]["body_part_prepared"] for rid in record_ids}
    for rid in record_ids:
        add_issue(rid, "repeated_normalized_name", "medium",
                  f"Name group contains {len(record_ids)} records",
                  "Review variations and preserve grouping evidence")

    if len(distinct_labels) > 1:
        for rid in record_ids:
            add_issue(rid, "conflicting_labels_for_name", "high",
                      f"Labels: {sorted(distinct_labels)}",
                      "Verify against original sources and team review")

for signature, record_ids in by_content.items():
    if len(record_ids) > 1:
        for rid in record_ids:
            add_issue(rid, "duplicate_meaningful_content", "medium",
                      f"Content group contains {len(record_ids)} records",
                      "Review before removal or consolidation")

LOW_SUPPORT_ALERT = 10  # Review threshold, not automatic exclusion.
for row in catalog:
    rid = row["catalog_id"]
    label = row["body_part_prepared"]
    if label and labels[label] < LOW_SUPPORT_ALERT:
        add_issue(rid, "low_support_label", "medium",
                  f"{label}: {labels[label]} examples",
                  "Review class eligibility in Step 15")
    row["review_flags"] = ";".join(sorted(issues_by_record[rid]))

# --------------------------------------------------
# 5. Verify existing outputs without erasing prior reviews
# --------------------------------------------------

catalog_path = root / "data/processed/exercise_catalog.csv"
log_path = root / "results/exercise_cleaning_log.csv"
queue_path = root / "results/exercise_review_queue.csv"

outputs = [
    (catalog_path, catalog, original_columns + added_columns, {"label_review_status"}),
    (log_path, cleaning_log, log_columns, set()),
    (queue_path, review_queue, queue_columns, {"review_status"}),
]


def compare_existing(path, generated, expected_columns, editable_columns):
    with path.open(newline="", encoding="utf-8-sig") as f:
        reader = csv.DictReader(f)
        existing_columns = list(reader.fieldnames or [])
        existing = list(reader)

    if not set(expected_columns).issubset(existing_columns):
        missing_cols = sorted(set(expected_columns) - set(existing_columns))
        return False, f"Missing output columns: {missing_cols}"
    if len(existing) != len(generated):
        return False, f"Row count differs: existing={len(existing)}, expected={len(generated)}"

    # Compare by row and field, preserving reviewer-editable statuses.
    for index, (actual, expected) in enumerate(zip(existing, generated), start=1):
        for column in expected_columns:
            if column in editable_columns:
                continue
            if actual.get(column, "") != str(expected.get(column, "")):
                return False, (
                    f"Difference at row {index}, field '{column}': "
                    f"existing={actual.get(column, '')!r}; "
                    f"recomputed={expected.get(column, '')!r}"
                )
    return True, "Recomputed records match; existing reviewer statuses preserved."


print("\n[1] VERIFYING STEP 10 OUTPUTS")
status = []
problems = []
missing_outputs = []

for path, generated, columns, editable in outputs:
    if path.exists():
        ok, note = compare_existing(path, generated, columns, editable)
        status.append((path, "VERIFIED" if ok else "REVIEW REQUIRED"))
        print(f"{'PASS' if ok else 'REVIEW'} | {path.relative_to(root)} | {note}")
        if not ok:
            problems.append((path, note))
    else:
        missing_outputs.append((path, generated, columns))

# Do not change ANY existing file, even if it differs from this re-computation.
# Also do not create partial new outputs if another output requires reconciliation.
if problems:
    print("\nExisting project outputs were preserved unchanged.")
    raise RuntimeError(
        "Step 10 existing outputs differ from the computed results. "
        "Inspect the REVIEW entries above before changing any files."
    )

for path, generated, columns in missing_outputs:
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("x", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=columns)
        writer.writeheader()
        writer.writerows(generated)
    status.append((path, "CREATED"))
    print(f"CREATED | {path.relative_to(root)} | {len(generated)} records")

# --------------------------------------------------
# 6. Final review and quality summary
# --------------------------------------------------

if len(catalog) != len(source_rows) or len(by_id) != len(catalog):
    raise RuntimeError("Record preservation or catalog ID uniqueness failed.")
if any(r["body_part"] != r["body_part_prepared"] for r in catalog):
    print("NOTE: Source and prepared label formatting differ; original labels preserved.")

issue_counts = Counter(r["issue_type"] for r in review_queue)
flagged_records = sum(bool(row["review_flags"]) for row in catalog)

print("\n[2] PREPARATION SUMMARY")
print("Source records:", len(source_rows))
print("Catalog records:", len(catalog))
print("Catalog columns:", len(original_columns + added_columns))
print("Preliminary classifier candidates:", sum(
    r["preliminary_classifier_candidate"] == "yes" for r in catalog
))
print("Formatting differences logged:", len(cleaning_log))
print("Records with review flags:", flagged_records)
print("Review queue entries:", len(review_queue))
print("Distinct body-part labels:", len(labels))
print("Repeated-name groups:", sum(len(v) > 1 for v in by_name.values()))
print("Duplicate-content groups:", sum(len(v) > 1 for v in by_content.values()))

print("\n[3] REVIEW QUEUE ISSUE COUNTS")
for name, count in issue_counts.most_common():
    print(f"  {name:<35} {count:>5}")

print("\n[4] BODY-PART LABEL DISTRIBUTION")
for name, count in labels.most_common():
    print(f"  {name:<22} {count:>5}")

print("\n[5] FINAL INTEGRITY AND PRESERVATION CHECK")
print("Exercise source SHA-256 unchanged:", sha256_file(src) == source_hash)
print("Every cleaned source row retained:", len(source_rows) == len(catalog))
print("Unique catalog identifiers:", len(by_id) == len(catalog))
print("Original body-part labels reassigned: No")
print("Existing Step 10 output files overwritten: No")
print("Git commits or pushes performed: No")
print("OVERALL: PASS - Step 10 verified, outstanding reviews tracked")
print("=" * 68)
```

```python
# STEP 11 — Integrated Exploratory Data Analysis and Data Summary
# One notebook cell. Preserves the eight original exercise EDA tables and four figures.
# New derived reports use safe, versioned updates. Does not modify source datasets.

import csv
import hashlib
import io
import json
import os
import re
import shutil
import tempfile
import unicodedata
from collections import Counter
from datetime import datetime
from difflib import SequenceMatcher
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

SHOW_FIGURES = True

root = next(
    (p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (p / '.git').exists()), None
)
if root is None:
    raise FileNotFoundError('Project Git root not found; open your VS Code project.')

eda_dir = root / 'results' / 'eda'
fig_dir = root / 'results' / 'figures'
backup_dir = root / 'results' / 'backups' / 'step11'
manifest_path = root / 'data' / 'source_manifest.csv'
if not manifest_path.is_file():
    raise FileNotFoundError('Step 8 manifest missing. Run Steps 8–10 first.')
manifest = pd.read_csv(manifest_path, dtype=str, keep_default_na=False)
source_map = manifest.set_index('source_id', verify_integrity=True)

sources = {
    'local_exercise_catalog': 'exercise',
    'aditi_jr_nba_drills': 'drills',
    'aditi_jr_nba_rag': 'rag',
    'catalog_basketball_v02': 'basketball_catalog',
    'aditi_usda_foundation_nutrition': 'nutrition',
    'aditi_synthetic_athletes': 'athletes',
    'aditi_recovery_guidance': 'recovery_guidance',
    'recovery_rules_v01': 'recovery_rules',
}


def digest(path):
    h = hashlib.sha256()
    with path.open('rb') as stream:
        for block in iter(lambda: stream.read(1024 * 1024), b''):
            h.update(block)
    return h.hexdigest()


def read_source(source_id):
    if source_id not in source_map.index:
        raise ValueError(f'Missing manifest source: {source_id}')
    row = source_map.loc[source_id]
    rel = row['relative_path']
    path = (root / rel).resolve()
    if not path.is_relative_to(root.resolve()) or not path.is_file():
        raise FileNotFoundError(f'Unsafe or unavailable input: {rel}')
    if digest(path).lower() != row['sha256'].lower():
        raise ValueError(f'SHA-256 mismatch; rerun Step 8 only after review: {rel}')
    if path.suffix.lower() == '.csv':
        df = pd.read_csv(path, dtype=str, keep_default_na=False)
        if df.columns.duplicated().any():
            raise ValueError(f'Duplicate field names: {rel}')
        expected = row.get('reported_records', '')
        if expected and len(df) != int(expected):
            raise ValueError(f'Row count differs from Step 8: {rel}')
        return df
    if path.suffix.lower() == '.json':
        with path.open(encoding='utf-8-sig') as f:
            data = json.load(f)
        if not isinstance(data, list):
            raise ValueError(f'Expected JSON list: {rel}')
        expected = row.get('reported_records', '')
        if expected and len(data) != int(expected):
            raise ValueError(f'JSON count differs from Step 8: {rel}')
        return data
    raise ValueError(f'Unsupported data file: {rel}')


def assert_columns(df, required, name):
    absent = sorted(set(required) - set(df.columns))
    if absent:
        raise ValueError(f'{name}: missing fields {absent}')


def normalize_name(value):
    s = unicodedata.normalize('NFKC', str(value)).casefold()
    return ' '.join(re.findall(r'[a-z0-9]+', s))


def nvalue(s):
    return pd.to_numeric(s.replace('', np.nan), errors='coerce')


print('=' * 72)
print('CHAMPIONS ATHLETIC MAP — STEP 11 INTEGRATED EDA')
print('=' * 72)
print('Project:', root)
print('Manifest entries:', len(manifest))

D = {short: read_source(sid) for sid, short in sources.items()}
exercise, drills, catalog_v02 = D['exercise'], D['drills'], D['basketball_catalog']
nutrition, athletes = D['nutrition'], D['athletes']
recovery_guidance, recovery_rules, rag = D['recovery_guidance'], D['recovery_rules'], D['rag']

queue_path = root / 'results/exercise_review_queue.csv'
log_path = root / 'results/exercise_cleaning_log.csv'
for p in (queue_path, log_path):
    if not p.is_file():
        raise FileNotFoundError(f'Existing Step 10 artifact missing: {p}')
queue = pd.read_csv(queue_path, dtype=str, keep_default_na=False)
changes = pd.read_csv(log_path, dtype=str, keep_default_na=False)

assert_columns(exercise, ['catalog_id', 'body_part_prepared', 'exercise_name_prepared',
                          'description_prepared', 'equipment_prepared', 'difficulty_level_prepared',
                          'preliminary_classifier_candidate', 'source_sha256'], 'Exercise')
assert_columns(drills, ['drill_id', 'drill_name', 'skill_category', 'curriculum_level',
                        'group_requirement', 'equipment', 'source_document'], 'JrNBA')
assert_columns(catalog_v02, ['catalog_candidate_id', 'drill_name', 'scope_candidate',
                             'catalog_approval_status', 'planner_eligible'], 'Basketball catalog')
assert_columns(nutrition, ['fdc_id', 'description', 'protein_g', 'fat_g', 'carbohydrate_g',
                           'fiber_g', 'calories_kcal', 'energy_general_kcal',
                           'energy_specific_kcal', 'energy_method'], 'Nutrition')
assert_columns(athletes, ['athlete_id', 'age', 'position', 'experience_level',
                          'primary_goal', 'available_days_per_week', 'equipment',
                          'injury_status', 'scenario_type', 'is_synthetic'], 'Athletes')
assert_columns(recovery_rules, ['rule_id', 'metric', 'segment', 'rule_type', 'source_url',
                                'value_min', 'value_max', 'unit'], 'Recovery rules')
assert_columns(recovery_guidance, ['topic', 'source', 'source_url'], 'Recovery guidance')
if not all(isinstance(x, dict) and x.get('drill_id') and x.get('text') for x in rag):
    raise ValueError('RAG records must contain nonempty drill_id and text.')
if set(drills['drill_id']) != {x['drill_id'] for x in rag}:
    raise ValueError('RAG document IDs do not match Jr. NBA drill IDs.')
if not exercise['catalog_id'].is_unique or not athletes['athlete_id'].is_unique:
    raise ValueError('Duplicate exercise or athlete identifiers detected.')
if not nutrition['fdc_id'].is_unique or not recovery_rules['rule_id'].is_unique:
    raise ValueError('Duplicate food or recovery rule identifiers detected.')
if not catalog_v02['catalog_candidate_id'].is_unique:
    raise ValueError('Duplicate basketball catalog candidate identifiers.')
if set(exercise['source_sha256']) != {source_map.loc['derived_exercise', 'sha256']}:
    raise ValueError('Exercise catalog does not match original cleaned source lineage.')
if catalog_v02['planner_eligible'].str.lower().eq('yes').any():
    raise ValueError('Review-only catalog contains planner-eligible drills; stop.')

print('\n[1] INPUTS VERIFIED AGAINST STEP 8')
for short, sid in ((v, k) for k, v in sources.items()):
    print(f'PASS | {short:<19} | {len(D[short]):>5} records')
print('PASS | drill-to-RAG source linkage')

# ----------------------------------------------------------------------
# 1. Original exercise EDA: reproduce calculations, preserve prior outputs
# ----------------------------------------------------------------------

ex = exercise.copy()
ex['body_part'] = ex['body_part_prepared']
ex['description_missing'] = ex['description_prepared'].str.strip().eq('')
ex['equipment_missing'] = ex['equipment_prepared'].str.strip().eq('')
ex['name_words'] = ex['exercise_name_prepared'].str.findall(r"\b[\w'-]+\b").str.len()
ex['description_words'] = ex['description_prepared'].str.findall(r"\b[\w'-]+\b").str.len()
n = len(ex)

classes = ex['body_part'].value_counts().rename_axis('body_part').reset_index(name='records')
classes['percent'] = (100 * classes['records'] / n).round(2)
coverage = ex.groupby('body_part', as_index=False).agg(
    records=('catalog_id', 'size'), missing_descriptions=('description_missing', 'sum'))
coverage['available_descriptions'] = coverage['records'] - coverage['missing_descriptions']
coverage['missing_percent'] = (
    coverage['missing_descriptions'] / coverage['records'] * 100).round(2)
coverage = coverage.sort_values('missing_percent', ascending=False)


def freq(series, key, population=None):
    population = len(series) if population is None else population
    table = series.replace('', '[missing/unknown]').value_counts().rename_axis(key).reset_index(name='records')
    table['percent'] = (table['records'] / population * 100).round(2) if population else 0
    return table


def word_stats(population, series):
    return dict(population=population, records=len(series),
                mean_words=round(series.mean(), 2), median_words=float(series.median()),
                p25_words=float(series.quantile(.25)), p75_words=float(series.quantile(.75)),
                p95_words=float(series.quantile(.95)), maximum_words=int(series.max()))

text_stats = pd.DataFrame([
    word_stats('exercise_name_all', ex['name_words']),
    word_stats('description_all_including_missing', ex['description_words']),
    word_stats('description_available_only', ex.loc[~ex['description_missing'], 'description_words']),
])
issues = queue['issue_type'].value_counts().rename_axis('issue_type').reset_index(name='queue_entries')
preparation = pd.DataFrame([
    ('catalog_records', n),
    ('preliminary_classifier_candidates', int(exercise['preliminary_classifier_candidate'].eq('yes').sum())),
    ('formatting_changes', len(changes)),
    ('flagged_distinct_records', queue['catalog_id'].nunique()),
    ('review_queue_entries', len(queue)),
    ('body_part_categories', len(classes)),
    ('missing_descriptions', int(ex['description_missing'].sum())),
    ('missing_equipment', int(ex['equipment_missing'].sum())),
], columns=['measure', 'count'])
findings_original = pd.DataFrame([
    ('EDA01', 'missing_descriptions', int(ex['description_missing'].sum()),
     'exercise_description_coverage.csv', 'Investigate description coverage by class; retain title-only candidates.'),
    ('EDA02', 'largest_to_smallest_class_ratio',
     round(classes['records'].max() / classes['records'].min(), 2),
     'exercise_body_part_distribution.csv', 'Review small-class eligibility and grouped splitting in Step 15.'),
    ('EDA03', 'records_with_review_flags', queue['catalog_id'].nunique(),
     'exercise_review_issues.csv', 'Track unresolved review flags; do not automatically exclude records.'),
    ('EDA04', 'formatting_changes', len(changes),
     'exercise_preparation_summary.csv', 'Preserve the preparation log and source lineage.'),
], columns=['finding_id', 'metric', 'value', 'evidence', 'follow_up'])

legacy_tables = {
    'exercise_preparation_summary.csv': preparation,
    'exercise_body_part_distribution.csv': classes,
    'exercise_description_coverage.csv': coverage,
    'exercise_text_length_statistics.csv': text_stats,
    'exercise_equipment_distribution.csv': freq(ex['equipment_prepared'], 'equipment'),
    'exercise_difficulty_distribution.csv': freq(ex['difficulty_level_prepared'], 'difficulty'),
    'exercise_review_issues.csv': issues,
    'exercise_eda_findings.csv': findings_original,
}

# Existing exercise outputs must agree with the new computation; don't overwrite.
for name, generated in legacy_tables.items():
    existing = eda_dir / name
    if existing.exists():
        old = pd.read_csv(existing)
        a = old.sort_values(list(old.columns)).reset_index(drop=True)
        b = generated.sort_values(list(generated.columns)).reset_index(drop=True)
        if list(a.columns) != list(b.columns) or len(a) != len(b):
            raise RuntimeError(f'Original EDA table schema/count differs: {name}')
        # Compare values rather than CSV formatting; float tolerance handles rounding.
        for c in a.columns:
            av, bv = pd.to_numeric(a[c], errors='coerce'), pd.to_numeric(b[c], errors='coerce')
            if av.notna().all() and bv.notna().all():
                match = np.allclose(av, bv, atol=.011, rtol=0)
            else:
                match = a[c].fillna('').astype(str).equals(b[c].fillna('').astype(str))
            if not match:
                raise RuntimeError(f'Original EDA table differs in {name}/{c}; review manually.')

# ----------------------------------------------------------------------
# 2. Statistical association helpers
# ----------------------------------------------------------------------

def cramers_v(x, y):
    tab = pd.crosstab(x, y)
    r, c = tab.shape
    total = int(tab.to_numpy().sum())
    if total < 4 or r < 2 or c < 2:
        return np.nan, total, 0.0
    obs = tab.to_numpy(dtype=float)
    expected = obs.sum(axis=1, keepdims=True) @ obs.sum(axis=0, keepdims=True) / total
    chisq = np.sum((obs - expected) ** 2 / expected)
    phi2 = chisq / total
    corrected = max(0, phi2 - (c - 1) * (r - 1) / (total - 1))
    rc = r - (r - 1) ** 2 / (total - 1)
    cc = c - (c - 1) ** 2 / (total - 1)
    denom = min(rc - 1, cc - 1)
    return (np.sqrt(corrected / denom) if denom > 0 else np.nan,
            total, round(float((expected < 5).mean() * 100), 2))

assoc_rows = []
def categorical_assoc(domain, label, x, y, note):
    strength, count, sparse = cramers_v(x, y)
    assoc_rows.append(dict(domain=domain, relationship=label, method='bias-corrected Cramers V',
                           effect_size=round(float(strength), 4) if pd.notna(strength) else '',
                           observations=count, expected_cells_below_5_pct=sparse, caveat=note))

categorical_assoc('exercise', 'body-part vs description available', ex['body_part'],
                  ~ex['description_missing'], 'Descriptive association; not classifier performance.')
categorical_assoc('exercise', 'body-part vs equipment', ex['body_part'],
                  ex['equipment_prepared'].replace('', '[missing]'),
                  'Review sparse equipment categories before inferring patterns.')
categorical_assoc('exercise', 'body-part vs difficulty', ex['body_part'],
                  ex['difficulty_level_prepared'].replace('', '[missing]'),
                  'Metadata categories may be incomplete.')

ex_groups = ex.groupby('body_part').agg(
    records=('catalog_id', 'size'), mean_name_words=('name_words', 'mean'),
    median_name_words=('name_words', 'median'),
    mean_description_words_including_missing=('description_words', 'mean'),
    missing_description_pct=('description_missing', 'mean')).reset_index()
ex_groups['missing_description_pct'] = (100 * ex_groups['missing_description_pct']).round(2)
ex_groups[['mean_name_words', 'mean_description_words_including_missing']] = (
    ex_groups[['mean_name_words', 'mean_description_words_including_missing']].round(2))

# ----------------------------------------------------------------------
# 3. Basketball activity, overlap and RAG coverage
# ----------------------------------------------------------------------

skill_freq = freq(drills['skill_category'], 'skill_category', len(drills))
source_freq = freq(drills['source_document'], 'source_document', len(drills))
level_freq = freq(drills['curriculum_level'], 'curriculum_level', len(drills))
participant_freq = freq(drills['group_requirement'], 'group_requirement', len(drills))

name_index = {}
for s in catalog_v02['drill_name']:
    name_index.setdefault(normalize_name(s), []).append(s)
overlap_rows = []
for name in sorted(drills['drill_name'].unique()):
    key = normalize_name(name)
    exact = name_index.get(key, [])
    nearest = sorted(((SequenceMatcher(None, key, k).ratio(), k)
                      for k in name_index if k != key), reverse=True)[:1]
    near_score, near_key = nearest[0] if nearest else (0.0, '')
    overlap_rows.append(dict(imported_drill=name,
                             imported_source_occurrences=int(drills['drill_name'].eq(name).sum()),
                             exact_catalog_match='yes' if exact else 'no',
                             matching_catalog_names='; '.join(exact),
                             nearest_catalog_name='; '.join(name_index.get(near_key, [])) if near_score >= .80 else '',
                             nearest_similarity=round(near_score, 3),
                             possible_near_match_review='yes' if not exact and near_score >= .88 else 'no'))
overlap_table = pd.DataFrame(overlap_rows)

cat_status = []
for col in ('scope_candidate', 'metadata_status', 'catalog_approval_status', 'planner_eligible', 'source_record_class'):
    if col in catalog_v02.columns:
        for val, count in catalog_v02[col].value_counts(dropna=False).items():
            cat_status.append(dict(field=col, value=val, records=int(count)))
cat_status = pd.DataFrame(cat_status)

rag_ids = {str(d['drill_id']) for d in rag}
rag_summary = pd.DataFrame([
    dict(drill_id=d['drill_id'], found_in_drill_master=d['drill_id'] in set(drills['drill_id']),
         text_characters=len(d['text']), text_words=len(d['text'].split()),
         source_text_present='Source:' in d['text']) for d in rag
])

categorical_assoc('basketball', 'skill category vs participant requirements',
                  drills['skill_category'], drills['group_requirement'],
                  'Small sample and sparse cells; exploratory only.')

# ----------------------------------------------------------------------
# 4. USDA food completeness, distributions, correlations, anomalies
# ----------------------------------------------------------------------

nut_cols = ['protein_g', 'fat_g', 'carbohydrate_g', 'fiber_g',
            'energy_general_kcal', 'energy_specific_kcal', 'calories_kcal']
num_food = nutrition[['fdc_id', 'description', 'energy_method']].copy()
for c in nut_cols:
    num_food[c] = nvalue(nutrition[c])

nutrition_completeness = pd.DataFrame([
    dict(nutrient=c, total_foods=len(num_food), nonmissing=int(num_food[c].notna().sum()),
         missing=int(num_food[c].isna().sum()),
         missing_pct=round(100 * num_food[c].isna().mean(), 2)) for c in nut_cols
])
core = ['protein_g', 'fat_g', 'carbohydrate_g', 'calories_kcal']
complete_core = int(num_food[core].notna().all(axis=1).sum())
neg_carbs = num_food['carbohydrate_g'].lt(0)

nutrition_stats = pd.DataFrame([
    dict(nutrient=c, n=int(num_food[c].count()), mean=round(num_food[c].mean(), 3),
         median=round(num_food[c].median(), 3), std=round(num_food[c].std(), 3),
         minimum=round(num_food[c].min(), 3), p25=round(num_food[c].quantile(.25), 3),
         p75=round(num_food[c].quantile(.75), 3), maximum=round(num_food[c].max(), 3))
    for c in nut_cols
])

anomaly_rows = []
for _, r in num_food.loc[neg_carbs].iterrows():
    anomaly_rows.append(dict(fdc_id=r['fdc_id'], food=r['description'],
                             field='carbohydrate_g', observed_value=r['carbohydrate_g'],
                             concern='negative measured/calculated carbohydrate',
                             action='verify original USDA nutrient definition and value'))
nutrition_anomalies = pd.DataFrame(anomaly_rows, columns=[
    'fdc_id', 'food', 'field', 'observed_value', 'concern', 'action'])

# Correlations use pairwise available, nonnegative quantities, and document exclusions.
food_clean = num_food.copy()
for c in nut_cols:
    food_clean.loc[food_clean[c] < 0, c] = np.nan
pairs = []
for i, a in enumerate(nut_cols):
    for b in nut_cols[i+1:]:
        valid = food_clean[[a, b]].dropna()
        count = len(valid)
        for method in ('pearson', 'spearman'):
            coeff = valid[a].corr(valid[b], method=method) if count >= 10 else np.nan
            pairs.append(dict(first_measure=a, second_measure=b, method=method,
                              valid_pairs=count, correlation=round(float(coeff), 4)
                              if pd.notna(coeff) else '',
                              excluded_from_complete_pairs=len(food_clean)-count,
                              interpretation_note='Association only; energy partly calculated from nutrients.'))
nutrition_correlations = pd.DataFrame(pairs)
energy_pair = num_food[['energy_general_kcal', 'energy_specific_kcal']].dropna().copy()
energy_pair['difference_specific_minus_general'] = (
    energy_pair['energy_specific_kcal'] - energy_pair['energy_general_kcal'])
energy_methods = freq(nutrition['energy_method'], 'energy_method', len(nutrition))
energy_comparison = pd.DataFrame([
    dict(metric='both_energy_methods_present', value=len(energy_pair)),
    dict(metric='mean_absolute_kcal_difference', value=round(energy_pair['difference_specific_minus_general'].abs().mean(), 3)),
    dict(metric='median_absolute_kcal_difference', value=round(energy_pair['difference_specific_minus_general'].abs().median(), 3)),
    dict(metric='differences_over_40_kcal', value=int(energy_pair['difference_specific_minus_general'].abs().gt(40).sum())),
])

# ----------------------------------------------------------------------
# 5. Synthetic athlete scenario coverage (not population inference)
# ----------------------------------------------------------------------

athlete_distributions = []
for field in ('age', 'position', 'experience_level', 'primary_goal',
              'available_days_per_week', 'equipment', 'injury_status', 'scenario_type'):
    for value, count in athletes[field].value_counts().items():
        athlete_distributions.append(dict(variable=field, category=value, records=int(count),
                                          percent=round(count * 100 / len(athletes), 2)))
athlete_distributions = pd.DataFrame(athlete_distributions)
athlete_goal_by_level = pd.crosstab(athletes['experience_level'], athletes['primary_goal'])
athlete_goal_by_level.index.name = 'experience_level'
athlete_goal_by_level = athlete_goal_by_level.reset_index()
athlete_days_by_level = pd.crosstab(athletes['experience_level'], athletes['available_days_per_week'])
athlete_days_by_level.index.name = 'experience_level'
athlete_days_by_level = athlete_days_by_level.reset_index()

categorical_assoc('synthetic_profiles', 'experience level vs available days',
                  athletes['experience_level'], athletes['available_days_per_week'],
                  'Generator-created associations are not real athlete evidence.')
categorical_assoc('synthetic_profiles', 'experience level vs primary goal',
                  athletes['experience_level'], athletes['primary_goal'],
                  'Generated sample, sparse contingency table; use for scenario coverage only.')

# ----------------------------------------------------------------------
# 6. Recovery source, segments, boundaries and rule applicability
# ----------------------------------------------------------------------

recovery_coverage = recovery_rules.groupby(
    ['segment', 'metric', 'rule_type', 'unit'], dropna=False).size().rename('rule_count').reset_index()
recovery_coverage['source_urls_present'] = recovery_coverage.apply(
    lambda r: int(recovery_rules.loc[
        (recovery_rules['segment'] == r['segment']) &
        (recovery_rules['metric'] == r['metric']) &
        (recovery_rules['rule_type'] == r['rule_type']), 'source_url'
    ].str.startswith('https://').sum()), axis=1)
recovery_topics = freq(recovery_guidance['topic'], 'topic', len(recovery_guidance))
recovery_type_freq = freq(recovery_rules['rule_type'], 'rule_type', len(recovery_rules))

assoc = pd.DataFrame(assoc_rows)

# ----------------------------------------------------------------------
# 7. Plain-language variable dictionary and documented model implications
# ----------------------------------------------------------------------

dictionary_spec = [
    ('exercise', 'exercise_name_prepared', 'Exercise name', 'Text describing the movement', 'text', 'classifier_input', 'May carry anatomical words naturally'),
    ('exercise', 'description_prepared', 'Exercise instructions', 'Written explanation of movement', 'text', 'classifier_input', 'Missing for many exercises; title-only possible'),
    ('exercise', 'body_part_prepared', 'Targeted body part', 'Primary source-assigned anatomical category', 'categorical', 'prediction_target', 'Never include as text input'),
    ('exercise', 'equipment_prepared', 'Exercise equipment', 'Equipment requirements when documented', 'categorical', 'catalog_metadata', 'Missing does not mean no equipment'),
    ('exercise', 'difficulty_level_prepared', 'Difficulty', 'Source difficulty category', 'categorical', 'catalog_metadata', 'Not youth safety clearance'),
    ('exercise', 'catalog_id', 'Exercise ID', 'Stable identifier for traceability', 'identifier', 'audit_only', 'Not predictive'),
    ('basketball', 'drill_name', 'Drill name', 'Name of basketball coaching activity', 'text', 'retrieval', 'Duplicate or variation review pending'),
    ('basketball', 'skill_category', 'Skill area', 'Basketball technique or skill category', 'categorical', 'retrieval_filter', 'Proposed grouping requires checking'),
    ('basketball', 'curriculum_level', 'Curriculum level', 'Jr NBA instructional level', 'categorical', 'retrieval_filter', 'Not independently age-appropriate'),
    ('basketball', 'equipment', 'Drill equipment', 'Resources required to conduct drill', 'categorical', 'planning_constraint', 'May be incomplete'),
    ('basketball', 'group_requirement', 'Participants', 'Individual/partner/group requirement', 'categorical', 'planning_constraint', 'Solo adaptations are proposals, not approved'),
    ('basketball', 'duration', 'Prescription', 'Minutes, shots, repetitions or game targets', 'mixed text', 'session_planning', 'Cannot convert all durations to minutes'),
    ('basketball', 'drill_id', 'Drill ID', 'Source-occurrence identifier', 'identifier', 'traceability', 'Do not use as continuous variable'),
    ('nutrition', 'description', 'Food description', 'Name and state of food', 'text', 'nutrition_lookup', 'Distinct FDC IDs may share name'),
    ('nutrition', 'fdc_id', 'FoodData Central ID', 'Publisher record identifier', 'identifier', 'source_key', 'Never numeric feature'),
    ('nutrition', 'protein_g', 'Protein', 'Protein quantity per source reference amount', 'numeric', 'nutrition_reference', 'Units/reference amount must be verified'),
    ('nutrition', 'fat_g', 'Fat', 'Fat quantity per source reference amount', 'numeric', 'nutrition_reference', 'Missing values retained'),
    ('nutrition', 'carbohydrate_g', 'Carbohydrates', 'Carbohydrate quantity per source reference amount', 'numeric', 'nutrition_reference', 'Negative values require source review'),
    ('nutrition', 'fiber_g', 'Dietary fiber', 'Fiber quantity per source reference amount', 'numeric', 'nutrition_reference', 'Substantial missingness'),
    ('nutrition', 'calories_kcal', 'Food energy', 'Selected kilocalorie value from source', 'numeric', 'nutrition_reference', 'Can be derived from macronutrients'),
    ('nutrition', 'energy_method', 'Energy calculation method', 'Formula convention for published energy', 'categorical', 'measurement_metadata', 'General/specific factor distinction'),
    ('athlete', 'age', 'Athlete age', 'Fictional athlete age in years', 'numeric', 'planning_constraint', 'Synthetic, not real outcomes'),
    ('athlete', 'position', 'Playing position', 'Basketball position of fictional athlete', 'categorical', 'planning_context', 'Synthetic'),
    ('athlete', 'experience_level', 'Experience', 'Skill-experience category', 'categorical', 'planning_constraint', 'Does not establish medical clearance'),
    ('athlete', 'primary_goal', 'Development goal', 'Basketball area athlete wants to improve', 'categorical', 'planning_goal', 'Synthetic'),
    ('athlete', 'available_days_per_week', 'Available training days', 'Number of days available weekly', 'integer', 'planning_constraint', 'Not exact day availability'),
    ('athlete', 'equipment', 'Accessible equipment', 'Reported training resources', 'categorical', 'planning_constraint', 'Do not assume unspecified equipment'),
    ('athlete', 'injury_status', 'Reported injury information', 'Self-reported/unknown injury status', 'categorical', 'safety_review', 'Unknown does not mean injury-free'),
    ('athlete', 'scenario_type', 'Evaluation scenario', 'Deliberately constructed planning situation', 'categorical', 'evaluation_design', 'Artificial prevalence'),
    ('recovery', 'segment', 'Guideline group', 'Age or grade population to which source refers', 'categorical', 'rule_applicability', 'Age/grade overlap may need review'),
    ('recovery', 'metric', 'Participation or recovery measure', 'Quantity controlled or recommended by rule', 'categorical', 'planning_constraint', 'Not a measured recovery observation'),
    ('recovery', 'rule_type', 'Rule direction', 'Minimum, maximum, or recommended range', 'categorical', 'constraint_logic', 'Validate before enforcing'),
    ('recovery', 'value_min', 'Lower bound', 'Minimum value if rule requires one', 'numeric', 'constraint_logic', 'Blank can mean inapplicable'),
    ('recovery', 'value_max', 'Upper bound', 'Maximum value if rule requires one', 'numeric', 'constraint_logic', 'Blank can mean inapplicable'),
]
data_dictionary = pd.DataFrame(dictionary_spec, columns=[
    'domain', 'source_field', 'plain_language_name', 'plain_language_definition',
    'variable_type', 'model_or_system_role', 'limitations_or_cautions'])

model_recommendations = pd.DataFrame([
    ('exercise_text', 'description missingness', int(ex['description_missing'].sum()),
     'Compare name-only vs name-plus-description on development split; no fabricated text'),
    ('exercise_labels', 'largest/smallest class ratio', round(classes['records'].max()/classes['records'].min(), 2),
     'Use grouped splits, macro-F1 and per-class recall; evaluate class weighting later'),
    ('exercise_review', 'flagged distinct catalog records', queue['catalog_id'].nunique(),
     'Preserve review queue and label provenance; do not auto-relabel'),
    ('basketball', 'exact normalized-name overlap', int(overlap_table['exact_catalog_match'].eq('yes').sum()),
     'Review source occurrence/variation before canonical catalog merge'),
    ('nutrition', 'incomplete core nutrition records', len(nutrition)-complete_core,
     'Require completeness checks for numerical nutrition comparisons'),
    ('nutrition', 'negative carbohydrate records', int(neg_carbs.sum()),
     'Source-check rather than silently zeroing; excluded from analysis-only correlations'),
    ('athlete', 'synthetic evaluation profiles', len(athletes),
     'Assess scenario coverage; never claim real-world prevalence'),
    ('recovery', 'structured rule records', len(recovery_rules),
     'Review rule applicability before activating planner rules'),
], columns=['domain', 'finding', 'observed_value', 'implication_and_next_action'])

quality_policy = pd.DataFrame([
    ('exercise', 'Descriptions missing', int(ex['description_missing'].sum()),
     'Cause unverified: source documentation or prior preparation may be incomplete',
     'Preserved title-only records and source labels',
     'Validate expected text coverage; record missingness by class; compare title-only model on development data'),
    ('exercise', 'Class imbalance', int(classes['records'].max() - classes['records'].min()),
     'Observed sampling/coverage imbalance; origin not established',
     'Preserved all original classes',
     'Grouped splitting; macro-F1 and per-class recall; evaluate class weighting on training data'),
    ('nutrition', 'Missing core measurements', len(nutrition) - complete_core,
     'Cause unverified: source nutrient availability and/or extraction pipeline',
     'Preserved missing values; identified complete-case subset',
     'Check required nutrient IDs, coverage thresholds, and reject incomplete numeric calculations'),
    ('nutrition', 'Negative carbohydrate measurements', int(neg_carbs.sum()),
     'Cause unverified: raw USDA nutrient or calculation conventions',
     'Preserved source values; excluded negative measurements only from exploratory correlation pairs',
     'Trace original FDC values and calculation method before introducing correction rules'),
    ('basketball', 'Exact normalized drill-name overlaps',
     int(overlap_table['exact_catalog_match'].eq('yes').sum()),
     'Multiple practice plans or duplicate/variation names; exact identity unverified',
     'Kept source occurrences and review-only catalog separate',
     'Source-aware canonicalization, human review, and permission/eligibility gates'),
    ('synthetic_profiles', 'Optional scenario/request blanks',
     int(athletes.reindex(columns=['scenario_note', 'athlete_request'], fill_value='').eq('').sum().sum()),
     'Optional fields or scenario generator design; confirm field requirements',
     'Preserved unknown and intentionally absent fields',
     'Separate required planner constraints from optional evaluation annotations'),
    ('recovery', 'Missing age/grade/numeric bounds',
     int((recovery_rules[['value_min', 'value_max']].eq('')).sum().sum()),
     'Some bounds are intentionally inapplicable depending on rule type',
     'Preserved empty bounds and source references',
     'Validate conditional schema by rule type and segment, then review before activation'),
], columns=['domain', 'issue', 'affected_count_or_indicator', 'source_cause_status',
            'current_handling', 'proposed_production_handling'])

summary = pd.DataFrame([
    ('exercise_records', len(exercise)),
    ('exercise_target_classes', len(classes)),
    ('exercise_missing_descriptions', int(ex['description_missing'].sum())),
    ('exercise_flagged_distinct_records', queue['catalog_id'].nunique()),
    ('imported_drill_occurrences', len(drills)),
    ('imported_unique_drill_names', drills['drill_name'].nunique()),
    ('basketball_catalog_candidate_groups', len(catalog_v02)),
    ('exact_normalized_drill_overlaps', int(overlap_table['exact_catalog_match'].eq('yes').sum())),
    ('rag_documents', len(rag)),
    ('nutrition_food_records', len(nutrition)),
    ('nutrition_complete_core_records', complete_core),
    ('nutrition_negative_carb_records', int(neg_carbs.sum())),
    ('synthetic_athlete_profiles', len(athletes)),
    ('recovery_general_guidance', len(recovery_guidance)),
    ('recovery_structured_rules', len(recovery_rules)),
], columns=['metric', 'value'])

new_tables = {
    'integrated_data_dictionary.csv': data_dictionary,
    'exercise_text_by_body_part.csv': ex_groups,
    'eda_categorical_associations.csv': assoc,
    'basketball_skill_coverage.csv': skill_freq,
    'basketball_source_coverage.csv': source_freq,
    'basketball_level_coverage.csv': level_freq,
    'basketball_participant_requirements.csv': participant_freq,
    'basketball_catalog_review_summary.csv': cat_status,
    'basketball_name_overlap_review.csv': overlap_table,
    'basketball_rag_traceability.csv': rag_summary,
    'nutrition_field_completeness.csv': nutrition_completeness,
    'nutrition_descriptive_statistics.csv': nutrition_stats,
    'nutrition_anomaly_review.csv': nutrition_anomalies,
    'nutrition_pairwise_correlations.csv': nutrition_correlations,
    'nutrition_energy_methods.csv': energy_methods,
    'nutrition_energy_comparison.csv': energy_comparison,
    'synthetic_athlete_distributions.csv': athlete_distributions,
    'synthetic_goal_by_experience.csv': athlete_goal_by_level,
    'synthetic_days_by_experience.csv': athlete_days_by_level,
    'recovery_guidance_topics.csv': recovery_topics,
    'recovery_rule_types.csv': recovery_type_freq,
    'recovery_rule_coverage.csv': recovery_coverage,
    'integrated_eda_findings.csv': model_recommendations,
    'eda_quality_causes_and_production_handling.csv': quality_policy,
    'integrated_eda_summary.csv': summary,
}

# ----------------------------------------------------------------------
# 8. Save legacy and new outputs safely
# ----------------------------------------------------------------------

def save_csv(path, frame, preserve_original=False):
    path.parent.mkdir(parents=True, exist_ok=True)
    content = frame.to_csv(index=False, lineterminator='\n').encode('utf-8')
    prior_exists = path.exists()
    if prior_exists:
        if preserve_original or path.read_bytes() == content:
            return 'PRESERVED' if preserve_original else 'UNCHANGED'
        backup_dir.mkdir(parents=True, exist_ok=True)
        backup = backup_dir / (path.stem + '_backup_' + datetime.now().strftime('%Y%m%d_%H%M%S_%f') + path.suffix)
        shutil.copy2(path, backup)
    temp = None
    try:
        with tempfile.NamedTemporaryFile(dir=path.parent, delete=False) as f:
            temp = Path(f.name)
            f.write(content)
        os.replace(temp, path)
    finally:
        if temp is not None and temp.exists():
            temp.unlink()
    return 'UPDATED' if prior_exists else 'CREATED'

print('\n[2] SAVING VERIFIED EDA TABLES')
for filename, frame in legacy_tables.items():
    print(f'{save_csv(eda_dir / filename, frame, preserve_original=(eda_dir / filename).exists()):<10} {filename}')
for filename, frame in new_tables.items():
    print(f'{save_csv(eda_dir / filename, frame):<10} {filename} ({len(frame)} rows)')

# ----------------------------------------------------------------------
# 9. Plots; retain existing four exercise figures
# ----------------------------------------------------------------------

def output_plot(fig, filename, keep_existing=False):
    fig.tight_layout()
    path = fig_dir / filename
    path.parent.mkdir(parents=True, exist_ok=True)
    if path.exists() and keep_existing:
        if path.stat().st_size < 100 or path.read_bytes()[:8] != b'\x89PNG\r\n\x1a\n':
            raise ValueError(f'Corrupted existing figure: {path}')
        print('PRESERVED FIGURE:', filename)
    else:
        rendered = io.BytesIO()
        fig.savefig(rendered, format='png', dpi=180, bbox_inches='tight')
        content = rendered.getvalue()
        if path.exists() and path.read_bytes() == content:
            print('UNCHANGED FIGURE:', filename)
        else:
            existed = path.exists()
            if existed:
                backup_dir.mkdir(parents=True, exist_ok=True)
                backup = backup_dir / (path.stem + '_backup_' + datetime.now().strftime('%Y%m%d_%H%M%S_%f') + '.png')
                shutil.copy2(path, backup)
            tmp = None
            try:
                with tempfile.NamedTemporaryFile(dir=path.parent, delete=False) as f:
                    tmp = Path(f.name)
                    f.write(content)
                os.replace(tmp, path)
            finally:
                if tmp is not None and tmp.exists():
                    tmp.unlink()
            print(('UPDATED' if existed else 'CREATED') + ' FIGURE:', filename)
    if SHOW_FIGURES:
        from IPython.display import Image, display
        display(Image(filename=str(path)))
    plt.close(fig)

fig, ax = plt.subplots(figsize=(10, 6))
sorted_classes = classes.sort_values('records')
ax.barh(sorted_classes['body_part'], sorted_classes['records'])
ax.set(xlabel='Exercise records', ylabel='Body-part category', title=f'Exercise classes (n={len(ex):,})')
output_plot(fig, 'exercise_body_part_distribution.png', keep_existing=True)

fig, ax = plt.subplots(figsize=(10, 6))
sorted_coverage = coverage.sort_values('missing_percent')
ax.barh(sorted_coverage['body_part'], sorted_coverage['missing_percent'])
ax.set(xlabel='Missing descriptions (%)', ylabel='Body-part category', xlim=(0, 100),
       title='Missing exercise descriptions by body part')
output_plot(fig, 'exercise_missing_description_by_class.png', keep_existing=True)

fig, ax = plt.subplots(figsize=(9, 4))
ax.hist(ex['name_words'], bins=min(30, max(2, int(ex['name_words'].max()) + 1)))
ax.set(xlabel='Words per exercise name', ylabel='Records', title='Exercise name length')
output_plot(fig, 'exercise_name_word_length.png', keep_existing=True)

fig, ax = plt.subplots(figsize=(9, 4))
ax.hist(ex.loc[~ex['description_missing'], 'description_words'], bins=30)
ax.set(xlabel='Words per available description', ylabel='Records', title='Description length for nonmissing records')
output_plot(fig, 'exercise_description_word_length.png', keep_existing=True)

fig, ax = plt.subplots(figsize=(10, 5))
nutrition_completeness.set_index('nutrient')['missing_pct'].sort_values().plot.barh(ax=ax)
ax.set(xlabel='Missing records (%)', ylabel='Nutrient', title='USDA Foundation Foods: missing nutrient values')
output_plot(fig, 'nutrition_missingness.png')

fig, ax = plt.subplots(figsize=(9, 5))
macro_names = ['protein_g', 'fat_g', 'carbohydrate_g', 'fiber_g', 'calories_kcal']
correlation_matrix = food_clean[macro_names].corr(method='spearman', min_periods=10)
heatmap = ax.imshow(correlation_matrix.to_numpy(), vmin=-1, vmax=1, cmap='coolwarm')
ax.set_xticks(range(len(macro_names)), [x.replace('_g', '').replace('_kcal', '') for x in macro_names], rotation=35, ha='right')
ax.set_yticks(range(len(macro_names)), [x.replace('_g', '').replace('_kcal', '') for x in macro_names])
for i in range(len(macro_names)):
    for j in range(len(macro_names)):
        value = correlation_matrix.iloc[i, j]
        if pd.notna(value):
            ax.text(j, i, f'{value:.2f}', ha='center', va='center', fontsize=8)
fig.colorbar(heatmap, ax=ax, label='Spearman correlation')
ax.set_title('USDA food composition: exploratory correlations')
output_plot(fig, 'nutrition_spearman_correlations.png')

fig, ax = plt.subplots(figsize=(9, 4))
age_counts = athletes['age'].value_counts().sort_index()
ax.bar(age_counts.index, age_counts.values)
ax.set(xlabel='Synthetic athlete age (years)', ylabel='Profiles', title='Synthetic scenario age coverage')
output_plot(fig, 'synthetic_athlete_age_coverage.png')

fig, ax = plt.subplots(figsize=(9, 5))
skill_plot = skill_freq.sort_values('records')
ax.barh(skill_plot['skill_category'], skill_plot['records'])
ax.set(xlabel='Jr. NBA drill occurrences', ylabel='Skill category', title='Imported basketball drill skill coverage')
output_plot(fig, 'basketball_imported_skill_coverage.png')

# ----------------------------------------------------------------------
# 10. Verify expected outputs and display interpretable findings
# ----------------------------------------------------------------------

missing_tables = [p for p in (eda_dir / f for f in (*legacy_tables, *new_tables)) if not p.is_file()]
expected_figures = [
    'exercise_body_part_distribution.png', 'exercise_missing_description_by_class.png',
    'exercise_name_word_length.png', 'exercise_description_word_length.png',
    'nutrition_missingness.png', 'nutrition_spearman_correlations.png',
    'synthetic_athlete_age_coverage.png', 'basketball_imported_skill_coverage.png'
]
missing_figures = [f for f in expected_figures if not (fig_dir / f).is_file()]
if missing_tables or missing_figures:
    raise RuntimeError(f'Missing reports={missing_tables}; missing figures={missing_figures}')

print('\n[3] VERIFIED ANALYSIS SUMMARY')
print(summary.to_string(index=False))
print('\n[4] SELECTED VARIABLE ASSOCIATIONS (EXPLORATORY)')
print(assoc.to_string(index=False))
print('\n[5] NUTRITION CORRELATION EXCERPT')
subset = nutrition_correlations.loc[
    nutrition_correlations['method'].eq('spearman') &
    nutrition_correlations['second_measure'].eq('calories_kcal')]
print(subset.to_string(index=False))
print('\n[6] OUTPUT VERIFICATION')
print('Original exercise tables:', len(legacy_tables), '/ 8')
print('Additional EDA tables:', len(new_tables))
print('Total named EDA tables verified:', len(legacy_tables) + len(new_tables))
print('Original exercise figures:', 4, '/ 4')
print('Additional figures:', len(expected_figures) - 4)
print('Total named figures verified:', len(expected_figures))
print('All source SHA-256 checks: PASS')
print('Original source files modified: No')
print('Existing four exercise figures overwritten: No')
print('Git commit/push performed: No')
print('OVERALL: PASS — EDA completed; review flags remain open')
print('=' * 72)
```

#### 3. Exercise Body-Part Distribution and Class Imbalance

We confirmed that our exercise dataset contains 2,909 records distributed across seventeen body-part categories.

The largest categories were:

| Body-Part Category | Records | Percentage |

| Shoulders | 338 | 11.62% |

The smallest categories included neck (8 records), adductors (17), abductors (21), and traps (24).

The largest-to-smallest class ratio was **82.5:1**.

Abdominals and quadriceps alone represented approximately 44.9% of all exercise records.

This substantial imbalance is important for our supervised classification system because a model may learn the patterns of frequently represented categories more effectively than those of minority categories.

We will therefore evaluate classification performance using macro-F1 and per-class precision and recall in addition to accuracy.

We will also consider class weighting and appropriate handling of low-support categories during model development.

We will not automatically remove or merge small classes based solely on the EDA results.

#### 5. Relationships Between Body-Part Labels and Data Completeness

We examined the association between exercise body-part categories and the availability of instructional descriptions.

The analysis produced a bias-corrected Cramér's V of **0.1492**, based on 2,909 observations.

This represents a relatively weak association between body-part category and description availability.

Nevertheless, the category-level missingness chart shows that some body-part categories have substantially higher missing-description rates than others.

The overall association measure summarizes differences across all categories and does not mean that every category has similar data completeness.

Approximately 5.88% of the contingency-table cells had expected counts below five, so the result should be interpreted cautiously, particularly for smaller classes.

This finding supports examining missing-description patterns within individual categories rather than relying only on the dataset-wide missingness percentage.

It also reinforces our decision to compare text-classification approaches that use exercise names alone with approaches that incorporate available descriptions.

The association does not establish why the descriptions are missing.

#### 12. USDA Nutrition: Correlations and Variable Relationships

We calculated exploratory numerical correlations between selected nutrient measurements.

The results revealed several moderate to strong positive relationships.

| Variable Relationship | Spearman Correlation | Valid Pairs |
|---|---:|---:|
| Protein and calories | 0.7046 | 321 |
| Fat and calories | 0.8020 | 321 |
| Carbohydrates and calories | 0.4101 | 313 |
| Fiber and calories | 0.7029 | 161 |
| General-factor energy and selected calories | 0.9956 | 322 |
| Specific-factor energy and selected calories | 1.0000 | 312 |

Fat had the strongest association with calories among the three principal macronutrients in this analysis.

Protein also showed a relatively strong positive association with calories.

Carbohydrates showed a more moderate positive association.

These relationships are consistent with the general dependence of food energy on macronutrient composition.

However, they should not be interpreted as independently discovered causal relationships.

The very high correlations between alternative energy measurements and selected calories require particular attention.

Our nutrition preparation selects the specific-factor energy measurement when available and otherwise uses the general-factor measurement.

Consequently, the perfect correlation between specific-factor energy and selected calories is expected from the data-construction process rather than evidence of an independently discovered relationship.

Including both a selected calorie value and the source measurement from which it was derived as separate model features could introduce unnecessary redundancy or leakage, depending on the prediction target.

The fiber-calorie association also requires caution because it was calculated using only 161 valid pairs.

More generally, the correlation calculations used different numbers of available observations, and missing data may affect the comparability of the coefficients.

These results will help us document measurement redundancy and the reliability of nutrition-reference fields.

We will not use these correlations to infer individual athletic performance or nutritional requirements.

Regularization and class weighting will be considered during model training and evaluation rather than treated as substitutes for source-data cleaning.

### Data Summary (Step 11)

#### Dataset Overview and Variables

Champions Athletic Map uses several complementary datasets to support supervised exercise classification, basketball coaching retrieval, personalized training-plan development, nutrition information, and youth participation and recovery constraints.

Our primary supervised machine learning dataset is a prepared version of the Gym Exercise Dataset, containing **2,909 exercise records across 17 body-part categories**. Each record includes the exercise name, available instructional description, targeted body-part category, exercise type, equipment requirements, and difficulty information. Exercise names and descriptions will serve as text inputs for predicting body-part categories, while equipment and difficulty information will remain supporting catalog metadata.

For basketball coaching, we use the Jr. NBA/Jr. WNBA instructional curriculum. Our current expanded basketball catalog contains **129 preliminary candidate groups**, including 117 basketball-specific candidates and twelve general warm-up candidates. We also integrated 23 structured Jr. NBA drill occurrences and 23 corresponding RAG documents. These records describe basketball skill categories, instructional levels, equipment, participant requirements, activity durations, and coaching instructions. They will support source-linked retrieval and activity selection after the necessary review and approval procedures.

Our nutrition component uses **411 USDA Foundation Foods records**, containing food descriptions, identifiers, protein, fat, carbohydrates, dietary fiber, and available energy measurements. These variables support general nutrition reference and food-composition comparisons.

For personalized planning and evaluation, we use **60 fictional basketball athlete profiles** representing ages 13–17. These profiles include athlete age, playing position, experience level, development goals, equipment availability, weekly training availability, and reported injury status.

Finally, five general recovery-guidance records and seventeen structured youth basketball recovery rules provide reference information for rest, sleep, training participation, and scheduling constraints.

These datasets have different purposes and structures. We will not treat their combined record counts as a single machine learning sample or assume that records from different sources can be joined without validated relationships.

#### Data Quality and Preparation

Our data-quality investigation identified several important limitations.

The exercise dataset exhibits substantial class imbalance. Abdominal and quadriceps exercises account for approximately **44.9%** of all records, while the smallest category, neck, contains only eight examples. The largest-to-smallest class ratio is **82.5:1**. This distribution may affect the classifier's ability to recognize underrepresented categories.

Missing exercise descriptions represent another important concern. Of the 2,909 records, **1,550 (53.3%)** lack instructional descriptions. We also identified 32 records with missing equipment information.

Our preparation process preserved all original cleaned exercise records and created separate normalized fields for text analysis. We maintained source identifiers and body-part labels, screened for duplicate candidates, and documented unresolved issues. No additional formatting changes were required under our implemented normalization rules.

The resulting exercise review queue contains 1,590 entries affecting 1,559 distinct records. We retained records with usable names and labels as preliminary classification candidates rather than automatically removing those with missing descriptions.

The nutrition dataset also contains incomplete information. Of the 411 food records, **321 (78.1%)** contain all four core measurements: calories, protein, fat, and carbohydrates. Dietary fiber is missing in approximately 49.6% of records, and eight records contain negative carbohydrate values requiring further verification.

We preserved these values and missing-data indicators rather than replacing them automatically. Their exact causes remain unresolved until we can compare them with the relevant original USDA records and nutrient definitions.

Our synthetic athlete profiles contain deliberately optional fields and explicitly unknown injury-status information. We preserved this uncertainty rather than interpreting missing information as an absence of injury.

We also identified seven exact normalized-name overlaps between the imported Jr. NBA drills and our existing basketball catalog. These records require reconciliation before we can determine the final number of distinct activities.

In a production machine learning pipeline, we would manage these issues through automated schema validation, source versioning, integrity checks, controlled preprocessing, duplicate screening, data-quality monitoring, and documented human review. Learned preprocessing transformations would be fitted using training data only, with validation and test data kept independent.

Regularization and class weighting would be evaluated during model training rather than used as substitutes for addressing unreliable data.

#### Variable Relationships and Exploratory Findings

Our EDA identified several relationships relevant to the design of our AI system.

For exercise classification, we examined the relationships between body-part categories, description availability, equipment requirements, and difficulty levels.

Description availability showed a relatively weak association with body-part category, with a bias-corrected Cramér's V of **0.1492** across 2,909 records. Body-part categories also showed associations with equipment requirements (0.1982) and difficulty levels (0.2172).

Although these relationships may provide useful descriptive information, several contingency tables contained sparse categories, limiting the strength of the conclusions.

We also examined exercise text lengths. Exercise names contained an average of **3.86 words**, while the 1,359 available descriptions averaged **54.48 words**. This difference suggests that the classifier must handle both short exercise names and substantially longer descriptions.

Our nutrition EDA identified meaningful numerical relationships. Spearman correlations between calories and selected macronutrients included:

| Variable Relationship | Spearman Correlation |
|---|---:|
| Fat and calories | 0.8020 |
| Protein and calories | 0.7046 |
| Carbohydrates and calories | 0.4101 |
| Fiber and calories | 0.7029 |

These correlations were calculated using available observations, so the number of records differed across comparisons.

We also observed extremely strong relationships between alternative energy measurements and the selected calorie field. General-factor energy and selected calories had a correlation of 0.9956, while specific-factor energy and selected calories had a correlation of 1.0000.

These results are partly explained by the dataset preparation process, which derives the selected calorie field from the available energy measurements. Therefore, the relationships represent expected measurement dependencies rather than independently discovered predictive effects.

This distinction is important because including derived and source energy measurements together in a predictive model could introduce redundant information or target leakage, depending on the modeling objective.

The basketball EDA showed that the smaller imported drill dataset is concentrated in shooting and ball-handling activities, with six occurrences in each category. Our larger catalog provides additional candidate coverage but still requires source verification and approval.

The sixty synthetic athlete profiles provide balanced representation across ages thirteen through seventeen. However, relationships within this dataset reflect its generation process and cannot be interpreted as evidence of real-world athlete behavior.

Our recovery analysis established coverage of guidance topics and structured planning constraints. These records are reference materials rather than measured recovery outcomes, so physiological correlations cannot be inferred from them.

#### Implications for Machine Learning and Application Development

Our findings will inform several important design decisions.

For the exercise classifier, substantial class imbalance supports the use of **macro-F1, per-class precision, and per-class recall**, rather than relying on accuracy alone. We will evaluate appropriate class-weighting and regularization strategies during model development.

The high proportion of missing descriptions supports comparing exercise-name-only inputs with combined name-and-description inputs. We will preserve title-only records where appropriate and avoid introducing unsupported synthetic descriptions.

We will also use grouped dataset splitting and additional similarity checks to reduce the risk of related exercises appearing in both training and evaluation data.

For basketball retrieval and planning, we will use skill categories, instructional requirements, equipment, participant availability, and athlete goals to support structured activity filtering. However, candidate records will not become planner eligible until their source information and suitability requirements have been reviewed.

For nutrition references, we will account for missing measurements, measurement units, and redundancy between calculated energy fields before using the data in retrieval or numerical comparisons.

The synthetic athlete profiles will support development and scenario testing. We will expand the available planning constraints and preserve a separate final evaluation set to avoid evaluation leakage.

Finally, recovery references will support structured scheduling checks. We will verify rule applicability and distinguish general participation guidance from individualized medical or training decisions.

#### Limitations and Conclusion

Our integrated EDA provides a stronger understanding of the variables available to Champions Athletic Map and their potential usefulness.

However, several limitations remain. The original publisher files for portions of our prepared dataset package are not yet available locally, some exercise descriptions are missing, minor body-part categories contain relatively few observations, nutrition records include incomplete or unusual measurements, and basketball coaching records require final approval and permissions review.

We will preserve these limitations in our project documentation and address them through subsequent cleaning, modeling, source verification, and evaluation procedures.

Overall, the EDA indicates that our datasets provide a practical foundation for developing an exercise classifier and a source-grounded basketball planning system. Their value will depend on careful preprocessing, appropriate model evaluation, reliable retrieval, explicit planning constraints, and continued data-quality verification.

Our completed analysis generated **33 EDA tables, eight figures, and a data dictionary describing 34 variables**. These outputs provide reproducible evidence for our modeling decisions and will support the Data Summary section of our final capstone report.

```python
# Step 12-1 - Basketball Drill Catalog and Coaching Source Register

import csv
from pathlib import Path
from collections import Counter

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - BASKETBALL DRILL CATALOG")
print("=" * 65)

# --------------------------------------------------
# 1. Locate Project
# --------------------------------------------------

cwd = Path.cwd().resolve()

root = next(
    (p for p in (cwd, *cwd.parents) if (p / ".git").exists()),
    None
)

if root is None:
    raise FileNotFoundError("Git project root not found.")

print("\n[1] PROJECT")
print("Root:", root)

# --------------------------------------------------
# 2. Register Official Coaching Sources
# --------------------------------------------------

source_urls = {
    "Rookie": (
        "https://jr.nba.com/basketball-practice-plans/rookie/"
    ),
    "Starter": (
        "https://jr.nba.com/basketball-practice-plans/starter/"
    ),
    "All-Star": (
        "https://jr.nba.com/basketball-practice-plans/all-star/"
    ),
    "MVP": (
        "https://jr.nba.com/basketball-practice-plans/mvp/"
    ),
}

source_index = []

for level, url in source_urls.items():

    source_index.append({
        "source_id": (
            f"jr_nba_{level.lower().replace('-', '_')}"
        ),
        "publisher": "Jr. NBA",
        "curriculum_level": level,
        "source_url": url,
        "source_location": "Skills Checklist",
        "practice_plans_published_for_level": 12,
        "title_review_date": "2026-11-05",
        "usage_permission": "pending_review",
        "stored_source_content": "no",
    })

# --------------------------------------------------
# 3. Initial Source-Linked Skill Candidates
# --------------------------------------------------

# Each entry includes:
# (publisher level, publisher checklist section,
#  source title, proposed team skill category)
#
# These are skills-checklist topics, NOT approved drills.
# We will verify full instructions and requirements later.

candidates = [

    # Rookie
    ("Rookie", "Ball Handling",
     "Stationary Crossover", "ball_handling"),

    ("Rookie", "Ball Handling",
     "Change of Direction Dribble", "ball_handling"),

    ("Rookie", "Passing",
     "Pivot & Pass", "passing"),

    ("Rookie", "Shooting",
     "Stationary Form Shooting", "shooting"),

    ("Rookie", "Shooting",
     "Using the Backboard", "shooting"),

    ("Rookie", "Footwork & Conditioning",
     "Two-Foot Jump Stop", "footwork"),

    ("Rookie", "Defense",
     "Defensive Stance", "defense"),

    ("Rookie", "Defense",
     "Defensive Slide", "defense"),

    # Starter
    ("Starter", "Ball Handling",
     "Pound Dribbling", "ball_handling"),

    ("Starter", "Ball Handling",
     "Running Crossover", "ball_handling"),

    ("Starter", "Passing",
     "Overhead Pass", "passing"),

    ("Starter", "Shooting",
     "Weak Hand Lay-Ups", "finishing"),

    ("Starter", "Shooting",
     "Free Throws", "shooting"),

    ("Starter", "Footwork & Conditioning",
     "Jab Step", "footwork"),

    ("Starter", "Defense",
     "On Ball Defense", "defense"),

    ("Starter", "Defense",
     "Closeout", "defense"),

    # All-Star
    ("All-Star", "Ball Handling",
     "Between the Legs Dribble", "ball_handling"),

    ("All-Star", "Passing",
     "Backdoor Pass", "passing"),

    ("All-Star", "Shooting",
     "Step-Back Shot", "shooting"),

    ("All-Star", "Shooting",
     "Reverse Lay-Up", "finishing"),

    ("All-Star", "Footwork & Conditioning",
     "Quick Feet", "footwork"),

    ("All-Star", "Footwork & Conditioning",
     "Backdoor Cut", "footwork"),

    ("All-Star", "Defense",
     "Shell Drill", "defense"),

    ("All-Star", "Defense",
     "Help the Helper", "defense"),

    # MVP
    ("MVP", "Ball Handling",
     "Rhythm Dribbling", "ball_handling"),

    ("MVP", "Passing",
     "Behind the Back Pass", "passing"),

    ("MVP", "Shooting",
     "Floaters", "finishing"),

    ("MVP", "Shooting",
     "Pick & Roll Shots", "shooting"),

    ("MVP", "Shooting",
     "Advanced Lay-Up Finishes", "finishing"),

    ("MVP", "Offense",
     "Reading the Pick & Roll", "offense"),

    ("MVP", "Defense",
     "Stunt & Recover", "defense"),

    ("MVP", "Defense",
     "Defending Cross Screens", "defense"),
]

# --------------------------------------------------
# 4. Define Output Paths
# --------------------------------------------------

source_index_path = (
    root / "data" / "references" / "basketball"
    / "source_index.csv"
)

catalog_path = (
    root / "data" / "processed"
    / "basketball_drills.csv"
)

review_path = (
    root / "results" / "basketball_drill_review.csv"
)

coverage_path = (
    root / "results" / "eda"
    / "basketball_drill_coverage.csv"
)

outputs = [
    source_index_path,
    catalog_path,
    review_path,
    coverage_path,
]

# Do not overwrite previously reviewed files.

if any(path.exists() for path in outputs):
    raise FileExistsError(
        "A Step 12 output already exists. "
        "Review the existing files before regenerating."
    )

# --------------------------------------------------
# 5. Construct Candidate Records and Review Queue
# --------------------------------------------------

catalog = []
review = []
seen = set()

for number, (level, section, title, skill) in enumerate(
    candidates, start=1
):

    key = (level.casefold(), title.casefold())

    if key in seen:
        raise ValueError(
            f"Duplicate source title within level: {key}"
        )

    seen.add(key)

    candidate_id = f"JR-TOPIC-{number:04d}"

    source_id = (
        f"jr_nba_{level.lower().replace('-', '_')}"
    )

    catalog.append({
        "candidate_id": candidate_id,
        "drill_id": "",
        "record_kind": "skills_checklist_topic",
        "source_item_title": title,
        "canonical_drill_name": "",
        "proposed_primary_skill": skill,
        "secondary_tags": "",
        "source_skill_section": section,
        "publisher_curriculum_level": level,
        "instructions_summary": "",
        "equipment_required": "",
        "space_required": "",
        "participant_format": "",
        "minimum_participants": "",
        "source_prescription_text": "",
        "source_duration_text": "",
        "interpreted_duration_minutes": "",
        "documented_variations": "",
        "media_url": "",
        "source_id": source_id,
        "source_url": source_urls[level],
        "source_locator": f"Skills Checklist > {section}",
        "title_review_date": "2026-11-05",
        "permission_status": "pending_review",
        "review_status": "candidate_needs_drill_verification",
        "planner_eligible": "no",
        "reviewer": "",
        "review_notes": (
            "Checklist topic; not yet verified as "
            "a complete, distinct basketball drill."
        ),
    })

    review.append({
        "candidate_id": candidate_id,
        "source_id": source_id,
        "issue_type": "drill_definition_and_requirements",
        "priority": "high",
        "issue": (
            "Checklist title alone does not establish "
            "a complete drill."
        ),
        "next_action": (
            "Verify practice-plan instructions, equipment, "
            "space, participants, timing, and distinctness."
        ),
        "review_status": "open",
    })

# Add permission reviews for the four source levels.

for source in source_index:

    review.append({
        "candidate_id": "",
        "source_id": source["source_id"],
        "issue_type": "permission_review",
        "priority": "high",
        "issue": (
            "Storage, transformation, and redistribution "
            "rights are not yet documented."
        ),
        "next_action": (
            "Review source terms before copying "
            "or redistributing instructional content."
        ),
        "review_status": "open",
    })

# --------------------------------------------------
# 6. Prepare Coverage Report
# --------------------------------------------------

coverage_counts = Counter(
    (
        row["publisher_curriculum_level"],
        row["proposed_primary_skill"]
    )
    for row in catalog
)

coverage = []

for (level, skill), count in sorted(
    coverage_counts.items()
):

    coverage.append({
        "publisher_curriculum_level": level,
        "proposed_primary_skill": skill,
        "candidate_topics": count,
        "approved_distinct_drills": 0,
        "target_approved_distinct_drills": 100,
    })

coverage.append({
    "publisher_curriculum_level": "ALL",
    "proposed_primary_skill": "ALL",
    "candidate_topics": len(catalog),
    "approved_distinct_drills": 0,
    "target_approved_distinct_drills": 100,
})

# --------------------------------------------------
# 7. Save Files Without Overwriting
# --------------------------------------------------

def create_csv(path, rows):

    path.parent.mkdir(parents=True, exist_ok=True)

    with path.open(
        "x", newline="", encoding="utf-8"
    ) as file:

        writer = csv.DictWriter(
            file,
            fieldnames=list(rows[0].keys())
        )

        writer.writeheader()
        writer.writerows(rows)

    print("CREATED:", path.relative_to(root))


print("\n[2] SAVING CATALOG FILES")

create_csv(source_index_path, source_index)
create_csv(catalog_path, catalog)
create_csv(review_path, review)
create_csv(coverage_path, coverage)

# --------------------------------------------------
# 8. Verification Summary
# --------------------------------------------------

print("\n[3] SOURCE SUMMARY")

print("Official source pages:", len(source_index))
print("Practice plans reported by publisher:", 48)
print("Source-linked candidate topics:", len(catalog))
print("Approved distinct drills:", 0)
print("Target approved distinct drills:", 100)

print("\n[4] CANDIDATES BY CURRICULUM LEVEL")

for level, count in Counter(
    row["publisher_curriculum_level"]
    for row in catalog
).items():

    print(f"{level:<15} {count}")

print("\n[5] CANDIDATES BY PROPOSED SKILL")

for skill, count in Counter(
    row["proposed_primary_skill"]
    for row in catalog
).most_common():

    print(f"{skill:<20} {count}")

print("\n[6] FINAL VERIFICATION")

print("Open review items:", len(review))

print(
    "All candidates excluded from planner:",
    all(row["planner_eligible"] == "no" for row in catalog)
)

print(
    "Original instructional materials downloaded: No"
)

print("Existing source datasets modified: No")

print("\n" + "=" * 65)
print("STEP 12 - INITIAL DRILL CATALOG CREATED")
print("=" * 65)
```

```python
# Step 12-5 - Consolidate and Review Basketball Drill Evidence

import csv
import re
from pathlib import Path
from collections import Counter, defaultdict

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - DRILL EVIDENCE AUDIT")
print("=" * 65)

root = next(
    p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (p / ".git").exists()
)

results = root / "results"

batch_files = [
    results / "basketball_drill_source_evidence.csv",
    results / "basketball_drill_source_evidence_batch2.csv",
    results / "basketball_drill_source_evidence_batch3.csv",
]

# --------------------------------------------------
# 1. Load and Validate the Three Batches
# --------------------------------------------------

evidence = []
expected_columns = None

for path in batch_files:

    if not path.is_file():
        raise FileNotFoundError(path)

    with path.open(newline="", encoding="utf-8-sig") as f:
        reader = csv.DictReader(f)
        columns = reader.fieldnames or []

        if expected_columns is None:
            expected_columns = columns
        elif set(columns) != set(expected_columns):
            raise ValueError("Evidence batch schema mismatch.")

        rows = list(reader)
        evidence.extend(rows)

        print(f"Loaded {path.name}: {len(rows)} records")

ids = [r["evidence_id"] for r in evidence]

if len(ids) != len(set(ids)):
    raise ValueError("Duplicate evidence IDs detected.")

if len(evidence) != 30:
    raise ValueError(
        f"Expected 30 source occurrences; found {len(evidence)}."
    )

# --------------------------------------------------
# 2. Normalize Names for Preliminary Grouping
# --------------------------------------------------

def name_key(value):
    return re.sub(
        r"[^a-z0-9]+",
        " ",
        (value or "").casefold()
    ).strip()

groups = defaultdict(list)

for row in evidence:
    groups[name_key(row["official_drill_name"])].append(row)

# Activities requiring closer classification.
# These are review flags, not automatic exclusions.

general_warmup_candidates = {
    "run to that line",
    "forward backward lateral march",
    "high knees",
    "quick feet",
    "side push",
    "the escape game",
}

source_description_review = {
    "quick feet",
    "side push",
}

# --------------------------------------------------
# 3. Create Consolidated Evidence Records
# --------------------------------------------------

consolidated = []
review_checks = []

for row in evidence:

    key = name_key(row["official_drill_name"])
    group = groups[key]

    flags = []

    if len(group) > 1:
        flags.append("cross_source_name_review")

    if key in general_warmup_candidates:
        flags.append("general_warmup_scope_review")

    if key in source_description_review:
        flags.append("source_description_overlap_review")

    if not row["source_timing"].strip() or (
        row["source_timing"].strip().lower()
        in {"not stated", "not specified"}
    ):
        flags.append("source_timing_not_verified")

    if not row["factual_summary"].strip():
        flags.append("summary_not_prepared")

    if not row["source_url"].startswith(
        "https://jr.nba.com/"
    ):
        flags.append("source_url_review")

    if row["permission_status"] != "pending_review":
        flags.append("permission_status_review")

    enriched = dict(row)

    enriched.update({
        "normalized_name": key,
        "name_group_occurrences": len(group),
        "scope_status": "pending_review",
        "source_validation_status": "pending_final_review",
        "review_flags": ";".join(flags),
    })

    # No automatic approval.
    enriched["planner_eligible"] = "no"
    enriched["final_review_status"] = "pending"

    consolidated.append(enriched)

    for flag in flags:
        review_checks.append({
            "evidence_id": row["evidence_id"],
            "official_drill_name": row["official_drill_name"],
            "check": flag,
            "status": "needs_review",
        })

# --------------------------------------------------
# 4. Create Preliminary Canonical-Name Groups
# --------------------------------------------------

canonical = []

for number, (key, rows) in enumerate(
    sorted(groups.items()), start=1
):

    canonical.append({
        "group_id": f"JR-GROUP-{number:04d}",
        "normalized_name": key,
        "display_name": rows[0]["official_drill_name"],
        "source_occurrences": len(rows),
        "evidence_ids": ";".join(
            r["evidence_id"] for r in rows
        ),
        "curriculum_levels": ";".join(sorted({
            r["source_level"] for r in rows
        })),
        "proposed_skills": ";".join(sorted({
            r["proposed_skill"] for r in rows
        })),
        "scope_review": (
            "general_warmup_candidate"
            if key in general_warmup_candidates
            else "basketball_drill_candidate"
        ),
        "distinctness_status": "pending_review",
        "permission_status": "pending_review",
        "planner_approved": "no",
    })

# --------------------------------------------------
# 5. Save New Reports Without Overwriting
# --------------------------------------------------

outputs = {
    results / "basketball_drill_evidence_consolidated.csv":
        consolidated,

    results / "basketball_canonical_review.csv":
        canonical,

    results / "basketball_evidence_validation.csv":
        review_checks,
}

if any(path.exists() for path in outputs):
    raise FileExistsError(
        "One or more consolidated outputs already exist. "
        "Existing files were preserved."
    )

for path, rows in outputs.items():

    if not rows:
        raise ValueError(f"No rows available for {path.name}")

    with path.open(
        "x", newline="", encoding="utf-8"
    ) as f:

        writer = csv.DictWriter(
            f, fieldnames=list(rows[0].keys())
        )

        writer.writeheader()
        writer.writerows(rows)

    print("CREATED:", path.relative_to(root))

# --------------------------------------------------
# 6. Final Verification
# --------------------------------------------------

print("\n[1] CONSOLIDATION SUMMARY")

print("Source occurrences:", len(consolidated))
print("Preliminary distinct name groups:", len(canonical))

print(
    "Repeated-name groups:",
    sum(len(g) > 1 for g in groups.values())
)

print(
    "General warm-up candidates:",
    sum(
        row["normalized_name"] in general_warmup_candidates
        for row in consolidated
    )
)

print("Review flags by type:")

for issue, count in Counter(
    r["check"] for r in review_checks
).most_common():

    print(f"  {issue}: {count}")

print("\n[2] PLANNER APPROVAL")

print(
    "Approved records:",
    sum(r["planner_eligible"] == "yes"
        for r in consolidated)
)

print("Original evidence batches modified: No")
print("Candidate catalog modified: No")

print("\n" + "=" * 65)
print("STEP 12 - EVIDENCE CONSOLIDATION COMPLETE")
print("=" * 65)
```

```python
# Step 12-7 - Consolidate Four Drill Evidence Batches

import csv
import re
from pathlib import Path
from collections import Counter, defaultdict
from urllib.parse import urlparse

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - FOUR-BATCH DRILL REVIEW")
print("=" * 65)

root = next(
    p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (p / ".git").exists()
)

results = root / "results"

batch_files = [
    results / "basketball_drill_source_evidence.csv",
    results / "basketball_drill_source_evidence_batch2.csv",
    results / "basketball_drill_source_evidence_batch3.csv",
    results / "basketball_drill_source_evidence_batch4.csv",
]

output_files = [
    results / "basketball_drill_evidence_consolidated_46.csv",
    results / "basketball_canonical_review_46.csv",
    results / "basketball_evidence_validation_46.csv",
]

# Protect previously created outputs.
if any(p.exists() for p in output_files):
    raise FileExistsError(
        "A consolidated output already exists. "
        "Review existing files before rerunning."
    )

# --------------------------------------------------
# 1. Load All Four Evidence Batches
# --------------------------------------------------

evidence = []
expected_columns = None

for path in batch_files:

    with path.open(newline="", encoding="utf-8-sig") as f:
        reader = csv.DictReader(f)
        columns = reader.fieldnames or []

        if expected_columns is None:
            expected_columns = columns
        elif set(columns) != set(expected_columns):
            raise ValueError("Evidence schemas differ.")

        rows = list(reader)
        evidence.extend(rows)

        print(f"Loaded {path.name}: {len(rows)}")

if len(evidence) != 46:
    raise ValueError("Expected 46 evidence occurrences.")

ids = [r["evidence_id"] for r in evidence]

if len(ids) != len(set(ids)):
    raise ValueError("Duplicate evidence identifiers.")

if any(r["planner_eligible"] != "no" for r in evidence):
    raise ValueError(
        "Unexpected planner-approved record. Review manually."
    )

# --------------------------------------------------
# 2. Normalize Names and Group Potential Duplicates
# --------------------------------------------------

def normalize_name(value):
    return re.sub(
        r"[^a-z0-9]+", " ",
        (value or "").casefold()
    ).strip()

groups = defaultdict(list)

for row in evidence:
    groups[normalize_name(row["official_drill_name"])].append(row)

# Preliminary review indicators, not automatic exclusions.
warmup_candidates = {
    "run to that line",
    "forward backward lateral march",
    "high knees",
    "quick feet",
    "side push",
    "the escape game",
}

description_overlap = {
    "quick feet",
    "side push",
}

# --------------------------------------------------
# 3. Build Consolidated Evidence and Review Flags
# --------------------------------------------------

consolidated = []
review_items = []

actions = {
    "repeated_name":
        "Compare source instructions and variations.",
    "general_warmup_scope":
        "Decide whether this counts toward the basketball-drill target.",
    "description_overlap":
        "Compare source descriptions before assigning distinct IDs.",
    "missing_summary":
        "Review the practice plan and prepare a permitted factual summary.",
    "timing_unverified":
        "Confirm the source prescription or retain unknown timing.",
    "equipment_ambiguous":
        "Verify the equipment requirements against the source.",
    "source_url_invalid":
        "Correct and verify the official source URL.",
}

for row in evidence:

    key = normalize_name(row["official_drill_name"])
    flags = []

    if len(groups[key]) > 1:
        flags.append("repeated_name")

    if key in warmup_candidates:
        flags.append("general_warmup_scope")

    if key in description_overlap:
        flags.append("description_overlap")

    if not row["factual_summary"].strip():
        flags.append("missing_summary")

    if row["source_timing"].strip().lower() in {
        "", "not stated", "not specified"
    }:
        flags.append("timing_unverified")

    if row["source_equipment"].strip().lower() in {
        "", "not stated", "not specified",
        "not explicitly required"
    }:
        flags.append("equipment_ambiguous")

    parsed = urlparse(row["source_url"])

    if (
        parsed.scheme != "https"
        or parsed.netloc != "jr.nba.com"
        or not parsed.path.startswith(
            "/basketball-practice-plans/"
        )
    ):
        flags.append("source_url_invalid")

    updated = dict(row)

    updated.update({
        "normalized_name": key,
        "name_group_size": len(groups[key]),
        "scope_status": "pending_review",
        "independent_source_audit": "pending",
        "distinctness_status": "pending",
        "approval_status": "not_approved",
        "review_flags": ";".join(flags),
    })

    consolidated.append(updated)

    for flag in flags:
        review_items.append({
            "evidence_id": row["evidence_id"],
            "drill_name": row["official_drill_name"],
            "issue_type": flag,
            "recommended_action": actions[flag],
            "status": "open",
        })

# --------------------------------------------------
# 4. Build Preliminary Canonical Groups
# --------------------------------------------------

canonical = []

for number, (key, rows) in enumerate(
    sorted(groups.items()), start=1
):

    canonical.append({
        "canonical_candidate_id": f"JR-CAN-{number:04d}",
        "normalized_name": key,
        "display_name": rows[0]["official_drill_name"],
        "occurrences": len(rows),
        "evidence_ids": ";".join(
            r["evidence_id"] for r in rows
        ),
        "curriculum_levels": ";".join(
            sorted({r["source_level"] for r in rows})
        ),
        "proposed_skills": ";".join(
            sorted({r["proposed_skill"] for r in rows})
        ),
        "scope_candidate": (
            "general_warmup"
            if key in warmup_candidates
            else "basketball_drill"
        ),
        "distinctness_status": "pending_review",
        "source_verification": "pending",
        "permission_status": "pending_review",
        "planner_approved": "no",
    })

# --------------------------------------------------
# 5. Save New Snapshot Files
# --------------------------------------------------

def save_new(path, rows):

    with path.open("x", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(
            f, fieldnames=list(rows[0].keys())
        )
        writer.writeheader()
        writer.writerows(rows)

    print("CREATED:", path.relative_to(root))

print("\n[1] SAVING CONSOLIDATED REPORTS")

save_new(output_files[0], consolidated)
save_new(output_files[1], canonical)
save_new(output_files[2], review_items)

# --------------------------------------------------
# 6. Verification
# --------------------------------------------------

flags = Counter(r["issue_type"] for r in review_items)

print("\n[2] CATALOG SUMMARY")
print("Total source occurrences:", len(consolidated))
print("Preliminary distinct names:", len(canonical))
print(
    "Repeated-name groups:",
    sum(len(v) > 1 for v in groups.values())
)
print(
    "Warm-up candidate occurrences:",
    sum(
        normalize_name(r["official_drill_name"])
        in warmup_candidates
        for r in evidence
    )
)

print("\n[3] REVIEW FLAGS")

for flag, count in flags.most_common():
    print(f"{flag:<30} {count}")

print("\n[4] FINAL STATUS")
print("Open review entries:", len(review_items))
print("Planner-approved drills:", 0)
print("Existing source files modified: No")
print("Earlier snapshots overwritten: No")

print("\n" + "=" * 65)
print("FOUR-BATCH CONSOLIDATION COMPLETE")
print("=" * 65)
```

```python
# Step 12-10 - Basketball Source Content Audit

import csv
import re
from pathlib import Path
from collections import Counter
from html.parser import HTMLParser
from urllib.request import Request, urlopen
from urllib.error import HTTPError, URLError

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - SOURCE CONTENT AUDIT")
print("=" * 65)

root = next(
    p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (p / ".git").exists()
)

results = root / "results"

batch_paths = [
    results / "basketball_drill_source_evidence.csv",
    *[
        results / f"basketball_drill_source_evidence_batch{i}.csv"
        for i in range(2, 6)
    ],
]

output_path = results / "basketball_source_content_audit.csv"

if output_path.exists():
    raise FileExistsError(
        "Audit output already exists. Existing file preserved."
    )

# --------------------------------------------------
# 1. Load All 70 Evidence Records
# --------------------------------------------------

evidence = []

for path in batch_paths:
    with path.open(newline="", encoding="utf-8-sig") as f:
        evidence.extend(csv.DictReader(f))

if len(evidence) != 70:
    raise ValueError("Expected 70 evidence records.")

# --------------------------------------------------
# 2. Extract Visible Text Without Saving Webpages
# --------------------------------------------------

class PageTextParser(HTMLParser):

    def __init__(self):
        super().__init__()
        self.parts = []
        self.skip = 0

    def handle_starttag(self, tag, attrs):
        if tag in {"script", "style", "noscript"}:
            self.skip += 1

    def handle_endtag(self, tag):
        if tag in {"script", "style", "noscript"}:
            self.skip = max(0, self.skip - 1)

    def handle_data(self, data):
        if self.skip == 0:
            self.parts.append(data)


def normalize(value):
    return re.sub(
        r"[^a-z0-9]+",
        " ",
        value.casefold()
    ).strip()


def fetch_page(url):
    request = Request(
        url,
        headers={
            "User-Agent": (
                "ChampionsAthleticMap/0.1 "
                "(educational source verification)"
            )
        }
    )

    try:
        with urlopen(request, timeout=20) as response:
            content = response.read(2_000_000)
            status = response.status
            final_url = response.geturl()

        parser = PageTextParser()
        parser.feed(
            content.decode("utf-8", errors="replace")
        )

        return {
            "fetch_status": "success",
            "http_status": status,
            "final_url": final_url,
            "normalized_text": normalize(
                " ".join(parser.parts)
            ),
            "error": "",
        }

    except (HTTPError, URLError, TimeoutError, OSError) as error:
        return {
            "fetch_status": "failed",
            "http_status": getattr(error, "code", ""),
            "final_url": "",
            "normalized_text": "",
            "error": str(error)[:150],
        }


# --------------------------------------------------
# 3. Fetch Each Unique Page Once
# --------------------------------------------------

urls = sorted({
    row["source_url"] for row in evidence
})

print("\n[1] SOURCE PAGE ACCESS")

page_cache = {}

for number, url in enumerate(urls, start=1):

    result = fetch_page(url)
    page_cache[url] = result

    print(
        f"{number:02d}/{len(urls):02d} "
        f"{result['fetch_status']}: {url}"
    )

# --------------------------------------------------
# 4. Check Activity Title Presence
# --------------------------------------------------

audit = []

for row in evidence:

    page = page_cache[row["source_url"]]
    title = normalize(row["official_drill_name"])

    if page["fetch_status"] != "success":
        title_status = "not_checked"
    elif title and title in page["normalized_text"]:
        title_status = "text_present"
    else:
        title_status = "not_found_in_extracted_text"

    audit.append({
        "evidence_id": row["evidence_id"],
        "source_level": row["source_level"],
        "practice_plan": row["practice_plan"],
        "drill_name": row["official_drill_name"],
        "source_url": row["source_url"],
        "http_status": page["http_status"],
        "fetch_status": page["fetch_status"],
        "title_check": title_status,
        "exact_drill_section_verified": "no",
        "equipment_verified": "pending",
        "participants_verified": "pending",
        "timing_verified": "pending",
        "permission_review": "pending",
        "planner_approved": "no",
        "error": page["error"],
    })

# --------------------------------------------------
# 5. Save Report Without Overwriting
# --------------------------------------------------

with output_path.open(
    "x", newline="", encoding="utf-8"
) as f:

    writer = csv.DictWriter(
        f, fieldnames=list(audit[0].keys())
    )

    writer.writeheader()
    writer.writerows(audit)

print("\nCREATED:", output_path.relative_to(root))

# --------------------------------------------------
# 6. Audit Summary
# --------------------------------------------------

print("\n[2] SOURCE CONTENT VERIFICATION")

print("Evidence records:", len(evidence))
print("Unique source pages:", len(urls))

print("Page fetch results:")
for status, count in Counter(
    p["fetch_status"] for p in page_cache.values()
).items():
    print(f"  {status}: {count}")

print("\nActivity title checks:")
for status, count in Counter(
    r["title_check"] for r in audit
).items():
    print(f"  {status}: {count}")

flagged = [
    r for r in audit
    if r["title_check"] != "text_present"
]

print("\nTitles requiring further investigation:", len(flagged))

for row in flagged[:15]:
    print(
        f"  {row['evidence_id']}: "
        f"{row['drill_name']} "
        f"({row['title_check']})"
    )

print("\nPlanner-approved records: 0")
print("Existing evidence files modified: No")
print("Full instructional webpages stored: No")

print("\n" + "=" * 65)
print("SOURCE CONTENT AUDIT COMPLETE")
print("=" * 65)
```

```python
# Step 12-11 - Verify Activity Headings and Source Plan Identity

import csv
import re
import unicodedata
from pathlib import Path
from collections import Counter
from html.parser import HTMLParser
from urllib.parse import urlsplit
from urllib.request import Request, urlopen
from urllib.error import HTTPError, URLError

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - DRILL HEADING VERIFICATION")
print("=" * 65)

# --------------------------------------------------
# 1. Locate Project and Load Evidence
# --------------------------------------------------

root = next(
    p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (p / ".git").exists()
)

results = root / "results"

batch_paths = [
    results / "basketball_drill_source_evidence.csv",
    *[
        results / f"basketball_drill_source_evidence_batch{i}.csv"
        for i in range(2, 6)
    ],
]

output_path = (
    results / "basketball_drill_heading_audit.csv"
)

evidence = []

for path in batch_paths:
    with path.open(newline="", encoding="utf-8-sig") as f:
        evidence.extend(csv.DictReader(f))

if len(evidence) != 70:
    raise ValueError("Expected 70 evidence records.")

# --------------------------------------------------
# 2. Normalize Activity Titles
# --------------------------------------------------

def normalize(value):
    value = unicodedata.normalize("NFKC", value or "")
    return re.sub(
        r"[^a-z0-9]+",
        " ",
        value.casefold()
    ).strip()

# --------------------------------------------------
# 3. Parse HTML Headings and Visible Text
# --------------------------------------------------

class PracticePageParser(HTMLParser):

    def __init__(self):
        super().__init__()
        self.headings = []
        self.visible_parts = []
        self.heading_tag = None
        self.heading_parts = []
        self.skip_depth = 0

    def handle_starttag(self, tag, attrs):
        if tag in {"script", "style", "noscript"}:
            self.skip_depth += 1
            return

        if self.skip_depth:
            return

        if tag in {"h1", "h2", "h3", "h4", "h5", "h6"}:
            self.heading_tag = tag
            self.heading_parts = []

    def handle_data(self, data):
        if self.skip_depth:
            return

        self.visible_parts.append(data)

        if self.heading_tag:
            self.heading_parts.append(data)

    def handle_endtag(self, tag):
        if tag in {"script", "style", "noscript"}:
            self.skip_depth = max(0, self.skip_depth - 1)
            return

        if tag == self.heading_tag:
            heading = " ".join(self.heading_parts).strip()

            if heading:
                self.headings.append(heading)

            self.heading_tag = None
            self.heading_parts = []

# --------------------------------------------------
# 4. Fetch Each Unique Official Source Once
# --------------------------------------------------

urls = sorted({r["source_url"] for r in evidence})
cache = {}

for index, url in enumerate(urls, start=1):

    parsed_url = urlsplit(url)

    if (
        parsed_url.scheme != "https"
        or parsed_url.hostname != "jr.nba.com"
    ):
        raise ValueError(f"Unexpected source URL: {url}")

    try:
        request = Request(
            url,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        with urlopen(request, timeout=25) as response:
            html = response.read(2_000_000).decode(
                "utf-8", errors="replace"
            )

        parser = PracticePageParser()
        parser.feed(html)

        cache[url] = {
            "status": "success",
            "headings": [
                normalize(h) for h in parser.headings
            ],
            "visible_text": normalize(
                " ".join(parser.visible_parts)
            ),
        }

        print(
            f"FETCHED {index:02d}/{len(urls):02d}: "
            f"{len(parser.headings)} headings"
        )

    except (HTTPError, URLError, TimeoutError, OSError) as error:

        cache[url] = {
            "status": "failed",
            "headings": [],
            "visible_text": "",
        }

        print(
            f"UNAVAILABLE {index:02d}/{len(urls):02d}: "
            f"{type(error).__name__}"
        )

# --------------------------------------------------
# 5. Check Heading and Practice Plan Identity
# --------------------------------------------------

audit = []

for row in evidence:

    page = cache[row["source_url"]]
    name = normalize(row["official_drill_name"])

    plan_marker = normalize(
        f"{row['source_level']} Practice Plan "
        f"{row['practice_plan']}"
    )

    if page["status"] != "success":
        heading_status = "not_checked"
        heading_matches = 0
        plan_match = "not_checked"

    else:
        heading_matches = page["headings"].count(name)

        heading_status = (
            "heading_found"
            if heading_matches > 0
            else "not_found_in_html_headings"
        )

        plan_match = (
            "yes"
            if plan_marker in page["visible_text"]
            else "no"
        )

    special_review = (
        "overlapping_source_description"
        if name in {"quick feet", "side push"}
        and row["source_level"] == "Rookie"
        and row["practice_plan"] == "2"
        else ""
    )

    audit.append({
        "evidence_id": row["evidence_id"],
        "activity_name": row["official_drill_name"],
        "source_level": row["source_level"],
        "practice_plan": row["practice_plan"],
        "source_url": row["source_url"],
        "fetch_status": page["status"],
        "heading_status": heading_status,
        "heading_matches": heading_matches,
        "practice_plan_marker": plan_match,
        "special_review_flag": special_review,
        "equipment_review": "pending",
        "timing_review": "pending",
        "participants_review": "pending",
        "distinctness_review": "pending",
        "permission_review": "pending",
        "planner_approved": "no",
    })

# --------------------------------------------------
# 6. Save New Report
# --------------------------------------------------

if output_path.exists():
    print("\nExisting report preserved:", output_path.name)

else:
    with output_path.open(
        "x", newline="", encoding="utf-8"
    ) as f:
        writer = csv.DictWriter(
            f, fieldnames=list(audit[0].keys())
        )
        writer.writeheader()
        writer.writerows(audit)

    print("\nCREATED:", output_path.relative_to(root))

# --------------------------------------------------
# 7. Verification Summary
# --------------------------------------------------

print("\n[1] ACTIVITY HEADING RESULTS")

for status, count in Counter(
    r["heading_status"] for r in audit
).items():
    print(f"  {status}: {count}")

print("\n[2] PRACTICE PLAN IDENTITY")

for status, count in Counter(
    r["practice_plan_marker"] for r in audit
).items():
    print(f"  {status}: {count}")

print("\n[3] HEADING MATCHES NEEDING REVIEW")

flagged = [
    r for r in audit
    if (
        r["heading_status"] != "heading_found"
        or r["practice_plan_marker"] != "yes"
    )
]

print("Records requiring investigation:", len(flagged))

for row in flagged[:20]:
    print(
        f"  {row['evidence_id']}: "
        f"{row['activity_name']} "
        f"({row['heading_status']})"
    )

print("\n[4] FINAL STATUS")
print("Evidence records:", len(audit))
print("Approved drills:", 0)
print("Source instructions copied to disk: No")
print("Existing source CSVs modified: No")

print("\n" + "=" * 65)
print("STEP 12 - HEADING VERIFICATION FINISHED")
print("=" * 65)
```

```python
# Step 12-12 - Drill-Level Source Section and Timing Audit

import csv
import re
import hashlib
import unicodedata
from pathlib import Path
from collections import Counter, defaultdict
from html.parser import HTMLParser
from urllib.request import Request, urlopen
from urllib.error import HTTPError, URLError

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - DRILL SECTION AUDIT")
print("=" * 65)

root = next(
    p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (p / ".git").exists()
)

results = root / "results"

batch_paths = [
    results / "basketball_drill_source_evidence.csv",
    *[
        results / f"basketball_drill_source_evidence_batch{i}.csv"
        for i in range(2, 6)
    ],
]

output_path = results / "basketball_drill_section_audit.csv"

if output_path.exists():
    raise FileExistsError(
        "Existing section audit preserved. Review before rerunning."
    )

evidence = []

for path in batch_paths:
    with path.open(newline="", encoding="utf-8-sig") as f:
        evidence.extend(csv.DictReader(f))

if len(evidence) != 70:
    raise ValueError("Expected 70 evidence records.")

# --------------------------------------------------
# 1. Normalize Text
# --------------------------------------------------

def normalize(value):
    value = unicodedata.normalize("NFKC", value or "")
    return re.sub(
        r"[^a-z0-9]+", " ",
        value.casefold()
    ).strip()


# --------------------------------------------------
# 2. Extract Text Associated With Each HTML Heading
# --------------------------------------------------

class SectionParser(HTMLParser):

    def __init__(self):
        super().__init__()
        self.sections = []
        self.current = None
        self.heading_tag = None
        self.heading_parts = []
        self.skip_depth = 0

    def handle_starttag(self, tag, attrs):

        if tag in {"script", "style", "noscript"}:
            self.skip_depth += 1
            return

        if self.skip_depth:
            return

        if re.fullmatch(r"h[1-6]", tag):
            self.heading_tag = tag
            self.heading_parts = []

    def handle_data(self, data):

        if self.skip_depth:
            return

        if self.heading_tag:
            self.heading_parts.append(data)

        elif self.current is not None:
            self.current["parts"].append(data)

    def handle_endtag(self, tag):

        if tag in {"script", "style", "noscript"}:
            self.skip_depth = max(0, self.skip_depth - 1)
            return

        if self.skip_depth:
            return

        if tag == self.heading_tag:

            heading = " ".join(self.heading_parts).strip()

            if heading:
                self.current = {
                    "heading": heading,
                    "parts": []
                }
                self.sections.append(self.current)

            self.heading_tag = None
            self.heading_parts = []

        elif tag in {"p", "li", "div", "br"}:
            if self.current is not None:
                self.current["parts"].append(" ")


# --------------------------------------------------
# 3. Fetch Official Pages
# --------------------------------------------------

pages = {}

for url in sorted({r["source_url"] for r in evidence}):

    try:
        request = Request(
            url,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        with urlopen(request, timeout=25) as response:
            html = response.read(2_000_000).decode(
                "utf-8", errors="replace"
            )

        parser = SectionParser()
        parser.feed(html)

        pages[url] = parser.sections

        print(
            "FETCHED:",
            url,
            "| sections:",
            len(parser.sections)
        )

    except (HTTPError, URLError, TimeoutError, OSError) as error:
        pages[url] = None
        print("FETCH FAILED:", url, type(error).__name__)


# --------------------------------------------------
# 4. Compare Each Recorded Activity
# --------------------------------------------------

audit = []

for row in evidence:

    sections = pages[row["source_url"]]
    title = normalize(row["official_drill_name"])

    if sections is None:
        matches = []
        status = "page_unavailable"
    else:
        matches = [
            section for section in sections
            if normalize(section["heading"]) == title
        ]
        status = (
            "section_found" if matches
            else "section_not_found"
        )

    # When several headings match, retain the largest
    # immediate section for preliminary analysis.
    section_text = ""

    if matches:
        section_text = max(
            (
                normalize(" ".join(s["parts"]))
                for s in matches
            ),
            key=len
        )

    section_words = len(section_text.split())

    # The fingerprint supports change and overlap detection.
    # The original instructional section is not saved.
    fingerprint = (
        hashlib.sha256(
            section_text.encode("utf-8")
        ).hexdigest()
        if section_text else ""
    )

    timing = normalize(row["source_timing"])

    if not timing or timing in {
        "not stated", "not specified"
    }:
        timing_status = "source_timing_unspecified"

    elif not section_text:
        timing_status = "section_unavailable"

    elif timing in section_text:
        timing_status = "timing_text_found"

    else:
        expected_numbers = set(re.findall(r"\d+", timing))
        source_numbers = set(re.findall(r"\d+", section_text))

        timing_status = (
            "numeric_overlap_only"
            if expected_numbers & source_numbers
            else "timing_not_confirmed"
        )

    audit.append({
        "evidence_id": row["evidence_id"],
        "activity_name": row["official_drill_name"],
        "source_level": row["source_level"],
        "practice_plan": row["practice_plan"],
        "source_url": row["source_url"],
        "section_status": status,
        "matching_heading_sections": len(matches),
        "section_word_count": section_words,
        "section_fingerprint": fingerprint,
        "recorded_timing": row["source_timing"],
        "timing_screen": timing_status,
        "equipment_verified": "pending",
        "participants_verified": "pending",
        "distinctness_verified": "pending",
        "permission_verified": "pending",
        "planner_approved": "no",
    })


# --------------------------------------------------
# 5. Detect Identical Source Section Text
# --------------------------------------------------

fingerprint_groups = defaultdict(list)

for row in audit:
    if row["section_fingerprint"]:
        fingerprint_groups[row["section_fingerprint"]].append(row)

identical_groups = [
    rows for rows in fingerprint_groups.values()
    if len(rows) > 1
    and len({
        normalize(r["activity_name"]) for r in rows
    }) > 1
]


# --------------------------------------------------
# 6. Save Read-Only Findings
# --------------------------------------------------

with output_path.open(
    "x", newline="", encoding="utf-8"
) as f:

    writer = csv.DictWriter(
        f, fieldnames=list(audit[0].keys())
    )
    writer.writeheader()
    writer.writerows(audit)

print("\nCREATED:", output_path.relative_to(root))

# --------------------------------------------------
# 7. Report
# --------------------------------------------------

print("\n[1] ACTIVITY SECTIONS")

for status, count in Counter(
    r["section_status"] for r in audit
).items():
    print(f"  {status}: {count}")

print("\n[2] TIMING SCREEN")

for status, count in Counter(
    r["timing_screen"] for r in audit
).items():
    print(f"  {status}: {count}")

print("\n[3] IDENTICAL SECTION TEXT")

print(
    "Groups with different activity names:",
    len(identical_groups)
)

for group in identical_groups[:10]:
    print(
        "  " + " / ".join(
            r["activity_name"] for r in group
        )
    )

print("\n[4] RECORDS WITH VERY SHORT SECTIONS")

short_sections = [
    r for r in audit
    if (
        r["section_status"] == "section_found"
        and r["section_word_count"] < 10
    )
]

print("Sections with fewer than 10 words:", len(short_sections))

for row in short_sections[:10]:
    print(
        f"  {row['evidence_id']}: {row['activity_name']}"
    )

print("\n[5] FINAL STATUS")
print("Source occurrences audited:", len(audit))
print("Approved drills:", 0)
print("Full instructional content saved: No")
print("Existing evidence files modified: No")

print("\n" + "=" * 65)
print("DRILL SECTION AUDIT COMPLETE")
print("=" * 65)
```

```python
# Step 12-17 - Formal Basketball Drill Approval Checklist

import csv
from pathlib import Path
from collections import defaultdict, Counter

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - DRILL APPROVAL CHECKLIST")
print("=" * 65)

# --------------------------------------------------
# 1. Locate Project and Existing Reports
# --------------------------------------------------

root = next(
    p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (p / ".git").exists()
)

results = root / "results"

files = {
    "worklist": results / "basketball_drill_review_worklist.csv",
    "headings": results / "basketball_drill_heading_audit.csv",
    "sections": results / "basketball_drill_section_audit.csv",
    "timing": results / "basketball_targeted_timing_source_review.csv",
    "requirements": results / "basketball_requirement_source_review.csv",
}

output_path = (
    results / "basketball_drill_approval_checklist.csv"
)

def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)

    with path.open(newline="", encoding="utf-8-sig") as f:
        return list(csv.DictReader(f))

data = {
    name: read_csv(path)
    for name, path in files.items()
}

worklist = data["worklist"]

if len(worklist) != 67:
    raise ValueError("Expected 67 candidate groups.")

# --------------------------------------------------
# 2. Index Source Verification Results
# --------------------------------------------------

headings = {
    r["evidence_id"]: r
    for r in data["headings"]
}

sections = {
    r["evidence_id"]: r
    for r in data["sections"]
}

timing_reviews = {
    r["evidence_id"]: r
    for r in data["timing"]
}

requirement_reviews = defaultdict(list)

for row in data["requirements"]:
    requirement_reviews[row["evidence_id"]].append(row)

# --------------------------------------------------
# 3. Create Approval Checklist
# --------------------------------------------------

checklist = []

for group in worklist:

    evidence_ids = [
        value.strip()
        for value in group["evidence_ids"].split(";")
        if value.strip()
    ]

    if not evidence_ids:
        raise ValueError("Candidate group has no evidence IDs.")

    if any(
        eid not in headings or eid not in sections
        for eid in evidence_ids
    ):
        raise ValueError(
            "Missing heading or section evidence."
        )

    heading_pass = all(
        headings[eid]["heading_status"] == "heading_found"
        and headings[eid]["practice_plan_marker"] == "yes"
        for eid in evidence_ids
    )

    section_pass = all(
        sections[eid]["section_status"] == "section_found"
        for eid in evidence_ids
    )

    timing_reviewed = sum(
        eid in timing_reviews for eid in evidence_ids
    )

    equipment_reviewed = sum(
        any(
            r["requirement_type"] == "equipment"
            for r in requirement_reviews[eid]
        )
        for eid in evidence_ids
    )

    participants_reviewed = sum(
        any(
            r["requirement_type"] == "participants"
            for r in requirement_reviews[eid]
        )
        for eid in evidence_ids
    )

    actions = []

    if int(group["source_occurrences"]) > 1:
        actions.append("Compare repeated activity variations")

    if group["scope_flag"] == "general_warmup_candidate":
        actions.append("Confirm basketball-drill scope")

    if group["normalized_name"] in {
        "quick feet", "side push"
    }:
        actions.append("Investigate identical descriptions")

    if timing_reviewed:
        actions.append("Confirm recorded timing decisions")

    if equipment_reviewed or participants_reviewed:
        actions.append("Confirm requirement-review conclusions")

    actions.append("Verify all source requirements and permissions")

    checklist.append({
        "candidate_group_id": group["candidate_group_id"],
        "drill_name": group["display_name"],
        "evidence_ids": group["evidence_ids"],
        "source_locations": group["source_locations"],
        "source_occurrences": group["source_occurrences"],
        "proposed_skills": group["proposed_skills"],
        "scope_candidate": group["scope_flag"],

        # Automated evidence checks
        "heading_screen": (
            "pass" if heading_pass else "needs_review"
        ),
        "section_screen": (
            "pass" if section_pass else "needs_review"
        ),

        # Prior targeted evidence
        "timing_occurrences_reviewed": timing_reviewed,
        "equipment_occurrences_reviewed": equipment_reviewed,
        "participant_occurrences_reviewed": participants_reviewed,

        # Required human review gates
        "instructions_review": "pending",
        "timing_review": "pending",
        "equipment_review": "pending",
        "participants_review": "pending",
        "distinctness_review": "pending",
        "scope_review": "pending",
        "permission_review": "pending",

        # Final decisions
        "reviewer": "",
        "review_date": "",
        "review_notes": "",
        "next_actions": "; ".join(actions),
        "catalog_approval_status": "pending",
        "planner_eligible": "no",
    })

# --------------------------------------------------
# 4. Save Without Overwriting
# --------------------------------------------------

if output_path.exists():
    print("Existing checklist preserved:", output_path.name)

else:
    with output_path.open(
        "x", newline="", encoding="utf-8"
    ) as f:

        writer = csv.DictWriter(
            f,
            fieldnames=list(checklist[0].keys())
        )

        writer.writeheader()
        writer.writerows(checklist)

    print("CREATED:", output_path.relative_to(root))

# --------------------------------------------------
# 5. Verification Summary
# --------------------------------------------------

print("\n[1] CATALOG REVIEW SCOPE")

print("Preliminary distinct-name groups:", len(checklist))

print(
    "Source occurrences represented:",
    sum(int(r["source_occurrences"]) for r in checklist)
)

print("\n[2] AUTOMATED SOURCE EVIDENCE")

print(
    "Groups passing heading screen:",
    sum(r["heading_screen"] == "pass" for r in checklist)
)

print(
    "Groups passing section screen:",
    sum(r["section_screen"] == "pass" for r in checklist)
)

print("\n[3] EXISTING TARGETED REVIEWS")

print("Timing records reviewed:", len(timing_reviews))
print("Requirement records reviewed:",
      len(data["requirements"]))

print("\n[4] APPROVAL STATUS")

print("Pending candidate groups:", sum(
    r["catalog_approval_status"] == "pending"
    for r in checklist
))

print("Catalog-approved groups:", sum(
    r["catalog_approval_status"] == "approved"
    for r in checklist
))

print("Planner-eligible groups:", sum(
    r["planner_eligible"] == "yes"
    for r in checklist
))

print("\n[5] FILE INTEGRITY")

print("Existing evidence reports modified: No")
print("Existing drill catalog modified: No")
print("Automatic approvals performed: No")

print("\n" + "=" * 65)
print("FORMAL APPROVAL CHECKLIST INITIALIZED")
print("=" * 65)
```

```python
# Step 12-19 - Integrate Five-Drill Pilot into Approval Checklist

import csv
import os
import shutil
from pathlib import Path

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - PILOT REVIEW INTEGRATION")
print("=" * 65)

root = next(
    p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (p / ".git").exists()
)

results = root / "results"

checklist_path = (
    results / "basketball_drill_approval_checklist.csv"
)

pilot_path = (
    results / "basketball_pilot_source_reviews.csv"
)

backup_path = (
    results / "basketball_drill_approval_checklist_before_pilot.csv"
)

def read_csv(path):
    with path.open(newline="", encoding="utf-8-sig") as f:
        reader = csv.DictReader(f)
        return list(reader), list(reader.fieldnames or [])

# --------------------------------------------------
# 1. Load and Validate
# --------------------------------------------------

checklist, columns = read_csv(checklist_path)
pilot, _ = read_csv(pilot_path)

if len(checklist) != 67 or len(pilot) != 5:
    raise ValueError("Unexpected checklist or pilot size.")

new_fields = [
    "pilot_evidence_id",
    "pilot_source_fact_status",
    "pilot_scope",
    "pilot_source_notes",
    "pilot_review_reference",
]

if any(field in columns for field in new_fields):
    print("Pilot integration already appears to exist.")
    print("No files modified.")

elif backup_path.exists():
    print("A prior backup exists.")
    print("Review it before attempting integration.")
    print("No files modified.")

else:
    pilot_by_group = {
        row["candidate_group_id"]: row
        for row in pilot
    }

    if len(pilot_by_group) != 5:
        raise ValueError("Pilot group identifiers are not unique.")

    checklist_groups = {
        row["candidate_group_id"] for row in checklist
    }

    if not set(pilot_by_group).issubset(checklist_groups):
        raise ValueError("Pilot groups missing from checklist.")

    # --------------------------------------------------
    # 2. Add Evidence Metadata
    # --------------------------------------------------

    updated_count = 0

    for row in checklist:

        gid = row["candidate_group_id"]

        for field in new_fields:
            row[field] = ""

        if gid not in pilot_by_group:
            continue

        review = pilot_by_group[gid]

        evidence_ids = {
            eid.strip()
            for eid in row["evidence_ids"].split(";")
        }

        if review["evidence_id"] not in evidence_ids:
            raise ValueError(
                f"Evidence mismatch for group {gid}"
            )

        row["pilot_evidence_id"] = review["evidence_id"]
        row["pilot_source_fact_status"] = (
            "source_metadata_documented"
        )
        row["pilot_scope"] = review["proposed_scope"]
        row["pilot_source_notes"] = review["source_notes"]
        row["pilot_review_reference"] = (
            "results/basketball_pilot_source_reviews.csv"
        )

        # Never automatically pass human review gates.
        row["catalog_approval_status"] = "pending"
        row["planner_eligible"] = "no"

        updated_count += 1

    if updated_count != 5:
        raise ValueError("Expected exactly five pilot updates.")

    # --------------------------------------------------
    # 3. Back Up and Update Working Checklist
    # --------------------------------------------------

    shutil.copy2(checklist_path, backup_path)

    temp_path = checklist_path.with_name(
        checklist_path.name + ".tmp"
    )

    try:
        with temp_path.open(
            "x", newline="", encoding="utf-8"
        ) as f:
            writer = csv.DictWriter(
                f,
                fieldnames=columns + new_fields
            )
            writer.writeheader()
            writer.writerows(checklist)

        os.replace(temp_path, checklist_path)

    finally:
        if temp_path.exists():
            temp_path.unlink()

    # --------------------------------------------------
    # 4. Verify Saved Data
    # --------------------------------------------------

    saved, _ = read_csv(checklist_path)

    print("\n[1] INTEGRATION")
    print("Main candidate groups:", len(saved))
    print("Pilot reviews integrated:", sum(
        bool(r["pilot_evidence_id"]) for r in saved
    ))

    print("\n[2] APPROVAL STATUS")
    print("Pending groups:", sum(
        r["catalog_approval_status"] == "pending"
        for r in saved
    ))
    print("Planner-eligible groups:", sum(
        r["planner_eligible"] == "yes"
        for r in saved
    ))

    print("\n[3] FILE MANAGEMENT")
    print("Backup:", backup_path.name)
    print("Updated:", checklist_path.name)
    print("Original evidence batches modified: No")
    print("GitHub modified: No")

print("\n" + "=" * 65)
print("PILOT INTEGRATION CHECK FINISHED")
print("=" * 65)
```

```python
# Step 12-20 - Integrate Source-Verified Pilot Decisions

import csv
import os
import shutil
import tempfile
from pathlib import Path
from datetime import date
from collections import Counter

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - PILOT SOURCE DECISIONS")
print("=" * 65)

root = next(
    p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (p / ".git").exists()
)

results = root / "results"

checklist_path = (
    results / "basketball_drill_approval_checklist.csv"
)

evidence_path = (
    results / "basketball_drill_source_evidence.csv"
)

heading_path = (
    results / "basketball_drill_heading_audit.csv"
)

section_path = (
    results / "basketball_drill_section_audit.csv"
)

backup_path = (
    results / "basketball_drill_approval_checklist"
    "_before_source_decisions.csv"
)

def read_csv(path):
    with path.open(newline="", encoding="utf-8-sig") as f:
        reader = csv.DictReader(f)
        return list(reader), list(reader.fieldnames or [])

checklist, columns = read_csv(checklist_path)
evidence, _ = read_csv(evidence_path)
headings, _ = read_csv(heading_path)
sections, _ = read_csv(section_path)

if len(checklist) != 67 or len(evidence) != 5:
    raise ValueError("Unexpected pilot or checklist size.")

# --------------------------------------------------
# 1. Define Source-Reviewed Pilot Facts
# --------------------------------------------------

# These are short factual records, not copied instructions.
# All rights and youth-planning approvals remain pending.

pilot_facts = {
    "JR-EVID-0001": {
        "scope": "general_warmup",
        "target": "supplemental_not_counted",
        "prescription": "1 x 3 minutes",
        "equipment": "No ball for base activity",
        "space": "Court or gym",
        "participants": "Coach-directed group",
        "note": "General movement and stopping activity."
    },
    "JR-EVID-0002": {
        "scope": "ball_handling_warmup",
        "target": "supplemental_not_counted",
        "prescription": "1 x 20 seconds",
        "equipment": "Basketball; sharing permitted",
        "space": "Designated practice area",
        "participants": "Individual base activity",
        "note": "Basketball familiarization warm-up."
    },
    "JR-EVID-0003": {
        "scope": "basketball_skill",
        "target": "candidate_pending_approval",
        "prescription": "3 x 20-second segments",
        "equipment": "Basketball; sharing permitted",
        "space": "Court sideline",
        "participants": (
            "Individual base activity; "
            "optional partner variation"
        ),
        "note": (
            "Low, middle, and high crossover variations. "
            "Source timing excludes unspecified transitions."
        )
    },
    "JR-EVID-0004": {
        "scope": "basketball_skill",
        "target": "candidate_pending_duplicate_review",
        "prescription": "1 x 90 seconds (Starter Plan 2)",
        "equipment": "Basketball",
        "space": "Half court",
        "participants": "Coach-directed group",
        "note": (
            "Also appears in Rookie Plan 2, with a "
            "different prescription; compare variations."
        )
    },
    "JR-EVID-0005": {
        "scope": "basketball_skill",
        "target": "candidate_pending_approval",
        "prescription": "3 x 15-second intervals",
        "equipment": "Basketball; sharing permitted",
        "space": "Court with baseline and other markings",
        "participants": "Individual base activity",
        "note": (
            "Retreat-dribble sequence; source time is "
            "not a complete session-duration estimate."
        )
    },
}

evidence_by_id = {
    r["evidence_id"]: r for r in evidence
}
heading_by_id = {
    r["evidence_id"]: r for r in headings
}
section_by_id = {
    r["evidence_id"]: r for r in sections
}

if set(pilot_facts) != set(evidence_by_id):
    raise ValueError("Pilot evidence IDs do not match.")

for eid in pilot_facts:
    if (
        heading_by_id[eid]["heading_status"] != "heading_found"
        or section_by_id[eid]["section_status"] != "section_found"
    ):
        raise ValueError(f"Source verification missing: {eid}")

# --------------------------------------------------
# 2. Define Additional Checklist Fields
# --------------------------------------------------

new_fields = [
    "pilot_source_review_version",
    "pilot_source_review_date",
    "pilot_reviewed_prescription",
    "pilot_reviewed_equipment",
    "pilot_reviewed_space",
    "pilot_reviewed_participants",
    "pilot_target_count_decision",
    "pilot_verified_source_url",
    "youth_planning_review",
]

already_integrated = "pilot_source_review_version" in columns

if already_integrated:
    print("\nPilot source decisions already integrated.")
    print("Existing checklist preserved.")

else:

    if backup_path.exists():
        raise FileExistsError(
            "A previous source-review backup exists. "
            "Review it before making further changes."
        )

    # Do not overwrite existing approval decisions.
    if any(
        r["catalog_approval_status"] != "pending"
        or r["planner_eligible"] != "no"
        for r in checklist
    ):
        raise ValueError(
            "Existing approval decisions detected. "
            "Manual reconciliation required."
        )

    updated = 0

    # --------------------------------------------------
    # 3. Integrate Five Pilot Reviews
    # --------------------------------------------------

    for row in checklist:

        for field in new_fields:
            row[field] = (
                "pending"
                if field == "youth_planning_review"
                else ""
            )

        group_ids = {
            eid.strip()
            for eid in row["evidence_ids"].split(";")
        }

        matching = group_ids & set(pilot_facts)

        if not matching:
            continue

        if len(matching) != 1:
            raise ValueError(
                "Unexpected multiple pilot records in one group."
            )

        eid = next(iter(matching))
        facts = pilot_facts[eid]
        source = evidence_by_id[eid]

        row["pilot_source_review_version"] = "v1"
        row["pilot_source_review_date"] = date.today().isoformat()

        row["pilot_source_fact_status"] = (
            "official_webpage_facts_checked"
        )

        row["pilot_scope"] = facts["scope"]
        row["pilot_reviewed_prescription"] = facts["prescription"]
        row["pilot_reviewed_equipment"] = facts["equipment"]
        row["pilot_reviewed_space"] = facts["space"]
        row["pilot_reviewed_participants"] = facts["participants"]
        row["pilot_target_count_decision"] = facts["target"]
        row["pilot_verified_source_url"] = source["source_url"]

        # Source review is separate from final team approval.
        row["instructions_review"] = (
            "source_checked_pending_team_signoff"
        )
        row["timing_review"] = (
            "source_checked_pending_team_signoff"
        )
        row["equipment_review"] = (
            "source_checked_pending_team_signoff"
        )
        row["participants_review"] = (
            "source_checked_pending_team_signoff"
        )
        row["scope_review"] = facts["scope"]

        row["distinctness_review"] = (
            "repeated_source_variation_pending"
            if int(row["source_occurrences"]) > 1
            else "preliminary_unique_name"
        )

        row["permission_review"] = "pending"
        row["youth_planning_review"] = "pending"
        row["catalog_approval_status"] = "pending"
        row["planner_eligible"] = "no"

        row["review_notes"] = (
            (row["review_notes"] + " | ")
            if row["review_notes"] else ""
        ) + facts["note"]

        updated += 1

    if updated != 5:
        raise ValueError(
            f"Expected five pilot updates, found {updated}."
        )

    # --------------------------------------------------
    # 4. Write Safely With Backup
    # --------------------------------------------------

    temporary = None

    try:
        with tempfile.NamedTemporaryFile(
            mode="w",
            newline="",
            encoding="utf-8",
            suffix=".tmp",
            dir=results,
            delete=False
        ) as f:

            temporary = Path(f.name)

            writer = csv.DictWriter(
                f,
                fieldnames=columns + new_fields
            )

            writer.writeheader()
            writer.writerows(checklist)

        # Validate staged file before changing checklist.
        staged, _ = read_csv(temporary)

        if len(staged) != 67:
            raise ValueError("Staged record count changed.")

        shutil.copy2(checklist_path, backup_path)
        os.replace(temporary, checklist_path)

        print("\nBackup created:", backup_path.name)
        print("Updated:", checklist_path.name)

    finally:
        if temporary and temporary.exists():
            temporary.unlink()

# --------------------------------------------------
# 5. Final Verification
# --------------------------------------------------

saved, _ = read_csv(checklist_path)

reviewed = [
    row for row in saved
    if row.get("pilot_source_review_version") == "v1"
]

scope_counts = Counter(
    row["scope_review"] for row in reviewed
)

print("\n[1] PILOT REVIEW SUMMARY")
print("Total candidate groups:", len(saved))
print("Source-reviewed pilot groups:", len(reviewed))

for scope, count in scope_counts.items():
    print(f"{scope}: {count}")

print("\n[2] TARGET STATUS")

print("Basketball-skill candidates:", sum(
    r["scope_review"] == "basketball_skill"
    for r in reviewed
))

print("Supplemental warm-ups:", sum(
    r["pilot_target_count_decision"] == "supplemental_not_counted"
    for r in reviewed
))

print("\n[3] APPROVAL GATES")

print("Pending groups:", sum(
    r["catalog_approval_status"] == "pending"
    for r in saved
))

print("Planner-approved groups:", sum(
    r["planner_eligible"] == "yes"
    for r in saved
))

print("Permission review: Pending")
print("Youth planning review: Pending")

print("\n[4] INTEGRITY")
print("Original evidence files modified: No")
print("GitHub modified: No")

print("\n" + "=" * 65)
print("PILOT SOURCE REVIEW INTEGRATION COMPLETE")
print("=" * 65)
```

```python
# Step 12-21 - Prioritize the Next Basketball Drill Review Batch

import csv
from pathlib import Path

root = next(
    p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (p / ".git").exists()
)

results = root / "results"

def read_csv(path):
    with path.open(newline="", encoding="utf-8-sig") as f:
        return list(csv.DictReader(f))

checklist = read_csv(
    results / "basketball_drill_approval_checklist.csv"
)

batch_files = [
    results / "basketball_drill_source_evidence.csv",
    *[
        results / f"basketball_drill_source_evidence_batch{i}.csv"
        for i in range(2, 6)
    ],
]

evidence = {}
for path in batch_files:
    for row in read_csv(path):
        evidence[row["evidence_id"]] = row

section_audit = {
    r["evidence_id"]: r
    for r in read_csv(
        results / "basketball_drill_section_audit.csv"
    )
}

review_candidates = []

for group in checklist:

    # Previously reviewed groups are excluded.
    if group.get("pilot_source_review_version") == "v1":
        continue

    # Start with basketball-specific, single-source records.
    if group["scope_candidate"] != "basketball_drill_candidate":
        continue

    if int(group["source_occurrences"]) != 1:
        continue

    eid = group["evidence_ids"].strip()

    if eid not in evidence or eid not in section_audit:
        continue

    source = evidence[eid]
    audit = section_audit[eid]

    concerns = []

    if audit["timing_screen"] != "timing_text_found":
        concerns.append("timing")

    if not source["factual_summary"].strip():
        concerns.append("summary")

    equipment = source["source_equipment"].lower()

    if any(word in equipment for word in [
        "not explicitly", "optional", "or shared",
        "not specified"
    ]):
        concerns.append("equipment")

    participants = source["participant_requirement"].lower()

    if any(word in participants for word in [
        "not stated", "or coach", "or receiver",
        "not specified"
    ]):
        concerns.append("participants")

    review_candidates.append({
        "group_id": group["candidate_group_id"],
        "evidence_id": eid,
        "name": source["official_drill_name"],
        "level": source["source_level"],
        "plan": source["practice_plan"],
        "skill": source["proposed_skill"],
        "source_url": source["source_url"],
        "concerns": concerns,
    })

# Prioritize records with fewer preliminary concerns.
review_candidates.sort(
    key=lambda r: (
        len(r["concerns"]),
        r["level"],
        r["name"]
    )
)

selected = review_candidates[:12]

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - NEXT DRILL REVIEW BATCH")
print("=" * 65)

print("\nEligible for prioritization:", len(review_candidates))
print("Selected for next source review:", len(selected))

for i, item in enumerate(selected, start=1):
    print(f"\n[{i}] {item['name']}")
    print("Evidence:", item["evidence_id"])
    print("Source:", item["level"], "Plan", item["plan"])
    print("Skill:", item["skill"])
    print("Preliminary concerns:",
          ", ".join(item["concerns"]) or "None flagged")
    print("URL:", item["source_url"])

print("\nNo existing files were modified.")
print("No drills were automatically approved.")
print("=" * 65)
```

```python
# Step 12-22 - Integrate 12 Source-Reviewed Basketball Drills

import csv
import os
import shutil
import tempfile
from pathlib import Path
from datetime import date

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - 12-DRILL SOURCE REVIEW")
print("=" * 65)

root = next(
    p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (p / ".git").exists()
)

results = root / "results"

checklist_path = results / "basketball_drill_approval_checklist.csv"
backup_path = results / "basketball_drill_approval_before_batch12.csv"

def read_csv(path):
    with path.open(newline="", encoding="utf-8-sig") as f:
        reader = csv.DictReader(f)
        return list(reader), list(reader.fieldnames or [])

checklist, columns = read_csv(checklist_path)

if len(checklist) != 67:
    raise ValueError("Expected 67 candidate groups.")

# --------------------------------------------------
# 1. Source-Reviewed Factual Metadata
# --------------------------------------------------

# Fields:
# skill, source prescription, equipment, participants,
# minimum active players, space, review note.
#
# These values reflect factual source review.
# They do not represent youth-safety or permission approval.

facts = {
    "JR-EVID-0031": (
        "ball_handling", "5 minutes",
        "Basketball", "Offense and defender", "2",
        "Half court", "Live defensive pressure."
    ),
    "JR-EVID-0056": (
        "ball_handling", "5-7 minutes",
        "Basketball", "One attacker and two defenders", "3",
        "Half court", "Two-on-one defensive pressure."
    ),
    "JR-EVID-0061": (
        "passing", "5 minutes",
        "Basketball and hoop", "Three attackers, one defender",
        "4", "Court with three-point line",
        "Passing decisions following a drive."
    ),
    "JR-EVID-0033": (
        "transition_offense", "5 minutes",
        "Basketball and hoop", "Three attackers, one defender",
        "4", "Court with transition space",
        "Transition attack against one defender."
    ),
    "JR-EVID-0037": (
        "team_offense", "Games to 3 makes",
        "Basketball and hoop", "Two teams of three",
        "6", "Half court",
        "Possessions begin with prescribed offensive action."
    ),
    "JR-EVID-0062": (
        "team_defense", "Games to 3",
        "Basketball and hoop", "Two teams of three",
        "6", "Half court",
        "Help-defense rotation before live play."
    ),
    "JR-EVID-0059": (
        "shooting", "2 x 10 shots each",
        "Basketball and hoop", "Shooter, passer, rebounder",
        "3", "Court with hoop",
        "Players rotate among three roles."
    ),
    "JR-EVID-0035": (
        "defense", "5-7 minutes",
        "Basketball and hoop",
        "Attacker, defender, passer", "3",
        "Court with wings and key",
        "Pass denial followed by live one-on-one."
    ),
    "JR-EVID-0038": (
        "team_offense", "Games to 4 makes",
        "Basketball and two hoops", "Two teams of five",
        "10", "Full court",
        "Four passes required before shooting."
    ),
    "JR-EVID-0058": (
        "finishing", "5 minutes",
        "Basketballs and hoop", "Multiple players in group",
        "", "Court with hoop",
        "Group competition for possession and finishing."
    ),
    "JR-EVID-0060": (
        "off_ball_movement", "3-4 minutes",
        "Basketballs and hoop",
        "Attacker, defender, coach or player passer",
        "3", "Court with baseline and wing",
        "Moving to receive a pass against a defender."
    ),
    "JR-EVID-0032": (
        "footwork", "6 minutes",
        "Basketball and hoop", "Attacker and defender",
        "2", "Court inside three-point line",
        "Pivoting to create scoring space."
    ),
}

# --------------------------------------------------
# 2. Validate Source Evidence and Selected Groups
# --------------------------------------------------

evidence = {}

for batch in range(1, 6):
    filename = (
        "basketball_drill_source_evidence.csv"
        if batch == 1
        else f"basketball_drill_source_evidence_batch{batch}.csv"
    )

    rows, _ = read_csv(results / filename)

    for row in rows:
        eid = row["evidence_id"]

        if eid in evidence:
            raise ValueError(f"Repeated evidence ID: {eid}")

        evidence[eid] = row

heading_rows, _ = read_csv(
    results / "basketball_drill_heading_audit.csv"
)

section_rows, _ = read_csv(
    results / "basketball_drill_section_audit.csv"
)

headings = {r["evidence_id"]: r for r in heading_rows}
sections = {r["evidence_id"]: r for r in section_rows}

group_by_eid = {}

for group in checklist:
    for eid in group["evidence_ids"].split(";"):
        eid = eid.strip()

        if eid:
            group_by_eid[eid] = group

for eid in facts:

    if eid not in evidence or eid not in group_by_eid:
        raise ValueError(f"Missing source/group: {eid}")

    if (
        headings[eid]["heading_status"] != "heading_found"
        or headings[eid]["practice_plan_marker"] != "yes"
        or sections[eid]["section_status"] != "section_found"
    ):
        raise ValueError(f"Source checks not passed: {eid}")

    group = group_by_eid[eid]

    if (
        group["catalog_approval_status"] != "pending"
        or group["planner_eligible"] != "no"
        or int(group["source_occurrences"]) != 1
    ):
        raise ValueError(f"Unexpected review state: {eid}")

# --------------------------------------------------
# 3. Add Shared Review Fields
# --------------------------------------------------

new_fields = [
    "source_review_stage",
    "reviewed_skill_area",
    "reviewed_prescription",
    "reviewed_equipment",
    "reviewed_participants",
    "reviewed_minimum_active_players",
    "reviewed_space",
    "reviewed_source_url",
    "reviewed_activity_scope",
    "reviewed_metadata_notes",
    "source_metadata_review_date",
]

if any(field in columns for field in new_fields):
    raise ValueError(
        "Shared review fields already exist. "
        "Inspect the checklist before rerunning."
    )

if backup_path.exists():
    raise FileExistsError(
        "Existing Batch 12 backup found. No changes made."
    )

# Initialize shared fields for every group.
for group in checklist:

    for field in new_fields:
        group[field] = ""

    group["source_review_stage"] = "not_reviewed"

# Preserve the five previously documented pilot reviews
# in the new shared review fields.

pilot_count = 0

for group in checklist:

    if group.get("pilot_source_review_version") != "v1":
        continue

    group["source_review_stage"] = (
        "metadata_reviewed_pending_signoff"
    )
    group["reviewed_skill_area"] = group["pilot_scope"]
    group["reviewed_prescription"] = group["pilot_reviewed_prescription"]
    group["reviewed_equipment"] = group["pilot_reviewed_equipment"]
    group["reviewed_participants"] = group["pilot_reviewed_participants"]
    group["reviewed_space"] = group["pilot_reviewed_space"]
    group["reviewed_source_url"] = group["pilot_verified_source_url"]
    group["reviewed_activity_scope"] = group["pilot_scope"]
    group["reviewed_metadata_notes"] = group["pilot_source_notes"]
    group["source_metadata_review_date"] = (
        group["pilot_source_review_date"]
    )

    pilot_count += 1

if pilot_count != 5:
    raise ValueError("Expected five existing pilot reviews.")

# --------------------------------------------------
# 4. Integrate Twelve New Source Reviews
# --------------------------------------------------

updated = 0

for eid, values in facts.items():

    (
        skill, prescription, equipment,
        participants, minimum_players, space, note
    ) = values

    group = group_by_eid[eid]
    source = evidence[eid]

    if group["source_review_stage"] != "not_reviewed":
        raise ValueError(f"Already reviewed: {eid}")

    group["source_review_stage"] = (
        "metadata_reviewed_pending_signoff"
    )
    group["reviewed_skill_area"] = skill
    group["reviewed_prescription"] = prescription
    group["reviewed_equipment"] = equipment
    group["reviewed_participants"] = participants
    group["reviewed_minimum_active_players"] = minimum_players
    group["reviewed_space"] = space
    group["reviewed_source_url"] = source["source_url"]
    group["reviewed_activity_scope"] = "basketball_skill_candidate"
    group["reviewed_metadata_notes"] = note
    group["source_metadata_review_date"] = date.today().isoformat()

    # Preserve human review and planner gates.
    group["catalog_approval_status"] = "pending"
    group["planner_eligible"] = "no"

    updated += 1

if updated != 12:
    raise ValueError("Expected twelve new source reviews.")

# --------------------------------------------------
# 5. Stage, Back Up, and Save Safely
# --------------------------------------------------

temp_path = None

try:
    with tempfile.NamedTemporaryFile(
        mode="w",
        newline="",
        encoding="utf-8",
        suffix=".tmp",
        dir=results,
        delete=False
    ) as f:

        temp_path = Path(f.name)

        writer = csv.DictWriter(
            f, fieldnames=columns + new_fields
        )
        writer.writeheader()
        writer.writerows(checklist)

    staged, _ = read_csv(temp_path)

    assert len(staged) == 67
    assert sum(
        r["source_review_stage"]
        == "metadata_reviewed_pending_signoff"
        for r in staged
    ) == 17

    assert all(
        r["catalog_approval_status"] == "pending"
        and r["planner_eligible"] == "no"
        for r in staged
    )

    shutil.copy2(checklist_path, backup_path)
    os.replace(temp_path, checklist_path)

finally:
    if temp_path and temp_path.exists():
        temp_path.unlink()

# --------------------------------------------------
# 6. Verify Updated Checklist
# --------------------------------------------------

saved, _ = read_csv(checklist_path)

reviewed_count = sum(
    r["source_review_stage"]
    == "metadata_reviewed_pending_signoff"
    for r in saved
)

print("\n[1] SOURCE REVIEW RESULTS")
print("Previously documented pilot reviews:", pilot_count)
print("New source reviews integrated:", updated)
print("Total metadata-reviewed groups:", reviewed_count)
print("Groups not yet metadata-reviewed:", 67 - reviewed_count)

print("\n[2] APPROVAL STATUS")
print("Candidate groups:", len(saved))
print("Pending approvals:", sum(
    r["catalog_approval_status"] == "pending"
    for r in saved
))
print("Planner-eligible groups:", sum(
    r["planner_eligible"] == "yes"
    for r in saved
))

print("\n[3] FILES")
print("Backup:", backup_path.name)
print("Updated:", checklist_path.name)
print("Original evidence files modified: No")
print("GitHub modified: No")

print("\n" + "=" * 65)
print("BATCH SOURCE REVIEW INTEGRATION COMPLETE")
print("=" * 65)
```

```python
# Step 12-23 - Prefill Remaining Drill Review Candidates

import csv
import json
import os
import shutil
import tempfile
from pathlib import Path
from collections import Counter

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - DRAFT REVIEW PREPARATION")
print("=" * 65)

# --------------------------------------------------
# 1. Locate Project and Read Existing Files
# --------------------------------------------------

root = next(
    p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (p / ".git").exists()
)

results = root / "results"

checklist_path = results / "basketball_drill_approval_checklist.csv"
backup_path = results / "basketball_drill_approval_before_draft_prefill.csv"

def load_csv(path):
    with path.open(newline="", encoding="utf-8-sig") as f:
        reader = csv.DictReader(f)
        return list(reader), list(reader.fieldnames or [])

checklist, original_columns = load_csv(checklist_path)

if len(checklist) != 67:
    raise ValueError("Expected 67 candidate groups.")

batch_paths = [
    results / "basketball_drill_source_evidence.csv",
    *[
        results / f"basketball_drill_source_evidence_batch{i}.csv"
        for i in range(2, 6)
    ]
]

evidence = {}

for path in batch_paths:
    rows, _ = load_csv(path)

    for row in rows:
        eid = row["evidence_id"]

        if eid in evidence:
            raise ValueError(f"Duplicate evidence ID: {eid}")

        evidence[eid] = row

if len(evidence) != 70:
    raise ValueError("Expected 70 source occurrences.")

# --------------------------------------------------
# 2. Define Draft Fields
# --------------------------------------------------

draft_fields = [
    "draft_skill_areas",
    "draft_source_prescriptions",
    "draft_equipment_requirements",
    "draft_participant_requirements",
    "draft_space_requirements",
    "draft_source_details_json",
    "draft_review_flags",
    "draft_review_status",
]

if any(field in original_columns for field in draft_fields):
    raise ValueError(
        "Draft fields already exist. "
        "Review the checklist before rerunning."
    )

if backup_path.exists():
    raise FileExistsError(
        "Existing prefill backup found. No changes made."
    )

# Keep copies for checking existing reviews.
existing_reviews = {
    r["candidate_group_id"]: dict(r)
    for r in checklist
    if r["source_review_stage"]
    == "metadata_reviewed_pending_signoff"
}

if len(existing_reviews) != 17:
    raise ValueError("Expected 17 existing metadata reviews.")

# --------------------------------------------------
# 3. Prepare Draft Information
# --------------------------------------------------

prefilled = 0
flag_counts = Counter()

def combine_values(rows, field):
    values = [
        row[field].strip()
        for row in rows
        if row.get(field, "").strip()
    ]
    return " | ".join(dict.fromkeys(values))

for group in checklist:

    for field in draft_fields:
        group[field] = ""

    if group["candidate_group_id"] in existing_reviews:
        group["draft_review_status"] = "not_required"
        continue

    if group["source_review_stage"] != "not_reviewed":
        raise ValueError(
            "Unexpected review stage for "
            + group["candidate_group_id"]
        )

    ids = [
        eid.strip()
        for eid in group["evidence_ids"].split(";")
        if eid.strip()
    ]

    if not ids or any(eid not in evidence for eid in ids):
        raise ValueError("Missing source evidence.")

    rows = [evidence[eid] for eid in ids]

    details = []

    for row in rows:
        details.append({
            "evidence_id": row["evidence_id"],
            "source_url": row["source_url"],
            "source_level": row["source_level"],
            "practice_plan": row["practice_plan"],
            "activity_name": row["official_drill_name"],
            "proposed_skill": row["proposed_skill"],
            "source_timing": row["source_timing"],
            "source_equipment": row["source_equipment"],
            "participant_requirement": row["participant_requirement"],
            "source_space": row["source_space"],
            "factual_summary": row["factual_summary"],
        })

    flags = []

    if len(rows) > 1:
        flags.append("multiple_source_occurrences")

    if group["scope_candidate"] == "general_warmup_candidate":
        flags.append("warmup_scope_review")

    if any(not r["factual_summary"].strip() for r in rows):
        flags.append("missing_summary")

    if any(
        r["source_timing"].strip().casefold()
        in {"", "not stated", "not specified"}
        for r in rows
    ):
        flags.append("timing_requires_review")

    if any(
        r["source_equipment"].strip().casefold()
        in {"", "not specified", "unknown"}
        for r in rows
    ):
        flags.append("equipment_requires_review")

    # The original evidence has not passed full human review.
    flags.append("metadata_pending_independent_verification")

    group["draft_skill_areas"] = combine_values(
        rows, "proposed_skill"
    )
    group["draft_source_prescriptions"] = combine_values(
        rows, "source_timing"
    )
    group["draft_equipment_requirements"] = combine_values(
        rows, "source_equipment"
    )
    group["draft_participant_requirements"] = combine_values(
        rows, "participant_requirement"
    )
    group["draft_space_requirements"] = combine_values(
        rows, "source_space"
    )
    group["draft_source_details_json"] = json.dumps(
        details, ensure_ascii=False
    )
    group["draft_review_flags"] = ";".join(flags)
    group["draft_review_status"] = "pending_source_review"

    flag_counts.update(flags)
    prefilled += 1

if prefilled != 50:
    raise ValueError(
        f"Expected 50 draft reviews; found {prefilled}."
    )

# --------------------------------------------------
# 4. Verify Existing Reviews Were Preserved
# --------------------------------------------------

for group in checklist:
    gid = group["candidate_group_id"]

    if gid not in existing_reviews:
        continue

    old = existing_reviews[gid]

    if any(
        group[col] != old[col]
        for col in original_columns
    ):
        raise ValueError(
            f"Existing review changed unexpectedly: {gid}"
        )

if any(
    row["catalog_approval_status"] != "pending"
    or row["planner_eligible"] != "no"
    for row in checklist
):
    raise ValueError("Unexpected approval status.")

# --------------------------------------------------
# 5. Stage and Save With Backup
# --------------------------------------------------

temporary = None

try:
    with tempfile.NamedTemporaryFile(
        mode="w",
        newline="",
        encoding="utf-8",
        suffix=".tmp",
        dir=results,
        delete=False
    ) as f:

        temporary = Path(f.name)
        writer = csv.DictWriter(
            f, fieldnames=original_columns + draft_fields
        )
        writer.writeheader()
        writer.writerows(checklist)

    staged, _ = load_csv(temporary)

    assert len(staged) == 67

    assert sum(
        r["draft_review_status"] == "pending_source_review"
        for r in staged
    ) == 50

    shutil.copy2(checklist_path, backup_path)
    os.replace(temporary, checklist_path)

finally:
    if temporary and temporary.exists():
        temporary.unlink()

# --------------------------------------------------
# 6. Final Verification
# --------------------------------------------------

saved, _ = load_csv(checklist_path)

print("\n[1] REVIEW PREPARATION")
print("Total candidate groups:", len(saved))
print("Existing metadata-reviewed groups:", len(existing_reviews))
print("New draft groups populated:", prefilled)

print("\n[2] DRAFT FLAGS")
for flag, count in flag_counts.most_common():
    print(f"{flag:<42} {count}")

print("\n[3] APPROVAL STATUS")
print("Catalog-approved groups:", sum(
    r["catalog_approval_status"] == "approved"
    for r in saved
))
print("Planner-eligible groups:", sum(
    r["planner_eligible"] == "yes"
    for r in saved
))

print("\n[4] FILE MANAGEMENT")
print("Backup:", backup_path.name)
print("Updated:", checklist_path.name)
print("Existing reviews preserved: Yes")
print("Original source evidence modified: No")
print("GitHub modified: No")

print("\n" + "=" * 65)
print("DRAFT REVIEW PREPARATION COMPLETE")
print("=" * 65)
```

```python

# Step 12-24 - Build Versioned Basketball Drill Catalog v0.1

import csv
import re
from pathlib import Path
from collections import Counter

print("=" * 65)
print("CHAMPIONS ATHLETIC MAP - BASKETBALL CATALOG v0.1")
print("=" * 65)

# --------------------------------------------------
# 1. Locate Project
# --------------------------------------------------

cwd = Path.cwd().resolve()

root = next(
    (p for p in (cwd, *cwd.parents) if (p / ".git").exists()),
    None
)

if root is None:
    raise FileNotFoundError("Git project root not found.")

results = root / "results"

checklist_path = (
    results / "basketball_drill_approval_checklist.csv"
)

output_path = (
    root / "data" / "processed"
    / "basketball_drill_catalog_v0_1.csv"
)


def load_csv(path):
    if not path.is_file():
        raise FileNotFoundError(f"Required file not found: {path}")

    with path.open(newline="", encoding="utf-8-sig") as f:
        reader = csv.DictReader(f)
        return list(reader), list(reader.fieldnames or [])


checklist, checklist_columns = load_csv(checklist_path)

# --------------------------------------------------
# 2. Validate Checklist Structure
# --------------------------------------------------

required_columns = {
    "candidate_group_id",
    "drill_name",
    "evidence_ids",
    "source_occurrences",
    "scope_candidate",
    "source_review_stage",
    "draft_review_status",
    "draft_skill_areas",
    "draft_source_prescriptions",
    "draft_equipment_requirements",
    "draft_participant_requirements",
    "draft_space_requirements",
    "reviewed_skill_area",
    "reviewed_prescription",
    "reviewed_equipment",
    "reviewed_participants",
    "reviewed_space",
    "heading_screen",
    "section_screen",
    "distinctness_review",
    "permission_review",
    "youth_planning_review",
    "catalog_approval_status",
    "planner_eligible",
}

missing_columns = required_columns - set(checklist_columns)

if missing_columns:
    raise ValueError(
        "Missing checklist columns: "
        + ", ".join(sorted(missing_columns))
    )

if len(checklist) != 67:
    raise ValueError(
        f"Expected 67 candidate groups, found {len(checklist)}."
    )

group_ids = [
    row["candidate_group_id"]
    for row in checklist
]

if len(group_ids) != len(set(group_ids)):
    raise ValueError("Candidate group IDs are not unique.")

print("\n[1] CHECKLIST VERIFICATION")
print("Candidate groups:", len(checklist))
print("Required columns verified: True")

# --------------------------------------------------
# 3. Load Original Evidence Records
# --------------------------------------------------

evidence = {}

for batch in range(1, 6):

    filename = (
        "basketball_drill_source_evidence.csv"
        if batch == 1
        else f"basketball_drill_source_evidence_batch{batch}.csv"
    )

    batch_path = results / filename

    rows, _ = load_csv(batch_path)

    for row in rows:

        eid = row["evidence_id"]

        if eid in evidence:
            raise ValueError(
                f"Duplicate evidence ID detected: {eid}"
            )

        evidence[eid] = row

if len(evidence) != 70:
    raise ValueError(
        f"Expected 70 evidence records, found {len(evidence)}."
    )

print("\n[2] SOURCE EVIDENCE")
print("Evidence records loaded:", len(evidence))
print("Evidence batches loaded: 5")

# --------------------------------------------------
# 4. Define Name Normalization
# --------------------------------------------------

def normalize_name(value):
    return re.sub(
        r"[^a-z0-9]+",
        " ",
        (value or "").casefold()
    ).strip()

# --------------------------------------------------
# 5. Build One Record per Preliminary Name Group
# --------------------------------------------------

catalog = []
used_evidence_ids = []

for group in checklist:

    # Generate the normalized name from drill_name.
    # The checklist does not contain a normalized_name column.

    normalized_name = normalize_name(
        group["drill_name"]
    )

    if not normalized_name:
        raise ValueError(
            "Empty normalized drill name for "
            + group["candidate_group_id"]
        )

    ids = [
        eid.strip()
        for eid in group["evidence_ids"].split(";")
        if eid.strip()
    ]

    if not ids:
        raise ValueError(
            "Missing evidence IDs for "
            + group["candidate_group_id"]
        )

    if any(eid not in evidence for eid in ids):
        raise ValueError(
            "Invalid evidence reference for "
            + group["candidate_group_id"]
        )

    used_evidence_ids.extend(ids)

    source_rows = [
        evidence[eid]
        for eid in ids
    ]

    if int(group["source_occurrences"]) != len(ids):
        raise ValueError(
            "Source occurrence count mismatch for "
            + group["candidate_group_id"]
        )

    # --------------------------------------------------
    # 6. Select Reviewed or Draft Metadata
    # --------------------------------------------------

    review_stage = group["source_review_stage"]

    if review_stage == "metadata_reviewed_pending_signoff":

        metadata_status = "reviewed_pending_signoff"

        skill = group["reviewed_skill_area"]
        prescription = group["reviewed_prescription"]
        equipment = group["reviewed_equipment"]
        participants = group["reviewed_participants"]
        space = group["reviewed_space"]

    elif review_stage == "not_reviewed":

        if group["draft_review_status"] != "pending_source_review":
            raise ValueError(
                "Draft metadata not prepared for "
                + group["candidate_group_id"]
            )

        metadata_status = "draft_pending_verification"

        skill = group["draft_skill_areas"]
        prescription = group["draft_source_prescriptions"]
        equipment = group["draft_equipment_requirements"]
        participants = group["draft_participant_requirements"]
        space = group["draft_space_requirements"]

    else:
        raise ValueError(
            f"Unexpected review stage: {review_stage}"
        )

    # --------------------------------------------------
    # 7. Verify Approval Restrictions
    # --------------------------------------------------

    if (
        group["catalog_approval_status"] != "pending"
        or group["planner_eligible"] != "no"
    ):
        raise ValueError(
            "Unexpected approval status for "
            + group["candidate_group_id"]
        )

    # --------------------------------------------------
    # 8. Build Review Flags
    # --------------------------------------------------

    review_flags = []

    if len(ids) > 1:
        review_flags.append(
            "duplicate_variation_review"
        )

    if group["scope_candidate"] == "general_warmup_candidate":
        review_flags.append(
            "warmup_scope_review"
        )

    if normalized_name in {"quick feet", "side push"}:
        review_flags.append(
            "source_description_overlap"
        )

    if metadata_status == "draft_pending_verification":
        review_flags.append(
            "metadata_not_independently_verified"
        )

    # --------------------------------------------------
    # 9. Construct Catalog Record
    # --------------------------------------------------

    catalog.append({
        "catalog_candidate_id": group["candidate_group_id"],
        "approved_drill_id": "",
        "drill_name": group["drill_name"],
        "normalized_name": normalized_name,
        "skill_area": skill,
        "scope_candidate": group["scope_candidate"],
        "source_prescription": prescription,
        "equipment": equipment,
        "participants": participants,
        "space": space,
        "source_occurrences": len(ids),
        "evidence_ids": ";".join(ids),

        "curriculum_levels": ";".join(
            sorted({
                row["source_level"]
                for row in source_rows
            })
        ),

        "source_urls": ";".join(
            dict.fromkeys(
                row["source_url"]
                for row in source_rows
            )
        ),

        "heading_screen": group["heading_screen"],
        "section_screen": group["section_screen"],
        "metadata_status": metadata_status,
        "distinctness_status": group["distinctness_review"],
        "permission_status": group["permission_review"],
        "youth_planning_status": group["youth_planning_review"],
        "catalog_approval_status": "pending",
        "planner_eligible": "no",
        "review_flags": ";".join(review_flags),
        "catalog_version": "0.1",
    })

# --------------------------------------------------
# 10. Validate Catalog Integrity
# --------------------------------------------------

if (
    len(used_evidence_ids) != 70
    or len(set(used_evidence_ids)) != 70
    or set(used_evidence_ids) != set(evidence)
):
    raise ValueError(
        "Source evidence was duplicated or omitted."
    )

if len(catalog) != 67:
    raise ValueError(
        f"Expected 67 catalog candidates, found {len(catalog)}."
    )

catalog_ids = [
    row["catalog_candidate_id"]
    for row in catalog
]

if len(catalog_ids) != len(set(catalog_ids)):
    raise ValueError("Catalog candidate IDs are not unique.")

normalized_names = [
    row["normalized_name"]
    for row in catalog
]

if len(normalized_names) != len(set(normalized_names)):
    raise ValueError(
        "Repeated normalized names found in catalog."
    )

if any(
    row["planner_eligible"] != "no"
    or row["catalog_approval_status"] != "pending"
    or row["approved_drill_id"]
    for row in catalog
):
    raise ValueError(
        "Unapproved records must remain restricted."
    )

print("\n[3] CATALOG INTEGRITY")
print("Catalog candidate records:", len(catalog))
print("Unique candidate IDs: True")
print("Unique normalized names: True")
print("All 70 evidence records represented: True")

# --------------------------------------------------
# 11. Save Versioned Catalog Without Overwriting
# --------------------------------------------------

output_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

if output_path.exists():
    raise FileExistsError(
        "Catalog v0.1 already exists. "
        "Existing file preserved."
    )

with output_path.open(
    "x",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.DictWriter(
        f,
        fieldnames=list(catalog[0].keys())
    )

    writer.writeheader()
    writer.writerows(catalog)

# --------------------------------------------------
# 12. Reload and Verify Saved Output
# --------------------------------------------------

saved, saved_columns = load_csv(output_path)

if len(saved) != 67:
    raise ValueError(
        "Saved catalog row count mismatch."
    )

if len(saved_columns) != len(catalog[0]):
    raise ValueError(
        "Saved catalog column count mismatch."
    )

status_counts = Counter(
    row["metadata_status"]
    for row in saved
)

# --------------------------------------------------
# 13. Display Final Results
# --------------------------------------------------

print("\n[4] CATALOG CREATED")
print("File:", output_path.relative_to(root))
print("Catalog version: 0.1")
print("Catalog candidate groups:", len(saved))
print("Source occurrences represented:", len(evidence))

print("\n[5] METADATA STATUS")

for status, count in status_counts.items():
    print(f"{status}: {count}")

print("\n[6] REVIEW AND APPROVAL")

print(
    "Repeated-name groups:",
    sum(
        int(row["source_occurrences"]) > 1
        for row in saved
    )
)

print(
    "Catalog-approved:",
    sum(
        row["catalog_approval_status"] == "approved"
        for row in saved
    )
)

print(
    "Planner-eligible:",
    sum(
        row["planner_eligible"] == "yes"
        for row in saved
    )
)

print("\n[7] SOURCE INTEGRITY")
print("All 70 source occurrences represented: Yes")
print("Original evidence files modified: No")
print("Working approval checklist modified: No")
print("GitHub modified: No")

print("\n" + "=" * 65)
print("BASKETBALL CATALOG v0.1 CREATED SUCCESSFULLY")
print("=" * 65)
```

```python
# Step 12-26 - Basketball Source Reconciliation and Versioned Catalog
# Integrates source records while preserving previous catalog versions.

import csv
import difflib
import hashlib
import io
import json
import os
import re
import tempfile
from collections import Counter, defaultdict
from pathlib import Path

root = next(
    (p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (p / ".git").exists()),
    None,
)
if root is None:
    raise FileNotFoundError("Project Git directory was not found.")

processed = root / "data" / "processed"
results = root / "results"
archive = root / "docs" / "step12_review_v0_2"

paths = {
    "catalog_v01": processed / "basketball_drill_catalog_v0_1.csv",
    "catalog_v02": processed / "basketball_drill_catalog_v0_2_REVIEW_ONLY.csv",
    "archive_v02": archive / "basketball_drill_catalog_v0_2_REVIEW_ONLY.csv",
    "expanded_evidence": archive / "new_source_heading_evidence_v0_2.csv",
    "checklist": results / "basketball_drill_approval_checklist.csv",
    "aditi_drills": processed / "jr_nba_drills_master.csv",
    "aditi_rag": processed / "jr_nba_rag_documents.json",
    "manifest": root / "data" / "source_manifest.csv",
}
for i in range(1, 6):
    suffix = "" if i == 1 else f"_batch{i}"
    paths[f"old_evidence_{i}"] = (
        results / f"basketball_drill_source_evidence{suffix}.csv"
    )

for label, path in paths.items():
    if not path.is_file():
        raise FileNotFoundError(f"Required Step 12 input missing ({label}): {path}")


def read_csv(path):
    with path.open("r", newline="", encoding="utf-8-sig") as f:
        reader = csv.DictReader(f)
        columns = list(reader.fieldnames or [])
        if not columns or len(columns) != len(set(columns)):
            raise ValueError(f"Invalid or repeated CSV headers: {path}")
        rows = list(reader)
    if any(None in r or any(v is None for v in r.values()) for r in rows):
        raise ValueError(f"Malformed CSV records: {path}")
    return rows, columns


def sha256(path):
    h = hashlib.sha256()
    with path.open("rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()


def normalize(text):
    return re.sub(r"[^a-z0-9]+", " ", str(text).casefold()).strip()


def parts(value):
    return [v.strip() for v in (value or "").split(";") if v.strip()]


def unique(values):
    return list(dict.fromkeys(v for v in values if v))


def joined(values):
    return ";".join(unique(values))


manifest, _ = read_csv(paths["manifest"])
manifest_by_id = {r["source_id"]: r for r in manifest}
if len(manifest_by_id) != len(manifest):
    raise ValueError("Repeated Step 8 source IDs.")
for source_id, key in [
    ("catalog_basketball_v01", "catalog_v01"),
    ("catalog_basketball_v02", "catalog_v02"),
    ("aditi_jr_nba_drills", "aditi_drills"),
    ("aditi_jr_nba_rag", "aditi_rag"),
]:
    entry = manifest_by_id.get(source_id)
    if not entry or entry.get("sha256", "").lower() != sha256(paths[key]):
        raise ValueError(f"SHA-256 verification failed for {source_id}.")

if sha256(paths["catalog_v02"]) != sha256(paths["archive_v02"]):
    raise ValueError("Catalog v0.2 differs from its archived copy.")

v01, _ = read_csv(paths["catalog_v01"])
v02, catalog_columns = read_csv(paths["catalog_v02"])
checklist, _ = read_csv(paths["checklist"])
expanded, _ = read_csv(paths["expanded_evidence"])
aditi, aditi_columns = read_csv(paths["aditi_drills"])
with paths["aditi_rag"].open("r", encoding="utf-8-sig") as f:
    rag = json.load(f)

required_aditi = {
    "drill_id", "drill_name", "skill_category", "curriculum_level",
    "duration", "equipment", "group_requirement", "instructions_summary",
    "source_document", "source_page", "solo_adaptation_possible",
    "solo_adaptation_note",
}
if not required_aditi.issubset(aditi_columns):
    raise ValueError(f"Aditi drill fields missing: {sorted(required_aditi - set(aditi_columns))}")
if not all(isinstance(r, dict) for r in rag):
    raise ValueError("RAG document list must contain objects.")

if not (len(v01) == 67 and len(v02) == 129 and len(checklist) == 67
        and len(expanded) == 63 and len(aditi) == len(rag) == 23):
    raise ValueError("Unexpected input version/count; review sources before integration.")

aditi_ids = [r["drill_id"] for r in aditi]
rag_ids = [r.get("drill_id", "") for r in rag]
if (len(set(aditi_ids)) != 23 or len(set(rag_ids)) != 23
        or set(aditi_ids) != set(rag_ids)
        or any(not str(r.get("text", "")).strip() for r in rag)):
    raise ValueError("Jr. NBA drill and RAG ID/text linkage failed.")

old_evidence = []
for i in range(1, 6):
    batch, _ = read_csv(paths[f"old_evidence_{i}"])
    old_evidence.extend(batch)
if len(old_evidence) != 70:
    raise ValueError("Expected 70 original evidence occurrences.")

old_ids = [r["evidence_id"] for r in old_evidence]
new_ids = [r["new_evidence_id"] for r in expanded]
if (len(set(old_ids)) != 70 or len(set(new_ids)) != 63
        or set(old_ids) & set(new_ids)):
    raise ValueError("Duplicate/missing historical evidence IDs.")

v02_ids = [id_ for row in v02 for id_ in parts(row["evidence_ids"])]
if (len(v02_ids) != 133 or len(set(v02_ids)) != 133
        or set(v02_ids) != set(old_ids) | set(new_ids)):
    raise ValueError("Catalog v0.2 evidence mapping is inconsistent.")

base_ids = [r["catalog_candidate_id"] for r in v02]
base_names = [normalize(r["drill_name"]) for r in v02]
if len(set(base_ids)) != 129 or len(set(base_names)) != 129:
    raise ValueError("Catalog v0.2 candidate IDs/names are not unique.")

if (Counter(r["source_record_class"] for r in v02)
        != Counter({"prior_candidate": 67, "expanded_candidate": 62})):
    raise ValueError("Unexpected v0.2 provenance groups.")
if (Counter(r["scope_candidate"] for r in v02)
        != Counter({"basketball_drill_candidate": 117,
                    "general_warmup_candidate": 12})):
    raise ValueError("Unexpected v0.2 scope counts.")

checklist_by_id = {r["candidate_group_id"]: r for r in checklist}
if (len(checklist_by_id) != 67
        or set(checklist_by_id) != set(r["catalog_candidate_id"] for r in v01)):
    raise ValueError("Original 67-group approval checklist does not match catalog v0.1.")

if any(r["catalog_approval_status"] != "pending" or r["planner_eligible"] != "no"
       for r in v02):
    raise ValueError("Existing approval decisions detected: manual reconciliation required.")

# Candidate links are provisional. Similar names are not canonicalized automatically.
base_by_name = {normalize(r["drill_name"]): r for r in v02}
imported_by_name = defaultdict(list)
for record in aditi:
    key = normalize(record["drill_name"])
    if not key:
        raise ValueError(f"Empty imported drill name: {record['drill_id']}")
    imported_by_name[key].append(record)

exact_names = set(imported_by_name) & set(base_by_name)
new_names = sorted(set(imported_by_name) - set(base_by_name))
if len(imported_by_name) != 22 or len(exact_names) != 7 or len(new_names) != 15:
    raise ValueError("Imported names differ from the previously reviewed baseline.")

candidate_by_name = dict(base_by_name)
new_group_id = {}
for index, key in enumerate(new_names, start=1):
    new_group_id[key] = f"JR-ADITI-{index:04d}"

for key in imported_by_name:
    if key not in candidate_by_name:
        candidate_by_name[key] = {"catalog_candidate_id": new_group_id[key]}

# Retain page references as source locators; constructed URLs require review.
def source_locator(record):
    match = re.search(
        r"(Rookie|Starter|All-Star|MVP)\s+Practice\s+Plan\s+(\d+)",
        record["source_document"], re.IGNORECASE
    )
    if not match:
        return "", "source_document_unparsed"
    slug = match.group(1).lower()
    plan = int(match.group(2))
    if plan < 1 or plan > 12:
        return "", "practice_plan_out_of_range"
    url = f"https://jr.nba.com/basketball-practice-plans/{slug}/"
    if plan != 1:
        url += f"page/{plan}/"
    return url, "inferred_from_document_not_reverified"


def imported_evidence_id(record):
    return "ADI-" + record["drill_id"]


# Near matches are suggestions only, never automatic merges.
near_matches = []
for key in new_names:
    candidates = []
    for base_key, prior in base_by_name.items():
        similarity = difflib.SequenceMatcher(None, key, base_key).ratio()
        a, b = set(key.split()), set(base_key.split())
        jaccard = len(a & b) / len(a | b) if a | b else 0
        if similarity >= 0.82 and jaccard >= 0.50:
            candidates.append((similarity, jaccard, prior))
    for similarity, jaccard, prior in sorted(
        candidates, key=lambda t: (-t[0], -t[1], t[2]["drill_name"])
    )[:3]:
        near_matches.append({
            "imported_group_id": new_group_id[key],
            "imported_name": imported_by_name[key][0]["drill_name"],
            "existing_candidate_id": prior["catalog_candidate_id"],
            "existing_name": prior["drill_name"],
            "name_similarity": round(similarity, 4),
            "token_jaccard": round(jaccard, 4),
            "decision": "pending_manual_distinctness_review",
        })
near_ids = {r["imported_group_id"] for r in near_matches}

# Construct v0.3 without changing any previously stored catalog/checklist.
extra_fields = [
    "aditi_evidence_ids", "aditi_source_records", "total_evidence_records",
    "integration_status", "possible_near_name_match", "imported_metadata_status",
    "source_locator_status",
]
output_columns = catalog_columns + [f for f in extra_fields if f not in catalog_columns]
new_catalog = []
for existing in v02:
    key = normalize(existing["drill_name"])
    record = dict(existing)
    incoming = imported_by_name.get(key, [])
    incoming_ids = [imported_evidence_id(r) for r in incoming]
    record.update({
        "catalog_version": "0.3",
        "aditi_evidence_ids": joined(incoming_ids),
        "aditi_source_records": str(len(incoming)),
        "total_evidence_records": str(len(parts(existing["evidence_ids"])) + len(incoming)),
        "integration_status": (
            "exact_normalized_name_match_pending_source_comparison" if incoming
            else "preserved_from_v0_2"
        ),
        "possible_near_name_match": "no",
        "imported_metadata_status": (
            "source_details_require_comparison" if incoming else "not_applicable"
        ),
        "source_locator_status": (
            "candidate_source_locations_unverified" if incoming else "not_applicable"
        ),
    })
    new_catalog.append(record)

skill_map = {
    "ball handling": "ball_handling",
    "shooting": "shooting",
    "passing": "passing",
    "footwork conditioning": "footwork",
    "defense": "defense",
    "offense": "offense",
    "team concepts": "team_concepts",
    "warm up": "warmup",
}
for key in new_names:
    imports = imported_by_name[key]
    group_id = new_group_id[key]
    first = imports[0]
    original_skills = unique(r["skill_category"] for r in imports)
    proposed_skills = unique(skill_map.get(normalize(s), normalize(s)) for s in original_skills)
    locators = [source_locator(r) for r in imports]
    row = {c: "" for c in output_columns}
    row.update({
        "catalog_candidate_id": group_id,
        "drill_name": first["drill_name"],
        "normalized_name": key,
        "skill_area": joined(proposed_skills),
        "scope_candidate": (
            "general_warmup_candidate" if all(normalize(s) == "warm up" for s in original_skills)
            else "basketball_drill_candidate"
        ),
        "source_prescription": joined(r["duration"] for r in imports),
        "equipment": joined(r["equipment"] for r in imports),
        "participants": joined(r["group_requirement"] for r in imports),
        "space": "",  # No independently verified space requirements in imported data.
        "source_occurrences": str(len(imports)),
        "evidence_ids": joined(imported_evidence_id(r) for r in imports),
        "curriculum_levels": joined(r["curriculum_level"] for r in imports),
        "source_urls": joined(url for url, _ in locators),
        "heading_screen": "not_checked",
        "section_screen": "not_checked",
        "metadata_status": "imported_metadata_requires_source_verification",
        "distinctness_status": "pending_source_and_variation_comparison",
        "permission_status": "pending",
        "youth_planning_status": "pending",
        "catalog_approval_status": "pending",
        "planner_eligible": "no",
        "review_flags": "source_details_pending;permissions_pending;distinctness_pending",
        "catalog_version": "0.3",
        "source_record_class": "aditi_import_candidate",
        "heading_verification": "not_checked_against_original_source",
        "source_plan": joined(r["source_document"] for r in imports),
        "provenance_review": "source_locator_pending_independent_verification",
        "requirements_review": "pending",
        "reproduction_permission": "pending",
        "release_status": "research_candidate_not_for_planner",
        "aditi_evidence_ids": joined(imported_evidence_id(r) for r in imports),
        "aditi_source_records": str(len(imports)),
        "total_evidence_records": str(len(imports)),
        "integration_status": "new_normalized_name_candidate_pending_verification",
        "possible_near_name_match": "yes" if group_id in near_ids else "no",
        "imported_metadata_status": "source_details_require_independent_verification",
        "source_locator_status": "inferred_from_source_document_not_reverified",
    })
    new_catalog.append(row)

if len(new_catalog) != 144 or len({r["catalog_candidate_id"] for r in new_catalog}) != 144:
    raise ValueError("Integrated catalog group count or ID uniqueness failed.")
if len({normalize(r["drill_name"]) for r in new_catalog}) != 144:
    raise ValueError("Imported normalized-name groups are not unique.")
if any(r["catalog_approval_status"] != "pending" or r["planner_eligible"] != "no"
       for r in new_catalog):
    raise ValueError("Unapproved drill became planner eligible.")
if new_catalog[:129] and any(
    any(new_catalog[i].get(c, "") != v02[i].get(c, "") for c in catalog_columns if c != "catalog_version")
    for i in range(129)
):
    raise ValueError("Historical v0.2 record unexpectedly modified.")

catalog_for_name = {normalize(r["drill_name"]): r for r in new_catalog}

# One traceable row per source-evidence record, not per distinct real-world activity.
registry_columns = [
    "evidence_id", "origin", "drill_name", "source_level", "practice_plan",
    "source_document", "source_page", "source_url", "source_locator_status",
    "source_prescription", "equipment", "participants", "space",
    "catalog_candidate_id", "integration_status", "permissions_status",
]
registry = []
for row in old_evidence:
    key = normalize(row["official_drill_name"])
    candidate = catalog_for_name.get(key)
    if not candidate:
        raise ValueError(f"Unmatched historical name: {row['official_drill_name']}")
    registry.append({
        "evidence_id": row["evidence_id"], "origin": "original_review_batches",
        "drill_name": row["official_drill_name"],
        "source_level": row["source_level"], "practice_plan": row["practice_plan"],
        "source_document": "", "source_page": "", "source_url": row["source_url"],
        "source_locator_status": "previously_audited_requires_final_signoff",
        "source_prescription": row["source_timing"],
        "equipment": row["source_equipment"],
        "participants": row["participant_requirement"], "space": row["source_space"],
        "catalog_candidate_id": candidate["catalog_candidate_id"],
        "integration_status": "linked_from_v0_2",
        "permissions_status": row["permission_status"],
    })
for row in expanded:
    key = normalize(row["drill_name"])
    candidate = catalog_for_name.get(key)
    if not candidate:
        raise ValueError(f"Unmatched expanded name: {row['drill_name']}")
    registry.append({
        "evidence_id": row["new_evidence_id"], "origin": "v0_2_heading_expansion",
        "drill_name": row["drill_name"], "source_level": row["source_level"],
        "practice_plan": row["practice_plan"], "source_document": "",
        "source_page": "", "source_url": row["source_url"],
        "source_locator_status": "source_title_reviewed_requirements_pending",
        "source_prescription": row["source_prescription"],
        "equipment": "", "participants": "", "space": "",
        "catalog_candidate_id": candidate["catalog_candidate_id"],
        "integration_status": "linked_from_v0_2",
        "permissions_status": row["reuse_permissions"],
    })

reconciliation = []
for row in aditi:
    key = normalize(row["drill_name"])
    candidate = catalog_for_name[key]
    url, locator_status = source_locator(row)
    exact_match = key in exact_names
    reconc_status = (
        "exact_name_match_not_yet_confirmed_same_activity" if exact_match
        else "new_name_candidate_not_yet_confirmed_distinct"
    )
    evid = imported_evidence_id(row)
    registry.append({
        "evidence_id": evid, "origin": "aditi_starter_drill_dataset",
        "drill_name": row["drill_name"], "source_level": row["curriculum_level"],
        "practice_plan": "", "source_document": row["source_document"],
        "source_page": row["source_page"], "source_url": url,
        "source_locator_status": locator_status,
        "source_prescription": row["duration"],
        "equipment": row["equipment"], "participants": row["group_requirement"],
        "space": "", "catalog_candidate_id": candidate["catalog_candidate_id"],
        "integration_status": reconc_status, "permissions_status": "pending_review",
    })
    reconciliation.append({
        "aditi_drill_id": row["drill_id"], "aditi_evidence_id": evid,
        "imported_drill_name": row["drill_name"], "normalized_name": key,
        "match_status": reconc_status,
        "candidate_group_id": candidate["catalog_candidate_id"],
        "prior_catalog_match": "yes" if exact_match else "no",
        "aditi_source_document": row["source_document"],
        "aditi_source_page": row["source_page"],
        "source_url_candidate": url, "url_verification": locator_status,
        "aditi_skill_category": row["skill_category"],
        "aditi_duration": row["duration"],
        "aditi_equipment": row["equipment"],
        "aditi_group_requirement": row["group_requirement"],
        "aditi_solo_adaptation": row["solo_adaptation_note"],
        "solo_adaptation_status": "team_proposal_requires_independent_review",
        "distinctness_review": "pending",
        "content_review": "pending",
        "permissions_review": "pending",
        "planner_eligible": "no",
    })

registry_ids = [r["evidence_id"] for r in registry]
if len(registry) != 156 or len(set(registry_ids)) != 156:
    raise ValueError("Provenance registry must contain 156 unique evidence records.")
assert Counter(r["origin"] for r in registry) == {
    "original_review_batches": 70,
    "v0_2_heading_expansion": 63,
    "aditi_starter_drill_dataset": 23,
}

lookup_checklist = {r["candidate_group_id"]: r for r in checklist}
queue_columns = [
    "candidate_group_id", "drill_name", "scope_candidate", "skill_area",
    "source_occurrences_previous_catalog", "new_imported_evidence_records",
    "total_evidence_records", "integration_status", "metadata_status",
    "heading_review", "instructions_review", "timing_review", "equipment_review",
    "participants_review", "distinctness_review", "permission_review",
    "youth_planning_review", "catalog_approval_status", "planner_eligible",
    "reviewer", "review_notes", "next_action",
]
worklist = []
for row in new_catalog:
    prior = lookup_checklist.get(row["catalog_candidate_id"], {})
    requires_reconciliation = row["integration_status"] != "preserved_from_v0_2"
    worklist.append({
        "candidate_group_id": row["catalog_candidate_id"],
        "drill_name": row["drill_name"],
        "scope_candidate": row["scope_candidate"],
        "skill_area": row["skill_area"],
        "source_occurrences_previous_catalog": row["source_occurrences"],
        "new_imported_evidence_records": row["aditi_source_records"],
        "total_evidence_records": row["total_evidence_records"],
        "integration_status": row["integration_status"],
        "metadata_status": row["metadata_status"],
        "heading_review": row["heading_screen"],
        "instructions_review": prior.get("instructions_review", "pending"),
        "timing_review": prior.get("timing_review", "pending"),
        "equipment_review": prior.get("equipment_review", "pending"),
        "participants_review": prior.get("participants_review", "pending"),
        "distinctness_review": row["distinctness_status"],
        "permission_review": prior.get("permission_review", "pending"),
        "youth_planning_review": prior.get("youth_planning_review", "pending"),
        "catalog_approval_status": row["catalog_approval_status"],
        "planner_eligible": row["planner_eligible"],
        "reviewer": prior.get("reviewer", ""),
        "review_notes": prior.get("review_notes", ""),
        "next_action": (
            "Compare source instructions, variations and practical requirements"
            if requires_reconciliation else
            "Complete source, distinctness, permission and youth planning reviews"
        ),
    })

coverage = []
for scope in ("basketball_drill_candidate", "general_warmup_candidate"):
    selections = [r for r in new_catalog if r["scope_candidate"] == scope]
    coverage.append({
        "coverage_dimension": "scope", "category": scope,
        "candidate_groups": len(selections),
        "approved_drills": sum(r["catalog_approval_status"] == "approved" for r in selections),
        "planner_eligible": sum(r["planner_eligible"] == "yes" for r in selections),
        "interpretation": "Provisional groups; distinctness and eligibility pending",
    })
for skill in sorted({r["skill_area"] or "[unspecified]" for r in new_catalog}):
    selections = [r for r in new_catalog if (r["skill_area"] or "[unspecified]") == skill]
    coverage.append({
        "coverage_dimension": "skill_metadata", "category": skill,
        "candidate_groups": len(selections),
        "approved_drills": 0, "planner_eligible": 0,
        "interpretation": "Raw or proposed skill label; review before treating as canonical",
    })
for level in ("Rookie", "Starter", "All-Star", "MVP"):
    selections = [r for r in new_catalog if level in parts(r["curriculum_levels"])]
    coverage.append({
        "coverage_dimension": "curriculum_level", "category": level,
        "candidate_groups": len(selections),
        "approved_drills": 0, "planner_eligible": 0,
        "interpretation": "Multi-level groups may appear in several categories",
    })

summary = {
    "catalog_version": "0.3_REVIEW_ONLY",
    "baseline_catalog_groups": len(v02),
    "baseline_basketball_candidates": 117,
    "baseline_warmup_candidates": 12,
    "existing_evidence_records": 70,
    "expanded_v02_evidence_records": 63,
    "imported_aditi_drill_records": len(aditi),
    "imported_aditi_unique_names": len(imported_by_name),
    "imported_exact_name_match_groups": len(exact_names),
    "new_normalized_name_candidate_groups": len(new_names),
    "provisional_v03_candidate_groups": len(new_catalog),
    "provisional_basketball_candidate_groups": sum(
        r["scope_candidate"] == "basketball_drill_candidate" for r in new_catalog
    ),
    "provisional_general_warmup_groups": sum(
        r["scope_candidate"] == "general_warmup_candidate" for r in new_catalog
    ),
    "combined_evidence_records_with_possible_duplicates": len(registry),
    "suggested_near_match_pairs": len(near_matches),
    "catalog_approved_drills": 0,
    "planner_eligible_drills": 0,
    "source_permissions_review": "pending",
    "activity_distinctness_review": "pending",
    "youth_planning_rule_review": "pending_step_13",
    "inputs_sha256": {
        key: sha256(paths[key]) for key in (
            "catalog_v01", "catalog_v02", "archive_v02", "expanded_evidence",
            "checklist", "aditi_drills", "aditi_rag",
        )
    },
}


# Source index separates previously screened URLs from URLs inferred from
# the imported source-document labels. Neither implies usage permission.
source_by_url = defaultdict(list)
for row in registry:
    if row["source_url"]:
        source_by_url[row["source_url"]].append(row)
source_index_columns = [
    "source_url", "curriculum_levels", "evidence_record_count",
    "evidence_origins", "source_location_status", "permission_review",
]
source_index = []
for url, linked in sorted(source_by_url.items()):
    inferred = any(
        r["source_locator_status"] == "inferred_from_document_not_reverified"
        for r in linked
    )
    source_index.append({
        "source_url": url,
        "curriculum_levels": joined(sorted({r["source_level"] for r in linked})),
        "evidence_record_count": len(linked),
        "evidence_origins": joined(sorted({r["origin"] for r in linked})),
        "source_location_status": (
            "contains_unverified_imported_document_locators" if inferred
            else "previously_screened_not_final_content_approval"
        ),
        "permission_review": "pending",
    })

# Validate all intended outputs before writing any of them.
output_specs = [
    (processed / "basketball_drill_catalog_v0_3_REVIEW_ONLY.csv", new_catalog, output_columns, set()),
    (results / "basketball_drill_reconciliation_v0_3.csv", reconciliation,
     list(reconciliation[0]), set()),
    (results / "basketball_drill_source_registry_v0_3.csv", registry,
     registry_columns, set()),
    (results / "basketball_drill_near_matches_v0_3.csv", near_matches,
     ["imported_group_id", "imported_name", "existing_candidate_id",
      "existing_name", "name_similarity", "token_jaccard", "decision"], set()),
    (results / "basketball_drill_approval_queue_v0_3.csv", worklist,
     queue_columns, {"reviewer", "review_notes"}),
    (results / "eda" / "basketball_drill_coverage_v0_3.csv", coverage,
     list(coverage[0]), set()),
    (root / "data" / "references" / "basketball" / "source_index_v0_3.csv",
     source_index, source_index_columns, set()),
]


def encode_csv(rows, columns):
    buffer = io.StringIO(newline="")
    writer = csv.DictWriter(buffer, fieldnames=columns)
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode("utf-8")


pending_writes = []
for path, rows, columns, editable in output_specs:
    if path.exists():
        actual_rows, actual_columns = read_csv(path)
        if actual_columns != columns or len(actual_rows) != len(rows):
            raise ValueError(f"Existing output does not match expected schema/count: {path}")
        for i, (current, generated) in enumerate(zip(actual_rows, rows), start=1):
            for column in columns:
                if column not in editable and current.get(column, "") != str(generated.get(column, "")):
                    raise ValueError(
                        f"Existing report differs: {path.name}, row {i}, column {column}. "
                        "No files were modified; investigate manually."
                    )
    else:
        pending_writes.append((path, encode_csv(rows, columns)))

summary_path = results / "basketball_drill_integration_summary_v0_3.json"
summary_content = (json.dumps(summary, indent=2, ensure_ascii=False) + "\n").encode("utf-8")
if summary_path.is_file() and json.loads(summary_path.read_text(encoding="utf-8")) != summary:
    raise ValueError("Existing v0.3 summary differs; no files were modified.")
if not summary_path.exists():
    pending_writes.append((summary_path, summary_content))

for path, content in pending_writes:
    path.parent.mkdir(parents=True, exist_ok=True)
    tmp_path = None
    try:
        with tempfile.NamedTemporaryFile(dir=path.parent, delete=False) as f:
            tmp_path = Path(f.name)
            f.write(content)
        if path.exists():
            raise FileExistsError(f"Refusing to replace existing output: {path}")
        os.replace(tmp_path, path)
    finally:
        if tmp_path is not None and tmp_path.exists():
            tmp_path.unlink()

for path, rows, _, _ in output_specs:
    verified_rows, _ = read_csv(path)
    if len(verified_rows) != len(rows):
        raise RuntimeError(f"Failed output count verification: {path}")
if json.loads(summary_path.read_text(encoding="utf-8")) != summary:
    raise RuntimeError("Summary verification failed.")

print("=" * 72)
print("STEP 12 — BASKETBALL DRILL CATALOG INTEGRATION")
print("=" * 72)
print("Original candidate groups:", len(v02))
print("Imported drill records / unique names:", len(aditi), "/", len(imported_by_name))
print("Existing normalized-name match groups:", len(exact_names))
print("New provisional name groups:", len(new_names))
print("Integrated v0.3 candidate groups:", len(new_catalog))
print("Basketball / warm-up candidate groups:",
      summary["provisional_basketball_candidate_groups"], "/",
      summary["provisional_general_warmup_groups"])
print("Traceable evidence records (not unique activities):", len(registry))
print("Near-name match suggestions (not automatic merges):", len(near_matches))
print("Approved / planner-eligible:", 0, "/", 0)
print("Drill-to-RAG identifiers:", "PASS")
print("Existing source and review files overwritten: No")
print("Files newly created:", len(pending_writes))
print("Outputs verified:", len(output_specs) + 1)
print("Git operations performed: None")
print("OVERALL: PASS — integration documented; final approvals pending")
```

```python
# Step 12-27 - Basketball Drill Overlap and Variation Review
# Compare imported records with previous evidence; preserve all approval gates.

import csv
import hashlib
import io
import json
import re
from collections import Counter, defaultdict
from difflib import SequenceMatcher
from pathlib import Path

root = next(
    (path for path in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (path / ".git").exists()),
    None,
)
if root is None:
    raise FileNotFoundError("Project Git root not found.")

processed = root / "data" / "processed"
results = root / "results"

paths = {
    "catalog": processed / "basketball_drill_catalog_v0_3_REVIEW_ONLY.csv",
    "imports": processed / "jr_nba_drills_master.csv",
    "registry": results / "basketball_drill_source_registry_v0_3.csv",
    "reconciliation": results / "basketball_drill_reconciliation_v0_3.csv",
    "near_matches": results / "basketball_drill_near_matches_v0_3.csv",
    "integration_summary": results / "basketball_drill_integration_summary_v0_3.json",
    "manifest_catalog": processed / "basketball_drill_catalog_v0_2_REVIEW_ONLY.csv",
}
for label, path in paths.items():
    if not path.is_file():
        raise FileNotFoundError(f"Required Step 12 input missing ({label}): {path}")


def read_csv(path):
    with path.open("r", newline="", encoding="utf-8-sig") as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        if not columns or len(columns) != len(set(columns)):
            raise ValueError(f"Missing or duplicate columns: {path}")
        rows = list(reader)
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f"Malformed CSV row in {path}")
    return rows, columns


def file_sha256(path):
    digest = hashlib.sha256()
    with path.open("rb") as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()


def normalized(value):
    return re.sub(r"[^a-z0-9]+", " ", (value or "").casefold()).strip()


def split_ids(value):
    return [item.strip() for item in (value or "").split(";") if item.strip()]


def compare_text(first, second):
    unknowns = {"", "not specified", "not stated", "unknown", "see source practice plan"}
    a, b = normalized(first), normalized(second)
    if a in unknowns and b in unknowns:
        return "both_unavailable"
    if a in unknowns or b in unknowns:
        return "one_unavailable"
    return "same_wording" if a == b else "different_wording_review"


def source_plan(document):
    match = re.search(
        r"(Rookie|Starter|All-Star|MVP)\s+Practice\s+Plan\s+(\d+)",
        document or "", flags=re.IGNORECASE,
    )
    if not match:
        return "", ""
    return normalized(match.group(1)), str(int(match.group(2)))


catalog, _ = read_csv(paths["catalog"])
imports, _ = read_csv(paths["imports"])
registry, _ = read_csv(paths["registry"])
recon, _ = read_csv(paths["reconciliation"])
near, _ = read_csv(paths["near_matches"])
summary = json.loads(paths["integration_summary"].read_text(encoding="utf-8"))

if (len(catalog), len(imports), len(registry), len(recon), len(near)) != (144, 23, 156, 23, 1):
    raise ValueError("The v0.3 integration inputs do not match the verified baseline.")

for key, path_key in (("catalog_v02", "manifest_catalog"), ("aditi_drills", "imports")):
    expected_sha = summary.get("inputs_sha256", {}).get(key)
    if not expected_sha or file_sha256(paths[path_key]) != expected_sha:
        raise ValueError(f"Source checksum differs from Step 12-26: {key}")

if (summary.get("provisional_v03_candidate_groups") != 144
        or summary.get("catalog_approved_drills") != 0):
    raise ValueError("The saved Step 12-26 integration summary has changed.")

catalog_by_id = {row["catalog_candidate_id"]: row for row in catalog}
import_by_id = {row["drill_id"]: row for row in imports}
recon_by_id = {row["aditi_drill_id"]: row for row in recon}
registry_by_id = {row["evidence_id"]: row for row in registry}
if (len(catalog_by_id) != 144 or len(import_by_id) != 23
        or len(recon_by_id) != 23 or len(registry_by_id) != 156):
    raise ValueError("Source identifiers are missing or repeated.")
if set(import_by_id) != set(recon_by_id):
    raise ValueError("Imported drill IDs do not match the v0.3 reconciliation.")
if any(row["planner_eligible"] != "no" or row["catalog_approval_status"] != "pending"
       for row in catalog):
    raise ValueError("An approved catalog status appeared; review manually before continuing.")

historical_by_group = defaultdict(list)
for row in registry:
    if row["origin"] != "aditi_starter_drill_dataset":
        historical_by_group[row["catalog_candidate_id"]].append(row)

# One review case per matching name/group pair, not per historical occurrence.
review_groups = defaultdict(list)
for record in recon:
    if record["prior_catalog_match"] == "yes":
        cid = record["candidate_group_id"]
        review_groups[("exact_normalized_name", cid, cid)].append(record)

for suggestion in near:
    imported_id = suggestion["imported_group_id"]
    existing_id = suggestion["existing_candidate_id"]
    linked = [row for row in recon if row["candidate_group_id"] == imported_id]
    if not linked or imported_id == existing_id:
        raise ValueError("Near-name suggestion does not link two distinct catalog groups.")
    review_groups[("near_name_suggestion", imported_id, existing_id)].extend(linked)

if (sum(key[0] == "exact_normalized_name" for key in review_groups) != 7
        or sum(key[0] == "near_name_suggestion" for key in review_groups) != 1):
    raise ValueError("Unexpected exact/near overlap group count.")

pair_rows = []
worklist = []
for case_number, ((match_type, imported_gid, existing_gid), records) in enumerate(
    sorted(review_groups.items(), key=lambda pair: (pair[0][0], pair[0][1], pair[0][2])),
    start=1,
):
    case_id = f"JR-MATCH-{case_number:04d}"
    existing_group = catalog_by_id.get(existing_gid)
    imported_group = catalog_by_id.get(imported_gid)
    historical = sorted(historical_by_group[existing_gid], key=lambda row: row["evidence_id"])
    if not existing_group or not imported_group or not historical:
        raise ValueError(f"Missing matched catalog group or historical evidence: {case_id}")
    comparison_rows = []

    for record in sorted(records, key=lambda row: row["aditi_drill_id"]):
        imp = import_by_id[record["aditi_drill_id"]]
        expected_evidence = "ADI-" + imp["drill_id"]
        if record["aditi_evidence_id"] != expected_evidence:
            raise ValueError(f"Imported evidence reference mismatch for {case_id}")
        imp_level, imp_plan = source_plan(imp["source_document"])
        for old in historical:
            old_level = normalized(old["source_level"])
            old_plan = old["practice_plan"].strip()
            same_level = (
                "same" if imp_level and old_level and imp_level == old_level
                else "different" if imp_level and old_level else "unknown"
            )
            same_plan = (
                "same" if imp_plan and old_plan and imp_plan == old_plan and same_level == "same"
                else "different" if imp_plan and old_plan and same_level != "unknown"
                else "unknown"
            )
            timing = compare_text(imp["duration"], old["source_prescription"])
            equipment = compare_text(imp["equipment"], old["equipment"])
            participants = compare_text(imp["group_requirement"], old["participants"])
            flags = []
            if same_plan == "different":
                flags.append("different_source_plans")
            if timing == "different_wording_review":
                flags.append("prescription_wording_differs")
            if equipment == "different_wording_review":
                flags.append("equipment_wording_differs")
            if participants == "different_wording_review":
                flags.append("participant_wording_differs")
            if old["origin"] == "v0_2_heading_expansion":
                flags.append("historical_record_title_only")

            pair = {
                "case_id": case_id,
                "match_type": match_type,
                "imported_evidence_id": expected_evidence,
                "imported_drill_id": imp["drill_id"],
                "imported_drill_name": imp["drill_name"],
                "historical_evidence_id": old["evidence_id"],
                "historical_name": old["drill_name"],
                "historical_origin": old["origin"],
                "imported_candidate_id": imported_gid,
                "historical_candidate_id": existing_gid,
                "imported_source_document": imp["source_document"],
                "imported_source_page": imp["source_page"],
                "historical_source_url": old["source_url"],
                "source_level_screen": same_level,
                "source_practice_plan_screen": same_plan,
                "imported_prescription": imp["duration"],
                "historical_prescription": old["source_prescription"],
                "prescription_wording_screen": timing,
                "imported_equipment": imp["equipment"],
                "historical_equipment": old["equipment"],
                "equipment_wording_screen": equipment,
                "imported_participants": imp["group_requirement"],
                "historical_participants": old["participants"],
                "participant_wording_screen": participants,
                "imported_summary_available": "yes" if imp["instructions_summary"].strip() else "no",
                "content_comparison": "pending_manual_source_review",
                "review_flags": ";".join(flags),
                "distinctness_decision": "pending",
                "approval_status": "pending",
                "planner_eligible": "no",
            }
            pair_rows.append(pair)
            comparison_rows.append(pair)

    imported_names = sorted({import_by_id[r["aditi_drill_id"]]["drill_name"] for r in records})
    exact_title_similarity = SequenceMatcher(
        None, normalized(imported_names[0]), normalized(existing_group["drill_name"])
    ).ratio()
    worklist.append({
        "case_id": case_id,
        "match_type": match_type,
        "imported_candidate_id": imported_gid,
        "historical_candidate_id": existing_gid,
        "imported_name": ";".join(imported_names),
        "historical_name": existing_group["drill_name"],
        "name_similarity": f"{exact_title_similarity:.4f}",
        "imported_drill_ids": ";".join(sorted({r["aditi_drill_id"] for r in records})),
        "imported_occurrences": str(len(records)),
        "historical_occurrences": str(len(historical)),
        "source_comparison_pairs": str(len(comparison_rows)),
        "same_practice_plan_pairs": str(sum(r["source_practice_plan_screen"] == "same" for r in comparison_rows)),
        "different_practice_plan_pairs": str(sum(r["source_practice_plan_screen"] == "different" for r in comparison_rows)),
        "prescription_wording_differences": str(sum(r["prescription_wording_screen"] == "different_wording_review" for r in comparison_rows)),
        "equipment_wording_differences": str(sum(r["equipment_wording_screen"] == "different_wording_review" for r in comparison_rows)),
        "participant_wording_differences": str(sum(r["participant_wording_screen"] == "different_wording_review" for r in comparison_rows)),
        "original_evidence_ids": ";".join(r["evidence_id"] for r in historical),
        "source_instruction_review": "pending",
        "distinctness_decision": "pending",
        "variation_decision": "pending",
        "permission_review": "pending",
        "catalog_approved": "no",
        "planner_eligible": "no",
        "reviewer": "",
        "review_date": "",
        "review_notes": "",
        "next_action": (
            "Compare the original activity sections; determine whether these names describe the same drill or different variations."
        ),
    })

assert len(worklist) == 8 and len(pair_rows) >= 8
if any(row["planner_eligible"] != "no" for row in worklist + pair_rows):
    raise ValueError("Unapproved source comparison unexpectedly marked planner-eligible.")

# Outputs are immutable except for the explicit human-review fields in the worklist.
outputs = [
    (results / "basketball_overlap_review_worklist_v0_3.csv", worklist,
     {"reviewer", "review_date", "review_notes", "source_instruction_review",
      "distinctness_decision", "variation_decision", "permission_review"}),
    (results / "basketball_overlap_source_comparisons_v0_3.csv", pair_rows, set()),
]


def csv_bytes(rows, columns):
    buffer = io.StringIO(newline="")
    writer = csv.DictWriter(buffer, fieldnames=columns)
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode("utf-8")


prepared = []
for path, rows, editable_columns in outputs:
    columns = list(rows[0])
    if path.exists():
        saved, saved_columns = read_csv(path)
        if saved_columns != columns or len(saved) != len(rows):
            raise ValueError(f"Existing review output has changed its schema or row count: {path}")
        for index, (actual, expected) in enumerate(zip(saved, rows), start=1):
            for column in columns:
                if column not in editable_columns and actual[column] != str(expected[column]):
                    raise ValueError(f"Existing {path.name} row {index} differs in {column}; no files modified.")
        prepared.append((path, None))
    else:
        prepared.append((path, csv_bytes(rows, columns)))

# All checks must finish before creating any file.
for path, content in prepared:
    if content is not None:
        path.parent.mkdir(parents=True, exist_ok=True)
        with path.open("xb") as handle:
            handle.write(content)

for path, rows, _ in outputs:
    saved, _ = read_csv(path)
    if len(saved) != len(rows):
        raise RuntimeError(f"Saved Step 12-27 report verification failed: {path}")

print("=" * 72)
print("STEP 12-27 — DRILL OVERLAP AND SOURCE COMPARISON")
print("=" * 72)
print("Existing v0.3 candidate groups:", len(catalog))
print("Exact-name review cases:", sum(r["match_type"] == "exact_normalized_name" for r in worklist))
print("Near-name review cases:", sum(r["match_type"] == "near_name_suggestion" for r in worklist))
print("Source-occurrence comparison pairs:", len(pair_rows))
print("Same practice-plan pairs:", sum(r["source_practice_plan_screen"] == "same" for r in pair_rows))
print("Different practice-plan pairs:", sum(r["source_practice_plan_screen"] == "different" for r in pair_rows))
print("Timing wording differences:", sum(r["prescription_wording_screen"] == "different_wording_review" for r in pair_rows))
print("Equipment wording differences:", sum(r["equipment_wording_screen"] == "different_wording_review" for r in pair_rows))
print("Participant wording differences:", sum(r["participant_wording_screen"] == "different_wording_review" for r in pair_rows))
print("\nREVIEW CASES")
for row in worklist:
    print(f"  {row['case_id']} | {row['match_type']} | {row['imported_name']} / {row['historical_name']}")
print("\nREPORTS")
for path, content in prepared:
    print("  ", "CREATED" if content is not None else "PRESERVED", path.relative_to(root))
print("\nAutomatic canonical merges: 0")
print("Automatic approvals / planner-eligible records: 0 / 0")
print("Existing catalog, source and approval files overwritten: No")
print("Git operations performed: None")
print("OVERALL: PASS — overlaps screened; original-source distinctness review remains pending")
print("=" * 72)
```

```python
# Step 12-28 - Original-Source Screening and Reconciliation Review
# Audit official source headings and prepare unresolved review decisions.

import csv
import hashlib
import io
import os
import json
import re
import unicodedata
from collections import Counter, defaultdict
from datetime import date
from difflib import SequenceMatcher
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

root = next(
    (p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (p / ".git").exists()),
    None,
)
if root is None:
    raise FileNotFoundError("Project Git root not found.")

results = root / "results"
processed = root / "data" / "processed"
inputs = {
    "worklist": results / "basketball_overlap_review_worklist_v0_3.csv",
    "comparisons": results / "basketball_overlap_source_comparisons_v0_3.csv",
    "catalog": processed / "basketball_drill_catalog_v0_3_REVIEW_ONLY.csv",
    "summary": results / "basketball_drill_integration_summary_v0_3.json",
}
outputs = {
    "source_checks": results / "basketball_overlap_source_heading_checks_v0_3.csv",
    "decisions": results / "basketball_overlap_reconciliation_decisions_v0_3.csv",
    "summary": results / "basketball_overlap_source_review_summary_v0_3.json",
}


def load_csv(path):
    with path.open(newline="", encoding="utf-8-sig") as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        if not columns or len(columns) != len(set(columns)):
            raise ValueError(f"Invalid CSV schema: {path}")
        rows = list(reader)
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f"Malformed CSV data: {path}")
    return rows, columns


def normalize(text):
    text = unicodedata.normalize("NFKC", text or "").casefold()
    return re.sub(r"[^a-z0-9]+", " ", text).strip()


def source_url_from_document(document):
    match = re.search(
        r"\b(rookie|starter|all[ -]star|mvp)\s+practice\s+plan\s+(\d+)\b",
        document or "", re.IGNORECASE,
    )
    if not match:
        return "", "", ""
    level = normalize(match.group(1)).replace(" ", "-")
    number = int(match.group(2))
    if number < 1 or number > 12:
        return "", "", ""
    base = f"https://jr.nba.com/basketball-practice-plans/{level}/"
    return (base if number == 1 else f"{base}page/{number}/"), level, str(number)


def valid_official_url(url):
    parsed = urlsplit(url)
    return (
        parsed.scheme == "https" and parsed.hostname == "jr.nba.com"
        and parsed.username is None and parsed.password is None
        and parsed.port in (None, 443)
        and bool(re.fullmatch(
            r"/basketball-practice-plans/(?:rookie|starter|all-star|mvp)/(?:page/\d+/)?",
            parsed.path,
        ))
        and not parsed.query and not parsed.fragment
    )


class HeadingSectionParser(HTMLParser):
    """Extract page headings and short-lived in-memory sections, without saving source text."""

    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.current = None
        self.active_heading = None
        self.heading_words = []
        self.skip_depth = 0

    def handle_starttag(self, tag, attrs):
        if tag in {"script", "style", "noscript"}:
            self.skip_depth += 1
            return
        if self.skip_depth:
            return
        if re.fullmatch(r"h[1-6]", tag):
            self.active_heading = tag
            self.heading_words = []
        elif tag in {"p", "li", "div", "br"} and self.current is not None:
            self.current["parts"].append(" ")

    def handle_data(self, data):
        if self.skip_depth:
            return
        if self.active_heading:
            self.heading_words.append(data)
        elif self.current is not None:
            self.current["parts"].append(data)

    def handle_endtag(self, tag):
        if tag in {"script", "style", "noscript"}:
            self.skip_depth = max(0, self.skip_depth - 1)
            return
        if self.skip_depth:
            return
        if tag == self.active_heading:
            heading = normalize(" ".join(self.heading_words))
            if heading:
                self.current = {"heading": heading, "parts": []}
                self.sections.append(self.current)
            self.active_heading = None
            self.heading_words = []


def fetch_source(url):
    if os.environ.get("STEP12_OFFLINE_REVIEW", "0") == "1":
        return {"status": "offline_screen_only", "http": "", "sections": [], "final_url": "", "error": "Live source check disabled"}
    if not valid_official_url(url):
        return {"status": "invalid_locator", "http": "", "sections": [], "final_url": "", "error": "Source URL not in allowed Jr. NBA practice-plan pattern"}
    request = Request(url, headers={
        "User-Agent": "Mozilla/5.0 (compatible; CapstoneSourceAudit/1.0)"
    })
    try:
        with urlopen(request, timeout=15) as response:
            final_url = response.geturl()
            html = response.read(2_000_000).decode("utf-8", errors="replace")
            http = str(response.status)
        if not valid_official_url(final_url):
            return {"status": "unexpected_redirect", "http": http, "sections": [], "final_url": final_url, "error": "Redirect outside Jr. NBA practice-plan URL pattern"}
        parser = HeadingSectionParser()
        parser.feed(html)
        return {"status": "fetched", "http": http, "sections": parser.sections, "final_url": final_url, "error": ""}
    except (HTTPError, URLError, TimeoutError, OSError, ValueError) as exc:
        return {"status": "unavailable", "http": str(getattr(exc, "code", "")), "sections": [], "final_url": "", "error": type(exc).__name__}


for label, path in inputs.items():
    if not path.is_file():
        raise FileNotFoundError(f"Missing Step 12 source ({label}): {path}")

cases, _ = load_csv(inputs["worklist"])
pairs, _ = load_csv(inputs["comparisons"])
catalog, _ = load_csv(inputs["catalog"])
integration = json.loads(inputs["summary"].read_text(encoding="utf-8"))

if (len(cases), len(pairs), len(catalog)) != (8, 11, 144):
    raise ValueError("Step 12-27 baseline counts have changed.")
if (len({r["case_id"] for r in cases}) != 8
        or set(r["case_id"] for r in pairs) != set(r["case_id"] for r in cases)):
    raise ValueError("Overlap case identifiers are inconsistent.")
if any(r["catalog_approved"] != "no" or r["planner_eligible"] != "no" for r in cases):
    raise ValueError("Unexpected approval status in overlap worklist.")
if any(r["approval_status"] != "pending" or r["planner_eligible"] != "no" for r in pairs):
    raise ValueError("Unexpected approval status in source comparisons.")
if any(r["catalog_approval_status"] != "pending" or r["planner_eligible"] != "no" for r in catalog):
    raise ValueError("Existing v0.3 catalog contains an approval; manual inspection required.")
if integration.get("provisional_v03_candidate_groups") != 144:
    raise ValueError("Step 12-26 integration summary does not match current catalog.")

# One source-heading check per unique case, source occurrence, role, and locator.
records = {}
for pair in pairs:
    imported_url, imported_level, imported_plan = source_url_from_document(pair["imported_source_document"])
    if not imported_url:
        imported_status = "source_document_unresolved"
    else:
        imported_status = "derived_from_source_document_not_source_verified"
    observations = (
        ("imported", pair["imported_evidence_id"], pair["imported_drill_name"], imported_url,
         imported_status, pair["imported_source_document"], pair["imported_source_page"],
         pair["imported_prescription"], imported_level, imported_plan),
        ("historical", pair["historical_evidence_id"], pair["historical_name"],
         pair["historical_source_url"], "recorded_in_historical_evidence", "", "",
         pair["historical_prescription"], "", ""),
    )
    for role, eid, name, url, locator_basis, document, document_page, prescription, level, plan in observations:
        key = (pair["case_id"], role, eid, url)
        if key in records:
            if records[key]["activity_name"] != name or records[key]["prescription_recorded"] != prescription:
                raise ValueError(f"Conflicting source occurrence in comparisons: {key}")
            continue
        records[key] = {
            "case_id": pair["case_id"],
            "record_role": role,
            "evidence_id": eid,
            "activity_name": name,
            "source_document": document,
            "source_document_page": document_page,
            "source_url": url,
            "locator_basis": locator_basis,
            "inferred_level": level,
            "inferred_plan": plan,
            "prescription_recorded": prescription,
        }

urls = sorted({r["source_url"] for r in records.values() if r["source_url"]})
cache = {}
print("=" * 72)
print("STEP 12-28 — OFFICIAL SOURCE SCREEN AND RECONCILIATION WORKLIST")
print("=" * 72)
print("Cases / source comparison pairs:", len(cases), "/", len(pairs))
print("Source occurrences to screen:", len(records))
print("Distinct candidate source URLs:", len(urls))
print("\nOFFICIAL PAGE ACCESS")
for number, url in enumerate(urls, start=1):
    response = fetch_source(url)
    cache[url] = response
    print(f"  {number:02d}/{len(urls):02d} | {response['status']:<19} | {url}")

source_checks = []
for key, row in sorted(records.items()):
    url = row["source_url"]
    response = cache.get(url, {"status": "unresolved_locator", "http": "", "sections": [], "final_url": "", "error": "Source document did not identify a practice-plan URL"})
    observed = response["sections"]
    target = normalize(row["activity_name"])
    exact = [section for section in observed if section["heading"] == target]
    similar = sorted(
        ((SequenceMatcher(None, target, s["heading"]).ratio(), s["heading"])
         for s in observed if target and s["heading"] != target), reverse=True,
    )
    if response["status"] != "fetched":
        title_screen = "not_checked"
    elif exact:
        title_screen = "exact_heading_present"
    elif similar and similar[0][0] >= 0.88:
        title_screen = "similar_heading_review"
    else:
        title_screen = "heading_not_found_in_html"

    selected = exact[0] if exact else None
    section_text = normalize(" ".join(selected["parts"])) if selected else ""
    source_hash = hashlib.sha256(section_text.encode("utf-8")).hexdigest() if section_text else ""
    recorded = normalize(row["prescription_recorded"])
    if not recorded or recorded in {"not specified", "not stated", "unknown"}:
        timing_screen = "recorded_prescription_unspecified"
    elif not section_text:
        timing_screen = "section_not_available_for_comparison"
    elif recorded in section_text:
        timing_screen = "recorded_phrase_present_in_section"
    else:
        timing_screen = "requires_manual_prescription_check"

    source_checks.append({
        **row,
        "fetch_status": response["status"],
        "http_status": response["http"],
        "resolved_url": response["final_url"],
        "activity_heading_screen": title_screen,
        "matching_exact_headings": str(len(exact)),
        "closest_heading_similarity": f"{similar[0][0]:.4f}" if similar else "",
        "source_section_word_count": str(len(section_text.split())) if section_text else "",
        "source_section_sha256": source_hash,
        "prescription_text_screen": timing_screen,
        "source_review_status": "pending_manual_verification",
        "permissions_review": "pending",
        "planner_eligible": "no",
        "fetch_error_category": response["error"],
    })

by_case = defaultdict(list)
for row in source_checks:
    by_case[row["case_id"]].append(row)

pair_cases = defaultdict(list)
for pair in pairs:
    pair_cases[pair["case_id"]].append(pair)

decisions = []
for case in cases:
    checks = by_case[case["case_id"]]
    these_pairs = pair_cases[case["case_id"]]
    access_issue = any(row["fetch_status"] != "fetched" for row in checks)
    heading_issue = any(row["activity_heading_screen"] != "exact_heading_present" for row in checks)
    different_plans = any(row["source_practice_plan_screen"] == "different" for row in these_pairs)
    prescription_issue = any(row["prescription_wording_screen"] != "same_wording" for row in these_pairs)
    unknown_requirements = any(
        row["equipment_wording_screen"] != "same_wording" or row["participant_wording_screen"] != "same_wording"
        for row in these_pairs
    )
    flags = []
    if access_issue:
        flags.append("source_access_or_locator_check")
    if heading_issue:
        flags.append("verify_activity_heading")
    if different_plans:
        flags.append("different_practice_plan_occurrences")
    if prescription_issue:
        flags.append("verify_prescriptions_and_variations")
    if unknown_requirements:
        flags.append("compare_equipment_and_participants")
    if case["match_type"] == "near_name_suggestion":
        flags.append("confirm_name_alias_or_distinct_drill")

    decisions.append({
        "case_id": case["case_id"],
        "match_type": case["match_type"],
        "imported_name": case["imported_name"],
        "historical_name": case["historical_name"],
        "imported_candidate_id": case["imported_candidate_id"],
        "historical_candidate_id": case["historical_candidate_id"],
        "source_occurrences_screened": str(len(checks)),
        "source_pages_fetched": str(sum(r["fetch_status"] == "fetched" for r in checks)),
        "exact_activity_headings_found": str(sum(r["activity_heading_screen"] == "exact_heading_present" for r in checks)),
        "comparison_pairs": str(len(these_pairs)),
        "review_flags": ";".join(flags),
        "instructions_decision": "pending_manual_review",
        "same_drill_decision": "pending_manual_review",
        "variation_decision": "pending_manual_review",
        "prescription_review": "pending_manual_review",
        "equipment_participants_review": "pending_manual_review",
        "permissions_review": "pending",
        "reviewer": "",
        "review_date": "",
        "review_evidence_notes": "",
        "catalog_approval_status": "pending",
        "planner_eligible": "no",
    })

if len(decisions) != 8 or len(source_checks) < 16:
    raise ValueError("Source screening worklist does not cover all overlap cases.")
if any(row["catalog_approval_status"] != "pending" or row["planner_eligible"] != "no" for row in decisions):
    raise ValueError("Pending review records cannot be automatically approved.")

summary = {
    "step": "12-28",
    "cases": len(decisions),
    "source_comparison_pairs": len(pairs),
    "source_occurrence_checks": len(source_checks),
    "unique_source_urls": len(urls),
    "official_page_fetch_counts": dict(sorted(Counter(response["status"] for response in cache.values()).items())),
    "activity_heading_screens": dict(sorted(Counter(row["activity_heading_screen"] for row in source_checks).items())),
    "prescription_text_screens": dict(sorted(Counter(row["prescription_text_screen"] for row in source_checks).items())),
    "manual_reconciliation_decisions_complete": 0,
    "auto_merges": 0,
    "auto_approvals": 0,
    "input_catalog_modified": False,
    "source_html_persisted": False,
    "source_review_limit": "HTML heading and phrase screens are not independent verification of instructional meaning, variation equivalence, or reuse permissions.",
}


def csv_bytes(rows):
    buffer = io.StringIO(newline="")
    columns = list(rows[0])
    writer = csv.DictWriter(buffer, fieldnames=columns)
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode("utf-8"), columns


# Preserve existing manual decisions on repeat execution.
editable = {
    "instructions_decision", "same_drill_decision", "variation_decision",
    "prescription_review", "equipment_participants_review", "permissions_review",
    "reviewer", "review_date", "review_evidence_notes",
}
files = {
    "source_checks": (outputs["source_checks"], source_checks),
    "decisions": (outputs["decisions"], decisions),
    "summary": (outputs["summary"], summary),
}

for path, _ in files.values():
    path.parent.mkdir(parents=True, exist_ok=True)

results_saved = []
for label, (path, content) in files.items():
    if label != "summary":
        expected_bytes, columns = csv_bytes(content)
        if path.exists():
            old, old_columns = load_csv(path)
            if old_columns != columns or len(old) != len(content):
                raise ValueError(f"Existing {path.name} schema/row count changed; no updates made.")
            for previous, current in zip(old, content):
                if previous["case_id"] != current["case_id"]:
                    raise ValueError(f"Existing case order changed in {path.name}.")
                for column in columns:
                    if label == "decisions" and column in editable:
                        continue
                    if label == "source_checks" and column in {
                        "fetch_status", "http_status", "resolved_url", "activity_heading_screen",
                        "matching_exact_headings", "closest_heading_similarity",
                        "source_section_word_count", "source_section_sha256", "prescription_text_screen",
                        "fetch_error_category",
                    }:
                        continue  # A source website can change between runs.
                    if previous[column] != str(current[column]):
                        raise ValueError(f"Existing {path.name} differs at {previous['case_id']} / {column}.")
            results_saved.append((path, "PRESERVED"))
        else:
            # Write only new reports; immutable source screens never contain full source text.
            with path.open("xb") as handle:
                handle.write(expected_bytes)
            results_saved.append((path, "CREATED"))
    elif path.exists():
        earlier = json.loads(path.read_text(encoding="utf-8"))
        for name in ("cases", "source_comparison_pairs", "source_occurrence_checks"):
            if earlier.get(name) != summary[name]:
                raise ValueError(f"Existing summary changed baseline field: {name}")
        results_saved.append((path, "PRESERVED"))
    else:
        with path.open("x", encoding="utf-8") as handle:
            json.dump(content, handle, indent=2)
            handle.write("\n")
        results_saved.append((path, "CREATED"))

# Verify outputs currently on disk.
stored_checks, _ = load_csv(outputs["source_checks"])
stored_decisions, _ = load_csv(outputs["decisions"])
if len(stored_checks) != len(source_checks) or len(stored_decisions) != 8:
    raise ValueError("Saved source review row counts did not match expected results.")
if any(row["catalog_approval_status"] != "pending" or row["planner_eligible"] != "no"
       for row in stored_decisions):
    raise ValueError("A manual review file is now marked approved. Inspect it before proceeding.")

print("\nSOURCE-HEADING SCREEN (LIVE RUN)")
for status, count in sorted(Counter(row["activity_heading_screen"] for row in source_checks).items()):
    print(f"  {status}: {count}")
print("\nCASE REVIEW FLAGS")
for row in decisions:
    print(f"  {row['case_id']} | {row['imported_name']} / {row['historical_name']} | {row['review_flags']}")
print("\nREPORTS")
for path, disposition in results_saved:
    print(f"  {disposition:<10} {path.relative_to(root)}")
print("\nSource pages fetched:", sum(item["status"] == "fetched" for item in cache.values()), "/", len(urls))
print("Cases awaiting human reconciliation:", len(stored_decisions))
print("Approved / planner-eligible:", 0, "/", 0)
print("Existing source files or catalogs overwritten: No")
print("Git operations performed: None")
print("OVERALL: PASS — evidence screening recorded; canonical decisions and permissions pending")
print("=" * 72)
```

```python
# Step 12-29 - Basketball Source Reconciliation and Metadata Quality Review
# Assess source-location matches, unresolved differences, and imported metadata patterns.

import csv
import hashlib
import io
import json
import re
import unicodedata
from collections import Counter, defaultdict
from pathlib import Path

root = next(
    (path for path in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (path / ".git").exists()),
    None,
)
if root is None:
    raise FileNotFoundError("Git repository root not found.")

results = root / "results"
processed = root / "data" / "processed"
inputs = {
    "pairs": results / "basketball_overlap_source_comparisons_v0_3.csv",
    "cases": results / "basketball_overlap_review_worklist_v0_3.csv",
    "checks": results / "basketball_overlap_source_heading_checks_v0_3.csv",
    "decisions": results / "basketball_overlap_reconciliation_decisions_v0_3.csv",
    "imported": processed / "jr_nba_drills_master.csv",
    "catalog": processed / "basketball_drill_catalog_v0_3_REVIEW_ONLY.csv",
    "integration": results / "basketball_drill_integration_summary_v0_3.json",
}
outputs = {
    "pairs": results / "basketball_overlap_pair_triage_v0_3.csv",
    "cases": results / "basketball_overlap_case_triage_v0_3.csv",
    "patterns": results / "basketball_imported_metadata_patterns_v0_3.csv",
    "summary": results / "basketball_overlap_triage_summary_v0_3.json",
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open("r", newline="", encoding="utf-8-sig") as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        if not columns or len(columns) != len(set(columns)):
            raise ValueError(f"CSV columns are missing or duplicated: {path.name}")
        rows = list(reader)
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f"Malformed CSV rows: {path.name}")
    return rows, columns


def normalized(value):
    return re.sub(
        r"[^a-z0-9]+", " ",
        unicodedata.normalize("NFKC", value or "").casefold(),
    ).strip()


def clean_url(value):
    return (value or "").strip().rstrip("/").lower()


def prescription_type(value):
    """Screen wording only; do not infer unsupplied units or durations."""
    value = normalized(value)
    if not value or value in {"not specified", "not stated", "unknown"}:
        return "unspecified"
    if re.search(r"\b(seconds?|secs?|minutes?|mins?|hours?)\b", value):
        return "time_measure"
    if re.search(r"\b(makes?|shots?|reps?|repetitions?|each|passes|stops|games|points|lines)\b", value):
        return "count_or_performance_target"
    if re.fullmatch(r"\d+\s*x\s*\d+", value):
        return "unlabeled_sets_or_counts"
    return "other_or_mixed_prescription"


for label, path in inputs.items():
    if not path.is_file():
        raise FileNotFoundError(f"Missing Step 12 input ({label}): {path}")

pairs, _ = read_csv(inputs["pairs"])
cases, _ = read_csv(inputs["cases"])
checks, _ = read_csv(inputs["checks"])
decisions, _ = read_csv(inputs["decisions"])
imported, _ = read_csv(inputs["imported"])
catalog, _ = read_csv(inputs["catalog"])
integration = json.loads(inputs["integration"].read_text(encoding="utf-8"))

expected_counts = {"pairs": 11, "cases": 8, "checks": 19,
                   "decisions": 8, "imported": 23, "catalog": 144}
actual_counts = {"pairs": len(pairs), "cases": len(cases),
                 "checks": len(checks), "decisions": len(decisions),
                 "imported": len(imported), "catalog": len(catalog)}
if actual_counts != expected_counts:
    raise ValueError(f"Step 12 baseline differs: {actual_counts}")
if integration.get("provisional_v03_candidate_groups") != 144:
    raise ValueError("Step 12-26 integration summary no longer matches.")

case_ids = [row["case_id"] for row in cases]
if len(set(case_ids)) != 8:
    raise ValueError("Duplicate overlap case IDs.")
if {row["case_id"] for row in decisions} != set(case_ids):
    raise ValueError("Step 12-28 decision worklist does not match case IDs.")
if {row["case_id"] for row in pairs} != set(case_ids):
    raise ValueError("A case has no comparison pair.")
if len({row["drill_id"] for row in imported}) != 23:
    raise ValueError("Imported drill IDs are not unique.")
if any(row["catalog_approval_status"] != "pending" or row["planner_eligible"] != "no"
       for row in catalog):
    raise ValueError("Catalog contains non-pending approval decisions.")
if any(row["catalog_approval_status"] != "pending" or row["planner_eligible"] != "no"
       for row in decisions):
    raise ValueError("Step 12-28 decision file contains an approval; inspect before continuing.")

# Each Step 12-28 source occurrence should have one unambiguous screen.
source_by_key = {}
for row in checks:
    key = (row["case_id"], row["record_role"], row["evidence_id"])
    if key in source_by_key:
        raise ValueError(f"Repeated source screen: {key}")
    source_by_key[key] = row

case_pairs = defaultdict(list)
pair_triage = []
for index, pair in enumerate(pairs, start=1):
    case_id = pair["case_id"]
    imported_key = (case_id, "imported", pair["imported_evidence_id"])
    historical_key = (case_id, "historical", pair["historical_evidence_id"])
    if imported_key not in source_by_key or historical_key not in source_by_key:
        raise ValueError(f"A comparison pair has no Step 12-28 source screen: {case_id}")

    i_source = source_by_key[imported_key]
    h_source = source_by_key[historical_key]
    i_url = clean_url(i_source["source_url"])
    h_url = clean_url(h_source["source_url"])
    same_location = bool(i_url and h_url and i_url == h_url)
    both_fetched = (
        i_source["fetch_status"] == "fetched"
        and h_source["fetch_status"] == "fetched"
    )
    both_exact_headings = (
        i_source["activity_heading_screen"] == "exact_heading_present"
        and h_source["activity_heading_screen"] == "exact_heading_present"
    )
    same_name_key = normalized(pair["imported_drill_name"]) == normalized(pair["historical_name"])
    i_type = prescription_type(pair["imported_prescription"])
    h_type = prescription_type(pair["historical_prescription"])
    different_wording = pair["prescription_wording_screen"] != "same_wording"

    flags = []
    if not both_fetched:
        flags.append("source_access_incomplete")
    if not both_exact_headings:
        flags.append("source_heading_review")
    if not same_location:
        flags.append("different_source_practice_plans")
    if not same_name_key:
        flags.append("name_alias_review")
    if different_wording:
        flags.append("different_prescription_wording")
    if "unlabeled_sets_or_counts" in (i_type, h_type):
        flags.append("unlabeled_prescription_units")
    if (i_type not in {"unspecified", "unlabeled_sets_or_counts"}
            and h_type not in {"unspecified", "unlabeled_sets_or_counts"}
            and i_type != h_type):
        flags.append("different_prescription_types")
    if pair["equipment_wording_screen"] != "same_wording":
        flags.append("equipment_metadata_comparison")
    if pair["participant_wording_screen"] != "same_wording":
        flags.append("participant_metadata_comparison")

    if same_location and both_fetched and both_exact_headings and same_name_key:
        location_result = "same_official_page_and_matching_heading_candidate"
    elif same_location:
        location_result = "same_official_page_requires_heading_review"
    else:
        location_result = "different_official_pages_requires_variation_review"

    if not both_fetched or not both_exact_headings or not same_name_key:
        priority = "high"
    elif different_wording or not same_location:
        priority = "medium"
    else:
        priority = "standard"

    record = {
        "comparison_id": f"JR-PAIR-{index:04d}",
        "case_id": case_id,
        "imported_evidence_id": pair["imported_evidence_id"],
        "historical_evidence_id": pair["historical_evidence_id"],
        "imported_drill_name": pair["imported_drill_name"],
        "historical_drill_name": pair["historical_name"],
        "imported_source_url": i_source["source_url"],
        "historical_source_url": h_source["source_url"],
        "same_source_page": "yes" if same_location else "no",
        "both_pages_fetched": "yes" if both_fetched else "no",
        "both_headings_exact": "yes" if both_exact_headings else "no",
        "normalized_names_match": "yes" if same_name_key else "no",
        "source_location_assessment": location_result,
        "imported_prescription": pair["imported_prescription"],
        "historical_prescription": pair["historical_prescription"],
        "imported_prescription_type": i_type,
        "historical_prescription_type": h_type,
        "prescription_wording_screen": pair["prescription_wording_screen"],
        "equipment_wording_screen": pair["equipment_wording_screen"],
        "participant_wording_screen": pair["participant_wording_screen"],
        "review_flags": ";".join(flags),
        "review_priority": priority,
        "canonical_distinctness_decision": "pending_manual_source_comparison",
        "variation_review": "pending_manual_source_comparison",
        "prescription_correction": "pending_source_validation" if different_wording else "not_required_by_wording_screen",
        "permissions_review": "pending",
        "planner_eligible": "no",
        "reviewer": "",
        "review_notes": "",
    }
    pair_triage.append(record)
    case_pairs[case_id].append(record)

if len(pair_triage) != 11 or set(case_pairs) != set(case_ids):
    raise ValueError("Pair triage did not represent every comparison.")

case_triage = []
for case in cases:
    case_id = case["case_id"]
    records = case_pairs[case_id]
    same_page = sum(r["same_source_page"] == "yes" for r in records)
    exact = sum(r["both_headings_exact"] == "yes" for r in records)
    high = sum(r["review_priority"] == "high" for r in records)
    flags = sorted({flag for r in records for flag in r["review_flags"].split(";") if flag})
    if case["match_type"] == "near_name_suggestion":
        priority = "high"
    elif high:
        priority = "high"
    elif any(r["review_priority"] == "medium" for r in records):
        priority = "medium"
    else:
        priority = "standard"

    if case["match_type"] == "near_name_suggestion":
        next_review = "Verify name alias and both instructional sections before canonicalization"
    elif same_page:
        next_review = "Review same-page source occurrence and reconcile recorded requirements"
    else:
        next_review = "Compare drill instructions across practice plans for distinct variations"

    case_triage.append({
        "case_id": case_id,
        "match_type": case["match_type"],
        "imported_name": case["imported_name"],
        "historical_name": case["historical_name"],
        "pair_count": len(records),
        "same_source_page_pairs": same_page,
        "different_source_page_pairs": len(records) - same_page,
        "both_headings_exact_pairs": exact,
        "review_priority": priority,
        "review_flags": ";".join(flags),
        "next_review": next_review,
        "same_drill_decision": "pending_manual_review",
        "variation_decision": "pending_manual_review",
        "requirements_decision": "pending_manual_review",
        "permissions_review": "pending",
        "catalog_approval_status": "pending",
        "planner_eligible": "no",
        "reviewer": "",
        "review_date": "",
        "review_notes": "",
    })

# Screen potential repeated or templated metadata, without declaring it erroneous.
pattern_fields = [
    "duration", "equipment", "group_requirement", "primary_goal",
    "skill_category", "instructions_summary", "solo_adaptation_note",
]
metadata_patterns = []
for field in pattern_fields:
    if field not in imported[0]:
        raise ValueError(f"Imported dataset is missing expected field: {field}")
    values = [normalized(row[field]) for row in imported]
    supplied = [v for v in values if v and v not in {"unknown", "not specified", "n a", "none"}]
    frequency = Counter(supplied)
    top, count = frequency.most_common(1)[0] if frequency else ("", 0)
    share = count / len(supplied) if supplied else 0
    flags = []
    if len(supplied) < len(imported):
        flags.append("incomplete_field_coverage")
    if field in {"duration", "primary_goal", "instructions_summary", "group_requirement"} and share >= 0.85:
        flags.append("highly_repeated_field_review")
    if field == "duration" and top and prescription_type(top) == "unlabeled_sets_or_counts":
        flags.append("unlabeled_duration_units_review")
    if field == "instructions_summary" and len(frequency) <= 2 and len(supplied) >= 10:
        flags.append("limited_instructional_diversity_review")
    metadata_patterns.append({
        "field_name": field,
        "total_imported_records": len(imported),
        "records_with_informative_values": len(supplied),
        "distinct_informative_values": len(frequency),
        "most_common_value_count": count,
        "most_common_value_fraction": f"{share:.4f}",
        "most_common_value_short": (
            top[:36] if field in {"duration", "equipment", "group_requirement"}
            and len(top) <= 36 else ""
        ),
        "most_common_value_sha256": hashlib.sha256(top.encode("utf-8")).hexdigest() if top else "",
        "review_flags": ";".join(flags),
        "quality_decision": "review_needed" if flags else "descriptive_only",
    })

summary = {
    "step": "12-29",
    "catalog_candidates": len(catalog),
    "review_cases": len(case_triage),
    "comparison_pairs": len(pair_triage),
    "same_source_page_pairs": sum(row["same_source_page"] == "yes" for row in pair_triage),
    "different_source_page_pairs": sum(row["same_source_page"] == "no" for row in pair_triage),
    "pairs_with_two_exact_headings": sum(row["both_headings_exact"] == "yes" for row in pair_triage),
    "cases_by_priority": dict(sorted(Counter(row["review_priority"] for row in case_triage).items())),
    "metadata_fields_screened": len(metadata_patterns),
    "metadata_fields_flagged": sum(row["quality_decision"] == "review_needed" for row in metadata_patterns),
    "source_location_equivalence_is_not_instructional_equivalence": True,
    "canonical_merges": 0,
    "automatic_approvals": 0,
    "planner_eligible": 0,
    "source_data_overwritten": False,
    "git_operations": False,
}


def serialized_csv(rows):
    output = io.StringIO(newline="")
    writer = csv.DictWriter(output, fieldnames=list(rows[0]))
    writer.writeheader()
    writer.writerows(rows)
    return output.getvalue().encode("utf-8"), list(rows[0])


# Existing review annotations are not replaced on subsequent runs.
manual_fields = {
    "canonical_distinctness_decision", "variation_review", "prescription_correction",
    "reviewer", "review_notes", "same_drill_decision", "variation_decision",
    "requirements_decision", "review_date", "permissions_review",
}
content_by_name = {
    "pairs": pair_triage,
    "cases": case_triage,
    "patterns": metadata_patterns,
}
print("=" * 72)
print("STEP 12-29 — SOURCE RECONCILIATION AND METADATA REVIEW")
print("=" * 72)

saved_statuses = []
for label, rows in content_by_name.items():
    path = outputs[label]
    path.parent.mkdir(parents=True, exist_ok=True)
    contents, column_names = serialized_csv(rows)
    if path.exists():
        prior, prior_columns = read_csv(path)
        if prior_columns != column_names or len(prior) != len(rows):
            raise ValueError(f"Existing report changed its structure: {path.name}")
        id_field = "comparison_id" if label == "pairs" else "case_id" if label == "cases" else "field_name"
        prior_by_id = {row[id_field]: row for row in prior}
        if len(prior_by_id) != len(rows):
            raise ValueError(f"Duplicate identifiers in previously saved report: {path.name}")
        for expected_row in rows:
            old = prior_by_id.get(expected_row[id_field])
            if old is None:
                raise ValueError(f"Previously saved report is missing {expected_row[id_field]}")
            for column in column_names:
                if column in manual_fields and label != "patterns":
                    continue
                if old[column] != str(expected_row[column]):
                    raise ValueError(
                        f"Existing {path.name} differs in non-editable field "
                        f"{column} for {expected_row[id_field]}"
                    )
        saved_statuses.append((path, "PRESERVED"))
    else:
        with path.open("xb") as handle:
            handle.write(contents)
        saved_statuses.append((path, "CREATED"))

summary_path = outputs["summary"]
if summary_path.exists():
    previous = json.loads(summary_path.read_text(encoding="utf-8"))
    if previous != summary:
        raise ValueError("Previously saved Step 12-29 summary differs; review before altering it.")
    saved_statuses.append((summary_path, "PRESERVED"))
else:
    with summary_path.open("x", encoding="utf-8") as handle:
        json.dump(summary, handle, indent=2, ensure_ascii=False)
        handle.write("\n")
    saved_statuses.append((summary_path, "CREATED"))

# Verify that the reports persisted with the expected record counts.
for label, rows in content_by_name.items():
    saved_rows, _ = read_csv(outputs[label])
    if len(saved_rows) != len(rows):
        raise ValueError(f"Saved {label} report has an unexpected number of rows.")

print("Cases / comparison pairs:", len(case_triage), "/", len(pair_triage))
print("Same official source-page pairs:", summary["same_source_page_pairs"])
print("Different official source-page pairs:", summary["different_source_page_pairs"])
print("Pairs with exact headings on both sources:", summary["pairs_with_two_exact_headings"])
print("\nCASE TRIAGE")
for row in case_triage:
    print(f"  {row['case_id']} | {row['review_priority']:<8} | "
          f"{row['imported_name']} / {row['historical_name']}")
print("\nIMPORTED METADATA PATTERNS")
for row in metadata_patterns:
    review = row["review_flags"] or "no automatic concern"
    print(f"  {row['field_name']:<22} "
          f"{row['distinct_informative_values']:>2} distinct | "
          f"{row['most_common_value_fraction']} top-value share | {review}")
print("\nREPORTS")
for path, status in saved_statuses:
    print(f"  {status:<9} {path.relative_to(root)}")
print("\nCanonical merges / approvals / planner eligible: 0 / 0 / 0")
print("Existing source data and review decisions overwritten: No")
print("Git operations performed: None")
print("OVERALL: PASS — review priorities documented; final source decisions pending")
print("=" * 72)
```

```python
# Step 12-30 - Drill Source-Section Comparison and Reconciliation Evidence
# Compare official drill sections; retain unresolved canonical and approval decisions.

import csv
import hashlib
import io
import json
import os
import re
import unicodedata
from collections import Counter, defaultdict
from difflib import SequenceMatcher
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

root = next(
    (p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (p / ".git").exists()), None
)
if root is None:
    raise FileNotFoundError("Git repository root not found.")
results = root / "results"
input_paths = {
    "pairs": results / "basketball_overlap_pair_triage_v0_3.csv",
    "cases": results / "basketball_overlap_case_triage_v0_3.csv",
    "source_checks": results / "basketball_overlap_source_heading_checks_v0_3.csv",
    "decisions": results / "basketball_overlap_reconciliation_decisions_v0_3.csv",
    "catalog": root / "data/processed/basketball_drill_catalog_v0_3_REVIEW_ONLY.csv",
}
output_paths = {
    "occurrences": results / "basketball_source_section_evidence_v0_3.csv",
    "pair_evidence": results / "basketball_overlap_content_comparison_v0_3.csv",
    "case_evidence": results / "basketball_overlap_distinctness_review_v0_3.csv",
    "summary": results / "basketball_overlap_content_summary_v0_3.json",
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open("r", newline="", encoding="utf-8-sig") as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        if not columns or len(columns) != len(set(columns)):
            raise ValueError(f"Missing or duplicate column names: {path.name}")
        rows = list(reader)
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f"Malformed CSV data in {path.name}")
    return rows, columns


def normalize(value):
    return re.sub(
        r"[^a-z0-9]+", " ",
        unicodedata.normalize("NFKC", value or "").casefold()
    ).strip()


def valid_source_url(url):
    split = urlsplit(url or "")
    return (
        split.scheme == "https"
        and split.hostname == "jr.nba.com"
        and split.port is None
        and bool(re.fullmatch(
            r"/basketball-practice-plans/(?:rookie|starter|all-star|mvp)/(?:page/\d+/)?",
            split.path,
        ))
        and not split.query and not split.fragment
    )


class PracticePageParser(HTMLParser):
    """Partition webpage text by headings without retaining original HTML."""

    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.active_heading = None
        self.heading_parts = []
        self.current_section = None
        self.skip_depth = 0

    def handle_starttag(self, tag, attrs):
        if tag in {"script", "style", "noscript", "svg"}:
            self.skip_depth += 1
            return
        if self.skip_depth:
            return
        if re.fullmatch(r"h[1-6]", tag):
            self.active_heading = tag
            self.heading_parts = []
        elif tag in {"div", "p", "li", "br", "td", "tr"} and self.current_section:
            self.current_section["parts"].append(" ")

    def handle_data(self, data):
        if self.skip_depth:
            return
        if self.active_heading is not None:
            self.heading_parts.append(data)
        elif self.current_section is not None:
            self.current_section["parts"].append(data)

    def handle_endtag(self, tag):
        if tag in {"script", "style", "noscript", "svg"}:
            self.skip_depth = max(0, self.skip_depth - 1)
            return
        if self.skip_depth:
            return
        if tag == self.active_heading:
            heading = normalize(" ".join(self.heading_parts))
            if heading:
                self.current_section = {"heading": heading, "parts": []}
                self.sections.append(self.current_section)
            self.active_heading = None
            self.heading_parts = []


def fetch_page(url):
    if os.environ.get("STEP12_OFFLINE_REVIEW", "0") == "1":
        return {"status": "offline_not_checked", "sections": [], "http": "", "error": ""}
    if not valid_source_url(url):
        return {"status": "invalid_url", "sections": [], "http": "", "error": "invalid_source_url"}
    request = Request(url, headers={"User-Agent": "Mozilla/5.0 (compatible; CapstoneEvidenceReview/1.0)"})
    try:
        with urlopen(request, timeout=20) as response:
            resolved = response.geturl()
            content = response.read(2_000_000)
            http = str(response.status)
        if not valid_source_url(resolved):
            return {"status": "unexpected_redirect", "sections": [], "http": http, "error": "unexpected_redirect"}
        parser = PracticePageParser()
        parser.feed(content.decode("utf-8", errors="replace"))
        return {"status": "fetched", "sections": parser.sections, "http": http, "error": ""}
    except (HTTPError, URLError, TimeoutError, OSError, ValueError) as error:
        return {"status": "unavailable", "sections": [], "http": str(getattr(error, "code", "")), "error": type(error).__name__}


for name, path in input_paths.items():
    if not path.is_file():
        raise FileNotFoundError(f"Missing Step 12 input ({name}): {path}")
pairs, _ = read_csv(input_paths["pairs"])
cases, _ = read_csv(input_paths["cases"])
source_checks, _ = read_csv(input_paths["source_checks"])
decisions, _ = read_csv(input_paths["decisions"])
catalog, _ = read_csv(input_paths["catalog"])

if (len(pairs), len(cases), len(source_checks), len(decisions), len(catalog)) != (11, 8, 19, 8, 144):
    raise ValueError("Step 12-29 baseline counts changed; inspect existing reports.")
if len({r["case_id"] for r in cases}) != 8 or {r["case_id"] for r in decisions} != {r["case_id"] for r in cases}:
    raise ValueError("Reconciliation case identifiers do not match.")
if any(r["catalog_approval_status"] != "pending" or r["planner_eligible"] != "no" for r in catalog):
    raise ValueError("Review approval changes in the catalog before proceeding.")
if any(r["catalog_approval_status"] != "pending" or r["planner_eligible"] != "no" for r in decisions):
    raise ValueError("Review prior manual approval decisions before proceeding.")
if len({r["comparison_id"] for r in pairs}) != len(pairs):
    raise ValueError("Repeated comparison identifiers.")

source_by_id = {}
for record in source_checks:
    key = (record["case_id"], record["record_role"], record["evidence_id"])
    if key in source_by_id:
        raise ValueError(f"Repeated source occurrence screen: {key}")
    source_by_id[key] = record

unique_urls = sorted({row["source_url"] for row in source_checks if row["source_url"]})
print("=" * 72)
print("STEP 12-30 — OFFICIAL DRILL CONTENT SIMILARITY REVIEW")
print("=" * 72)
print("Review cases / source comparison pairs:", len(cases), "/", len(pairs))
print("Source occurrences:", len(source_checks))
print("Official source URLs:", len(unique_urls))

page_cache = {}
print("\nOFFICIAL SOURCE ACCESS")
for number, url in enumerate(unique_urls, start=1):
    page_cache[url] = fetch_page(url)
    print(f"  {number:02d}/{len(unique_urls):02d} | {page_cache[url]['status']:<20} | {url}")

# Source instructional text remains transient in memory; only fingerprints and similarity measures are saved.
source_text_by_key = {}
source_evidence = []
for key, record in sorted(source_by_id.items()):
    page = page_cache.get(record["source_url"], {"status": "missing_locator", "sections": [], "http": "", "error": ""})
    target = normalize(record["activity_name"])
    headings = page["sections"]
    exact = [section for section in headings if section["heading"] == target]
    aliases = sorted(
        ((SequenceMatcher(None, target, s["heading"]).ratio(), s["heading"], s)
         for s in headings if target and s["heading"] != target),
        key=lambda match: (-match[0], match[1]),
    )
    if page["status"] != "fetched":
        heading_status = "not_checked"
        selected = None
    elif len(exact) == 1:
        heading_status = "exact_unique_heading"
        selected = exact[0]
    elif len(exact) > 1:
        heading_status = "multiple_exact_headings"
        selected = None
    elif aliases and aliases[0][0] >= 0.88:
        heading_status = "similar_heading_requires_review"
        selected = None
    else:
        heading_status = "heading_not_found"
        selected = None
    content = normalize(" ".join(selected["parts"])) if selected else ""
    words = content.split()
    if selected and len(words) < 8:
        section_status = "too_short_for_content_comparison"
    elif selected and len(words) > 1200:
        section_status = "long_section_requires_manual_check"
    elif selected:
        section_status = "section_available_for_screening"
    else:
        section_status = "section_unavailable"
    normalized_prescription = normalize(record["prescription_recorded"])
    if not normalized_prescription or normalized_prescription in {"not specified", "not stated", "unknown"}:
        prescription_screen = "recorded_prescription_unspecified"
    elif not content:
        prescription_screen = "section_unavailable"
    elif normalized_prescription in content:
        prescription_screen = "exact_recorded_phrase_present"
    elif set(re.findall(r"\b\d+\b", normalized_prescription)) & set(re.findall(r"\b\d+\b", content)):
        prescription_screen = "numbers_overlap_only"
    else:
        prescription_screen = "phrase_not_found_in_section"
    source_text_by_key[key] = content if section_status == "section_available_for_screening" else ""
    source_evidence.append({
        "case_id": record["case_id"],
        "record_role": record["record_role"],
        "evidence_id": record["evidence_id"],
        "activity_name": record["activity_name"],
        "source_url": record["source_url"],
        "fetch_status": page["status"],
        "http_status": page["http"],
        "heading_status": heading_status,
        "exact_heading_count": len(exact),
        "closest_alternative_heading": aliases[0][1] if aliases and aliases[0][0] >= 0.88 else "",
        "closest_alternative_similarity": f"{aliases[0][0]:.4f}" if aliases and aliases[0][0] >= 0.88 else "",
        "source_section_word_count": len(words),
        "source_section_sha256": hashlib.sha256(content.encode("utf-8")).hexdigest() if content else "",
        "section_status": section_status,
        "recorded_prescription": record["prescription_recorded"],
        "prescription_screen": prescription_screen,
        "source_verified_for_final_approval": "no",
        "review_required": "yes",
    })

pairs_by_case = defaultdict(list)
pair_evidence = []
for pair in pairs:
    case_id = pair["case_id"]
    i_key = (case_id, "imported", pair["imported_evidence_id"])
    h_key = (case_id, "historical", pair["historical_evidence_id"])
    if i_key not in source_by_id or h_key not in source_by_id:
        raise ValueError(f"Source screen absent for {pair['comparison_id']}")
    i_text, h_text = source_text_by_key[i_key], source_text_by_key[h_key]
    i_source, h_source = source_by_id[i_key], source_by_id[h_key]
    same_page = i_source["source_url"].rstrip("/") == h_source["source_url"].rstrip("/")
    if i_text and h_text:
        exact_text = i_text == h_text
        similarity = SequenceMatcher(None, i_text.split()[:800], h_text.split()[:800], autojunk=False).ratio()
        if exact_text:
            signal = "identical_extracted_sections"
        elif similarity >= 0.85:
            signal = "high_text_similarity_review"
        elif similarity >= 0.65:
            signal = "moderate_text_similarity_review"
        else:
            signal = "different_extracted_sections_review"
        ratio = f"{similarity:.4f}"
    else:
        signal = "insufficient_section_evidence"
        exact_text = False
        ratio = ""
    # Similarity is diagnostic evidence, not a decision that two drills are interchangeable.
    pair_row = {
        "comparison_id": pair["comparison_id"],
        "case_id": case_id,
        "imported_evidence_id": i_source["evidence_id"],
        "historical_evidence_id": h_source["evidence_id"],
        "imported_drill_name": pair["imported_drill_name"],
        "historical_drill_name": pair["historical_drill_name"],
        "imported_source_url": i_source["source_url"],
        "historical_source_url": h_source["source_url"],
        "same_source_page": "yes" if same_page else "no",
        "content_evidence_signal": signal,
        "extracted_sections_identical": "yes" if exact_text and ratio else "no" if ratio else "not_checked",
        "source_text_similarity": ratio,
        "imported_prescription": pair["imported_prescription"],
        "historical_prescription": pair["historical_prescription"],
        "prescription_wording_screen": pair["prescription_wording_screen"],
        "equipment_wording_screen": pair["equipment_wording_screen"],
        "participant_wording_screen": pair["participant_wording_screen"],
        "recommended_review": (
            "Verify one activity occurrence and reconcile metadata against its source"
            if signal == "identical_extracted_sections" and same_page else
            "Compare instructions and requirements before determining canonical identity"
        ),
        "canonical_merge_authorized": "no",
        "final_approval_authorized": "no",
    }
    pairs_by_case[case_id].append(pair_row)
    pair_evidence.append(pair_row)

case_evidence = []
for case in cases:
    rows = pairs_by_case.get(case["case_id"], [])
    if not rows:
        raise ValueError("Case missing pair evidence: " + case["case_id"])
    signals = Counter(row["content_evidence_signal"] for row in rows)
    needs_source = signals["insufficient_section_evidence"] > 0
    has_name_alias = case["match_type"] == "near_name_suggestion"
    if needs_source:
        review_path = "review_unavailable_or_ambiguous_source_sections"
    elif has_name_alias:
        review_path = "verify_heading_alias_and_drill_distinctness"
    elif any(row["same_source_page"] == "no" for row in rows):
        review_path = "compare_cross_plan_variations_and_prescriptions"
    else:
        review_path = "resolve_same_plan_metadata_and_canonical_identity"
    case_evidence.append({
        "case_id": case["case_id"],
        "match_type": case["match_type"],
        "imported_name": case["imported_name"],
        "historical_name": case["historical_name"],
        "review_priority": case["review_priority"],
        "source_comparison_pairs": len(rows),
        "identical_section_pairs": signals["identical_extracted_sections"],
        "high_similarity_pairs": signals["high_text_similarity_review"],
        "moderate_similarity_pairs": signals["moderate_text_similarity_review"],
        "different_section_pairs": signals["different_extracted_sections_review"],
        "unavailable_section_pairs": signals["insufficient_section_evidence"],
        "review_path": review_path,
        "source_comparison_complete": "yes" if not needs_source else "no",
        "canonical_distinctness_decision": "pending_manual_source_review",
        "variation_decision": "pending_manual_source_review",
        "source_prescription_decision": "pending_manual_source_review",
        "permissions_review": "pending",
        "catalog_approval_status": "pending",
        "planner_eligible": "no",
        "reviewer": "",
        "review_date": "",
        "review_notes": "",
    })

summary = {
    "step": "12-30",
    "catalog_candidate_groups": len(catalog),
    "review_cases": len(case_evidence),
    "source_comparison_pairs": len(pair_evidence),
    "unique_source_urls": len(unique_urls),
    "fetched_source_pages": sum(page["status"] == "fetched" for page in page_cache.values()),
    "source_occurrences_screened": len(source_evidence),
    "source_section_statuses": dict(sorted(Counter(row["section_status"] for row in source_evidence).items())),
    "pair_content_signals": dict(sorted(Counter(row["content_evidence_signal"] for row in pair_evidence).items())),
    "manual_distinctness_decisions_pending": len(case_evidence),
    "canonical_merges": 0,
    "catalog_approvals": 0,
    "planner_eligible": 0,
    "original_files_overwritten": False,
    "git_operations": False,
}


def serialize_csv(rows):
    buffer = io.StringIO(newline="")
    writer = csv.DictWriter(buffer, fieldnames=list(rows[0]))
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode("utf-8")


files = {
    "occurrences": source_evidence,
    "pair_evidence": pair_evidence,
    "case_evidence": case_evidence,
}
print("\nSOURCE-SECTION SCREENING")
for label, count in Counter(row["section_status"] for row in source_evidence).most_common():
    print(f"  {label:<40} {count}")
print("\nPAIR CONTENT COMPARISONS")
for row in pair_evidence:
    print(f"  {row['comparison_id']} | {row['case_id']} | {row['content_evidence_signal']:<36} | similarity={row['source_text_similarity'] or 'N/A'}")
print("\nCASE REVIEW PATHS")
for row in case_evidence:
    print(f"  {row['case_id']} | {row['review_path']}")
print("\nREPORTS")
for label, rows in files.items():
    path = output_paths[label]
    expected = serialize_csv(rows)
    if path.exists():
        saved, cols = read_csv(path)
        if cols != list(rows[0]) or len(saved) != len(rows):
            raise ValueError(f"Existing evidence report has changed schema or row count: {path.name}")
        key_name = {"occurrences": ("case_id", "record_role", "evidence_id"),
                    "pair_evidence": ("comparison_id",),
                    "case_evidence": ("case_id",)}[label]
        saved_by_key = {tuple(r[k] for k in key_name): r for r in saved}
        if len(saved_by_key) != len(rows):
            raise ValueError(f"Repeated row identifiers in {path.name}")
        manual_fields = {"reviewer", "review_date", "review_notes", "canonical_distinctness_decision",
                         "variation_decision", "source_prescription_decision", "permissions_review",
                         "catalog_approval_status", "planner_eligible"} if label == "case_evidence" else set()
        for expected_row in rows:
            prior_row = saved_by_key.get(tuple(str(expected_row[k]) for k in key_name))
            if prior_row is None:
                raise ValueError(f"Missing source evidence identifier in {path.name}")
            for column, value in expected_row.items():
                if column in manual_fields:
                    continue
                if prior_row[column] != str(value):
                    raise ValueError(f"Existing evidence report differs: {path.name} / {column}")
        status = "PRESERVED"
    else:
        path.parent.mkdir(parents=True, exist_ok=True)
        with path.open("xb") as handle:
            handle.write(expected)
        status = "CREATED"
    print(f"  {status:<10} {path.relative_to(root)}")

summary_path = output_paths["summary"]
if summary_path.exists():
    previous = json.loads(summary_path.read_text(encoding="utf-8"))
    if previous != summary:
        raise ValueError("The existing Step 12-30 summary differs; review source changes before proceeding.")
    status = "PRESERVED"
else:
    with summary_path.open("x", encoding="utf-8") as handle:
        json.dump(summary, handle, indent=2, ensure_ascii=False)
        handle.write("\n")
    status = "CREATED"
print(f"  {status:<10} {summary_path.relative_to(root)}")

for name, rows in files.items():
    saved, _ = read_csv(output_paths[name])
    if len(saved) != len(rows):
        raise ValueError("Saved output verification failed: " + name)

print("\nApproval and planner eligibility changes: None")
print("Existing source/catalog files overwritten: No")
print("Git operations performed: None")
if summary["fetched_source_pages"] < len(unique_urls):
    print("SOURCE ACCESS: INCOMPLETE — unavailable pages require manual verification")
print("OVERALL: PASS — evidence screening recorded; final manual review remains open")
print("=" * 72)
```

```python
# Step 12-31 - Crossover Alias Verification and Canonical Review Packet
# Review source wording and prepare documented decisions without modifying catalogs.

import csv
import hashlib
import io
import json
import os
import re
import unicodedata
from collections import Counter, defaultdict
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

root = next(
    (p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (p / ".git").exists()), None
)
if root is None:
    raise FileNotFoundError("Git project root not found.")
results = root / "results"
inputs = {
    "catalog": root / "data/processed/basketball_drill_catalog_v0_3_REVIEW_ONLY.csv",
    "cases": results / "basketball_overlap_distinctness_review_v0_3.csv",
    "pairs": results / "basketball_overlap_content_comparison_v0_3.csv",
    "occurrences": results / "basketball_source_section_evidence_v0_3.csv",
    "source_locations": results / "basketball_overlap_source_heading_checks_v0_3.csv",
    "prior_decisions": results / "basketball_overlap_reconciliation_decisions_v0_3.csv",
}
outputs = {
    "alias": results / "basketball_crossover_alias_source_screen_v0_3.csv",
    "packet": results / "basketball_canonical_review_packet_v0_3.csv",
    "summary": results / "basketball_reconciliation_readiness_summary_v0_3.json",
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline="", encoding="utf-8-sig") as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or len(columns) != len(set(columns)):
        raise ValueError(f"Invalid column names: {path.name}")
    if any(None in r or any(v is None for v in r.values()) for r in rows):
        raise ValueError(f"Malformed CSV: {path.name}")
    return rows, columns


def norm(text):
    return re.sub(
        r"[^a-z0-9]+", " ",
        unicodedata.normalize("NFKC", text or "").casefold(),
    ).strip()


def synonym_key(text):
    # Only used to test the documented Mid/Middle alias; not general merging.
    return re.sub(r"\bmid\b", "middle", norm(text))


def valid_url(url):
    parts = urlsplit(url or "")
    return (
        parts.scheme == "https" and parts.hostname == "jr.nba.com"
        and parts.port is None and not parts.query and not parts.fragment
        and bool(re.fullmatch(
            r"/basketball-practice-plans/(?:rookie|starter|all-star|mvp)/(?:page/\d+/)?",
            parts.path,
        ))
    )


class DrillSectionParser(HTMLParser):
    """Locate heading sections; webpage text is used transiently only."""

    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.tag = None
        self.heading_parts = []
        self.current = None
        self.skip = 0

    def handle_starttag(self, tag, attrs):
        if tag in {"script", "style", "noscript", "svg"}:
            self.skip += 1
            return
        if self.skip:
            return
        if re.fullmatch(r"h[1-6]", tag):
            self.tag = tag
            self.heading_parts = []
        elif tag in {"p", "li", "div", "br", "tr", "td"} and self.current is not None:
            self.current["parts"].append(" ")

    def handle_data(self, data):
        if self.skip:
            return
        if self.tag is not None:
            self.heading_parts.append(data)
        elif self.current is not None:
            self.current["parts"].append(data)

    def handle_endtag(self, tag):
        if tag in {"script", "style", "noscript", "svg"}:
            self.skip = max(0, self.skip - 1)
            return
        if self.skip:
            return
        if tag == self.tag:
            heading = norm(" ".join(self.heading_parts))
            if heading:
                self.current = {"heading": heading, "parts": []}
                self.sections.append(self.current)
            self.tag = None
            self.heading_parts = []


def fetch_sections(url):
    if not valid_url(url):
        return "invalid_url", [], ""
    if os.environ.get("STEP12_OFFLINE_REVIEW") == "1":
        return "offline_not_checked", [], ""
    request = Request(url, headers={"User-Agent": "Mozilla/5.0"})
    try:
        with urlopen(request, timeout=25) as response:
            if not valid_url(response.geturl()):
                return "unexpected_redirect", [], ""
            page_html = response.read(2_000_000).decode("utf-8", errors="replace")
        parser = DrillSectionParser()
        parser.feed(page_html)
        return "fetched", parser.sections, ""
    except (HTTPError, URLError, TimeoutError, OSError, ValueError) as error:
        return "unavailable", [], type(error).__name__


sources = {name: read_csv(path)[0] for name, path in inputs.items()}
catalog = sources["catalog"]
cases = sources["cases"]
pairs = sources["pairs"]
occ = sources["occurrences"]
locations = sources["source_locations"]
prior = sources["prior_decisions"]

if tuple(map(len, (catalog, cases, pairs, occ, locations, prior))) != (144, 8, 11, 19, 19, 8):
    raise ValueError("Step 12 baseline counts changed; review prior outputs.")
if any(r["catalog_approval_status"] != "pending" or r["planner_eligible"] != "no" for r in catalog):
    raise ValueError("Catalog approvals or eligibility have changed; reconcile first.")
if {r["case_id"] for r in cases} != {r["case_id"] for r in prior}:
    raise ValueError("Prior reconciliation case IDs do not match.")
if any(r["canonical_merge_authorized"] != "no" or r["final_approval_authorized"] != "no" for r in pairs):
    raise ValueError("Comparison approval gates changed.")
if any(r["catalog_approval_status"] != "pending" or r["planner_eligible"] != "no" for r in prior):
    raise ValueError("Existing review decisions require separate reconciliation.")

cases_by_id = {r["case_id"]: r for r in cases}
if len(cases_by_id) != 8:
    raise ValueError("Duplicate review case IDs.")
pairs_by_case = defaultdict(list)
for pair in pairs:
    if pair["case_id"] not in cases_by_id:
        raise ValueError("Pair references unknown case.")
    pairs_by_case[pair["case_id"]].append(pair)

# Check extraction fingerprints for accidental reuse across unrelated headings.
hashes = defaultdict(list)
for row in occ:
    sha = row["source_section_sha256"].strip()
    if sha:
        hashes[sha].append(row)
collisions = [
    rows for rows in hashes.values()
    if len({synonym_key(r["activity_name"]) for r in rows}) > 1
]

# Recheck only the unresolved crossover-name alias, retaining all other analyses.
alias_case_id = "JR-MATCH-0008"
alias_locations = [r for r in locations if r["case_id"] == alias_case_id]
if len(alias_locations) != 2 or {r["record_role"] for r in alias_locations} != {"historical", "imported"}:
    raise ValueError("Expected imported and historical crossover source occurrences.")

page_cache = {}
for url in sorted({r["source_url"] for r in alias_locations}):
    page_cache[url] = fetch_sections(url)

alias_rows = []
for location in sorted(alias_locations, key=lambda r: r["record_role"]):
    status, sections, error = page_cache[location["source_url"]]
    source_title = location["activity_name"]
    candidate_matches = [s for s in sections if synonym_key(s["heading"]) == synonym_key(source_title)]
    if status != "fetched":
        match_status = "source_unavailable"
        selected = None
    elif len(candidate_matches) == 1:
        selected = candidate_matches[0]
        match_status = (
            "exact_heading" if norm(source_title) == selected["heading"]
            else "documented_mid_middle_alias"
        )
    elif len(candidate_matches) > 1:
        match_status = "ambiguous_heading_matches"
        selected = None
    else:
        match_status = "no_matching_heading"
        selected = None

    # A short prescription immediately after the heading is a screen, not full instruction validation.
    text = " ".join(selected["parts"]) if selected else ""
    prescriptions = re.findall(r"\(([^)]{1,65})\)", text[:350])
    prescription = next((p.strip() for p in prescriptions if re.search(r"\d", p)), "")
    raw_recorded = location["prescription_recorded"].strip()
    if not prescription:
        prescription_result = "source_prescription_not_extracted"
    elif norm(prescription) == norm(raw_recorded):
        prescription_result = "same_wording"
    else:
        prescription_result = "recorded_prescription_differs_check_original"
    alias_rows.append({
        "case_id": alias_case_id,
        "record_role": location["record_role"],
        "evidence_id": location["evidence_id"],
        "recorded_activity_name": source_title,
        "source_url": location["source_url"],
        "page_access": status,
        "heading_screen": match_status,
        "source_heading_normalized": selected["heading"] if selected else "",
        "recorded_prescription": raw_recorded,
        "source_prescription_screen": prescription,
        "prescription_comparison": prescription_result,
        "source_section_sha256": hashlib.sha256(norm(text).encode()).hexdigest() if text else "",
        "review_required": "yes",
        "canonical_decision": "pending",
        "planner_eligible": "no",
        "error_category": error,
    })

alias_status = {r["heading_screen"] for r in alias_rows}
alias_supported = ("documented_mid_middle_alias" in alias_status and
                   "exact_heading" in alias_status)

# The packet makes the eight decisions explicit; no canonical identity is inferred automatically.
packet = []
for case_id, case in sorted(cases_by_id.items()):
    associated = pairs_by_case[case_id]
    if not associated:
        raise ValueError(f"No source comparisons for {case_id}")
    signals = Counter(p["content_evidence_signal"] for p in associated)
    different_page = any(p["same_source_page"] == "no" for p in associated)
    metadata_diff = sorted({
        field for p in associated for field, col in [
            ("prescription", "prescription_wording_screen"),
            ("equipment", "equipment_wording_screen"),
            ("participants", "participant_wording_screen"),
        ] if p.get(col, "") not in {"same_wording", "both_missing", ""}
    })
    if case_id == alias_case_id:
        next_action = "Verify Mid/Middle alias and source prescription; resolve canonical identity"
    elif any(p["content_evidence_signal"] == "insufficient_section_evidence" for p in associated):
        next_action = "Inspect unavailable source instructions before any distinctness decision"
    elif different_page:
        next_action = "Compare source prescriptions and identify cross-plan variations"
    else:
        next_action = "Resolve same-plan identity and conflicting metadata"
    source_links = ";".join(sorted({u for p in associated for u in
        [p["imported_source_url"], p["historical_source_url"]] if u}))
    packet.append({
        "case_id": case_id,
        "priority": case["review_priority"],
        "imported_name": case["imported_name"],
        "historical_name": case["historical_name"],
        "source_pairs": len(associated),
        "same_page_pairs": sum(p["same_source_page"] == "yes" for p in associated),
        "cross_page_pairs": sum(p["same_source_page"] == "no" for p in associated),
        "identical_extracted_pairs": signals["identical_extracted_sections"],
        "high_similarity_pairs": signals["high_text_similarity_review"],
        "pairs_lacking_section": signals["insufficient_section_evidence"],
        "metadata_fields_to_review": ";".join(metadata_diff),
        "reference_urls": source_links,
        "review_next_action": next_action,
        "crossover_name_alias_screen": (
            "supported_by_official_heading_screen" if alias_supported else
            "needs_review"
        ) if case_id == alias_case_id else "not_applicable",
        "manual_canonical_decision": "pending",
        "manual_variation_decision": "pending",
        "manual_prescription_decision": "pending",
        "manual_requirements_decision": "pending",
        "permissions_review": "pending",
        "reviewer": "",
        "review_date": "",
        "evidence_notes": "",
        "catalog_approval_status": "pending",
        "planner_eligible": "no",
    })

summary = {
    "step": "12-31",
    "catalog_candidate_groups": len(catalog),
    "open_review_cases": len(packet),
    "source_comparison_pairs": len(pairs),
    "source_occurrences_in_prior_review": len(occ),
    "cross_title_fingerprint_collision_groups": len(collisions),
    "alias_heading_screen_supported": alias_supported,
    "alias_source_page_access": dict(sorted(Counter(r["page_access"] for r in alias_rows).items())),
    "alias_prescription_screens": dict(sorted(Counter(r["prescription_comparison"] for r in alias_rows).items())),
    "manual_canonical_decisions_pending": len(packet),
    "catalog_approvals": 0,
    "planner_eligible": 0,
    "source_files_modified": False,
    "git_operations": False,
}


def csv_bytes(rows):
    buffer = io.StringIO(newline="")
    writer = csv.DictWriter(buffer, fieldnames=list(rows[0]))
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode("utf-8")


manual_fields = {"manual_canonical_decision", "manual_variation_decision",
                 "manual_prescription_decision", "manual_requirements_decision",
                 "permissions_review", "reviewer", "review_date", "evidence_notes"}


def safe_save_csv(path, rows, allow_manual_edits=False):
    expected = csv_bytes(rows)
    if path.exists():
        existing, cols = read_csv(path)
        if cols != list(rows[0]) or len(existing) != len(rows):
            raise ValueError(f"Existing output schema/count differs: {path.name}")
        key_fields = ("case_id", "record_role") if "record_role" in cols else ("case_id",)
        keyed = {tuple(row[k] for k in key_fields): row for row in existing}
        if len(keyed) != len(rows):
            raise ValueError(f"Existing output has duplicate IDs: {path.name}")
        for expected_row in rows:
            actual = keyed.get(tuple(str(expected_row[k]) for k in key_fields))
            if actual is None:
                raise ValueError("Existing output case identity has changed")
            for name, value in expected_row.items():
                if allow_manual_edits and name in manual_fields:
                    continue
                if actual[name] != str(value):
                    raise ValueError(f"Existing output differs: {path.name} / {name}")
        result = "PRESERVED"
    else:
        with path.open("xb") as handle:
            handle.write(expected)
        result = "CREATED"
    return result


print("=" * 72)
print("STEP 12-31 — CROSSOVER ALIAS AND CANONICAL REVIEW PACKET")
print("=" * 72)
print("Catalog candidates:", len(catalog))
print("Open cases / comparison pairs:", len(packet), "/", len(pairs))
print("Cross-title source-fingerprint collision groups:", len(collisions))
if collisions:
    print("  WARNING: different activity titles share extracted text fingerprints; inspect extraction quality")

print("\nTARGETED OFFICIAL HEADING CHECK")
for row in alias_rows:
    print(
        f"  {row['record_role']:<10} | {row['page_access']:<12} | "
        f"{row['heading_screen']:<32} | recorded={row['recorded_prescription'] or 'N/A'} "
        f"| source={row['source_prescription_screen'] or 'not extracted'}"
    )
print("Mid/Middle alias heading supported:", "YES" if alias_supported else "NOT YET")

print("\nMANUAL RECONCILIATION PRIORITIES")
for row in packet:
    print(f"  {row['case_id']} | {row['priority']:<8} | {row['review_next_action']}")

print("\nOUTPUTS")
for name, rows in [("alias", alias_rows), ("packet", packet)]:
    print(" ", safe_save_csv(outputs[name], rows, allow_manual_edits=(name == "packet")),
          outputs[name].relative_to(root))
summary_path = outputs["summary"]
if summary_path.exists():
    if json.loads(summary_path.read_text(encoding="utf-8")) != summary:
        raise ValueError("Previous summary differs; review source changes.")
    state = "PRESERVED"
else:
    summary_path.write_text(json.dumps(summary, indent=2) + "\n", encoding="utf-8")
    state = "CREATED"
print(" ", state, summary_path.relative_to(root))

print("\nCanonical merges / approvals / planner-eligible: 0 / 0 / 0")
print("Existing catalogs and evidence overwritten: No")
print("Git operations performed: None")
print("OVERALL: PASS — alias checked; manual canonical decisions still pending")
print("=" * 72)
```

```python
# Step 12-32 - Evidence-Based Drill Identity and Metadata Reconciliation
# Document supported source relationships and proposed corrections without changing approval status.

import csv
import io
import json
import re
from collections import Counter, defaultdict
from pathlib import Path

root = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if root is None:
    raise FileNotFoundError('Git project root not found.')
results = root / 'results'
inputs = {
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_REVIEW_ONLY.csv',
    'packet': results / 'basketball_canonical_review_packet_v0_3.csv',
    'pairs': results / 'basketball_overlap_content_comparison_v0_3.csv',
    'sections': results / 'basketball_source_section_evidence_v0_3.csv',
    'alias': results / 'basketball_crossover_alias_source_screen_v0_3.csv',
    'summary': results / 'basketball_reconciliation_readiness_summary_v0_3.json',
}
outputs = {
    'pair_decisions': results / 'basketball_source_identity_assessments_v0_3.csv',
    'case_decisions': results / 'basketball_canonical_resolution_proposals_v0_3.csv',
    'corrections': results / 'basketball_source_metadata_corrections_proposed_v0_3.csv',
    'summary': results / 'basketball_resolution_proposal_summary_v0_3.json',
}

def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or len(columns) != len(set(columns)):
        raise ValueError(f'Invalid CSV schema: {path.name}')
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f'Malformed CSV: {path.name}')
    return rows, columns

def normalize(value):
    return re.sub(r'[^a-z0-9]+', ' ', (value or '').casefold()).strip()

def text_alias(value):
    return re.sub(r'\bmid\b', 'middle', normalize(value))

def pairs_unique(rows, id_field, expected_count):
    if len(rows) != expected_count or len({r[id_field] for r in rows}) != expected_count:
        raise ValueError(f'Unexpected count or duplicate IDs: {id_field}')

catalog, _ = read_csv(inputs['catalog'])
packet, _ = read_csv(inputs['packet'])
pairs, _ = read_csv(inputs['pairs'])
sections, _ = read_csv(inputs['sections'])
alias, _ = read_csv(inputs['alias'])
prior_summary = json.loads(inputs['summary'].read_text(encoding='utf-8'))

pairs_unique(catalog, 'catalog_candidate_id', 144)
pairs_unique(packet, 'case_id', 8)
pairs_unique(pairs, 'comparison_id', 11)
if len(sections) != 19 or len(alias) != 2:
    raise ValueError('Step 12-30/31 source evidence counts have changed.')
if {r['case_id'] for r in pairs} != {r['case_id'] for r in packet}:
    raise ValueError('Source comparisons and review cases do not agree.')
if any(r['catalog_approval_status'] != 'pending' or r['planner_eligible'] != 'no'
       for r in catalog):
    raise ValueError('Catalog approval status changed; review before proceeding.')
if any(r['catalog_approval_status'] != 'pending' or r['planner_eligible'] != 'no'
       for r in packet):
    raise ValueError('Review packet approval status changed.')
if any(r['canonical_merge_authorized'] != 'no' or r['final_approval_authorized'] != 'no'
       for r in pairs):
    raise ValueError('Prior source comparison now authorizes approval or merging.')
if prior_summary.get('alias_heading_screen_supported') is not True:
    raise ValueError('The Step 12-31 alias heading screen did not pass.')

section_index = {}
for row in sections:
    key = (row['case_id'], row['record_role'], row['evidence_id'])
    if key in section_index:
        raise ValueError(f'Duplicate source-section identity: {key}')
    section_index[key] = row
if len(section_index) != 19:
    raise ValueError('Source-section evidence contains duplicate identities.')

packet_index = {r['case_id']: r for r in packet}
pairs_by_case = defaultdict(list)
pair_assessments = []
for row in sorted(pairs, key=lambda r: r['comparison_id']):
    case = row['case_id']
    imp_key = (case, 'imported', row['imported_evidence_id'])
    old_key = (case, 'historical', row['historical_evidence_id'])
    if imp_key not in section_index or old_key not in section_index:
        raise ValueError(f'Incomplete source-evidence linkage: {row["comparison_id"]}')
    imp, old = section_index[imp_key], section_index[old_key]
    same_page = row['same_source_page'] == 'yes'
    fp_equal = (bool(imp['source_section_sha256'])
                and imp['source_section_sha256'] == old['source_section_sha256'])
    screen = row['content_evidence_signal']
    imp_found = imp['section_status'] == 'section_available_for_screening'
    old_found = old['section_status'] == 'section_available_for_screening'
    if screen == 'identical_extracted_sections' and not fp_equal:
        raise ValueError(f'Identical-section flag conflicts with fingerprints: {row["comparison_id"]}')
    if same_page and fp_equal and imp_found and old_found:
        conclusion = 'same_published_section_supported'
        action = 'reconcile_metadata_for_shared_source_section'
    elif fp_equal and imp_found and old_found:
        conclusion = 'identical_text_across_source_pages'
        action = 'compare_cross_plan_prescriptions_and_variations'
    elif imp_found and old_found and screen == 'high_text_similarity_review':
        conclusion = 'similar_published_sections'
        action = 'compare_cross_plan_prescriptions_and_variations'
    else:
        conclusion = 'insufficient_section_evidence'
        action = 'locate_or_review_official_instructions'
    differing = [label for label, column in (
        ('prescription', 'prescription_wording_screen'),
        ('equipment', 'equipment_wording_screen'),
        ('participants', 'participant_wording_screen'))
        if row.get(column, '') not in {'same_wording', 'both_unavailable', 'both_missing', ''}]
    result = {
        'comparison_id': row['comparison_id'],
        'case_id': case,
        'imported_evidence_id': row['imported_evidence_id'],
        'historical_evidence_id': row['historical_evidence_id'],
        'imported_drill_name': row['imported_drill_name'],
        'historical_drill_name': row['historical_drill_name'],
        'imported_source_url': row['imported_source_url'],
        'historical_source_url': row['historical_source_url'],
        'same_source_page': 'yes' if same_page else 'no',
        'source_section_screen': screen,
        'shared_section_fingerprint': 'yes' if fp_equal else 'no',
        'source_identity_assessment': conclusion,
        'metadata_fields_to_reconcile': ';'.join(differing),
        'imported_recorded_prescription': row['imported_prescription'],
        'historical_recorded_prescription': row['historical_prescription'],
        'next_action': action,
        'canonical_merge_authorized': 'no',
        'planner_eligible': 'no',
    }
    pair_assessments.append(result)
    pairs_by_case[case].append(result)

# A source-heading alias is a naming fact, not authorization to merge two catalog groups.
alias_map = {row['record_role']: row for row in alias}
if set(alias_map) != {'historical', 'imported'}:
    raise ValueError('Expected one historical and one imported alias record.')
if ({r['case_id'] for r in alias} != {'JR-MATCH-0008'}
        or alias_map['historical']['heading_screen'] != 'exact_heading'
        or alias_map['imported']['heading_screen'] != 'documented_mid_middle_alias'
        or text_alias(alias_map['historical']['recorded_activity_name']) !=
           text_alias(alias_map['imported']['recorded_activity_name'])):
    raise ValueError('Mid/Middle alias evidence requires reassessment.')
if any(r['page_access'] != 'fetched' for r in alias):
    raise ValueError('The official alias source pages were not retrieved.')

# Create a source-backed correction proposal rather than editing the original imported data.
corrections = []
for row in sorted(alias, key=lambda r: r['record_role']):
    recorded = row['recorded_prescription'].strip()
    source = row['source_prescription_screen'].strip()
    if source and normalize(recorded) != normalize(source):
        corrections.append({
            'case_id': row['case_id'],
            'evidence_id': row['evidence_id'],
            'record_role': row['record_role'],
            'field': 'source_prescription',
            'existing_recorded_value': recorded,
            'proposed_source_value': source,
            'official_source_url': row['source_url'],
            'evidence_basis': 'official_heading_adjacent_prescription_screen',
            'proposed_status': 'source_text_supported_pending_review',
            'source_dataset_updated': 'no',
            'review_decision': 'pending',
            'reviewer': '',
            'review_notes': '',
        })

case_proposals = []
for case_id, case in sorted(packet_index.items()):
    matches = pairs_by_case[case_id]
    if not matches:
        raise ValueError(f'Case has no source comparison pairs: {case_id}')
    conclusions = Counter(r['source_identity_assessment'] for r in matches)
    shared_page = conclusions['same_published_section_supported']
    cross_page = sum(r['same_source_page'] == 'no' for r in matches)
    if case_id == 'JR-MATCH-0008':
        proposal = 'documented_name_alias_pending_canonical_signoff'
        review_action = 'Confirm source prescription; decide whether one canonical drill has plan variations'
    elif conclusions['insufficient_section_evidence']:
        proposal = 'hold_for_source_section_review'
        review_action = 'Inspect missing or ambiguous source sections before identity decisions'
    elif cross_page:
        proposal = 'shared_drill_family_candidate_cross_plan_variation_review'
        review_action = 'Compare official prescriptions and determine whether plan variations are material'
    elif shared_page == len(matches):
        proposal = 'shared_source_section_candidate_pending_metadata_reconciliation'
        review_action = 'Resolve metadata discrepancies; validate canonical source identity'
    else:
        proposal = 'source_content_similarity_candidate_pending_review'
        review_action = 'Review source content and practical requirements before canonical decision'
    issue_fields = sorted({name for row in matches
                           for name in row['metadata_fields_to_reconcile'].split(';') if name})
    case_proposals.append({
        'case_id': case_id,
        'imported_name': case['imported_name'],
        'historical_name': case['historical_name'],
        'case_priority': case['priority'],
        'comparison_pairs': len(matches),
        'same_published_section_pairs': shared_page,
        'cross_plan_comparison_pairs': cross_page,
        'similar_or_identical_pairs': sum(
            r['source_identity_assessment'] in {
                'same_published_section_supported',
                'identical_text_across_source_pages',
                'similar_published_sections'} for r in matches),
        'insufficient_evidence_pairs': conclusions['insufficient_section_evidence'],
        'documented_alias_supported': 'yes' if case_id == 'JR-MATCH-0008' else 'not_applicable',
        'metadata_fields_to_reconcile': ';'.join(issue_fields),
        'proposed_identity_review_path': proposal,
        'proposed_next_action': review_action,
        'canonical_decision': 'pending',
        'variation_decision': 'pending',
        'metadata_decision': 'pending',
        'permission_review': 'pending',
        'reviewer': '',
        'review_date': '',
        'review_notes': '',
        'catalog_approval_status': 'pending',
        'planner_eligible': 'no',
    })

summary = {
    'step': '12-32',
    'catalog_candidate_groups_unchanged': len(catalog),
    'review_cases': len(case_proposals),
    'source_comparison_pairs': len(pair_assessments),
    'same_published_section_pairs': sum(
        r['source_identity_assessment'] == 'same_published_section_supported'
        for r in pair_assessments),
    'identical_text_across_source_pages': sum(
        r['source_identity_assessment'] == 'identical_text_across_source_pages'
        for r in pair_assessments),
    'similar_source_section_pairs': sum(
        r['source_identity_assessment'] == 'similar_published_sections'
        for r in pair_assessments),
    'insufficient_source_section_pairs': sum(
        r['source_identity_assessment'] == 'insufficient_section_evidence'
        for r in pair_assessments),
    'mid_middle_heading_alias_supported': True,
    'proposed_prescription_corrections': len(corrections),
    'canonical_decisions_pending': len(case_proposals),
    'catalog_approved': 0,
    'planner_eligible': 0,
    'source_datasets_modified': False,
    'git_operations': False,
}

editable = {'canonical_decision', 'variation_decision', 'metadata_decision',
            'permission_review', 'reviewer', 'review_date', 'review_notes', 'review_decision'}

def serialize_csv(rows):
    if not rows:
        raise ValueError('Cannot serialize an empty output.')
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=list(rows[0].keys()))
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')

# Verify any pre-existing results before writing missing files; preserve manual review cells.
expected_rows = {'pair_decisions': pair_assessments,
                 'case_decisions': case_proposals,
                 'corrections': corrections}
for label, path in outputs.items():
    if not path.exists():
        continue
    if label == 'summary':
        previous = json.loads(path.read_text(encoding='utf-8'))
        if previous != summary:
            raise ValueError(f'Existing output differs: {path.name}')
        continue
    old, columns = read_csv(path)
    generated = expected_rows[label]
    if len(old) != len(generated) or columns != list(generated[0].keys()):
        raise ValueError(f'Existing output schema/count changed: {path.name}')
    id_cols = ('comparison_id',) if label == 'pair_decisions' else (
        ('case_id',) if label == 'case_decisions' else ('case_id', 'evidence_id', 'field'))
    indexed = {tuple(row[c] for c in id_cols): row for row in old}
    if len(indexed) != len(old):
        raise ValueError(f'Duplicate output keys: {path.name}')
    for row in generated:
        found = indexed.get(tuple(str(row[c]) for c in id_cols))
        if found is None:
            raise ValueError(f'Existing case identities differ: {path.name}')
        for col, val in row.items():
            if col in editable:
                continue
            if found[col] != str(val):
                raise ValueError(f'Existing output changed: {path.name} / {col}')

print('=' * 72)
print('STEP 12-32 — EVIDENCE-BASED RECONCILIATION PROPOSALS')
print('=' * 72)
print('Catalog candidate groups:', len(catalog))
print('Review cases / source comparisons:', len(case_proposals), '/', len(pair_assessments))
print('\nSOURCE IDENTITY ASSESSMENTS')
for signal, count in Counter(r['source_identity_assessment'] for r in pair_assessments).items():
    print(f'  {signal:<43} {count}')
print('\nCASE-LEVEL PROPOSALS')
for row in case_proposals:
    print(f'  {row["case_id"]} | {row["proposed_identity_review_path"]}')
print('\nPROPOSED METADATA CORRECTIONS')
if corrections:
    for row in corrections:
        print(f'  {row["evidence_id"]} | {row["existing_recorded_value"] or "(missing)"} '
              f'-> {row["proposed_source_value"]} | decision pending')
else:
    print('  None requiring a recorded-value correction')
print('\nOUTPUTS')
for label, path in outputs.items():
    if path.exists():
        status = 'PRESERVED'
    elif label == 'summary':
        with path.open('x', encoding='utf-8') as handle:
            json.dump(summary, handle, indent=2)
            handle.write('\n')
        status = 'CREATED'
    else:
        with path.open('xb') as handle:
            handle.write(serialize_csv(expected_rows[label]))
        status = 'CREATED'
    print(f'  {status:<10} {path.relative_to(root)}')

for label, path in outputs.items():
    if label == 'summary':
        assert json.loads(path.read_text(encoding='utf-8')) == summary
    else:
        assert len(read_csv(path)[0]) == len(expected_rows[label])
print('\nCanonical merges / approvals / planner-eligible: 0 / 0 / 0')
print('Existing catalog or source files overwritten: No')
print('Git operations performed: None')
print('OVERALL: PASS — proposals recorded; final human sign-off remains pending')
print('=' * 72)
```

```python
# Step 12-33 - Source Identity Reconciliation Ledger and Catalog Review Snapshot
# Preserve original evidence, distinguish source identity from approval, and track corrections.

import csv
import hashlib
import io
import json
from collections import Counter, defaultdict
from pathlib import Path

root = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if root is None:
    raise FileNotFoundError('Git project root not found.')
results = root / 'results'
inputs = {
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_REVIEW_ONLY.csv',
    'registry': results / 'basketball_drill_source_registry_v0_3.csv',
    'pairs': results / 'basketball_source_identity_assessments_v0_3.csv',
    'cases': results / 'basketball_canonical_resolution_proposals_v0_3.csv',
    'corrections': results / 'basketball_source_metadata_corrections_proposed_v0_3.csv',
}
outputs = {
    'pair_ledger': results / 'basketball_pair_relationship_evidence_v0_3.csv',
    'case_ledger': results / 'basketball_canonical_decision_ledger_v0_3.csv',
    'correction_plan': results / 'basketball_metadata_correction_review_plan_v0_3.csv',
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_1_REVIEW_ONLY.csv',
    'summary': results / 'basketball_step12_33_summary_v0_3.json',
}

def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as stream:
        reader = csv.DictReader(stream)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or len(columns) != len(set(columns)) or any(not c for c in columns):
        raise ValueError(f'Invalid CSV columns: {path}')
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f'Malformed CSV record: {path}')
    return rows, columns

def unique_index(rows, field, expected=None):
    if expected is not None and len(rows) != expected:
        raise ValueError(f'{field}: expected {expected} rows, got {len(rows)}')
    values = [r[field].strip() for r in rows]
    if not all(values) or len(set(values)) != len(values):
        raise ValueError(f'Duplicate or missing {field}')
    return {r[field]: r for r in rows}

def digest(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()

def csv_bytes(rows, fields):
    temp = io.StringIO(newline='')
    writer = csv.DictWriter(temp, fieldnames=fields, lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return temp.getvalue().encode('utf-8')

def ensure_output(path, content):
    if path.exists():
        if path.read_bytes() != content:
            raise FileExistsError(f'Conflicting existing output, preserved: {path.name}')
        return 'PRESERVED'
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open('xb') as f:
        f.write(content)
    return 'CREATED'

catalog, catalog_columns = read_csv(inputs['catalog'])
registry, _ = read_csv(inputs['registry'])
pairs, _ = read_csv(inputs['pairs'])
cases, _ = read_csv(inputs['cases'])
corrections, _ = read_csv(inputs['corrections'])

catalog_by_id = unique_index(catalog, 'catalog_candidate_id', 144)
registry_by_id = unique_index(registry, 'evidence_id', 156)
unique_index(pairs, 'comparison_id', 11)
cases_by_id = unique_index(cases, 'case_id', 8)
if len(corrections) != 1 or corrections[0]['evidence_id'] != 'ADI-JRNBA002':
    raise ValueError('The previously reviewed metadata-correction proposal changed.')
if {r['case_id'] for r in pairs} != set(cases_by_id):
    raise ValueError('Comparison pair references do not match the eight review cases.')
if any(r['catalog_approval_status'] != 'pending' or r['planner_eligible'] != 'no'
       or r['approved_drill_id'].strip() for r in catalog):
    raise ValueError('Catalog approval state changed; do not overwrite reviewed decisions.')
if any(r['canonical_merge_authorized'] != 'no' or r['planner_eligible'] != 'no'
       for r in pairs):
    raise ValueError('Unexpected source comparison approval/merge authorization.')
if any(r['canonical_decision'] != 'pending' or r['variation_decision'] != 'pending'
       or r['catalog_approval_status'] != 'pending' or r['planner_eligible'] != 'no'
       for r in cases):
    raise ValueError('Existing human review decisions may have changed.')

input_hashes = {k: digest(p) for k, p in inputs.items()}
pair_ledger = []
pairs_by_case = defaultdict(list)
groups_by_case = defaultdict(set)

for row in sorted(pairs, key=lambda r: r['comparison_id']):
    imported = registry_by_id.get(row['imported_evidence_id'])
    historical = registry_by_id.get(row['historical_evidence_id'])
    if imported is None or historical is None:
        raise ValueError(f'Missing source ID for {row["comparison_id"]}')
    if imported['origin'] != 'aditi_starter_drill_dataset':
        raise ValueError(f'Unexpected imported source lineage: {row["imported_evidence_id"]}')
    if historical['origin'] == 'aditi_starter_drill_dataset':
        raise ValueError(f'Unexpected historical source lineage: {row["historical_evidence_id"]}')
    if (imported['source_url'].rstrip('/') != row['imported_source_url'].rstrip('/')
        or historical['source_url'].rstrip('/') != row['historical_source_url'].rstrip('/')):
        raise ValueError(f'Source URL mismatch for {row["comparison_id"]}')
    imported_gid = imported['catalog_candidate_id']
    historical_gid = historical['catalog_candidate_id']
    if imported_gid not in catalog_by_id or historical_gid not in catalog_by_id:
        raise ValueError(f'Unknown candidate group in {row["comparison_id"]}')
    same_group = imported_gid == historical_gid
    same_page_supported = row['source_identity_assessment'] == 'same_published_section_supported'
    if same_page_supported and (row['same_source_page'] != 'yes'
                               or row['shared_section_fingerprint'] != 'yes'
                               or not same_group):
        raise ValueError(f'Conflicting same-section evidence: {row["comparison_id"]}')
    state = row['source_identity_assessment']
    if state == 'same_published_section_supported':
        link = 'same_published_section_evidence'
        next_review = 'reconcile_recorded_requirements_and_prescription'
    elif state == 'identical_text_across_source_pages':
        link = 'same_text_on_different_practice_plan_pages'
        next_review = 'review_publisher_plan_context_and_variations'
    elif state == 'similar_published_sections':
        link = 'similar_source_sections_cross_plan'
        next_review = 'compare_original_instructions_and_variations'
    elif state == 'insufficient_section_evidence':
        link = 'insufficient_section_evidence'
        next_review = 'resolve_source_section_and_alias_evidence'
    else:
        raise ValueError(f'Unexpected source identity classification: {state}')
    pair_ledger.append({
        'comparison_id': row['comparison_id'],
        'case_id': row['case_id'],
        'imported_evidence_id': row['imported_evidence_id'],
        'historical_evidence_id': row['historical_evidence_id'],
        'imported_candidate_group_id': imported_gid,
        'historical_candidate_group_id': historical_gid,
        'already_in_same_candidate_group': 'yes' if same_group else 'no',
        'source_identity_evidence': link,
        'recorded_metadata_fields_to_review': row['metadata_fields_to_reconcile'],
        'review_action': next_review,
        'source_comparison_id': row['comparison_id'],
        'canonical_identity_decision': 'pending',
        'variation_decision': 'pending',
        'human_signoff': 'pending',
        'planner_eligible': 'no',
    })
    pairs_by_case[row['case_id']].append(pair_ledger[-1])
    groups_by_case[row['case_id']].update({imported_gid, historical_gid})

case_ledger = []
for case_id, case in sorted(cases_by_id.items()):
    reviewed_pairs = pairs_by_case[case_id]
    gids = sorted(groups_by_case[case_id])
    if len(reviewed_pairs) != int(case['comparison_pairs']):
        raise ValueError(f'Pair count mismatch for {case_id}')
    if case_id == 'JR-MATCH-0008':
        if case['documented_alias_supported'] != 'yes' or len(gids) != 2:
            raise ValueError('The Mid/Middle alias relationship changed.')
        relationship = 'documented_name_alias_pending_identity_signoff'
        action = 'confirm_alias_equivalence_and_source_prescription'
    elif len(gids) != 1:
        raise ValueError(f'Unexpected group membership for {case_id}: {gids}')
    elif int(case['cross_plan_comparison_pairs']) > 0:
        relationship = 'existing_name_group_with_cross_plan_variations_pending'
        action = 'compare_each_practice_plan_and_preserve_distinct_prescriptions'
    else:
        relationship = 'shared_published_section_in_existing_name_group'
        action = 'reconcile_source_specific_equipment_and_participants'
    case_ledger.append({
        'case_id': case_id,
        'imported_name': case['imported_name'],
        'historical_name': case['historical_name'],
        'review_priority': case['case_priority'],
        'candidate_group_ids': ';'.join(gids),
        'group_count': len(gids),
        'source_comparison_pairs': len(reviewed_pairs),
        'supported_same_section_pairs': case['same_published_section_pairs'],
        'cross_plan_comparison_pairs': case['cross_plan_comparison_pairs'],
        'relationship_evidence_status': relationship,
        'metadata_fields_pending': case['metadata_fields_to_reconcile'],
        'next_review_action': action,
        'canonical_identity_decision': 'pending',
        'variation_decision': 'pending',
        'metadata_decision': 'pending',
        'permissions_review': 'pending',
        'reviewer': '',
        'review_date': '',
        'review_notes': '',
        'approved_drill_id': '',
        'catalog_approval_status': 'pending',
        'planner_eligible': 'no',
    })

correction_plan = []
for correction in corrections:
    evidence = registry_by_id.get(correction['evidence_id'])
    if evidence is None or correction['source_dataset_updated'] != 'no' or correction['review_decision'] != 'pending':
        raise ValueError('Proposed correction was already processed or source identity changed.')
    if evidence['source_prescription'].strip() != correction['existing_recorded_value'].strip():
        raise ValueError('Imported source prescription changed since the proposed correction.')
    if correction['case_id'] != 'JR-MATCH-0008':
        raise ValueError('Unexpected correction case.')
    correction_plan.append({
        'case_id': correction['case_id'],
        'evidence_id': correction['evidence_id'],
        'candidate_group_id': evidence['catalog_candidate_id'],
        'field': correction['field'],
        'original_imported_value': correction['existing_recorded_value'],
        'proposed_source_value': correction['proposed_source_value'],
        'source_url': correction['official_source_url'],
        'evidence_basis': correction['evidence_basis'],
        'review_status': 'source_supported_pending_human_signoff',
        'record_update_mode': 'proposed_review_annotation_only',
        'source_dataset_changed': 'no',
        'catalog_value_changed': 'no',
        'reviewer': '',
        'review_date': '',
    })

case_by_group = defaultdict(list)
for case in case_ledger:
    for gid in case['candidate_group_ids'].split(';'):
        case_by_group[gid].append(case)

new_fields = [
    'source_identity_review_cases',
    'source_identity_evidence_status',
    'canonical_linkage_review_status',
    'metadata_reconciliation_fields',
    'source_correction_proposal_ids',
    'source_review_next_action',
]
if set(new_fields) & set(catalog_columns):
    raise ValueError('The review annotation fields are already in Catalog v0.3.')
annotated_catalog = []
for original in catalog:
    row = dict(original)
    row['catalog_version'] = '0.3.1'
    connected = case_by_group.get(row['catalog_candidate_id'], [])
    row.update({
        'source_identity_review_cases': ';'.join(c['case_id'] for c in connected),
        'source_identity_evidence_status': ';'.join(c['relationship_evidence_status'] for c in connected),
        'canonical_linkage_review_status': 'pending_human_review' if connected else 'not_in_eight_overlap_cases',
        'metadata_reconciliation_fields': ';'.join(sorted({f for c in connected
                                          for f in c['metadata_fields_pending'].split(';') if f})),
        'source_correction_proposal_ids': ';'.join(c['evidence_id'] for c in correction_plan
                                        if c['candidate_group_id'] == row['catalog_candidate_id']),
        'source_review_next_action': ';'.join(c['next_review_action'] for c in connected),
    })
    annotated_catalog.append(row)

if len(annotated_catalog) != 144 or len(pair_ledger) != 11 or len(case_ledger) != 8:
    raise ValueError('The source review scope was not preserved.')
if sum(int(c['group_count']) == 2 for c in case_ledger) != 1:
    raise ValueError('Unexpected alias-to-candidate-group mapping.')
for previous, current in zip(catalog, annotated_catalog):
    for col in catalog_columns:
        if col != 'catalog_version' and previous[col] != current[col]:
            raise ValueError(f'Unexpected catalog mutation in field {col}')
    if current['catalog_approval_status'] != 'pending' or current['planner_eligible'] != 'no':
        raise ValueError('Approval gate must remain closed.')
if any(digest(path) != input_hashes[name] for name, path in inputs.items()):
    raise ValueError('An input file changed while preparing the reconciliation snapshot.')

summary = {
    'stage': '12-33',
    'catalog_version': '0.3.1-review-only',
    'original_catalog_candidates': len(catalog),
    'annotated_catalog_candidates': len(annotated_catalog),
    'source_registry_records': len(registry),
    'source_comparison_pairs': len(pair_ledger),
    'review_cases': len(case_ledger),
    'same_published_section_supported_pairs': sum(
        p['source_identity_evidence'] == 'same_published_section_evidence' for p in pair_ledger),
    'identical_cross_page_text_pairs': sum(
        p['source_identity_evidence'] == 'same_text_on_different_practice_plan_pages' for p in pair_ledger),
    'similar_cross_page_section_pairs': sum(
        p['source_identity_evidence'] == 'similar_source_sections_cross_plan' for p in pair_ledger),
    'insufficient_section_pairs': sum(
        p['source_identity_evidence'] == 'insufficient_section_evidence' for p in pair_ledger),
    'catalog_groups_linked_to_review_cases': len(case_by_group),
    'name_alias_cases': sum(int(c['group_count']) == 2 for c in case_ledger),
    'source_metadata_correction_proposals': len(correction_plan),
    'human_canonical_signoffs': 0,
    'automatic_merges': 0,
    'catalog_approvals': 0,
    'planner_eligible': 0,
    'original_source_files_modified': False,
    'git_operations': False,
}

payload = {
    'pair_ledger': csv_bytes(pair_ledger, list(pair_ledger[0])),
    'case_ledger': csv_bytes(case_ledger, list(case_ledger[0])),
    'correction_plan': csv_bytes(correction_plan, list(correction_plan[0])),
    'catalog': csv_bytes(annotated_catalog, catalog_columns + new_fields),
    'summary': (json.dumps(summary, indent=2, ensure_ascii=False) + '\n').encode('utf-8'),
}
# Check existing files before creating anything; divergent outputs are never overwritten.
for name, path in outputs.items():
    if path.exists() and path.read_bytes() != payload[name]:
        raise FileExistsError(f'Existing output differs; preserved without change: {path}')
print('=' * 72)
print('STEP 12-33 — SOURCE IDENTITY LEDGER AND REVIEW CATALOG')
print('=' * 72)
print('Catalog groups:             ', summary['annotated_catalog_candidates'])
print('Source evidence records:    ', summary['source_registry_records'])
print('Pair assessments:           ', summary['source_comparison_pairs'])
print('Same published sections:    ', summary['same_published_section_supported_pairs'])
print('Identical cross-page text:  ', summary['identical_cross_page_text_pairs'])
print('Similar cross-page sections:', summary['similar_cross_page_section_pairs'])
print('Insufficient section pairs: ', summary['insufficient_section_pairs'])
print('Review cases:               ', summary['review_cases'])
print('Catalog groups in cases:    ', summary['catalog_groups_linked_to_review_cases'])
print('Name-alias cases:           ', summary['name_alias_cases'])
print('Proposed source corrections:', summary['source_metadata_correction_proposals'])
print('\nCASE REVIEW LEDGER')
for case in case_ledger:
    print(f"  {case['case_id']} | groups={case['candidate_group_ids']} | {case['relationship_evidence_status']}")
print('\nOUTPUTS')
for name, path in outputs.items():
    state = ensure_output(path, payload[name])
    print(f'  {state:<9} {path.relative_to(root)}')
for name, path in inputs.items():
    if digest(path) != input_hashes[name]:
        raise ValueError('Input integrity changed during execution.')
for name, path in outputs.items():
    if path.read_bytes() != payload[name]:
        raise ValueError(f'Output verification failed: {path}')
print('\nCanonical decisions signed: 0')
print('Approval / planner eligible: 0 / 0')
print('Source datasets and prior catalogs overwritten: No')
print('Git operations performed: None')
print('OVERALL: PASS — identity evidence linked; human decisions remain open')
print('=' * 72)
```

```python
# Step 12-35 - Verify Basketball Source Locations and Prepare Actionable Reviews
# Recheck pending official activity headings; preserve existing catalogs and approvals.

import csv
import hashlib
import io
import json
import re
import unicodedata
from collections import Counter, defaultdict
from datetime import date
from difflib import SequenceMatcher
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

root = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if root is None:
    raise FileNotFoundError('Git project root not found.')

inputs = {
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_1_REVIEW_ONLY.csv',
    'registry': root / 'results/basketball_drill_source_registry_v0_3.csv',
    'source_gaps': root / 'results/basketball_source_requirement_gaps_v0_3_1.csv',
    'readiness': root / 'results/basketball_catalog_readiness_worklist_v0_3_1.csv',
}
outputs = {
    'page_access': root / 'results/basketball_source_locator_page_access_v0_3_1.csv',
    'checks': root / 'results/basketball_source_locator_checks_v0_3_1.csv',
    'worklist': root / 'results/basketball_source_locator_followups_v0_3_1.csv',
    'coverage': root / 'results/eda/basketball_source_locator_coverage_v0_3_1.csv',
    'summary': root / 'results/basketball_step12_35_summary_v0_3_1.json',
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as file:
        reader = csv.DictReader(file)
        fields = list(reader.fieldnames or [])
        rows = list(reader)
    if not fields or len(set(fields)) != len(fields):
        raise ValueError(f'Invalid CSV headings: {path.name}')
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f'Malformed CSV: {path.name}')
    return rows


def sha(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def unique(rows, field, total):
    if len(rows) != total:
        raise ValueError(f'Expected {total} rows for {field}, found {len(rows)}')
    keys = [row[field] for row in rows]
    if not all(keys) or len(set(keys)) != len(keys):
        raise ValueError(f'Missing or duplicate {field}')
    return {row[field]: row for row in rows}


def normalized(value):
    value = unicodedata.normalize('NFKC', value or '').casefold()
    return re.sub(r'[^a-z0-9]+', ' ', value).strip()


def csv_payload(rows):
    if not rows:
        raise ValueError('Refusing to write an empty report')
    output = io.StringIO(newline='')
    writer = csv.DictWriter(output, fieldnames=list(rows[0]), lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return output.getvalue().encode('utf-8')


class DrillHeadingParser(HTMLParser):
    """Keep only headings and counts, never source instructional text on disk."""

    def __init__(self):
        super().__init__()
        self.headings = []
        self.visible_parts = []
        self.heading_depth = 0
        self.heading_parts = []
        self.skip_tag = None
        self.skip_depth = 0

    def handle_starttag(self, tag, attrs):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth += 1
            return
        if tag in {'script', 'style', 'noscript'}:
            self.skip_tag = tag
            self.skip_depth = 1
            return
        if re.fullmatch(r'h[1-6]', tag):
            self.heading_depth += 1
            if self.heading_depth == 1:
                self.heading_parts = []

    def handle_endtag(self, tag):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth -= 1
                if self.skip_depth == 0:
                    self.skip_tag = None
            return
        if re.fullmatch(r'h[1-6]', tag) and self.heading_depth:
            self.heading_depth -= 1
            if not self.heading_depth:
                heading = ' '.join(self.heading_parts).strip()
                if heading:
                    self.headings.append(heading)
                self.heading_parts = []

    def handle_data(self, text):
        if self.skip_depth:
            return
        if text.strip():
            self.visible_parts.append(text.strip())
            if self.heading_depth:
                self.heading_parts.append(text.strip())


def source_plan(url):
    path = urlsplit(url).path.rstrip('/')
    found = re.search(r'/page/(\d+)$', path)
    return int(found.group(1)) if found else 1


def fetch_page(url):
    try:
        request = Request(url, headers={
            'User-Agent': 'Mozilla/5.0 (compatible; educational reference-check/1.0)',
            'Accept': 'text/html',
        })
        with urlopen(request, timeout=22) as response:
            final_url = response.geturl()
            body = response.read(2_000_000)
            status = response.status
        if urlsplit(final_url).hostname != 'jr.nba.com':
            return {'fetch_status': 'unexpected_redirect', 'http_status': status,
                    'final_url': final_url, 'headings': [], 'visible': '',
                    'error': 'Redirect left the official publisher domain'}
        parser = DrillHeadingParser()
        parser.feed(body.decode('utf-8', errors='replace'))
        return {'fetch_status': 'fetched', 'http_status': status,
                'final_url': final_url, 'headings': parser.headings,
                'visible': normalized(' '.join(parser.visible_parts)), 'error': ''}
    except (HTTPError, URLError, TimeoutError, OSError, ValueError) as exc:
        return {'fetch_status': 'unavailable', 'http_status': getattr(exc, 'code', ''),
                'final_url': '', 'headings': [], 'visible': '',
                'error': f'{type(exc).__name__}: {str(exc)[:120]}'}


catalog = read_csv(inputs['catalog'])
registry = read_csv(inputs['registry'])
gaps = read_csv(inputs['source_gaps'])
readiness = read_csv(inputs['readiness'])
cat_by_id = unique(catalog, 'catalog_candidate_id', 144)
reg_by_id = unique(registry, 'evidence_id', 156)
gap_by_id = unique(gaps, 'evidence_id', 156)
ready_by_id = unique(readiness, 'candidate_group_id', 144)
if set(ready_by_id) != set(cat_by_id) or set(gap_by_id) != set(reg_by_id):
    raise ValueError('Source or catalog identifiers do not align.')
if any(r['catalog_approval_status'] != 'pending' or r['planner_eligible'] != 'no'
       or r['approved_drill_id'].strip() for r in catalog):
    raise ValueError('Approval state changed. Inspect current records before continuing.')

flagged = [r for r in registry if 'source_locator_needs_independent_review' in
           gap_by_id[r['evidence_id']]['data_gap_flags'].split(';')]
flagged.sort(key=lambda row: (int(ready_by_id[row['catalog_candidate_id']]['review_wave']),
                              row['origin'], row['evidence_id']))
if len(flagged) != 86:
    raise ValueError(f'Expected 86 source locations, got {len(flagged)}.')
origin_counts = Counter(row['origin'] for row in flagged)
if origin_counts != {'v0_2_heading_expansion': 63, 'aditi_starter_drill_dataset': 23}:
    raise ValueError(f'Unexpected source origin distribution: {origin_counts}')
urls = sorted({r['source_url'] for r in flagged})
if any(urlsplit(url).scheme != 'https' or urlsplit(url).hostname != 'jr.nba.com'
       or not urlsplit(url).path.startswith('/basketball-practice-plans/') for url in urls):
    raise ValueError('Unexpected source URL among pending source locations.')
input_hashes = {name: sha(path) for name, path in inputs.items()}

print('=' * 72)
print('STEP 12-35 — INDEPENDENT BASKETBALL SOURCE-LOCATION SCREEN')
print('=' * 72)
print('Pending source occurrences:', len(flagged))
print('Expanded catalog sources:', origin_counts['v0_2_heading_expansion'])
print('Imported Jr. NBA sources:', origin_counts['aditi_starter_drill_dataset'])
print('Distinct official source URLs:', len(urls))

existing = [path.exists() for path in outputs.values()]
if any(existing):
    if not all(existing):
        raise FileExistsError('Some Step 12-35 reports already exist. No files were changed.')
    prior = json.loads(outputs['summary'].read_text(encoding='utf-8'))
    if prior.get('input_sha256') != input_hashes:
        raise ValueError('Inputs changed since Step 12-35. Preserve prior reports; review changes.')
    for name in ('page_access', 'checks', 'worklist', 'coverage'):
        if sha(outputs[name]) != prior['report_sha256'][name]:
            raise ValueError(f'Existing Step 12-35 report altered: {name}')
    summary = prior
    print('\nEXISTING AUDIT VERIFIED — original report preserved, pages not refetched.')
    for name, path in outputs.items():
        print('  PRESERVED ', path.relative_to(root))
else:
    cache = {}
    print('\nOFFICIAL SOURCE PAGE ACCESS')
    for i, url in enumerate(urls, 1):
        cache[url] = fetch_page(url)
        print(f"  {i:02d}/{len(urls):02d} | {cache[url]['fetch_status']:<19} | {url}")

    page_rows = [{
        'source_url': url,
        'fetch_status': cache[url]['fetch_status'],
        'http_status': cache[url]['http_status'],
        'final_url': cache[url]['final_url'],
        'html_headings_found': len(cache[url]['headings']),
        'referenced_source_occurrences': sum(r['source_url'] == url for r in flagged),
        'fetch_error': cache[url]['error'],
    } for url in urls]

    checks = []
    for source in flagged:
        eid = source['evidence_id']
        page = cache[source['source_url']]
        recorded = normalized(source['drill_name'])
        headings = [(h, normalized(h)) for h in page['headings']]
        matches = [h for h, n in headings if n == recorded]
        alias = [h for h, n in headings if
                 n == re.sub(r'\bmid\b', 'middle', recorded) and n != recorded]
        best = max(headings, key=lambda pair: SequenceMatcher(None, recorded, pair[1]).ratio(),
                   default=('', ''))
        best_ratio = SequenceMatcher(None, recorded, best[1]).ratio() if best[0] else 0
        if page['fetch_status'] != 'fetched':
            status, matched_heading = 'page_unavailable', ''
        elif matches:
            status, matched_heading = 'exact_heading', matches[0]
        elif alias:
            status, matched_heading = 'documented_mid_middle_alias', alias[0]
        elif best_ratio >= 0.82 and len(recorded) >= 10:
            status, matched_heading = 'similar_heading_needs_review', best[0]
        elif recorded and recorded in page['visible']:
            status, matched_heading = 'visible_text_only', ''
        else:
            status, matched_heading = 'heading_not_found', ''

        # A plan URL establishes a candidate location, not independent verification
        # of the PDF page number recorded in the imported source data.
        url_level = re.search(r'/basketball-practice-plans/([^/]+)/', source['source_url'])
        level_match = (url_level is not None and
                       normalized(url_level.group(1)) == normalized(source['source_level']))
        plan_in_url = source_plan(source['source_url'])
        plan_reported = source['practice_plan'].strip()
        doc_marker = re.search(r'practice\s+plan\s+(\d+)', source['source_document'], flags=re.I)
        expected_plan = plan_reported or (doc_marker.group(1) if doc_marker else '')
        plan_match = ('yes' if expected_plan and int(expected_plan) == plan_in_url else
                      'no' if expected_plan else 'not_available')
        if not level_match or plan_match == 'no':
            status = 'source_location_mismatch_review'
        review_wave = ready_by_id[source['catalog_candidate_id']]['review_wave']
        fields = gap_by_id[eid]['data_gap_flags'].split(';')
        field_needs = [f for f in fields if f and f != 'source_locator_needs_independent_review']
        checks.append({
            'evidence_id': eid,
            'candidate_group_id': source['catalog_candidate_id'],
            'review_wave': review_wave,
            'source_origin': source['origin'],
            'drill_name': source['drill_name'],
            'source_level': source['source_level'],
            'recorded_plan': expected_plan,
            'plan_from_url': plan_in_url,
            'reported_source_page': source['source_page'],
            'source_url': source['source_url'],
            'level_url_matches': 'yes' if level_match else 'no',
            'plan_url_matches': plan_match,
            'heading_check': status,
            'matched_heading': matched_heading,
            'closest_heading_similarity': f'{best_ratio:.4f}',
            'other_requirement_flags': ';'.join(field_needs),
            'source_instructions_verified': 'no',
            'source_pdf_page_verified': 'no',
            'permissions_approved': 'no',
            'planner_eligible': 'no',
        })

    actions = {
        'exact_heading': 'Compare instructional section, prescription and missing requirements against original source.',
        'documented_mid_middle_alias': 'Confirm alias identity; verify each-level prescription and pending correction.',
        'similar_heading_needs_review': 'Inspect publisher headings to verify activity name and plan context.',
        'visible_text_only': 'Locate the activity section; visible text does not confirm heading.',
        'heading_not_found': 'Find official activity heading or correct the recorded source location.',
        'page_unavailable': 'Retry official source access and document an alternative permitted source.',
        'source_location_mismatch_review': 'Resolve curriculum level or practice-plan URL mismatch before trusting heading.',
    }
    followups = [{
        'review_rank': i,
        'evidence_id': row['evidence_id'],
        'candidate_group_id': row['candidate_group_id'],
        'review_wave': row['review_wave'],
        'drill_name': row['drill_name'],
        'heading_check': row['heading_check'],
        'source_url': row['source_url'],
        'matched_heading': row['matched_heading'],
        'recorded_source_page': row['reported_source_page'],
        'remaining_metadata_flags': row['other_requirement_flags'],
        'next_action': actions[row['heading_check']],
        'source_instructions_signoff': 'pending',
        'equipment_participant_space_review': 'pending',
        'permission_review': 'pending',
        'canonical_identity_review': 'pending',
        'reviewer': '',
        'review_date': '',
        'review_notes': '',
        'approval_status': 'pending',
        'planner_eligible': 'no',
    } for i, row in enumerate(sorted(checks, key=lambda r: (
        0 if r['heading_check'] not in {'exact_heading', 'documented_mid_middle_alias'} else 1,
        int(r['review_wave']), r['evidence_id'])), 1)]

    coverage = []
    for origin in ['v0_2_heading_expansion', 'aditi_starter_drill_dataset']:
        selected = [r for r in checks if r['source_origin'] == origin]
        coverage.append({
            'source_origin': origin,
            'source_occurrences': len(selected),
            'exact_headings': sum(r['heading_check'] == 'exact_heading' for r in selected),
            'documented_alias_headings': sum(r['heading_check'] == 'documented_mid_middle_alias' for r in selected),
            'needs_heading_or_location_review': sum(r['heading_check'] not in
                                                    {'exact_heading', 'documented_mid_middle_alias'} for r in selected),
            'plan_url_mismatches': sum(r['plan_url_matches'] == 'no' for r in selected),
            'full_source_instructions_approved': 0,
            'planner_eligible': 0,
        })
    coverage.append({
        'source_origin': 'ALL',
        'source_occurrences': len(checks),
        'exact_headings': sum(r['heading_check'] == 'exact_heading' for r in checks),
        'documented_alias_headings': sum(r['heading_check'] == 'documented_mid_middle_alias' for r in checks),
        'needs_heading_or_location_review': sum(r['heading_check'] not in
                                                {'exact_heading', 'documented_mid_middle_alias'} for r in checks),
        'plan_url_mismatches': sum(r['plan_url_matches'] == 'no' for r in checks),
        'full_source_instructions_approved': 0,
        'planner_eligible': 0,
    })
    if coverage[-1]['source_occurrences'] != 86:
        raise ValueError('Some source locations were omitted.')

    report_rows = {'page_access': page_rows, 'checks': checks, 'worklist': followups, 'coverage': coverage}
    payloads = {name: csv_payload(rows) for name, rows in report_rows.items()}
    summary = {
        'step': '12-35',
        'run_date': date.today().isoformat(),
        'baseline_catalog_groups': 144,
        'approved_drills': 0,
        'planner_eligible_drills': 0,
        'sources_targeted': len(flagged),
        'source_origins': dict(sorted(origin_counts.items())),
        'official_pages_requested': len(urls),
        'official_pages_fetched': sum(r['fetch_status'] == 'fetched' for r in page_rows),
        'heading_checks': dict(sorted(Counter(r['heading_check'] for r in checks).items())),
        'plan_location_mismatches': sum(r['plan_url_matches'] == 'no' for r in checks),
        'original_source_values_changed': False,
        'catalog_or_approval_decisions_changed': False,
        'git_operations_performed': False,
        'input_sha256': input_hashes,
        'report_sha256': {name: hashlib.sha256(payload).hexdigest() for name, payload in payloads.items()},
        'interpretation': 'Heading checks establish candidate source locations only; instructional, requirement, copyright, and eligibility reviews remain pending.',
    }
    payloads['summary'] = (json.dumps(summary, indent=2, ensure_ascii=False) + '\n').encode('utf-8')
    if any(sha(path) != input_hashes[name] for name, path in inputs.items()):
        raise RuntimeError('An input changed during source verification.')
    if any(path.exists() for path in outputs.values()):
        raise FileExistsError('Step 12-35 outputs appeared during analysis; none were overwritten.')
    for name, path in outputs.items():
        path.parent.mkdir(parents=True, exist_ok=True)
        with path.open('xb') as file:
            file.write(payloads[name])
    for name, path in outputs.items():
        if path.read_bytes() != payloads[name]:
            raise RuntimeError(f'Output verification failed: {name}')
    print('\nHEADING AND SOURCE-LOCATION RESULTS')
    for key, count in sorted(summary['heading_checks'].items()):
        print(f'  {key:<36} {count:>3}')
    print('  plan_location_mismatches:', summary['plan_location_mismatches'])
    for name, path in outputs.items():
        print('  CREATED  ', path.relative_to(root))

print('\nSUMMARY')
print('Official pages fetched:', summary['official_pages_fetched'], '/', summary['official_pages_requested'])
print('Source locations screened:', summary['sources_targeted'])
print('Source instructions approved: 0')
print('Catalog approval / planner eligible: 0 / 0')
print('Existing catalogs and input records overwritten: No')
print('Git operations performed: None')
print('OVERALL: PASS — official source locator checks recorded; instruction and permission signoff pending')
print('=' * 72)
```

```python
# Step 12-36 - Source-Section Requirement Evidence and Targeted Review
# Screen source sections for factual requirement cues while preserving original data.

import csv
import hashlib
import io
import json
import re
import unicodedata
from collections import Counter
from datetime import date
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

root = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if root is None:
    raise FileNotFoundError('Project root containing .git was not found.')

inputs = {
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_1_REVIEW_ONLY.csv',
    'registry': root / 'results/basketball_drill_source_registry_v0_3.csv',
    'gaps': root / 'results/basketball_source_requirement_gaps_v0_3_1.csv',
    'readiness': root / 'results/basketball_catalog_readiness_worklist_v0_3_1.csv',
    'locator_checks': root / 'results/basketball_source_locator_checks_v0_3_1.csv',
    'locator_summary': root / 'results/basketball_step12_35_summary_v0_3_1.json',
}
outputs = {
    'screen': root / 'results/basketball_requirement_section_screen_v0_3_1.csv',
    'worklist': root / 'results/basketball_requirement_source_review_worklist_v0_3_1.csv',
    'coverage': root / 'results/eda/basketball_requirement_section_coverage_v0_3_1.csv',
    'summary': root / 'results/basketball_step12_36_summary_v0_3_1.json',
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        fields = list(reader.fieldnames or [])
        rows = list(reader)
    if not fields or len(set(fields)) != len(fields):
        raise ValueError(f'Invalid or repeated CSV columns: {path.name}')
    if any(None in row or any(v is None for v in row.values()) for row in rows):
        raise ValueError(f'Malformed CSV row in {path.name}')
    return rows


def index(rows, key, expected):
    if len(rows) != expected:
        raise ValueError(f'Unexpected number of {key} records: {len(rows)} != {expected}')
    values = [row[key] for row in rows]
    if any(not x for x in values) or len(values) != len(set(values)):
        raise ValueError(f'Empty or repeated {key} values')
    return {row[key]: row for row in rows}


def sha(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def normalized(text):
    return re.sub(r'[^a-z0-9]+', ' ',
                  unicodedata.normalize('NFKC', text or '').casefold()).strip()


def csv_bytes(rows):
    if not rows:
        raise ValueError('Cannot write empty result table.')
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=list(rows[0]), lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')


class SectionParser(HTMLParser):
    """Keep source sections temporarily in memory; write only derived evidence."""

    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.skip_tag = ''
        self.skip_depth = 0
        self.heading_tag = ''
        self.heading_text = []
        self.current = None

    def handle_starttag(self, tag, attrs):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth += 1
            return
        if tag in {'script', 'style', 'noscript', 'svg'}:
            self.skip_tag = tag
            self.skip_depth = 1
            return
        if re.fullmatch(r'h[1-6]', tag):
            self.heading_tag = tag
            self.heading_text = []
        elif self.current is not None and tag in {'p', 'div', 'li', 'br', 'td', 'tr'}:
            self.current['parts'].append(' ')

    def handle_data(self, data):
        if self.skip_depth:
            return
        if self.heading_tag:
            self.heading_text.append(data)
        elif self.current is not None:
            self.current['parts'].append(data)

    def handle_endtag(self, tag):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth -= 1
                if not self.skip_depth:
                    self.skip_tag = ''
            return
        if tag == self.heading_tag:
            heading = normalized(' '.join(self.heading_text))
            if heading:
                self.current = {'heading': heading, 'parts': []}
                self.sections.append(self.current)
            self.heading_tag = ''
            self.heading_text = []


def official_url(url):
    parsed = urlsplit(url)
    return (parsed.scheme == 'https' and parsed.hostname == 'jr.nba.com'
            and parsed.port is None
            and re.fullmatch(r'/basketball-practice-plans/(rookie|starter|all-star|mvp)/(page/\d+/)?',
                             parsed.path) is not None
            and not parsed.query and not parsed.fragment)


def fetch_sections(url):
    if not official_url(url):
        raise ValueError(f'Unexpected publisher URL: {url}')
    try:
        request = Request(url, headers={
            'User-Agent': 'Mozilla/5.0 (compatible; AcademicSourceReview/1.0)',
            'Accept': 'text/html',
        })
        with urlopen(request, timeout=22) as response:
            resolved = response.geturl()
            html = response.read(2_000_000).decode('utf-8', errors='replace')
        if not official_url(resolved):
            return {'status': 'unexpected_redirect', 'sections': []}
        parser = SectionParser()
        parser.feed(html)
        return {'status': 'fetched', 'sections': parser.sections}
    except (HTTPError, URLError, TimeoutError, OSError, ValueError):
        return {'status': 'unavailable', 'sections': []}


# Terms appearing in a section are possible evidence cues, not verified requirements.
KEYWORDS = {
    'equipment': {
        'basketball': r'\b(?:basketball|basketballs|ball|balls)\b',
        'cones': r'\b(?:cone|cones)\b',
        'markers': r'\b(?:marker|markers)\b',
        'chair': r'\b(?:chair|chairs)\b',
        'hoop': r'\b(?:hoop|hoops|rim|rims|basket|baskets)\b',
    },
    'space': {
        'court': r'\b(?:court|courts)\b',
        'baseline': r'\b(?:baseline|baselines)\b',
        'sideline': r'\b(?:sideline|sidelines)\b',
        'half_court': r'\bhalf court\b',
        'full_court': r'\bfull court\b',
        'paint_or_key': r'\b(?:paint|key|lane)\b',
        'wing': r'\b(?:wing|wings)\b',
        'free_throw': r'\bfree throw\b',
    },
    'participants': {
        'coach': r'\b(?:coach|coaches)\b',
        'partner': r'\b(?:partner|partners|pair|pairs)\b',
        'defender': r'\b(?:defender|defenders|defense|defensive)\b',
        'team': r'\b(?:team|teams)\b',
        'group': r'\b(?:group|groups)\b',
        'player': r'\b(?:player|players)\b',
    },
    'prescription': {
        'seconds': r'\b(?:second|seconds|sec|secs)\b',
        'minutes': r'\b(?:minute|minutes|min|mins)\b',
        'repetitions': r'\b(?:repetition|repetitions|rep|reps|times)\b',
        'made_shots': r'\b(?:make|makes|made)\b',
        'rounds': r'\b(?:round|rounds|set|sets)\b',
    },
}


def keyword_labels(section, category):
    return ';'.join(label for label, pattern in KEYWORDS[category].items()
                    if re.search(pattern, section))


catalog = read_csv(inputs['catalog'])
registry = read_csv(inputs['registry'])
gaps = read_csv(inputs['gaps'])
readiness = read_csv(inputs['readiness'])
locators = read_csv(inputs['locator_checks'])
locator_summary = json.loads(inputs['locator_summary'].read_text(encoding='utf-8'))
cat_by_id = index(catalog, 'catalog_candidate_id', 144)
reg_by_id = index(registry, 'evidence_id', 156)
gap_by_id = index(gaps, 'evidence_id', 156)
ready_by_id = index(readiness, 'candidate_group_id', 144)
locator_by_id = index(locators, 'evidence_id', 86)
if set(reg_by_id) != set(gap_by_id) or set(cat_by_id) != set(ready_by_id):
    raise ValueError('Source registry, gap report or catalog IDs have changed.')
if locator_summary.get('sources_targeted') != 86:
    raise ValueError('Step 12-35 summary does not match the expected source count.')
if any(row['catalog_approval_status'] != 'pending' or row['planner_eligible'] != 'no'
       or row['approved_drill_id'].strip() for row in catalog):
    raise ValueError('Approval decisions changed; inspect before proceeding.')
if any(row['planner_eligible'] != 'no' for row in locators):
    raise ValueError('Unexpected approved record among reviewed source locations.')
for eid, loc in locator_by_id.items():
    if (eid not in reg_by_id or loc['candidate_group_id'] != reg_by_id[eid]['catalog_candidate_id']
            or loc['source_url'] != reg_by_id[eid]['source_url']):
        raise ValueError(f'Source location mismatch in evidence record {eid}.')

input_hashes = {name: sha(path) for name, path in inputs.items()}
print('=' * 72)
print('STEP 12-36 — DRILL SOURCE-SECTION REQUIREMENT EVIDENCE')
print('=' * 72)
print('Catalog groups:', len(catalog))
print('Source records:', len(registry))
print('Source locations selected for section review:', len(locators))

existing = [path.exists() for path in outputs.values()]
if any(existing):
    if not all(existing):
        raise FileExistsError('Partial Step 12-36 outputs found; original files preserved.')
    summary = json.loads(outputs['summary'].read_text(encoding='utf-8'))
    if summary.get('input_sha256') != input_hashes:
        raise ValueError('Inputs changed after Step 12-36; inspect before rerunning.')
    for name in ('screen', 'worklist', 'coverage'):
        if sha(outputs[name]) != summary['report_sha256'][name]:
            raise ValueError(f'Existing Step 12-36 report was modified: {name}')
    print('EXISTING REPORTS VERIFIED — source pages not refetched.')
    for name, path in outputs.items():
        print('  PRESERVED', path.relative_to(root))
else:
    urls = sorted({row['source_url'] for row in locators})
    if len(urls) != 9:
        raise ValueError(f'Source URL baseline changed: expected 9, found {len(urls)}.')
    pages = {}
    print('\nOFFICIAL SOURCE ACCESS')
    for number, url in enumerate(urls, 1):
        pages[url] = fetch_sections(url)
        print(f'  {number:02d}/{len(urls):02d} | {pages[url]["status"]:<16} | {url}')

    screened = []
    for loc in sorted(locators, key=lambda row: (int(row['review_wave']), row['evidence_id'])):
        eid = loc['evidence_id']
        source = reg_by_id[eid]
        gap = gap_by_id[eid]
        page = pages[loc['source_url']]
        heading_status = loc['heading_check']
        candidate_heading = normalized(loc['matched_heading'])
        section = ''
        section_status = 'not_checked'
        if heading_status not in {'exact_heading', 'documented_mid_middle_alias'}:
            section_status = 'heading_not_verified'
        elif page['status'] != 'fetched':
            section_status = 'page_unavailable'
        elif not candidate_heading:
            section_status = 'source_heading_missing'
        else:
            matched = [row for row in page['sections'] if row['heading'] == candidate_heading]
            if len(matched) != 1:
                section_status = 'ambiguous_or_missing_section'
            else:
                section = normalized(' '.join(matched[0]['parts']))
                word_count = len(section.split())
                section_status = ('short_section_requires_review' if word_count < 8
                                  else 'long_section_requires_review' if word_count > 1200
                                  else 'section_available_for_keyword_screen')
        screenable = section_status == 'section_available_for_keyword_screen'
        text = section if screenable else ''
        flags = list(filter(None, gap['data_gap_flags'].split(';')))
        missing_categories = [category for category in ('equipment', 'participants', 'space', 'prescription')
                              if any(f.startswith(category) or
                                     (category == 'prescription' and f.startswith('source_prescription'))
                                     for f in flags)]
        cues = {category: keyword_labels(text, category) for category in KEYWORDS}
        action = []
        if heading_status not in {'exact_heading', 'documented_mid_middle_alias'}:
            action.append('resolve_source_heading')
        if section_status != 'section_available_for_keyword_screen':
            action.append('inspect_original_section_or_source_pdf')
        for category in missing_categories:
            action.append(f'verify_{category}_from_publisher')
        if 'source_metadata_correction_proposed' in flags:
            action.append('review_pending_metadata_correction')
        action.extend(['review_full_instructions', 'review_content_permissions'])
        screened.append({
            'evidence_id': eid,
            'candidate_group_id': loc['candidate_group_id'],
            'review_wave': loc['review_wave'],
            'source_origin': loc['source_origin'],
            'drill_name': loc['drill_name'],
            'source_url': loc['source_url'],
            'locator_heading_status': heading_status,
            'source_heading': loc['matched_heading'],
            'source_section_status': section_status,
            'source_section_word_count': len(section.split()) if section else 0,
            'source_section_sha256': hashlib.sha256(section.encode('utf-8')).hexdigest() if section else '',
            'equipment_terms_mentioned': cues['equipment'],
            'space_terms_mentioned': cues['space'],
            'participant_terms_mentioned': cues['participants'],
            'prescription_terms_mentioned': cues['prescription'],
            'recorded_equipment': source['equipment'],
            'recorded_space': source['space'],
            'recorded_participants': source['participants'],
            'recorded_prescription': source['source_prescription'],
            'prior_data_gap_flags': gap['data_gap_flags'],
            'source_action_items': ';'.join(dict.fromkeys(action)),
            'instructions_verified': 'no',
            'requirements_verified': 'no',
            'permissions_approved': 'no',
            'planner_eligible': 'no',
        })

    screen_by_id = index(screened, 'evidence_id', 86)
    review_worklist = []
    for pos, loc in enumerate(sorted(locators, key=lambda row: (
            0 if row['heading_check'] not in {'exact_heading', 'documented_mid_middle_alias'} else 1,
            int(row['review_wave']), row['evidence_id'])), 1):
        screened_row = screen_by_id[loc['evidence_id']]
        review_worklist.append({
            'review_rank': pos,
            'evidence_id': loc['evidence_id'],
            'candidate_group_id': loc['candidate_group_id'],
            'review_wave': loc['review_wave'],
            'drill_name': loc['drill_name'],
            'source_url': loc['source_url'],
            'original_heading_status': loc['heading_check'],
            'matched_heading_or_suggestion': loc['matched_heading'],
            'source_section_status': screened_row['source_section_status'],
            'equipment_evidence_terms': screened_row['equipment_terms_mentioned'],
            'space_evidence_terms': screened_row['space_terms_mentioned'],
            'participant_evidence_terms': screened_row['participant_terms_mentioned'],
            'prescription_evidence_terms': screened_row['prescription_terms_mentioned'],
            'outstanding_requirement_flags': screened_row['prior_data_gap_flags'],
            'next_actions': screened_row['source_action_items'],
            'reviewed_equipment': '',
            'reviewed_space': '',
            'reviewed_participants': '',
            'reviewed_prescription': '',
            'reviewed_source_locator': '',
            'reviewer': '',
            'review_date': '',
            'review_notes': '',
            'independent_source_review': 'pending',
            'permission_review': 'pending',
            'canonical_review': 'pending',
            'planner_eligible': 'no',
        })

    coverage = []
    for origin in ['v0_2_heading_expansion', 'aditi_starter_drill_dataset', 'ALL']:
        rows = screened if origin == 'ALL' else [r for r in screened if r['source_origin'] == origin]
        coverage.append({
            'source_origin': origin,
            'records': len(rows),
            'source_sections_screenable': sum(r['source_section_status'] == 'section_available_for_keyword_screen' for r in rows),
            'headings_still_unverified': sum(r['locator_heading_status'] not in {'exact_heading','documented_mid_middle_alias'} for r in rows),
            'source_sections_unavailable_or_ambiguous': sum(r['source_section_status'] != 'section_available_for_keyword_screen' for r in rows),
            'equipment_cues_found': sum(bool(r['equipment_terms_mentioned']) for r in rows),
            'space_cues_found': sum(bool(r['space_terms_mentioned']) for r in rows),
            'participant_cues_found': sum(bool(r['participant_terms_mentioned']) for r in rows),
            'prescription_cues_found': sum(bool(r['prescription_terms_mentioned']) for r in rows),
            'requirement_reviews_approved': 0,
            'permission_approvals': 0,
        })
    if coverage[-1]['records'] != 86:
        raise ValueError('Some source records were omitted from the coverage report.')

    data_tables = {'screen': screened, 'worklist': review_worklist, 'coverage': coverage}
    payloads = {name: csv_bytes(rows) for name, rows in data_tables.items()}
    summary = {
        'step': '12-36',
        'run_date': date.today().isoformat(),
        'catalog_groups': 144,
        'registered_source_occurrences': 156,
        'sources_screened': 86,
        'official_source_urls': len(urls),
        'pages_fetched': sum(p['status'] == 'fetched' for p in pages.values()),
        'heading_checks': dict(Counter(r['locator_heading_status'] for r in screened)),
        'section_status': dict(sorted(Counter(r['source_section_status'] for r in screened).items())),
        'text_evidence_signals': {
            label: sum(bool(r[column]) for r in screened) for label, column in (
                ('equipment', 'equipment_terms_mentioned'), ('space', 'space_terms_mentioned'),
                ('participants', 'participant_terms_mentioned'), ('prescription', 'prescription_terms_mentioned'))},
        'input_sha256': input_hashes,
        'report_sha256': {name: hashlib.sha256(value).hexdigest() for name, value in payloads.items()},
        'source_files_changed': False,
        'catalog_approvals_changed': False,
        'git_operations_performed': False,
        'interpretation': 'Keywords indicate only presence of terms, not verified requirements or permission to reuse instructions.',
    }
    payloads['summary'] = (json.dumps(summary, indent=2) + '\n').encode('utf-8')
    if any(sha(path) != input_hashes[name] for name, path in inputs.items()):
        raise RuntimeError('Inputs changed during source-section screening.')
    if any(path.exists() for path in outputs.values()):
        raise FileExistsError('New output appeared during processing. Existing files preserved.')
    for name, path in outputs.items():
        path.parent.mkdir(parents=True, exist_ok=True)
        with path.open('xb') as handle:
            handle.write(payloads[name])
    for name, path in outputs.items():
        if path.read_bytes() != payloads[name]:
            raise RuntimeError(f'Saved output differs from verified results: {name}')
    print('\nSOURCE-SECTION RESULTS')
    for status, count in sorted(summary['section_status'].items()):
        print(f'  {status:<40} {count}')
    print('\nSOURCE-TEXT EVIDENCE CUES (NOT APPROVALS)')
    for item, count in summary['text_evidence_signals'].items():
        print(f'  {item:<20} {count}')
    pending = [r for r in screened if r['locator_heading_status'] not in {'exact_heading','documented_mid_middle_alias'}]
    print('\nUNRESOLVED SOURCE HEADINGS')
    for row in pending:
        print(f'  {row["evidence_id"]} | {row["drill_name"]} | suggested={row["source_heading"]} | {row["source_url"]}')
    for name, path in outputs.items():
        print('  CREATED ', path.relative_to(root))

print('\nSUMMARY')
print('Source records screened:', summary['sources_screened'])
print('Official pages retrieved:', summary['pages_fetched'], '/', summary['official_source_urls'])
print('Source and catalog approvals updated: No')
print('Existing source and catalog files overwritten: No')
print('Git operations performed: None')
print('OVERALL: PASS — source-section evidence screened; requirement and permission review pending')
print('=' * 72)
```

```python
# Step 12-37 - Source-Verified Metadata Reconciliation: Shield Steal Game
# Verify publisher facts and create versioned review-only corrections without modifying source records.

import csv
import hashlib
import io
import json
import re
import unicodedata
from datetime import date
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

root = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if root is None:
    raise FileNotFoundError('Project root containing .git was not found.')

inputs = {
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_1_REVIEW_ONLY.csv',
    'registry': root / 'results/basketball_drill_source_registry_v0_3.csv',
    'section_screen': root / 'results/basketball_requirement_section_screen_v0_3_1.csv',
    'locator_checks': root / 'results/basketball_source_locator_checks_v0_3_1.csv',
    'imported_drills': root / 'data/processed/jr_nba_drills_master.csv',
}
outputs = {
    'verification': root / 'results/basketball_shield_steal_source_verification_v0_3_2.csv',
    'changes': root / 'results/basketball_source_metadata_change_log_v0_3_2.csv',
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_2_REVIEW_ONLY.csv',
    'registry': root / 'results/basketball_drill_source_registry_v0_3_2.csv',
    'summary': root / 'results/basketball_step12_37_summary_v0_3_2.json',
}

EVIDENCE_ID = 'ADI-JRNBA016'
CANDIDATE_ID = 'JR-ADITI-0013'
SOURCE_URL = 'https://jr.nba.com/basketball-practice-plans/starter/page/2/'
PUBLISHED_NAME = 'Shield Steal Game'


def read_table(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or len(set(columns)) != len(columns):
        raise ValueError(f'Invalid CSV schema: {path.name}')
    if any(None in row or any(v is None for v in row.values()) for row in rows):
        raise ValueError(f'Malformed CSV rows: {path.name}')
    return rows, columns


def indexed(rows, field, count):
    if len(rows) != count:
        raise ValueError(f'{field} count differs from expected baseline: {len(rows)} != {count}')
    keys = [r[field] for r in rows]
    if any(not v for v in keys) or len(set(keys)) != len(keys):
        raise ValueError(f'Non-unique or missing {field} values')
    return dict(zip(keys, rows))


def sha(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def normalized(text):
    return re.sub(r'[^a-z0-9]+', ' ', unicodedata.normalize('NFKC', text or '').casefold()).strip()


def csv_payload(rows, columns):
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=columns, lineterminator='\n', extrasaction='raise')
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')


class HeadingSectionParser(HTMLParser):
    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.skip = ''
        self.skip_count = 0
        self.heading = ''
        self.heading_parts = []
        self.current = None

    def handle_starttag(self, tag, attrs):
        if self.skip_count:
            if tag == self.skip:
                self.skip_count += 1
            return
        if tag in {'script', 'style', 'noscript', 'svg'}:
            self.skip, self.skip_count = tag, 1
        elif re.fullmatch(r'h[1-6]', tag):
            self.heading, self.heading_parts = tag, []
        elif self.current is not None and tag in {'p', 'li', 'div', 'br'}:
            self.current['parts'].append(' ')

    def handle_data(self, data):
        if self.skip_count:
            return
        if self.heading:
            self.heading_parts.append(data)
        elif self.current is not None:
            self.current['parts'].append(data)

    def handle_endtag(self, tag):
        if self.skip_count:
            if tag == self.skip:
                self.skip_count -= 1
                if not self.skip_count:
                    self.skip = ''
            return
        if tag == self.heading:
            heading = normalized(' '.join(self.heading_parts))
            if heading:
                self.current = {'heading': heading, 'parts': []}
                self.sections.append(self.current)
            self.heading, self.heading_parts = '', []


def read_published_section():
    request = Request(SOURCE_URL, headers={'User-Agent': 'Mozilla/5.0', 'Accept': 'text/html'})
    try:
        with urlopen(request, timeout=25) as response:
            destination = urlsplit(response.geturl())
            if destination.scheme != 'https' or destination.hostname != 'jr.nba.com':
                raise RuntimeError('Official source redirected to an unexpected website.')
            html = response.read(2_000_000).decode('utf-8', errors='replace')
    except (HTTPError, URLError, TimeoutError, OSError) as error:
        raise RuntimeError('Official page unavailable; no metadata changes created.') from error
    parser = HeadingSectionParser()
    parser.feed(html)
    matching = [s for s in parser.sections if s['heading'] == normalized(PUBLISHED_NAME)]
    if len(matching) != 1:
        raise RuntimeError('Published Shield Steal Game heading is missing or ambiguous.')
    content = normalized(' '.join(matching[0]['parts']))
    # Guard against agreeing only on a title while the underlying section has changed.
    support = {
        'four_minute_prescription': bool(re.search(r'\b4 minutes\b', content)),
        'pairs': 'divide the group into pairs' in content,
        'half_court': 'half court size space' in content,
        'boundary_markers': 'cones or other items' in content,
        'ball': 'one player has a ball' in content,
        'offense_and_defense': 'other is on defense' in content,
        'coach_direction': 'coach says go' in content,
    }
    if not all(support.values()):
        raise RuntimeError('Published instructions changed or could not be confirmed: '
                           + ', '.join(k for k, ok in support.items() if not ok))
    return support, hashlib.sha256(content.encode('utf-8')).hexdigest()


catalog, catalog_cols = read_table(inputs['catalog'])
registry, registry_cols = read_table(inputs['registry'])
screen, _ = read_table(inputs['section_screen'])
locator, _ = read_table(inputs['locator_checks'])
imported, _ = read_table(inputs['imported_drills'])
cat_by_id = indexed(catalog, 'catalog_candidate_id', 144)
reg_by_id = indexed(registry, 'evidence_id', 156)
screen_by_id = indexed(screen, 'evidence_id', 86)
locator_by_id = indexed(locator, 'evidence_id', 86)
imported_by_id = indexed(imported, 'drill_id', 23)

original = reg_by_id[EVIDENCE_ID]
item = cat_by_id[CANDIDATE_ID]

# Verify identity and approval state without assuming that the preliminary
# prescription strings have exactly the same formatting in both files.
identity_checks = [
    ('Registry candidate link', original.get('catalog_candidate_id') == CANDIDATE_ID,
     original.get('catalog_candidate_id'), CANDIDATE_ID),
    ('Registry source URL', original.get('source_url', '').rstrip('/') == SOURCE_URL.rstrip('/'),
     original.get('source_url'), SOURCE_URL),
    ('Registry activity name', normalized(original.get('drill_name')) == 'shield steal',
     original.get('drill_name'), 'Shield Steal'),
    ('Catalog activity name', normalized(item.get('drill_name')) == 'shield steal',
     item.get('drill_name'), 'Shield Steal'),
    ('Catalog normalized name', normalized(item.get('normalized_name')) == 'shield steal',
     item.get('normalized_name'), 'shield steal'),
    ('Catalog source evidence link', EVIDENCE_ID in item.get('evidence_ids', '').split(';'),
     item.get('evidence_ids'), EVIDENCE_ID),
    ('Catalog approval status', item.get('catalog_approval_status') == 'pending',
     item.get('catalog_approval_status'), 'pending'),
    ('Catalog planner eligibility', item.get('planner_eligible') == 'no',
     item.get('planner_eligible'), 'no'),
    ('No approved drill ID', not item.get('approved_drill_id', '').strip(),
     item.get('approved_drill_id'), '[empty]'),
]

failed_checks = [(label, actual, expected) for label, passed, actual, expected
                 in identity_checks if not passed]
if failed_checks:
    print('\nSOURCE RECORD VALIDATION — unresolved differences:')
    for label, actual, expected in failed_checks:
        print(f'  {label}: actual={actual!r}; expected={expected!r}')
    raise ValueError('Source identity or approval-state check failed. No files changed.')

# Preliminary prescription strings are recorded as found and independently
# compared with the official source; differences alone are not an error.
if original['source_prescription'] != item['source_prescription']:
    print('SOURCE PRESCRIPTION DIFFERENCE (review-only):')
    print('  Registry:', repr(original['source_prescription']))
    print('  Catalog: ', repr(item['source_prescription']))
if imported_by_id['JRNBA016']['drill_name'] != 'Shield Steal':
    raise ValueError('Imported source record changed.')
if screen_by_id[EVIDENCE_ID]['candidate_group_id'] != CANDIDATE_ID:
    raise ValueError('Section-screen record does not link to the expected candidate.')
if locator_by_id[EVIDENCE_ID]['source_url'] != SOURCE_URL:
    raise ValueError('Locator record does not match the source URL.')
if any(r['catalog_approval_status'] != 'pending' or r['planner_eligible'] != 'no'
       or r['approved_drill_id'] for r in catalog):
    raise ValueError('Existing catalog contains approved records; do not overwrite decisions.')
if normalized(PUBLISHED_NAME) in {r['normalized_name'] for r in catalog}:
    raise ValueError('Corrected published name already exists as a separate catalog candidate.')

input_hashes = {label: sha(path) for label, path in inputs.items()}
print('=' * 72)
print('STEP 12-37 — SOURCE-VERIFIED SHIELD STEAL METADATA RECONCILIATION')
print('=' * 72)

existing = [p.exists() for p in outputs.values()]
if any(existing):
    if not all(existing):
        raise FileExistsError('Incomplete Step 12-37 outputs exist; existing data preserved.')
    prior = json.loads(outputs['summary'].read_text(encoding='utf-8'))
    if prior.get('input_sha256') != input_hashes:
        raise RuntimeError('Source inputs changed after the previous execution.')
    for label in ('verification', 'changes', 'catalog', 'registry'):
        if sha(outputs[label]) != prior['report_sha256'][label]:
            raise RuntimeError(f'Existing output was modified: {label}')
    if len(read_table(outputs['catalog'])[0]) != 144 or len(read_table(outputs['registry'])[0]) != 156:
        raise RuntimeError('Previously saved catalog or registry count has changed.')
    summary = prior
    print('EXISTING OUTPUTS VERIFIED — no network request or rewrite.')
else:
    support, section_hash = read_published_section()

    # Only publisher-supported factual fields are revised; all approval gates remain closed.
    updates = {
        'drill_name': PUBLISHED_NAME,
        'source_prescription': '4 minutes',
        'equipment': 'Basketball; cones or equivalent boundary markers',
        'participants': 'Pairs (2 active players: attacker and defender); coach gives start cue',
        'space': 'Half-court area; marked square for each pair',
    }
    changes = []
    for field, corrected in updates.items():
        changes.append({
            'change_id': f'S12-37-{len(changes) + 1:02d}',
            'evidence_id': EVIDENCE_ID,
            'catalog_candidate_id': CANDIDATE_ID,
            'field': field,
            'original_value': original[field],
            'catalog_original_value': item[field],
            'publisher_supported_value': corrected,
            'official_source_url': SOURCE_URL,
            'review_basis': 'official_starter_practice_plan_2_activity_section',
            'source_check': 'passed',
            'team_signoff': 'pending',
            'permission_review': 'pending',
        })

    verification = [{
        'evidence_id': EVIDENCE_ID,
        'catalog_candidate_id': CANDIDATE_ID,
        'source_title_recorded': original['drill_name'],
        'publisher_heading': PUBLISHED_NAME,
        'recorded_prescription': original['source_prescription'],
        'publisher_prescription': updates['source_prescription'],
        'publisher_section_sha256': section_hash,
        'official_source_url': SOURCE_URL,
        'source_heading_verified': 'yes',
        'source_fact_checks_passed': len(support),
        'source_fact_checks_total': len(support),
        'source_pdf_page_verified': 'no',
        'source_instruction_signoff': 'pending',
        'permission_review': 'pending',
        'planner_eligible': 'no',
    }]

    new_registry = [dict(r) for r in registry]
    for r in new_registry:
        if r['evidence_id'] == EVIDENCE_ID:
            r.update(updates)
            r['source_locator_status'] = 'official_html_heading_and_requirements_checked'
    extra_reg = [
        'original_source_drill_name', 'original_source_prescription',
        'source_metadata_review_status', 'source_metadata_review_reference',
    ]
    for r in new_registry:
        for field in extra_reg:
            r[field] = ''
        if r['evidence_id'] == EVIDENCE_ID:
            r['original_source_drill_name'] = original['drill_name']
            r['original_source_prescription'] = original['source_prescription']
            r['source_metadata_review_status'] = 'publisher_facts_checked_pending_team_signoff'
            r['source_metadata_review_reference'] = 'results/basketball_shield_steal_source_verification_v0_3_2.csv'

    new_catalog = [dict(r) for r in catalog]
    extra_catalog = [
        'publisher_verified_heading', 'original_imported_drill_name',
        'original_imported_prescription', 'publisher_metadata_review_status',
        'publisher_metadata_review_reference',
    ]
    for r in new_catalog:
        for field in extra_catalog:
            r[field] = ''
        r['catalog_version'] = '0.3.2'
        if r['catalog_candidate_id'] == CANDIDATE_ID:
            r.update(updates)
            r['normalized_name'] = normalized(PUBLISHED_NAME)
            r['publisher_verified_heading'] = PUBLISHED_NAME
            r['original_imported_drill_name'] = original['drill_name']
            r['original_imported_prescription'] = original['source_prescription']
            r['publisher_metadata_review_status'] = 'publisher_facts_checked_pending_team_signoff'
            r['publisher_metadata_review_reference'] = 'results/basketball_shield_steal_source_verification_v0_3_2.csv'
            r['heading_verification'] = 'official_html_heading_confirmed'
            r['source_locator_status'] = 'official_html_heading_confirmed'
            flags = set(filter(None, r['review_flags'].split(';')))
            flags.add('source_metadata_reviewed_pending_signoff')
            r['review_flags'] = ';'.join(sorted(flags))

    if len(new_catalog) != 144 or len(new_registry) != 156:
        raise RuntimeError('Unexpected output counts.')
    if len({r['normalized_name'] for r in new_catalog}) != 144:
        raise RuntimeError('Corrected name collides with another candidate.')
    if any(r['catalog_approval_status'] != 'pending' or r['planner_eligible'] != 'no'
           or r['approved_drill_id'] for r in new_catalog):
        raise RuntimeError('An approval restriction was changed.')
    # No other catalog record may change beyond the catalog version and new empty columns.
    for before, after in zip(catalog, new_catalog):
        if before['catalog_candidate_id'] != CANDIDATE_ID:
            if any(before[key] != after[key] for key in catalog_cols if key != 'catalog_version'):
                raise RuntimeError('An unrelated catalog record changed.')
    for before, after in zip(registry, new_registry):
        if before['evidence_id'] != EVIDENCE_ID and any(before[k] != after[k] for k in registry_cols):
            raise RuntimeError('An unrelated source record changed.')

    payloads = {
        'verification': csv_payload(verification, list(verification[0])),
        'changes': csv_payload(changes, list(changes[0])),
        'catalog': csv_payload(new_catalog, catalog_cols + extra_catalog),
        'registry': csv_payload(new_registry, registry_cols + extra_reg),
    }
    summary = {
        'step': '12-37',
        'run_date': date.today().isoformat(),
        'catalog_version': '0.3.2',
        'catalog_candidates': len(new_catalog),
        'registry_source_occurrences': len(new_registry),
        'source_record_corrected': EVIDENCE_ID,
        'candidate_group_corrected': CANDIDATE_ID,
        'publisher_heading': PUBLISHED_NAME,
        'metadata_fields_reconciled': sorted(updates),
        'source_fact_checks_passed': len(support),
        'source_fact_checks_total': len(support),
        'original_imported_dataset_modified': False,
        'prior_catalog_modified': False,
        'human_signoff_completed': False,
        'drills_approved': 0,
        'planner_eligible': 0,
        'git_operations_performed': False,
        'input_sha256': input_hashes,
        'report_sha256': {key: hashlib.sha256(value).hexdigest() for key, value in payloads.items()},
    }
    payloads['summary'] = (json.dumps(summary, indent=2) + '\n').encode('utf-8')
    if any(sha(path) != digest for (key, path), digest in zip(inputs.items(), input_hashes.values())):
        raise RuntimeError('Source inputs changed during verification.')
    if any(p.exists() for p in outputs.values()):
        raise FileExistsError('An output appeared during processing; source files preserved.')
    created = []
    try:
        for key, path in outputs.items():
            path.parent.mkdir(parents=True, exist_ok=True)
            with path.open('xb') as handle:
                handle.write(payloads[key])
            created.append(path)
        for key, path in outputs.items():
            if path.read_bytes() != payloads[key]:
                raise RuntimeError(f'Saved file verification failed: {key}')
    except Exception:
        for path in created:
            path.unlink(missing_ok=True)
        raise
    print('\nPUBLISHER SOURCE CONFIRMATION')
    print('  Recorded title:        ', original['drill_name'])
    print('  Official title:        ', PUBLISHED_NAME)
    print('  Recorded prescription: ', original['source_prescription'])
    print('  Official prescription: ', updates['source_prescription'])
    print('  Requirement fact checks:', len(support), '/', len(support))
    print('\nRECONCILED METADATA')
    for row in changes:
        print(f'  {row["field"]:<20} | {row["publisher_supported_value"]}')
    print('\nOUTPUTS')
    for key, path in outputs.items():
        print('  CREATED', path.relative_to(root))

print('\nSUMMARY')
print('Catalog candidates:', summary['catalog_candidates'])
print('Source evidence records:', summary['registry_source_occurrences'])
print('Publisher-supported fields reconciled:', len(summary['metadata_fields_reconciled']))
print('Team signoff / permissions: Pending / Pending')
print('Approved / planner-eligible:', summary['drills_approved'], '/', summary['planner_eligible'])
print('Existing input datasets and prior catalogs modified: No')
print('Git operations performed: None')
print('OVERALL: PASS — source-backed metadata revised in v0.3.2; approval gates remain closed')
print('=' * 72)
```

```python
# Step 12-38 - Batch Publisher Verification for Metadata-Reviewed Basketball Drills
# Examine Wave 2 source evidence without modifying the existing catalog or approvals.

import csv
import hashlib
import io
import json
import re
import tempfile
import unicodedata
from collections import Counter, defaultdict
from datetime import date
from difflib import SequenceMatcher
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

root = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if root is None:
    raise FileNotFoundError('Project root was not found.')

inputs = {
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_2_REVIEW_ONLY.csv',
    'registry': root / 'results/basketball_drill_source_registry_v0_3_2.csv',
    'readiness': root / 'results/basketball_catalog_readiness_worklist_v0_3_1.csv',
}
outputs = {
    'source_checks': root / 'results/basketball_wave2_publisher_checks_v0_3_2.csv',
    'field_comparisons': root / 'results/basketball_wave2_field_comparisons_v0_3_2.csv',
    'review_queue': root / 'results/basketball_wave2_review_queue_v0_3_2.csv',
    'coverage': root / 'results/eda/basketball_wave2_source_coverage_v0_3_2.csv',
    'summary': root / 'results/basketball_step12_38_summary_v0_3_2.json',
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or len(columns) != len(set(columns)):
        raise ValueError(f'Invalid CSV column names: {path.name}')
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f'Malformed CSV rows: {path.name}')
    return rows


def indexed(rows, key, count):
    if len(rows) != count:
        raise ValueError(f'Unexpected {key} count: {len(rows)}; expected {count}')
    keys = [row[key].strip() for row in rows]
    if not all(keys) or len(keys) != len(set(keys)):
        raise ValueError(f'Blank or duplicate {key} identifiers')
    return {row[key]: row for row in rows}


def sha(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def norm(value):
    value = unicodedata.normalize('NFKC', value or '').casefold()
    return re.sub(r'[^a-z0-9]+', ' ', value).strip()


def csv_payload(rows):
    if not rows:
        raise ValueError('Cannot save an empty report.')
    stream = io.StringIO(newline='')
    writer = csv.DictWriter(stream, fieldnames=list(rows[0]),
                            lineterminator='\n', extrasaction='raise')
    writer.writeheader()
    writer.writerows(rows)
    return stream.getvalue().encode('utf-8')


class PracticeSectionParser(HTMLParser):
    """Read heading-scoped sections in memory; never save source instructions."""

    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.skip_tag = None
        self.skip_depth = 0
        self.heading_tag = None
        self.heading_parts = []
        self.current = None

    def handle_starttag(self, tag, attrs):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth += 1
            return
        if tag in {'script', 'style', 'noscript', 'svg'}:
            self.skip_tag, self.skip_depth = tag, 1
        elif re.fullmatch(r'h[1-6]', tag):
            self.heading_tag, self.heading_parts = tag, []
        elif self.current is not None and tag in {'p', 'li', 'br', 'div', 'td', 'tr'}:
            self.current['parts'].append(' ')

    def handle_data(self, data):
        if self.skip_depth:
            return
        if self.heading_tag:
            self.heading_parts.append(data)
        elif self.current is not None:
            self.current['parts'].append(data)

    def handle_endtag(self, tag):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth -= 1
                if not self.skip_depth:
                    self.skip_tag = None
            return
        if tag == self.heading_tag:
            heading = norm(' '.join(self.heading_parts))
            if heading:
                self.current = {'heading': heading, 'parts': []}
                self.sections.append(self.current)
            self.heading_tag, self.heading_parts = None, []


def official_page(url):
    parsed = urlsplit(url)
    return (parsed.scheme == 'https' and parsed.hostname == 'jr.nba.com'
            and parsed.port is None and not parsed.query and not parsed.fragment
            and re.fullmatch(
                r'/basketball-practice-plans/(rookie|starter|all-star|mvp)/(page/\d+/)?',
                parsed.path) is not None)


def fetch(url):
    if not official_page(url):
        raise ValueError(f'Non-official or unexpected source URL: {url}')
    try:
        request = Request(url, headers={
            'User-Agent': 'Mozilla/5.0 (compatible; AcademicSourceAudit/1.0)',
            'Accept': 'text/html',
        })
        with urlopen(request, timeout=25) as response:
            final_url = response.geturl()
            if not official_page(final_url):
                return {'status': 'unexpected_redirect', 'sections': []}
            html = response.read(2_000_000).decode('utf-8', errors='replace')
        parser = PracticeSectionParser()
        parser.feed(html)
        return {'status': 'fetched', 'sections': parser.sections}
    except (HTTPError, URLError, TimeoutError, OSError, ValueError):
        return {'status': 'unavailable', 'sections': []}


CUES = {
    'equipment': {
        'basketball': r'\b(?:basketball|basketballs|ball|balls)\b',
        'hoop': r'\b(?:hoop|hoops|basket|baskets|rim|rims)\b',
        'cones': r'\b(?:cone|cones)\b',
        'markers': r'\b(?:marker|markers)\b',
        'chair': r'\b(?:chair|chairs)\b',
    },
    'participants': {
        'coach': r'\b(?:coach|coaches)\b',
        'partner_or_pair': r'\b(?:partner|partners|pair|pairs)\b',
        'defender': r'\b(?:defender|defenders|defense|defensive)\b',
        'team': r'\b(?:team|teams)\b',
        'group': r'\b(?:group|groups)\b',
        'player': r'\b(?:player|players)\b',
    },
    'space': {
        'full_court': r'\bfull court\b',
        'half_court': r'\bhalf court\b',
        'baseline': r'\b(?:baseline|baselines)\b',
        'sideline': r'\b(?:sideline|sidelines)\b',
        'three_point_line': r'\bthree point (?:line|arc)\b',
        'wing': r'\b(?:wing|wings)\b',
        'key_or_paint': r'\b(?:key|paint|lane)\b',
        'court': r'\b(?:court|courts)\b',
        'gym': r'\b(?:gym|gymnasium)\b',
    },
    'prescription': {
        'minutes': r'\b(?:minutes|minute|mins|min)\b',
        'seconds': r'\b(?:seconds|second|secs|sec)\b',
        'repetitions': r'\b(?:repetitions|repetition|reps|rep|times)\b',
        'makes': r'\b(?:makes|made)\b',
        'games': r'\b(?:game|games)\b',
        'shots': r'\b(?:shot|shots)\b',
    },
}


def labels(text, field):
    return ';'.join(key for key, pattern in CUES[field].items()
                    if re.search(pattern, text))


def time_screen(recorded, text):
    expected = norm(recorded)
    if not text:
        return 'source_section_unavailable'
    if not expected or expected in {'not stated', 'not specified'}:
        return 'recorded_prescription_unspecified'
    if expected in text:
        return 'exact_recorded_phrase_present'
    numbers = set(re.findall(r'\b\d+\b', expected))
    present = set(re.findall(r'\b\d+\b', text))
    if numbers and numbers <= present:
        return 'all_numbers_present_context_unverified'
    if numbers & present:
        return 'some_numbers_present_context_unverified'
    return 'recorded_prescription_not_confirmed'


catalog = read_csv(inputs['catalog'])
registry = read_csv(inputs['registry'])
readiness = read_csv(inputs['readiness'])
cat_by_id = indexed(catalog, 'catalog_candidate_id', 144)
reg_by_id = indexed(registry, 'evidence_id', 156)
ready_by_id = indexed(readiness, 'candidate_group_id', 144)

if set(cat_by_id) != set(ready_by_id):
    raise ValueError('Readiness and catalog candidate IDs differ.')
if any(row['catalog_approval_status'] != 'pending'
       or row['planner_eligible'] != 'no' or row['approved_drill_id'].strip()
       for row in catalog):
    raise ValueError('One or more existing approval decisions have changed.')
if any(row['catalog_candidate_id'] not in cat_by_id for row in registry):
    raise ValueError('Source registry contains an unknown catalog candidate.')

selected = sorted([row for row in readiness if row['review_wave'] == '2'],
                  key=lambda row: int(row['review_rank']))
if len(selected) != 14:
    raise ValueError(f'Wave 2 expected 14 candidate groups; found {len(selected)}.')

sources_by_candidate = defaultdict(list)
for source in registry:
    sources_by_candidate[source['catalog_candidate_id']].append(source)

for row in selected:
    cid = row['candidate_group_id']
    item = cat_by_id[cid]
    sources = sources_by_candidate[cid]
    if row['metadata_stage'] != 'reviewed_metadata_requires_final_signoff':
        raise ValueError(f'Unexpected review stage: {cid}')
    if item['metadata_status'] != 'reviewed_pending_signoff':
        raise ValueError(f'Catalog review stage changed: {cid}')
    if len(sources) != 1 or sources[0]['origin'] != 'original_review_batches':
        raise ValueError(f'Wave 2 requires one historical source occurrence: {cid}')
    if norm(sources[0]['drill_name']) != norm(item['drill_name']):
        raise ValueError(f'Catalog and registry drill name differ: {cid}')
    if row['outstanding_overlap_case_ids'].strip():
        raise ValueError(f'Unexpected unresolved overlap in Wave 2: {cid}')
    if not official_page(sources[0]['source_url']):
        raise ValueError(f'Invalid official source: {cid}')

input_hashes = {name: sha(path) for name, path in inputs.items()}
print('=' * 72)
print('STEP 12-38 — WAVE 2 PUBLISHER METADATA VERIFICATION')
print('=' * 72)
print('Catalog candidates:', len(catalog))
print('Selected metadata-reviewed candidates:', len(selected))

exists = [path.exists() for path in outputs.values()]
if any(exists):
    if not all(exists):
        raise FileExistsError('Partial Step 12-38 outputs exist. Inspect before rerunning.')
    summary = json.loads(outputs['summary'].read_text(encoding='utf-8'))
    if summary.get('input_sha256') != input_hashes:
        raise ValueError('Input files changed after Step 12-38; outputs preserved.')
    for key in ('source_checks', 'field_comparisons', 'review_queue', 'coverage'):
        if sha(outputs[key]) != summary['report_sha256'].get(key):
            raise ValueError(f'Existing report changed: {outputs[key].name}')
    print('EXISTING OUTPUTS VERIFIED — no additional webpage requests.')
    for key, path in outputs.items():
        print(f'  VERIFIED {path.relative_to(root)}')
    print('Approved / planner-eligible: 0 / 0')
    print('OVERALL: PASS — saved publisher-review evidence preserved')
else:
    source_urls = sorted({sources_by_candidate[row['candidate_group_id']][0]['source_url']
                          for row in selected})
    print('Distinct official webpages:', len(source_urls))
    print('\nOFFICIAL PAGE ACCESS')
    pages = {}
    for i, url in enumerate(source_urls, 1):
        pages[url] = fetch(url)
        print(f"  {i:02d}/{len(source_urls):02d} | {pages[url]['status']:<19} | {url}")

    if not any(page['status'] == 'fetched' for page in pages.values()):
        raise RuntimeError('All official publisher pages were unavailable; no reports saved.')

    checks, fields, queue = [], [], []
    for row in selected:
        cid = row['candidate_group_id']
        item = cat_by_id[cid]
        source = sources_by_candidate[cid][0]
        page = pages[source['source_url']]
        expected_heading = norm(source['drill_name'])
        headings = [section['heading'] for section in page['sections']]
        exact = [section for section in page['sections']
                 if section['heading'] == expected_heading]
        suggestion = ''
        status = 'not_checked_source_unavailable'
        content = ''
        if page['status'] == 'fetched':
            if len(exact) == 1:
                status = 'exact_heading_found'
                content = norm(' '.join(exact[0]['parts']))
            elif len(exact) > 1:
                status = 'ambiguous_repeated_heading'
            else:
                ranked = sorted(((SequenceMatcher(None, expected_heading, h).ratio(), h)
                                 for h in set(headings)), reverse=True)
                if ranked and ranked[0][0] >= 0.83:
                    status = 'similar_heading_pending_review'
                    suggestion = ranked[0][1]
                else:
                    status = 'heading_not_found'

        prescription = time_screen(source['source_prescription'], content)
        source_cues = {field: labels(content, field) for field in CUES}
        check = {
            'candidate_group_id': cid,
            'evidence_id': source['evidence_id'],
            'drill_name': item['drill_name'],
            'curriculum_level': source['source_level'],
            'practice_plan': source['practice_plan'],
            'official_source_url': source['source_url'],
            'page_access_status': page['status'],
            'heading_screen': status,
            'similar_heading_suggestion': suggestion,
            'section_word_count': len(content.split()),
            'section_sha256': hashlib.sha256(content.encode('utf-8')).hexdigest() if content else '',
            'source_prescription_screen': prescription,
            'source_equipment_cues': source_cues['equipment'],
            'source_participant_cues': source_cues['participants'],
            'source_space_cues': source_cues['space'],
            'source_prescription_cues': source_cues['prescription'],
            'independent_instruction_signoff': 'pending',
            'independent_metadata_signoff': 'pending',
            'permission_review': 'pending',
            'planner_eligible': 'no',
        }
        checks.append(check)

        issues = []
        if status != 'exact_heading_found':
            issues.append('source_heading_needs_manual_review')
        if not content:
            issues.append('source_instruction_section_unavailable')
        if prescription != 'exact_recorded_phrase_present':
            issues.append('source_prescription_needs_manual_review')
        if source['space'].strip().casefold() in {'', 'see source practice plan', 'unknown', 'not specified'}:
            issues.append('registry_space_requires_source_confirmation')

        for field in ('equipment', 'participants', 'space', 'prescription'):
            registry_field = 'source_prescription' if field == 'prescription' else field
            fields.append({
                'candidate_group_id': cid,
                'evidence_id': source['evidence_id'],
                'drill_name': item['drill_name'],
                'field': field,
                'catalog_value': item['source_prescription'] if field == 'prescription' else item[field],
                'registry_value': source[registry_field],
                'source_section_cues': source_cues[field],
                'source_text_screen': (prescription if field == 'prescription'
                                       else 'cues_present_not_approval' if source_cues[field]
                                       else 'no_cues_detected_or_section_unavailable'),
                'independent_field_signoff': 'pending',
                'reviewer': '',
                'review_notes': '',
            })
            if field != 'prescription' and not source_cues[field]:
                issues.append(f'{field}_requires_source_review')

        if item['source_prescription'].strip() != source['source_prescription'].strip():
            issues.append('catalog_registry_prescription_wording_differs')
        if item['space'].strip() != source['space'].strip():
            issues.append('catalog_registry_space_wording_differs')

        queue.append({
            'review_rank': row['review_rank'],
            'candidate_group_id': cid,
            'evidence_id': source['evidence_id'],
            'drill_name': item['drill_name'],
            'review_wave': '2',
            'heading_screen': status,
            'prescription_screen': prescription,
            'registry_space_unspecified': 'yes' if 'registry_space_requires_source_confirmation' in issues else 'no',
            'review_issues': ';'.join(dict.fromkeys(issues)),
            'next_action': 'Compare the official drill section with all four recorded requirement fields; document reviewer decision and copyright permissions.',
            'instructions_review': 'pending',
            'requirements_review': 'pending',
            'distinctness_review': 'pending',
            'permission_review': 'pending',
            'youth_planning_review': 'pending',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
            'reviewer': '',
            'review_notes': '',
        })

    coverage = []
    for key in sorted({(r['curriculum_level'], r['practice_plan']) for r in checks}):
        group = [r for r in checks if (r['curriculum_level'], r['practice_plan']) == key]
        coverage.append({
            'curriculum_level': key[0],
            'practice_plan': key[1],
            'source_occurrences': len(group),
            'heading_exact_count': sum(r['heading_screen'] == 'exact_heading_found' for r in group),
            'sections_available': sum(bool(r['section_sha256']) for r in group),
            'recorded_prescription_exact_text_count': sum(r['source_prescription_screen'] == 'exact_recorded_phrase_present' for r in group),
            'instructional_signoffs': 0,
            'approved_drills': 0,
        })
    coverage.append({
        'curriculum_level': 'ALL',
        'practice_plan': 'ALL',
        'source_occurrences': len(checks),
        'heading_exact_count': sum(r['heading_screen'] == 'exact_heading_found' for r in checks),
        'sections_available': sum(bool(r['section_sha256']) for r in checks),
        'recorded_prescription_exact_text_count': sum(r['source_prescription_screen'] == 'exact_recorded_phrase_present' for r in checks),
        'instructional_signoffs': 0,
        'approved_drills': 0,
    })

    assert len(checks) == 14 and len(fields) == 56 and len(queue) == 14
    if {r['candidate_group_id'] for r in checks} != {r['candidate_group_id'] for r in selected}:
        raise ValueError('Some selected candidates were not represented.')
    if any(r['catalog_approval_status'] != 'pending' or r['planner_eligible'] != 'no' for r in queue):
        raise ValueError('Unexpected approval change in review queue.')

    reports = {'source_checks': checks, 'field_comparisons': fields,
               'review_queue': queue, 'coverage': coverage}
    payloads = {key: csv_payload(rows) for key, rows in reports.items()}
    report_hashes = {key: hashlib.sha256(data).hexdigest() for key, data in payloads.items()}
    summary = {
        'step': '12-38',
        'catalog_version': '0.3.2',
        'screen_execution_date': date.today().isoformat(),
        'catalog_candidates_preserved': len(catalog),
        'registry_source_occurrences_preserved': len(registry),
        'selected_candidate_groups': len(checks),
        'comparison_fields': len(fields),
        'source_pages_attempted': len(source_urls),
        'source_pages_fetched': sum(p['status'] == 'fetched' for p in pages.values()),
        'exact_heading_matches': sum(r['heading_screen'] == 'exact_heading_found' for r in checks),
        'source_sections_available': sum(bool(r['section_sha256']) for r in checks),
        'exact_prescription_phrase_matches': sum(r['source_prescription_screen'] == 'exact_recorded_phrase_present' for r in checks),
        'registry_space_unspecified': sum(r['registry_space_unspecified'] == 'yes' for r in queue),
        'catalog_approval_changes': 0,
        'planner_eligibility_changes': 0,
        'input_sha256': input_hashes,
        'report_sha256': report_hashes,
        'limitations': 'HTML heading matches, numbers and keyword cues are preliminary evidence; each instruction and requirement needs source-level reviewer signoff. Publisher permissions and youth planning checks remain pending.',
    }
    payloads['summary'] = (json.dumps(summary, indent=2, sort_keys=True) + '\n').encode('utf-8')

    if any(path.exists() for path in outputs.values()):
        raise FileExistsError('A Step 12-38 output appeared during processing. No files changed.')
    for path in outputs.values():
        path.parent.mkdir(parents=True, exist_ok=True)
    with tempfile.TemporaryDirectory(prefix='step12_38_', dir=root / 'results') as folder:
        staged = {}
        for key, path in outputs.items():
            temporary = Path(folder) / path.name
            temporary.write_bytes(payloads[key])
            if hashlib.sha256(temporary.read_bytes()).hexdigest() != hashlib.sha256(payloads[key]).hexdigest():
                raise RuntimeError(f'Failed to verify staged output: {key}')
            staged[key] = temporary
        for key, path in outputs.items():
            # Exclusive creation prevents silently replacing a previous report.
            with path.open('xb') as target:
                target.write(staged[key].read_bytes())

    if {name: sha(path) for name, path in inputs.items()} != input_hashes:
        raise RuntimeError('An input file changed during execution. Inspect source history.')
    for key in reports:
        if sha(outputs[key]) != report_hashes[key]:
            raise RuntimeError(f'Output hash verification failed: {key}')

    print('\nSOURCE SCREENING RESULTS')
    for status, count in Counter(r['heading_screen'] for r in checks).items():
        print(f'  {status:<42} {count}')
    for status, count in Counter(r['source_prescription_screen'] for r in checks).items():
        print(f'  {status:<42} {count}')
    print('  Registry space fields unspecified:', summary['registry_space_unspecified'])
    print('\nREVIEW QUEUE')
    for row in queue:
        print(f"  {row['candidate_group_id']} | {row['drill_name']} | "
              f"{row['heading_screen']} | {row['prescription_screen']}")
    print('\nOUTPUTS')
    for path in outputs.values():
        print('  CREATED ', path.relative_to(root))
    print('\nCatalog, source registry, and approval decisions modified: No')
    print('Git operations performed: None')
    print('OVERALL: PASS — publisher evidence recorded; reviewer signoff pending')
```

```python
# Step 12-39 - Wave 2 Source-Space Reconciliation
# Verify source-specific space requirements; retain pending review and approvals.

import csv
import hashlib
import io
import json
import re
import tempfile
import unicodedata
from collections import Counter
from datetime import date
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

root = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if root is None:
    raise FileNotFoundError('Project root not found.')

inputs = {
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_2_REVIEW_ONLY.csv',
    'registry': root / 'results/basketball_drill_source_registry_v0_3_2.csv',
    'readiness': root / 'results/basketball_catalog_readiness_worklist_v0_3_1.csv',
    'wave2_checks': root / 'results/basketball_wave2_publisher_checks_v0_3_2.csv',
}
outputs = {
    'evidence': root / 'results/basketball_wave2_space_source_evidence_v0_3_3.csv',
    'revisions': root / 'results/basketball_wave2_space_revision_log_v0_3_3.csv',
    'registry': root / 'results/basketball_drill_source_registry_v0_3_3.csv',
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_3_REVIEW_ONLY.csv',
    'summary': root / 'results/basketball_step12_39_summary_v0_3_3.json',
}

# Each source-location conclusion is reviewed separately from planner eligibility.
# The sixth activity describes roles but does not state a specific court layout.
SOURCE_SPACE = {
    'JR-EVID-0056': {
        'space': 'Half court, with sideline and baseline markings',
        'anchors': ('half court', 'sideline', 'baseline'),
        'type': 'explicit_location',
        'rationale': 'The published activity places three participants near half court and the sideline, with movement toward the baseline.',
    },
    'JR-EVID-0058': {
        'space': 'Basket and adjacent painted area',
        'anchors': ('basket', 'paint'),
        'type': 'explicit_location',
        'rationale': 'The published activity locates players at a basket and around the paint.',
    },
    'JR-EVID-0062': {
        'space': 'Basketball court with basket, baseline, and top-of-key markings',
        'anchors': ('baseline', 'top of the key', 'basket'),
        'type': 'explicit_location',
        'rationale': 'The published game references a baseline drive, top of the key, and use of baskets.',
    },
    'JR-EVID-0060': {
        'space': 'Basketball court with baseline, lane line, and wing area',
        'anchors': ('baseline', 'lane line', 'wing'),
        'type': 'explicit_location',
        'rationale': 'The published instruction specifies baseline, lane-line, and wing positions.',
    },
    'JR-EVID-0061': {
        'space': 'Basketball court with hoop and three-point-line wing positions',
        'anchors': ('three point line', 'wing', 'basket'),
        'type': 'explicit_location',
        'rationale': 'The published activity places players around the three-point line and wings and includes an attack toward the basket.',
    },
    'JR-EVID-0059': {
        'space': 'Hoop area with space for shooter, passer, and rebounder',
        'anchors': ('shoot', 'pass', 'rebound'),
        'type': 'inferred_location',
        'rationale': 'The source confirms three shooting roles but does not specify a formal court area; space description remains a proposal.',
    },
}


def load_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as f:
        reader = csv.DictReader(f)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or len(columns) != len(set(columns)) or any(
            None in row or any(v is None for v in row.values()) for row in rows):
        raise ValueError(f'Invalid CSV content: {path.name}')
    return rows, columns


def unique_index(rows, key, expected):
    if len(rows) != expected:
        raise ValueError(f'Unexpected {key} count: {len(rows)}; expected {expected}')
    values = [r[key].strip() for r in rows]
    if not all(values) or len(set(values)) != expected:
        raise ValueError(f'Nonunique or blank {key}')
    return dict(zip(values, rows))


def sha(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def norm(value):
    return re.sub(r'[^a-z0-9]+', ' ',
                  unicodedata.normalize('NFKC', value or '').casefold()).strip()


def csv_bytes(rows, columns):
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=columns, lineterminator='\n',
                            extrasaction='raise')
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')


class DrillSections(HTMLParser):
    """Parse heading-scoped text in memory; do not save publisher instructions."""
    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.skip_tag = None
        self.skip_depth = 0
        self.header = None
        self.header_parts = []
        self.current = None

    def handle_starttag(self, tag, attrs):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth += 1
            return
        if tag in {'script', 'style', 'noscript', 'svg'}:
            self.skip_tag, self.skip_depth = tag, 1
        elif re.fullmatch(r'h[1-6]', tag):
            self.header, self.header_parts = tag, []
        elif self.current is not None and tag in {'p', 'li', 'div', 'br', 'td', 'tr'}:
            self.current['parts'].append(' ')

    def handle_data(self, data):
        if self.skip_depth:
            return
        if self.header:
            self.header_parts.append(data)
        elif self.current is not None:
            self.current['parts'].append(data)

    def handle_endtag(self, tag):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth -= 1
                if not self.skip_depth:
                    self.skip_tag = None
            return
        if tag == self.header:
            header = norm(' '.join(self.header_parts))
            if header:
                self.current = {'heading': header, 'parts': []}
                self.sections.append(self.current)
            self.header, self.header_parts = None, []


def fetch_official_sections(url):
    parsed = urlsplit(url)
    if (parsed.scheme != 'https' or parsed.hostname != 'jr.nba.com'
            or parsed.query or parsed.fragment or parsed.port is not None
            or not re.fullmatch(r'/basketball-practice-plans/all-star/page/3/',
                                parsed.path)):
        raise ValueError(f'Unexpected publisher URL: {url}')
    request = Request(url, headers={'User-Agent': 'Mozilla/5.0',
                                    'Accept': 'text/html'})
    try:
        with urlopen(request, timeout=25) as response:
            final = urlsplit(response.geturl())
            if final.hostname != 'jr.nba.com' or final.path != parsed.path:
                raise ValueError('Publisher page redirected to a different location.')
            text = response.read(2_000_000).decode('utf-8', errors='replace')
        parser = DrillSections()
        parser.feed(text)
        return parser.sections
    except (HTTPError, URLError, TimeoutError, OSError) as error:
        raise RuntimeError('Official publisher page unavailable; no files were changed.') from error


print('=' * 72)
print('STEP 12-39 — WAVE 2 SOURCE SPACE RECONCILIATION')
print('=' * 72)

present = {key: path.exists() for key, path in outputs.items()}
if any(present.values()):
    if not all(present.values()):
        raise FileExistsError('Partial Step 12-39 outputs found. Inspect before rerunning.')
    summary = json.loads(outputs['summary'].read_text(encoding='utf-8'))
    for key, path in outputs.items():
        if key != 'summary' and sha(path) != summary['output_sha256'][key]:
            raise RuntimeError(f'Existing output checksum mismatch: {key}')
    print('Existing Step 12-39 outputs verified and preserved.')
    print('Source-supported space revisions:', summary['explicit_source_space_updates'])
    print('Space proposals still pending:', summary['inferred_space_proposals'])
    print('Catalog candidates / source records: 144 / 156')
    print('Approved / planner-eligible: 0 / 0')
    print('OVERALL: PASS — existing outputs preserved')
else:
    original_sha = {key: sha(path) for key, path in inputs.items()}
    catalog, cat_cols = load_csv(inputs['catalog'])
    registry, reg_cols = load_csv(inputs['registry'])
    ready, _ = load_csv(inputs['readiness'])
    wave2, _ = load_csv(inputs['wave2_checks'])
    cat_idx = unique_index(catalog, 'catalog_candidate_id', 144)
    reg_idx = unique_index(registry, 'evidence_id', 156)
    ready_idx = unique_index(ready, 'candidate_group_id', 144)
    wave_idx = unique_index(wave2, 'evidence_id', 14)

    if set(cat_idx) != set(ready_idx):
        raise ValueError('Catalog and readiness IDs differ.')
    if any(r['catalog_approval_status'] != 'pending'
           or r['planner_eligible'] != 'no'
           or r['approved_drill_id'].strip() for r in catalog):
        raise ValueError('Catalog contains previously approved or eligible records.')
    if any(r['catalog_candidate_id'] not in cat_idx for r in registry):
        raise ValueError('Unknown source registry candidate ID.')
    if len(SOURCE_SPACE) != 6 or not set(SOURCE_SPACE).issubset(wave_idx):
        raise ValueError('Unexpected Wave 2 source-space review selection.')

    for eid in SOURCE_SPACE:
        r = reg_idx[eid]
        c = cat_idx[r['catalog_candidate_id']]
        w = ready_idx[r['catalog_candidate_id']]
        a = wave_idx[eid]
        if (r['origin'] != 'original_review_batches'
                or r['source_level'] != 'All-Star'
                or r['practice_plan'] != '3'
                or r['source_url'] != 'https://jr.nba.com/basketball-practice-plans/all-star/page/3/'
                or r['space'].strip().casefold() not in {'', 'see source practice plan', 'not specified', 'unknown'}
                or w['review_wave'] != '2'
                or c['metadata_status'] != 'reviewed_pending_signoff'
                or a['heading_screen'] != 'exact_heading_found'
                or a['candidate_group_id'] != r['catalog_candidate_id']):
            raise ValueError(f'Source identity, review status, or prior audit changed: {eid}')

    official_url = 'https://jr.nba.com/basketball-practice-plans/all-star/page/3/'
    sections = fetch_official_sections(official_url)
    print('Official source page: fetched')
    by_heading = {}
    for section in sections:
        by_heading.setdefault(section['heading'], []).append(
            norm(' '.join(section['parts'])))

    evidence = []
    revisions = []
    explicit_updates = 0
    inference_pending = 0
    extra_reg_cols = ['source_space_review_status', 'source_space_review_reference',
                      'source_space_proposal', 'source_space_evidence_note']
    extra_cat_cols = ['wave2_space_review_status', 'wave2_space_review_evidence_id',
                      'wave2_source_space_proposal']
    if set(extra_reg_cols) & set(reg_cols) or set(extra_cat_cols) & set(cat_cols):
        raise ValueError('The v0.3.3 metadata columns already exist in input.')
    for record in registry:
        for field in extra_reg_cols:
            record[field] = ''
    for record in catalog:
        for field in extra_cat_cols:
            record[field] = ''
        record['catalog_version'] = '0.3.3'

    for eid, info in SOURCE_SPACE.items():
        record = reg_idx[eid]
        previous_registry_space = record['space']
        candidate = cat_idx[record['catalog_candidate_id']]
        matching = by_heading.get(norm(record['drill_name']), [])
        sections_present = len(matching)
        passage = max(matching, key=len) if matching else ''
        anchors_found = [anchor for anchor in info['anchors'] if anchor in passage]
        verified = bool(passage) and len(anchors_found) == len(info['anchors'])
        if not verified:
            status = 'publisher_section_or_anchor_review_required'
            action = 'no_revision'
            proposed = ''
        elif info['type'] == 'inferred_location':
            status = 'inferred_space_needs_reviewer_confirmation'
            action = 'proposal_only'
            proposed = info['space']
            inference_pending += 1
        else:
            status = 'source_location_evidence_confirmed_pending_team_signoff'
            action = 'reconcile_source_space_in_review_registry'
            proposed = info['space']
            record['space'] = proposed
            explicit_updates += 1

        record['source_space_review_status'] = status
        record['source_space_review_reference'] = ('results/basketball_wave2_space_source_evidence_v0_3_3.csv')
        record['source_space_proposal'] = proposed
        record['source_space_evidence_note'] = info['rationale'] if verified else 'Source heading or expected location cues require investigation.'
        candidate['wave2_space_review_status'] = status
        candidate['wave2_space_review_evidence_id'] = eid
        candidate['wave2_source_space_proposal'] = proposed

        evidence.append({
            'evidence_id': eid,
            'candidate_group_id': record['catalog_candidate_id'],
            'drill_name': record['drill_name'],
            'source_url': record['source_url'],
            'heading_matches': sections_present,
            'section_available': 'yes' if passage else 'no',
            'location_fact_anchors': ';'.join(info['anchors']),
            'location_fact_anchors_found': ';'.join(anchors_found),
            'source_location_evidence_status': status,
            'reviewer': '',
            'reviewer_signoff': 'pending',
            'permission_review': 'pending',
            'planner_eligible': 'no',
        })
        revisions.append({
            'evidence_id': eid,
            'candidate_group_id': record['catalog_candidate_id'],
            'drill_name': record['drill_name'],
            'previous_registry_space': previous_registry_space,
            'catalog_space_before_reconciliation': candidate['space'],
            'proposed_or_reconciled_source_space': proposed,
            'action': action,
            'evidence_status': status,
            'source_rationale': info['rationale'] if verified else 'Source evidence unavailable or incomplete.',
            'team_signoff': 'pending',
            'planner_eligible': 'no',
        })

    if any(r['catalog_approval_status'] != 'pending' or r['planner_eligible'] != 'no'
           or r['approved_drill_id'].strip() for r in catalog):
        raise ValueError('Approval flags changed unexpectedly.')
    if len(catalog) != 144 or len(registry) != 156:
        raise ValueError('Record counts changed during reconciliation.')
    if not (len(evidence) == len(revisions) == 6):
        raise ValueError('Expected six review records.')

    reports = {
        'evidence': csv_bytes(evidence, list(evidence[0])),
        'revisions': csv_bytes(revisions, list(revisions[0])),
        'registry': csv_bytes(registry, reg_cols + extra_reg_cols),
        'catalog': csv_bytes(catalog, cat_cols + extra_cat_cols),
    }
    summary = {
        'step': '12-39',
        'catalog_version': '0.3.3',
        'analysis_date': date.today().isoformat(),
        'catalog_candidates': 144,
        'registry_source_occurrences': 156,
        'wave2_review_records': 14,
        'targeted_source_space_records': len(evidence),
        'explicit_source_space_updates': explicit_updates,
        'inferred_space_proposals': inference_pending,
        'source_heading_or_fact_review_remaining': len(evidence) - explicit_updates - inference_pending,
        'approved_drills': 0,
        'planner_eligible_drills': 0,
        'input_sha256': original_sha,
        'output_sha256': {key: hashlib.sha256(value).hexdigest() for key, value in reports.items()},
        'limitations': 'Section headings and location cues are evidence for metadata preparation only. Editorial signoff, usage permissions, drill distinctness, and youth suitability remain pending.',
    }
    reports['summary'] = (json.dumps(summary, sort_keys=True, indent=2) + '\n').encode('utf-8')

    if any(path.exists() for path in outputs.values()):
        raise FileExistsError('An output appeared during processing; no existing files replaced.')
    for path in outputs.values():
        path.parent.mkdir(parents=True, exist_ok=True)
    with tempfile.TemporaryDirectory(prefix='step12_39_', dir=root / 'results') as tmp:
        for key, path in outputs.items():
            staged = Path(tmp) / path.name
            staged.write_bytes(reports[key])
            if hashlib.sha256(staged.read_bytes()).hexdigest() != hashlib.sha256(reports[key]).hexdigest():
                raise RuntimeError(f'Staging verification failed: {key}')
        for key, path in outputs.items():
            with path.open('xb') as destination:
                destination.write((Path(tmp) / path.name).read_bytes())
    if {key: sha(path) for key, path in inputs.items()} != original_sha:
        raise RuntimeError('An input changed during execution.')
    for key, path in outputs.items():
        if sha(path) != hashlib.sha256(reports[key]).hexdigest():
            raise RuntimeError(f'Output verification failed: {key}')

    print('\nSOURCE-SPACE REVIEW')
    for record in evidence:
        print(f"  {record['evidence_id']} | {record['drill_name']} | {record['source_location_evidence_status']}")
    print('\nSUMMARY')
    print('Explicit source-space revisions:', explicit_updates)
    print('Inferred space proposals requiring review:', inference_pending)
    print('Records requiring another source check:', summary['source_heading_or_fact_review_remaining'])
    print('Catalog candidates / source occurrences: 144 / 156')
    print('Approved / planner-eligible: 0 / 0')
    print('\nOUTPUTS')
    for path in outputs.values():
        print('  CREATED', path.relative_to(root))
    print('Source input files and earlier catalogs overwritten: No')
    print('Git operations performed: None')
    print('OVERALL: PASS — source-space evidence reconciled; signoff pending')
```

```python
# Step 12-40 - Wave 2 Evidence Consolidation and Signoff Packet
# Consolidate prior publisher checks and space reviews without changing approval states.

import csv
import hashlib
import io
import json
import tempfile
from collections import Counter, defaultdict
from datetime import date
from pathlib import Path

root = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if root is None:
    raise FileNotFoundError('Project Git root not found.')

inputs = {
    'catalog_previous': root / 'data/processed/basketball_drill_catalog_v0_3_2_REVIEW_ONLY.csv',
    'catalog_current': root / 'data/processed/basketball_drill_catalog_v0_3_3_REVIEW_ONLY.csv',
    'registry_previous': root / 'results/basketball_drill_source_registry_v0_3_2.csv',
    'registry_current': root / 'results/basketball_drill_source_registry_v0_3_3.csv',
    'readiness': root / 'results/basketball_catalog_readiness_worklist_v0_3_1.csv',
    'publisher_checks': root / 'results/basketball_wave2_publisher_checks_v0_3_2.csv',
    'publisher_fields': root / 'results/basketball_wave2_field_comparisons_v0_3_2.csv',
    'space_evidence': root / 'results/basketball_wave2_space_source_evidence_v0_3_3.csv',
    'space_revisions': root / 'results/basketball_wave2_space_revision_log_v0_3_3.csv',
}
outputs = {
    'matrix': root / 'results/basketball_wave2_evidence_matrix_v0_3_3.csv',
    'fields': root / 'results/basketball_wave2_field_signoff_v0_3_3.csv',
    'worklist': root / 'results/basketball_wave2_signoff_worklist_v0_3_3.csv',
    'coverage': root / 'results/eda/basketball_wave2_review_coverage_v0_3_3.csv',
    'summary': root / 'results/basketball_step12_40_summary_v0_3_3.json',
}


def load_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or len(columns) != len(set(columns)):
        raise ValueError(f'Invalid CSV column names: {path.name}')
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f'Malformed CSV rows in {path.name}')
    return rows, columns


def unique_index(rows, field, n):
    if len(rows) != n:
        raise ValueError(f'Unexpected {field} record count: {len(rows)}; expected {n}')
    values = [r[field].strip() for r in rows]
    if not all(values) or len(values) != len(set(values)):
        raise ValueError(f'Missing or duplicate identifiers: {field}')
    return {r[field]: r for r in rows}


def checksum(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def csv_data(rows):
    if not rows:
        raise ValueError('Cannot produce an empty output table.')
    buf = io.StringIO(newline='')
    writer = csv.DictWriter(buf, fieldnames=list(rows[0]), lineterminator='\n',
                            extrasaction='raise')
    writer.writeheader()
    writer.writerows(rows)
    return buf.getvalue().encode('utf-8')


def pending_approval(row, label):
    if (row['catalog_approval_status'] != 'pending'
            or row['planner_eligible'] != 'no'
            or row.get('approved_drill_id', '').strip()):
        raise ValueError(f'Unexpected existing approval status: {label}')


print('=' * 72)
print('STEP 12-40 — WAVE 2 EVIDENCE CONSOLIDATION AND SIGNOFF')
print('=' * 72)

original_hashes = {key: checksum(path) for key, path in inputs.items()}
existing = [path.exists() for path in outputs.values()]
if any(existing):
    if not all(existing):
        raise FileExistsError('Some Step 12-40 outputs exist; inspect before rerunning.')
    previous_summary = json.loads(outputs['summary'].read_text(encoding='utf-8'))
    if previous_summary.get('input_sha256') != original_hashes:
        raise RuntimeError('Inputs changed after Step 12-40; existing reports preserved.')
    for name, path in outputs.items():
        if name != 'summary' and checksum(path) != previous_summary['output_sha256'][name]:
            raise RuntimeError(f'Existing report changed: {path.name}')
    print('Previous output set verified; files preserved without modification.')
    print('Wave 2 candidate groups:', previous_summary['wave2_candidates'])
    print('Confirmed source-location evidence:', previous_summary['space_evidence_explicit'])
    print('Inferred space proposals:', previous_summary['space_proposals_inferred'])
    print('Approved / planner-eligible: 0 / 0')
    print('OVERALL: PASS — verified existing reports preserved')
else:
    datasets = {key: load_csv(path)[0] for key, path in inputs.items()}
    old_cat = unique_index(datasets['catalog_previous'], 'catalog_candidate_id', 144)
    cat = unique_index(datasets['catalog_current'], 'catalog_candidate_id', 144)
    old_reg = unique_index(datasets['registry_previous'], 'evidence_id', 156)
    reg = unique_index(datasets['registry_current'], 'evidence_id', 156)
    ready = unique_index(datasets['readiness'], 'candidate_group_id', 144)
    publisher = unique_index(datasets['publisher_checks'], 'evidence_id', 14)
    space = unique_index(datasets['space_evidence'], 'evidence_id', 6)
    revisions = unique_index(datasets['space_revisions'], 'evidence_id', 6)
    if set(old_cat) != set(cat) or set(old_reg) != set(reg) or set(cat) != set(ready):
        raise ValueError('Input catalog, registry, or readiness identifiers changed.')
    if set(space) != set(revisions):
        raise ValueError('Space evidence and revision records do not agree.')

    for cid, current in cat.items():
        pending_approval(current, cid)
        pending_approval(old_cat[cid], cid)
        if current['catalog_version'] != '0.3.3':
            raise ValueError(f'Catalog version not updated: {cid}')
        for field, value in old_cat[cid].items():
            if field != 'catalog_version' and current.get(field) != value:
                raise ValueError(f'Unexpected existing catalog field change: {cid}.{field}')

    source_space_changes = []
    for eid, current in reg.items():
        old = old_reg[eid]
        if current['catalog_candidate_id'] not in cat:
            raise ValueError(f'Unknown candidate in source registry: {eid}')
        for field, value in old.items():
            if current.get(field) != value:
                if field != 'space' or eid not in space:
                    raise ValueError(f'Unexpected source field change: {eid}.{field}')
                if space[eid]['source_location_evidence_status'] != 'source_location_evidence_confirmed_pending_team_signoff':
                    raise ValueError(f'Unconfirmed source-space revision: {eid}')
                if revisions[eid]['action'] != 'reconcile_source_space_in_review_registry':
                    raise ValueError(f'Unexpected source-space revision action: {eid}')
                if current[field] != revisions[eid]['proposed_or_reconciled_source_space']:
                    raise ValueError(f'New registry space disagrees with review: {eid}')
                source_space_changes.append(eid)
    if len(source_space_changes) != 5:
        raise ValueError(f'Expected five explicit source-space changes, got {len(source_space_changes)}.')
    if Counter(x['source_location_evidence_status'] for x in space.values()) != {
        'source_location_evidence_confirmed_pending_team_signoff': 5,
        'inferred_space_needs_reviewer_confirmation': 1,
    }:
        raise ValueError('Unexpected space-review status distribution.')

    selected = sorted((r for r in ready.values() if r['review_wave'] == '2'),
                      key=lambda r: int(r['review_rank']))
    if len(selected) != 14:
        raise ValueError('Wave 2 must contain 14 selected candidate groups.')
    reg_by_cid = defaultdict(list)
    for record in reg.values():
        reg_by_cid[record['catalog_candidate_id']].append(record)

    field_lookup = {}
    for comparison in datasets['publisher_fields']:
        key = (comparison['evidence_id'], comparison['field'])
        if key in field_lookup:
            raise ValueError(f'Duplicate Wave 2 field comparison: {key}')
        field_lookup[key] = comparison
    expected_keys = {(eid, field) for eid in publisher for field in (
        'equipment', 'participants', 'space', 'prescription')}
    if set(field_lookup) != expected_keys:
        raise ValueError('Publisher field comparisons must cover four fields for all 14 source records.')

    matrix, field_signoffs, worklist = [], [], []
    for selected_row in selected:
        cid = selected_row['candidate_group_id']
        record = cat[cid]
        sources = reg_by_cid[cid]
        if len(sources) != 1 or selected_row['outstanding_overlap_case_ids']:
            raise ValueError(f'Wave 2 candidate unexpectedly has multiple sources or an overlap case: {cid}')
        source = sources[0]
        eid = source['evidence_id']
        if eid not in publisher or selected_row['metadata_stage'] != 'reviewed_metadata_requires_final_signoff':
            raise ValueError(f'Wave 2 source or stage does not match prior audit: {cid}')
        check = publisher[eid]
        if (check['candidate_group_id'] != cid
                or check['heading_screen'] != 'exact_heading_found'
                or check['source_prescription_screen'] != 'exact_recorded_phrase_present'
                or check['official_source_url'] != source['source_url']):
            raise ValueError(f'Wave 2 publisher evidence mismatch: {cid}')
        if record['metadata_status'] != 'reviewed_pending_signoff':
            raise ValueError(f'Unexpected catalog metadata stage: {cid}')

        if eid in space:
            space_status = space[eid]['source_location_evidence_status']
            if space[eid]['candidate_group_id'] != cid:
                raise ValueError(f'Source-space candidate mismatch: {eid}')
        else:
            space_status = 'not_targeted_in_step12_39'

        if space_status == 'inferred_space_needs_reviewer_confirmation':
            priority, focus = 'high', 'Confirm shooting-space requirement against the source or retain it as an explicit inference.'
        elif space_status == 'source_location_evidence_confirmed_pending_team_signoff':
            priority, focus = 'medium', 'Review the source-supported space description and sign off its intended interpretation.'
        else:
            priority, focus = 'standard', 'Check equipment, participant, and space semantics against the original instructions.'

        cues_by_field = {
            'equipment': check['source_equipment_cues'],
            'participants': check['source_participant_cues'],
            'space': check['source_space_cues'],
            'prescription': check['source_prescription_cues'],
        }
        for field in ('equipment', 'participants', 'space', 'prescription'):
            compare = field_lookup[(eid, field)]
            value = source['source_prescription'] if field == 'prescription' else source[field]
            original_value = old_reg[eid]['source_prescription'] if field == 'prescription' else old_reg[eid][field]
            if compare['registry_value'].strip() != original_value.strip():
                raise ValueError(f'Step 12-38 field evidence does not match historical source registry: {eid}.{field}')
            if field == 'prescription':
                evidence_level = 'recorded_phrase_found_in_source_section'
                review_action = 'Validate units, work/rest meaning, and applicability.'
            elif field == 'space' and space_status == 'source_location_evidence_confirmed_pending_team_signoff':
                evidence_level = 'explicit_location_cues_reviewed_pending_signoff'
                review_action = 'Confirm source-grounded interpretation and the finalized space requirement.'
            elif field == 'space' and space_status == 'inferred_space_needs_reviewer_confirmation':
                evidence_level = 'inferred_space_not_publisher_requirement'
                review_action = 'Confirm practical minimum space independently; do not label as a publisher requirement.'
            elif cues_by_field[field]:
                evidence_level = 'publisher_keywords_found_not_requirements_verified'
                review_action = f'Read the official section and verify {field} requirements and exceptions.'
            else:
                evidence_level = 'no_publisher_keyword_cue'
                review_action = f'Check the source manually for {field} requirements.'
            field_signoffs.append({
                'candidate_group_id': cid,
                'evidence_id': eid,
                'drill_name': record['drill_name'],
                'field': field,
                'catalog_value': record['source_prescription'] if field == 'prescription' else record[field],
                'previous_registry_value': original_value,
                'current_registry_value': value,
                'publisher_text_screen': compare['source_text_screen'],
                'publisher_keyword_cues': cues_by_field[field],
                'evidence_level': evidence_level,
                'review_action': review_action,
                'reviewer_decision': 'pending',
                'verified_value': '',
                'reviewer': '',
                'review_date': '',
                'review_notes': '',
            })

        matrix.append({
            'review_rank': selected_row['review_rank'],
            'candidate_group_id': cid,
            'evidence_id': eid,
            'drill_name': record['drill_name'],
            'curriculum_level': source['source_level'],
            'practice_plan': source['practice_plan'],
            'official_source_url': source['source_url'],
            'source_heading_screen': check['heading_screen'],
            'source_prescription_screen': check['source_prescription_screen'],
            'source_section_words': check['section_word_count'],
            'source_equipment_cues': cues_by_field['equipment'],
            'source_participant_cues': cues_by_field['participants'],
            'source_space_cues': cues_by_field['space'],
            'current_equipment': source['equipment'],
            'current_participants': source['participants'],
            'current_space': source['space'],
            'current_prescription': source['source_prescription'],
            'space_source_evidence_status': space_status,
            'source_space_changed_in_v033': 'yes' if eid in source_space_changes else 'no',
            'instruction_signoff': 'pending',
            'requirements_signoff': 'pending',
            'distinctness_signoff': 'pending',
            'permission_review': 'pending',
            'youth_rule_review': 'pending',
            'catalog_approval_status': record['catalog_approval_status'],
            'planner_eligible': record['planner_eligible'],
        })
        worklist.append({
            'priority': priority,
            'review_rank': selected_row['review_rank'],
            'candidate_group_id': cid,
            'evidence_id': eid,
            'drill_name': record['drill_name'],
            'official_source_url': source['source_url'],
            'space_evidence_status': space_status,
            'next_review_action': focus,
            'instructional_content_signoff': 'pending',
            'equipment_signoff': 'pending',
            'participant_signoff': 'pending',
            'space_signoff': 'pending',
            'prescription_signoff': 'pending',
            'canonical_identity_signoff': 'pending',
            'copyright_permission_review': 'pending',
            'youth_planning_review': 'pending',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
            'reviewer': '',
            'review_date': '',
            'decision_notes': '',
        })

    if not (len(matrix) == 14 and len(worklist) == 14 and len(field_signoffs) == 56):
        raise ValueError('Wave 2 packet has incomplete records.')
    priority_order = {'high': 0, 'medium': 1, 'standard': 2}
    worklist.sort(key=lambda r: (priority_order[r['priority']], int(r['review_rank'])))
    coverage = []
    combinations = sorted({(r['curriculum_level'], r['practice_plan']) for r in matrix})
    for level, plan in combinations:
        rows = [r for r in matrix if (r['curriculum_level'], r['practice_plan']) == (level, plan)]
        coverage.append({
            'curriculum_level': level,
            'practice_plan': plan,
            'candidates': len(rows),
            'publisher_heading_matches': sum(r['source_heading_screen'] == 'exact_heading_found' for r in rows),
            'publisher_prescription_phrase_matches': sum(r['source_prescription_screen'] == 'exact_recorded_phrase_present' for r in rows),
            'explicit_space_evidence_pending_signoff': sum(r['space_source_evidence_status'] == 'source_location_evidence_confirmed_pending_team_signoff' for r in rows),
            'inferred_space_requires_review': sum(r['space_source_evidence_status'] == 'inferred_space_needs_reviewer_confirmation' for r in rows),
            'approved': 0,
        })
    coverage.append({
        'curriculum_level': 'ALL', 'practice_plan': 'ALL', 'candidates': 14,
        'publisher_heading_matches': 14, 'publisher_prescription_phrase_matches': 14,
        'explicit_space_evidence_pending_signoff': 5,
        'inferred_space_requires_review': 1, 'approved': 0,
    })

    reports = {
        'matrix': csv_data(matrix),
        'fields': csv_data(field_signoffs),
        'worklist': csv_data(worklist),
        'coverage': csv_data(coverage),
    }
    summary = {
        'step': '12-40',
        'analysis_date': date.today().isoformat(),
        'catalog_version': '0.3.3',
        'catalog_candidate_groups': 144,
        'source_occurrences': 156,
        'wave2_candidates': 14,
        'wave2_field_decisions_pending': len(field_signoffs),
        'source_headings_present': 14,
        'source_prescription_phrases_present': 14,
        'space_evidence_explicit': 5,
        'space_proposals_inferred': 1,
        'wave2_other_space_records': 8,
        'registry_space_changes_verified': 5,
        'review_priorities': dict(Counter(r['priority'] for r in worklist)),
        'approved': 0,
        'planner_eligible': 0,
        'input_sha256': original_hashes,
        'output_sha256': {name: hashlib.sha256(payload).hexdigest() for name, payload in reports.items()},
        'notes': 'Publisher heading and phrase matches are evidence screens. All field signoffs, permissions, canonical decisions, and youth suitability remain pending.',
    }
    reports['summary'] = (json.dumps(summary, indent=2, sort_keys=True) + '\n').encode('utf-8')

    if any(path.exists() for path in outputs.values()):
        raise FileExistsError('Output created during execution; prior files preserved.')
    for path in outputs.values():
        path.parent.mkdir(parents=True, exist_ok=True)
    with tempfile.TemporaryDirectory(prefix='step12_40_', dir=root / 'results') as directory:
        staged = {}
        for name, payload in reports.items():
            stage_path = Path(directory) / outputs[name].name
            stage_path.write_bytes(payload)
            if hashlib.sha256(stage_path.read_bytes()).hexdigest() != hashlib.sha256(payload).hexdigest():
                raise RuntimeError(f'Failed staged checksum check for {name}')
            staged[name] = stage_path
        for name, path in outputs.items():
            with path.open('xb') as handle:
                handle.write(staged[name].read_bytes())
    if {key: checksum(path) for key, path in inputs.items()} != original_hashes:
        raise RuntimeError('An input file changed during execution.')
    if any(checksum(path) != hashlib.sha256(reports[key]).hexdigest()
           for key, path in outputs.items()):
        raise RuntimeError('Saved output checksum verification failed.')

    print('Catalog candidates / source records: 144 / 156')
    print('Wave 2 candidates / field reviews: 14 / 56')
    print('Original v0.3.2 → v0.3.3 source-space changes verified:', len(source_space_changes))
    print('\nWAVE 2 SOURCE EVIDENCE')
    print('  Publisher headings found:', 14)
    print('  Recorded prescription phrases found:', 14)
    print('  Explicit space evidence pending signoff:', 5)
    print('  Inferred space proposals pending confirmation:', 1)
    print('  Other Wave 2 space records requiring signoff:', 8)
    print('\nREVIEW PRIORITIES')
    for priority in ('high', 'medium', 'standard'):
        print(f'  {priority}:', sum(r['priority'] == priority for r in worklist))
    print('\nOUTPUTS')
    for path in outputs.values():
        print('  CREATED', path.relative_to(root))
    print('\nCanonical signoffs / approvals / planner-eligible: 0 / 0 / 0')
    print('Earlier catalogs and datasets overwritten: No')
    print('Git operations performed: None')
    print('OVERALL: PASS — Wave 2 evidence consolidated; reviewer decisions pending')
```

```python
# Step 12-41 - Wave 2 Source-Fact Reconciliation and Review Decisions
# Prepare source-linked metadata proposals without changing catalog approvals.

from collections import Counter, defaultdict
from pathlib import Path
import csv
import io
import json


def project_root():
    for directory in (Path.cwd().resolve(), *Path.cwd().resolve().parents):
        if (directory / '.git').exists():
            return directory
    raise FileNotFoundError('Project Git root not found.')


ROOT = project_root()
RESULTS = ROOT / 'results'
DATA = ROOT / 'data' / 'processed'

FILES = {
    'matrix': RESULTS / 'basketball_wave2_evidence_matrix_v0_3_3.csv',
    'fields': RESULTS / 'basketball_wave2_field_signoff_v0_3_3.csv',
    'worklist': RESULTS / 'basketball_wave2_signoff_worklist_v0_3_3.csv',
    'catalog': DATA / 'basketball_drill_catalog_v0_3_3_REVIEW_ONLY.csv',
    'registry': RESULTS / 'basketball_drill_source_registry_v0_3_3.csv',
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open('r', newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        return list(reader), list(reader.fieldnames or [])


inputs = {name: read_csv(path)[0] for name, path in FILES.items()}
matrix = inputs['matrix']
field_rows = inputs['fields']
worklist = inputs['worklist']
catalog = inputs['catalog']
registry = inputs['registry']

if (len(matrix), len(field_rows), len(worklist), len(catalog), len(registry)) != (14, 56, 14, 144, 156):
    raise ValueError('Unexpected Step 12 baseline counts; no review files were written.')


def unique_index(rows, key):
    keys = [row[key] for row in rows]
    if any(not k for k in keys) or len(keys) != len(set(keys)):
        raise ValueError(f'Missing or repeated {key}.')
    return {row[key]: row for row in rows}


matrix_by_id = unique_index(matrix, 'candidate_group_id')
work_by_id = unique_index(worklist, 'candidate_group_id')
cat_by_id = unique_index(catalog, 'catalog_candidate_id')
reg_by_id = unique_index(registry, 'evidence_id')

if set(matrix_by_id) != set(work_by_id):
    raise ValueError('Wave 2 matrix and review worklist have different candidate IDs.')

if any(gid not in cat_by_id for gid in matrix_by_id):
    raise ValueError('Wave 2 group missing from current catalog.')

if any(row['evidence_id'] not in reg_by_id for row in matrix):
    raise ValueError('Wave 2 source occurrence missing from current registry.')

if any(row['catalog_approval_status'] != 'pending' or row['planner_eligible'] != 'no' for row in catalog):
    raise ValueError('Unexpected catalog approval or eligibility decision; stop for review.')

if any(row['catalog_approval_status'] != 'pending' or row['planner_eligible'] != 'no' for row in worklist):
    raise ValueError('Unexpected worklist approval or eligibility decision; stop for review.')

if any(row['reviewer_decision'] != 'pending' for row in field_rows):
    raise ValueError('Existing reviewer decisions detected; preserve them and reconcile manually.')

for row in matrix:
    gid, eid = row['candidate_group_id'], row['evidence_id']
    if reg_by_id[eid]['catalog_candidate_id'] != gid:
        raise ValueError(f'Catalog/source linkage inconsistent for {eid}.')
    for field, current in (
        ('equipment', 'current_equipment'),
        ('participants', 'current_participants'),
        ('space', 'current_space'),
        ('source_prescription', 'current_prescription'),
    ):
        if reg_by_id[eid][field].strip() != row[current].strip():
            raise ValueError(f'Source registry differs from Wave 2 evidence: {eid} / {field}.')

fields_by_group = defaultdict(dict)
for row in field_rows:
    gid, field = row['candidate_group_id'], row['field']
    if field not in {'equipment', 'participants', 'space', 'prescription'} or field in fields_by_group[gid]:
        raise ValueError(f'Unexpected or duplicate review field: {gid} / {field}.')
    fields_by_group[gid][field] = row
    if gid not in matrix_by_id or row['evidence_id'] != matrix_by_id[gid]['evidence_id']:
        raise ValueError(f'Field review does not match evidence: {gid} / {field}.')
if any(set(v) != {'equipment', 'participants', 'space', 'prescription'} for v in fields_by_group.values()):
    raise ValueError('Every Wave 2 candidate must have exactly four field reviews.')

# Compact editorial source-fact proposals. These are not independent team signoffs.
# Values distinguish explicit source instructions from practical inferences.
# All source URLs are supplied by the project review matrix.
SOURCE_FACTS = {
    'JR-EVID-0031': {
        'equipment': ('Basketball required for dribbling; marked court belongs under space.', 'activity_implied'),
        'participants': ('Two active players per pair: attacker and defender.', 'source_explicit'),
        'space': ('Half-court with baseline; up to four pairs per half-court.', 'source_explicit'),
        'prescription': ('5 minutes.', 'source_explicit'),
        'note': 'Separate marked court from equipment; the source specifies pair roles and an eight-second crossing objective.'
    },
    'JR-EVID-0035': {
        'equipment': ('Basketball used for the pass; hoop is part of the scoring environment.', 'activity_implied'),
        'participants': ('Attacker, defender, and passer (coach or another player).', 'source_explicit'),
        'space': ('Court with wings and top of key; two groups may use one half-court.', 'source_explicit'),
        'prescription': ('5–7 minutes.', 'source_explicit'),
        'note': 'The source allows a coach or another player to act as passer; do not assume a two-person solo version.'
    },
    'JR-EVID-0032': {
        'equipment': ('Basketball for passing and shooting; basket used for scoring.', 'activity_implied'),
        'participants': ('Two active players: attacker and defender.', 'source_explicit'),
        'space': ('Playing space inside the three-point line.', 'source_explicit'),
        'prescription': ('6 minutes.', 'source_explicit'),
        'note': 'The source specifies no-dribble pivoting before a later one-dribble variation.'
    },
    'JR-EVID-0033': {
        'equipment': ('Basketball and baskets for transition scoring.', 'source_explicit'),
        'participants': ('Four active players: three attackers and one defender, with rotating court groups.', 'source_explicit'),
        'space': ('Court with half-court, four corner starting areas, and a basket.', 'source_explicit'),
        'prescription': ('5 minutes.', 'source_explicit'),
        'note': 'The source uses players in the four court corners and a rotating transition game.'
    },
    'JR-EVID-0037': {
        'equipment': ('Basketball and basket(s) used in the game.', 'activity_implied'),
        'participants': ('Two teams of three active players; more teams may rotate.', 'source_explicit'),
        'space': ('Court with top-of-key restart; both baskets when four or more teams participate.', 'source_explicit'),
        'prescription': ('Games to three made baskets; this is not a three-minute duration.', 'source_explicit'),
        'note': 'Both baskets are conditional on team count; do not mark two baskets as universally required.'
    },
    'JR-EVID-0038': {
        'equipment': ('Basketball required for gameplay; full court is a space requirement.', 'activity_implied'),
        'participants': ('Two teams of five active players.', 'source_explicit'),
        'space': ('Full basketball court.', 'source_explicit'),
        'prescription': ('Games to four made baskets; four passes before a shot.', 'source_explicit'),
        'note': 'Remove the full-court concept from the equipment field; do not convert made baskets into minutes.'
    },
    'JR-EVID-0001': {
        'equipment': ('No basketball for base activity; adding a ball is an optional variation.', 'source_explicit'),
        'participants': ('Coach-directed group with verbal movement commands.', 'source_explicit'),
        'space': ('Gym or basketball court, starting at the baseline.', 'source_explicit'),
        'prescription': ('1 × 3 minutes.', 'source_explicit'),
        'note': 'A basketball keyword occurs in an optional modification, not the base requirement.'
    },
    'JR-EVID-0002': {
        'equipment': ('Basketball; sharing a ball is permitted.', 'source_explicit'),
        'participants': ('Individual base activity; multiple players can perform it at once.', 'source_explicit'),
        'space': ('Designated practice area; no specific court geometry required in the source.', 'source_explicit'),
        'prescription': ('1 × 20 seconds.', 'source_explicit'),
        'note': 'Restore source-described ball sharing; no mandatory partner should be inferred.'
    },
    'JR-EVID-0056': {
        'equipment': ('Basketball for the dribbler; half-court belongs under space.', 'activity_implied'),
        'participants': ('Three active players: one attacker and two defenders.', 'source_explicit'),
        'space': ('Half-court with sideline and baseline; up to two groups per half-court.', 'source_explicit'),
        'prescription': ('5–7 minutes.', 'source_explicit'),
        'note': 'Source requires a second defender in the base activity, so this is not one-on-one solo practice.'
    },
    'JR-EVID-0058': {
        'equipment': ('Basketballs and basket; some players start without a ball.', 'source_explicit'),
        'participants': ('Multiple players at each basket; group size is not specified numerically.', 'source_explicit'),
        'space': ('Basket and nearby painted area; no numeric minimum floor dimensions stated.', 'source_explicit'),
        'prescription': ('5 minutes.', 'source_explicit'),
        'note': 'Avoid inventing an exact group minimum or floor-size requirement.'
    },
    'JR-EVID-0062': {
        'equipment': ('Basketball and basket(s) for play.', 'activity_implied'),
        'participants': ('Two teams of three active players.', 'source_explicit'),
        'space': ('Court with basket, baseline, and top-of-key; both baskets when four or more teams play.', 'source_explicit'),
        'prescription': ('Games to three made baskets; no standalone minute duration supplied.', 'source_explicit'),
        'note': 'Retain the conditional use of both baskets rather than treating it as a fixed requirement.'
    },
    'JR-EVID-0060': {
        'equipment': ('Basketballs; source calls for 2–3 balls per baseline line, plus a scoring basket.', 'source_explicit'),
        'participants': ('Attacker, defender, and coach or player passer, with rotating lines.', 'source_explicit'),
        'space': ('Court with baseline lines, lane lines, wing areas, and passer near top of key.', 'source_explicit'),
        'prescription': ('3–4 minutes.', 'source_explicit'),
        'note': 'The source allows a coach or another player to act as passer.'
    },
    'JR-EVID-0061': {
        'equipment': ('Basketball and scoring basket used in the activity.', 'source_explicit'),
        'participants': ('Four active players: three attackers and one defender.', 'source_explicit'),
        'space': ('Three-point line with top and wing spots plus access to the basket.', 'source_explicit'),
        'prescription': ('5 minutes.', 'source_explicit'),
        'note': 'The source positions the three attackers around the three-point line before driving and passing.'
    },
    'JR-EVID-0059': {
        'equipment': ('Basketball and basket inferred from shooting, passing, and rebounding roles.', 'operational_inference'),
        'participants': ('Three active players: shooter, passer, and rebounder, rotating roles.', 'source_explicit'),
        'space': ('Court area with a basket is an operational inference; minimum dimensions are not stated.', 'operational_inference'),
        'prescription': ('2 × 10 shots each player; not a 20-minute duration.', 'source_explicit'),
        'note': 'Keep exact minimum space unresolved; an inferred hoop requirement must not be labeled an explicit source rule.'
    },
}

if set(SOURCE_FACTS) != {row['evidence_id'] for row in matrix}:
    raise ValueError('Source-fact drafts do not cover the exact 14 Wave 2 records.')

field_proposals = []
candidate_proposals = []
for work in sorted(worklist, key=lambda row: ({'high': 0, 'medium': 1, 'standard': 2}.get(row['priority'], 3), int(row['review_rank']))):
    gid = work['candidate_group_id']
    source = matrix_by_id[gid]
    eid = source['evidence_id']
    facts = SOURCE_FACTS[eid]
    disagreement_fields = []
    unresolved_fields = []

    for field in ['equipment', 'participants', 'space', 'prescription']:
        raw = fields_by_group[gid][field]
        proposal, source_basis = facts[field]
        catalog_value, registry_value = raw['catalog_value'].strip(), raw['current_registry_value'].strip()
        wording_diff = catalog_value.casefold() != registry_value.casefold()
        if wording_diff:
            disagreement_fields.append(field)
        if source_basis == 'operational_inference':
            unresolved_fields.append(field)

        field_proposals.append({
            'candidate_group_id': gid,
            'evidence_id': eid,
            'drill_name': source['drill_name'],
            'field': field,
            'catalog_value': catalog_value,
            'source_registry_value': registry_value,
            'catalog_registry_wording_differs': 'yes' if wording_diff else 'no',
            'source_screen': raw['publisher_text_screen'],
            'evidence_basis': source_basis,
            'proposed_normalized_value': proposal,
            'official_source_url': source['official_source_url'],
            'source_fact_review': 'draft_pending_independent_team_review',
            'reviewer_decision': 'pending',
            'final_verified_value': '',
            'reviewer': '',
            'review_date': '',
            'review_notes': '',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
        })

    candidate_proposals.append({
        'priority': work['priority'],
        'review_rank': work['review_rank'],
        'candidate_group_id': gid,
        'evidence_id': eid,
        'drill_name': source['drill_name'],
        'source_level': source['curriculum_level'],
        'practice_plan': source['practice_plan'],
        'official_source_url': source['official_source_url'],
        'heading_screen': source['source_heading_screen'],
        'prescription_screen': source['source_prescription_screen'],
        'catalog_registry_wording_differences': ';'.join(disagreement_fields),
        'operational_inferences_to_review': ';'.join(unresolved_fields),
        'source_fact_note': facts['note'],
        'proposed_next_action': work['next_review_action'],
        'field_evidence_review': 'pending_team_review',
        'instructional_content_signoff': 'pending',
        'canonical_identity_signoff': 'pending',
        'permission_review': 'pending',
        'youth_planning_review': 'pending',
        'catalog_approval_status': 'pending',
        'planner_eligible': 'no',
        'reviewer': '',
        'review_date': '',
        'review_notes': '',
    })

if len(field_proposals) != 56 or len(candidate_proposals) != 14:
    raise ValueError('Draft record counts changed unexpectedly.')

summary = {
    'step': '12-41',
    'catalog_version_read_only': '0.3.3',
    'catalog_candidate_count': 144,
    'source_occurrence_count': 156,
    'wave2_candidates': 14,
    'field_decision_drafts': 56,
    'candidate_priorities': dict(Counter(row['priority'] for row in candidate_proposals)),
    'field_wording_differences': dict(Counter(row['field'] for row in field_proposals if row['catalog_registry_wording_differs'] == 'yes')),
    'source_basis_counts': dict(Counter(row['evidence_basis'] for row in field_proposals)),
    'operational_inference_rows': [f"{row['evidence_id']}:{row['field']}" for row in field_proposals if row['evidence_basis'] == 'operational_inference'],
    'reviewer_decisions_completed': 0,
    'catalog_approved': 0,
    'planner_eligible': 0,
    'source_documents_copied': False,
    'note': 'Factual metadata proposals for team review, not signed approvals or finalized permissions.',
}

OUTPUTS = {
    RESULTS / 'basketball_wave2_field_decision_drafts_v0_3_3.csv': field_proposals,
    RESULTS / 'basketball_wave2_candidate_decision_packet_v0_3_3.csv': candidate_proposals,
}
SUMMARY_PATH = RESULTS / 'basketball_step12_41_summary_v0_3_3.json'


def encode_csv(rows):
    handle = io.StringIO(newline='')
    writer = csv.DictWriter(handle, fieldnames=list(rows[0]))
    writer.writeheader()
    writer.writerows(rows)
    return handle.getvalue().encode('utf-8')


payloads = {path: encode_csv(rows) for path, rows in OUTPUTS.items()}
payloads[SUMMARY_PATH] = (json.dumps(summary, indent=2, ensure_ascii=False) + '\n').encode('utf-8')

# Preflight every output before writing any files. Exact reruns are safe.
for path, payload in payloads.items():
    if path.exists() and path.read_bytes() != payload:
        raise FileExistsError(f'Existing output differs from current proposal: {path}. No files changed.')

for path, payload in payloads.items():
    path.parent.mkdir(parents=True, exist_ok=True)
    if not path.exists():
        with path.open('xb') as handle:
            handle.write(payload)
        state = 'CREATED  '
    else:
        state = 'PRESERVED'
    if path.read_bytes() != payload:
        raise RuntimeError(f'Output integrity verification failed: {path}')
    print(f'{state} {path.relative_to(ROOT)}')

print('\n' + '=' * 72)
print('STEP 12-41 — WAVE 2 FIELD-LEVEL DECISION DRAFTS')
print('=' * 72)
print(f'Catalog candidates / source records: {len(catalog)} / {len(registry)}')
print(f'Wave 2 drills / field proposals: {len(candidate_proposals)} / {len(field_proposals)}')
print('Review priorities:', dict(Counter(r['priority'] for r in candidate_proposals)))
print('Source basis:', summary['source_basis_counts'])
print('Catalog/registry wording differences:', summary['field_wording_differences'])
print('Operational inference checks:', summary['operational_inference_rows'])
print('\nIMPORTANT SOURCE-FACT FOLLOWUPS')
for eid in ['JR-EVID-0059', 'JR-EVID-0038', 'JR-EVID-0001', 'JR-EVID-0002']:
    row = next(r for r in matrix if r['evidence_id'] == eid)
    print(f"  {eid} | {row['drill_name']}: {SOURCE_FACTS[eid]['note']}")
print('\nExisting catalog and approval decisions modified: No')
print('Human signoffs / catalog approvals / planner eligible: 0 / 0 / 0')
print('Git operations performed: None')
print('OVERALL: PASS — review proposals saved; team decisions and permissions pending')
print('=' * 72)
```

```python
# Step 12-42 - Wave 2 Metadata Decision Sheet and Versioned Review Catalog
# Stage field-level review decisions; preserve source data and approval controls.

import csv
import hashlib
import io
import json
import tempfile
from collections import Counter, defaultdict
from datetime import date
from pathlib import Path


ROOT = next(
    (p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (p / '.git').exists()), None
)
if ROOT is None:
    raise FileNotFoundError('Project Git root not found.')

INPUTS = {
    'catalog': ROOT / 'data/processed/basketball_drill_catalog_v0_3_3_REVIEW_ONLY.csv',
    'registry': ROOT / 'results/basketball_drill_source_registry_v0_3_3.csv',
    'fields': ROOT / 'results/basketball_wave2_field_decision_drafts_v0_3_3.csv',
    'packet': ROOT / 'results/basketball_wave2_candidate_decision_packet_v0_3_3.csv',
}
OUTPUTS = {
    'field_sheet': ROOT / 'results/basketball_wave2_field_decision_sheet_v0_3_4.csv',
    'candidate_sheet': ROOT / 'results/basketball_wave2_candidate_signoff_sheet_v0_3_4.csv',
    'catalog': ROOT / 'data/processed/basketball_drill_catalog_v0_3_4_REVIEW_ONLY.csv',
    'summary': ROOT / 'results/basketball_step12_42_summary_v0_3_4.json',
}


def digest(data):
    return hashlib.sha256(data).hexdigest()


def file_digest(path):
    return digest(path.read_bytes())


def load_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open('r', newline='', encoding='utf-8-sig') as stream:
        reader = csv.DictReader(stream)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or len(columns) != len(set(columns)):
        raise ValueError(f'Invalid or repeated CSV headers: {path.name}')
    if any(None in row or any(value is None for value in row.values())
           for row in rows):
        raise ValueError(f'Malformed CSV rows: {path.name}')
    return rows, columns


def unique_rows(rows, field, expected):
    if len(rows) != expected:
        raise ValueError(f'{field}: expected {expected} rows, found {len(rows)}.')
    ids = [row[field].strip() for row in rows]
    if not all(ids) or len(set(ids)) != len(ids):
        raise ValueError(f'Duplicate or blank {field}.')
    return dict(zip(ids, rows))


def encode_csv(rows, columns):
    buf = io.StringIO(newline='')
    writer = csv.DictWriter(buf, fieldnames=columns, lineterminator='\n',
                            extrasaction='raise')
    writer.writeheader()
    writer.writerows(rows)
    return buf.getvalue().encode('utf-8')


print('=' * 72)
print('STEP 12-42 — WAVE 2 METADATA REVIEW DECISION SHEETS')
print('=' * 72)

existing = {key: path.exists() for key, path in OUTPUTS.items()}
if any(existing.values()):
    if not all(existing.values()):
        raise FileExistsError('Incomplete Step 12-42 outputs exist; inspect before rerunning.')
    old = json.loads(OUTPUTS['summary'].read_text(encoding='utf-8'))
    for key, path in OUTPUTS.items():
        if key != 'summary' and file_digest(path) != old['output_sha256'][key]:
            raise RuntimeError(f'Existing review output has changed: {path.name}')
    print('Existing Step 12-42 outputs verified and preserved.')
    print('Field decisions:', old['field_decisions'])
    print('Candidate reviews:', old['wave2_candidates'])
    print('No new approvals or catalog changes.')
    print('OVERALL: PASS — existing reviewed outputs preserved')
else:
    input_digests = {key: file_digest(path) for key, path in INPUTS.items()}
    catalog, catalog_columns = load_csv(INPUTS['catalog'])
    registry, _ = load_csv(INPUTS['registry'])
    drafts, draft_columns = load_csv(INPUTS['fields'])
    packet, packet_columns = load_csv(INPUTS['packet'])

    required_catalog = {
        'catalog_candidate_id', 'approved_drill_id', 'catalog_approval_status',
        'planner_eligible', 'catalog_version', 'equipment', 'participants',
        'space', 'source_prescription',
    }
    required_registry = {
        'evidence_id', 'catalog_candidate_id', 'source_url',
        'source_prescription', 'equipment', 'participants', 'space',
    }
    required_draft = {
        'candidate_group_id', 'evidence_id', 'drill_name', 'field',
        'catalog_value', 'source_registry_value', 'source_screen',
        'evidence_basis', 'proposed_normalized_value', 'official_source_url',
        'source_fact_review', 'reviewer_decision', 'catalog_approval_status',
        'planner_eligible',
    }
    required_packet = {
        'candidate_group_id', 'evidence_id', 'priority', 'review_rank',
        'drill_name', 'official_source_url', 'field_evidence_review',
        'instructional_content_signoff', 'canonical_identity_signoff',
        'permission_review', 'youth_planning_review',
        'catalog_approval_status', 'planner_eligible',
    }
    for label, columns, required in [
        ('catalog', catalog_columns, required_catalog),
        ('registry', list(registry[0]) if registry else [], required_registry),
        ('drafts', draft_columns, required_draft),
        ('packet', packet_columns, required_packet),
    ]:
        missing = required - set(columns)
        if missing:
            raise ValueError(f'Missing {label} fields: {sorted(missing)}')

    catalog_index = unique_rows(catalog, 'catalog_candidate_id', 144)
    registry_index = unique_rows(registry, 'evidence_id', 156)
    packet_index = unique_rows(packet, 'candidate_group_id', 14)
    if len(drafts) != 56 or any(
        row['catalog_approval_status'] != 'pending'
        or row['planner_eligible'] != 'no'
        or row['reviewer_decision'] != 'pending'
        for row in drafts
    ):
        raise ValueError('Expected 56 undecided proposals with all approvals restricted.')
    if any(
        row['catalog_version'] != '0.3.3'
        or row['catalog_approval_status'] != 'pending'
        or row['planner_eligible'] != 'no'
        or row['approved_drill_id'].strip()
        for row in catalog
    ):
        raise ValueError('Catalog version or approval flags differ from v0.3.3 baseline.')
    if any(
        row['catalog_approval_status'] != 'pending'
        or row['planner_eligible'] != 'no'
        or row['field_evidence_review'] != 'pending_team_review'
        for row in packet
    ):
        raise ValueError('Existing candidate reviews or approvals need reconciliation.')
    if any(row['catalog_candidate_id'] not in catalog_index for row in registry):
        raise ValueError('Registry contains unknown catalog groups.')

    field_names = {'equipment', 'participants', 'space', 'prescription'}
    catalog_field_map = {
        'equipment': 'equipment',
        'participants': 'participants',
        'space': 'space',
        'prescription': 'source_prescription',
    }
    by_group = defaultdict(list)
    source_bases = Counter()
    screening = Counter()
    field_sheet = []
    flagged_items = []
    seen_pairs = set()

    for draft in drafts:
        gid, eid, field = (draft['candidate_group_id'].strip(),
                           draft['evidence_id'].strip(), draft['field'].strip())
        if gid not in packet_index or gid not in catalog_index or eid not in registry_index:
            raise ValueError(f'Unknown review linkage: {gid} / {eid}')
        if field not in field_names or (gid, field) in seen_pairs:
            raise ValueError(f'Duplicate or invalid field proposal: {gid} / {field}')
        seen_pairs.add((gid, field))
        parent = packet_index[gid]
        catalog_row = catalog_index[gid]
        source = registry_index[eid]
        if (parent['evidence_id'] != eid
                or source['catalog_candidate_id'] != gid
                or draft['official_source_url'].rstrip('/') != source['source_url'].rstrip('/')
                or parent['official_source_url'].rstrip('/') != source['source_url'].rstrip('/')
                or draft['catalog_value'].strip() != catalog_row[catalog_field_map[field]].strip()
                or draft['source_registry_value'].strip() != source[catalog_field_map[field]].strip()):
            raise ValueError(f'Source, registry, or catalog values changed: {gid}/{field}')

        basis, screen = draft['evidence_basis'], draft['source_screen']
        if basis == 'operational_inference':
            review_class = 'operational_inference_requires_confirmation'
        elif basis == 'activity_implied':
            review_class = 'activity_implied_requires_source_review'
        elif basis == 'source_explicit' and screen == 'exact_recorded_phrase_present':
            review_class = 'exact_phrase_seen_pending_signoff'
        elif basis == 'source_explicit' and screen == 'cues_present_not_approval':
            review_class = 'keyword_evidence_pending_source_review'
        elif basis == 'source_explicit' and screen == 'no_cues_detected_or_section_unavailable':
            review_class = 'claimed_explicit_but_no_source_cue'
        else:
            raise ValueError(f'Unexpected source evidence classification for {gid}/{field}')

        source_bases[basis] += 1
        screening[review_class] += 1
        if review_class == 'claimed_explicit_but_no_source_cue' or basis == 'operational_inference':
            flagged_items.append((eid, field, review_class))

        field_sheet.append({
            'priority': parent['priority'],
            'candidate_group_id': gid,
            'evidence_id': eid,
            'drill_name': parent['drill_name'],
            'field': field,
            'current_catalog_value': draft['catalog_value'],
            'current_registry_value': draft['source_registry_value'],
            'proposed_value': draft['proposed_normalized_value'],
            'evidence_basis_claim': basis,
            'automatic_source_screen': screen,
            'review_evidence_class': review_class,
            'source_url': source['source_url'],
            'reviewer_decision': 'pending',
            'decision_options': 'accept_proposal;retain_current;revise_value;defer',
            'final_verified_value': '',
            'reviewer': '',
            'review_date': '',
            'review_notes': '',
            'source_permission_review': 'pending',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
        })
        by_group[gid].append(field_sheet[-1])

    if set(by_group) != set(packet_index) or any(
        {r['field'] for r in rows} != field_names or len(rows) != 4
        for rows in by_group.values()
    ):
        raise ValueError('Expected four unique field proposals for each of 14 groups.')

    priority_order = {'high': 0, 'medium': 1, 'standard': 2}
    field_order = {'equipment': 0, 'participants': 1, 'space': 2, 'prescription': 3}
    field_sheet.sort(key=lambda r: (
        priority_order.get(r['priority'], 9),
        int(packet_index[r['candidate_group_id']]['review_rank']),
        field_order[r['field']],
    ))

    candidate_sheet = []
    for item in sorted(packet, key=lambda r: (
        priority_order.get(r['priority'], 9), int(r['review_rank']))):
        gid = item['candidate_group_id']
        cases = by_group[gid]
        flagged = [r['field'] for r in cases
                   if r['review_evidence_class'] in {
                       'operational_inference_requires_confirmation',
                       'claimed_explicit_but_no_source_cue',
                   }]
        candidate_sheet.append({
            'priority': item['priority'],
            'candidate_group_id': gid,
            'evidence_id': item['evidence_id'],
            'drill_name': item['drill_name'],
            'source_url': item['official_source_url'],
            'field_proposals': 4,
            'fields_needing_special_attention': ';'.join(flagged),
            'source_fact_note': item.get('source_fact_note', ''),
            'proposed_next_action': item.get('proposed_next_action', ''),
            'field_reviews_completed': 0,
            'field_review_status': 'pending_team_review',
            'instructional_content_signoff': 'pending',
            'canonical_identity_signoff': 'pending',
            'permission_review': 'pending',
            'youth_planning_review': 'pending',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
            'reviewer': '',
            'review_date': '',
            'review_notes': '',
        })

    new_catalog_columns = [
        'wave2_field_proposal_count',
        'wave2_review_priority',
        'wave2_special_evidence_flags',
        'wave2_metadata_decision_status',
        'wave2_field_sheet_reference',
    ]
    if set(new_catalog_columns) & set(catalog_columns):
        raise ValueError('The proposed v0.3.4 metadata already exists in source catalog.')

    for row in catalog:
        gid = row['catalog_candidate_id']
        proposals = by_group.get(gid, [])
        row.update({
            'catalog_version': '0.3.4',
            'wave2_field_proposal_count': str(len(proposals)),
            'wave2_review_priority': (packet_index[gid]['priority'] if proposals else ''),
            'wave2_special_evidence_flags': ';'.join(
                r['field'] for r in proposals if r['review_evidence_class'] in {
                    'operational_inference_requires_confirmation',
                    'claimed_explicit_but_no_source_cue',
                }),
            'wave2_metadata_decision_status': ('pending_team_review' if proposals else 'not_in_wave2'),
            'wave2_field_sheet_reference': (
                'results/basketball_wave2_field_decision_sheet_v0_3_4.csv'
                if proposals else ''
            ),
        })

    if len(catalog) != 144 or len(registry) != 156 or len(field_sheet) != 56 or len(candidate_sheet) != 14:
        raise ValueError('Output cardinality verification failed.')
    if any(r['catalog_approval_status'] != 'pending'
           or r['planner_eligible'] != 'no' or r['approved_drill_id'].strip()
           for r in catalog):
        raise ValueError('Approval settings changed unexpectedly.')
    if sum(int(r['wave2_field_proposal_count']) for r in catalog) != 56:
        raise ValueError('Catalog-to-proposal linkage is incomplete.')

    report_bytes = {
        'field_sheet': encode_csv(field_sheet, list(field_sheet[0])),
        'candidate_sheet': encode_csv(candidate_sheet, list(candidate_sheet[0])),
        'catalog': encode_csv(catalog, catalog_columns + new_catalog_columns),
    }
    summary = {
        'step': '12-42',
        'catalog_version': '0.3.4',
        'created_on': date.today().isoformat(),
        'catalog_candidates': 144,
        'source_occurrences': 156,
        'wave2_candidates': 14,
        'field_decisions': 56,
        'evidence_basis_claims': dict(sorted(source_bases.items())),
        'review_evidence_classes': dict(sorted(screening.items())),
        'priority_counts': dict(sorted(Counter(r['priority'] for r in candidate_sheet).items())),
        'special_attention_fields': [
            {'evidence_id': eid, 'field': field, 'reason': reason}
            for eid, field, reason in flagged_items
        ],
        'metadata_values_applied': 0,
        'reviewer_decisions_signed': 0,
        'approved_drills': 0,
        'planner_eligible_drills': 0,
        'input_sha256': input_digests,
        'output_sha256': {key: digest(value) for key, value in report_bytes.items()},
        'limits': ('All proposals require independent review. A keyword cue or source-explicit '
                   'label is not itself evidence of a signed verification. No baseline '
                   'metadata fields or approval decisions were changed.'),
    }
    report_bytes['summary'] = (json.dumps(summary, indent=2, sort_keys=True) + '\n').encode('utf-8')

    if any(path.exists() for path in OUTPUTS.values()):
        raise FileExistsError('An output appeared during staging; nothing overwritten.')
    for path in OUTPUTS.values():
        path.parent.mkdir(parents=True, exist_ok=True)

    with tempfile.TemporaryDirectory(prefix='step12_42_', dir=ROOT / 'results') as work:
        stage = Path(work)
        for key, payload in report_bytes.items():
            tmp = stage / OUTPUTS[key].name
            tmp.write_bytes(payload)
            if file_digest(tmp) != digest(payload):
                raise RuntimeError(f'Staged write failed for {key}')
        for key, output in OUTPUTS.items():
            with output.open('xb') as writer:
                writer.write((stage / output.name).read_bytes())

    if {key: file_digest(path) for key, path in INPUTS.items()} != input_digests:
        raise RuntimeError('An input file changed during output creation.')
    for key, path in OUTPUTS.items():
        if file_digest(path) != digest(report_bytes[key]):
            raise RuntimeError(f'Output verification failed: {key}')

    print('Catalog candidates / registered source occurrences: 144 / 156')
    print('Wave 2 candidates / field decisions: 14 / 56')
    print('\nEVIDENCE CLASSIFICATION')
    for status, count in sorted(screening.items()):
        print(f'  {status:<48} {count}')
    print('\nFIELDS REQUIRING EXTRA ATTENTION')
    for eid, field, reason in flagged_items:
        print(f'  {eid} | {field} | {reason}')
    print('\nPRIORITY COUNTS')
    for priority, count in sorted(summary['priority_counts'].items(),
                                  key=lambda kv: priority_order.get(kv[0], 9)):
        print(f'  {priority}: {count}')
    print('\nOUTPUTS')
    for path in OUTPUTS.values():
        print('  CREATED', path.relative_to(ROOT))
    print('\nMetadata values applied / reviewer signoffs: 0 / 0')
    print('Approved / planner eligible: 0 / 0')
    print('Prior catalogs, registry, and source files overwritten: No')
    print('Git operations performed: None')
    print('OVERALL: PASS — Wave 2 review decisions staged, human signoffs pending')
```

```python
# Step 12-43 - Wave 2 Source Exceptions and Human Decision Preparation
# Recheck targeted publisher evidence and prepare editable, unapproved review decisions.

import csv
import json
import re
import unicodedata
from collections import Counter, defaultdict
from datetime import date
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen


def project_root():
    cwd = Path.cwd().resolve()
    for location in (cwd, *cwd.parents):
        if (location / '.git').exists():
            return location
    raise FileNotFoundError('Git project root not found.')


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(f'Required input not found: {path}')
    with path.open(newline='', encoding='utf-8-sig') as file:
        reader = csv.DictReader(file)
        rows = list(reader)
        columns = reader.fieldnames or []
    if not rows:
        raise ValueError(f'Empty input file: {path.name}')
    return rows, columns


def require_columns(columns, required, name):
    absent = set(required) - set(columns)
    if absent:
        raise ValueError(f'{name} missing columns: {sorted(absent)}')


def normalized(text):
    text = unicodedata.normalize('NFKC', str(text or '')).casefold()
    return re.sub(r'\s+', ' ', text).strip()


class OfficialSections(HTMLParser):
    """Read h2 activity sections in memory; do not save publisher text."""

    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.active_heading = None
        self.heading_parts = []
        self.in_heading = False
        self.skipping = 0

    def handle_starttag(self, tag, attrs):
        if tag in {'script', 'style', 'noscript'}:
            self.skipping += 1
            return
        if self.skipping:
            return
        if tag == 'h2':
            self.in_heading = True
            self.heading_parts = []
        elif tag in {'br', 'p', 'li'} and self.active_heading is not None:
            self.active_heading['pieces'].append(' ')

    def handle_data(self, data):
        if self.skipping:
            return
        if self.in_heading:
            self.heading_parts.append(data)
        elif self.active_heading is not None:
            self.active_heading['pieces'].append(data)

    def handle_endtag(self, tag):
        if tag in {'script', 'style', 'noscript'}:
            self.skipping = max(0, self.skipping - 1)
            return
        if self.skipping:
            return
        if tag == 'h2' and self.in_heading:
            heading = normalized(' '.join(self.heading_parts))
            if heading:
                self.active_heading = {'heading': heading, 'pieces': []}
                self.sections.append(self.active_heading)
            self.in_heading = False
            self.heading_parts = []
        elif tag in {'p', 'li', 'div'} and self.active_heading is not None:
            self.active_heading['pieces'].append(' ')


def get_activity_section(url, activity):
    parts = urlsplit(url)
    if (parts.scheme != 'https' or parts.hostname != 'jr.nba.com'
            or not parts.path.startswith('/basketball-practice-plans/')):
        raise ValueError(f'Unexpected source URL: {url}')
    try:
        request = Request(url, headers={
            'User-Agent': 'Mozilla/5.0 (educational source metadata review)'
        })
        with urlopen(request, timeout=25) as response:
            if urlsplit(response.geturl()).hostname != 'jr.nba.com':
                raise ValueError('Source redirected outside official Jr. NBA domain.')
            content = response.read(2_000_000).decode('utf-8', errors='replace')
        parser = OfficialSections()
        parser.feed(content)
        hits = [section for section in parser.sections
                if section['heading'] == normalized(activity)]
        if len(hits) != 1:
            return {'status': 'heading_unresolved', 'text': '',
                    'matches': len(hits)}
        return {'status': 'heading_found',
                'text': normalized(' '.join(hits[0]['pieces'])),
                'matches': 1}
    except (HTTPError, URLError, TimeoutError, OSError) as error:
        return {'status': 'page_unavailable', 'text': '',
                'matches': 0, 'error': type(error).__name__}


root = project_root()
results = root / 'results'
original_sheet = results / 'basketball_wave2_field_decision_sheet_v0_3_4.csv'
original_signoff = results / 'basketball_wave2_candidate_signoff_sheet_v0_3_4.csv'
catalog_file = root / 'data/processed/basketball_drill_catalog_v0_3_4_REVIEW_ONLY.csv'
registry_file = results / 'basketball_drill_source_registry_v0_3_3.csv'

fields, field_columns = read_csv(original_sheet)
signoffs, signoff_columns = read_csv(original_signoff)
catalog, catalog_columns = read_csv(catalog_file)
registry, registry_columns = read_csv(registry_file)

require_columns(field_columns, {
    'candidate_group_id', 'evidence_id', 'drill_name', 'field',
    'current_catalog_value', 'current_registry_value', 'proposed_value',
    'review_evidence_class', 'source_url', 'reviewer_decision',
    'final_verified_value', 'catalog_approval_status', 'planner_eligible'
}, original_sheet.name)
require_columns(signoff_columns, {
    'candidate_group_id', 'evidence_id', 'drill_name',
    'catalog_approval_status', 'planner_eligible', 'field_reviews_completed'
}, original_signoff.name)
require_columns(catalog_columns, {
    'catalog_candidate_id', 'catalog_approval_status', 'planner_eligible',
    'approved_drill_id', 'catalog_version'
}, catalog_file.name)
require_columns(registry_columns, {
    'evidence_id', 'catalog_candidate_id'
}, registry_file.name)

if (len(fields), len(signoffs), len(catalog), len(registry)) != (56, 14, 144, 156):
    raise ValueError('Unexpected source totals. Inspect latest catalog and sheets.')
if any(row['catalog_approval_status'] != 'pending'
       or row['planner_eligible'] != 'no' for row in fields + signoffs + catalog):
    raise ValueError('Some approval state has changed; review before proceeding.')
if any(row.get('approved_drill_id', '').strip() for row in catalog):
    raise ValueError('A canonical approved drill ID is present; review required.')
if any(row['reviewer_decision'] != 'pending' or row['final_verified_value'].strip()
       for row in fields):
    raise ValueError('Input field decisions have already been edited; preserve human choices.')
if any(row['field_reviews_completed'] != '0' for row in signoffs):
    raise ValueError('Human field reviews already recorded; review before preparing a new packet.')
if any(row.get('catalog_version') != '0.3.4' for row in catalog):
    raise ValueError('Input catalog is not v0.3.4.')

catalog_by_id = {row['catalog_candidate_id']: row for row in catalog}
registry_by_id = {row['evidence_id']: row for row in registry}
signoff_by_id = {row['candidate_group_id']: row for row in signoffs}
if len(catalog_by_id) != 144 or len(registry_by_id) != 156 or len(signoff_by_id) != 14:
    raise ValueError('Duplicate group or evidence ID detected.')

field_keys = set()
for row in fields:
    key = (row['candidate_group_id'], row['evidence_id'], row['field'])
    if key in field_keys:
        raise ValueError(f'Duplicate field review record: {key}')
    field_keys.add(key)
    if row['field'] not in {'equipment', 'participants', 'space', 'prescription'}:
        raise ValueError(f'Unexpected field: {row["field"]}')
    if row['candidate_group_id'] not in signoff_by_id:
        raise ValueError(f'Field group not in signoff sheet: {key}')
    if row['candidate_group_id'] not in catalog_by_id:
        raise ValueError(f'Field group not in catalog: {key}')
    if row['evidence_id'] not in registry_by_id:
        raise ValueError(f'Evidence missing from registry: {key}')
    if registry_by_id[row['evidence_id']]['catalog_candidate_id'] != row['candidate_group_id']:
        raise ValueError(f'Source registry linkage mismatch: {key}')
    if signoff_by_id[row['candidate_group_id']]['evidence_id'] != row['evidence_id']:
        raise ValueError(f'Candidate signoff linkage mismatch: {key}')
    source_group = catalog_by_id[row['candidate_group_id']]
    if source_group[row['field'] if row['field'] != 'prescription'
                    else 'source_prescription'] != row['current_catalog_value']:
        raise ValueError(f'Catalog field changed since decision sheet: {key}')

field_counts = Counter(row['candidate_group_id'] for row in fields)
if set(field_counts) != set(signoff_by_id) or set(field_counts.values()) != {4}:
    raise ValueError('Each selected candidate must have exactly four field reviews.')

print('=' * 72)
print('STEP 12-43 — SOURCE EXCEPTIONS AND REVIEWER DECISION PACKET')
print('=' * 72)
print(f'Catalog candidates / source occurrences: {len(catalog)} / {len(registry)}')
print(f'Wave 2 candidates / field decisions: {len(signoffs)} / {len(fields)}')

# Check two exact source sections for the three special fields.
exception_cases = [
    ('JR-EVID-0059', 'equipment', '3-Person Shooting',
     'https://jr.nba.com/basketball-practice-plans/all-star/page/3/'),
    ('JR-EVID-0059', 'space', '3-Person Shooting',
     'https://jr.nba.com/basketball-practice-plans/all-star/page/3/'),
    ('JR-EVID-0002', 'space', 'Smacks',
     'https://jr.nba.com/basketball-practice-plans/rookie/page/3/'),
]
source_checks = {}
for evidence_id, field_name, name, url in exception_cases:
    if (evidence_id, field_name) not in {
        (r['evidence_id'], r['field']) for r in fields
    }:
        raise ValueError(f'Missing expected special review: {evidence_id}:{field_name}')
    if url not in {r['source_url'] for r in fields if r['evidence_id'] == evidence_id}:
        raise ValueError(f'Official URL mismatch: {evidence_id}')
    key = (url, name)
    if key not in source_checks:
        source_checks[key] = get_activity_section(url, name)

exception_rows = []
exception_lookup = {}
for evidence_id, field_name, name, url in exception_cases:
    section = source_checks[url, name]
    text = section['text']
    if name == 'Smacks':
        cues = ('designated area' in text and
                'basketball' in text and 'share' in text)
        fact = ('Publisher explicitly describes a designated area and permits '
                'sharing a basketball; no specific court dimensions established.')
        recommendation = ('Use source-described designated practice area; '
                          'do not add a precise court geometry.')
        value = 'Designated practice area'
    elif field_name == 'equipment':
        cues = (('shoot' in text) and ('pass' in text) and ('rebound' in text))
        fact = ('Publisher describes shooting, passing and rebounding roles; '
                'the necessary equipment is implied, not enumerated in the text.')
        recommendation = ('Retain the existing equipment value as provisional '
                          'unless the reviewer verifies an explicit source requirement.')
        value = ''
    else:
        cues = ('groups of 3' in text and 'shoot' in text and 'rebound' in text)
        fact = ('Publisher specifies a three-player shooting activity but '
                'does not provide a numeric minimum playing-area dimension.')
        recommendation = ('Keep minimum space unresolved; do not present '
                          'a derived court size as publisher guidance.')
        value = ''

    status = ('source_section_cues_present'
              if section['status'] == 'heading_found' and cues
              else 'not_verified')
    exception_rows.append({
        'evidence_id': evidence_id,
        'drill_name': name,
        'field': field_name,
        'source_url': url,
        'heading_status': section['status'],
        'evidence_status': status,
        'source_fact_interpretation': fact if status == 'source_section_cues_present' else '',
        'review_recommendation': recommendation,
        'suggested_value': value if status == 'source_section_cues_present' else '',
        'human_review_status': 'pending',
        'permission_review': 'pending',
    })
    exception_lookup[(evidence_id, field_name)] = exception_rows[-1]

# Build a practical editable decision packet without modifying the original sheets.
review_sheet = []
for row in fields:
    key = (row['evidence_id'], row['field'])
    special = exception_lookup.get(key)
    evidence_class = row['review_evidence_class']
    if special:
        if special['evidence_status'] == 'source_section_cues_present':
            basis = 'targeted_publisher_section_screened_pending_human_review'
            recommended_action = special['review_recommendation']
            suggested_value = special['suggested_value']
        else:
            basis = 'source_exception_still_unverified'
            recommended_action = 'Recheck publisher instructions before selecting a final value.'
            suggested_value = ''
    else:
        basis = evidence_class
        suggested_value = ''
        if evidence_class == 'exact_phrase_seen_pending_signoff':
            recommended_action = ('Confirm the prescription is a duration, repetition, '
                                  'shot or game target, then sign off its meaning.')
        elif evidence_class == 'keyword_evidence_pending_source_review':
            recommended_action = 'Read the cited source section and verify the full requirement.'
        elif evidence_class == 'activity_implied_requires_source_review':
            recommended_action = ('Identify whether the requirement is stated or inferred; '
                                  'retain uncertainty where appropriate.')
        else:
            recommended_action = 'Review source evidence and retain uncertainty if unresolved.'
    review_sheet.append({
        'priority': row['priority'],
        'candidate_group_id': row['candidate_group_id'],
        'evidence_id': row['evidence_id'],
        'drill_name': row['drill_name'],
        'field': row['field'],
        'current_catalog_value': row['current_catalog_value'],
        'current_registry_value': row['current_registry_value'],
        'proposed_value': row['proposed_value'],
        'targeted_source_suggestion': suggested_value,
        'source_review_basis': basis,
        'source_url': row['source_url'],
        'review_guidance': recommended_action,
        'decision_options': 'accept_proposal;retain_current;revise_value;defer',
        'reviewer_decision': 'pending',
        'final_verified_value': '',
        'reviewer': '',
        'review_date': '',
        'review_notes': '',
        'source_permission_review': 'pending',
        'catalog_approval_status': 'pending',
        'planner_eligible': 'no',
    })

candidate_sheet = []
for row in signoffs:
    gid = row['candidate_group_id']
    selected = [r for r in review_sheet if r['candidate_group_id'] == gid]
    special = [r['field'] for r in selected
               if (r['evidence_id'], r['field']) in exception_lookup]
    candidate_sheet.append({
        'priority': row['priority'],
        'candidate_group_id': gid,
        'evidence_id': row['evidence_id'],
        'drill_name': row['drill_name'],
        'source_url': row['source_url'],
        'fields_to_review': 'equipment;participants;space;prescription',
        'targeted_exception_fields': ';'.join(special),
        'field_reviews_pending': 4,
        'field_reviews_signed': 0,
        'source_instruction_signoff': 'pending',
        'canonical_identity_signoff': 'pending',
        'source_permission_review': 'pending',
        'youth_planning_review': 'pending',
        'catalog_approval_status': 'pending',
        'planner_eligible': 'no',
        'reviewer': '',
        'review_date': '',
        'review_notes': '',
    })

outfiles = {
    results / 'basketball_wave2_targeted_source_exceptions_v0_3_4.csv': exception_rows,
    results / 'basketball_wave2_reviewer_adjudication_v0_3_4.csv': review_sheet,
    results / 'basketball_wave2_candidate_review_tracker_v0_3_4.csv': candidate_sheet,
}


def preserve_or_create(path, rows, key_fields):
    if path.exists():
        old, _ = read_csv(path)
        old_keys = {tuple(r.get(k, '') for k in key_fields) for r in old}
        new_keys = {tuple(r.get(k, '') for k in key_fields) for r in rows}
        if old_keys != new_keys or len(old) != len(rows):
            raise ValueError(f'Existing review file differs; inspect before continuing: {path}')
        print('PRESERVED', path.relative_to(root))
        return 'preserved'
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open('x', newline='', encoding='utf-8') as file:
        writer = csv.DictWriter(file, fieldnames=list(rows[0]))
        writer.writeheader()
        writer.writerows(rows)
    print('CREATED  ', path.relative_to(root))
    return 'created'

print('\nTARGETED OFFICIAL SOURCE RESULTS')
for row in exception_rows:
    print(f'  {row["evidence_id"]} | {row["field"]:<12} | {row["evidence_status"]}')

print('\nREVIEW PACKET OUTPUTS')
for path, rows in outfiles.items():
    keys = (['evidence_id', 'field'] if 'exceptions' in path.name
            else ['candidate_group_id', 'evidence_id', 'field']
            if 'adjudication' in path.name else ['candidate_group_id'])
    preserve_or_create(path, rows, keys)

summary_path = results / 'basketball_step12_43_summary_v0_3_4.json'
summary = {
    'step': '12-43',
    'catalog_version_read': '0.3.4',
    'catalog_candidates': len(catalog),
    'source_occurrences': len(registry),
    'wave2_candidates': len(candidate_sheet),
    'reviewer_field_decisions': len(review_sheet),
    'targeted_exceptions': len(exception_rows),
    'targeted_exception_cues_supported': sum(
        r['evidence_status'] == 'source_section_cues_present'
        for r in exception_rows
    ),
    'review_decisions_signed': 0,
    'catalog_approvals': 0,
    'planner_eligible': 0,
    'upstream_catalog_modified': False,
}
if summary_path.exists():
    with summary_path.open(encoding='utf-8') as file:
        previous = json.load(file)
    if previous['catalog_candidates'] != 144 or previous['reviewer_field_decisions'] != 56:
        raise ValueError('Existing summary has different scope; inspect before continuing.')
    print('PRESERVED', summary_path.relative_to(root))
else:
    with summary_path.open('x', encoding='utf-8') as file:
        json.dump(summary, file, indent=2)
        file.write('\n')
    print('CREATED  ', summary_path.relative_to(root))

print('\nSUMMARY')
print(f'Field decisions staged / signed: {len(review_sheet)} / 0')
print(f'Candidate signoffs pending: {len(candidate_sheet)}')
print(f'Catalog / source records unchanged: {len(catalog)} / {len(registry)}')
print('Approved / planner-eligible: 0 / 0')
print('Git operations performed: None')
print('OVERALL: PASS — targeted source evidence recorded; human decisions remain pending')
```

```python
# Step 12-45 - Apply Confirmed Wave 2 Metadata Decisions
# Preserve source lineage, retain unresolved values, and maintain approval restrictions.

import csv
import hashlib
import json
from collections import Counter, defaultdict
from pathlib import Path


PROJECT_ROOT = next(
    (path for path in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (path / '.git').exists()),
    None,
)
if PROJECT_ROOT is None:
    raise FileNotFoundError('Cannot locate the Git project root.')

DATA_DIR = PROJECT_ROOT / 'data' / 'processed'
RESULTS_DIR = PROJECT_ROOT / 'results'
VERSION = '0.3.5'
DECISION_DATE = '2026-11-07'
DECISION_PROVENANCE = 'user_confirmed_step_12_44_recommendations'

INPUTS = {
    'catalog': DATA_DIR / 'basketball_drill_catalog_v0_3_4_REVIEW_ONLY.csv',
    'registry': RESULTS_DIR / 'basketball_drill_source_registry_v0_3_3.csv',
    'recommendations': RESULTS_DIR / 'Step_12_44_Field_Recommendations_REVIEW_ONLY.csv',
}
SUMMARY_ALTERNATIVES = [
    RESULTS_DIR / 'Step12_44_Candidate_Review_Summary.csv',
    RESULTS_DIR / 'Step_12_44_Candidate_Review_Summary_REVIEW_ONLY.csv',
]
summary_files = [p for p in SUMMARY_ALTERNATIVES if p.exists()]
if not summary_files:
    raise FileNotFoundError(
        'Save the downloaded Step 12-44 Candidate Review Summary CSV in results/ first.'
    )
INPUTS['candidate_summary'] = summary_files[0]

OUTPUTS = {
    'catalog': DATA_DIR / 'basketball_drill_catalog_v0_3_5_REVIEW_ONLY.csv',
    'registry': RESULTS_DIR / 'basketball_drill_source_registry_v0_3_5.csv',
    'field_decisions': RESULTS_DIR / 'basketball_wave2_confirmed_field_decisions_v0_3_5.csv',
    'candidate_decisions': RESULTS_DIR / 'basketball_wave2_confirmed_candidate_decisions_v0_3_5.csv',
    'change_log': RESULTS_DIR / 'basketball_wave2_metadata_change_log_v0_3_5.csv',
    'summary': RESULTS_DIR / 'basketball_step12_45_summary_v0_3_5.json',
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(f'Missing input: {path}')
    with path.open('r', encoding='utf-8-sig', newline='') as file:
        reader = csv.DictReader(file)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or not rows:
        raise ValueError(f'Empty CSV: {path}')
    return rows, columns


def require_columns(actual, required, label):
    absent = set(required) - set(actual)
    if absent:
        raise ValueError(f'{label} is missing required columns: {sorted(absent)}')


def unique_index(rows, key, label):
    values = [row[key] for row in rows]
    if len(values) != len(set(values)) or not all(values):
        raise ValueError(f'{label}: missing or duplicate {key} values.')
    return {row[key]: row for row in rows}


def write_csv(path, rows, columns):
    if not rows:
        raise ValueError(f'Cannot write empty output: {path.name}')
    with path.open('x', encoding='utf-8', newline='') as file:
        writer = csv.DictWriter(file, fieldnames=columns, extrasaction='raise')
        writer.writeheader()
        writer.writerows(rows)


def sha256(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


print('=' * 72)
print('STEP 12-45 — CONFIRMED WAVE 2 METADATA RECONCILIATION')
print('=' * 72)

# Existing outputs are never replaced; partial output sets require inspection.
existing = {label: path for label, path in OUTPUTS.items() if path.exists()}
if existing:
    if len(existing) != len(OUTPUTS):
        raise FileExistsError(
            'Some Step 12-45 outputs already exist. Review the partial output set '
            'before rerunning; no files were modified.'
        )
    saved = json.loads(OUTPUTS['summary'].read_text(encoding='utf-8'))
    for label, path in OUTPUTS.items():
        if not path.is_file() or not path.stat().st_size:
            raise ValueError(f'Existing output is missing or empty: {path}')
        if label != 'summary' and sha256(path) != saved['output_sha256'][label]:
            raise ValueError(f'Existing output changed after creation: {path.name}')
    print('Existing Step 12-45 outputs verified and PRESERVED.')
    print('Catalog candidates:', saved['catalog_candidates'])
    print('Source occurrences:', saved['registry_source_records'])
    print('Decisions:', saved['decision_counts'])
    print('Actual field changes:', saved['actual_field_changes'])
    print('Approved / planner eligible: 0 / 0')
    print('OVERALL: PASS — original Step 12-45 outputs preserved')
else:
    inputs_hash_before = {key: sha256(path) for key, path in INPUTS.items()}
    catalog, catalog_columns = read_csv(INPUTS['catalog'])
    registry, registry_columns = read_csv(INPUTS['registry'])
    recommendations, rec_columns = read_csv(INPUTS['recommendations'])
    candidate_summary, candidate_columns = read_csv(INPUTS['candidate_summary'])

    require_columns(catalog_columns, [
        'catalog_candidate_id', 'approved_drill_id', 'drill_name',
        'evidence_ids', 'source_prescription', 'equipment', 'participants', 'space',
        'catalog_version', 'catalog_approval_status', 'planner_eligible',
    ], 'Catalog')
    require_columns(registry_columns, [
        'evidence_id', 'catalog_candidate_id', 'source_prescription',
        'equipment', 'participants', 'space',
    ], 'Source registry')
    require_columns(rec_columns, [
        'candidate_group_id', 'evidence_id', 'drill_name', 'field',
        'current_catalog_value', 'current_registry_value',
        'recommended_decision', 'recommended_value_if_accepted',
        'reviewer_decision', 'human_signoff_status',
        'independent_source_basis', 'independent_source_note', 'source_url',
    ], 'Recommendations')
    require_columns(candidate_columns, [
        'candidate_group_id', 'evidence_id', 'drill_name',
        'recommend_accept_proposal', 'recommend_retain_current',
        'recommend_revise_value', 'recommend_defer',
    ], 'Candidate review summary')

    if len(catalog) != 144 or len(registry) != 156:
        raise ValueError('Catalog or source-registry counts differ from verified baseline.')
    if len(recommendations) != 56 or len(candidate_summary) != 14:
        raise ValueError('Unexpected Step 12-44 review package size.')
    if {row['catalog_version'] for row in catalog} != {'0.3.4'}:
        raise ValueError('Expected a v0.3.4 source catalog.')

    catalog_by_id = unique_index(catalog, 'catalog_candidate_id', 'Catalog')
    registry_by_id = unique_index(registry, 'evidence_id', 'Source registry')
    summary_by_id = unique_index(candidate_summary, 'candidate_group_id', 'Review summary')
    if any(
        row['catalog_approval_status'] != 'pending'
        or row['planner_eligible'] != 'no'
        or row['approved_drill_id'].strip()
        for row in catalog
    ):
        raise ValueError('Existing catalog contains approval decisions; manual review required.')

    fields = {'equipment', 'participants', 'space', 'prescription'}
    decisions = {'accept_proposal', 'retain_current', 'revise_value', 'defer'}
    field_map = {'prescription': 'source_prescription',
                 'equipment': 'equipment', 'participants': 'participants', 'space': 'space'}
    counts_expected = {'accept_proposal': 27, 'retain_current': 26,
                       'revise_value': 2, 'defer': 1}
    counts = Counter(row['recommended_decision'] for row in recommendations)
    if dict(counts) != counts_expected:
        raise ValueError(f'Recommendation counts differ from agreed decisions: {dict(counts)}')

    grouped = defaultdict(list)
    recommendation_keys = set()
    for row in recommendations:
        group_id, evidence_id, field = (
            row['candidate_group_id'], row['evidence_id'], row['field']
        )
        key = (group_id, evidence_id, field)
        if key in recommendation_keys:
            raise ValueError(f'Repeated field-review key: {key}')
        recommendation_keys.add(key)
        if field not in fields or row['recommended_decision'] not in decisions:
            raise ValueError(f'Unexpected field or decision: {key}')
        if row['reviewer_decision'] != 'pending' or row['human_signoff_status'] != 'pending':
            raise ValueError(f'Previous reviewer decision detected: {key}')
        if group_id not in catalog_by_id or evidence_id not in registry_by_id:
            raise ValueError(f'Unknown catalog or source reference: {key}')
        item = catalog_by_id[group_id]
        source = registry_by_id[evidence_id]
        if source['catalog_candidate_id'] != group_id:
            raise ValueError(f'Incorrect source-to-catalog relationship: {key}')
        # This batch contains single-source groups; do not spread a source-specific
        # metadata decision to other source occurrences without reconciliation.
        if item['evidence_ids'].split(';') != [evidence_id]:
            raise ValueError(f'Multi-source candidate needs separate review: {group_id}')
        if item['drill_name'] != row['drill_name']:
            raise ValueError(f'Drill name changed for {key}')
        col = field_map[field]
        if item[col] != row['current_catalog_value']:
            raise ValueError(
                f'Catalog {group_id}/{field} differs from the reviewed value. '
                f'Expected {row["current_catalog_value"]!r}; found {item[col]!r}'
            )
        if source[col] != row['current_registry_value']:
            raise ValueError(
                f'Registry {evidence_id}/{field} differs from the reviewed value. '
                f'Expected {row["current_registry_value"]!r}; found {source[col]!r}'
            )
        if row['recommended_decision'] in {'accept_proposal', 'revise_value'} \
                and not row['recommended_value_if_accepted'].strip():
            raise ValueError(f'Missing accepted/revised value for {key}')
        if row['recommended_decision'] == 'defer' and row['recommended_value_if_accepted'].strip():
            raise ValueError(f'Deferred field unexpectedly has replacement value: {key}')
        grouped[group_id].append(row)

    if set(grouped) != set(summary_by_id) or len(grouped) != 14:
        raise ValueError('Field decisions do not match the 14 candidate summaries.')
    for group_id, rows in grouped.items():
        if {row['field'] for row in rows} != fields:
            raise ValueError(f'Expected four distinct fields: {group_id}')
        info = summary_by_id[group_id]
        expected = {
            'accept_proposal': int(info['recommend_accept_proposal']),
            'retain_current': int(info['recommend_retain_current']),
            'revise_value': int(info['recommend_revise_value']),
            'defer': int(info['recommend_defer']),
        }
        actual = Counter(row['recommended_decision'] for row in rows)
        if any(actual.get(k, 0) != v for k, v in expected.items()):
            raise ValueError(f'Candidate summary and recommendation mismatch: {group_id}')

    # Only reviewed Wave 2 metadata may change. All candidate identity, source
    # linkage, approval, and permission fields remain unchanged.
    review_columns = [
        'wave2_user_decision_status', 'wave2_user_decision_date',
        'wave2_deferred_fields', 'wave2_metadata_decision_reference',
    ]
    for col in review_columns:
        if col in catalog_columns or col in registry_columns:
            raise ValueError(f'Output review field already exists: {col}')
        for item in catalog:
            item[col] = ''
        for item in registry:
            item[col] = ''
    new_catalog_columns = catalog_columns + review_columns
    new_registry_columns = registry_columns + review_columns

    reviewed_rows = []
    change_log = []
    for row in recommendations:
        group_id, evidence_id, field = (
            row['candidate_group_id'], row['evidence_id'], row['field']
        )
        col = field_map[field]
        cat = catalog_by_id[group_id]
        reg = registry_by_id[evidence_id]
        old_cat = cat[col]
        old_reg = reg[col]
        decision = row['recommended_decision']
        replacement = row['recommended_value_if_accepted']
        applied = decision in ('accept_proposal', 'revise_value')
        # Retain and defer ALWAYS leave their source and catalog values untouched.
        if applied:
            cat[col] = replacement
            reg[col] = replacement
        else:
            if cat[col] != old_cat or reg[col] != old_reg:
                raise AssertionError('Retained or deferred field was altered.')
        changed_cat = cat[col] != old_cat
        changed_reg = reg[col] != old_reg

        reviewed_rows.append({
            'candidate_group_id': group_id,
            'evidence_id': evidence_id,
            'drill_name': row['drill_name'],
            'field': field,
            'user_confirmed_decision': decision,
            'previous_catalog_value': old_cat,
            'previous_registry_value': old_reg,
            'recommended_value': replacement,
            'resulting_catalog_value': cat[col],
            'resulting_registry_value': reg[col],
            'catalog_value_changed': 'yes' if changed_cat else 'no',
            'registry_value_changed': 'yes' if changed_reg else 'no',
            'evidence_basis': row['independent_source_basis'],
            'source_url': row['source_url'],
            'source_review_note': row['independent_source_note'],
            'user_decision_record': DECISION_PROVENANCE,
            'user_decision_date': DECISION_DATE,
            'source_instruction_signoff': 'pending',
            'team_signoff': 'pending',
            'permission_review': 'pending',
            'youth_planning_review': 'pending',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
        })
        if changed_cat or changed_reg:
            change_log.append({
                'candidate_group_id': group_id,
                'evidence_id': evidence_id,
                'drill_name': row['drill_name'],
                'field': field,
                'decision': decision,
                'catalog_before': old_cat,
                'catalog_after': cat[col],
                'registry_before': old_reg,
                'registry_after': reg[col],
                'evidence_basis': row['independent_source_basis'],
                'source_url': row['source_url'],
                'change_date': DECISION_DATE,
                'review_status': 'user_confirmed_metadata_pending_other_signoffs',
            })

    candidate_decisions = []
    for group_id, rows in grouped.items():
        item = catalog_by_id[group_id]
        source = registry_by_id[rows[0]['evidence_id']]
        outstanding = ';'.join(sorted(
            r['field'] for r in rows if r['recommended_decision'] == 'defer'
        ))
        status = (
            'user_confirmed_with_deferred_metadata_field'
            if outstanding else 'user_confirmed_metadata_pending_other_signoffs'
        )
        for entry in (item, source):
            entry['wave2_user_decision_status'] = status
            entry['wave2_user_decision_date'] = DECISION_DATE
            entry['wave2_deferred_fields'] = outstanding
            entry['wave2_metadata_decision_reference'] = (
                'results/basketball_wave2_confirmed_field_decisions_v0_3_5.csv'
            )
        selections = Counter(r['recommended_decision'] for r in rows)
        candidate_decisions.append({
            'candidate_group_id': group_id,
            'evidence_id': rows[0]['evidence_id'],
            'drill_name': item['drill_name'],
            'accept_proposal': selections['accept_proposal'],
            'retain_current': selections['retain_current'],
            'revise_value': selections['revise_value'],
            'defer': selections['defer'],
            'deferred_fields': outstanding,
            'metadata_decision_status': status,
            'source_instruction_signoff': 'pending',
            'canonical_identity_signoff': 'pending',
            'team_signoff': 'pending',
            'permission_review': 'pending',
            'youth_planning_review': 'pending',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
            'decision_provenance': DECISION_PROVENANCE,
            'decision_date': DECISION_DATE,
        })

    for item in catalog:
        item['catalog_version'] = VERSION

    if len(catalog) != 144 or len(registry) != 156 or len(reviewed_rows) != 56 \
            or len(candidate_decisions) != 14:
        raise AssertionError('An output record count changed unexpectedly.')
    if sum(r['user_confirmed_decision'] == 'defer' for r in reviewed_rows) != 1:
        raise AssertionError('Deferred decision was not preserved.')
    if not all(
        r['catalog_approval_status'] == 'pending' and r['planner_eligible'] == 'no'
        and not r['approved_drill_id'].strip()
        for r in catalog
    ):
        raise AssertionError('An unapproved candidate became eligible.')
    if set(row['evidence_id'] for row in registry) != set(registry_by_id):
        raise AssertionError('Source evidence identifiers changed.')

    # Integrity verification: compare every cell to its original value except
    # for precisely authorized metadata edits, version, and new review columns.
    original_catalog, _ = read_csv(INPUTS['catalog'])
    original_registry, _ = read_csv(INPUTS['registry'])
    authorized_changes = {
        (r['candidate_group_id'], field_map[r['field']])
        for r in reviewed_rows if r['catalog_value_changed'] == 'yes'
    }
    authorized_source_changes = {
        (r['evidence_id'], field_map[r['field']])
        for r in reviewed_rows if r['registry_value_changed'] == 'yes'
    }
    for before, after in zip(original_catalog, catalog):
        for col in catalog_columns:
            if col == 'catalog_version' or (before['catalog_candidate_id'], col) in authorized_changes:
                continue
            if before[col] != after[col]:
                raise AssertionError(f'Unexpected catalog edit: {before["catalog_candidate_id"]}/{col}')
    for before, after in zip(original_registry, registry):
        for col in registry_columns:
            if (before['evidence_id'], col) in authorized_source_changes:
                continue
            if before[col] != after[col]:
                raise AssertionError(f'Unexpected registry edit: {before["evidence_id"]}/{col}')

    # Write all artifacts only after successful in-memory validation. No input
    # files, earlier versions, or previously completed reviewer sheets are edited.
    for path in OUTPUTS.values():
        path.parent.mkdir(parents=True, exist_ok=True)
    created = []
    try:
        specifications = (
            ('catalog', catalog, new_catalog_columns),
            ('registry', registry, new_registry_columns),
            ('field_decisions', reviewed_rows, list(reviewed_rows[0])),
            ('candidate_decisions', candidate_decisions, list(candidate_decisions[0])),
            ('change_log', change_log, list(change_log[0]) if change_log else []),
        )
        for key, rows, cols in specifications:
            # Include a header-only change log when no values differ from source.
            if key == 'change_log' and not rows:
                cols = ['candidate_group_id', 'evidence_id', 'drill_name', 'field',
                        'decision', 'catalog_before', 'catalog_after',
                        'registry_before', 'registry_after', 'evidence_basis',
                        'source_url', 'change_date', 'review_status']
                with OUTPUTS[key].open('x', encoding='utf-8', newline='') as f:
                    csv.DictWriter(f, fieldnames=cols).writeheader()
            else:
                write_csv(OUTPUTS[key], rows, cols)
            created.append(OUTPUTS[key])
        summary = {
            'step': '12-45',
            'version': VERSION,
            'catalog_candidates': len(catalog),
            'registry_source_records': len(registry),
            'wave2_candidates': len(candidate_decisions),
            'field_decisions': len(reviewed_rows),
            'decision_counts': dict(counts),
            'actual_field_changes': len(change_log),
            'catalog_fields_changed': sum(r['catalog_value_changed'] == 'yes' for r in reviewed_rows),
            'registry_fields_changed': sum(r['registry_value_changed'] == 'yes' for r in reviewed_rows),
            'deferred_field': {
                'evidence_id': 'JR-EVID-0059', 'field': 'space',
                'status': 'exact_minimum_space_pending_review',
            },
            'metadata_review_confirmation': DECISION_PROVENANCE,
            'confirmation_date': DECISION_DATE,
            'team_signoff': 'pending',
            'canonical_signoff': 'pending',
            'permissions': 'pending',
            'youth_planning_review': 'pending',
            'approved_count': 0,
            'planner_eligible_count': 0,
            'input_sha256': inputs_hash_before,
            'output_sha256': {
                key: sha256(OUTPUTS[key]) for key in OUTPUTS if key != 'summary'
            },
        }
        with OUTPUTS['summary'].open('x', encoding='utf-8') as f:
            json.dump(summary, f, indent=2)
            f.write('\n')
        created.append(OUTPUTS['summary'])
    except Exception:
        for path in created:
            if path.exists():
                path.unlink()
        raise

    for key, path in INPUTS.items():
        if sha256(path) != inputs_hash_before[key]:
            raise AssertionError(f'Input file changed unexpectedly: {key}')
    if json.loads(OUTPUTS['summary'].read_text(encoding='utf-8'))['output_sha256'] != {
        key: sha256(OUTPUTS[key]) for key in OUTPUTS if key != 'summary'
    }:
        raise AssertionError('Saved output checksums do not match summary.')

    print('Catalog candidates / registered sources:', len(catalog), '/', len(registry))
    print('Wave 2 candidates / reviewed fields:', len(candidate_decisions), '/', len(reviewed_rows))
    print('\nCONFIRMED METADATA DECISIONS')
    for name in ('accept_proposal', 'retain_current', 'revise_value', 'defer'):
        print(f'  {name:<18}: {counts[name]}')
    print('\nACTUAL FIELD EDITS')
    print('  Changed catalog fields:', summary['catalog_fields_changed'])
    print('  Changed source-registry fields:', summary['registry_fields_changed'])
    print('  Deferred: JR-EVID-0059 / 3-Person Shooting / space')
    print('\nOUTPUTS')
    for key, path in OUTPUTS.items():
        print('  CREATED', path.relative_to(PROJECT_ROOT))
    print('\nTeam signoff / permissions / youth rules: PENDING')
    print('Approved / planner-eligible: 0 / 0')
    print('Input and earlier catalog files overwritten: No')
    print('Git operations performed: None')
    print('OVERALL: PASS — user-confirmed metadata decisions recorded in v0.3.5; final approvals pending')
```

```python
# Step 12-46 - Wave 3 Publisher Verification for Draft Basketball Drill Metadata
# Examine 42 draft-metadata candidates while preserving confirmed Wave 2 metadata decisions and all approval gates.

import csv
import hashlib
import io
import json
import re
import tempfile
import unicodedata
from collections import Counter, defaultdict
from datetime import date
from difflib import SequenceMatcher
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

root = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if root is None:
    raise FileNotFoundError('Project root was not found.')

inputs = {
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_5_REVIEW_ONLY.csv',
    'registry': root / 'results/basketball_drill_source_registry_v0_3_5.csv',
    'readiness': root / 'results/basketball_catalog_readiness_worklist_v0_3_1.csv',
}
outputs = {
    'source_checks': root / 'results/basketball_wave3_publisher_checks_v0_3_5.csv',
    'field_comparisons': root / 'results/basketball_wave3_field_comparisons_v0_3_5.csv',
    'review_queue': root / 'results/basketball_wave3_review_queue_v0_3_5.csv',
    'coverage': root / 'results/eda/basketball_wave3_source_coverage_v0_3_5.csv',
    'summary': root / 'results/basketball_step12_46_summary_v0_3_5.json',
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or len(columns) != len(set(columns)):
        raise ValueError(f'Invalid CSV column names: {path.name}')
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f'Malformed CSV rows: {path.name}')
    return rows


def indexed(rows, key, count):
    if len(rows) != count:
        raise ValueError(f'Unexpected {key} count: {len(rows)}; expected {count}')
    keys = [row[key].strip() for row in rows]
    if not all(keys) or len(keys) != len(set(keys)):
        raise ValueError(f'Blank or duplicate {key} identifiers')
    return {row[key]: row for row in rows}


def sha(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def norm(value):
    value = unicodedata.normalize('NFKC', value or '').casefold()
    return re.sub(r'[^a-z0-9]+', ' ', value).strip()


def csv_payload(rows):
    if not rows:
        raise ValueError('Cannot save an empty report.')
    stream = io.StringIO(newline='')
    writer = csv.DictWriter(stream, fieldnames=list(rows[0]),
                            lineterminator='\n', extrasaction='raise')
    writer.writeheader()
    writer.writerows(rows)
    return stream.getvalue().encode('utf-8')


class PracticeSectionParser(HTMLParser):
    """Read heading-scoped sections in memory; never save source instructions."""

    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.skip_tag = None
        self.skip_depth = 0
        self.heading_tag = None
        self.heading_parts = []
        self.current = None

    def handle_starttag(self, tag, attrs):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth += 1
            return
        if tag in {'script', 'style', 'noscript', 'svg'}:
            self.skip_tag, self.skip_depth = tag, 1
        elif re.fullmatch(r'h[1-6]', tag):
            self.heading_tag, self.heading_parts = tag, []
        elif self.current is not None and tag in {'p', 'li', 'br', 'div', 'td', 'tr'}:
            self.current['parts'].append(' ')

    def handle_data(self, data):
        if self.skip_depth:
            return
        if self.heading_tag:
            self.heading_parts.append(data)
        elif self.current is not None:
            self.current['parts'].append(data)

    def handle_endtag(self, tag):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth -= 1
                if not self.skip_depth:
                    self.skip_tag = None
            return
        if tag == self.heading_tag:
            heading = norm(' '.join(self.heading_parts))
            if heading:
                self.current = {'heading': heading, 'parts': []}
                self.sections.append(self.current)
            self.heading_tag, self.heading_parts = None, []


def official_page(url):
    parsed = urlsplit(url)
    return (parsed.scheme == 'https' and parsed.hostname == 'jr.nba.com'
            and parsed.port is None and not parsed.query and not parsed.fragment
            and re.fullmatch(
                r'/basketball-practice-plans/(rookie|starter|all-star|mvp)/(page/\d+/)?',
                parsed.path) is not None)


def fetch(url):
    if not official_page(url):
        raise ValueError(f'Non-official or unexpected source URL: {url}')
    try:
        request = Request(url, headers={
            'User-Agent': 'Mozilla/5.0 (compatible; AcademicSourceAudit/1.0)',
            'Accept': 'text/html',
        })
        with urlopen(request, timeout=25) as response:
            final_url = response.geturl()
            if not official_page(final_url):
                return {'status': 'unexpected_redirect', 'sections': []}
            html = response.read(2_000_000).decode('utf-8', errors='replace')
        parser = PracticeSectionParser()
        parser.feed(html)
        return {'status': 'fetched', 'sections': parser.sections}
    except (HTTPError, URLError, TimeoutError, OSError, ValueError):
        return {'status': 'unavailable', 'sections': []}


CUES = {
    'equipment': {
        'basketball': r'\b(?:basketball|basketballs|ball|balls)\b',
        'hoop': r'\b(?:hoop|hoops|basket|baskets|rim|rims)\b',
        'cones': r'\b(?:cone|cones)\b',
        'markers': r'\b(?:marker|markers)\b',
        'chair': r'\b(?:chair|chairs)\b',
    },
    'participants': {
        'coach': r'\b(?:coach|coaches)\b',
        'partner_or_pair': r'\b(?:partner|partners|pair|pairs)\b',
        'defender': r'\b(?:defender|defenders|defense|defensive)\b',
        'team': r'\b(?:team|teams)\b',
        'group': r'\b(?:group|groups)\b',
        'player': r'\b(?:player|players)\b',
    },
    'space': {
        'full_court': r'\bfull court\b',
        'half_court': r'\bhalf court\b',
        'baseline': r'\b(?:baseline|baselines)\b',
        'sideline': r'\b(?:sideline|sidelines)\b',
        'three_point_line': r'\bthree point (?:line|arc)\b',
        'wing': r'\b(?:wing|wings)\b',
        'key_or_paint': r'\b(?:key|paint|lane)\b',
        'court': r'\b(?:court|courts)\b',
        'gym': r'\b(?:gym|gymnasium)\b',
    },
    'prescription': {
        'minutes': r'\b(?:minutes|minute|mins|min)\b',
        'seconds': r'\b(?:seconds|second|secs|sec)\b',
        'repetitions': r'\b(?:repetitions|repetition|reps|rep|times)\b',
        'makes': r'\b(?:makes|made)\b',
        'games': r'\b(?:game|games)\b',
        'shots': r'\b(?:shot|shots)\b',
    },
}


def labels(text, field):
    return ';'.join(key for key, pattern in CUES[field].items()
                    if re.search(pattern, text))


def time_screen(recorded, text):
    expected = norm(recorded)
    if not text:
        return 'source_section_unavailable'
    if not expected or expected in {'not stated', 'not specified'}:
        return 'recorded_prescription_unspecified'
    if expected in text:
        return 'exact_recorded_phrase_present'
    numbers = set(re.findall(r'\b\d+\b', expected))
    present = set(re.findall(r'\b\d+\b', text))
    if numbers and numbers <= present:
        return 'all_numbers_present_context_unverified'
    if numbers & present:
        return 'some_numbers_present_context_unverified'
    return 'recorded_prescription_not_confirmed'


catalog = read_csv(inputs['catalog'])
registry = read_csv(inputs['registry'])
readiness = read_csv(inputs['readiness'])
cat_by_id = indexed(catalog, 'catalog_candidate_id', 144)
reg_by_id = indexed(registry, 'evidence_id', 156)
ready_by_id = indexed(readiness, 'candidate_group_id', 144)

if set(cat_by_id) != set(ready_by_id):
    raise ValueError('Readiness and catalog candidate IDs differ.')
if any(row['catalog_approval_status'] != 'pending'
       or row['planner_eligible'] != 'no' or row['approved_drill_id'].strip()
       for row in catalog):
    raise ValueError('One or more existing approval decisions have changed.')
if any(row['catalog_candidate_id'] not in cat_by_id for row in registry):
    raise ValueError('Source registry contains an unknown catalog candidate.')

selected = sorted([row for row in readiness if row['review_wave'] == '3'],
                  key=lambda row: int(row['review_rank']))
if len(selected) != 42:
    raise ValueError(f'Wave 3 expected 42 candidate groups; found {len(selected)}.')

sources_by_candidate = defaultdict(list)
for source in registry:
    sources_by_candidate[source['catalog_candidate_id']].append(source)

for row in selected:
    cid = row['candidate_group_id']
    item = cat_by_id[cid]
    sources = sources_by_candidate[cid]
    if row['metadata_stage'] != 'draft_metadata_requires_source_verification':
        raise ValueError(f'Unexpected review stage: {cid}')
    if item['metadata_status'] != 'draft_pending_verification':
        raise ValueError(f'Catalog review stage changed: {cid}')
    if len(sources) != 1 or sources[0]['origin'] != 'original_review_batches':
        raise ValueError(f'Wave 3 requires one historical source occurrence: {cid}')
    if norm(sources[0]['drill_name']) != norm(item['drill_name']):
        raise ValueError(f'Catalog and registry drill name differ: {cid}')
    if row['outstanding_overlap_case_ids'].strip():
        raise ValueError(f'Unexpected unresolved overlap in Wave 3: {cid}')
    if not official_page(sources[0]['source_url']):
        raise ValueError(f'Invalid official source: {cid}')

# Step 12-45 has already recorded the confirmed Wave 2 changes.
# Verify that the current v0.3.5 inputs are exactly those validated outputs.
previous_summary_path = root / 'results/basketball_step12_45_summary_v0_3_5.json'
if not previous_summary_path.is_file():
    raise FileNotFoundError(f'Missing Step 12-45 integrity summary: {previous_summary_path}')
previous_summary = json.loads(previous_summary_path.read_text(encoding='utf-8'))
for key in ('catalog', 'registry'):
    expected_digest = previous_summary.get('output_sha256', {}).get(key)
    if not expected_digest or sha(inputs[key]) != expected_digest:
        raise ValueError(f'Current {key} input does not match Step 12-45 output checksum.')
if int(previous_summary.get('actual_field_changes', 0)) != 29:
    raise ValueError('Step 12-45 verified change count differs from 29.')

input_hashes = {name: sha(path) for name, path in inputs.items()}
print('=' * 72)
print('STEP 12-46 — WAVE 3 PUBLISHER METADATA VERIFICATION')
print('=' * 72)
print('Catalog candidates:', len(catalog))
print('Selected draft-metadata candidates:', len(selected))

exists = [path.exists() for path in outputs.values()]
if any(exists):
    if not all(exists):
        raise FileExistsError('Partial Step 12-46 outputs exist. Inspect before rerunning.')
    summary = json.loads(outputs['summary'].read_text(encoding='utf-8'))
    if summary.get('input_sha256') != input_hashes:
        raise ValueError('Input files changed after Step 12-46; outputs preserved.')
    for key in ('source_checks', 'field_comparisons', 'review_queue', 'coverage'):
        if sha(outputs[key]) != summary['report_sha256'].get(key):
            raise ValueError(f'Existing report changed: {outputs[key].name}')
    print('EXISTING OUTPUTS VERIFIED — no additional webpage requests.')
    for key, path in outputs.items():
        print(f'  VERIFIED {path.relative_to(root)}')
    print('Approved / planner-eligible: 0 / 0')
    print('OVERALL: PASS — saved Wave 3 publisher-review evidence preserved')
else:
    source_urls = sorted({sources_by_candidate[row['candidate_group_id']][0]['source_url']
                          for row in selected})
    print('Distinct official webpages:', len(source_urls))
    print('\nOFFICIAL PAGE ACCESS')
    pages = {}
    for i, url in enumerate(source_urls, 1):
        pages[url] = fetch(url)
        print(f"  {i:02d}/{len(source_urls):02d} | {pages[url]['status']:<19} | {url}")

    if not any(page['status'] == 'fetched' for page in pages.values()):
        raise RuntimeError('All official publisher pages were unavailable; no reports saved.')

    checks, fields, queue = [], [], []
    for row in selected:
        cid = row['candidate_group_id']
        item = cat_by_id[cid]
        source = sources_by_candidate[cid][0]
        page = pages[source['source_url']]
        expected_heading = norm(source['drill_name'])
        headings = [section['heading'] for section in page['sections']]
        exact = [section for section in page['sections']
                 if section['heading'] == expected_heading]
        suggestion = ''
        status = 'not_checked_source_unavailable'
        content = ''
        if page['status'] == 'fetched':
            if len(exact) == 1:
                status = 'exact_heading_found'
                content = norm(' '.join(exact[0]['parts']))
            elif len(exact) > 1:
                status = 'ambiguous_repeated_heading'
            else:
                ranked = sorted(((SequenceMatcher(None, expected_heading, h).ratio(), h)
                                 for h in set(headings)), reverse=True)
                if ranked and ranked[0][0] >= 0.83:
                    status = 'similar_heading_pending_review'
                    suggestion = ranked[0][1]
                else:
                    status = 'heading_not_found'

        prescription = time_screen(source['source_prescription'], content)
        source_cues = {field: labels(content, field) for field in CUES}
        check = {
            'candidate_group_id': cid,
            'evidence_id': source['evidence_id'],
            'drill_name': item['drill_name'],
            'curriculum_level': source['source_level'],
            'practice_plan': source['practice_plan'],
            'official_source_url': source['source_url'],
            'page_access_status': page['status'],
            'heading_screen': status,
            'similar_heading_suggestion': suggestion,
            'section_word_count': len(content.split()),
            'section_sha256': hashlib.sha256(content.encode('utf-8')).hexdigest() if content else '',
            'source_prescription_screen': prescription,
            'source_equipment_cues': source_cues['equipment'],
            'source_participant_cues': source_cues['participants'],
            'source_space_cues': source_cues['space'],
            'source_prescription_cues': source_cues['prescription'],
            'independent_instruction_signoff': 'pending',
            'independent_metadata_signoff': 'pending',
            'permission_review': 'pending',
            'planner_eligible': 'no',
        }
        checks.append(check)

        issues = []
        if status != 'exact_heading_found':
            issues.append('source_heading_needs_manual_review')
        if not content:
            issues.append('source_instruction_section_unavailable')
        if prescription != 'exact_recorded_phrase_present':
            issues.append('source_prescription_needs_manual_review')
        if source['space'].strip().casefold() in {'', 'see source practice plan', 'unknown', 'not specified'}:
            issues.append('registry_space_requires_source_confirmation')

        for field in ('equipment', 'participants', 'space', 'prescription'):
            registry_field = 'source_prescription' if field == 'prescription' else field
            fields.append({
                'candidate_group_id': cid,
                'evidence_id': source['evidence_id'],
                'drill_name': item['drill_name'],
                'field': field,
                'catalog_value': item['source_prescription'] if field == 'prescription' else item[field],
                'registry_value': source[registry_field],
                'source_section_cues': source_cues[field],
                'source_text_screen': (prescription if field == 'prescription'
                                       else 'cues_present_not_approval' if source_cues[field]
                                       else 'no_cues_detected_or_section_unavailable'),
                'independent_field_signoff': 'pending',
                'reviewer': '',
                'review_notes': '',
            })
            if field != 'prescription' and not source_cues[field]:
                issues.append(f'{field}_requires_source_review')

        if item['source_prescription'].strip() != source['source_prescription'].strip():
            issues.append('catalog_registry_prescription_wording_differs')
        if item['space'].strip() != source['space'].strip():
            issues.append('catalog_registry_space_wording_differs')

        queue.append({
            'review_rank': row['review_rank'],
            'candidate_group_id': cid,
            'evidence_id': source['evidence_id'],
            'drill_name': item['drill_name'],
            'review_wave': '3',
            'heading_screen': status,
            'prescription_screen': prescription,
            'registry_space_unspecified': 'yes' if 'registry_space_requires_source_confirmation' in issues else 'no',
            'review_issues': ';'.join(dict.fromkeys(issues)),
            'next_action': 'Compare the official drill section with all four recorded requirement fields; document reviewer decision and copyright permissions.',
            'instructions_review': 'pending',
            'requirements_review': 'pending',
            'distinctness_review': 'pending',
            'permission_review': 'pending',
            'youth_planning_review': 'pending',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
            'reviewer': '',
            'review_notes': '',
        })

    coverage = []
    for key in sorted({(r['curriculum_level'], r['practice_plan']) for r in checks}):
        group = [r for r in checks if (r['curriculum_level'], r['practice_plan']) == key]
        coverage.append({
            'curriculum_level': key[0],
            'practice_plan': key[1],
            'source_occurrences': len(group),
            'heading_exact_count': sum(r['heading_screen'] == 'exact_heading_found' for r in group),
            'sections_available': sum(bool(r['section_sha256']) for r in group),
            'recorded_prescription_exact_text_count': sum(r['source_prescription_screen'] == 'exact_recorded_phrase_present' for r in group),
            'instructional_signoffs': 0,
            'approved_drills': 0,
        })
    coverage.append({
        'curriculum_level': 'ALL',
        'practice_plan': 'ALL',
        'source_occurrences': len(checks),
        'heading_exact_count': sum(r['heading_screen'] == 'exact_heading_found' for r in checks),
        'sections_available': sum(bool(r['section_sha256']) for r in checks),
        'recorded_prescription_exact_text_count': sum(r['source_prescription_screen'] == 'exact_recorded_phrase_present' for r in checks),
        'instructional_signoffs': 0,
        'approved_drills': 0,
    })

    assert len(checks) == 42 and len(fields) == 168 and len(queue) == 42
    if {r['candidate_group_id'] for r in checks} != {r['candidate_group_id'] for r in selected}:
        raise ValueError('Some selected candidates were not represented.')
    if any(r['catalog_approval_status'] != 'pending' or r['planner_eligible'] != 'no' for r in queue):
        raise ValueError('Unexpected approval change in review queue.')

    reports = {'source_checks': checks, 'field_comparisons': fields,
               'review_queue': queue, 'coverage': coverage}
    payloads = {key: csv_payload(rows) for key, rows in reports.items()}
    report_hashes = {key: hashlib.sha256(data).hexdigest() for key, data in payloads.items()}
    summary = {
        'step': '12-46',
        'catalog_version': '0.3.5',
        'screen_execution_date': date.today().isoformat(),
        'catalog_candidates_preserved': len(catalog),
        'registry_source_occurrences_preserved': len(registry),
        'selected_candidate_groups': len(checks),
        'comparison_fields': len(fields),
        'source_pages_attempted': len(source_urls),
        'source_pages_fetched': sum(p['status'] == 'fetched' for p in pages.values()),
        'exact_heading_matches': sum(r['heading_screen'] == 'exact_heading_found' for r in checks),
        'source_sections_available': sum(bool(r['section_sha256']) for r in checks),
        'exact_prescription_phrase_matches': sum(r['source_prescription_screen'] == 'exact_recorded_phrase_present' for r in checks),
        'registry_space_unspecified': sum(r['registry_space_unspecified'] == 'yes' for r in queue),
        'catalog_approval_changes': 0,
        'wave2_confirmed_metadata_edits_preserved': 29,
        'planner_eligibility_changes': 0,
        'input_sha256': input_hashes,
        'report_sha256': report_hashes,
        'limitations': 'HTML heading matches, numbers and keyword cues are preliminary evidence; each instruction and requirement needs source-level reviewer signoff. Publisher permissions and youth planning checks remain pending.',
    }
    payloads['summary'] = (json.dumps(summary, indent=2, sort_keys=True) + '\n').encode('utf-8')

    if any(path.exists() for path in outputs.values()):
        raise FileExistsError('A Step 12-46 output appeared during processing. No files changed.')
    for path in outputs.values():
        path.parent.mkdir(parents=True, exist_ok=True)
    with tempfile.TemporaryDirectory(prefix='step12_46_', dir=root / 'results') as folder:
        staged = {}
        for key, path in outputs.items():
            temporary = Path(folder) / path.name
            temporary.write_bytes(payloads[key])
            if hashlib.sha256(temporary.read_bytes()).hexdigest() != hashlib.sha256(payloads[key]).hexdigest():
                raise RuntimeError(f'Failed to verify staged output: {key}')
            staged[key] = temporary
        for key, path in outputs.items():
            # Exclusive creation prevents silently replacing a previous report.
            with path.open('xb') as target:
                target.write(staged[key].read_bytes())

    if {name: sha(path) for name, path in inputs.items()} != input_hashes:
        raise RuntimeError('An input file changed during execution. Inspect source history.')
    for key in reports:
        if sha(outputs[key]) != report_hashes[key]:
            raise RuntimeError(f'Output hash verification failed: {key}')

    print('\nSOURCE SCREENING RESULTS')
    for status, count in Counter(r['heading_screen'] for r in checks).items():
        print(f'  {status:<42} {count}')
    for status, count in Counter(r['source_prescription_screen'] for r in checks).items():
        print(f'  {status:<42} {count}')
    print('  Registry space fields unspecified:', summary['registry_space_unspecified'])
    print('\nREVIEW QUEUE')
    for row in queue:
        print(f"  {row['candidate_group_id']} | {row['drill_name']} | "
              f"{row['heading_screen']} | {row['prescription_screen']}")
    print('\nOUTPUTS')
    for path in outputs.values():
        print('  CREATED ', path.relative_to(root))
    print('\nCatalog, source registry, and approval decisions modified: No')
    print('Git operations performed: None')
    print('OVERALL: PASS — publisher evidence recorded; reviewer signoff pending')
```

```python
# Step 12-47 - Wave 3 Prescription Reconciliation and Space Follow-ups
# Verify publisher prescriptions in the 12 flagged records; retain all approval restrictions.

import csv
import hashlib
import io
import json
import re
import tempfile
import unicodedata
from collections import Counter
from datetime import date
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

root = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if root is None:
    raise FileNotFoundError('Project Git root not found.')

inputs = {
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_5_REVIEW_ONLY.csv',
    'registry': root / 'results/basketball_drill_source_registry_v0_3_5.csv',
    'source_checks': root / 'results/basketball_wave3_publisher_checks_v0_3_5.csv',
    'field_comparisons': root / 'results/basketball_wave3_field_comparisons_v0_3_5.csv',
    'review_queue': root / 'results/basketball_wave3_review_queue_v0_3_5.csv',
    'step46_summary': root / 'results/basketball_step12_46_summary_v0_3_5.json',
}
outputs = {
    'prescription_evidence': root / 'results/basketball_wave3_prescription_source_review_v0_3_5.csv',
    'prescription_decisions': root / 'results/basketball_wave3_prescription_decision_sheet_v0_3_5.csv',
    'space_followups': root / 'results/basketball_wave3_space_followups_v0_3_5.csv',
    'summary': root / 'results/basketball_step12_47_summary_v0_3_5.json',
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as stream:
        reader = csv.DictReader(stream)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or len(columns) != len(set(columns)):
        raise ValueError(f'Invalid CSV schema: {path.name}')
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f'Malformed CSV row: {path.name}')
    return rows


def unique_index(rows, column, count):
    if len(rows) != count:
        raise ValueError(f'Unexpected record count for {column}: {len(rows)} (expected {count})')
    values = [row[column].strip() for row in rows]
    if not all(values) or len(set(values)) != len(values):
        raise ValueError(f'Blank or duplicate {column}')
    return {row[column]: row for row in rows}


def digest(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def norm(value):
    return re.sub(r'[^a-z0-9]+', ' ', unicodedata.normalize('NFKC', value or '').casefold()).strip()


def normalize_prescription(value):
    text = unicodedata.normalize('NFKC', value or '').casefold().replace('×', 'x')
    text = re.sub(r'(?<=\d)\s*x\s*(?=\d)', ' x ', text)
    text = re.sub(r'\bsecs?\b', 'seconds', text)
    text = re.sub(r'\bmins?\b', 'minutes', text)
    text = re.sub(r'\breps?\b', 'repetitions', text)
    return re.sub(r'[^a-z0-9]+', ' ', text).strip()


def csv_bytes(rows):
    if not rows:
        raise ValueError('Cannot create an empty report')
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=list(rows[0]),
                            extrasaction='raise', lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')


class SectionParser(HTMLParser):
    """Extract individual publisher heading sections in memory."""

    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.skip_tag = None
        self.skip_depth = 0
        self.heading_tag = None
        self.heading_parts = []
        self.current = None

    def handle_starttag(self, tag, attrs):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth += 1
            return
        if tag in {'script', 'style', 'noscript', 'svg'}:
            self.skip_tag, self.skip_depth = tag, 1
        elif re.fullmatch(r'h[1-6]', tag):
            self.heading_tag, self.heading_parts = tag, []
        elif self.current is not None and tag in {'p', 'li', 'div', 'br', 'td', 'tr'}:
            self.current['parts'].append(' ')

    def handle_data(self, data):
        if self.skip_depth:
            return
        if self.heading_tag:
            self.heading_parts.append(data)
        elif self.current is not None:
            self.current['parts'].append(data)

    def handle_endtag(self, tag):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth -= 1
                if not self.skip_depth:
                    self.skip_tag = None
            return
        if tag == self.heading_tag:
            heading = norm(' '.join(self.heading_parts))
            if heading:
                self.current = {'heading': heading, 'parts': []}
                self.sections.append(self.current)
            self.heading_tag, self.heading_parts = None, []


def official_url(url):
    parsed = urlsplit(url)
    return (parsed.scheme == 'https' and parsed.hostname == 'jr.nba.com'
            and parsed.port is None and not parsed.query and not parsed.fragment
            and re.fullmatch(r'/basketball-practice-plans/(rookie|starter|all-star|mvp)/(page/\d+/)?',
                             parsed.path) is not None)


def fetch(url):
    if not official_url(url):
        raise ValueError(f'Unexpected official source URL: {url}')
    try:
        request = Request(url, headers={'User-Agent': 'Mozilla/5.0 (Academic Basketball Source Review)',
                                        'Accept': 'text/html'})
        with urlopen(request, timeout=25) as response:
            if not official_url(response.geturl()):
                return {'status': 'unexpected_redirect', 'sections': []}
            html = response.read(2_000_000).decode('utf-8', errors='replace')
        parser = SectionParser()
        parser.feed(html)
        return {'status': 'fetched', 'sections': parser.sections}
    except (HTTPError, URLError, TimeoutError, OSError, ValueError):
        return {'status': 'unavailable', 'sections': []}


def leading_source_prescription(section):
    """Read only the short parenthetical prescription immediately after a heading."""
    text = ' '.join(section['parts']).strip()
    # Prevent extracting a number mentioned later in the coaching instructions.
    match = re.search(r'\(\s*([^()]{1,90})\s*\)', text[:180])
    if match and match.start() <= 45:
        phrase = ' '.join(match.group(1).split())
        # Ignore non-prescription parentheticals (e.g., explanatory labels).
        if re.search(r'\d', phrase):
            return phrase
    return ''


catalog = read_csv(inputs['catalog'])
registry = read_csv(inputs['registry'])
checks = read_csv(inputs['source_checks'])
fields = read_csv(inputs['field_comparisons'])
queue = read_csv(inputs['review_queue'])
if not inputs['step46_summary'].is_file():
    raise FileNotFoundError(inputs['step46_summary'])
previous = json.loads(inputs['step46_summary'].read_text(encoding='utf-8'))

catalog_by_id = unique_index(catalog, 'catalog_candidate_id', 144)
registry_by_id = unique_index(registry, 'evidence_id', 156)
check_by_eid = unique_index(checks, 'evidence_id', 42)
queue_by_eid = unique_index(queue, 'evidence_id', 42)
if len(fields) != 168 or len({(r['evidence_id'], r['field']) for r in fields}) != 168:
    raise ValueError('Wave 3 field comparison rows are missing or duplicated.')
if any({r['field'] for r in fields if r['evidence_id'] == eid} !=
       {'equipment', 'participants', 'space', 'prescription'} for eid in check_by_eid):
    raise ValueError('Four required Wave 3 fields not present for each source.')
if set(check_by_eid) != set(queue_by_eid):
    raise ValueError('Step 12-46 source checks and review queue disagree.')
if any(r['catalog_approval_status'] != 'pending' or r['planner_eligible'] != 'no'
       or r['approved_drill_id'].strip() for r in catalog):
    raise ValueError('A catalog approval status changed; review before continuing.')
if any(check['candidate_group_id'] != registry_by_id[eid]['catalog_candidate_id']
       or check['candidate_group_id'] not in catalog_by_id
       or check['official_source_url'] != registry_by_id[eid]['source_url']
       for eid, check in check_by_eid.items()):
    raise ValueError('Wave 3 source linkage does not match current registry.')
for key in ('source_checks', 'field_comparisons', 'review_queue'):
    if digest(inputs[key]) != previous.get('report_sha256', {}).get(key):
        raise ValueError(f'Step 12-46 report hash changed: {key}')
for key in ('catalog', 'registry'):
    if digest(inputs[key]) != previous.get('input_sha256', {}).get(key):
        raise ValueError(f'Catalog or registry changed after Step 12-46: {key}')
if previous.get('selected_candidate_groups') != 42:
    raise ValueError('Step 12-46 did not verify 42 Wave 3 candidates.')

flagged = [r for r in checks if r['source_prescription_screen'] != 'exact_recorded_phrase_present']
unspecified_space = [r for r in queue if r['registry_space_unspecified'] == 'yes']
if len(flagged) != 12 or len(unspecified_space) != 18:
    raise ValueError(f'Unexpected Wave 3 gaps: {len(flagged)} prescriptions, '
                     f'{len(unspecified_space)} space fields; review changed inputs.')
input_hashes = {key: digest(path) for key, path in inputs.items()}

print('=' * 72)
print('STEP 12-47 — WAVE 3 PRESCRIPTION AND SPACE REVIEW')
print('=' * 72)
print('Catalog candidates / source occurrences:', len(catalog), '/', len(registry))
print('Wave 3 candidates:', len(checks))
print('Flagged prescriptions / unspecified source-space fields:',
      len(flagged), '/', len(unspecified_space))

existing = [path.exists() for path in outputs.values()]
if any(existing):
    if not all(existing):
        raise FileExistsError('Only some Step 12-47 reports exist; inspect before rerunning.')
    prior_output = json.loads(outputs['summary'].read_text(encoding='utf-8'))
    if prior_output.get('input_sha256') != input_hashes:
        raise ValueError('Step 12-47 source inputs have changed; preserve and inspect saved results.')
    for key in ('prescription_evidence', 'space_followups'):
        if digest(outputs[key]) != prior_output.get('report_sha256', {}).get(key):
            raise ValueError(f'Saved evidence report changed: {outputs[key].name}')
    decision_rows = read_csv(outputs['prescription_decisions'])
    if len(decision_rows) != 12 or {r['evidence_id'] for r in decision_rows} != {r['evidence_id'] for r in flagged}:
        raise ValueError('Saved decision sheet has unexpected source identifiers.')
    print('EXISTING REPORTS PRESERVED — any reviewer edits remain intact.')
    for path in outputs.values():
        print('  VERIFIED', path.relative_to(root))
    print('OVERALL: PASS — existing Wave 3 review artifacts preserved')
else:
    urls = sorted({r['official_source_url'] for r in flagged})
    print('Official pages requested for flagged prescriptions:', len(urls))
    pages = {}
    for index, url in enumerate(urls, 1):
        pages[url] = fetch(url)
        print(f"  {index:02d}/{len(urls):02d} | {pages[url]['status']:<12} | {url}")
    if not any(page['status'] == 'fetched' for page in pages.values()):
        raise RuntimeError('No source pages were available. No reports created.')

    evidence = []
    decision_sheet = []
    for check in sorted(flagged, key=lambda row: row['evidence_id']):
        eid = check['evidence_id']
        source = registry_by_id[eid]
        page = pages[check['official_source_url']]
        sections = [item for item in page['sections'] if item['heading'] == norm(source['drill_name'])]
        heading_status = ('single_exact_heading' if len(sections) == 1 else
                          'ambiguous_multiple_headings' if len(sections) > 1 else
                          'not_checked_page_unavailable' if page['status'] != 'fetched' else
                          'heading_not_found')
        published = leading_source_prescription(sections[0]) if len(sections) == 1 else ''
        recorded = source['source_prescription'].strip()
        if heading_status != 'single_exact_heading':
            comparison = 'source_heading_requires_review'
            recommendation = 'defer'
        elif published and not recorded:
            comparison = 'publisher_value_present_registry_missing'
            recommendation = 'review_explicit_source_value'
        elif published and normalize_prescription(published) == normalize_prescription(recorded):
            comparison = 'same_prescription_after_notation_normalization'
            recommendation = 'retain_current'
        elif published:
            comparison = 'publisher_recorded_prescription_difference'
            recommendation = 'review_explicit_source_value'
        elif recorded.casefold() in {'', 'not specified', 'not stated'}:
            comparison = 'no_standalone_prescription_visible_in_html'
            recommendation = 'retain_unspecified_pending_full_source_review'
        else:
            comparison = 'no_leading_publisher_prescription_detected'
            recommendation = 'defer'

        evidence.append({
            'candidate_group_id': check['candidate_group_id'],
            'evidence_id': eid,
            'drill_name': check['drill_name'],
            'source_url': check['official_source_url'],
            'step46_prescription_screen': check['source_prescription_screen'],
            'publisher_heading_screen': heading_status,
            'recorded_registry_prescription': recorded,
            'short_publisher_prescription': published,
            'normalized_recorded_prescription': normalize_prescription(recorded),
            'normalized_publisher_prescription': normalize_prescription(published),
            'source_comparison_status': comparison,
            'source_section_full_instructions_saved': 'no',
        })
        decision_sheet.append({
            'candidate_group_id': check['candidate_group_id'],
            'evidence_id': eid,
            'drill_name': check['drill_name'],
            'source_url': check['official_source_url'],
            'existing_catalog_prescription': catalog_by_id[check['candidate_group_id']]['source_prescription'],
            'existing_registry_prescription': recorded,
            'publisher_prescription_for_review': published,
            'evidence_status': comparison,
            'suggested_review_action': recommendation,
            'reviewer_decision': 'pending',
            'reviewed_prescription': '',
            'reviewer': '',
            'review_date': '',
            'review_notes': '',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
        })

    space_followups = []
    for row in sorted(unspecified_space, key=lambda item: int(item['review_rank'])):
        eid = row['evidence_id']
        check = check_by_eid[eid]
        source = registry_by_id[eid]
        item = catalog_by_id[check['candidate_group_id']]
        space_followups.append({
            'review_rank': row['review_rank'],
            'candidate_group_id': check['candidate_group_id'],
            'evidence_id': eid,
            'drill_name': item['drill_name'],
            'source_url': source['source_url'],
            'current_catalog_space': item['space'],
            'current_registry_space': source['space'],
            'step46_source_space_cues': check['source_space_cues'],
            'required_action': 'Confirm source-stated space separately from feasible operational space.',
            'reviewer_decision': 'pending',
            'verified_source_space': '',
            'reviewer': '',
            'review_notes': '',
            'planner_eligible': 'no',
        })

    assert len(evidence) == len(decision_sheet) == 12 and len(space_followups) == 18
    payloads = {'prescription_evidence': csv_bytes(evidence),
                'prescription_decisions': csv_bytes(decision_sheet),
                'space_followups': csv_bytes(space_followups)}
    report_hashes = {key: hashlib.sha256(data).hexdigest() for key, data in payloads.items()}
    summary = {
        'step': '12-47',
        'catalog_version': '0.3.5',
        'source_review_date': date.today().isoformat(),
        'catalog_groups_preserved': 144,
        'source_occurrences_preserved': 156,
        'wave3_candidates': 42,
        'prescriptions_flagged_in_step46': 12,
        'space_followups': 18,
        'publisher_pages_requested': len(urls),
        'publisher_pages_fetched': sum(page['status'] == 'fetched' for page in pages.values()),
        'prescription_results': dict(Counter(row['source_comparison_status'] for row in evidence)),
        'reviewer_approvals_recorded': 0,
        'catalog_changes': 0,
        'planner_eligibility_changes': 0,
        'input_sha256': input_hashes,
        'report_sha256': report_hashes,
        'limitations': 'Short published prescriptions are preliminary source evidence; no inferred timing or court space is promoted. Instructions, permissions, canonical identity, and youth planning checks require separate review.',
    }
    payloads['summary'] = (json.dumps(summary, indent=2, sort_keys=True) + '\n').encode('utf-8')

    if any(path.exists() for path in outputs.values()):
        raise FileExistsError('Step 12-47 outputs appeared during processing; no files overwritten.')
    for path in outputs.values():
        path.parent.mkdir(parents=True, exist_ok=True)
    with tempfile.TemporaryDirectory(prefix='step12_47_', dir=root / 'results') as tempdir:
        for key, path in outputs.items():
            staged = Path(tempdir) / path.name
            staged.write_bytes(payloads[key])
            if hashlib.sha256(staged.read_bytes()).hexdigest() != hashlib.sha256(payloads[key]).hexdigest():
                raise RuntimeError(f'Failed to stage: {path.name}')
        for key, path in outputs.items():
            with path.open('xb') as target:
                target.write(payloads[key])
    if {key: digest(path) for key, path in inputs.items()} != input_hashes:
        raise RuntimeError('Input files changed during execution; inspect before proceeding.')
    for key in ('prescription_evidence', 'prescription_decisions', 'space_followups'):
        if digest(outputs[key]) != report_hashes[key]:
            raise RuntimeError(f'Output verification failed: {key}')

    print('\nPUBLISHER PRESCRIPTION FINDINGS')
    for kind, count in Counter(row['source_comparison_status'] for row in evidence).items():
        print(f'  {kind:<52} {count}')
    print('\nTARGETED PRESCRIPTION RECORDS')
    for row in evidence:
        print(f"  {row['evidence_id']} | {row['drill_name']} | {row['source_comparison_status']}")
    print('\nSPACE REVIEW')
    print('  Source-space fields awaiting review:', len(space_followups))
    print('\nOUTPUTS')
    for path in outputs.values():
        print('  CREATED', path.relative_to(root))
    print('\nReviewer decisions / catalog approvals / planner-eligible: 0 / 0 / 0')
    print('Previous catalog, source registry, and Wave 2 decisions overwritten: No')
    print('Git operations performed: None')
    print('OVERALL: PASS — publisher prescription evidence recorded; reviewer signoff pending')
```

```python
# Step 12-48 - Wave 3 Prescription Corrections and Space Adjudication Packet
# Prepare evidence-linked reviewer decisions without changing the catalog or source registry.

import csv
import hashlib
import io
import json
from collections import Counter
from datetime import date
from pathlib import Path
from urllib.parse import urlsplit

root = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if root is None:
    raise FileNotFoundError('Project Git root not found.')

inputs = {
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_5_REVIEW_ONLY.csv',
    'registry': root / 'results/basketball_drill_source_registry_v0_3_5.csv',
    'publisher_evidence': root / 'results/basketball_wave3_prescription_source_review_v0_3_5.csv',
    'prescription_decisions': root / 'results/basketball_wave3_prescription_decision_sheet_v0_3_5.csv',
    'space_followups': root / 'results/basketball_wave3_space_followups_v0_3_5.csv',
    'step47_summary': root / 'results/basketball_step12_47_summary_v0_3_5.json',
}
outputs = {
    'prescription_review': root / 'results/basketball_wave3_prescription_adjudication_v0_3_5.csv',
    'correction_proposals': root / 'results/basketball_wave3_prescription_correction_candidates_v0_3_5.csv',
    'space_review': root / 'results/basketball_wave3_space_adjudication_v0_3_5.csv',
    'summary': root / 'results/basketball_step12_48_summary_v0_3_5.json',
}


def rows_from_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as stream:
        reader = csv.DictReader(stream)
        columns = list(reader.fieldnames or [])
        result = list(reader)
    if not columns or len(columns) != len(set(columns)):
        raise ValueError(f'Invalid CSV schema: {path.name}')
    if any(None in row or any(value is None for value in row.values()) for row in result):
        raise ValueError(f'Malformed CSV row: {path.name}')
    return result


def unique(rows, key, count):
    if len(rows) != count:
        raise ValueError(f'Expected {count} rows for {key}, found {len(rows)}')
    vals = [row[key].strip() for row in rows]
    if not all(vals) or len(set(vals)) != len(vals):
        raise ValueError(f'Duplicate or blank {key}')
    return {row[key]: row for row in rows}


def sha(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def csv_payload(rows):
    if not rows:
        raise ValueError('Refusing to save an empty review table')
    stream = io.StringIO(newline='')
    writer = csv.DictWriter(stream, fieldnames=list(rows[0]),
                            extrasaction='raise', lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return stream.getvalue().encode('utf-8')


def official_url(url):
    p = urlsplit(url)
    return p.scheme == 'https' and p.hostname == 'jr.nba.com' and p.port is None

catalog = rows_from_csv(inputs['catalog'])
registry = rows_from_csv(inputs['registry'])
evidence = rows_from_csv(inputs['publisher_evidence'])
decisions = rows_from_csv(inputs['prescription_decisions'])
spaces = rows_from_csv(inputs['space_followups'])

catalog_by_id = unique(catalog, 'catalog_candidate_id', 144)
registry_by_id = unique(registry, 'evidence_id', 156)
evidence_by_id = unique(evidence, 'evidence_id', 12)
decision_by_id = unique(decisions, 'evidence_id', 12)
space_by_id = unique(spaces, 'evidence_id', 18)

if set(evidence_by_id) != set(decision_by_id):
    raise ValueError('The prescription source review and decision sheet do not agree.')
if not inputs['step47_summary'].is_file():
    raise FileNotFoundError(inputs['step47_summary'])
step47 = json.loads(inputs['step47_summary'].read_text(encoding='utf-8'))
if step47.get('catalog_groups_preserved') != 144 or step47.get('source_occurrences_preserved') != 156:
    raise ValueError('Step 12-47 does not match the current baseline.')
if step47.get('prescriptions_flagged_in_step46') != 12 or step47.get('space_followups') != 18:
    raise ValueError('Step 12-47 summary counts have changed.')
for source_key, expected_key in [('publisher_evidence', 'prescription_evidence'),
                                 ('space_followups', 'space_followups')]:
    if sha(inputs[source_key]) != step47.get('report_sha256', {}).get(expected_key):
        raise ValueError(f'Step 12-47 evidence changed: {source_key}')
for source_key in ('catalog', 'registry'):
    if sha(inputs[source_key]) != step47.get('input_sha256', {}).get(source_key):
        raise ValueError(f'Current {source_key} differs from Step 12-47 verified input.')
if any(row['catalog_approval_status'] != 'pending' or
       row['planner_eligible'] != 'no' or row['approved_drill_id'].strip()
       for row in catalog):
    raise ValueError('Catalog approval state has changed; inspect before continuing.')

allowed = {
    'same_prescription_after_notation_normalization',
    'no_standalone_prescription_visible_in_html',
    'publisher_recorded_prescription_difference',
}
counts = Counter(row['source_comparison_status'] for row in evidence)
if set(counts) - allowed:
    raise ValueError(f'Unexpected review status: {set(counts) - allowed}')
if counts != {
    'same_prescription_after_notation_normalization': 3,
    'no_standalone_prescription_visible_in_html': 3,
    'publisher_recorded_prescription_difference': 6,
}:
    raise ValueError(f'Unexpected Step 12-47 prescription status counts: {dict(counts)}')

review = []
corrections = []
for eid, row in sorted(evidence_by_id.items()):
    initial = decision_by_id[eid]
    source = registry_by_id[eid]
    candidate = catalog_by_id[row['candidate_group_id']]
    recorded = row['recorded_registry_prescription'].strip()
    proposed = row['short_publisher_prescription'].strip()
    status = row['source_comparison_status']

    if (row['candidate_group_id'] != source['catalog_candidate_id'] or
        row['candidate_group_id'] != initial['candidate_group_id'] or
        candidate['catalog_candidate_id'] != row['candidate_group_id'] or
        source['source_prescription'].strip() != recorded or
        initial['publisher_prescription_for_review'].strip() != proposed or
        not official_url(row['source_url'])):
        raise ValueError(f'Conflicting source metadata for {eid}')
    if initial['reviewer_decision'] not in {'pending', 'defer', 'retain_current', 'accept_source'}:
        raise ValueError(f'Unexpected pre-existing reviewer decision for {eid}')
    if initial['catalog_approval_status'] != 'pending' or initial['planner_eligible'] != 'no':
        raise ValueError(f'Approval state changed for {eid}')

    if status == 'publisher_recorded_prescription_difference':
        if not proposed or row['normalized_publisher_prescription'] == row['normalized_recorded_prescription']:
            raise ValueError(f'Publisher correction lacks distinct evidence: {eid}')
        suggested = 'review_publisher_value'
        reason = 'Short leading source prescription differs from recorded value; confirm against the full activity section.'
    elif status == 'same_prescription_after_notation_normalization':
        suggested = 'retain_current'
        reason = 'Prescriptions agree after notation normalization; avoid unnecessary edits.'
    else:
        suggested = 'retain_unspecified_pending_source_review'
        reason = 'No standalone prescription found in the publisher HTML; do not invent a duration.'

    record = {
        'evidence_id': eid,
        'candidate_group_id': row['candidate_group_id'],
        'drill_name': row['drill_name'],
        'source_url': row['source_url'],
        'existing_catalog_prescription': candidate['source_prescription'],
        'existing_registry_prescription': recorded,
        'publisher_prescription_proposed': proposed,
        'evidence_classification': status,
        'proposed_action': suggested,
        'evidence_review_note': reason,
        'previous_review_decision': initial['reviewer_decision'],
        'reviewer_decision': 'pending',
        'confirmed_prescription': '',
        'reviewer': '',
        'review_date': '',
        'review_notes': '',
        'catalog_approval_status': 'pending',
        'planner_eligible': 'no',
    }
    review.append(record)
    if status == 'publisher_recorded_prescription_difference':
        corrections.append(dict(record))

space_review = []
for eid, item in sorted(space_by_id.items(), key=lambda kv: int(kv[1]['review_rank'])):
    source = registry_by_id[eid]
    candidate = catalog_by_id[item['candidate_group_id']]
    if (source['catalog_candidate_id'] != item['candidate_group_id'] or
        not official_url(item['source_url']) or
        item['current_registry_space'] != source['space'] or
        item['current_catalog_space'] != candidate['space'] or
        item['planner_eligible'] != 'no'):
        raise ValueError(f'Inconsistent space review record: {eid}')
    space_review.append({
        'review_rank': item['review_rank'],
        'evidence_id': eid,
        'candidate_group_id': item['candidate_group_id'],
        'drill_name': item['drill_name'],
        'source_url': item['source_url'],
        'catalog_space': item['current_catalog_space'],
        'registry_space': item['current_registry_space'],
        'source_text_cues': item['step46_source_space_cues'],
        'source_cues_are_not_verified_requirements': 'yes',
        'required_review': 'Confirm exact source-stated space; keep planning inference separate.',
        'reviewer_decision': 'pending',
        'verified_source_space': '',
        'operational_space_proposal': '',
        'reviewer': '',
        'review_date': '',
        'review_notes': '',
        'catalog_approval_status': 'pending',
        'planner_eligible': 'no',
    })

assert len(review) == 12 and len(corrections) == 6 and len(space_review) == 18
input_hashes = {key: sha(path) for key, path in inputs.items()}

print('=' * 72)
print('STEP 12-48 — WAVE 3 PRESCRIPTION CORRECTION PROPOSALS')
print('=' * 72)
print('Catalog candidate groups / source records:', len(catalog), '/', len(registry))
print('Prescription review records:', len(review))
print('Publisher-versus-recorded differences:', len(corrections))
print('Source-space follow-ups:', len(space_review))

existing = [path.exists() for path in outputs.values()]
if any(existing):
    if not all(existing):
        raise FileExistsError('Partial Step 12-48 outputs found; inspect before rerunning.')
    previous = json.loads(outputs['summary'].read_text(encoding='utf-8'))
    if previous.get('input_sha256') != input_hashes:
        raise ValueError('Source inputs changed since Step 12-48 reports were written.')
    for key, expected_count in [('prescription_review', 12), ('correction_proposals', 6), ('space_review', 18)]:
        if len(rows_from_csv(outputs[key])) != expected_count:
            raise ValueError(f'Existing review file has unexpected row count: {key}')
    print('EXISTING OUTPUTS PRESERVED — reviewer edits have not been overwritten.')
    for file in outputs.values():
        print('  VERIFIED', file.relative_to(root))
else:
    payloads = {
        'prescription_review': csv_payload(review),
        'correction_proposals': csv_payload(corrections),
        'space_review': csv_payload(space_review),
    }
    summary = {
        'step': '12-48',
        'catalog_version': '0.3.5',
        'run_date': date.today().isoformat(),
        'catalog_groups': 144,
        'source_records': 156,
        'prescription_review_count': len(review),
        'publisher_difference_count': len(corrections),
        'notation_equivalence_count': counts['same_prescription_after_notation_normalization'],
        'unspecified_prescription_count': counts['no_standalone_prescription_visible_in_html'],
        'space_followups': len(space_review),
        'reviewer_signoffs_performed': 0,
        'catalog_updates': 0,
        'approvals': 0,
        'planner_eligible': 0,
        'input_sha256': input_hashes,
        'report_sha256': {key: hashlib.sha256(payload).hexdigest()
                          for key, payload in payloads.items()},
        'limitations': 'Publisher prescription extracted from a short heading-adjacent segment; every proposed correction requires reviewer confirmation. Space keyword cues do not establish exact requirements. No metadata applied or source content copied.',
    }
    payloads['summary'] = (json.dumps(summary, indent=2, sort_keys=True) + '\n').encode('utf-8')
    for path in outputs.values():
        path.parent.mkdir(parents=True, exist_ok=True)
    if any(path.exists() for path in outputs.values()):
        raise FileExistsError('New outputs appeared during preparation; no files written.')
    for key, path in outputs.items():
        with path.open('xb') as stream:
            stream.write(payloads[key])
    if {key: sha(path) for key, path in inputs.items()} != input_hashes:
        raise RuntimeError('Input files changed during execution.')
    for key, path in outputs.items():
        if sha(path) != hashlib.sha256(payloads[key]).hexdigest():
            raise RuntimeError(f'Output verification failed: {path.name}')

    print('\nSIX SOURCE-PRESCRIPTION DIFFERENCES')
    for row in corrections:
        print(f"  {row['evidence_id']} | {row['drill_name']}")
        print(f"    recorded: {row['existing_registry_prescription']}")
        print(f"    proposed: {row['publisher_prescription_proposed']}")
    print('\nPRESCRIPTION REVIEW OUTCOMES')
    for key, count in counts.items():
        print(f'  {key:<54} {count}')
    print('\nOUTPUTS')
    for path in outputs.values():
        print('  CREATED', path.relative_to(root))

print('Catalog changes / approvals / planner-eligible: 0 / 0 / 0')
print('Original source and earlier catalog files overwritten: No')
print('Git operations performed: None')
print('OVERALL: PASS — six source differences staged for review; space requirements remain pending')
```

```python
# Step 12-50 - Wave 3 Source-Space Verification and Prescription Decision Packet
# Document source-supported space descriptions and prescription equivalences without changing approvals.

import csv
import hashlib
import io
import json
import re
import unicodedata
from collections import Counter
from datetime import date
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

root = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if root is None:
    raise FileNotFoundError('Project Git root not found.')

inputs = {
    'catalog': root / 'data/processed/basketball_drill_catalog_v0_3_5_REVIEW_ONLY.csv',
    'registry': root / 'results/basketball_drill_source_registry_v0_3_5.csv',
    'prescriptions': root / 'results/basketball_wave3_prescription_equivalence_review_v0_3_5.csv',
    'spaces': root / 'results/basketball_wave3_space_resolution_tracker_v0_3_5.csv',
    'prior_summary': root / 'results/basketball_step12_49_summary_v0_3_5.json',
}
outputs = {
    'source_evidence': root / 'results/basketball_wave3_space_source_evidence_v0_3_5.csv',
    'space_decisions': root / 'results/basketball_wave3_space_recommendations_v0_3_5.csv',
    'prescription_decisions': root / 'results/basketball_wave3_prescription_recommendations_v0_3_5.csv',
    'summary': root / 'results/basketball_step12_50_summary_v0_3_5.json',
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as stream:
        reader = csv.DictReader(stream)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or len(columns) != len(set(columns)):
        raise ValueError(f'Invalid CSV columns: {path.name}')
    if any(None in row or any(v is None for v in row.values()) for row in rows):
        raise ValueError(f'Malformed CSV row: {path.name}')
    return rows


def index_by(rows, field, count):
    if len(rows) != count:
        raise ValueError(f'{field}: expected {count} rows; found {len(rows)}')
    keys = [r[field].strip() for r in rows]
    if not all(keys) or len(keys) != len(set(keys)):
        raise ValueError(f'Missing or duplicated {field} values')
    return dict(zip(keys, rows))


def sha(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def norm(value):
    return re.sub(r'[^a-z0-9]+', ' ',
                  unicodedata.normalize('NFKC', value or '').casefold()).strip()


def csv_bytes(rows):
    if not rows:
        raise ValueError('Cannot export an empty report')
    buf = io.StringIO(newline='')
    writer = csv.DictWriter(buf, fieldnames=list(rows[0]),
                            extrasaction='raise', lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return buf.getvalue().encode('utf-8')


class DrillSections(HTMLParser):
    """Read official HTML heading sections in memory; do not store publisher prose."""
    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.skip_tag = None
        self.skip_depth = 0
        self.heading_tag = None
        self.heading_parts = []
        self.current = None

    def handle_starttag(self, tag, attrs):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth += 1
            return
        if tag in {'script', 'style', 'noscript', 'svg'}:
            self.skip_tag, self.skip_depth = tag, 1
        elif re.fullmatch(r'h[1-6]', tag):
            self.heading_tag, self.heading_parts = tag, []
        elif self.current is not None and tag in {'p', 'li', 'div', 'br', 'td', 'tr'}:
            self.current['parts'].append(' ')

    def handle_data(self, data):
        if self.skip_depth:
            return
        if self.heading_tag:
            self.heading_parts.append(data)
        elif self.current is not None:
            self.current['parts'].append(data)

    def handle_endtag(self, tag):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth -= 1
                if not self.skip_depth:
                    self.skip_tag = None
            return
        if tag == self.heading_tag:
            heading = norm(' '.join(self.heading_parts))
            if heading:
                self.current = {'heading': heading, 'parts': []}
                self.sections.append(self.current)
            self.heading_tag, self.heading_parts = None, []


def is_official(url):
    parsed = urlsplit(url)
    return (parsed.scheme == 'https' and parsed.hostname == 'jr.nba.com'
            and not parsed.query and not parsed.fragment and parsed.port is None
            and re.fullmatch(r'/basketball-practice-plans/(rookie|starter|all-star|mvp)/(page/\d+/)?',
                             parsed.path) is not None)


def fetch_sections(url):
    if not is_official(url):
        raise ValueError(f'Unexpected source URL: {url}')
    try:
        request = Request(url, headers={'User-Agent': 'Mozilla/5.0',
                                        'Accept': 'text/html'})
        with urlopen(request, timeout=25) as response:
            if not is_official(response.geturl()):
                return 'unexpected_redirect', []
            html = response.read(2_000_000).decode('utf-8', errors='replace')
        parser = DrillSections()
        parser.feed(html)
        return 'fetched', parser.sections
    except (HTTPError, URLError, TimeoutError, OSError, ValueError):
        return 'unavailable', []


# Each description refers to observed source locations, not verified minimum court dimensions.
# The pattern checks must pass against the specific activity section before suggesting any value.
space_rules = {
    'JR-EVID-0064': ('Baseline start with movement toward half court.',
                     ('baseline', r'half\s?court'), 'source_locations_reported'),
    'JR-EVID-0063': ('Players spread along the baseline; coach near the free-throw line.',
                     ('baseline', 'free throw line'), 'source_locations_reported'),
    'JR-EVID-0047': ('Starts inside the key, with defenders along the three-point line.',
                     ('key', 'three point line'), 'source_locations_reported'),
    'JR-EVID-0053': ('Baseline lines and wing positions around a basket.',
                     ('baseline', 'wing'), 'source_locations_reported'),
    'JR-EVID-0067': ('Baseline and lane-line start, progressing through elbow and basket areas.',
                     ('baseline', 'lane line', 'elbow'), 'source_locations_reported'),
    'JR-EVID-0052': ('Under the basket, with play extending behind the three-point line.',
                     ('under the basket', 'three point line'), 'source_locations_reported'),
    'JR-EVID-0054': ('Half-court start, with play transitioning between both baskets.',
                     (r'half\s?court', 'other basket'), 'source_locations_reported'),
    'JR-EVID-0069': ('Baseline start and movement to the opposite basket and back.',
                     ('baseline', 'other basket'), 'source_locations_reported'),
    'JR-EVID-0070': ('Baseline start with players advancing down the court.',
                     ('baseline', 'down the court'), 'source_locations_reported'),
    'JR-EVID-0066': ('Wing and corner positions with a baseline-side drive toward the basket.',
                     ('wing', 'corner', 'baseline'), 'source_locations_reported'),
    'JR-EVID-0049': ('Half-court starting lines with a defender near the three-point line.',
                     (r'half\s?court', 'three point line'), 'source_locations_reported'),
    'JR-EVID-0068': ('Post players on the block, guards above the opposite wing, finishing in the paint.',
                     ('block', 'wing', 'paint'), 'source_locations_reported'),
    'JR-EVID-0057': ('Players spaced at least a lane width apart on the court; hoop attack follows.',
                     ('lane width', 'court'), 'source_locations_reported'),
    'JR-EVID-0050': ('Wing and weak-side corner positions with drives toward the basket.',
                     ('wing', 'corner', 'basket'), 'source_locations_reported'),
    'JR-EVID-0051': ('Free-throw shooting at available baskets; minimum court dimensions not specified.',
                     ('baskets', 'free throws'), 'source_locations_reported'),
    'JR-EVID-0065': ('Two lines near the lane-line/sideline intersection, using the basket and block.',
                     ('lane line', 'sideline', 'block'), 'source_locations_reported'),
    'JR-EVID-0055': ('Space for paired dribbling activity; exact court location not specified.',
                     ('partner', 'dribble'), 'operational_space_inference'),
    'JR-EVID-0048': ('Open space on the court for triple-threat footwork.',
                     ('open space', 'court'), 'source_locations_reported'),
}

catalog = read_csv(inputs['catalog'])
registry = read_csv(inputs['registry'])
prescriptions = read_csv(inputs['prescriptions'])
spaces = read_csv(inputs['spaces'])
if not inputs['prior_summary'].is_file():
    raise FileNotFoundError(inputs['prior_summary'])
prior = json.loads(inputs['prior_summary'].read_text(encoding='utf-8'))
catalog_by_id = index_by(catalog, 'catalog_candidate_id', 144)
registry_by_id = index_by(registry, 'evidence_id', 156)
prescription_by_id = index_by(prescriptions, 'evidence_id', 6)
space_by_id = index_by(spaces, 'evidence_id', 18)
if set(space_by_id) != set(space_rules):
    raise ValueError('Space review identifiers differ from the 18 documented source checks.')
if set(prescription_by_id) != {
    'JR-EVID-0034', 'JR-EVID-0036', 'JR-EVID-0041',
    'JR-EVID-0044', 'JR-EVID-0045', 'JR-EVID-0069',
}:
    raise ValueError('Unexpected prescription-equivalence identifiers.')
for key in ('catalog', 'registry'):
    if prior.get('input_sha256', {}).get(key) != sha(inputs[key]):
        raise ValueError(f'{key} changed from the previously verified Step 12-49 baseline.')
for key, prior_key in (('prescriptions', 'equivalence_review'),
                       ('spaces', 'space_tracker')):
    if prior.get('report_sha256', {}).get(prior_key) != sha(inputs[key]):
        raise ValueError(f'{key} changed since Step 12-49; reconcile reviewer edits first.')
if prior.get('semantic_equivalence_proposals') != 6 or prior.get('space_review_records') != 18:
    raise ValueError('Step 12-49 summary has unexpected record counts.')
if any(r['catalog_approval_status'] != 'pending' or r['planner_eligible'] != 'no'
       or r['approved_drill_id'].strip() for r in catalog):
    raise ValueError('Catalog approval gates changed; review before proceeding.')
for eid, row in space_by_id.items():
    source = registry_by_id[eid]
    candidate = catalog_by_id[row['candidate_group_id']]
    if (row['candidate_group_id'] != source['catalog_candidate_id'] or
            row['source_url'] != source['source_url'] or
            row['drill_name'] != source['drill_name'] or
            row['reviewer_decision'] != 'pending' or
            row['catalog_approval_status'] != 'pending' or
            row['planner_eligible'] != 'no' or
            candidate['catalog_approval_status'] != 'pending'):
        raise ValueError(f'Unresolved source link or review status changed: {eid}')
for eid, row in prescription_by_id.items():
    source = registry_by_id[eid]
    candidate = catalog_by_id[row['candidate_group_id']]
    if (row['candidate_group_id'] != source['catalog_candidate_id'] or
            row['source_url'] != source['source_url'] or
            row['existing_registry_prescription'] != source['source_prescription'] or
            row['reviewer_decision'] != 'pending' or
            row['catalog_approval_status'] != 'pending' or
            candidate['catalog_approval_status'] != 'pending'):
        raise ValueError(f'Prescription identity or pending review status changed: {eid}')

print('=' * 72)
print('STEP 12-50 — WAVE 3 SPACE VERIFICATION AND PRESCRIPTION DECISIONS')
print('=' * 72)
print('Catalog candidates / registered sources:', len(catalog), '/', len(registry))
print('Prescription recommendations / space records:', len(prescriptions), '/', len(spaces))

if any(path.exists() for path in outputs.values()):
    if not all(path.exists() for path in outputs.values()):
        raise FileExistsError('Partial Step 12-50 output set exists; inspect before rerun.')
    existing_summary = json.loads(outputs['summary'].read_text(encoding='utf-8'))
    if existing_summary.get('input_sha256') != {k: sha(p) for k, p in inputs.items()}:
        raise ValueError('Existing Step 12-50 reports refer to different inputs.')
    for key in ('source_evidence', 'space_decisions', 'prescription_decisions'):
        rows = read_csv(outputs[key])
        expected = 18 if key != 'prescription_decisions' else 6
        index_by(rows, 'evidence_id', expected)
        print('PRESERVED', outputs[key].relative_to(root))
    print('PRESERVED', outputs['summary'].relative_to(root))
    print('OVERALL: PASS — previous reports and any reviewer edits preserved')
else:
    urls = sorted({r['source_url'] for r in spaces})
    if len(urls) != 3:
        raise ValueError(f'Expected three official practice plan URLs; found {len(urls)}.')
    pages = {}
    print('\nOFFICIAL SOURCE ACCESS')
    for i, url in enumerate(urls, start=1):
        status, sections = fetch_sections(url)
        pages[url] = (status, sections)
        print(f'  {i:02d}/{len(urls):02d} | {status:<20} | {url}')

    evidence_rows, space_decisions = [], []
    for row in sorted(spaces, key=lambda r: int(r['review_rank'])):
        eid = row['evidence_id']
        description, phrases, basis = space_rules[eid]
        page_status, sections = pages[row['source_url']]
        heading_key = norm(row['drill_name'])
        matches = [s for s in sections if s['heading'] == heading_key]
        if page_status != 'fetched':
            section_status, text = 'page_unavailable', ''
        elif len(matches) != 1:
            section_status, text = 'heading_not_unique_or_missing', ''
        else:
            section_status, text = 'section_found', norm(' '.join(matches[0]['parts']))
        matched = [bool(re.search(pattern, text)) for pattern in phrases]
        supported = section_status == 'section_found' and all(matched)
        if not supported:
            recommendation_status = 'source_evidence_incomplete'
            recommended_space = ''
        elif basis == 'operational_space_inference':
            recommendation_status = 'inference_requires_confirmation'
            recommended_space = ''
        else:
            recommendation_status = 'source_locations_supported_pending_signoff'
            recommended_space = description
        evidence_rows.append({
            'evidence_id': eid,
            'candidate_group_id': row['candidate_group_id'],
            'drill_name': row['drill_name'],
            'source_url': row['source_url'],
            'page_status': page_status,
            'heading_status': section_status,
            'cue_checks_passed': sum(matched) if section_status == 'section_found' else 0,
            'cue_checks_expected': len(phrases),
            'source_location_basis': basis,
            'evidence_status': recommendation_status,
            'source_text_saved': 'no',
        })
        space_decisions.append({
            'evidence_id': eid,
            'candidate_group_id': row['candidate_group_id'],
            'drill_name': row['drill_name'],
            'source_url': row['source_url'],
            'existing_catalog_space': row['existing_catalog_space'],
            'existing_registry_space': row['existing_registry_space'],
            'source_location_proposal': recommended_space,
            'source_location_basis': basis,
            'evidence_status': recommendation_status,
            'minimum_space_dimensions_verified': 'no',
            'review_recommendation': (
                'consider_source_location_summary_without_inventing_dimensions'
                if supported and basis == 'source_locations_reported'
                else 'defer_space_requirement_pending_human_review'
            ),
            'reviewer_decision': 'pending',
            'reviewer': '',
            'review_date': '',
            'review_notes': '',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
        })

    prescription_decisions = []
    for row in prescriptions:
        prescription_decisions.append({
            'evidence_id': row['evidence_id'],
            'candidate_group_id': row['candidate_group_id'],
            'drill_name': row['drill_name'],
            'source_url': row['source_url'],
            'existing_catalog_prescription': row['existing_catalog_prescription'],
            'existing_registry_prescription': row['existing_registry_prescription'],
            'published_prescription_notation': row['published_prescription_notation'],
            'semantic_assessment': row['semantic_assessment'],
            'review_recommendation': 'retain_existing_and_document_source_notation',
            'reviewer_decision': 'pending',
            'reviewer': '',
            'review_date': '',
            'review_notes': '',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
        })

    input_hashes = {k: sha(p) for k, p in inputs.items()}
    payloads = {
        'source_evidence': csv_bytes(evidence_rows),
        'space_decisions': csv_bytes(space_decisions),
        'prescription_decisions': csv_bytes(prescription_decisions),
    }
    summary = {
        'step': '12-50', 'catalog_version': '0.3.5',
        'run_date': date.today().isoformat(),
        'catalog_candidates': len(catalog), 'source_records': len(registry),
        'space_records': len(space_decisions), 'prescription_recommendations': len(prescription_decisions),
        'publisher_pages_fetched': sum(v[0] == 'fetched' for v in pages.values()),
        'publisher_pages_requested': len(pages),
        'space_status_counts': dict(Counter(r['evidence_status'] for r in space_decisions)),
        'space_review_signoffs': 0, 'prescription_review_signoffs': 0,
        'metadata_changes': 0, 'approvals': 0, 'planner_eligible': 0,
        'input_sha256': input_hashes,
        'output_sha256_at_creation': {k: hashlib.sha256(b).hexdigest() for k, b in payloads.items()},
        'limitations': ('Source-location cues are screening evidence only. Minimum dimensions, '
                        'human reviewer decisions, permissions, and youth-planning approval remain pending.'),
    }
    payloads['summary'] = (json.dumps(summary, indent=2, sort_keys=True) + '\n').encode('utf-8')
    for key, data in payloads.items():
        path = outputs[key]
        path.parent.mkdir(parents=True, exist_ok=True)
        with path.open('xb') as stream:
            stream.write(data)
        if path.read_bytes() != data:
            raise RuntimeError(f'Saved output failed verification: {path.name}')
        print('CREATED', path.relative_to(root))

    print('\nSPACE SOURCE FINDINGS')
    for status, count in Counter(r['evidence_status'] for r in space_decisions).items():
        print(f'  {status}: {count}')
    print('\nUNRESOLVED SPACE RECORDS')
    for row in space_decisions:
        if row['evidence_status'] != 'source_locations_supported_pending_signoff':
            print(' ', row['evidence_id'], '|', row['drill_name'], '|', row['evidence_status'])
    print('\nPRESCRIPTION REVIEW')
    print('  Equivalence proposals retained for human confirmation:', len(prescription_decisions))
    print('\nFINAL STATUS')
    print('  Catalog metadata changed: 0')
    print('  Reviewer signoffs / approvals / planner-eligible: 0 / 0 / 0')
    print('  Prior catalog and registry files overwritten: No')
    print('  Git operations performed: None')
    print('OVERALL: PASS — source-location evidence documented; final review pending')
```

### Step 12-51 — Wave 3 Metadata Review and Reconciliation

We will review the source-supported metadata recommendations
prepared during Step 12-50.

The review includes 24 field-level decisions across 23
distinct basketball drill candidates:

- Six prescriptions recommended for retention, with the
  publisher's original notation documented separately.
- Seventeen source-supported court-location descriptions
  recommended for incorporation after review.
- One unresolved space requirement recommended for deferral.

We will distinguish published court locations from verified
minimum-space requirements and planning feasibility.

We will document the review decisions before applying changes
to a new versioned catalog.

We will preserve all existing source records, previous catalog
versions, and approval restrictions.

```python
# Step 12-52 - Confirmed Wave 3 Metadata Reconciliation
# Preserve source prescriptions, document court locations, and retain unresolved constraints.

import csv
import hashlib
import json
from collections import Counter, defaultdict
from datetime import date
from pathlib import Path

VERSION = "0.3.6"
EXPECTED_REVIEW_DIGEST = "0d08965200fa6742a777591debac3b1dc1f321d7b4acf177626febb3c1b8470d"

root = next(
    (p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (p / ".git").exists()),
    None,
)
if root is None:
    raise FileNotFoundError("Run this cell from the Champions Athletic Map Git repository.")

catalog_input = root / "data/processed/basketball_drill_catalog_v0_3_5_REVIEW_ONLY.csv"
registry_input = root / "results/basketball_drill_source_registry_v0_3_5.csv"
review_input = root / "results/Step_12_51_Wave3_Metadata_Recommendations_REVIEW_ONLY.csv"

outputs = {
    "catalog": root / "data/processed/basketball_drill_catalog_v0_3_6_REVIEW_ONLY.csv",
    "registry": root / "results/basketball_drill_source_registry_v0_3_6.csv",
    "decisions": root / "results/basketball_wave3_confirmed_metadata_decisions_v0_3_6.csv",
    "changes": root / "results/basketball_wave3_metadata_change_log_v0_3_6.csv",
    "unresolved": root / "results/basketball_wave3_unresolved_metadata_v0_3_6.csv",
    "summary": root / "results/basketball_step12_52_summary_v0_3_6.json",
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(f"Required input not found: {path.relative_to(root)}")
    with path.open("r", encoding="utf-8-sig", newline="") as stream:
        reader = csv.DictReader(stream)
        rows = list(reader)
        columns = list(reader.fieldnames or [])
    if not rows or not columns or len(columns) != len(set(columns)):
        raise ValueError(f"Invalid or empty CSV: {path.name}")
    return rows, columns


def file_sha(path):
    digest = hashlib.sha256()
    with path.open("rb") as stream:
        for chunk in iter(lambda: stream.read(1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()


def require_columns(columns, required, label):
    missing = set(required) - set(columns)
    if missing:
        raise ValueError(f"{label} missing columns: {', '.join(sorted(missing))}")


def single_index(rows, key, label):
    index = {}
    for row in rows:
        identifier = row.get(key, "").strip()
        if not identifier or identifier in index:
            raise ValueError(f"{label}: missing or duplicated {key}: {identifier!r}")
        index[identifier] = row
    return index


catalog, catalog_columns = read_csv(catalog_input)
registry, registry_columns = read_csv(registry_input)
review, review_columns = read_csv(review_input)

require_columns(catalog_columns, [
    "catalog_candidate_id", "catalog_version", "evidence_ids", "source_prescription",
    "space", "catalog_approval_status", "planner_eligible", "approved_drill_id",
    "permission_status", "youth_planning_status",
], "Catalog")
require_columns(registry_columns, [
    "evidence_id", "catalog_candidate_id", "source_prescription", "space",
    "source_url", "permissions_status",
], "Source registry")
require_columns(review_columns, [
    "review_id", "evidence_id", "candidate_group_id", "drill_name", "field",
    "current_catalog_value", "current_registry_value", "proposed_value",
    "recommendation", "recommended_catalog_value", "review_basis", "verification_limit",
    "source_url", "reviewer_decision", "team_signoff", "permission_review",
    "youth_planning_review", "catalog_approval_status", "planner_eligible",
], "Wave 3 recommendations")

if len(catalog) != 144 or len(registry) != 156 or len(review) != 24:
    raise ValueError("Unexpected input counts; verify Catalog v0.3.5 and the Step 12-51 review file.")

catalog_by_id = single_index(catalog, "catalog_candidate_id", "Catalog")
registry_by_id = single_index(registry, "evidence_id", "Source registry")
review_by_id = single_index(review, "review_id", "Wave 3 recommendations")

if any(r["catalog_version"] != "0.3.5" for r in catalog):
    raise ValueError("Catalog input is not version 0.3.5.")
if any(
    r["catalog_approval_status"] != "pending"
    or r["planner_eligible"] != "no"
    or r["approved_drill_id"]
    for r in catalog
):
    raise ValueError("An existing catalog approval decision requires separate reconciliation.")

# Reconfirm the identity and exact approved values of the review packet.
review_digest_columns = [
    "evidence_id", "candidate_group_id", "field", "current_catalog_value",
    "current_registry_value", "recommendation", "recommended_catalog_value",
    "review_basis", "source_url",
]
review_snapshot = json.dumps(
    [[r.get(field, "") for field in review_digest_columns]
     for r in sorted(review, key=lambda r: (r["evidence_id"], r["field"]))],
    ensure_ascii=False,
    separators=(",", ":"),
)
review_digest = hashlib.sha256(review_snapshot.encode("utf-8")).hexdigest()
if review_digest != EXPECTED_REVIEW_DIGEST:
    raise ValueError(
        "Step 12-51 recommendations differ from the reviewed 24-decision snapshot. "
        "Do not apply changes until the edited recommendations are reconciled."
    )

recommended_counts = Counter(r["recommendation"] for r in review)
expected_counts = {
    "retain_current_and_record_publisher_notation": 6,
    "accept_source_location_description_pending_signoff": 17,
    "defer_unverified_space_requirement": 1,
}
if dict(recommended_counts) != expected_counts:
    raise ValueError(f"Unexpected Wave 3 recommendation counts: {dict(recommended_counts)}")

review_groups = defaultdict(list)
for record in review:
    eid = record["evidence_id"]
    gid = record["candidate_group_id"]
    field = record["field"]
    if eid not in registry_by_id or gid not in catalog_by_id:
        raise ValueError(f"Unregistered review relationship: {eid} / {gid}")
    cat_row, reg_row = catalog_by_id[gid], registry_by_id[eid]
    linked_evidence = [x.strip() for x in cat_row["evidence_ids"].split(";") if x.strip()]
    if linked_evidence != [eid] or reg_row["catalog_candidate_id"] != gid:
        raise ValueError(f"Expected one-to-one source relationship for {eid} / {gid}.")
    if field not in {"space", "source_prescription"}:
        raise ValueError(f"Unexpected review field: {field}")
    if cat_row[field] != record["current_catalog_value"]:
        raise ValueError(f"Catalog value changed after Step 12-51: {eid} / {field}")
    if reg_row[field] != record["current_registry_value"]:
        raise ValueError(f"Registry value changed after Step 12-51: {eid} / {field}")
    if reg_row["source_url"] != record["source_url"]:
        raise ValueError(f"Source URL changed: {eid}")
    if record["reviewer_decision"] not in {"", "pending"}:
        raise ValueError(f"Existing individual field decision detected: {eid} / {field}")
    if (
        record["catalog_approval_status"] != "pending"
        or record["planner_eligible"] != "no"
        or record["team_signoff"] != "pending"
        or record["permission_review"] != "pending"
        or record["youth_planning_review"] != "pending"
    ):
        raise ValueError(f"Unexpected approval or sign-off state: {eid} / {field}")
    review_groups[gid].append(record)

if len(review_groups) != 23 or any(len(entries) > 2 for entries in review_groups.values()):
    raise ValueError("Expected 24 decisions affecting 23 candidate groups.")
if any(
    r["recommendation"] == "defer_unverified_space_requirement"
    and (r["evidence_id"] != "JR-EVID-0055" or r["field"] != "space")
    for r in review
):
    raise ValueError("Unexpected deferred field identity.")

input_checksums = {
    str(path.relative_to(root)): file_sha(path)
    for path in (catalog_input, registry_input, review_input)
}

# A repeated run verifies the existing immutable outputs rather than overwriting them.
existing = [path for path in outputs.values() if path.exists()]
if existing:
    if len(existing) != len(outputs):
        raise FileExistsError("Partial Step 12-52 outputs exist; inspect before rerunning.")
    with outputs["summary"].open("r", encoding="utf-8") as stream:
        previous = json.load(stream)
    if previous.get("input_sha256") != input_checksums:
        raise ValueError("Inputs differ from those used for the saved Step 12-52 outputs.")
    expected_sha = previous.get("output_sha256", {})
    for name, path in outputs.items():
        if name == "summary":
            continue
        if expected_sha.get(str(path.relative_to(root))) != file_sha(path):
            raise ValueError(f"Saved Step 12-52 output changed: {path.name}")
    print("STEP 12-52 — EXISTING OUTPUTS VERIFIED")
    print("Catalog candidates / registered sources: 144 / 156")
    print("Confirmed decisions: 6 retain / 17 source locations / 1 defer")
    print("All six outputs preserved; no files modified.")
    print("OVERALL: PASS — existing versioned outputs verified")
else:
    decision_date = date.today().isoformat()
    decision_rows = []
    change_rows = []
    unresolved_rows = []
    applied_groups = set()
    statuses_by_gid = defaultdict(list)
    statuses_by_eid = defaultdict(list)

    for r in review:
        gid, eid, field = r["candidate_group_id"], r["evidence_id"], r["field"]
        action = r["recommendation"]
        catalog_row, registry_row = catalog_by_id[gid], registry_by_id[eid]
        before_catalog, before_registry = catalog_row[field], registry_row[field]
        final_value = r["recommended_catalog_value"]

        if action == "accept_source_location_description_pending_signoff":
            if field != "space" or not final_value.strip():
                raise ValueError(f"Invalid source-location proposal: {eid}")
            catalog_row[field] = final_value
            registry_row[field] = final_value
            applied_groups.add(gid)
            status = "location_documented_team_signoff_pending"
        elif action == "retain_current_and_record_publisher_notation":
            if field != "source_prescription" or final_value != before_catalog:
                raise ValueError(f"Prescription retention mismatch: {eid}")
            status = "prescription_equivalence_documented"
        elif action == "defer_unverified_space_requirement":
            if field != "space" or final_value != before_catalog:
                raise ValueError(f"Deferred field would change: {eid}")
            status = "minimum_space_unverified_deferred"
            unresolved_rows.append({
                "candidate_group_id": gid,
                "evidence_id": eid,
                "drill_name": r["drill_name"],
                "field": field,
                "retained_value": before_catalog,
                "reason": r["verification_limit"],
                "next_action": "Verify feasible play-area requirements before planning eligibility.",
                "source_url": r["source_url"],
                "review_status": "deferred_pending_additional_evidence",
            })
        else:
            raise ValueError(f"Unrecognized review action: {action}")

        statuses_by_gid[gid].append(status)
        statuses_by_eid[eid].append(status)

        if action == "retain_current_and_record_publisher_notation":
            catalog_row["wave3_publisher_prescription_notation"] = r["proposed_value"]
            registry_row["wave3_publisher_prescription_notation"] = r["proposed_value"]
        if action == "accept_source_location_description_pending_signoff":
            catalog_row["wave3_space_location_evidence_basis"] = r["review_basis"]
            registry_row["wave3_space_location_evidence_basis"] = r["review_basis"]
            catalog_row["wave3_minimum_space_dimensions_verified"] = "no"
            registry_row["wave3_minimum_space_dimensions_verified"] = "no"
        if action == "defer_unverified_space_requirement":
            catalog_row["wave3_space_deferred"] = "yes"
            registry_row["wave3_space_deferred"] = "yes"

        decision_rows.append({
            "review_id": r["review_id"],
            "candidate_group_id": gid,
            "evidence_id": eid,
            "drill_name": r["drill_name"],
            "field": field,
            "user_confirmed_decision": action,
            "catalog_before": before_catalog,
            "catalog_after": catalog_row[field],
            "registry_before": before_registry,
            "registry_after": registry_row[field],
            "published_notation_or_proposal": r["proposed_value"],
            "evidence_basis": r["review_basis"],
            "verification_limit": r["verification_limit"],
            "source_url": r["source_url"],
            "decision_recorded_date": decision_date,
            "decision_authority": "user_confirmed_metadata_review",
            "team_signoff": "pending",
            "permission_review": "pending",
            "youth_planning_review": "pending",
            "catalog_approval_status": "pending",
            "planner_eligible": "no",
        })

        if catalog_row[field] != before_catalog or registry_row[field] != before_registry:
            change_rows.append({
                "candidate_group_id": gid,
                "evidence_id": eid,
                "drill_name": r["drill_name"],
                "field": field,
                "catalog_before": before_catalog,
                "catalog_after": catalog_row[field],
                "registry_before": before_registry,
                "registry_after": registry_row[field],
                "source_url": r["source_url"],
                "evidence_basis": r["review_basis"],
                "team_signoff": "pending",
                "changed_on": decision_date,
            })

    # Add explicit review metadata without changing unrelated catalog fields.
    extra_catalog_columns = [
        "wave3_metadata_decision_status", "wave3_metadata_decision_reference",
        "wave3_publisher_prescription_notation", "wave3_space_location_evidence_basis",
        "wave3_minimum_space_dimensions_verified", "wave3_space_deferred",
    ]
    extra_registry_columns = extra_catalog_columns[:]
    for row in catalog:
        for field in extra_catalog_columns:
            row.setdefault(field, "")
        row["catalog_version"] = VERSION
        gid = row["catalog_candidate_id"]
        if gid in statuses_by_gid:
            row["wave3_metadata_decision_status"] = ";".join(statuses_by_gid[gid])
            row["wave3_metadata_decision_reference"] = (
                "results/basketball_wave3_confirmed_metadata_decisions_v0_3_6.csv"
            )
    for row in registry:
        for field in extra_registry_columns:
            row.setdefault(field, "")
        eid = row["evidence_id"]
        if eid in statuses_by_eid:
            row["wave3_metadata_decision_status"] = ";".join(statuses_by_eid[eid])
            row["wave3_metadata_decision_reference"] = (
                "results/basketball_wave3_confirmed_metadata_decisions_v0_3_6.csv"
            )

    # Verify the intended 17 value changes, 6 retained prescriptions, and 1 deferral.
    if (len(decision_rows), len(change_rows), len(unresolved_rows), len(applied_groups)) != (24, 17, 1, 17):
        raise ValueError("Reconciled metadata counts differ from the confirmed decisions.")
    if any(
        row["catalog_approval_status"] != "pending"
        or row["planner_eligible"] != "no"
        or row["approved_drill_id"]
        or row["permission_status"] != "pending"
        for row in catalog
    ):
        raise ValueError("Approval or permission values changed unexpectedly.")
    if any(r["permissions_status"] not in {"pending_review", "pending"} for r in registry):
        raise ValueError("Source permissions state changed unexpectedly.")

    # Save new artifacts only. Previous catalog, registry, and review files remain immutable.
    for path in outputs.values():
        path.parent.mkdir(parents=True, exist_ok=True)
    csv_artifacts = {
        "catalog": (catalog, catalog_columns + [x for x in extra_catalog_columns if x not in catalog_columns]),
        "registry": (registry, registry_columns + [x for x in extra_registry_columns if x not in registry_columns]),
        "decisions": (decision_rows, list(decision_rows[0])),
        "changes": (change_rows, list(change_rows[0])),
        "unresolved": (unresolved_rows, list(unresolved_rows[0])),
    }
    for name, (rows, columns) in csv_artifacts.items():
        with outputs[name].open("x", newline="", encoding="utf-8") as stream:
            writer = csv.DictWriter(stream, fieldnames=columns, extrasaction="raise")
            writer.writeheader()
            writer.writerows(rows)

    summary = {
        "step": "12-52",
        "catalog_version": VERSION,
        "decision_date": decision_date,
        "catalog_groups": len(catalog),
        "source_occurrences": len(registry),
        "affected_candidate_groups": len(review_groups),
        "confirmed_decisions": dict(recommended_counts),
        "confirmed_prescription_retentions": 6,
        "confirmed_source_location_descriptions": 17,
        "deferred_space_requirements": 1,
        "changed_catalog_fields": 17,
        "changed_registry_fields": 17,
        "team_signoffs": 0,
        "catalog_approved": 0,
        "planner_eligible": 0,
        "input_sha256": input_checksums,
        "review_snapshot_sha256": review_digest,
        "output_sha256": {
            str(path.relative_to(root)): file_sha(path)
            for name, path in outputs.items() if name != "summary"
        },
        "pending_gates": ["team_signoff", "permissions", "youth_planning", "canonical_review"],
    }
    with outputs["summary"].open("x", encoding="utf-8") as stream:
        json.dump(summary, stream, indent=2, ensure_ascii=False)
        stream.write("\n")

    for name, (expected_rows, _) in csv_artifacts.items():
        saved, _ = read_csv(outputs[name])
        if len(saved) != len(expected_rows):
            raise ValueError(f"Output verification failed: {outputs[name].name}")

    print("=" * 72)
    print("STEP 12-52 — CONFIRMED WAVE 3 METADATA RECONCILIATION")
    print("=" * 72)
    print("Catalog candidates / source occurrences: 144 / 156")
    print("Affected candidate groups / field decisions: 23 / 24")
    print("\nCONFIRMED DECISIONS")
    print("  Retain existing source prescriptions: 6")
    print("  Apply source-supported location descriptions: 17")
    print("  Defer uncertain space requirement: 1")
    print("\nACTUAL METADATA CHANGES")
    print("  Catalog space values updated: 17")
    print("  Source registry space values updated: 17")
    print("  Publisher prescription notations recorded: 6")
    print("  Deferred: JR-EVID-0055 | 1 on 1 Dribble Battle | space")
    print("\nOUTPUTS")
    for name, path in outputs.items():
        print(f"  CREATED {path.relative_to(root)}")
    print("\nTeam signoff / permissions / youth-planning: PENDING")
    print("Approved / planner-eligible: 0 / 0")
    print("Earlier catalog versions and source datasets modified: No")
    print("Git operations performed: None")
    print("OVERALL: PASS — Wave 3 decisions documented in v0.3.6; approvals pending")
    print("=" * 72)
```

```python
# Step 12-53 - Wave 3 Equipment and Participant Evidence Reliability
# Reconcile current metadata with historical source screens; keep every field pending independent review.

import csv
import hashlib
import io
import json
from collections import Counter, defaultdict
from datetime import date
from pathlib import Path

ROOT = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / ".git").exists()), None)
if ROOT is None:
    raise FileNotFoundError("Run this cell from the Champions Athletic Map Git repository.")

inputs = {
    "catalog": ROOT / "data/processed/basketball_drill_catalog_v0_3_6_REVIEW_ONLY.csv",
    "registry": ROOT / "results/basketball_drill_source_registry_v0_3_6.csv",
    "readiness": ROOT / "results/basketball_catalog_readiness_worklist_v0_3_1.csv",
    "publisher_checks": ROOT / "results/basketball_wave3_publisher_checks_v0_3_5.csv",
    "field_comparisons": ROOT / "results/basketball_wave3_field_comparisons_v0_3_5.csv",
    "version_summary": ROOT / "results/basketball_step12_52_summary_v0_3_6.json",
}
outputs = {
    "evidence": ROOT / "results/basketball_wave3_requirement_evidence_reliability_v0_3_6.csv",
    "decision_sheet": ROOT / "results/basketball_wave3_requirement_adjudication_v0_3_6.csv",
    "worklist": ROOT / "results/basketball_wave3_requirement_review_queue_v0_3_6.csv",
    "summary": ROOT / "results/basketball_step12_53_summary_v0_3_6.json",
}


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(f"Missing input: {path.relative_to(ROOT)}")
    with path.open(newline="", encoding="utf-8-sig") as stream:
        reader = csv.DictReader(stream)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or len(columns) != len(set(columns)) or not rows:
        raise ValueError(f"Invalid or empty CSV: {path.name}")
    if any(None in row or any(value is None for value in row.values()) for row in rows):
        raise ValueError(f"Malformed CSV: {path.name}")
    return rows, columns


def require(rows, columns, expected, label):
    missing = set(expected) - set(columns)
    if missing:
        raise ValueError(f"{label}: missing columns: {', '.join(sorted(missing))}")
    return rows


def unique_index(rows, field, label):
    identifiers = [r[field].strip() for r in rows]
    if not all(identifiers) or len(identifiers) != len(set(identifiers)):
        raise ValueError(f"{label}: blank or duplicated {field}")
    return {r[field]: r for r in rows}


def sha(path):
    digest = hashlib.sha256()
    with path.open("rb") as stream:
        for block in iter(lambda: stream.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()


def encode_csv(rows):
    if not rows:
        raise ValueError("An empty review report cannot be saved.")
    stream = io.StringIO(newline="")
    writer = csv.DictWriter(stream, fieldnames=list(rows[0]), lineterminator="\n")
    writer.writeheader()
    writer.writerows(rows)
    return stream.getvalue().encode("utf-8")


catalog, cat_cols = read_csv(inputs["catalog"])
registry, reg_cols = read_csv(inputs["registry"])
readiness, ready_cols = read_csv(inputs["readiness"])
publisher, pub_cols = read_csv(inputs["publisher_checks"])
comparisons, comp_cols = read_csv(inputs["field_comparisons"])

require(catalog, cat_cols, ["catalog_candidate_id", "catalog_version", "drill_name", "evidence_ids", "equipment", "participants", "catalog_approval_status", "planner_eligible", "approved_drill_id"], "Catalog")
require(registry, reg_cols, ["evidence_id", "catalog_candidate_id", "drill_name", "source_url", "equipment", "participants", "permissions_status"], "Registry")
require(readiness, ready_cols, ["candidate_group_id", "review_wave", "review_rank"], "Readiness")
require(publisher, pub_cols, ["candidate_group_id", "evidence_id", "drill_name", "official_source_url", "heading_screen", "section_word_count", "source_equipment_cues", "source_participant_cues"], "Publisher checks")
require(comparisons, comp_cols, ["candidate_group_id", "evidence_id", "field", "catalog_value", "registry_value", "source_section_cues", "source_text_screen", "independent_field_signoff"], "Field comparisons")

if len(catalog) != 144 or len(registry) != 156 or len(readiness) != 144:
    raise ValueError("Unexpected catalog/registry/readiness counts; inspect the current version.")
if len(publisher) != 42 or len(comparisons) != 168:
    raise ValueError("Unexpected Wave 3 publisher/field comparison counts.")
if any(r["catalog_version"] != "0.3.6" for r in catalog):
    raise ValueError("Expected Catalog version 0.3.6.")
if any(r["catalog_approval_status"] != "pending" or r["planner_eligible"] != "no" or r["approved_drill_id"].strip() for r in catalog):
    raise ValueError("An existing approval status changed; stop for review.")

# Match the checksums from Step 12-52 regardless of Windows or POSIX path separators.
# The file contents are still verified byte-for-byte using SHA-256.
previous = json.loads(inputs["version_summary"].read_text(encoding="utf-8"))
recorded_hashes = previous.get("output_sha256", {})
if not isinstance(recorded_hashes, dict) or not recorded_hashes:
    raise ValueError("Step 12-52 summary has no output checksums; review the summary before proceeding.")
normalized_hashes = {}
for recorded_path, recorded_digest in recorded_hashes.items():
    normalized_path = str(recorded_path).replace("\\", "/")
    if normalized_path in normalized_hashes and normalized_hashes[normalized_path] != recorded_digest:
        raise ValueError(f"Conflicting Step 12-52 checksums for {normalized_path}.")
    normalized_hashes[normalized_path] = recorded_digest

for key in ("catalog", "registry"):
    identifier = inputs[key].relative_to(ROOT).as_posix()
    expected_digest = normalized_hashes.get(identifier)
    if expected_digest is None:
        raise ValueError(f"{key}: checksum entry not found in Step 12-52 summary ({identifier}).")
    current_digest = sha(inputs[key])
    if current_digest != expected_digest:
        raise ValueError(
            f"{key}: actual SHA-256 mismatch after path normalization. "
            f"Expected {expected_digest[:16]}..., got {current_digest[:16]}...; "
            "do not overwrite the original file."
        )
print("STEP 12-52 INPUT INTEGRITY: PASS — catalog and registry SHA-256 match.")

cat_by = unique_index(catalog, "catalog_candidate_id", "Catalog")
reg_by = unique_index(registry, "evidence_id", "Registry")
pub_by = unique_index(publisher, "evidence_id", "Publisher checks")
selected = sorted((r for r in readiness if r["review_wave"] == "3"),
                  key=lambda r: int(r["review_rank"]))
if len(selected) != 42:
    raise ValueError("Expected exactly 42 Wave 3 candidate groups.")
expected_ids = set()
for wave in selected:
    group_id = wave["candidate_group_id"]
    if group_id not in cat_by:
        raise ValueError(f"Missing Wave 3 catalog group: {group_id}")
    linked = [s.strip() for s in cat_by[group_id]["evidence_ids"].split(";") if s.strip()]
    if len(linked) != 1 or linked[0] not in reg_by or reg_by[linked[0]]["catalog_candidate_id"] != group_id:
        raise ValueError(f"Expected one historical source occurrence for {group_id}")
    expected_ids.add(linked[0])
if expected_ids != set(pub_by):
    raise ValueError("Wave 3 source records differ from publisher-check report.")

field_by = {}
for rec in comparisons:
    key = (rec["evidence_id"], rec["field"])
    if key in field_by:
        raise ValueError(f"Duplicate field comparison: {key}")
    field_by[key] = rec
if set(field_by) != {(eid, field) for eid in expected_ids for field in ("equipment", "participants", "space", "prescription")}:
    raise ValueError("Incomplete Wave 3 field-comparison coverage.")

# Repetition in the earlier cue extract is a diagnostic indicator, not proof of an error.
cue_counts = {name: Counter(pub_by[eid][column].strip().casefold() for eid in expected_ids)
              for name, column in (("equipment", "source_equipment_cues"),
                                   ("participants", "source_participant_cues"))}
cue_page_counts = defaultdict(set)
for eid in expected_ids:
    row = pub_by[eid]
    for field, col in (("equipment", "source_equipment_cues"),
                       ("participants", "source_participant_cues")):
        cue_page_counts[(field, row[col].strip().casefold())].add(row["official_source_url"])

# Output records are informational, with no automatic requirement changes.
evidence_rows, decision_rows, worklist_rows = [], [], []
for selected_item in selected:
    gid = selected_item["candidate_group_id"]
    cat = cat_by[gid]
    eid = cat["evidence_ids"].strip()
    reg = reg_by[eid]
    page = pub_by[eid]
    if page["candidate_group_id"] != gid or page["official_source_url"] != reg["source_url"]:
        raise ValueError(f"Publisher/source relationship changed: {gid} / {eid}")
    try:
        word_count = int(page["section_word_count"])
    except ValueError:
        word_count = 0
    group_flags = []
    field_risk = {}
    for field in ("equipment", "participants"):
        old = field_by[(eid, field)]
        cat_value, reg_value = cat[field].strip(), reg[field].strip()
        if old["candidate_group_id"] != gid:
            raise ValueError(f"Comparison group mismatch: {eid} / {field}")
        if old["catalog_value"] != cat[field] or old["registry_value"] != reg[field]:
            raise ValueError(f"Current {field} differs from saved source comparison: {eid}")
        if old["independent_field_signoff"].strip().casefold() not in ("pending", ""):
            raise ValueError(f"Field contains a prior signed review: {eid} / {field}")
        cue = old["source_section_cues"].strip()
        if cue != page["source_equipment_cues" if field == "equipment" else "source_participant_cues"]:
            raise ValueError(f"Recorded source cues conflict: {eid} / {field}")
        used_by = cue_counts[field][cue.casefold()]
        across_pages = len(cue_page_counts[(field, cue.casefold())])
        reused_very_broadly = bool(cue and used_by >= 10 and across_pages >= 3)
        discrepancy = cat_value.casefold() != reg_value.casefold()
        vague = not cat_value or cat_value.casefold() in {
            "not specified", "unknown", "n/a", "none stated", "see source practice plan"}
        small_section = word_count < 20
        flags = []
        if reused_very_broadly:
            flags.append("generic_cues_repeated_across_source_pages")
        if small_section:
            flags.append("brief_section_extract_needs_independent_context")
        if not cue:
            flags.append("no_keyword_evidence")
        if discrepancy:
            flags.append("catalog_registry_wording_difference")
        if vague:
            flags.append("missing_or_unspecified_requirement")
        if field == "equipment" and "cones" in cue.casefold() and "cone" not in cat_value.casefold():
            flags.append("cone_keyword_not_in_recorded_equipment_do_not_auto_add")
        if field == "participants" and cue.casefold() == "player" and any(
            x in cat_value.casefold() for x in ("pair", "team", "coach", "defender", "group")):
            flags.append("specific_participant_role_not_established_by_generic_cue")
        if discrepancy or vague:
            priority = "high"
        elif reused_very_broadly or small_section or not cue:
            priority = "medium"
        else:
            priority = "standard"
        field_risk[field] = priority
        group_flags.extend(flags)
        item = {
            "candidate_group_id": gid,
            "evidence_id": eid,
            "drill_name": cat["drill_name"],
            "field": field,
            "current_catalog_value": cat_value,
            "current_registry_value": reg_value,
            "source_url": reg["source_url"],
            "publisher_heading_screen": page["heading_screen"],
            "source_section_word_count": word_count,
            "previous_source_cues": cue,
            "same_cues_across_drills": used_by,
            "same_cues_across_pages": across_pages,
            "reliability_flags": ";".join(dict.fromkeys(flags)),
            "automated_evidence_status": "screening_only_not_requirement_verification",
            "review_priority": priority,
        }
        evidence_rows.append(item)
        decision_rows.append({
            **item,
            "recommended_review_action": (
                "Check the specific original drill instructions and distinguish "
                "mandatory requirements from optional variations; do not rely "
                "on repeated page-level keywords."
            ),
            "reviewer_decision": "pending",
            "reviewed_value": "",
            "independent_source_evidence_note": "",
            "reviewer": "",
            "review_date": "",
            "team_signoff": "pending",
            "permission_review": "pending",
            "youth_planning_review": "pending",
            "catalog_approval_status": "pending",
            "planner_eligible": "no",
        })
    group_priority = "high" if "high" in field_risk.values() else (
        "medium" if "medium" in field_risk.values() else "standard")
    worklist_rows.append({
        "review_rank": selected_item["review_rank"],
        "candidate_group_id": gid,
        "evidence_id": eid,
        "drill_name": cat["drill_name"],
        "curriculum_level": reg["source_level"],
        "practice_plan": reg["practice_plan"],
        "source_url": reg["source_url"],
        "current_equipment": cat["equipment"],
        "current_participants": cat["participants"],
        "equipment_review_priority": field_risk["equipment"],
        "participant_review_priority": field_risk["participants"],
        "overall_priority": group_priority,
        "review_flags": ";".join(dict.fromkeys(group_flags)),
        "review_status": "pending_independent_source_review",
        "team_signoff": "pending",
        "planner_eligible": "no",
    })

assert len(evidence_rows) == len(decision_rows) == 84 and len(worklist_rows) == 42
worklist_rows.sort(key=lambda r: ({"high": 0, "medium": 1, "standard": 2}[r["overall_priority"]], int(r["review_rank"])))

input_hashes = {str(path.relative_to(ROOT)).replace("\\", "/"): sha(path)
                for path in inputs.values()}
report_bytes = {
    "evidence": encode_csv(evidence_rows),
    "decision_sheet": encode_csv(decision_rows),
    "worklist": encode_csv(worklist_rows),
}

summary = {
    "step": "12-53",
    "catalog_version": "0.3.6",
    "run_date": date.today().isoformat(),
    "catalog_candidates": 144,
    "source_occurrences": 156,
    "wave3_candidates": 42,
    "requirement_field_reviews": 84,
    "publisher_heading_matches": Counter(r["heading_screen"] for r in publisher),
    "equipment_cue_distribution": dict(cue_counts["equipment"]),
    "participant_cue_distribution": dict(cue_counts["participants"]),
    "evidence_flags": dict(Counter(flag for row in evidence_rows
        for flag in row["reliability_flags"].split(";") if flag)),
    "review_priorities": dict(Counter(r["overall_priority"] for r in worklist_rows)),
    "signoffs": 0,
    "approved": 0,
    "planner_eligible": 0,
    "input_sha256": input_hashes,
    "report_sha256": {key: hashlib.sha256(payload).hexdigest()
                      for key, payload in report_bytes.items()},
    "interpretation": "Repeated keyword cues are screening flags, not drill-specific evidence of requirements.",
}
summary_bytes = (json.dumps(summary, indent=2, ensure_ascii=False, default=dict) + "\n").encode("utf-8")

existing = [p for p in outputs.values() if p.exists()]
print("=" * 72)
print("STEP 12-53 — WAVE 3 REQUIREMENT EVIDENCE RELIABILITY")
print("=" * 72)
print("Catalog candidates / source occurrences: 144 / 156")
print("Wave 3 candidates / equipment-participant fields: 42 / 84")

if existing:
    if len(existing) != len(outputs):
        raise FileExistsError("Partial Step 12-53 outputs exist. Inspect before rerunning.")
    prior = json.loads(outputs["summary"].read_text(encoding="utf-8"))
    if prior.get("input_sha256") != input_hashes:
        raise ValueError("Inputs changed since the Step 12-53 outputs were created.")
    for key in report_bytes:
        if sha(outputs[key]) != prior["report_sha256"].get(key):
            raise ValueError(f"Step 12-53 file changed: {outputs[key].name}; preserve reviewer edits.")
    print("EXISTING OUTPUTS VERIFIED — no files overwritten.")
else:
    for key, payload in report_bytes.items():
        outputs[key].parent.mkdir(parents=True, exist_ok=True)
        with outputs[key].open("xb") as handle:
            handle.write(payload)
    with outputs["summary"].open("xb") as handle:
        handle.write(summary_bytes)
    for key, path in outputs.items():
        print(f"  CREATED  {path.relative_to(ROOT)}")

print("\nSOURCE CUE RELIABILITY")
for label in ("equipment", "participants"):
    top, count = cue_counts[label].most_common(1)[0]
    print(f"  {label:12s} | most reused: {top!r} | {count}/42 records")
print("  Cue values repeated across multiple unrelated drills require review;")
print("  matching source headings alone do not verify requirements.")
print("\nREVIEW FLAGS")
for flag, number in Counter(f for row in evidence_rows for f in row["reliability_flags"].split(";") if f).most_common():
    print(f"  {flag:<58} {number}")
print("\nREVIEW PRIORITIES (CANDIDATES)")
for priority in ("high", "medium", "standard"):
    print(f"  {priority}: {sum(r['overall_priority']==priority for r in worklist_rows)}")
print("\nReviewer decisions / catalog approvals / planner eligible: 0 / 0 / 0")
print("Source datasets and Catalog v0.3.6 modified: No")
print("Git operations performed: None")
print("OVERALL: PASS — evidence reliability assessed; individual source reviews remain pending")
print("=" * 72)
```

```python
# Step 12-54 - Wave 3 Drill-Specific Equipment and Participant Review
# Verify source-scoped evidence and prepare field decisions without changing approvals.

import csv
import hashlib
import io
import json
import re
import unicodedata
from collections import Counter, defaultdict
from datetime import date
from html.parser import HTMLParser
from pathlib import Path
from urllib.parse import urlsplit
from urllib.request import Request, urlopen
from urllib.error import HTTPError, URLError

ROOT = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / ".git").exists()), None)
if ROOT is None:
    raise FileNotFoundError("Run the cell inside the Champions Athletic Map Git repository.")

INPUTS = {
    "catalog": ROOT / "data/processed/basketball_drill_catalog_v0_3_6_REVIEW_ONLY.csv",
    "registry": ROOT / "results/basketball_drill_source_registry_v0_3_6.csv",
    "version_summary": ROOT / "results/basketball_step12_52_summary_v0_3_6.json",
    "evidence": ROOT / "results/basketball_wave3_requirement_evidence_reliability_v0_3_6.csv",
    "adjudication": ROOT / "results/basketball_wave3_requirement_adjudication_v0_3_6.csv",
    "worklist": ROOT / "results/basketball_wave3_requirement_review_queue_v0_3_6.csv",
}
OUTPUTS = {
    "source_checks": ROOT / "results/basketball_wave3_drill_specific_source_evidence_v0_3_6.csv",
    "field_review": ROOT / "results/basketball_wave3_requirement_field_review_v0_3_6.csv",
    "candidate_review": ROOT / "results/basketball_wave3_requirement_reviewer_worklist_v0_3_6.csv",
    "summary": ROOT / "results/basketball_step12_54_summary_v0_3_6.json",
}


def load_csv(path, required):
    if not path.is_file():
        raise FileNotFoundError(f"Missing input: {path.relative_to(ROOT)}")
    with path.open(newline="", encoding="utf-8-sig") as stream:
        reader = csv.DictReader(stream)
        fields = list(reader.fieldnames or [])
        records = list(reader)
    if not records or len(fields) != len(set(fields)) or set(required) - set(fields):
        raise ValueError(f"CSV schema or rows invalid: {path.name}; missing {sorted(set(required) - set(fields))}")
    if any(None in row or any(value is None for value in row.values()) for row in records):
        raise ValueError(f"Malformed CSV row in {path.name}")
    return records


def index_unique(records, field, label):
    ids = [r[field].strip() for r in records]
    if not all(ids) or len(ids) != len(set(ids)):
        raise ValueError(f"{label}: duplicate or blank {field}")
    return {row[field].strip(): row for row in records}


def file_sha(path):
    with path.open("rb") as stream:
        digest = hashlib.sha256()
        for block in iter(lambda: stream.read(1_048_576), b""):
            digest.update(block)
    return digest.hexdigest()


def normalize(text):
    text = unicodedata.normalize("NFKC", str(text or "")).casefold()
    return re.sub(r"[^a-z0-9]+", " ", text).strip()


def encode_csv(rows):
    buffer = io.StringIO(newline="")
    writer = csv.DictWriter(buffer, fieldnames=list(rows[0]), lineterminator="\n", extrasaction="raise")
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode("utf-8")


class DrillSectionParser(HTMLParser):
    """Extract only text between official h2 activity headings, in memory."""

    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.current = None
        self.h2 = False
        self.heading_parts = []
        self.skip_tag = None
        self.skip_depth = 0

    def handle_starttag(self, tag, attrs):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth += 1
            return
        if tag in {"script", "style", "noscript", "svg"}:
            self.skip_tag, self.skip_depth = tag, 1
        elif tag == "h2":
            self.h2, self.heading_parts = True, []
        elif self.current is not None and tag in {"p", "li", "div", "br", "td", "tr"}:
            self.current["parts"].append(" ")

    def handle_data(self, data):
        if self.skip_depth:
            return
        if self.h2:
            self.heading_parts.append(data)
        elif self.current is not None:
            self.current["parts"].append(data)

    def handle_endtag(self, tag):
        if self.skip_depth:
            if tag == self.skip_tag:
                self.skip_depth -= 1
                if self.skip_depth == 0:
                    self.skip_tag = None
            return
        if tag == "h2" and self.h2:
            heading = " ".join(self.heading_parts).strip()
            self.h2 = False
            if heading:
                self.current = {"heading": heading, "parts": []}
                self.sections.append(self.current)


def specific_cues(text, field):
    # A cue documents source-section language, not a mandatory requirement.
    checks = {
        "equipment": {
            "basketball_or_ball": r"\b(?:basketball|basketballs|ball|balls)\b",
            "basket_or_hoop": r"\b(?:basket|baskets|hoop|hoops|rim)\b",
            "cones_or_markers": r"\b(?:cones?|markers?)\b",
            "chair_or_screen": r"\b(?:chairs?|screens?)\b",
            "optional_or_shared": r"\b(?:share|shared|optional|variation|variations)\b",
            "two_ball_description": r"\b(?:two|2)\s+(?:basketballs?|balls?)\b",
        },
        "participants": {
            "pairs_or_partners": r"\b(?:pairs?|partners?|partnered)\b",
            "teams_or_groups": r"\b(?:teams?|groups?)\b",
            "coach": r"\b(?:coach|coaches)\b",
            "defender": r"\b(?:defender|defenders|defense|defensive)\b",
            "passer_or_receiver": r"\b(?:passer|passers|receiver|receivers)\b",
            "five_lines": r"\b(?:five|5)\s+lines\b",
            "multiple_lines": r"\b(?:two|2|three|3|four|4)\s+(?:or\s+more\s+)?(?:group\s+)?lines\b",
            "roles_rotate": r"\b(?:rotate|rotation|switch roles|next player)\b",
        },
    }
    return [label for label, pattern in checks[field].items() if re.search(pattern, text)]


catalog = load_csv(INPUTS["catalog"], [
    "catalog_candidate_id", "drill_name", "evidence_ids", "equipment", "participants",
    "catalog_approval_status", "planner_eligible", "approved_drill_id",
])
registry = load_csv(INPUTS["registry"], [
    "evidence_id", "catalog_candidate_id", "source_url", "equipment", "participants",
])
evidence = load_csv(INPUTS["evidence"], [
    "candidate_group_id", "evidence_id", "field", "current_catalog_value",
    "current_registry_value", "source_url", "reliability_flags",
])
adjudication = load_csv(INPUTS["adjudication"], [
    "candidate_group_id", "evidence_id", "field", "reviewer_decision",
    "team_signoff", "permission_review", "youth_planning_review",
    "catalog_approval_status", "planner_eligible",
])
worklist = load_csv(INPUTS["worklist"], [
    "candidate_group_id", "evidence_id", "drill_name", "source_url",
    "current_equipment", "current_participants", "overall_priority",
])
if (len(catalog), len(registry), len(evidence), len(adjudication), len(worklist)) != (144, 156, 84, 84, 42):
    raise ValueError("Unexpected Step 12-53 input counts; inspect saved reports.")

# Verify catalog and registry against Step 12-52 without OS-dependent key separators.
summary_path = INPUTS["version_summary"]
if not summary_path.is_file():
    raise FileNotFoundError(summary_path)
previous = json.loads(summary_path.read_text(encoding="utf-8"))
saved_hashes = {
    str(name).replace("\\", "/"): value
    for name, value in previous.get("output_sha256", {}).items()
}
for key in ("catalog", "registry"):
    path = INPUTS[key]
    relative = path.relative_to(ROOT).as_posix()
    if file_sha(path) != saved_hashes.get(relative):
        raise ValueError(f"Input integrity check failed: {relative}")
print("INPUT INTEGRITY: PASS — current Catalog v0.3.6 and registry match Step 12-52.")

cat_by = index_unique(catalog, "catalog_candidate_id", "Catalog")
reg_by = index_unique(registry, "evidence_id", "Source registry")
queue_by = index_unique(worklist, "evidence_id", "Wave 3 worklist")
field_by = {}
for record in evidence:
    key = (record["evidence_id"], record["field"])
    if key in field_by:
        raise ValueError(f"Duplicate evidence field: {key}")
    field_by[key] = record
review_by = {}
for record in adjudication:
    key = (record["evidence_id"], record["field"])
    if key in review_by:
        raise ValueError(f"Duplicate reviewer field: {key}")
    review_by[key] = record
expected = {(eid, fld) for eid in queue_by for fld in ("equipment", "participants")}
if set(field_by) != expected or set(review_by) != expected:
    raise ValueError("Wave 3 field coverage is incomplete or contains unexpected IDs.")
if any(r["catalog_approval_status"] != "pending" or r["planner_eligible"] != "no" or r["approved_drill_id"].strip() for r in catalog):
    raise ValueError("Unexpected catalog approval state; stop before preparing new review records.")
if any(r["reviewer_decision"] != "pending" or r["team_signoff"] != "pending" or
       r["catalog_approval_status"] != "pending" or r["planner_eligible"] != "no"
       for r in adjudication):
    raise ValueError("Existing reviewer decisions detected; preserve and review before proceeding.")

for eid, item in queue_by.items():
    group = item["candidate_group_id"]
    if group not in cat_by or eid not in reg_by:
        raise ValueError(f"Missing source linkage: {eid}")
    cat, reg = cat_by[group], reg_by[eid]
    if eid not in cat["evidence_ids"].split(";") or reg["catalog_candidate_id"] != group:
        raise ValueError(f"Candidate/source linkage mismatch: {eid}")
    if item["source_url"] != reg["source_url"]:
        raise ValueError(f"Official source URL changed: {eid}")
    for field in ("equipment", "participants"):
        old = field_by[(eid, field)]
        if (old["candidate_group_id"] != group or
            old["current_catalog_value"] != cat[field] or
            old["current_registry_value"] != reg[field]):
            raise ValueError(f"Field input differs from verified Step 12-53 data: {eid}/{field}")

# Fetch only official webpages, and only once per unique URL.
urls = sorted({item["source_url"] for item in worklist})
page_sections = {}
print("=" * 72)
print("STEP 12-54 — WAVE 3 DRILL-SPECIFIC REQUIREMENT SOURCE REVIEW")
print("=" * 72)
print(f"Wave 3 drills: {len(worklist)} | Fields: {len(expected)} | Official pages: {len(urls)}")
for pos, url in enumerate(urls, 1):
    parsed = urlsplit(url)
    if parsed.scheme != "https" or parsed.hostname != "jr.nba.com" or not parsed.path.startswith("/basketball-practice-plans/"):
        raise ValueError(f"Unapproved source hostname or path: {url}")
    try:
        request = Request(url, headers={"User-Agent": "Mozilla/5.0 (compatible; ChampionsAthleticMap/1.0)"})
        with urlopen(request, timeout=25) as response:
            actual = urlsplit(response.geturl())
            if actual.hostname != "jr.nba.com" or actual.scheme != "https":
                raise ValueError("Official source redirected to a different hostname.")
            html = response.read(2_000_000).decode("utf-8", errors="replace")
        parser = DrillSectionParser()
        parser.feed(html)
        page_sections[url] = parser.sections
        print(f"  {pos:02d}/{len(urls):02d} | fetched | {url}")
    except (HTTPError, URLError, TimeoutError, OSError) as exc:
        # Do not write partial review files when online verification is unavailable.
        raise RuntimeError(f"Official source unavailable: {url}; {type(exc).__name__}: {exc}") from exc

source_rows = []
field_rows = []
candidate_rows = []
section_status_counts = Counter()
focus_checks = {}
for index, item in enumerate(worklist, 1):
    eid = item["evidence_id"]
    group = item["candidate_group_id"]
    cat, reg = cat_by[group], reg_by[eid]
    key = normalize(item["drill_name"])
    matches = [section for section in page_sections[item["source_url"]]
               if normalize(section["heading"]) == key]
    section = max(matches, key=lambda s: len(" ".join(s["parts"]))) if matches else None
    content = normalize(" ".join(section["parts"])) if section else ""
    word_count = len(content.split())
    if not section:
        status = "heading_or_section_unavailable"
    elif word_count < 12:
        status = "section_too_short_for_independent_review"
    else:
        status = "drill_section_available"
    section_status_counts[status] += 1
    equip = specific_cues(content, "equipment") if content else []
    participants = specific_cues(content, "participants") if content else []
    source_rows.append({
        "candidate_group_id": group, "evidence_id": eid,
        "drill_name": item["drill_name"], "source_url": item["source_url"],
        "heading_matches": len(matches), "section_status": status,
        "section_word_count": word_count,
        "section_sha256": hashlib.sha256(content.encode("utf-8")).hexdigest() if content else "",
        "drill_specific_equipment_cues": ";".join(equip),
        "drill_specific_participant_cues": ";".join(participants),
        "raw_instructional_content_saved": "no",
    })

    followups = []
    for field in ("equipment", "participants"):
        old = field_by[(eid, field)]
        review = review_by[(eid, field)]
        cues = equip if field == "equipment" else participants
        basis = "drill_section_cue_requires_interpretation" if cues else "no_drill_specific_cue_confirmed"
        recommendation = "compare_current_value_to_original_section"
        note = "Source cues are indicators only; separate base requirements from optional variations."
        if status != "drill_section_available":
            basis = "section_unavailable_or_incomplete"
            recommendation = "defer_until_source_section_is_available"
        if eid == "JR-EVID-0029" and field == "equipment":
            supported = all(part in content for part in (
                "two or more group lines", "designated shooting spot", "get their rebound"))
            focus_checks["shooting_line_equipment"] = supported
            basis = "activity_supported_requirement_pending_signoff" if supported else "source_phrase_review_needed"
            recommendation = "retain_current_basketball_and_hoop_pending_signoff" if supported else "defer_original_source_review"
            note = ("Official drill instructs groups to shoot and rebound; basketball and hoop are necessary for the described activity, "
                    "but the text is not a formal equipment list.") if supported else "The expected source details were not all found."
        if eid == "JR-EVID-0069" and field == "participants":
            supported = all(part in content for part in (
                "start in 5 lines on the baseline", "first two players in the middle line", "all the players will run forward"))
            focus_checks["fast_break_five_players"] = supported
            basis = "source_supports_five_active_roles_pending_signoff" if supported else "source_phrase_review_needed"
            recommendation = "retain_groups_of_five_pending_signoff" if supported else "defer_original_source_review"
            note = ("Official drill describes five baseline lines and a group moving together; "
                    "do not interpret this as a total practice roster size.") if supported else "Five-player role evidence could not be confirmed."
        if eid == "JR-EVID-0048" and field == "equipment":
            note = "Basketball is optional in the source base activity; do not turn optional equipment into a requirement."
        flags = sorted(set(filter(None, old["reliability_flags"].split(";"))))
        if status != "drill_section_available":
            flags.append("drill_section_verification_pending")
        if basis in ("source_phrase_review_needed", "no_drill_specific_cue_confirmed"):
            flags.append("further_evidence_needed")
        if flags:
            followups.append(f"{field}:" + ",".join(sorted(set(flags))))
        field_rows.append({
            "candidate_group_id": group, "evidence_id": eid, "drill_name": item["drill_name"],
            "field": field, "current_catalog_value": cat[field],
            "current_registry_value": reg[field], "source_url": item["source_url"],
            "drill_specific_source_cues": ";".join(cues),
            "source_evidence_status": basis, "recommended_review_action": recommendation,
            "source_note": note, "previous_reliability_flags": old["reliability_flags"],
            "reviewer_decision": review["reviewer_decision"], "reviewed_value": "",
            "independent_evidence_note": "", "reviewer": "", "review_date": "",
            "team_signoff": "pending", "permissions": "pending", "youth_planning": "pending",
            "catalog_approval_status": "pending", "planner_eligible": "no",
        })
    candidate_rows.append({
        "review_rank": index, "candidate_group_id": group, "evidence_id": eid,
        "drill_name": item["drill_name"], "source_url": item["source_url"],
        "source_section_status": status,
        "equipment_current": cat["equipment"], "participants_current": cat["participants"],
        "review_priority": "targeted" if eid in {"JR-EVID-0029", "JR-EVID-0069"} else item["overall_priority"],
        "review_questions": " | ".join(followups),
        "human_review_status": "pending", "team_signoff": "pending", "planner_eligible": "no",
    })

if len(source_rows) != 42 or len(field_rows) != 84 or len(candidate_rows) != 42:
    raise ValueError("Unexpected review-record coverage.")
if set(focus_checks) != {"shooting_line_equipment", "fast_break_five_players"}:
    raise ValueError("Required focused source cases are missing from Wave 3.")

candidate_rows.sort(key=lambda r: (r["review_priority"] != "targeted", int(r["review_rank"])))
summary = {
    "step": "12-54", "catalog_version": "0.3.6", "catalog_candidates": 144,
    "registered_source_occurrences": 156, "wave3_candidate_groups": len(source_rows),
    "field_reviews": len(field_rows), "official_pages_accessed": len(urls),
    "source_section_status_counts": dict(section_status_counts),
    "focus_source_checks": focus_checks, "human_decisions_completed": 0,
    "catalog_approvals": 0, "planner_eligible": 0,
    "input_sha256": {path.relative_to(ROOT).as_posix(): file_sha(path)
                     for key, path in INPUTS.items() if key != "version_summary"},
    "note": "Source-section evidence is preliminary; no automatic requirements, merging, permissions or youth-planning approvals.",
}

payloads = {
    OUTPUTS["source_checks"]: encode_csv(source_rows),
    OUTPUTS["field_review"]: encode_csv(field_rows),
    OUTPUTS["candidate_review"]: encode_csv(candidate_rows),
    OUTPUTS["summary"]: (json.dumps(summary, indent=2, ensure_ascii=False) + "\n").encode("utf-8"),
}
existing = [p for p in payloads if p.exists()]
if existing and len(existing) != len(payloads):
    raise FileExistsError("Some Step 12-54 outputs already exist. No files changed; inspect before rerunning.")
for path, content_bytes in payloads.items():
    path.parent.mkdir(parents=True, exist_ok=True)
    if path.exists():
        print("  PRESERVED", path.relative_to(ROOT))
    else:
        with path.open("xb") as stream:
            stream.write(content_bytes)
        if path.read_bytes() != content_bytes:
            raise IOError(f"Saved output verification failed: {path.name}")
        print("  CREATED  ", path.relative_to(ROOT))

print("\nSOURCE-SECTION VERIFICATION")
for status, total in section_status_counts.items():
    print(f"  {status:<47} {total}")
print("\nTARGETED DRILL CHECKS")
print("  Shooting Line Game — equipment implication:", "SUPPORTED FOR REVIEW" if focus_checks["shooting_line_equipment"] else "NEEDS SOURCE REVIEW")
print("  5 on 0 Fast Break — five active roles:", "SUPPORTED FOR REVIEW" if focus_checks["fast_break_five_players"] else "NEEDS SOURCE REVIEW")
print("\nFINAL STATUS")
print("  Candidate groups / registered sources: 144 / 156")
print("  Wave 3 field review records:", len(field_rows))
print("  Reviewer decisions / catalog approvals / planner eligible: 0 / 0 / 0")
print("  Existing catalog and prior source reports overwritten: No")
print("  Git operations performed: None")
print("OVERALL: PASS — drill-specific evidence recorded; human review and approvals remain pending")
```

```python
# Step 12-55 - Wave 3 Equipment and Participant Metadata Review Packet
# Prepare evidence-linked recommendations while preserving existing catalog and approval decisions.

import csv
import hashlib
import io
import json
from collections import Counter
from pathlib import Path

ROOT = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Run this cell from the Champions Athletic Map Git repository.')

INPUT = {
    'catalog': ROOT / 'data/processed/basketball_drill_catalog_v0_3_6_REVIEW_ONLY.csv',
    'registry': ROOT / 'results/basketball_drill_source_registry_v0_3_6.csv',
    'version_summary': ROOT / 'results/basketball_step12_52_summary_v0_3_6.json',
    'source_evidence': ROOT / 'results/basketball_wave3_drill_specific_source_evidence_v0_3_6.csv',
    'field_review': ROOT / 'results/basketball_wave3_requirement_field_review_v0_3_6.csv',
    'worklist': ROOT / 'results/basketball_wave3_requirement_reviewer_worklist_v0_3_6.csv',
}
OUTPUT = {
    'field_recommendations': ROOT / 'results/basketball_wave3_requirement_recommendations_v0_3_6.csv',
    'exception_review': ROOT / 'results/basketball_wave3_requirement_exceptions_v0_3_6.csv',
    'candidate_packet': ROOT / 'results/basketball_wave3_candidate_decision_packet_v0_3_6.csv',
    'summary': ROOT / 'results/basketball_step12_55_summary_v0_3_6.json',
}


def sha256(path):
    digest = hashlib.sha256()
    with path.open('rb') as stream:
        for chunk in iter(lambda: stream.read(1024 * 1024), b''):
            digest.update(chunk)
    return digest.hexdigest()


def read_csv(path, required):
    if not path.is_file():
        raise FileNotFoundError(f'Missing input: {path.relative_to(ROOT)}')
    with path.open('r', newline='', encoding='utf-8-sig') as stream:
        reader = csv.DictReader(stream)
        fields = list(reader.fieldnames or [])
        rows = list(reader)
    absent = set(required) - set(fields)
    if absent or not rows or len(fields) != len(set(fields)):
        raise ValueError(f'Invalid CSV {path.name}; missing columns: {sorted(absent)}')
    if any(None in row or any(v is None for v in row.values()) for row in rows):
        raise ValueError(f'Malformed CSV rows in {path.name}')
    return rows


def by_id(rows, field, label):
    names = [row[field].strip() for row in rows]
    if not all(names) or len(names) != len(set(names)):
        raise ValueError(f'{label} has blank or duplicate {field}')
    return {row[field].strip(): row for row in rows}


def as_csv_bytes(rows):
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=list(rows[0]), lineterminator='\n', extrasaction='raise')
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')


catalog = read_csv(INPUT['catalog'], [
    'catalog_candidate_id', 'drill_name', 'evidence_ids', 'equipment', 'participants',
    'catalog_approval_status', 'planner_eligible', 'approved_drill_id',
])
registry = read_csv(INPUT['registry'], [
    'evidence_id', 'catalog_candidate_id', 'source_url', 'equipment', 'participants',
])
source = read_csv(INPUT['source_evidence'], [
    'candidate_group_id', 'evidence_id', 'drill_name', 'source_url', 'section_status',
    'section_sha256', 'drill_specific_equipment_cues', 'drill_specific_participant_cues',
])
fields = read_csv(INPUT['field_review'], [
    'candidate_group_id', 'evidence_id', 'drill_name', 'field', 'current_catalog_value',
    'current_registry_value', 'source_url', 'source_evidence_status',
    'drill_specific_source_cues', 'reviewer_decision', 'team_signoff', 'planner_eligible',
])
worklist = read_csv(INPUT['worklist'], [
    'candidate_group_id', 'evidence_id', 'drill_name', 'source_url', 'review_priority',
    'human_review_status', 'team_signoff', 'planner_eligible',
])

if (len(catalog), len(registry), len(source), len(fields), len(worklist)) != (144, 156, 42, 84, 42):
    raise ValueError('Unexpected input counts; check Step 12-54 outputs before proceeding.')

version_summary = json.loads(INPUT['version_summary'].read_text(encoding='utf-8'))
saved_hashes = {str(k).replace('\\', '/'): v for k, v in version_summary.get('output_sha256', {}).items()}
for key in ('catalog', 'registry'):
    relative = INPUT[key].relative_to(ROOT).as_posix()
    if sha256(INPUT[key]) != saved_hashes.get(relative):
        raise ValueError(f'Integrity mismatch: {relative}; preserve and investigate the file.')

catalog_map = by_id(catalog, 'catalog_candidate_id', 'Catalog')
registry_map = by_id(registry, 'evidence_id', 'Registry')
source_map = by_id(source, 'evidence_id', 'Drill source evidence')
work_map = by_id(worklist, 'evidence_id', 'Review worklist')
field_map = {}
for row in fields:
    key = (row['evidence_id'], row['field'])
    if key in field_map or row['field'] not in {'equipment', 'participants'}:
        raise ValueError(f'Duplicate or invalid field: {key}')
    field_map[key] = row
if set(source_map) != set(work_map) or set(field_map) != {
    (eid, field) for eid in work_map for field in ('equipment', 'participants')
}:
    raise ValueError('Wave 3 source, field, and worklist identifiers do not align.')
if any(row['catalog_approval_status'] != 'pending' or row['planner_eligible'] != 'no'
       or row['approved_drill_id'].strip() for row in catalog):
    raise ValueError('Unexpected catalog approval status: stop before preparing decisions.')
if any(row['reviewer_decision'] != 'pending' or row['team_signoff'] != 'pending'
       or row['planner_eligible'] != 'no' for row in fields):
    raise ValueError('Existing reviewer decisions found; preserve and reconcile them manually.')

for eid, work in work_map.items():
    cid = work['candidate_group_id']
    cat, reg, src = catalog_map[cid], registry_map[eid], source_map[eid]
    if reg['catalog_candidate_id'] != cid or src['candidate_group_id'] != cid:
        raise ValueError(f'Source/candidate link mismatch: {eid}')
    if work['source_url'] != reg['source_url'] or work['source_url'] != src['source_url']:
        raise ValueError(f'Source URL mismatch: {eid}')
    if src['section_status'] != 'drill_section_available' or not src['section_sha256']:
        raise ValueError(f'Unavailable source section: {eid}')
    for field in ('equipment', 'participants'):
        f = field_map[(eid, field)]
        if (f['candidate_group_id'] != cid or
            f['source_url'] != work['source_url'] or
            f['current_catalog_value'] != cat[field] or
            f['current_registry_value'] != reg[field] or
            cat[field] != reg[field]):
            raise ValueError(f'Catalog/registry/field value mismatch: {eid} / {field}')

# Targeted publisher facts are stated separately from automated source-keyword cues.
# These proposed values remain subject to reviewer confirmation.
SPECIAL = {
    ('JR-EVID-0029', 'equipment'): (
        'retain_current_pending_signoff', '', 'publisher_activity_implied',
        'Rookie Plan 2 describes players shooting and rebounding at a designated spot. '
        'Basketball and hoop are supported by the activity, although the source-section keyword screen missed them.'
    ),
    ('JR-EVID-0069', 'participants'): (
        'retain_current_pending_signoff', '', 'publisher_source_explicit',
        'MVP Plan 3 describes five starting lines and a five-position transition sequence. '
        'Groups of five is consistent with the documented drill format.'
    ),
    ('JR-EVID-0069', 'equipment'): (
        'review_equipment_space_separation',
        'Basketballs (two initial ball handlers); access to hoop for layup finish',
        'publisher_source_explicit_and_activity_implied',
        'MVP Plan 3 distinguishes the two initial ball handlers from full-court travel. '
        'Court requirements belong in the space field, not the equipment field.'
    ),
    ('JR-EVID-0054', 'equipment'): (
        'review_equipment_space_separation', 'Basketball and hoops',
        'catalog_field_classification_review',
        'The recorded equipment value includes court space. Verify required hoops and '
        'keep court information in the separately reviewed space field.'
    ),
    ('JR-EVID-0070', 'equipment'): (
        'review_equipment_space_separation', 'Basketball',
        'catalog_field_classification_review',
        'The equipment field currently includes court space. Confirm court and hoop '
        'requirements in the source before standardizing the equipment field.'
    ),
    ('JR-EVID-0008', 'participants'): (
        'review_base_vs_optional_participation', '', 'source_keyword_ambiguous',
        'Source-section partner wording does not by itself establish that a partner '
        'is mandatory for the base around-the-body ball movement.'
    ),
    ('JR-EVID-0024', 'participants'): (
        'review_base_vs_optional_participation', '', 'source_keyword_ambiguous',
        'Partner terminology may describe a variation or sharing arrangement. '
        'Confirm the base activity before changing individual participation.'
    ),
    ('JR-EVID-0023', 'participants'): (
        'review_base_vs_optional_participation', '', 'source_keyword_ambiguous',
        'Coach or partner cues may describe alternatives; do not change the base activity '
        'from individual to group without source-level confirmation.'
    ),
}

recommendations = []
exceptions = []
for work in sorted(worklist, key=lambda r: (
    {'targeted': 0, 'medium': 1, 'standard': 2}.get(r['review_priority'], 3),
    r['evidence_id']
)):
    eid = work['evidence_id']
    for field in ('equipment', 'participants'):
        current = field_map[(eid, field)]
        cues = current['drill_specific_source_cues']
        if (eid, field) in SPECIAL:
            action, suggested, basis, note = SPECIAL[(eid, field)]
            review_priority = 'targeted'
        elif not cues.strip():
            action, suggested, basis = 'source_section_detail_review', '', 'no_matching_automated_cue'
            note = ('The drill section was found, but the existing keyword rules did not '
                    'identify a field-specific cue. This does not prove the requirement is absent.')
            review_priority = 'medium'
        elif 'court' in current['current_catalog_value'].casefold() and field == 'equipment':
            action, suggested, basis = 'review_equipment_space_separation', '', 'metadata_field_mixing'
            note = 'Equipment includes a location term. Recheck the source before separating requirements.'
            review_priority = 'medium'
        else:
            action, suggested, basis = 'retain_current_pending_source_signoff', '', 'source_scoped_cues_only'
            note = ('Existing catalog and registry values match. Drill-scoped source cues '
                    'are present but do not independently establish mandatory requirements.')
            review_priority = work['review_priority']
        rec = {
            'candidate_group_id': work['candidate_group_id'],
            'evidence_id': eid,
            'drill_name': work['drill_name'],
            'field': field,
            'current_value': current['current_catalog_value'],
            'recommended_action': action,
            'suggested_value_if_revised': suggested,
            'evidence_basis': basis,
            'source_cues': cues,
            'source_section_sha256': source_map[eid]['section_sha256'],
            'source_url': work['source_url'],
            'source_review_note': note,
            'review_priority': review_priority,
            'reviewer_decision': 'pending',
            'reviewed_value': '',
            'reviewer': '',
            'review_date': '',
            'team_signoff': 'pending',
            'permission_review': 'pending',
            'youth_planning_review': 'pending',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
        }
        recommendations.append(rec)
        if action not in {'retain_current_pending_source_signoff'}:
            exceptions.append(dict(rec))

candidate_packet = []
for work in sorted(worklist, key=lambda r: (r['evidence_id'])):
    eid = work['evidence_id']
    subset = [r for r in recommendations if r['evidence_id'] == eid]
    if len(subset) != 2:
        raise ValueError(f'Incomplete candidate field records: {eid}')
    candidate_packet.append({
        'candidate_group_id': work['candidate_group_id'],
        'evidence_id': eid,
        'drill_name': work['drill_name'],
        'source_url': work['source_url'],
        'equipment_current': field_map[(eid, 'equipment')]['current_catalog_value'],
        'participants_current': field_map[(eid, 'participants')]['current_catalog_value'],
        'equipment_recommendation': subset[0]['recommended_action'],
        'participants_recommendation': subset[1]['recommended_action'],
        'exception_fields': ';'.join(r['field'] for r in subset
                                     if r['recommended_action'] != 'retain_current_pending_source_signoff'),
        'review_priority': 'targeted' if any(r['review_priority'] == 'targeted' for r in subset)
                           else work['review_priority'],
        'reviewer_decision_status': 'pending',
        'team_signoff': 'pending',
        'permission_review': 'pending',
        'youth_planning_review': 'pending',
        'catalog_approval_status': 'pending',
        'planner_eligible': 'no',
    })

assert len(recommendations) == 84 and len(candidate_packet) == 42
summary = {
    'step': '12-55',
    'catalog_version': '0.3.6',
    'catalog_candidates': len(catalog),
    'registered_source_occurrences': len(registry),
    'wave3_candidates': len(candidate_packet),
    'equipment_participant_field_recommendations': len(recommendations),
    'recommendations_by_action': dict(sorted(Counter(
        row['recommended_action'] for row in recommendations).items())),
    'targeted_exception_fields': [
        {'evidence_id': eid, 'field': field} for eid, field in SPECIAL
    ],
    'human_signoffs': 0,
    'catalog_approvals': 0,
    'planner_eligible': 0,
    'input_sha256': {
        INPUT[key].relative_to(ROOT).as_posix(): sha256(INPUT[key])
        for key in ('catalog', 'registry', 'source_evidence', 'field_review', 'worklist')
    },
    'interpretation': ('These are evidence-linked review recommendations, not independently '
                       'approved equipment/participant requirements or final drill approvals.'),
}

payloads = {
    'field_recommendations': as_csv_bytes(recommendations),
    'exception_review': as_csv_bytes(exceptions),
    'candidate_packet': as_csv_bytes(candidate_packet),
    'summary': (json.dumps(summary, indent=2, ensure_ascii=False) + '\n').encode('utf-8'),
}

# Existing outputs may contain reviewer edits; never overwrite them.
# When a file already exists, preserve it, validate its basic schema, and report its status.
print('INPUT INTEGRITY: PASS — Catalog v0.3.6 and registry match Step 12-52 checksums.')
print('=' * 72)
print('STEP 12-55 — WAVE 3 EQUIPMENT AND PARTICIPANT DECISION PACKET')
print('=' * 72)
print('Candidate groups / source records:', len(catalog), '/', len(registry))
print('Wave 3 candidates / field recommendations:', len(candidate_packet), '/', len(recommendations))
print('\nRECOMMENDATIONS BY REVIEW ACTION')
for label, n in Counter(r['recommended_action'] for r in recommendations).most_common():
    print(f'  {label:<49} {n}')
print('\nTARGETED REQUIREMENT FOLLOW-UPS')
for r in exceptions:
    if (r['evidence_id'], r['field']) in SPECIAL:
        print(f"  {r['evidence_id']} | {r['drill_name']} | {r['field']} | {r['recommended_action']}")
print('\nOUTPUTS')
for key, path in OUTPUT.items():
    if path.exists():
        if key == 'summary':
            prior = json.loads(path.read_text(encoding='utf-8'))
            if prior.get('step') != '12-55' or prior.get('input_sha256') != summary['input_sha256']:
                raise ValueError(f'Existing output summary has a different input snapshot: {path.name}')
        else:
            expected_ids = ({(r['evidence_id'], r['field']) for r in (recommendations if key == 'field_recommendations' else exceptions)}
                            if key != 'candidate_packet' else {r['evidence_id'] for r in candidate_packet})
            with path.open(newline='', encoding='utf-8-sig') as f:
                saved = list(csv.DictReader(f))
            actual_ids = ({(r['evidence_id'], r['field']) for r in saved}
                          if key != 'candidate_packet' else {r['evidence_id'] for r in saved})
            if expected_ids != actual_ids:
                raise ValueError(f'Existing output records differ from expected identifiers: {path.name}')
        status = 'PRESERVED'
    else:
        path.parent.mkdir(parents=True, exist_ok=True)
        with path.open('xb') as file:
            file.write(payloads[key])
        if path.read_bytes() != payloads[key]:
            raise ValueError(f'Output verification failed: {path.name}')
        status = 'CREATED  '
    print(f'  {status} {path.relative_to(ROOT)}')

print('\nFINAL STATUS')
print('Reviewer decisions / team signoffs / catalog approvals: 0 / 0 / 0')
print('Original catalog or registry modified: No')
print('Git operations performed: None')
print('OVERALL: PASS — 84 review recommendations prepared; publisher checks and signoffs remain open')
```

```python
# Step 12-56 - Wave 3 Equipment and Participant Source Decisions
# Verify targeted source claims and stage metadata decisions without changing catalog approvals.

import csv
import hashlib
import json
import re
import unicodedata
from collections import Counter
from html.parser import HTMLParser
from pathlib import Path
from urllib.error import HTTPError, URLError
from urllib.parse import urlsplit
from urllib.request import Request, urlopen


def project_root():
    current = Path.cwd().resolve()
    for folder in (current, *current.parents):
        if (folder / ".git").exists():
            return folder
    raise FileNotFoundError("Git project root not found")


ROOT = project_root()
RESULTS = ROOT / "results"
source_path = RESULTS / "basketball_wave3_requirement_recommendations_v0_3_6.csv"
exception_path = RESULTS / "basketball_wave3_requirement_exceptions_v0_3_6.csv"


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(encoding="utf-8-sig", newline="") as stream:
        return list(csv.DictReader(stream))


def checksum(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def normalized(value):
    value = unicodedata.normalize("NFKC", value or "")
    return re.sub(r"[^a-z0-9]+", " ", value.casefold()).strip()


recommendations = read_csv(source_path)
exceptions = read_csv(exception_path)
original_checksums = {path: checksum(path) for path in (source_path, exception_path)}
required = {
    "candidate_group_id", "evidence_id", "drill_name", "field",
    "current_value", "recommended_action", "source_url",
    "reviewer_decision", "catalog_approval_status", "planner_eligible",
}
if not recommendations or required - set(recommendations[0]):
    raise ValueError("Wave 3 recommendations have missing required columns")
if len(recommendations) != 84 or len(exceptions) != 10:
    raise ValueError("Expected 84 Wave 3 fields and 10 targeted exceptions")
keys = [(r["evidence_id"], r["field"]) for r in recommendations]
if len(keys) != len(set(keys)):
    raise ValueError("Duplicate source ID + field in recommendations")
if len({r["evidence_id"] for r in recommendations}) != 42:
    raise ValueError("Expected 42 distinct Wave 3 source records")
exception_keys = {(r["evidence_id"], r["field"]) for r in exceptions}
if len(exception_keys) != len(exceptions) or not exception_keys.issubset(keys):
    raise ValueError("Exception records are duplicated or missing from the source")
if any(r["reviewer_decision"] != "pending" for r in recommendations):
    raise ValueError("Existing reviewer decisions detected; reconcile before generating proposals")
if any(r["catalog_approval_status"] != "pending" or r["planner_eligible"] != "no" for r in recommendations):
    raise ValueError("Unexpected approval state")

# Drill-specific checks derived from official Jr. NBA practice-plan sections.
# A matching passage supports review; it does not establish copyright permission,
# athlete-specific suitability, or team sign-off.
TARGETS = {
    ("JR-EVID-0029", "equipment"): {
        "decision": "retain_current", "value": None,
        "terms": ["players will shoot one time", "get their rebound"],
        "finding": "Shooting and rebounding imply basketball and hoop; preserve the current equipment label.",
        "classification": "activity_implied",
    },
    ("JR-EVID-0069", "equipment"): {
        "decision": "revise_value",
        "value": "Basketballs (two initial ball handlers); hoop for layup finish",
        "terms": ["first two players in the middle line will need a ball", "other basket"],
        "finding": "Move full-court requirements to the space field, separate from basketball and hoop equipment.",
        "classification": "source_explicit_plus_activity_implied",
    },
    ("JR-EVID-0069", "participants"): {
        "decision": "retain_current", "value": None,
        "terms": ["players start in 5 lines on the baseline", "each player should rotate through each of the starting points"],
        "finding": "Five player roles are consistent with groups of five; do not infer fewer active players.",
        "classification": "source_explicit",
    },
    ("JR-EVID-0007", "participants"): {
        "decision": "retain_current", "value": None,
        "terms": ["all players need a basketball or share", "designated area"],
        "finding": "Base activity is individually performed; sharing the ball or following a leader does not make a partner mandatory.",
        "classification": "source_explicit_base_and_optional",
    },
    ("JR-EVID-0008", "participants"): {
        "decision": "retain_current", "value": None,
        "terms": ["players start with the ball in front of their waist", "pair players up"],
        "finding": "Individual ball wraps are the base exercise; partner competition is an optional modification.",
        "classification": "source_explicit_base_and_optional",
    },
    ("JR-EVID-0023", "participants"): {
        "decision": "retain_current", "value": None,
        "terms": ["players need a basketball or share", "have the players get into pairs"],
        "finding": "Tossing and catching alone is the base exercise; pair exchange appears as a social variation.",
        "classification": "source_explicit_base_and_optional",
    },
    ("JR-EVID-0024", "participants"): {
        "decision": "retain_current", "value": None,
        "terms": ["all players need a basketball or share", "have the players get into pairs"],
        "finding": "Individual dribbling is the base exercise; paired mirror/competition variants remain optional.",
        "classification": "source_explicit_base_and_optional",
    },
    ("JR-EVID-0040", "participants"): {
        "decision": "retain_current", "value": None,
        "terms": ["all the players should have a ball", "find some space on the court"],
        "finding": "Rhythm Dribbling is individually performed; the section does not require a partner.",
        "classification": "source_explicit",
    },
    ("JR-EVID-0054", "equipment"): {
        "decision": "revise_value", "value": "Basketball; two hoops",
        "terms": ["there will be one team at half court", "other basket"],
        "finding": "Separate the two-basket activity equipment from full-court playing-space requirements.",
        "classification": "source_explicit_plus_activity_implied",
    },
    ("JR-EVID-0070", "equipment"): {
        "decision": "revise_value", "value": "Basketball",
        "terms": ["first group up will give the coach a ball", "players will balance the floor"],
        "finding": "Basketball is documented; court access remains a separate space requirement.",
        "classification": "source_explicit",
    },
}
if set(TARGETS) != exception_keys:
    raise ValueError("Targeted case list differs from uploaded exceptions")


class SectionExtractor(HTMLParser):
    def __init__(self):
        super().__init__()
        self.sections = []
        self.heading_open = False
        self.heading_parts = []
        self.current = None
        self.skip_depth = 0

    def handle_starttag(self, tag, attrs):
        if tag in {"script", "style", "noscript"}:
            self.skip_depth += 1
            return
        if self.skip_depth:
            return
        if tag == "h2":
            self.heading_open = True
            self.heading_parts = []
        elif self.current is not None and tag in {"p", "li", "br", "div"}:
            self.current["parts"].append(" ")

    def handle_endtag(self, tag):
        if tag in {"script", "style", "noscript"}:
            self.skip_depth = max(0, self.skip_depth - 1)
            return
        if self.skip_depth:
            return
        if tag == "h2" and self.heading_open:
            title = " ".join(self.heading_parts).strip()
            if title:
                self.current = {"heading": title, "parts": []}
                self.sections.append(self.current)
            self.heading_open = False
            self.heading_parts = []
        elif self.current is not None and tag in {"p", "li", "div"}:
            self.current["parts"].append(" ")

    def handle_data(self, data):
        if self.skip_depth:
            return
        if self.heading_open:
            self.heading_parts.append(data)
        elif self.current is not None:
            self.current["parts"].append(data)


def fetch_sections(url):
    parts = urlsplit(url)
    if parts.scheme != "https" or parts.hostname != "jr.nba.com" or not parts.path.startswith("/basketball-practice-plans/"):
        raise ValueError("Unexpected or untrusted basketball source URL")
    request = Request(url, headers={"User-Agent": "Mozilla/5.0"})
    try:
        with urlopen(request, timeout=20) as response:
            raw = response.read(2_000_000).decode("utf-8", errors="replace")
            if urlsplit(response.geturl()).hostname != "jr.nba.com":
                raise ValueError("Unexpected redirect hostname")
        parser = SectionExtractor()
        parser.feed(raw)
        return "fetched", parser.sections
    except (HTTPError, URLError, TimeoutError, OSError, UnicodeError) as err:
        return f"unavailable:{type(err).__name__}", []


by_key = {(r["evidence_id"], r["field"]): r for r in recommendations}
urls = sorted({by_key[k]["source_url"] for k in TARGETS})
print("=" * 72)
print("STEP 12-56 — WAVE 3 EQUIPMENT AND PARTICIPANT TARGETED REVIEW")
print("=" * 72)
print("Source records / field proposals:", 42, "/", len(recommendations))
print("Targeted exception fields:", len(TARGETS))
print("Official source pages:", len(urls))

pages = {}
for pos, url in enumerate(urls, 1):
    status, sections = fetch_sections(url)
    pages[url] = (status, sections)
    print(f"  {pos:02d}/{len(urls):02d} | {status:<20} | {url}")

checks = []
for key, config in TARGETS.items():
    row = by_key[key]
    status, sections = pages[row["source_url"]]
    candidates = [s for s in sections if normalized(s["heading"]) == normalized(row["drill_name"])]
    if status != "fetched":
        result = "page_unavailable"
        terms_found = 0
    elif len(candidates) != 1:
        result = "heading_not_unique_or_missing"
        terms_found = 0
    else:
        section = normalized(" ".join(candidates[0]["parts"]))
        terms_found = sum(normalized(term) in section for term in config["terms"])
        result = "source_cues_supported_for_review" if terms_found == len(config["terms"]) else "source_cues_incomplete_review_required"

    checks.append({
        "evidence_id": key[0], "candidate_group_id": row["candidate_group_id"],
        "drill_name": row["drill_name"], "field": key[1],
        "source_url": row["source_url"], "page_status": status,
        "source_screen_status": result,
        "required_cues_found": terms_found,
        "required_cues_total": len(config["terms"]),
        "source_evidence_classification": config["classification"],
        "finding": config["finding"],
        "catalog_approval_status": "pending", "planner_eligible": "no",
    })

check_by = {(r["evidence_id"], r["field"]): r for r in checks}
field_proposals = []
exception_packet = []
for original in recommendations:
    row = original.copy()
    key = (row["evidence_id"], row["field"])
    if key in TARGETS:
        config = TARGETS[key]
        checked = check_by[key]
        row["proposed_review_action"] = config["decision"]
        row["proposed_review_value"] = (
            row["current_value"] if config["value"] is None else config["value"]
        )
        row["source_screen_status"] = checked["source_screen_status"]
        row["source_evidence_classification"] = config["classification"]
        row["decision_rationale"] = config["finding"]
        exception_packet.append({
            "evidence_id": key[0], "drill_name": row["drill_name"], "field": key[1],
            "current_value": row["current_value"],
            "proposed_review_action": config["decision"],
            "proposed_review_value": row["proposed_review_value"],
            "source_screen_status": checked["source_screen_status"],
            "source_evidence_classification": config["classification"],
            "decision_rationale": config["finding"],
            "source_url": row["source_url"], "reviewer_decision": "pending",
            "reviewer": "", "review_date": "", "team_signoff": "pending",
        })
    else:
        row["proposed_review_action"] = "retain_current_pending_source_signoff"
        row["proposed_review_value"] = row["current_value"]
        row["source_screen_status"] = "not_checked_in_step12_56"
        row["source_evidence_classification"] = "prior_screen_only"
        row["decision_rationale"] = "The prior metadata matches the registry; drill-specific approval remains pending."
    row["reviewer_decision"] = "pending"
    field_proposals.append(row)

assert len(field_proposals) == 84 and len(exception_packet) == 10
assert len([r for r in field_proposals if r["proposed_review_action"] == "revise_value"]) == 3
assert all(r["planner_eligible"] == "no" and r["catalog_approval_status"] == "pending" for r in field_proposals)

outputs = {
    RESULTS / "basketball_wave3_targeted_source_checks_v0_3_6.csv": checks,
    RESULTS / "basketball_wave3_field_decision_proposals_v0_3_6.csv": field_proposals,
    RESULTS / "basketball_wave3_targeted_reconciliation_packet_v0_3_6.csv": exception_packet,
}
summary = {
    "step": "12-56", "catalog_version": "0.3.6", "wave3_candidates": 42,
    "total_field_review_records": 84, "targeted_exception_fields": 10,
    "reviewed_source_urls": len(urls),
    "source_check_results": dict(Counter(r["source_screen_status"] for r in checks)),
    "proposed_field_actions": dict(Counter(r["proposed_review_action"] for r in field_proposals)),
    "actual_reviewer_decisions": 0, "catalog_edits": 0,
    "catalog_approvals": 0, "planner_eligible": 0,
    "note": "Source cues require reviewer interpretation; no automatic field signoffs or approvals.",
}
summary_path = RESULTS / "basketball_step12_56_summary_v0_3_6.json"

# Existing review documents may contain reviewer edits: never overwrite them.
existing = [p for p in [*outputs, summary_path] if p.exists()]
if existing:
    if len(existing) != len(outputs) + 1:
        raise FileExistsError("Partial Step 12-56 outputs exist; inspect before rerunning")
    for path, rows in outputs.items():
        saved = read_csv(path)
        if len(saved) != len(rows):
            raise ValueError(f"Existing Step 12-56 record count differs: {path.name}")
    print("\nExisting Step 12-56 outputs preserved (may contain reviewer edits):")
    for path in [*outputs, summary_path]:
        print("  PRESERVED ", path.relative_to(ROOT))
else:
    for path, rows in outputs.items():
        with path.open("x", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=list(rows[0].keys()))
            writer.writeheader()
            writer.writerows(rows)
        print("  CREATED  ", path.relative_to(ROOT))
    with summary_path.open("x", encoding="utf-8") as f:
        json.dump(summary, f, indent=2)
    print("  CREATED  ", summary_path.relative_to(ROOT))

for path, old_hash in original_checksums.items():
    if checksum(path) != old_hash:
        raise RuntimeError(f"Input file changed unexpectedly: {path}")

print("\nTARGETED SOURCE RESULTS")
for status, count in Counter(r["source_screen_status"] for r in checks).items():
    print(f"  {status:<45} {count}")
print("\nPROPOSED ACTIONS (NOT SIGNOFF)")
for action, count in Counter(r["proposed_review_action"] for r in field_proposals).items():
    print(f"  {action:<45} {count}")
print("\nReviewer decisions / catalog edits / planner approvals: 0 / 0 / 0")
print("Input review sheets modified: No")
print("Git operations performed: None")
print("OVERALL: PASS — evidence and decisions staged; approval gates remain pending")
```

```python
# Step 12-58 - Apply Confirmed Wave 3 Equipment and Participant Decisions
# Preserve existing approvals and create a new versioned catalog with a complete decision log.

import csv
import hashlib
import io
import json
from collections import Counter
from pathlib import Path


def project_root():
    cwd = Path.cwd().resolve()
    for path in (cwd, *cwd.parents):
        if (path / '.git').exists():
            return path
    raise FileNotFoundError('Project repository (.git) not found.')


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or not rows:
        raise ValueError(f'Empty table or missing header: {path}')
    return rows, columns


def unique_index(rows, column, name):
    keys = [row.get(column, '').strip() for row in rows]
    if any(not key for key in keys) or len(keys) != len(set(keys)):
        raise ValueError(f'{name}: missing or repeated {column}.')
    return {row[column].strip(): row for row in rows}


def digest(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def csv_bytes(columns, rows):
    output = io.StringIO(newline='')
    writer = csv.DictWriter(output, fieldnames=columns, lineterminator='\n',
                            extrasaction='raise')
    writer.writeheader()
    writer.writerows(rows)
    return output.getvalue().encode('utf-8')


def verify_and_save(payloads, root):
    # A second run may verify identical outputs; altered review files are never overwritten.
    for path, content in payloads.items():
        if path.is_file() and path.read_bytes() != content:
            raise FileExistsError(f'Existing output differs; preserved without overwrite: {path}')
    for path, content in payloads.items():
        path.parent.mkdir(parents=True, exist_ok=True)
        if path.exists():
            label = 'VERIFIED '
        else:
            try:
                with path.open('xb') as handle:
                    handle.write(content)
            except FileExistsError:
                if path.read_bytes() != content:
                    raise
            label = 'CREATED  '
        if path.read_bytes() != content:
            raise RuntimeError(f'Output verification failed: {path}')
        print(f'  {label} {path.relative_to(root)}')


ROOT = project_root()
DATA = ROOT / 'data' / 'processed'
RESULTS = ROOT / 'results'

catalog_path = DATA / 'basketball_drill_catalog_v0_3_6_REVIEW_ONLY.csv'
registry_path = RESULTS / 'basketball_drill_source_registry_v0_3_6.csv'
checksum_path = RESULTS / 'basketball_step12_52_summary_v0_3_6.json'
recommendations_path = RESULTS / 'Step_12_57_Targeted_Review_Recommendations_REVIEW_ONLY.csv'
proposals_path = RESULTS / 'basketball_wave3_field_decision_proposals_v0_3_6.csv'
source_packet_path = RESULTS / 'basketball_wave3_targeted_reconciliation_packet_v0_3_6.csv'

catalog, catalog_columns = read_csv(catalog_path)
registry, registry_columns = read_csv(registry_path)
recommendations, _ = read_csv(recommendations_path)
proposals, _ = read_csv(proposals_path)
source_packet, _ = read_csv(source_packet_path)

with checksum_path.open(encoding='utf-8') as handle:
    previous_summary = json.load(handle)

# Normalize Windows and POSIX separator conventions, but never bypass hash verification.
expected_sha = {
    str(key).replace('\\', '/'): value
    for key, value in previous_summary.get('output_sha256', {}).items()
}
for path in (catalog_path, registry_path):
    key = path.relative_to(ROOT).as_posix()
    if not expected_sha.get(key) or expected_sha[key] != digest(path):
        raise ValueError(f'Input integrity mismatch: {key} is not the Step 12-52 output.')
print('INPUT INTEGRITY: PASS — Catalog v0.3.6 and source registry match Step 12-52 checksums.')

if len(catalog) != 144 or len(registry) != 156:
    raise ValueError('Unexpected Catalog v0.3.6 or source-registry record count.')

cat_by = unique_index(catalog, 'catalog_candidate_id', 'Catalog')
reg_by = unique_index(registry, 'evidence_id', 'Source registry')

for row in catalog:
    if (row.get('catalog_approval_status') != 'pending'
            or row.get('planner_eligible') != 'no'
            or row.get('approved_drill_id', '').strip()):
        raise ValueError('Existing catalog contains an unexpected approval state.')

field_key = lambda row: (row.get('evidence_id', '').strip(), row.get('field', '').strip())
recommendations_by_key = {
    field_key(row): row for row in recommendations
}
if len(recommendations) != 10 or len(recommendations_by_key) != 10:
    raise ValueError('Expected ten unique Step 12-57 recommendations.')

packet_by_key = {field_key(row): row for row in source_packet}
if len(source_packet) != 10 or len(packet_by_key) != 10:
    raise ValueError('Expected ten unique Step 12-56 source-packet records.')
if recommendations_by_key.keys() != packet_by_key.keys():
    raise ValueError('Source review packet and confirmed recommendations refer to different fields.')

for key, rec in recommendations_by_key.items():
    packet = packet_by_key[key]
    for field in ('drill_name', 'current_value', 'proposed_review_action',
                  'proposed_review_value', 'source_url'):
        if rec.get(field, '') != packet.get(field, ''):
            raise ValueError(f'Recommendation differs from original review packet: {key} / {field}')
    if rec.get('review_recommendation') != rec.get('proposed_review_action'):
        raise ValueError(f'Unexpected recommendation override: {key}')
    if rec.get('reviewer_decision') != 'pending' or rec.get('team_signoff') != 'pending':
        raise ValueError(f'Unexpected pre-existing reviewer sign-off: {key}')
    if rec.get('reviewer_confirmation') not in ('', 'pending'):
        raise ValueError(f'Unexpected reviewer confirmation field: {key}')

counts = Counter(row['review_recommendation'] for row in recommendations)
if counts != Counter({'retain_current': 7, 'revise_value': 3}):
    raise ValueError(f'Confirmed 7/3 decision distribution has changed: {counts}')

proposal_by_key = {field_key(row): row for row in proposals}
if len(proposals) != 84 or len(proposal_by_key) != 84:
    raise ValueError('Expected 84 unique Wave 3 field-review records.')
if not set(recommendations_by_key).issubset(proposal_by_key):
    raise ValueError('A confirmed recommendation is missing from Wave 3 review records.')
pending = [row for row in proposals if field_key(row) not in recommendations_by_key]
if len(pending) != 74 or any(row.get('proposed_review_action') !=
                                  'retain_current_pending_source_signoff' for row in pending):
    raise ValueError('The 74 other Wave 3 fields must remain pending source review.')

confirmed_rows = []
change_rows = []
updated_groups = set()
for key in sorted(recommendations_by_key):
    rec = recommendations_by_key[key]
    evidence_id, field = key
    if field not in {'equipment', 'participants'}:
        raise ValueError(f'Unexpected metadata field: {key}')
    registry_row = reg_by.get(evidence_id)
    if registry_row is None:
        raise ValueError(f'Unregistered source: {evidence_id}')
    group_id = registry_row['catalog_candidate_id']
    catalog_row = cat_by.get(group_id)
    if catalog_row is None or evidence_id not in {
        x.strip() for x in catalog_row.get('evidence_ids', '').split(';') if x.strip()
    }:
        raise ValueError(f'Source-to-catalog mapping mismatch: {key}')
    if ((registry_row.get('drill_name') and registry_row['drill_name'] != rec['drill_name'])
            or registry_row.get('source_url') != rec['source_url']):
        raise ValueError(f'Source identity mismatch: {key}')
    old = rec['current_value']
    if registry_row.get(field) != old or catalog_row.get(field) != old:
        raise ValueError(f'Catalog or registry value differs from reviewed value: {key}')
    if proposal_by_key[key].get('current_value') != old:
        raise ValueError(f'Wave 3 original review value differs: {key}')

    decision = rec['review_recommendation']
    new = rec['proposed_review_value'] if decision == 'revise_value' else old
    if not new.strip() or (decision == 'revise_value' and new == old):
        raise ValueError(f'Invalid proposed revision: {key}')
    if decision == 'retain_current' and rec['proposed_review_value'] != old:
        raise ValueError(f'Retained value unexpectedly differs: {key}')

    updated_groups.add(group_id)
    confirmed_rows.append({
        'candidate_group_id': group_id,
        'evidence_id': evidence_id,
        'drill_name': rec['drill_name'],
        'field': field,
        'confirmed_decision': decision,
        'prior_value': old,
        'confirmed_value': new,
        'source_url': rec['source_url'],
        'evidence_classification': rec['source_evidence_classification'],
        'review_note': rec['source_review_note'],
        'confirmation_basis': 'user_confirmed_metadata_decision',
        'team_signoff': 'pending',
        'permission_review': 'pending',
        'youth_planning_review': 'pending',
        'catalog_approval_status': 'pending',
        'planner_eligible': 'no',
    })
    if decision == 'revise_value':
        catalog_row[field] = new
        registry_row[field] = new
        change_rows.append({
            'candidate_group_id': group_id,
            'evidence_id': evidence_id,
            'drill_name': rec['drill_name'],
            'field': field,
            'old_value': old,
            'new_value': new,
            'change_reason': rec['decision_rationale'],
            'source_url': rec['source_url'],
            'review_basis': 'confirmed_metadata_only',
        })

if len(change_rows) != 3 or len(updated_groups) != 9:
    raise ValueError('Unexpected number of metadata revisions or affected groups.')

for row in catalog:
    row['catalog_version'] = '0.3.7'

# All approved-status fields remain exactly as they were in v0.3.6.
assert all(row['catalog_approval_status'] == 'pending' and
           row['planner_eligible'] == 'no' and
           not row.get('approved_drill_id', '').strip() for row in catalog)

# Preserve the 74 unreviewed values as a separate follow-up report.
pending_columns = ['candidate_group_id', 'evidence_id', 'drill_name', 'field',
                   'current_value', 'source_url', 'required_next_action',
                   'reviewer_decision', 'team_signoff']
pending_rows = [{
    'candidate_group_id': row['candidate_group_id'],
    'evidence_id': row['evidence_id'],
    'drill_name': row['drill_name'],
    'field': row['field'],
    'current_value': row['current_value'],
    'source_url': row['source_url'],
    'required_next_action': 'individual_source_verification_and_signoff',
    'reviewer_decision': 'pending',
    'team_signoff': 'pending',
} for row in pending]

paths = {
    'catalog': DATA / 'basketball_drill_catalog_v0_3_7_REVIEW_ONLY.csv',
    'registry': RESULTS / 'basketball_drill_source_registry_v0_3_7.csv',
    'decisions': RESULTS / 'basketball_wave3_confirmed_requirement_decisions_v0_3_7.csv',
    'changes': RESULTS / 'basketball_wave3_requirement_change_log_v0_3_7.csv',
    'pending': RESULTS / 'basketball_wave3_pending_requirement_reviews_v0_3_7.csv',
    'summary': RESULTS / 'basketball_step12_58_summary_v0_3_7.json',
}

payloads = {
    paths['catalog']: csv_bytes(catalog_columns, catalog),
    paths['registry']: csv_bytes(registry_columns, registry),
    paths['decisions']: csv_bytes(list(confirmed_rows[0]), confirmed_rows),
    paths['changes']: csv_bytes(list(change_rows[0]), change_rows),
    paths['pending']: csv_bytes(pending_columns, pending_rows),
}
summary = {
    'step': '12-58',
    'catalog_version': '0.3.7',
    'input_catalog_version': '0.3.6',
    'catalog_candidates': len(catalog),
    'source_occurrences': len(registry),
    'confirmed_targeted_field_decisions': len(confirmed_rows),
    'retained_fields': counts['retain_current'],
    'revised_equipment_fields': counts['revise_value'],
    'changed_catalog_fields': len(change_rows),
    'changed_registry_fields': len(change_rows),
    'distinct_candidate_groups_in_targeted_review': len(updated_groups),
    'wave3_fields_still_pending': len(pending_rows),
    'team_signoff': 'pending',
    'permission_review': 'pending',
    'youth_planning_review': 'pending',
    'approved_drills': 0,
    'planner_eligible_drills': 0,
    'source_inputs_preserved': True,
    'git_operations_performed': False,
    'input_sha256': {path.relative_to(ROOT).as_posix(): digest(path)
                     for path in (catalog_path, registry_path, recommendations_path,
                                  proposals_path, source_packet_path)},
    'output_sha256': {path.relative_to(ROOT).as_posix(): hashlib.sha256(content).hexdigest()
                      for path, content in payloads.items()},
}
payloads[paths['summary']] = (json.dumps(summary, indent=2, ensure_ascii=False,
                                  sort_keys=True) + '\n').encode('utf-8')

print('=' * 72)
print('STEP 12-58 — CONFIRMED WAVE 3 EQUIPMENT AND PARTICIPANT DECISIONS')
print('=' * 72)
print('Catalog candidate groups / source records:', len(catalog), '/', len(registry))
print('Confirmed targeted decisions: 7 retain, 3 revise')
print('Source equipment revisions:')
for change in change_rows:
    print(f"  {change['evidence_id']} | {change['drill_name']}")
    print(f"    {change['old_value']} -> {change['new_value']}")
print('Other Wave 3 equipment/participant fields still awaiting signoff:', len(pending_rows))
print('\nOUTPUTS')
verify_and_save(payloads, ROOT)
print('\nAPPROVAL GATES')
print('Team sign-off / permissions / youth planning: PENDING / PENDING / PENDING')
print('Approved / planner-eligible: 0 / 0')
print('Input data and previous catalogs overwritten: No')
print('Git operations performed: None')
print('OVERALL: PASS — v0.3.7 reconciled; remaining review gates open')
```

```python
# Step 12-59 - Wave 3 Remaining Requirement Review and Signoff Ledger
# Consolidate pending equipment and participant checks without changing catalog approval status.

from __future__ import annotations

import csv
import hashlib
import json
import re
from collections import Counter, defaultdict
from datetime import date
from pathlib import Path


def project_root() -> Path:
    here = Path.cwd().resolve()
    for directory in (here, *here.parents):
        if (directory / '.git').exists():
            return directory
    raise FileNotFoundError('Project root (.git) not found.')


ROOT = project_root()
RESULTS = ROOT / 'results'
SOURCES = {
    'catalog': ROOT / 'data/processed/basketball_drill_catalog_v0_3_7_REVIEW_ONLY.csv',
    'registry': RESULTS / 'basketball_drill_source_registry_v0_3_7.csv',
    'pending': RESULTS / 'basketball_wave3_pending_requirement_reviews_v0_3_7.csv',
    'source_sections': RESULTS / 'basketball_wave3_drill_specific_source_evidence_v0_3_6.csv',
    'summary_58': RESULTS / 'basketball_step12_58_summary_v0_3_7.json',
}
OUTPUTS = {
    'field_ledger': RESULTS / 'basketball_wave3_remaining_field_review_v0_3_7.csv',
    'drill_packet': RESULTS / 'basketball_wave3_drill_signoff_packet_v0_3_7.csv',
    'page_plan': RESULTS / 'basketball_wave3_source_page_review_plan_v0_3_7.csv',
    'summary': RESULTS / 'basketball_step12_59_summary_v0_3_7.json',
}


def csv_rows(path: Path) -> tuple[list[dict], list[str]]:
    if not path.is_file():
        raise FileNotFoundError(f'Required file not found: {path}')
    with path.open('r', newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns or not rows:
        raise ValueError(f'Empty or invalid CSV: {path.name}')
    return rows, columns


def unique_by(rows: list[dict], key: str, label: str) -> dict[str, dict]:
    output = {}
    for row in rows:
        value = str(row.get(key, '')).strip()
        if not value or value in output:
            raise ValueError(f'{label}: empty or repeated {key}: {value!r}')
        output[value] = row
    return output


def digest(path: Path) -> str:
    return hashlib.sha256(path.read_bytes()).hexdigest()


def normalized_key(key: str) -> str:
    return str(key).replace('\\', '/').casefold()


for name, path in SOURCES.items():
    if not path.is_file():
        raise FileNotFoundError(f'{name}: {path}')

catalog, _ = csv_rows(SOURCES['catalog'])
registry, _ = csv_rows(SOURCES['registry'])
pending, _ = csv_rows(SOURCES['pending'])
sections, _ = csv_rows(SOURCES['source_sections'])
summary_58 = json.loads(SOURCES['summary_58'].read_text(encoding='utf-8'))
expected_hashes = {
    normalized_key(key): value.lower()
    for key, value in summary_58['output_sha256'].items()
}

for name in ('catalog', 'registry', 'pending'):
    path = SOURCES[name]
    key = normalized_key(path.relative_to(ROOT))
    expected = expected_hashes.get(key)
    actual = digest(path)
    if not expected:
        raise ValueError(f'{name}: checksum reference missing from Step 12-58 summary.')
    if actual.lower() != expected:
        raise ValueError(f'{name}: file differs from the verified Step 12-58 output.')

if len(catalog) != 144 or len(registry) != 156 or len(pending) != 74 or len(sections) != 42:
    raise ValueError('Unexpected record counts in verified Step 12 inputs.')

cat_map = unique_by(catalog, 'catalog_candidate_id', 'Catalog')
reg_map = unique_by(registry, 'evidence_id', 'Source registry')
section_map = unique_by(sections, 'evidence_id', 'Source-section audit')
if any(row.get('planner_eligible') != 'no' or
       row.get('catalog_approval_status') != 'pending'
       for row in catalog):
    raise ValueError('Unexpected candidate approval status; inspect before proceeding.')

unique_fields = set()
for row in pending:
    ident = (row['evidence_id'], row['field'])
    if ident in unique_fields:
        raise ValueError(f'Duplicate pending review: {ident}')
    unique_fields.add(ident)
    if row['field'] not in ('equipment', 'participants'):
        raise ValueError(f'Unexpected pending field: {row["field"]}')
    eid, gid = row['evidence_id'], row['candidate_group_id']
    if eid not in reg_map or gid not in cat_map or eid not in section_map:
        raise ValueError(f'Missing linked source record for {eid}')
    c, r = cat_map[gid], reg_map[eid]
    if r.get('catalog_candidate_id') != gid or c['evidence_ids'].split(';').count(eid) != 1:
        raise ValueError(f'Catalog-source linkage differs for {eid}')
    if row['current_value'] != c[row['field']] or row['current_value'] != r[row['field']]:
        raise ValueError(f'{eid} {row["field"]}: catalog / registry mismatch')
    if row['source_url'] != r['source_url'] or row['source_url'] != section_map[eid]['source_url']:
        raise ValueError(f'{eid}: official source URL mismatch')
    if section_map[eid]['section_status'] != 'drill_section_available':
        raise ValueError(f'{eid}: source section not available for review')
    if row.get('reviewer_decision', 'pending') != 'pending':
        raise ValueError(f'{eid}: pre-existing decisions require reconciliation')

if Counter(row['field'] for row in pending) != Counter({'equipment': 38, 'participants': 36}):
    raise ValueError('Unexpected remaining field mix.')


def field_questions(field: str, value: str, cues: str) -> tuple[list[str], str]:
    text_value = value.strip().casefold()
    questions = []
    if field == 'equipment':
        if any(term in text_value for term in ('court', 'half-court', 'full-court', 'sideline', 'baseline', 'space')):
            questions.append('separate_space_from_equipment')
        if any(term in text_value for term in ('optional', 'or shared', 'sharing', 'or coach', 'not explicitly', 'unknown', 'not stated')):
            questions.append('verify_optional_or_unknown_equipment')
        if re.search(r'\b(two|three|multiple|per person|per player|per pair|[2-9])\b', text_value):
            questions.append('verify_equipment_quantities')
        if not cues:
            questions.append('no_source_equipment_cues')
    else:
        if any(term in text_value for term in (' or ', 'optional', 'coach', 'group', 'team', 'multiple', 'substitute')):
            questions.append('verify_roles_minimum_and_optional_variations')
        if not any(term in text_value for term in ('individual', 'pair', 'two', 'three', 'four', 'five', 'team', 'group', 'coach', 'players', 'one ')):
            questions.append('clarify_minimum_active_participants')
        if not cues:
            questions.append('no_source_participant_cues')
    if not questions:
        questions.append('check_value_against_drill_specific_source')
    priority = 'high' if ('separate_space_from_equipment' in questions or
                          'no_source_participant_cues' in questions or
                          'no_source_equipment_cues' in questions) else (
        'medium' if len(questions) > 1 or questions[0] != 'check_value_against_drill_specific_source'
        else 'standard'
    )
    return questions, priority


fields = []
by_drill = defaultdict(list)
for row in pending:
    eid, gid, field = row['evidence_id'], row['candidate_group_id'], row['field']
    audit = section_map[eid]
    reg = reg_map[eid]
    cues = audit['drill_specific_equipment_cues' if field == 'equipment' else 'drill_specific_participant_cues']
    questions, priority = field_questions(field, row['current_value'], cues)
    fields.append({
        'candidate_group_id': gid,
        'evidence_id': eid,
        'drill_name': row['drill_name'],
        'source_level': reg.get('source_level', ''),
        'practice_plan': reg.get('practice_plan', ''),
        'field': field,
        'current_catalog_value': row['current_value'],
        'source_cues_not_requirements': cues,
        'review_questions': ';'.join(questions),
        'priority': priority,
        'source_url': row['source_url'],
        'source_section_sha256': audit.get('section_sha256', ''),
        'proposed_action': 'review_source_then_retain_revise_or_defer',
        'reviewer_decision': 'pending',
        'final_value': '',
        'source_evidence_note': '',
        'reviewer': '',
        'review_date': '',
        'team_signoff': 'pending',
        'permission_review': 'pending',
        'youth_planning_review': 'pending',
        'catalog_approval_status': 'pending',
        'planner_eligible': 'no',
    })
    by_drill[eid].append(fields[-1])

priority_rank = {'high': 0, 'medium': 1, 'standard': 2}
fields.sort(key=lambda r: (r['source_url'], priority_rank[r['priority']], r['drill_name'], r['field']))
packet = []
for eid, records in by_drill.items():
    reg = reg_map[eid]
    fieldset = {r['field']: r for r in records}
    packet.append({
        'candidate_group_id': reg['catalog_candidate_id'],
        'evidence_id': eid,
        'drill_name': reg['drill_name'],
        'source_level': reg.get('source_level', ''),
        'practice_plan': reg.get('practice_plan', ''),
        'source_url': reg['source_url'],
        'pending_field_count': len(records),
        'equipment_value': reg.get('equipment', ''),
        'equipment_review': 'pending' if 'equipment' in fieldset else 'handled_prior_step',
        'participants_value': reg.get('participants', ''),
        'participants_review': 'pending' if 'participants' in fieldset else 'handled_prior_step',
        'highest_priority': min((r['priority'] for r in records), key=lambda p: priority_rank[p]),
        'open_questions': ';'.join(sorted({q for r in records for q in r['review_questions'].split(';')})),
        'independent_source_review': 'pending',
        'distinctness_and_variation_review': 'pending',
        'team_signoff': 'pending',
        'planner_eligible': 'no',
    })
packet.sort(key=lambda r: (r['source_url'], priority_rank[r['highest_priority']], r['drill_name']))

by_page = defaultdict(list)
for row in packet:
    by_page[row['source_url']].append(row)
page_plan = []
for url, records in sorted(by_page.items()):
    page_plan.append({
        'source_url': url,
        'curriculum_level': ';'.join(sorted({r['source_level'] for r in records})),
        'practice_plan': ';'.join(sorted({r['practice_plan'] for r in records})),
        'candidate_drills_to_review': len(records),
        'pending_fields_on_page': sum(int(r['pending_field_count']) for r in records),
        'high_priority_drills': sum(r['highest_priority'] == 'high' for r in records),
        'reviewer_status': 'pending',
    })

summary = {
    'step': '12-59',
    'catalog_version': '0.3.7',
    'catalog_candidates': len(catalog),
    'registered_source_occurrences': len(registry),
    'pending_wave3_requirement_fields': len(fields),
    'fields_by_type': dict(sorted(Counter(r['field'] for r in fields).items())),
    'distinct_drills_with_pending_fields': len(packet),
    'official_source_pages': len(page_plan),
    'review_priority_by_field': dict(sorted(Counter(r['priority'] for r in fields).items())),
    'review_priority_by_drill': dict(sorted(Counter(r['highest_priority'] for r in packet).items())),
    'source_cues_are_not_approval': True,
    'approved': 0,
    'planner_eligible': 0,
    'git_operations_performed': False,
    'previous_catalog_and_registry_verified': True,
}


def serialized_csv(rows: list[dict]) -> bytes:
    import io
    output = io.StringIO(newline='')
    writer = csv.DictWriter(output, fieldnames=list(rows[0]))
    writer.writeheader()
    writer.writerows(rows)
    return output.getvalue().encode('utf-8')


def serialized_json(data: dict) -> bytes:
    return (json.dumps(data, indent=2, sort_keys=True, ensure_ascii=False) + '\n').encode('utf-8')

content = {
    OUTPUTS['field_ledger']: serialized_csv(fields),
    OUTPUTS['drill_packet']: serialized_csv(packet),
    OUTPUTS['page_plan']: serialized_csv(page_plan),
    OUTPUTS['summary']: serialized_json(summary),
}
existing = [p for p in content if p.exists()]
if existing and len(existing) != len(content):
    raise FileExistsError('Incomplete prior Step 12-59 outputs; review them before running again.')

if existing:
    print('Existing Step 12-59 review files preserved; reviewer edits were not overwritten.')
else:
    for path, data in content.items():
        path.parent.mkdir(parents=True, exist_ok=True)
        with path.open('xb') as f:
            f.write(data)
        print('CREATED', str(path.relative_to(ROOT)))

# Confirm outputs exist, and all saved review sheets retain expected record counts.
for path, expected in [(OUTPUTS['field_ledger'], 74), (OUTPUTS['drill_packet'], len(packet)), (OUTPUTS['page_plan'], len(page_plan))]:
    rows, _ = csv_rows(path)
    if len(rows) != expected:
        raise ValueError(f'Saved output count changed: {path.name}')

print('=' * 72)
print('STEP 12-59 — WAVE 3 REMAINING REQUIREMENTS REVIEW LEDGER')
print('=' * 72)
print('INPUT INTEGRITY: PASS — Catalog v0.3.7, registry, and pending queue match Step 12-58.')
print(f'Catalog candidates / source occurrences: {len(catalog)} / {len(registry)}')
print(f'Pending equipment / participant fields: 38 / 36')
print(f'Drills needing one or two remaining field reviews: {len(packet)}')
print(f'Official source pages grouped for review: {len(page_plan)}')
print('REVIEW PRIORITY BY DRILL:', dict(Counter(r['highest_priority'] for r in packet)))
print('\nSOURCE-PAGE WORKLOAD')
for row in page_plan:
    print(f"  {row['candidate_drills_to_review']:>2} drills | {row['pending_fields_on_page']:>2} fields | {row['source_url']}")
print('\nDECISIONS / APPROVALS')
print('  Remaining 74 fields: pending independent source review')
print('  Team signoffs / approved / planner-eligible: 0 / 0 / 0')
print('  Original catalogs and registries modified: No')
print('  Git operations performed: None')
print('OVERALL: PASS — review work grouped; no unsupported approvals or metadata changes')
```

```python
# Step 12-60 - Wave 3 Source-Specific Requirement Adjudication (All-Star Plan 1)
# Prepare source-grounded field recommendations without modifying catalog or approval data.

import csv
import hashlib
import json
import re
import unicodedata
from collections import Counter, defaultdict
from html.parser import HTMLParser
from pathlib import Path
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

ROOT = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Project root not found (missing .git directory).')
RESULTS = ROOT / 'results'
DATA = ROOT / 'data/processed'
SOURCE_URL = 'https://jr.nba.com/basketball-practice-plans/all-star/'

INPUTS = {
    'field_ledger': RESULTS / 'basketball_wave3_remaining_field_review_v0_3_7.csv',
    'drill_packet': RESULTS / 'basketball_wave3_drill_signoff_packet_v0_3_7.csv',
    'catalog': DATA / 'basketball_drill_catalog_v0_3_7_REVIEW_ONLY.csv',
    'registry': RESULTS / 'basketball_drill_source_registry_v0_3_7.csv',
    'previous_summary': RESULTS / 'basketball_step12_58_summary_v0_3_7.json',
}
OUTPUTS = {
    'decisions': RESULTS / 'basketball_wave3_plan1_field_decisions_v0_3_7.csv',
    'candidates': RESULTS / 'basketball_wave3_plan1_candidate_summary_v0_3_7.csv',
    'remaining': RESULTS / 'basketball_wave3_remaining_after_plan1_v0_3_7.csv',
    'summary': RESULTS / 'basketball_step12_60_summary_v0_3_7.json',
}

def sha256(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()

def load_csv(path):
    if not path.is_file():
        raise FileNotFoundError(path)
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        fields = list(reader.fieldnames or [])
        return list(reader), fields

def unique_index(rows, field, name):
    values = [row.get(field, '') for row in rows]
    if any(not value for value in values) or len(values) != len(set(values)):
        raise ValueError(f'{name}: blank or repeated {field}.')
    return {row[field]: row for row in rows}

for input_name, path in INPUTS.items():
    if not path.is_file():
        raise FileNotFoundError(f'{input_name}: {path}')

ledger, ledger_fields = load_csv(INPUTS['field_ledger'])
packet, _ = load_csv(INPUTS['drill_packet'])
catalog, _ = load_csv(INPUTS['catalog'])
registry, _ = load_csv(INPUTS['registry'])
summary58 = json.loads(INPUTS['previous_summary'].read_text(encoding='utf-8'))

if (len(ledger), len(packet), len(catalog), len(registry)) != (74, 41, 144, 156):
    raise ValueError('The Step 12-59 review scope or current catalog counts have changed.')

hashes = {key.replace('\\', '/'): value for key, value
          in summary58.get('output_sha256', {}).items()}
for name in ('catalog', 'registry'):
    path = INPUTS[name]
    key = path.relative_to(ROOT).as_posix()
    if hashes.get(key) != sha256(path):
        raise ValueError(f'{name}: SHA-256 differs from the verified Step 12-58 output.')

cat_by = unique_index(catalog, 'catalog_candidate_id', 'Catalog')
reg_by = unique_index(registry, 'evidence_id', 'Source registry')
packet_by = unique_index(packet, 'candidate_group_id', 'Drill packet')
field_keys = [(r['evidence_id'], r['field']) for r in ledger]
if len(field_keys) != len(set(field_keys)):
    raise ValueError('Repeated evidence/field review records.')

for row in ledger:
    eid, cid, field = row['evidence_id'], row['candidate_group_id'], row['field']
    if field not in {'equipment', 'participants'}:
        raise ValueError(f'Unexpected field: {field}')
    if cid not in cat_by or eid not in reg_by or cid not in packet_by:
        raise ValueError(f'Unlinked review record: {eid}/{cid}')
    if reg_by[eid]['catalog_candidate_id'] != cid:
        raise ValueError(f'Registry candidate mismatch: {eid}')
    if row['current_catalog_value'] != cat_by[cid][field]:
        raise ValueError(f'Catalog value changed for {eid}/{field}.')
    if row['current_catalog_value'] != reg_by[eid][field]:
        raise ValueError(f'Registry value changed for {eid}/{field}.')
    if row['source_url'].rstrip('/') != reg_by[eid]['source_url'].rstrip('/'):
        raise ValueError(f'Source URL mismatch for {eid}.')
    if (row['reviewer_decision'] != 'pending'
            or row['catalog_approval_status'] != 'pending'
            or row['planner_eligible'] != 'no'):
        raise ValueError(f'Existing reviewer or approval status must be preserved: {eid}')

if any(row['catalog_approval_status'] != 'pending' or row['planner_eligible'] != 'no'
       for row in catalog):
    raise ValueError('Non-pending catalog approvals detected; review before proceeding.')

print('INPUT INTEGRITY: PASS — Catalog v0.3.7 and source registry verified.')

# Factual source anchors: these are checks within each individual drill section,
# not evidence that an entire candidate is approved or safe for every athlete.
REVIEW = {
    ('JR-EVID-0012', 'equipment'): {
        'action': 'retain_current',
        'new_value': None,
        'anchors': [r'\bcones\b', r'\bball\b', r'\bshot\b'],
        'note': 'Source describes a ball handler moving through cone-marked lanes and finishing with a shot.',
    },
    ('JR-EVID-0012', 'participants'): {
        'action': 'revise_value',
        'new_value': 'One offensive ball handler and 2-3 defenders per lane; players rotate roles',
        'anchors': [r'\b2\s*[-–]\s*3\s+defenders\b', r'\boffensive player\b'],
        'note': 'Source specifies 2-3 defenders in each lane, one active ball handler, and role rotations.',
    },
    ('JR-EVID-0013', 'equipment'): {
        'action': 'retain_current',
        'new_value': None,
        'anchors': [r'\bbasketball\b', r'\bhoop\b'],
        'note': 'Published base activity uses one basketball, with participants facing the basket.',
    },
    ('JR-EVID-0013', 'participants'): {
        'action': 'retain_current',
        'new_value': None,
        'anchors': [r'\bpartner\b', r'\bone player\b'],
        'note': 'Published activity uses a pair, with a dribbler and a partner providing visual cues.',
    },
    ('JR-EVID-0014', 'equipment'): {
        'action': 'retain_current',
        'new_value': None,
        'anchors': [r'\bball\b', r'\bbasket\b'],
        'note': 'Published activity uses basketballs and a basket for the two-on-one decision.',
    },
    ('JR-EVID-0014', 'participants'): {
        'action': 'revise_value',
        'new_value': 'Two active attackers and one defender; additional players in wing passing lines',
        'anchors': [r'\bone defender\b', r'\b2\s+on\s+1\b', r'\bwing\b'],
        'note': 'Published setup includes wing passing lines and one defender in the live two-on-one.',
    },
}

current = {(r['evidence_id'], r['field']): r for r in ledger}
if set(REVIEW) - set(current):
    raise ValueError('One or more targeted source IDs are not in the 74-field ledger.')
if any(current[key]['source_url'].rstrip('/') != SOURCE_URL.rstrip('/') for key in REVIEW):
    raise ValueError('Targeted records are not all from All-Star Practice Plan 1.')

class HeadingSectionParser(HTMLParser):
    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.heading_tag = None
        self.heading_parts = []
        self.skip_stack = []
        self.current = None

    def handle_starttag(self, tag, attrs):
        if tag in {'script', 'style', 'noscript'}:
            self.skip_stack.append(tag)
            return
        if self.skip_stack:
            return
        if re.fullmatch(r'h[1-6]', tag):
            self.heading_tag = tag
            self.heading_parts = []

    def handle_data(self, data):
        if self.skip_stack:
            return
        if self.heading_tag:
            self.heading_parts.append(data)
        elif self.current is not None:
            self.current['text_parts'].append(data)

    def handle_endtag(self, tag):
        if self.skip_stack:
            if tag == self.skip_stack[-1]:
                self.skip_stack.pop()
            return
        if tag == self.heading_tag:
            name = ' '.join(self.heading_parts).strip()
            if name:
                self.current = {'heading': name, 'text_parts': []}
                self.sections.append(self.current)
            self.heading_tag = None
            self.heading_parts = []
        elif tag in {'p', 'li', 'br', 'div'} and self.current is not None:
            self.current['text_parts'].append(' ')

def normalize(text):
    return re.sub(r'[^a-z0-9]+', ' ',
                  unicodedata.normalize('NFKC', str(text)).casefold()).strip()

url_info = urlsplit(SOURCE_URL)
if url_info.scheme != 'https' or url_info.hostname != 'jr.nba.com':
    raise ValueError('Official source URL validation failed.')

existing = [path.exists() for path in OUTPUTS.values()]
if any(existing) and not all(existing):
    raise FileExistsError('Only part of the Step 12-60 output set exists; inspect before rerunning.')

if not all(existing):
    request = Request(SOURCE_URL, headers={'User-Agent': 'Mozilla/5.0'})
    with urlopen(request, timeout=30) as response:
        if response.status != 200:
            raise RuntimeError(f'Official source returned HTTP {response.status}.')
        html = response.read(2_000_000).decode('utf-8', errors='replace')

    parser = HeadingSectionParser()
    parser.feed(html)
    names = {name for name in ('Gauntlet Dribbling',
                               'Partner Decision Dribbling Game',
                               'Skip Pass 2 on 1')}
    sections = defaultdict(list)
    for section in parser.sections:
        sections[normalize(section['heading'])].append(
            re.sub(r'\s+', ' ', ' '.join(section['text_parts'])).strip()
        )
    if any(len(sections[normalize(name)]) != 1 for name in names):
        raise ValueError('Official source drill heading changed or is ambiguous. No files created.')

    decisions = []
    for (eid, field), recommendation in REVIEW.items():
        source = current[(eid, field)]
        text = sections[normalize(source['drill_name'])][0]
        source_checks = [bool(re.search(pat, text, flags=re.I))
                         for pat in recommendation['anchors']]
        supported = all(source_checks)
        action = recommendation['action'] if supported else 'defer_source_evidence'
        proposed = ((source['current_catalog_value'] if recommendation['new_value'] is None
                     else recommendation['new_value']) if supported else '')
        decisions.append({
            'evidence_id': eid,
            'candidate_group_id': source['candidate_group_id'],
            'drill_name': source['drill_name'],
            'field': field,
            'current_value': source['current_catalog_value'],
            'proposed_action': action,
            'proposed_value': proposed,
            'source_evidence_note': recommendation['note'] if supported else 'Source anchors need manual review.',
            'heading_found': 'yes',
            'source_anchor_checks': f'{sum(source_checks)}/{len(source_checks)}',
            'section_fingerprint_sha256': hashlib.sha256(
                normalize(text).encode('utf-8')).hexdigest(),
            'source_url': SOURCE_URL,
            'reviewer_decision': 'pending',
            'reviewer': '',
            'review_date': '',
            'team_signoff': 'pending',
            'permission_review': 'pending',
            'youth_planning_review': 'pending',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
        })

    decisions.sort(key=lambda r: (r['evidence_id'], r['field']))
    candidate_rows = []
    for eid in ('JR-EVID-0012', 'JR-EVID-0013', 'JR-EVID-0014'):
        rows = [r for r in decisions if r['evidence_id'] == eid]
        candidate_rows.append({
            'candidate_group_id': rows[0]['candidate_group_id'],
            'evidence_id': eid,
            'drill_name': rows[0]['drill_name'],
            'reviewed_fields': len(rows),
            'retain_recommendations': sum(r['proposed_action'] == 'retain_current' for r in rows),
            'revision_recommendations': sum(r['proposed_action'] == 'revise_value' for r in rows),
            'evidence_gaps': sum(r['proposed_action'] == 'defer_source_evidence' for r in rows),
            'source_url': SOURCE_URL,
            'human_review_status': 'pending',
            'catalog_approved': 'no',
            'planner_eligible': 'no',
        })

    remaining = [r for r in ledger if (r['evidence_id'], r['field']) not in REVIEW]
    if len(decisions) != 6 or len(candidate_rows) != 3 or len(remaining) != 68:
        raise ValueError('Review partition counts are incorrect; no files created.')

    summary = {
        'step': '12-60', 'catalog_version': '0.3.7',
        'catalog_candidate_groups': len(catalog),
        'registered_source_occurrences': len(registry),
        'initial_pending_field_reviews': len(ledger),
        'source_page': SOURCE_URL,
        'first_batch_candidate_groups': len(candidate_rows),
        'first_batch_field_recommendations': len(decisions),
        'source_anchor_review_needed': sum(r['proposed_action'] == 'defer_source_evidence' for r in decisions),
        'proposed_retain': sum(r['proposed_action'] == 'retain_current' for r in decisions),
        'proposed_revise': sum(r['proposed_action'] == 'revise_value' for r in decisions),
        'remaining_field_reviews_not_yet_adjudicated': len(remaining),
        'reviewer_signoffs': 0, 'catalog_approvals': 0,
        'planner_eligible': 0,
        'input_sha256': {path.relative_to(ROOT).as_posix(): sha256(path)
                         for path in INPUTS.values()},
        'original_inputs_modified': False,
        'git_operations_performed': False,
    }

    def write_new_csv(path, rows, columns):
        with path.open('x', newline='', encoding='utf-8') as file:
            writer = csv.DictWriter(file, fieldnames=columns)
            writer.writeheader()
            writer.writerows(rows)

    write_new_csv(OUTPUTS['decisions'], decisions, list(decisions[0]))
    write_new_csv(OUTPUTS['candidates'], candidate_rows, list(candidate_rows[0]))
    write_new_csv(OUTPUTS['remaining'], remaining, ledger_fields)
    with OUTPUTS['summary'].open('x', encoding='utf-8') as file:
        json.dump(summary, file, indent=2)
else:
    decisions, _ = load_csv(OUTPUTS['decisions'])
    candidate_rows, _ = load_csv(OUTPUTS['candidates'])
    remaining, _ = load_csv(OUTPUTS['remaining'])
    summary = json.loads(OUTPUTS['summary'].read_text(encoding='utf-8'))
    for path in INPUTS.values():
        key = path.relative_to(ROOT).as_posix()
        if summary.get('input_sha256', {}).get(key) != sha256(path):
            raise ValueError(f'Previously saved Step 12-60 inputs changed: {key}')
    if len(decisions) != 6 or len(candidate_rows) != 3 or len(remaining) != 68:
        raise ValueError('Existing Step 12-60 outputs have unexpected counts.')

print('=' * 72)
print('STEP 12-60 — ALL-STAR PLAN 1 REQUIREMENT REVIEW')
print('=' * 72)
print('Catalog candidates / source occurrences:', len(catalog), '/', len(registry))
print('Initial Wave 3 pending fields:', len(ledger))
print('First-batch drills / field recommendations:', len(candidate_rows), '/', len(decisions))
print('RECOMMENDATIONS (PENDING YOUR REVIEW)')
for row in decisions:
    print(f"  {row['evidence_id']} | {row['drill_name']} | {row['field']} | {row['proposed_action']} | anchors={row['source_anchor_checks']}")
print('Still pending in other source pages:', len(remaining))
print('Reviewer signoffs / approved / planner-eligible: 0 / 0 / 0')
print('\nOUTPUTS')
for name, path in OUTPUTS.items():
    print(f"  {'PRESERVED' if all(existing) else 'CREATED  '} {path.relative_to(ROOT)}")
print('Source catalogs and registries modified: No')
print('Git operations performed: None')
print('OVERALL: PASS — first source-page decisions proposed; signoff remains pending')
print('=' * 72)
```

```python
# Step 12-61 - Confirmed All-Star Plan 1 Metadata Reconciliation
# Apply reviewed participant descriptions while preserving source and approval controls.

import csv
import hashlib
import json
import re
from collections import Counter
from datetime import datetime
from pathlib import Path
from tempfile import TemporaryDirectory
from zoneinfo import ZoneInfo

ROOT = next(
    (path for path in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (path / '.git').exists()),
    None,
)
if ROOT is None:
    raise FileNotFoundError('Git project root not found.')

RESULTS = ROOT / 'results'
PROCESSED = ROOT / 'data' / 'processed'
INPUT = {
    'catalog': PROCESSED / 'basketball_drill_catalog_v0_3_7_REVIEW_ONLY.csv',
    'registry': RESULTS / 'basketball_drill_source_registry_v0_3_7.csv',
    'field_decisions': RESULTS / 'basketball_wave3_plan1_field_decisions_v0_3_7.csv',
    'remaining': RESULTS / 'basketball_wave3_remaining_after_plan1_v0_3_7.csv',
    'original_pending': RESULTS / 'basketball_wave3_remaining_field_review_v0_3_7.csv',
    'step60_summary': RESULTS / 'basketball_step12_60_summary_v0_3_7.json',
    'step58_summary': RESULTS / 'basketball_step12_58_summary_v0_3_7.json',
}
OUTPUT = {
    'catalog': PROCESSED / 'basketball_drill_catalog_v0_3_8_REVIEW_ONLY.csv',
    'registry': RESULTS / 'basketball_drill_source_registry_v0_3_8.csv',
    'confirmed': RESULTS / 'basketball_wave3_plan1_confirmed_field_decisions_v0_3_8.csv',
    'changes': RESULTS / 'basketball_wave3_plan1_metadata_change_log_v0_3_8.csv',
    'pending': RESULTS / 'basketball_wave3_remaining_requirement_reviews_v0_3_8.csv',
    'summary': RESULTS / 'basketball_step12_61_summary_v0_3_8.json',
}


def sha256(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def rel(path):
    return path.relative_to(ROOT).as_posix()


def read_csv(path):
    if not path.is_file():
        raise FileNotFoundError(f'Required input not found: {path}')
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    if not columns:
        raise ValueError(f'File has no CSV header: {path}')
    return rows, columns


def index_unique(rows, name, field):
    keys = [row.get(field, '') for row in rows]
    if any(not key for key in keys) or len(keys) != len(set(keys)):
        raise ValueError(f'{name}: missing or duplicate {field}.')
    return {row[field]: row for row in rows}


def normalized_hashes(mapping):
    return {key.replace('\\', '/'): value for key, value in mapping.items()}


def csv_bytes(rows, columns):
    import io
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=columns, lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')


for name, path in INPUT.items():
    if not path.is_file():
        raise FileNotFoundError(f'{name}: {path}')

catalog, catalog_columns = read_csv(INPUT['catalog'])
registry, registry_columns = read_csv(INPUT['registry'])
proposals, _ = read_csv(INPUT['field_decisions'])
remaining, remaining_columns = read_csv(INPUT['remaining'])
original_pending, _ = read_csv(INPUT['original_pending'])
summary60 = json.loads(INPUT['step60_summary'].read_text(encoding='utf-8'))
summary58 = json.loads(INPUT['step58_summary'].read_text(encoding='utf-8'))

if (len(catalog), len(registry), len(proposals), len(remaining), len(original_pending)) != (144, 156, 6, 68, 74):
    raise ValueError('Input record counts differ from the verified Step 12-60 scope.')
if summary60.get('step') != '12-60' or summary60.get('source_anchor_review_needed') != 0:
    raise ValueError('Step 12-60 source-review summary does not support this reconciliation.')

checks58 = normalized_hashes(summary58.get('output_sha256', {}))
checks60 = normalized_hashes(summary60.get('input_sha256', {}))
for name in ('catalog', 'registry'):
    path = INPUT[name]
    if checks58.get(rel(path)) != sha256(path):
        raise ValueError(f'{name} differs from the verified Step 12-58 source snapshot.')
for name in ('catalog', 'registry', 'original_pending', 'step58_summary'):
    path = INPUT[name]
    if checks60.get(rel(path)) != sha256(path):
        raise ValueError(f'{name} differs from the inputs used in Step 12-60.')

cat_by = index_unique(catalog, 'Catalog', 'catalog_candidate_id')
reg_by = index_unique(registry, 'Registry', 'evidence_id')
proposal_by = {
    (row['evidence_id'], row['field']): row for row in proposals
}
if len(proposal_by) != 6:
    raise ValueError('Duplicate evidence/field combinations in Step 12-60 recommendations.')

# Confirmed on Step 12-60: four retained values and two participant revisions.
DECISIONS = {
    ('JR-EVID-0012', 'equipment'): 'retain_current',
    ('JR-EVID-0012', 'participants'): 'revise_value',
    ('JR-EVID-0013', 'equipment'): 'retain_current',
    ('JR-EVID-0013', 'participants'): 'retain_current',
    ('JR-EVID-0014', 'equipment'): 'retain_current',
    ('JR-EVID-0014', 'participants'): 'revise_value',
}
REVISED_PARTICIPANTS = {
    'JR-EVID-0012': 'One offensive ball handler and 2-3 defenders per lane; players rotate roles',
    'JR-EVID-0014': 'Two active attackers and one defender; additional players in wing passing lines',
}
if set(proposal_by) != set(DECISIONS):
    raise ValueError('The source-review proposals differ from the confirmed six decisions.')
if (Counter(row['proposed_action'] for row in proposals)
        != Counter({'retain_current': 4, 'revise_value': 2})):
    raise ValueError('Unexpected retain/revise recommendation counts.')

if any(row.get('catalog_version') != '0.3.7' for row in catalog):
    raise ValueError('Catalog version or source group state has changed.')
if any(row.get('catalog_approval_status') != 'pending'
       or row.get('planner_eligible') != 'no'
       or row.get('approved_drill_id', '')
       for row in catalog):
    raise ValueError('Catalog contains approval decisions; cannot apply this update automatically.')
if any(row.get('planner_eligible', '') not in {'', 'no'} for row in registry):
    raise ValueError('Source registry includes unexpected planner eligibility.')

confirmed = []
changes = []
review_date = datetime.now(ZoneInfo('America/Los_Angeles')).date().isoformat()
for key, action in DECISIONS.items():
    evidence_id, field = key
    proposal = proposal_by[key]
    if evidence_id not in reg_by:
        raise ValueError(f'Unknown source occurrence: {evidence_id}.')
    source = reg_by[evidence_id]
    cid = source['catalog_candidate_id']
    if cid not in cat_by:
        raise ValueError(f'Source record missing catalog group: {evidence_id}.')
    item = cat_by[cid]
    if (item['evidence_ids'].split(';') != [evidence_id]
            or int(item['source_occurrences']) != 1):
        raise ValueError(f'{evidence_id} is not a single-source catalog group; manual review required.')
    if (proposal['candidate_group_id'] != cid
            or proposal['proposed_action'] != action
            or proposal['reviewer_decision'] != 'pending'
            or proposal['team_signoff'] != 'pending'
            or proposal['permission_review'] != 'pending'
            or proposal['youth_planning_review'] != 'pending'
            or proposal['catalog_approval_status'] != 'pending'
            or proposal['planner_eligible'] != 'no'):
        raise ValueError(f'Review status or group linkage changed for {evidence_id}/{field}.')
    old = item[field]
    if old != source[field] or old != proposal['current_value']:
        raise ValueError(f'Catalog/registry/proposal value mismatch for {evidence_id}/{field}.')
    if (source['source_url'].rstrip('/') != proposal['source_url'].rstrip('/')
            or proposal['heading_found'] != 'yes'
            or not re.fullmatch(r'[0-9a-f]{64}', proposal['section_fingerprint_sha256'])):
        raise ValueError(f'Insufficient source traceability for {evidence_id}/{field}.')
    anchor = re.fullmatch(r'(\d+)/(\d+)', proposal['source_anchor_checks'])
    if not anchor or int(anchor[1]) != int(anchor[2]) or int(anchor[2]) == 0:
        raise ValueError(f'Source anchor checks are incomplete for {evidence_id}/{field}.')
    new = (REVISED_PARTICIPANTS[evidence_id]
           if action == 'revise_value' else old)
    if proposal['proposed_value'] != new:
        raise ValueError(f'Proposed final value changed for {evidence_id}/{field}.')
    if action == 'revise_value' and (field != 'participants' or new == old):
        raise ValueError(f'Unsupported or redundant revision: {evidence_id}/{field}.')
    confirmed.append({
        'evidence_id': evidence_id,
        'candidate_group_id': cid,
        'drill_name': proposal['drill_name'],
        'field': field,
        'decision': action,
        'original_value': old,
        'confirmed_value': new,
        'value_changed': 'yes' if old != new else 'no',
        'decision_basis': 'user_confirmed_after_source_evidence_review',
        'decision_date': review_date,
        'source_url': proposal['source_url'],
        'source_section_sha256': proposal['section_fingerprint_sha256'],
        'source_evidence_note': proposal['source_evidence_note'],
        'team_signoff': 'pending',
        'permission_review': 'pending',
        'youth_planning_review': 'pending',
        'catalog_approval_status': 'pending',
        'planner_eligible': 'no',
    })
    if old != new:
        item[field] = new
        source[field] = new
        changes.append({
            'evidence_id': evidence_id,
            'candidate_group_id': cid,
            'drill_name': proposal['drill_name'],
            'field': field,
            'previous_value': old,
            'new_value': new,
            'decision_reference': rel(OUTPUT['confirmed']),
            'source_url': proposal['source_url'],
            'team_signoff': 'pending',
        })

if (len(confirmed), len(changes),
        sum(row['decision'] == 'retain_current' for row in confirmed)) != (6, 2, 4):
    raise ValueError('Confirmed decision totals or actual metadata changes differ from expectation.')

# All other catalog and source fields remain unchanged except the catalog version.
for item in catalog:
    item['catalog_version'] = '0.3.8'
if len({row['evidence_id'] for row in registry}) != 156:
    raise ValueError('Source registry identifiers changed.')

remaining_keys = {(row['evidence_id'], row['field']) for row in remaining}
original_keys = {(row['evidence_id'], row['field']) for row in original_pending}
if (len(remaining_keys) != 68 or len(original_keys) != 74
        or remaining_keys & set(DECISIONS)
        or remaining_keys | set(DECISIONS) != original_keys):
    raise ValueError('The remaining 68 review fields do not partition the original 74 correctly.')
if any(row['reviewer_decision'] != 'pending' or row['team_signoff'] != 'pending'
       or row['catalog_approval_status'] != 'pending' or row['planner_eligible'] != 'no'
       for row in remaining):
    raise ValueError('An unprocessed field contains an unexpected review or approval state.')
if any(row['catalog_approval_status'] != 'pending' or row['planner_eligible'] != 'no'
       or row['approved_drill_id'] for row in catalog):
    raise ValueError('An approval restriction was unexpectedly changed.')

print('INPUT INTEGRITY: PASS — v0.3.7 inputs match Steps 12-58 and 12-60.')

input_checksums = {rel(path): sha256(path) for path in INPUT.values()}
output_existing = [path.exists() for path in OUTPUT.values()]
if any(output_existing) and not all(output_existing):
    raise FileExistsError('Only part of the Step 12-61 outputs exists. Inspect before rerunning.')

if not any(output_existing):
    new_files = {
        'catalog': csv_bytes(catalog, catalog_columns),
        'registry': csv_bytes(registry, registry_columns),
        'confirmed': csv_bytes(confirmed, list(confirmed[0])),
        'changes': csv_bytes(changes, list(changes[0])),
        'pending': INPUT['remaining'].read_bytes(),
    }
    output_hashes = {rel(OUTPUT[name]): hashlib.sha256(contents).hexdigest()
                     for name, contents in new_files.items()}
    summary = {
        'step': '12-61',
        'catalog_version': '0.3.8',
        'catalog_candidate_groups': len(catalog),
        'registered_source_occurrences': len(registry),
        'source_plan': 'All-Star Practice Plan 1',
        'confirmed_metadata_decisions': 6,
        'retain_current': 4,
        'revised_participant_fields': 2,
        'pending_other_wave3_fields': len(remaining),
        'team_signoffs': 0,
        'catalog_approvals': 0,
        'planner_eligible': 0,
        'decision_authority': 'user_confirmed_metadata_only',
        'input_sha256': input_checksums,
        'output_sha256': output_hashes,
        'original_inputs_modified': False,
        'git_operations_performed': False,
    }
    new_files['summary'] = (json.dumps(summary, indent=2, ensure_ascii=False) + '\n').encode('utf-8')
    # Stage and validate before creating any final destination.
    with TemporaryDirectory(prefix='step12_61_', dir=ROOT) as staging_dir:
        staged = {}
        for name, contents in new_files.items():
            staged_file = Path(staging_dir) / OUTPUT[name].name
            staged_file.write_bytes(contents)
            staged[name] = staged_file
            if staged_file.read_bytes() != contents:
                raise RuntimeError(f'Staging integrity failure: {name}.')
        for path in OUTPUT.values():
            if path.exists():
                raise FileExistsError(f'Output appeared during staging: {path}.')
        created = []
        try:
            for name, target in OUTPUT.items():
                target.parent.mkdir(parents=True, exist_ok=True)
                with target.open('xb') as file:
                    file.write(staged[name].read_bytes())
                created.append(target)
        except Exception:
            for path in created:
                path.unlink(missing_ok=True)
            raise
    mode = 'CREATED'
else:
    summary = json.loads(OUTPUT['summary'].read_text(encoding='utf-8'))
    if summary.get('step') != '12-61' or summary.get('catalog_version') != '0.3.8':
        raise ValueError('Existing Step 12-61 summary is not the expected version.')
    checks = normalized_hashes(summary.get('input_sha256', {}))
    if any(checks.get(path) != digest for path, digest in input_checksums.items()):
        raise ValueError('Existing Step 12-61 outputs were created from different input files.')
    output_checks = normalized_hashes(summary.get('output_sha256', {}))
    if any(output_checks.get(rel(OUTPUT[name])) != sha256(OUTPUT[name])
           for name in OUTPUT if name != 'summary'):
        raise ValueError('Existing Step 12-61 output differs from its verified snapshot; preserved.')
    mode = 'PRESERVED'

# Verify outputs and the exact changes to previously unmodified files.
saved_catalog, _ = read_csv(OUTPUT['catalog'])
saved_registry, _ = read_csv(OUTPUT['registry'])
saved_decisions, _ = read_csv(OUTPUT['confirmed'])
saved_changes, _ = read_csv(OUTPUT['changes'])
saved_pending, _ = read_csv(OUTPUT['pending'])
if (len(saved_catalog), len(saved_registry), len(saved_decisions),
        len(saved_changes), len(saved_pending)) != (144, 156, 6, 2, 68):
    raise ValueError('Saved output record counts did not pass verification.')
if INPUT['remaining'].read_bytes() != OUTPUT['pending'].read_bytes():
    raise ValueError('The 68 pending review records were altered.')
if any(row['catalog_approval_status'] != 'pending' or row['planner_eligible'] != 'no'
       for row in saved_catalog):
    raise ValueError('Saved catalog contains an unexpected approval status.')
if any(row['team_signoff'] != 'pending' for row in saved_decisions):
    raise ValueError('A field decision was incorrectly recorded as team-approved.')

print('=' * 72)
print('STEP 12-61 — CONFIRMED ALL-STAR PLAN 1 METADATA RECONCILIATION')
print('=' * 72)
print(f'Catalog candidate groups / source occurrences: {len(saved_catalog)} / {len(saved_registry)}')
print('Confirmed metadata decisions: 4 retained, 2 participant revisions')
for row in saved_changes:
    print(f"  {row['evidence_id']} | {row['drill_name']} | {row['field']}")
    print(f"    {row['previous_value']} -> {row['new_value']}")
print('Remaining Wave 3 equipment/participant fields: 68 (unchanged)')
print('OUTPUTS')
for name, path in OUTPUT.items():
    print(f'  {mode:<9} {rel(path)}')
print('Team signoff / permissions / youth planning: PENDING / PENDING / PENDING')
print('Approved / planner-eligible: 0 / 0')
print('Previous catalogs, source inputs, and remaining review records overwritten: No')
print('Git operations performed: None')
print('OVERALL: PASS — v0.3.8 prepared; remaining approval gates preserved')
print('=' * 72)
```

```python
# Step 12-62 - Wave 3 Requirement Review: All-Star Practice Plan 2
# Prepare four source-grounded recommendations; all decisions remain pending.

import csv
import hashlib
import io
import json
import re
import unicodedata
from collections import Counter, defaultdict
from datetime import datetime, timezone
from html.parser import HTMLParser
from pathlib import Path
from tempfile import TemporaryDirectory
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

ROOT = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Run this cell inside the Champions Athletic Map Git project.')

RESULTS = ROOT / 'results'
INPUT = {
    'catalog': ROOT / 'data/processed/basketball_drill_catalog_v0_3_8_REVIEW_ONLY.csv',
    'registry': RESULTS / 'basketball_drill_source_registry_v0_3_8.csv',
    'pending': RESULTS / 'basketball_wave3_remaining_requirement_reviews_v0_3_8.csv',
    'previous_summary': RESULTS / 'basketball_step12_61_summary_v0_3_8.json',
}
OUTPUT = {
    'decisions': RESULTS / 'basketball_wave3_plan2_field_decisions_v0_3_8.csv',
    'candidates': RESULTS / 'basketball_wave3_plan2_candidate_summary_v0_3_8.csv',
    'other_pending': RESULTS / 'basketball_wave3_other_fields_after_plan2_proposals_v0_3_8.csv',
    'source_snapshot': RESULTS / 'basketball_wave3_plan2_source_snapshot_v0_3_8.json',
    'summary': RESULTS / 'basketball_step12_62_summary_v0_3_8.json',
}
SOURCE_URL = 'https://jr.nba.com/basketball-practice-plans/all-star/page/2/'


def sha256(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def rel(path):
    return path.relative_to(ROOT).as_posix()


def require(condition, message):
    if not condition:
        raise ValueError(message)


def read_csv(path):
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    require(columns and len(columns) == len(set(columns)), f'Invalid CSV header: {path.name}')
    require(all(None not in row and all(v is not None for v in row.values()) for row in rows),
            f'Malformed CSV record: {path.name}')
    return rows, columns


def unique_index(rows, field, label):
    keys = [row[field] for row in rows]
    require(all(keys) and len(keys) == len(set(keys)), f'{label}: missing or repeated {field}.')
    return {row[field]: row for row in rows}


def normalize(text):
    return re.sub(r'[^a-z0-9]+', ' ', unicodedata.normalize('NFKC', text).casefold()).strip()


def fingerprint(text):
    return hashlib.sha256(normalize(text).encode('utf-8')).hexdigest()


def json_bytes(value):
    return (json.dumps(value, ensure_ascii=False, indent=2) + '\n').encode('utf-8')


def csv_bytes(rows, columns):
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=columns, lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')


for name, path in INPUT.items():
    if not path.is_file():
        raise FileNotFoundError(f'Missing {name}: {path}')
input_hashes = {rel(path): sha256(path) for path in INPUT.values()}
previous = json.loads(INPUT['previous_summary'].read_text(encoding='utf-8'))
require(previous.get('step') == '12-61' and previous.get('catalog_version') == '0.3.8',
        'Expected the verified Step 12-61 v0.3.8 summary.')
require(previous.get('pending_other_wave3_fields') == 68
        and previous.get('catalog_candidate_groups') == 144
        and previous.get('registered_source_occurrences') == 156,
        'The Step 12-61 review scope has changed.')
previous_hashes = {key.replace('\\', '/'): value
                   for key, value in previous.get('output_sha256', {}).items()}
for name in ('catalog', 'registry', 'pending'):
    path = INPUT[name]
    require(previous_hashes.get(rel(path)) == input_hashes[rel(path)],
            f'Input integrity mismatch: {rel(path)}. Preserve the file and inspect it.')

catalog, _ = read_csv(INPUT['catalog'])
registry, _ = read_csv(INPUT['registry'])
pending, pending_columns = read_csv(INPUT['pending'])
require((len(catalog), len(registry), len(pending)) == (144, 156, 68),
        'Unexpected catalog, source registry, or pending-review counts.')
cat_by = unique_index(catalog, 'catalog_candidate_id', 'Catalog')
reg_by = unique_index(registry, 'evidence_id', 'Registry')
pending_by = {(row['evidence_id'], row['field']): row for row in pending}
require(len(pending_by) == 68, 'Repeated evidence/field review records.')
require(all(row['catalog_version'] == '0.3.8'
            and row['catalog_approval_status'] == 'pending'
            and row['planner_eligible'] == 'no'
            and not row['approved_drill_id'] for row in catalog),
        'Catalog version or approval state differs from the expected review-only state.')

GATES = ('team_signoff', 'permission_review', 'youth_planning_review', 'catalog_approval_status')
for row in pending:
    eid, cid, field = row['evidence_id'], row['candidate_group_id'], row['field']
    require(field in {'equipment', 'participants'} and eid in reg_by and cid in cat_by,
            f'Unknown field or linkage: {eid}/{field}.')
    source, item = reg_by[eid], cat_by[cid]
    require(source['catalog_candidate_id'] == cid
            and eid in item['evidence_ids'].split(';')
            and source['drill_name'] == row['drill_name'] == item['drill_name']
            and source['source_url'].rstrip('/') == row['source_url'].rstrip('/')
            and source['source_level'] == row['source_level']
            and source['practice_plan'] == row['practice_plan'],
            f'Candidate or source linkage mismatch: {eid}/{field}.')
    require(row['current_catalog_value'] == source[field] == item[field],
            f'Current catalog/registry value mismatch: {eid}/{field}.')
    require(row['reviewer_decision'] == 'pending' and row['planner_eligible'] == 'no'
            and not row['final_value'] and not row['reviewer'] and not row['review_date']
            and all(row[gate] == 'pending' for gate in GATES),
            f'An existing review decision must be preserved: {eid}/{field}.')

# These recommendations concern source metadata only. No team, permission,
# youth-planning, distinctness, or final catalog approval is implied.
REVIEW = {
    ('JR-EVID-0034', 'equipment'): {
        'name': 'Progression Shooting Competition',
        'action': 'revise_value',
        'value': 'One basketball per team and access to a basket',
        'anchors': [r'each team should have 1 ball', r'at a basket'],
        'note': 'The setup assigns a ball to each team and uses baskets; the proposed value clarifies ball allocation.',
    },
    ('JR-EVID-0034', 'participants'): {
        'name': 'Progression Shooting Competition',
        'action': 'revise_value',
        'value': 'Teams in shooting lines; each shooter rebounds, passes to the next teammate, and rejoins the line',
        'anchors': [r'divide the players into teams', r'get the rebound',
                    r'next person in line', r'go to the back of the line'],
        'note': 'The activity uses team shooting lines with rebounding and passing rotations. The source does not specify a numeric team size.',
    },
    ('JR-EVID-0036', 'equipment'): {
        'name': 'Jump to the Ball Drill',
        'action': 'retain_current',
        'value': 'Basketball and hoop',
        'anchors': [r'coach at the top with the ball', r'lay up', r'under the basket'],
        'note': 'Passing, dribbling, and finishing at the basket support retaining the existing equipment description.',
    },
    ('JR-EVID-0036', 'participants'): {
        'name': 'Jump to the Ball Drill',
        'action': 'revise_value',
        'value': 'Two wing players, one rotating defender from the baseline line, and a coach passing from the top',
        'anchors': [r'1 player on each wing', r'coach at the top with the ball',
                    r'first player in line on the baseline', r'fill the open space on the wing'],
        'note': 'The source describes two wing positions, a defender rotating from the baseline line, and a coach initiating the pass; queue length is not fixed.',
    },
}
selected_keys = {key for key, row in pending_by.items()
                 if row['source_url'].rstrip('/') == SOURCE_URL.rstrip('/')}
require(selected_keys == set(REVIEW), 'All-Star Plan 2 no longer contains exactly the four expected fields.')
for key, recommendation in REVIEW.items():
    row = pending_by[key]
    require(row['drill_name'] == recommendation['name']
            and row['source_level'] == 'All-Star' and row['practice_plan'] == '2',
            f'Target drill identity changed: {key}.')
    require(cat_by[row['candidate_group_id']]['evidence_ids'].split(';') == [row['evidence_id']],
            'A target now has multiple sources; compare the source variations before proceeding.')


class HeadingSections(HTMLParser):
    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.sections = []
        self.visible = []
        self.skipping = []
        self.heading_tag = None
        self.heading_parts = []
        self.current = None

    def handle_starttag(self, tag, attrs):
        if tag in {'script', 'style', 'noscript'}:
            self.skipping.append(tag)
            return
        if self.skipping:
            return
        if re.fullmatch(r'h[1-6]', tag):
            self.heading_tag = tag
            self.heading_parts = []
        elif tag in {'p', 'li', 'br', 'div'} and self.current is not None:
            self.current['parts'].append(' ')

    def handle_data(self, data):
        if self.skipping:
            return
        self.visible.append(data)
        if self.heading_tag:
            self.heading_parts.append(data)
        elif self.current is not None:
            self.current['parts'].append(data)

    def handle_endtag(self, tag):
        if self.skipping:
            if tag == self.skipping[-1]:
                self.skipping.pop()
            return
        if tag == self.heading_tag:
            heading = ' '.join(self.heading_parts).strip()
            if heading:
                self.current = {'heading': heading, 'parts': []}
                self.sections.append(self.current)
            self.heading_tag = None
            self.heading_parts = []
        elif tag in {'p', 'li', 'br', 'div'} and self.current is not None:
            self.current['parts'].append(' ')


def build_proposals(snapshot):
    require(snapshot.get('source_url') == SOURCE_URL
            and snapshot.get('practice_plan') == 'All-Star Practice Plan 2',
            'Source snapshot identity is incorrect.')
    by_name = unique_index(snapshot['sections'], 'heading', 'Source snapshot')
    require(set(by_name) == {r['name'] for r in REVIEW.values()},
            'Source snapshot has missing or unexpected drill sections.')
    decisions = []
    for key, recommendation in REVIEW.items():
        row = pending_by[key]
        section = by_name[row['drill_name']]
        text = section['text']
        require(fingerprint(text) == section['normalized_section_sha256'],
                'Saved source-section fingerprint mismatch.')
        checks = [bool(re.search(pattern, normalize(text))) for pattern in recommendation['anchors']]
        supported = all(checks)
        action = recommendation['action'] if supported else 'defer_source_evidence'
        value = recommendation['value'] if supported else ''
        if supported and action == 'retain_current':
            require(value == row['current_catalog_value'], 'Retain recommendation would change a value.')
        if supported and action == 'revise_value':
            require(value != row['current_catalog_value'], 'Revision recommendation is redundant.')
        decisions.append({
            'evidence_id': row['evidence_id'],
            'candidate_group_id': row['candidate_group_id'],
            'drill_name': row['drill_name'],
            'field': row['field'],
            'current_value': row['current_catalog_value'],
            'proposed_action': action,
            'proposed_value': value,
            'source_evidence_note': recommendation['note'] if supported else 'Expected source details were not found; manual source review is needed.',
            'heading_found': 'yes',
            'source_anchor_checks': f'{sum(checks)}/{len(checks)}',
            'section_fingerprint_sha256': section['normalized_section_sha256'],
            'source_snapshot_reference': rel(OUTPUT['source_snapshot']),
            'prior_ledger_section_sha256': row['source_section_sha256'],
            'source_url': SOURCE_URL,
            'reviewer_decision': 'pending',
            'reviewer': '',
            'review_date': '',
            'team_signoff': 'pending',
            'permission_review': 'pending',
            'youth_planning_review': 'pending',
            'catalog_approval_status': 'pending',
            'planner_eligible': 'no',
        })
    return decisions


print('INPUT INTEGRITY: PASS - v0.3.8 files match the Step 12-61 summary.')
existing = [path.exists() for path in OUTPUT.values()]
if any(existing) and not all(existing):
    raise FileExistsError('Only part of the Step 12-62 output set exists. Inspect it before rerunning.')

if all(existing):
    # Reruns validate and preserve the original evidence without another download.
    summary = json.loads(OUTPUT['summary'].read_text(encoding='utf-8'))
    require(summary.get('step') == '12-62' and summary.get('catalog_version') == '0.3.8'
            and summary.get('input_sha256') == input_hashes,
            'Existing Step 12-62 files came from different inputs or a different step.')
    output_hashes = summary.get('output_sha256', {})
    for name, path in OUTPUT.items():
        if name != 'summary':
            require(output_hashes.get(rel(path)) == sha256(path),
                    f'Saved output changed: {rel(path)}. It has been preserved.')
    snapshot = json.loads(OUTPUT['source_snapshot'].read_text(encoding='utf-8'))
    mode = 'PRESERVED'
else:
    request = Request(SOURCE_URL, headers={'User-Agent': 'Mozilla/5.0'})
    with urlopen(request, timeout=30) as response:
        final_url = response.geturl()
        url = urlsplit(final_url)
        require(response.status == 200 and url.scheme == 'https' and url.hostname == 'jr.nba.com'
                and url.path.rstrip('/') == urlsplit(SOURCE_URL).path.rstrip('/'),
                'The official source did not return the expected practice-plan page.')
        raw_html = response.read(3_000_001)
        require(len(raw_html) <= 3_000_000, 'Source page exceeded the supported download size.')
        html = raw_html.decode(response.headers.get_content_charset() or 'utf-8', errors='replace')
    parser = HeadingSections()
    parser.feed(html)
    parser.close()
    require('all star practice plan 2' in normalize(' '.join(parser.visible)),
            'Practice-plan identity was not found on the source page.')
    by_heading = defaultdict(list)
    for position, section in enumerate(parser.sections):
        by_heading[normalize(section['heading'])].append((position, section))
    sections = []
    for name in sorted({r['name'] for r in REVIEW.values()}):
        matches = by_heading[normalize(name)]
        require(len(matches) == 1, f'Missing or ambiguous source heading: {name}. No files created.')
        position, section = matches[0]
        text = re.sub(r'\s+', ' ', ' '.join(section['parts'])).strip()
        require(text, f'Empty source drill section: {name}.')
        sections.append({
            'heading': name,
            'publisher_heading': section['heading'],
            'next_heading': parser.sections[position + 1]['heading'] if position + 1 < len(parser.sections) else '',
            'text': text,
            'normalized_section_sha256': fingerprint(text),
        })
    snapshot = {
        'source_url': SOURCE_URL,
        'resolved_url': final_url,
        'practice_plan': 'All-Star Practice Plan 2',
        'retrieved_at_utc': datetime.now(timezone.utc).isoformat(),
        'page_sha256': hashlib.sha256(raw_html).hexdigest(),
        'section_boundary_rule': 'Text after a drill heading up to the next HTML heading.',
        'fingerprint_method': 'NFKC, casefold, replace non-ASCII-alphanumeric runs with spaces, UTF-8 SHA-256.',
        'sections': sections,
    }
    mode = 'CREATED'

decisions = build_proposals(snapshot)
other_pending = [row for row in pending if (row['evidence_id'], row['field']) not in REVIEW]
candidates = []
for eid in sorted({key[0] for key in REVIEW}):
    rows = [row for row in decisions if row['evidence_id'] == eid]
    counts = Counter(row['proposed_action'] for row in rows)
    candidates.append({
        'candidate_group_id': rows[0]['candidate_group_id'],
        'evidence_id': eid,
        'drill_name': rows[0]['drill_name'],
        'fields_with_proposals': len(rows),
        'retain_recommendations': counts['retain_current'],
        'revision_recommendations': counts['revise_value'],
        'source_evidence_gaps': counts['defer_source_evidence'],
        'source_url': SOURCE_URL,
        'reviewer_decision': 'pending',
        'team_signoff': 'pending',
        'permission_review': 'pending',
        'youth_planning_review': 'pending',
        'catalog_approval_status': 'pending',
        'planner_eligible': 'no',
    })
require((len(decisions), len(candidates), len(other_pending)) == (4, 2, 64),
        'The proposed batch does not partition the original pending reviews correctly.')
require({(r['evidence_id'], r['field']) for r in other_pending} | set(REVIEW) == set(pending_by),
        'The review partition lost or introduced a field.')
proposed_counts = Counter(row['proposed_action'] for row in decisions)
expected_summary = {
    'step': '12-62',
    'catalog_version': '0.3.8',
    'catalog_candidate_groups': 144,
    'registered_source_occurrences': 156,
    'source_plan': 'All-Star Practice Plan 2',
    'source_page': SOURCE_URL,
    'initial_pending_field_reviews': 68,
    'selected_candidate_groups': 2,
    'selected_field_recommendations': 4,
    'proposed_retain': proposed_counts['retain_current'],
    'proposed_revise': proposed_counts['revise_value'],
    'source_anchor_review_needed': proposed_counts['defer_source_evidence'],
    'selected_fields_awaiting_user_review': 4,
    'other_fields_without_this_batch_proposals': 64,
    'total_fields_still_pending_confirmation': 68,
    'reviewer_signoffs': 0,
    'catalog_approvals': 0,
    'planner_eligible': 0,
    'original_inputs_modified': False,
    'git_operations_performed': False,
    'input_sha256': input_hashes,
}
file_bytes = {
    'decisions': csv_bytes(decisions, list(decisions[0])),
    'candidates': csv_bytes(candidates, list(candidates[0])),
    'other_pending': csv_bytes(other_pending, pending_columns),
    'source_snapshot': json_bytes(snapshot),
}
if mode == 'CREATED':
    summary = dict(expected_summary)
    summary['output_sha256'] = {rel(OUTPUT[name]): hashlib.sha256(data).hexdigest()
                                for name, data in file_bytes.items()}
    file_bytes['summary'] = json_bytes(summary)
    # Validate and stage the complete batch before creating final destinations.
    with TemporaryDirectory(prefix='step12_62_', dir=ROOT) as staging_dir:
        staged = {}
        for name, data in file_bytes.items():
            staged[name] = Path(staging_dir) / OUTPUT[name].name
            staged[name].write_bytes(data)
            require(staged[name].read_bytes() == data, f'Staging integrity failed: {name}.')
        require(all(sha256(path) == input_hashes[rel(path)] for path in INPUT.values()),
                'An input changed during source review. No outputs created.')
        if any(path.exists() for path in OUTPUT.values()):
            raise FileExistsError('A destination appeared during staging; preserve and inspect it.')
        created = []
        try:
            for name, path in OUTPUT.items():
                path.parent.mkdir(parents=True, exist_ok=True)
                with path.open('xb') as handle:
                    created.append(path)
                    handle.write(staged[name].read_bytes())
        except Exception:
            for path in created:
                path.unlink(missing_ok=True)
            raise
else:
    require(all(summary.get(key) == value for key, value in expected_summary.items()),
            'Saved summary differs from the verified review state.')
    require(all(OUTPUT[name].read_bytes() == data for name, data in file_bytes.items()),
            'Saved proposals differ from the current recommendation specification; preserved.')

require(all(sha256(path) == input_hashes[rel(path)] for path in INPUT.values()),
        'An original input changed during execution.')
for name, path in OUTPUT.items():
    if name != 'summary':
        require(summary['output_sha256'].get(rel(path)) == sha256(path),
                f'Saved output verification failed: {name}.')

print('=' * 72)
print('STEP 12-62 - ALL-STAR PLAN 2 REQUIREMENT REVIEW')
print('=' * 72)
print('Catalog candidate groups / source occurrences: 144 / 156')
print('Selected drills / proposed field decisions: 2 / 4')
print('RECOMMENDATIONS - PENDING YOUR REVIEW')
for row in decisions:
    print(f"  {row['evidence_id']} | {row['drill_name']} | {row['field']} | {row['proposed_action']} | anchors={row['source_anchor_checks']}")
    print(f"    Current:  {row['current_value']}")
    print(f"    Proposed: {row['proposed_value'] or '(deferred - inspect source evidence)'}")
print(f"Source evidence gaps: {summary['source_anchor_review_needed']}")
print('Fields awaiting confirmation: 68 (4 in this batch + 64 other fields)')
print('Reviewer signoffs / approved / planner-eligible: 0 / 0 / 0')
print('OUTPUTS')
for name, path in OUTPUT.items():
    print(f'  {mode:<9} {rel(path)}')
print('Catalog, source registry, and original pending ledger modified: No')
print('Git operations performed: None')
if summary['source_anchor_review_needed']:
    print('OVERALL: REVIEW NEEDED - evidence gaps recorded; approval gates preserved')
else:
    print('OVERALL: PASS - four recommendations prepared; confirmation remains pending')
print('=' * 72)
```

```python
# Step 12-63 - Consolidated Wave 3 Equipment and Participant Review
# Cover all 68 pending fields in one review packet. No decisions are auto-approved.

import csv
import hashlib
import io
import json
import re
import unicodedata
from collections import Counter, defaultdict
from concurrent.futures import ThreadPoolExecutor
from datetime import datetime, timezone
from html.parser import HTMLParser
from pathlib import Path
from tempfile import TemporaryDirectory
from urllib.parse import urlsplit
from urllib.request import Request, urlopen

ROOT = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Run this cell inside the Champions Athletic Map Git project.')
RESULTS = ROOT / 'results'
INPUT = {
    'catalog': ROOT / 'data/processed/basketball_drill_catalog_v0_3_8_REVIEW_ONLY.csv',
    'registry': RESULTS / 'basketball_drill_source_registry_v0_3_8.csv',
    'pending': RESULTS / 'basketball_wave3_remaining_requirement_reviews_v0_3_8.csv',
    'summary61': RESULTS / 'basketball_step12_61_summary_v0_3_8.json',
    'summary62': RESULTS / 'basketball_step12_62_summary_v0_3_8.json',
    'plan2_decisions': RESULTS / 'basketball_wave3_plan2_field_decisions_v0_3_8.csv',
    'plan2_candidates': RESULTS / 'basketball_wave3_plan2_candidate_summary_v0_3_8.csv',
    'plan2_other': RESULTS / 'basketball_wave3_other_fields_after_plan2_proposals_v0_3_8.csv',
    'plan2_snapshot': RESULTS / 'basketball_wave3_plan2_source_snapshot_v0_3_8.json',
}
OUTPUT = {
    'recommendations': RESULTS / 'basketball_wave3_consolidated_field_review_v0_3_8.csv',
    'candidates': RESULTS / 'basketball_wave3_consolidated_candidate_review_v0_3_8.csv',
    'exceptions': RESULTS / 'basketball_wave3_consolidated_source_exceptions_v0_3_8.csv',
    'snapshot': RESULTS / 'basketball_wave3_consolidated_source_snapshot_v0_3_8.json',
    'summary': RESULTS / 'basketball_step12_63_summary_v0_3_8.json',
}

# Reviewed factual proposals and per-drill fingerprints; contains no final signoffs.
REVIEW_SPEC = [{'evidence_id': 'JR-EVID-0034',
  'field': 'equipment',
  'drill_name': 'Progression Shooting Competition',
  'proposed_action': 'revise_value',
  'proposed_value': 'One basketball per team and access to a basket',
  'source_evidence_note': 'The setup assigns a ball to each team and uses baskets; the proposed value '
                          'clarifies ball allocation.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/all-star/page/2/',
  'expected_section_sha256': '8e359003e8dc3972f54f290c8e598bff6e26eb8c9a96e698b4b0a164bdb4eb42',
  'proposal_origin': 'step12_62_carried_forward'},
 {'evidence_id': 'JR-EVID-0034',
  'field': 'participants',
  'drill_name': 'Progression Shooting Competition',
  'proposed_action': 'revise_value',
  'proposed_value': 'Teams in shooting lines; each shooter rebounds, passes to the next teammate, and '
                    'rejoins the line',
  'source_evidence_note': 'The activity uses team shooting lines with rebounding and passing rotations. '
                          'The source does not specify a numeric team size.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/all-star/page/2/',
  'expected_section_sha256': '8e359003e8dc3972f54f290c8e598bff6e26eb8c9a96e698b4b0a164bdb4eb42',
  'proposal_origin': 'step12_62_carried_forward'},
 {'evidence_id': 'JR-EVID-0036',
  'field': 'equipment',
  'drill_name': 'Jump to the Ball Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball and hoop',
  'source_evidence_note': 'Passing, dribbling, and finishing at the basket support retaining the '
                          'existing equipment description.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/all-star/page/2/',
  'expected_section_sha256': 'cb5768ea3182c37e6736d9f5c04737ea9ae8da8e06a0ec2c0a37a6ae63fae1ea',
  'proposal_origin': 'step12_62_carried_forward'},
 {'evidence_id': 'JR-EVID-0036',
  'field': 'participants',
  'drill_name': 'Jump to the Ball Drill',
  'proposed_action': 'revise_value',
  'proposed_value': 'Two wing players, one rotating defender from the baseline line, and a coach '
                    'passing from the top',
  'source_evidence_note': 'The source describes two wing positions, a defender rotating from the '
                          'baseline line, and a coach initiating the pass; queue length is not fixed.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/all-star/page/2/',
  'expected_section_sha256': 'cb5768ea3182c37e6736d9f5c04737ea9ae8da8e06a0ec2c0a37a6ae63fae1ea',
  'proposal_origin': 'step12_62_carried_forward'},
 {'evidence_id': 'JR-EVID-0055',
  'field': 'equipment',
  'drill_name': '1 on 1 Dribble Battle',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball per player',
  'source_evidence_note': 'The published base setup supplies a ball to every player.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/all-star/page/3/',
  'expected_section_sha256': '11a2f8025cc82a72161433938b226222d9c106bd03322b790ab005ecb3463f54',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0055',
  'field': 'participants',
  'drill_name': '1 on 1 Dribble Battle',
  'proposed_action': 'retain_current',
  'proposed_value': 'Pairs',
  'source_evidence_note': 'Pairs are the base setup; a three-player shielding variation is separate.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/all-star/page/3/',
  'expected_section_sha256': '11a2f8025cc82a72161433938b226222d9c106bd03322b790ab005ecb3463f54',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0057',
  'field': 'equipment',
  'drill_name': 'Pound Pass 1 on 1',
  'proposed_action': 'revise_value',
  'proposed_value': 'One basketball per pair and access to a hoop',
  'source_evidence_note': 'Live scoring requires a hoop; the two-ball setup is an optional variation.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/all-star/page/3/',
  'expected_section_sha256': '6e25262c3cb211259b7094889c8ac63501d547ec3ec951c5af77929d5e940bc7',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0057',
  'field': 'participants',
  'drill_name': 'Pound Pass 1 on 1',
  'proposed_action': 'retain_current',
  'proposed_value': 'Pairs',
  'source_evidence_note': 'The base setup uses attacking and defending partners who exchange roles.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/all-star/page/3/',
  'expected_section_sha256': '6e25262c3cb211259b7094889c8ac63501d547ec3ec951c5af77929d5e940bc7',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0016',
  'field': 'equipment',
  'drill_name': 'Pound Pass Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'One basketball per pair',
  'source_evidence_note': 'The published partners share one ball.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/',
  'expected_section_sha256': '77f7cab74fa1cce796816c893290ec68d892da49ad1594dd12be5ba52e18b07d',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0016',
  'field': 'participants',
  'drill_name': 'Pound Pass Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Two players per pair',
  'source_evidence_note': 'The drill describes two partners exchanging passes.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/',
  'expected_section_sha256': '77f7cab74fa1cce796816c893290ec68d892da49ad1594dd12be5ba52e18b07d',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0017',
  'field': 'equipment',
  'drill_name': 'Slip Pass Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball and hoop',
  'source_evidence_note': 'Passes and finishes use a ball and basket; no chair is prescribed.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/',
  'expected_section_sha256': '25f285d1e38b9b67fad6c67b5524aa59b668cc66c2c561aba329f658a722ad79',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0017',
  'field': 'participants',
  'drill_name': 'Slip Pass Drill',
  'proposed_action': 'revise_value',
  'proposed_value': 'Passing, screening, and shooting lines, plus a coach or player defending the '
                    'screen',
  'source_evidence_note': 'Players rotate across three roles; a player may replace the coach as '
                          'defender.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/',
  'expected_section_sha256': '25f285d1e38b9b67fad6c67b5524aa59b668cc66c2c561aba329f658a722ad79',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0039',
  'field': 'equipment',
  'drill_name': 'Catch a Second Ball Series',
  'proposed_action': 'retain_current',
  'proposed_value': 'Two basketballs per pair',
  'source_evidence_note': 'Tennis-ball substitution is an optional variation.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': '34f30e86f97febba8fe016ad2bf246d43471dcb3bd464e531047dd1b4c7118f0',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0039',
  'field': 'participants',
  'drill_name': 'Catch a Second Ball Series',
  'proposed_action': 'retain_current',
  'proposed_value': 'Two players per pair',
  'source_evidence_note': 'Partner roles alternate.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': '34f30e86f97febba8fe016ad2bf246d43471dcb3bd464e531047dd1b4c7118f0',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0040',
  'field': 'equipment',
  'drill_name': 'Rhythm Dribbling',
  'proposed_action': 'revise_value',
  'proposed_value': 'One basketball per player',
  'source_evidence_note': 'Each dribbler has a ball.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': '5568746cf192627c3498f47d1c824152af77ec34c2c45ada20d8b51da40cdeb6',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0041',
  'field': 'equipment',
  'drill_name': 'Hook Pass Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball, hoop, cone or chair',
  'source_evidence_note': 'Screen marker can be a cone or chair.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': '4e095034620ddbe38bfb47fc01a18f3570f21329a605b8a9e8eac23e77732722',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0041',
  'field': 'participants',
  'drill_name': 'Hook Pass Drill',
  'proposed_action': 'revise_value',
  'proposed_value': 'Players in a wing line and a coach receiving and returning passes',
  'source_evidence_note': 'Alternative receiver role is not established.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': '4e095034620ddbe38bfb47fc01a18f3570f21329a605b8a9e8eac23e77732722',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0042',
  'field': 'equipment',
  'drill_name': 'Floater Series',
  'proposed_action': 'revise_value',
  'proposed_value': 'One basketball per group and a basket',
  'source_evidence_note': 'Published setup specifies shared ball allocation.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': 'd564c02b749c6983ce3ab13245540f0883c048e10198cd2ff6ea6b20b2df5e3d',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0042',
  'field': 'participants',
  'drill_name': 'Floater Series',
  'proposed_action': 'retain_current',
  'proposed_value': 'Small groups',
  'source_evidence_note': 'Players share a group station.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': 'd564c02b749c6983ce3ab13245540f0883c048e10198cd2ff6ea6b20b2df5e3d',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0043',
  'field': 'equipment',
  'drill_name': 'Pick & Pop Shooting',
  'proposed_action': 'revise_value',
  'proposed_value': 'One basketball per player and a hoop',
  'source_evidence_note': 'Each player starts with a ball.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': 'a55b18c7587be4d2bff4f6bb1b58626bf9d4207dcf86c4d69302aa0954eae839',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0043',
  'field': 'participants',
  'drill_name': 'Pick & Pop Shooting',
  'proposed_action': 'retain_current',
  'proposed_value': 'Guards, post players, and coach',
  'source_evidence_note': 'Guard and post lines rotate.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': 'a55b18c7587be4d2bff4f6bb1b58626bf9d4207dcf86c4d69302aa0954eae839',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0044',
  'field': 'equipment',
  'drill_name': 'Rebound & Score',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball and hoop',
  'source_evidence_note': 'Rebounding and scoring use the basket.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': 'f65b3879ba2233a568cb8a342fdee2b9cac79b15311a4e03eb9123e94019ffc3',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0044',
  'field': 'participants',
  'drill_name': 'Rebound & Score',
  'proposed_action': 'revise_value',
  'proposed_value': 'Three active players, a coach initiating the shot, and a baseline rotation line',
  'source_evidence_note': 'Coach starts the live rebounding contest.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': 'f65b3879ba2233a568cb8a342fdee2b9cac79b15311a4e03eb9123e94019ffc3',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0045',
  'field': 'equipment',
  'drill_name': '1 on 1 – Stunt & Recover',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball and hoop',
  'source_evidence_note': 'Live possession follows the coach’s pass.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': '1a7e292dbe36b353c16b768b5a1a9a86c4a55976ee7cc2e1fccb24d5955e224e',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0045',
  'field': 'participants',
  'drill_name': '1 on 1 – Stunt & Recover',
  'proposed_action': 'retain_current',
  'proposed_value': 'Players and coach',
  'source_evidence_note': 'Coach initiates; offense and defense rotate.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': '1a7e292dbe36b353c16b768b5a1a9a86c4a55976ee7cc2e1fccb24d5955e224e',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0046',
  'field': 'equipment',
  'drill_name': '4 on 4 Shell to Rebound',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball and hoop',
  'source_evidence_note': 'Possession continues through a defensive rebound.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': 'e67c55d9b14afdaae0e0c21910c89d2df9b9b69835666d4dabc120b2df9a9a75',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0046',
  'field': 'participants',
  'drill_name': '4 on 4 Shell to Rebound',
  'proposed_action': 'retain_current',
  'proposed_value': 'Two teams of four',
  'source_evidence_note': 'Four attackers oppose four defenders.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/2/',
  'expected_section_sha256': 'e67c55d9b14afdaae0e0c21910c89d2df9b9b69835666d4dabc120b2df9a9a75',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0063',
  'field': 'equipment',
  'drill_name': '2-Ball Jab Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Two basketballs per active player or shared',
  'source_evidence_note': 'Partner sharing is explicitly supported.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': '73453df78a945837180026430a96b8edccbafc00563c87db233f5b14d6f44139',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0063',
  'field': 'participants',
  'drill_name': '2-Ball Jab Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Coach-directed players',
  'source_evidence_note': 'Players respond to coaching cues.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': '73453df78a945837180026430a96b8edccbafc00563c87db233f5b14d6f44139',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0064',
  'field': 'equipment',
  'drill_name': '2 Ball Alternate & React',
  'proposed_action': 'retain_current',
  'proposed_value': 'Two basketballs per active player or shared',
  'source_evidence_note': 'Partner sharing is explicitly supported.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': 'b073150614dea27b61b20910f2bebc944fc6831a6132f9b877564d3390cfd34a',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0064',
  'field': 'participants',
  'drill_name': '2 Ball Alternate & React',
  'proposed_action': 'retain_current',
  'proposed_value': 'Coach-directed players',
  'source_evidence_note': 'Players respond to coaching cues.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': 'b073150614dea27b61b20910f2bebc944fc6831a6132f9b877564d3390cfd34a',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0065',
  'field': 'equipment',
  'drill_name': 'Loop Shooting Drill',
  'proposed_action': 'revise_value',
  'proposed_value': 'Two basketballs per line and a basket',
  'source_evidence_note': 'Each of two lines receives two balls.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': 'efd2d39932677ca638f2a8764e242a0c20bcd06658902671120af20b466f82cf',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0065',
  'field': 'participants',
  'drill_name': 'Loop Shooting Drill',
  'proposed_action': 'revise_value',
  'proposed_value': 'Two passing lines and one initial shooter; players rotate',
  'source_evidence_note': 'The setup specifies both lines and an initial on-court shooter.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': 'efd2d39932677ca638f2a8764e242a0c20bcd06658902671120af20b466f82cf',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0066',
  'field': 'equipment',
  'drill_name': 'Filling the Corner Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball and hoop',
  'source_evidence_note': 'Passing and shooting use the listed resources.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': '115388c6a2e787b8515f35f3a2b966029628c15664eec32f97de02418bfb27f1',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0066',
  'field': 'participants',
  'drill_name': 'Filling the Corner Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Two lines of players',
  'source_evidence_note': 'The published setup uses two wing lines.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': '115388c6a2e787b8515f35f3a2b966029628c15664eec32f97de02418bfb27f1',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0067',
  'field': 'equipment',
  'drill_name': 'Spin 5 Series',
  'proposed_action': 'revise_value',
  'proposed_value': 'One basketball per player and a basket',
  'source_evidence_note': 'Individual balls support the passing and finishing sequence.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': 'e187d781604bd135b2e514474330554ebaceaaf818bb31061f3902b3ba8c4b81',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0067',
  'field': 'participants',
  'drill_name': 'Spin 5 Series',
  'proposed_action': 'retain_current',
  'proposed_value': 'Players and coach',
  'source_evidence_note': 'Players pass to the coach before finishing.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': 'e187d781604bd135b2e514474330554ebaceaaf818bb31061f3902b3ba8c4b81',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0068',
  'field': 'equipment',
  'drill_name': 'Pocket Pass & Shoot',
  'proposed_action': 'revise_value',
  'proposed_value': 'One basketball per player and a basket',
  'source_evidence_note': 'Source assigns each player a ball.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': 'dc51b0d7c568dafce2553e3bae2d719adf6aae5f8439acecd24f9ae9191fbaf3',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0068',
  'field': 'participants',
  'drill_name': 'Pocket Pass & Shoot',
  'proposed_action': 'retain_current',
  'proposed_value': 'Guards, post players, coach',
  'source_evidence_note': 'These roles match the published setup.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': 'dc51b0d7c568dafce2553e3bae2d719adf6aae5f8439acecd24f9ae9191fbaf3',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0070',
  'field': 'participants',
  'drill_name': 'Balance the Floor Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Players and coach',
  'source_evidence_note': 'The coach directs players from five starting lines.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/mvp/page/3/',
  'expected_section_sha256': '1c1e23647de1c5fb6d44eebe749ade65910b16910ab19265f0315c94bce52b17',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0007',
  'field': 'equipment',
  'drill_name': 'Hot Ball (Taps)',
  'proposed_action': 'revise_value',
  'proposed_value': 'Basketball per active player; sharing permitted',
  'source_evidence_note': 'Ball sharing is explicitly permitted.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/',
  'expected_section_sha256': '26a29f9466a84ae94f0f34872521c54ddd309526bf70031a2d0553267f595d5f',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0008',
  'field': 'equipment',
  'drill_name': 'Around The World Wraps',
  'proposed_action': 'revise_value',
  'proposed_value': 'Basketball per active player; sharing permitted',
  'source_evidence_note': 'Base activity explicitly permits shared basketballs.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/',
  'expected_section_sha256': '17a5be1ce1c06c01baa8223c84ed1e9b0482c01733b61068ccc17a0fc51dbe3a',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0023',
  'field': 'equipment',
  'drill_name': 'Toss & Catch',
  'proposed_action': 'revise_value',
  'proposed_value': 'Basketball per active player; sharing permitted',
  'source_evidence_note': 'Base instructions permit sharing; partner interaction is a variation.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/page/2/',
  'expected_section_sha256': '58b545b4baa3a79e57d63fc4c07e00754de13dd1311448840819e378c19b9c1b',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0024',
  'field': 'equipment',
  'drill_name': 'Stationary Dribble Low/Middle',
  'proposed_action': 'revise_value',
  'proposed_value': 'Basketball per active player; sharing permitted in base activity',
  'source_evidence_note': 'Base sharing is allowed; competitive partners each use a ball.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/page/2/',
  'expected_section_sha256': 'b4317b598cef6bf232b8edd2547da5f82672df035e1111b05e309798de56b9c0',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0025',
  'field': 'equipment',
  'drill_name': 'Up or Down Game',
  'proposed_action': 'revise_value',
  'proposed_value': 'One basketball per player and cones',
  'source_evidence_note': 'Every player dribbles while changing cone orientation.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/page/2/',
  'expected_section_sha256': '23ddc564c4e704483f8df602f65a7705b8a7cef2e49076790c40f46b5e48dc9e',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0025',
  'field': 'participants',
  'drill_name': 'Up or Down Game',
  'proposed_action': 'retain_current',
  'proposed_value': 'Two teams',
  'source_evidence_note': 'The base activity divides players into two teams.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/page/2/',
  'expected_section_sha256': '23ddc564c4e704483f8df602f65a7705b8a7cef2e49076790c40f46b5e48dc9e',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0027',
  'field': 'equipment',
  'drill_name': 'Gates Passing Game',
  'proposed_action': 'revise_value',
  'proposed_value': 'One basketball per pair; paired cones or other gate markers',
  'source_evidence_note': 'Partner units share a ball; gates may use alternative objects.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/page/2/',
  'expected_section_sha256': '3b932945f416c6dbd85b0deeb3a3acebcd84b6300034e595466e0f53f10b5132',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0027',
  'field': 'participants',
  'drill_name': 'Gates Passing Game',
  'proposed_action': 'retain_current',
  'proposed_value': 'Partners',
  'source_evidence_note': 'Each passing unit consists of partners.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/page/2/',
  'expected_section_sha256': '3b932945f416c6dbd85b0deeb3a3acebcd84b6300034e595466e0f53f10b5132',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0028',
  'field': 'equipment',
  'drill_name': 'Red Light',
  'proposed_action': 'revise_value',
  'proposed_value': 'One basketball per player',
  'source_evidence_note': 'Each player dribbles an individual ball.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/page/2/',
  'expected_section_sha256': '43783663e5d5cd4469cb3b9afe10025f799143db3ef67b391f16dedd49bf4b67',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0028',
  'field': 'participants',
  'drill_name': 'Red Light',
  'proposed_action': 'retain_current',
  'proposed_value': 'Coach-directed group',
  'source_evidence_note': "The base game follows a coach's movement signals.",
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/page/2/',
  'expected_section_sha256': '43783663e5d5cd4469cb3b9afe10025f799143db3ef67b391f16dedd49bf4b67',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0029',
  'field': 'participants',
  'drill_name': 'Shooting Line Game',
  'proposed_action': 'revise_value',
  'proposed_value': 'Two or more team lines; players rotate through shooting and rebounding',
  'source_evidence_note': 'Team size is unspecified; shooting, rebounding, and rejoining are described.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/page/2/',
  'expected_section_sha256': '9fdffb2b05ff5df31c617c1a95579858a6af365600019d65de0711b96263cdb4',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0030',
  'field': 'equipment',
  'drill_name': 'Advantage Numbers Game',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball and hoop',
  'source_evidence_note': 'The coach presents one ball for possession play toward the hoop.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/page/2/',
  'expected_section_sha256': '4769abb6f1375916a201dc0136de1ab1ffaf7caf78d40c963e98ef31b01af504',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0030',
  'field': 'participants',
  'drill_name': 'Advantage Numbers Game',
  'proposed_action': 'retain_current',
  'proposed_value': 'Two teams and coach',
  'source_evidence_note': 'Two opposing teams and a coach coordinate numbered matchups.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/rookie/page/2/',
  'expected_section_sha256': '4769abb6f1375916a201dc0136de1ab1ffaf7caf78d40c963e98ef31b01af504',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0047',
  'field': 'equipment',
  'drill_name': 'Jail Break',
  'proposed_action': 'revise_value',
  'proposed_value': 'One basketball per dribbler',
  'source_evidence_note': 'Dribblers operate simultaneously.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': '86a118068fed6e66f66c3f8be129a45a11a25a96d6d6154f2bf96202777aea23',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0047',
  'field': 'participants',
  'drill_name': 'Jail Break',
  'proposed_action': 'revise_value',
  'proposed_value': 'Two teams: dribblers and defenders; escaping dribblers become defenders',
  'source_evidence_note': 'Roles change during play.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': '86a118068fed6e66f66c3f8be129a45a11a25a96d6d6154f2bf96202777aea23',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0048',
  'field': 'equipment',
  'drill_name': 'Tuck Back Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball optional for base movement',
  'source_evidence_note': 'Source permits miming a ball.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': '4f7250e78cfaecc5624e181e6e5878cfc0a8ef92004c8f5d148cfcf4e6e79189',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0048',
  'field': 'participants',
  'drill_name': 'Tuck Back Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Coach-directed players',
  'source_evidence_note': 'Coach cues the movement.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': '4f7250e78cfaecc5624e181e6e5878cfc0a8ef92004c8f5d148cfcf4e6e79189',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0049',
  'field': 'equipment',
  'drill_name': 'Rooted 1 on 1',
  'proposed_action': 'revise_value',
  'proposed_value': 'Basketballs for the first attacker in each of three lines, plus a hoop',
  'source_evidence_note': 'Three simultaneous starting balls.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': '7f3e11d76fde0f361da2b6866e1ce6ad8de1845af666701bc0688038dccc10d3',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0049',
  'field': 'participants',
  'drill_name': 'Rooted 1 on 1',
  'proposed_action': 'revise_value',
  'proposed_value': 'One attacker and one defender per line, using three lines; optional second '
                    'defender',
  'source_evidence_note': 'Base and variation separated.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': '7f3e11d76fde0f361da2b6866e1ce6ad8de1845af666701bc0688038dccc10d3',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0050',
  'field': 'equipment',
  'drill_name': 'The Kick Pass Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball and hoop',
  'source_evidence_note': 'Drives finish at the basket.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': 'ed97a8a4c225ccea1bfaaad11a74783d5e2634547260fd44fb7e46dd47085d58',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0050',
  'field': 'participants',
  'drill_name': 'The Kick Pass Drill',
  'proposed_action': 'revise_value',
  'proposed_value': 'Two attackers and two defenders; optional three-on-three variation',
  'source_evidence_note': 'Offense rotates to defense.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': 'ed97a8a4c225ccea1bfaaad11a74783d5e2634547260fd44fb7e46dd47085d58',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0051',
  'field': 'equipment',
  'drill_name': 'Free Throw Golf',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball and hoop',
  'source_evidence_note': 'Free throws use baskets.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': '2888d4d9862af3c18d38393ba19c76249fb00e6380db905606fcd3c4daedae03',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0051',
  'field': 'participants',
  'drill_name': 'Free Throw Golf',
  'proposed_action': 'revise_value',
  'proposed_value': 'Rotating shooters taking turns at baskets; optional partner competition',
  'source_evidence_note': 'No fixed group size.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': '2888d4d9862af3c18d38393ba19c76249fb00e6380db905606fcd3c4daedae03',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0052',
  'field': 'equipment',
  'drill_name': '2 on 2 Box Out Game',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball and hoop',
  'source_evidence_note': 'Rim rebounds and scoring.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': '45a4d31631160d0eb4bf2318be855bbf5b1b858aa6d4605c5c4b355b217f575d',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0052',
  'field': 'participants',
  'drill_name': '2 on 2 Box Out Game',
  'proposed_action': 'revise_value',
  'proposed_value': 'Two teams of two plus a coach or another player initiating rebounds',
  'source_evidence_note': 'Includes the rebound initiator.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': '45a4d31631160d0eb4bf2318be855bbf5b1b858aa6d4605c5c4b355b217f575d',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0053',
  'field': 'equipment',
  'drill_name': '2-Person Closeout Drill',
  'proposed_action': 'retain_current',
  'proposed_value': 'Basketball and hoop',
  'source_evidence_note': 'Live scoring uses the basket.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': 'd577c8a61853c9b5d14618bc48967d75321c1f66c457d08356cbf078f51933be',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0053',
  'field': 'participants',
  'drill_name': '2-Person Closeout Drill',
  'proposed_action': 'revise_value',
  'proposed_value': 'Two wing passers (a coach and another coach or player), with two baseline lines '
                    'rotating defenders',
  'source_evidence_note': 'Includes live two-on-two play.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': 'd577c8a61853c9b5d14618bc48967d75321c1f66c457d08356cbf078f51933be',
  'proposal_origin': 'step12_63_source_review'},
 {'evidence_id': 'JR-EVID-0054',
  'field': 'participants',
  'drill_name': '4 on 4 Continuous with Constraints',
  'proposed_action': 'retain_current',
  'proposed_value': 'Three teams of four in source example',
  'source_evidence_note': 'Substitutes may be added.',
  'source_url': 'https://jr.nba.com/basketball-practice-plans/starter/page/3/',
  'expected_section_sha256': 'dce7a298533c9a082fde97e4ce40dccfe5544c98cf2fc3d36f264de8f41ac865',
  'proposal_origin': 'step12_63_source_review'}]


def require(condition, message):
    if not condition:
        raise ValueError(message)


def sha256(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def rel(path):
    return path.relative_to(ROOT).as_posix()


def read_csv(path):
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    require(columns and len(columns) == len(set(columns)), f'Invalid CSV header: {path.name}')
    require(all(None not in row and all(v is not None for v in row.values()) for row in rows),
            f'Malformed CSV record: {path.name}')
    return rows, columns


def unique_index(rows, field, label):
    keys = [row[field] for row in rows]
    require(all(keys) and len(keys) == len(set(keys)), f'{label}: duplicate or empty {field}.')
    return dict(zip(keys, rows))


def json_bytes(value):
    return (json.dumps(value, ensure_ascii=False, indent=2) + '\n').encode('utf-8')


def csv_bytes(rows, columns):
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=columns, lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')


import re
import unicodedata
import hashlib
from html.parser import HTMLParser

def normalize(text):
    return re.sub(r'[^a-z0-9]+', ' ', unicodedata.normalize('NFKC', text).casefold()).strip()

def fingerprint(text):
    return hashlib.sha256(normalize(text).encode('utf-8')).hexdigest()

class SourceSections(HTMLParser):
    def __init__(self):
        super().__init__(convert_charrefs=True)
        self.blocks = []
        self.skip = []
        self.heading = None
        self.parts = []
        self.current = None
        self.visible = []

    def handle_starttag(self, tag, attrs):
        if tag in {'script', 'style', 'noscript'}:
            self.skip.append(tag)
        elif not self.skip:
            if re.fullmatch(r'h[1-6]', tag):
                self.heading = tag
                self.parts = []
            elif self.current is not None and tag in {'p', 'li', 'br', 'div'}:
                self.current['parts'].append(' ')

    def handle_data(self, text):
        if self.skip:
            return
        self.visible.append(text)
        if self.heading:
            self.parts.append(text)
        elif self.current is not None:
            self.current['parts'].append(text)

    def handle_endtag(self, tag):
        if self.skip:
            if tag == self.skip[-1]:
                self.skip.pop()
            return
        if tag == self.heading:
            self.current = {'heading': ' '.join(self.parts).strip(),
                            'level': int(tag[1]), 'parts': []}
            self.blocks.append(self.current)
            self.heading = None
            self.parts = []
        elif self.current is not None and tag in {'p', 'li', 'br', 'div'}:
            self.current['parts'].append(' ')

    def section(self, name):
        matches = [(i, b) for i, b in enumerate(self.blocks)
                   if normalize(b['heading']) == normalize(name)]
        if len(matches) != 1:
            raise ValueError(f'Missing or ambiguous heading: {name}')
        start, block = matches[0]
        parts = list(block['parts'])
        end = start + 1
        while end < len(self.blocks) and self.blocks[end]['level'] > block['level']:
            parts.append(self.blocks[end]['heading'])
            parts.extend(self.blocks[end]['parts'])
            end += 1
        text = re.sub(r'\s+', ' ', ' '.join(parts)).strip()
        if not text:
            raise ValueError(f'Empty drill section: {name}')
        return {'heading': block['heading'], 'text': text,
                'next_heading': self.blocks[end]['heading'] if end < len(self.blocks) else '',
                'section_sha256': fingerprint(text)}



for name, path in INPUT.items():
    if not path.is_file():
        raise FileNotFoundError(f'Missing {name}: {path}')
input_hashes = {rel(path): sha256(path) for path in INPUT.values()}
summary61 = json.loads(INPUT['summary61'].read_text(encoding='utf-8'))
summary62 = json.loads(INPUT['summary62'].read_text(encoding='utf-8'))
require(summary61.get('step') == '12-61' and summary61.get('catalog_version') == '0.3.8'
        and summary61.get('pending_other_wave3_fields') == 68,
        'Expected the verified Step 12-61 v0.3.8 review scope.')
require(summary62.get('step') == '12-62' and summary62.get('catalog_version') == '0.3.8'
        and summary62.get('selected_field_recommendations') == 4
        and summary62.get('total_fields_still_pending_confirmation') == 68
        and summary62.get('source_anchor_review_needed') == 0,
        'Step 12-62 differs from the successful run supplied for continuation.')
for name in ('catalog', 'registry', 'pending'):
    path = INPUT[name]
    checks = {k.replace('\\', '/'): v for k, v in summary61.get('output_sha256', {}).items()}
    require(checks.get(rel(path)) == input_hashes[rel(path)], f'Step 12-61 input integrity mismatch: {name}.')
for name in ('catalog', 'registry', 'pending', 'summary61'):
    path = INPUT[name]
    checks = {k.replace('\\', '/'): v for k, v in summary62.get('input_sha256', {}).items()}
    require(checks.get(rel(path)) == input_hashes[rel(path)], f'Step 12-62 input integrity mismatch: {name}.')
for name in ('plan2_decisions', 'plan2_candidates', 'plan2_other', 'plan2_snapshot'):
    path = INPUT[name]
    checks = {k.replace('\\', '/'): v for k, v in summary62.get('output_sha256', {}).items()}
    require(checks.get(rel(path)) == input_hashes[rel(path)], f'Step 12-62 output integrity mismatch: {name}.')

catalog, _ = read_csv(INPUT['catalog'])
registry, _ = read_csv(INPUT['registry'])
pending, _ = read_csv(INPUT['pending'])
plan2, _ = read_csv(INPUT['plan2_decisions'])
plan2_other, _ = read_csv(INPUT['plan2_other'])
snapshot62 = json.loads(INPUT['plan2_snapshot'].read_text(encoding='utf-8'))
require((len(catalog), len(registry), len(pending), len(plan2), len(plan2_other)) == (144, 156, 68, 4, 64),
        'Catalog, source registry, or review record counts changed.')
cat_by = unique_index(catalog, 'catalog_candidate_id', 'Catalog')
reg_by = unique_index(registry, 'evidence_id', 'Registry')
pending_by = {(r['evidence_id'], r['field']): r for r in pending}
spec_by = {(r['evidence_id'], r['field']): r for r in REVIEW_SPEC}
plan2_by = {(r['evidence_id'], r['field']): r for r in plan2}
require(len(pending_by) == len(spec_by) == 68 and set(pending_by) == set(spec_by),
        'The consolidated review must cover the exact original 68 fields.')
require(len(plan2_by) == 4 and set(plan2_by) <= set(pending_by)
        and {(r['evidence_id'], r['field']) for r in plan2_other} == set(pending_by) - set(plan2_by),
        'The Step 12-62 review partition changed.')
require(len({r['evidence_id'] for r in pending}) == 38
        and len({r['source_url'] for r in pending}) == 8,
        'Expected 38 source-linked drills on eight source pages.')
require(all(r['catalog_version'] == '0.3.8' and r['catalog_approval_status'] == 'pending'
            and r['planner_eligible'] == 'no' and not r['approved_drill_id'] for r in catalog),
        'Catalog version or approval state changed; preserve the current catalog.')
GATES = ('team_signoff', 'permission_review', 'youth_planning_review', 'catalog_approval_status')
for key, row in pending_by.items():
    eid, field = key
    cid = row['candidate_group_id']
    require(field in {'equipment', 'participants'} and eid in reg_by and cid in cat_by,
            f'Unknown source or field: {key}.')
    source, item, spec = reg_by[eid], cat_by[cid], spec_by[key]
    require(source['catalog_candidate_id'] == cid
            and item['evidence_ids'].split(';') == [eid]
            and item['source_occurrences'] == '1'
            and row['drill_name'] == source['drill_name'] == item['drill_name'] == spec['drill_name']
            and row['source_url'] == source['source_url'] == spec['source_url']
            and row['source_level'] == source['source_level']
            and row['practice_plan'] == source['practice_plan'],
            f'Source linkage changed: {key}.')
    require(row['current_catalog_value'] == source[field] == item[field], f'Current value mismatch: {key}.')
    require(row['reviewer_decision'] == 'pending' and row['planner_eligible'] == 'no'
            and all(row[gate] == 'pending' for gate in GATES)
            and not row['final_value'] and not row['reviewer'] and not row['review_date'],
            f'An existing reviewer decision must be preserved: {key}.')
    require(spec['proposed_action'] in {'retain_current', 'revise_value', 'defer_source_evidence'},
            f'Invalid recommendation action: {key}.')
    if spec['proposed_action'] == 'retain_current':
        require(spec['proposed_value'] == row['current_catalog_value'], f'Retain value changed: {key}.')
    if spec['proposed_action'] == 'revise_value':
        require(spec['proposed_value'] and spec['proposed_value'] != row['current_catalog_value'],
                f'Invalid or redundant revision: {key}.')
    if key in plan2_by:
        old = plan2_by[key]
        require(all(old[k] == spec[k] for k in ('proposed_action', 'proposed_value', 'source_evidence_note'))
                and old['current_value'] == row['current_catalog_value']
                and old['reviewer_decision'] == 'pending' and old['planner_eligible'] == 'no'
                and all(old[gate] == 'pending' for gate in GATES),
                f'A Step 12-62 proposal or decision changed: {key}.')

PLAN2_URL = 'https://jr.nba.com/basketball-practice-plans/all-star/page/2/'
require(snapshot62.get('source_url') == PLAN2_URL
        and snapshot62.get('practice_plan') == 'All-Star Practice Plan 2',
        'Step 12-62 source snapshot identity changed.')
sections62 = unique_index(snapshot62['sections'], 'heading', 'Step 12-62 snapshot')
for key, row in plan2_by.items():
    section = sections62[row['drill_name']]
    require(fingerprint(section['text']) == section['normalized_section_sha256']
            == row['section_fingerprint_sha256'] == spec_by[key]['expected_section_sha256'],
            'Step 12-62 source evidence differs from the reviewed section.')


def fetch_page(url):
    url_info = urlsplit(url)
    require(url_info.scheme == 'https' and url_info.hostname == 'jr.nba.com'
            and url_info.path.startswith('/basketball-practice-plans/'), 'Non-publisher source URL.')
    page_rows = [r for r in pending if r['source_url'] == url]
    result = {'source_url': url, 'retrieved_at_utc': datetime.now(timezone.utc).isoformat(),
              'page_status': 'fetch_failed', 'error': '', 'sections': {}}
    try:
        with urlopen(Request(url, headers={'User-Agent': 'Mozilla/5.0'}), timeout=25) as response:
            resolved = urlsplit(response.geturl())
            require(response.status == 200 and resolved.scheme == 'https'
                    and resolved.hostname == 'jr.nba.com'
                    and resolved.path.rstrip('/') == url_info.path.rstrip('/'), 'Unexpected publisher response.')
            raw = response.read(3_000_001)
            require(len(raw) <= 3_000_000, 'Source page exceeds the supported size.')
            html = raw.decode(response.headers.get_content_charset() or 'utf-8', errors='replace')
        parser = SourceSections()
        parser.feed(html)
        parser.close()
        expected_plan = f"{page_rows[0]['source_level']} Practice Plan {page_rows[0]['practice_plan']}"
        require(normalize(expected_plan) in normalize(' '.join(parser.visible)), 'Practice-plan identity mismatch.')
        result.update(page_status='fetched', page_sha256=hashlib.sha256(raw).hexdigest())
        for name in sorted({r['drill_name'] for r in page_rows}):
            try:
                result['sections'][name] = dict(parser.section(name), extraction_status='found')
            except ValueError as error:
                result['sections'][name] = {'heading': name, 'text': '', 'section_sha256': '',
                                             'extraction_status': 'missing_or_ambiguous', 'error': str(error)}
    except Exception as error:
        result['error'] = f'{type(error).__name__}: {error}'
    return result


print('INPUT INTEGRITY: PASS - Steps 12-61 and 12-62 inputs and outputs verified.')
existing = [path.exists() for path in OUTPUT.values()]
if any(existing) and not all(existing):
    raise FileExistsError('Only part of the Step 12-63 output set exists. Inspect it before rerunning.')
if all(existing):
    summary = json.loads(OUTPUT['summary'].read_text(encoding='utf-8'))
    require(summary.get('step') == '12-63' and summary.get('input_sha256') == input_hashes,
            'Existing Step 12-63 outputs came from different inputs.')
    for name, path in OUTPUT.items():
        if name != 'summary':
            require(summary.get('output_sha256', {}).get(rel(path)) == sha256(path),
                    f'Existing output changed: {rel(path)}. It has been preserved.')
    snapshot = json.loads(OUTPUT['snapshot'].read_text(encoding='utf-8'))
    mode = 'PRESERVED'
else:
    other_urls = sorted({r['source_url'] for r in pending} - {PLAN2_URL})
    print('Checking seven publisher pages; reusing the verified Step 12-62 evidence for Plan 2...')
    with ThreadPoolExecutor(max_workers=4) as pool:
        pages = list(pool.map(fetch_page, other_urls))
    carried = {'source_url': PLAN2_URL, 'page_status': 'verified_step12_62_snapshot',
               'retrieved_at_utc': snapshot62['retrieved_at_utc'],
               'page_sha256': snapshot62['page_sha256'], 'error': '', 'sections': {}}
    for name, section in sections62.items():
        carried['sections'][name] = {'heading': name, 'text': section['text'],
                                     'section_sha256': section['normalized_section_sha256'],
                                     'next_heading': section['next_heading'], 'extraction_status': 'found'}
    pages.append(carried)
    snapshot = {'step': '12-63', 'pages': sorted(pages, key=lambda r: r['source_url']),
                'section_boundary_rule': 'Drill heading through its subsections, stopping before the next heading of the same or higher level.',
                'fingerprint_method': 'NFKC/casefold; replace non-ASCII-alphanumeric runs with spaces; UTF-8 SHA-256.',
                'plan2_evidence_reference': rel(INPUT['plan2_snapshot'])}
    mode = 'CREATED'

page_by = unique_index(snapshot['pages'], 'source_url', 'Consolidated source snapshot')
require(set(page_by) == {r['source_url'] for r in pending}, 'Source snapshot page coverage changed.')
recommendations = []
for spec in REVIEW_SPEC:
    key = spec['evidence_id'], spec['field']
    old = pending_by[key]
    page = page_by[old['source_url']]
    section = page['sections'].get(old['drill_name'], {})
    current_fingerprint = fingerprint(section['text']) if section.get('text') else ''
    supported = (page['page_status'] in {'fetched', 'verified_step12_62_snapshot'}
                 and section.get('extraction_status') == 'found'
                 and current_fingerprint == section.get('section_sha256') == spec['expected_section_sha256'])
    if supported:
        status = 'matches_reviewed_section'
        action, value, note = spec['proposed_action'], spec['proposed_value'], spec['source_evidence_note']
    else:
        status = ('page_unavailable' if page['page_status'] == 'fetch_failed' else
                  'section_missing_or_ambiguous' if section.get('extraction_status') != 'found' else 'source_section_changed')
        action, value = 'defer_source_evidence', ''
        note = 'Source evidence needs another review; the stored recommendation was not applied.'
    recommendations.append({
        'candidate_group_id': old['candidate_group_id'], 'evidence_id': old['evidence_id'],
        'drill_name': old['drill_name'], 'source_level': old['source_level'],
        'practice_plan': old['practice_plan'], 'field': old['field'],
        'current_value': old['current_catalog_value'], 'proposed_action': action, 'proposed_value': value,
        'source_evidence_note': note, 'source_evidence_status': status,
        'expected_section_sha256': spec['expected_section_sha256'],
        'observed_section_sha256': current_fingerprint, 'source_url': old['source_url'],
        'proposal_origin': spec['proposal_origin'], 'source_snapshot_reference': rel(OUTPUT['snapshot']),
        'reviewer_decision': 'pending', 'final_value': '', 'reviewer': '', 'review_date': '',
        'team_signoff': 'pending', 'permission_review': 'pending', 'youth_planning_review': 'pending',
        'catalog_approval_status': 'pending', 'planner_eligible': 'no',
    })
exceptions = [r for r in recommendations if r['proposed_action'] == 'defer_source_evidence']
candidates = []
for eid in sorted({r['evidence_id'] for r in recommendations}):
    rows = [r for r in recommendations if r['evidence_id'] == eid]
    actions = Counter(r['proposed_action'] for r in rows)
    candidates.append({'candidate_group_id': rows[0]['candidate_group_id'], 'evidence_id': eid,
                       'drill_name': rows[0]['drill_name'], 'source_url': rows[0]['source_url'],
                       'pending_fields': len(rows), 'proposed_retain': actions['retain_current'],
                       'proposed_revise': actions['revise_value'],
                       'source_exceptions': actions['defer_source_evidence'],
                       'reviewer_decision': 'pending', 'catalog_approval_status': 'pending',
                       'planner_eligible': 'no'})
require(len(recommendations) == 68 and len(candidates) == 38
        and sum(r['pending_fields'] for r in candidates) == 68, 'Consolidated coverage is incomplete.')
actions = Counter(r['proposed_action'] for r in recommendations)
expected_summary = {
    'step': '12-63', 'catalog_version': '0.3.8',
    'catalog_candidate_groups': 144, 'registered_source_occurrences': 156,
    'reviewed_source_pages': 8, 'selected_candidate_groups': 38, 'pending_fields_covered': 68,
    'carried_forward_step12_62_fields': 4, 'additional_fields_covered': 64,
    'proposed_retain': actions['retain_current'], 'proposed_revise': actions['revise_value'],
    'source_exceptions': len(exceptions), 'fields_without_review_packet_entry': 0,
    'fields_still_pending_confirmation': 68, 'reviewer_signoffs': 0,
    'catalog_approvals': 0, 'planner_eligible': 0,
    'original_inputs_modified': False, 'git_operations_performed': False,
    'input_sha256': input_hashes,
}
new_files = {
    'recommendations': csv_bytes(recommendations, list(recommendations[0])),
    'candidates': csv_bytes(candidates, list(candidates[0])),
    'exceptions': csv_bytes(exceptions, list(recommendations[0])),
    'snapshot': json_bytes(snapshot),
}
if mode == 'CREATED':
    summary = dict(expected_summary)
    summary['output_sha256'] = {rel(OUTPUT[name]): hashlib.sha256(data).hexdigest() for name, data in new_files.items()}
    new_files['summary'] = json_bytes(summary)
    with TemporaryDirectory(prefix='step12_63_', dir=ROOT) as staging_dir:
        staged = {}
        for name, data in new_files.items():
            staged[name] = Path(staging_dir) / OUTPUT[name].name
            staged[name].write_bytes(data)
            require(staged[name].read_bytes() == data, f'Staging verification failed: {name}.')
        require(all(sha256(path) == input_hashes[rel(path)] for path in INPUT.values()),
                'An input changed while gathering evidence. No outputs created.')
        if any(path.exists() for path in OUTPUT.values()):
            raise FileExistsError('An output appeared during staging. Preserve and inspect it.')
        created = []
        try:
            for name, path in OUTPUT.items():
                path.parent.mkdir(parents=True, exist_ok=True)
                with path.open('xb') as handle:
                    created.append(path)
                    handle.write(staged[name].read_bytes())
        except Exception:
            for path in created:
                path.unlink(missing_ok=True)
            raise
else:
    require(all(summary.get(key) == value for key, value in expected_summary.items()),
            'Saved summary does not match the verified review state.')
    require(all(OUTPUT[name].read_bytes() == data for name, data in new_files.items()),
            'Saved review packet differs from the current specification. It has been preserved.')
require(all(sha256(path) == input_hashes[rel(path)] for path in INPUT.values()),
        'An original input changed during execution.')
for name, path in OUTPUT.items():
    if name != 'summary':
        require(summary['output_sha256'].get(rel(path)) == sha256(path), f'Saved output verification failed: {name}.')

print('=' * 76)
print('STEP 12-63 - CONSOLIDATED WAVE 3 REQUIREMENT REVIEW')
print('=' * 76)
print('Catalog candidate groups / source occurrences: 144 / 156')
print('Pending fields / drill candidates / source pages covered: 68 / 38 / 8')
print('Carried forward from Step 12-62 / additional fields: 4 / 64')
print(f"Recommendations: retain={actions['retain_current']}, revise={actions['revise_value']}, defer={actions['defer_source_evidence']}")
print('Fields without a review-packet entry: 0')
print('Fields still awaiting confirmation: 68')
print('Reviewer signoffs / approved / planner-eligible: 0 / 0 / 0')
print('PROPOSED VALUE CHANGES - PENDING YOUR REVIEW')
for row in recommendations:
    if row['proposed_action'] == 'revise_value':
        print(f"  {row['evidence_id']} | {row['drill_name']} | {row['field']}")
        print(f"    {row['current_value']} -> {row['proposed_value']}")
if exceptions:
    print('SOURCE EXCEPTIONS - ORIGINAL VALUES PRESERVED')
    for row in exceptions:
        print(f"  {row['evidence_id']} | {row['field']} | {row['source_evidence_status']}")
print('OUTPUTS')
for name, path in OUTPUT.items():
    print(f'  {mode:<9} {rel(path)}')
print('Catalog, registry, original review ledger, and Step 12-62 files modified: No')
print('Git operations performed: None')
print('Next: review the consolidated field CSV before Step 12-64 applies any confirmed changes.')
print('OVERALL: ' + ('REVIEW NEEDED - source exceptions recorded; all approvals remain pending'
                      if exceptions else 'PASS - all 68 fields covered; confirmation remains pending'))
print('=' * 76)
```

```python
# Step 12-64 - Consolidated Draft Catalog and Source-Link Reconciliation
# Prepare v0.3.9 REVIEW_ONLY using the source-matched Step 12-63 proposals.
# Draft metadata application is not human confirmation or final drill approval.

import csv
import hashlib
import io
import json
import re
import unicodedata
from collections import Counter, defaultdict
from pathlib import Path
from tempfile import TemporaryDirectory

ROOT = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Run this cell inside the Champions Athletic Map Git project.')
RESULTS = ROOT / 'results'
INPUT = {
    'catalog': ROOT / 'data/processed/basketball_drill_catalog_v0_3_8_REVIEW_ONLY.csv',
    'registry': RESULTS / 'basketball_drill_source_registry_v0_3_8.csv',
    'pending': RESULTS / 'basketball_wave3_remaining_requirement_reviews_v0_3_8.csv',
    'summary61': RESULTS / 'basketball_step12_61_summary_v0_3_8.json',
    'summary63': RESULTS / 'basketball_step12_63_summary_v0_3_8.json',
    'review': RESULTS / 'basketball_wave3_consolidated_field_review_v0_3_8.csv',
    'candidate_review': RESULTS / 'basketball_wave3_consolidated_candidate_review_v0_3_8.csv',
    'exceptions': RESULTS / 'basketball_wave3_consolidated_source_exceptions_v0_3_8.csv',
    'snapshot': RESULTS / 'basketball_wave3_consolidated_source_snapshot_v0_3_8.json',
}
OUTPUT = {
    'catalog': ROOT / 'data/processed/basketball_drill_catalog_v0_3_9_REVIEW_ONLY.csv',
    'registry': RESULTS / 'basketball_drill_source_registry_v0_3_9.csv',
    'draft_review': RESULTS / 'basketball_wave3_draft_metadata_review_v0_3_9.csv',
    'changes': RESULTS / 'basketball_wave3_draft_metadata_change_log_v0_3_9.csv',
    'linkage_audit': RESULTS / 'basketball_source_linkage_audit_v0_3_9.csv',
    'canonical_status': RESULTS / 'basketball_canonical_reconciliation_status_v0_3_9.csv',
    'summary': RESULTS / 'basketball_step12_64_summary_v0_3_9.json',
}
EXPECTED_REVIEW_SHA256 = 'ff8409eec4b2b324750205952d3644ca43a340d7a438f2764359a0c8edb8ffc0'
DRAFT_STATUS = 'source_matched_proposal_applied_to_draft'
TRACE_FIELDS = ('wave3_requirement_draft_status', 'wave3_requirement_draft_fields',
                'wave3_requirement_draft_reference')


def require(condition, message):
    if not condition:
        raise ValueError(message)


def sha256(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def rel(path):
    return path.relative_to(ROOT).as_posix()


def normalized_hashes(mapping):
    return {key.replace('\\', '/'): value for key, value in mapping.items()}


def read_csv(path):
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    require(columns and len(columns) == len(set(columns)), f'Invalid CSV header: {path.name}')
    require(all(None not in row and all(v is not None for v in row.values()) for row in rows),
            f'Malformed CSV records: {path.name}')
    return rows, columns


def unique_index(rows, field, label):
    keys = [row[field] for row in rows]
    require(all(keys) and len(keys) == len(set(keys)), f'{label}: missing or repeated {field}.')
    return dict(zip(keys, rows))


def split_ids(value):
    return [part.strip() for part in value.split(';') if part.strip()]


def fingerprint(text):
    normalized = re.sub(r'[^a-z0-9]+', ' ', unicodedata.normalize('NFKC', text).casefold()).strip()
    return hashlib.sha256(normalized.encode('utf-8')).hexdigest()


def json_bytes(value):
    return (json.dumps(value, ensure_ascii=False, indent=2) + '\n').encode('utf-8')


def csv_bytes(rows, columns):
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=columns, lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')


for name, path in INPUT.items():
    if not path.is_file():
        raise FileNotFoundError(f'Missing {name}: {path}')
input_hashes = {rel(path): sha256(path) for path in INPUT.values()}
summary61 = json.loads(INPUT['summary61'].read_text(encoding='utf-8'))
summary63 = json.loads(INPUT['summary63'].read_text(encoding='utf-8'))
require(summary61.get('step') == '12-61' and summary61.get('catalog_version') == '0.3.8',
        'Expected the Step 12-61 v0.3.8 catalog summary.')
require(summary63.get('step') == '12-63' and summary63.get('catalog_version') == '0.3.8'
        and summary63.get('pending_fields_covered') == 68
        and summary63.get('selected_candidate_groups') == 38
        and summary63.get('proposed_retain') == 38 and summary63.get('proposed_revise') == 30
        and summary63.get('source_exceptions') == 0
        and summary63.get('fields_still_pending_confirmation') == 68,
        'Step 12-63 does not match the successful consolidated review supplied for continuation.')
checks61 = normalized_hashes(summary61.get('output_sha256', {}))
checks63_in = normalized_hashes(summary63.get('input_sha256', {}))
checks63_out = normalized_hashes(summary63.get('output_sha256', {}))
for name in ('catalog', 'registry', 'pending'):
    path = INPUT[name]
    require(checks61.get(rel(path)) == input_hashes[rel(path)], f'Step 12-61 integrity mismatch: {name}.')
for name in ('catalog', 'registry', 'pending', 'summary61'):
    path = INPUT[name]
    require(checks63_in.get(rel(path)) == input_hashes[rel(path)], f'Step 12-63 input mismatch: {name}.')
for name in ('review', 'candidate_review', 'exceptions', 'snapshot'):
    path = INPUT[name]
    require(checks63_out.get(rel(path)) == input_hashes[rel(path)], f'Step 12-63 output mismatch: {name}.')
require(sha256(INPUT['review']) == EXPECTED_REVIEW_SHA256,
        'The review packet differs from the exact 68-field packet inspected for Step 12-64.')

catalog, catalog_columns = read_csv(INPUT['catalog'])
registry, registry_columns = read_csv(INPUT['registry'])
pending, _ = read_csv(INPUT['pending'])
review, review_columns = read_csv(INPUT['review'])
candidate_review, _ = read_csv(INPUT['candidate_review'])
exceptions, _ = read_csv(INPUT['exceptions'])
snapshot = json.loads(INPUT['snapshot'].read_text(encoding='utf-8'))
require((len(catalog), len(registry), len(pending), len(review), len(candidate_review), len(exceptions))
        == (144, 156, 68, 68, 38, 0), 'Unexpected input record counts.')
cat_by = unique_index(catalog, 'catalog_candidate_id', 'Catalog')
reg_by = unique_index(registry, 'evidence_id', 'Registry')
candidate_by = unique_index(candidate_review, 'candidate_group_id', 'Candidate review')
pending_by = {(r['evidence_id'], r['field']): r for r in pending}
review_by = {(r['evidence_id'], r['field']): r for r in review}
require(len(pending_by) == len(review_by) == 68 and set(pending_by) == set(review_by),
        'Review keys do not match the original 68 pending fields.')
require(set(candidate_by) == {r['candidate_group_id'] for r in review}, 'Candidate review linkage mismatch.')
require(not (set(TRACE_FIELDS) & set(catalog_columns)) and not (set(TRACE_FIELDS) & set(registry_columns)),
        'Draft reconciliation trace fields already exist in an input; preserve and inspect it.')
require(all(r['catalog_version'] == '0.3.8' and r['catalog_approval_status'] == 'pending'
            and r['planner_eligible'] == 'no' and not r['approved_drill_id'] for r in catalog),
        'Catalog version or final approval state changed.')
source_pages = unique_index(snapshot['pages'], 'source_url', 'Source snapshot')
require(snapshot.get('step') == '12-63' and len(source_pages) == 8, 'Source snapshot identity changed.')

# Reconcile source membership structurally. Imported aliases and source-specific
# metadata are preserved; this audit does not decide semantic equivalence.
registry_by_candidate = defaultdict(set)
for source in registry:
    require(source['catalog_candidate_id'] in cat_by, f"Unlinked registry record: {source['evidence_id']}.")
    registry_by_candidate[source['catalog_candidate_id']].add(source['evidence_id'])
linkage_audit = []
canonical_status = []
for item in catalog:
    cid = item['catalog_candidate_id']
    primary = split_ids(item['evidence_ids'])
    imported = split_ids(item['aditi_evidence_ids'])
    require(len(primary) == len(set(primary)) and len(imported) == len(set(imported)),
            f'Duplicate evidence identifiers within a source list: {cid}.')
    combined = set(primary) | set(imported)
    require(combined == registry_by_candidate[cid]
            and int(item['source_occurrences']) == len(primary)
            and int(item['aditi_source_records']) == len(imported)
            and int(item['total_evidence_records']) == len(combined),
            f'Catalog/registry source membership or count mismatch: {cid}.')
    for eid in sorted(combined):
        source = reg_by[eid]
        linkage_audit.append({
            'catalog_candidate_id': cid, 'catalog_drill_name': item['drill_name'],
            'evidence_id': eid, 'source_drill_name': source['drill_name'],
            'source_origin': source['origin'], 'source_url': source['source_url'],
            'primary_evidence_member': 'yes' if eid in primary else 'no',
            'imported_evidence_member': 'yes' if eid in imported else 'no',
            'primary_source_occurrences': len(primary), 'total_linked_evidence_records': len(combined),
            'structural_linkage_status': 'pass',
            'source_integration_status': source['integration_status'],
            'source_locator_status': source['source_locator_status'],
            'canonical_identity_status': item['canonical_linkage_review_status'],
            'semantic_equivalence_decided_by_this_audit': 'no',
            'catalog_approval_status': 'pending', 'planner_eligible': 'no',
        })
    canonical_status.append({
        'catalog_candidate_id': cid, 'drill_name': item['drill_name'],
        'scope_candidate': item['scope_candidate'], 'skill_area': item['skill_area'],
        'structural_linkage_status': 'pass', 'linked_evidence_records': len(combined),
        'source_identity_review_cases': item['source_identity_review_cases'],
        'source_identity_evidence_status': item['source_identity_evidence_status'],
        'canonical_linkage_review_status': item['canonical_linkage_review_status'],
        'distinctness_status': item['distinctness_status'], 'metadata_status': item['metadata_status'],
        'permission_status': item['permission_status'], 'reproduction_permission': item['reproduction_permission'],
        'youth_planning_status': item['youth_planning_status'], 'requirements_review': item['requirements_review'],
        'next_recorded_identity_action': item['source_review_next_action'],
        'identity_resolution_performed_in_this_step': 'no',
        'catalog_approval_status': 'pending', 'planner_eligible': 'no',
    })
require(len(linkage_audit) == 156 and len({r['evidence_id'] for r in linkage_audit}) == 156,
        'The linkage audit does not cover all 156 source records exactly once.')
open_identity = [r for r in canonical_status if r['canonical_linkage_review_status'] == 'pending_human_review']
open_cases = {case for r in open_identity for case in split_ids(r['source_identity_review_cases'])}
require((len(open_identity), len(open_cases)) == (9, 8), 'Known unresolved identity-case scope changed.')

GATES = ('team_signoff', 'permission_review', 'youth_planning_review', 'catalog_approval_status')
for key, row in review_by.items():
    eid, field = key
    require(field in {'equipment', 'participants'}, f'Unexpected metadata field: {field}.')
    cid = row['candidate_group_id']
    require(eid in reg_by and cid in cat_by, f'Unknown review linkage: {key}.')
    source, item, original = reg_by[eid], cat_by[cid], pending_by[key]
    require(item['evidence_ids'].split(';') == [eid]
            and registry_by_candidate[cid] == {eid}
            and source['catalog_candidate_id'] == cid
            and row['drill_name'] == source['drill_name'] == item['drill_name'] == original['drill_name']
            and row['source_url'] == source['source_url'] == original['source_url']
            and row['source_level'] == source['source_level'] == original['source_level']
            and row['practice_plan'] == source['practice_plan'] == original['practice_plan'],
            f'Single-source drill linkage changed: {key}.')
    require(row['current_value'] == item[field] == source[field] == original['current_catalog_value'],
            f'Original value mismatch: {key}.')
    require(row['reviewer_decision'] == original['reviewer_decision'] == 'pending'
            and row['planner_eligible'] == original['planner_eligible'] == 'no'
            and all(row[g] == original[g] == 'pending' for g in GATES)
            and not row['final_value'] and not row['reviewer'] and not row['review_date'],
            f'A human review or approval decision must be preserved: {key}.')
    require(row['source_evidence_status'] == 'matches_reviewed_section'
            and row['expected_section_sha256'] == row['observed_section_sha256']
            and re.fullmatch(r'[0-9a-f]{64}', row['observed_section_sha256']),
            f'Source evidence is incomplete: {key}.')
    page = source_pages[row['source_url']]
    section = page['sections'].get(row['drill_name'], {})
    require(page['page_status'] in {'fetched', 'verified_step12_62_snapshot'}
            and section.get('extraction_status') == 'found'
            and fingerprint(section.get('text', '')) == section.get('section_sha256')
            == row['observed_section_sha256'], f'Saved source evidence does not support the draft: {key}.')
    action = row['proposed_action']
    require(action in {'retain_current', 'revise_value'} and row['proposed_value'],
            f'Unsupported or empty proposal: {key}.')
    require((action == 'retain_current') == (row['proposed_value'] == row['current_value']),
            f'Action/value mismatch: {key}.')
require(Counter(r['proposed_action'] for r in review) == Counter({'retain_current': 38, 'revise_value': 30}),
        'Expected exactly 38 retains and 30 proposed revisions.')

draft_catalog = [dict(row) for row in catalog]
draft_registry = [dict(row) for row in registry]
draft_cat_by = {row['catalog_candidate_id']: row for row in draft_catalog}
draft_reg_by = {row['evidence_id']: row for row in draft_registry}
fields_by_candidate = defaultdict(set)
fields_by_evidence = defaultdict(set)
draft_review = []
changes = []
for row in review:
    eid, cid, field = row['evidence_id'], row['candidate_group_id'], row['field']
    new_value = row['proposed_value']
    draft_cat_by[cid][field] = new_value
    draft_reg_by[eid][field] = new_value
    fields_by_candidate[cid].add(field)
    fields_by_evidence[eid].add(field)
    changed = new_value != row['current_value']
    draft_review.append(dict(row, original_catalog_version='0.3.8', draft_catalog_version='0.3.9',
                             draft_catalog_value=new_value, draft_value_changed='yes' if changed else 'no',
                             metadata_application_status=DRAFT_STATUS,
                             human_confirmation_recorded='no'))
    if changed:
        changes.append({
            'candidate_group_id': cid, 'evidence_id': eid, 'drill_name': row['drill_name'],
            'field': field, 'original_v0_3_8_value': row['current_value'], 'draft_v0_3_9_value': new_value,
            'application_basis': DRAFT_STATUS, 'source_url': row['source_url'],
            'source_section_sha256': row['observed_section_sha256'],
            'review_reference': rel(OUTPUT['draft_review']), 'reviewer_decision': 'pending',
            'team_signoff': 'pending', 'permission_review': 'pending', 'youth_planning_review': 'pending',
            'catalog_approval_status': 'pending', 'planner_eligible': 'no',
        })
for rows, identifier, fields_map in ((draft_catalog, 'catalog_candidate_id', fields_by_candidate),
                                     (draft_registry, 'evidence_id', fields_by_evidence)):
    for row in rows:
        fields = fields_map.get(row[identifier], set())
        row[TRACE_FIELDS[0]] = DRAFT_STATUS if fields else ''
        row[TRACE_FIELDS[1]] = ';'.join(sorted(fields))
        row[TRACE_FIELDS[2]] = rel(OUTPUT['draft_review']) if fields else ''
for row in draft_catalog:
    row['catalog_version'] = '0.3.9'

# Verify every existing field against the exact expected delta. This protects
# identity, provenance, previous decisions, permissions, and release controls.
changed_catalog_keys = set()
changed_registry_keys = set()
for before, after in zip(catalog, draft_catalog):
    cid = before['catalog_candidate_id']
    for column in catalog_columns:
        expected = '0.3.9' if column == 'catalog_version' else before[column]
        matching = [r for r in review if r['candidate_group_id'] == cid and r['field'] == column]
        if matching:
            expected = matching[0]['proposed_value']
        require(after[column] == expected, f'Unexpected catalog change: {cid}/{column}.')
        if column != 'catalog_version' and before[column] != after[column]:
            changed_catalog_keys.add((cid, column))
for before, after in zip(registry, draft_registry):
    eid = before['evidence_id']
    for column in registry_columns:
        expected = review_by[(eid, column)]['proposed_value'] if (eid, column) in review_by else before[column]
        require(after[column] == expected, f'Unexpected registry change: {eid}/{column}.')
        if before[column] != after[column]:
            changed_registry_keys.add((eid, column))
require(len(changes) == len(changed_catalog_keys) == len(changed_registry_keys) == 30,
        'Expected exactly 30 draft changes in both the catalog and source registry.')
require(len(fields_by_candidate) == len(fields_by_evidence) == 38, 'Expected 38 affected drill candidates.')

expected_summary = {
    'step': '12-64', 'catalog_version': '0.3.9', 'previous_catalog_version': '0.3.8',
    'catalog_state': 'review_only_draft_pending_human_confirmation',
    'catalog_candidate_groups': 144, 'registered_source_occurrences': 156,
    'basketball_candidate_groups': sum(r['scope_candidate'] == 'basketball_drill_candidate' for r in catalog),
    'general_warmup_candidate_groups': sum(r['scope_candidate'] == 'general_warmup_candidate' for r in catalog),
    'requirement_fields_in_draft_review': 68, 'affected_candidate_groups': 38,
    'retained_values': 38, 'draft_metadata_changes': 30,
    'confirmed_metadata_decisions_recorded_by_this_step': 0,
    'human_confirmations_pending': 68, 'source_exceptions_in_selected_packet': 0,
    'source_linkage_audit_rows': 156, 'source_linkage_failures': 0,
    'canonical_status_report_rows': 144, 'explicit_identity_case_candidate_groups_pending': 9,
    'explicit_identity_cases_pending': 8, 'semantic_identity_resolutions_in_this_step': 0,
    'team_signoffs_recorded_by_this_step': 0, 'catalog_approvals': 0, 'planner_eligible': 0,
    'original_inputs_modified': False, 'git_operations_performed': False,
    'input_sha256': input_hashes,
}
new_files = {
    'catalog': csv_bytes(draft_catalog, catalog_columns + list(TRACE_FIELDS)),
    'registry': csv_bytes(draft_registry, registry_columns + list(TRACE_FIELDS)),
    'draft_review': csv_bytes(draft_review, review_columns + ['original_catalog_version', 'draft_catalog_version',
                             'draft_catalog_value', 'draft_value_changed', 'metadata_application_status',
                             'human_confirmation_recorded']),
    'changes': csv_bytes(changes, list(changes[0])),
    'linkage_audit': csv_bytes(linkage_audit, list(linkage_audit[0])),
    'canonical_status': csv_bytes(canonical_status, list(canonical_status[0])),
}
print('INPUT INTEGRITY: PASS - v0.3.8 and the exact Step 12-63 review packet verified.')
existing = [path.exists() for path in OUTPUT.values()]
if any(existing) and not all(existing):
    raise FileExistsError('Only part of the Step 12-64 output set exists. Inspect it before rerunning.')
if all(existing):
    summary = json.loads(OUTPUT['summary'].read_text(encoding='utf-8'))
    require(all(summary.get(key) == value for key, value in expected_summary.items()),
            'Existing Step 12-64 summary differs from the verified draft state.')
    for name, data in new_files.items():
        path = OUTPUT[name]
        require(summary.get('output_sha256', {}).get(rel(path)) == sha256(path)
                and path.read_bytes() == data, f'Existing Step 12-64 output changed: {rel(path)}. Preserved.')
    mode = 'PRESERVED'
else:
    summary = dict(expected_summary)
    summary['output_sha256'] = {rel(OUTPUT[name]): hashlib.sha256(data).hexdigest() for name, data in new_files.items()}
    new_files['summary'] = json_bytes(summary)
    with TemporaryDirectory(prefix='step12_64_', dir=ROOT) as staging_dir:
        staged = {}
        for name, data in new_files.items():
            staged[name] = Path(staging_dir) / OUTPUT[name].name
            staged[name].write_bytes(data)
            require(staged[name].read_bytes() == data, f'Staging verification failed: {name}.')
        require(all(sha256(path) == input_hashes[rel(path)] for path in INPUT.values()),
                'An input changed during draft preparation. No outputs created.')
        if any(path.exists() for path in OUTPUT.values()):
            raise FileExistsError('A destination appeared during staging. Preserve and inspect it.')
        created = []
        try:
            for name, path in OUTPUT.items():
                path.parent.mkdir(parents=True, exist_ok=True)
                with path.open('xb') as handle:
                    created.append(path)
                    handle.write(staged[name].read_bytes())
        except Exception:
            for path in created:
                path.unlink(missing_ok=True)
            raise
    mode = 'CREATED'
require(all(sha256(path) == input_hashes[rel(path)] for path in INPUT.values()),
        'An original input changed during execution.')
for name, path in OUTPUT.items():
    if name != 'summary':
        require(summary['output_sha256'].get(rel(path)) == sha256(path), f'Saved output verification failed: {name}.')

print('=' * 76)
print('STEP 12-64 - CONSOLIDATED DRAFT CATALOG AND SOURCE-LINK RECONCILIATION')
print('=' * 76)
print('Draft catalog version / candidate groups / source occurrences: 0.3.9 / 144 / 156')
print('Basketball candidates / general warm-ups: 132 / 12')
print('Draft review fields / affected candidates: 68 / 38')
print('Retained values / draft metadata revisions: 38 / 30')
print('Source-link integrity: PASS - all 156 source records linked exactly once')
print('Explicit unresolved identity cases / candidate groups: 8 / 9 (preserved)')
print('Canonical identities merged or resolved by this step: 0')
print('Human field confirmations still pending: 68')
print('Team signoffs / catalog approvals / planner-eligible: 0 / 0 / 0')
print('OUTPUTS')
for name, path in OUTPUT.items():
    print(f'  {mode:<9} {rel(path)}')
print('Original catalogs, source inputs, and previous review decisions modified: No')
print('Git operations performed: None')
print('Next: Step 12-65 coverage, completeness, and approval-readiness reports.')
print('OVERALL: PASS - v0.3.9 draft prepared; human confirmation and final approval remain pending')
print('=' * 76)
```

```python
# Step 12-65 - Catalog Coverage, Completeness, and Approval-Readiness Reports
# Read-only analysis of v0.3.9. Metadata presence does not establish verification.

import csv
import hashlib
import io
import json
import re
from collections import Counter, defaultdict
from pathlib import Path
from tempfile import TemporaryDirectory

ROOT = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Run this cell inside the Champions Athletic Map Git project.')
RESULTS = ROOT / 'results'
INPUT = {
    'catalog': ROOT / 'data/processed/basketball_drill_catalog_v0_3_9_REVIEW_ONLY.csv',
    'registry': RESULTS / 'basketball_drill_source_registry_v0_3_9.csv',
    'draft_review': RESULTS / 'basketball_wave3_draft_metadata_review_v0_3_9.csv',
    'changes': RESULTS / 'basketball_wave3_draft_metadata_change_log_v0_3_9.csv',
    'linkage': RESULTS / 'basketball_source_linkage_audit_v0_3_9.csv',
    'canonical': RESULTS / 'basketball_canonical_reconciliation_status_v0_3_9.csv',
    'previous_summary': RESULTS / 'basketball_step12_64_summary_v0_3_9.json',
}
OUTPUT = {
    'coverage': RESULTS / 'eda/basketball_drill_coverage_v0_3_9.csv',
    'completeness': RESULTS / 'eda/basketball_metadata_completeness_v0_3_9.csv',
    'readiness': RESULTS / 'basketball_candidate_readiness_v0_3_9.csv',
    'schema': RESULTS / 'basketball_catalog_schema_review_v0_3_9.csv',
    'summary': RESULTS / 'basketball_step12_65_summary_v0_3_9.json',
}
PRIMARY_SKILLS = ('ball_handling', 'shooting', 'passing', 'finishing', 'footwork', 'defense')
CORE_FIELDS = ('equipment', 'participants', 'space', 'source_prescription')
SCOPES = ('all_candidates', 'basketball_drill_candidate', 'general_warmup_candidate')
STATES = ('blank_in_catalog', 'recorded_unspecified', 'reference_only',
          'recorded_requirement_uncertain', 'recorded_value_present')


def require(condition, message):
    if not condition:
        raise ValueError(message)


def sha256(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def rel(path):
    return path.relative_to(ROOT).as_posix()


def read_csv(path):
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    require(columns and len(columns) == len(set(columns)), f'Invalid CSV header: {path.name}')
    require(all(None not in row and all(v is not None for v in row.values()) for row in rows),
            f'Malformed CSV records: {path.name}')
    return rows, columns


def unique_index(rows, field, label):
    keys = [row[field] for row in rows]
    require(all(keys) and len(keys) == len(set(keys)), f'{label}: missing or repeated {field}.')
    return dict(zip(keys, rows))


def ids(value):
    return {part.strip() for part in value.split(';') if part.strip()}


def record_state(value):
    text = re.sub(r'\s+', ' ', value).strip().casefold().rstrip('.')
    if not text:
        return 'blank_in_catalog'
    if text in {'not specified', 'not stated', 'unspecified', 'unknown', 'not provided', 'n/a', 'na'}:
        return 'recorded_unspecified'
    if text in {'see source practice plan', 'see source', 'refer to source', 'see practice plan'}:
        return 'reference_only'
    if text == 'not explicitly required' or text.endswith('| not explicitly required'):
        return 'recorded_requirement_uncertain'
    return 'recorded_value_present'


def json_bytes(value):
    return (json.dumps(value, ensure_ascii=False, indent=2) + '\n').encode('utf-8')


def csv_bytes(rows):
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=list(rows[0]), lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')


for name, path in INPUT.items():
    if not path.is_file():
        raise FileNotFoundError(f'Missing {name}: {path}')
input_hashes = {rel(path): sha256(path) for path in INPUT.values()}
previous = json.loads(INPUT['previous_summary'].read_text(encoding='utf-8'))
require(previous.get('step') == '12-64' and previous.get('catalog_version') == '0.3.9'
        and previous.get('catalog_state') == 'review_only_draft_pending_human_confirmation'
        and previous.get('draft_metadata_changes') == 30
        and previous.get('human_confirmations_pending') == 68,
        'Expected the successful Step 12-64 review-only draft.')
checks = {k.replace('\\', '/'): v for k, v in previous.get('output_sha256', {}).items()}
for name, path in INPUT.items():
    if name != 'previous_summary':
        require(checks.get(rel(path)) == input_hashes[rel(path)], f'Step 12-64 output integrity mismatch: {name}.')
catalog, catalog_columns = read_csv(INPUT['catalog'])
registry, registry_columns = read_csv(INPUT['registry'])
draft_review, _ = read_csv(INPUT['draft_review'])
changes, _ = read_csv(INPUT['changes'])
linkage, _ = read_csv(INPUT['linkage'])
canonical, _ = read_csv(INPUT['canonical'])
require(tuple(map(len, (catalog, registry, draft_review, changes, linkage, canonical))) == (144, 156, 68, 30, 156, 144),
        'Input counts differ from the verified Step 12-64 scope.')
cat_by = unique_index(catalog, 'catalog_candidate_id', 'Catalog')
reg_by = unique_index(registry, 'evidence_id', 'Registry')
link_by = unique_index(linkage, 'evidence_id', 'Source linkage audit')
canonical_by = unique_index(canonical, 'catalog_candidate_id', 'Canonical status report')
require(set(reg_by) == set(link_by) and set(cat_by) == set(canonical_by), 'Audit coverage mismatch.')
require(all(r['catalog_version'] == '0.3.9' and r['catalog_approval_status'] == 'pending'
            and r['planner_eligible'] == 'no' and not r['approved_drill_id'] for r in catalog),
        'The catalog version or approval state changed; preserve and inspect it.')
linked = defaultdict(set)
for row in registry:
    cid, eid = row['catalog_candidate_id'], row['evidence_id']
    require(cid in cat_by and link_by[eid]['catalog_candidate_id'] == cid
            and link_by[eid]['structural_linkage_status'] == 'pass', f'Source audit mismatch: {eid}.')
    linked[cid].add(eid)
for row in catalog:
    cid = row['catalog_candidate_id']
    require(ids(row['evidence_ids']) | ids(row['aditi_evidence_ids']) == linked[cid]
            and int(row['total_evidence_records']) == len(linked[cid]), f'Source membership changed: {cid}.')
confirmation_by_candidate = Counter()
review_keys = set()
for row in draft_review:
    eid, field, cid = row['evidence_id'], row['field'], row['candidate_group_id']
    require((eid, field) not in review_keys and field in {'equipment', 'participants'}
            and eid in reg_by and cid in cat_by and reg_by[eid]['catalog_candidate_id'] == cid,
            'Draft-review record linkage changed.')
    review_keys.add((eid, field))
    require(row['reviewer_decision'] == 'pending' and row['human_confirmation_recorded'] == 'no'
            and not row['final_value'] and row['catalog_approval_status'] == 'pending'
            and row['planner_eligible'] == 'no'
            and row['draft_catalog_value'] == cat_by[cid][field] == reg_by[eid][field],
            f'Draft metadata or reviewer state changed: {eid}/{field}.')
    confirmation_by_candidate[cid] += 1
require(len(confirmation_by_candidate) == 38 and sum(confirmation_by_candidate.values()) == 68,
        'Expected 68 unconfirmed fields across 38 candidates.')
scope_rows = {scope: [r for r in catalog if scope == 'all_candidates' or r['scope_candidate'] == scope]
              for scope in SCOPES}
require(tuple(len(scope_rows[s]) for s in SCOPES) == (144, 132, 12), 'Candidate scope counts changed.')

# Coverage counts preserve the existing labels. Curriculum counts can overlap
# because a candidate can have source occurrences at multiple levels.
coverage = []
for scope, rows in scope_rows.items():
    dimensions = {
        'primary_skill_raw_label': sorted(set(PRIMARY_SKILLS) | {r['skill_area'] for r in rows}),
        'curriculum_level': ('Rookie', 'Starter', 'All-Star', 'MVP'),
        'catalog_approval_status': sorted({r['catalog_approval_status'] for r in rows}),
        'planner_eligible': sorted({r['planner_eligible'] for r in rows}),
    }
    for dimension, categories in dimensions.items():
        for category in categories:
            if dimension == 'primary_skill_raw_label':
                selected = [r for r in rows if r['skill_area'] == category]
            elif dimension == 'curriculum_level':
                selected = [r for r in rows if category in ids(r['curriculum_levels'])]
            else:
                selected = [r for r in rows if r[dimension] == category]
            coverage.append({
                'scope': scope, 'dimension': dimension, 'category': category,
                'candidate_groups': len(selected), 'scope_total_candidate_groups': len(rows),
                'approved_candidate_groups': sum(r['catalog_approval_status'] == 'approved' for r in selected),
                'planner_eligible_candidate_groups': sum(r['planner_eligible'] == 'yes' for r in selected),
                'count_basis': 'multiple_curriculum_memberships_allowed' if dimension == 'curriculum_level'
                               else 'one_existing_label_per_candidate',
                'limitation': 'Candidate coverage does not establish distinctness, approval, or athlete suitability.',
            })
    raw_skill_total = sum(r['candidate_groups'] for r in coverage
                          if r['scope'] == scope and r['dimension'] == 'primary_skill_raw_label')
    require(raw_skill_total == len(rows), f'Primary-skill coverage does not reconcile: {scope}.')

completeness = []
for scope, rows in scope_rows.items():
    for field in CORE_FIELDS:
        counts = Counter(record_state(r[field]) for r in rows)
        require(sum(counts.values()) == len(rows), f'Completeness counts do not reconcile: {scope}/{field}.')
        for state in STATES:
            completeness.append({
                'scope': scope, 'catalog_field': field, 'record_state': state,
                'candidate_groups': counts[state], 'scope_total_candidate_groups': len(rows),
                'verification_interpretation': 'Stored text classification only; source verification is assessed separately.',
            })

readiness = []
for row in catalog:
    cid = row['catalog_candidate_id']
    states = {field: record_state(row[field]) for field in CORE_FIELDS}
    nonspecific = [field for field, state in states.items() if state != 'recorded_value_present']
    blockers = []
    for field in ('requirements_review', 'permission_status', 'reproduction_permission',
                  'youth_planning_status', 'catalog_approval_status'):
        if row[field] == 'pending':
            blockers.append(field + '_pending')
    if row['distinctness_status'] in {
        'pending', 'repeated_source_variation_pending', 'preliminary_unique_name',
        'provisional_name_unique_semantic_review_pending', 'pending_source_and_variation_comparison'}:
        blockers.append('distinctness_not_finalized')
    if row['canonical_linkage_review_status'] == 'pending_human_review':
        blockers.append('explicit_identity_case_pending')
    if nonspecific:
        blockers.append('core_metadata_missing_or_nonspecific')
    if confirmation_by_candidate[cid]:
        blockers.append('wave3_field_confirmation_pending')
    readiness.append({
        'catalog_candidate_id': cid, 'drill_name': row['drill_name'],
        'scope_candidate': row['scope_candidate'], 'primary_skill_raw_label': row['skill_area'],
        'curriculum_levels': row['curriculum_levels'],
        **{field + '_record_state': states[field] for field in CORE_FIELDS},
        'nonspecific_core_fields': ';'.join(nonspecific),
        'equipment_text_may_describe_space': 'yes' if row['equipment'].strip().casefold()
            in {'basketball court', 'court', 'half court', 'full court', 'gym'} else 'no',
        'metadata_status': row['metadata_status'], 'heading_screen': row['heading_screen'],
        'section_screen': row['section_screen'], 'provenance_review': row['provenance_review'],
        'requirements_review': row['requirements_review'], 'distinctness_status': row['distinctness_status'],
        'source_identity_review_cases': row['source_identity_review_cases'],
        'canonical_linkage_review_status': row['canonical_linkage_review_status'],
        'permission_status': row['permission_status'], 'reproduction_permission': row['reproduction_permission'],
        'youth_planning_status': row['youth_planning_status'],
        'wave3_field_confirmations_pending': confirmation_by_candidate[cid],
        'catalog_approval_status': row['catalog_approval_status'], 'planner_eligible': row['planner_eligible'],
        'readiness_status': 'excluded_pending_review', 'recorded_review_blockers': ';'.join(blockers),
    })

# Schema coverage concerns these two tables only. An absent dedicated column
# does not establish that the information is absent from the source/project.
schema_specs = [
    ('candidate_identity', 'catalog', 'catalog_candidate_id', 'dedicated_column'),
    ('approved_drill_identity', 'catalog', 'approved_drill_id', 'dedicated_column'),
    ('source_identity', 'registry', 'evidence_id', 'dedicated_column'),
    ('primary_skill', 'catalog', 'skill_area', 'dedicated_column'),
    ('equipment', 'catalog', 'equipment', 'dedicated_column'),
    ('participants', 'catalog', 'participants', 'dedicated_column'),
    ('space', 'catalog', 'space', 'dedicated_column'),
    ('source_prescription', 'catalog', 'source_prescription', 'mixed_prescription_text'),
    ('source_url', 'registry', 'source_url', 'dedicated_column'),
    ('source_instructions', 'catalog', 'instructions', 'dedicated_column'),
    ('primary_goal', 'catalog', 'primary_goal', 'dedicated_column'),
    ('secondary_skills', 'catalog', 'secondary_skills', 'dedicated_column'),
    ('numeric_minimum_participants', 'catalog', 'minimum_participants', 'dedicated_column'),
    ('separate_duration', 'catalog', 'duration', 'dedicated_column'),
    ('separate_rest_interval', 'catalog', 'rest', 'dedicated_column'),
    ('difficulty_modifications', 'catalog', 'difficulty_modification', 'dedicated_column'),
    ('solo_adaptations', 'catalog', 'solo_adaptation', 'dedicated_column'),
    ('adaptation_provenance', 'catalog', 'adaptation_provenance', 'dedicated_column'),
]
schema = []
for concept, table, column, representation in schema_specs:
    rows, columns = (catalog, catalog_columns) if table == 'catalog' else (registry, registry_columns)
    present = column in columns
    schema.append({
        'planned_concept': concept, 'table_audited': table, 'expected_column': column,
        'representation': representation if present else 'no_dedicated_column_in_this_table',
        'column_present': 'yes' if present else 'no',
        'rows_in_table': len(rows),
        'nonblank_values': sum(bool(r[column].strip()) for r in rows) if present else '',
        'interpretation': ('Text presence does not establish verification or numeric duration.' if present
                           else 'May exist in source or other project artifacts; do not invent missing values.'),
    })

basketball = scope_rows['basketball_drill_candidate']
skill_counts = {skill: sum(r['skill_area'] == skill for r in basketball) for skill in PRIMARY_SKILLS}
blank_counts = {scope: {field: sum(not r[field].strip() for r in rows) for field in CORE_FIELDS}
                for scope, rows in scope_rows.items()}
unspecified_counts = {scope: {field: sum(record_state(r[field]) == 'recorded_unspecified' for r in rows)
                            for field in CORE_FIELDS} for scope, rows in scope_rows.items()}
open_identity = [r for r in catalog if r['canonical_linkage_review_status'] == 'pending_human_review']
open_cases = {case for r in open_identity for case in ids(r['source_identity_review_cases'])}
require((len(open_identity), len(open_cases)) == (9, 8), 'Explicit unresolved identity-case counts changed.')
expected_summary = {
    'step': '12-65', 'catalog_version': '0.3.9',
    'catalog_state': 'review_only_draft_pending_human_confirmation',
    'catalog_candidate_groups': 144, 'registered_source_occurrences': 156,
    'basketball_candidate_groups': len(basketball), 'general_warmup_candidate_groups': 12,
    'basketball_primary_skill_candidate_counts': skill_counts,
    'primary_skill_areas_with_candidates': sum(count > 0 for count in skill_counts.values()),
    'basketball_candidates_in_six_primary_labels': sum(skill_counts.values()),
    'basketball_candidates_with_other_existing_skill_labels': len(basketball) - sum(skill_counts.values()),
    'blank_core_metadata_by_scope': blank_counts,
    'recorded_unspecified_metadata_by_scope': unspecified_counts,
    'planned_concepts_without_dedicated_catalog_columns': [r['planned_concept'] for r in schema if r['column_present'] == 'no'],
    'candidate_readiness_rows': len(readiness), 'coverage_report_rows': len(coverage),
    'completeness_report_rows': len(completeness), 'schema_report_rows': len(schema),
    'human_field_confirmations_pending': 68, 'candidate_groups_with_wave3_confirmations_pending': 38,
    'explicit_identity_cases_pending': 8, 'explicit_identity_case_candidate_groups_pending': 9,
    'approved_drill_target': 100, 'approved_distinct_drills_verified': 0,
    'catalog_approvals': 0, 'planner_eligible': 0, 'final_approval_target_met': False,
    'catalog_modified': False, 'original_inputs_modified': False, 'git_operations_performed': False,
    'limitations': [
        'Candidate counts are not verified distinct or approved drill counts.',
        'Warm-up scope is excluded from basketball-only skill coverage and the 100-drill target.',
        'Curriculum memberships overlap; raw skill labels are preserved without automatic remapping.',
        'Nonblank fields and recorded absence assertions do not establish independent source verification.',
        'Mixed prescriptions are not converted into numeric durations or minimum participant counts.',
        'Schema gaps apply to these tables; source evidence may be available elsewhere.',
        'The 68 human confirmations concern the current Wave 3 packet, not all outstanding catalog work.',
        'These reports do not authorize retrieval, public reuse, or athlete-specific recommendations.',
    ],
    'input_sha256': input_hashes,
}
new_files = {'coverage': csv_bytes(coverage), 'completeness': csv_bytes(completeness),
             'readiness': csv_bytes(readiness), 'schema': csv_bytes(schema)}
print('INPUT INTEGRITY: PASS - all six Step 12-64 outputs verified.')
existing = [path.exists() for path in OUTPUT.values()]
if any(existing) and not all(existing):
    raise FileExistsError('Only part of the Step 12-65 output set exists. Inspect it before rerunning.')
if all(existing):
    summary = json.loads(OUTPUT['summary'].read_text(encoding='utf-8'))
    require(all(summary.get(key) == value for key, value in expected_summary.items()),
            'Existing Step 12-65 summary differs from the verified report state.')
    for name, data in new_files.items():
        path = OUTPUT[name]
        require(summary.get('output_sha256', {}).get(rel(path)) == sha256(path)
                and path.read_bytes() == data, f'Existing report changed: {rel(path)}. It has been preserved.')
    mode = 'PRESERVED'
else:
    summary = dict(expected_summary)
    summary['output_sha256'] = {rel(OUTPUT[name]): hashlib.sha256(data).hexdigest() for name, data in new_files.items()}
    new_files['summary'] = json_bytes(summary)
    with TemporaryDirectory(prefix='step12_65_', dir=ROOT) as staging_dir:
        staged = {}
        for name, data in new_files.items():
            staged[name] = Path(staging_dir) / OUTPUT[name].name
            staged[name].write_bytes(data)
            require(staged[name].read_bytes() == data, f'Staging verification failed: {name}.')
        require(all(sha256(path) == input_hashes[rel(path)] for path in INPUT.values()),
                'An input changed during report preparation. No outputs created.')
        if any(path.exists() for path in OUTPUT.values()):
            raise FileExistsError('A destination appeared during staging. Preserve and inspect it.')
        created = []
        try:
            for name, path in OUTPUT.items():
                path.parent.mkdir(parents=True, exist_ok=True)
                with path.open('xb') as handle:
                    created.append(path)
                    handle.write(staged[name].read_bytes())
        except Exception:
            for path in created:
                path.unlink(missing_ok=True)
            raise
    mode = 'CREATED'
require(all(sha256(path) == input_hashes[rel(path)] for path in INPUT.values()), 'An original input changed.')
for name, path in OUTPUT.items():
    if name != 'summary':
        require(summary['output_sha256'].get(rel(path)) == sha256(path), f'Saved report verification failed: {name}.')

print('=' * 76)
print('STEP 12-65 - COVERAGE, COMPLETENESS, AND APPROVAL-READINESS REPORTS')
print('=' * 76)
print('Catalog candidates / basketball candidates / warm-ups: 144 / 132 / 12')
print('BASKETBALL-ONLY PRIMARY SKILL COVERAGE - CANDIDATE GROUPS')
for skill, count in skill_counts.items():
    print(f'  {skill:<16} {count}')
print(f'Primary skill areas represented: {sum(c > 0 for c in skill_counts.values())}/6')
print(f'Candidates in those six labels / other basketball skill labels: {sum(skill_counts.values())} / {len(basketball) - sum(skill_counts.values())}')
print('BLANK FIELDS IN BASKETBALL CANDIDATES - OUT OF 132')
for field, count in blank_counts['basketball_drill_candidate'].items():
    print(f'  {field:<22} {count}')
print('Recorded unspecified basketball prescriptions: ' + str(unspecified_counts['basketball_drill_candidate']['source_prescription']))
print('Text presence is not verification; numeric durations and participant minima were not inferred.')
print('Human field confirmations / explicit identity cases pending: 68 / 8')
print('Approved distinct drills / target / planner-eligible: 0 / 100 / 0')
print('Final 100-approved-drill target: NOT YET MET')
print('OUTPUTS')
for name, path in OUTPUT.items():
    print(f'  {mode:<9} {rel(path)}')
print('Catalog, source registry, previous reports, and approval decisions modified: No')
print('Git operations performed: None')
print('Next: Step 12-66 final integrity audit, observations, and Step 13 handoff.')
print('OVERALL: PASS - reports verified; final drill approvals remain open')
print('=' * 76)
```

```python
# Step 12-66 - Final Integration Checkpoint Audit, Observations, and Step 13 Handoff
# This audits saved artifacts. It does not approve drills or rerun the whole notebook.

import csv
import hashlib
import io
import json
import re
import unicodedata
from collections import Counter, defaultdict
from pathlib import Path
from tempfile import TemporaryDirectory

ROOT = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Run this cell inside the Champions Athletic Map Git project.')
RESULTS = ROOT / 'results'
MANIFESTS = {
    '12-63': RESULTS / 'basketball_step12_63_summary_v0_3_8.json',
    '12-64': RESULTS / 'basketball_step12_64_summary_v0_3_9.json',
    '12-65': RESULTS / 'basketball_step12_65_summary_v0_3_9.json',
}
OUTPUT = {
    'audit': RESULTS / 'basketball_step12_final_audit_v0_3_9.csv',
    'observations': RESULTS / 'basketball_step12_observations_v0_3_9.md',
    'handoff': RESULTS / 'basketball_step12_step13_handoff_v0_3_9.json',
    'summary': RESULTS / 'basketball_step12_66_summary_v0_3_9.json',
}
SKILLS = ('ball_handling', 'shooting', 'passing', 'finishing', 'footwork', 'defense')
CORE = ('equipment', 'participants', 'space', 'source_prescription')
SCOPES = ('all_candidates', 'basketball_drill_candidate', 'general_warmup_candidate')
STATES = ('blank_in_catalog', 'recorded_unspecified', 'reference_only',
          'recorded_requirement_uncertain', 'recorded_value_present')
TRACE = ('wave3_requirement_draft_status', 'wave3_requirement_draft_fields', 'wave3_requirement_draft_reference')
audit_rows = []


def require(condition, message):
    if not condition:
        raise ValueError(message)


def sha256(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def rel(path):
    return path.relative_to(ROOT).as_posix()


def local_path(reference):
    reference = reference.replace('\\', '/')
    require(not Path(reference).is_absolute(), 'Manifest paths must be relative to the project.')
    path = (ROOT / reference).resolve()
    require(path.is_relative_to(ROOT) and path.is_file(), f'Missing or invalid manifest artifact: {reference}.')
    return path


def check(check_id, description, observed, expected, evidence):
    require(observed == expected, f'Final audit failed [{check_id}]: {description}')
    audit_rows.append({'check_id': check_id, 'description': description,
                       'observed': json.dumps(observed, sort_keys=True, ensure_ascii=False),
                       'expected': json.dumps(expected, sort_keys=True, ensure_ascii=False),
                       'status': 'PASS', 'evidence_reference': evidence})


def read_csv(reference):
    path = local_path(reference)
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    require(columns and len(columns) == len(set(columns)), f'Invalid CSV header: {reference}.')
    require(all(None not in row and all(v is not None for v in row.values()) for row in rows),
            f'Malformed CSV record: {reference}.')
    return rows, columns


def unique(rows, field):
    keys = [r[field] for r in rows]
    require(all(keys) and len(keys) == len(set(keys)), f'Missing or duplicate {field}.')
    return dict(zip(keys, rows))


def ids(text):
    return {part.strip() for part in text.split(';') if part.strip()}


def fingerprint(text):
    value = re.sub(r'[^a-z0-9]+', ' ', unicodedata.normalize('NFKC', text).casefold()).strip()
    return hashlib.sha256(value.encode('utf-8')).hexdigest()


def state(value):
    text = re.sub(r'\s+', ' ', value).strip().casefold().rstrip('.')
    if not text:
        return 'blank_in_catalog'
    if text in {'not specified', 'not stated', 'unspecified', 'unknown', 'not provided', 'n/a', 'na'}:
        return 'recorded_unspecified'
    if text in {'see source practice plan', 'see source', 'refer to source', 'see practice plan'}:
        return 'reference_only'
    if text == 'not explicitly required' or text.endswith('| not explicitly required'):
        return 'recorded_requirement_uncertain'
    return 'recorded_value_present'


def json_bytes(value):
    return (json.dumps(value, ensure_ascii=False, indent=2) + '\n').encode('utf-8')


def csv_bytes(rows):
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=list(rows[0]), lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')


# Verify every input/output declared by the three scoped manifests. Earlier
# historical cells are outside this audit, even when their saved files are inputs.
manifests = {}
expected_hashes = {}
for step, path in MANIFESTS.items():
    require(path.is_file(), f'Missing summary: {path}.')
    manifest = json.loads(path.read_text(encoding='utf-8'))
    require(manifest.get('step') == step, f'Manifest step mismatch: {step}.')
    manifests[step] = manifest
    for kind in ('input_sha256', 'output_sha256'):
        mapping = manifest.get(kind, {})
        require(mapping, f'Missing {kind}: {step}.')
        for reference, digest in mapping.items():
            reference = reference.replace('\\', '/')
            require(re.fullmatch(r'[0-9a-f]{64}', digest), f'Invalid checksum: {reference}.')
            require(reference not in expected_hashes or expected_hashes[reference] == digest,
                    f'Conflicting manifest hashes: {reference}.')
            expected_hashes[reference] = digest
tracked = {reference: local_path(reference) for reference in expected_hashes}
for index, reference in enumerate(sorted(tracked), start=1):
    check(f'ARTIFACT_{index:03}', 'Manifest-declared artifact matches SHA-256',
          sha256(tracked[reference]), expected_hashes[reference], reference)
for path in MANIFESTS.values():
    tracked[rel(path)] = path
input_hashes = {reference: sha256(path) for reference, path in sorted(tracked.items())}
require(not (set(input_hashes) & {rel(p) for p in OUTPUT.values()}), 'Audit inputs overlap output destinations.')
check('MANIFEST_VERSIONS', 'Scoped manifests identify the expected catalog versions',
      [manifests[s].get('catalog_version') for s in MANIFESTS], ['0.3.8', '0.3.9', '0.3.9'],
      ';'.join(rel(p) for p in MANIFESTS.values()))

catalog_ref = 'data/processed/basketball_drill_catalog_v0_3_9_REVIEW_ONLY.csv'
registry_ref = 'results/basketball_drill_source_registry_v0_3_9.csv'
review_ref = 'results/basketball_wave3_draft_metadata_review_v0_3_9.csv'
old_catalog_ref = 'data/processed/basketball_drill_catalog_v0_3_8_REVIEW_ONLY.csv'
old_registry_ref = 'results/basketball_drill_source_registry_v0_3_8.csv'
proposal_ref = 'results/basketball_wave3_consolidated_field_review_v0_3_8.csv'
catalog, cat_columns = read_csv(catalog_ref)
registry, reg_columns = read_csv(registry_ref)
old_catalog, old_cat_columns = read_csv(old_catalog_ref)
old_registry, old_reg_columns = read_csv(old_registry_ref)
proposals, proposal_columns = read_csv(proposal_ref)
draft_review, _ = read_csv(review_ref)
changes, _ = read_csv('results/basketball_wave3_draft_metadata_change_log_v0_3_9.csv')
linkage, _ = read_csv('results/basketball_source_linkage_audit_v0_3_9.csv')
canonical, _ = read_csv('results/basketball_canonical_reconciliation_status_v0_3_9.csv')
coverage, _ = read_csv('results/eda/basketball_drill_coverage_v0_3_9.csv')
completeness, _ = read_csv('results/eda/basketball_metadata_completeness_v0_3_9.csv')
readiness, _ = read_csv('results/basketball_candidate_readiness_v0_3_9.csv')
schema, _ = read_csv('results/basketball_catalog_schema_review_v0_3_9.csv')
cat_by, reg_by = unique(catalog, 'catalog_candidate_id'), unique(registry, 'evidence_id')
link_by, canonical_by = unique(linkage, 'evidence_id'), unique(canonical, 'catalog_candidate_id')
ready_by = unique(readiness, 'catalog_candidate_id')
check('RECORD_COUNTS', 'Current catalog, source, field, change, and report row counts',
      [len(catalog), len(registry), len(proposals), len(draft_review), len(changes), len(linkage),
       len(canonical), len(readiness)], [144, 156, 68, 68, 30, 156, 144, 144], catalog_ref)
check('VERSIONS', 'All current catalog rows are review-only v0.3.9',
      all(r['catalog_version'] == '0.3.9' for r in catalog), True, catalog_ref)
check('REPORT_IDENTIFIERS', 'Linkage, canonical, and readiness reports cover the same identifiers',
      set(reg_by) == set(link_by) and set(cat_by) == set(canonical_by) == set(ready_by), True, registry_ref)

# Deterministically reconstruct the new catalog and registry from saved inputs.
# This replay uses no network, source extraction, model calls, or approval edits.
replay_catalog = [dict(r) for r in old_catalog]
replay_registry = [dict(r) for r in old_registry]
replay_cat_by = unique(replay_catalog, 'catalog_candidate_id')
replay_reg_by = unique(replay_registry, 'evidence_id')
proposal_by = {(r['evidence_id'], r['field']): r for r in proposals}
require(len(proposal_by) == 68, 'Duplicate proposal fields.')
source_snapshot_ref = 'results/basketball_wave3_consolidated_source_snapshot_v0_3_8.json'
snapshot = json.loads(local_path(source_snapshot_ref).read_text(encoding='utf-8'))
pages = unique(snapshot['pages'], 'source_url')
fields_by_candidate, fields_by_source = defaultdict(set), defaultdict(set)
expected_changes = set()
for row in proposals:
    eid, cid, field = row['evidence_id'], row['candidate_group_id'], row['field']
    require(eid in replay_reg_by and cid in replay_cat_by and field in {'equipment', 'participants'},
            'Invalid proposal linkage during replay.')
    require(replay_reg_by[eid]['catalog_candidate_id'] == cid
            and row['current_value'] == replay_cat_by[cid][field] == replay_reg_by[eid][field],
            f'Replay original value mismatch: {eid}/{field}.')
    section = pages[row['source_url']]['sections'][row['drill_name']]
    require(row['source_evidence_status'] == 'matches_reviewed_section'
            and fingerprint(section['text']) == section['section_sha256']
            == row['expected_section_sha256'] == row['observed_section_sha256'],
            f'Source fingerprint mismatch during replay: {eid}/{field}.')
    require(row['proposed_action'] in {'retain_current', 'revise_value'} and row['proposed_value'],
            'Unsupported replay action.')
    require((row['proposed_action'] == 'retain_current') == (row['current_value'] == row['proposed_value']),
            'Replay action/value mismatch.')
    replay_cat_by[cid][field] = row['proposed_value']
    replay_reg_by[eid][field] = row['proposed_value']
    fields_by_candidate[cid].add(field)
    fields_by_source[eid].add(field)
    if row['proposed_value'] != row['current_value']:
        expected_changes.add((cid, eid, field, row['current_value'], row['proposed_value']))
for rows, identity, field_map in ((replay_catalog, 'catalog_candidate_id', fields_by_candidate),
                                   (replay_registry, 'evidence_id', fields_by_source)):
    for row in rows:
        fields = field_map.get(row[identity], set())
        row[TRACE[0]] = 'source_matched_proposal_applied_to_draft' if fields else ''
        row[TRACE[1]] = ';'.join(sorted(fields))
        row[TRACE[2]] = review_ref if fields else ''
for row in replay_catalog:
    row['catalog_version'] = '0.3.9'
check('CATALOG_REPLAY', 'Every catalog value reconstructs from v0.3.8 and the saved proposals',
      replay_catalog == catalog and old_cat_columns + list(TRACE) == cat_columns, True, catalog_ref)
check('REGISTRY_REPLAY', 'Every registry value reconstructs while preserving original source metadata',
      replay_registry == registry and old_reg_columns + list(TRACE) == reg_columns, True, registry_ref)
actual_changes = {(r['candidate_group_id'], r['evidence_id'], r['field'],
                   r['original_v0_3_8_value'], r['draft_v0_3_9_value']) for r in changes}
check('EXACT_CHANGES', 'The change log records exactly the 30 draft revisions',
      actual_changes == expected_changes and len(actual_changes) == 30, True,
      'results/basketball_wave3_draft_metadata_change_log_v0_3_9.csv')
check('ACTION_COUNTS', 'The source review contains 38 retains and 30 revisions',
      dict(Counter(r['proposed_action'] for r in proposals)), {'retain_current': 38, 'revise_value': 30}, proposal_ref)
confirmation_counts = Counter()
draft_keys = [(r['evidence_id'], r['field']) for r in draft_review]
check('DRAFT_REVIEW_KEYS', 'The draft ledger preserves every proposed field exactly once',
      len(set(draft_keys)) == 68 and set(draft_keys) == set(proposal_by), True, review_ref)
for row in draft_review:
    key = row['evidence_id'], row['field']
    require(key in proposal_by and all(row[column] == proposal_by[key][column] for column in proposal_columns),
            'Draft-review ledger differs from the saved proposal.')
    require(row['draft_catalog_value'] == cat_by[row['candidate_group_id']][row['field']]
            and row['reviewer_decision'] == 'pending' and row['human_confirmation_recorded'] == 'no'
            and not row['final_value'] and not row['reviewer'] and not row['review_date'],
            'A draft value or human-confirmation status changed.')
    confirmation_counts[row['candidate_group_id']] += 1
check('PENDING_CONFIRMATIONS', 'All 68 field confirmations remain pending across 38 candidates',
      [sum(confirmation_counts.values()), len(confirmation_counts)], [68, 38], review_ref)

source_membership = defaultdict(set)
for row in registry:
    cid, eid = row['catalog_candidate_id'], row['evidence_id']
    require(cid in cat_by and link_by[eid]['catalog_candidate_id'] == cid
            and link_by[eid]['structural_linkage_status'] == 'pass', 'Registry/audit linkage mismatch.')
    source_membership[cid].add(eid)
for row in catalog:
    cid = row['catalog_candidate_id']
    require(ids(row['evidence_ids']) | ids(row['aditi_evidence_ids']) == source_membership[cid]
            and int(row['source_occurrences']) == len(ids(row['evidence_ids']))
            and int(row['total_evidence_records']) == len(source_membership[cid]),
            f'Primary/imported evidence union mismatch: {cid}.')
check('SOURCE_UNION', 'Deduplicated primary/imported membership covers all 156 source occurrences',
      sum(len(v) for v in source_membership.values()), 156, registry_ref)
identity_rows = [r for r in catalog if r['canonical_linkage_review_status'] == 'pending_human_review']
identity_cases = {case for row in identity_rows for case in ids(row['source_identity_review_cases'])}
check('IDENTITY_CASES', 'Existing unresolved identity cases remain explicit',
      [len(identity_cases), len(identity_rows)], [8, 9], 'results/basketball_canonical_reconciliation_status_v0_3_9.csv')
for row in catalog:
    old_status = canonical_by[row['catalog_candidate_id']]
    require(all(old_status[field] == row[field] for field in ('distinctness_status', 'canonical_linkage_review_status',
            'permission_status', 'reproduction_permission', 'youth_planning_status', 'requirements_review')),
            'Canonical status report differs from preserved catalog review states.')

s65 = manifests['12-65']
scope_rows = {scope: [r for r in catalog if scope == 'all_candidates' or r['scope_candidate'] == scope]
              for scope in SCOPES}
check('SCOPES', 'Basketball and warm-up scopes remain separate',
      [len(scope_rows[s]) for s in SCOPES], [144, 132, 12], catalog_ref)
basketball = scope_rows['basketball_drill_candidate']
skills = {skill: sum(r['skill_area'] == skill for r in basketball) for skill in SKILLS}
check('SKILLS', 'Basketball-only primary-skill counts reproduce the coverage summary',
      skills, s65['basketball_primary_skill_candidate_counts'], 'results/eda/basketball_drill_coverage_v0_3_9.csv')
check('SKILL_CANDIDATE_GROUPS', 'Raw skill labels reconcile without counting warm-ups',
      [sum(skills.values()), len(basketball) - sum(skills.values()), sum(n > 0 for n in skills.values())],
      [97, 35, 6], catalog_ref)
coverage_keys = {(r['scope'], r['dimension'], r['category']) for r in coverage}
require(len(coverage_keys) == len(coverage), 'Duplicate coverage rows.')
expected_coverage_keys = set()
for scope, rows in scope_rows.items():
    dimensions = {
        'primary_skill_raw_label': sorted(set(SKILLS) | {r['skill_area'] for r in rows}),
        'curriculum_level': ('Rookie', 'Starter', 'All-Star', 'MVP'),
        'catalog_approval_status': sorted({r['catalog_approval_status'] for r in rows}),
        'planner_eligible': sorted({r['planner_eligible'] for r in rows}),
    }
    for dimension, categories in dimensions.items():
        for category in categories:
            expected_coverage_keys.add((scope, dimension, category))
for row in coverage:
    scope, dimension, category = row['scope'], row['dimension'], row['category']
    rows = scope_rows[scope]
    if dimension == 'primary_skill_raw_label':
        selected = [r for r in rows if r['skill_area'] == category]
    elif dimension == 'curriculum_level':
        selected = [r for r in rows if category in ids(r['curriculum_levels'])]
    else:
        require(dimension in {'catalog_approval_status', 'planner_eligible'}, 'Unknown coverage dimension.')
        selected = [r for r in rows if r[dimension] == category]
    require(int(row['candidate_groups']) == len(selected) and int(row['scope_total_candidate_groups']) == len(rows)
            and int(row['approved_candidate_groups']) == sum(r['catalog_approval_status'] == 'approved' for r in selected)
            and int(row['planner_eligible_candidate_groups']) == sum(r['planner_eligible'] == 'yes' for r in selected),
            'Coverage report denominator or count mismatch.')
check('COVERAGE_REPORT', 'All coverage categories and counts reconstruct from the catalog',
      coverage_keys == expected_coverage_keys, True, 'results/eda/basketball_drill_coverage_v0_3_9.csv')
blank_counts = {scope: {field: sum(not r[field].strip() for r in rows) for field in CORE}
                for scope, rows in scope_rows.items()}
unspecified = {scope: {field: sum(state(r[field]) == 'recorded_unspecified' for r in rows) for field in CORE}
               for scope, rows in scope_rows.items()}
check('BLANK_METADATA', 'Blank-field counts reproduce across all three scopes', blank_counts,
      s65['blank_core_metadata_by_scope'], 'results/eda/basketball_metadata_completeness_v0_3_9.csv')
check('UNSPECIFIED_METADATA', 'Recorded unspecified assertions stay distinct from blank fields', unspecified,
      s65['recorded_unspecified_metadata_by_scope'], 'results/eda/basketball_metadata_completeness_v0_3_9.csv')
expected_completeness = {(scope, field, label) for scope in SCOPES for field in CORE for label in STATES}
actual_completeness = {(r['scope'], r['catalog_field'], r['record_state']) for r in completeness}
check('COMPLETENESS_COVERAGE', 'The completeness report covers every field/state/scope combination once',
      actual_completeness == expected_completeness and len(completeness) == len(expected_completeness), True,
      'results/eda/basketball_metadata_completeness_v0_3_9.csv')
for row in completeness:
    rows = scope_rows[row['scope']]
    require(int(row['candidate_groups']) == sum(state(r[row['catalog_field']]) == row['record_state'] for r in rows)
            and int(row['scope_total_candidate_groups']) == len(rows), 'Completeness report value mismatch.')
for row in catalog:
    ready = ready_by[row['catalog_candidate_id']]
    require(all(ready[field + '_record_state'] == state(row[field]) for field in CORE)
            and int(ready['wave3_field_confirmations_pending']) == confirmation_counts[row['catalog_candidate_id']]
            and ready['readiness_status'] == 'excluded_pending_review'
            and all(ready[field] == row[field] for field in ('catalog_approval_status', 'planner_eligible',
                'permission_status', 'reproduction_permission', 'youth_planning_status', 'requirements_review',
                'distinctness_status', 'canonical_linkage_review_status')),
            'Candidate-readiness report differs from the catalog or pending field ledger.')
check('READINESS_REPORT', 'All candidate-readiness rows preserve recorded gates and field states',
      len(readiness), 144, 'results/basketball_candidate_readiness_v0_3_9.csv')
missing_concepts = []
for row in schema:
    rows, columns = (catalog, cat_columns) if row['table_audited'] == 'catalog' else (registry, reg_columns)
    present = row['expected_column'] in columns
    require(row['column_present'] == ('yes' if present else 'no') and int(row['rows_in_table']) == len(rows),
            'Schema report presence or denominator mismatch.')
    require(row['nonblank_values'] == (str(sum(bool(r[row['expected_column']].strip()) for r in rows))
                                      if present else ''), 'Schema report nonblank count mismatch.')
    if not present:
        missing_concepts.append(row['planned_concept'])
check('SCHEMA_GAPS', 'Schema gaps are documented without inventing dedicated values', missing_concepts,
      s65['planned_concepts_without_dedicated_catalog_columns'], 'results/basketball_catalog_schema_review_v0_3_9.csv')
approved = sum(r['catalog_approval_status'] == 'approved' for r in catalog)
eligible = sum(r['planner_eligible'] == 'yes' for r in catalog)
check('APPROVAL_GATES', 'No record is approved, planner eligible, or assigned an approved identity',
      [approved, eligible, sum(bool(r['approved_drill_id']) for r in catalog)], [0, 0, 0], catalog_ref)
check('PENDING_GATES', 'Requirements, permission, reproduction, and youth-planning reviews remain pending',
      all(r[field] == 'pending' for r in catalog for field in
          ('requirements_review', 'permission_status', 'reproduction_permission', 'youth_planning_status')),
      True, catalog_ref)
check('TARGET_STATUS', 'Final approval target is explicitly unmet',
      [s65['approved_drill_target'], s65['approved_distinct_drills_verified'], s65['final_approval_target_met']],
      [100, 0, False], rel(MANIFESTS['12-65']))
check('INPUT_PRESERVATION', 'All scoped input files retain their initial checksums',
      all(sha256(path) == input_hashes[reference] for reference, path in tracked.items()), True,
      'All manifest-scoped input artifacts')

pending_tasks = [
    {'task': 'Confirm the 68 Wave 3 field proposals', 'count': 68, 'unit': 'fields',
     'status': 'pending_human_confirmation', 'evidence': review_ref,
     'required_before': 'Adopting the draft metadata as confirmed; approval gates remain separate.'},
    {'task': 'Resolve documented identity comparisons', 'count': 8, 'unit': 'cases_across_9_candidate_groups',
     'status': 'pending', 'evidence': 'results/basketball_canonical_reconciliation_status_v0_3_9.csv',
     'required_before': 'Claiming a final distinct canonical drill set.'},
    {'task': 'Complete source-backed instructional and practical metadata review',
     'status': 'pending', 'evidence': 'results/eda/basketball_metadata_completeness_v0_3_9.csv',
     'blank_basketball_fields': blank_counts['basketball_drill_candidate'],
     'required_before': 'Using any affected record for planning; preserve genuinely unspecified information.'},
    {'task': 'Link essential planning fields to existing source evidence or add appropriate schema fields',
     'status': 'pending', 'evidence': 'results/basketball_catalog_schema_review_v0_3_9.csv',
     'concepts_without_dedicated_columns': missing_concepts,
     'required_before': 'Planner/RAG integration; absent columns do not establish absent source information.'},
    {'task': 'Finalize broader distinctness, source permissions, and intended-use reviews',
     'status': 'pending', 'evidence': 'results/basketball_candidate_readiness_v0_3_9.csv',
     'required_before': 'Final catalog approval and any intended public reuse.'},
    {'task': 'Prepare and review youth participation and planning rules', 'status': 'next_step_13',
     'required_before': 'Evaluating youth-rule compatibility or enabling planner use.'},
    {'task': 'Establish at least 100 distinct, source-verified, approved basketball drills',
     'target': 100, 'current_approved': approved, 'status': 'not_achieved',
     'required_before': 'Declaring the full Step 12 drill-approval objective complete.'},
]
handoff = {
    'from_step': '12', 'to_step': '13 - Youth Participation Guidelines and Planning Rules',
    'catalog_version': '0.3.9', 'integration_checkpoint_status': 'passed',
    'full_step12_complete': False, 'final_approval_target_met': False,
    'rule_development_can_proceed': True, 'planner_use_of_unapproved_catalog_allowed': False,
    'catalog_approved_by_this_step': False, 'rules_adopted_by_this_step': 0,
    'catalog_reference': catalog_ref, 'source_registry_reference': registry_ref,
    'coverage_reference': 'results/eda/basketball_drill_coverage_v0_3_9.csv',
    'readiness_reference': 'results/basketball_candidate_readiness_v0_3_9.csv',
    'observations_reference': rel(OUTPUT['observations']), 'audit_reference': rel(OUTPUT['audit']),
    'pending_tasks': pending_tasks,
    'step13_work': [
        'Inventory existing guideline records and verify their sources and interpretations.',
        'Keep published guidance, user constraints, and project policies distinguishable.',
        'Record applicability, units, qualifications, provenance, and review status for each rule.',
        'Handle age/grade ambiguities and missing information without an automatic pass.',
        'Plan boundary, conflict, missing-data, and schedule-accounting tests.',
        'Keep catalog candidates excluded from planner use until required approvals are complete.',
    ],
    'audit_scope': 'Saved Steps 12-63 through 12-65 manifest-declared artifacts, draft replay, and report checks.',
    'not_performed': ['Full notebook clean execution', 'Live source refresh',
                      'Rerunning historical Steps 12-1 through 12-62', 'Final content/permission/youth approval'],
}
skill_table = '\n'.join(f"| {skill.replace('_', ' ').title()} | {count} |" for skill, count in skills.items())
blank_table = '\n'.join(f'| {field} | {count} |'
                         for field, count in blank_counts['basketball_drill_candidate'].items())
observations = f'''### Observations - Step 12: Audited Catalog Integration Checkpoint

We prepared and audited Basketball Drill Catalog v0.3.9 as a **review-only draft**. It contains **144 candidate groups**, comprising **132 basketball candidates** and **12 general warm-up candidates**, linked to **156 source occurrences**. These are candidate and source counts; they are not counts of approved, distinct drills.

We checked the saved input/output checksums declared by Steps 12-63 through 12-65, reconstructed the draft catalog and registry from v0.3.8 and the saved proposals, and reconciled the coverage, completeness, schema, and readiness reports. The reconstruction reproduced all 30 draft metadata changes and preserved existing identities, source information, earlier review states, and approval controls.

#### Consolidated requirement review

The consolidated Wave 3 review covers **68 equipment/participant fields across 38 candidates**, with **38 retain recommendations** and **30 proposed revisions**. The revisions are present in the draft catalog and registry. All **68 human field confirmations remain pending**. Draft application does not establish team signoff, permission approval, youth-rule compatibility, or final catalog approval.

The source-membership audit accounts for all **156 source occurrences** using the deduplicated union of primary and imported evidence identifiers. Structural consistency does not establish semantic equivalence. **Eight explicit identity cases across nine candidate groups** remain unresolved, alongside the broader distinctness work recorded in the catalog.

#### Basketball skill coverage

All six primary skill areas have basketball candidates. Warm-ups are excluded from this table.

| Existing primary skill label | Basketball candidate groups |
|---|---:|
{skill_table}

The six labels account for **97 basketball candidate groups**. Another **35 basketball candidates** retain other existing skill labels. We did not automatically remap these labels or count warm-ups toward the 100-drill target. Curriculum-level coverage permits multiple memberships and should not be added as though the categories were exclusive.

#### Metadata and schema limitations

The basketball-only completeness audit records the following blank fields out of 132 candidate groups:

| Catalog field | Candidates with a blank value |
|---|---:|
{blank_table}

Another **three basketball prescription fields** record unspecified information. These assertions are distinct from blanks and remain subject to source review. Nonblank text does not establish verification. We did not infer numeric durations, rest intervals, or minimum participant counts from ambiguous descriptions.

Nine planned concepts have no dedicated column in the audited catalog schema: instructions, primary goal, secondary skills, numeric minimum participants, separate duration and rest, difficulty modifications, solo adaptations, and adaptation provenance. Relevant information may exist in other project artifacts or sources. The schema report documents these boundaries without inventing values.

#### Completion status and next work

The **saved-artifact integration checkpoint passed**, and we can proceed to **Step 13: Youth Participation Guidelines and Planning Rules**. The full Step 12 approval objective remains open: **0 approved drills out of the target of at least 100**, and **0 planner-eligible records**.

Requirements, permissions, reproduction permissions, youth-planning compatibility, and final approval remain pending. The handoff preserves the field confirmations, identity cases, missing metadata, schema work, and final approval target. Step 13 rule development does not authorize planner use of this draft catalog.

This audit establishes reproducibility from the scoped saved artifacts and deterministic draft reconstruction. It does not establish a complete clean notebook run, a fresh source-site review, a rerun of all historical cells, or independent coaching/content approval. Previous notebook observations describing v0.2 are historical and should not be used as the current catalog status.
'''
expected_summary = {
    'step': '12-66', 'catalog_version': '0.3.9',
    'audit_protocol_version': 1, 'audit_scope': handoff['audit_scope'],
    'integration_checkpoint_status': 'passed', 'step13_rule_development_can_proceed': True,
    'full_step12_complete': False, 'final_approval_target_met': False,
    'audit_checks_passed': len(audit_rows), 'audit_checks_failed': 0,
    'manifest_scoped_artifacts_tracked': len(tracked),
    'draft_reconstruction_verified': True, 'coverage_and_completeness_recomputed': True,
    'full_notebook_clean_execution_performed': False, 'live_source_refresh_performed': False,
    'catalog_candidate_groups': 144, 'basketball_candidate_groups': 132, 'general_warmup_candidate_groups': 12,
    'registered_source_occurrences': 156, 'draft_metadata_changes': 30,
    'human_field_confirmations_pending': 68, 'explicit_identity_cases_pending': 8,
    'explicit_identity_case_candidate_groups_pending': 9,
    'basketball_primary_skill_candidate_counts': skills,
    'blank_basketball_core_fields': blank_counts['basketball_drill_candidate'],
    'catalog_approvals': approved, 'approved_drill_target': 100, 'planner_eligible': eligible,
    'catalog_modified': False, 'original_inputs_modified': False, 'git_operations_performed': False,
    'input_sha256': input_hashes,
}
new_files = {'audit': csv_bytes(audit_rows), 'observations': observations.encode('utf-8'),
             'handoff': json_bytes(handoff)}
existing = [path.exists() for path in OUTPUT.values()]
if any(existing) and not all(existing):
    raise FileExistsError('Only part of the Step 12-66 output set exists. Inspect it before rerunning.')
if all(existing):
    summary = json.loads(OUTPUT['summary'].read_text(encoding='utf-8'))
    require(all(summary.get(key) == value for key, value in expected_summary.items()),
            'Existing final-audit summary differs from the verified checkpoint state.')
    for name, data in new_files.items():
        path = OUTPUT[name]
        require(summary.get('output_sha256', {}).get(rel(path)) == sha256(path)
                and path.read_bytes() == data, f'Existing final-audit output changed: {rel(path)}. Preserved.')
    mode = 'PRESERVED'
else:
    summary = dict(expected_summary)
    summary['output_sha256'] = {rel(OUTPUT[name]): hashlib.sha256(data).hexdigest() for name, data in new_files.items()}
    new_files['summary'] = json_bytes(summary)
    with TemporaryDirectory(prefix='step12_66_', dir=ROOT) as staging_dir:
        staged = {}
        for name, data in new_files.items():
            staged[name] = Path(staging_dir) / OUTPUT[name].name
            staged[name].write_bytes(data)
            require(staged[name].read_bytes() == data, f'Staging verification failed: {name}.')
        require(all(sha256(path) == input_hashes[reference] for reference, path in tracked.items()),
                'An input changed during audit preparation. No outputs created.')
        if any(path.exists() for path in OUTPUT.values()):
            raise FileExistsError('An output appeared during staging. Preserve and inspect it.')
        created = []
        try:
            for name, path in OUTPUT.items():
                path.parent.mkdir(parents=True, exist_ok=True)
                with path.open('xb') as handle:
                    created.append(path)
                    handle.write(staged[name].read_bytes())
        except Exception:
            for path in created:
                path.unlink(missing_ok=True)
            raise
    mode = 'CREATED'
require(all(sha256(path) == input_hashes[reference] for reference, path in tracked.items()),
        'An original input changed during final audit execution.')
for name, path in OUTPUT.items():
    if name != 'summary':
        require(summary['output_sha256'].get(rel(path)) == sha256(path), f'Saved final-audit output failed: {name}.')

print('=' * 76)
print('STEP 12-66 - FINAL INTEGRATION CHECKPOINT AUDIT AND STEP 13 HANDOFF')
print('=' * 76)
print(f'Verified scoped artifacts / passed audit checks: {len(tracked)} / {len(audit_rows)}')
print('Draft reconstruction: PASS - 30 metadata revisions reproduced from saved inputs')
print('Catalog candidates / basketball candidates / warm-ups / source occurrences: 144 / 132 / 12 / 156')
print('Basketball primary skill areas represented: 6 / 6')
print('Human field confirmations / explicit identity cases still pending: 68 / 8')
print('Approved distinct drills / target / planner-eligible: 0 / 100 / 0')
print('Saved-artifact integration checkpoint: PASS')
print('Full Step 12 approval objective: OPEN')
print('Next development step: 13 - Youth Participation Guidelines and Planning Rules')
print('Full notebook clean execution / live source refresh performed: No / No')
print('OUTPUTS')
for name, path in OUTPUT.items():
    print(f'  {mode:<9} {rel(path)}')
print('Catalogs, source files, previous reports, and approval decisions modified: No')
print('Git operations performed: None')
print('OVERALL: PASS - integration checkpoint audited; proceed to Step 13 with open reviews preserved')
print('=' * 76)
```

## 13. Youth Participation Guidelines and Planning Rules

### 13.1 Purpose

We will prepare a reviewed set of planning rules that our application can use to check weekly basketball plans. These rules will help us identify scheduling conflicts, missing information, and recommendations that do not follow the selected guidance.

We will build on the guideline preparation described in Aditi’s dataset review. Before using any prepared records, we will check their meaning and source references.

Our checks will measure adherence to documented guidance. Passing these checks will not establish that a plan is medically safe or suitable for every athlete.

### 13.2 Separate Published Guidance from Our Project Rules

We will keep three types of rules clearly identified.

| Rule type | What it covers | Example |
|---|---|---|
| Published guidance | Recommendations and limits from an identified reference | Weekly participation and rest guidance |
| User constraints | Requirements supplied in the athlete’s profile | Available days, equipment, and existing commitments |
| Project policies | Boundaries and application behavior defined by our team | Request clarification when essential information is missing |

This distinction will let us explain where each check comes from. We will not present a decision made by our team as an official recommendation.

### 13.3 Review the Youth Participation Guidance

Our primary reference will be the NBA and USA Basketball Youth Basketball Guidelines. We will preserve the distinction between recommended participation and maximum participation.

The following source categories are relevant to our project’s age range:

| Source category | Maximum organized basketball per week | Minimum rest days per week |
|---|---:|---:|
| Ages 12–14 | 10 hours | 1 |
| Grades 9–12 | 14 hours | 1 |

These maximums will not become training targets. The source cautions that participation below a maximum may still exceed advisable recommendations. Its qualifications about multiple games, other organized sports, camps, and recovery will remain attached to the relevant records.

We will collect both age and school grade. Where categories overlap or applicability is unclear, we will document the interpretation or flag the case for review rather than automatically selecting the more permissive limit.

We will also preserve the source’s definition of organized basketball, which includes games, practices, and applicable structured training. We will not relabel assigned training as informal activity simply to exclude it from a check.

### 13.4 Create a Structured Rule Register

We will organize the reviewed guidance into records that can later support both document retrieval and coded checks.

| Information | What we will record |
|---|---|
| Rule identity | Stable rule ID, category, and short description |
| Applicability | Age or grade group, required inputs, and relevant activity types |
| Check definition | Measurement, comparison, value or range, unit, and time period |
| Guidance type | Recommendation, maximum, minimum, or qualitative caution |
| Qualifications | Definitions, footnotes, exceptions, and unresolved interpretation |
| Source evidence | Source ID, URL, section or table location, and verification date |
| Implementation decision | Our interpretation, intended response, reviewer, and rule version |

Some guidance will support a numerical comparison. Other guidance will require an explanation or review flag. We will not invent a numerical threshold when the source provides only a qualitative caution.

AI may help us organize candidate records, but we will verify every adopted rule against its source. We will keep our implementation decisions distinguishable from the published guidance.

### 13.5 Check the Complete Weekly Schedule

We will evaluate existing commitments together with the sessions proposed by our application. The planner will not assume that the athlete starts with an empty week.

For each scheduled activity, we will record its day, duration, activity type, and whether it counts toward the relevant participation check. We will use consistent duration units and an explicitly defined seven-day planning window.

We will avoid counting the same activity twice. For example, a drill already included within a practice will not be added again as a separate session.

Our checks will also examine whether the proposed plan:

- Fits the athlete’s available days and time.
- Avoids overlapping activities.
- Uses available equipment, space, and participants.
- Accounts for existing practices and games.
- Preserves the required opportunities for rest from organized basketball.

Other sports and general conditioning will remain visible in the schedule. We will not automatically count every physical activity as organized basketball.

If existing commitments already conflict with the selected guidance, we will flag the conflict and explain it. We will not silently remove those commitments or add more training to an already conflicting schedule.

### 13.6 Handle Missing Information and Requests Outside Our Scope

Each check will return one of four results:

| Result | Meaning |
|---|---|
| Met | Available information supports the condition |
| Violated | Available information shows a conflict |
| Unknown | Information or rule interpretation is insufficient |
| Not applicable | The rule does not apply to this case |

Each result will include a short explanation. Missing information will not produce an automatic pass.

We cannot verify annual participation or actual sleep from a weekly plan alone. Without the necessary history or reported information, these checks will remain unknown or provide clearly identified reminders.

As a project policy, we will not generate injury rehabilitation programs, supplement recommendations, or individualized dietary prescriptions. Requests involving pain, injury, or medical restrictions will receive an appropriate clarification or referral response.

### 13.7 Review and Test the Rules

Before implementing the checks, one teammate will prepare each rule and another will review its source, interpretation, and expected behavior.

Later, we will test cases involving:

- Values below, equal to, and above a numerical limit.
- Missing or conflicting age and grade information.
- Existing commitments that leave insufficient time for additional sessions.
- Incomplete activity durations or participation history.
- Incorrect activity classification and double counting.
- Camps, tournaments, and other cases requiring qualified interpretation.
- Changes made by the user after a plan is generated.

We will document the expected result for each test before using it to assess the implementation.

We will use the same independently defined scoring rules to evaluate all four planner versions. Version 4 may also use checks during generation, but its internal evaluator will not determine the final research scores.

### 13.8 Expected Outputs

After we complete this step, we expect to save:

- `data/processed/planning_rules.json`: reviewed rules, applicability conditions, and implementation decisions.
- `data/references/guidelines/`: permitted reference material with source metadata.
- `results/planning_rule_review.csv`: reviewer decisions, unresolved questions, and changes.

We will version these records and freeze the adopted rules before final evaluation. Any later correction will be documented so we can identify which results it affects.

```python
# Step 13-01 - Consolidated Planning-Rule Inventory and Interpretation Review
# Preserve 17 structured references and 5 guidance records; activate no rules.

import csv
import hashlib
import io
import json
from collections import Counter, defaultdict
from decimal import Decimal, InvalidOperation
from pathlib import Path
from tempfile import TemporaryDirectory
from urllib.parse import urlsplit

ROOT = next((p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
             if (p / '.git').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Run this cell inside the Champions Athletic Map Git project.')
RESULTS = ROOT / 'results'
INPUT = {
    'structured_rules': ROOT / 'data/references/recovery/youth_basketball_recovery_rules_v0_1.csv',
    'general_guidance': ROOT / 'data/processed/recovery_guidance.csv',
    'readme': ROOT / 'data/references/recovery/YOUTH_BASKETBALL_RECOVERY_RULES_README.md',
    'catalog': ROOT / 'data/processed/basketball_drill_catalog_v0_3_9_REVIEW_ONLY.csv',
    'checkpoint': RESULTS / 'basketball_step12_66_summary_v0_3_9.json',
    'handoff': RESULTS / 'basketball_step12_step13_handoff_v0_3_9.json',
    'audit': RESULTS / 'basketball_step12_final_audit_v0_3_9.csv',
    'observations': RESULTS / 'basketball_step12_observations_v0_3_9.md',
}
OUTPUT = {
    'candidates': ROOT / 'data/processed/planning_rule_candidates_v0_1_REVIEW_ONLY.json',
    'review': RESULTS / 'planning_rule_review_v0_1.csv',
    'sources': ROOT / 'data/references/guidelines/planning_rule_source_index_v0_1.csv',
    'audit': RESULTS / 'planning_rule_inventory_audit_v0_1.csv',
    'summary': RESULTS / 'planning_step13_01_summary_v0_1.json',
}
EXPECTED_HASHES = {
    'structured_rules': 'd4ef1ab35329e747b980d19fb0339c28ee3dff41e5b60724490193b0dab46ac6',
    'general_guidance': '292a48fd2841edc2be6c05a74b7e2fa3586760a5e9e048844bf7590a0c6725c1',
}
STRUCTURED_COLUMNS = ['rule_id', 'domain', 'segment', 'age_min', 'age_max', 'grade_min', 'grade_max',
                      'metric', 'rule_type', 'value_min', 'value_max', 'unit', 'source_organization',
                      'source_url', 'source_section', 'applicability_notes', 'planner_handling', 'verification_status']
GENERAL_COLUMNS = ['topic', 'guidance', 'source', 'source_url']
audit = []


def require(condition, message):
    if not condition:
        raise ValueError(message)


def check(identity, observed, expected, note):
    require(observed == expected, f'Inventory check failed: {identity}. {note}')
    audit.append({'check_id': identity, 'status': 'PASS', 'observed': json.dumps(observed, sort_keys=True),
                  'expected': json.dumps(expected, sort_keys=True), 'note': note})


def sha256(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()


def rel(path):
    return path.relative_to(ROOT).as_posix()


def read_csv(path):
    with path.open(newline='', encoding='utf-8-sig') as handle:
        reader = csv.DictReader(handle)
        columns = list(reader.fieldnames or [])
        rows = list(reader)
    require(columns and len(columns) == len(set(columns)), f'Invalid CSV header: {path.name}')
    require(all(None not in r and all(v is not None for v in r.values()) for r in rows),
            f'Malformed CSV records: {path.name}')
    return rows, columns


def number(value, label):
    if not value.strip():
        return None
    try:
        result = Decimal(value)
    except InvalidOperation as error:
        raise ValueError(f'Invalid number: {label}.') from error
    require(result.is_finite() and result >= 0, f'Invalid numeric bound: {label}.')
    return result


def validate_pair(row, low_field, high_field, identity, integer=False):
    low, high = number(row[low_field], identity + '/' + low_field), number(row[high_field], identity + '/' + high_field)
    if low is not None and high is not None:
        require(low <= high, f'Reversed bounds: {identity}/{low_field}.')
    if integer:
        require((low is None) == (high is None), f'Incomplete age/grade pair: {identity}.')
        require(all(v == v.to_integral_value() for v in (low, high) if v is not None),
                f'Age/grade bound is not an integer: {identity}.')
    return low, high


def json_bytes(value):
    return (json.dumps(value, ensure_ascii=False, indent=2) + '\n').encode('utf-8')


def csv_bytes(rows):
    buffer = io.StringIO(newline='')
    writer = csv.DictWriter(buffer, fieldnames=list(rows[0]), lineterminator='\n')
    writer.writeheader()
    writer.writerows(rows)
    return buffer.getvalue().encode('utf-8')


for name, path in INPUT.items():
    if not path.is_file():
        raise FileNotFoundError(f'Missing {name}: {path}')
input_hashes = {rel(path): sha256(path) for path in INPUT.values()}
for name, expected in EXPECTED_HASHES.items():
    inspected_hash = (hashlib.sha256(INPUT[name].read_text(encoding='utf-8-sig').encode('utf-8')).hexdigest()
                      if name == 'general_guidance' else input_hashes[rel(INPUT[name])])
    check('input_' + name, inspected_hash, expected,
          'General guidance comparison tolerates line-ending/BOM differences only; original byte hashes are retained.')
checkpoint = json.loads(INPUT['checkpoint'].read_text(encoding='utf-8'))
handoff = json.loads(INPUT['handoff'].read_text(encoding='utf-8'))
check('step12_checkpoint', [checkpoint.get('step'), checkpoint.get('integration_checkpoint_status'),
      checkpoint.get('step13_rule_development_can_proceed')], ['12-66', 'passed', True],
      'Rule development proceeds from the saved integration checkpoint.')
for name in ('handoff', 'audit', 'observations'):
    path = INPUT[name]
    require(checkpoint.get('output_sha256', {}).get(rel(path)) == input_hashes[rel(path)],
            f'Step 12-66 output integrity mismatch: {name}.')
require(checkpoint.get('input_sha256', {}).get(rel(INPUT['catalog'])) == input_hashes[rel(INPUT['catalog'])],
        'The catalog changed after the Step 12 checkpoint.')
check('handoff_exclusions', [handoff.get('full_step12_complete'), handoff.get('planner_use_of_unapproved_catalog_allowed')],
      [False, False], 'Open catalog approvals are preserved.')
catalog, _ = read_csv(INPUT['catalog'])
check('catalog_gates', [len(catalog), sum(r['catalog_approval_status'] == 'approved' for r in catalog),
      sum(r['planner_eligible'] == 'yes' for r in catalog)], [144, 0, 0], 'No candidate is enabled for the planner.')
rules, rule_columns = read_csv(INPUT['structured_rules'])
guidance, guidance_columns = read_csv(INPUT['general_guidance'])
check('structured_schema', rule_columns, STRUCTURED_COLUMNS, 'Imported numeric and applicability fields are preserved.')
check('general_schema', guidance_columns, GENERAL_COLUMNS, 'General guidance remains distinct from structured limits.')
check('input_counts', [len(rules), len(guidance)], [17, 5], 'Inventory all 22 existing records in one batch.')
check('unique_rule_ids', len({r['rule_id'] for r in rules if r['rule_id']}), 17, 'Preserve stable imported rule identities.')
check('unique_guidance_topics', len({r['topic'] for r in guidance if r['topic']}), 5, 'Maintain distinct source guidance records.')
require(INPUT['readme'].read_text(encoding='utf-8-sig').strip(), 'Reference README is empty.')
for row in rules:
    identity = row['rule_id']
    require(all(row[k].strip() for k in ('segment', 'metric', 'rule_type', 'unit', 'source_organization',
                                       'source_url', 'source_section', 'applicability_notes')), f'Incomplete rule: {identity}.')
    validate_pair(row, 'age_min', 'age_max', identity, integer=True)
    validate_pair(row, 'grade_min', 'grade_max', identity, integer=True)
    low, high = validate_pair(row, 'value_min', 'value_max', identity)
    kind = row['rule_type']
    require(kind in {'maximum', 'minimum', 'recommended_range', 'recommended_exact', 'published_range_9_to_10'},
            f'Unreviewed source rule type: {identity}.')
    require((kind != 'maximum' or (low is None and high is not None))
            and (kind != 'minimum' or (low is not None and high is None))
            and (kind not in {'recommended_range', 'recommended_exact', 'published_range_9_to_10'}
                 or (low is not None and high is not None))
            and (kind != 'recommended_exact' or low == high), f'Unexpected bound shape: {identity}.')
check('structured_bound_shapes', len(rules), 17, 'Intentional one-sided and inapplicable bounds are not missing-data errors.')
for row in guidance:
    require(all(row[k].strip() for k in GENERAL_COLUMNS), 'Incomplete general guidance record.')
all_urls = sorted({r['source_url'] for r in rules + guidance})
for url in all_urls:
    parsed = urlsplit(url)
    require(parsed.scheme == 'https' and parsed.hostname, f'Invalid source URL: {url}.')
source_ids = {url: f'GUID-SRC-{i:04d}' for i, url in enumerate(all_urls, start=1)}


def interpretation_flags(row, general=False):
    flags = ['current_source_and_qualifications_review_pending', 'peer_review_and_adoption_pending']
    if general:
        flags.append('unstructured_guidance_do_not_infer_numeric_cutoff')
        if row['topic'] == 'Sleep':
            flags += ['reported_sleep_history_required', 'review_relation_to_structured_sleep_records']
        if row['source'] == 'AAP':
            flags.append('broad_sport_guidance_not_basketball_specific')
        if row['topic'] == 'Rest':
            flags += ['review_relation_to_AAP_structured_rest_record', 'do_not_treat_rest_range_as_ceiling']
        return flags
    metric = row['metric']
    if row['rule_type'].startswith('recommended'):
        flags.append('recommendation_not_mandatory_participation_requirement')
    if row['rule_type'] == 'maximum':
        flags.append('maximum_is_not_training_target_or_clearance')
    if row['source_organization'] == 'NBA and USA Basketball':
        flags += ['review_age_grade_overlap_without_assuming_precedence', 'review_publisher_table_coverage_and_footnotes']
    if 'resolve using published player segmentation' in row['applicability_notes']:
        flags.append('upstream_segmentation_precedence_note_needs_interpretation_review')
    if metric == 'organized_basketball_hours_per_week':
        flags.append('structured_training_classification_and_complete_schedule_required')
    if metric == 'organized_basketball_months_per_year':
        flags.append('annual_history_required_not_inferable_from_weekly_plan')
    if row['rule_id'] == 'NBA_G9_12_MAX_MONTHS':
        flags.append('published_maximum_range_lower_endpoint_is_not_minimum_participation')
    if metric.startswith('sleep_'):
        flags += ['reported_sleep_history_required', 'preserve_nightly_versus_daily_measurement_period']
    if row['rule_id'] == 'NBA_A12_SLEEP':
        flags.append('outside_project_target_ages_13_to_17')
    if row['rule_id'] == 'AAP_WEEKLY_SPORT_REST':
        flags += ['broad_sport_guidance_not_basketball_specific', 'do_not_treat_upper_rest_value_as_ceiling']
    if row['rule_id'] == 'NBA_A12_14_MAX_HOURS':
        flags.append('review_concurrent_sport_qualification')
    return flags


candidates = []
for index, row in enumerate(rules, start=1):
    candidates.append({
        'candidate_rule_id': row['rule_id'], 'record_kind': 'structured_reference',
        'rule_class': 'published_guidance_candidate', 'input_reference': rel(INPUT['structured_rules']),
        'input_row_number': index, 'source_id': source_ids[row['source_url']],
        'metric_or_topic': row['metric'], 'source_rule_type': row['rule_type'], 'source_unit': row['unit'],
        'imported_record': dict(row), 'upstream_verification_assertion': row['verification_status'],
        'interpretation_flags': interpretation_flags(row), 'current_source_review_status': 'pending',
        'interpretation_review_status': 'pending', 'peer_review_status': 'pending', 'adoption_status': 'not_adopted',
        'enabled_for_planner': False, 'comparison_operator': None, 'adopted_threshold': None,
    })
for index, row in enumerate(guidance, start=1):
    candidates.append({
        'candidate_rule_id': f'RECOVERY-GUIDANCE-{index:04d}', 'record_kind': 'general_guidance',
        'rule_class': 'published_guidance_candidate', 'input_reference': rel(INPUT['general_guidance']),
        'input_row_number': index, 'source_id': source_ids[row['source_url']],
        'metric_or_topic': row['topic'], 'source_rule_type': 'unstructured_guidance', 'source_unit': '',
        'imported_record': dict(row), 'upstream_verification_assertion': '',
        'interpretation_flags': interpretation_flags(row, general=True), 'current_source_review_status': 'pending',
        'interpretation_review_status': 'pending', 'peer_review_status': 'pending', 'adoption_status': 'not_adopted',
        'enabled_for_planner': False, 'comparison_operator': None, 'adopted_threshold': None,
    })
check('candidate_count_and_identity', [len(candidates), len({r['candidate_rule_id'] for r in candidates})],
      [22, 22], 'No imported records are merged or dropped.')
check('preserved_structured_records', [r['imported_record'] for r in candidates[:17]], rules,
      'Every imported value and original verification assertion is retained.')
check('preserved_general_records', [r['imported_record'] for r in candidates[17:]], guidance,
      'General source statements are retained without numeric extraction.')
check('no_active_rules', sum(r['enabled_for_planner'] for r in candidates), 0, 'Source references are not adopted rules.')
source_rows = []
for url in all_urls:
    selected = [r for r in candidates if r['source_id'] == source_ids[url]]
    organizations = sorted({r['imported_record'].get('source_organization', r['imported_record'].get('source', ''))
                            for r in selected})
    source_rows.append({'source_id': source_ids[url], 'source_url': url,
                        'source_organization_labels': ';'.join(organizations), 'candidate_records': len(selected),
                        'current_source_review_status': 'pending', 'source_retrieval_performed_by_this_cell': 'no',
                        'alias_or_duplicate_source_review': 'pending',
                        'intended_use_review': 'pending', 'source_coverage_complete': 'not_established'})
check('source_inventory', [len(source_rows), sum(r['candidate_records'] for r in source_rows)],
      [4, 22], 'Keep both AAP source URLs; URL identity does not resolve source equivalence.')
review_rows = []
for row in candidates:
    raw = row['imported_record']
    review_rows.append({
        'candidate_rule_id': row['candidate_rule_id'], 'record_kind': row['record_kind'],
        'source_id': row['source_id'], 'source_url': raw['source_url'], 'metric_or_topic': row['metric_or_topic'],
        'source_rule_type': row['source_rule_type'], 'source_unit': row['source_unit'],
        'upstream_verification_assertion': row['upstream_verification_assertion'],
        'interpretation_flags': ';'.join(row['interpretation_flags']), 'source_review_status': 'pending',
        'interpretation_review_status': 'pending', 'peer_review_status': 'pending',
        'reviewer_decision': 'pending', 'reviewer': '', 'review_date': '', 'review_evidence_note': '',
        'adoption_status': 'not_adopted', 'enabled_for_planner': 'no',
    })
register = {
    'register_version': '0.1', 'status': 'candidate_inventory_only', 'target_age_min': 13, 'target_age_max': 17,
    'rule_classes_planned': ['published_guidance', 'user_constraint', 'project_policy'],
    'classes_imported_in_this_cell': ['published_guidance_candidate'],
    'user_constraints_or_project_policies_created_by_this_cell': False,
    'current_source_verification_performed_by_this_cell': False, 'source_coverage_complete': False,
    'records': candidates,
    'interpretation_boundaries': [
        'Original verification assertions do not constitute current Step 13 review or adoption.',
        'Recommendation ranges and exact recommendations are not mandatory training minima.',
        'Do not infer participation-table precedence from competition player segmentation.',
        'An endpoint in a published maximum range is not a minimum participation requirement.',
        'A rest recommendation upper value must not become a prohibition on more rest.',
        'Weekly plans cannot establish annual participation or actual sleep without additional history.',
        'Preserve nightly and daily measurement periods and separate sport-specific scopes.',
        'The imported records do not exhaust publisher tables, footnotes, or camp/tournament qualifications.',
    ],
}
expected_summary = {
    'step': '13-01', 'register_version': '0.1', 'structured_reference_records': 17,
    'general_guidance_records': 5, 'candidate_records': 22, 'source_urls': 4,
    'source_reviews_pending': 22, 'interpretation_reviews_pending': 22, 'peer_reviews_pending': 22,
    'adopted_rules': 0, 'enabled_rules': 0, 'source_coverage_complete': False,
    'current_source_verification_performed_by_this_cell': False,
    'outside_target_age_exception_rule_ids': ['NBA_A12_SLEEP'],
    'upstream_segmentation_note_review_records': sum('upstream_segmentation_precedence_note_needs_interpretation_review'
                                                  in r['interpretation_flags'] for r in candidates),
    'rule_kind_counts': dict(Counter(r['record_kind'] for r in candidates)),
    'catalog_approvals': 0, 'planner_eligible_drills': 0,
    'step12_human_field_confirmations_still_pending': 68, 'step12_explicit_identity_cases_still_pending': 8,
    'input_files_modified': False, 'catalog_modified': False, 'git_operations_performed': False,
    'inventory_checks_passed': len(audit), 'input_sha256': input_hashes,
}
new_files = {'candidates': json_bytes(register), 'review': csv_bytes(review_rows),
             'sources': csv_bytes(source_rows), 'audit': csv_bytes(audit)}
existing = [p.exists() for p in OUTPUT.values()]
if any(existing) and not all(existing):
    raise FileExistsError('Only part of the Step 13-01 output set exists. Inspect it before rerunning.')
if all(existing):
    summary = json.loads(OUTPUT['summary'].read_text(encoding='utf-8'))
    require(all(summary.get(k) == v for k, v in expected_summary.items()), 'Existing inventory summary differs from current inputs.')
    for name, data in new_files.items():
        path = OUTPUT[name]
        require(summary.get('output_sha256', {}).get(rel(path)) == sha256(path) and path.read_bytes() == data,
                f'Existing inventory output changed: {rel(path)}. Preserved.')
    mode = 'PRESERVED'
else:
    summary = dict(expected_summary)
    summary['output_sha256'] = {rel(OUTPUT[name]): hashlib.sha256(data).hexdigest() for name, data in new_files.items()}
    new_files['summary'] = json_bytes(summary)
    with TemporaryDirectory(prefix='step13_01_', dir=ROOT) as staging_dir:
        staged = {}
        for name, data in new_files.items():
            staged[name] = Path(staging_dir) / OUTPUT[name].name
            staged[name].write_bytes(data)
            require(staged[name].read_bytes() == data, f'Staging verification failed: {name}.')
        require(all(sha256(p) == input_hashes[rel(p)] for p in INPUT.values()), 'An input changed; no outputs created.')
        if any(p.exists() for p in OUTPUT.values()):
            raise FileExistsError('An output appeared during staging. Preserve and inspect it.')
        created = []
        try:
            for name, path in OUTPUT.items():
                path.parent.mkdir(parents=True, exist_ok=True)
                with path.open('xb') as handle:
                    created.append(path)
                    handle.write(staged[name].read_bytes())
        except Exception:
            for path in created:
                path.unlink(missing_ok=True)
            raise
    mode = 'CREATED'
require(all(sha256(p) == input_hashes[rel(p)] for p in INPUT.values()), 'An original input changed during inventory.')
for name, path in OUTPUT.items():
    if name != 'summary':
        require(summary['output_sha256'].get(rel(path)) == sha256(path), f'Saved output verification failed: {name}.')

print('=' * 76)
print('STEP 13-01 - CONSOLIDATED PLANNING-RULE INVENTORY AND REVIEW')
print('=' * 76)
print('Step 12 handoff / source input integrity: PASS / PASS')
print('Structured reference records / general guidance / total candidates: 17 / 5 / 22')
print('Registered source URLs: 4 (source aliases remain under review)')
print(f'Inventory checks passed: {len(audit)}')
print('Source / interpretation / peer reviews pending: 22 / 22 / 22')
print('Adopted rules / enabled rules: 0 / 0')
print('REVIEW PRIORITIES')
print('  Age/grade applicability and imported player-segmentation interpretation')
print('  Recommendations versus limits; annual maximum ranges; rest-range semantics')
print('  Sleep measurement periods; annual/sleep history; source qualifications')
print('  Age-12 exception retained outside the target ages 13-17')
print('  General guidance and related structured references preserved separately')
print('Catalog approvals / planner-eligible drills: 0 / 0 (unchanged)')
print('OUTPUTS')
for name, path in OUTPUT.items():
    print(f'  {mode:<9} {rel(path)}')
print('Original reference files, README, catalog, and Step 12 artifacts modified: No')
print('Source refresh / Git operations performed by this cell: No / No')
print('Next: source-grounded rule definitions and applicability review before adoption.')
print('OVERALL: PASS - all 22 reference records inventoried; no rules activated')
print('=' * 76)
```

## 14. USDA Food Facts and Nutrient Data Preparation

### 14.1 Purpose

We will prepare a reliable food reference table for the optional nutrition feature in our application. This feature will answer factual questions about the nutrients in a selected food.

For example, we could show the recorded protein, carbohydrate, and energy values for a clearly identified food and reference amount.

We will use this data for food facts. We will not use it to calculate an athlete’s required calorie intake, prescribe a diet, or recommend supplements.

### 14.2 Confirm the Source and Complete the Required Files

We plan to use the April 2018 SR Legacy release available through USDA FoodData Central. We selected this fixed release so our team can work with a consistent reference throughout development and evaluation.

Our earlier review of Beakal’s package identified useful supporting food records, but the files did not provide the complete connections between food descriptions, nutrient amounts, and units.

We will preserve his work and reuse records where their origin and compatibility can be verified. We will not label the existing files as SR Legacy unless we establish that they belong to that release.

Before preparing the final table, we will record:

- The dataset name and release.
- The original download URL and archive name.
- Our download date and file hashes.
- Any supporting files and their versions.
- The preparation script and resulting output version.

Some older USDA archives require separate supporting files. We will check the archive contents and USDA instructions rather than assume that every required table is included.

We will use a local download for the initial implementation. An API connection will not be required for the application to reproduce these food facts.

### 14.3 Identify the Required Tables

For the CSV preparation approach, we will verify the following tables.

| Table | Purpose |
|---|---|
| `food.csv` | Food identifiers, descriptions, and data types |
| `food_nutrient.csv` | Nutrient amounts associated with each food |
| `nutrient.csv` | Nutrient identifiers, names, and units |
| `food_category.csv` | Category descriptions, when available and useful |
| `food_portion.csv` | Food-specific measures and their gram weights |
| `measure_unit.csv` | Measure definitions where the selected records use them |

The first three tables will support the basic food lookup. Household portion displays will require additional usable portion information.

We will check the selected release’s documentation before interpreting each field. In particular, SR Legacy may describe a household measure in its `modifier` field rather than through a meaningful measure-unit lookup.

We will not treat laboratory result records as substitutes for the final food nutrient table.

### 14.4 Select and Describe Our Food Subset

We will prepare a manageable subset of ordinary foods for the prototype. Our selection will cover several food categories and include enough variation to demonstrate meaningful lookups.

We will document the selection criteria and report the actual number of retained foods after preparation. We will not describe the subset as a complete nutrition database.

For every selected food, we will preserve:

- The original USDA identifier and description.
- The source data type and release.
- Preparation details such as raw, cooked, drained, or dried.
- The category, where available.
- A source record link.

We will keep foods with different preparation methods separate. Similar names alone will not justify merging records.

### 14.5 Join and Validate the Nutrient Records

We will connect the tables using their documented identifiers.

We will join food records to nutrient amounts using `fdc_id`, then connect nutrient amounts to nutrient definitions using `nutrient_id`.

Before and after joining, we will check:

- Whether the lookup-table identifiers are unique.
- Whether every retained nutrient record matches a food and nutrient definition.
- Whether unexpected duplicate food–nutrient combinations appear.
- Whether joins unexpectedly multiply or remove records.
- Whether the resulting amounts have usable units and a known reference basis.

We will investigate conflicting records instead of averaging them automatically or keeping whichever appears first.

We will retain nutrient identifiers alongside readable names so that similar labels do not cause us to combine different measurements.

Our initial display will focus on energy, protein, carbohydrate, total fat, and dietary fiber, where reported. We may include selected minerals if their values and units are available.

We will document the exact nutrient definitions used, including the energy measure selected for display.

### 14.6 Preserve Units, Missing Values, and Reference Amounts

We will keep each nutrient amount together with its unit and reference amount.

For the selected SR Legacy data, we will use the documented per-100-gram edible-food basis. We will clearly distinguish this from a household portion.

We will not confuse:

- Kilocalories with kilojoules.
- Grams with milligrams or micrograms.
- A per-100-gram value with a per-portion value.
- A missing measurement with a recorded zero.

We will preserve missing values as unavailable. We will not fill missing nutrient amounts using AI-generated estimates.

We will flag invalid amounts or unexpected units for review. A flagged record will not automatically become a corrected record.

### 14.7 Calculate Portions Only When Supported

Our default display will show values per 100 grams.

Where we have a verified food-specific portion weight, our code can calculate:

**Nutrient amount for the selected portion = nutrient amount per 100 grams × portion weight in grams ÷ 100**

We will interpret the portion’s amount, description, and gram weight together. The listed gram weight may already represent multiple measure units, so we will avoid multiplying it twice.

We will not use a generic conversion from cups or tablespoons to grams. If a usable portion weight is unavailable, we will retain the per-100-gram display.

A displayed portion will describe the reference quantity used in the calculation. It will not imply that we recommend that quantity for the athlete.

### 14.8 Prepare the Lookup Behavior and Review Its Accuracy

Later, our food lookup tool will retrieve the selected records and perform any arithmetic in code. The language model may explain the returned facts, but it will not invent nutrient values or perform unchecked calculations.

Each response will identify the food, reference amount, nutrient values, units, and source.

If a request matches several foods, we will ask the user to select or clarify the intended item. We will not silently choose between raw and cooked versions.

Before enabling the feature, we will compare a reviewed sample of outputs with the source records. Our checks will cover missing values, ambiguous food names, unit handling, and portion calculations.

These records will not establish allergy safety, dietary suitability, or an individual athlete’s nutritional needs.

If essential links or units remain unresolved, we will disable the affected lookup rather than return unsupported facts.

### 14.9 Expected Outputs

After completing this step, we expect to save:

- `data/processed/usda_foods.csv`: selected food identities and source information.
- `data/processed/usda_food_nutrients.csv`: validated nutrient amounts, units, and reference basis.
- `data/processed/usda_food_portions.csv`: verified portion records, if supported.

## 15. Exercise Duplicate Grouping and Training, Validation, and Test Splits

### 15.1 Purpose

We will divide the exercise dataset into separate training, validation, and test sets for our body-part classifier.

Before splitting, we will group duplicate and closely related records. Otherwise, nearly identical exercises could appear in both training and testing, making the model’s performance look better than its ability to classify genuinely different examples.

This step applies to our exercise classifier. The fictional athlete profiles used to evaluate the basketball planners will have their own development and test separation in Step 16.

### 15.2 Identify Related Exercise Records

We will start with the cleaned exercise catalog from Step 10 and the quality findings from Step 11.

We will use consistent comparison rules to identify candidate matches, including normalized exercise names and repeated meaningful descriptions. These comparison fields will support duplicate review; they will not replace the original exercise text.

Our review will distinguish between:

| Record relationship | Planned treatment |
|---|---|
| Confirmed duplicate records | Retain one canonical example and preserve its source references |
| Different wording for the same exercise | Assign the records to the same group |
| Closely related variants with substantially repeated content | Review whether they should share a group |
| Different exercises sharing generic instructional text | Keep separate unless other evidence supports grouping |
| Matching inputs with conflicting labels | Investigate before including them in the supervised benchmark |

We will not group every exercise targeting the same muscle. Our groups will represent duplicate or closely related examples, rather than entire body-part categories.

Repeated boilerplate will not automatically connect unrelated exercises. We will also review unusually large groups to check whether a chain of weak similarities has combined distinct movements.

Every retained exercise will receive a stable group ID. An exercise without a related record will form its own group.

### 15.3 Review Label Conflicts and Small Classes

We will check label disagreements before freezing the split.

Some related variants may legitimately have different body-part labels. We will preserve justified differences rather than force every member of a group to have the same label.

If identical model inputs have conflicting labels and we cannot resolve the disagreement from reliable evidence, we will exclude those ambiguous examples from the supervised benchmark and document the reason.

We will assess class support using both record counts and independent group counts. Several nearly identical records will not count as several independent examples.

Our earlier review identified very small classes, including neck exercises. We will verify their remaining support after cleaning and grouping before deciding which classes the classifier can reasonably cover.

Before training, we will document the minimum group support required for each split and freeze the eligible class list. Unsupported classes can remain in the broader catalog for review without becoming classifier output categories.

We will not merge unrelated anatomical categories simply to increase their counts.

### 15.4 Create the Three Splits

We will initially target the following allocation:

| Split | Approximate share | Purpose |
|---|---:|---|
| Training | 70% | Learn text representations and model parameters |
| Validation | 15% | Compare settings, select models, and make development decisions |
| Test | 15% | Assess the selected models after development is complete |

These percentages are targets. Keeping groups intact and maintaining usable class coverage will take priority over matching the percentages exactly.

We will use a reproducible group-aware allocation that considers the class distribution. Every group will belong entirely to one split.

We will record the splitting method, random seed, software version, and resulting allocation. We will not assume that a group-aware method automatically produces balanced classes.

If the allocation is unsuitable, we will resolve the data-design issue before training. We will not move individual records out of their groups or choose a split because it produces better model scores.

### 15.5 Keep Inputs Separate from Labels and Metadata

Our classifier input will contain only the exercise name and available description.

The target will be the reviewed body-part label.

We will exclude fields that directly reveal or were constructed from the target, including target-bearing `agent_text` and derived fields such as `goal_tag`.

Exercise IDs, group IDs, source information, and review decisions will remain available for auditing, but they will not become model features.

Naturally occurring muscle names in the original exercise text will remain valid input. We will not remove useful source wording simply because it resembles a class name.

### 15.6 Verify Separation and Coverage

Before accepting the split, we will check that:

- No exercise ID appears in more than one split.
- No group ID appears in more than one split.
- No duplicate normalized model-input records cross split boundaries.
- Every eligible record has exactly one split assignment.
- Each included class has the required training and held-out support.
- Record totals reconcile with the eligible cleaned dataset.
- Exclusions and unresolved records are accounted for.

We will report record counts and independent group counts by split and class.

We will also summarize description availability and text lengths within each split. This will help us identify cases where one split contains disproportionately more title-only examples.

Small held-out class counts will remain a limitation. We will report their support rather than present a result based on very few examples as strong evidence.

### 15.7 Freeze the Split and Protect the Test Set

Once the checks pass, we will save a split manifest connecting each exercise ID to its group, label, eligibility status, and split.

All classifier models will use the same approved split. This includes the majority-class baseline, TF-IDF logistic regression, and neural classifier.

In Step 18, we will fit learned preprocessing only on training data. This includes vocabulary selection, TF-IDF statistics, tokenizer vocabulary where applicable, and any learned feature transformations.

Class weights or training-data resampling will also use training data only. We will not oversample the validation or test sets.

We will use validation results for development and reserve test performance for the final evaluation in Step 32.

Our earlier dataset-wide review will be described as data inspection and benchmark preparation. After freezing, model-driven exploration and adjustments will use training and validation data.

If we later discover a genuine data error, we will document the correction and version the affected dataset and split. We will not silently change the benchmark after seeing test results.

### 15.8 Expected Outputs

After completing this step, we expect to save:

- `data/processed/exercise_groups.csv`: exercise IDs, group assignments, and grouping decisions.
- `data/processed/exercise_split_manifest.csv`: eligibility, labels, and split membership.
- `results/exercise_split_summary.csv`: record and group counts by class and split.
- `results/exercise_split_checks.csv`: overlap, completeness, and coverage checks.
- `results/exercise_class_eligibility.csv`: included classes, support criteria, and exclusion reasons.

These files will let every teammate reproduce the same classifier experiment without independently creating a different split.

## 16. Fictional Athlete Profiles and Development/Test Separation

### 16.1 Purpose

We will create fictional athlete profiles to develop and evaluate our basketball planning application.

Each profile will describe a planning situation, including the athlete’s experience, goals, available resources, and existing commitments. These details will let us check whether the application produces a useful plan for the information provided.

Our profiles will be synthetic test cases. They will not represent observed training outcomes or establish that our recommendations improve athletic performance.

### 16.2 Review the Existing Profiles

We will first review the 30 profiles included in Beakal’s dataset package.

Before reusing them, we will confirm how they were created, identify missing fields, and check whether their requests fit our project’s scope. We will document their origin rather than assume that historical NBA records establish realistic teenage athlete profiles.

Approved examples may support development. We will keep them and any closely related variations out of the final test set.

We will also clarify ambiguous labels. For example, a general recovery goal will not automatically mean that an athlete has an injury, and missing restriction information will not automatically mean that no restrictions exist.

### 16.3 Define the Profile Structure

We will use a consistent structure so that all four planner versions receive comparable information.

| Profile component | Information we will include |
|---|---|
| Identity | Stable fictional profile ID |
| Age and grade | Age within our 13–17 scope and school grade |
| Experience | Basketball experience and current skill level |
| Goals | Specific priorities such as ball handling, shooting, or general conditioning |
| Basketball context | Position, where relevant, and individual or team participation |
| Available time | Planning week, available days, and session time limits |
| Existing commitments | Practices, games, other sports, and their scheduled durations |
| Resources | Equipment, court or hoop access, and available space |
| Participants | Whether the athlete trains alone or has partners or a group |
| Relevant limitations | Clearly stated constraints or requests requiring clarification |
| Optional history | Annual participation or reported sleep information when a scenario requires it |

We will distinguish information that is missing, explicitly unknown, or confirmed absent.

We will not infer grade from age, assume access to a court, or invent available time. We will also avoid adding personal details that do not help us evaluate the planning task.

### 16.4 Plan the Development and Final Test Sets

We plan to prepare 300 profiles: 100 for development and 200 for final evaluation.

| Age | Development profiles | Final test profiles |
|---|---:|---:|
| 13 | 20 | 40 |
| 14 | 20 | 40 |
| 15 | 20 | 40 |
| 16 | 20 | 40 |
| 17 | 20 | 40 |
| **Total** | **100** | **200** |

We will use the development profiles to improve prompts, retrieval, coordination, application behavior, and evaluation procedures.

The final profiles will support the comparison after those development decisions are complete.

These balanced age counts are an experimental design choice. We will not claim that they represent the distribution of real teenage basketball players.

### 16.5 Include Useful Planning Cases and Challenging Requests

Most final cases will involve feasible requests, so the application must demonstrate that it can produce useful plans.

Our planned final composition is:

| Scenario category | Final cases | Expected behavior |
|---|---:|---|
| Feasible planning request | 150 | Produce a relevant plan that respects the supplied constraints |
| Missing essential information | 20 | Identify the missing information and request clarification |
| Conflicting constraints | 15 | Explain the conflict and offer an appropriate next step |
| Request outside our scope | 15 | Explain the boundary and provide an appropriate response |

Within each age group, we will include 30 feasible cases, four missing-information cases, three conflicting cases, and three requests outside our scope.

The development set will also cover all four categories.

We will distinguish essential missing information from optional details. A missing preferred position, for example, should not automatically prevent a general skills plan.

Missing-age and outside-age-range inputs will be additional application tests in Step 30. They will not replace the age-balanced final cases above.

### 16.6 Create Varied and Coherent Scenarios

Before generating profiles, we will prepare a coverage matrix describing the combinations we want to test.

Our cases will vary across experience, goals, available time, equipment, court access, training partners, existing commitments, and relevant age–grade combinations.

We will include situations such as an athlete with limited equipment, an athlete who can train only briefly, and an athlete whose team schedule leaves little room for additional sessions.

We will check that ordinary planning cases are internally coherent. Deliberately conflicting cases will be clearly identified in the evaluation records.

Python and AI may help us draft profiles, but a teammate will review each case. Another teammate will check its expected behavior and important constraints before approval.

This review will check scenario quality and consistency with our references. It will not certify the medical suitability of a training plan.

### 16.7 Keep Related Profiles in the Same Partition

Changing a name, rewording a request, or slightly adjusting one field will not create an independent test case.

We will assign related profiles a shared family ID. Every variation of the same underlying profile will remain entirely within development or final testing.

Where possible, we will create distinct base profiles rather than fill the benchmark with paraphrases.

Our checks will examine exact duplicates, similar wording, and repeated combinations of profile facts. Shared formatting or broad scenario categories alone will not make profiles duplicates.

We will report both profile counts and distinct family counts. If related variants remain, we will account for that relationship when interpreting results.

### 16.8 Prepare Expected Behaviors Separately

Each profile will have an evaluation record describing what a reasonable response should accomplish.

This record may identify:

- The expected response category.
- Availability and resource constraints that must be respected.
- Relevant planning-rule IDs from Step 13.
- Information that requires clarification.
- Unsupported assumptions the planner should avoid.
- The source or project policy supporting the expectation.

We will prepare these expectations before examining final planner outputs.

We will not require one exact weekly plan when several plans could satisfy the request. Our evaluation will focus on the documented requirements.

Expected behaviors, scenario labels, split assignments, and reviewer notes will remain separate from model-visible profile inputs. We will not place them in prompts or the RAG knowledge base.

AI-generated ratings alone will not establish that a response is correct. Step 17 will define the scoring rules and human review rubric.

### 16.9 Freeze the Final Profiles Before Evaluation

Before the final experiment, we will review and version the profiles, family assignments, scenario categories, and expected behaviors.

We will then freeze the final set. We will not use it to revise prompts, tune the classifier, select retrieval settings, or improve planning rules.

All four planner versions will receive the same final profile inputs. We will evaluate their generated responses before user edits.

Our planned 200 profiles across four versions will produce 800 planner runs. Agent calls, evaluator calls, retries, and development runs will add to the total language-model usage.

If a genuine error is discovered after freezing, we will document the correction and identify which results require rerunning.

Our conclusions will describe performance on this synthetic benchmark. They will not establish real-world athlete outcomes.

## 17. Evaluation Metrics, Scoring Rules, and Human Review Rubric

### 17.1 Purpose

We will define our evaluation approach before comparing models or examining final test results.

Our evaluation will answer two separate questions:

1. How well does our exercise classifier predict the reviewed body-part labels?
2. How well do the four planner versions respond to the athlete profiles?

We will keep these results separate. Correctly classifying an exercise’s body part does not establish that the exercise is suitable for a particular teenage athlete.

For the planners, we will combine automated checks with human review. Each method will have a clearly stated role and limitation.

### 17.2 Define the Classifier Metrics

We will compare the majority-class baseline, TF-IDF logistic regression, and neural classifier using the same frozen split and approved class list from Step 15.

| Measure | What it tells us |
|---|---|
| Macro-F1 — primary measure | The average of the individual class F1 scores, giving each class equal weight |
| Accuracy | The proportion of exercise records classified correctly |
| Per-class precision | How often predictions for a particular class are correct |
| Per-class recall | How many examples belonging to a class the model identifies |
| Per-class F1 | The balance between precision and recall for each class |
| Confusion matrix | Which body-part categories the model commonly confuses |
| Support | The number of records and independent groups supporting each result |

We will use validation macro-F1 for model selection. Final test metrics will be calculated after selecting the models and settings.

We will keep the class list fixed and document zero-division handling. Our planned implementation will use `zero_division=0`, while reporting affected classes and their support.

We will not remove a poorly performing class after seeing its test result.

Where a class has few independent examples, we will explain the uncertainty. We will report whether the neural classifier improves on the baselines rather than assume that it must perform better.

### 17.3 Define the Primary Planner Measure

Our primary planner measure will be the **automated feasible-plan success rate**.

For each version, we will calculate:

**Success rate = feasible cases receiving a complete plan that passes every required automated check ÷ all feasible cases attempted**

Under our current design, the denominator will be the 150 feasible final profiles.

A successful case must produce a usable response in the shared output structure, include the required planning information, and satisfy its predefined required checks.

These checks may cover schedule conflicts, available time, resource requirements, and applicable numerical planning rules.

We will decide which checks are required for each case before running the final experiment. Optional reminders and conditions that cannot be assessed from the supplied information will remain separately identified.

Timeouts, missing outputs, inappropriate refusals, and unparseable responses will remain in the denominator. An unknown result on a required check will not count as a pass.

This measure will describe success against our automated contract. It will not establish clinical safety or prove that every instruction in a plan is appropriate.

### 17.4 Report Additional Planner Measures

We will report the following measures alongside the primary result.

| Measure | Planned reporting |
|---|---|
| Plan completion | Feasible cases containing all required planning fields |
| Constraint results | Met, violated, unknown, and not-applicable results by rule |
| Check coverage | How many expected applicable checks could actually be evaluated |
| Missing-information handling | Whether the response requests the information identified as essential |
| Conflict handling | Whether the response acknowledges the documented conflict |
| Scope handling | Whether the response respects the project’s defined boundaries |
| Evidence quality | Whether factual claims and cited references are supported |
| Reliability and efficiency | Failures, response time, calls, and resource usage |

We will preserve the separate denominators for the 20 missing-information cases, 15 conflicting cases, and 15 requests outside our scope.

Automatic response-type checks will not, by themselves, establish that a clarification or boundary response is appropriate. Human review will assess that distinction on the selected review sample.

We will not combine all categories into one headline score that could reward refusing feasible requests.

### 17.5 Make Rule Results Traceable

Each automated check will record its rule ID, applicability, expected condition, observed information, result, and explanation.

Applicability will come from the reviewed profile and rule definitions. The planner will not decide which requirements it can omit from scoring.

We will retain the four statuses defined in Step 13:

- **Met:** available information supports the condition.
- **Violated:** available information shows a conflict.
- **Unknown:** information or interpretation is insufficient.
- **Not applicable:** the condition does not apply.

We will report unknown results separately rather than quietly exclude them to improve a pass rate.

Annual participation and actual sleep, for example, cannot be established from a weekly schedule alone. Their absence will not automatically make an otherwise complete weekly plan fail unless the case explicitly requires that information.

### 17.6 Evaluate Evidence Beyond Citation Counts

We will distinguish three questions:

1. Does the cited source or record exist?
2. Does it support the associated claim?
3. Are important source-dependent claims supported at all?

A valid link will not automatically establish that a recommendation is supported. We will check relevant details, including age groups, units, quantities, and qualifications.

We will also distinguish published guidance from the planner’s own scheduling choices. Allocating time to a drill does not mean that the source prescribed that exact duration.

For source-dependent claims, we will record missing, incorrect, or misleading attribution. An answer containing no citations will not receive a perfect citation score simply because none can be marked incorrect.

Human reviewers will use the approved references to assess factual support across all four versions, including the version without retrieval.

### 17.7 Define the Human Review Rubric

We will use a simple three-point scale with written examples prepared during development.

| Dimension | 0 — Does not meet expectations | 1 — Partly meets expectations | 2 — Meets expectations |
|---|---|---|---|
| Response suitability | Gives the wrong response type or ignores the central issue | Addresses the issue incompletely | Provides the appropriate plan, clarification, conflict explanation, or boundary response |
| Profile fit | Ignores or contradicts important profile information | Uses some details but misses relevant constraints or goals | Clearly reflects the relevant goals, experience, schedule, and resources |
| Practicality and completeness | Provides an unusable or substantially incomplete plan | Provides a partly usable plan with important gaps | Provides an understandable plan that can be followed with the stated resources |
| Evidence accuracy | Makes materially unsupported or misleading claims | Contains partially supported claims or important omissions | Claims agree with the reviewed evidence and clearly identify uncertainty |
| Clarity | Is confusing or internally inconsistent | Is understandable but contains ambiguity | Gives clear, consistent instructions and explanations |

We will define which dimensions apply to each scenario category. A request needing clarification will not be penalized for lacking a complete weekly plan.

An empty response to a feasible request will count as a failure, rather than disappear from the reviewed results.

We will report scores by dimension and identify not-applicable ratings. We will not hide serious issues inside an overall average.

Reviewers will also record problems that automated checks miss, such as contradictory instructions or unsupported assumptions.

### 17.8 Select a Paired, Blinded Review Sample

We plan to select 40 final profiles for human review before examining their generated responses.

| Scenario category | Profiles selected |
|---|---:|
| Feasible planning requests | 25 |
| Missing essential information | 5 |
| Conflicting constraints | 5 |
| Requests outside our scope | 5 |
| **Total** | **40** |

We will select eight profiles per age: five feasible cases and one from each challenge category.

For every selected profile, we will review the outputs from all four versions. This will produce 160 responses, each independently scored by two reviewers.

We will replace version labels with anonymous identifiers and randomize presentation order. We will preserve the content needed to judge the response, while recognizing that writing style may still provide clues about its origin.

Before final review, we will practice with development examples and resolve ambiguities in the rubric.

We will save both reviewers’ original ratings before discussing disagreements. We will report exact agreement and linearly weighted Cohen’s kappa for each ordinal dimension where it can be calculated.

Undefined agreement statistics will be reported with an explanation. Any later consensus ratings will remain separate from the original ratings.

Our reviewers will be team members, so we will describe this as structured team review rather than independent clinical or coaching validation.

### 17.9 Keep the Comparison Consistent

All four versions will receive the same final profile inputs and shared output requirements.

We will hold the base language model and common generation settings consistent. Versions with retrieval will use the same approved reference collection and lookup data.

Separate scoring code will evaluate saved outputs. We will not accept a planner’s self-reported success as the final result.

Version 4’s internal evaluator may help revise its output, but it will not serve as the final research judge.

We will score responses before user edits and preserve failures. Human-rated results will be reported for the reviewed sample, not presented as judgments of all 800 responses.

In Step 35, we will use paired comparisons and account for exercise groups or related profile families where appropriate. Four versions answering one profile will not be treated as four independent users.

### 17.10 Measure Runtime and Resource Usage

For each planner run, we will record wall-clock time, model calls, retrieval or lookup calls, revisions, retries, and completion status.

Where available, we will also record input and output tokens and clearly labeled usage-cost estimates.

The measurements will include coordinator, specialist, evaluator, and revision calls. We will not report only the final response call for a multi-agent workflow.

We will establish timeouts and retry limits during development and freeze them before final testing. Failed attempts will remain visible in both reliability and resource summaries.

### 17.11 Expected Outputs

After completing this step, we expect to save:

- `docs/evaluation_protocol.md`: metric definitions, denominators, aggregation rules, and failure handling.
- `docs/human_review_rubric.md`: scoring anchors, applicability rules, and development examples.
- `data/evaluation/human_review_sample.csv`: selected profile IDs and sampling information.
- `data/evaluation/human_review_template.csv`: anonymous response IDs and reviewer scoring fields.

Actual scores and experiment logs will be produced in the later evaluation steps.

We will finalize and version this protocol after the development pilot, before accessing final test performance.

## 18. Training-Only Text Preprocessing and Feature Preparation

### 18.1 Purpose

We will prepare the exercise text for our baseline and neural classifiers.

Our goal is to create reproducible model inputs while preserving useful information and keeping the held-out data separate from training.

We will distinguish fixed text preparation from learned preprocessing. Fixed preparation will follow the same documented rules for every record. Vocabulary, word-frequency statistics, and other learned settings will come from training data only.

### 18.2 Define the Model Input

We will construct each input from the exercise name and its available description, using a consistent separator.

| Field | Role |
|---|---|
| Exercise name | Model input |
| Available description | Model input |
| Reviewed body-part label | Prediction target, stored separately |
| Exercise ID and group ID | Alignment and auditing |
| Source and review metadata | Traceability |
| Target-bearing `agent_text` or derived `goal_tag` | Excluded from model features |

We will explicitly select the permitted input fields rather than combine every text column.

Naturally occurring muscle names in the original exercise text will remain valid information. We will not remove them simply because they resemble a target label.

The classifier will not receive athlete profiles, nutrition records, retrieved coaching documents, or evaluation expectations as training features.

### 18.3 Apply Consistent, Minimal Text Preparation

We will preserve the original text and create a separate version for model preparation.

Our initial preparation will normalize whitespace and letter case. We will inspect tokenization to make sure it handles exercise names, abbreviations, hyphenated terms, and useful numbers sensibly.

We will avoid aggressive cleaning that removes information such as movement direction, equipment, anatomy, or words that change an instruction’s meaning.

We will begin without automatic stop-word removal or stemming. Any later change will be documented and assessed using validation results.

A missing description will become empty text for input construction. It will not become the literal string `"nan"` or an invented description.

Title-only examples will remain eligible when they meet the quality requirements established earlier. We will not generate replacement descriptions to make the dataset appear more complete.

### 18.4 Separate Fitting from Transformation

We will follow the same boundary for every candidate preprocessing configuration.

| Partition | Permitted use |
|---|---|
| Training | Fit vocabulary and other learned preprocessing; train model parameters |
| Validation | Apply fitted preprocessing and evaluate development choices |
| Test | Apply the frozen preprocessing during final evaluation in Step 32 |

Validation and test records will not add words to a learned vocabulary or change feature statistics.

We may use validation performance to choose between predefined settings. For each candidate, however, its learned preprocessing will be fitted again using training records only.

We will not inspect test vocabulary coverage or text-length patterns to select a better preprocessing configuration.

### 18.5 Prepare TF-IDF Features for Logistic Regression

Our logistic regression baseline will use TF-IDF features representing words or short word combinations in the exercise text.

We will fit the vectorizer on training inputs. It will establish the vocabulary and inverse-document-frequency values used for subsequent transformations.

We will document settings such as:

- Tokenization rules.
- Word or phrase lengths included.
- Minimum document frequency.
- Maximum vocabulary size.
- Feature normalization.

We will keep the resulting feature matrix sparse rather than unnecessarily converting it into a large dense matrix.

The fitted vectorizer and classifier will ultimately be saved together so that later predictions use the same preparation as training.

The majority-class baseline will not require TF-IDF features. It will determine its prediction from training labels.

### 18.6 Prepare Inputs for the Neural Classifier

Our initial neural approach will use tokens mapped to integer IDs through a vocabulary built from training text.

We will reserve separate IDs for:

- **Padding:** positions added to make sequences suitable for batching.
- **Unknown tokens:** words that are not in the training vocabulary.

Validation and test text will use the existing vocabulary. Unseen words will map to the unknown-token ID rather than expand the vocabulary.

We will use training text lengths to propose sequence-length settings and validation results to assess the practical tradeoff between retained information and computation.

The exercise name will appear first in the combined input. We will document truncation behavior so we know which information may be removed from long descriptions.

If the network combines token representations through pooling, padding positions will be excluded from that calculation.

We will train the neural model’s embedding weights from scratch in Step 20. We will confirm the final architecture there and document any corresponding change to this input preparation.

### 18.7 Preserve the Label Mapping and Training Boundaries

We will create a stable mapping between the approved body-part labels and their numerical class IDs.

This mapping will follow the class list frozen in Step 15. It will remain separate from learning the text vocabulary.

We will check that every included class has training support and that labels map back to their original names correctly.

Unexpected labels will trigger a data check rather than silently create a new output category.

If we use class weights, we will calculate them from training labels only. Any training-data resampling will also remain within the training partition.

We will not change the validation or test distributions to make evaluation easier.

### 18.8 Check the Prepared Features

Before model training, we will verify that:

- Exercise IDs, inputs, and labels remain aligned.
- Record counts agree with the frozen split manifest.
- Excluded metadata and target-derived fields are absent from model inputs.
- TF-IDF feature columns remain consistent across transformations.
- Neural token IDs fall within the vocabulary range.
- Padding and unknown tokens use different IDs.
- Sequence lengths and truncation follow the recorded configuration.
- Repeated preparation with the same configuration produces consistent inputs.

We will examine vocabulary coverage and truncation on training and validation data during development.

Some valid inputs may produce an all-zero TF-IDF vector or mostly unknown neural tokens. We will record these cases and define predictable handling rather than silently remove difficult examples.

A completely empty source input will be treated as a data-quality issue. We will investigate it without quietly changing the approved dataset.

### 18.9 Keep the Model Comparison Consistent

The logistic regression and neural classifiers may use different numerical representations, but they will receive the same underlying text fields and approved records.

We will document their representation differences so readers can understand the comparison.

We will not give one model additional descriptions, body-part tags, or reference information unavailable to the other.

After selecting the preprocessing configuration, we will freeze it with the corresponding model. Application predictions will use that saved preparation rather than rebuild a vocabulary from incoming requests.

## 19. Majority-Class and TF-IDF Logistic Regression Baselines

### 19.1 Purpose

We will build two baseline classifiers before training our neural model.

The first will always predict the most common training label. The second will learn relationships between exercise text and body-part labels using TF-IDF and logistic regression.

These baselines will help us determine whether learning from text improves performance and whether the neural model offers an advantage over a simpler approach.

We will report the results honestly, including cases where logistic regression performs better than the neural classifier.

### 19.2 Confirm the Training and Validation Inputs

We will use the approved exercise records, group assignments, and split manifest from Step 15.

Our input will contain only the exercise name and available description, prepared according to Step 18. The reviewed body-part label will remain separate as the prediction target.

Before fitting either baseline, we will confirm that:

- Training and validation records match the frozen manifest.
- Exercise IDs, text, and labels remain aligned.
- Every included class has training support.
- Target-derived fields are excluded from the input.
- Title-only examples remain included where eligible.

Both baselines and the later neural classifier will be evaluated on the same validation records.

We will not load the final test set for model evaluation in this step. Final test comparisons will take place in Step 32.

### 19.3 Build the Majority-Class Baseline

We will use `DummyClassifier` with the `most_frequent` strategy.

This baseline will identify the most common label in the training partition and predict that label for every validation example. It will ignore the exercise text.

We will record the selected label and its training count. If classes tie, we will verify and document the deterministic choice under our fixed class ordering.

The selected label will remain unchanged during validation. We will not recalculate the majority class using validation or test labels.

We will calculate validation metrics from the actual predictions rather than assume that validation accuracy equals the training majority-class proportion.

This baseline will show how much performance we can obtain from class frequency alone.

### 19.4 Build the TF-IDF Logistic Regression Pipeline

We will combine the TF-IDF vectorizer and logistic regression classifier in one scikit-learn pipeline.

The vectorizer will convert exercise text into numerical features. Logistic regression will use those features to predict one of our approved body-part categories.

We will fit the entire pipeline on training records only. During validation, it will apply the existing vocabulary and feature statistics without learning from the validation text.

Our initial configuration will be:

| Component | Starting setting |
|---|---|
| Input | Exercise name and available description |
| Features | Word TF-IDF |
| Word combinations | Unigrams and bigrams |
| Maximum features | 10,000 |
| Minimum document frequency | 1 |
| Automatic stop-word removal | None |
| Regularization | L2, with `C = 1.0` |
| Solver | `lbfgs` |
| Class weighting | None initially |
| Maximum solver iterations | 1,000 |

These settings are a starting configuration, not a claim that they are optimal.

We will verify the implementation against our installed scikit-learn version and save the actual parameters used. The final vocabulary may contain fewer features than the configured maximum.

We will keep the TF-IDF matrix sparse.

### 19.5 Check Training Behavior and Convergence

We will record fitting time, solver iteration information, and any warnings.

A larger iteration limit will allow more optimization steps, but it will not automatically establish that training converged. We will inspect the reported behavior before accepting a run.

If training raises a warning or fails, we will investigate the cause and document any change. We will not suppress warnings simply to make the notebook output look cleaner.

We will also confirm that every prediction maps back to an approved body-part label.

Nonempty inputs that produce an all-zero TF-IDF vector will remain in evaluation. We will flag them for analysis rather than remove difficult records.

### 19.6 Evaluate Both Baselines on Validation Data

We will use the metrics and scoring conventions defined in Step 17.

| Measure | Planned use |
|---|---|
| Macro-F1 | Primary measure for comparing classifier configurations |
| Accuracy | Overall proportion of correct predictions |
| Per-class precision, recall, and F1 | Identify strengths and weaknesses across categories |
| Confusion matrix | Examine commonly confused labels |
| Record and group support | Show how much independent evidence supports each result |

We will use the same approved label list and `zero_division=0` policy for both models.

The comparison table will identify the model, configuration, dataset version, split version, and validation metrics.

We will not report training performance as evidence of generalization.

### 19.7 Review Validation Errors and Plan Limited Tuning

We will review validation mistakes to understand whether problems involve similar movements, short descriptions, uncommon vocabulary, or ambiguous labels.

We will distinguish a model error from a possible source-label problem. A disagreement with the model will not, by itself, justify changing a label.

Any genuine dataset correction will follow our documented review and versioning process.

In Step 21, we will compare a small, predefined set of settings using validation macro-F1. Candidate changes may include regularization strength, class weighting, and limited TF-IDF settings.

If we test balanced class weights, they will be calculated from training labels. We will not assume that weighting improves every class or the overall result.

For each candidate, we will fit the complete pipeline using training data only and retain its configuration and results.

### 19.8 Preserve Predictions and Interpret the Results

We will save validation predictions with:

- Exercise ID and group ID.
- True label and predicted label.
- Model and configuration identifier.
- Relevant preprocessing flags.

If we save predicted probabilities, we will preserve their class ordering and describe them as model outputs rather than verified confidence estimates.

Our interpretation will address whether logistic regression improves on the majority baseline, which categories remain difficult, and whether limited class support affects the comparison.

These results will evaluate body-part classification. They will not establish that an exercise is appropriate for a teenage athlete or that a personalized training plan is effective.

## 20. Neural Exercise Classifier Design and Training from Scratch

### 20.1 Purpose

We will build a compact neural text classifier that predicts the reviewed body-part category from an exercise name and its available description.

We will train its embeddings and classification layers from scratch using our approved training records.

This will be the trainable exercise-classification component of our project. We will evaluate whether it adds value compared with the majority-class and TF-IDF logistic regression baselines.

### 20.2 Confirm the Inputs and Training Boundaries

We will use the frozen split from Step 15 and the input preparation from Step 18.

Before training, we will confirm that:

- Inputs contain only the exercise name and available description.
- The vocabulary was built from training text only.
- Labels follow the approved class-to-index mapping.
- Title-only examples remain included where eligible.
- Group assignments remain separate across partitions.
- Validation data does not update vocabulary or model parameters.

We will reserve the final test set for Step 32.

The output layer will match the number of approved classes. We will not hard-code all original categories if the eligibility review changes the supported class list.

### 20.3 Define the Initial Network

Our initial model will use trainable token embeddings followed by masked mean pooling and a small classification network.

| Component | Initial design | Purpose |
|---|---|---|
| Token input | Integer IDs from the training vocabulary | Represent the prepared exercise text |
| Embedding layer | 64 dimensions per token | Learn numerical representations of words |
| Masked mean pooling | Average non-padding token representations | Create one vector for each exercise |
| Hidden layer | 64 units with ReLU activation | Learn combinations of text features |
| Dropout | 0.2 | Apply regularization during training |
| Output layer | One raw score per approved class | Produce the classification scores |

Mean pooling will exclude padding from both the sum and the number of tokens used in the average.

We selected this compact starting architecture to keep the training procedure manageable and reproducible. We will report its actual vocabulary size and parameter count after implementation.

Because averaging loses word order, we will describe this as a bag-of-tokens model. We will not claim that it understands detailed exercise instructions.

### 20.4 Handle Padding, Unknown Tokens, and Text Length

We will use separate IDs for padding and unknown tokens.

The sequence-length setting and truncation behavior will follow the configuration prepared in Step 18. We will not change them after inspecting final test text.

Before batching, we will check that every included input produces a valid token sequence. We will investigate genuinely empty source inputs rather than silently remove them.

Valid text containing unseen words will use the unknown-token ID. We will record low vocabulary coverage, including inputs that become entirely unknown tokens.

We will also record whether the unknown token occurs during training. If it does not, its embedding will not receive a training signal, which will limit how we interpret predictions for unfamiliar text.

### 20.5 Define the Loss and Initial Training Settings

We will use multiclass cross-entropy loss with the reviewed numerical class IDs.

The network will return raw class scores, also called logits. We will pass those scores directly to the loss function without applying softmax first.

Our initial settings will be:

| Setting | Starting value |
|---|---|
| Optimizer | AdamW |
| Learning rate | 0.001 |
| Weight decay | 0.0001 |
| Batch size | 32 |
| Maximum epochs | 30 |
| Early-stopping patience | 5 validation checks without improvement |
| Checkpoint selection | Highest validation macro-F1 |
| Class weighting | None initially |
| Initial random seed | 42 |

These are starting choices. We will assess a limited set of alternatives in Step 21 using validation data.

If we later introduce class weights, we will calculate them from training labels only and document the loss configuration.

We will use the working device verified earlier and record whether the run used a CPU or GPU.

### 20.6 Check the Implementation Before a Full Run

We will first inspect a small batch of training examples.

We will check that:

- Token IDs and class IDs are valid.
- Input and target tensors use the expected types.
- The output contains one score per class for every example.
- Padding is excluded from pooling.
- Loss and gradients contain finite values.
- Model parameters receive updates during a training step.

If necessary, we will use a small training-only diagnostic to investigate whether the model can learn from a limited set of examples.

After diagnostic updates, we will initialize a fresh model and optimizer for the recorded experiment. We will not carry those updates into the main run without accounting for them.

### 20.7 Run Separate Training and Validation Loops

During training, we will enable training mode and shuffle the training examples.

For each batch, we will clear previous gradients, calculate predictions and loss, backpropagate the loss, and update the model parameters.

During validation, we will switch to evaluation mode and disable gradient recording. Dropout will be inactive, and validation will not update any model weights.

We will calculate validation metrics across the complete validation partition using the fixed class list and scoring conventions from Step 17.

For the initial unweighted loss, we will calculate epoch averages using the actual number of examples in each batch rather than give a smaller final batch equal weight.

We will record training loss, validation loss, validation macro-F1, validation accuracy, and epoch duration.

If a run produces invalid values or fails, we will preserve the failure information and investigate before continuing.

### 20.8 Select and Restore the Best Checkpoint

After each epoch, we will compare validation macro-F1 with the best value recorded so far.

We will save a new best checkpoint only when the score strictly improves. Equal scores will retain the earlier checkpoint.

A strict improvement will reset the early-stopping counter. Otherwise, we will increase the counter and stop after five consecutive validation checks without improvement, or after reaching the maximum epoch count.

We will restore the best checkpoint before producing the initial model’s validation report.

The last training epoch will not automatically become the selected model.

### 20.9 Review Learning Behavior and Compare with the Baselines

We will plot training and validation loss together and show validation macro-F1 across epochs.

These plots will help us identify learning problems, unstable behavior, or signs of overfitting. A decreasing training loss alone will not establish better performance on unseen exercises.

We will compare the selected neural checkpoint with both baselines on the same validation records.

Our comparison will include macro-F1, accuracy, per-class results, and support. We will examine which categories improve and which remain difficult.

Predictions will use the highest-scoring class. If we display softmax probabilities, we will describe them as uncalibrated model outputs rather than verified confidence.

Body-part predictions will support exercise tagging. They will not establish youth suitability or training effectiveness.

### 20.10 Preserve the Run and Expected Outputs

For each recorded run, we will retain its configuration, random seeds, device, software versions, data and split identifiers, vocabulary version, label mapping, and selected epoch.

After completing this step, we expect to save:

- `artifacts/neural/initial_model_state.pt`: the best checkpoint from the initial run.
- `artifacts/neural/initial_run_config.json`: architecture, preprocessing references, and training settings.
- `results/neural/initial_training_history.csv`: epoch-level losses, metrics, and timings.
- `results/neural/initial_validation_predictions.csv`: predictions linked to exercise and group IDs.
- `results/figures/neural_initial_learning_curves.png`: learning-curve figures.

We will save the checkpoint with enough information to reconstruct the network and its inputs. Step 22 will package the selected model and preprocessing together.

We will report reproducibility settings without promising identical numerical results across different hardware or software versions.

Actual training outcomes will be added after execution. Step 21 will cover controlled tuning and model selection.

## 21. Validation, Hyperparameter Tuning, and Model Selection

### 21.1 Purpose

We will compare a small, predefined set of model settings and select the configurations we will carry into final evaluation.

Our primary selection measure will remain validation macro-F1, as defined in Step 17.

We will record every attempted configuration, including unsuccessful runs. This will let us explain our choices without presenting only the strongest results.

### 21.2 Establish the Comparison Rules

All candidates will use the same approved exercise records, class list, and grouped split.

We will keep the input fields and scoring conventions consistent. Vocabulary, feature statistics, and class weights will be learned from training data only.

For this initial search, we will hold the following settings fixed:

- Exercise text construction and cleanup.
- TF-IDF feature settings from Step 19.
- Neural vocabulary and sequence preparation.
- Neural embedding and hidden-layer dimensions.
- Batch size, optimizer, and maximum epoch count.
- Early-stopping and checkpoint-selection rules.

We will change only the parameters listed below.

Every candidate will be evaluated against the existing validation partition. We will record this as a fixed-holdout search.

### 21.3 Compare Six Logistic Regression Configurations

We will compare three regularization settings with two class-weight options.

| Parameter | Candidate values |
|---|---|
| Regularization strength parameter, `C` | 0.1, 1.0, 10.0 |
| Class weighting | None, balanced |

This will produce six configurations.

We will evaluate candidates in the listed `C` order, testing no weighting before balanced weighting for each value.

For each candidate, we will fit the complete TF-IDF and logistic regression pipeline on training data. Balanced weights will be calculated from training labels.

We will select the configuration with the highest validation macro-F1. On an exact tie, we will retain the first candidate in the predefined order.

The initial configuration from Step 19 is included in this comparison. We can reuse that run if its data, preprocessing, software, and model settings match the recorded experiment.

### 21.4 Compare Four Initial Neural Configurations

We will begin with the model from Step 20 and assess three targeted changes.

| Candidate | Learning rate | Dropout | Class weighting |
|---|---:|---:|---|
| N0 — Initial model | 0.001 | 0.2 | None |
| N1 — Lower learning rate | 0.0003 | 0.2 | None |
| N2 — Higher dropout | 0.001 | 0.4 | None |
| N3 — Weighted loss | 0.001 | 0.2 | Training-frequency weights |

Each alternative changes one setting relative to N0. This will help us interpret the comparison, although it will not test every possible interaction between settings.

For N3, each class weight will be calculated as:

**Training record count ÷ (number of approved classes × training record count for that class)**

We will train all four candidates with seed 42 and the same stopping procedure. Every run will start with a fresh model and optimizer.

The matching initial run from Step 20 may be reused.

We will rank these candidates by their best validation macro-F1, using the order N0 through N3 to resolve exact ties.

### 21.5 Check Training Variability for the Leading Neural Candidates

We will take the two leading neural configurations and repeat each with seeds 17 and 73.

Each shortlisted configuration will therefore have three runs: seeds 42, 17, and 73.

The seed changes will affect training randomness while keeping split membership unchanged.

We will compare the shortlisted configurations using their mean validation macro-F1 across the three runs. We will also report the individual scores and their standard deviation.

On an exact mean-score tie, we will retain the earlier candidate in the predefined order.

Only the shortlisted configurations will receive these additional runs. We will describe this as a practical, limited selection procedure.

This plan produces eight neural runs in total: four initial runs and four additional runs. It will not produce eight additional runs.

The variation across seeds will describe training variability on this split. It will not measure uncertainty across different athlete populations or independent datasets.

### 21.6 Select the Checkpoints and Application Candidate

We will retain the best-epoch checkpoint from each neural run using the rule established in Step 20.

After selecting the neural configuration, we will use its seed-42 checkpoint as the primary neural artifact. We will not choose whichever seed happened to score highest.

Our comparison table will distinguish:

- The selected logistic regression model’s validation score.
- The selected neural configuration’s three-seed mean and variation.
- The seed-42 neural checkpoint’s own validation score.
- The majority-class baseline.

For application integration, we will compare the selected logistic regression artifact with the selected seed-42 neural artifact using validation macro-F1.

On an exact tie, we will prefer logistic regression because it has a simpler inference pipeline.

If neither learned model exceeds the majority baseline’s macro-F1, we will document that limitation and retain reviewed catalog labels for the application.

The trained neural classifier will remain part of our research comparison regardless of which model we integrate.

### 21.7 Review Results Without Changing the Selection Measure

Alongside macro-F1, we will examine accuracy, per-class precision and recall, confusion patterns, and class support.

These results will help us explain strengths and weaknesses. They will not become alternative selection measures chosen after we see which model performs best.

We will use loss curves to diagnose training behavior. We will not directly rank weighted and unweighted candidates by loss values produced under different loss definitions.

We will also examine whether apparent improvements depend on very small classes or a few related exercise groups.

Repeated validation use for early stopping and tuning means these scores are development evidence. The final test set will provide a separate assessment in Step 32.

### 21.8 Record Failures and Control the Search Budget

We will preserve failed runs, convergence warnings, invalid losses, and incomplete experiments in the run log.

A neural configuration will need valid results from all three scheduled shortlist runs to qualify for selection by its three-seed mean. We will not average only its successful runs.

If an implementation error requires a correction, we will document the change and rerun the affected comparisons consistently.

We will not keep adding configurations until the neural model wins.

Any necessary amendment to this search plan will be documented before final testing, including its reason and additional runs.

### 21.9 Freeze the Selected Models

After selection, we will freeze:

- The approved data and split versions.
- The selected configurations and checkpoint identifiers.
- The vocabulary, vectorizer, and label mapping.
- The preprocessing and sequence settings.
- The selection rules and supporting validation results.

We will retain the models fitted on the original training partition for the planned final comparison. We will not silently retrain them on combined training and validation data.

Any classifier used by the planning workflows will remain fixed during their final comparison.

We will not switch the application’s classifier after inspecting final test results. Any later revision will be documented as a separate development cycle.

After completing this step, we expect to save:

Athlete profiles, evaluation expectations, reviewer notes, and generated plans will remain outside the index. Held-out classifier records will remain outside the development retrieval collection under the boundary established in Step 22.

### 23.3 Create Passages That Preserve Meaning

We will divide permitted reference text into passages that can be retrieved and cited.

We will organize passages around meaningful units, such as a drill record or a guideline statement with its qualifications.

For drills, we will preserve the relationship between instructions, equipment, participants, and progression.

For guidelines, we will retain the relevant age or grade group, measurement units, definitions, and footnotes.

Our initial target will be approximately 150–200 embedding-model tokens per passage, including headings. We will check the actual tokenizer limit and prevent silent truncation.

For the initial MiniLM candidate, the published default truncation limit is 256 word pieces.

Where a complete record is longer, we will create linked passages pointing back to the full record. Retrieval will be able to recover the required supporting context.

We will not separate a numerical limit from a qualification that changes its interpretation.

### 23.4 Preserve Source and Passage Metadata

Each passage will include:

- A stable source ID and passage ID.
- Its parent drill record or source section.
- Publisher, title, URL, and page or section locator.
- Source version and verification date.
- Relevant skill, resource, or applicability metadata.
- Permission and review status.
- Content hash and corpus version.

We will distinguish source-supported fields from our own interpretations.

Jr. NBA skill levels will remain skill levels. We will not convert them into invented age bands.

Where several passages belong to the same underlying evidence unit, we will preserve that relationship so repeated fragments do not appear to be independent evidence.

### 23.5 Compare Two Retrieval Approaches

We will begin with two approaches over the same approved passages.

| Approach | Initial implementation |
|---|---|
| Lexical retrieval | TF-IDF similarity based on words and short word combinations |
| Semantic retrieval | Similarity between pretrained passage and query embeddings |

Our initial semantic candidate will be `sentence-transformers/all-MiniLM-L6-v2`.

We will verify access, license information, local feasibility, and the model revision before adoption. Its pretrained weights will remain fixed during this initial comparison.

The retrieval vectorizer will be fitted on the approved reference corpus. It will be separate from the classifier’s training-only TF-IDF pipeline.

For semantic retrieval, we will use consistent query and passage encoding, normalization, and similarity settings. We will record these settings with the index.

For our initial collection, we will rank saved embeddings directly using cosine similarity. We can introduce a dedicated vector index later if collection size or measured performance justifies it.

### 23.6 Define Retrieval and Context Assembly

Each retrieval request will include the relevant goal or question and any explicit profile constraints.

We will apply filters only when their meaning is supported by the source metadata. Unknown equipment requirements or age applicability will remain visible rather than become invented matches.

Our starting retrieval depth will be five distinct evidence units. We will remove repeated results from the same unit while preserving the ranking rule.

Before passing results to the language model, we will recover linked definitions, footnotes, or parent-record details needed to interpret them.

We will record both the ranked results and the context actually delivered after expansion and context-budget handling.

If required evidence cannot fit, we will record incomplete context rather than silently remove an important qualification.

Mandatory planning checks will access the structured rule register independently of search ranking. USDA calculations will use the structured food lookup.

Similarity scores will describe ranking. They will not become verified confidence scores or proof that a passage answers the question.

### 23.7 Prepare a Separate Retrieval Benchmark

We plan to create 60 reviewed retrieval questions: 30 for development and 30 held out for final assessment.

This question-to-reference benchmark will evaluate retrieval separately from our athlete-profile planning benchmark.

Each partition will initially contain:

| Question category | Development | Held out |
|---|---:|---:|
| Answerable drill and resource questions | 10 | 10 |
| Answerable guideline and qualification questions | 10 | 10 |
| Requests unsupported by the approved corpus | 10 | 10 |
| **Total** | **30** | **30** |

We will include questions requiring multiple pieces of evidence, such as a participation limit together with its definition or qualification.

Paraphrases of the same underlying question will remain in one partition.

Two teammates will verify the expected evidence against the original records. We will identify acceptable source units and any complete evidence bundles needed to support an answer.

The original reference passages will remain searchable. Questions, relevance judgments, and expected responses will be stored separately and excluded from retrieval.

We will report actual approved counts after review.

### 23.8 Define Retrieval Metrics

For answerable questions, we will evaluate results at the same retrieval depths: 1, 3, and 5.

| Measure | Meaning |
|---|---|
| Hit@k | Whether at least one reviewed relevant evidence unit appears in the first k results |
| Recall@k | The proportion of reviewed relevant evidence units recovered in the first k results |
| MRR@5 | The average reciprocal rank of the first relevant result, with zero when none appears in the first five |
| Complete-evidence coverage | Whether the delivered context contains an accepted complete evidence bundle |
| Retrieval time | Time required to retrieve and assemble the context |

We will map passage results to their underlying evidence units and count repeated units once.

A relevant fragment may satisfy Hit@k while still omit information needed for a complete answer. Complete-evidence coverage will help us identify that difference.

We will state that recall is measured against our reviewed relevance judgments. An unlisted passage will require review before we automatically describe it as irrelevant.

For unsupported questions, we will separately examine whether retrieval presents misleading candidates as sufficient evidence.

Any automatic rejection threshold will require development-set validation and a documented decision rule. Without that validation, returned passages will remain candidates requiring an evidence-adequacy check.

### 23.9 Select and Freeze the Retrieval Configuration

We will select the retriever using development Recall@5, followed by MRR@5 to resolve equal scores.

If both measures tie exactly, we will retain lexical retrieval as the simpler implementation.

We will also report complete-evidence coverage, unsupported-question behavior, latency, and context size. These findings will guide any documented development correction.

Before final assessment, we will freeze:

- Approved corpus and passage versions.
- Evidence-unit mappings.
- Retrieval method and model revision.
- Encoding and similarity settings.
- Filtering, deduplication, and context-expansion rules.
- Retrieval depth and context budget.
- Any adopted rejection policy.

We will run the held-out retrieval assessment after the development protocol is frozen in Step 31. Its results will not be used for further tuning within the same reported experiment.

Planner Versions 2–4 will use the same adopted retrieval configuration and reference collection.

### 23.10 Expected Outputs

After completing development, we expect to save:

- `data/processed/rag_passages.jsonl`: approved passages and source metadata.
- `artifacts/retrieval/`: the versioned retriever, index data, and configuration.
- `data/evaluation/retrieval_questions_development.jsonl`: development questions.
- `data/evaluation/retrieval_questions_test.jsonl`: held-out questions.
- `data/evaluation/retrieval_relevance.jsonl`: reviewed evidence mappings and required bundles.
- `results/retrieval/development_metrics.csv`: development comparisons.
- `results/retrieval/development_results.jsonl`: ranked results and delivered-context records.
- `docs/retrieval_design.md`: corpus decisions, retrieval settings, and limitations.

Held-out results will be saved separately when the final assessment is run.

## 24. Language Model Integration and Reference Lookup Tools

### 24.1 Purpose

We will connect our language model to the approved reference collection and structured lookup functions.

This integration will provide a shared foundation for the four planner versions. It will handle message formatting, generation settings, tool access, output parsing, and execution records.

We will keep these shared functions separate from the specialist coordination and evaluator workflows introduced in Steps 25–28.

### 24.2 Confirm the Model and Runtime

Our initial language-model candidate remains `meta-llama/Llama-3.1-8B-Instruct`, subject to the access and hardware checks established in Step 7.

Before adopting it, we will verify that the selected runtime can handle representative planning requests, reference context, and the required output format.

We will record:

- Model identifier and exact revision.
- Tokenizer and chat-template version.
- Runtime and dependency versions.
- Execution device.
- Numerical precision or quantization settings.
- Supported context length and our working context budget.
- Effective decoding settings and output-token limit.

We will use the same base model, runtime, and common generation settings across all four planner versions.

Our initial development approach will use greedy decoding where supported. We will record the effective settings rather than assume that a seed guarantees identical outputs across environments.

If we change the model or runtime during development, we will update the shared configuration and repeat affected checks before final evaluation.

### 24.3 Define the Shared Input Contract

Each planner will receive the same model-visible athlete information for a given case.

Our input adapter will standardize field names and data types while preserving missing information, conflicting commitments, and uncertainty.

It will not infer an athlete’s grade, invent available equipment, or remove commitments to make a request easier.

The common task instructions will describe our planning scope, response requirements, and expected handling of uncertainty.

Retrieved passages and lookup results will be added only where the planner version permits them.

We will start a fresh conversation and reset workflow state for every profile and planner version. Generated responses from earlier runs will not become reference material for later cases.

### 24.4 Implement Narrow Reference Tools

We will prepare the following read-only functions.

| Tool | Purpose | Main result |
|---|---|---|
| `search_references` | Find relevant approved passages | Ranked evidence, source identifiers, and context status |
| `get_drill` | Retrieve an approved basketball drill by ID | Instructions, skill category, resources, and source |
| `get_exercise` | Retrieve an eligible exercise record | Exercise details, label origins, and review status |
| `get_planning_rules` | Retrieve potentially applicable participation guidance | Rule values, applicability, definitions, and qualifications |
| `lookup_food` | Retrieve verified food facts | Food identity, nutrient values, units, and reference amount |

Each tool will have documented argument types, required fields, return fields, and error behavior.

Results will distinguish statuses such as found, ambiguous, not found, unavailable, and invalid arguments.

Every factual result will retain its source and data version. A failed lookup will remain visible rather than become an invented answer.

### 24.5 Preserve the Meaning of Structured Lookups

The planning-rule tool will use the reviewed rule register from Step 13. It will preserve age and grade applicability, definitions, and exceptions.

If applicability is unresolved, the result will explain that uncertainty.

Retrieving a rule will be separate from checking whether a generated plan follows it. Generation-time coded checking will belong to Version 4’s workflow.

Food lookup will use the prepared local USDA records from Step 14. Nutrient calculations will be performed in code using verified units and food-specific weights.

An ambiguous food name will return candidates or request clarification. We will not silently choose between different preparation methods.

Exercise lookup will preserve source labels, reviewed labels, and classifier predictions as separate information. It will not turn a predicted body-part tag into a youth-suitability judgment.

### 24.6 Keep Tool Execution Under Application Control

Our first implementation will let the Python workflow invoke the required lookup functions.

If we enable model-requested tool calls, we will first verify the model’s request format and the runtime’s support. The adapter will validate the tool name and arguments before executing an approved function.

We will use an explicit list of permitted tools. Requests will not become arbitrary Python commands, file operations, or unrestricted web searches.

Retrieved text will be supplied as reference content with source identifiers. It will not replace our application instructions.

Invalid requests and unsuccessful calls will produce structured results. Any follow-up call or retry will count toward the recorded execution budget.

We will document and freeze the adopted calling approach before the final comparison.

### 24.7 Preserve the Four Planner Configurations

We will enforce access through the workflow configuration.

| Version | Reference access during generation | Planning workflow |
|---|---|---|
| 1 — LLM alone | No retrieval or lookup results | Single planner |
| 2 — LLM with references | Approved retrieval and lookup tools | Single planner using reference information |
| 3 — Specialist workflow | Same reference collection and tools | Specialists coordinated into a final response |
| 4 — Evaluated specialist workflow | Same reference collection and tools | Specialists, internal evaluation, coded checks, and bounded revisions |

Versions 2–4 will use the same adopted corpus, retriever, lookup implementations, and data versions.

They may use different numbers of calls because their workflows differ. We will record the complete activity and resource usage.

Independent final evaluation will apply to every version after generation. Its feedback will not be returned to Versions 1–3 to improve their responses.

Version 4’s internal evaluation results will remain separate from the final research scores.

### 24.8 Define a Shared Output Structure

We will request a structured response that supports the different scenario categories.

| Output component | Expected content |
|---|---|
| Response type | Plan, clarification, constraint conflict, or outside scope |
| Weekly schedule | Planned activities and rest arrangements when a plan is appropriate |
| Activity details | Day, activity, allocated duration, and relevant resource requirements |
| Existing commitments | How supplied practices, games, and other commitments are accounted for |
| Questions and unresolved issues | Missing information or conflicts needing attention |
| Explanation | A brief explanation of the proposed response |
| Evidence references | Source or record identifiers when available |
| Optional food facts | Verified facts returned by the food lookup |

Required fields will depend on the response type. A clarification response will not need a complete weekly schedule.

Catalog and source identifiers will be optional where unavailable. Version 1 will not be required to invent identifiers from a collection it cannot access.

We will evaluate evidence quality separately from basic output structure.

We will preserve the raw generated response and its parsing status. Parsing will not silently add missing durations, replace activities, or correct unsupported claims.

Any evaluator-driven revision will follow Version 4’s explicit procedure.

### 24.9 Manage Context, Limits, and Execution Records

We will measure the complete input, including instructions, profile information, reference passages, and tool results.

Our context assembly will preserve the qualifications identified in Step 23. If essential material cannot fit, we will record the limitation.

We will define maximum model calls, tool calls, timeouts, output lengths, and revision limits for each workflow. Steps 25–28 will describe these budgets, and Step 31 will confirm them through development runs.

Execution records will include:

- Model, prompt, corpus, and tool versions.
- Observable model responses and parsing outcomes.
- Tool names, validated arguments, results, and statuses.
- Timing, token usage where available, and cache status.
- Retries, revisions, timeouts, and failures.

We will use short decision summaries where needed to explain workflow actions.

Credentials will remain outside saved prompts and logs. Reference caches will be versioned and kept separate from generated plans and evaluation records.

### 24.10 Verify Integration with Development Cases

Before building the complete planners, we will check the integration using development examples.

Our checks will cover:

- A complete and feasible planning request.
- Missing essential profile information.
- A request containing conflicting commitments.
- An unknown drill or exercise ID.
- An ambiguous food query.
- A reference lookup returning no suitable result.
- Invalid tool arguments.

- Workflow reset between independent cases.

We will also verify that each planner receives only its permitted tools and reference material.

These checks will use development profiles and development retrieval questions. Final athlete profiles and held-out retrieval questions will remain reserved for their planned assessments.

## 26. Planner Version 2 — Language Model with RAG

### 26.1 Purpose

### 26.2 Define the Execution Contract

| Component | Initial configuration |
|---|---|
| Language-model generation | One logical generation request per profile |
| Reference searches | Up to two predefined searches |
| Results per search | Up to five distinct evidence units |

| Food lookup | Optional, when explicitly relevant to the request |
| Specialist or evaluator calls | None |

Two searches may return up to ten distinct evidence units before deduplication. Required rule information and linked qualifications will be accounted for separately.

We will keep the base model, runtime, precision, and common generation settings consistent with Version 1.

We will confirm context, timeout, and optional lookup limits during the development pilot in Step 31.

### 26.3 Prepare References from the Supplied Profile

We will use a reusable Python function to prepare the reference packet.

The preparation process will:

1. Read the approved model-visible profile.
2. Construct search queries using predefined templates.
3. Retrieve candidate passages from the frozen reference collection.
4. Fetch potentially applicable participation rules.
5. Retrieve full approved drill or exercise records where needed.
6. Add any explicitly requested food facts.
7. Assemble the available evidence within the context budget.

Our initial search templates will cover basketball goals and resources, followed by relevant supporting guidance.

Query construction will use supplied profile facts. It will preserve missing information and avoid inventing age, grade, equipment, or available time.

We will perform this preparation without an additional language-model call for query rewriting.

The preparation function will collect evidence. It will not construct the weekly schedule, remove conflicting commitments, or choose the final response category.

### 26.4 Assemble a Traceable Reference Packet

Each included reference will retain its source ID, record or passage ID, version, and relevant locator.

The packet will distinguish:

- Basketball drill instructions and resource requirements.
- Approved exercise information and label origins.
- Participation guidance and applicability conditions.
- Definitions and qualifications needed to interpret that guidance.
- Optional food facts with units and reference amounts.
- Missing, ambiguous, or unavailable information.

We will remove repeated evidence units while retaining the source relationships needed for citation.

We will use the context-assembly rules from Step 23. Complete evidence bundles will take priority over isolated fragments that lose essential qualifications.

If the context budget prevents adequate coverage, we will record the omitted material and indicate that the evidence is incomplete.

The reference packet will contain source material and lookup results. Evaluation expectations and answer keys will remain outside it.

### 26.5 Keep Rule Lookup Separate from Plan Checking

The planning-rule lookup will retrieve the guidance and qualifications potentially relevant to the supplied age, grade, and activity context.

If applicability is unresolved, that uncertainty will remain in the packet. We will not automatically choose whichever category permits more activity.

This lookup will occur before the plan exists. It will not provide a verdict on whether a generated schedule passes our checks.

Version 2 will receive the guidance but will be responsible for applying it when generating its response.

Independent evaluation will check the saved output afterward. That feedback will not be returned to Version 2 for correction.

### 26.6 Handle Catalog and Food Lookups Carefully

We will expand only approved drill or exercise IDs linked to the retained search results, within the recorded lookup limit.

The model will receive available instructions, resources, skill categories, and review status. A classifier-generated body-part tag will retain its prediction origin.

We will preserve the distinction between a source prescription and a planning choice. For example, time allocated to a drill will not automatically become a duration endorsed by its source.

Food lookup will run only when the supplied request makes it relevant. It will use the structured records and calculations prepared in Step 14.

Ambiguous food matches will remain ambiguous. Missing nutrient values or portion weights will remain unavailable rather than become generated estimates.

We will record the request-based food-lookup limit in the configuration before final evaluation.

### 26.7 Generate One Response from the Profile and References

We will construct the model input from:

1. The shared task instructions.
2. The unchanged model-visible profile.
3. The assembled reference packet.
4. The shared output requirements.

The prompt will ask the planner to use the supplied evidence accurately, identify unresolved information, and distinguish source-supported statements from its own scheduling choices.

When citing supplied material, it will use the corresponding reference identifiers.

We will then make one generation request and preserve the returned response.

Version 2 will not request specialist opinions, receive internal evaluator feedback, or make post-generation lookups to revise its answer.

We will reset conversation state between profiles. Previous plans will not become reference material for subsequent cases.

### 26.8 Preserve Missing Evidence and Output Failures

A search may return weak matches, no suitable evidence, or records with unresolved applicability.

We will pass those statuses clearly to the planner. A high similarity score alone will not establish that a passage supports the requested recommendation.

The model may need to ask a focused question or explain a limitation. We will evaluate whether that response is appropriate for the case.

We will preserve the raw output before applying the shared parser.

Malformed output, missing fields, truncation, unsupported citations, and unexpected tool requests will remain visible in the run record.

Parsing will not repair the schedule or replace unsupported claims. Version 2 will retain its single-generation execution contract.

We will use development profiles and development retrieval questions to check:

- Whether queries reflect the supplied goals and constraints.

- Whether duplicate passages are handled consistently.
- Whether ambiguous or missing results remain visible.

## 27. Planner Version 3 — RAG with Specialist Agents

### 27.1 Purpose

We will extend the RAG planner by dividing the preparation work among specialist roles.

Each specialist will contribute focused recommendations or considerations. A coordinator will then combine those contributions into one response using the original athlete profile and supporting references.

These agents will be role-specific calls to our shared language model. We will document their instructions and evaluate whether this organization improves the resulting plans.

Version 3 will follow a fixed sequence. Evaluation feedback and revisions to a completed plan will be introduced in Version 4.

### 27.2 Define the Specialist Roles

Our initial workflow will contain two core specialists, an optional food-facts specialist, and a coordinator.

| Role | Responsibility | Expected contribution |
|---|---|---|
| Training specialist | Consider basketball goals, experience, resources, and eligible conditioning activities | Candidate activities, progression considerations, resource requirements, and evidence references |
| Participation and recovery specialist | Interpret applicable participation guidance and existing commitments | Relevant limits, rest considerations, qualifications, and unresolved information |
| Food-facts specialist | Address an explicit food question using the prepared lookup results | Verified food facts, units, source references, and limitations |
| Coordinator | Combine the original information and specialist contributions | One response using the shared output structure |

The training specialist will use approved drill and exercise information.

The participation and recovery specialist will work before the coordinator drafts the weekly plan. It will provide prospective guidance rather than score or critique a finished schedule.

The food-facts specialist will stay within the factual scope established in Step 14.

### 27.3 Reuse the Shared Reference Packet

For a matched profile and corpus version, we will use the same prepared reference packet as Version 2.

This will preserve the same source collection, query templates, retrieval settings, rule lookups, and available catalog records.

Each specialist will receive the original profile and the shared references. Its role instructions will identify the contribution it should prepare.

The coordinator will receive those original inputs together with the specialist notes.

Our initial Version 3 workflow will not add further searches between agent calls. This will keep reference access consistent while we examine the additional coordination stages.

We will record the packet version and hash so we can verify which material each role received.

During the development pilot, we will reserve sufficient context space for compact specialist notes while preserving the shared reference packet and its essential qualifications.

### 27.4 Use a Fixed Execution Sequence

We will implement the initial workflow through the shared Python model adapter.

For each profile, we will:

1. Prepare the approved reference packet.
2. Generate the training specialist’s contribution.
3. Generate the participation and recovery contribution.
4. Generate the food-facts contribution when the routing rule activates it.
5. Give the original inputs and available specialist contributions to the coordinator.
6. Save the coordinator’s response and the complete execution record.

We will initially execute the specialist calls sequentially to keep resource use predictable.

The specialists will not receive one another’s generated notes. The coordinator will be the stage where their contributions are combined.

We will reset conversation and workflow state between profiles.

### 27.5 Define When the Food-Facts Role Runs

We will activate the food-facts specialist through a predefined rule identifying an explicit food question in the model-visible request.

A general goal such as improving basketball fitness will not automatically trigger dietary recommendations.

The routing decision will use the supplied request, not its evaluation category or expected answer.

If the role is activated, it will receive the available food lookup results, including any ambiguous or unavailable status.

It will explain the verified facts and their limitations. Nutrient and portion calculations will remain the responsibility of the structured tools.

We will use the same routing rule in the corresponding Version 4 workflow.

### 27.6 Request Compact Specialist Contributions

We will ask each specialist for structured notes rather than a separate complete weekly plan.

The notes will include:

- The role’s relevant recommendations or considerations.
- Candidate record IDs and activity details where available.
- Required equipment, space, or participants.
- Supporting source identifiers.
- Applicable conditions and qualifications.
- Missing information, conflicts, or evidence limitations.

We will set compact output limits during development and record them in the configuration.

Specialist statements will remain model-generated suggestions. They will not become new source records or verified facts simply because an agent produced them.

We will preserve raw specialist responses alongside their parsed forms.

### 27.7 Define the Coordinator’s Responsibilities

The coordinator will produce one final response using the shared conditional schema from Step 24.

It will consider the specialist contributions alongside the original profile and primary reference material.

Its responsibilities will include:

- Keeping the supplied commitments, availability, and resources intact.
- Selecting activities relevant to the stated goals.
- Combining compatible recommendations into a coherent schedule.
- Resolving conflicting suggestions using the original facts and references.
- Preserving qualifications attached to published guidance.
- Identifying essential uncertainty or an unresolved conflict.
- Explaining the resulting response clearly.

Agreement between specialists will not establish that a claim is correct. Their suggestions cannot override verified food values, source qualifications, or the athlete’s supplied information.

Where the available information does not support a complete plan, the coordinator may produce the appropriate clarification, conflict, or boundary response.

The coordinator will generate its response once. It will not receive post-generation checking feedback in this version.

### 27.8 Set Call Limits and Failure Handling

Our nominal generation budget will be:

| Stage | Model calls |
|---|---:|
| Training specialist | 1 |
| Participation and recovery specialist | 1 |
| Food-facts specialist | 0 or 1 |
| Coordinator | 1 |
| **Total** | **3 or 4** |

This count excludes reference preparation and any separately recorded transport attempts.

A valid specialist response reporting insufficient evidence will remain a usable contribution. A failed call or malformed mandatory contribution will be recorded as an execution failure.

If the optional food role is activated, its contribution will be required for that run’s intended workflow.

We will not silently replace a failed specialist workflow with a single-planner response and label it as a completed Version 3 run.

The shared transport-retry policy will apply consistently, with every attempt recorded.

### 27.9 Keep Evaluation Outside the Generation Workflow

We will parse and preserve the coordinator’s output using the shared rules.

Independent evaluation will take place after generation. Its findings will not trigger another specialist call, coordinator pass, or content revision within Version 3.

Each profile will produce at most one final coordinator response for the comparison. Intermediate specialist notes will support workflow analysis rather than count as additional independent planning results.

We will apply the same feasible-plan success measure, challenge-category measures, evidence review, and paired human-review process used for the earlier versions.

A run that fails before producing a final response will remain included in the relevant outcome summaries.

### 27.10 Verify the Workflow and Measure Its Cost

We will use development cases to check:

- Whether each specialist stays within its assigned role.
- Whether source references remain traceable through the notes.
- Whether uncertainty survives the handoff to the coordinator.
- Whether the coordinator preserves important profile constraints.
- Whether conflicting suggestions are handled transparently.
- Whether the optional routing rule behaves consistently.
- Whether context limits preserve essential evidence.
- Whether failure handling matches the recorded policy.

We will measure reference preparation, every specialist call, coordination, parsing, retries, and failures.

Run records will include role and prompt versions, reference-packet identifiers, raw contributions, final responses, timings, and token usage where available.

Version 3 adds both role decomposition and additional model computation. We will interpret its results alongside the complete runtime and resource usage rather than attribute every difference solely to specialization.

### 27.11 Freeze the Workflow and Save Its Outputs

Before final evaluation, we will freeze the role prompts, routing rule, execution order, context limits, call budgets, and failure policy.

The model and reference versions will remain consistent with the shared configurations established earlier.

Final evaluation on the 200 frozen profiles will take place in Step 33.

We expect to prepare:

- `src/champion_athletic_map/planners/specialists.py`: the Version 3 workflow.
- `docs/prompts/planner_v3/`: the specialist and coordinator prompt templates.
- `artifacts/planners/version3_config.json`: role, routing, generation, and access settings.
- `results/planner_v3/development_agent_outputs.jsonl`: specialist contributions and handoff records.
- `results/planner_v3/development_runs.jsonl`: final responses and complete execution metadata.
- `results/planner_v3/development_scores.csv`: independent development scores.
- `docs/planner_v3.md`: workflow decisions and limitations.

We will add actual findings after implementation.

## 28. Planner Version 4 — Specialist Agents with an Evaluator and Automated Checks

### 28.1 Purpose

We will extend the specialist workflow by checking its draft and allowing a limited number of revisions.

Version 4 will combine coded checks for conditions we can verify directly with a language-model evaluator for issues that require interpreting the response.

We will test whether this process improves the final output and whether any improvement justifies the additional calls, runtime, and complexity.

Internal approval will describe the outcome of our configured checking procedure. It will not establish medical safety or real-world training effectiveness.

### 28.2 Start with the Version 3 Drafting Procedure

We will generate the initial draft using the same specialist roles, coordinator instructions, reference packet, and shared model settings as Version 3.

The initial workflow will contain:

- The training specialist.
- The participation and recovery specialist.
- The optional food-facts specialist.
- The coordinator.

The evaluator will then inspect the generated coordinator response.

This will distinguish its role from the participation specialist, which supplies guidance before the weekly plan is drafted.

We will reuse the original specialist notes during revisions. We will not restart the specialists or introduce additional searches within the initial Version 4 design.

### 28.3 Run the Automated Checks

We will implement checks using the reviewed rules, approved catalog metadata, and shared output requirements.

| Check area | What we will assess |
|---|---|
| Output structure | Required fields, types, and conditional response requirements |
| Schedule consistency | Durations, overlapping activities, availability, and existing commitments |
| Resource requirements | Equipment, space, and participant requirements where documented |
| Participation guidance | Applicable numerical limits and rest conditions |
| Reference identifiers | Whether supplied IDs correspond to available records |
| Food facts | Whether reported values, units, reference amounts, and calculations match the structured data |

These checks will cover conditions that can be assessed from the available structured information.

An existing source ID will establish that a record exists. Whether it supports a particular claim will require a separate assessment.

Every check will preserve its rule ID, applicability, observed information, result, and explanation.

The check results will use the statuses established in Step 13: met, violated, unknown, and not applicable.

### 28.4 Distinguish Required Conditions from Informational Findings

We will define the required checks, issue categories, and severity rules during development.

A violated required condition will prevent internal acceptance. An unknown required condition will also remain unresolved.

Informational gaps will be treated according to their relevance to the request. Missing annual participation or actual sleep history, for example, will not automatically make every weekly plan fail.

Applicability will come from the public rule definitions and supplied profile facts. The model’s declared response type will not, by itself, establish that requirements are not applicable.

Our internal checks will not read benchmark-specific expected behaviors, scenario labels, or answer keys.

We will keep conclusive coded findings separate from language-model judgments. The evaluator will not be allowed to waive a numerical conflict or convert a checker execution error into a pass.

### 28.5 Define the Evaluator’s Role

The evaluator will receive:

- The original model-visible athlete profile.
- The current candidate response.
- The same approved reference packet.
- The coded check results.
- The predefined evaluation instructions.

It will examine issues such as goal relevance, completeness, contradictory instructions, unsupported claims, and appropriate handling of uncertainty.

Its findings will use a compact structure containing:

- The affected field or activity.
- The issue category.
- A brief description of the observed problem.
- The supporting rule or source, where available.
- Whether the issue requires correction under our policy.
- A requested correction or clarification.

We will ask for actionable findings rather than an unsupported overall quality score.

The evaluator will use the same base model as the other roles. Its judgments may be incorrect, so we will preserve them for review rather than treat them as verified ground truth.

### 28.6 Apply a Bounded Revision Procedure

Our initial revision limit will be two coordinator revisions.

For each run, we will:

1. Generate the initial specialist-based draft.
2. Parse the candidate and run the coded checks.
3. Ask the evaluator to assess that candidate.
4. Accept it internally if the required conditions are met and no required evaluator issue remains.
5. Otherwise, give the coordinator the current candidate, findings, original profile, reference packet, and specialist notes.
6. Ask the coordinator to produce a complete revised response.
7. Check and evaluate the revised candidate again.
8. Stop when acceptance is reached, the revision limit is reached, or an execution error prevents completion.

Every revised candidate will receive its own checks and evaluation before the run ends.

A malformed candidate may receive a recorded format finding and use an available revision. Parsing will not repair it silently.

The coordinator will retain the original profile facts and source qualifications. Feedback will not authorize it to invent missing information or change existing commitments.

We will preserve the full revision history in the logs while keeping each revision prompt focused on the current candidate and relevant findings.

### 28.7 Define the Model-Call Budget

The initial drafting procedure will use three calls, or four when the food-facts role is active.

| Stage | Maximum calls without food role | Maximum calls with food role |
|---|---:|---:|
| Initial specialist and coordinator workflow | 3 | 4 |
| Evaluator assessments | 3 | 3 |
| Coordinator revisions | 2 | 2 |
| **Maximum logical model calls** | **8** | **9** |

The three possible evaluator assessments correspond to the initial draft and the two revised candidates.

If the first candidate is accepted, the run will use four calls, or five with the food-facts role.

Reference preparation, coded checks, and transport attempts will be recorded separately. Any permitted retry will follow the shared policy and count toward actual resource usage.

We will confirm these budgets during the development pilot in Step 31 and freeze them before final evaluation.

### 28.8 Define Terminal States and the Returned Response

Each run will finish with an explicit workflow status.

| Status | Meaning |
|---|---|
| Internally accepted | The candidate satisfies the configured acceptance procedure |
| Revision limit reached | The allowed revisions are exhausted and issues remain |
| Execution error | A required generation, evaluator, or checking stage cannot complete |

We will return the first internally accepted candidate.

If the revision limit is reached, we will retain the last checked candidate as the research response and attach its unresolved status and findings.

We will not select an earlier candidate using external benchmark scores.

For an execution error, earlier drafts will remain available for diagnosis. We will record the workflow as failed rather than silently substitute a completed Version 3 response.

In the application, an unresolved candidate will remain identifiable as a draft requiring attention. Its presentation will not imply that it passed the configured checks.

### 28.9 Preserve Independent Research Evaluation

The internal evaluator and automated checks will support generation. They will not determine the final research scores.

The independent evaluation process will assess the returned response using the same protocol applied to the other versions.

A revision-limit result will remain in the experiment and receive independent assessment. An execution failure will remain a failed case in the relevant completion measures.

All 150 feasible profiles will remain in the feasible-plan success-rate denominator.

We will report internal acceptance separately from independent performance. We will also examine cases where revisions introduce new errors or where the evaluator raises an incorrect concern.

Final human review will use the anonymous final responses. Internal self-ratings will not become human-review scores.

### 28.10 Verify the Revision Workflow During Development

Our development checks will include:

- A draft that already meets the required conditions.
- A correctable schedule or resource conflict.
- A missing required field or malformed response.
- A required condition that remains unknown.
- An informational gap that should not block a weekly plan.
- A disagreement between the evaluator and a conclusive coded result.
- A revision that introduces a new problem.
- A run that exhausts its revision budget.
- A checker, evaluator, or revision-call failure.

We will verify that every revised candidate is checked, call counts respect the limits, and original profile information remains unchanged.

We will also inspect whether the evaluator’s findings are specific enough to support useful revisions.

These checks will use development cases. Final profiles and expected responses will remain outside the workflow.

### 28.11 Record Outcomes and Freeze the Configuration

Run records will preserve the initial draft, specialist contributions, coded findings, evaluator outputs, revisions, terminal status, and returned response.

We will measure the complete workflow, including reference preparation, all model calls, coded checks, retries, failures, and total elapsed time.

Before final evaluation, we will freeze the prompts, rule versions, issue policy, acceptance conditions, revision budget, and returned-candidate policy.

- `src/champion_athletic_map/planning_checks.py`: the coded checks.

- `results/planner_v4/development_revision_history.jsonl`: candidates, findings, and revision events.
- `results/planner_v4/development_runs.jsonl`: execution records and terminal outcomes.

## 29. Prototype Application, Athlete Intake, Editable Plans, and Progress Checklist

### 29.1 Purpose

We will build a prototype that lets users enter an athlete profile, generate a weekly basketball plan, review its sources, make changes, and record completed activities.

Our initial application will use Streamlit and reuse the Python modules developed in earlier steps.

The main planning action will run Version 4. The four-version research comparison will remain in the separate experiment runner.

We will use approved fictional development profiles for testing and portfolio demonstrations.

### 29.2 Organize the User Flow

We will keep the application focused on a small number of clear actions.

| Area | What the user can do |
|---|---|
| Athlete profile | Enter or update planning information |
| Weekly plan | Generate and inspect a response |
| Plan editor | Adjust the proposed schedule |
| Review | Check the saved plan and understand unresolved issues |
| Progress checklist | Record completed, pending, and skipped activities |
| Save and restore | Download a snapshot or reopen a previously saved one |

We will use plain labels and short explanations.

Source links and relevant limitations will be available alongside the plan. Technical execution details will remain in development records rather than dominate the user interface.

### 29.3 Build the Athlete Intake Form

We will collect the information needed for our planning task.

| Information | Planned fields |
|---|---|
| Profile basics | A non-identifying label, age, and school grade |
| Basketball background | Experience, skill level, and position where relevant |
| Goals | Selected basketball skills or general conditioning priorities |
| Planning period | The actual week being planned |
| Availability | Available days, time windows, and session limits |
| Existing commitments | Practices, games, other sports, and their durations |
| Resources | Equipment, space, court or hoop access, and training partners |
| Additional request | Relevant constraints, questions, or information needing clarification |

We will preserve the distinction between an unknown value and a confirmed absence.

For example, no existing practices and an unknown practice schedule will remain different inputs.

We will validate field types and impossible values while preserving legitimate uncertainty and conflicting commitments.

Our intake will not infer grade from age or assume that an athlete has equipment because of their experience level.

Loading a demonstration profile will be an explicit action. It will not silently replace information the user has already entered.

### 29.4 Make Generation an Explicit Action

We will submit the intake through a form and start generation only when the user selects the planning action.

Changing a field, opening a section, downloading a file, or updating a checklist will not automatically start another planning run.

While a request is active, we will show its progress and prevent duplicate submissions within the session.

If generation fails, we will preserve the submitted inputs and provide a clear explanation of the outcome.

We will distinguish a valid clarification or conflict response from an execution failure. When the planner needs more information, the application will show the relevant questions rather than an empty weekly table.

### 29.5 Display the Plan and Its Current Status

For a planning response, we will display the week in a readable schedule.

Each activity will show its day, allocated duration, relevant instructions, resources, and available source links.

Existing commitments and rest arrangements will remain visible so users can understand the complete week.

We will present the workflow status in plain language, including:

- Checks completed.
- Changes need review.
- More information needed.
- A conflict remains.
- Generation could not complete.

Unresolved drafts will retain their outstanding issues.

We will explain what was checked rather than present internal acceptance as a guarantee about the plan.

Optional food facts will show the selected food, quantity or reference basis, nutrient units, and source.

### 29.6 Preserve the Original and Support Controlled Editing

We will keep the returned Version 4 response separate from the working copy used for edits.

The initial editor will support changes to scheduling and allocated duration, together with activity substitutions from the approved catalog.

Source identifiers, original source instructions, and verification metadata will remain controlled by the application.

If an activity is replaced, we will attach the replacement’s own source information rather than retain an unrelated citation.

We will preserve the distinction between time allocated by the user or planner and a duration prescribed by a source.

Each saved edit will create a new plan version with stable activity identifiers and an edit record.

We will use those identifiers rather than table row positions to connect activities, sources, and checklist entries.

### 29.7 Recheck Changes and Invalidate Outdated Assessments

When a user saves a planning edit, we will rerun the relevant coded checks.

The earlier language-model assessment will become outdated because it evaluated a different plan version.

We will provide a separate review action that assesses the saved edited plan using its current profile and relevant references. That action will report findings without silently rewriting the user’s changes.

A fresh assessment will use the same acceptance definitions established in Step 28.

If the user requests a new generated plan, we will run the bounded Version 4 workflow and preserve the earlier version.

Changes to planning facts, such as availability or team commitments, will also mark the existing plan as belonging to an earlier profile version.

Every assessment will be attached to the exact profile, plan, and rule versions it examined.

### 29.8 Add a Progress Checklist

We will let users mark trackable activities as pending, completed, or skipped.

Checklist records will include the activity ID, plan version, reported status, and update time.

We will show completed and skipped counts together with the total number of trackable activities. Rest days will not appear as missed training sessions.

Skipped activities will remain visible in the total rather than disappear from the completion summary.

Checklist updates will not change the planned schedule, invalidate its assessment, or trigger a model call.

When an activity is replaced or substantially changed, we will preserve its previous history and reset progress for the changed activity.

These records will describe self-reported completion. They will not establish training quality or athletic improvement.

### 29.9 Manage Session State and Model Resources Separately

We will use session state for the current profile, returned response, editable plan, review status, and checklist.

Streamlit session state is temporary and may reset when the browser connection is replaced. We will provide a clear save-and-restore route rather than promise permanent account history.

We may cache loaded model resources and versioned reference artifacts to avoid unnecessary loading.

Profiles, conversations, generated plans, and checklist records will remain separate from globally shared model resources.

Our initial local prototype will allow one active generation at a time against a shared model resource.

We will ensure that ordinary interface reruns do not repeat completed model requests or carry another session’s information into a prompt.

### 29.10 Export and Restore a Coherent Snapshot

Our initial downloads will include a readable schedule in CSV format and a structured JSON snapshot.

The snapshot will contain:

- The minimal planning profile.
- The selected plan version.
- Activity identifiers and source references.
- Current check results and review status.
- Relevant edit history.
- Checklist progress.
- Data and application version information.

Credentials, benchmark expectations, and internal reference collections will remain outside the export.

When restoring a JSON snapshot, we will validate its structure and resolve identifiers against the available reference versions.

We will recompute the coded checks. An imported assessment will remain historical until its status has been verified for the restored content and current configuration.

A restored file will not automatically start generation or overwrite an active plan.

### 29.11 Keep Application Activity Separate from Research Results

The application will use the shared planning modules, but its interactive sessions will remain distinct from the controlled experiment.

The experiment runner will preserve the model’s generated response before user edits.

Edited plans and checklist events will not replace benchmark outputs, become new evaluation labels, or automatically enter model training.

Our demonstration profiles will come from approved development examples rather than the held-out final set.

This separation will let us demonstrate an interactive product while keeping the research comparison reproducible.

### 29.12 Verify the Complete User Flow

We will check that a user can:

- Load a fictional example or enter a profile.
- Generate a response without duplicate model requests.
- Understand clarification, conflict, and failure outcomes.
- Inspect a plan and its supporting sources.
- Save an edit and see that review is needed.
- Request a fresh assessment of the edited version.
- Update progress without regenerating the plan.
- Download and restore a consistent snapshot.
- Reset the session without mixing old and new profile information.

We will also check that changing planning facts invalidates outdated assessments and that replaced activities do not inherit unrelated completion records.

Step 30 will test these interactions alongside the backend integration and constraint checks.

## 30. Integration Tests, Invalid Requests, and Planning Constraint Checks

### 30.1 Purpose

We will consolidate the checks introduced in earlier steps and test how the project’s components work together.

Our tests will cover input handling, reference access, planner execution, constraint calculations, application state, and failure handling.

We will distinguish software correctness from model quality. A workflow can execute correctly while producing an inaccurate recommendation, and both outcomes need to remain visible.

### 30.2 Organize the Testing Approach

We will use a small set of complementary test types.

| Test type | Main purpose |
|---|---|
| Focused function tests | Verify calculations, applicability rules, parsing, and state updates |
| Integration tests | Verify communication between reference tools, planners, checks, and records |
| Application tests | Verify supported user interactions and session behavior |
| Manual browser checks | Inspect editing, downloads, restoration, and other interactions needing direct inspection |
| Real-backend checks | Verify that the selected model and runtime work with the integration |

Most routine tests will use controlled model responses and small reference fixtures.

These substitutes will let us reproduce specific outcomes without repeatedly downloading models or making expensive generation calls.

A few development-only runs will exercise the actual backend. We will report those observations separately from substitute-based test results.

### 30.3 Prepare Independent Test Fixtures

We will create small, versioned examples containing profiles, reference records, candidate plans, tool results, and expected outcomes.

Expected results will come from reviewed requirements and hand-worked examples. We will not calculate a test’s expected answer using the same function being tested.

For example, we will manually establish the total duration and expected conflict status of a small schedule before checking the implementation.

Fixtures will include successful and failing cases.

We will use dedicated examples and approved development records. The final 200 athlete profiles and held-out retrieval questions will remain outside development testing.

### 30.4 Test Input Handling and Invalid Requests

We will check that the application handles:

- Invalid field types and impossible values.
- Ages outside the prototype’s supported range.
- Dates outside the selected planning week.
- Negative or inconsistent activity durations.
- Missing essential planning information.
- Conflicting availability and existing commitments.
- Unknown equipment or participant availability.
- Ambiguous food descriptions.
- Requests outside the project’s planning scope.

We will distinguish invalid structure from a legitimate request requiring clarification.

Unknown information will not become a confirmed absence. An unknown practice schedule, for example, will remain different from a confirmed schedule containing no practices.

Application validation and benchmark handling will remain distinct. The research runner will preserve its planned challenging cases and record their outcomes rather than remove them through frontend filtering.

### 30.5 Test Planning Rules at Their Boundaries

For numerical rules, we will test values below, equal to, and above the relevant boundary.

The expected outcome will follow the documented operator and qualifications for that rule.

| Area | Important cases |
|---|---|
| Participation totals | Existing commitments combined with proposed sessions |
| Activity counting | Structured basketball, informal activity, and unrelated activity handled according to the reviewed definitions |
| Duplicate entries | The same commitment or drill is not counted twice |
| Rest arrangements | A claimed rest day conflicts with an existing basketball commitment |
| Rule applicability | Age and grade overlap, missing information, or unresolved interpretation |
| Resources | Required equipment, space, or participants are unavailable or unknown |
| Units | Minutes, hours, grams, and nutrient units remain consistent |
| Optional history | Missing sleep or annual participation information is handled according to its relevance |

A passing maximum-limit comparison may still coexist with a recommendation warning. We will preserve those separate findings.

We will also test that required unknowns remain unresolved and that not-applicable results have a valid applicability explanation.

A model’s chosen response label will not automatically waive relevant requirements.

### 30.6 Verify Reference and Model Boundaries

Our integration tests will check that:

- Version 1 receives no retrieved passages or lookup results.
- Versions 2–4 use the adopted reference collection and lookup configuration.
- Versions 1–3 receive no post-generation checking feedback.
- Retrieved qualifications remain attached to the relevant guidance.
- Optional source IDs may be absent under the shared schema.
- Supplied identifiers can be checked against the available records.
- Identifier existence remains separate from whether a claim is supported.

We will verify that reference indexing accepts approved source records and excludes athlete profiles, benchmark questions, answer keys, and generated plans.

Data-integrity checks will use manifests and identifiers to confirm group separation and version consistency.

We will also check that classifier inputs exclude target-derived fields, saved preprocessing matches its model, and catalog predictions do not overwrite benchmark labels.

### 30.7 Test the Version 4 Revision Procedure

We will use a scripted backend to reproduce the important execution paths.

These tests will include:

- Immediate internal acceptance.
- A correctable issue followed by a successful revision.
- A required unknown that remains unresolved.
- A malformed candidate that uses an available revision.
- An evaluator judgment that conflicts with a conclusive coded finding.
- A revision that introduces another problem.
- Exhaustion of the revision limit.
- Failure of a specialist, evaluator, revision call, or checker.

We will verify that each revised candidate receives another check and evaluation.

The initial drafting stage will follow its three-call or four-call contract. The complete Version 4 workflow will respect its maximum of eight or nine logical model calls.

Transport attempts will be tracked separately.

We will test that revision-limit candidates remain available for independent assessment and that execution failures are not silently replaced with Version 3 outputs.

Earlier diagnostic drafts will remain distinguishable from the returned response.

### 30.8 Test Application State and User Edits

We will test the state transitions defined in Step 29.

Important behaviors will include:

- Submitting a request starts one intended workflow.
- Ordinary widget changes do not start another generation.
- Duplicate submissions do not duplicate an active request.
- A failed request preserves the submitted inputs.
- Planning edits create a new version and rerun coded checks.
- Earlier evaluator assessments become outdated after substantive edits.
- Changes to planning facts invalidate the previous plan’s current status.
- Checklist updates make no model calls.
- Replaced activities do not inherit unrelated completion records.
- Independent sessions retain separate profile and plan information.

We will also verify that source-controlled fields remain protected during editing.

For JSON restoration, we will check that plan content and progress remain correctly associated with their identifiers. Current checks will be recomputed, and imported approval fields will not automatically establish a current assessment.

We will combine supported AppTest interactions with focused tests of the state helpers and a short manual browser checklist.

### 30.9 Test Failure Handling and Result Accounting

We will simulate unavailable references, missing model artifacts, malformed tool results, timeouts, context-budget failures, and backend errors.

We will check that each failure produces an explicit status and preserves the information needed to diagnose it.

We will also test the result aggregation using small, hand-worked examples.

These checks will confirm that:

- Failed feasible cases remain in the success-rate denominator.
- Inappropriate refusals do not count as completed feasible plans.
- Unknown required checks do not become passes.
- Challenge categories retain their separate denominators.
- Internal evaluator approval is not substituted for independent scoring.
- User-edited responses do not replace generated benchmark outputs.

Skipped or unrun checks will remain distinguishable from passed checks.

### 30.10 Run a Small Set of Real-Backend Checks

After the controlled tests pass, we will exercise the actual selected backend using a few development cases.

We will inspect:

- Model and tokenizer loading.
- Message formatting.
- Reference-context handling.
- Output parsing.
- Tool integration.
- Runtime and context limits.
- Error reporting and execution records.

We will assess observable behavior rather than require one exact wording or weekly schedule.

A content mistake will become a model-development finding. An integration defect, such as incorrectly formatted messages or missing context, will require a software correction.

These checks will establish what actually ran on the recorded environment. They will not replace the final model and planner evaluations.

### 30.11 Review Failures and Prepare for the Pilot

When a test reveals a defect, we will document the issue, correct its cause, and add a focused regression case where useful.

We will run affected tests after changes and complete the required routine suite before freezing the experiment.

Blocking defects in calculations, workflow access, termination, state isolation, and result accounting will need resolution before final runs.

We will record the code revision, dependency versions, executed tests, outcomes, and remaining limitations.

Step 31 will use the verified integration to assess runtime, resource usage, and the final experimental configuration.

## 31. Development Pilot, Runtime and Cost Checks, and Final Evaluation Protocol

### 31.1 Purpose

We will run a small development pilot before beginning the held-out evaluations.

The pilot will help us confirm that all four planner versions execute correctly, fit our available resources, and produce complete records for analysis.

We will use its findings to resolve development issues and finalize the operating settings.

We will then freeze the evaluation protocol before examining final test performance.

### 31.2 Select the Pilot Profiles

We plan to select 20 approved profiles from the development set.

The sample will include four profiles from each age, with two feasible planning requests per age.

| Scenario category | Planned pilot profiles |
|---|---:|
| Feasible planning requests | 10 |
| Missing essential information | 4 |
| Conflicting constraints | 3 |
| Requests outside our scope | 3 |
| **Total** | **20** |

The sample will include different experience levels, resource restrictions, commitment loads, and context lengths. At least two profiles will contain explicit food questions so we can exercise the optional specialist route.

We will select these cases before reviewing their pilot outputs.

### 31.3 Run All Four Versions on the Same Cases

Each of the 20 profiles will be processed by all four versions, producing 80 planned planner runs.

We will preserve the same model-visible inputs and common generation configuration.

Versions 2–4 will use the same reference preparation and approved reference collection. We will verify the reference-packet identifiers for matched cases.

We will run workflows sequentially on the recorded environment and balance their execution order across profiles.

Fresh conversations and workflow state will be used for each run.

If a development correction changes the configuration, we will retain the earlier records and repeat the affected pilot checks. We will not pool incompatible configurations into one comparison.

### 31.4 Count Planner Runs and Model Calls Separately

A planner run may contain several language-model calls.

| Version | Logical model calls per completed run |
|---|---|
| Version 1 | 1 |
| Version 2 | 1, with reference preparation performed separately |
| Version 3 | 3, or 4 when the food-facts role runs |
| Version 4 | 4–8, or 5–9 when the food-facts role runs |

Version 4’s count will depend on whether it requires revisions.

We will record actual calls, including the stage and purpose of each call. Transport attempts, warm-up activity, and failed executions will also remain visible.

The final comparison will contain 800 planner runs. We will estimate its model-call workload from the observed role usage and revision frequency.

### 31.5 Measure Runtime Consistently

We will separate one-time setup from request processing.

Setup measurements will include model loading and reference-index preparation where relevant.

Request measurements will include:

- Reference searches and structured lookups.
- Context assembly.
- Every generation call.
- Coded checks and evaluator calls.
- Revisions and retries.
- Total elapsed time and completion status.

For the primary timing comparison, we will keep model and index resources loaded and use a consistent application-level cache policy.

Reference contents will remain consistent for matched cases, while preparation time will be measured under the same policy for each applicable version.

We will record provider caching behavior where available.

Where local GPU execution requires synchronization for accurate timing, we will account for it in the measurement implementation.

We will summarize observed runtime with the sample count, median, range, and failure information. Pilot timing will remain an estimate based on a small development sample.

### 31.6 Check Context, Memory, and Operating Limits

We will record input and output token counts where available, including reference packets, specialist notes, candidate plans, and evaluator findings.

We will verify that the coordinator and revision stages retain the information needed to interpret the references.

Our checks will cover:

- Model and tokenizer revisions.
- Available hardware and execution device.
- Precision or quantization settings.
- Working context budget.
- Common output-token limits.
- Per-call and per-workflow time limits.
- Maximum reference and food lookups.
- Transport-retry policy.
- Revision and acceptance settings.

We will inspect available memory measurements and identify what each measurement represents. Unavailable measurements will remain explicitly unavailable.

We will also verify timeout and cancellation behavior. A timeout reported to the caller will not automatically establish that backend computation has stopped.

### 31.7 Project the Final Runtime and Cost

We will use observed usage to prepare a typical workload estimate and a maximum-budget scenario.

The projection will account for the final scenario distribution rather than simply multiply the unweighted pilot average.

It will include:

- The number of final cases in each category.
- Observed model calls and tokens by workflow stage.
- Optional food-role activation.
- Version 4 revision frequency.
- Time and usage consumed by failed attempts.
- Permitted retries.
- Separate setup and preparation costs.

For a hosted backend, we will use the selected provider’s dated published rates and identify the assumptions behind the estimate.

For local execution, we will report resource time and available usage measurements. We will not describe local computation as having zero operating cost.

Actual dollar costs will remain unavailable until we have a verified basis for calculating them.

If the projected workload does not fit our resources, we will adjust the common operating configuration during development and repeat the affected checks before freezing the protocol.

## 32. Final Exercise Classifier Evaluation and Error Analysis

### 32.1 Purpose

We will evaluate the frozen exercise classifiers on the held-out test partition.

This assessment will show how well the selected models predict reviewed body-part labels for exercises outside their training groups.

We will compare the majority-class baseline, selected logistic regression pipeline, and selected neural classifier.

After scoring, we will examine mistakes and limitations. These findings will inform our discussion and future work without changing the reported models.

### 32.2 Verify the Frozen Evaluation Inputs

Before generating predictions, we will verify the protocol manifest and the saved artifacts from earlier steps.

| Model | Artifact used for final evaluation |
|---|---|
| Majority-class baseline | The fitted baseline with its training-derived prediction |
| Logistic regression | The selected fitted TF-IDF and classifier pipeline |
| Neural classifier | The selected configuration’s best seed-42 checkpoint |

We will confirm that:

- All models use the same held-out exercise IDs.
- Test groups remain separate from training and validation groups.
- Targets match the frozen reviewed labels.
- The approved class list and label mapping are unchanged.
- Model, preprocessing, and split versions match the protocol.
- The neural checkpoint is the one selected before testing.

We will report the actual test record count, independent group count, and class coverage.

We will not assume that every original dataset category remains included after the eligibility review.

### 32.3 Generate Predictions Without Updating the Models

We will construct inputs from the exercise name and available description using the saved preparation rules.

The logistic regression pipeline will apply its fitted vocabulary and TF-IDF statistics.

The neural classifier will use its saved vocabulary, sequence settings, and label mapping. We will enable evaluation mode and disable gradient recording.

The majority baseline will retain the class selected from training labels.

Title-only exercises, unfamiliar vocabulary, and valid inputs producing all-zero TF-IDF features will remain in the evaluated population.

We will not expand the vocabulary, change truncation settings, introduce a new confidence threshold, or remove difficult records after inspecting predictions.

Each prediction will remain linked to its exercise ID, group ID, and model version.

### 32.4 Verify Prediction Completeness

Before calculating headline metrics, we will check that every model produced one valid prediction for every expected test record.

We will check for missing predictions, duplicate IDs, unexpected records, invalid class IDs, and incorrect label decoding.

An incomplete prediction set will be recorded as an execution problem. We will preserve the affected attempts and follow the correction policy established in Step 31.

We will not calculate a primary comparison using only the rows that happened to succeed.

Any necessary technical correction will be documented and verified using development checks, with affected comparisons rerun consistently.

A low score by itself will not justify changing the implementation or repeating the experiment.

### 32.5 Calculate the Final Metrics

We will use the definitions and conventions established in Step 17.

| Measure | Planned interpretation |
|---|---|
| Macro-F1 — primary measure | Average F1 across the fixed approved classes |
| Accuracy | Proportion of correctly classified test records |
| Per-class precision | How often predictions for a class are correct |
| Per-class recall | How many reference examples of a class are identified |
| Per-class F1 | Precision and recall summarized for each class |
| Support | Record and independent-group counts supporting each result |

We will use the same label list and `zero_division=0` policy for every model.

The results table will identify the evaluated artifact, data version, record count, group count, and metrics.

We will retain full numerical precision in the saved results and apply consistent rounding only for presentation.

If a class has no test support, we will identify its recall as unestimable. A numerical zero produced by our scoring convention will not be described as observed failure on examples that do not exist.

### 32.6 Examine Confusion Patterns

We will produce confusion matrices using the same class ordering for all models.

Rows will represent reference labels and columns will represent predictions.

We will save both:

- Raw counts, showing how many records fall into each outcome.
- Row-normalized values, showing the prediction distribution within each reference class.

Classes with no test examples will be clearly marked in the normalized presentation.

We will use these figures to identify common confusions and compare the models’ behavior. We will show class support alongside the plots so that a pattern based on very few examples is not overstated.

### 32.7 Review Errors and Difficult Inputs

Our error analysis will examine:

- Classes that are frequently confused.
- Exercises with a title but no description.
- Inputs with low vocabulary coverage.
- Short and long descriptions.
- Records describing related or multi-muscle movements.
- Cases where the learned models disagree.
- Errors concentrated within a small number of exercise groups.

We will record the exercise ID, reference label, predictions, relevant text, and review notes.

Naturally occurring muscle names will remain legitimate input features. Their presence will not automatically be treated as leakage.

If we retain prediction scores, we will preserve their meaning and class ordering. High scores will not be presented as calibrated confidence.

We will document how illustrative examples were selected. Newly identified subgroup analyses will be labeled exploratory and include their sample counts.

### 32.8 Handle Disputed Reference Labels Transparently

Some reviewed errors may suggest an ambiguous or incorrect reference label.

A model disagreement alone will not establish that the label is wrong. We will check the original source and document any independent review.

The primary results will retain the frozen benchmark targets.

If a correction is substantiated, we may prepare a separately versioned sensitivity analysis. That analysis will recompute results for every comparison model using the same corrected targets.

We will preserve the original results and explain what changed.

The corrected analysis will not silently replace the original benchmark or become an opportunity to retrain the evaluated models.

### 32.9 Preserve the Inputs for Paired Uncertainty Analysis

We will save aligned reference labels, exercise IDs, group IDs, and predictions for all three models.

Step 35 will apply the uncertainty procedure frozen in Step 31: 2,000 paired group-level bootstrap resamples, seed 42, and individual 95% percentile intervals.

Related records will remain together during resampling, and matched model predictions will be compared on the same sampled groups.

We will retain the approved class list throughout the calculations.

Small class support and limited independent groups will remain part of the interpretation. We will not describe an unstable estimate as strong evidence of superiority.

### 32.10 Interpret Results Within the Task’s Scope

We will report whether each learned classifier improves on the majority baseline and how the neural model compares with logistic regression.

We will also report differences between validation and test rankings.

The application’s selected classifier will remain fixed for the ongoing experiment. We will not switch models after seeing which one wins on the test set.

Our conclusions will concern body-part classification within the evaluated exercise population.

They will not establish youth suitability, transfer to basketball drills, or the effectiveness of the generated weekly plans.

Error-analysis findings will contribute to the Results and Limitations sections of our report.

If we prepare predictions for previously held-out catalog records after scoring, those outputs will remain separate from the frozen retrieval collection used in the experiment.

## 34. Independent Blind Plan Review and Reviewer Agreement

### Purpose

We will use independent human review to examine qualities that automated checks cannot fully assess, including whether a response addresses the athlete’s situation, provides practical instructions, and accurately uses its supporting evidence.

We will also measure how consistently our reviewers apply the rubric. Reviewer agreement will help us understand the reliability of our assessment, but agreement alone will not establish that a recommendation is correct or appropriate for real-world use.

### 34.1. Use the Selected Review Sample

We will use the review sample selected before the final experiment and recorded in our evaluation protocol.

The sample will contain 40 held-out athlete profiles:

| Scenario type | Number of profiles |
|---|---:|
| Feasible planning requests | 25 |
| Requests missing essential information | 5 |
| Requests containing conflicting constraints | 5 |
| Requests outside our application’s scope | 5 |
| **Total** | **40** |

Each age from 13 through 17 will have eight profiles: five feasible requests and one from each challenge category.

We will review the responses from all four planner versions for every selected profile. This will produce 160 response records, each reviewed independently by the same two teammates.

We will retain failed runs in the review sample. We will not replace them with successful responses or select only the strongest plans.

Because this sample includes a larger proportion of challenge cases than the full final benchmark, we will report findings by scenario type. We will not treat its pooled scores as representative of all requests.

### 34.2. Assign Reviewers and Explain Their Role

We will identify the two reviewers before final review begins. Both reviewers will assess every response separately using the same instructions and rubric.

Our reviewers will be members of the project team. We will describe them accurately as team reviewers rather than independent external coaches, clinicians, or youth training experts.

In this project, independent review means that reviewers will complete their initial assessments without seeing each other’s ratings or discussing individual responses.

The third teammate may help resolve disagreements after the original ratings have been saved.

### 34.3. Prepare Anonymized Review Materials

We will use the anonymized materials prepared in Step 33. Each response will have an identifier that does not reveal its planner version or scenario category.

For each review assignment, we will provide:

- The athlete’s original profile and request.
- The complete response, including any citations it contains.
- The common review rubric.
- Neutral access to the approved reference materials needed to check claims.

We will keep reference materials clearly separate from the generated response.

We will hide planner names, internal agent messages, automated scores, evaluator comments, runtime information, and expected scenario labels. We will keep the identity mapping outside the reviewer materials until the initial ratings are locked.

We will preserve the actual response content. We will not improve its wording, repair its recommendations, or remove citations to make the versions appear more similar.

If no response was produced, we will include a clearly identified missing-output record. If malformed response text exists, reviewers will receive that text without content repairs.

This process will conceal version labels, but it cannot guarantee complete blinding. A reviewer may infer a version from its wording, structure, or use of references. We will acknowledge this limitation.

### 34.4. Apply the Existing Review Rubric

We will use the rubric developed in Step 17 and finalized before the held-out experiment in Step 31.

Any reviewer practice and clarification of scoring rules will take place using development examples before the final rubric is frozen. We will not change the rubric after seeing which final planner version performs better.

We will score five dimensions using the following anchors:

| Dimension | 0 | 1 | 2 |
|---|---|---|---|
| Response suitability | Gives the wrong type of response or ignores the central issue. | Addresses the issue partly but misses an important requirement. | Provides an appropriate plan, clarification, explanation of conflicting constraints, or scope boundary. |
| Profile fit | Ignores or contradicts important profile information. | Uses some profile information but misses relevant goals or constraints. | Reflects the relevant goals, experience, schedule, and available resources. |
| Practicality and completeness | Is unusable or substantially incomplete for the request. | Is partly usable but leaves important gaps. | Provides understandable and sufficiently complete guidance for the request. |
| Evidence accuracy | Contains materially unsupported or misleading claims. | Has partial support but includes important omissions or qualifications. | Agrees with the reviewed evidence and communicates relevant uncertainty. |
| Clarity | Contains confusing or inconsistent instructions. | Is understandable but contains noticeable ambiguity. | Provides clear and consistent instructions and explanations. |

We will judge completeness according to what the request needs. For example, an appropriate clarification response should identify the essential missing information rather than invent a complete weekly plan.

We will retain the five dimensions separately. We will not introduce a new overall score or passing threshold after reviewing the final responses.

### 34.5. Distinguish Failure, Non-Applicability, and Missing Ratings

We will keep these three situations separate:

| Rating status | Meaning | Treatment |
|---|---|---|
| Valid numeric rating | The dimension applies and has been assessed. | Record 0, 1, or 2. |
| Not applicable | The dimension does not apply under the frozen rubric. | Record a separate status and a reason. |
| Missing rating | The reviewer has not completed the assessment. | Return it to the original reviewer before unblinding. |

A score of 0 will represent an applicable failure. It will not mean that a rating is missing.

We will determine applicability from the original request and the frozen rubric. We will not let a planner avoid a poor score simply by choosing an inappropriate response type.

For example, if a feasible request receives no usable response, we will assign 0 for response suitability and practicality and completeness. We will assess the remaining dimensions according to their applicability rules. Evidence accuracy may be not applicable when there are no substantive claims to check.

We will record rating status explicitly so that an intentional “not applicable” entry cannot be confused with an unfinished CSV field.

### 34.6. Complete and Lock the Independent Ratings

Each reviewer will read the profile, examine the response, check relevant evidence, and record their own ratings.

We will save the following information:

- Reviewer identifier and anonymous response identifier.
- Dimension, numeric rating, and rating status.
- A reason for any not-applicable entry.
- Brief comments explaining important weaknesses or concerns.
- Source identifiers or locations used to check disputed claims.
- Rubric version and submission timestamp.

Reviewers will check whether the cited material actually supports the response. A valid source identifier will not automatically establish evidence accuracy.

When relevant, we will check whether the response preserves the source’s age group, units, conditions, and qualifications. We will also distinguish a model’s proposed activity allocation from a recommendation explicitly stated by a source.

We will verify that both reviewers have completed their assignments before comparing ratings. We will not fill missing ratings using another reviewer’s judgment or a language model.

We will preserve and lock the original independent ratings before any discussion.

### 34.7. Measure Reviewer Agreement

We will calculate agreement separately for each of the five dimensions.

For responses where both reviewers provide numeric ratings, we will report:

1. The number of jointly rated responses.
2. Exact agreement: the percentage receiving the same numeric score.
3. Linearly weighted Cohen’s kappa, using the fixed ordinal order 0, 1, and 2.

Weighted kappa will account for the difference between a one-point and a two-point disagreement. We will calculate it from the original independent ratings.

We will not treat “not applicable” as a fourth score. We will report separately how often both reviewers selected not applicable and how often only one reviewer did so.

We will also report missing-rating counts and each reviewer’s score distribution. These details will help us interpret agreement when most responses receive the same score or only a small number have applicable ratings.

If kappa is undefined, we will report it as unavailable and explain why. We will not replace it with zero or perfect agreement. For example, identical constant ratings can produce 100% exact agreement while leaving kappa undefined.

We will interpret agreement alongside reviewer comments and evidence checks rather than relying on a single threshold.

### 34.8. Discuss Disagreements Without Replacing the Original Evidence

After locking the initial ratings, we will compare disagreements and review the relevant comments and sources.

Where useful, the third teammate will help adjudicate a disagreement. We will keep any agreed assessment in a separate record linked to the original ratings.

We will calculate reviewer agreement from the independent ratings, not from the reconciled assessments. We will identify clearly whether any reported finding uses independent ratings or later adjudication.

If review reveals a possible benchmark or reference error, we will document it and follow our evaluation amendment policy. We will not silently change the expected answer or overwrite the original experiment.

### 34.9. Save the Review Results

We will save the review records under `results/human_review/`, including:

- `independent_ratings.csv`
- `review_completeness.csv`
- `agreement_by_dimension.csv`
- `rating_distributions.csv`
- `applicability_checks.csv`
- `adjudication.csv`, if adjudication is needed.

We will summarize the findings and limitations in `docs/human_review_findings.md`.

Our summary will retain the individual dimensions, show both reviewers’ assessments, and separate the scenario categories. We will preserve the profile and family mappings needed for paired analysis in Step 35.

### Completion Criteria

We will consider this step complete when both reviewers have assessed all 160 response records, missing entries have been resolved, the original ratings have been preserved, and agreement results have been calculated with their applicable denominators.

Our findings will describe response quality within this fictional benchmark. They will not establish that the application improves athletic performance or replaces professional coaching or clinical judgment.

Scikit-learn developers. (n.d.). *cohen_kappa_score*. Scikit-learn documentation. https://scikit-learn.org/stable/modules/generated/sklearn.metrics.cohen_kappa_score.html

## 35. Statistical Comparisons, Uncertainty, and Failure Analysis

### Purpose

We will bring together the results from our classifier, retrieval system, planner experiment, and human review.

We want to understand how much performance changes between versions, how uncertain those differences are, and where each approach fails. We will also examine whether any improvement justifies the additional runtime and resource use.

We will follow the evaluation protocol frozen in Step 31. We will not change our primary metric, comparison groups, or scoring rules after seeing the final results.

### 35.1. Verify the Analysis Inputs

We will use the saved results from Steps 32 through 34 rather than generate new responses during analysis.

Before calculating comparisons, we will confirm that:

- Classifier predictions match the frozen test examples and class labels.
- Every final profile has an accounted-for outcome for each planner version.
- Profile, family, exercise-group, and version identifiers are consistent.
- Automated results retain their original checks and failure statuses.
- Human ratings match the locked independent review records.
- Runtime and usage records include the attempts associated with each job.

We will distinguish an attempted job that failed from a job that has not been executed. An unexecuted job represents incomplete evaluation work. We will not invent its outcome or silently remove it from the planned experiment.

We will resolve missing records before presenting the final comparison, or clearly identify the evaluation as incomplete.

### 35.2. Keep Each Evaluation Measure Separate

We will organize the results around the following measures:

| Component | Main analysis | Important boundary |
|---|---|---|
| Exercise classifier | Test macro-F1, accuracy, and per-class performance. | Predicting an exercise category does not establish youth suitability. |
| Document retrieval | Held-out retrieval and evidence-coverage measures from Step 23. | Finding a relevant passage does not establish that the final response uses it correctly. |
| Feasible planning requests | Automated feasible-plan success across all 150 feasible profiles. | Passing coded checks does not establish overall recommendation quality. |
| Challenge requests | Results for missing information, conflicting constraints, and out-of-scope requests. | A matching response label alone does not establish an appropriate explanation. |
| Human review | Ratings for each dimension and reviewer, with agreement results. | Findings apply to the reviewed sample and team reviewers. |
| Runtime and resources | Recorded elapsed time, model calls, tokens, retries, and resource use. | Fast failure is not successful service. |

We will show the numerator and denominator for every reported rate.

For the primary planner metric, the denominator will remain all 150 feasible profiles. Recorded execution failures, inappropriate refusals, and unresolved required checks will remain unsuccessful outcomes.

### 35.3. Compare Planner Versions on the Same Profiles

We will use the four comparisons specified in our evaluation protocol:

| Comparison | What we will examine |
|---|---|
| Version 2 minus Version 1 | The change when we add retrieval and reference information. |
| Version 3 minus Version 2 | The change when we organize planning through specialist roles. |
| Version 4 minus Version 3 | The change when we add evaluation, automated checks, and bounded revision. |
| Version 4 minus Version 1 | The overall difference between the complete workflow and the language-model-only baseline. |

We will calculate each difference using matched outcomes from the same profiles.

For success rates, we will report differences in percentage points. We will also count profiles where:

- Both versions succeed.
- Only the comparison version succeeds.
- Only the later version succeeds.
- Neither version succeeds.

These paired counts will show whether a later version resolves earlier failures, introduces new failures, or does both.

We will describe these as comparisons between our implemented workflows. Because the workflows also differ in model-call counts and computation, we will not attribute every difference solely to one mechanism.

### 35.4. Estimate Uncertainty with Paired Group Bootstrap Resampling

We will use the procedure recorded in Step 31:

- 2,000 bootstrap resamples.
- Random seed 42.
- Paired resampling.
- Individual 95% percentile intervals.

In plain language, we will repeatedly sample evaluation groups with replacement and recalculate the comparison. The resulting distribution will show how sensitive the observed difference is to the mix of cases in our benchmark.

For planner analysis, we will resample profile families. We will keep related profiles together and retain all four versions’ outcomes for every selected profile.

For classifier analysis, we will resample exercise groups and retain all models’ predictions for the selected examples.

We will recalculate the same metric in every resample. For example, a profile-level success rate will remain a profile-level rate; we will not accidentally replace it with an equally weighted average of family rates.

We will report the number of independent groups supporting each analysis. We will follow the frozen sampling scheme rather than introduce stratification after inspecting the results.

Any necessary clarification or correction will be documented through our protocol amendment process. We will not choose a method because it produces narrower intervals.

### 35.5. Interpret Uncertainty Carefully

Our intervals will describe resampling uncertainty within this designed benchmark and its grouping structure.

They will not measure every source of uncertainty. In particular, they will not establish performance across real athletes, different training datasets, different model versions, or repeated generations.

We will report each paired difference with its interval. We will not decide whether versions differ by checking whether their separate intervals overlap.

The intervals will be individual intervals, not a simultaneous guarantee across all comparisons. We will avoid claims of statistical significance or universal superiority.

We will explain undefined or degenerate results. For example, an interval with no observed variation will not prove that the system has no possible failures.

We will not silently discard problematic resamples or replace unavailable intervals with zero uncertainty.

### 35.6. Analyze Classifier Performance

We will compare the majority-class baseline, logistic regression model, and neural network using their saved predictions on the same test examples.

Macro-F1 will remain our primary classifier metric. We will also examine accuracy, per-class precision and recall, support, and the confusion matrices from Step 32.

During bootstrap calculations, we will retain the frozen class list and zero-division policy. We will not remove a difficult class because it is absent from a particular resample.

We will report limitations caused by rare classes and small numbers of independent exercise groups.

We will distinguish test-sample uncertainty from variation caused by retraining a model. Resampling saved predictions will not measure training-seed variability.

We will preserve the application model selected through validation. A different ranking on the test set will become a reported finding rather than a reason to select a new model using held-out results.

### 35.7. Analyze Human Review and Subgroup Results

We will report the 0, 1, and 2 rating distributions separately for each dimension and reviewer. We will retain the agreement findings from Step 34 alongside these results.

Where we compare ratings between planner versions, we will use matched responses from the same profile and reviewer. We will show the number of applicable pairs and distinguish higher, equal, and lower ratings.

We will keep not-applicable entries and missing ratings separate. We will not introduce a new combined quality score or passing threshold.

We will examine results by age and scenario type, with the number of profiles and families shown for each subgroup.

Our human review sample deliberately includes more challenge cases than the full benchmark. We will therefore emphasize scenario-specific findings rather than treat its pooled scores as overall application performance.

Small subgroup findings will be descriptive. We will not present two reviewers or multiple responses from one profile as additional independent athletes.

### 35.8. Examine Runtime and Resource Tradeoffs

We will compare quality alongside the actual work required by each version.

We will summarize the recorded elapsed time, model calls, token usage, retries, and available resource measurements. We will retain resource use from failed attempts.

If we report response times only for completed jobs, we will label that result clearly and show which matched cases were included. We will also report failures separately so that a version that fails quickly does not appear preferable.

We will preserve the startup, caching, and timing boundaries defined in our protocol.

Where monetary cost can be calculated, we will document the applicable dated rates and assumptions. Missing usage information will remain unknown rather than being treated as zero.

We will discuss whether observed improvements appear worthwhile for our prototype without inventing a new weighted score that combines quality and cost.

### 35.9. Investigate Failures and Regressions

We will begin with the failure categories defined in our evaluation protocol.

Our investigation will distinguish:

- Execution, timeout, and resource failures.
- Invalid or incomplete output structures.
- Inappropriate response choices.
- Missed profile constraints or impractical instructions.
- Failed required planning checks.
- Missing or irrelevant retrieved evidence.
- Unsupported claims or misleading citations identified during review.
- Incorrect units or reference-data handling.
- Version 4 checks or revisions that miss a problem or introduce a new one.
- Exercise-classification errors.

We will allow multiple diagnostic labels when a case has several problems. We will explain that category percentages may therefore add to more than 100%.

Each finding will link to its case identifier, saved response, relevant checks, and reviewer comments where available.

We will distinguish an observed symptom from a confirmed cause. We will label newly discovered patterns as exploratory.

We will examine both improvements and regressions. We will preserve Version 4’s actual returned outcome rather than select an earlier draft because it looks better.

Human findings about evidence accuracy will apply to the responses actually reviewed. We will not extend them automatically to all generated plans.

### 35.10. Preserve Findings and Document Limitations

We will save the analysis under `results/statistical_analysis/`, including:

- `paired_comparisons.csv`
- `bootstrap_intervals.csv`
- `bootstrap_settings.json`
- `subgroup_summary.csv`
- `failure_analysis.csv`

We will summarize the findings in `docs/statistical_analysis_findings.md` and prepare clearly labeled figures for the report.

We will preserve original results if we discover a software bug, source error, or benchmark problem. Any corrected analysis will have a documented version and explanation.

We will use final findings to identify future improvements. We will not tune the current system on these held-out cases and then report the revised results as an untouched final evaluation.

### Completion Criteria

We will consider this step complete when our comparisons have traceable inputs, consistent denominators, paired uncertainty estimates, and documented failure patterns.

Our conclusions will explain the observed benefits, limitations, and resource tradeoffs of each workflow within our benchmark. We will not interpret these results as proof of improved athletic performance or professional validation of the recommendations.

## 37. Results, Limitations, and Future Improvements

### Purpose

We will bring our findings together and explain what they mean for the Champions Athletic Map prototype.

We will organize this discussion around our research questions, using the verified results from Steps 32 through 36. We will distinguish what we measured, how we interpret those measurements, and what still needs investigation.

At this planning stage, our results are pending. We will add numerical findings and conclusions after completing the experiments and checking the analysis.

### 37.1. Build a Traceable Results Summary

We will connect each research question to its evaluation measure, recorded result, and conclusion.

Our summary will cover:

| Area | What we will explain |
|---|---|
| Exercise classification | How the baseline and neural network models perform on held-out exercise examples. |
| Document retrieval | Whether the retrieval system finds the evidence needed for the evaluation questions. |
| Planner comparison | How the four workflows differ in automated planning success and challenge-case handling. |
| Human review | How reviewers assess response suitability, profile fit, practicality, evidence accuracy, and clarity. |
| Runtime and resources | What additional time and computation each workflow requires. |
| Application behavior | Which prototype functions and user interactions we have verified. |
| Reproducibility | Which results and procedures we have successfully reproduced. |

We will use the existing evaluation definitions. We will keep each measure separate rather than combine unrelated results into a new overall score.

Every numerical claim will link to a saved table, figure, or evaluation record.

### 37.2. Summarize the Exercise Classifier Results

We will report the majority-class baseline, logistic regression model, and neural network results on the same held-out test examples.

We will include:

- Macro-F1 as the primary measure.
- Accuracy and per-class precision, recall, and F1.
- Class support and the number of independent exercise groups.
- Important confusion patterns.
- The uncertainty estimates calculated in Step 35.

We will explain where the classifier performs well and where it struggles, including limitations associated with rare classes, missing descriptions, or ambiguous exercise names.

We will keep the classifier’s purpose clear: it predicts an exercise category from exercise text. Its performance will not establish whether an exercise is suitable for a particular teenager.

We will also distinguish classifier accuracy from its contribution to the complete planning application. A strong classification result alone will not demonstrate better personalized plans.

### 37.3. Summarize Retrieval Performance

We will report the held-out retrieval findings using the measures defined in Step 23.

We will distinguish the 20 answerable held-out questions from the 10 unsupported questions. Measures requiring relevant evidence will use the appropriate answerable-question denominator.

Our discussion will examine whether:

- Relevant evidence appears among the retrieved results.
- The delivered context contains the complete evidence needed.
- Important qualifications or linked guidance are missing.
- Unsupported questions receive the behavior defined in our protocol.

We will connect retrieval errors to planning failures only where the saved records support that connection.

We will treat retrieval performance and response evidence accuracy as separate findings. A system can retrieve a useful passage and still misinterpret or ignore it.

### 37.4. Summarize the Four Planner Versions

We will present each version’s automated feasible-plan success as a count and percentage out of all 150 feasible profiles.

We will retain recorded failures, inappropriate refusals, and unresolved required checks in that denominator.

We will report the four planned paired comparisons:

- Version 2 minus Version 1.
- Version 3 minus Version 2.
- Version 4 minus Version 3.
- Version 4 minus Version 1.

For each comparison, we will show the difference in percentage points, its individual uncertainty interval, and the paired success and failure counts from Step 35.

We will report missing-information, conflicting-constraint, and out-of-scope cases separately. These requests require different responses from a feasible planning request.

We will explain both improvements and regressions. If a later version resolves some problems but introduces others, our discussion will describe both findings.

Version 4’s internal acceptance status will remain separate from the independent final evaluation.

### 37.5. Summarize Human Review and Prototype Behavior

We will present the five human-review dimensions separately, showing each reviewer’s rating distribution and the applicable number of responses.

We will include the agreement findings from Step 34 and explain where reviewers reached different assessments.

Our discussion will connect ratings to the reviewer comments and source checks where appropriate. Agreement between reviewers will describe consistency, while evidence checks will address whether particular claims are supported.

We will preserve the distinction between independent ratings and later adjudication.

For the application, we will summarize verified behavior such as athlete intake, plan display, editing, rechecking, and progress tracking.

We will identify whether each application finding comes from a software test, a manual check, or a genuine model demonstration.

A functioning interface will demonstrate that the prototype operates as described. It will not, by itself, establish usability or effectiveness for real athletes.

### 37.6. Explain the Practical Tradeoffs

We will discuss the quality findings alongside recorded runtime, model calls, token usage, retries, and resource measurements.

We will examine whether the additional work required by specialist roles and evaluation produces a useful improvement in our benchmark.

Our conclusions will reflect the size of the observed differences and the problems they address. A ranking alone will not explain whether an improvement matters.

Where a simpler workflow performs similarly or better, we will report that result clearly.

We will describe observations and interpretations separately. For example, fewer recorded constraint violations would be an observed result. A proposed explanation involving the evaluator would remain an interpretation unless our experiment directly establishes that explanation.

### 37.7. State the Main Limitations

We will describe limitations that affect how our findings can be interpreted.

| Limitation | Effect on our conclusions |
|---|---|
| Fictional athlete profiles | We can evaluate responses to designed scenarios, but we do not observe real training participation or athletic outcomes. |
| Related profiles and repeated versions | The 800 planner jobs represent responses to 200 profiles, with family relationships accounted for in analysis. |
| Limited human-review sample | Human findings come from 40 profiles and deliberately oversample challenge cases. |
| Two project-team reviewers | Their assessments do not represent a wider population of coaches, clinicians, or users. |
| Imperfect blinding | Response structure and citations may reveal clues about the planner version. |
| Source coverage and preparation | Missing guidance, extraction errors, or incomplete metadata can affect retrieval and recommendations. |
| Imperfect evaluation labels | Manually prepared cases and expected behaviors may contain ambiguity or mistakes. |
| Rare exercise categories | Small class support can make classifier results unstable. |
| Tested model and hardware configuration | Findings may change with another model, backend, precision setting, or computing environment. |
| Different workflow budgets | Comparisons include differences in computation as well as planning structure. |
| Automated checks cover defined rules | Passing those checks does not establish that every recommendation is correct or suitable. |

We will also describe any unresolved data-provenance, permission, or reproducibility issues that remain at the end of the project.

Our conclusions will stay within the application functions and benchmark outcomes we actually evaluated.

### 37.8. Select Representative Examples Fairly

We will use examples to make the findings understandable.

We will document how examples were selected and include both successful responses and important failures. Examples will illustrate the broader results rather than replace them.

We will preserve the original generated text and identify any shortening used for presentation.

We will link each example to its profile, planner version, and evaluation record. We will avoid choosing only unusually strong outputs.

### 37.9. Prioritize Future Improvements from the Evidence

We will build our improvement list after reviewing the actual findings.

Possible directions include:

| Observed weakness | Possible future improvement |
|---|---|
| Missing basketball-specific evidence | Expand the permitted, verified drill and guidance collection. |
| Incorrect or incomplete evidence use | Improve retrieval, context preparation, or citation checking. |
| Poor handling of missing information | Improve clarification behavior and intake validation. |
| Repeated planning-constraint failures | Investigate the relevant rules, checks, and revision behavior. |
| Weak personalization | Improve how the workflow uses goals, schedules, and available resources. |
| Unstable classifier categories | Review labels and collect better examples for affected categories. |
| High runtime or resource use | Investigate fewer calls, smaller models, or more efficient execution. |
| Unclear recommendations | Seek structured feedback from intended users and qualified reviewers. |

These are candidate improvements, not commitments to add every feature.

For each selected improvement, we will record the supporting finding, expected benefit, required work, and proposed evaluation.

Changes informed by the final benchmark will belong to a future system version. We will evaluate them using a new protocol and fresh held-out cases.

Expert review and real-user usability studies may strengthen a later version. Commercial readiness and improved athletic performance will remain untested until appropriate additional work is completed.

### 37.10. Prepare the Findings for the Final Report

We will organize our synthesis into results, interpretation, limitations, and future improvements.

We will cite external sources where we discuss datasets, guidance, or methods. Our own experimental findings will link to the corresponding saved results.

We will keep the working synthesis in `docs/results_and_limitations.md`, including a prioritized future-improvement table.

We will prepare clear tables and figures with readable labels, stated denominators, and appropriate uncertainty information. Final template formatting and deliverable preparation will take place in Step 38.

### Completion Criteria

We will consider this step complete when each conclusion is linked to a recorded result, important limitations are explained, and future improvements follow from the evidence.

Our final discussion will give an honest account of what the prototype demonstrates, where it falls short, and what further evaluation would be needed.
