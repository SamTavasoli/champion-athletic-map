```python

from pathlib import Path

# Define our project folder
PROJECT_DIR = Path(
    "/content/gdrive/MyDrive/ChampionsAthleticMap"
)

# Define the folder for processed data
OUTPUT_DIR = PROJECT_DIR/"data"/"processed"

# Create the folder if it doesn't exist
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Define the CSV file path
OUTPUT_PATH = OUTPUT_DIR/"jr_nba_starter_practice1.csv"

# Save the DataFrame
drills_df.to_csv(OUTPUT_PATH, index=False)

print("Dataset saved to:", OUTPUT_PATH)
print("Total drills saved:", len(drills_df))
```

```python

# Review drills by participation requirement

participation_review = drills_df[
    [
        "drill_id",
        "drill_name",
        "group_requirement",
        "equipment"
    ]
].copy()

# Identify activities that are not currently marked Individual
participation_review["needs_review"] = (
    participation_review["group_requirement"] != "Individual"
)

display(participation_review)

print(
    "Drills requiring participation review:",
    participation_review["needs_review"].sum()
)
```

```python

# Proposed solo adaptations for our 13 drills
solo_adaptations = {
    "JRNBA001": (True, "Practice passing against a wall"),
    "JRNBA002": (True, "Practice stationary crossovers independently"),
    "JRNBA003": (True, "Dribble through cones independently"),
    "JRNBA004": (True, "Practice one-handed form shooting independently"),
    "JRNBA005": (True, "Perform court-line movements independently"),
    "JRNBA006": (False, "Chase game requires other players"),
    "JRNBA007": (False, "Original decision drill requires a defender and coach"),
    "JRNBA008": (True, "Shoot, rebound, and relocate independently"),
    "JRNBA009": (True, "Practice jump stops using predetermined court markers"),
    "JRNBA010": (True, "Practice defensive slides using fixed court markers"),
    "JRNBA011": (False, "3-on-2 game requires multiple players"),
    "JRNBA012": (True, "Perform timed dribbling runs independently"),
    "JRNBA013": (False, "3-on-3 game requires multiple players")
}

# Add the proposed metadata
drills_df["solo_adaptation_possible"] = (
    drills_df["drill_id"].map(
        lambda drill_id: solo_adaptations[drill_id][0]
    )
)

drills_df["solo_adaptation_note"] = (
    drills_df["drill_id"].map(
        lambda drill_id: solo_adaptations[drill_id][1]
    )
)

# Display the results
display(drills_df[[
    "drill_name",
    "group_requirement",
    "solo_adaptation_possible",
    "solo_adaptation_note"
]])
```

```python

# Validate the updated dataset
assert drills_df.shape == (13, 14)
assert drills_df["drill_id"].is_unique
assert drills_df.isna().sum().sum() == 0

# Count proposed solo adaptations
print("Total drills:", len(drills_df))

print(
    "Solo adaptations possible:",
    drills_df["solo_adaptation_possible"].sum()
)

print(
    "No direct solo adaptation:",
    (~drills_df["solo_adaptation_possible"]).sum()
)

# Save the updated dataset
OUTPUT_PATH = (
    PROJECT_DIR/"data"/"processed"
    / "jr_nba_starter_practice1.csv"
)

drills_df.to_csv(OUTPUT_PATH, index=False)

print("\nUpdated dataset saved to:", OUTPUT_PATH)
```
